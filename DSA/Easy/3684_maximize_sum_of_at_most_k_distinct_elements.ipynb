## 4. Ý tưởng nền tảng / cách nghĩ trực giác

Giả sử bạn nhặt các tờ tiền (toàn số dương), mỗi **mệnh giá chỉ được lấy một tờ**, tối đa `k` tờ. Muốn nhiều tiền nhất bạn nhặt tờ **mệnh giá cao nhất** trước, rồi cao thứ hai, ... Làm tay `nums = [84, 93, 100, 77, 93]`, `k = 3`:

```
các mệnh giá khác nhau : 77, 84, 93, 100
xếp giảm dần           : 100, 93, 84, 77
lấy 3 đầu              : 100, 93, 84     tổng 277
```

Hai nhận xét:

1. **Toàn số dương** -> thêm một phần tử luôn làm tổng **tăng**, nên nên chọn đủ `min(k, số giá trị khác nhau)` phần tử.
2. **Chọn lớn nhất trước:** nếu bỏ một giá trị lớn `a` để lấy giá trị nhỏ `b < a`, đổi lại `b` bằng `a` thì tổng chỉ tăng. Đây là **lập luận trao đổi** của thuật toán tham lam.

## 5. Cách 1 — Brute Force / baseline

Lặp `k` lần: mỗi lần tìm **giá trị lớn nhất chưa chọn** (`max` trên các giá trị còn lại), thêm vào kết quả rồi loại **mọi** bản sao của nó khỏi tập các giá trị còn lại. Dừng sớm nếu hết giá trị.

Nút thắt: mỗi lần `max` quét lại toàn bộ tập còn lại, tổng `O(k * n)`; và ta phải cẩn thận loại hết bản sao. Cách này đúng nhưng ta lặp lại việc "tìm lớn nhất" mà sắp xếp một lần giải quyết gọn hơn.

````python
class SolutionBrute:
    def maxKDistinct(self, nums, k):
        remaining = set(nums)            # mỗi giá trị một lần
        result = []
        for _ in range(k):
            if not remaining:
                break
            best = max(remaining)        # giá trị lớn nhất còn lại: O(n)
            result.append(best)
            remaining.remove(best)       # set nên đã loại hết bản sao
        return result

print(SolutionBrute().maxKDistinct([84, 93, 100, 77, 90], 3))
print(SolutionBrute().maxKDistinct([84, 93, 100, 77, 93], 3))
print(SolutionBrute().maxKDistinct([1, 1, 1, 2, 2, 2], 6))
````

**Độ phức tạp:** `n = len(nums)`. Time `O(n + k * n)` (dựng `set` rồi `k` lần `max`), Space `O(n)`.

## 6. Ý tưởng thuật toán / bước đột phá

**Bước đột phá:** "chọn `k` lớn nhất" = **sắp xếp một lần** rồi lấy `k` phần tử đầu — không cần `max` lặp lại. Kết hợp với `set` để khử trùng, ta có một dòng:

```python
sorted(set(nums), reverse=True)[:k]
```

**Quy tắc "nếu… thì…":**

- Nếu một giá trị xuất hiện nhiều lần -> chỉ tính **một** (dùng `set`).
- Nếu số giá trị khác nhau `< k` -> chọn **tất cả** (cắt lát `[:k]` tự xử lý).
- Kết quả đã sắp **giảm dần ngặt** vì các giá trị đã khác nhau và ta sắp giảm dần.

Invariant: danh sách `unique_desc[:i]` chứa `i` giá trị khác nhau lớn nhất, nên với mọi `i <= k` tổng là lớn nhất có thể ứng với `i` phần tử.

````python
def max_k_distinct(nums, k):
    return sorted(set(nums), reverse=True)[:k]

print(max_k_distinct([84, 93, 100, 77, 90], 3))
print(max_k_distinct([84, 93, 100, 77, 93], 3))
print(max_k_distinct([1, 1, 1, 2, 2, 2], 6))
````

## 7. Các cách tiếp theo

### Cách 2 — `sorted(set(...))` rồi cắt `k` (cách nộp bài)

Đúng ý tưởng mục 6. Dùng `set` khử trùng, `sorted(reverse=True)` sắp giảm dần, `[:k]` giữ tối đa `k` phần tử.

````python
class SolutionSortSet:
    def maxKDistinct(self, nums, k):
        return sorted(set(nums), reverse=True)[:k]

print(SolutionSortSet().maxKDistinct([84, 93, 100, 77, 90], 3), SolutionSortSet().maxKDistinct([1, 1, 1, 2, 2, 2], 6))
````

**Độ phức tạp:** Time `O(n log n)` (sắp xếp), Space `O(n)`.

### Cách 3 — Sắp xếp giảm dần rồi quét, bỏ qua phần tử trùng

Sắp **toàn bộ** `nums` giảm dần (các giá trị bằng nhau nằm liền nhau). Quét một lượt: lấy phần tử nếu **khác** phần tử vừa lấy; dừng khi đủ `k`. Cách này không cần `set` — chỉ cần so sánh với phần tử trước (kỹ thuật "bỏ trùng trong mảng đã sắp").

````python
class SolutionSortScan:
    def maxKDistinct(self, nums, k):
        result = []
        for v in sorted(nums, reverse=True):
            if len(result) == k:
                break
            if not result or v != result[-1]:     # chỉ lấy khi khác phần tử đã lấy gần nhất
                result.append(v)
        return result

print(SolutionSortScan().maxKDistinct([84, 93, 100, 77, 93], 3), SolutionSortScan().maxKDistinct([1, 1, 1, 2, 2, 2], 6))
````

**Độ phức tạp:** Time `O(n log n)`, Space `O(k)` cho kết quả (cộng bộ nhớ của `sorted`).

### Cách 4 — `heapq.nlargest` trên tập giá trị khác nhau

Khử trùng bằng `set`, rồi `heapq.nlargest(k, ...)` trả `k` phần tử lớn nhất theo thứ tự giảm dần. Time `O(n log k)` — tốt khi `k` nhỏ so với `n`.

````python
import heapq

class SolutionHeap:
    def maxKDistinct(self, nums, k):
        return heapq.nlargest(k, set(nums))

print(SolutionHeap().maxKDistinct([84, 93, 100, 77, 90], 3), SolutionHeap().maxKDistinct([1, 1, 1, 2, 2, 2], 6))
````

**Độ phức tạp:** Time `O(n + u log k)` với `u` là số giá trị khác nhau (`u <= n`), Space `O(n)` cho `set`, `O(k)` heap.

## 8. TRACE từng bước

Trace `nums = [84, 93, 100, 77, 93]`, `k = 3` bằng cách quét mảng đã sắp giảm dần (Cách 3).

Sắp giảm dần: `[100, 93, 93, 84, 77]`.

| Bước | v | result trước | `v != result[-1]`? | result sau |
|---|---|---|---|---|
| 1 | 100 | `[]` | rỗng -> lấy | `[100]` |
| 2 | 93 | `[100]` | `93 != 100` -> lấy | `[100, 93]` |
| 3 | 93 | `[100, 93]` | `93 == 93` -> bỏ | `[100, 93]` |
| 4 | 84 | `[100, 93]` | `84 != 93` -> lấy | `[100, 93, 84]` |
| 5 | 77 | `len == k` | dừng | `[100, 93, 84]` |

````python
def trace_scan(nums, k):
    arr = sorted(nums, reverse=True)
    print("sau khi sắp giảm dần:", arr, "| k =", k)
    result = []
    for step, v in enumerate(arr, start=1):
        if len(result) == k:
            print(f"bước {step}: đã đủ k={k} phần tử -> dừng")
            break
        if not result or v != result[-1]:
            result.append(v)
            action = "lấy"
        else:
            action = "bỏ (trùng)"
        print(f"bước {step}: v={v:>4} {action:<11} -> result={result}")
    print("kết quả:", result, "| tổng =", sum(result))
    return result

trace_scan([84, 93, 100, 77, 93], 3)
trace_scan([1, 1, 1, 2, 2, 2], 6)
````

## 9. Code nộp LeetCode

````python
from typing import List, Optional

class Solution:
    def maxKDistinct(self, nums: List[int], k: int) -> List[int]:
        return sorted(set(nums), reverse=True)[:k]
````

## 10. Vì sao thuật toán đúng

- **Khử trùng đúng đề:** các số được chọn phải khác nhau về **giá trị**. `set(nums)` chính là tập các giá trị có thể chọn, mỗi giá trị một lần; bất kỳ lựa chọn hợp lệ nào cũng là một tập con của nó.
- **Chọn nhiều hơn luôn tốt hơn:** mọi số đều dương nên tổng của một tập con thật sự lớn hơn tổng của tập con của nó. Vậy lời giải tối ưu dùng đủ `min(k, u)` phần tử (`u` = số giá trị khác nhau).
- **Lập luận trao đổi:** giả sử lời giải tối ưu chứa giá trị `b` nhưng bỏ sót một giá trị `a > b` (có trong `set`). Thay `b` bằng `a` vẫn hợp lệ (vẫn khác nhau) và tổng **tăng** (`a > b`), mâu thuẫn tính tối ưu. Vậy lời giải tối ưu là `min(k, u)` giá trị **lớn nhất**.
- **Thứ tự:** sắp giảm dần các giá trị đôi một khác nhau cho dãy **giảm ngặt**, đúng yêu cầu đầu ra.
- **`k` lớn hơn `u`:** `[:k]` cho toàn bộ danh sách, tương ứng "tối đa `k`" phần tử.

## 11. Phân tích Time / Space Complexity

`n = len(nums) <= 100`, `u` = số giá trị khác nhau, `k <= n`.

| Cách | Time | Space |
|---|---|---|
| Chọn `max` lặp `k` lần | `O(n + k n)` | `O(n)` |
| `sorted(set(...))[:k]` | `O(n log n)` | `O(n)` |
| Sắp xếp + quét bỏ trùng | `O(n log n)` | `O(k)` + sắp xếp |
| `heapq.nlargest` | `O(n + u log k)` | `O(n)` |

Với `n <= 100` mọi cách tức thì. Nếu `n` rất lớn và `k` nhỏ thì cách `heap` có lợi thế.

## 12. Bộ test

Ví dụ của đề, các biên (`k = 1`, mọi phần tử bằng nhau, `k = n`, giá trị tới `10^9`) và đối chiếu **tất cả cách**, kể cả **vét cạn mọi tổ hợp** để kiểm tra tính tối ưu.

````python
impls = [SolutionBrute(), SolutionSortSet(), SolutionSortScan(), SolutionHeap(), Solution()]

cases = [
    ([84, 93, 100, 77, 90], 3, [100, 93, 90]),
    ([84, 93, 100, 77, 93], 3, [100, 93, 84]),
    ([1, 1, 1, 2, 2, 2], 6, [2, 1]),
    ([5], 1, [5]),                          # n = 1
    ([7, 7, 7], 1, [7]),                    # mọi phần tử bằng nhau
    ([7, 7, 7], 3, [7]),                    # k = n nhưng chỉ có 1 giá trị
    ([1, 2, 3, 4, 5], 5, [5, 4, 3, 2, 1]),  # k = n, tất cả khác nhau
    ([10 ** 9, 1, 10 ** 9], 2, [10 ** 9, 1]),   # biên giá trị
    ([3, 1, 2], 1, [3]),                    # k = 1 -> phần tử lớn nhất
]
for nums, k, exp in cases:
    assert 1 <= len(nums) <= 100 and 1 <= k <= len(nums) and all(1 <= v <= 10 ** 9 for v in nums)
    for impl in impls:
        got = impl.maxKDistinct(nums[:], k)
        assert got == exp, (type(impl).__name__, nums, k, got, exp)
print("ví dụ + edge OK:", len(cases), "ca")
````

````python
import random
from itertools import combinations
random.seed(3684)

def exhaustive_best_sum(nums, k):
    vals = sorted(set(nums))
    best = 0
    for r in range(1, min(k, len(vals)) + 1):
        for comb in combinations(vals, r):
            best = max(best, sum(comb))
    return best

for _ in range(1500):
    n = random.randint(1, 9)
    nums = [random.randint(1, 8) for _ in range(n)]       # nhiều trùng lặp
    k = random.randint(1, n)
    results = [impl.maxKDistinct(nums[:], k) for impl in impls]
    assert all(r == results[0] for r in results)
    ans = results[0]
    assert len(ans) == min(k, len(set(nums)))
    assert all(a > b for a, b in zip(ans, ans[1:]))      # giảm dần ngặt
    assert set(ans) <= set(nums)
    assert sum(ans) == exhaustive_best_sum(nums, k)      # tổng tối ưu theo vét cạn
print("1500 test nhỏ + vét cạn tổ hợp (seed=3684) OK")
````

````python
# Test lớn đúng constraints: n <= 100, giá trị tới 10^9
for _ in range(300):
    n = random.randint(1, 100)
    hi = random.choice([5, 50, 10 ** 9])
    nums = [random.randint(1, hi) for _ in range(n)]
    k = random.randint(1, n)
    exp = sorted(set(nums), reverse=True)[:k]
    for impl in impls:
        assert impl.maxKDistinct(nums[:], k) == exp
print("300 test lớn OK")
# Không làm thay đổi input
nums = [84, 93, 100, 77, 93]
copy = nums[:]
Solution().maxKDistinct(nums, 3)
assert nums == copy
````

## 13. Các lỗi người mới hay gặp

### Lỗi 1: quên khử trùng

Sắp giảm dần `nums` rồi lấy `k` đầu sẽ chọn **cùng một giá trị hai lần**.

````python
def wrong_no_dedupe(nums, k):
    return sorted(nums, reverse=True)[:k]

nums, k = [84, 93, 100, 77, 93], 3
print(wrong_no_dedupe(nums, k), "!=", Solution().maxKDistinct(nums, k))
assert wrong_no_dedupe(nums, k) != Solution().maxKDistinct(nums, k)       # [100,93,93] != [100,93,84]
````

### Lỗi 2: trả thứ tự tăng dần

Đề yêu cầu **giảm dần**. Dùng `sorted(set(nums))[-k:]` cho đúng tập giá trị nhưng **ngược thứ tự**.

````python
def wrong_ascending(nums, k):
    return sorted(set(nums))[-k:]                         # SAI thứ tự: tăng dần

nums, k = [84, 93, 100, 77, 90], 3
print(wrong_ascending(nums, k), "!=", Solution().maxKDistinct(nums, k))
assert wrong_ascending(nums, k) != Solution().maxKDistinct(nums, k)       # [90,93,100] != [100,93,90]
````

### Lỗi 3: khử trùng **sau** khi đã cắt `k` phần tử

Cắt `k` trước rồi mới khử trùng thì mất cơ hội lấy các giá trị nhỏ hơn thay cho bản sao.

````python
def wrong_dedupe_after(nums, k):
    top = sorted(nums, reverse=True)[:k]
    return sorted(set(top), reverse=True)

nums, k = [84, 93, 100, 77, 93], 3
print(wrong_dedupe_after(nums, k), "!=", Solution().maxKDistinct(nums, k))
assert wrong_dedupe_after(nums, k) != Solution().maxKDistinct(nums, k)    # [100,93] != [100,93,84]
````

### Lỗi 4: cắt lát trực tiếp một `set` (không có thứ tự)

`list(set(nums))` có thứ tự **không xác định** (phụ thuộc băm). Lấy `[:k]` mà chưa sắp xếp thì không đảm bảo là `k` giá trị lớn nhất.

````python
def wrong_set_slice(nums, k):
    return list(set(nums))[:k]

nums, k = [84, 93, 100, 77, 90], 3
print(wrong_set_slice(nums, k), "!=", Solution().maxKDistinct(nums, k))
assert wrong_set_slice(nums, k) != Solution().maxKDistinct(nums, k)
````

## 14. Mẹo làm bài nhanh & nhận dạng pattern

- "Chọn tối đa `k` phần tử, **tổng lớn nhất**, toàn số **dương**" -> **tham lam: lấy `k` lớn nhất**. Nếu có số âm/không dương thì chỉ lấy các số dương.
- "**Khác nhau** về giá trị" -> `set(...)` trước, rồi mới chọn.
- "k lớn nhất" -> **sắp xếp giảm dần + cắt lát** (đơn giản) hoặc **`heapq.nlargest`** (khi `k` nhỏ).
- `[:k]` an toàn khi `k` vượt độ dài, rất hợp với "tối đa `k`".
- Đừng dựa vào thứ tự của `set`: luôn sắp xếp trước khi cắt.
- Khi bí, tự kiểm tra bằng **vét cạn tổ hợp** trên mảng nhỏ (cách làm trong bộ test).

## 15. Biến thể của bài

1. **Kth Largest Element in an Array** (LeetCode 215): phần tử lớn thứ `k` (có thể trùng).
2. **Third Maximum Number** (414): số lớn thứ ba **khác nhau**, nếu không có thì trả số lớn nhất.
3. Cho phép số **âm**: chỉ lấy các số dương (tối đa `k`), nếu không có số dương thì lấy số lớn nhất.
4. **Tổng lớn nhất của `k` phần tử khác nhau về giá trị** nhưng chỉ trả **tổng** (không trả mảng).
5. Mỗi giá trị được chọn **tối đa `c` lần** thay vì một lần (dùng `Counter` + tham lam).

## 16. So sánh các cách làm

| Cách | Ý tưởng | Time | Space | Khi nào dùng |
|---|---|---|---|---|
| Chọn `max` lặp | `k` lần lấy lớn nhất còn lại | `O(n + k n)` | `O(n)` | Dễ hiểu, làm bản đối chiếu |
| `sorted(set)[:k]` | khử trùng, sắp giảm, cắt | `O(n log n)` | `O(n)` | Cách nộp chuẩn |
| Sắp xếp + quét | bỏ trùng bằng so sánh với phần tử trước | `O(n log n)` | `O(k)` | Khi không muốn dùng `set` |
| `heapq.nlargest` | heap kích thước `k` | `O(n + u log k)` | `O(n)` | `k` nhỏ, `n` lớn |

Với giới hạn của đề, chọn `sorted(set(nums), reverse=True)[:k]`.

## 19. Đáp án tham khảo cho bài tập

````python
# Bài 1
assert Solution().maxKDistinct([5, 5, 4, 4, 3], 2) == [5, 4]

# Bài 2
assert Solution().maxKDistinct([9, 1, 9, 8, 2], 4) == [9, 8, 2, 1]

# Bài 3
def unique_desc(nums):
    return sorted(set(nums), reverse=True)
assert unique_desc([9, 1, 9, 8, 2]) == [9, 8, 2, 1]
assert unique_desc([7, 7]) == [7]

# Bài 4
def max_sum_k_distinct(nums, k):
    return sum(unique_desc(nums)[:k])
assert max_sum_k_distinct([84, 93, 100, 77, 90], 3) == 283
assert max_sum_k_distinct([84, 93, 100, 77, 93], 3) == 277
assert max_sum_k_distinct([1, 1, 1, 2, 2, 2], 6) == 3

# Bài 5
def max_k_distinct_signed(nums, k):
    pos = sorted({v for v in nums if v > 0}, reverse=True)
    if not pos:
        return [max(nums)]
    return pos[:k]
assert max_k_distinct_signed([3, -1, 3, 2], 5) == [3, 2]
assert max_k_distinct_signed([-5, -2, -9], 2) == [-2]
assert max_k_distinct_signed([0, -1], 1) == [0]
print("đáp án bài tập OK")
````
