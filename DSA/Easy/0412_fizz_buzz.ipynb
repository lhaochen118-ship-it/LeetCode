## 3. Cách trực giác (baseline) — và cái bẫy

Người mới thường viết theo đúng thứ tự đọc đề: 3 trước, 5 sau, 15 sau cùng. Hãy xem chuyện gì xảy ra.

````python
def fizz_buzz_naive_wrong(n: int) -> list:
    answer = []
    for i in range(1, n + 1):
        if i % 3 == 0:
            answer.append("Fizz")
        elif i % 5 == 0:
            answer.append("Buzz")
        elif i % 15 == 0:
            answer.append("FizzBuzz")
        else:
            answer.append(str(i))
    return answer

res = fizz_buzz_naive_wrong(15)
print(res)
print("phần tử số 15:", res[14], "-> sai, phải là FizzBuzz")
````

## 4. Nút thắt và insight

**Nút thắt:** số 15 thỏa **cả ba** điều kiện "chia hết cho 3", "cho 5", "cho 15". Với `if/elif`, nhánh đầu tiên đúng sẽ "ăn" mất, nên nhánh 15 không bao giờ chạy.

## 5. Lời giải chính — `if/elif` với thứ tự đúng

**Đoạn 1 — chuẩn bị:** `answer = []`.

**Đoạn 2 — duyệt từ 1 đến n:** `for i in range(1, n + 1):`

**Đoạn 3 — 4 nhánh, cụ thể nhất trước:**
```python
if i % 15 == 0:   answer.append("FizzBuzz")
elif i % 3 == 0:  answer.append("Fizz")
elif i % 5 == 0:  answer.append("Buzz")
else:             answer.append(str(i))
```

**Đoạn 4 — trả về** `answer`.

````python
def fizz_buzz_if(n: int) -> list:
    answer = []
    for i in range(1, n + 1):
        if i % 15 == 0:
            answer.append("FizzBuzz")
        elif i % 3 == 0:
            answer.append("Fizz")
        elif i % 5 == 0:
            answer.append("Buzz")
        else:
            answer.append(str(i))
    return answer

print(fizz_buzz_if(3))
print(fizz_buzz_if(5))
print(fizz_buzz_if(15))
````

## 6. TRACE từng bước

| i | i%3 | i%5 | Nhánh chạy | Thêm vào answer |
|:---:|:---:|:---:|---|---|
| 1 | 1 | 1 | else | `"1"` |
| 3 | 0 | 3 | `% 3` | `"Fizz"` |
| 5 | 2 | 0 | `% 5` | `"Buzz"` |
| 9 | 0 | 4 | `% 3` | `"Fizz"` |
| 10 | 1 | 0 | `% 5` | `"Buzz"` |
| 15 | 0 | 0 | `% 15` (đầu tiên!) | `"FizzBuzz"` |

````python
def trace_fizz_buzz(n: int) -> list:
    answer = []
    for i in range(1, n + 1):
        if i % 15 == 0:
            branch, val = "%15", "FizzBuzz"
        elif i % 3 == 0:
            branch, val = "%3", "Fizz"
        elif i % 5 == 0:
            branch, val = "%5", "Buzz"
        else:
            branch, val = "else", str(i)
        answer.append(val)
        print(f"i={i:>2} | i%3={i % 3} i%5={i % 5} | nhánh {branch:<4} | thêm {val!r}")
    return answer

_ = trace_fizz_buzz(15)
````

Cú pháp `branch, val = "%15", "FizzBuzz"` là **gán nhiều biến cùng lúc** (tuple unpacking). `_ = ...` là quy ước "kết quả này không dùng tới" — tránh notebook in thêm list dài.

## 7. Các lời giải khác

### 7.1. Cách 3 — ghép chuỗi (không cần nhánh 15)

```python
s = ""
if i % 3 == 0: s += "Fizz"
if i % 5 == 0: s += "Buzz"     # chú ý: if, KHÔNG phải elif
answer.append(s or str(i))
```

`s or str(i)`: nếu `s` là chuỗi rỗng (được coi là "False") thì lấy `str(i)`, ngược lại lấy `s`.

````python
print(repr("" or "7"), repr("Fizz" or "7"))

def fizz_buzz_concat(n: int) -> list:
    answer = []
    for i in range(1, n + 1):
        s = ""
        if i % 3 == 0:
            s += "Fizz"
        if i % 5 == 0:
            s += "Buzz"
        answer.append(s or str(i))
    return answer

print(fizz_buzz_concat(15))
````

### 7.2. Cách 4 — bảng luật (mở rộng được)

Nếu sếp bảo thêm luật "chia hết cho 7 → Jazz", cách if/elif phải viết thêm nhiều nhánh tổ hợp (21, 35, 105...). Cách ghép chuỗi tổng quát thành **danh sách luật** `[(3, "Fizz"), (5, "Buzz")]`: duyệt từng luật, chia hết thì nối từ.

````python
def fizz_buzz_rules(n: int, rules=((3, "Fizz"), (5, "Buzz"))) -> list:
    answer = []
    for i in range(1, n + 1):
        s = "".join(word for d, word in rules if i % d == 0)
        answer.append(s or str(i))
    return answer

print(fizz_buzz_rules(15))
print(fizz_buzz_rules(21, rules=((3, "Fizz"), (5, "Buzz"), (7, "Jazz")))[-1])   # 21 -> FizzJazz
````

### 7.3. Cách 5 — bộ đếm, không dùng `%`

Phép `%` tương đối "đắt" trên một số phần cứng. Ta giữ hai bộ đếm `c3`, `c5`, tăng mỗi bước; khi chạm 3 (hoặc 5) thì reset về 0 — đó chính là lúc chia hết.

````python
def fizz_buzz_counters(n: int) -> list:
    answer = []
    c3 = c5 = 0
    for i in range(1, n + 1):
        c3 += 1
        c5 += 1
        s = ""
        if c3 == 3:
            s += "Fizz"
            c3 = 0
        if c5 == 5:
            s += "Buzz"
            c5 = 0
        answer.append(s or str(i))
    return answer

print(fizz_buzz_counters(15))
````

### 7.4. Cách 6 — list comprehension một dòng

**List comprehension** `[biểu_thức for x in dãy]` tạo list mới gọn gàng. Ví dụ nhỏ trước:

````python
print([x * x for x in range(1, 6)])
print([str(x) for x in range(1, 4)])

def fizz_buzz_oneliner(n: int) -> list:
    return ["Fizz" * (i % 3 == 0) + "Buzz" * (i % 5 == 0) or str(i) for i in range(1, n + 1)]

print(fizz_buzz_oneliner(15))
print("Fizz" * True, "|", repr("Fizz" * False))   # True = 1, False = 0
````

`"Fizz" * (i % 3 == 0)`: `True` được coi như `1`, `False` như `0`; chuỗi nhân 0 là rỗng. Cách này ngắn nhưng khó đọc với người mới — khi phỏng vấn nên ưu tiên cách 1 hoặc 3.

## 8. Code nộp LeetCode

Bản chính dùng `if/elif` (thứ tự 15 → 3 → 5); bản phụ dùng ghép chuỗi. `List[str]` từ `typing` tương đương `list[str]` trong skeleton.

````python
from typing import List

class Solution:
    def fizzBuzz(self, n: int) -> List[str]:
        answer = []
        for i in range(1, n + 1):
            if i % 15 == 0:
                answer.append("FizzBuzz")
            elif i % 3 == 0:
                answer.append("Fizz")
            elif i % 5 == 0:
                answer.append("Buzz")
            else:
                answer.append(str(i))
        return answer


sol = Solution()
print(sol.fizzBuzz(3))
print(sol.fizzBuzz(5))
````

````python
class SolutionConcat:
    def fizzBuzz(self, n: int) -> List[str]:
        answer = []
        for i in range(1, n + 1):
            s = ""
            if i % 3 == 0:
                s += "Fizz"
            if i % 5 == 0:
                s += "Buzz"
            answer.append(s or str(i))
        return answer

print(SolutionConcat().fizzBuzz(15)[-1])
````

## 9. Vì sao đúng?

Xét một số `i` bất kỳ trong `1..n`. Có đúng 4 trường hợp **loại trừ nhau** theo `(3 | i, 5 | i)`:

| 3 chia hết i? | 5 chia hết i? | Đề yêu cầu | if/elif chọn nhánh | Ghép chuỗi cho |
|:---:|:---:|---|---|---|
| có | có | FizzBuzz | `% 15` (vì 15 \| i ⟺ 3 \| i và 5 \| i) | "Fizz"+"Buzz" |
| có | không | Fizz | `% 15` sai → `% 3` đúng | "Fizz" |
| không | có | Buzz | `% 15`, `% 3` sai → `% 5` | "Buzz" |
| không | không | str(i) | else | "" → `str(i)` |

Mọi trường hợp đều cho đúng giá trị. Vòng `for` thêm đúng **một** phần tử cho mỗi `i` theo thứ tự tăng ⇒ `answer[i-1]` ứng với `i`. ∎

## 10. Độ phức tạp

Gọi `n` là input.

| Cách | Time | Space (không tính output) |
|---|:---:|:---:|
| if/elif | `O(n)` | `O(1)` |
| Ghép chuỗi | `O(n)` | `O(1)` |
| Bảng luật với `k` luật | `O(n·k)` | `O(k)` |
| Bộ đếm | `O(n)` | `O(1)` |
| Comprehension | `O(n)` | `O(1)` |

Output có `n` chuỗi, mỗi chuỗi dài tối đa `max(8, số chữ số của n)` ⇒ bộ nhớ output `O(n)` (độ dài chuỗi bị chặn với `n <= 10^4`). Không thể nhanh hơn `O(n)` vì phải tạo `n` phần tử.


### Lỗi 3: `range(n)` thay vì `range(1, n + 1)`
Bắt đầu từ 0 (`0 % 3 == 0` → `"FizzBuzz"` cho số 0!) và thiếu số `n`.

### Lỗi 4: Append số thay vì chuỗi
`answer.append(i)` → `[1, 2, "Fizz"]` khác `["1", "2", "Fizz"]`.

````python
def wrong_elif_concat(n):
    out = []
    for i in range(1, n + 1):
        s = ""
        if i % 3 == 0:
            s += "Fizz"
        elif i % 5 == 0:
            s += "Buzz"
        out.append(s or str(i))
    return out

def wrong_range(n):
    return ["FizzBuzz" if i % 15 == 0 else "Fizz" if i % 3 == 0 else "Buzz" if i % 5 == 0 else str(i)
            for i in range(n)]

def wrong_int(n):
    return ["Fizz" if i % 3 == 0 else i for i in range(1, n + 1)]

print("lỗi 2:", wrong_elif_concat(15)[-1])
print("lỗi 3:", wrong_range(5))
print("lỗi 4:", wrong_int(3), "==", ["1", "2", "Fizz"], "?", wrong_int(3) == ["1", "2", "Fizz"])
````

- Nhiều điều kiện **chồng lấn** trong `if/elif` → đặt điều kiện **hẹp nhất** trước, hoặc biến thành các `if` **độc lập** và **gộp kết quả**.
- "Chia hết cho a và b" → chia hết cho `lcm(a, b)` (với 3 và 5 là 15; nhưng với 4 và 6 là **12**, không phải 24!).
- Bài mô phỏng (simulation) đơn giản: làm đúng **từng chữ** của đề, cẩn thận biên `1..n`.
- Muốn dễ mở rộng → dữ liệu hóa luật (bảng luật) thay vì viết cứng nhánh.

````python
from math import lcm
print("lcm(4, 6) =", lcm(4, 6), "| 12 % 4 =", 12 % 4, "| 12 % 6 =", 12 % 6)
````

````python
def count_labels(n: int) -> dict:
    fb = n // 15
    return {"FizzBuzz": fb, "Fizz": n // 3 - fb, "Buzz": n // 5 - fb,
            "number": n - (n // 3 + n // 5 - fb)}

print(count_labels(100))
r = Solution().fizzBuzz(100)
print({"FizzBuzz": r.count("FizzBuzz"), "Fizz": r.count("Fizz"), "Buzz": r.count("Buzz"),
       "number": sum(x.isdigit() for x in r)})
````

````python
def answer_fizz_buzz(n: int) -> list:
    answer = []
    for i in range(1, n + 1):
        s = ""
        if i % 3 == 0:
            s += "Fizz"
        if i % 5 == 0:
            s += "Buzz"
        answer.append(s or str(i))
    return answer

def answer_fizz_buzz_range(a: int, b: int) -> list:
    out = []
    for i in range(a, b + 1):
        s = ("Fizz" if i % 3 == 0 else "") + ("Buzz" if i % 5 == 0 else "")
        out.append(s or str(i))
    return out

def answer_fizz_buzz_jazz(n: int) -> list:
    rules = ((3, "Fizz"), (5, "Buzz"), (7, "Jazz"))
    return ["".join(w for d, w in rules if i % d == 0) or str(i) for i in range(1, n + 1)]

def answer_count_numbers(n: int) -> int:
    return n - (n // 3 + n // 5 - n // 15)

assert answer_fizz_buzz(15) == Solution().fizzBuzz(15)
assert answer_fizz_buzz_range(14, 16) == ["14", "FizzBuzz", "16"]
assert answer_fizz_buzz_range(1, 50) == Solution().fizzBuzz(50)
assert answer_fizz_buzz_jazz(105)[-1] == "FizzBuzzJazz" and answer_fizz_buzz_jazz(21)[-1] == "FizzJazz"
assert answer_count_numbers(15) == 8
for n in range(1, 300):
    assert answer_count_numbers(n) == sum(x.isdigit() for x in Solution().fizzBuzz(n))
print("Đáp án bài 3–6 PASS; n = 10^18 ->", answer_count_numbers(10 ** 18))
````
