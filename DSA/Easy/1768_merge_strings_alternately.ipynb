## 1. Bài toán yêu cầu gì?

Cho hai chuỗi `word1`, `word2`. Trộn chúng bằng cách lấy ký tự **xen kẽ**, **bắt đầu từ `word1`**. Nếu một chuỗi dài hơn, nối phần còn lại của nó vào cuối.

| word1 | word2 | Kết quả |
|---|---|---|
| `"abc"` | `"pqr"` | `"apbqcr"` |
| `"ab"` | `"pqrs"` | `"apbqrs"` |
| `"abcd"` | `"pq"` | `"apbqcd"` |

Constraints: `1 <= word1.length, word2.length <= 100`; chỉ chữ thường tiếng Anh.

```text
word1:  a   b   c   d
word2:    p   q
merged: a p b q c   d
```

````python
word1, word2 = "abcd", "pq"
print("độ dài:", len(word1), len(word2))
print("phần ghép xen kẽ dùng", min(len(word1), len(word2)), "cặp, phần dư:", word1[len(word2):])
````

## 2. Nền tảng Python cần biết

### 2.1. Chỉ số và độ dài chuỗi

`word[i]` là ký tự thứ `i`; hợp lệ khi `0 <= i < len(word)`. Đọc ngoài phạm vi → `IndexError`.

````python
w = "pq"
print(w[0], w[1])
try:
    print(w[2])
except IndexError as e:
    print("IndexError:", e)
````

### 2.2. `while` với hai điều kiện

`while i < len(a) and j < len(b):` chạy khi **cả hai** con trỏ còn hợp lệ. `or` thì chạy khi **ít nhất một** còn hợp lệ.

````python
i, j = 0, 0
a, b = "abc", "pq"
while i < len(a) and j < len(b):
    print("cặp:", a[i], b[j])
    i += 1
    j += 1
print("dừng tại i =", i, ", j =", j)
````

### 2.3. Slicing phần còn lại: `word[k:]`

`word[k:]` lấy từ chỉ số `k` đến hết. Nếu `k >= len(word)` thì trả `""` — **không lỗi**. Đây là điểm rất tiện để xử lý phần dư.

````python
print(repr("abcd"[2:]), repr("pq"[2:]), repr("pq"[5:]))
````

### 2.4. Cộng chuỗi vs `list` + `"".join`

Chuỗi Python **bất biến**: `res += ch` tạo chuỗi **mới** mỗi lần (về lý thuyết có thể tốn `O(độ dài)` mỗi lần). Cách chuẩn là gom vào list (`append` nhanh) rồi `"".join(list)` **một lần** ở cuối.

(CPython có tối ưu riêng cho `+=` trên chuỗi nên thực tế thường vẫn nhanh, nhưng không nên dựa vào điều đó.)

````python
parts = []
for ch in "abc":
    parts.append(ch)
    parts.append("-")
print(parts)
print("".join(parts))
````

### 2.5. `zip(a, b)` dừng ở chuỗi ngắn hơn

````python
print(list(zip("abcd", "pq")))
````

### 2.6. `itertools.zip_longest` và `fillvalue`

Ngược với `zip`, nó chạy tới chuỗi **dài hơn**, chỗ thiếu điền `fillvalue`.

````python
from itertools import zip_longest

print(list(zip_longest("abcd", "pq")))
print(list(zip_longest("abcd", "pq", fillvalue="")))
````

## 3. Cách trực giác — duyệt chỉ số đến độ dài lớn nhất

Người mới thường nghĩ: "Đi `i = 0, 1, 2, …` tới độ dài lớn nhất. Ở mỗi `i`, nếu `word1` còn ký tự thì lấy, rồi nếu `word2` còn ký tự thì lấy."

````python
def merge_index(word1: str, word2: str) -> str:
    res = []
    for i in range(max(len(word1), len(word2))):
        if i < len(word1):
            res.append(word1[i])
        if i < len(word2):
            res.append(word2[i])
    return "".join(res)

print(merge_index("abc", "pqr"), merge_index("ab", "pqrs"), merge_index("abcd", "pq"))
````

### 3.1. Một cách "vụng" hay gặp: cộng chuỗi và xử lý phần dư bằng `if` dài dòng

````python
def merge_clumsy(word1: str, word2: str) -> str:
    res = ""
    n = min(len(word1), len(word2))
    for i in range(n):
        res = res + word1[i] + word2[i]
    if len(word1) > len(word2):
        for i in range(n, len(word1)):
            res = res + word1[i]
    elif len(word2) > len(word1):
        for i in range(n, len(word2)):
            res = res + word2[i]
    return res

print(merge_clumsy("abc", "pqr"), merge_clumsy("ab", "pqrs"), merge_clumsy("abcd", "pq"))
````

## 4. Nút thắt và insight

**Nút thắt:**

- Cách "vụng" viết 3 vòng lặp, nhiều nhánh `if` → dễ sai chỉ số, khó đọc.
- Cộng chuỗi `res = res + …` tạo chuỗi mới liên tục.
- Cách duyệt chỉ số phải kiểm tra `i < len(...)` **hai lần mỗi vòng**, kể cả khi một chuỗi đã hết từ lâu.

**Insight (hai con trỏ):** quá trình có **hai pha rõ ràng**:

1. **Pha xen kẽ:** khi **cả hai** chuỗi còn ký tự, lấy `word1[i]` rồi `word2[j]`, tiến cả hai con trỏ.
2. **Pha đuôi:** khi một chuỗi hết, phần còn lại của chuỗi kia được nối **nguyên khối**: `word1[i:] + word2[j:]` — một trong hai là `""`.

Không cần `if` để biết chuỗi nào dài hơn: slicing ngoài phạm vi trả `""` tự động lo việc đó.

````python
i = j = 2   # giả sử pha xen kẽ dừng ở đây với word1="abcd", word2="pq"
print(repr("abcd"[i:]), "+", repr("pq"[j:]), "=", repr("abcd"[i:] + "pq"[j:]))
````

## 5. Lời giải chính — hai con trỏ

### 5.1. Khởi tạo

- `i`: con trỏ vào `word1`; `j`: con trỏ vào `word2`. Cả hai bắt đầu ở `0`.
- `res`: list chứa các ký tự kết quả.

````python
word1, word2 = "ab", "pqrs"
i, j, res = 0, 0, []
print(i, j, res)
````

### 5.2. Pha xen kẽ

Mỗi vòng: thêm `word1[i]` (vì bắt đầu từ `word1`), rồi `word2[j]`, rồi tiến cả hai.

````python
while i < len(word1) and j < len(word2):
    res.append(word1[i])
    res.append(word2[j])
    i += 1
    j += 1
print("sau pha xen kẽ:", res, "| i =", i, "j =", j)
````

### 5.3. Pha đuôi

````python
res.append(word1[i:])
res.append(word2[j:])
print("sau pha đuôi:", res, "->", "".join(res))
````

`"".join` chấp nhận phần tử là chuỗi dài hơn 1 ký tự (như `"rs"`) và chuỗi rỗng `""` — ghép như bình thường.

### 5.4. Ghép thành hàm

````python
def merge_two_pointers(word1: str, word2: str) -> str:
    i, j = 0, 0
    res = []
    while i < len(word1) and j < len(word2):   # pha xen kẽ
        res.append(word1[i])
        res.append(word2[j])
        i += 1
        j += 1
    res.append(word1[i:])                     # pha đuôi: một trong hai là ""
    res.append(word2[j:])
    return "".join(res)

print(merge_two_pointers("abc", "pqr"), merge_two_pointers("ab", "pqrs"), merge_two_pointers("abcd", "pq"))
````

## 6. TRACE từng bước

Trace cho `word1 = "abcd"`, `word2 = "pq"`:

| bước | i | j | điều kiện while | hành động | res |
|---:|---:|---:|---|---|---|
| 1 | 0 | 0 | 0<4 và 0<2 ✔ | thêm `a`, `p` | `a p` |
| 2 | 1 | 1 | 1<4 và 1<2 ✔ | thêm `b`, `q` | `a p b q` |
| 3 | 2 | 2 | 2<4 và 2<2 ✘ | thoát vòng | |
| đuôi | 2 | 2 | — | thêm `word1[2:]="cd"`, `word2[2:]=""` | `a p b q cd` |

Kết quả `"apbqcd"`.

````python
def trace_merge(word1: str, word2: str) -> str:
    i, j, res = 0, 0, []
    step = 1
    print(f"word1={word1!r}, word2={word2!r}")
    while i < len(word1) and j < len(word2):
        res.append(word1[i])
        res.append(word2[j])
        print(f"  bước {step}: i={i}, j={j} -> thêm {word1[i]!r}, {word2[j]!r} | res={''.join(res)!r}")
        i += 1
        j += 1
        step += 1
    print(f"  thoát while: i={i} (len1={len(word1)}), j={j} (len2={len(word2)})")
    print(f"  đuôi: word1[{i}:]={word1[i:]!r}, word2[{j}:]={word2[j:]!r}")
    res.append(word1[i:])
    res.append(word2[j:])
    out = "".join(res)
    print("  => kết quả:", repr(out))
    return out

trace_merge("abcd", "pq")
````

## 7. Code nộp LeetCode

````python
class Solution:
    def mergeAlternately(self, word1: str, word2: str) -> str:
        i, j = 0, 0
        res = []
        while i < len(word1) and j < len(word2):
            res.append(word1[i])
            res.append(word2[j])
            i += 1
            j += 1
        res.append(word1[i:])
        res.append(word2[j:])
        return "".join(res)

sol = Solution()
print(sol.mergeAlternately("abc", "pqr"), sol.mergeAlternately("ab", "pqrs"), sol.mergeAlternately("abcd", "pq"))
````

Nhận xét: vì `i` và `j` luôn bằng nhau trong bài này, có thể dùng **một** chỉ số. Ta vẫn giữ hai con trỏ vì đây là khuôn tổng quát (ví dụ bài trộn hai mảng đã sắp xếp — con trỏ tiến không đều).

## 8. Các lời giải khác

### 8.1. `zip` + phần đuôi bằng slicing

`zip` lo pha xen kẽ; phần đuôi là `word1[k:] + word2[k:]` với `k = min(len1, len2)`.

````python
class SolutionZip:
    def mergeAlternately(self, word1: str, word2: str) -> str:
        k = min(len(word1), len(word2))
        return "".join(a + b for a, b in zip(word1, word2)) + word1[k:] + word2[k:]

print(SolutionZip().mergeAlternately("abcd", "pq"), SolutionZip().mergeAlternately("ab", "pqrs"))
````

### 8.2. `zip_longest` với `fillvalue=""` — một dòng

Chỗ thiếu được điền `""`, nên `a + b` vẫn đúng.

````python
from itertools import zip_longest

class SolutionZipLongest:
    def mergeAlternately(self, word1: str, word2: str) -> str:
        return "".join(a + b for a, b in zip_longest(word1, word2, fillvalue=""))

print(SolutionZipLongest().mergeAlternately("abcd", "pq"), SolutionZipLongest().mergeAlternately("ab", "pqrs"))
````

### 8.3. Duyệt chỉ số tới `max` — bản nộp

````python
class SolutionIndex:
    def mergeAlternately(self, word1: str, word2: str) -> str:
        res = []
        for i in range(max(len(word1), len(word2))):
            if i < len(word1):
                res.append(word1[i])
            if i < len(word2):
                res.append(word2[i])
        return "".join(res)

print(SolutionIndex().mergeAlternately("abcd", "pq"))
````

### 8.4. Gán xen kẽ bằng slicing có bước (sáng tạo)

Tạo list độ dài `2k` (k = độ dài chung), gán `res[0::2] = word1[:k]` (vị trí chẵn) và `res[1::2] = word2[:k]` (vị trí lẻ), rồi nối đuôi.

````python
lst = [None] * 6
lst[0::2] = "abc"
lst[1::2] = "pqr"
print(lst)

class SolutionStride:
    def mergeAlternately(self, word1: str, word2: str) -> str:
        k = min(len(word1), len(word2))
        res = [""] * (2 * k)
        res[0::2] = word1[:k]
        res[1::2] = word2[:k]
        return "".join(res) + word1[k:] + word2[k:]

print(SolutionStride().mergeAlternately("abcd", "pq"), SolutionStride().mergeAlternately("ab", "pqrs"))
````

### 8.5. Bảng so sánh

Gọi `m = len(word1)`, `n = len(word2)`.

| Cách | Ý tưởng | Time | Space phụ | Ghi chú |
|---|---|---:|---:|---|
| Cộng chuỗi "vụng" | 3 vòng + nhiều `if` | `O(m + n)`* | `O(m + n)` | *có thể tệ hơn do cộng chuỗi |
| Duyệt chỉ số tới max | kiểm tra biên 2 lần/vòng | `O(m + n)` | `O(m + n)` | Dễ nghĩ |
| Hai con trỏ | xen kẽ + đuôi | `O(m + n)` | `O(m + n)` | **Lời giải chính**, khuôn tổng quát |
| `zip` + đuôi | dựng sẵn | `O(m + n)` | `O(m + n)` | Gọn |
| `zip_longest` | điền `""` | `O(m + n)` | `O(m + n)` | Một dòng |
| Slicing bước 2 | gán chẵn/lẻ | `O(m + n)` | `O(m + n)` | Sáng tạo |

Bộ nhớ `O(m + n)` là của **kết quả** — không tránh được.

## 9. Vì sao lời giải đúng?

**Bất biến của pha xen kẽ:** trước mỗi vòng, `i == j == t` (số vòng đã chạy) và `res` chứa đúng `word1[0], word2[0], …, word1[t-1], word2[t-1]` theo thứ tự — tức là kết quả trộn của `t` cặp đầu.

- Vòng lặp chạy khi cả hai còn ký tự → mỗi vòng thêm đúng cặp tiếp theo, bắt đầu từ `word1`. Bất biến giữ nguyên.
- Vòng lặp dừng khi `t = min(m, n)`: một chuỗi đã hết.

**Pha đuôi:** nếu `m > n` thì `word1[t:]` là phần dư của `word1`, `word2[t:] = ""`; ngược lại tương tự; nếu `m == n` cả hai là `""`. Nối hai đuôi đúng bằng "nối phần còn lại của chuỗi dài hơn".

Mỗi ký tự của hai chuỗi xuất hiện **đúng một lần** trong kết quả → độ dài kết quả = `m + n`.

````python
import random, string

rng = random.Random(1768)
for _ in range(300):
    w1 = "".join(rng.choice(string.ascii_lowercase) for _ in range(rng.randint(1, 15)))
    w2 = "".join(rng.choice(string.ascii_lowercase) for _ in range(rng.randint(1, 15)))
    out = merge_two_pointers(w1, w2)
    assert len(out) == len(w1) + len(w2)
    k = min(len(w1), len(w2))
    assert out[0:2 * k:2] == w1[:k] and out[1:2 * k:2] == w2[:k]     # vị trí chẵn từ word1, lẻ từ word2
    assert out[2 * k:] == w1[k:] + w2[k:]                            # đuôi
print("Kiểm chứng tính chất: độ dài, vị trí chẵn/lẻ và phần đuôi đều đúng")
````

`string.ascii_lowercase` là chuỗi `"abcdefghijklmnopqrstuvwxyz"` có sẵn trong thư viện chuẩn.

## 10. Độ phức tạp

Gọi `m = len(word1)`, `n = len(word2)`.

- **Time `O(m + n)`:** pha xen kẽ `min(m, n)` vòng; slicing đuôi `O(|m - n|)`; `join` `O(m + n)`.
- **Space `O(m + n)`:** list `res` và chuỗi kết quả. Không tính output thì bộ nhớ phụ là `O(m + n)` cho list (hoặc `O(1)` nếu chỉ tính con trỏ).

Không thể nhanh hơn: kết quả dài `m + n`, phải ghi từng ký tự.

## 12. Lỗi người mới hay gặp

### Lỗi 1: Dùng `zip` mà quên phần đuôi

````python
def wrong_zip_only(word1, word2):
    return "".join(a + b for a, b in zip(word1, word2))

print("sai:", wrong_zip_only("abcd", "pq"), "| đúng:", merge_two_pointers("abcd", "pq"))
````

**Sửa:** thêm `word1[k:] + word2[k:]`, hoặc dùng `zip_longest(..., fillvalue="")`.

### Lỗi 2: Dùng `or` trong điều kiện `while` → `IndexError`

````python
def wrong_or(word1, word2):
    i, res = 0, []
    while i < len(word1) or i < len(word2):
        res.append(word1[i])      # có thể vượt biên
        res.append(word2[i])
        i += 1
    return "".join(res)

try:
    wrong_or("abcd", "pq")
except IndexError as e:
    print("IndexError:", e)
````

**Sửa:** `and` cho pha xen kẽ; nếu dùng `or` thì phải kiểm tra `i < len(...)` trước mỗi lần đọc (như cách duyệt chỉ số).

### Lỗi 3: Bắt đầu từ `word2`

Đảo thứ tự `append` → `"pqr"` lên trước.

````python
def wrong_order(word1, word2):
    return "".join(b + a for a, b in zip_longest(word1, word2, fillvalue=""))

print("sai:", wrong_order("abc", "pqr"), "| đúng:", merge_two_pointers("abc", "pqr"))
````

### Lỗi 4: Chỉ nối đuôi của một chuỗi

Viết `res.append(word1[i:])` mà quên `word2[j:]` → sai khi `word2` dài hơn.

````python
def wrong_one_tail(word1, word2):
    i, res = 0, []
    while i < len(word1) and i < len(word2):
        res += [word1[i], word2[i]]
        i += 1
    res.append(word1[i:])
    return "".join(res)

print("sai:", wrong_one_tail("ab", "pqrs"), "| đúng:", merge_two_pointers("ab", "pqrs"))
````

### Lỗi 5: `zip_longest` mà không đặt `fillvalue`

Mặc định điền `None` → `a + b` gây `TypeError`.

````python
try:
    "".join(a + b for a, b in zip_longest("abc", "p"))
except TypeError as e:
    print("TypeError:", e)
````

## 13. Mẹo nhận dạng pattern

- **"Trộn / xen kẽ / ghép lần lượt"** hai dãy → **hai con trỏ**, mỗi dãy một con trỏ.
- Luôn tách thành **pha chung** (cả hai còn) và **pha đuôi** (một bên hết).
- Trong Python, `seq[k:]` an toàn khi `k` vượt độ dài → dùng để nối đuôi không cần `if`.
- Xây chuỗi kết quả: **list + `"".join`**.
- Khuôn hai con trỏ này dùng lại cho: trộn hai mảng sắp xếp (88), trộn hai danh sách liên kết (21), so sánh phiên bản (165).

Bài cùng họ: 88 Merge Sorted Array, 21 Merge Two Sorted Lists, 281 Zigzag Iterator, 1528 Shuffle String.

## 14. Biến thể

### 14.1. Trộn **k** chuỗi xen kẽ

`zip_longest(*words, fillvalue="")` — dấu `*` "rải" list thành nhiều đối số.

````python
def merge_k(words):
    return "".join("".join(group) for group in zip_longest(*words, fillvalue=""))

print(merge_k(["abc", "pq", "xyzw"]))   # a p x b q y c z w
````

### 14.2. Trộn theo **khối** kích thước `b`

Lấy `b` ký tự từ `word1`, rồi `b` ký tự từ `word2`, lặp lại.

````python
def merge_blocks(word1, word2, b):
    i, res = 0, []
    while i < len(word1) or i < len(word2):
        res.append(word1[i:i + b])     # slicing an toàn kể cả khi vượt biên
        res.append(word2[i:i + b])
        i += b
    return "".join(res)

print(merge_blocks("abcdef", "PQRS", 2))   # ab PQ cd RS ef
````

### 14.3. Bài 88 — trộn hai mảng **đã sắp xếp** (con trỏ tiến không đều)

Khác bài gốc: mỗi bước chỉ tiến **một** con trỏ — con trỏ trỏ tới phần tử nhỏ hơn.

````python
def merge_sorted(a, b):
    i = j = 0
    res = []
    while i < len(a) and j < len(b):
        if a[i] <= b[j]:
            res.append(a[i]); i += 1
        else:
            res.append(b[j]); j += 1
    return res + a[i:] + b[j:]

print(merge_sorted([1, 4, 7], [2, 3, 8, 9]))
````

````python
def answer_merge(word1, word2):
    i = j = 0
    res = []
    while i < len(word1) and j < len(word2):
        res.append(word1[i])
        res.append(word2[j])
        i += 1
        j += 1
    res.append(word1[i:])
    res.append(word2[j:])
    return "".join(res)


def answer_unmerge(merged, m):
    n = len(merged) - m
    k = min(m, n)
    w1 = merged[0:2 * k:2]
    w2 = merged[1:2 * k:2]
    tail = merged[2 * k:]
    if m > n:
        w1 += tail
    else:
        w2 += tail
    return w1, w2


def answer_merge_lists_start_second(a, b):
    i = j = 0
    res = []
    while i < len(a) and j < len(b):
        res.append(b[j])
        res.append(a[i])
        i += 1
        j += 1
    return res + b[j:] + a[i:]


def answer_merge_k(words):
    ptr = [0] * len(words)
    res = []
    while True:
        took = False
        for t, w in enumerate(words):
            if ptr[t] < len(w):
                res.append(w[ptr[t]])
                ptr[t] += 1
                took = True
        if not took:
            break
    return "".join(res)


assert answer_merge("abcd", "pq") == "apbqcd"
assert answer_merge("ab", "pqrs") == "apbqrs"
assert answer_unmerge("apbqcd", 4) == ("abcd", "pq")
assert answer_unmerge("apbqrs", 2) == ("ab", "pqrs")
assert answer_merge_lists_start_second([1, 2, 3], [9]) == [9, 1, 2, 3]
assert answer_merge_lists_start_second([1], [7, 8]) == [7, 1, 8]
assert answer_merge_k(["abc", "pq", "xyzw"]) == "apxbqyczw"
assert answer_merge_k(["a"]) == "a"
for _ in range(300):
    w1 = "".join(rng.choice("abc") for _ in range(rng.randint(1, 10)))
    w2 = "".join(rng.choice("xyz") for _ in range(rng.randint(1, 10)))
    assert answer_unmerge(answer_merge(w1, w2), len(w1)) == (w1, w2)
    assert answer_merge_k([w1, w2]) == answer_merge(w1, w2)
print("Đáp án tham khảo PASS toàn bộ test!")
````
