## 1. Đề bài

## 2. Nền tảng: đếm tần suất bằng `Counter` và viết hàm kiểm tra

Trước khi giải, ta viết **validator**: hàm nhận `n` và chuỗi `s`, trả `True` nếu `s` đúng yêu cầu. Với bài có nhiều đáp án, validator là cách test đúng đắn duy nhất.

`collections.Counter(s)` đếm số lần mỗi ký tự: `Counter("pppz") == {'p': 3, 'z': 1}`. `.values()` lấy các số đếm.

````python
from collections import Counter

c = Counter("holasss")
print(c)
print(list(c.values()))
````

Hàm `all(iterable)` trả `True` nếu **mọi** phần tử đều đúng. `str.islower()` kiểm tra chuỗi chỉ có chữ thường; kết hợp `str.isalpha()` để loại số/ký hiệu.

````python
def is_valid(n, s):
    if len(s) != n:
        return False
    if not (s.isalpha() and s.islower() and s.isascii()):
        return False
    return all(cnt % 2 == 1 for cnt in Counter(s).values())

print(is_valid(4, "pppz"), is_valid(4, "ohhh"), is_valid(4, "love"))   # True True True
print(is_valid(4, "aabb"), is_valid(4, "abc"), is_valid(2, "AB"))      # False False False
````

## 3. Nền tảng: tính chẵn lẻ (parity)

- Số **chẵn**: chia hết cho 2 → `n % 2 == 0`. Số **lẻ**: `n % 2 == 1`.
- Cách khác: `n & 1` lấy bit cuối của `n` (1 nếu lẻ, 0 nếu chẵn).

**Quy tắc cộng:** lẻ + lẻ = chẵn; chẵn + lẻ = lẻ; chẵn + chẵn = chẵn. Tổng quát: **tổng của `k` số lẻ có cùng tính chẵn lẻ với `k`**.

````python
for n in [1, 2, 7, 10]:
    print(n, "lẻ" if n % 2 == 1 else "chẵn", "| n & 1 =", n & 1)

print("3 + 5 =", 3 + 5, "(lẻ + lẻ = chẵn)")
print("3 + 5 + 7 =", 3 + 5 + 7, "(3 số lẻ → lẻ)")
````

### Kiểm chứng quy tắc "tổng của k số lẻ"
Thử ngẫu nhiên nhiều bộ số lẻ để thấy quy tắc luôn đúng (minh hoạ, không thay chứng minh).

````python
import random
rng = random.Random(1374)
for _ in range(5):
    k = rng.randint(1, 6)
    odds = [2 * rng.randint(0, 20) + 1 for _ in range(k)]
    print(f"k={k}, các số lẻ={odds}, tổng={sum(odds)}, tổng % 2 = {sum(odds) % 2}, k % 2 = {k % 2}")
````

**Hệ quả cho bài:** nếu chuỗi dùng `k` chữ cái khác nhau, mỗi chữ xuất hiện lẻ lần, thì `n` (tổng số lần) có cùng tính chẵn lẻ với `k`.
- `n` lẻ → cần **số lẻ** chữ cái khác nhau (1, 3, 5, …). Đơn giản nhất: **1 chữ**.
- `n` chẵn → cần **số chẵn** chữ cái khác nhau (2, 4, …). Không thể dùng 1 chữ! Đơn giản nhất: **2 chữ**.

## 4. Cách 1 — Vét cạn (brute force) để "nhìn" lời giải với `n` nhỏ

Khi chưa nghĩ ra công thức, hãy **thử mọi chuỗi** trên bảng chữ cái nhỏ (ví dụ `"ab"`) cho `n` nhỏ và xem chuỗi hợp lệ đầu tiên.

`itertools.product("ab", repeat=n)` sinh mọi tuple độ dài `n` gồm `'a'`/`'b'`: có `2^n` tuple.

````python
from itertools import product

print(list(product("ab", repeat=2)))

def brute_force(n, alphabet="ab"):
    for chars in product(alphabet, repeat=n):
        s = "".join(chars)
        if is_valid(n, s):
            return s
    return None

for n in range(1, 9):
    print(n, brute_force(n))
````

Quan sát output: với `n` lẻ, vét cạn tìm ra `"aaa...a"`; với `n` chẵn, tìm ra dạng `"a...ab"` (`n-1` chữ `a` và 1 chữ `b`). Khớp phân tích chẵn lẻ ở mục 3!

**Nút thắt:** vét cạn tốn `O(2^n · n)` — với `n = 500` là con số khổng lồ. Ta chỉ dùng nó để **gợi ý** công thức và để **đối chiếu** với `n` nhỏ.

````python
for n in [10, 20, 60, 500]:
    print(f"n={n}: 2^n có {len(str(2 ** n))} chữ số")
````

## 5. Cách 2 (chính) — Dựng trực tiếp theo chẵn lẻ

- `n` lẻ: `"a" * n` — chữ `a` xuất hiện `n` lần (lẻ).
- `n` chẵn: `"a" * (n - 1) + "b"` — `a` xuất hiện `n - 1` lần (lẻ vì `n` chẵn), `b` 1 lần (lẻ).

Phép `*` với chuỗi: `"a" * 3 == "aaa"`; phép `+` nối chuỗi. `"a" * 0 == ""` (chuỗi rỗng).

````python
print("a" * 3, "|", "a" * 0 + "b", "|", "a" * 3 + "b")

def generate(n):
    if n % 2 == 1:
        return "a" * n
    return "a" * (n - 1) + "b"

for n in [1, 2, 4, 7]:
    s = generate(n)
    print(n, repr(s), Counter(s), is_valid(n, s))
````

### 5.1. TRACE quyết định cho một vài `n`

| n | n % 2 | Nhánh | Chuỗi | Đếm | Hợp lệ? |
|---|---|---|---|---|---|
| 1 | 1 | lẻ | `a` | a:1 | ✓ |
| 2 | 0 | chẵn | `ab` | a:1, b:1 | ✓ |
| 4 | 0 | chẵn | `aaab` | a:3, b:1 | ✓ |
| 7 | 1 | lẻ | `aaaaaaa` | a:7 | ✓ |

Trường hợp `n = 2` đáng chú ý: `n - 1 = 1` → `"a" + "b"`.

````python
def trace_generate(n):
    parity = n % 2
    if parity == 1:
        s = "a" * n
        branch = f"lẻ  → 'a' * {n}"
    else:
        s = "a" * (n - 1) + "b"
        branch = f"chẵn → 'a' * {n - 1} + 'b'"
    counts = dict(Counter(s))
    print(f"n={n:>3} | n % 2 = {parity} | {branch:<22} | đếm = {counts} | hợp lệ: {is_valid(n, s)}")
    return s

for n in [1, 2, 4, 7, 500]:
    trace_generate(n)
````

## 6. Cách 3 — Viết gọn một dòng

### 6.1. Biểu thức điều kiện (ternary)
`A if điều_kiện else B` — trả `A` nếu điều kiện đúng, ngược lại `B`.

````python
def generate_ternary(n):
    return "a" * n if n % 2 else "a" * (n - 1) + "b"

print([generate_ternary(n) for n in range(1, 6)])
````

`if n % 2` — số `1` được coi là `True`, `0` là `False`.

### 6.2. Mẹo không rẽ nhánh: dùng `n % 2`
Gọi `e = 1 - n % 2` (1 nếu `n` chẵn, 0 nếu lẻ). Chuỗi `"a" * (n - e) + "b" * e` đúng cho cả hai trường hợp.

````python
def generate_branchless(n):
    e = 1 - n % 2
    return "a" * (n - e) + "b" * e

print([generate_branchless(n) for n in range(1, 6)])
assert all(generate_branchless(n) == generate(n) for n in range(1, 501))
````

### 6.3. Cách khác cũng đúng: `"b"` ở đầu
Đáp án không duy nhất: `"b" + "a" * (n - 1)` với `n` chẵn cũng hợp lệ. LeetCode chấm bằng validator nên cả hai đều Accepted.

````python
def generate_b_first(n):
    return "a" * n if n % 2 else "b" + "a" * (n - 1)

print(generate_b_first(4), is_valid(4, generate_b_first(4)))
print("Khác chuỗi nhưng đều hợp lệ:", generate(4) != generate_b_first(4))
````

### 6.4. Phiên bản dùng toán tử bit `n & 1` và cách LeetCode chấm bài nhiều đáp án

`n & 1` cho bit cuối của `n`: `1` nếu lẻ, `0` nếu chẵn — nhanh và hay gặp trong code thi đấu. Ta viết lại lời giải bằng `n & 1`, và mô phỏng cách một hệ thống chấm bài "nhiều đáp án" làm việc: nó **không** so chuỗi với đáp án mẫu, mà gọi một **checker** (chính là validator) trên output của bạn.

Bảng so sánh vài đáp án hợp lệ cho `n = 4` (mọi dòng đều được chấp nhận):

| Output | Đếm | Checker |
|---|---|---|
| `aaab` | a:3, b:1 | ✓ |
| `baaa` | b:1, a:3 | ✓ |
| `pppz` | p:3, z:1 | ✓ |
| `love` | l,o,v,e mỗi chữ 1 | ✓ |
| `aabb` | a:2, b:2 | ✗ |

````python
def generate_bit(n):
    return "a" * n if n & 1 else "a" * (n - 1) + "b"

def judge(n, output):
    # mô phỏng checker của hệ thống chấm: chỉ kiểm tính chất, không so với đáp án mẫu
    return "Accepted" if is_valid(n, output) else "Wrong Answer"

for out in ["aaab", "baaa", "pppz", "love", "aabb", generate_bit(4)]:
    print(f"n=4, output={out!r:>7} → {judge(4, out)}")
````

## 7. Cách 4 — Tổng quát: dùng `k` chữ cái khác nhau

Câu hỏi mở rộng: dựng chuỗi độ dài `n` dùng **đúng `k`** chữ cái khác nhau, mỗi chữ lẻ lần. Điều kiện cần và đủ (từ mục 3):
1. `k <= n` (mỗi chữ ít nhất 1 lần),
2. `k ≡ n (mod 2)` (cùng chẵn lẻ),
3. `1 <= k <= 26`.

Cách dựng: `k - 1` chữ đầu mỗi chữ **1** lần, chữ cuối nhận phần còn lại `n - (k - 1)` lần — phần này lẻ vì `n - k + 1` lẻ khi `n ≡ k`.

````python
import string
print(string.ascii_lowercase)

def generate_k(n, k):
    if not (1 <= k <= 26 and k <= n and (n - k) % 2 == 0):
        return None
    letters = string.ascii_lowercase
    return letters[: k - 1] + letters[k - 1] * (n - k + 1)

for n, k in [(7, 1), (7, 3), (8, 2), (8, 4), (8, 3), (3, 5)]:
    s = generate_k(n, k)
    print(f"n={n}, k={k} → {s!r}", "" if s is None else f"| đếm {dict(Counter(s))} | hợp lệ {is_valid(n, s)}")
````

`string.ascii_lowercase` là hằng chuỗi `"abc...z"` có sẵn trong module chuẩn `string`. `letters[: k - 1]` lấy `k - 1` chữ đầu.

## 8. Code nộp LeetCode

````python
class Solution:
    def generateTheString(self, n: int) -> str:
        if n % 2 == 1:
            return "a" * n
        return "a" * (n - 1) + "b"

sol = Solution()
for n in [4, 2, 7]:
    s = sol.generateTheString(n)
    print(n, s, is_valid(n, s))
````

## 9. Vì sao đúng?

**Trường hợp `n` lẻ:** chuỗi `"a" * n` dài đúng `n`, chỉ có chữ `a` với số lần `n` — lẻ. ✓

**Trường hợp `n` chẵn (`n >= 2`):** chuỗi `"a" * (n-1) + "b"` dài `(n - 1) + 1 = n`. Chữ `a` xuất hiện `n - 1` lần: `n` chẵn nên `n - 1` lẻ. Chữ `b` xuất hiện 1 lần — lẻ. ✓ (Với `n = 2`, `n - 1 = 1 >= 1` nên `a` thật sự có mặt; nếu `n - 1 = 0` thì `a` không xuất hiện và cũng không vi phạm gì — nhưng trường hợp đó không xảy ra vì `n` chẵn ≥ 2.)

**Vì sao không thể dùng 1 chữ khi `n` chẵn?** Một chữ duy nhất phải xuất hiện `n` lần — chẵn → vi phạm. Đây là lý do bắt buộc phải có chữ thứ hai.

````python
# Kiểm cho cách tổng quát: generate_k trả None đúng lúc, hợp lệ khi khác None
for n in range(1, 60):
    for k in range(1, 27):
        s = generate_k(n, k)
        possible = k <= n and (n - k) % 2 == 0
        assert (s is not None) == possible, (n, k)
        if s is not None:
            assert is_valid(n, s) and len(set(s)) == k
print("generate_k: đúng điều kiện tồn tại và luôn hợp lệ — PASS")
````

### Lỗi 2: Dùng `n` chữ khác nhau
`"abcd..."` đúng với `n <= 26`, nhưng `n` tới 500 → hết chữ cái (và `chr` sẽ sinh ký tự **không** phải chữ thường).

````python
def wrong_distinct(n):
    return "".join(chr(ord("a") + i) for i in range(n))

print(is_valid(5, wrong_distinct(5)), is_valid(27, wrong_distinct(27)), repr(wrong_distinct(28)[-2:]))
````

### Lỗi 3: Nhầm `n - 1` với `n // 2`
Ý "chia đôi" (`"a" * (n//2) + "b" * (n//2)`) cho hai số bằng nhau — nếu `n//2` chẵn thì cả hai chữ đều chẵn lần.

````python
def wrong_half(n):
    return "a" * (n // 2) + "b" * (n - n // 2)

print([(n, wrong_half(n), is_valid(n, wrong_half(n))) for n in [2, 4, 6]])
````

Thấy không: `n = 2` và `n = 6` đúng **tình cờ** (vì `n//2` lẻ), `n = 4` sai. Test nhiều `n` mới lộ lỗi — một lý do để kiểm **toàn bộ** miền.

### Lỗi 4: Test bằng so sánh chuỗi cố định
`assert sol.generateTheString(4) == "pppz"` sẽ **fail** dù lời giải đúng. Bài nhiều đáp án → test bằng **tính chất** (validator).

### Mẹo nhận dạng pattern

- Đề nói **"return any of them"** → bài constructive: tìm cách dựng **đơn giản nhất**, test bằng validator.
- Có chữ **odd/even** → nghĩ ngay tới **tính chẵn lẻ của tổng**: tổng `k` số lẻ ≡ `k (mod 2)`.
- Không nghĩ ra công thức → **vét cạn với `n` nhỏ** và quan sát pattern.
- Bài anh em: *Find N Unique Integers Sum up to Zero* (1304), *Construct K Palindrome Strings* (1400), *Palindrome Permutation* (266).

## 13. Biến thể

### 13.1. Mỗi chữ xuất hiện **chẵn** lần
Chỉ có lời giải khi `n` chẵn: `"a" * n`. Với `n` lẻ thì **không thể** (tổng các số chẵn luôn chẵn).

````python
def generate_even_counts(n):
    return "a" * n if n % 2 == 0 else None

print([generate_even_counts(n) for n in range(1, 7)])
````

### 13.2. Find N Unique Integers Sum up to Zero (LeetCode 1304)
Cũng là bài constructive dùng **đối xứng**: lấy các cặp `±i`, thêm `0` nếu `n` lẻ.

````python
def sum_zero(n):
    res = []
    for i in range(1, n // 2 + 1):
        res += [i, -i]
    if n % 2:
        res.append(0)
    return res

for n in [1, 4, 5]:
    r = sum_zero(n)
    print(n, r, "tổng =", sum(r), "| phân biệt:", len(set(r)) == n)
````

### 13.3. Có thể hoán vị chuỗi thành palindrome? (LeetCode 266)
Palindrome cho phép **tối đa một** chữ xuất hiện lẻ lần — lại là đếm chẵn lẻ.

````python
def can_permute_palindrome(s):
    odd = sum(1 for c in Counter(s).values() if c % 2)
    return odd <= 1

print(can_permute_palindrome("code"), can_permute_palindrome("aab"), can_permute_palindrome("carerac"))
````

## 16. Đáp án tham khảo

**Bài 1:** `"aaaaab"` (`a`:5, `b`:1). `"a" * 6` **không** hợp lệ.

````python
assert Solution().generateTheString(6) == "aaaaab" and not is_valid(6, "a" * 6)

def answer_generate(n):
    return "a" * n if n % 2 else "a" * (n - 1) + "b"

def answer_is_valid(n, s):
    if len(s) != n:
        return False
    cnt = [0] * 26
    for ch in s:
        if not ("a" <= ch <= "z"):
            return False
        cnt[ord(ch) - 97] += 1
    return all(c % 2 == 1 for c in cnt if c > 0)

def answer_three_letters(n):
    if n < 3 or n % 2 == 0:
        return None
    return "ab" + "c" * (n - 2)

def answer_min_deletions(s):
    return sum(1 for c in Counter(s).values() if c % 2 == 0)

def answer_lexi_smallest(n):
    return "a" * n if n % 2 else "a" * (n - 1) + "b"

assert all(is_valid(n, answer_generate(n)) for n in range(1, 501))
assert answer_is_valid(4, "pppz") is True and answer_is_valid(4, "aabb") is False
assert answer_is_valid(2, "a1") is False and answer_is_valid(3, "ab") is False
assert answer_three_letters(2) is None and answer_three_letters(4) is None
assert all(is_valid(n, answer_three_letters(n)) and len(set(answer_three_letters(n))) == 3
           for n in range(3, 200, 2))
assert answer_min_deletions("aabbb") == 1
assert answer_min_deletions("abc") == 0
assert answer_min_deletions("aabbcc") == 3

# Đối chiếu bài 6 với vét cạn: product duyệt theo thứ tự từ điển → chuỗi hợp lệ đầu tiên là nhỏ nhất
for n in range(1, 11):
    assert brute_force(n, "abc") == answer_lexi_smallest(n), n

# Đối chiếu validator tự viết với is_valid trên chuỗi ngẫu nhiên
rng = random.Random(9)
for _ in range(500):
    m = rng.randint(1, 12)
    s = "".join(rng.choice("abc") for _ in range(m))
    assert answer_is_valid(m, s) == is_valid(m, s)
print("Đáp án bài tập: PASS toàn bộ")
````
