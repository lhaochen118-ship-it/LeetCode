| Ví dụ | num | Output | Giải thích |

| 1 | `"51230100"` | `"512301"` | có hai chữ số `0` ở cuối; bỏ đi còn `"512301"` (số `0` ở giữa `"…301…"` giữ nguyên) |
| 2 | `"123"` | `"123"` | không có số `0` ở cuối nên giữ nguyên |

Constraints: `1 <= num.length <= 1000`; `num` chỉ gồm các chữ số; `num` **không có chữ số 0 ở đầu** (và là số dương, nên chữ số đầu khác `0`).

````python
num = "51230100"
print(num, "->", num[:-2])             # 512301
print("123", "-> giữ nguyên vì không có 0 ở cuối")
print(num.count("0"), "chữ số 0 trong chuỗi, nhưng chỉ 2 cái ở cuối")
````

## 3. Kiến thức nền

### 3.1. Chuỗi là dãy kí tự: chỉ số dương, chỉ số âm, `len`

Chuỗi `"51230100"` có `len = 8`. Kí tự đầu `num[0]`, kí tự cuối `num[7]` hay `num[-1]`. Chỉ số âm đếm từ cuối: `num[-1]` cuối cùng, `num[-2]` kề cuối...

````python
num = "51230100"
print(len(num))               # 8
print(num[0], num[7], num[-1], num[-2])       # 5 0 0 0
for i in range(len(num) - 1, -1, -1):          # đi ngược từ cuối: 7, 6, ..., 0
    print(i, num[i], end=" | ")
print()
````

### 3.2. Cắt lát (slicing): `s[a:b]`

`s[a:b]` lấy các kí tự từ chỉ số `a` đến `b - 1` (**không gồm `b`**). `s[:k]` lấy `k` kí tự đầu. `s[:-1]` bỏ kí tự cuối. Nếu `b` quá lớn thì tự cắt về hết chuỗi, không lỗi.

````python
num = "51230100"
print(num[:6])        # 512301 : 6 kí tự đầu
print(num[:-1])       # 5123010 : bỏ kí tự cuối
print(num[2:5])       # 230
print(num[:100])      # 51230100 : không lỗi dù 100 > len
````

### 3.3. `endswith` và `rstrip`

- `s.endswith("0")` kiểm tra chuỗi **kết thúc** bằng `"0"`.
- `s.rstrip(chars)` **cắt bỏ ở bên phải** mọi kí tự thuộc tập `chars` **cho đến khi gặp kí tự không thuộc tập**. `"51230100".rstrip("0")` cho `"512301"`. Lưu ý đối số là **tập các kí tự**, không phải chuỗi con cố định (ở đây chỉ có `"0"`).
- Tương tự: `lstrip` (bên trái), `strip` (cả hai bên).

````python
print("51230100".endswith("0"))        # True
print("123".endswith("0"))             # False
print("51230100".rstrip("0"))          # 512301
print("00120".lstrip("0"))             # 120 (bỏ 0 ĐẦU)
print("00120".strip("0"))              # 12  (cả hai bên)
print("1 0 1".replace("0", ""))         # '1  1' : replace bỏ MỌI số 0 (khác rstrip)
````

### 3.4. Vòng `while` và điều kiện dừng

`while điều_kiện:` lặp **chừng nào** điều kiện còn đúng. Ta dùng nó khi không biết trước số vòng lặp (ở đây: số số 0 ở cuối). Luôn kiểm tra: (1) biến điều kiện có thay đổi trong thân vòng lặp? (2) có điều kiện chặn khi chỉ số "rớt khỏi" chuỗi không?

````python
num = "51230100"
i = len(num) - 1
while num[i] == "0":      # lùi chừng nào còn gặp '0'
    i -= 1
print("chỉ số của chữ số khác 0 cuối cùng:", i)       # 5
print(num[: i + 1])                                    # 512301
````

### 3.5. Số học: `% 10`, `// 10` và `int` / `str`

Với số nguyên: `x % 10` là chữ số cuối, `x // 10` bỏ chữ số cuối. Số có `0` cuối khi `x % 10 == 0`. `int("51230100")` đổi chuỗi sang số (Python số nguyên không giới hạn độ lớn), `str(x)` đổi ngược lại. Chú ý `int("0012")` bỏ **0 ở đầu**, không bỏ 0 ở cuối.

````python
x = 51230100
print(x % 10, x // 10)              # 0 5123010
while x % 10 == 0:
    x //= 10
print(x, str(x))                    # 512301 '512301'
print(int("0012"), int("120"))      # 12 120   : int chỉ bỏ 0 ĐẦU, không bỏ 0 CUỐI
````

## 4. Ý tưởng nền tảng / cách nghĩ trực giác

Ta chỉ cần làm một việc: **nhìn từ cuối chuỗi, lùi dần chừng nào còn gặp `0`; dừng ở chữ số khác `0` đầu tiên — phần còn lại từ đầu đến chữ số đó là đáp án.**

Tính tay `"51230100"`:

| vị trí (từ phải) | kí tự | hành động |
|---|---|---|
| chỉ số 7 | `0` | bỏ |
| chỉ số 6 | `0` | bỏ |
| chỉ số 5 | `1` | khác 0 ⇒ dừng; giữ `num[0..5]` = `"512301"` |

Ba điều cần nhớ: (1) chỉ nhìn **cuối chuỗi**; (2) **dừng ngay** khi gặp chữ số khác `0` — các số `0` nằm sâu bên trong (như `0` ở chỉ số 4) không đụng tới; (3) vì đề đảm bảo số dương không có số 0 ở đầu nên chắc chắn có ít nhất một chữ số khác `0` để dừng lại.

````python
num = "51230100"
end = len(num)                  # end là "chỉ số vượt qua kí tự giữ lại cuối"
while end > 0 and num[end - 1] == "0":
    end -= 1
    print("bỏ một số 0, còn giữ", num[:end])
print("kết quả:", num[:end])
````

## 5. Cách 1 — Brute Force / baseline: cắt từng kí tự cuối bằng `while`

**Ý tưởng:** chừng nào chuỗi còn kết thúc bằng `"0"` thì **bỏ kí tự cuối** (`num = num[:-1]`). Đơn giản, dịch đúng nghĩa "bỏ các số 0 ở cuối". Mỗi lần cắt tạo ra một chuỗi mới (chuỗi Python bất biến).

````python
def remove_zeros_loop(num: str) -> str:
    while num.endswith("0"):
        num = num[:-1]
    return num

print(remove_zeros_loop("51230100"))    # 512301
print(remove_zeros_loop("123"))         # 123
print(remove_zeros_loop("1000"))        # 1
````

**Độ phức tạp:** gọi `n = len(num)`, `z` là số số 0 ở cuối (`z < n`). Mỗi lần cắt sao chép tối đa `n` kí tự ⇒ Time `O(z · n)`, xấu nhất `O(n²)` (ví dụ `"1" + "0" * 999`). Space `O(n)` (mỗi lần tạo chuỗi mới; các chuỗi cũ được thu hồi).

**Nút thắt:** ta sao chép chuỗi **nhiều lần**. Chỉ cần tìm ra **vị trí** cắt rồi cắt **một lần**.

## 6. Ý tưởng thuật toán / bước đột phá

Tách bài thành hai bước: (1) **tìm chỉ số `i` của chữ số khác `0` cuối cùng** bằng cách chạy chỉ số từ cuối lùi về đầu; (2) **cắt một lần** `num[: i + 1]`.

Bước (1) chỉ đọc kí tự (không sao chép) nên rẻ: `O(z)` lần so sánh. Bước (2) sao chép đúng một lần: `O(n)`. Tổng `O(n)`.

Ngoài ra Python có sẵn `rstrip("0")` làm đúng việc này bằng mã C nhanh, và cách số học `// 10` cho thấy cùng ý tưởng nhưng trên số thay vì chuỗi.

````python
num = "51230100"
i = len(num) - 1
while num[i] == "0":
    i -= 1
print("i =", i, "-> cắt num[:i+1] =", num[: i + 1])
````

## 7. Các cách tiếp theo

Ngoài Cách 1 (cắt từng kí tự), ta xem 3 cách: (2) quét chỉ số ngược + cắt một lần, (3) `rstrip("0")`, (4) số học `// 10`.

### Cách 2 — Quét chỉ số ngược rồi cắt một lần

Giải thích từng đoạn:

- `i = len(num) - 1`: bắt đầu từ kí tự cuối.
- `while i >= 0 and num[i] == "0"`: lùi chừng nào còn `0`. **Điều kiện `i >= 0` đứng trước** để không đọc `num[-1]` (là kí tự cuối do chỉ số âm!) khi `i` đã rớt xuống `-1`. (Với đề này luôn có chữ số khác `0` nên `i >= 0` không bao giờ bị phá, nhưng viết bảo vệ vẫn là thói quen tốt.)
- `return num[: i + 1]`: giữ từ đầu tới `i` (bao gồm `i`) — vì cắt lát **không gồm** chỉ số cuối nên phải cộng `1`.

````python
def remove_zeros_scan(num: str) -> str:
    i = len(num) - 1
    while i >= 0 and num[i] == "0":
        i -= 1
    return num[: i + 1]

print(remove_zeros_scan("51230100"), remove_zeros_scan("123"), remove_zeros_scan("1000"))
````

````python
class SolutionScan:
    def removeTrailingZeros(self, num: str) -> str:
        i = len(num) - 1
        while i >= 0 and num[i] == "0":
            i -= 1
        return num[: i + 1]

print(SolutionScan().removeTrailingZeros("51230100"))
````

**Độ phức tạp:** vòng `while` chạy `z` lần `O(1)`, cắt lát `O(n)` ⇒ Time `O(n)`, Space `O(n)` cho chuỗi kết quả (không tính chuỗi đầu vào). Đếm lại: chỉ sao chép **một** lần.

### Cách 3 — `rstrip("0")` (hàm có sẵn)

`num.rstrip("0")` bỏ mọi `"0"` ở **đầu phải** của chuỗi, đúng thứ ta cần. Trong LeetCode đây là cách ngắn nhất. Vì đề đảm bảo số dương (có ít nhất một chữ số khác `0`) nên kết quả **không bao giờ rỗng**.

````python
def remove_zeros_rstrip(num: str) -> str:
    return num.rstrip("0")

print(remove_zeros_rstrip("51230100"), remove_zeros_rstrip("123"), remove_zeros_rstrip("1000"))
````

````python
class SolutionRstrip:
    def removeTrailingZeros(self, num: str) -> str:
        return num.rstrip("0")

print(SolutionRstrip().removeTrailingZeros("51230100"))
````

**Độ phức tạp:** Time `O(n)` (duyệt từ phải sang trái đến khi gặp kí tự khác `0`, rồi tạo chuỗi mới), Space `O(n)`.

### Cách 4 — Số học: chia `10` khi chia hết

Nghĩ chuỗi như một **số**: bỏ một chữ số `0` ở cuối tương đương `x //= 10`. Ta lặp `while x % 10 == 0: x //= 10` rồi đổi lại `str(x)`.

Điểm cần lưu ý: Python cho phép số nguyên rất lớn (tối đa 1000 chữ số ở đây là ổn; từ Python 3.11 có giới hạn mặc định 4300 chữ số khi đổi `int` ↔ `str`). Ở ngôn ngữ có số nguyên cố định 32/64 bit **cách này không dùng được**, vì số 1000 chữ số tràn bộ nhớ — đó cũng là lý do đề cho `num` dưới dạng **chuỗi**.

````python
def remove_zeros_arith(num: str) -> str:
    x = int(num)
    while x % 10 == 0:
        x //= 10
    return str(x)

print(remove_zeros_arith("51230100"), remove_zeros_arith("123"), remove_zeros_arith("1000"))
````

````python
class SolutionArith:
    def removeTrailingZeros(self, num: str) -> str:
        x = int(num)
        while x % 10 == 0:
            x //= 10
        return str(x)

print(SolutionArith().removeTrailingZeros("51230100"))
````

**Độ phức tạp:** mỗi `//= 10` trên số `n` chữ số tốn `O(n)`; làm `z` lần ⇒ Time `O(z · n)` (xấu nhất `O(n²)`, như Cách 1), cộng chi phí `int`/`str` ≈ `O(n)` (chỉ khi số chữ số nhỏ). Space `O(n)`. Kém hơn Cách 2, 3 nhưng cho thấy "ý tưởng giống nhau, cấu trúc dữ liệu khác nhau".

## 8. TRACE từng bước

Trace Cách 2 (quét chỉ số ngược) với `num = "51230100"` (`len = 8`):

| vòng | `i` | `num[i]` | `num[i] == "0"`? | hành động |
|---|---|---|---|---|
| (đầu) | 7 | `0` | có | `i = 6` |
| 1 | 6 | `0` | có | `i = 5` |
| 2 | 5 | `1` | không | dừng |

Sau vòng lặp `i = 5` ⇒ `num[:6] = "512301"`.

````python
def trace_scan(num):
    i = len(num) - 1
    print(f"num = {num!r}, len = {len(num)}, bắt đầu i = {i}")
    while i >= 0 and num[i] == "0":
        print(f"  i={i}: num[{i}]='0' -> lùi")
        i -= 1
    print(f"  i={i}: num[{i}]={num[i]!r} khác '0' -> dừng")
    ans = num[: i + 1]
    print(f"  cắt num[:{i + 1}] = {ans!r}")
    return ans

trace_scan("51230100")
````

````python
# Trace: không có số 0 cuối, và số có 0 ở giữa + dày đặc 0 cuối
trace_scan("123")
trace_scan("1000")
trace_scan("100100")
````

## 9. Code nộp LeetCode

Dùng Cách 2 (quét chỉ số ngược + cắt một lần), đúng class `Solution` và method `removeTrailingZeros`. (Cách `num.rstrip("0")` một dòng cũng được chấp nhận.)

````python
class Solution:
    def removeTrailingZeros(self, num: str) -> str:
        i = len(num) - 1
        while i >= 0 and num[i] == "0":
            i -= 1
        return num[: i + 1]
````

````python
print(Solution().removeTrailingZeros("51230100"))    # 512301
print(Solution().removeTrailingZeros("123"))         # 123
````

## 10. Vì sao thuật toán đúng

Ký hiệu `n = len(num)`. Đề đảm bảo `num` là số dương không có 0 đầu ⇒ `num[0] != "0"` ⇒ tồn tại ít nhất một chỉ số có chữ số khác `0`.

**Invariant của vòng `while`.** Tại mọi thời điểm, các kí tự `num[i + 1], ..., num[n - 1]` **đều bằng `"0"`** (đã được "lùi qua"). Ban đầu `i = n - 1` nên tập này rỗng ⇒ đúng. Mỗi lần lặp, ta chỉ lùi `i` khi `num[i] == "0"`, nên khi `i` giảm một đơn vị thì tính chất vẫn đúng.

**Khi dừng.** Vòng dừng vì `num[i] != "0"` (không phải vì `i < 0`, theo mục trên). Khi đó `num[i]` là chữ số khác `0`, còn `num[i + 1:]` toàn `0`: `i` chính là **chữ số khác `0` cuối cùng**. Mọi số 0 ở cuối chuỗi chính là `num[i + 1:]`, và đề bảo bỏ chúng đi.

**Kết quả.** `num[: i + 1]` giữ nguyên `num[0..i]`, trong đó có các số `0` ở giữa chuỗi (nằm trước `i`) — đúng yêu cầu "chỉ bỏ số 0 ở **cuối**". Không bỏ gì thừa vì `num[i] != "0"`, không bỏ thiếu vì toàn bộ `num[i + 1:]` đã bị cắt.

## 11. Phân tích Time / Space Complexity

Gọi `n = len(num)` (tối đa 1000), `z` = số số 0 ở cuối.

| Cách | Time | Space |
|---|---|---|
| 1. cắt từng kí tự bằng `while` | `O(z · n)`, xấu nhất `O(n²)` | `O(n)` |
| 2. quét chỉ số ngược + cắt một lần | `O(n)` | `O(n)` |
| 3. `rstrip("0")` | `O(n)` | `O(n)` |
| 4. số học `// 10` | `O(z · n)` (xấu nhất `O(n²)`) | `O(n)` |

Trong Cách 2, vòng `while` tốn `O(z)` còn cắt lát tốn `O(n)`; tổng `O(n)`. Space `O(n)` chủ yếu cho chuỗi kết quả (chuỗi Python bất biến nên không thể bỏ kí tự tại chỗ).

### 12.1. Ví dụ của đề

````python
sols = [SolutionScan(), SolutionRstrip(), SolutionArith(), Solution()]
funcs = [remove_zeros_loop, remove_zeros_scan, remove_zeros_rstrip, remove_zeros_arith] + [s.removeTrailingZeros for s in sols]

for f in funcs:
    assert f("51230100") == "512301"
    assert f("123") == "123"
print("Ví dụ của đề: tất cả cách đều đúng")
````

### 12.2. Edge case

Chỉ dùng input **hợp lệ** (số dương, không có 0 đầu):

- Một chữ số: `"5"` → `"5"`, `"1"` → `"1"`.
- Một chữ số khác `0` rồi toàn `0`: `"10"` → `"1"`, `"1000"` → `"1"`, `"900000"` → `"9"`.
- `0` ở giữa không bị đụng: `"101"` → `"101"`, `"1001"` → `"1001"`, `"100100"` → `"1001"`.
- `0` ở cả giữa và cuối: `"50505000"` → `"50505"`.
- Chuỗi dài 1000: `"1" + "0" * 999` → `"1"`; `"9" * 1000` → chính nó; `"1" * 500 + "0" * 500` → `"1" * 500`.

````python
cases = {
    "5": "5", "1": "1", "10": "1", "1000": "1", "900000": "9",
    "101": "101", "1001": "1001", "100100": "1001", "50505000": "50505",
    "1" + "0" * 999: "1", "9" * 1000: "9" * 1000, "1" * 500 + "0" * 500: "1" * 500,
}
for f in funcs:
    for s, want in cases.items():
        assert f(s) == want, (f, s[:20])
print("Edge case OK:", len(cases), "trường hợp x", len(funcs), "cách")
````

### 12.3. Phản ví dụ cho cách làm sai

Hai cách sai hay gặp: `replace("0", "")` (bỏ **mọi** số 0) và `str(int(num))` (chỉ bỏ 0 **đầu**). Ta xác nhận chúng sai (mục 13 giải thích kỹ).

### 12.4. Đối chiếu nhiều input ngẫu nhiên (có seed)

Sinh `num` hợp lệ: chữ số đầu từ `1..9`, độ dài `1..1000`, các chữ số sau lấy **thiên về `0`** (xác suất `0` khác nhau mỗi test) để thường xuyên có chuỗi `0` dài ở cuối. Oracle **độc lập** (không dùng cách nào ở trên): lật chuỗi bằng list, `pop` đến khi gặp chữ số khác `0`.

````python
import random

def oracle(num):
    digits = list(num)
    while digits and digits[-1] == "0":
        digits.pop()
    return "".join(digits)

random.seed(2710)
for _ in range(2000):
    n = random.randint(1, 1000) if random.random() < 0.3 else random.randint(1, 30)
    p0 = random.choice([0.1, 0.5, 0.9])               # xác suất sinh '0'
    first = random.choice("123456789")
    rest = ["0" if random.random() < p0 else random.choice("123456789") for _ in range(n - 1)]
    num = first + "".join(rest)
    assert len(num) == n and num[0] != "0" and num.isdigit()
    want = oracle(num)
    for f in funcs:
        assert f(num) == want, (num[:40], f)
    assert want == "" or want[-1] != "0"
    assert num.startswith(want) and set(num[len(want):]) <= {"0"}
print("OK: 2000 test ngẫu nhiên; mọi cách khớp oracle")
````

## 13. Các lỗi người mới hay gặp

### Lỗi 1: Dùng `replace("0", "")`

````python
def sai_replace_het(num: str) -> str:
    return num.replace("0", "")

print("sai :", sai_replace_het("51230100"))                 # 51231 (mất cả 0 giữa)
print("đúng:", Solution().removeTrailingZeros("51230100"))   # 512301
assert sai_replace_het("51230100") != Solution().removeTrailingZeros("51230100")
````

**Vì sao sai:** `replace` bỏ **mọi** số `0` trong chuỗi, kể cả `0` ở giữa. Đề chỉ muốn bỏ các số `0` **liền nhau ở cuối**. Cần hàm chỉ động đến phần cuối (`rstrip`) hoặc dừng khi gặp chữ số khác `0`.

### Lỗi 2: Đổi sang `int` rồi đổi lại

````python
def sai_qua_int(num: str) -> str:
    return str(int(num))

print("sai :", sai_qua_int("51230100"))        # 51230100 (không đổi gì)
print("đúng:", Solution().removeTrailingZeros("51230100"))
assert sai_qua_int("51230100") == "51230100"
````

**Vì sao sai:** `int("51230100")` là số `51230100` — chuyển lại thành chuỗi vẫn `"51230100"`. `int` chỉ làm mất **0 ở đầu** (`int("0012") = 12`), không đụng đến 0 ở cuối. Để bỏ 0 cuối bằng số học phải **chia `10`** như Cách 4.

### Lỗi 3: Quên `+ 1` khi cắt lát

````python
def sai_thieu_cong_1(num: str) -> str:
    i = len(num) - 1
    while i >= 0 and num[i] == "0":
        i -= 1
    return num[:i]                  # SAI: thiếu + 1, mất chữ số khác 0 cuối cùng

print("sai :", sai_thieu_cong_1("51230100"))        # 51230
print("đúng:", Solution().removeTrailingZeros("51230100"))
assert sai_thieu_cong_1("51230100") != Solution().removeTrailingZeros("51230100")
````

**Vì sao sai:** `num[:i]` **không gồm** chỉ số `i`, nên mất chữ số khác `0` cuối cùng (`1` ở chỉ số 5). Cắt lát `a:b` lấy tới `b - 1`; muốn giữ cả `num[i]` phải viết `num[: i + 1]`.

### Lỗi 4: Chỉ bỏ một chữ số `0` (dùng `if` thay vì `while`)

````python
def sai_if_thay_while(num: str) -> str:
    if num.endswith("0"):          # SAI: chỉ kiểm tra một lần
        num = num[:-1]
    return num

print("sai :", sai_if_thay_while("51230100"))        # 5123010 (còn 1 số 0 ở cuối)
print("đúng:", Solution().removeTrailingZeros("51230100"))
assert sai_if_thay_while("51230100") != Solution().removeTrailingZeros("51230100")
````

**Vì sao sai:** số số `0` ở cuối có thể là nhiều; phải **lặp** (`while`) đến khi chữ số cuối khác `0`. `if` chỉ xử lý được 1 chữ số.

### Lỗi 5: Quét tiến từ đầu thay vì lùi từ cuối

Nếu bạn quét `i` từ `0` tới cuối rồi dừng ở số `0` đầu tiên, bạn cắt nhầm tại `0` giữa chuỗi (`"1001"` ⇒ `"1"` — sai, vì `0` ở giữa không phải số 0 ở cuối). Luôn **lùi từ cuối**.

````python
def sai_quet_tien(num: str) -> str:
    for i, ch in enumerate(num):
        if ch == "0":
            return num[:i]          # SAI: gặp 0 đầu tiên đã cắt luôn
    return num

print("sai :", sai_quet_tien("1001"))           # '1'
print("đúng:", Solution().removeTrailingZeros("1001"))   # '1001'
assert sai_quet_tien("1001") != Solution().removeTrailingZeros("1001")
````

## 14. Mẹo làm bài nhanh & nhận dạng pattern

- "Bỏ ký tự **cuối** thoả điều kiện" ⇒ **quét từ cuối**, dừng ở kí tự đầu tiên không thoả. Tương tự đầu chuỗi ⇒ `lstrip`.
- Hai bước **tìm vị trí → cắt một lần** thường tốt hơn **cắt từng kí tự** nhiều lần.
- Python: `rstrip(ký_tự)` / `lstrip` / `strip` làm việc này sẵn; nhớ đối số là **tập kí tự**.
- Khi số quá lớn không nằm trong `int` của ngôn ngữ khác (C++/Java) nên giữ dạng **chuỗi** — đề cho `num` là `str` vì lẽ đó.
- Cẩn thận `num[i]` khi `i` có thể bằng `-1` (chỉ số âm **không báo lỗi** mà đọc kí tự cuối!). Luôn đặt `i >= 0` **trước** điều kiện đọc `num[i]` trong `and`.
- Hãy phân biệt: **bỏ mọi số 0** (`replace`), **bỏ 0 đầu** (`lstrip`), **bỏ 0 cuối** (`rstrip`).

## 15. Biến thể của bài

1. **Đếm số số 0 ở cuối** (LeetCode 172 Factorial Trailing Zeroes là biến thể toán học nổi tiếng): `len(num) - len(num.rstrip("0"))`.
2. **Bỏ số 0 ở đầu**: `num.lstrip("0")` (và nếu rỗng thì `"0"`).
3. **Bỏ 0 cả hai đầu**: `strip("0")`.
4. **Bỏ trailing zero sau dấu thập phân** (ví dụ `"3.1400"` → `"3.14"`, `"5.000"` → `"5"`): `rstrip("0").rstrip(".")` nếu có dấu chấm.
5. **Số có thể là 0** (không đảm bảo dương): kết quả rỗng phải thay bằng `"0"`.
6. **Dùng regex**: `re.sub(r"0+$", "", num)`.

````python
import re

def count_trailing_zeros(num):
    return len(num) - len(num.rstrip("0"))

def strip_decimal(s):
    return s.rstrip("0").rstrip(".") if "." in s else s

def remove_zeros_safe(num):
    r = num.rstrip("0")
    return r if r else "0"

print(count_trailing_zeros("51230100"), count_trailing_zeros("123"))      # 2 0
print(strip_decimal("3.1400"), strip_decimal("5.000"), strip_decimal("100"))   # 3.14 5 100
print(remove_zeros_safe("000"), remove_zeros_safe("1200"))                 # 0 12
print(re.sub(r"0+$", "", "51230100"))                                       # 512301
````

## 19. Đáp án tham khảo cho bài tập

````python
# Bài 1
assert "7000"[:-3] == "7"
assert "0100".rstrip("0") == "01"
assert "0100".lstrip("0") == "100"
assert "90".replace("0", "") == "9"

# Bài 2: "3040500": i = 6 (0), 5 (0), 4 ('5' dừng) -> num[:5] = "30405"; bỏ 2 số 0
num = "3040500"
i = len(num) - 1
trace_i = []
while i >= 0 and num[i] == "0":
    trace_i.append(i)
    i -= 1
assert trace_i == [6, 5] and i == 4
assert num[: i + 1] == "30405" and len(num) - len(num[: i + 1]) == 2

# Bài 3
def remove_trailing_zeros_todo_ans(num):
    i = len(num) - 1
    while i >= 0 and num[i] == "0":
        i -= 1
    return num[: i + 1]

assert remove_trailing_zeros_todo_ans("51230100") == "512301"
assert remove_trailing_zeros_todo_ans("123") == "123"

# Bài 4
def remove_zeros_arith2(num):
    x = int(num)
    while x % 10 == 0:
        x //= 10
    return str(x)

assert remove_zeros_arith2("51230100") == "512301"
assert remove_zeros_arith2("123") == "123"

# Bài 5
def trim_zeros_both(num):
    l, r = 0, len(num) - 1
    while l <= r and num[l] == "0":
        l += 1
    while r >= l and num[r] == "0":
        r -= 1
    return num[l: r + 1] if l <= r else "0"

assert trim_zeros_both("00120300") == "1203"
assert trim_zeros_both("0000") == "0"
assert trim_zeros_both("5") == "5"

def normalize_decimal(s):
    if "." not in s:
        return s.lstrip("0") or "0"
    whole, frac = s.split(".")
    whole = whole.lstrip("0") or "0"
    frac = frac.rstrip("0")
    return whole + ("." + frac if frac else "")

assert normalize_decimal("003.1400") == "3.14"
assert normalize_decimal("5.000") == "5"
assert normalize_decimal("0.50") == "0.5"
assert normalize_decimal("100") == "100"
print("Tất cả đáp án đúng")
````
