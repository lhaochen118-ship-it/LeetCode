## 1. Mục tiêu học

Sau notebook này bạn sẽ:

1. Hiểu **"độ rộng của một số nguyên"** theo đề: số chữ số, **cộng thêm 1 nếu số âm** (dấu `-` chiếm một ký tự), và **độ rộng của cột** là độ rộng lớn nhất trong cột đó.
2. Duyệt ma trận theo **cột** thay vì theo hàng: dùng hai vòng lặp lồng nhau với chỉ số `[hàng][cột]`, hoặc **chuyển vị** bằng `zip(*grid)`.
3. Đếm chữ số bằng `len(str(x))` và bằng **phép chia nguyên liên tiếp** `x //= 10` (không cần chuỗi), kể cả các trường hợp `0` và số âm.
4. Viết 5 cách: duyệt theo cột, `zip(*grid)`, một lượt theo hàng cập nhật mảng `ans`, đếm chữ số bằng phép chia, và mẹo "chỉ cần xét **max** và **min** của cột".
5. Tránh các lỗi: quên dấu `-`, đếm `0` có `0` chữ số, nhầm số hàng với số cột, và ghi đè `ans[j]` thay vì lấy max.

## 2. Đề bài & ví dụ

Cho ma trận số nguyên `grid` cỡ `m × n` (đánh chỉ số từ 0). **Độ rộng của một cột** là **độ dài lớn nhất** của các số nguyên trong cột đó. Ví dụ `grid = [[-10], [3], [12]]` thì độ rộng của cột duy nhất là `3`, vì `-10` có độ dài `3`.

Độ dài của số nguyên `x` có `len` chữ số bằng `len` nếu `x` **không âm**, và `len + 1` nếu `x` **âm** (cộng thêm dấu trừ).

Trả về mảng `ans` có kích thước `n`, trong đó `ans[i]` là độ rộng của cột thứ `i`.

| Ví dụ | grid | Output | Giải thích |
|---|---|---|---|
| 1 | `[[1],[22],[333]]` | `[3]` | Có một cột; độ dài các số là `1, 2, 3` ⇒ lớn nhất `3`. |
| 2 | `[[-15,1,3],[15,7,12],[5,6,-2]]` | `[3,1,2]` | Cột 0: `-15`(3), `15`(2), `5`(1) ⇒ 3. Cột 1: `1, 7, 6` ⇒ 1. Cột 2: `3`(1), `12`(2), `-2`(2) ⇒ 2. |

Constraints:

- `1 <= m, n <= 100` (`m = grid.length`, `n = grid[i].length`)
- `-10^9 <= grid[r][c] <= 10^9`

Lưu ý: giá trị lớn nhất `±10^9` nên độ rộng tối đa là `11` (`-1000000000` gồm dấu `-` và 10 chữ số).

````python
grid = [[-15, 1, 3], [15, 7, 12], [5, 6, -2]]
m, n = len(grid), len(grid[0])
print("số hàng m =", m, ", số cột n =", n)
for r in range(m):
    print([len(str(x)) for x in grid[r]])   # độ dài của từng ô, theo từng hàng
````

## 3. Kiến thức nền

### 3.1. Ma trận là "danh sách các danh sách"

`grid[r]` là hàng `r` (một danh sách); `grid[r][c]` là ô ở hàng `r`, cột `c`. `len(grid)` là **số hàng**; `len(grid[0])` là **số cột** (mọi hàng cùng độ dài). Hai chỉ số hay bị nhầm: `r` chạy `0..m-1`, còn `c` chạy `0..n-1`.

````python
grid = [[1, 2, 3],
        [4, 5, 6]]
print(len(grid), len(grid[0]))        # 2 3  (2 hàng, 3 cột)
print(grid[1][2])                     # 6: hàng 1, cột 2
print([grid[r][1] for r in range(len(grid))])   # cột 1: [2, 5]
````

### 3.2. Đổi số thành chuỗi và lấy độ dài

`str(x)` đổi số nguyên thành chuỗi **kể cả dấu trừ**: `str(-15)` là `"-15"`, độ dài `3`. Chính định nghĩa độ rộng của đề trùng với `len(str(x))`: số âm có thêm một ký tự `-`.

````python
for x in [0, 7, 15, -15, -2, 1000000000, -1000000000]:
    print(f"x={x:>12}  str={str(x)!r:>15}  len={len(str(x))}")
````

### 3.3. `max` với danh sách và biểu thức sinh (generator)

- `max([1, 5, 3])` trả `5`.
- `max(len(str(x)) for x in col)` là **biểu thức sinh**: không tạo danh sách trung gian, duyệt từng giá trị một.
- `max(a, b)` cho hai số: dùng khi cập nhật "lớn nhất đến giờ": `best = max(best, new)`.

````python
col = [-15, 15, 5]
print(max([len(str(x)) for x in col]))   # 3 (có tạo list)
print(max(len(str(x)) for x in col))     # 3 (không tạo list)
best = 0
for x in col:
    best = max(best, len(str(x)))
    print("x =", x, "-> best =", best)
````

### 3.4. Chuyển vị bằng `zip(*grid)`

`*grid` "bung" các hàng thành các đối số riêng; `zip` ghép phần tử cùng vị trí của mỗi hàng ⇒ mỗi bộ trả về là **một cột**. Đây là cách ngắn nhất để duyệt theo cột. Lưu ý `zip` trả các `tuple`.

````python
grid = [[-15, 1, 3], [15, 7, 12], [5, 6, -2]]
for j, col in enumerate(zip(*grid)):
    print("cột", j, "=", col)
print(list(zip(*grid)))
````

### 3.5. Đếm chữ số bằng phép chia nguyên

Với `x >= 0`: `x // 10` bỏ chữ số cuối. Đếm xem bỏ được bao nhiêu lần để về `0` ⇒ số chữ số. Ví dụ `123 → 12 → 1 → 0` là ba lần ⇒ `3` chữ số. Trường hợp riêng: `0` có **một** chữ số, nhưng vòng `while x > 0` sẽ cho `0` lần ⇒ cần xử lý riêng.

````python
x = 123
count = 0
while x > 0:
    x //= 10
    count += 1
    print("sau khi bỏ một chữ số: x =", x, ", count =", count)

# trường hợp đặc biệt x = 0
x, count = 0, 0
while x > 0:
    x //= 10
    count += 1
print("với x = 0, vòng while cho count =", count, "(nhưng số 0 có 1 chữ số)")
````

## 4. Ý tưởng nền tảng / cách nghĩ trực giác

Nhìn ma trận như một **bảng in ra màn hình**, ta cần cho mỗi cột một "bề rộng" vừa đủ để chứa số dài nhất. Với `[[-15,1,3],[15,7,12],[5,6,-2]]`:

```
 -15   1   3
  15   7  12
   5   6  -2
```

- Cột 0 chứa `-15` (3 ký tự) ⇒ rộng 3.
- Cột 1 chỉ có chữ số đơn ⇒ rộng 1.
- Cột 2 có `12` và `-2` (đều 2 ký tự) ⇒ rộng 2.

Công việc là: **với mỗi cột, tính độ dài của từng ô, lấy lớn nhất.** Việc khó duy nhất là "độ dài" của số âm và của số `0`.

````python
grid = [[-15, 1, 3], [15, 7, 12], [5, 6, -2]]
n = len(grid[0])
widths = []
for j in range(n):
    lengths = [len(str(row[j])) for row in grid]
    print(f"cột {j}: độ dài từng ô = {lengths} -> rộng {max(lengths)}")
    widths.append(max(lengths))
print(widths)
````

## 5. Cách 1 — Brute Force / baseline: duyệt theo cột, hai vòng lặp lồng nhau

**Ý tưởng:** vòng ngoài `j` chạy qua từng **cột**, vòng trong `i` chạy qua từng **hàng**; dùng `len(str(grid[i][j]))` và cập nhật `width = max(width, ...)`. Đúng theo đề, dễ hiểu.

````python
from typing import List

class SolutionColumns:
    def findColumnWidth(self, grid: List[List[int]]) -> List[int]:
        m, n = len(grid), len(grid[0])
        ans = []
        for j in range(n):                 # từng cột
            width = 0
            for i in range(m):             # từng hàng trong cột đó
                width = max(width, len(str(grid[i][j])))
            ans.append(width)
        return ans
````

````python
sol = SolutionColumns()
print(sol.findColumnWidth([[1], [22], [333]]))                       # [3]
print(sol.findColumnWidth([[-15, 1, 3], [15, 7, 12], [5, 6, -2]]))   # [3, 1, 2]
````

**Độ phức tạp:** `m` hàng, `n` cột. Mỗi ô được xem đúng một lần, mỗi lần đổi sang chuỗi tốn `O(d)` với `d <= 11` (hằng số) ⇒ Time `O(m · n)`. Space `O(1)` phụ (ngoài mảng kết quả `O(n)`).

**Nút thắt:** thật ra không có nút thắt lớn về độ phức tạp — mọi ô đều phải được nhìn ít nhất một lần nên `O(m · n)` là tối ưu. Việc còn lại là **viết gọn hơn** và **tránh đổi sang chuỗi** nếu muốn.

## 6. Ý tưởng thuật toán / bước đột phá

Có ba hướng cải tiến, đều giữ `O(m · n)`:

1. **Chuyển vị**: `zip(*grid)` cho ta từng cột; mỗi cột chỉ cần `max(len(str(x)) for x in col)` ⇒ code một dòng.
2. **Một lượt theo hàng**: duyệt `grid` theo hàng (kiểu truy cập tự nhiên của Python, nhanh hơn duyệt theo cột), giữ mảng `ans` kích thước `n` và cập nhật `ans[j] = max(ans[j], len(str(x)))`.
3. **Chỉ xét `max` và `min` của cột.** Trong một cột, số **dài nhất** chỉ có thể là số **lớn nhất** (nhiều chữ số nhất trong các số dương) hoặc số **nhỏ nhất** (nhiều chữ số nhất trong các số âm, cộng dấu `-`). Vì vậy `width = max(len(str(max(col))), len(str(min(col))))` — chỉ đổi sang chuỗi hai lần mỗi cột. Phần chứng minh ở mục 10.

````python
col = [-15, 15, 5, -3]
print("max =", max(col), "min =", min(col))
print("đáp án vét cạn :", max(len(str(x)) for x in col))
print("chỉ max và min :", max(len(str(max(col))), len(str(min(col)))))
````

## 7. Các cách tiếp theo

### Cách 2 — Chuyển vị bằng `zip(*grid)`

`zip(*grid)` cho các cột. Với mỗi cột, độ rộng là `max(len(str(x)) for x in col)`. Cả hàm gói trong một list comprehension.

````python
class SolutionZip:
    def findColumnWidth(self, grid: List[List[int]]) -> List[int]:
        return [max(len(str(x)) for x in col) for col in zip(*grid)]

print(SolutionZip().findColumnWidth([[1], [22], [333]]))                       # [3]
print(SolutionZip().findColumnWidth([[-15, 1, 3], [15, 7, 12], [5, 6, -2]]))   # [3, 1, 2]
````

**Độ phức tạp:** `zip(*grid)` tạo các tuple cột, tổng kích thước `m · n` ⇒ Time `O(m · n)`, Space `O(m · n)` nếu tạo toàn bộ cột cùng lúc (thực tế `zip` sinh lười từng cột nên bộ nhớ đỉnh `O(m)` ở mỗi lần, nhưng `*grid` mở `m` đối số).

### Cách 3 — Một lượt theo hàng, cập nhật mảng `ans`

Khởi tạo `ans = [0] * n`. Duyệt từng hàng, từng ô `x` ở cột `j`: `ans[j] = max(ans[j], len(str(x)))`. Không cần tạo cột, không cần chỉ số `[i][j]` theo kiểu cột trước hàng sau.

````python
class SolutionRowWise:
    def findColumnWidth(self, grid: List[List[int]]) -> List[int]:
        ans = [0] * len(grid[0])
        for row in grid:
            for j, x in enumerate(row):
                ans[j] = max(ans[j], len(str(x)))
        return ans

print(SolutionRowWise().findColumnWidth([[1], [22], [333]]))                       # [3]
print(SolutionRowWise().findColumnWidth([[-15, 1, 3], [15, 7, 12], [5, 6, -2]]))   # [3, 1, 2]
````

**Độ phức tạp:** Time `O(m · n)`, Space `O(1)` phụ ngoài `ans` (`O(n)`). Đây là cách ta nộp: duyệt theo hàng thân thiện với bộ nhớ, không cần bản sao.

### Cách 4 — Không dùng chuỗi: đếm chữ số bằng phép chia

Hàm `digit_len(x)`:

- Nếu `x < 0`: độ rộng cần cộng dấu, vậy `1 + (số chữ số của |x|)`.
- Nếu `x == 0`: trả `1`.
- Ngược lại: đếm số lần chia `10` để về `0`.

````python
def digit_len(x: int) -> int:
    if x == 0:
        return 1
    length = 0
    if x < 0:
        length = 1          # chỗ cho dấu '-'
        x = -x
    while x > 0:
        x //= 10
        length += 1
    return length

print([digit_len(v) for v in [0, 7, 15, -15, -2, 1000000000, -1000000000]])   # [1, 1, 2, 3, 2, 10, 11]

class SolutionDigits:
    def findColumnWidth(self, grid: List[List[int]]) -> List[int]:
        ans = [0] * len(grid[0])
        for row in grid:
            for j, x in enumerate(row):
                ans[j] = max(ans[j], digit_len(x))
        return ans

print(SolutionDigits().findColumnWidth([[-15, 1, 3], [15, 7, 12], [5, 6, -2]]))   # [3, 1, 2]
````

**Độ phức tạp:** mỗi ô tốn tối đa `10` lần chia (hằng số) ⇒ Time `O(m · n)`; Space `O(1)` phụ. Thường **chậm hơn** `len(str(x))` trong Python (vòng lặp thuần Python so với hàm viết bằng C), nhưng cho thấy ý nghĩa của "số chữ số".

### Cách 5 — Chỉ xét `max` và `min` của từng cột

Dựa trên nhận xét ở mục 6: độ dài lớn nhất trong cột chỉ có thể đạt tại phần tử lớn nhất hoặc nhỏ nhất của cột. Mỗi cột chỉ cần hai phép `len(str(...))`.

````python
class SolutionMinMax:
    def findColumnWidth(self, grid: List[List[int]]) -> List[int]:
        return [max(len(str(max(col))), len(str(min(col)))) for col in zip(*grid)]

print(SolutionMinMax().findColumnWidth([[1], [22], [333]]))                       # [3]
print(SolutionMinMax().findColumnWidth([[-15, 1, 3], [15, 7, 12], [5, 6, -2]]))   # [3, 1, 2]
````

**Độ phức tạp:** vẫn phải đọc cả cột để tìm `max`/`min` ⇒ Time `O(m · n)`, nhưng số lần đổi chuỗi giảm từ `m · n` xuống `2n`. Space `O(m)` cho mỗi cột (`zip`). Đây là mẹo thú vị, không bắt buộc.

## 8. TRACE từng bước

Trace Cách 3 (một lượt theo hàng) trên `[[-15,1,3],[15,7,12],[5,6,-2]]`. Cột "độ dài" là `len(str(x))`; cột `ans[j]` sau là giá trị sau khi cập nhật:

| hàng `i` | cột `j` | `x` | `len(str(x))` | `ans[j]` trước | `ans[j]` sau |
|---|---|---|---|---|---|
| 0 | 0 | `-15` | 3 | 0 | 3 |
| 0 | 1 | `1` | 1 | 0 | 1 |
| 0 | 2 | `3` | 1 | 0 | 1 |
| 1 | 0 | `15` | 2 | 3 | 3 |
| 1 | 1 | `7` | 1 | 1 | 1 |
| 1 | 2 | `12` | 2 | 1 | 2 |
| 2 | 0 | `5` | 1 | 3 | 3 |
| 2 | 1 | `6` | 1 | 1 | 1 |
| 2 | 2 | `-2` | 2 | 2 | 2 |

Kết quả `[3, 1, 2]`.

````python
def trace_row_wise(grid):
    ans = [0] * len(grid[0])
    for i, row in enumerate(grid):
        for j, x in enumerate(row):
            length = len(str(x))
            before = ans[j]
            ans[j] = max(ans[j], length)
            print(f"i={i} j={j} x={x:>4} len={length} ans[{j}]: {before} -> {ans[j]}   ans={ans}")
    print("kết quả:", ans)
    return ans

trace_row_wise([[-15, 1, 3], [15, 7, 12], [5, 6, -2]])
````

````python
def trace_digit_len(x):
    print(f"digit_len({x}):")
    if x == 0:
        print("  x == 0 -> 1")
        return 1
    length = 0
    if x < 0:
        length, x = 1, -x
        print(f"  số âm: length = 1 (dấu '-'), x đổi thành {x}")
    while x > 0:
        print(f"  x = {x}, bỏ chữ số cuối -> {x // 10}")
        x //= 10
        length += 1
    print("  length =", length)
    return length

trace_digit_len(-203)
trace_digit_len(0)
````

## 9. Code nộp LeetCode

Dùng Cách 3 (một lượt theo hàng, `len(str(x))`): ngắn, dễ kiểm tra, `O(m · n)`.

````python
from typing import List

class Solution:
    def findColumnWidth(self, grid: List[List[int]]) -> List[int]:
        ans = [0] * len(grid[0])
        for row in grid:
            for j, x in enumerate(row):
                ans[j] = max(ans[j], len(str(x)))
        return ans
````

````python
print(Solution().findColumnWidth([[1], [22], [333]]))                       # [3]
print(Solution().findColumnWidth([[-15, 1, 3], [15, 7, 12], [5, 6, -2]]))   # [3, 1, 2]
````

## 10. Vì sao thuật toán đúng

1. **Định nghĩa khớp.** `len(str(x))` là số ký tự khi in `x`: với `x >= 0` đúng bằng số chữ số (`0` → `"0"` có độ dài 1); với `x < 0` là số chữ số của `|x|` cộng một dấu `-`. Điều này trùng khớp đúng định nghĩa "độ dài" của đề.
2. **Invariant của vòng lặp.** Sau khi xử lý xong các hàng `0..i`, `ans[j]` bằng độ dài lớn nhất của các ô `grid[0][j], ..., grid[i][j]`. Ban đầu (`i = -1`, chưa xử lý hàng nào) `ans[j] = 0` hợp lý vì độ dài thực luôn `>= 1`. Mỗi ô mới `x` cập nhật `ans[j] = max(ans[j], len(str(x)))` giữ nguyên invariant.
3. **Kết quả.** Sau hàng cuối, `ans[j]` là độ dài lớn nhất cả cột `j`, đúng yêu cầu.
4. **Vì sao mẹo `min` / `max` (Cách 5) đúng.** Trong các số `>= 0`, độ dài không giảm khi giá trị tăng (nhiều chữ số hơn ⇒ lớn hơn) nên độ dài lớn nhất của nhóm này đạt ở số **lớn nhất**. Trong các số `< 0`, độ dài là `1 + số chữ số của |x|`, tăng khi `|x|` tăng tức khi `x` **nhỏ** đi, nên đạt ở số **nhỏ nhất**. Nếu cột có cả hai nhóm thì `max(col)` là max của nhóm dương (hoặc là số âm gần `0` nhất khi toàn âm — đã có `min(col)` lo), còn `min(col)` là min của nhóm âm (hoặc là số dương nhỏ nhất khi toàn dương, độ dài không vượt max). Vậy lấy `max(len(str(max)), len(str(min)))` luôn bao được độ dài lớn nhất.

## 11. Phân tích Time / Space Complexity

`m` = số hàng, `n` = số cột; mỗi số có tối đa 11 ký tự (hằng số).

| Cách | Time | Space (phụ) |
|---|---|---|
| 1. Duyệt theo cột | `O(m · n)` | `O(1)` |
| 2. `zip(*grid)` | `O(m · n)` | `O(m)` đến `O(m · n)` |
| 3. Một lượt theo hàng | `O(m · n)` | `O(1)` |
| 4. Đếm chữ số bằng phép chia | `O(m · n)` (hằng số lớn hơn) | `O(1)` |
| 5. `max` và `min` của cột | `O(m · n)` | `O(m)` |

Mọi cách đều phải đọc cả `m · n` ô nên không thể tốt hơn `O(m · n)` về lý thuyết. Với `m, n <= 100`, tối đa `10^4` ô — chạy tức thì.

### 12.2. Đối chiếu ngẫu nhiên (có seed)

Sinh ma trận đúng constraints: `1 <= m, n <= 100`, giá trị `[-10^9, 10^9]`. Để cột có độ dài đa dạng, mỗi lần sinh chọn ngẫu nhiên một "bậc độ lớn" cho từng ô. Oracle viết bằng định nghĩa: `len(str(x))` theo từng cột bằng chỉ số rõ ràng.

````python
import random
random.seed(2639)

def oracle(grid):
    m, n = len(grid), len(grid[0])
    out = []
    for j in range(n):
        best = 0
        for i in range(m):
            x = grid[i][j]
            digits = 1 if x == 0 else len(str(abs(x)))
            length = digits + (1 if x < 0 else 0)
            best = max(best, length)
        out.append(best)
    return out

def rand_val():
    k = random.randint(0, 9)              # số chữ số tối đa
    v = random.randint(0, 10 ** k)
    v = min(v, 10 ** 9)
    return v if random.random() < 0.5 else -v

for _ in range(500):
    m, n = random.randint(1, 8), random.randint(1, 8)
    g = [[rand_val() for _ in range(n)] for _ in range(m)]
    exp = oracle(g)
    for s in sols:
        got = s.findColumnWidth([row[:] for row in g])
        assert got == exp, (type(s).__name__, g, got, exp)
print("500 ma trận ngẫu nhiên: mọi cách đều khớp oracle")
````

````python
random.seed(7)
g = [[rand_val() for _ in range(100)] for _ in range(100)]
res = Solution().findColumnWidth(g)
assert len(res) == 100
assert all(1 <= w <= 11 for w in res)
assert res == oracle(g) == SolutionMinMax().findColumnWidth(g)
print("100x100 OK, vài cột đầu:", res[:10])
````

## 13. Các lỗi người mới hay gặp

### Lỗi 1: Chỉ đếm chữ số, quên dấu `-` của số âm

````python
from typing import List

def sai_quen_dau(grid: List[List[int]]) -> List[int]:
    ans = [0] * len(grid[0])
    for row in grid:
        for j, x in enumerate(row):
            ans[j] = max(ans[j], len(str(abs(x))))     # abs làm mất dấu '-'
    return ans

g = [[-15, 1, 3], [15, 7, 12], [5, 6, -2]]
print("sai :", sai_quen_dau(g))                         # [2, 1, 2]
print("đúng:", Solution().findColumnWidth(g))           # [3, 1, 2]
````

**Vì sao sai:** đề tính độ dài của `-15` là `3`, tính cả dấu. `abs(x)` bỏ dấu nên cột 0 chỉ còn `2`. Dùng thẳng `len(str(x))` là an toàn nhất.

### Lỗi 2: Đếm số `0` là có `0` chữ số

````python
def sai_chia_cho_zero(grid: List[List[int]]) -> List[int]:
    def digits(x):
        x = abs(x)
        c = 0
        while x > 0:          # với x = 0 vòng lặp không chạy
            x //= 10
            c += 1
        return c
    ans = [0] * len(grid[0])
    for row in grid:
        for j, x in enumerate(row):
            ans[j] = max(ans[j], digits(x) + (1 if x < 0 else 0))
    return ans

print("sai :", sai_chia_cho_zero([[0, 5], [0, 7]]))     # [0, 1]: cột 0 phải rộng 1
print("đúng:", Solution().findColumnWidth([[0, 5], [0, 7]]))   # [1, 1]
````

**Vì sao sai:** `0` được in là `"0"` — một ký tự. Vòng `while x > 0` không chạy với `x = 0` nên cho `0`. Phải xử lý `x == 0` riêng (như hàm `digit_len` ở Cách 4) hoặc dùng `len(str(x))`.

### Lỗi 3: Ghi đè thay vì lấy `max`

````python
def sai_ghi_de(grid: List[List[int]]) -> List[int]:
    ans = [0] * len(grid[0])
    for row in grid:
        for j, x in enumerate(row):
            ans[j] = len(str(x))        # lấy giá trị của hàng cuối cùng
    return ans

g = [[1000], [5]]
print("sai :", sai_ghi_de(g))                           # [1]
print("đúng:", Solution().findColumnWidth(g))           # [4]
````

**Vì sao sai:** đề yêu cầu độ dài **lớn nhất** của cột. Gán trực tiếp chỉ giữ lại độ dài của hàng cuối. Luôn `ans[j] = max(ans[j], ...)`.

### Lỗi 4: Nhầm số hàng và số cột

````python
def sai_nham_kich_thuoc(grid: List[List[int]]) -> List[int]:
    ans = [0] * len(grid)            # len(grid) là số HÀNG, không phải số cột
    for row in grid:
        for j, x in enumerate(row):
            ans[j] = max(ans[j], len(str(x)))
    return ans

try:
    sai_nham_kich_thuoc([[1, 22, 333]])      # 1 hàng, 3 cột -> ans chỉ có 1 phần tử
except IndexError as e:
    print("lỗi:", e)
print("đúng:", Solution().findColumnWidth([[1, 22, 333]]))   # [1, 2, 3]
````

**Vì sao sai:** kết quả có `n` phần tử (số **cột**), nên phải là `len(grid[0])`. Khi `m != n` lỗi này làm `IndexError` hoặc kết quả sai độ dài. Hãy thử một ma trận không vuông.

## 14. Mẹo làm bài nhanh & nhận dạng pattern

- "Độ dài lớn nhất của từng cột/hàng" ⇒ vòng lặp lồng nhau với `max`, hoặc `zip(*grid)` để lấy cột.
- Chuyển vị ma trận trong Python: `list(zip(*grid))`. Lưu ý nó trả **tuple**; cần list thì `list(map(list, zip(*grid)))`.
- Số ký tự khi in số nguyên = `len(str(x))` (đã gồm dấu trừ) — không cần tự cộng `1` cho số âm.
- Mọi bài ma trận: viết sẵn `m, n = len(grid), len(grid[0])` để tránh nhầm hàng/cột; kiểm tra bằng ma trận **không vuông**.
- Khi muốn căn lề cột đẹp: `str(x).rjust(width)` hoặc `f"{x:>{width}}"`.

## 15. Biến thể của bài

1. **In ma trận thẳng cột:** dùng `ans` để `rjust` từng ô rồi `print` — đúng ứng dụng thực tế của "độ rộng cột".
2. **Độ rộng hàng:** thay `zip(*grid)` bằng duyệt từng hàng.
3. **Maximum Number of Words Found in Sentences (LeetCode 2114)** và **Longest Word in Dictionary**: "độ dài lớn nhất" trên danh sách.
4. **Transpose Matrix (LeetCode 867):** chuyển vị chính thức.
5. **Cộng tổng độ rộng:** tổng các phần tử `ans` cho chiều rộng cả bảng.

````python
# Bài 1
assert len(str(-100)) == 4
assert len(str(0)) == 1
assert len(str(10**9)) == 10
assert list(zip(*[[1, 2], [3, 4]])) == [(1, 3), (2, 4)]

# Bài 2
g = [[0, -100], [999, 5]]
assert [[len(str(x)) for x in row] for row in g] == [[1, 4], [3, 1]]
assert Solution().findColumnWidth(g) == [3, 4]

# Bài 3
def digit_len_ans(x):
    if x == 0:
        return 1
    length = 0
    if x < 0:
        length = 1
        x = -x
    while x > 0:
        x //= 10
        length += 1
    return length

for v in [0, 5, -5, 10, -10, 99, 100, -100, 10**9, -10**9]:
    assert digit_len_ans(v) == len(str(v)), v

# Bài 4
def column_widths_zip(grid):
    return [max(len(str(x)) for x in col) for col in zip(*grid)]

assert column_widths_zip([[1], [22], [333]]) == [3]
assert column_widths_zip([[-15, 1, 3], [15, 7, 12], [5, 6, -2]]) == [3, 1, 2]

# Bài 5
def format_table(grid):
    widths = Solution().findColumnWidth(grid)
    lines = []
    for row in grid:
        lines.append(" ".join(str(x).rjust(w) for x, w in zip(row, widths)))
    return "\n".join(lines)

expected = "-15 1  3\n 15 7 12\n  5 6 -2"
assert format_table([[-15, 1, 3], [15, 7, 12], [5, 6, -2]]) == expected

def row_widths(grid):
    return [max(len(str(x)) for x in row) for row in grid]

assert row_widths([[-15, 1, 3], [15, 7, 12], [5, 6, -2]]) == [3, 2, 2]
print(format_table([[-15, 1, 3], [15, 7, 12], [5, 6, -2]]))
print("Tất cả đáp án đúng")
````
