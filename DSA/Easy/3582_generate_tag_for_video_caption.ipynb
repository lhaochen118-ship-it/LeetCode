## 1. Mục tiêu học

Sau notebook này bạn sẽ:

1. Hiểu đề: biến một câu mô tả video thành **hashtag dạng camelCase**: bắt đầu bằng `#`, từ đầu viết thường hết, các từ sau viết hoa chữ đầu và thường các chữ còn lại, rồi **cắt tối đa 100 ký tự** (tính cả `#`).
2. Thành thạo các hàm chuỗi: `split()`, `lower()`, `upper()`, `capitalize()`, `title()`, `join`, và cắt lát `s[:100]`.
3. Hiểu sự khác nhau giữa `split()` (bỏ mọi khoảng trắng thừa) và `split(" ")` (sinh ra chuỗi rỗng khi có nhiều dấu cách).
4. Viết 4 cách: `split` + vòng lặp, **một lượt duyệt ký tự** (máy trạng thái, dừng sớm khi đủ 100), `capitalize` + `join`, và mẹo `title()`.
5. Tránh các lỗi: quên cắt 100 ký tự (hoặc cắt sai chỗ `#`), viết hoa luôn từ đầu, không hạ chữ thường phần còn lại của từ, và `split(" ")` gây lỗi với nhiều dấu cách.

## 2. Đề bài & ví dụ

Cho chuỗi `caption` mô tả một video. Tạo **tag hợp lệ** bằng cách thực hiện các bước sau **theo thứ tự**:

1. **Ghép các từ** thành một chuỗi camelCase, đặt `#` ở đầu. Từ đầu tiên viết **thường hoàn toàn**; mỗi từ sau viết **hoa chữ cái đầu**, các chữ còn lại **thường**.
2. **Xóa** mọi ký tự không phải chữ cái tiếng Anh (trừ `#` ở đầu).
3. **Cắt** kết quả còn tối đa **100 ký tự**.

Trả về tag đã tạo.

| Ví dụ | caption | Output | Giải thích |
|---|---|---|---|
| 1 | `"Leetcode daily streak achieved"` | `"#leetcodeDailyStreakAchieved"` | từ đầu `leetcode` thường; ba từ sau viết hoa chữ đầu |
| 2 | `"can I Go There"` | `"#canIGoThere"` | `I` -> `I`, `Go` -> `Go`, `There` -> `There`; `can` thường |
| 3 | một từ gồm `101` chữ `h` | `"#"` + `99` chữ `h` | từ dài `101` nên bị cắt: tag tối đa `100` ký tự **kể cả `#`** |

Constraints: `1 <= caption.length <= 150`; `caption` chỉ gồm chữ cái tiếng Anh (hoa/thường) và dấu cách `' '`.

Chú ý: vì `caption` chỉ có chữ cái và dấu cách, bước "xóa ký tự không phải chữ cái" thực chất **chỉ xóa dấu cách** (khi ghép từ). Caption có thể có **dấu cách thừa** ở đầu, cuối, hoặc nhiều dấu cách liền nhau.

````python
examples = [("Leetcode daily streak achieved", "#leetcodeDailyStreakAchieved"), ("can I Go There", "#canIGoThere")]
for caption, expected in examples:
    words = caption.split()
    tag = "#"
    for idx, w in enumerate(words):
        tag += w.lower() if idx == 0 else w[0].upper() + w[1:].lower()
    print(f"{caption!r} -> {words} -> {tag!r} (mong đợi {expected!r})")
long_caption = "h" * 101
print(len(long_caption), "chữ h ->", len(("#" + long_caption)[:100]), "ký tự sau khi cắt")
````

## 3. Kiến thức nền

### 3.1. `split()` và `split(" ")`

`s.split()` (không đối số) tách theo **mọi dãy khoảng trắng** và **bỏ** phần rỗng ở đầu/cuối, nên `"  a   b ".split()` cho `['a', 'b']`. Còn `s.split(" ")` tách theo **từng** dấu cách một và sinh ra chuỗi rỗng `''` cho mỗi dấu cách thừa: `"  a   b ".split(" ")` cho `['', '', 'a', '', '', 'b', '']`. Bài này nên dùng `split()`.

````python
s = "  a   b "
print(s.split())
print(s.split(" "))
````

### 3.2. `lower`, `upper`, `capitalize`, `title`

- `s.lower()` đổi mọi chữ thành thường, `s.upper()` thành hoa.
- `s.capitalize()` viết hoa chữ **đầu**, và hạ **thường** mọi chữ còn lại: `"hELLO".capitalize()` là `"Hello"`.
- `s.title()` viết hoa chữ đầu của **mỗi từ** (từ tách bởi ký tự không phải chữ), hạ thường phần còn lại.

````python
print("hELLO".lower(), "hello".upper(), "hELLO".capitalize())
print("hELLO wORLD".title())
print("I".capitalize(), "i".capitalize())
````

### 3.3. Cắt lát để giới hạn độ dài

`s[:100]` lấy tối đa 100 ký tự đầu; nếu `s` ngắn hơn 100 thì trả nguyên `s` (không lỗi). Ở bài này `#` **được tính vào** 100 ký tự, nên cắt trên **cả chuỗi đã có `#`**: `("#" + body)[:100]`, không phải `"#" + body[:100]` (sẽ dài `101`).

````python
body = "h" * 101
print(len(("#" + body)[:100]), len("#" + body[:100]))
print(repr("abc"[:100]))
````

### 3.4. `"".join(list)` và danh sách gom từng phần

Ghép nhiều chuỗi bằng `"".join(parts)` nhanh hơn `+=` lặp lại (nhất là chuỗi dài). Thường ta dựng `parts = [...]` rồi `join` một lần. Ở đây `parts` là các từ đã chuẩn hóa.

````python
parts = ["leetcode", "Daily", "Streak"]
print("".join(parts), "#" + "".join(parts))
````

### 3.5. Máy trạng thái đơn giản khi duyệt ký tự

Thay vì tách từ, ta có thể duyệt từng ký tự và nhớ hai cờ: `started` (đã gặp chữ cái đầu tiên chưa) và `capital_next` (ký tự chữ kế tiếp có cần viết hoa không). Gặp dấu cách (sau khi đã `started`) thì bật `capital_next`. Cách này dừng được sớm ngay khi đủ 100 ký tự.

````python
caption = "ab  cd"
out, started, capital_next = [], False, False
for ch in caption:
    if ch == " ":
        capital_next = started
        continue
    out.append(ch.upper() if capital_next else ch.lower())
    started, capital_next = True, False
print("".join(out))
````

## 4. Ý tưởng nền tảng / cách nghĩ trực giác

Làm tay với `"can I Go There"`:

```
tách từ        : can | I | Go | There
chuẩn hóa      : can (thường) | I (hoa đầu, còn lại thường) | Go | There
ghép + thêm #  : #canIGoThere
độ dài 12 <= 100 -> giữ nguyên
```

Quy trình gồm ba bước độc lập: (1) **tách từ**; (2) **chuẩn hóa từng từ** (từ đầu thường, từ sau `capitalize`); (3) **ghép, thêm `#`, cắt 100**. Không có thuật toán hay cấu trúc dữ liệu đặc biệt; điểm quan trọng là các **chi tiết nhỏ**: `#` có tính vào giới hạn 100, từ đầu viết **thường**, phần còn lại của mỗi từ phải về **thường**, và dấu cách thừa không được tạo ra từ rỗng.

## 5. Cách 1 — Brute Force / baseline

Tách từ bằng `split()`; duyệt từng từ với chỉ số `idx`: từ đầu (`idx == 0`) thì `w.lower()`; từ sau thì `w[0].upper() + w[1:].lower()`; nối vào chuỗi kết quả bắt đầu bằng `"#"`; cuối cùng cắt `[:100]`. Đây là bản dịch trực tiếp của đề.

````python
class SolutionSplit:
    def generateTag(self, caption):
        words = caption.split()
        tag = "#"
        for idx, w in enumerate(words):
            if idx == 0:
                tag += w.lower()
            else:
                tag += w[0].upper() + w[1:].lower()
        return tag[:100]

print(SolutionSplit().generateTag("Leetcode daily streak achieved"))
print(SolutionSplit().generateTag("can I Go There"))
print(len(SolutionSplit().generateTag("h" * 101)))
````

**Độ phức tạp:** gọi `n = len(caption)`. `split` `O(n)`, mỗi từ xử lý `O(độ dài từ)` -> tổng `O(n)`. Nối chuỗi bằng `+=` trong CPython thường được tối ưu nên vẫn `O(n)` ở đây (`n <= 150`). Bộ nhớ `O(n)`.

## 6. Ý tưởng thuật toán / bước đột phá

Không có "bước đột phá" về độ phức tạp (mọi cách đều `O(n)`); ta cải thiện **độ gọn và độ an toàn**:

- **`capitalize` làm đúng việc của từ sau:** viết hoa chữ đầu, hạ thường phần còn lại, ngắn gọn hơn `w[0].upper() + w[1:].lower()` và không lỗi với chuỗi ngắn.
- **Một lượt duyệt ký tự:** không cần tạo danh sách từ; dừng ngay khi đủ 100 ký tự (hữu ích nếu chuỗi rất dài).
- **`title()`:** viết hoa chữ đầu mỗi từ, rồi bỏ dấu cách và hạ thường **một** chữ đầu của kết quả.

**Quy tắc "nếu… thì…":**

- Nếu là **từ đầu tiên** thì viết thường hoàn toàn; nếu là từ sau thì chữ đầu hoa, phần còn lại thường.
- Nếu `caption` chỉ gồm dấu cách thì không có từ nào, tag là `"#"`.
- Sau khi ghép, nếu dài hơn 100 (tính cả `#`) thì cắt còn 100.

**Invariant (máy trạng thái):** `len(out)` luôn là độ dài tag hiện tại (đã tính `#`); khi `len(out) == 100` ta dừng.

````python
caption = "  hELLO   wORLD  "
words = caption.split()
print(words)
print([w.capitalize() for w in words])
print("#" + words[0].lower() + "".join(w.capitalize() for w in words[1:]))
````

## 7. Các cách tiếp theo

Ta có 4 cách: (1) `split` + vòng lặp; (2) một lượt duyệt ký tự (máy trạng thái); (3) `capitalize` + `join`; (4) mẹo `title()`.

### Cách 2 — Một lượt duyệt ký tự với dừng sớm

Duyệt từng ký tự. Dấu cách: nếu đã bắt đầu thì bật `capital_next`, rồi bỏ qua. Chữ cái: viết hoa nếu `capital_next`, ngược lại viết thường; tắt `capital_next`, bật `started`. Khi `len(out) == 100` thì `break`. Chữ cái đầu tiên luôn thường vì `capital_next` chỉ bật **sau** khi `started`.

````python
class SolutionStateMachine:
    def generateTag(self, caption):
        out = ["#"]
        started = False
        capital_next = False
        for ch in caption:
            if ch == " ":
                capital_next = started          # dấu cách trước từ đầu không làm gì
                continue
            out.append(ch.upper() if capital_next else ch.lower())
            started = True
            capital_next = False
            if len(out) == 100:                 # đã đủ 100 ký tự kể cả '#'
                break
        return "".join(out)

print(SolutionStateMachine().generateTag("Leetcode daily streak achieved"))
print(SolutionStateMachine().generateTag("can I Go There"))
print(len(SolutionStateMachine().generateTag("h" * 101)))
````

Độ phức tạp Cách 2: `O(n)` thời gian (dừng sớm nếu đủ 100), `O(1)` bộ nhớ ngoài kết quả (tối đa 100 ký tự).

### Cách 3 — `capitalize` + `join`

Từ mỗi từ trong `caption.split()` lấy `w.capitalize()` (hoa đầu, thường phần còn lại). Từ đầu tiên hạ thường hoàn toàn bằng `parts[0].lower()`. Ghép, thêm `#`, cắt.

````python
class SolutionCapitalize:
    def generateTag(self, caption):
        parts = [w.capitalize() for w in caption.split()]
        if parts:
            parts[0] = parts[0].lower()
        return ("#" + "".join(parts))[:100]

print(SolutionCapitalize().generateTag("Leetcode daily streak achieved"))
print(SolutionCapitalize().generateTag("can I Go There"))
print(len(SolutionCapitalize().generateTag("h" * 101)))
````

Độ phức tạp Cách 3: `O(n)` thời gian, `O(n)` bộ nhớ cho danh sách từ.

### Cách 4 — Mẹo `title()`

`caption.title()` viết hoa chữ đầu mỗi từ (và hạ thường phần còn lại), `"".join(... .split())` bỏ hết khoảng trắng. Kết quả bắt đầu bằng chữ **hoa**; hạ thường đúng chữ đầu: `body[:1].lower() + body[1:]`. Chú ý: `title()` coi mọi ký tự không phải chữ là ranh giới từ; ở bài này `caption` chỉ có chữ và dấu cách nên an toàn (nếu có chữ số hay dấu nháy thì `title()` có thể cho kết quả khác mong đợi).

````python
class SolutionTitle:
    def generateTag(self, caption):
        body = "".join(caption.title().split())
        body = body[:1].lower() + body[1:]
        return ("#" + body)[:100]

print(SolutionTitle().generateTag("Leetcode daily streak achieved"))
print(SolutionTitle().generateTag("can I Go There"))
print(len(SolutionTitle().generateTag("h" * 101)))
````

Độ phức tạp Cách 4: `O(n)` thời gian và bộ nhớ.

## 8. TRACE từng bước

Trace Cách 2 (máy trạng thái) với `caption = "can I Go"`:

| bước | ký tự | `started` | `capital_next` | ký tự thêm vào | `out` |
|---|---|---|---|---|---|
| 1 | `c` | không | không | `c` (thường) | `#c` |
| 2 | `a` | có | không | `a` | `#ca` |
| 3 | `n` | có | không | `n` | `#can` |
| 4 | ` ` | có | **bật** | (bỏ qua) | `#can` |
| 5 | `I` | có | có | `I` (hoa) | `#canI` |
| 6 | ` ` | có | **bật** | (bỏ qua) | `#canI` |
| 7 | `G` | có | có | `G` (hoa) | `#canIG` |
| 8 | `o` | có | không | `o` (thường) | `#canIGo` |

Kết quả `#canIGo`.

````python
def trace_state_machine(caption):
    out = ["#"]
    started = capital_next = False
    for step, ch in enumerate(caption, start=1):
        if ch == " ":
            capital_next = started
            print(f"bước {step}: dấu cách  -> capital_next={capital_next}")
            continue
        add = ch.upper() if capital_next else ch.lower()
        out.append(add)
        print(f"bước {step}: ký tự {ch!r} (capital_next={capital_next}) -> thêm {add!r} -> {''.join(out)!r}")
        started, capital_next = True, False
    return "".join(out)

print("kết quả:", trace_state_machine("can I Go"))
print("kết quả:", trace_state_machine("  hELLO  wORLD"))
````

## 9. Code nộp LeetCode

````python
class Solution:
    def generateTag(self, caption: str) -> str:
        parts = [w.capitalize() for w in caption.split()]
        if parts:
            parts[0] = parts[0].lower()
        return ("#" + "".join(parts))[:100]

print(Solution().generateTag("Leetcode daily streak achieved"), Solution().generateTag("can I Go There"), len(Solution().generateTag("h" * 101)))
````

## 10. Vì sao thuật toán đúng

- **Tách từ:** `caption` chỉ có chữ cái và dấu cách; `split()` trả đúng các dãy chữ cái liên tiếp (từ), bỏ mọi dấu cách thừa nên không có từ rỗng.
- **Chuẩn hóa:** từ đầu `lower()` thỏa "viết thường hoàn toàn"; từ sau `capitalize()` thỏa "chữ đầu hoa, còn lại thường".
- **Xóa ký tự không phải chữ:** khi chỉ có chữ và dấu cách, thao tác này tương đương xóa dấu cách; `join("")` đã loại hết dấu cách.
- **Cắt 100:** đề yêu cầu tag (có `#`) dài tối đa 100, nên cắt `("#" + body)[:100]`; nếu ngắn hơn thì giữ nguyên.
- **Caption chỉ có dấu cách:** không có từ -> `parts` rỗng -> kết quả `"#"` (hợp lệ), và code không bị lỗi vì kiểm tra `if parts` trước khi hạ thường từ đầu.
- **Cách 2:** `capital_next` chỉ bật khi `started` nên chữ cái đầu tiên luôn thường; các chữ còn lại trong từ nhận `lower()`, chữ đầu từ sau nhận `upper()`; `len(out)` tính cả `#` nên dừng ở `100`.
- **Cách 4:** `title()` cho mỗi từ dạng `Hoa + thường`; bỏ khoảng trắng rồi hạ thường chữ đầu cho ra cùng kết quả.

## 11. Phân tích Time / Space Complexity

Gọi `n = len(caption) <= 150`.

| Cách | Time | Space |
|---|---|---|
| 1. `split` + vòng lặp | `O(n)` | `O(n)` |
| 2. Một lượt duyệt ký tự | `O(n)` (dừng sớm khi đủ 100) | `O(1)` ngoài kết quả (<= 100) |
| 3. `capitalize` + `join` | `O(n)` | `O(n)` |
| 4. `title()` | `O(n)` | `O(n)` |

Với `n <= 150` mọi cách tức thì. Cách 2 có lợi thế khi chuỗi rất dài (không cần tạo danh sách từ).

## 12. Bộ test

Test 1: ví dụ đề và các edge case (một từ, nhiều dấu cách, dấu cách đầu/cuối, chỉ có dấu cách, chữ hoa lẫn thường, từ dài `101`, tag đúng `100` và `99` ký tự). Mọi input tuân thủ `1 <= len <= 150`, chỉ chữ cái và dấu cách.

````python
impls = [SolutionSplit(), SolutionStateMachine(), SolutionCapitalize(), SolutionTitle(), Solution()]
tests = [
    ("Leetcode daily streak achieved", "#leetcodeDailyStreakAchieved"),
    ("can I Go There", "#canIGoThere"),
    ("h" * 101, "#" + "h" * 99),                          # cắt còn 100 ký tự kể cả '#'
    ("a", "#a"),                                          # một chữ
    ("A", "#a"),                                          # từ đầu luôn thường
    ("Hello", "#hello"),
    ("HELLO WORLD", "#helloWorld"),                       # hạ thường phần còn lại của mỗi từ
    ("hELLO wORLD", "#helloWorld"),
    ("  a   b  ", "#aB"),                                 # nhiều dấu cách thừa
    (" ", "#"),                                           # chỉ có dấu cách -> không có từ
    ("   ", "#"),
    ("a b c d e", "#aBCDE"),                              # từ một chữ
    ("h" * 99, "#" + "h" * 99),                           # đúng 100 ký tự sau khi thêm '#'
    ("h" * 100, "#" + "h" * 99),                          # 101 ký tự -> cắt bớt một
    ("a " * 75, "#a" + "A" * 74),                         # 150 ký tự vào, 75 từ 'a' -> '#a' + 74 'A' = 76 ký tự
]
for caption, exp in tests:
    assert 1 <= len(caption) <= 150 and all(c.isalpha() or c == " " for c in caption)
    assert len(exp) <= 100
    for im in impls:
        got = im.generateTag(caption)
        assert got == exp, (type(im).__name__, caption[:30], got[:30], exp[:30])
print("tất cả test cố định OK:", len(tests))
````

Test 2: ngẫu nhiên có seed. Sinh `caption` gồm các từ ngẫu nhiên (chữ hoa/thường lẫn lộn) cách nhau bởi `1..3` dấu cách, có thể có dấu cách đầu/cuối, độ dài `1..150`. Oracle độc lập dùng **regex** `re.findall` để lấy các từ (thay vì `split`) và dựng chuỗi từng ký tự.

````python
import random
import re
import string
random.seed(3582)

def oracle(caption):
    words = re.findall(r"[A-Za-z]+", caption)
    out = "#"
    for idx, w in enumerate(words):
        for j, ch in enumerate(w):
            if idx > 0 and j == 0:
                out += ch.upper()
            else:
                out += ch.lower()
    return out[:100]

def random_caption():
    while True:
        target = random.choice([10, 40, 150])
        s = " " * random.randint(0, 2)
        while len(s) < target:
            wl = random.choice([1, 2, 3, 5, 12, 101]) if random.random() < 0.05 else random.randint(1, 8)
            s += "".join(random.choice(string.ascii_letters) for _ in range(wl)) + " " * random.randint(1, 3)
        s = s[:random.randint(1, 150)] if random.random() < 0.5 else s[:150]
        if 1 <= len(s) <= 150:
            return s

long_cases = 0
for t in range(3000):
    caption = random_caption()
    assert 1 <= len(caption) <= 150 and all(c.isalpha() or c == " " for c in caption)
    ref = oracle(caption)
    long_cases += len(ref) == 100
    for im in impls:
        assert im.generateTag(caption) == ref, (type(im).__name__, caption)
print("3000 test ngẫu nhiên OK; số ca bị cắt đúng 100 ký tự:", long_cases)
assert long_cases > 50
````

## 13. Các lỗi người mới hay gặp

### Lỗi 1: quên cắt 100 ký tự, hoặc cắt sai chỗ `#`

````python
def wrong_no_truncate(caption):
    parts = [w.capitalize() for w in caption.split()]
    if parts:
        parts[0] = parts[0].lower()
    return "#" + "".join(parts)                           # SAI: không cắt 100

def wrong_truncate_body(caption):
    parts = [w.capitalize() for w in caption.split()]
    if parts:
        parts[0] = parts[0].lower()
    return "#" + "".join(parts)[:100]                     # SAI: cắt phần thân 100 rồi mới thêm '#'

caption = "h" * 101
print("không cắt   :", len(wrong_no_truncate(caption)))
print("cắt sai chỗ :", len(wrong_truncate_body(caption)))
print("đúng        :", len(Solution().generateTag(caption)))
assert len(wrong_no_truncate(caption)) == 102
assert len(wrong_truncate_body(caption)) == 101
assert len(Solution().generateTag(caption)) == 100
````

Vì sao sai: đề giới hạn **cả tag** (kể cả `#`) tối đa 100 ký tự. Không cắt cho `102`; cắt thân rồi mới thêm `#` cho `101`. **Sửa:** `("#" + body)[:100]`.

### Lỗi 2: viết hoa luôn từ đầu tiên

````python
def wrong_capitalize_first(caption):
    return ("#" + "".join(w.capitalize() for w in caption.split()))[:100]   # SAI: từ đầu cũng bị viết hoa

caption = "Leetcode daily streak achieved"
print("sai :", wrong_capitalize_first(caption))
print("đúng:", Solution().generateTag(caption))
assert wrong_capitalize_first(caption) != Solution().generateTag(caption)
````

Vì sao sai: camelCase yêu cầu **từ đầu viết thường hoàn toàn**: `"#leetcode..."`, không phải `"#Leetcode..."`. **Sửa:** hạ thường từ đầu (`parts[0] = parts[0].lower()`).

### Lỗi 3: không hạ chữ thường phần còn lại của từ

````python
def wrong_keep_rest(caption):
    words = caption.split()
    parts = [w[0].upper() + w[1:] for w in words]          # SAI: phần sau chữ đầu giữ nguyên hoa/thường
    if parts:
        parts[0] = words[0].lower()
    return ("#" + "".join(parts))[:100]

caption = "hELLO wORLD"
print("sai :", wrong_keep_rest(caption))
print("đúng:", Solution().generateTag(caption))
assert wrong_keep_rest(caption) != Solution().generateTag(caption)
````

Vì sao sai: từ sau phải có **các chữ còn lại viết thường**: `"wORLD"` -> `"World"`. Code sai giữ `"ORLD"` nên ra `"#helloWORLD"`. **Sửa:** dùng `capitalize()` (hoặc `w[0].upper() + w[1:].lower()`).

### Lỗi 4: dùng `split(" ")` và truy cập `w[0]` khi có nhiều dấu cách

````python
def wrong_split_space(caption):
    parts = [w[0].upper() + w[1:].lower() for w in caption.split(" ")]   # SAI: có từ rỗng ''
    parts[0] = parts[0].lower()
    return ("#" + "".join(parts))[:100]

caption = "a  b"                                  # hai dấu cách liền nhau
try:
    wrong_split_space(caption)
    crashed = False
except IndexError:
    crashed = True
print("code sai có bị IndexError với", repr(caption), ":", crashed)
print("đúng:", Solution().generateTag(caption))
assert crashed and Solution().generateTag(caption) == "#aB"
print("split(' ') của 'a  b':", "a  b".split(" "), "| split():", "a  b".split())
````

Vì sao sai: `"a  b".split(" ")` cho `['a', '', 'b']`; từ rỗng `''` làm `w[0]` ném `IndexError`. **Sửa:** dùng `split()` không đối số (bỏ mọi khoảng trắng thừa), hoặc lọc `if w`.

## 14. Mẹo làm bài nhanh & nhận dạng pattern

- Bài "định dạng chuỗi theo quy tắc" -> làm **từng bước đúng thứ tự đề** và test các chi tiết nhỏ.
- `split()` không đối số để tách từ an toàn; `split(" ")` chỉ dùng khi thật sự muốn giữ từ rỗng.
- `capitalize()` = hoa đầu + thường phần còn lại; `title()` tương tự cho mọi từ nhưng coi mọi ký tự không phải chữ là ranh giới.
- Giới hạn độ dài **bao gồm cả `#`** -> cắt trên chuỗi đã có `#`.
- Từ đầu của camelCase **thường hoàn toàn**.
- Caption chỉ có dấu cách -> trả `"#"`, đừng để `parts[0]` gây `IndexError`.
- Bài họ hàng: `2129. Capitalize the Title`, `824. Goat Latin`, `1816. Truncate Sentence`, `709. To Lower Case`.

## 15. Biến thể của bài

1. **snake_case / kebab-case:** nối bằng `_` hoặc `-` thay vì viết hoa chữ đầu.
2. **Giới hạn độ dài tùy ý** `L`, hoặc **chỉ giữ nguyên từ** (không cắt giữa từ): bỏ các từ không vừa.
3. Caption có **chữ số và dấu câu**: phải xóa ký tự không phải chữ (hoặc giữ chữ số), và chú ý `title()` coi chúng là ranh giới từ.
4. **PascalCase:** cả từ đầu cũng viết hoa chữ đầu.
5. Tạo tag cho **nhiều** caption và loại tag **trùng** (dùng `set`).

## 16. So sánh các cách làm

| Cách | Ý tưởng | Time | Space | Khi nào dùng |
|---|---|---|---|---|
| 1. `split` + vòng lặp | chuẩn hóa từng từ bằng `lower`/`upper` | `O(n)` | `O(n)` | rõ ràng, dễ debug |
| 2. Một lượt duyệt ký tự | máy trạng thái, dừng khi đủ 100 | `O(n)` | `O(1)` ngoài kết quả | chuỗi rất dài, stream |
| 3. `capitalize` + `join` | `capitalize` từng từ, hạ thường từ đầu | `O(n)` | `O(n)` | **gọn, nên nộp** |
| 4. `title()` | `title`, bỏ khoảng trắng, hạ thường chữ đầu | `O(n)` | `O(n)` | mẹo ngắn (cẩn thận ký tự lạ) |
