## 2. Đề bài & ví dụ

- `sumEven`: tổng của `n` số nguyên **chẵn dương** đầu tiên.

Trả về `GCD(sumOdd, sumEven)`.

| Ví dụ | n | Output | Giải thích |

````python
import math
n = 4
odds = [2 * i + 1 for i in range(n)]
evens = [2 * i + 2 for i in range(n)]
print("lẻ :", odds, "tổng =", sum(odds))
print("chẵn:", evens, "tổng =", sum(evens))
print("gcd =", math.gcd(sum(odds), sum(evens)))
n = 5
print(sum(range(1, 2 * n, 2)), sum(range(2, 2 * n + 1, 2)), math.gcd(sum(range(1, 2 * n, 2)), sum(range(2, 2 * n + 1, 2))))
````

## 3. Kiến thức nền

### 3.1. "`n` số lẻ đầu tiên" nghĩa là gì?

Số lẻ thứ `i` (đếm từ `0`) là `2i + 1`; số chẵn dương thứ `i` là `2i + 2`. "`n` số đầu tiên" là `n` **phần tử**, **không phải** các số `<= n`. Ví dụ `n = 4`: `1, 3, 5, 7` (đến `7 = 2n - 1`) và `2, 4, 6, 8` (đến `8 = 2n`).

````python
n = 4
print([2 * i + 1 for i in range(n)], "-> số lẻ cuối =", 2 * n - 1)
print([2 * i + 2 for i in range(n)], "-> số chẵn cuối =", 2 * n)
print(list(range(1, 2 * n, 2)), list(range(2, 2 * n + 1, 2)))     # range với bước 2
````

`range(1, 2*n, 2)` cho `1, 3, ..., 2n-1` (đúng `n` số). `range(2, 2*n + 1, 2)` cho `2, 4, ..., 2n`. Cận `stop` **không** được lấy nên số chẵn phải dừng ở `2n + 1`. Đây là chỗ hay lệch một.

````python
n = 5
print(len(range(1, 2 * n, 2)), len(range(2, 2 * n + 1, 2)))
print(len(range(2, 2 * n, 2)), "<- thiếu một số chẵn (lỗi lệch một)")
````

### 3.3. Ước chung lớn nhất (GCD) và thuật toán Euclid

`gcd(a, b)` là số nguyên dương lớn nhất chia hết cả `a` lẫn `b`. **Euclid:** `gcd(a, b) = gcd(b, a % b)` và `gcd(a, 0) = a`. Lý do: mọi ước chung của `a` và `b` cũng chia hết `a - k*b`, tức `a % b`, và ngược lại. Python có sẵn `math.gcd(a, b)`.

````python
def my_gcd(a, b):
    while b:
        a, b = b, a % b
    return a

print(my_gcd(16, 20), math.gcd(16, 20), my_gcd(25, 30), my_gcd(7, 0))
````

### 3.4. Hai tính chất của GCD dùng trong bài

1. **Đưa thừa số chung ra ngoài:** `gcd(k*a, k*b) = k * gcd(a, b)`.
2. **Hai số liên tiếp nguyên tố cùng nhau:** `gcd(n, n + 1) = 1`. Chứng minh: nếu `d` chia hết cả `n` và `n + 1` thì `d` chia hết hiệu `(n + 1) - n = 1`, nên `d = 1`.

````python
print(math.gcd(6 * 4, 6 * 10), 6 * math.gcd(4, 10))
print(all(math.gcd(k, k + 1) == 1 for k in range(1, 1001)))
````

### 3.5. Tổng cấp số cộng và mẹo ghép cặp của Gauss

Với dãy cách đều `a1, a2, ..., ak`, tổng là `k * (a1 + ak) / 2`: ghép đầu với cuối, mỗi cặp cho cùng một tổng `a1 + ak`, có `k/2` cặp. Áp dụng: số lẻ `1..(2n-1)` có `n` số, tổng `n * (1 + 2n - 1) / 2 = n * 2n / 2 = n²`. Số chẵn `2..2n` có `n` số, tổng `n * (2 + 2n) / 2 = n(n+1)`. Hình học: xếp `1, 3, 5, ...` ô thành các **chữ L** lồng nhau sẽ lấp đầy hình vuông `n × n`.

````python
def arith_sum(first, last, count):
    return count * (first + last) // 2

n = 6
print(arith_sum(1, 2 * n - 1, n), n * n)            # tổng lẻ
print(arith_sum(2, 2 * n, n), n * (n + 1))          # tổng chẵn
# hình vuông n x n được lấp bởi các chữ L có 1, 3, 5, ... ô
for k in range(1, 5):
    print(f"chữ L thứ {k} có {2 * k - 1} ô, tổng tích lũy = {sum(2 * j - 1 for j in range(1, k + 1))} = {k}^2")
````

## 4. Ý tưởng nền tảng / cách nghĩ trực giác

Tính tay vài giá trị nhỏ:

```
n=1: lẻ 1             -> 1 ; chẵn 2           -> 2  ; gcd(1, 2)   = 1
n=2: lẻ 1+3 = 4       ; chẵn 2+4 = 6          ; gcd(4, 6)   = 2
n=3: lẻ 1+3+5 = 9     ; chẵn 2+4+6 = 12       ; gcd(9, 12)  = 3
n=4: lẻ 16            ; chẵn 20               ; gcd(16, 20) = 4
n=5: lẻ 25            ; chẵn 30               ; gcd(25, 30) = 5
```

Quan sát: tổng lẻ luôn là `n²` (1, 4, 9, 16, 25) và tổng chẵn là `n(n+1)` (2, 6, 12, 20, 30); GCD luôn **bằng đúng `n`**. Giả thuyết: đáp án là `n`. Phần dưới sẽ **chứng minh** giả thuyết này thay vì tin vào quy luật.

### Bảng quy luật cho `n = 1..10`

In bảng `n`, `sumOdd`, `sumEven`, `gcd` để **thấy** quy luật trước khi chứng minh. Cột `sumOdd / n` và `sumEven / n` lần lượt là `n` và `n + 1` — hai số liên tiếp, đây chính là chìa khóa của lời giải.

````python
print(f"{'n':>3} {'sumOdd':>7} {'sumEven':>8} {'gcd':>4} {'sumOdd/n':>9} {'sumEven/n':>10}")
for n in range(1, 11):
    so = sum(range(1, 2 * n, 2))
    se = sum(range(2, 2 * n + 1, 2))
    g = math.gcd(so, se)
    print(f"{n:>3} {so:>7} {se:>8} {g:>4} {so // n:>9} {se // n:>10}")
    assert g == n and so // n == n and se // n == n + 1
````

**Một cách chứng minh khác cho `gcd(n, n+1) = 1`:** giả sử số nguyên tố `q` chia hết cả `n` và `n+1`. Khi đó `n = q*a`, `n + 1 = q*b` nên `1 = q*(b - a)`, vô lý vì `q >= 2`. Vậy không có số nguyên tố nào chung, tức `gcd = 1`. Chứng minh này không dùng thuật toán Euclid mà dùng **tính chia hết** thuần túy.

## 5. Cách 1 — Brute Force / baseline

Làm đúng như đề: tính từng tổng bằng vòng lặp, rồi gọi `math.gcd`.

Nút thắt: tốn `O(n)` để cộng, trong khi hai tổng có công thức đóng. Với `n <= 1000` vẫn tức thì, nhưng nếu `n` lên `10^12` thì vòng lặp không còn khả thi.

````python
import math

class SolutionBrute:
    def gcdOfOddEvenSums(self, n):
        sum_odd = 0
        sum_even = 0
        for i in range(n):
            sum_odd += 2 * i + 1
            sum_even += 2 * i + 2
        return math.gcd(sum_odd, sum_even)

print(SolutionBrute().gcdOfOddEvenSums(4), SolutionBrute().gcdOfOddEvenSums(5))
````

**Độ phức tạp:** Time `O(n)`, Space `O(1)`.

## 6. Ý tưởng thuật toán / bước đột phá

**Bước đột phá:** tìm công thức đóng cho hai tổng.

- **Tổng `n` số lẻ đầu:** ghép cặp `(1 + (2n-1)) = 2n`, `(3 + (2n-3)) = 2n`, ... có `n/2` cặp nên tổng `= n/2 * 2n = n²`. Hoặc hình học: hình vuông `n × n` gồm các "chữ L" có `1, 3, 5, ..., 2n-1` ô.
- **Tổng `n` số chẵn đầu:** `2 + 4 + ... + 2n = 2 * (1 + 2 + ... + n) = 2 * n(n+1)/2 = n(n+1)`.

Từ đó: `gcd(n², n(n+1)) = gcd(n * n, n * (n+1)) = n * gcd(n, n+1) = n * 1 = n`.

**Quy tắc "nếu… thì…":**

- Muốn tính GCD hai tổng -> rút thừa số chung `n` trước.
- Phần còn lại `gcd(n, n+1)` luôn là `1` (hai số liên tiếp).
- Do đó đáp án **không cần** tính các tổng, chỉ cần `return n`.

````python
# Kiểm chứng công thức đóng với mọi n trong giới hạn của đề
for n in range(1, 1001):
    assert sum(range(1, 2 * n, 2)) == n * n
    assert sum(range(2, 2 * n + 1, 2)) == n * (n + 1)
    assert math.gcd(n * n, n * (n + 1)) == n
print("công thức n^2, n(n+1) và gcd = n đúng với n = 1..1000")
````

## 7. Các cách tiếp theo

### Cách 2 — Công thức đóng rồi `math.gcd`

Tính `sum_odd = n * n`, `sum_even = n * (n + 1)` trong `O(1)` rồi gọi `math.gcd`. Cách này **vẫn tính GCD thật sự** nên là bước đệm an toàn nếu bạn chưa tin kết luận `= n`.

````python
class SolutionFormula:
    def gcdOfOddEvenSums(self, n):
        sum_odd = n * n
        sum_even = n * (n + 1)
        return math.gcd(sum_odd, sum_even)

print(SolutionFormula().gcdOfOddEvenSums(4), SolutionFormula().gcdOfOddEvenSums(5))
````

**Độ phức tạp:** Time `O(log n)` cho `gcd` (hằng số nhỏ), Space `O(1)`.

### Cách 3 — Tự cài thuật toán Euclid trên hai tổng

Giống cách 2 nhưng tự viết `my_gcd(a, b)` bằng vòng `while b: a, b = b, a % b`. Phù hợp khi đề **cấm** `math.gcd` hoặc khi bạn muốn nắm cơ chế. Ta còn thấy ngay Euclid kết thúc sau **một bước**: `n(n+1) % n² = n` rồi `n² % n = 0`.

````python
class SolutionEuclid:
    def gcdOfOddEvenSums(self, n):
        a, b = n * n, n * (n + 1)
        while b:
            a, b = b, a % b
        return a

print(SolutionEuclid().gcdOfOddEvenSums(4), SolutionEuclid().gcdOfOddEvenSums(5))
````

**Độ phức tạp:** Time `O(log n)` (số bước Euclid theo định lý Lamé là `O(log min(a, b))`), Space `O(1)`.

### Cách 4 — Trả thẳng `n` (đáp án `O(1)`)

Sau khi chứng minh `gcd(n², n(n+1)) = n`, bài chỉ còn là `return n`. Đây là kiểu bài "tìm quy luật rồi chứng minh".

````python
class SolutionDirect:
    def gcdOfOddEvenSums(self, n):
        return n

print(SolutionDirect().gcdOfOddEvenSums(4), SolutionDirect().gcdOfOddEvenSums(5))
````

**Độ phức tạp:** Time `O(1)`, Space `O(1)`.

## 8. TRACE từng bước

Trace `n = 4` theo hai kiểu: (a) cộng dồn hai tổng, (b) thuật toán Euclid trên `(16, 20)`.

| i | số lẻ `2i+1` | sumOdd | số chẵn `2i+2` | sumEven |
|---|---|---|---|---|
| 0 | 1 | 1 | 2 | 2 |
| 1 | 3 | 4 | 4 | 6 |
| 2 | 5 | 9 | 6 | 12 |
| 3 | 7 | 16 | 8 | 20 |

Euclid: `(16, 20) -> (20, 16) -> (16, 4) -> (4, 0)` => `gcd = 4`.

````python
def trace_sums(n):
    s_odd = s_even = 0
    for i in range(n):
        o, e = 2 * i + 1, 2 * i + 2
        s_odd += o
        s_even += e
        print(f"i={i}: lẻ={o:>3} sumOdd={s_odd:>4} | chẵn={e:>3} sumEven={s_even:>4}")
    print(f"sumOdd={s_odd}, sumEven={s_even}; n^2={n * n}, n(n+1)={n * (n + 1)}")
    a, b = s_odd, s_even
    step = 0
    while b:
        step += 1
        print(f"Euclid bước {step}: gcd({a}, {b}) -> gcd({b}, {a % b})")
        a, b = b, a % b
    print("gcd =", a)
    return a

trace_sums(4)
trace_sums(5)
````

````python
from typing import List, Optional

class Solution:
    def gcdOfOddEvenSums(self, n: int) -> int:
        # sumOdd = n^2, sumEven = n(n+1)  =>  gcd = n * gcd(n, n+1) = n
        return n
````

## 10. Vì sao thuật toán đúng

- **Tổng lẻ:** quy nạp: `n = 1` -> `1 = 1²`. Giả sử tổng `n` số lẻ đầu là `n²`; số lẻ kế tiếp là `2n + 1`, nên tổng `n+1` số là `n² + 2n + 1 = (n+1)²`.
- **Tổng chẵn:** mỗi số chẵn thứ `i` bằng `2i`, nên tổng `= 2 * (1 + ... + n) = 2 * n(n+1)/2 = n(n+1)`.
- **GCD:** `gcd(n², n(n+1)) = gcd(n*n, n*(n+1)) = n * gcd(n, n+1)`. Nếu `d | n` và `d | (n+1)` thì `d | 1`, nên `gcd(n, n+1) = 1`. Suy ra GCD `= n`.
- **Kiểm chứng:** cell kiểm tra ở mục 6 và bộ test ở mục 12 thử **mọi** `n` trong `1..1000` và so với bản mô phỏng.

## 13. Các lỗi người mới hay gặp

### Lỗi 1: lệch một khi dùng `range`

`range(1, n)` chỉ có `n - 1` số, thiếu số lẻ thứ `n`.

````python
def wrong_off_by_one(n):
    sum_odd = sum(2 * i + 1 for i in range(1, n))        # SAI: i chạy từ 1 -> mất số lẻ đầu tiên
    sum_even = sum(2 * i + 2 for i in range(n))
    return math.gcd(sum_odd, sum_even)

print(wrong_off_by_one(4), "!=", Solution().gcdOfOddEvenSums(4))
assert wrong_off_by_one(4) != Solution().gcdOfOddEvenSums(4)
````

### Lỗi 2: nhầm "`n` số đầu tiên" với "các số không vượt quá `n`"

Đề cần `n` **phần tử** chứ không phải các số lẻ/chẵn `<= n`.

````python
def wrong_upto_n(n):
    sum_odd = sum(x for x in range(1, n + 1) if x % 2 == 1)
    sum_even = sum(x for x in range(1, n + 1) if x % 2 == 0)
    return math.gcd(sum_odd, sum_even)

print(wrong_upto_n(4), "!=", Solution().gcdOfOddEvenSums(4))       # 1+3=4 ; 2+4=6 -> 2
assert wrong_upto_n(4) != Solution().gcdOfOddEvenSums(4)
````

### Lỗi 3: nhầm công thức tổng chẵn

`n(n+1)/2` là tổng `1 + 2 + ... + n` (chưa nhân đôi). Tổng `n` số chẵn đầu là `n(n+1)`.

````python
def wrong_formula(n):
    sum_odd = n * n
    sum_even = n * (n + 1) // 2                           # SAI: thiếu nhân 2
    return math.gcd(sum_odd, sum_even)

print(wrong_formula(4), "!=", Solution().gcdOfOddEvenSums(4))       # gcd(16, 10) = 2
assert wrong_formula(4) != Solution().gcdOfOddEvenSums(4)
````

### Lỗi 4: trả nhầm một trong hai tổng hoặc dùng bội chung

Đề hỏi **ƯCLN** chứ không phải tổng lẻ `n²` hay bội chung nhỏ nhất.

````python
def wrong_return_sum(n):
    return n * n

def wrong_lcm(n):
    a, b = n * n, n * (n + 1)
    return a * b // math.gcd(a, b)

print(wrong_return_sum(4), wrong_lcm(4), "!=", Solution().gcdOfOddEvenSums(4))
assert wrong_return_sum(4) != Solution().gcdOfOddEvenSums(4)
assert wrong_lcm(4) != Solution().gcdOfOddEvenSums(4)
````

## 14. Mẹo làm bài nhanh & nhận dạng pattern

- Đề có "GCD" của hai biểu thức theo `n` -> thử **rút thừa số chung** và tìm đẳng thức `gcd(k, k+1) = 1`.
- Hãy **tính tay `n = 1..5`** rồi tìm quy luật; sau đó **chứng minh** để chắc chắn.
- Thuộc hai công thức: `1 + 3 + ... + (2n-1) = n²` và `1 + 2 + ... + n = n(n+1)/2`.
- Nếu cần tự cài GCD: `while b: a, b = b, a % b`.
- Khi `n` nhỏ như ở đây, viết **bản mô phỏng** để so với công thức; đó là cách tự kiểm tra tốt nhất.
- Để ý "`n` số đầu tiên" khác "các số `<= n`".

## 15. Biến thể của bài

1. **GCD của hai tổng khác:** tổng `n` số lẻ đầu và tổng `m` số chẵn đầu: `gcd(n², m(m+1))` (không còn đẹp).
2. **LCM của hai tổng** (cũng dùng `gcd`): `lcm = a * b // gcd(a, b)` = `n²(n+1)`.
3. **Find Greatest Common Divisor of Array** (LeetCode 1979): GCD của số nhỏ nhất và lớn nhất.
4. **GCD of Strings** (1071): Euclid trên độ dài chuỗi.
5. Tổng `n` số lẻ/chẵn đầu tiên **modulo** một số lớn (khi `n` rất lớn): dùng công thức đóng và `%`.

## 16. So sánh các cách làm

| Cách | Ý tưởng | Time | Space | Khi nào dùng |
|---|---|---|---|---|
| Mô phỏng + `gcd` | cộng từng số rồi `gcd` | `O(n)` | `O(1)` | Viết test đối chiếu |
| Công thức + `math.gcd` | `n²`, `n(n+1)` rồi `gcd` | `O(log n)` | `O(1)` | Khi chưa chắc đáp án `= n` |
| Công thức + Euclid | tự cài `while b` | `O(log n)` | `O(1)` | Khi cấm `math.gcd` / ôn Euclid |
| Trả `n` | chứng minh `gcd = n` | `O(1)` | `O(1)` | Nộp bài, ngắn nhất |

Nộp `return n`, nhưng giữ bản mô phỏng trong notebook để **kiểm chứng**.

## 19. Đáp án tham khảo cho bài tập
