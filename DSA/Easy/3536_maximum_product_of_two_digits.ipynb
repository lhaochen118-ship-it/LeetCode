## 1. Mục tiêu học

Sau notebook này bạn sẽ:

1. Hiểu đề: cho số nguyên dương `n` (có ít nhất 2 chữ số); chọn **hai chữ số ở hai vị trí khác nhau** của `n` sao cho **tích** lớn nhất; trả về tích đó.
2. Tách chữ số của một số bằng `str(n)` hoặc bằng `% 10` / `// 10` (không dùng chuỗi).
3. Nhận ra bước đột phá: tích lớn nhất đến từ **hai chữ số lớn nhất** (các chữ số đều không âm), kể cả khi hai chữ số đó bằng nhau (`22 -> 4`).
4. Viết 4 cách: brute force mọi cặp, sắp xếp chữ số, một lượt theo dõi hai số lớn nhất, và bảng đếm `cnt[0..9]`; đối chiếu bằng test ngẫu nhiên có `random.seed`.
5. Tránh các lỗi: dùng cùng một chữ số hai lần, lấy hai chữ số đầu, nhân lớn nhất với nhỏ nhất, và quên đẩy số lớn nhất cũ xuống làm số lớn thứ hai.

## 2. Đề bài & ví dụ

Cho số nguyên dương `n`. Hãy trả về **tích lớn nhất** của **hai chữ số bất kỳ** trong `n`. "Hai chữ số" nghĩa là hai **vị trí khác nhau** trong biểu diễn thập phân của `n` (nếu `n = 22` thì hai chữ `2` ở hai vị trí khác nhau, tích `2 * 2 = 4`; nhưng với `n = 31` không thể lấy chữ `3` hai lần).

| Ví dụ | n | Output | Giải thích |
|---|---|---|---|
| 1 | `31` | 3 | chỉ có cặp `(3, 1)` -> `3 * 1 = 3` |
| 2 | `22` | 4 | hai chữ số `2` và `2` -> `2 * 2 = 4` |
| 3 | `124` | 8 | các cặp: `1*2=2`, `1*4=4`, `2*4=8` -> lớn nhất `8` |

Constraints: `10 <= n <= 10^9`.

````python
n = 124
digits = [int(ch) for ch in str(n)]
print("các chữ số:", digits)
best = 0
for i in range(len(digits)):
    for j in range(i + 1, len(digits)):
        print(f"{digits[i]} * {digits[j]} = {digits[i] * digits[j]}")
        best = max(best, digits[i] * digits[j])
print("tích lớn nhất =", best)
````

### 3.2. Tách chữ số bằng toán học: `% 10` và `// 10`

`n % 10` là **chữ số cuối**; `n // 10` bỏ chữ số cuối. Lặp `while n > 0` lấy hết các chữ số (từ phải sang trái), không cần chuỗi. Cẩn thận: cách này cho chữ số theo thứ tự **ngược**, nhưng với bài này thứ tự không quan trọng.

````python
n = 90315
digits = []
while n > 0:
    digits.append(n % 10)
    n //= 10
print(digits)
````

### 3.3. Hai vòng lặp chọn cặp `i < j`

Để xét **mọi cặp hai vị trí khác nhau** mỗi cặp đúng một lần, dùng `for i in range(m)` và `for j in range(i + 1, m)`. Có `m * (m - 1) / 2` cặp (`m` chữ số). Với `m <= 10` chỉ tối đa `45` cặp.

````python
m = 4
pairs = [(i, j) for i in range(m) for j in range(i + 1, m)]
print(pairs, len(pairs), m * (m - 1) // 2)
````

### 3.4. `sorted(..., reverse=True)` và cắt lát

`sorted(digits, reverse=True)` sắp xếp **giảm dần**; hai phần tử đầu `d[0]`, `d[1]` là hai chữ số lớn nhất (có thể bằng nhau nếu chữ số đó xuất hiện hai lần). Tích: `d[0] * d[1]`.

````python
d = sorted([int(ch) for ch in "124"], reverse=True)
print(d, d[0] * d[1])
d = sorted([int(ch) for ch in "22"], reverse=True)
print(d, d[0] * d[1])
````

````python
cnt = [0] * 10
for ch in "2252":
    cnt[int(ch)] += 1
print(cnt)
````

## 4. Ý tưởng nền tảng / cách nghĩ trực giác

Với `n = 124`, chữ số là `1, 2, 4`. Ba cặp:

```
(1, 2) -> 2
(1, 4) -> 4
(2, 4) -> 8   <- lớn nhất, dùng hai chữ số LỚN NHẤT
```

Vì mọi chữ số `>= 0`, tăng một thừa số không làm tích giảm; nên tích lớn nhất dùng **hai chữ số lớn nhất**. Nếu số lớn nhất xuất hiện **hai lần** (như `22`, `2252`) thì ta dùng chính nó hai lần (hai vị trí khác nhau). Nếu có chữ `0` thì tích có thể bằng `0` (`10 -> 1 * 0 = 0`).

## 5. Cách 1 — Brute Force / baseline

Tách chữ số; thử **mọi cặp** `i < j`, tính tích và lấy lớn nhất.

Nút thắt: `O(m^2)` cặp với `m` là số chữ số (`m <= 10` nên chỉ `45` cặp) — rất rẻ; nhưng ta có thể làm mà không cần xét mọi cặp.

````python
class SolutionBrute:
    def maxProduct(self, n):
        digits = [int(ch) for ch in str(n)]
        best = 0
        for i in range(len(digits)):
            for j in range(i + 1, len(digits)):
                best = max(best, digits[i] * digits[j])
        return best

print(SolutionBrute().maxProduct(31))
print(SolutionBrute().maxProduct(22))
print(SolutionBrute().maxProduct(124))
````

**Độ phức tạp:** Time `O(m^2)` với `m = số chữ số` (tối đa `10`), Space `O(m)`.

## 6. Ý tưởng thuật toán / bước đột phá

**Bước đột phá:** tích hai số không âm lớn nhất khi cả hai thừa số lớn nhất -> chỉ cần **hai chữ số lớn nhất** (tính theo vị trí, nên giá trị lớn nhất có thể được dùng hai lần nếu nó xuất hiện hai lần).

**Quy tắc "nếu… thì…" khi theo dõi `first >= second`:**

- Nếu `d > first` -> `second = first`, `first = d` (đẩy số lớn nhất cũ xuống làm số lớn thứ hai!).
- Ngược lại nếu `d > second` -> `second = d`. (Với `d == first` ta có `d > second` nếu `second < first`, nên `second = d = first`: đúng cho trường hợp lặp.)

Invariant: sau mỗi chữ số, `first` và `second` là hai giá trị lớn nhất (theo vị trí) trong các chữ số đã xét, `first >= second`.

````python
def top_two(n):
    first = second = 0
    while n > 0:
        d = n % 10
        n //= 10
        if d > first:
            second = first
            first = d
        elif d > second:
            second = d
    return first, second

print(top_two(124), top_two(22), top_two(31), top_two(2252))
````

## 7. Các cách tiếp theo

Ta có 4 cách: (1) brute mọi cặp; (2) sắp xếp chữ số giảm dần; (3) một lượt theo dõi hai số lớn nhất (không chuỗi); (4) bảng đếm `cnt`.

### Cách 2 — Sắp xếp chữ số giảm dần

`d = sorted(digits, reverse=True)`; trả về `d[0] * d[1]`. Vì `n >= 10`, luôn có ít nhất hai chữ số nên `d[1]` hợp lệ.

````python
class SolutionSort:
    def maxProduct(self, n):
        d = sorted((int(ch) for ch in str(n)), reverse=True)
        return d[0] * d[1]

print(SolutionSort().maxProduct(31))
print(SolutionSort().maxProduct(22))
print(SolutionSort().maxProduct(124))
````

Độ phức tạp Cách 2: Time `O(m log m)` (`m <= 10`), Space `O(m)`.

### Cách 3 — Một lượt theo dõi `first`, `second` (không chuỗi)

Dùng `n % 10` và `n // 10` để lấy từng chữ số, cập nhật `first`/`second` như quy tắc mục 6. Chỉ `O(m)` thời gian và `O(1)` bộ nhớ.

````python
class SolutionTopTwo:
    def maxProduct(self, n):
        first = second = 0
        while n > 0:
            d = n % 10
            n //= 10
            if d > first:
                second = first
                first = d
            elif d > second:
                second = d
        return first * second

print(SolutionTopTwo().maxProduct(31))
print(SolutionTopTwo().maxProduct(22))
print(SolutionTopTwo().maxProduct(124))
````

Độ phức tạp Cách 3: Time `O(m)`, Space `O(1)`.

### Cách 4 — Bảng đếm `cnt[0..9]`

Đếm số lần xuất hiện của mỗi chữ số. Rồi lấy "hai chữ số lớn nhất": với `d` từ `9` xuống `0`, lấy `min(cnt[d], còn_cần)` lần chữ số `d` vào danh sách `top` cho tới khi đủ `2`. Tích `top[0] * top[1]`.

````python
class SolutionCount:
    def maxProduct(self, n):
        cnt = [0] * 10
        for ch in str(n):
            cnt[int(ch)] += 1
        top = []
        for d in range(9, -1, -1):
            take = min(cnt[d], 2 - len(top))
            top.extend([d] * take)
            if len(top) == 2:
                break
        return top[0] * top[1]

print(SolutionCount().maxProduct(31))
print(SolutionCount().maxProduct(22))
print(SolutionCount().maxProduct(124))
````

Độ phức tạp Cách 4: Time `O(m + 10)`, Space `O(10) = O(1)`.

## 8. TRACE từng bước

Trace Cách 3 với `n = 124` (lấy chữ số từ phải sang trái: `4, 2, 1`):

| bước | `n` trước | `d = n % 10` | so sánh | `first` | `second` |
|---|---|---|---|---|---|
| bắt đầu | 124 | | | 0 | 0 |
| 1 | 124 | 4 | `4 > 0` -> `second = 0`, `first = 4` | 4 | 0 |
| 2 | 12 | 2 | `2 > 4`? không; `2 > 0` -> `second = 2` | 4 | 2 |
| 3 | 1 | 1 | `1 > 4`? không; `1 > 2`? không | 4 | 2 |

Kết thúc `n = 0`: `first * second = 4 * 2 = 8`.

````python
def trace_digits(n):
    first = second = 0
    step = 0
    print(f"bắt đầu: n={n} first={first} second={second}")
    while n > 0:
        step += 1
        d = n % 10
        n //= 10
        if d > first:
            second, first = first, d
            note = f"{d} > first -> đẩy first cũ xuống second"
        elif d > second:
            second = d
            note = f"{d} > second -> cập nhật second"
        else:
            note = f"{d} không lớn hơn second -> bỏ qua"
        print(f"bước {step}: d={d} | {note} | first={first} second={second}")
    print("tích =", first * second)
    return first * second

trace_digits(124)
trace_digits(22)
````

## 9. Code nộp LeetCode

````python
from typing import List

class Solution:
    def maxProduct(self, n: int) -> int:
        first = second = 0
        while n > 0:
            d = n % 10
            n //= 10
            if d > first:
                second = first
                first = d
            elif d > second:
                second = d
        return first * second

print(Solution().maxProduct(31), Solution().maxProduct(22), Solution().maxProduct(124))
````

## 10. Vì sao thuật toán đúng

- **Tích tăng theo từng thừa số (số không âm):** nếu `0 <= a <= a'` và `b >= 0` thì `a * b <= a' * b`. Do đó với cặp `(a, b)` bất kỳ, thay bằng hai chữ số lớn nhất không làm tích giảm: tích lớn nhất đạt khi dùng hai chữ số lớn nhất.
- **"Hai chữ số" theo vị trí:** nếu giá trị lớn nhất xuất hiện hai lần thì được dùng cả hai (hai vị trí khác nhau); nếu chỉ một lần thì chữ số thứ hai là giá trị lớn nhì (có thể nhỏ hơn). Sắp xếp giảm dần theo từng chữ số (kể cả lặp) cho đúng hai chữ số này.
- **Cách 3:** quy tắc `if d > first: second = first; first = d; elif d > second: second = d` giữ invariant `first >= second` là hai giá trị lớn nhất đã thấy. Khi `d == first`, `d > first` sai nhưng `d > second` đúng (nếu `second < first`) nên `second = first` — đúng với chữ số lặp. Khi `second == first` đã bằng nhau thì `d` không thể lớn hơn.
- **Cách 4:** lấy lần lượt từ `9` xuống `0`, đủ `2` chữ số theo số lần xuất hiện -> chính là hai chữ số lớn nhất theo vị trí.
- **Số lượng chữ số `>= 2`** vì `n >= 10`, nên luôn có cặp hợp lệ; `second` luôn được gán giá trị thật (có thể `0` nếu chữ số khác là `0`, ví dụ `10 -> 1 * 0 = 0`).

## 11. Phân tích Time / Space Complexity

Gọi `m = số chữ số của n` (`m <= 10`), tức `m = O(log n)`.

| Cách | Time | Space |
|---|---|---|
| 1. Brute mọi cặp | `O(m^2)` | `O(m)` |
| 2. Sắp xếp chữ số | `O(m log m)` | `O(m)` |
| 3. Một lượt `first/second` | `O(m)` | `O(1)` |
| 4. Bảng đếm `cnt[10]` | `O(m)` | `O(1)` (10 ô) |

Cách 3 và 4 là tối ưu. Với `m <= 10`, mọi cách đều chạy tức thì.

## 12. Bộ test

Test 1: ví dụ đề và edge case (hai chữ số, chữ số lặp, có `0`, toàn `9`, `n = 10^9`, chữ số lớn nhất lặp ba lần).

````python
impls = [SolutionBrute(), SolutionSort(), SolutionTopTwo(), SolutionCount(), Solution()]
tests = [
    (31, 3),
    (22, 4),
    (124, 8),
    (10, 0),                       # biên dưới, có chữ số 0
    (99, 81),                      # hai chữ số 9
    (90, 0),                       # 9 * 0
    (100, 0),                      # 1*0, 1*0, 0*0 -> 0
    (1000000000, 0),               # biên trên 10^9 (1 và chín số 0)
    (999999999, 81),               # toàn 9
    (123456789, 72),               # 9 * 8
    (918273645, 72),               # đảo thứ tự, vẫn 9 * 8
    (2252, 10),                    # 5 * 2 (chữ số lớn nhất xuất hiện một lần)
    (5500, 25),                    # 5 * 5
    (1011, 1),                     # 1 * 1 (hai chữ số 1)
]
for n, exp in tests:
    assert 10 <= n <= 10**9
    for im in impls:
        got = im.maxProduct(n)
        assert got == exp, (type(im).__name__, n, got, exp)
print("tất cả test cố định OK:", len(tests))
````

Test 2: ngẫu nhiên có seed, đúng constraints (`10 <= n <= 10^9`). Oracle độc lập dùng `itertools.combinations` trên danh sách chữ số (chọn 2 vị trí khác nhau).

````python
import random
from itertools import combinations
random.seed(3536)

def oracle(n):
    digs = [int(ch) for ch in str(n)]
    return max(a * b for a, b in combinations(digs, 2))

for t in range(3000):
    kind = random.choice(["small", "mid", "big", "lowdigits"])
    if kind == "small":
        n = random.randint(10, 99)
    elif kind == "mid":
        n = random.randint(10, 10**5)
    elif kind == "big":
        n = random.randint(10, 10**9)
    else:
        m = random.randint(2, 9)
        n = int("".join(random.choice("0012") for _ in range(m)).lstrip("0") or "10")
        n = max(n, 10)
    assert 10 <= n <= 10**9
    exp = oracle(n)
    for im in impls:
        assert im.maxProduct(n) == exp, (type(im).__name__, n)
print("3000 test ngẫu nhiên OK")
````

Test 3: tính chất — kết quả `<= 81`; không phụ thuộc thứ tự chữ số (hoán vị chữ số, bỏ qua số có chữ số đầu `0`); kết quả `0` khi và chỉ khi có `<= 1` chữ số khác `0`; kết quả `d1 * d2` với `d1 >= d2` là hai chữ số lớn nhất.

````python
random.seed(536)
for _ in range(1000):
    n = random.randint(10, 10**9)
    digs = [int(ch) for ch in str(n)]
    ans = Solution().maxProduct(n)
    assert 0 <= ans <= 81
    perm = digs[:]
    random.shuffle(perm)
    if perm[0] != 0:
        n2 = int("".join(map(str, perm)))
        assert Solution().maxProduct(n2) == ans
    nonzero = sum(1 for d in digs if d != 0)
    assert (ans == 0) == (nonzero <= 1)
    top = sorted(digs, reverse=True)
    assert ans == top[0] * top[1]
print("tính chất (<= 81, hoán vị, ans = 0 khi <= 1 chữ số khác 0, top 2): OK")
````

## 13. Các lỗi người mới hay gặp

### Lỗi 1: dùng **cùng một** chữ số hai lần (bình phương chữ số lớn nhất)

````python
def wrong_square_max(n):
    m = max(int(ch) for ch in str(n))
    return m * m                                # SAI: chỉ được dùng chữ số lặp nếu nó thật sự xuất hiện hai lần

print("sai :", wrong_square_max(31))
print("đúng:", Solution().maxProduct(31))
assert wrong_square_max(31) != Solution().maxProduct(31)
````

Vì sao sai: `31` chỉ có một chữ `3`; không thể nhân `3 * 3`. Hai chữ số phải ở **hai vị trí khác nhau**. **Sửa:** lấy hai chữ số lớn nhất **theo vị trí** (sắp xếp cả lặp hoặc theo dõi `first/second`).

### Lỗi 2: lấy hai chữ số đầu tiên

````python
def wrong_first_two(n):
    s = str(n)
    return int(s[0]) * int(s[1])                # SAI: không phải hai chữ số lớn nhất

print("sai :", wrong_first_two(124))
print("đúng:", Solution().maxProduct(124))
assert wrong_first_two(124) != Solution().maxProduct(124)
````

Vì sao sai: `1 * 2 = 2`, nhưng `2 * 4 = 8` lớn hơn. Vị trí của chữ số **không** liên quan; chỉ giá trị mới quan trọng. **Sửa:** chọn theo giá trị lớn nhất.

### Lỗi 3: nhân chữ số lớn nhất với chữ số **nhỏ nhất**

````python
def wrong_max_times_min(n):
    digs = [int(ch) for ch in str(n)]
    return max(digs) * min(digs)                # SAI: tích lớn nhất cần HAI chữ số lớn nhất

print("sai :", wrong_max_times_min(124))
print("đúng:", Solution().maxProduct(124))
assert wrong_max_times_min(124) != Solution().maxProduct(124)
````

Vì sao sai: `4 * 1 = 4` trong khi `4 * 2 = 8`. Chữ số nhỏ nhất kéo tích xuống. **Sửa:** dùng hai chữ số lớn nhất.

### Lỗi 4: quên đẩy chữ số lớn nhất cũ xuống làm lớn thứ hai

````python
def wrong_no_shift(n):
    first = second = 0
    digs = [int(ch) for ch in str(n)]           # duyệt từ chữ số đầu tới cuối
    for d in digs:
        if d > first:
            first = d                           # SAI: bỏ mất first cũ, không gán second = first
        elif d > second:
            second = d
    return first * second

print("sai :", wrong_no_shift(124))
print("đúng:", Solution().maxProduct(124))
assert wrong_no_shift(124) != Solution().maxProduct(124)
````

Vì sao sai: với `1, 2, 4` (tăng dần): `first` thành `1`, rồi `2`, rồi `4` mà `second` luôn `0` nên tích `0`. Khi gặp số lớn hơn `first` phải gán `second = first` **trước** khi cập nhật `first`. **Sửa:** `second, first = first, d`.

## 14. Mẹo làm bài nhanh & nhận dạng pattern

- "Tích/tổng lớn nhất của hai phần tử" với các phần tử **không âm** -> lấy **hai phần tử lớn nhất**. (Nếu có số âm thì phải xét thêm hai số nhỏ nhất.)
- Đề nói "hai chữ số" -> hai **vị trí** khác nhau, nên chữ số lặp được dùng lặp.
- Tách chữ số: `str(n)` (dễ đọc) hoặc `% 10`, `// 10` (không chuỗi). `n` tới `10^9` chỉ 10 chữ số.
- Theo dõi hai số lớn nhất một lượt: nhớ **đẩy `first` xuống `second`** khi tìm được số lớn hơn.
- Bảng đếm 10 ô là cách rất sạch cho bài về chữ số.
- Kiểm tra: `n = 10` (có `0`), `22` (lặp), `31` (không được bình phương).

## 15. Biến thể của bài

1. **Tích của ba chữ số** lớn nhất (lấy ba chữ số lớn nhất theo vị trí).
2. **Tích nhỏ nhất** của hai chữ số (hai chữ số nhỏ nhất).
3. **Mảng có số âm** (`628. Maximum Product of Three Numbers`, `1464. Maximum Product of Two Elements in an Array`): xét hai số lớn nhất và hai số nhỏ nhất.
4. **Tổng lớn nhất** của hai chữ số hoặc **chênh lệch lớn nhất** (`max digit - min digit`).
5. Chọn hai chữ số ở **hai vị trí kề nhau** (cửa sổ độ dài 2).

## 16. So sánh các cách làm

| Cách | Ý tưởng | Time | Space | Khi nào dùng |
|---|---|---|---|---|
| 1. Brute mọi cặp | thử mọi `(i, j)` | `O(m^2)` | `O(m)` | hiểu đề, oracle |
| 2. Sắp xếp giảm dần | `d[0] * d[1]` | `O(m log m)` | `O(m)` | ngắn gọn, dễ nhớ |
| 3. Một lượt `first/second` | theo dõi hai số lớn nhất | `O(m)` | `O(1)` | **chuẩn, nên nộp** |
| 4. Bảng đếm `cnt[10]` | counting sort 10 ô | `O(m)` | `O(1)` | bài về chữ số |

## 18. Gợi ý 3 mức cho bài tập

**Mức 1 (nhẹ):** liệt kê các chữ số, chọn hai chữ số lớn nhất theo vị trí rồi nhân.

**Mức 2 (vừa):** Bài 3: `while n > 0: out.append(n % 10); n //= 10`. Bài 4: sắp xếp giảm dần rồi nhân `d[0] * d[1] * d[2]`. Bài 5: sắp xếp tăng dần rồi nhân hai chữ số đầu `d[0] * d[1]`.

## 19. Đáp án tham khảo cho bài tập

````python
from itertools import combinations

# Bài 1: 8 * 7
assert Solution().maxProduct(5873) == 56

# Bài 2: 7, 7 -> 49
assert Solution().maxProduct(7007) == 49

# Bài 3
def digits_of(n):
    out = []
    while n > 0:
        out.append(n % 10)
        n //= 10
    return out
assert sorted(digits_of(124)) == [1, 2, 4]
assert sorted(digits_of(7007)) == [0, 0, 7, 7]
assert digits_of(5) == [5]

# Bài 4
def max_product_three(n):
    d = sorted(digits_of(n), reverse=True)
    return d[0] * d[1] * d[2]
assert max_product_three(124) == 8
assert max_product_three(9511) == 45
assert max_product_three(999) == 729

# Bài 5
def min_product_two(n):
    d = sorted(digits_of(n))
    return d[0] * d[1]
for n in (31, 22, 124, 7007, 5873, 918273645):
    digs = digits_of(n)
    assert min_product_two(n) == min(a * b for a, b in combinations(digs, 2))
assert min_product_two(124) == 2
assert min_product_two(7007) == 0
print("đáp án bài tập OK")
````
