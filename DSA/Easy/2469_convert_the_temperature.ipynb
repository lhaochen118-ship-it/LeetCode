## 2. Đề bài & ví dụ

Cho số thực **không âm** `celsius`, là nhiệt độ theo độ C, làm tròn đến hai chữ số thập phân. Bạn cần đổi sang **Kelvin** và **Fahrenheit** rồi trả về mảng `ans = [kelvin, fahrenheit]`.

Công thức:

- `Kelvin = Celsius + 273.15`
- `Fahrenheit = Celsius * 1.80 + 32.00`

Câu trả lời được chấp nhận nếu sai số **không quá `10^-5`** so với đáp án đúng.

| Ví dụ | celsius | Output | Giải thích |
|---|---|---|---|
| 1 | `36.50` | `[309.65000, 97.70000]` | `36.5 + 273.15 = 309.65`; `36.5 * 1.8 + 32 = 97.7` |
| 2 | `122.11` | `[395.26000, 251.79800]` | `122.11 + 273.15 = 395.26`; `122.11 * 1.8 + 32 = 251.798` |

Constraints: `0 <= celsius <= 1000`.

````python
celsius = 36.50
kelvin = celsius + 273.15
fahrenheit = celsius * 1.80 + 32.00
print("Celsius   :", celsius)
print("Kelvin    :", kelvin)
print("Fahrenheit:", fahrenheit)
print([kelvin, fahrenheit])
````

## 3. Kiến thức nền

### 3.1. Phép tính số thực và công thức bậc nhất

`y = a * x + b` là hàm bậc nhất: nhân với hệ số `a` rồi cộng hằng số `b`. Với Kelvin, `a = 1`, `b = 273.15`. Với Fahrenheit, `a = 1.8`, `b = 32`. Python tính `*` trước `+` (thứ tự ưu tiên phép toán).

````python
x = 100
print(1 * x + 273.15)       # Kelvin
print(1.8 * x + 32)         # Fahrenheit
print(2 + 3 * 4, (2 + 3) * 4)   # ưu tiên phép nhân
````

### 3.2. `float` không biểu diễn chính xác mọi số thập phân

Số `float` lưu dưới dạng nhị phân nên nhiều số thập phân (như `0.1`) **không** biểu diễn chính xác. Hệ quả: `0.1 + 0.2` ra `0.30000000000000004`. Vì vậy đề cho phép sai số `10^-5`, và khi viết test ta **không** dùng `==` giữa hai số thực.

````python
print(0.1 + 0.2)
print(0.1 + 0.2 == 0.3)
print(abs((0.1 + 0.2) - 0.3) < 1e-9)
````

### 3.3. `math.isclose` để so sánh số thực

`math.isclose(a, b, abs_tol=1e-5)` trả `True` nếu hai số cách nhau không quá `1e-5`. Tham số `rel_tol` mặc định là `1e-9`; khi muốn chỉ dùng dung sai tuyệt đối như đề, ta đặt `rel_tol=0`.

````python
import math
print(math.isclose(0.1 + 0.2, 0.3))
print(math.isclose(309.65, 309.650004, rel_tol=0, abs_tol=1e-5))
print(math.isclose(309.65, 309.66, rel_tol=0, abs_tol=1e-5))
````

### 3.4. Trả về `list` gồm hai số và thứ tự quan trọng

Hàm trả `[kelvin, fahrenheit]`: **phần tử đầu là Kelvin**, phần tử sau là Fahrenheit. Python cho phép gán bộ `k, f = result` để tách. Đảo thứ tự là lỗi rất dễ mắc.

````python
result = [309.65, 97.7]
k, f = result
print("Kelvin =", k, "| Fahrenheit =", f)
````

### 3.5. `decimal.Decimal` và `fractions.Fraction`

- `Decimal("36.50")` là số thập phân **chính xác** (tính theo cơ số 10). Tạo từ **chuỗi** để tránh sai số của `float`.
- `Fraction("36.50")` là số hữu tỉ chính xác (`73/2`). Phép `+ - * /` cho kết quả chính xác, rồi `float(...)` mới đổi sang số thực.

````python
from decimal import Decimal
from fractions import Fraction

print(Decimal("0.1") + Decimal("0.2"))
print(Fraction("0.1") + Fraction("0.2"))
print(float(Fraction("36.50") * Fraction("1.8") + 32))
````

## 4. Ý tưởng nền tảng / cách nghĩ trực giác

Bài này **không có thuật toán phức tạp**: chỉ cần áp dụng hai công thức đúng và trả về theo đúng thứ tự. Điểm cần chú ý nằm ở các chi tiết:

1. **Nhớ hệ số:** `K = C + 273.15`; `F = C * 1.8 + 32`. Có thể kiểm tra nhanh: nước đóng băng ở `0 °C = 273.15 K = 32 °F`; nước sôi ở `100 °C = 373.15 K = 212 °F`.
2. **Thứ tự kết quả:** `[Kelvin, Fahrenheit]`.
3. **Kiểu dữ liệu:** `celsius` là số thực; kết quả là `list` các số thực. Không làm tròn thành số nguyên.

Hãy tự kiểm tra hai điểm "mốc" để chắc công thức đúng.

## 5. Cách 1 — Brute Force / baseline: viết thẳng công thức

**Ý tưởng:** không có cách "vét cạn" thực sự cho bài này; baseline đơn giản nhất là áp dụng công thức. Ta viết các bước thật rõ ràng thành hai biến trung gian.

````python
from typing import List

class SolutionBrute:
    def convertTemperature(self, celsius: float) -> List[float]:
        kelvin = celsius + 273.15
        fahrenheit = celsius * 1.80 + 32.00
        return [kelvin, fahrenheit]
````

````python
sol = SolutionBrute()
print(sol.convertTemperature(36.50))     # [309.65, 97.7]
print(sol.convertTemperature(122.11))    # [395.26, 251.798]
````

**Độ phức tạp:** Time `O(1)` (hai phép tính), Space `O(1)` (một danh sách hai phần tử). **Nút thắt:** không có; nhưng kết quả `float` có thể sai lệch rất nhỏ (cỡ `10^-13`) so với số thập phân chính xác, hoàn toàn nằm trong dung sai `10^-5` của đề.

## 6. Ý tưởng thuật toán / bước đột phá

Không có "bước đột phá" thuật toán, nhưng có hai cách nhìn giúp code gọn và dễ mở rộng:

- **Hàm bậc nhất tổng quát:** mọi phép đổi đơn vị tuyến tính đều có dạng `y = a·x + b`. Ta có thể lưu danh sách hệ số `[(1, 273.15), (1.8, 32)]` rồi áp dụng chung một công thức (dễ thêm Rankine, Réaumur, ...).
- **Tính chính xác khi cần:** nếu muốn kết quả chính xác từng chữ số (ví dụ để hiển thị), dùng số thập phân/hữu tỉ rồi mới đổi sang `float`.
- **Quy tắc:** *`celsius` luôn đi qua cùng hai phép biến đổi và kết quả giữ nguyên thứ tự `[K, F]`.*
- **Invariant kiểm tra:** biến đổi ngược phải trả lại `celsius`: `celsius == K - 273.15 == (F - 32) / 1.8` (trong dung sai).

## 7. Các cách tiếp theo

### Cách 2 — Bảng hệ số `(a, b)`

Mỗi đơn vị là một cặp `(a, b)`; dùng list comprehension áp dụng `a * celsius + b`.

````python
from typing import List

class SolutionTable:
    COEFFS = [(1.0, 273.15), (1.8, 32.0)]       # (Kelvin, Fahrenheit)

    def convertTemperature(self, celsius: float) -> List[float]:
        return [a * celsius + b for a, b in self.COEFFS]
````

````python
print(SolutionTable().convertTemperature(36.50))     # [309.65, 97.7]
print(SolutionTable().convertTemperature(122.11))    # [395.26, 251.798]
````

**Độ phức tạp:** Time `O(1)` (số đơn vị cố định), Space `O(1)`. Tiện khi mở rộng nhiều đơn vị: Time `O(số đơn vị)`.

### Cách 3 — `Decimal` (số thập phân chính xác)

Chuyển `celsius` thành chuỗi rồi `Decimal` để các phép `+`, `*` chính xác theo thập phân; cuối cùng đổi sang `float`.

````python
from typing import List
from decimal import Decimal

class SolutionDecimal:
    def convertTemperature(self, celsius: float) -> List[float]:
        c = Decimal(str(celsius))
        kelvin = c + Decimal("273.15")
        fahrenheit = c * Decimal("1.80") + Decimal("32.00")
        return [float(kelvin), float(fahrenheit)]
````

````python
print(SolutionDecimal().convertTemperature(36.50))     # [309.65, 97.7]
print(SolutionDecimal().convertTemperature(122.11))    # [395.26, 251.798]
````

**Độ phức tạp:** Time `O(1)` (số chữ số cố định), Space `O(1)`. Chậm hơn `float` thuần nhưng chính xác. **Lưu ý:** phải dùng `Decimal(str(celsius))`; `Decimal(0.1)` sao chép sai số nhị phân của `float`.

### Cách 4 — `Fraction` (số hữu tỉ chính xác)

Với `Fraction`, hệ số `1.80` là `9/5` chính xác. Tạo `Fraction(str(celsius))` rồi tính.

````python
from typing import List
from fractions import Fraction

class SolutionFraction:
    def convertTemperature(self, celsius: float) -> List[float]:
        c = Fraction(str(celsius))
        kelvin = c + Fraction(27315, 100)
        fahrenheit = c * Fraction(9, 5) + 32
        return [float(kelvin), float(fahrenheit)]
````

````python
print(SolutionFraction().convertTemperature(36.50))     # [309.65, 97.7]
print(SolutionFraction().convertTemperature(122.11))    # [395.26, 251.798]
````

**Độ phức tạp:** Time `O(1)`, Space `O(1)`. Đây là "oracle" đáng tin cậy để đối chiếu các cách `float`, vì mọi phép toán trung gian đều chính xác.

## 8. TRACE từng bước

Trace tính `celsius = 122.11`:

| Bước | Biểu thức | Giá trị |
|---|---|---|
| 1 | `kelvin = 122.11 + 273.15` | `395.26` |
| 2 | `tmp = 122.11 * 1.80` | `219.798` |
| 3 | `fahrenheit = tmp + 32.00` | `251.798` |
| 4 | trả về `[kelvin, fahrenheit]` | `[395.26, 251.798]` |

````python
def trace_convert(celsius):
    print("celsius =", celsius)
    kelvin = celsius + 273.15
    print(f"bước 1: kelvin = {celsius} + 273.15 = {kelvin}")
    tmp = celsius * 1.80
    print(f"bước 2: tmp = {celsius} * 1.80 = {tmp}")
    fahrenheit = tmp + 32.00
    print(f"bước 3: fahrenheit = {tmp} + 32.00 = {fahrenheit}")
    result = [kelvin, fahrenheit]
    print("bước 4: trả về", result)
    return result

trace_convert(122.11)
trace_convert(36.50)
````

## 12. Bộ test

### 12.1. Ví dụ của đề và edge case

Dùng hàm `close(got, want)` so sánh hai danh sách với dung sai `1e-5` đúng như đề.

````python
import math

def close(got, want, tol=1e-5):
    return len(got) == len(want) and all(math.isclose(g, w, rel_tol=0, abs_tol=tol) for g, w in zip(got, want))

sols = [SolutionBrute(), SolutionTable(), SolutionDecimal(), SolutionFraction(), Solution()]
cases = [
    (36.50, [309.65, 97.70]),            # ví dụ 1
    (122.11, [395.26, 251.798]),         # ví dụ 2
    (0, [273.15, 32.0]),                 # biên nhỏ nhất, điểm đóng băng
    (0.0, [273.15, 32.0]),               # số thực 0
    (0.01, [273.16, 32.018]),            # hai chữ số thập phân nhỏ nhất
    (100, [373.15, 212.0]),              # nước sôi
    (37, [310.15, 98.6]),                # thân nhiệt
    (1000, [1273.15, 1832.0]),           # biên lớn nhất
    (999.99, [1273.14, 1831.982]),       # sát biên lớn nhất
    (25.5, [298.65, 77.9]),
]
for c, want in cases:
    for s in sols:
        got = s.convertTemperature(c)
        assert close(got, want), (type(s).__name__, c, got, want)
print("tất cả test cố định đều đúng với cả 5 cách:", len(cases))
````

Kiểm tra thủ công vài số: `0.01 * 1.8 = 0.018`, cộng `32` được `32.018`; `999.99 * 1.8 = 1799.982`, cộng `32` được `1831.982`; `25.5 * 1.8 = 45.9`, cộng `32` được `77.9`. Các số trong bảng khớp.

### 12.2. Đối chiếu ngẫu nhiên (có seed)

Sinh `celsius` đúng constraints: số thực không âm tối đa `1000` với **hai chữ số thập phân**, tức `random.randint(0, 100000) / 100`. Dùng `Fraction` làm chuẩn chính xác.

````python
import random
from fractions import Fraction
random.seed(2469)

def exact(c_text):
    c = Fraction(c_text)
    return [float(c + Fraction(27315, 100)), float(c * Fraction(9, 5) + 32)]

for _ in range(3000):
    cents = random.randint(0, 100000)
    celsius = cents / 100                              # 0.00 .. 1000.00, hai chữ số thập phân
    assert 0 <= celsius <= 1000
    want = exact(str(celsius))
    for s in sols:
        got = s.convertTemperature(celsius)
        assert close(got, want), (type(s).__name__, celsius, got, want)
print("3000 test ngẫu nhiên: 5 cách khớp giá trị chính xác trong dung sai 1e-5")
````

### 12.3. Tính chất: biến đổi ngược và khoảng cách giữa hai giá trị

Từ Kelvin và Fahrenheit ta khôi phục `celsius`; ngoài ra `F` và `K` luôn thỏa quan hệ `F = (K - 273.15) * 1.8 + 32`.

````python
random.seed(11)
for _ in range(2000):
    celsius = random.randint(0, 100000) / 100
    k, f = Solution().convertTemperature(celsius)
    assert math.isclose(k - 273.15, celsius, rel_tol=0, abs_tol=1e-9)
    assert math.isclose((f - 32) / 1.8, celsius, rel_tol=0, abs_tol=1e-9)
    assert math.isclose(f, (k - 273.15) * 1.8 + 32, rel_tol=0, abs_tol=1e-9)
    assert k >= 273.15 and f >= 32.0                       # vì celsius >= 0
print("các tính chất nghịch đảo đều đúng")
````

## 13. Các lỗi người mới hay gặp

### Lỗi 1: Nhầm hệ số `9/5` thành `5/9`

````python
from typing import List

def sai_ti_le(celsius: float) -> List[float]:
    return [celsius + 273.15, celsius * 5 / 9 + 32]            # SAI: 5/9 dùng để đổi F -> C

c = 36.50
print("đúng:", Solution().convertTemperature(c), "| sai:", sai_ti_le(c))
assert not close(Solution().convertTemperature(c), sai_ti_le(c))
````

**Vì sao sai:** đổi **C → F** nhân với `9/5 = 1.8`; còn `5/9` dùng cho chiều ngược lại (**F → C**: `(F - 32) * 5/9`). Kiểm tra nhanh bằng mốc `100 °C = 212 °F`: dùng `5/9` ra `87.5...`.

### Lỗi 2: Đảo thứ tự `[Fahrenheit, Kelvin]`

````python
def sai_thu_tu(celsius: float) -> List[float]:
    return [celsius * 1.80 + 32.00, celsius + 273.15]            # SAI: Kelvin phải đứng trước

c = 36.50
print("đúng:", Solution().convertTemperature(c), "| sai:", sai_thu_tu(c))
assert not close(Solution().convertTemperature(c), sai_thu_tu(c))
````

**Vì sao sai:** đề quy định `ans = [kelvin, fahrenheit]`. Dù mỗi giá trị đúng, đảo vị trí vẫn bị chấm sai. Hãy đặt tên biến `kelvin`, `fahrenheit` rồi `return [kelvin, fahrenheit]` để khó nhầm.

### Lỗi 3: Chia nguyên `9 // 5` (kết quả `1`)

````python
def sai_chia_nguyen(celsius: float) -> List[float]:
    return [celsius + 273.15, celsius * (9 // 5) + 32]          # SAI: 9 // 5 == 1

print(9 // 5, 9 / 5)
c = 36.50
print("đúng:", Solution().convertTemperature(c), "| sai:", sai_chia_nguyen(c))
assert not close(Solution().convertTemperature(c), sai_chia_nguyen(c))
````

**Vì sao sai:** toán tử `//` là **chia nguyên** (làm tròn xuống). `9 // 5 = 1`, mất phần `0.8`. Với số thực dùng `/` hoặc viết thẳng `1.8`.

### Lỗi 4: Làm tròn thành số nguyên (`round`, `int`)

````python
def sai_lam_tron(celsius: float) -> List[float]:
    return [round(celsius + 273.15), round(celsius * 1.80 + 32.00)]   # SAI: mất phần thập phân

c = 36.50
print("đúng:", Solution().convertTemperature(c), "| sai:", sai_lam_tron(c))
assert not close(Solution().convertTemperature(c), sai_lam_tron(c))
````

**Vì sao sai:** đề chỉ chấp nhận sai số `10^-5`; làm tròn đến số nguyên làm sai tới `0.35`. Hãy giữ nguyên số thực, và nếu cần hiển thị đẹp thì định dạng lúc `print` (ví dụ `f"{x:.5f}"`) chứ không làm tròn giá trị trả về.

### Lỗi 5: So sánh số thực bằng `==` trong test

````python
k_f = Solution().convertTemperature(122.11)
print(k_f)
print(k_f == [395.26, 251.798])          # có thể False vì sai số nhị phân
print(close(k_f, [395.26, 251.798]))     # so sánh với dung sai: True
````

**Vì sao nguy hiểm:** `==` giữa hai số `float` yêu cầu **trùng từng bit**; kết quả phép tính thường lệch ở chữ số cuối. Dùng `math.isclose(..., abs_tol=1e-5)` như trong bộ test của notebook.

- Dùng `1.80` hoặc `1.8` đều được; tránh `9 // 5`.
- Trong test, so sánh số thực bằng `math.isclose` với `abs_tol`.
- Khi muốn chính xác tuyệt đối để làm "oracle" cho test, dùng `Fraction(str(x))`.
- Nếu đề cho `celsius` là chuỗi, dùng `float(s)` (hoặc `Decimal(s)` nếu cần chính xác).

## 15. Biến thể của bài

1. **Đổi ngược:** từ Kelvin hoặc Fahrenheit về Celsius: `C = K - 273.15`; `C = (F - 32) / 1.8`.
2. **Thêm đơn vị Rankine, Réaumur:** `Rankine = (C + 273.15) * 1.8`; `Réaumur = C * 0.8`.
3. **Cho danh sách nhiệt độ:** đổi hàng loạt bằng list comprehension.
4. **Nhiệt độ ở đầu vào có đơn vị** (`"36.5C"`): tách chuỗi, đổi theo đơn vị.
5. **In đẹp:** định dạng 5 chữ số thập phân `f"{x:.5f}"`.
6. **Điểm giao nhau của hai thang đo:** `C = F` tại `-40`, giải `1.8·x + 32 = x`.

## 19. Đáp án tham khảo cho bài tập

Chạy kiểm tra bằng `assert`.

````python
import math

# Bài 1
assert close(Solution().convertTemperature(0), [273.15, 32.0])
assert close(Solution().convertTemperature(100), [373.15, 212.0])

# Bài 2: 50 -> 323.15 K, 122 F
assert close(Solution().convertTemperature(50), [323.15, 122.0])

# Bài 3
def fahrenheit_to_celsius(f):
    return (f - 32) * 5 / 9

# Bài 4
def convert_many(temps):
    return [Solution().convertTemperature(t) for t in temps]

# Bài 5
def convert_all_units(celsius):
    kelvin, fahrenheit = Solution().convertTemperature(celsius)
    return {
        "kelvin": kelvin,
        "fahrenheit": fahrenheit,
        "rankine": (celsius + 273.15) * 1.8,
        "reaumur": celsius * 0.8,
    }

assert math.isclose(fahrenheit_to_celsius(212), 100.0)
assert math.isclose(fahrenheit_to_celsius(32), 0.0, abs_tol=1e-12)
assert math.isclose(fahrenheit_to_celsius(-40), -40.0)
out = convert_many([0, 100])
assert close(out[0], [273.15, 32.0]) and close(out[1], [373.15, 212.0])
assert convert_many([]) == []
d = convert_all_units(100)
assert math.isclose(d["rankine"], 671.67)
assert math.isclose(d["reaumur"], 80.0)
assert math.isclose(d["rankine"], d["fahrenheit"] + 459.67)
assert set(d) == {"kelvin", "fahrenheit", "rankine", "reaumur"}
print("tất cả bài tập OK")
````
