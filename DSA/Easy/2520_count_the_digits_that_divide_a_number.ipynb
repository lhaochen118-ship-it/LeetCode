## 2. Đề bài & ví dụ

Cho số nguyên `num`. Trả về **số chữ số** của `num` mà **chia hết** `num`. Một chữ số `d` chia hết `num` nếu `num % d == 0`. Chữ số xuất hiện nhiều lần được đếm nhiều lần.

| Ví dụ | num | Output | Giải thích |
|---|---|---|---|
| 1 | `7` | `1` | `7` chia hết `7` |
| 2 | `121` | `2` | `121 % 1 == 0` (hai lần), `121 % 2 == 1` nên chữ số `2` không chia hết; đếm hai chữ số `1` |
| 3 | `1248` | `4` | `1`, `2`, `4`, `8` đều chia hết `1248` |

Constraints: `1 <= num <= 10^9`; `num` **không chứa** chữ số `0`.

````python
examples = [(7, 1), (121, 2), (1248, 4)]
for num, expected in examples:
    digits = [int(ch) for ch in str(num)]
    good = [d for d in digits if num % d == 0]
    print(f"num={num}: chữ số {digits}, chia hết: {good} -> {len(good)} (mong đợi {expected})")
````

## 3. Kiến thức nền

### 3.1. Phép chia hết bằng `%`

`a % b` là số dư của `a` chia `b`. `a` chia hết cho `b` khi `a % b == 0`. Lưu ý `a % 0` gây lỗi `ZeroDivisionError`, vì vậy đề đảm bảo không có chữ số `0`.

````python
print(1248 % 4, 1248 % 8, 121 % 2)
try:
    print(5 % 0)
except ZeroDivisionError as e:
    print("ZeroDivisionError:", e)
````

### 3.2. Tách chữ số bằng chuỗi

`str(num)` đổi số sang chuỗi; duyệt từng ký tự rồi `int(ch)` để đổi lại thành số nguyên. Ký tự `'7'` và số `7` là hai kiểu khác nhau, nên `num % '7'` gây `TypeError`.

````python
num = 1248
print(str(num), [ch for ch in str(num)])
print([int(ch) for ch in str(num)])
try:
    num % "4"
except TypeError as e:
    print("TypeError:", e)
````

### 3.3. Tách chữ số bằng toán: `% 10`, `// 10`, `divmod`

`num % 10` là chữ số cuối; `num // 10` bỏ chữ số cuối. `divmod(num, 10)` trả `(num // 10, num % 10)` trong một lần gọi. Lặp đến khi `num` bằng `0`.

````python
num = 1248
while num > 0:
    num, d = divmod(num, 10)
    print("chữ số cuối:", d, "| phần còn lại:", num)
````

### 3.4. `sum` với điều kiện và mảng đếm tần suất

`sum(1 for x in xs if cond)` đếm số phần tử thỏa điều kiện (vì `True` cộng như `1`). Mảng `cnt[d]` ghi số lần chữ số `d` xuất hiện; tổng các `cnt[d]` với `d` thỏa điều kiện là đáp án.

````python
xs = [1, 2, 1]
print(sum(1 for x in xs if 121 % x == 0))
cnt = [0] * 10
for x in xs:
    cnt[x] += 1
print(cnt)
````

## 4. Ý tưởng nền tảng / cách nghĩ trực giác

Tính tay `121`: các chữ số `1`, `2`, `1`.

- `121 % 1 = 0`: chia hết, đếm 1.
- `121 % 2 = 1`: không chia hết, bỏ.
- `121 % 1 = 0`: chia hết, đếm thêm 1.

Tổng `2`. Chú ý rằng ta luôn chia **số gốc `121`**, không phải phần đã bị cắt.

Quan sát nhanh: chữ số `1` luôn chia hết mọi số; chữ số `d` chia hết `num` khi `num` là bội của `d`. Số `num` có tối đa `9` chữ số (vì `<= 10^9` và không có `0`), nên chỉ có tối đa `9` phép kiểm tra.

````python
for num in (7, 121, 1248, 35, 98):
    digits = [int(c) for c in str(num)]
    print(num, [(d, num % d == 0) for d in digits])
````

## 5. Cách 1 — Brute Force / baseline: duyệt chuỗi

**Ý tưởng:** đổi `num` thành chuỗi, với mỗi ký tự đổi thành số nguyên `d` rồi kiểm tra `num % d == 0`.

````python
class SolutionBrute:
    def countDigits(self, num: int) -> int:
        count = 0
        for ch in str(num):
            d = int(ch)
            if num % d == 0:
                count += 1
        return count
````

````python
sol = SolutionBrute()
print(sol.countDigits(7), sol.countDigits(121), sol.countDigits(1248))    # 1 2 4
````

**Độ phức tạp:** Time `O(log num)` (số chữ số là `~log10(num)`), Space `O(log num)` cho chuỗi. **Nút thắt:** tạo chuỗi trung gian; có thể tách chữ số bằng toán để chỉ dùng `O(1)` bộ nhớ.

## 6. Ý tưởng thuật toán / bước đột phá

Ta không cần chuỗi: lặp `temp = num`, lấy chữ số cuối `d = temp % 10`, kiểm tra `num % d == 0` (**luôn dùng `num` gốc**), rồi `temp //= 10`. Dừng khi `temp == 0`.

Ngoài ra vì chữ số chỉ có `9` giá trị `1..9`, ta có thể:

- Đếm tần suất `cnt[d]` của từng chữ số, rồi với mỗi `d` từ `1` đến `9`, nếu `cnt[d] > 0` và `num % d == 0` thì cộng `cnt[d]`. Số phép chia hết tối đa là `9` bất kể độ dài số.
- Hoặc viết đệ quy: `f(num) = f(num // 10) + (1 nếu chữ số cuối chia hết)`.

**Quy tắc:** *với mỗi chữ số `d` của `num` (kể cả lặp), cộng 1 nếu `num % d == 0`; luôn chia số gốc.*

**Invariant:** ở mỗi bước, `temp` chứa các chữ số chưa xét (phần đầu của số) và `count` là số chữ số đã xét thỏa điều kiện.

## 7. Các cách tiếp theo

### Cách 2 — Tách chữ số bằng `divmod`

````python
class SolutionDivmod:
    def countDigits(self, num: int) -> int:
        count = 0
        temp = num
        while temp > 0:
            temp, d = divmod(temp, 10)       # d là chữ số cuối
            if num % d == 0:                 # chia số GỐC, không chia temp
                count += 1
        return count
````

````python
sol = SolutionDivmod()
print(sol.countDigits(7), sol.countDigits(121), sol.countDigits(1248))    # 1 2 4
````

**Độ phức tạp:** Time `O(log num)`, Space `O(1)`.

### Cách 3 — Mảng đếm tần suất chữ số `1..9`

Đếm số lần mỗi chữ số xuất hiện, sau đó chỉ cần tối đa 9 phép kiểm tra chia hết.

````python
class SolutionFreq:
    def countDigits(self, num: int) -> int:
        cnt = [0] * 10
        temp = num
        while temp > 0:
            cnt[temp % 10] += 1
            temp //= 10
        return sum(cnt[d] for d in range(1, 10) if cnt[d] and num % d == 0)
````

````python
sol = SolutionFreq()
print(sol.countDigits(7), sol.countDigits(121), sol.countDigits(1248))    # 1 2 4
````

**Độ phức tạp:** Time `O(log num + 9)`, Space `O(10) = O(1)`. Chỗ lợi: số phép `%` chia hết chỉ tối đa 9, không phụ thuộc số chữ số lặp.

### Cách 4 — Đệ quy

Hàm phụ `rec(rest)`: nếu `rest == 0` trả `0`; ngược lại tách `rest, d = divmod(rest, 10)`, cộng `1` nếu `num % d == 0`, rồi gọi tiếp. Độ sâu đệ quy chỉ bằng số chữ số (`<= 9`).

````python
class SolutionRecursive:
    def countDigits(self, num: int) -> int:
        def rec(rest: int) -> int:
            if rest == 0:
                return 0
            rest, d = divmod(rest, 10)
            return (1 if num % d == 0 else 0) + rec(rest)
        return rec(num)
````

````python
sol = SolutionRecursive()
print(sol.countDigits(7), sol.countDigits(121), sol.countDigits(1248))    # 1 2 4
````

**Độ phức tạp:** Time `O(log num)`, Space `O(log num)` cho ngăn xếp đệ quy (tối đa 9 tầng).

## 8. TRACE từng bước

Trace Cách 2 với `num = 121`:

| Vòng | `temp` trước | chữ số `d` | `temp` sau | `121 % d` | `count` |
|---|---|---|---|---|---|
| 1 | 121 | 1 | 12 | 0 | 1 |
| 2 | 12 | 2 | 1 | 1 | 1 |
| 3 | 1 | 1 | 0 | 0 | 2 |

Kết quả `2`.

````python
def trace_divmod(num):
    print("num =", num)
    count, temp, step = 0, num, 0
    while temp > 0:
        step += 1
        before = temp
        temp, d = divmod(temp, 10)
        ok = (num % d == 0)
        if ok:
            count += 1
        print(f"  vòng {step}: temp={before} -> d={d}, temp mới={temp}, {num} % {d} = {num % d}, chia hết? {ok}, count={count}")
    print("kết quả:", count)
    return count

trace_divmod(121)
trace_divmod(1248)
trace_divmod(35)
````

## 9. Code nộp LeetCode

Cách `divmod`, chỉ dùng `O(1)` bộ nhớ:

````python
class Solution:
    def countDigits(self, num: int) -> int:
        count = 0
        temp = num
        while temp > 0:
            temp, d = divmod(temp, 10)
            if num % d == 0:
                count += 1
        return count
````

````python
print(Solution().countDigits(7))
print(Solution().countDigits(121))
print(Solution().countDigits(1248))
````

## 10. Vì sao thuật toán đúng

1. **Tách chữ số đúng.** Với `temp > 0`, `temp % 10` là chữ số cuối của `temp` và `temp // 10` là phần còn lại. Lặp cho tới khi `temp = 0` đi qua **đúng một lần** từng chữ số của `num` (từ phải sang trái), kể cả chữ số lặp.
2. **Kiểm tra đúng số gốc.** Điều kiện của đề là `num` chia hết cho chữ số `d`, nên ta dùng `num % d` chứ không dùng `temp % d`.
3. **An toàn khi chia.** Đề đảm bảo `num` không chứa chữ số `0`, nên `d >= 1` và `num % d` không gây `ZeroDivisionError`.
4. **Kết thúc.** `temp` giảm ít nhất 10 lần mỗi vòng, nên dừng sau tối đa `9` vòng (do `num <= 10^9` và không có `0`).
5. **Đếm lặp.** Mỗi lần gặp một chữ số thỏa điều kiện, `count` tăng 1, nên chữ số xuất hiện `k` lần được tính `k` lần, đúng yêu cầu.
6. **Các cách khác tương đương.** Cách chuỗi duyệt cùng dãy chữ số; cách tần suất gom `cnt[d]` lần cùng điều kiện `num % d == 0` nên cộng `cnt[d]` cho từng `d` thỏa.

| 3. Mảng tần suất | `O(log num)` | `O(1)` (mảng 10 ô) |

## 12. Bộ test

### 12.1. Ví dụ của đề, edge case và phản ví dụ

Gồm số một chữ số, chữ số lặp, số mà không chữ số nào (ngoài `1`) chia hết, và biên `999999999`. Mỗi kỳ vọng tính tay (ghi chú bên cạnh).

````python
sols = [SolutionBrute(), SolutionDivmod(), SolutionFreq(), SolutionRecursive(), Solution()]
cases = [
    (7, 1),                  # ví dụ 1
    (121, 2),                # ví dụ 2
    (1248, 4),               # ví dụ 3
    (1, 1),                  # nhỏ nhất
    (9, 1),                  # một chữ số
    (11, 2),                 # 1 và 1
    (12, 2),                 # 1 và 2
    (13, 1),                 # 1 chia hết, 3 không (13 % 3 = 1)
    (15, 2),                 # 1 và 5
    (35, 1),                 # 3 không, 5 có
    (98, 0),                 # 9 và 8 không chia hết 98
    (99, 2),                 # 9 chia hết 99 hai lần
    (111, 3),                # ba chữ số 1
    (128, 3),                # 1, 2, 8 đều chia hết 128
    (123, 2),                # 1 và 3 chia hết, 2 không
    (999999999, 9),          # biên trên hợp lệ (không có số 0): chín chữ số 9
]
for num, want in cases:
    for s in sols:
        got = s.countDigits(num)
        assert got == want, (type(s).__name__, num, got, want)
print("tất cả test cố định đều đúng với cả 5 cách:", len(cases))
````

Một vài kiểm tra tay: `98 % 9 = 8`, `98 % 8 = 2` nên cả hai không chia hết, đáp án `0` (nhắc rằng `1` không có mặt trong số này). `99 % 9 = 0` hai lần nên `2`. `128`: `128 % 1 = 0`, `128 % 2 = 0`, `128 % 8 = 0` nên `3`. Số `98` cho thấy đáp án có thể là `0`.

### 12.2. Đối chiếu ngẫu nhiên (có seed)

Sinh `num` đúng constraints: gồm `1..9` chữ số, mỗi chữ số lấy từ `1..9` (không có chữ số `0`), nên `1 <= num <= 999999999 <= 10^9`.

````python
import random
random.seed(2520)

def gen_num():
    length = random.randint(1, 9)
    return int("".join(random.choice("123456789") for _ in range(length)))

def oracle(num):
    return sum(1 for ch in str(num) if num % int(ch) == 0)

for _ in range(4000):
    num = gen_num()
    assert 1 <= num <= 10**9 and "0" not in str(num)
    ans = [s.countDigits(num) for s in sols]
    assert len(set(ans)) == 1 and ans[0] == oracle(num), (num, ans)
print("4000 test ngẫu nhiên: 5 cách khớp nhau và khớp oracle")
````

### 12.3. Quét toàn bộ số nhỏ và tính chất

Duyệt mọi `num` từ `1` đến `20000` **không chứa chữ số 0** và so sánh các cách. Tính chất: đáp án `>= số lượng chữ số 1`; đáp án `<= số chữ số của num`; số có toàn chữ số `1` thì đáp án bằng độ dài.

````python
checked = 0
for num in range(1, 20001):
    if "0" in str(num):
        continue
    checked += 1
    ans = [s.countDigits(num) for s in sols]
    assert len(set(ans)) == 1, (num, ans)
    r = ans[0]
    assert str(num).count("1") <= r <= len(str(num))
for k in range(1, 10):
    assert Solution().countDigits(int("1" * k)) == k
print("đã kiểm tra", checked, "số; các tính chất đều đúng")
````

## 13. Các lỗi người mới hay gặp

### Lỗi 1: Dùng `set` làm mất chữ số lặp

````python
def sai_set(num):
    return sum(1 for d in set(int(c) for c in str(num)) if num % d == 0)     # SAI: bỏ trùng

print("đúng:", Solution().countDigits(121), "| sai:", sai_set(121))
assert Solution().countDigits(121) != sai_set(121)
````

**Vì sao sai:** đề đếm cả chữ số lặp lại. `121` có hai chữ số `1` nên đáp án `2`; `set` chỉ giữ `{1, 2}` nên chỉ đếm được `1`.

### Lỗi 2: Sửa `num` trong vòng lặp rồi vẫn dùng `num % d`

````python
def sai_sua_num(num):
    count = 0
    while num > 0:
        d = num % 10
        if num % d == 0:                 # SAI: num đã bị cắt dần, không còn là số gốc
            count += 1
        num //= 10
    return count

print("đúng:", Solution().countDigits(121), "| sai:", sai_sua_num(121))
assert Solution().countDigits(121) != sai_sua_num(121)
print("đúng:", Solution().countDigits(35), "| sai:", sai_sua_num(35))
````

**Vì sao sai:** sau mỗi vòng `num` mất chữ số cuối, nên ta kiểm tra chia hết cho một số **khác** số gốc. Ở `121`: vòng 2 kiểm tra `12 % 2 == 0` và đếm sai chữ số `2`. Luôn giữ `num` gốc và dùng biến phụ `temp` để cắt.

### Lỗi 3: Đảo thứ tự `d % num`

````python
def sai_dao_thu_tu(num):
    return sum(1 for c in str(num) if int(c) % num == 0)          # SAI: d chia hết num hay num chia hết d?

print("đúng:", Solution().countDigits(121), "| sai:", sai_dao_thu_tu(121))
assert Solution().countDigits(121) != sai_dao_thu_tu(121)
print("đúng:", Solution().countDigits(1248), "| sai:", sai_dao_thu_tu(1248))
````

**Vì sao sai:** "`d` chia hết `num`" nghĩa là `num % d == 0` (số bị chia là `num`). Viết `d % num` hỏi ngược lại: `num` có chia hết `d` không; với `121` mọi chữ số `d < 121` nên `d % 121 = d != 0`, kết quả sai là `0` thay vì `2`.

### Lỗi 4: Quên đổi ký tự sang số nguyên

````python
def sai_quen_int(num):
    count = 0
    for ch in str(num):
        if num % ch == 0:               # SAI: ch là chuỗi, không phải số
            count += 1
    return count

try:
    sai_quen_int(121)
except TypeError as e:
    print("TypeError:", e)
assert Solution().countDigits(121) == 2
````

**Cách tránh:** luôn viết `d = int(ch)` trước khi tính toán. Hoặc dùng cách toán học với `divmod` để không cần đổi kiểu.

### Lỗi 5: Đếm chữ số **không** chia hết (đảo điều kiện)

````python
def sai_dao_dieu_kien(num):
    return sum(1 for c in str(num) if num % int(c) != 0)          # SAI: đếm chữ số không chia hết

print("đúng:", Solution().countDigits(35), "| sai:", sai_dao_dieu_kien(35))
assert Solution().countDigits(123) != sai_dao_dieu_kien(123)         # đúng 2, sai 1
assert Solution().countDigits(98) != sai_dao_dieu_kien(98)
````

**Vì sao sai:** đảo `==` thành `!=` đếm phần bù. Với `98` đáp án đúng `0` còn phiên bản sai ra `2`; với `123` đúng `2`, sai `1`. Mẹo: đọc lại câu "chữ số **chia hết** `num`" khi viết điều kiện.

## 14. Mẹo làm bài nhanh & nhận dạng pattern

- Thấy "chữ số của một số" ⇒ nghĩ ngay **`divmod(x, 10)`** hoặc **`str(x)`**.
- "Chia hết" ⇒ `%`; nhớ số bị chia là `num` gốc.
- Chữ số `1` luôn chia hết mọi số; có thể dùng để kiểm tra nhanh kết quả.
- Vì `num <= 10^9` nên số chữ số rất ít: mọi cách đều nhanh, ưu tiên cách dễ đọc.
- Một dòng nộp bài: `return sum(num % int(c) == 0 for c in str(num))`.
- Đề cam đoan không có chữ số `0`, nên không cần `if d == 0: continue`.

````python
def one_line(num):
    return sum(num % int(c) == 0 for c in str(num))

assert [one_line(7), one_line(121), one_line(1248)] == [1, 2, 4]
print("phiên bản một dòng đúng")
````

class Solution:
    def countDigits(self, num: int) -> int:
        count, temp = 0, num
        while temp > 0:
            temp, d = divmod(temp, 10)
            if num % d == 0:

        return count
