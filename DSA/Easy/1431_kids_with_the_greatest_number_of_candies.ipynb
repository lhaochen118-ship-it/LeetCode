## 1. Đề bài

Có `n` bạn nhỏ; `candies[i]` là số kẹo của bạn thứ `i`. Ta có `extraCandies` viên kẹo thêm.

Trả về mảng `result` gồm `n` giá trị `bool`: `result[i] = True` nếu **đưa hết** `extraCandies` cho bạn `i` thì bạn ấy có số kẹo **nhiều nhất** trong tất cả các bạn (được phép **bằng** người khác), ngược lại `False`.

| `candies` | `extra` | Output |
|---|---:|---|
| `[2,3,5,1,3]` | 3 | `[True, True, True, False, True]` |
| `[4,2,1,1,2]` | 1 | `[True, False, False, False, False]` |
| `[12,1,12]` | 10 | `[True, False, True]` |

**Constraints:** `2 <= n <= 100`, `1 <= candies[i] <= 100`, `1 <= extraCandies <= 50`.

Ví dụ 1, bạn thứ 4 (`candies = 1`): `1 + 3 = 4 < 5` ⇒ `False`. Bạn thứ 2 (`3`): `3 + 3 = 6 >= 5` ⇒ `True`.

````python
examples = [
    ([2, 3, 5, 1, 3], 3, [True, True, True, False, True]),
    ([4, 2, 1, 1, 2], 1, [True, False, False, False, False]),
    ([12, 1, 12], 10, [True, False, True]),
]
for candies, extra, expected in examples:
    print(f"candies={candies}, extra={extra} -> kỳ vọng {expected}")
````

## 2. Nền tảng Python cần biết

### 2.1. Kiểu `bool` và phép so sánh

`a >= b` trả về `True` hoặc `False` — chính là giá trị cần đưa vào `result`. Không cần viết `if ...: True else: False`.

````python
print(6 >= 5, 4 >= 5, 5 >= 5)
print(type(6 >= 5))
flag = 3 + 3 >= 5
print("flag =", flag)
````

### 2.2. `max(list)` — giá trị lớn nhất

`max` duyệt cả list một lần, `O(n)`. Gọi `max` **bên trong** vòng lặp sẽ thành `O(n²)` — cần tránh.

````python
candies = [2, 3, 5, 1, 3]
print("max =", max(candies))
````

### 2.3. List comprehension tạo `list[bool]`

`[biểu_thức for x in list]` tạo list mới. Nếu biểu thức là phép so sánh thì ta được list `True/False`.

````python
nums = [2, 3, 5, 1, 3]
print([x * 10 for x in nums])
print([x >= 3 for x in nums])
````

### 2.4. `enumerate` và `all`

- `enumerate(list)` cho cặp `(i, x)`.
- `all(điều_kiện for ...)` → `True` nếu **mọi** phần tử thỏa (dùng ở cách trực giác: "nhiều hơn hoặc bằng **tất cả** bạn khác").

````python
for i, x in enumerate([2, 3, 5]):
    print(i, x)
print(all(v <= 6 for v in [2, 3, 5]), all(v <= 4 for v in [2, 3, 5]))
````

## 3. Cách trực giác (baseline) — mô phỏng thật: đưa kẹo rồi so với mọi bạn

Với mỗi bạn `i`:
1. Tạo bản sao `after = candies[:]` và cộng `after[i] += extraCandies`.
2. Kiểm tra `after[i] == max(after)`.

Cách này **đúng nghĩa đen** với đề, dễ tin tưởng — ta dùng làm **trọng tài** khi test.

````python
def kids_bruteforce(candies: list[int], extra: int) -> list[bool]:
    result = []
    for i in range(len(candies)):
        after = candies[:]              # bản sao, không sửa mảng gốc
        after[i] += extra
        result.append(after[i] == max(after))
    return result

for candies, extra, expected in examples:
    print(candies, extra, "->", kids_bruteforce(candies, extra), "| khớp?", kids_bruteforce(candies, extra) == expected)
````

`candies[:]` là slicing toàn bộ ⇒ tạo **bản sao nông** (shallow copy). Nếu viết `after = candies` thì `after` chỉ là **tên khác** của cùng một list — sửa `after` sẽ sửa luôn `candies`:

````python
a = [1, 2, 3]
alias = a
copy = a[:]
alias[0] = 99
print("a =", a, "| copy =", copy)
````

### 3.1. Nút thắt

Mỗi bạn: sao chép `O(n)` + `max` `O(n)` ⇒ tổng `O(n²)`. Quan sát: ta lặp đi lặp lại việc tìm max của **gần như cùng một mảng**. Nhưng thật ra:
- Khi đưa kẹo cho bạn `i`, các bạn **khác** không đổi;
- Bạn `i` chỉ cần **không thua** người nhiều kẹo nhất trong số **những người khác**.

````python
ops = 0
def count_ops(candies, extra):
    global ops
    for i in range(len(candies)):
        ops += 2 * len(candies)          # sao chép + max
for n in [10, 100]:
    ops = 0
    count_ops([1] * n, 1)
    print(f"n = {n:>3}: baseline ≈ {ops} thao tác; cách tối ưu ≈ {2 * n}")
````

## 4. Insight — chỉ cần so với **max của mảng gốc**

Gọi `M = max(candies)`. Bạn `i` sau khi nhận kẹo có `candies[i] + extra`.

- Nếu bạn `i` **chính là** người có `M` (hoặc một trong số đó): `candies[i] + extra > M` ⇒ chắc chắn nhiều nhất ⇒ `True`.
- Nếu không: người nhiều nhất trong số bạn khác vẫn là `M` ⇒ cần `candies[i] + extra >= M`.

Cả hai trường hợp gộp thành **một điều kiện**: `candies[i] + extra >= M`.

Tương đương: `candies[i] >= M − extra` — một **ngưỡng** cố định.

````python
candies, extra = [2, 3, 5, 1, 3], 3
M = max(candies)
threshold = M - extra
print("M =", M, "| ngưỡng = M - extra =", threshold)
for i, c in enumerate(candies):
    print(f"bạn {i}: {c} + {extra} = {c + extra} >= {M}? {c + extra >= M} | {c} >= {threshold}? {c >= threshold}")
````

## 5. Lời giải chính — giảng từng đoạn

**Đoạn 1:** `M = max(candies)` — **một lần**, ngoài vòng lặp.

**Đoạn 2:** duyệt từng `c` trong `candies`.

**Đoạn 3:** thêm `c + extraCandies >= M` vào `result`.

````python
def kids_max(candies: list[int], extra: int) -> list[bool]:
    M = max(candies)                 # Đoạn 1
    result = []
    for c in candies:                # Đoạn 2
        result.append(c + extra >= M)  # Đoạn 3
    return result

for candies, extra, expected in examples:
    print(candies, extra, "->", kids_max(candies, extra), "| khớp?", kids_max(candies, extra) == expected)
````

### 5.1. Bản list comprehension

````python
def kids_comprehension(candies: list[int], extra: int) -> list[bool]:
    M = max(candies)
    return [c + extra >= M for c in candies]

print([kids_comprehension(c, e) for c, e, _ in examples])
````

## 6. Cách 3 — tự tìm max bằng vòng lặp (không dùng `max`)

Hữu ích khi phỏng vấn yêu cầu không dùng hàm có sẵn, hoặc khi chuyển sang ngôn ngữ khác. Hai lượt duyệt: lượt 1 tìm `M`, lượt 2 so sánh.

````python
def kids_manual(candies: list[int], extra: int) -> list[bool]:
    M = candies[0]
    for c in candies[1:]:
        if c > M:
            M = c
    return [c + extra >= M for c in candies]

print([kids_manual(c, e) for c, e, _ in examples])
````

## 7. Cách 4 — so với max của **những người khác** (prefix max + suffix max)

Đây là cách "trung thực" hơn với đề: bạn `i` cần `candies[i] + extra >= max(các bạn khác)`. Dùng **prefix max** (max từ đầu tới `i − 1`) và **suffix max** (max từ `i + 1` tới cuối) để mỗi bạn tốn `O(1)`.

Cách này dài hơn và cho cùng kết quả — nhưng kỹ thuật prefix/suffix max rất hữu ích ở bài khó hơn (ví dụ LeetCode 238 *Product of Array Except Self*, LeetCode 42 *Trapping Rain Water*).

````python
def kids_prefix_suffix(candies: list[int], extra: int) -> list[bool]:
    n = len(candies)
    NEG = float("-inf")
    prefix = [NEG] * n               # prefix[i] = max(candies[0..i-1])
    suffix = [NEG] * n               # suffix[i] = max(candies[i+1..n-1])
    for i in range(1, n):
        prefix[i] = max(prefix[i - 1], candies[i - 1])
    for i in range(n - 2, -1, -1):
        suffix[i] = max(suffix[i + 1], candies[i + 1])
    return [candies[i] + extra >= max(prefix[i], suffix[i]) for i in range(n)]

print([kids_prefix_suffix(c, e) for c, e, _ in examples])
````

`range(n - 2, -1, -1)` đi lùi từ `n − 2` về `0`.

## 8. TRACE từng bước

Trace lời giải chính với `candies = [4, 2, 1, 1, 2]`, `extra = 1` (`M = 4`):

| i | candies[i] | + extra | >= 4? | result |
|---:|---:|---:|:---:|---|
| 0 | 4 | 5 | ✔ | `[T]` |
| 1 | 2 | 3 | ✘ | `[T, F]` |
| 2 | 1 | 2 | ✘ | `[T, F, F]` |
| 3 | 1 | 2 | ✘ | `[T, F, F, F]` |
| 4 | 2 | 3 | ✘ | `[T, F, F, F, F]` |

Bạn 0 vốn đã nhiều nhất nên luôn `True`; các bạn khác thiếu quá nhiều.

````python
def trace_kids(candies: list[int], extra: int) -> list[bool]:
    M = max(candies)
    print(f"candies = {candies}, extra = {extra}, M = {M}")
    result = []
    for i, c in enumerate(candies):
        ok = c + extra >= M
        result.append(ok)
        print(f"  i={i}: {c} + {extra} = {c + extra:>3} >= {M}? {ok!s:<5} -> result = {result}")
    return result

trace_kids([4, 2, 1, 1, 2], 1)
print()
trace_kids([12, 1, 12], 10)
````

Ví dụ `[12, 1, 12]` cho thấy nhánh **hòa**: có hai bạn cùng `12`; bạn 0 nhận kẹo thì `22 > 12` ⇒ `True`; bạn 1 được `11 < 12` ⇒ `False`.

Trace prefix/suffix max cho `[2, 3, 5, 1, 3]`:

````python
candies, extra = [2, 3, 5, 1, 3], 3
n = len(candies)
prefix = [float("-inf")] * n
suffix = [float("-inf")] * n
for i in range(1, n):
    prefix[i] = max(prefix[i - 1], candies[i - 1])
for i in range(n - 2, -1, -1):
    suffix[i] = max(suffix[i + 1], candies[i + 1])
for i in range(n):
    others = max(prefix[i], suffix[i])
    print(f"i={i}: c={candies[i]}, max trái={prefix[i]}, max phải={suffix[i]}, max người khác={others}, "
          f"{candies[i]}+{extra}={candies[i] + extra} >= {others}? {candies[i] + extra >= others}")
````

## 9. Code nộp LeetCode

````python
class Solution:
    def kidsWithCandies(self, candies: list[int], extraCandies: int) -> list[bool]:
        M = max(candies)
        return [c + extraCandies >= M for c in candies]

for c, e, expected in examples:
    print(Solution().kidsWithCandies(c, e), expected)
````

### Bản ngưỡng (một phép trừ thay cho `n` phép cộng)

````python
class SolutionThreshold:
    def kidsWithCandies(self, candies: list[int], extraCandies: int) -> list[bool]:
        need = max(candies) - extraCandies
        return [c >= need for c in candies]

print([SolutionThreshold().kidsWithCandies(c, e) for c, e, _ in examples])
````

Trên LeetCode, output `[true, true, ...]` là cách JSON in `bool`; Python trả `True/False` là đúng.

### Bảng so sánh

| Cách | Ý tưởng | Time | Space | Ghi chú |
|---|---|---|---|---|
| 1. Mô phỏng | sao chép + `max` cho từng bạn | `O(n²)` | `O(n)` | đúng nghĩa đen, làm oracle |
| 2. Max gốc | `c + extra >= max` | `O(n)` | `O(1)` phụ (ngoài kết quả) | chuẩn |
| 3. Tự tìm max | 2 lượt duyệt | `O(n)` | `O(1)` phụ | không dùng built-in |
| 4. Prefix/suffix max | max của người khác | `O(n)` | `O(n)` | kỹ thuật cho bài khó hơn |

## 10. Vì sao đúng?

Gọi `M = max(candies)`, `O_i` = max của các bạn **khác** `i`. Ta có `O_i <= M`.

Sau khi bạn `i` nhận kẹo, bạn ấy nhiều nhất ⇔ `candies[i] + extra >= O_i`.

- **(⇐)** Nếu `candies[i] + extra >= M` thì `>= O_i` (vì `O_i <= M`) ⇒ `True` đúng.
- **(⇒)** Giả sử `candies[i] + extra >= O_i`. Nếu `O_i = M` thì xong. Nếu `O_i < M` thì `M` không đến từ người khác ⇒ `M = candies[i]` ⇒ `candies[i] + extra = M + extra > M` ✔.

Vậy `candies[i] + extra >= O_i` ⇔ `candies[i] + extra >= M`. ∎

Lưu ý phần (⇒) dùng `extra >= 1` (đề cho `extraCandies >= 1`); thực ra chỉ cần `extra >= 0` là đủ (`M + 0 >= M`).

## 13. Lỗi người mới hay gặp

### Lỗi 1: Dùng `>` thay vì `>=` (quên trường hợp hòa)

````python
def wrong_strict(candies, extra):
    M = max(candies)
    return [c + extra > M for c in candies]        # SAI: hòa vẫn tính là nhiều nhất

print("sai:", wrong_strict([5, 10], 5), "| đúng:", kids_max([5, 10], 5))
````

### Lỗi 2: Gọi `max` trong vòng lặp → `O(n²)` (vẫn đúng, nhưng chậm)

`[c + extra >= max(candies) for c in candies]` tính lại `max` cho **mỗi** phần tử. Với `n = 100` không sao, nhưng là thói quen xấu.

### Lỗi 3: Sửa trực tiếp `candies` khi mô phỏng

Cộng `candies[i] += extra` để thử rồi **quên trừ lại** ⇒ các lượt sau sai.

````python
def wrong_mutate(candies, extra):
    result = []
    for i in range(len(candies)):
        candies[i] += extra                   # SAI: sửa mảng gốc, không hoàn tác
        result.append(candies[i] == max(candies))
    return result

print("sai:", wrong_mutate([2, 3, 5, 1, 3], 3), "| đúng:", kids_max([2, 3, 5, 1, 3], 3))
````

### Lỗi 4: Cộng extra vào **max** rồi so sánh

Nhầm đề thành "mỗi bạn được thêm extra **và** người giỏi nhất cũng được thêm" ⇒ so `c + extra >= M + extra` ⇔ `c >= M` ⇒ chỉ những người vốn đã max mới `True`.

````python
def wrong_extra_to_max(candies, extra):
    M = max(candies) + extra                   # SAI: extra chỉ đưa cho 1 bạn đang xét
    return [c + extra >= M for c in candies]

print("sai:", wrong_extra_to_max([2, 3, 5, 1, 3], 3), "| đúng:", kids_max([2, 3, 5, 1, 3], 3))
````

## 14. Mẹo nhận dạng pattern

- "So sánh mỗi phần tử với **cực trị** của cả mảng" ⇒ tính cực trị **một lần** trước vòng lặp.
- Điều kiện dạng `x + k >= M` ⇔ `x >= M − k` ⇒ chuyển thành **ngưỡng**.
- "Max của mọi phần tử **trừ** phần tử `i`" ⇒ prefix max + suffix max, hoặc lưu **max và max nhì**.
- Trả về `list[bool]` ⇒ list comprehension với phép so sánh.

## 15. Biến thể

1. **Bắt buộc nhiều hơn hẳn** (không cho hòa) — khi đó phải so với max của **người khác**, không còn đơn giản là `> M` (nếu bạn `i` là max duy nhất thì luôn đúng).
2. Đếm **số bạn** có thể trở thành nhiều nhất.
3. Mỗi bạn nhận thêm `extra[i]` riêng.
4. Các bài "trừ chính nó" (except self), ví dụ LeetCode 238 *Product of Array Except Self*: dùng prefix/suffix hoặc max + max nhì.

````python
def kids_strict(candies, extra):
    # nhiều hơn HẲN mọi bạn khác: dùng max và max nhì
    first = second = float("-inf")
    for c in candies:
        if c > first:
            first, second = c, first
        elif c > second:
            second = c
    res = []
    for c in candies:
        others = second if c == first else first     # nếu c là max (có thể trùng) -> max người khác
        res.append(c + extra > others)
    return res

def kids_strict_brute(candies, extra):
    return [candies[i] + extra > max(candies[:i] + candies[i + 1:]) for i in range(len(candies))]

print(kids_strict([5, 10], 5), kids_strict_brute([5, 10], 5))
print(kids_strict([9, 9, 1], 0), kids_strict_brute([9, 9, 1], 0))
print("số bạn có thể nhiều nhất:", sum(kids_max([2, 3, 5, 1, 3], 3)))
````

`candies[:i] + candies[i + 1:]` = mảng bỏ phần tử `i`. Với `[9, 9, 1]`, `extra = 0`: bạn 0 có max người khác là `9` (bạn 1) ⇒ `9 > 9` sai. Hàm `kids_strict` xử lý đúng vì khi gặp số `9` thứ hai, nhánh `elif c > second` cập nhật `second = 9` — tức **max nhì được phép bằng max** khi max xuất hiện nhiều lần.

````python
def answer_kids(candies, extra):
    M = max(candies)
    return [c + extra >= M for c in candies]

def answer_count_kids(candies, extra):
    return sum(answer_kids(candies, extra))

def answer_kids_each(candies, extras):
    M = max(candies)
    return [c + e >= M for c, e in zip(candies, extras)]

def answer_kids_strict(candies, extra):
    first = max(candies)
    count_first = candies.count(first)
    smaller = [c for c in candies if c < first]
    second = max(smaller) if smaller else float("-inf")
    res = []
    for c in candies:
        others = second if (c == first and count_first == 1) else first
        res.append(c + extra > others)
    return res

def brute_each(candies, extras):
    out = []
    for i in range(len(candies)):
        after = candies[:]
        after[i] += extras[i]
        out.append(after[i] == max(after))
    return out

a = [1]; b2 = a; b2.append(2)
assert answer_kids([1, 6, 4, 6], 2) == [False, True, True, True]
assert [x > 2 for x in [1, 2, 3]] == [False, False, True] and max([4, 9, 2]) - 3 == 6
assert sum([True, False, True]) == 2 and a == [1, 2]
assert answer_kids([2, 3, 5, 1, 3], 3) == [True, True, True, False, True]
assert answer_count_kids([4, 2, 1, 1, 2], 1) == 1
assert answer_kids_each([1, 5, 3], [5, 0, 1]) == [True, True, False]
assert answer_kids_strict([9, 9, 1], 0) == [False, False, False]
random.seed(2)
for _ in range(2000):
    candies = [random.randint(1, 8) for _ in range(random.randint(2, 8))]
    extra = random.randint(0, 6)
    extras = [random.randint(0, 6) for _ in candies]
    assert answer_kids_strict(candies, extra) == kids_strict_brute(candies, extra)
    assert answer_kids_each(candies, extras) == brute_each(candies, extras)
print("Tất cả đáp án PASS!")
print("so sánh kids_strict (mục 15) với vét cạn trên ca khó [9, 9, 1], extra=0:",
      kids_strict([9, 9, 1], 0), "vs", kids_strict_brute([9, 9, 1], 0))
````
