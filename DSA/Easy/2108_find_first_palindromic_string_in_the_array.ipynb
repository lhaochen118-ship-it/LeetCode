## 4. Ý tưởng nền tảng — tính bằng tay

`words = ["abc","car","ada","racecar","cool"]`. Xét từng từ theo thứ tự:

- `"abc"` → ngược `"cba"` ≠ → không.
- `"car"` → ngược `"rac"` ≠ → không.
- `"ada"` → ngược `"ada"` = → **có**, dừng, trả `"ada"`.

Các từ sau (`"racecar"` cũng là palindrome) không cần xét: đề muốn **đầu tiên**.

````python
for w in ["abc", "car", "ada"]:
    print(w, "->", w[::-1], "| palindrome?", w == w[::-1])
````

## 5. Cách 1 — Brute Force: tự dựng chuỗi đảo ngược

Cách nguyên thủy nhất: xây chuỗi đảo ngược bằng vòng lặp rồi so sánh.

````python
def first_pal_manual(words):
    for w in words:
        rev = ""
        for ch in w:
            rev = ch + rev        # ghép ký tự mới vào đầu
        if rev == w:
            return w
    return ""

for words, expected in examples:
    print(repr(first_pal_manual(words)), repr(expected))
````

Độ phức tạp: với `L` là độ dài lớn nhất, mỗi từ tốn `O(L²)` do ghép chuỗi (mỗi lần `ch + rev` sao chép chuỗi). Tổng `O(n·L²)`. Nút thắt: tạo chuỗi trung gian lãng phí, trong khi chỉ cần so sánh từng cặp ký tự đối xứng.

## 6. Ý tưởng thuật toán / bước đột phá

Palindrome đối xứng qua tâm: `s[i] == s[len(s)-1-i]` với mọi `i`. Chỉ cần so **nửa đầu** với nửa cuối (`i < len(s)//2`), ký tự giữa (nếu độ dài lẻ) khỏi cần so.

Quy tắc "nếu… thì…":
- Nếu có một cặp `s[i] != s[n-1-i]` → **không** palindrome, dừng sớm.
- Nếu mọi cặp đều khớp → palindrome.
- Duyệt `words` từ trái sang phải, từ palindrome đầu tiên → `return`; hết vòng lặp → `""`.

## 7. Các cách tiếp theo

### Cách 2 — Slicing `s == s[::-1]`

Cách Pythonic nhất: một dòng kiểm tra.

````python
def first_pal_slice(words):
    for w in words:
        if w == w[::-1]:
            return w
    return ""

for words, expected in examples:
    print(repr(first_pal_slice(words)), repr(expected))
````

````python
from typing import List

class SolutionSlice:
    def firstPalindrome(self, words: List[str]) -> str:
        for w in words:
            if w == w[::-1]:
                return w
        return ""

print(SolutionSlice().firstPalindrome(["abc", "car", "ada", "racecar", "cool"]))
````

Độ phức tạp: **Time `O(n·L)`** (`n` từ, mỗi từ dài tối đa `L`), **Space `O(L)`** cho chuỗi đảo ngược.

### Cách 3 — Hai con trỏ (không tạo chuỗi mới)

Đặt `i = 0`, `j = len(w) - 1`, so `w[i]` với `w[j]`, rồi `i += 1`, `j -= 1` cho tới khi `i >= j`.

````python
def is_pal_two_pointers(w):
    i, j = 0, len(w) - 1
    while i < j:
        if w[i] != w[j]:
            return False
        i += 1
        j -= 1
    return True

def first_pal_two(words):
    for w in words:
        if is_pal_two_pointers(w):
            return w
    return ""

for words, expected in examples:
    print(repr(first_pal_two(words)), repr(expected))
````

````python
from typing import List

class SolutionTwoPointers:
    def firstPalindrome(self, words: List[str]) -> str:
        def is_pal(w: str) -> bool:
            i, j = 0, len(w) - 1
            while i < j:
                if w[i] != w[j]:
                    return False
                i += 1
                j -= 1
            return True
        for w in words:
            if is_pal(w):
                return w
        return ""

print(SolutionTwoPointers().firstPalindrome(["notapalindrome", "racecar"]))
````

Độ phức tạp: **Time `O(n·L)`**, **Space `O(1)`**. Dừng ngay khi gặp cặp khác nhau nên thường nhanh hơn slicing trên từ dài không đối xứng.

### Cách 4 — `next()` với generator

Gom logic "tìm phần tử đầu tiên thỏa điều kiện" vào một dòng: `next((w for w in words if w == w[::-1]), "")`. Generator **lười** (lazy): dừng ngay khi tìm được từ đầu tiên.

````python
def first_pal_next(words):
    return next((w for w in words if w == w[::-1]), "")

for words, expected in examples:
    print(repr(first_pal_next(words)), repr(expected))
````

````python
from typing import List

class SolutionNext:
    def firstPalindrome(self, words: List[str]) -> str:
        return next((w for w in words if w == w[::-1]), "")

print(repr(SolutionNext().firstPalindrome(["def", "ghi"])))
````

Độ phức tạp: **Time `O(n·L)`**, **Space `O(L)`**. Dùng `""` làm giá trị mặc định đúng yêu cầu đề.

## 8. TRACE từng bước

Trace Cách 3 (hai con trỏ) với `words = ["abc", "car", "ada"]`:

| từ | i | j | w[i] vs w[j] | kết luận |
|---|---|---|---|---|
| `abc` | 0 | 2 | `a` vs `c` | khác → không |
| `car` | 0 | 2 | `c` vs `r` | khác → không |
| `ada` | 0 | 2 | `a` vs `a` | khớp |
| `ada` | 1 | 1 | `i >= j` | dừng → palindrome |

````python
def trace_two(words):
    for w in words:
        print("xét từ", repr(w))
        i, j = 0, len(w) - 1
        ok = True
        while i < j:
            print(f"  i={i}, j={j}: {w[i]!r} vs {w[j]!r}")
            if w[i] != w[j]:
                ok = False
                break
            i += 1
            j -= 1
        print("  ->", "palindrome, DỪNG" if ok else "không")
        if ok:
            return w
    return ""

print(repr(trace_two(["abc", "car", "ada", "racecar"])))
````

````python
def trace_slice(words):
    for k, w in enumerate(words):
        rev = w[::-1]
        print(f"words[{k}]={w!r}, đảo={rev!r}, bằng nhau? {w == rev}")
        if w == rev:
            return w
    return ""

print(repr(trace_slice(["xy", "zz", "abc"])))
print(repr(trace_slice(["xy", "ab"])))
````

## 9. Code nộp LeetCode

````python
from typing import List

class Solution:
    def firstPalindrome(self, words: List[str]) -> str:
        for w in words:
            if w == w[::-1]:
                return w
        return ""

print(Solution().firstPalindrome(["abc", "car", "ada", "racecar", "cool"]))
````

## 10. Vì sao thuật toán đúng?

- **Tính đúng của kiểm tra palindrome.** Chuỗi `s` đối xứng ⇔ ký tự thứ `i` từ đầu bằng ký tự thứ `i` từ cuối với mọi `i`. `s[::-1]` đặt ký tự thứ `i` từ cuối vào vị trí `i`, nên `s == s[::-1]` đúng bằng điều kiện đó. Cách hai con trỏ chỉ kiểm tra `i < n-1-i`: cặp `(i, n-1-i)` và `(n-1-i, i)` là một, ký tự giữa tự khớp chính nó.
- **Tính "đầu tiên".** Ta duyệt theo thứ tự chỉ số tăng và thoát khi gặp từ đầu tiên thỏa; mọi từ trước đó đã được kiểm và loại. Nếu duyệt hết mà không thoát, mọi từ đều không palindrome → trả `""`.

### Ví dụ chạy thử thêm: đối xứng lẻ và chẵn

Ta thử các độ dài khác nhau để chắc chắn hàm kiểm tra không phụ thuộc độ dài lẻ hay chẵn. Chuỗi rỗng (không xuất hiện trong đề vì mỗi từ dài ít nhất 1) cũng được coi là palindrome theo quy ước toán học, còn chuỗi một ký tự luôn là palindrome.

````python
tests = ["a", "aa", "ab", "aba", "abba", "abca", "abcba", "abccba", "abcdba"]
for t in tests:
    print(f"{t:8s} dài {len(t)} -> palindrome? {t == t[::-1]}")
````

## 11. Phân tích Time / Space Complexity

Gọi `n = len(words)`, `L` là độ dài từ lớn nhất.

| Cách | Time | Space |
|---|---|---|
| 1. Tự dựng chuỗi đảo | `O(n·L²)` | `O(L)` |
| 2. Slicing | `O(n·L)` | `O(L)` |
| 3. Hai con trỏ | `O(n·L)` | `O(1)` |
| 4. `next` + generator | `O(n·L)` | `O(L)` |

Worst case: không có palindrome nào, phải duyệt hết `n` từ. Best case: từ đầu đã là palindrome. Với giới hạn `100 × 100` mọi cách đều rất nhanh.

Không. Trong trường hợp tốt nhất (từ đầu tiên đã là palindrome) ta chỉ làm việc `O(L)`. Đây chính là lý do ta ưu tiên `return` sớm hay `next(...)` lười thay vì lọc cả danh sách bằng list comprehension rồi lấy phần tử `[0]` (cách đó luôn tốn `O(n·L)` và còn phải xử lý trường hợp danh sách rỗng, nếu không sẽ gặp `IndexError`).

````python
words = ["aba", "xyz", "pqr"]
filtered = [w for w in words if w == w[::-1]]      # luôn kiểm tra TẤT CẢ các từ
print("lọc hết:", filtered, "-> lấy [0]:", filtered[0])
print("dừng sớm:", first_pal_slice(words))
try:
    print([w for w in ["xyz"] if w == w[::-1]][0])
except IndexError as e:
    print("IndexError khi không có palindrome:", e)
````

## 12. Bộ test

Kiểm ví dụ đề, edge case (một từ một ký tự, mọi từ đều là palindrome, palindrome ở cuối, từ không đối xứng gần giống), và đối chiếu mọi cách với input ngẫu nhiên có `random.seed`.

````python
cases = [
    (["abc", "car", "ada", "racecar", "cool"], "ada"),
    (["notapalindrome", "racecar"], "racecar"),
    (["def", "ghi"], ""),
    (["a"], "a"),
    (["ab"], ""),
    (["aa", "b", "c"], "aa"),
    (["ab", "ba", "aba"], "aba"),
    (["abca", "abcba"], "abcba"),
    (["xyz", "zyx", "xyx", "zzz"], "xyx"),
    (["abba", "aba"], "abba"),
]
impls = [first_pal_manual, first_pal_slice, first_pal_two, first_pal_next,
         Solution().firstPalindrome, SolutionSlice().firstPalindrome,
         SolutionTwoPointers().firstPalindrome, SolutionNext().firstPalindrome]
for words, expected in cases:
    for f in impls:
        assert f(list(words)) == expected, (f.__name__, words)
print("Tất cả", len(cases), "ca x", len(impls), "cách đều đúng")
````

## 13. Các lỗi người mới hay gặp

### Lỗi 1: Trả về `None` (hoặc `-1`) khi không có

Quên dòng `return ""` cuối hàm → Python mặc định trả về `None`, LeetCode sẽ báo sai.

````python
def wrong1(words):
    for w in words:
        if w == w[::-1]:
            return w
    # thiếu return "" ở cuối

print("sai:", wrong1(["def", "ghi"]))
print("đúng:", repr(first_pal_slice(["def", "ghi"])))
````

### Lỗi 2: `return` ngay trong nhánh `else` của vòng lặp

Đặt `return ""` bên trong vòng lặp (thụt vào quá sâu) làm hàm thoát ở từ đầu tiên không đối xứng, bỏ qua các từ sau.

````python
def wrong2(words):
    for w in words:
        if w == w[::-1]:
            return w
        else:
            return ""      # SAI: thoát luôn ở từ không-palindrome đầu tiên

print("sai:", repr(wrong2(["abc", "ada"])), "| đúng:", repr(first_pal_slice(["abc", "ada"])))
````

### Lỗi 3: Lấy palindrome **cuối cùng** hoặc **dài nhất** thay vì đầu tiên

Nếu quên `return` sớm và ghi đè biến kết quả, bạn nhận từ palindrome *cuối* cùng.

````python
def wrong3(words):
    ans = ""
    for w in words:
        if w == w[::-1]:
            ans = w            # SAI: không dừng nên bị ghi đè
    return ans

words = ["abc", "ada", "racecar"]
print("sai:", wrong3(words), "| đúng:", first_pal_slice(words))
````

### Lỗi 4: So sánh `w == reversed(w)`

`reversed(w)` là iterator, không phải chuỗi, nên phép so sánh luôn `False`.

````python
w = "ada"
print(w == reversed(w))
print(w == "".join(reversed(w)))
````

## 14. Mẹo làm bài nhanh & nhận dạng pattern

- Chữ "đầu tiên" → duyệt xuôi và `return` sớm (hoặc `next(...)`).
- Kiểm tra palindrome: `s == s[::-1]` cho bài đơn giản; hai con trỏ khi cần `O(1)` bộ nhớ hoặc khi muốn "bỏ qua ký tự" (bài Valid Palindrome).
- Giá trị mặc định khi không tìm thấy phải đúng kiểu trả về (`""` cho chuỗi).
- Khi thấy "đối xứng", "đọc ngược" → nghĩ ngay tới `[::-1]` hoặc hai đầu tiến vào giữa.
- Giới hạn nhỏ (`100 × 100`) nên không cần tối ưu cực đoan; ưu tiên code rõ ràng.
- Khi tự test, luôn thêm ca "palindrome xuất hiện ở **nhiều** vị trí" (như `["abc","ada","racecar"]`) để chắc chắn bạn trả về cái đầu chứ không phải cái cuối.
- Độ dài lẻ hay chẵn không ảnh hưởng: với độ dài chẵn hai con trỏ gặp nhau sau cùng (`i > j`), với độ dài lẻ chúng gặp nhau tại ký tự giữa (`i == j`) — cả hai đều dừng vòng `while`.

## 15. Biến thể của bài

1. Trả về **chỉ số** của palindrome đầu tiên (hoặc `-1`).
2. Đếm số từ palindrome trong mảng.
3. Trả về palindrome **dài nhất** (ưu tiên từ đứng trước nếu hòa).
4. Bỏ qua chữ hoa/thường và ký tự không phải chữ (Valid Palindrome).
5. Cho phép xóa tối đa một ký tự (Valid Palindrome II).

````python
def first_pal_index(words):
    for i, w in enumerate(words):
        if w == w[::-1]:
            return i
    return -1

def count_pal(words):
    return sum(1 for w in words if w == w[::-1])

def longest_pal(words):
    best = ""
    for w in words:
        if w == w[::-1] and len(w) > len(best):
            best = w
    return best

ws = ["abc", "ada", "racecar", "cool", "xyzzyx"]
print(first_pal_index(ws), count_pal(ws), longest_pal(ws))
````

## 16. So sánh các cách làm

| Cách | Ý tưởng | Time | Space | Khi nào dùng |
|---|---|---|---|---|
| 1. Tự dựng chuỗi đảo | `rev = ch + rev` | `O(n·L²)` | `O(L)` | chỉ để hiểu bản chất |
| 2. Slicing `[::-1]` | so với chuỗi đảo | `O(n·L)` | `O(L)` | nộp bài, ngắn gọn |
| 3. Hai con trỏ | so cặp đối xứng | `O(n·L)` | `O(1)` | tiết kiệm bộ nhớ, dừng sớm |
| 4. `next` + generator | lười, một dòng | `O(n·L)` | `O(L)` | code ngắn nhất |

## 19. Đáp án tham khảo cho bài tập

Bài 1: `"xx"`. Bài 2: cặp `(l,l)` rồi `(e,e)`; ký tự giữa `v` không cần so.

````python
def exercise_is_pal(w):
    i, j = 0, len(w) - 1
    while i < j:
        if w[i] != w[j]:
            return False
        i += 1
        j -= 1
    return True

def exercise_first(words):
    return next((w for w in words if w == w[::-1]), "")

def exercise_last(words):
    for w in reversed(words):
        if w == w[::-1]:
            return w
    return ""

assert first_pal_slice(["ab", "ba", "xx", "yy"]) == "xx"
assert exercise_is_pal("level") and exercise_is_pal("noon") and exercise_is_pal("a")
assert not exercise_is_pal("ab")
assert exercise_first(["ab", "aba"]) == "aba"
assert exercise_first(["ab"]) == ""
assert exercise_last(["aa", "ab", "bb"]) == "bb"
assert exercise_last(["ab"]) == ""
print("Đáp án bài tập đều đúng")
````
