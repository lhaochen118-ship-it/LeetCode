## 1. Đề bài và cách hiểu đúng

Cho mảng chuỗi `words`. Trả về **chuỗi palindrome đầu tiên** (theo thứ tự trong mảng). Nếu không có, trả `""`.

Một chuỗi là **palindrome** nếu đọc xuôi và đọc ngược giống nhau: `"ada"`, `"racecar"`, `"a"`, `"abba"`.

| `words` | Kết quả | Giải thích |

Hai ý chính: (1) cách **kiểm tra một chuỗi** có palindrome không; (2) **duyệt theo thứ tự** và trả về **ngay** khi gặp.

### 2.4. Vòng `while` với hai con trỏ

`left` bắt đầu từ đầu, `right` từ cuối, tiến vào giữa. Điều kiện `left < right`: khi hai con trỏ gặp nhau hoặc vượt qua nhau thì đã so hết các cặp.

````python
s = "abcba"
left, right = 0, len(s) - 1
while left < right:
    print(f"so s[{left}]={s[left]!r} với s[{right}]={s[right]!r}")
    left += 1
    right -= 1
print("dừng: left =", left, "right =", right)
````

### 2.5. `return` bên trong vòng lặp — dừng sớm

`return` kết thúc hàm **ngay lập tức**, vòng lặp không chạy tiếp. Đây chính là cách lấy phần tử "đầu tiên" thỏa điều kiện.

````python
def first_even(nums):
    for x in nums:
        print("  xét", x)
        if x % 2 == 0:
            return x
    return None

print("kết quả:", first_even([3, 5, 8, 10, 12]))
````

### 2.6. `next(generator, default)`

`(w for w in words if cond(w))` là **generator**: tạo phần tử dần dần khi được hỏi. `next(gen, default)` lấy phần tử **đầu tiên**, hết thì trả `default`. Không tạo cả danh sách → cũng dừng sớm.

````python
gen = (x for x in [3, 5, 8, 10] if x % 2 == 0)
print(next(gen, None))    # 8
print(next(gen, None))    # 10
print(next(gen, None))    # 12? không có -> None
print(next(gen, "hết"))
````

## 4. Nút thắt của cách gom tất cả

1. **Làm thừa:** nếu chuỗi đầu tiên đã là palindrome, ta vẫn kiểm tra 99 chuỗi còn lại.
2. **Tốn bộ nhớ:** list `pals` và mỗi `w[::-1]` tạo một chuỗi mới dài `len(w)`.

Hai cải tiến độc lập:
- **Dừng sớm:** trả về ngay khi gặp palindrome đầu tiên.
- **Kiểm tra không tạo chuỗi mới:** hai con trỏ so `w[left]` với `w[right]`, và **dừng ngay ở cặp sai đầu tiên** (ví dụ `"abcdefg...z"` sai ngay ở bước 1, trong khi `w[::-1]` phải tạo hết chuỗi đảo).

````python
calls = 0
def is_pal_count(w):
    global calls
    calls += 1
    return w == w[::-1]

words = ["aba"] + ["xyz"] * 99
calls = 0; _ = [w for w in words if is_pal_count(w)]
print("Gom tất cả: số lần kiểm tra =", calls)
calls = 0
for w in words:
    if is_pal_count(w):
        break
print("Dừng sớm  : số lần kiểm tra =", calls)
````

## 5. Bước đột phá — Hai con trỏ: so từng cặp đối xứng, sai là dừng

Chuỗi `w` là palindrome ⇔ với mọi `i`, `w[i] == w[len(w) - 1 - i]`. Chỉ cần kiểm tra **nửa đầu** (vì cặp `(i, n-1-i)` và `(n-1-i, i)` là một).

- Nếu gặp một cặp khác nhau → **không** palindrome, dừng ngay.
- Nếu hai con trỏ gặp nhau mà chưa thấy cặp sai → palindrome.
- Với độ dài lẻ, ký tự giữa không cần so (nó so với chính nó).

````python
w = "racecar"
n = len(w)
for i in range(n // 2):
    print(f"cặp ({i}, {n-1-i}): {w[i]!r} vs {w[n-1-i]!r} -> {'khớp' if w[i] == w[n-1-i] else 'KHÁC'}")
print("ký tự giữa (không cần so):", w[n // 2])
````

## 6. Lời giải chính — mổ xẻ rồi ghép

### 6.1. Đoạn 1: hàm kiểm tra palindrome bằng hai con trỏ

````python
def is_palindrome(w: str) -> bool:
    left, right = 0, len(w) - 1
    while left < right:
        if w[left] != w[right]:
            return False          # cặp sai đầu tiên -> dừng
        left += 1
        right -= 1
    return True

for w in ["ada", "abba", "a", "ab", "abca"]:
    print(f"{w!r:7} -> {is_palindrome(w)}")
````

### 6.2. Đoạn 2: duyệt theo thứ tự, trả về ngay

````python
def first_palindrome(words: list[str]) -> str:
    for w in words:
        if is_palindrome(w):
            return w
    return ""                     # chỉ chạy tới đây khi không có palindrome nào

for w, exp in examples:
    print(repr(first_palindrome(w)), "kỳ vọng", repr(exp))
````

### 6.3. Tại sao `return ""` nằm **ngoài** vòng for?

Nếu đặt `else: return ""` bên trong vòng lặp, hàm sẽ trả `""` ngay khi chuỗi **đầu tiên** không phải palindrome — sai (xem Lỗi 1 ở mục 12). Chỉ khi đã duyệt **hết** mà không tìm thấy mới được kết luận "không có".

## 7. TRACE từng bước

### 7.1. Bảng trace — `words = ["abc", "car", "ada", "racecar", "cool"]`

| Chuỗi | left, right | So sánh | Kết luận |
|---|---|---|---|
| `"abc"` | 0, 2 | `a` vs `c` → khác | không phải, xét tiếp |
| `"car"` | 0, 2 | `c` vs `r` → khác | không phải, xét tiếp |
| `"ada"` | 0, 2 | `a` vs `a` → khớp | |
| | 1, 1 | `left < right` sai → dừng | **palindrome → return "ada"** |
| `"racecar"` | — | không được xét | — |

````python
def trace_first_palindrome(words):
    for idx, w in enumerate(words):
        print(f"[{idx}] xét {w!r}")
        left, right = 0, len(w) - 1
        ok = True
        while left < right:
            same = w[left] == w[right]
            print(f"     left={left}, right={right}: {w[left]!r} vs {w[right]!r} -> {'khớp' if same else 'KHÁC'}")
            if not same:
                ok = False
                break
            left += 1; right -= 1
        if ok:
            print(f"     => palindrome, trả về {w!r} (không xét các chuỗi sau)")
            return w
        print("     => không phải, xét chuỗi tiếp theo")
    print("Hết mảng, trả về ''")
    return ""

trace_first_palindrome(["abc", "car", "ada", "racecar", "cool"])
````

## 8. Code nộp LeetCode (Python 3) — nhiều cách

### Cách A — Gom tất cả palindrome rồi lấy cái đầu

````python
class Solution:
    def firstPalindrome(self, words: list[str]) -> str:
        pals = [w for w in words if w == w[::-1]]
        return pals[0] if pals else ""

print([Solution().firstPalindrome(w) for w, _ in examples])
````

### Cách B — Duyệt + `w[::-1]` + dừng sớm (ngắn, phổ biến nhất)

````python
class Solution:
    def firstPalindrome(self, words: list[str]) -> str:
        for w in words:
            if w == w[::-1]:
                return w
        return ""

print([Solution().firstPalindrome(w) for w, _ in examples])
````

### Cách C — Hai con trỏ (lời giải chính, `O(1)` bộ nhớ phụ)

````python
class Solution:
    def firstPalindrome(self, words: list[str]) -> str:
        for w in words:
            left, right = 0, len(w) - 1
            while left < right and w[left] == w[right]:
                left += 1
                right -= 1
            if left >= right:          # không gặp cặp sai nào
                return w
        return ""

print([Solution().firstPalindrome(w) for w, _ in examples])
````

Chú ý cách viết gộp: `while left < right and w[left] == w[right]` dừng khi **hoặc** đã gặp nhau (palindrome) **hoặc** gặp cặp sai. Sau vòng lặp, `left >= right` nghĩa là trường hợp thứ nhất.

### Cách D — `next` + generator + `all` (kiểu Pythonic)

`all(w[i] == w[-1 - i] for i in range(len(w) // 2))` cũng dừng ở cặp sai đầu tiên (short-circuit).

````python
class Solution:
    def firstPalindrome(self, words: list[str]) -> str:
        return next((w for w in words
                     if all(w[i] == w[-1 - i] for i in range(len(w) // 2))), "")

print([Solution().firstPalindrome(w) for w, _ in examples])
````

### Cách E — Đệ quy kiểm tra palindrome (minh họa tư duy)

`w` palindrome ⇔ ký tự đầu = ký tự cuối **và** phần giữa `w[1:-1]` palindrome. Điểm dừng: độ dài ≤ 1. (Đệ quy tạo nhiều chuỗi con nên không tối ưu, nhưng giúp hiểu cấu trúc bài toán.)

````python
class Solution:
    def firstPalindrome(self, words: list[str]) -> str:
        def is_pal(s: str) -> bool:
            if len(s) <= 1:
                return True
            return s[0] == s[-1] and is_pal(s[1:-1])
        for w in words:
            if is_pal(w):
                return w
        return ""

print([Solution().firstPalindrome(w) for w, _ in examples])
````

## 9. Vì sao thuật toán đúng?

**Hai con trỏ — invariant:** trước mỗi lần kiểm tra điều kiện `while`, mọi cặp `(i, n-1-i)` với `i < left` đều đã khớp. Đồng thời luôn có `left + right == n - 1`.
- Nếu gặp cặp sai → tồn tại `i` với `w[i] != w[n-1-i]` → chắc chắn không phải palindrome.
- Nếu `left >= right` → mọi cặp của nửa đầu đã khớp; nửa sau là cùng các cặp đó; ký tự giữa (nếu có) so với chính nó → palindrome.

**"Đầu tiên":** ta duyệt `words` theo thứ tự chỉ số tăng dần và `return` ở chuỗi palindrome **đầu tiên gặp**. Mọi chuỗi trước nó đã được kiểm tra là không palindrome ⇒ đúng là chuỗi đầu tiên. Nếu duyệt hết không trả về ⇒ không có palindrome ⇒ `""`.

## 10. Phân tích Time / Space Complexity

Gọi `n = len(words)`, `L` = độ dài tối đa một chuỗi, `S = tổng độ dài` (≤ `n·L`).

| Cách | Time (tệ nhất) | Space phụ | Ghi chú |
|---|---|---|---|
| A. Gom tất cả | `Θ(S)` luôn luôn | `O(L + n)` | không dừng sớm |
| B. `w[::-1]` + dừng sớm | `O(S)` | `O(L)` chuỗi đảo | ngắn gọn |
| C. Hai con trỏ | `O(S)` | `O(1)` | dừng ở cặp sai đầu tiên |
| D. `next` + `all` | `O(S)` | `O(1)` | Pythonic |
| E. Đệ quy | `O(Σ len²)` | `O(L²)` | do cắt chuỗi con mỗi tầng |

Trong thực tế, `w[::-1]` chạy bằng C nên rất nhanh; hai con trỏ thắng về **bộ nhớ** và khi chuỗi sai sớm.

````python
import time
long_bad = ["a" + "b" * 100000 + "c"] * 50
for name, f in [("đảo chuỗi", lambda ws: [w == w[::-1] for w in ws]),
                ("hai con trỏ", lambda ws: [is_palindrome(w) for w in ws])]:
    t = time.perf_counter(); f(long_bad); dt = time.perf_counter() - t
    print(f"{name:12}: {dt*1000:.2f} ms  (chuỗi sai ngay ở cặp đầu)")
````

## 11. Kiểm thử: mẫu đề, biên, phản ví dụ, đối chiếu

````python
import random

def sol_A(ws): return first_palindrome_collect(ws)
def sol_B(ws): return next((w for w in ws if w == w[::-1]), "")
def sol_C(ws): return first_palindrome(ws)
def sol_D(ws): return next((w for w in ws if all(w[i] == w[-1 - i] for i in range(len(w) // 2))), "")
def sol_E(ws):
    def p(s): return len(s) <= 1 or (s[0] == s[-1] and p(s[1:-1]))
    return next((w for w in ws if p(w)), "")

SOLS = [sol_A, sol_B, sol_C, sol_D, sol_E]

def check(name, actual, expected):
    assert actual == expected, (name, actual, expected)
    print(f"{name:26} PASS ({actual!r})")

tests = examples + [
    (["a"], "a"),                       # một ký tự luôn palindrome
    (["ab"], ""),
    (["ab", "cc", "aa"], "cc"),         # độ dài chẵn
    (["abcba", "x"], "abcba"),          # palindrome ngay đầu
    (["abca", "abcda", "zz"], "zz"),    # gần đúng nhưng sai ở giữa
    (["racecar", "racecar"], "racecar"),
]
for i, (ws, exp) in enumerate(tests):
    for f in SOLS:
        check(f"test {i} {f.__name__}", f(ws), exp)
````

````python
rng = random.Random(2108)
for _ in range(500):
    ws = ["".join(rng.choice("ab") for _ in range(rng.randint(1, 5))) for _ in range(rng.randint(1, 6))]
    outs = [f(ws) for f in SOLS]
    assert len(set(outs)) == 1, (ws, outs)
print("500 test ngẫu nhiên: 5 cách khớp nhau")
````

## 12. Lỗi người mới hay gặp — code sai, vì sao sai, cách sửa

### Lỗi 1: `return ""` đặt trong vòng lặp

````python
def wrong_early_empty(words):
    for w in words:
        if w == w[::-1]:
            return w
        else:
            return ""          # SAI: kết luận quá sớm

print("Sai :", repr(wrong_early_empty(["abc", "ada"])), "| Đúng:", repr(first_palindrome(["abc", "ada"])))
````

**Sửa:** đưa `return ""` ra **sau** vòng for.

### Lỗi 2: Trả palindrome **cuối cùng** (ghi đè biến kết quả)

````python
def wrong_last(words):
    ans = ""
    for w in words:
        if w == w[::-1]:
            ans = w            # SAI: bị ghi đè bởi palindrome sau
    return ans

print("Sai :", repr(wrong_last(["abc", "ada", "racecar"])), "| Đúng:", repr(first_palindrome(["abc", "ada", "racecar"])))
````

**Sửa:** `return` ngay, hoặc `break` sau khi gán lần đầu.

### Lỗi 3: Hai con trỏ dùng `while left <= right` và `right = len(w)` (vượt chỉ số)

````python
def wrong_index(w):
    left, right = 0, len(w)          # SAI: phải là len(w) - 1
    while left < right:
        if w[left] != w[right]:
            return False
        left += 1; right -= 1
    return True

try:
    wrong_index("aba")
except IndexError as e:
    print("IndexError:", e, "-> sửa: right = len(w) - 1")
````

## 14. Biến thể

### 14.1. Trả về **chỉ số** của palindrome đầu tiên (hoặc -1)

````python
def first_palindrome_index(words):
    for i, w in enumerate(words):
        if is_palindrome(w):
            return i
    return -1

print(first_palindrome_index(["abc", "car", "ada"]), first_palindrome_index(["ab"]))
````

### 14.2. Palindrome **dài nhất** trong mảng (nếu bằng nhau lấy cái xuất hiện trước)

````python
def longest_palindrome_word(words):
    best = ""
    for w in words:
        if len(w) > len(best) and is_palindrome(w):
            best = w
    return best

print(repr(longest_palindrome_word(["ada", "racecar", "abba", "xyzzyxq"])))
````

### 14.3. Bỏ qua ký tự không phải chữ/số và không phân biệt hoa thường (kiểu LeetCode 125)

````python
def is_palindrome_clean(s):
    left, right = 0, len(s) - 1
    while left < right:
        if not s[left].isalnum():
            left += 1
        elif not s[right].isalnum():
            right -= 1
        elif s[left].lower() != s[right].lower():
            return False
        else:
            left += 1; right -= 1
    return True

print(is_palindrome_clean("A man, a plan, a canal: Panama"), is_palindrome_clean("race a car"))
````

## 17. Đáp án tham khảo

````python
def count_palindromes(words):
    return sum(1 for w in words if w == w[::-1])

def is_pal_two_pointers(w):
    left, right = 0, len(w) - 1
    while left < right:
        if w[left] != w[right]:
            return False
        left += 1; right -= 1
    return True

def first_palindrome_ex(words):
    for w in words:
        if is_pal_two_pointers(w):
            return w
    return ""

def last_palindrome(words):
    for w in reversed(words):
        if is_pal_two_pointers(w):
            return w
    return ""

my_tests = [["ab", "aa"], ["abc", "ada"], ["xy", "z"]]

def valid_palindrome_ii(s):
    def pal_range(l, r):
        while l < r:
            if s[l] != s[r]:
                return False
            l += 1; r -= 1
        return True
    l, r = 0, len(s) - 1
    while l < r:
        if s[l] != s[r]:
            return pal_range(l + 1, r) or pal_range(l, r - 1)
        l += 1; r -= 1
    return True

assert count_palindromes(["abc", "ada", "racecar", "x"]) == 3
assert is_pal_two_pointers("abba") and not is_pal_two_pointers("abca")
assert first_palindrome_ex(["abc", "car", "ada", "racecar", "cool"]) == "ada"
assert first_palindrome_ex(["def", "ghi"]) == ""
assert last_palindrome(["abc", "ada", "racecar", "cool"]) == "racecar"
for ws in my_tests:
    assert wrong_early_empty(ws) != first_palindrome(ws)
assert valid_palindrome_ii("abca") and valid_palindrome_ii("aba") and not valid_palindrome_ii("abc")
print("Tất cả đáp án PASS")
````

**Giải thích bằng lời:**
- Bài 1–3: kết hợp hàm kiểm tra + duyệt theo thứ tự + dừng sớm.
- Bài 4: duyệt ngược bằng `reversed(words)` biến "cuối cùng" thành "đầu tiên".
- Bài 5: Lỗi 1 chỉ lộ khi chuỗi **đầu** không palindrome nhưng có palindrome **phía sau**.
- Bài 6: ở cặp sai đầu tiên, chỉ có hai lựa chọn xóa (trái hoặc phải); phần trước đã khớp nên chỉ cần kiểm tra đoạn còn lại → `O(n)`.
