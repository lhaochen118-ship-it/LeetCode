## 1. Mục tiêu học

Sau notebook này bạn sẽ:

1. Hiểu khái niệm **câu vòng tròn**: ký tự cuối của mỗi từ phải trùng ký tự đầu của từ kế tiếp, và từ cuối nối ngược về từ đầu.
2. Biết cắt chuỗi thành từ bằng `split()`, truy cập ký tự đầu `w[0]` và ký tự cuối `w[-1]`.
3. Hiểu cách **chỉ số âm** `-1` và phép `%` giúp xử lý "nối vòng" mà không cần viết nhánh riêng cho từ cuối.
4. Viết 4 cách: duyệt từng cặp từ kề nhau, quét các dấu cách trên chuỗi gốc, so sánh danh sách ký tự đầu/cuối xoay vòng, và dùng `zip` với `all`; đối chiếu bằng test ngẫu nhiên.
5. Tránh các lỗi: quên kiểm tra cặp (từ cuối, từ đầu), so sánh không phân biệt hoa thường, và truy cập `s[i + 1]` vượt chỉ số.

## 2. Đề bài & ví dụ

Một **câu** là danh sách các từ cách nhau bằng **đúng một dấu cách**, không có dấu cách ở đầu hoặc cuối. Mỗi từ chỉ gồm chữ cái hoa và thường (hoa và thường là **khác nhau**).

Câu gọi là **vòng tròn** nếu:

- Ký tự **cuối** của mỗi từ bằng ký tự **đầu** của từ **kế tiếp**.
- Ký tự **cuối** của **từ cuối cùng** bằng ký tự **đầu** của **từ đầu tiên**.

Cho `sentence`, trả về `True` nếu câu là vòng tròn, ngược lại `False`.

| Ví dụ | sentence | Output | Giải thích |
|---|---|---|---|
| 1 | `"leetcode exercises sound delightful"` | `True` | `e-e`, `s-s`, `d-d`, và `l` (cuối) nối về `l` (đầu) |
| 2 | `"eetcode"` | `True` | chỉ một từ: ký tự cuối `e` bằng ký tự đầu `e` |
| 3 | `"Leetcode is cool"` | `False` | `e` cuối của `Leetcode` khác `i` đầu của `is` |

Constraints: `1 <= sentence.length <= 500`; `sentence` chỉ gồm chữ cái hoa/thường và dấu cách; các từ cách nhau bằng đúng một dấu cách; không có dấu cách ở đầu/cuối.

````python
examples = [("leetcode exercises sound delightful", True), ("eetcode", True), ("Leetcode is cool", False)]
for s, expected in examples:
    words = s.split()
    pairs = [(words[i][-1], words[(i + 1) % len(words)][0]) for i in range(len(words))]
    ok = all(a == b for a, b in pairs)
    print(f"{s!r}: các cặp (cuối, đầu kế tiếp) = {pairs} -> {ok} (mong đợi {expected})")
````

## 3. Kiến thức nền

### 3.1. `split()` và chỉ số ký tự

`s.split()` cắt chuỗi theo khoảng trắng thành list các từ. Với từ `w`: `w[0]` là ký tự đầu, `w[-1]` là ký tự cuối (chỉ số âm đếm từ cuối).

````python
s = "leetcode exercises sound delightful"
words = s.split()
print(words)
for w in words:
    print(w, "-> đầu:", w[0], "| cuối:", w[-1])
````

### 3.2. Chỉ số âm và phép `%` để "nối vòng"

Với danh sách `words` có `m` từ, từ kế tiếp của từ `i` là `words[(i + 1) % m]`. Khi `i = m - 1`, `(i + 1) % m = 0` tức là quay về từ đầu. Chỉ số `-1` cũng cho phần tử cuối: `words[i - 1]` với `i = 0` chính là từ cuối.

````python
words = ["a", "b", "c"]
m = len(words)
for i in range(m):
    print(i, "-> kế tiếp:", words[(i + 1) % m], "| trước đó:", words[i - 1])
````

### 3.3. `all()` và `zip()`

`all(iterable)` trả `True` nếu mọi phần tử đúng (và `True` với dãy rỗng). `zip(a, b)` ghép hai dãy theo vị trí, dừng ở dãy ngắn hơn. Kết hợp lại: `all(x == y for x, y in zip(a, b))`.

````python
a = [1, 2, 3]
b = [1, 2, 4]
print(list(zip(a, b)))
print(all(x == y for x, y in zip(a, b)))
print(all(x == y for x, y in zip(a[:2], b[:2])))
print(all([]))
````

### 3.4. Duyệt chuỗi theo chỉ số và tìm dấu cách

`for i, ch in enumerate(s)` cho cả chỉ số lẫn ký tự. Một dấu cách ở vị trí `i` luôn nằm **giữa** hai từ, nên `s[i - 1]` là ký tự cuối từ trước và `s[i + 1]` là ký tự đầu từ sau (vì đề đảm bảo không có dấu cách ở đầu/cuối và không có hai dấu cách liên tiếp).

````python
s = "ab ba"
for i, ch in enumerate(s):
    if ch == " ":
        print("dấu cách ở vị trí", i, "| ký tự trước:", s[i - 1], "| ký tự sau:", s[i + 1])
print(s[0], s[-1])
````

## 4. Ý tưởng nền tảng / cách nghĩ trực giác

Hãy tưởng tượng các từ là các toa tàu xếp thành vòng tròn: đuôi toa này phải khớp với đầu toa sau.

Tính tay `"Leetcode is cool"`:

- `Leetcode` kết thúc bằng `e`, từ kế tiếp `is` bắt đầu bằng `i`. `e != i` ngay từ cặp đầu nên kết luận `False`.

Tính tay `"abc cda"`:

- `abc` kết thúc `c`, `cda` bắt đầu `c`: khớp.
- `cda` kết thúc `a`, nối vòng về từ đầu `abc` bắt đầu `a`: khớp. Kết quả `True`.

Tính tay `"ab bc"`: cặp 1 khớp (`b`-`b`) nhưng cặp vòng `c` (cuối `bc`) so với `a` (đầu `ab`) khác nhau nên `False`. Nếu quên cặp vòng sẽ nhầm thành `True`.

**Nhận xét:** nếu có `m` từ thì phải kiểm tra **đúng `m` cặp** (`m - 1` cặp liền kề + 1 cặp nối vòng). Với `m = 1` chỉ có đúng một cặp: (cuối từ, đầu từ).

````python
for s in ("abc cda", "ab bc", "a"):
    w = s.split()
    m = len(w)
    print(f"{s!r}: {m} từ -> {m} cặp:", [(w[i][-1], w[(i + 1) % m][0]) for i in range(m)])
````

## 5. Cách 1 — Brute Force / baseline: duyệt từng cặp từ

**Ý tưởng:** tách thành từ, lặp `i` từ `0` đến `m - 1`, lấy `w = words[i]` và từ kế tiếp `nxt = words[(i + 1) % m]`; nếu `w[-1] != nxt[0]` thì trả `False`. Hết vòng lặp mà không lỗi thì `True`.

````python
class SolutionBrute:
    def isCircularSentence(self, sentence: str) -> bool:
        words = sentence.split()
        m = len(words)
        for i in range(m):
            cur = words[i]
            nxt = words[(i + 1) % m]          # % m để nối vòng về từ đầu
            if cur[-1] != nxt[0]:
                return False
        return True
````

````python
sol = SolutionBrute()
print(sol.isCircularSentence("leetcode exercises sound delightful"))   # True
print(sol.isCircularSentence("eetcode"))                               # True
print(sol.isCircularSentence("Leetcode is cool"))                      # False
````

**Độ phức tạp:** Time `O(L)` với `L = len(sentence)` (`split` đọc cả chuỗi, rồi mỗi từ được xét `O(1)`), Space `O(L)` cho list các từ. **Nút thắt:** ta tạo list từ chỉ để lấy hai ký tự mỗi từ; có thể làm trực tiếp trên chuỗi gốc.

## 6. Ý tưởng thuật toán / bước đột phá

Mỗi từ có hai "đầu mối": ký tự đầu và ký tự cuối. Trên chuỗi gốc, **ranh giới giữa hai từ chính là dấu cách**: ký tự ngay trước dấu cách là cuối từ trước, ký tự ngay sau là đầu từ sau. Do đó:

- Với mỗi dấu cách tại `i`: yêu cầu `s[i - 1] == s[i + 1]`.
- Cặp nối vòng là `s[-1] == s[0]` (cuối câu và đầu câu).

Câu có `m` từ có `m - 1` dấu cách nên tổng cộng `m` điều kiện, đúng như số cặp ở mục 4. Không cần `split`, không cần list phụ: Space `O(1)`.

**Quy tắc:** *nếu có một dấu cách mà hai ký tự hai bên khác nhau, hoặc `s[0] != s[-1]`, thì `False`; ngược lại `True`.*

**Invariant:** sau khi xét xong các dấu cách ở vị trí `<= i`, mọi ranh giới từ trong đoạn `s[:i + 1]` đã khớp.

## 7. Các cách tiếp theo

### Cách 2 — Quét dấu cách trên chuỗi gốc

Chỉ dùng vòng lặp trên `sentence`, không tạo list từ.

````python
class SolutionScan:
    def isCircularSentence(self, sentence: str) -> bool:
        if sentence[0] != sentence[-1]:
            return False                      # cặp nối vòng
        for i in range(1, len(sentence) - 1):
            if sentence[i] == " " and sentence[i - 1] != sentence[i + 1]:
                return False
        return True
````

````python
sol = SolutionScan()
print(sol.isCircularSentence("leetcode exercises sound delightful"))   # True
print(sol.isCircularSentence("eetcode"))                               # True
print(sol.isCircularSentence("Leetcode is cool"))                      # False
````

Vòng lặp chạy `i` từ `1` đến `len - 2` vì đề đảm bảo dấu cách không ở đầu/cuối, nên `i - 1` và `i + 1` luôn hợp lệ (không vượt chỉ số).

**Độ phức tạp:** Time `O(L)`, Space `O(1)`.

### Cách 3 — So sánh danh sách ký tự cuối với danh sách ký tự đầu xoay vòng

Gọi `ends = [w[-1] for w in words]` và `starts = [w[0] for w in words]`. Câu vòng tròn khi `ends[i] == starts[i + 1]` với mọi `i` (vòng quanh), tức `ends == starts[1:] + starts[:1]`.

````python
class SolutionRotate:
    def isCircularSentence(self, sentence: str) -> bool:
        words = sentence.split()
        starts = [w[0] for w in words]
        ends = [w[-1] for w in words]
        return ends == starts[1:] + starts[:1]       # xoay `starts` sang trái 1 vị trí
````

````python
sol = SolutionRotate()
print(sol.isCircularSentence("leetcode exercises sound delightful"))   # True
print(sol.isCircularSentence("eetcode"))                               # True
print(sol.isCircularSentence("Leetcode is cool"))                      # False
````

**Độ phức tạp:** Time `O(L)`, Space `O(m)` với `m` là số từ (hai list ký tự).

### Cách 4 — `all` + `zip` trên cặp (từ, từ kế tiếp)

Ghép `words` với `words[1:] + words[:1]` (xoay trái một từ), rồi kiểm tra từng cặp bằng `all`.

````python
class SolutionZip:
    def isCircularSentence(self, sentence: str) -> bool:
        words = sentence.split()
        nxt = words[1:] + words[:1]
        return all(a[-1] == b[0] for a, b in zip(words, nxt))
````

````python
sol = SolutionZip()
print(sol.isCircularSentence("leetcode exercises sound delightful"))   # True
print(sol.isCircularSentence("eetcode"))                               # True
print(sol.isCircularSentence("Leetcode is cool"))                      # False
````

**Độ phức tạp:** Time `O(L)`, Space `O(L)` (list từ và list đã xoay). Gọn và dễ đọc nhất khi không cần tối ưu bộ nhớ.

## 8. TRACE từng bước

Trace Cách 2 trên `s = "abc cda"` (độ dài 7):

| Bước | Điều kiện | Kết quả |
|---|---|---|
| 0 | `s[0]='a'`, `s[-1]='a'` | bằng nhau, tiếp tục |
| i=1..2 | `s[1]='b'`, `s[2]='c'` | không phải dấu cách |
| i=3 | `s[3]=' '`, `s[2]='c'`, `s[4]='c'` | khớp |
| i=4..5 | `c`, `d` | không phải dấu cách |

Kết quả `True`. Trace trên `"ab bc"` dừng ngay ở bước 0 với `s[0]='a'`, `s[-1]='c'` nên trả `False`.

````python
def trace_scan(s):
    print("sentence =", repr(s))
    print(f"  cặp nối vòng: s[0]={s[0]!r}, s[-1]={s[-1]!r}")
    if s[0] != s[-1]:
        print("  khác nhau -> False")
        return False
    for i in range(1, len(s) - 1):
        if s[i] == " ":
            print(f"  i={i}: dấu cách, trước={s[i-1]!r}, sau={s[i+1]!r}", end=" ")
            if s[i - 1] != s[i + 1]:
                print("-> khác nhau, False")
                return False
            print("-> khớp")
    print("  kết quả: True")
    return True

trace_scan("abc cda")
trace_scan("ab bc")
trace_scan("Leetcode is cool")
trace_scan("eetcode")
````

## 9. Code nộp LeetCode

Cách quét dấu cách (`O(L)` thời gian, `O(1)` bộ nhớ):

````python
class Solution:
    def isCircularSentence(self, sentence: str) -> bool:
        if sentence[0] != sentence[-1]:
            return False
        for i in range(1, len(sentence) - 1):
            if sentence[i] == " " and sentence[i - 1] != sentence[i + 1]:
                return False
        return True
````

````python
print(Solution().isCircularSentence("leetcode exercises sound delightful"))
print(Solution().isCircularSentence("eetcode"))
print(Solution().isCircularSentence("Leetcode is cool"))
````

## 10. Vì sao thuật toán đúng

1. **Ranh giới từ = dấu cách.** Đề đảm bảo các từ cách nhau đúng một dấu cách và không có dấu cách thừa ở hai đầu. Vì vậy mỗi dấu cách ngăn đúng hai từ liên tiếp: `s[i - 1]` là ký tự cuối từ trước, `s[i + 1]` là ký tự đầu từ sau.
2. **Đủ điều kiện kề nhau.** Điều kiện "cuối từ `k` bằng đầu từ `k + 1`" tương ứng đúng với dấu cách giữa hai từ đó; xét mọi dấu cách là xét mọi cặp kề nhau.
3. **Điều kiện nối vòng.** Ký tự cuối của từ cuối chính là `s[-1]`, ký tự đầu của từ đầu chính là `s[0]`. Kiểm tra `s[0] == s[-1]` bao phủ điều kiện còn lại.
4. **Từ duy nhất.** Khi chỉ có một từ, không có dấu cách; điều kiện duy nhất là `s[0] == s[-1]`, và hàm xét đúng điều kiện đó.
5. **Không vượt chỉ số.** Dấu cách nếu có luôn ở vị trí `1 <= i <= len - 2`, nên `s[i - 1]`, `s[i + 1]` hợp lệ; vòng lặp từ `1` đến `len - 2` chứa mọi dấu cách.
6. **Phân biệt hoa/thường.** So sánh `!=` giữa hai ký tự Python là so sánh mã ký tự nên `'A' != 'a'`, đúng với yêu cầu của đề.

## 12. Bộ test

### 12.1. Ví dụ của đề, edge case và phản ví dụ

Gồm: một ký tự, một từ không khớp đầu/cuối, hai từ khớp kề nhưng không khớp vòng, phân biệt hoa/thường.

````python
sols = [SolutionBrute(), SolutionScan(), SolutionRotate(), SolutionZip(), Solution()]
# Mỗi kỳ vọng đều được tính tay (ghi chú bên cạnh)
cases = [
    ("leetcode exercises sound delightful", True),     # ví dụ 1
    ("eetcode", True),                                 # ví dụ 2
    ("Leetcode is cool", False),                       # ví dụ 3
    ("a", True),                                       # một ký tự
    ("ab", False),                                     # một từ, đầu a khác cuối b
    ("ab bc", False),                                  # b-b khớp nhưng c != a ở chỗ nối vòng
    ("ab ba", True),                                   # b-b, a-a
    ("abc cda", True),                                 # c-c, a-a
    ("Aa", False),                                     # hoa khác thường
    ("a A", False),                                    # cuối 'a' != đầu 'A'
    ("Ab bA", True),                                   # b-b, A-A
    ("aba abc cba", True),                             # a-a, c-c, a-a
    ("aba abc cbd", False),                            # nối vòng: d != a
    ("ab ba ab", False),                               # b-b, a-a, nhưng cuối b != đầu a
    ("ab bc ca", True),                                # b-b, c-c, a-a
]
for s, want in cases:
    for sol in sols:
        got = sol.isCircularSentence(s)
        assert got == want, (type(sol).__name__, s, got, want)
print("tất cả test cố định đều đúng với cả 5 cách:", len(cases))
````

### 12.2. Đối chiếu ngẫu nhiên (có seed)

Sinh câu **đúng ràng buộc**: số từ 1..20, mỗi từ có độ dài 1..8 chỉ gồm chữ hoa/thường, các từ cách nhau đúng một dấu cách; độ dài tổng `<= 500`. Một nửa số câu được xây dựng để **là vòng tròn** (chọn ký tự đầu của từ kế tiếp = ký tự cuối của từ trước), nửa còn lại hoàn toàn ngẫu nhiên.

````python
import random
import string
random.seed(2490)

LET = string.ascii_letters       # chữ hoa và thường

def rand_word(first=None, last=None):
    n = random.randint(1, 8)
    w = [random.choice(LET) for _ in range(n)]
    if first is not None:
        w[0] = first
    if last is not None:
        w[-1] = last
    return "".join(w)

def gen_circular():
    m = random.randint(1, 20)
    chars = [random.choice(LET) for _ in range(m)]            # chars[i] = ký tự đầu của từ i
    words = []
    for i in range(m):
        first = chars[i]
        last = chars[(i + 1) % m]                              # cuối từ i = đầu từ i+1
        if m == 1:
            first = last = chars[0]
        w = rand_word(first, last)
        if len(w) == 1 and first != last:
            w = first + last                                    # từ 1 ký tự cần đầu = cuối
        words.append(w)
    return " ".join(words)

def gen_random():
    m = random.randint(1, 20)
    return " ".join(rand_word() for _ in range(m))

def check_constraints(s):
    assert 1 <= len(s) <= 500
    assert s == s.strip() and "  " not in s
    assert all(ch.isalpha() or ch == " " for ch in s)

cnt_true = 0
for t in range(4000):
    s = gen_circular() if t % 2 == 0 else gen_random()
    check_constraints(s)
    ans = [sol.isCircularSentence(s) for sol in sols]
    assert len(set(ans)) == 1, (s, ans)
    cnt_true += ans[0]
print("4000 test ngẫu nhiên: 5 cách khớp nhau; số câu vòng tròn:", cnt_true)
assert cnt_true > 1500
````

Nhận xét: ở `gen_circular`, từ có đúng 1 ký tự thì phải có đầu = cuối nên với trường hợp lệch ta nối thành từ 2 ký tự `first + last`. Cách sinh này giúp bộ test có đủ cả câu `True` lẫn `False`, nếu không câu ngẫu nhiên gần như luôn `False`.

### 12.3. Kiểm chứng độc lập bằng định nghĩa

Viết một "oracle" mô tả đúng định nghĩa bằng vòng lặp từng ký tự trong cặp từ để so với `Solution` (kể cả câu rất dài gần 500 ký tự).

````python
def oracle(s):
    w = s.split(" ")
    for i in range(len(w)):
        if w[i][len(w[i]) - 1] != w[(i + 1) % len(w)][0]:
            return False
    return True

random.seed(7)
for t in range(1500):
    s = gen_circular() if t % 3 else gen_random()
    assert Solution().isCircularSentence(s) == oracle(s), s

# câu dài gần 500 ký tự: 'a' lặp, khoảng cách đều
long_true = " ".join(["a" * 9] * 50)            # 50 từ, mỗi từ 9 ký tự 'a', độ dài 50*9 + 49 = 499
long_false = long_true[:-1] + "b"
assert len(long_true) == 499 and Solution().isCircularSentence(long_true) is True
assert Solution().isCircularSentence(long_false) is False
print("oracle khớp 1500 test; câu dài 499 ký tự đúng")
````

## 13. Các lỗi người mới hay gặp

### Lỗi 1: Quên cặp nối vòng (từ cuối → từ đầu)

````python
def sai_quen_noi_vong(sentence):
    words = sentence.split()
    for i in range(len(words) - 1):                   # SAI: thiếu cặp (từ cuối, từ đầu)
        if words[i][-1] != words[i + 1][0]:
            return False
    return True

s = "ab bc"
print("đúng:", Solution().isCircularSentence(s), "| sai:", sai_quen_noi_vong(s))
assert Solution().isCircularSentence(s) != sai_quen_noi_vong(s)
````

**Vì sao sai:** câu có `m` từ cần `m` kiểm tra; vòng lặp `range(m - 1)` chỉ có `m - 1` cặp. `"ab bc"` có `b = b` nhưng `c != a` ở chỗ nối vòng. Cũng với một từ `"ab"`: vòng lặp rỗng nên trả `True` sai.

### Lỗi 2: So sánh không phân biệt hoa thường

````python
def sai_lower(sentence):
    words = sentence.lower().split()                  # SAI: đề phân biệt hoa/thường
    m = len(words)
    return all(words[i][-1] == words[(i + 1) % m][0] for i in range(m))

s = "a A"
print("đúng:", Solution().isCircularSentence(s), "| sai:", sai_lower(s))
assert Solution().isCircularSentence(s) != sai_lower(s)
````

**Vì sao sai:** đề nói rõ chữ hoa và chữ thường là **khác nhau**. `"a A"` có `a` (cuối từ 1) khác `A` (đầu từ 2) nên đáp án `False`; `lower()` biến cả hai thành `a`.

### Lỗi 3: Chỉ so sánh ký tự đầu của câu với ký tự cuối của câu, bỏ qua giữa câu

````python
def sai_chi_dau_cuoi(sentence):
    return sentence[0] == sentence[-1]                # SAI: chỉ kiểm tra nối vòng

s = "Leetcode is cool"
print("đúng:", Solution().isCircularSentence(s), "| sai:", sai_chi_dau_cuoi(s))
s2 = "ab cd ba"
print("đúng:", Solution().isCircularSentence(s2), "| sai:", sai_chi_dau_cuoi(s2))
assert Solution().isCircularSentence(s2) != sai_chi_dau_cuoi(s2)
````

**Vì sao sai:** điều kiện nối vòng chỉ là một trong `m` điều kiện. `"ab cd ba"` có `a == a` ở hai đầu nhưng `b != c` ở giữa.

### Lỗi 4: Truy cập `s[i + 1]` vượt chỉ số khi quét

````python
def sai_vuot_chi_so(sentence):
    for i in range(len(sentence)):                    # SAI: i chạy tới len - 1, s[i + 1] vượt biên
        if sentence[i] == " " and sentence[i - 1] != sentence[i + 1]:
            return False
    return sentence[0] == sentence[-1]

print(sai_vuot_chi_so("ab ba"))     # tình cờ không lỗi vì dấu cách không ở cuối
try:
    bad = "ab "                      # chuỗi sai ràng buộc: dấu cách ở cuối
    sai_vuot_chi_so(bad)
except IndexError as e:
    print("IndexError:", e)
````

**Vì sao nguy hiểm:** nếu vòng lặp đi đến cuối chuỗi mà gặp dấu cách (không xảy ra với dữ liệu hợp lệ nhưng thể hiện thói quen xấu), `s[i + 1]` lỗi `IndexError`. Hãy giới hạn `range(1, len(s) - 1)` hoặc dùng `split()`.

### Lỗi 5: Dùng `split(" ")` mà quên rằng đề đã đảm bảo một dấu cách

Với dữ liệu hợp lệ `split()` và `split(" ")` cho cùng kết quả; nhưng nếu tự ý sinh test có hai dấu cách thì `split(" ")` tạo từ rỗng và `w[0]` lỗi.

````python
print("a  b".split())        # ['a', 'b']
print("a  b".split(" "))     # ['a', '', 'b']
try:
    [w[0] for w in "a  b".split(" ")]
except IndexError as e:
    print("IndexError với từ rỗng:", e)
````

**Cách tránh:** dùng `split()` không đối số (tự bỏ khoảng trắng thừa), và chỉ sinh test theo đúng ràng buộc của đề.

## 14. Mẹo làm bài nhanh & nhận dạng pattern

- Thấy "ký tự cuối của phần tử này = ký tự đầu của phần tử kế" + "nối vòng" ⇒ pattern **mảng vòng**: dùng `(i + 1) % m` hoặc chỉ số `-1`.
- `m` từ cần `m` điều kiện (kể cả điều kiện nối vòng); đếm lại bằng ví dụ một từ.
- Muốn không dùng bộ nhớ phụ: nhìn dấu cách như **ranh giới** `s[i-1]`, `s[i+1]`.
- Hai dòng nộp bài ngắn: `w = s.split(); return all(w[i-1][-1] == w[i][0] for i in range(len(w)))` (dùng `w[-1]` khi `i = 0`).
- Nhớ đề phân biệt hoa/thường, không `lower()`.
- Test nhanh với: một từ, hai từ, từ một ký tự, ví dụ có `False` do cặp giữa câu và do cặp nối vòng.

````python
# Phiên bản siêu ngắn nhờ chỉ số âm: w[i - 1] với i = 0 chính là từ cuối
def short(s):
    w = s.split()
    return all(w[i - 1][-1] == w[i][0] for i in range(len(w)))

for s, want in [("leetcode exercises sound delightful", True), ("eetcode", True), ("Leetcode is cool", False), ("ab bc", False)]:
    assert short(s) == want
print("phiên bản w[i-1][-1] == w[i][0] chạy đúng")
````

## 15. Biến thể của bài

1. **Không phân biệt hoa/thường:** chuẩn hóa bằng `.lower()` rồi làm như cũ.
2. **Câu có thể có nhiều dấu cách:** dùng `split()` thay vì dựa vào dấu cách đơn.
3. **Đếm số cặp "đứt gãy":** đếm số vị trí mà cuối từ trước khác đầu từ sau (kể cả nối vòng).
4. **Sắp xếp lại từ để thành câu vòng tròn:** bài đồ thị Euler (đỉnh = chữ cái, cạnh = từ).
5. **Chuỗi nối từ cuối-đầu (word chain):** trò chơi "nối từ", kiểm tra chuỗi liên tiếp không cần nối vòng.
6. **Đếm số cách quay vòng làm câu thành vòng tròn:** thử dịch vòng và kiểm tra.

````python
# Bài 1
assert Solution().isCircularSentence("ab bc ca") is True
assert Solution().isCircularSentence("ab bc cb") is False

# Bài 2: star rats sun
w = "star rats sun".split()
assert [(w[i][-1], w[(i + 1) % 3][0]) for i in range(3)] == [("r", "r"), ("s", "s"), ("n", "s")]
assert Solution().isCircularSentence("star rats sun") is False

# Bài 3
def count_breaks(sentence):
    words = sentence.split()
    m = len(words)
    return sum(1 for i in range(m) if words[i][-1] != words[(i + 1) % m][0])

# Bài 4
def is_word_chain(sentence):
    words = sentence.split()
    return all(words[i][-1] == words[i + 1][0] for i in range(len(words) - 1))

# Bài 5
def rotations_that_work(sentence):
    words = sentence.split()
    m = len(words)
    res = []
    for k in range(m):
        rot = words[k:] + words[:k]
        if Solution().isCircularSentence(" ".join(rot)):
            res.append(k)
    return res

assert count_breaks("ab bc ca") == 0
assert count_breaks("ab bc cb") == 1
assert count_breaks("star rats sun") == 1
assert (count_breaks("ab bc ca") == 0) == Solution().isCircularSentence("ab bc ca")
assert is_word_chain("ab bc") is True          # chuỗi nối nhưng không vòng tròn
assert Solution().isCircularSentence("ab bc") is False
assert is_word_chain("ab cd") is False
assert rotations_that_work("ab bc ca") == [0, 1, 2]
assert rotations_that_work("ab bc cb") == []
random.seed(13)
for _ in range(300):
    s = gen_circular() if random.random() < 0.5 else gen_random()
    r = rotations_that_work(s)
    assert r == [] or len(r) == len(s.split())
print("tất cả bài tập OK")
````
