## 1. Mục tiêu học

Sau notebook này bạn sẽ:

1. Hiểu đề: trong mảng chuỗi `words`, tìm **khoảng cách lớn nhất** `j - i + 1` giữa hai chỉ số `i < j` mà `words[i] != words[j]`; nếu mọi từ giống nhau thì trả `0`.
2. Hiểu "khoảng cách" ở đây **đếm cả hai đầu mút** (`+ 1`), tức là **số phần tử** từ `i` đến `j`, khác với hiệu chỉ số `j - i`.
3. Nhận ra **tính chất đầu mút**: cặp tối ưu luôn có thể đẩy ra sát **đầu** hoặc **cuối** mảng, nên chỉ cần so sánh với `words[0]` và `words[-1]`, thay vì mọi cặp.
4. Viết 4 cách (vét cạn mọi cặp, thu hẹp cửa sổ từ dài xuống ngắn, từ điển lần xuất hiện đầu/cuối, và quét theo đầu mút `O(n)`) và đối chiếu bằng test ngẫu nhiên có `random.seed`.
5. Tránh các lỗi: quên `+ 1`, trả `n` khi chỉ cần "có hai từ khác nhau", chỉ so với `words[0]`, và chỉ xét các cặp liền kề.

## 2. Đề bài & ví dụ

Cho mảng chuỗi `words`. Tìm **khoảng cách lớn nhất** giữa hai chỉ số **khác nhau** `i` và `j` sao cho `words[i] != words[j]`. Khoảng cách giữa `i` và `j` (với `i < j`) được định nghĩa là `j - i + 1` (tính cả hai đầu). Trả về khoảng cách lớn nhất; nếu **không có** cặp nào hợp lệ thì trả `0`.

| Ví dụ | words | Output | Giải thích |
|---|---|---|---|
| 1 | `["leetcode","leetcode","codeforces"]` | `3` | `words[0] != words[2]`, khoảng cách `2 - 0 + 1 = 3` |
| 2 | `["a","b","c","a","a"]` | `4` | `words[1] != words[4]`, khoảng cách `4 - 1 + 1 = 4` |
| 3 | `["z","z","z"]` | `0` | mọi từ bằng nhau nên không có cặp hợp lệ |

Constraints: `1 <= words.length <= 100`, `1 <= words[i].length <= 10`, `words[i]` gồm chữ cái thường tiếng Anh.

````python
words = ["a", "b", "c", "a", "a"]
n = len(words)
best = 0
for i in range(n):
    for j in range(i + 1, n):
        if words[i] != words[j]:
            best = max(best, j - i + 1)
print("khoảng cách lớn nhất =", best)
print([ (i, j, j - i + 1) for i in range(n) for j in range(i + 1, n) if words[i] != words[j] ][:6], "...")
````

## 4. Ý tưởng nền tảng / cách nghĩ trực giác

Làm tay `words = ["a","b","c","a","a"]`, `n = 5`:

```
các cặp khác nhau (i, j) và khoảng cách j - i + 1:
(0,1)=2  (0,2)=3  (1,2)=2  (1,3)=3  (1,4)=4  (2,3)=2  (2,4)=3
lớn nhất = 4 tại (1, 4)
```

Quan sát: hãy thử **kéo hai đầu ra xa hơn**. Cặp `(1, 4)` đã có `j = n - 1 = 4` ở sát cuối mảng; còn `i = 1` thì không kéo thêm được vì `words[0] = "a"` bằng `words[4] = "a"` (cặp `(0, 4)` không hợp lệ). Ý chính:

- Nếu `words[0] != words[n-1]` thì chọn `(0, n-1)` -> khoảng cách tối đa **`n`**. Kết thúc.
- Nếu `words[0] == words[n-1] = w`, mọi cặp khác nhau `(i, j)` phải có **ít nhất một** đầu khác `w`. Nếu `words[i] != w` thì đổi `j` thành `n-1` (vẫn khác vì `words[n-1] = w`) và khoảng cách **không giảm**; nếu `words[j] != w` thì đổi `i` thành `0`.

Vậy đáp án luôn nằm trong hai ứng viên: `(i*, n-1)` với `i*` là chỉ số **đầu tiên** khác `words[n-1]`, hoặc `(0, j*)` với `j*` là chỉ số **cuối cùng** khác `words[0]`.

## 5. Cách 1 — Brute Force / baseline

Xét **mọi cặp** `i < j`; nếu `words[i] != words[j]` thì cập nhật đáp án với `j - i + 1`. Đúng theo định nghĩa, không cần suy nghĩ.

Nút thắt: `O(n²)` cặp. Với `n <= 100` chỉ khoảng `5000` phép so sánh chuỗi, thoải mái; nhưng khi `n` lên `10^5` thì không dùng được — cần khai thác tính chất đầu mút.

````python
class SolutionBrute:
    def maxDistance(self, words):
        n = len(words)
        best = 0
        for i in range(n):
            for j in range(i + 1, n):
                if words[i] != words[j]:
                    best = max(best, j - i + 1)
        return best

print(SolutionBrute().maxDistance(["leetcode", "leetcode", "codeforces"]))
print(SolutionBrute().maxDistance(["a", "b", "c", "a", "a"]))
print(SolutionBrute().maxDistance(["z", "z", "z"]))
````

**Độ phức tạp:** `n = len(words)`, `L <= 10` là độ dài từ. Time `O(n² * L)` (mỗi so sánh chuỗi `O(L)`), Space `O(1)`.

## 6. Ý tưởng thuật toán / bước đột phá

**Bước đột phá:** cặp tối ưu luôn có thể "kéo" ra **sát một đầu mảng**. Do đó chỉ cần xét hai họ ứng viên, mỗi họ chỉ phụ thuộc **một vòng quét**:

1. **Ghim đầu phải `j = n - 1`:** tìm `i` **nhỏ nhất** sao cho `words[i] != words[n-1]`; khoảng cách `n - i`.
2. **Ghim đầu trái `i = 0`:** tìm `j` **lớn nhất** sao cho `words[j] != words[0]`; khoảng cách `j + 1`.

Đáp án = `max` của hai ứng viên (nếu không tồn tại ứng viên nào thì `0`).

**Quy tắc "nếu… thì…":**

- Nếu `words[0] != words[n-1]` -> `i = 0` đã thỏa ở họ 1, khoảng cách `n` (tối đa).
- Nếu mọi từ bằng nhau -> không có `i` hay `j` nào -> `0`.
- Ngược lại: hai ứng viên khác nhau, lấy `max`.

Invariant: mọi cặp hợp lệ `(i, j)` đều có một ứng viên (họ 1 hoặc họ 2) với khoảng cách **không nhỏ hơn**.

````python
def max_distance(words):
    n = len(words)
    best = 0
    for i in range(n):                      # họ 1: ghim j = n - 1, i nhỏ nhất khác words[-1]
        if words[i] != words[-1]:
            best = max(best, n - i)
            break
    for j in range(n - 1, -1, -1):          # họ 2: ghim i = 0, j lớn nhất khác words[0]
        if words[j] != words[0]:
            best = max(best, j + 1)
            break
    return best

print(max_distance(["leetcode", "leetcode", "codeforces"]), max_distance(["a", "b", "c", "a", "a"]), max_distance(["z", "z", "z"]))
````

## 7. Các cách tiếp theo

### Cách 2 — Thu hẹp cửa sổ từ dài xuống ngắn

Thử khoảng cách `d = n, n-1, ..., 2`. Với mỗi `d`, duyệt mọi đoạn `[s, s + d - 1]` và kiểm tra hai đầu. **Gặp đoạn đầu tiên có hai đầu khác nhau thì trả luôn `d`** (vì ta đi từ dài nhất). Nếu không có thì `0`. Tương tự vét cạn nhưng **dừng sớm**; xấu nhất vẫn `O(n²)` khi mọi từ bằng nhau.

````python
class SolutionWindow:
    def maxDistance(self, words):
        n = len(words)
        for d in range(n, 1, -1):                 # d = số phần tử của đoạn
            for s in range(0, n - d + 1):
                if words[s] != words[s + d - 1]:
                    return d
        return 0

print(SolutionWindow().maxDistance(["leetcode", "leetcode", "codeforces"]), SolutionWindow().maxDistance(["a", "b", "c", "a", "a"]), SolutionWindow().maxDistance(["z", "z", "z"]))
````

**Độ phức tạp:** Time `O(n² * L)` xấu nhất (khi mọi từ bằng nhau), tốt nhất `O(L)`. Space `O(1)`.

### Cách 3 — Từ điển lần xuất hiện đầu / cuối

Một lượt tạo `first[w]` và `last[w]`. Với mọi cặp từ **khác nhau** `(a, b)`, cặp `(first[a], last[b])` cho khoảng cách `last[b] - first[a] + 1` nếu dương. Lấy `max`. Số từ khác nhau `u <= n`, nên tốn `O(u²)`; hợp khi các từ **ít loại**.

````python
class SolutionFirstLast:
    def maxDistance(self, words):
        first, last = {}, {}
        for i, w in enumerate(words):
            first.setdefault(w, i)
            last[w] = i
        best = 0
        for a in first:
            for b in last:
                if a != b and last[b] >= first[a]:
                    best = max(best, last[b] - first[a] + 1)
        return best

print(SolutionFirstLast().maxDistance(["leetcode", "leetcode", "codeforces"]), SolutionFirstLast().maxDistance(["a", "b", "c", "a", "a"]), SolutionFirstLast().maxDistance(["z", "z", "z"]))
````

**Độ phức tạp:** Time `O(n + u²)`, Space `O(u)`.

### Cách 4 — Quét theo đầu mút `O(n)` (cách nộp bài)

Đúng ý tưởng mục 6: hai vòng `for` tách rời, mỗi vòng dừng ngay khi gặp phần tử đầu tiên khác đầu đối diện.

````python
class SolutionEnds:
    def maxDistance(self, words):
        n = len(words)
        best = 0
        for i in range(n):
            if words[i] != words[-1]:
                best = n - i
                break
        for j in range(n - 1, -1, -1):
            if words[j] != words[0]:
                best = max(best, j + 1)
                break
        return best

print(SolutionEnds().maxDistance(["leetcode", "leetcode", "codeforces"]), SolutionEnds().maxDistance(["a", "b", "c", "a", "a"]), SolutionEnds().maxDistance(["z", "z", "z"]))
````

**Độ phức tạp:** Time `O(n * L)` (mỗi vòng quét tối đa `n` so sánh chuỗi dài `<= L`), Space `O(1)`.

## 8. TRACE từng bước

Trace `words = ["a","b","c","a","a"]`, `n = 5`.

**Họ 1 (ghim `j = 4`, `words[-1] = "a"`), quét `i` từ trái:**

| i | words[i] | `!= "a"`? | hành động |
|---|---|---|---|
| 0 | `a` | không | tiếp tục |
| 1 | `b` | có | `best = 5 - 1 = 4`, dừng |

**Họ 2 (ghim `i = 0`, `words[0] = "a"`), quét `j` từ phải:**

| j | words[j] | `!= "a"`? | hành động |
|---|---|---|---|
| 4 | `a` | không | tiếp tục |
| 3 | `a` | không | tiếp tục |
| 2 | `c` | có | `best = max(4, 2 + 1) = 4`, dừng |

Đáp án `4`.

````python
def trace_ends(words):
    n = len(words)
    print("words =", words)
    best = 0
    for i in range(n):
        diff = words[i] != words[-1]
        print(f"  [họ 1] i={i}: {words[i]!r} != {words[-1]!r}? {diff}")
        if diff:
            best = n - i
            print(f"    -> khoảng cách n - i = {best}")
            break
    for j in range(n - 1, -1, -1):
        diff = words[j] != words[0]
        print(f"  [họ 2] j={j}: {words[j]!r} != {words[0]!r}? {diff}")
        if diff:
            print(f"    -> khoảng cách j + 1 = {j + 1}")
            best = max(best, j + 1)
            break
    print("  => đáp án", best)
    return best

trace_ends(["a", "b", "c", "a", "a"])
trace_ends(["z", "z", "z"])
trace_ends(["x", "y"])
````

## 9. Code nộp LeetCode

````python
from typing import List, Optional

class Solution:
    def maxDistance(self, words: List[str]) -> int:
        n = len(words)
        best = 0
        # ghim đầu phải: i nhỏ nhất mà words[i] != words[-1]
        for i in range(n):
            if words[i] != words[-1]:
                best = n - i
                break
        # ghim đầu trái: j lớn nhất mà words[j] != words[0]
        for j in range(n - 1, -1, -1):
            if words[j] != words[0]:
                best = max(best, j + 1)
                break
        return best
````

## 10. Vì sao thuật toán đúng

- **Cận trên bằng ứng viên:** lấy một cặp hợp lệ bất kỳ `(i, j)` với `i < j`, `words[i] != words[j]`. Xét `w = words[n-1]`.
  - Nếu `words[i] != w`: cặp `(i, n-1)` hợp lệ và khoảng cách `n - i >= j - i + 1` (vì `j <= n - 1`). Ứng viên họ 1 (với `i` nhỏ nhất như vậy) còn **không kém hơn**.
  - Nếu `words[i] == w`: vì `words[j] != words[i] = w` nên `words[j] != words[n-1]`; khi đó xét `words[0]`: nếu `words[j] != words[0]` thì cặp `(0, j)` có khoảng cách `j + 1 >= j - i + 1`, thuộc họ 2; nếu `words[j] == words[0]` thì `words[0] = words[j] != w = words[n-1]`, cặp `(0, n-1)` hợp lệ và có khoảng cách `n` (tối đa), thuộc họ 1 với `i = 0`.
- **Mọi ứng viên đều hợp lệ:** `(i, n-1)` có `words[i] != words[n-1]` và `i < n - 1` (vì `i = n - 1` không thể khác chính nó); tương tự `(0, j)`.
- **Dừng ở phần tử đầu tiên:** ở họ 1 `i` nhỏ nhất cho khoảng cách `n - i` lớn nhất; ở họ 2 `j` lớn nhất cho `j + 1` lớn nhất.
- **Không có cặp nào:** nếu mọi từ bằng nhau thì không vòng nào cập nhật, `best = 0`.

## 11. Phân tích Time / Space Complexity

`n = len(words) <= 100`, `L <= 10` độ dài từ, `u` số từ khác nhau.

| Cách | Time | Space |
|---|---|---|
| Vét cạn mọi cặp | `O(n² L)` | `O(1)` |
| Thu hẹp cửa sổ | `O(n² L)` xấu nhất | `O(1)` |
| Từ điển đầu/cuối | `O(n L + u²)` | `O(u)` |
| Quét theo đầu mút | `O(n L)` | `O(1)` |

Cách quét theo đầu mút là tối ưu: một lượt tuyến tính, không bộ nhớ phụ.

## 12. Bộ test

Ví dụ của đề, các biên (`n = 1`, `n = 2` khác nhau/giống nhau, đầu = cuối, đầu ≠ cuối) và đối chiếu **tất cả cách** trên input ngẫu nhiên tuân thủ constraints.

````python
impls = [SolutionBrute(), SolutionWindow(), SolutionFirstLast(), SolutionEnds(), Solution()]

cases = [
    (["leetcode", "leetcode", "codeforces"], 3),
    (["a", "b", "c", "a", "a"], 4),
    (["z", "z", "z"], 0),
    (["a"], 0),                           # n = 1
    (["a", "a"], 0),
    (["a", "b"], 2),
    (["a", "b", "a"], 2),                 # đầu = cuối, giữa khác -> (0,1) hoặc (1,2)
    (["a", "a", "b", "a"], 3),            # (0, 2) = 3 hoặc (2,3) = 2 -> 3
    (["b", "a", "a", "a"], 4),            # đầu != cuối -> n
    (["a", "a", "a", "b"], 4),
    (["ab", "ba", "ab"], 2),              # khác nhau theo thứ tự ký tự
]
for words, exp in cases:
    assert 1 <= len(words) <= 100 and all(1 <= len(w) <= 10 and w.islower() and w.isalpha() for w in words)
    for impl in impls:
        got = impl.maxDistance(words[:])
        assert got == exp, (type(impl).__name__, words, got, exp)
print("ví dụ + edge OK:", len(cases), "ca")
````

````python
import random
random.seed(3696)

def ref(words):
    n = len(words)
    return max([j - i + 1 for i in range(n) for j in range(i + 1, n) if words[i] != words[j]], default=0)

alphabet_small = ["a", "b", "c"]
for _ in range(3000):
    n = random.randint(1, 12)
    k = random.choice([1, 2, 3])
    words = [random.choice(alphabet_small[:k]) for _ in range(n)]       # ít loại -> nhiều ca bằng nhau
    exp = ref(words)
    for impl in impls:
        assert impl.maxDistance(words[:]) == exp, (words, type(impl).__name__)
print("3000 test ngẫu nhiên nhỏ (seed=3696) OK")
````

## 13. Các lỗi người mới hay gặp

### Lỗi 1: quên `+ 1` trong khoảng cách

Đề định nghĩa khoảng cách `j - i + 1` (số phần tử). Dùng `j - i` thì lệch một.

````python
def wrong_no_plus1(words):
    n = len(words)
    return max([j - i for i in range(n) for j in range(i + 1, n) if words[i] != words[j]], default=0)

words = ["leetcode", "leetcode", "codeforces"]
print(wrong_no_plus1(words), "!=", Solution().maxDistance(words))
assert wrong_no_plus1(words) != Solution().maxDistance(words)           # 2 != 3
````

### Lỗi 2: trả `n` ngay khi có hai từ khác nhau

"Có hai từ khác nhau" không có nghĩa hai **đầu mảng** khác nhau. Khoảng cách tối đa `n` chỉ khi `words[0] != words[-1]`.

````python
def wrong_return_n(words):
    return len(words) if len(set(words)) > 1 else 0

words = ["a", "b", "c", "a", "a"]
print(wrong_return_n(words), "!=", Solution().maxDistance(words))
assert wrong_return_n(words) != Solution().maxDistance(words)           # 5 != 4
````

### Lỗi 3: chỉ so sánh với `words[0]`

Bỏ họ ứng viên ghim đầu phải. Với `["a","b","c","a","a"]` chỉ nhìn đầu trái cho `3`, bỏ lỡ cặp `(1, 4)` có khoảng cách `4`.

````python
def wrong_only_first(words):
    n = len(words)
    for j in range(n - 1, -1, -1):
        if words[j] != words[0]:
            return j + 1
    return 0

words = ["a", "b", "c", "a", "a"]
print(wrong_only_first(words), "!=", Solution().maxDistance(words))
assert wrong_only_first(words) != Solution().maxDistance(words)         # 3 != 4
````

### Lỗi 4: chỉ xét các cặp liền kề

Đề hỏi cặp **xa nhất**, không phải cặp kề nhau khác nhau.

````python
def wrong_adjacent(words):
    return max([2 for a, b in zip(words, words[1:]) if a != b], default=0)

words = ["a", "b", "c", "a", "a"]
print(wrong_adjacent(words), "!=", Solution().maxDistance(words))
assert wrong_adjacent(words) != Solution().maxDistance(words)           # 2 != 4
````

## 14. Mẹo làm bài nhanh & nhận dạng pattern

- "Khoảng cách **lớn nhất** giữa hai phần tử thỏa điều kiện (khác nhau / bằng nhau)" -> hãy nghĩ ngay đến **đầu mút mảng**: kéo `i` về `0` hoặc `j` về `n-1` mà vẫn hợp lệ.
- Bài song sinh: **Two Furthest Houses With Different Colors** (LeetCode 2078) — cùng ý tưởng, khoảng cách là `j - i`.
- Cặp tốt nhất thường chỉ cần so với **`a[0]`** hoặc **`a[-1]`**, đừng vét cạn nếu `n` lớn.
- Luôn kiểm tra định nghĩa khoảng cách: `j - i` hay `j - i + 1`.
- Khi mọi phần tử giống nhau -> trả `0`; đừng quên ca này.
- Viết bản **vét cạn** để làm chuẩn kiểm tra cho bản tối ưu (cách làm trong bộ test).

## 15. Biến thể của bài

1. **Two Furthest Houses With Different Colors** (LeetCode 2078): cùng bài nhưng khoảng cách là `|i - j|`.
2. **Maximum Distance Between Unequal Words in Array II** (bài kế tiếp trong dãy): phiên bản khó hơn (đề riêng, cần đọc kỹ ràng buộc mới).
3. Khoảng cách lớn nhất giữa hai từ **bằng nhau** (dùng `first[w]` và `last[w]`).
4. Trả về **cặp chỉ số** `(i, j)` thay vì khoảng cách.
5. Đếm **số cặp** `(i, j)` đạt khoảng cách lớn nhất.

## 16. So sánh các cách làm

| Cách | Ý tưởng | Time | Space | Khi nào dùng |
|---|---|---|---|---|
| Vét cạn mọi cặp | thử mọi `i < j` | `O(n² L)` | `O(1)` | Bản chuẩn để đối chiếu |
| Thu hẹp cửa sổ | từ khoảng cách dài nhất xuống | `O(n² L)` xấu nhất | `O(1)` | Dừng sớm khi dữ liệu "dễ" |
| Đầu/cuối bằng từ điển | `first[w]`, `last[w]`, xét cặp từ | `O(n L + u²)` | `O(u)` | Ít loại từ, cần tổng quát hóa |
| Quét theo đầu mút | ghim `j = n-1` và `i = 0` | `O(n L)` | `O(1)` | Cách nộp chuẩn |

Chọn quét theo đầu mút để nộp; vét cạn dùng để kiểm chứng.

## 18. Gợi ý 3 mức cho bài tập

**Mức 1 (nhẹ):** Bài 1: đầu và cuối đều `"x"`, nhìn xem từ khác nằm ở đâu. Bài 2: `words[0] = "a"`, `words[-1] = "c"`.

**Mức 2 (vừa):** Bài 3: `for i, w in enumerate(words): if w != target: return i`. Bài 4: giống quét theo đầu mút nhưng nhớ cả chỉ số. Bài 5: dùng `first[w]` và `last[w]` cho từng từ.

**Mức 3 (gần đáp án):** Bài 1: `"y"` ở chỉ số `2`: cặp `(0, 2)` khoảng cách `3`, cặp `(2, 3)` khoảng cách `2` -> `3`. Bài 2: đầu `"a"` != cuối `"c"` nên `(0, 4)` khoảng cách `5`. Bài 5: với mỗi từ, `last[w] - first[w] + 1` nếu từ xuất hiện từ hai lần trở lên.

## 19. Đáp án tham khảo cho bài tập

````python
# Bài 1
assert Solution().maxDistance(["x", "x", "y", "x"]) == 3

# Bài 2
assert Solution().maxDistance(["a", "a", "b", "c", "c"]) == 5

# Bài 3
def first_diff_index(words, target):
    for i, w in enumerate(words):
        if w != target:
            return i
    return -1
assert first_diff_index(["a", "a", "b"], "a") == 2
assert first_diff_index(["a", "a"], "a") == -1

# Bài 4
def max_distance_pair(words):
    n = len(words)
    best, pair = 0, None
    i = next((k for k in range(n) if words[k] != words[-1]), None)
    if i is not None:
        best, pair = n - i, (i, n - 1)
    j = next((k for k in range(n - 1, -1, -1) if words[k] != words[0]), None)
    if j is not None and j + 1 > best:
        best, pair = j + 1, (0, j)
    return pair
assert max_distance_pair(["a", "b", "c", "a", "a"]) == (1, 4)
assert max_distance_pair(["x", "x", "y", "x"]) == (0, 2)
assert max_distance_pair(["z", "z"]) is None
assert max_distance_pair(["a", "b"]) == (0, 1)

# Bài 5
def max_distance_equal(words):
    first, last = {}, {}
    for i, w in enumerate(words):
        first.setdefault(w, i)
        last[w] = i
    return max((last[w] - first[w] + 1 for w in first if last[w] > first[w]), default=0)
assert max_distance_equal(["a", "b", "a"]) == 3
assert max_distance_equal(["a", "b", "c"]) == 0
assert max_distance_equal(["z", "z", "z"]) == 3
print("đáp án bài tập OK")
````

## 20. Template nhớ nhanh + Checklist tự kiểm tra trước khi submit

**Template "khoảng cách lớn nhất giữa hai phần tử khác nhau — chỉ xét đầu mút":**

```python
n, best = len(words), 0
for i in range(n):                       # ghim đầu phải
    if words[i] != words[-1]:
        best = n - i; break
for j in range(n - 1, -1, -1):           # ghim đầu trái
    if words[j] != words[0]:
        best = max(best, j + 1); break
return best
```

- [ ] Trả `0` khi mọi từ bằng nhau (và khi `n = 1`)?
- [ ] Đã thử `["a","b"]`, `["a","b","a"]`, `["a","a","b","a"]`, `["z","z","z"]`?
- [ ] Không lẫn "có hai từ khác nhau" với "hai đầu mảng khác nhau"?
