## 2. Đề bài & ví dụ

Cho số nguyên dương `n`. Xác định xem `n` có **chia hết** cho tổng của hai giá trị sau không:

- **Tổng chữ số** của `n`.
- **Tích chữ số** của `n`.

Trả về `true` nếu `n` chia hết cho tổng đó, ngược lại trả về `false`.

| Ví dụ | n | Output | Giải thích |
|---|---|---|---|
| 1 | `99` | `true` | tổng = `9+9 = 18`, tích = `9*9 = 81`, `18 + 81 = 99`, và `99` chia hết cho `99` |
| 2 | `23` | `false` | tổng = `5`, tích = `6`, `5 + 6 = 11`, `23` không chia hết cho `11` |

Constraints: `1 <= n <= 10^6`.

````python
n = 99
digits = [int(ch) for ch in str(n)]
s = sum(digits)
p = 1
for d in digits:
    p *= d
print("chữ số:", digits, "| tổng:", s, "| tích:", p, "| s+p =", s + p)
print("n % (s+p) =", n % (s + p), "->", n % (s + p) == 0)
n = 23
digits = [int(ch) for ch in str(n)]
print(sum(digits), eval("*".join(map(str, digits))), n % (sum(digits) + 6))
````

### 3.2. Tách chữ số bằng toán học: `% 10` và `// 10`

`n % 10` là **chữ số hàng đơn vị**; `n // 10` là số sau khi **bỏ** chữ số đó. Lặp đến khi `n == 0`. `divmod(n, 10)` trả cả hai: `n, d = divmod(n, 10)`. Chữ số lấy ra theo thứ tự từ phải sang trái, nhưng tổng và tích **không phụ thuộc thứ tự**, nên không cần đảo.

````python
n = 4096
out = []
while n > 0:
    n, d = divmod(n, 10)
    out.append(d)
print(out)           # [6, 9, 0, 4]
print(divmod(4096, 10), 4096 % 10, 4096 // 10)
````

### 3.3. Tổng, tích và phần tử trung hòa

Phép cộng có phần tử trung hòa `0` (`x + 0 = x`) nên tổng khởi tạo `0`. Phép nhân có phần tử trung hòa `1` (`x * 1 = x`) nên tích khởi tạo **`1`**. Nếu khởi tạo tích bằng `0` thì mọi tích đều bằng `0` — đây là lỗi hay gặp nhất. `math.prod(iterable)` tự khởi tạo `1` giúp bạn.

````python
import math
print(math.prod([9, 9]), math.prod([]), sum([]))
print(math.prod([4, 0, 9, 6]))      # có chữ số 0 -> tích = 0
````

### 3.4. Phép chia hết và toán tử `%`

`a % b == 0` nghĩa là `a` chia hết cho `b`. Cẩn thận **độ ưu tiên**: `%` ưu tiên **cao hơn** `+`, nên `n % s + p` nghĩa là `(n % s) + p`, **không** phải `n % (s + p)`. Luôn thêm ngoặc. Ở đây `s >= 1` (vì `n >= 1` có ít nhất một chữ số khác `0`) nên `s + p >= 1`, không bao giờ chia cho `0`.

````python
n, s, p = 99, 18, 81
print(n % s + p, "<- sai ngoặc")
print(n % (s + p), "<- đúng")
print(10 % 3, 10 % 5, 0 % 7)
````

### 3.5. Generator expression và `sum(... for ...)`

`sum(int(ch) for ch in str(n))` là **generator expression**: không tạo danh sách trung gian mà đưa từng giá trị cho `sum`. Tương tự `math.prod(int(ch) for ch in str(n))`. Cú pháp gọn hơn nhưng với người mới hãy nhớ nó tương đương một vòng `for` cộng dồn. Ngoài ra `all(...)` và `any(...)` cũng nhận generator, rất tiện cho các phép kiểm tra trong test.

````python
n = 4096
print(sum(int(ch) for ch in str(n)), math.prod(int(ch) for ch in str(n)))
print(all(int(ch) > 0 for ch in "4096"), any(int(ch) == 0 for ch in "4096"))
print(sum([int(ch) for ch in str(n)]) == sum(int(ch) for ch in str(n)))
````

## 4. Ý tưởng nền tảng / cách nghĩ trực giác

Làm tay `n = 99`:

```
chữ số: 9, 9
tổng  = 9 + 9 = 18
tích  = 9 * 9 = 81
t = tổng + tích = 99
99 % 99 = 0  -> True
```

Làm tay `n = 23`: tổng `5`, tích `6`, `t = 11`, `23 % 11 = 1` nên **False**.

Thêm hai trường hợp đáng nhớ:

- `n = 10`: chữ số `1, 0` -> tổng `1`, tích `0`, `t = 1`; mọi số đều chia hết cho `1` -> **True**.
- `n = 7` (một chữ số `d`): tổng `7`, tích `7`, `t = 14 > 7` -> **False**. Nói chung mọi số một chữ số `d` cho `t = 2d > d`, nên luôn **False**.

Bài này không có "mẹo" thuật toán: chỉ cần **tính đúng hai đại lượng** rồi kiểm tra `%`.

## 5. Cách 1 — Brute Force / baseline

Biến `n` thành chuỗi, lấy danh sách chữ số, rồi dùng `sum` và `math.prod`. Đây chính là mô tả đề viết thành code.

Nút thắt: không có nút thắt thật sự về tốc độ (chỉ `<= 7` chữ số). Chỗ cần chú ý là **đúng** chứ không phải **nhanh**: tích phải khởi tạo `1`, và phải kiểm tra `n % (s + p)`.

````python
import math

class SolutionString:
    def checkDivisibility(self, n):
        digits = [int(ch) for ch in str(n)]
        total = sum(digits) + math.prod(digits)
        return n % total == 0

print(SolutionString().checkDivisibility(99))
print(SolutionString().checkDivisibility(23))
print(SolutionString().checkDivisibility(10))
````

**Độ phức tạp:** `d` = số chữ số của `n` (`d = O(log n)`). Time `O(d)`, Space `O(d)` cho danh sách chữ số.

## 6. Ý tưởng thuật toán / bước đột phá

**Bước đột phá:** không cần danh sách chữ số — mỗi chữ số chỉ cần **một lần** để cập nhật hai biến chạy: `s += d` và `p *= d`. Dùng `divmod(n, 10)` để bóc chữ số, không đổi sang chuỗi, Space còn `O(1)`.

**Quy tắc "nếu… thì…":**

- Bóc chữ số cuối `d`: `s += d`, `p *= d`.
- Nếu có chữ số `0` -> `p` thành `0` và **ở lại `0`** (nhân gì cũng `0`), khi đó điều kiện là `n % s == 0`.
- Cuối cùng so `n % (s + p) == 0`.

Invariant: sau khi bóc `k` chữ số cuối, `s` và `p` đúng bằng tổng và tích của `k` chữ số đó.

````python
def digit_sum_product(n):
    s, p = 0, 1
    while n > 0:
        n, d = divmod(n, 10)
        s += d
        p *= d
    return s, p

print(digit_sum_product(99), digit_sum_product(23), digit_sum_product(4096))
````

## 7. Các cách tiếp theo

### Cách 2 — Liệt kê các bội của `s + p`

Thay vì tính `n % t`, ta **đếm bội**: bắt đầu `m = t`, cộng `t` đến khi `m >= n`, rồi xem `m == n`. Cách này không dùng toán tử `%` — chỉ để thấy "chia hết" nghĩa là "`n` nằm trong dãy `t, 2t, 3t, ...`". Với `n <= 10^6` và `t >= 1` có tối đa `10^6` bước — vẫn chạy được (khá chậm nếu `t = 1`, nhưng đúng).

````python
class SolutionMultiples:
    def checkDivisibility(self, n):
        s, p = 0, 1
        for ch in str(n):
            d = int(ch)
            s += d
            p *= d
        t = s + p
        m = t
        while m < n:
            m += t
        return m == n

print(SolutionMultiples().checkDivisibility(99), SolutionMultiples().checkDivisibility(23))
````

**Độ phức tạp:** Time `O(d + n / t)` (xấu nhất `O(n)` khi `t` nhỏ), Space `O(1)`. Chỉ dùng để học: cách `%` mới là cách nên nộp.

### Cách 3 — Vòng `divmod` một lượt (cách nộp bài)

Bóc chữ số bằng `divmod`, cập nhật `s`, `p`, rồi kiểm tra `n % (s + p)`. Cần giữ lại bản gốc của `n` vì ta đang biến đổi một biến tạm `x`.

````python
class SolutionDivmod:
    def checkDivisibility(self, n):
        x, s, p = n, 0, 1
        while x > 0:
            x, d = divmod(x, 10)
            s += d
            p *= d
        return n % (s + p) == 0

print(SolutionDivmod().checkDivisibility(99), SolutionDivmod().checkDivisibility(23))
````

**Độ phức tạp:** Time `O(d) = O(log n)`, Space `O(1)`.

### Cách 4 — Đệ quy trả về cặp `(tổng, tích)`

Hàm `go(x)` trả `(s, p)` của các chữ số của `x`: nếu `x == 0` thì `(0, 1)` (**phần tử trung hòa**), ngược lại gộp `(s', p')` của `x // 10` với chữ số `x % 10`. Đây là mẫu "gộp kết quả con" hay gặp trong cây và danh sách.

````python
class SolutionRecursive:
    def go(self, x):
        if x == 0:
            return 0, 1
        s, p = self.go(x // 10)
        d = x % 10
        return s + d, p * d

    def checkDivisibility(self, n):
        s, p = self.go(n)
        return n % (s + p) == 0

print(SolutionRecursive().checkDivisibility(99), SolutionRecursive().checkDivisibility(23))
````

**Độ phức tạp:** Time `O(d)`, Space `O(d)` (ngăn xếp đệ quy, `d <= 7`).

## 8. TRACE từng bước

Trace `n = 99` và `n = 23` — in biến qua từng vòng lặp `divmod`.

| Vòng | x trước | d = x % 10 | s | p |
|---|---|---|---|---|
| 0 | 99 | - | 0 | 1 |
| 1 | 99 | 9 | 9 | 9 |
| 2 | 9 | 9 | 18 | 81 |
| kết thúc | 0 | - | 18 | 81 |

Cuối cùng `t = 18 + 81 = 99`, `99 % 99 = 0` -> `True`.

````python
def trace_check(n):
    print(f"=== n = {n} ===")
    x, s, p = n, 0, 1
    step = 0
    while x > 0:
        step += 1
        before = x
        x, d = divmod(x, 10)
        s += d
        p *= d
        print(f"vòng {step}: x={before:>7} d={d} -> s={s:>3} p={p:>6} x_mới={x}")
    t = s + p
    print(f"s={s}, p={p}, t=s+p={t}, n % t = {n % t}  ->  {n % t == 0}")
    return n % t == 0

trace_check(99)
trace_check(23)
trace_check(4096)
````

## 9. Code nộp LeetCode

````python
from typing import List, Optional

class Solution:
    def checkDivisibility(self, n: int) -> bool:
        x = n
        digit_sum, digit_product = 0, 1
        while x > 0:
            x, d = divmod(x, 10)
            digit_sum += d
            digit_product *= d
        return n % (digit_sum + digit_product) == 0
````

## 10. Vì sao thuật toán đúng

- **Tách chữ số đúng:** `divmod(x, 10)` cho `x = 10*q + d` với `0 <= d <= 9`; `d` là chữ số hàng đơn vị, `q` là các chữ số còn lại. Lặp cho đến khi `x = 0` thì mọi chữ số đã được xét **đúng một lần**.
- **Khởi tạo đúng:** `s = 0` và `p = 1` là các phần tử trung hòa nên sau khi xét đủ chữ số, `s` là tổng và `p` là tích của các chữ số. Nếu có chữ số `0`, `p = 0` đúng với định nghĩa tích.
- **Không chia cho 0:** `n >= 1` nên có ít nhất một chữ số khác `0`, suy ra `s >= 1`; `p >= 0`, vậy `s + p >= 1`.
- **Điều kiện:** đề yêu cầu `n` chia hết cho `s + p`, chính là `n % (s + p) == 0`.

## 11. Phân tích Time / Space Complexity

`d` = số chữ số của `n` = `floor(log10 n) + 1 <= 7`.

| Cách | Time | Space |
|---|---|---|
| Chuỗi + `math.prod` | `O(d)` | `O(d)` |
| Liệt kê bội | `O(d + n/t)` | `O(1)` |
| `divmod` một lượt | `O(d)` | `O(1)` |
| Đệ quy | `O(d)` | `O(d)` |

Cách tốt nhất: `divmod` một lượt, `O(log n)` thời gian và `O(1)` bộ nhớ. Trên LeetCode mọi cách trừ "liệt kê bội" đều tức thì.

## 12. Bộ test

Ví dụ của đề, các biên (`1`, `10`, `10^6`), mọi số một chữ số (luôn `False`), rồi đối chiếu **tất cả cách** trên nhiều giá trị `n`.

````python
impls = [SolutionString(), SolutionMultiples(), SolutionDivmod(), SolutionRecursive(), Solution()]

def ref(n):
    digits = list(map(int, str(n)))
    p = 1
    for d in digits:
        p *= d
    return n % (sum(digits) + p) == 0

cases = [(99, True), (23, False), (10, True), (1, False), (1000000, True), (12, False), (24, False)]
for n, exp in cases:
    for impl in impls:
        got = impl.checkDivisibility(n)
        assert got == exp, (type(impl).__name__, n, got, exp)
print("ví dụ + edge OK")
# 24: s=6, p=8, t=14 -> 24 % 14 = 10 nên False
print([ref(n) for n, _ in cases])
````

Kiểm tra tay `24`: tổng `6`, tích `8`, `t = 14`, `24 % 14 = 10` nên `False`. Và mọi số tròn chục `10, 20, ..., 90` đều `True` vì tích `0` nên `t = s = d` (chữ số hàng chục) và `10*d % d == 0`.

````python
print("24:", sum([2, 4]), 2 * 4, sum([2, 4]) + 2 * 4, 24 % (sum([2, 4]) + 2 * 4))
good = [n for n in range(10, 100) if ref(n)]
print("các số 2 chữ số thỏa:", good)
assert all(n in good for n in range(10, 100, 10))
````

### Phân tích thêm: khi nào `s + p` lớn hơn `n`?

Nếu `s + p > n` thì `n` **không thể** chia hết (vì `n >= 1`, bội dương nhỏ nhất của `t` là `t` chính nó). Với số `d` chữ số, `s <= 9d` còn `p` có thể lên tới `9^d`. Ví dụ `99`: `p = 81` và `s + p = 99 = n` nên vừa khít. Với `999`: `s = 27`, `p = 729`, `t = 756 < 999` nhưng `999 % 756 = 243` nên `False`. Ta có thể dùng nhận xét này như một **kiểm tra nhanh** (early exit): `if t > n: return False`.

````python
def quick(n):
    s, p = digit_sum_product(n)
    t = s + p
    if t > n:
        return False        # bội dương nhỏ nhất của t là t > n
    return n % t == 0

def digit_sum_product(n):
    s, p = 0, 1
    while n > 0:
        n, d = divmod(n, 10)
        s += d
        p *= d
    return s, p

for n in range(1, 20001):
    assert quick(n) == ref(n)
print(quick(999), digit_sum_product(999), "early exit khớp ref trên 1..20000")
````

````python
# Số một chữ số luôn False; mọi cách giống ref trên 1..5000
for d in range(1, 10):
    for impl in impls:
        assert impl.checkDivisibility(d) is False
for n in range(1, 5001):
    exp = ref(n)
    assert all(impl.checkDivisibility(n) == exp for impl in impls[1:]) , n
print("1..5000: tất cả cách khớp ref")
````

## 13. Các lỗi người mới hay gặp

### Lỗi 1: khởi tạo tích bằng `0`

`p = 0` rồi `p *= d` thì `p` luôn `0`, tức là đang bỏ mất tích chữ số.

````python
def wrong_p0(n):
    x, s, p = n, 0, 0                  # SAI: p phải bắt đầu từ 1
    while x:
        x, d = divmod(x, 10)
        s += d
        p *= d
    return n % (s + p) == 0

print(wrong_p0(99), "!=", Solution().checkDivisibility(99))
assert wrong_p0(99) != Solution().checkDivisibility(99)        # False != True
````

### Lỗi 2: kiểm tra chia hết **riêng** cho tổng và cho tích

Đề hỏi `n` chia hết cho **tổng** `s + p`, không phải cho `s` **và** cho `p`. Hơn nữa nếu `p = 0` thì `n % 0` còn gây lỗi chia cho `0`.

````python
def wrong_separate(n):
    digits = list(map(int, str(n)))
    s, p = sum(digits), math.prod(digits)
    return p != 0 and n % s == 0 and n % p == 0

print(wrong_separate(99), "!=", Solution().checkDivisibility(99))
assert wrong_separate(99) != Solution().checkDivisibility(99)  # 99 % 18 != 0 nên False, trong khi đáp án True
````

### Lỗi 3: chỉ dùng tổng chữ số

Quên cộng tích vào mẫu số. Với `n = 12`: tổng `3` nên `12 % 3 == 0` (sai kết luận), trong khi `s + p = 3 + 2 = 5` và `12 % 5 != 0`.

````python
def wrong_sum_only(n):
    s = sum(map(int, str(n)))
    return n % s == 0

print(wrong_sum_only(12), "!=", Solution().checkDivisibility(12))
assert wrong_sum_only(12) != Solution().checkDivisibility(12)  # True != False
````

### Lỗi 4: sai độ ưu tiên toán tử

`n % s + p == 0` được hiểu là `(n % s) + p == 0` chứ không phải `n % (s + p) == 0`.

````python
def wrong_precedence(n):
    digits = list(map(int, str(n)))
    s, p = sum(digits), math.prod(digits)
    return n % s + p == 0              # SAI: thiếu ngoặc

print(wrong_precedence(99), "!=", Solution().checkDivisibility(99))
assert wrong_precedence(99) != Solution().checkDivisibility(99)  # False != True
````

## 14. Mẹo làm bài nhanh & nhận dạng pattern

- "Tổng chữ số / tích chữ số / đảo chữ số" -> vòng `while n: n, d = divmod(n, 10)`.
- Luôn tự hỏi **phần tử trung hòa**: tổng `0`, tích `1`, `min` là `+inf`, `max` là `-inf`, AND là toàn bit `1`.
- Với chữ số `0` hãy nghĩ ngay "tích bằng `0`" — ca biên mà đề thích kiểm tra (`10`, `100`, `1000000`).
- Nhớ "mọi số một chữ số cho `False`" (vì `s + p = 2d > d`), dùng làm sanity check nhanh.
- Trước khi viết `%` hãy tự kiểm tra có thể chia cho `0` không; ở bài này **không** vì `s >= 1`.
- Dùng `math.prod` (Python 3.8+) cho gọn, nhưng khi cần tối ưu hoặc ôn thuật toán thì viết vòng lặp.

## 15. Biến thể của bài

1. **Harshad / Niven number** (LeetCode 3099): `n` chia hết cho tổng chữ số, trả tổng hoặc `-1`.
2. Chia hết cho **tích** chữ số (cần xử lý chữ số `0`).
3. Đếm bao nhiêu số trong `[1, N]` thỏa điều kiện (duyệt hết nếu `N` nhỏ, digit DP nếu `N` rất lớn).
4. Dùng **tổng bình phương** chữ số thay cho tổng (Happy Number, LeetCode 202).
5. Cơ số khác 10: tính tổng chữ số của `n` ở cơ số `b` (`divmod(n, b)`).

## 16. So sánh các cách làm

| Cách | Ý tưởng | Time | Space | Khi nào dùng |
|---|---|---|---|---|
| Chuỗi + `math.prod` | `str(n)`, `sum`, `prod` | `O(d)` | `O(d)` | Làm nhanh, dễ đọc |

Mọi cách chấp nhận được với `n <= 10^6`; chọn `divmod` để tránh đổi kiểu và dùng `O(1)` bộ nhớ.

## 19. Đáp án tham khảo cho bài tập

````python
# Bài 1
assert Solution().checkDivisibility(10) is True

# Bài 2: 36 -> s=9, p=18, t=27 ; 36 % 27 = 9
assert 36 % 27 == 9 and Solution().checkDivisibility(36) is False

# Bài 3
def digit_sum_product(n):
    s, p = 0, 1
    while n > 0:
        n, d = divmod(n, 10)
        s += d
        p *= d
    return s, p
assert digit_sum_product(99) == (18, 81)
assert digit_sum_product(23) == (5, 6)
assert digit_sum_product(104) == (5, 0)

# Bài 4
def is_harshad(n):
    s, _ = digit_sum_product(n)
    return n % s == 0
assert is_harshad(18) is True and is_harshad(19) is False and is_harshad(12) is True

# Bài 5
def count_good(lo, hi):
    return sum(1 for n in range(lo, hi + 1) if Solution().checkDivisibility(n))
assert count_good(1, 9) == 0
assert count_good(10, 99) == len([n for n in range(10, 100) if ref(n)])
assert Solution().checkDivisibility(99) and count_good(99, 99) == 1
print("đáp án bài tập OK")
````

x, s, p = n, 0, 1            # tổng bắt đầu 0, TÍCH bắt đầu 1
