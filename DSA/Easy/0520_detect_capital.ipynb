## 3. Cách 1 — Trực giác: so với 3 "phiên bản chuẩn"

Một từ đúng ⇔ nó **bằng** một trong ba bản chuẩn của chính nó: `upper()`, `lower()`, `capitalize()`.

````python
def detect_capital_compare(word):
    return word == word.upper() or word == word.lower() or word == word.capitalize()

for w in ["USA", "leetcode", "Google", "FlaG", "mL", "g", "G"]:
    print(f"{w:9s} -> {detect_capital_compare(w)}")
````

### Nút thắt nhỏ

Cách 1 rất dễ đọc nhưng **tạo tới 3 chuỗi mới** (mỗi chuỗi `O(n)` bộ nhớ) và quét từ 3–6 lượt. Với `n ≤ 100` hoàn toàn ổn, nhưng ta muốn tư duy được lời giải **một lượt, O(1) bộ nhớ** — kỹ năng quan trọng với chuỗi dài.

## 4. Insight — chỉ cần **đếm số chữ hoa**

Gọi `cap` = số chữ hoa, `n = len(word)`:
- Toàn hoa ⇔ `cap == n`.
- Toàn thường ⇔ `cap == 0`.
- Chỉ đầu hoa ⇔ `cap == 1` **và** `word[0]` là hoa.

Ba điều kiện trên mô tả **đúng** ba trường hợp đề cho (và chỉ chúng).

````python
w = "Google"
cap = sum(ch.isupper() for ch in w)
print("cap =", cap, "| n =", len(w), "| chữ đầu hoa?", w[0].isupper())
````

## 5. Cách 2 — Lời giải chính: đếm chữ hoa

Các đoạn:
1. `cap = 0`; duyệt từng ký tự, nếu hoa thì `cap += 1`.
2. `return cap == 0 or cap == len(word) or (cap == 1 and word[0].isupper())`.

````python
def detect_capital_count(word):
    cap = 0
    for ch in word:
        if ch.isupper():
            cap += 1
    return cap == 0 or cap == len(word) or (cap == 1 and word[0].isupper())

for w in ["USA", "leetcode", "Google", "FlaG", "mL", "g", "G"]:
    print(f"{w:9s} -> {detect_capital_count(w)}")
````

## 6. TRACE từng bước

`word = "FlaG"`:

| i | ch | hoa? | cap sau |
|---:|---|---|---:|
| 0 | F | ✔ | 1 |
| 1 | l | | 1 |
| 2 | a | | 1 |
| 3 | G | ✔ | 2 |

`cap = 2`: không bằng 0, không bằng 4, không bằng 1 → `False`.

`word = "mL"`: `cap = 1` nhưng `word[0] = 'm'` thường → `False`. Đây là lý do cần điều kiện "**và chữ đầu hoa**".

````python
def trace_detect_capital(word):
    cap = 0
    for i, ch in enumerate(word):
        if ch.isupper():
            cap += 1
        print(f"  i={i} ch='{ch}' hoa={ch.isupper()!s:5} cap={cap}")
    n = len(word)
    c1, c2, c3 = cap == n, cap == 0, cap == 1 and word[0].isupper()
    print(f"'{word}': toàn hoa={c1}, toàn thường={c2}, chỉ đầu hoa={c3} -> {c1 or c2 or c3}")
    return c1 or c2 or c3

trace_detect_capital("FlaG")
print()
trace_detect_capital("mL")
print()
trace_detect_capital("Google")
````

## 7. Cách 3 — Dùng method có sẵn

```python
word.isupper() or word.islower() or word.istitle()
```
Đúng với đề này **vì `word` chỉ gồm chữ cái và là một từ**. Lưu ý `istitle()` xét **từng từ** trong chuỗi, nên với chuỗi có khoảng trắng `"Hello World"` cũng trả `True` — một khác biệt nếu đề mở rộng.

````python
def detect_capital_builtin(word):
    return word.isupper() or word.islower() or word.istitle()

for w in ["USA", "leetcode", "Google", "FlaG", "mL", "g", "G"]:
    print(f"{w:9s} -> {detect_capital_builtin(w)}")
print("'Hello World'.istitle() =", "Hello World".istitle(), "(xét từng từ!)")
````

## 8. Cách 4 — Một lượt, dừng sớm: quy tắc "ký tự thứ 2 quyết định"

Quan sát:
- Nếu từ dài ≥ 2, **ký tự thứ 2** quyết định kiểu của phần còn lại: hoa thì mọi ký tự từ vị trí 1 trở đi phải hoa, thường thì phải thường.
- Nếu `word[0]` thường mà `word[1]` hoa (`"mL"`) → sai ngay.

Lợi ích: gặp ký tự sai là **dừng ngay** (không cần đếm hết).

````python
def detect_capital_scan(word):
    if len(word) < 2:
        return True
    if word[0].islower() and word[1].isupper():
        return False
    second_upper = word[1].isupper()
    for ch in word[2:]:
        if ch.isupper() != second_upper:
            return False
    return True

for w in ["USA", "leetcode", "Google", "FlaG", "mL", "g", "G"]:
    print(f"{w:9s} -> {detect_capital_scan(w)}")
````

`word[2:]` tạo chuỗi con mới; để đúng O(1) bộ nhớ có thể duyệt chỉ số `for i in range(2, len(word))`. Với Python và `n ≤ 100` thì khác biệt không đáng kể.

### Cách 5 (tham khảo) — Regex
`[A-Z]*|[a-z]*|[A-Z][a-z]*` với `re.fullmatch`.

````python
import re
CAP_RE = re.compile(r"[A-Z]*|[a-z]*|[A-Z][a-z]*")

def detect_capital_regex(word):
    return CAP_RE.fullmatch(word) is not None

for w in ["USA", "leetcode", "Google", "FlaG", "mL"]:
    print(f"{w:9s} -> {detect_capital_regex(w)}")
````

## 10. Vì sao đúng? (cách đếm)

Gọi `n = len(word)`, `cap` = số chữ hoa.
- Toàn hoa ⇔ không có chữ thường ⇔ `cap = n`.
- Toàn thường ⇔ `cap = 0`.
- Chỉ đầu hoa ⇔ `word[0]` hoa và `word[1:]` toàn thường ⇔ `word[0]` hoa và `cap = 1` (chữ hoa duy nhất chính là `word[0]`).

Ba vế **tương đương** ba trường hợp đúng của đề → hàm trả `True` đúng khi và chỉ khi cách dùng chữ hoa hợp lệ.

**Cách quét:** với `n ≥ 2`, ba trường hợp đúng tương ứng với: (hoa, hoa, hoa...), (thường, thường...), (hoa, thường, thường...). Trong cả ba, `word[1:]` đồng nhất kiểu chữ, và cặp (thường, hoa) ở đầu không bao giờ xuất hiện — đó chính là những gì vòng quét kiểm tra.

## 11. Độ phức tạp

`n = len(word)`.

| Cách | Time | Space phụ | Dừng sớm? |
|---|---:|---:|---|
| So với 3 bản chuẩn | `O(n)` | `O(n)` (chuỗi mới) | không |
| Đếm chữ hoa | `O(n)` | `O(1)` | không |
| Method có sẵn | `O(n)` | `O(1)` | có (bên trong C) |
| Quét quy tắc ký tự thứ 2 | `O(n)` | `O(1)` | có |
| Regex | `O(n)` | `O(1)` | có |

"Dừng sớm" khác nhau thế nào trong thực tế? Ta đếm số ký tự mà mỗi cách phải đọc với một từ **sai ngay từ đầu** nhưng rất dài:

````python
def visits_count(word):
    visits = 0
    for ch in word:            # luôn đọc hết
        visits += 1
    return visits

def visits_scan(word):
    if len(word) < 2:
        return 1
    visits = 2
    if word[0].islower() and word[1].isupper():
        return visits          # sai ngay ở 2 ký tự đầu
    second_upper = word[1].isupper()
    for ch in word[2:]:
        visits += 1
        if ch.isupper() != second_upper:
            break
    return visits

long_bad = "mL" + "x" * 98          # 100 ký tự
long_bad2 = "Ab" + "C" + "d" * 97
print("cách đếm đọc:", visits_count(long_bad), "ký tự | cách quét đọc:", visits_scan(long_bad), "ký tự")
print("cách đếm đọc:", visits_count(long_bad2), "ký tự | cách quét đọc:", visits_scan(long_bad2), "ký tự")
````

Kết luận: cùng `O(n)` trong trường hợp xấu nhất, nhưng cách quét có thể dừng rất sớm với input sai.

## 12. Bộ test — kể cả **vét cạn** mọi chuỗi ngắn

Với chuỗi chỉ gồm 2 chữ cái `{a, A}` hoặc `{a, B}`... kiểu chữ là thứ duy nhất quan trọng. Ta sinh **mọi** chuỗi độ dài 1..8 trên bảng chữ `"aB"` (2⁸ = 256 chuỗi với độ dài 8) và so sánh tất cả các cách với định nghĩa gốc.

````python
def check(name, actual, expected):
    assert actual == expected, (name, actual, expected)
    print(f"PASS {name}: {actual}")

s = Solution()
check("đề 1 USA", s.detectCapitalUse("USA"), True)
check("đề 2 FlaG", s.detectCapitalUse("FlaG"), False)
check("toàn thường", s.detectCapitalUse("leetcode"), True)
check("chỉ đầu hoa", s.detectCapitalUse("Google"), True)
check("1 ký tự thường", s.detectCapitalUse("g"), True)
check("1 ký tự hoa", s.detectCapitalUse("G"), True)
check("thường rồi hoa", s.detectCapitalUse("mL"), False)
check("hoa hoa thường", s.detectCapitalUse("GOogle"), False)
check("hoa cuối", s.detectCapitalUse("googlE"), False)
````

````python
from itertools import product

def by_definition(word):
    all_up = all(c.isupper() for c in word)
    all_low = all(c.islower() for c in word)
    title = word[0].isupper() and all(c.islower() for c in word[1:])
    return all_up or all_low or title

impls = [detect_capital_compare, detect_capital_count, detect_capital_builtin,
         detect_capital_scan, detect_capital_regex, Solution().detectCapitalUse]
total = 0
for n in range(1, 9):
    for tup in product("aB", repeat=n):     # mọi tổ hợp hoa/thường độ dài n
        w = "".join(tup)
        exp = by_definition(w)
        assert all(f(w) == exp for f in impls), w
        total += 1
print(f"vét cạn {total} chuỗi (độ dài 1..8): 6 cài đặt khớp định nghĩa — PASS")
````

`itertools.product("aB", repeat=n)` sinh mọi bộ `n` ký tự lấy từ `"aB"` (tích Descartes) — công cụ rất tiện để **vét cạn** input nhỏ khi kiểm thử.

## 13. Lỗi người mới hay gặp

### Lỗi 1: chỉ kiểm tra `cap == 1` mà quên "chữ đầu"
`"mL"` có `cap = 1` nhưng sai.

### Lỗi 2: quên trường hợp từ 1 ký tự trong cách quét
Truy cập `word[1]` với `"g"` → `IndexError`.

### Lỗi 3: dùng `istitle()` một mình
`"USA".istitle()` là `False` → phải kết hợp với `isupper()`/`islower()`.

### Lỗi 4: nhầm `isupper()` của **chuỗi** với của **ký tự**
`"Google".isupper()` hỏi "toàn bộ có hoa không" (False), khác với `"Google"[0].isupper()` (True).

````python
def buggy_forget_first(word):
    cap = sum(ch.isupper() for ch in word)
    return cap == 0 or cap == len(word) or cap == 1

def buggy_no_len_check(word):
    if word[0].islower() and word[1].isupper():
        return False
    return all(ch.isupper() == word[1].isupper() for ch in word[2:])

print("quên chữ đầu  : 'mL' ->", buggy_forget_first("mL"), "| đúng:", Solution().detectCapitalUse("mL"))
try:
    buggy_no_len_check("g")
except IndexError as e:
    print("không xét len : 'g' -> IndexError:", e)
print("chỉ istitle   : 'USA' ->", "USA".istitle(), "| đúng: True")
print("chuỗi vs ký tự:", "Google".isupper(), "Google"[0].isupper())
````

## 14. Mẹo nhận dạng pattern

- Đề liệt kê **vài trường hợp hợp lệ** → viết mỗi trường hợp thành một biểu thức boolean, nối bằng `or`; kiểm tra 3 biểu thức có **phủ đúng** đề không bằng bảng phân loại.
- Muốn kiểm tra "mọi phần tử thoả" → `all(...)`; "đếm số phần tử thoả" → `sum(...)` trên boolean.
- Bài chuỗi nhỏ, bảng chữ hẹp → **vét cạn bằng `itertools.product`** để kiểm thử.
- Cần dừng sớm/một lượt → tìm "phần tử quyết định" (ở đây là ký tự thứ 2).

## 15. Biến thể

### 15.1. Sửa từ về dạng hợp lệ gần nhất
Nếu từ sai, chuyển về: toàn hoa nếu đa số là hoa, ngược lại `capitalize` nếu chữ đầu hoa, còn lại `lower`.

````python
def fix_capital(word):
    if Solution().detectCapitalUse(word):
        return word
    cap = sum(ch.isupper() for ch in word)
    if cap * 2 > len(word):
        return word.upper()
    return word.capitalize() if word[0].isupper() else word.lower()

print([fix_capital(w) for w in ["FlaG", "mL", "GOOgLE", "Google"]])
````

### 15.2. Kiểm tra cả một câu
Mỗi từ (tách bằng `split()`) phải hợp lệ.

````python
def sentence_ok(sentence):
    return all(Solution().detectCapitalUse(w) for w in sentence.split())

print(sentence_ok("I love NASA and Google"), sentence_ok("I lOve python"))
````
