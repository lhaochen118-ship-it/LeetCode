## 1. Mục tiêu học

Sau notebook này bạn sẽ:

1. Hiểu đề: chia các phần tử của `nums` vào hai mảng `arr1` và `arr2` theo **luật so sánh phần tử cuối**, rồi **nối `arr1` với `arr2`** làm kết quả.
2. Nắm công cụ Python: `append`, chỉ số âm `arr[-1]` (phần tử cuối), nối list bằng `+` / `extend`, cắt lát `nums[2:]`.

5. Tránh các lỗi kinh điển: bắt đầu vòng lặp sai chỗ (đưa lại `nums[0]`, `nums[1]`), so sánh **phần tử đầu** thay vì **phần tử cuối**, so sánh nhầm giá trị mới `nums[i]`, và quên nối **`arr1` rồi mới tới `arr2`**.

## 2. Đề bài & ví dụ

Cho mảng `nums` gồm các số nguyên **đôi một khác nhau**, **đánh số từ 1**, có độ dài `n`. Ta phân phối các phần tử vào hai mảng `arr1` và `arr2` qua `n` thao tác:

- Thao tác 1: thêm `nums[1]` vào `arr1`.
- Thao tác 2: thêm `nums[2]` vào `arr2`.
- Thao tác thứ `i` (với `i >= 3`): nếu **phần tử cuối của `arr1` lớn hơn phần tử cuối của `arr2`**, thêm `nums[i]` vào `arr1`; ngược lại, thêm `nums[i]` vào `arr2`.

Mảng kết quả `result` là **`arr1` nối tiếp `arr2`**. Hãy trả về `result`.

(Lưu ý: đề đánh số từ 1; trong code Python `nums` đánh số từ 0, nên `nums[0]` ↔ "`nums[1]` của đề".)

**Ví dụ 1**

- Input: `nums = [2, 1, 3]`
- Output: `[2, 3, 1]`
- Giải thích: sau hai thao tác đầu `arr1 = [2]`, `arr2 = [1]`. Thao tác 3: phần tử cuối `arr1` là `2 > 1` (cuối `arr2`) nên thêm `3` vào `arr1`: `arr1 = [2, 3]`, `arr2 = [1]`. Kết quả `[2, 3, 1]`.

**Ví dụ 2**

- Input: `nums = [5, 4, 3, 8]`
- Output: `[5, 3, 4, 8]`
- Giải thích: sau hai thao tác đầu `arr1 = [5]`, `arr2 = [4]`. Thao tác 3: `5 > 4` nên `3` vào `arr1`: `arr1 = [5, 3]`. Thao tác 4: cuối `arr1` là `3`, cuối `arr2` là `4`, `3 > 4` sai nên `8` vào `arr2`: `arr2 = [4, 8]`. Kết quả `[5, 3, 4, 8]`.

**Ràng buộc:** `3 <= nums.length <= 50`; `1 <= nums[i] <= 100`; mọi phần tử của `nums` **khác nhau**.

````python
from typing import List
import random

examples = [
    ([2, 1, 3], [2, 3, 1]),
    ([5, 4, 3, 8], [5, 3, 4, 8]),
    ([3, 1, 2, 4], [3, 2, 4, 1]),
]
for nums, expected in examples:
    print("nums =", nums, "-> kỳ vọng", expected)
````

## 3. Kiến thức nền

### 3.1. `append` và chỉ số âm `arr[-1]`

- `arr.append(x)` thêm `x` vào **cuối** list.
- `arr[-1]` là **phần tử cuối cùng** (chỉ số âm đếm từ cuối: `-1` là cuối, `-2` là kề cuối). Nếu list rỗng thì `arr[-1]` báo `IndexError` — bài này luôn có sẵn một phần tử trong mỗi mảng trước khi so sánh nên an toàn.

````python
arr = [2]
arr.append(3)
arr.append(7)
print(arr)          # [2, 3, 7]
print(arr[-1])      # 7  (phần tử cuối)
print(arr[-2])      # 3
try:
    [][-1]
except IndexError as e:
    print("IndexError:", e)
````

### 3.2. Đánh số từ 1 (đề) và từ 0 (code)

Đề viết "thao tác 1, 2, 3, ..." và "`nums[1]`, `nums[2]`". Python đánh số từ `0`, nên:

| theo đề | thao tác 1 | thao tác 2 | thao tác 3 | thao tác `i` |
|---|---|---|---|---|
| theo code | `nums[0]` | `nums[1]` | `nums[2]` | `nums[i - 1]` |

Vì vậy vòng lặp "thao tác thứ `i >= 3`" trong code chạy với **chỉ số từ `2`** trở đi (`range(2, n)`).

````python
nums = [5, 4, 3, 8]
print("nums[0] =", nums[0], "(thao tác 1 -> arr1)")
print("nums[1] =", nums[1], "(thao tác 2 -> arr2)")
print("phần còn lại để so sánh:", list(range(2, len(nums))), "->", nums[2:])
````

### 3.3. Nối hai list: `+` và `extend`

- `a + b` tạo **list mới** gồm `a` rồi `b`.
- `a.extend(b)` **thêm** mọi phần tử của `b` vào cuối `a` (sửa `a` tại chỗ).

Thứ tự quan trọng: kết quả phải là `arr1 + arr2`, **không phải** `arr2 + arr1`.

````python
arr1, arr2 = [2, 3], [1]
print(arr1 + arr2)       # [2, 3, 1]
print(arr2 + arr1)       # [1, 2, 3]  (sai thứ tự!)
c = list(arr1)
c.extend(arr2)
print(c)                 # [2, 3, 1]
````

### 3.4. Cắt lát `nums[2:]` và `enumerate`

`nums[2:]` là list con **từ chỉ số 2 tới hết** (bỏ hai phần tử đầu). Ta duyệt các phần tử còn lại mà không đụng tới hai phần tử đầu, rất hợp cho bước "thao tác 3 trở đi". Ngoài ra `enumerate(nums)` cho từng cặp `(chỉ số, giá trị)`.

````python
nums = [5, 4, 3, 8]
for x in nums[2:]:
    print("xử lý", x)
for i, x in enumerate(nums):
    print(i, x)
````

## 4. Ý tưởng nền tảng / cách nghĩ trực giác (tính tay)

Lấy `nums = [5, 4, 3, 8, 2]`. Ta ghi lại từng thao tác:

| thao tác | phần tử | quy tắc | arr1 | arr2 |
|---|---|---|---|---|
| 1 | 5 | luôn vào arr1 | [5] | [] |
| 2 | 4 | luôn vào arr2 | [5] | [4] |
| 3 | 3 | cuối arr1 = 5 > cuối arr2 = 4 ⇒ arr1 | [5, 3] | [4] |
| 4 | 8 | cuối arr1 = 3 > 4? không ⇒ arr2 | [5, 3] | [4, 8] |
| 5 | 2 | cuối arr1 = 3 > cuối arr2 = 8? không ⇒ arr2 | [5, 3] | [4, 8, 2] |

Kết quả `[5, 3] + [4, 8, 2] = [5, 3, 4, 8, 2]`.

Điều cần để ý: **quyết định chỉ phụ thuộc vào hai phần tử cuối hiện tại của hai mảng**. Mỗi lần thêm phần tử, phần tử cuối của mảng đó đổi. Đề cho các số đều khác nhau nên hai phần tử cuối **không bao giờ bằng nhau** (không có chuyện hòa).

````python
nums = [5, 4, 3, 8, 2]
arr1, arr2 = [nums[0]], [nums[1]]
for x in nums[2:]:
    if arr1[-1] > arr2[-1]:
        arr1.append(x)
    else:
        arr2.append(x)
    print(f"x={x}: arr1={arr1} arr2={arr2}")
print(arr1 + arr2)
````

## 5. Cách 1 — Brute Force / baseline: mô phỏng với hai list

**Ý tưởng.** Làm **đúng từng chữ** của đề: tạo `arr1`, `arr2`; đặt hai phần tử đầu; với mỗi phần tử còn lại so sánh `arr1[-1]` và `arr2[-1]`, thêm vào mảng tương ứng; cuối cùng nối `arr1 + arr2`.

Cách này vừa là lời giải vừa là "bản đúng đề" để kiểm chứng các cách khác.

````python
def result_array_simulate(nums: List[int]) -> List[int]:
    arr1 = [nums[0]]                      # thao tác 1
    arr2 = [nums[1]]                      # thao tác 2
    for i in range(2, len(nums)):         # thao tác 3.. n
        if arr1[-1] > arr2[-1]:
            arr1.append(nums[i])
        else:
            arr2.append(nums[i])
    return arr1 + arr2


for nums, expected in examples:
    print(result_array_simulate(nums), "(kỳ vọng", expected, ")")
````

**Độ phức tạp (Cách 1).** Mỗi phần tử được xử lý một lần với `O(1)` thao tác ⇒ **Time `O(n)`**. Hai list con lưu lại toàn bộ phần tử ⇒ **Space `O(n)`**, cũng là kích thước kết quả nên khó tránh.

**Nút thắt:** về thời gian không còn gì để cải thiện (phải đọc mọi phần tử). Ta chỉ có thể làm code **gọn hơn** hoặc **tiết kiệm bộ nhớ phụ**.

## 6. Ý tưởng thuật toán / bước đột phá

Để **quyết định** thêm phần tử vào mảng nào, ta chỉ cần **hai phần tử cuối** của hai mảng. Do đó ta có thể:

- Giữ hai biến `last1`, `last2` (phần tử cuối hiện tại) thay vì đọc `arr[-1]` mỗi lần (ý nghĩa như nhau, ghi rõ hơn).
- **Tách hai việc:** (1) quyết định nhãn của mỗi phần tử (`1` hay `2`) chỉ cần `last1`, `last2`; (2) dựng kết quả bằng cách lấy phần tử có nhãn `1` rồi phần tử có nhãn `2` theo thứ tự ban đầu.

Cách nhìn "gán nhãn rồi lọc" cho thấy kết quả là **một hoán vị ổn định (stable partition)** của `nums`: giữ nguyên thứ tự tương đối trong mỗi nhóm.

````python
nums = [5, 4, 3, 8, 2]
labels = [1, 2]                         # nhãn của 2 phần tử đầu
last1, last2 = nums[0], nums[1]
for x in nums[2:]:
    if last1 > last2:
        labels.append(1)
        last1 = x
    else:
        labels.append(2)
        last2 = x
print("nhãn:", labels)                  # [1, 2, 1, 2, 2]
print("arr1:", [x for x, t in zip(nums, labels) if t == 1])
print("arr2:", [x for x, t in zip(nums, labels) if t == 2])
````

## 7. Các cách tiếp theo

### Cách 2 — Hai biến `last1`, `last2` (không đọc `arr[-1]`)

**Giải thích.** Ta vẫn có `arr1`, `arr2` để dựng kết quả, nhưng quyết định dựa vào hai biến `last1`, `last2`. Mỗi lần thêm phần tử vào một mảng, **cập nhật biến cuối tương ứng**.

````python
def result_array_last(nums: List[int]) -> List[int]:
    arr1, arr2 = [nums[0]], [nums[1]]
    last1, last2 = nums[0], nums[1]
    for x in nums[2:]:
        if last1 > last2:
            arr1.append(x)
            last1 = x                  # cập nhật phần tử cuối của arr1
        else:
            arr2.append(x)
            last2 = x                  # cập nhật phần tử cuối của arr2
    return arr1 + arr2


for nums, expected in examples:
    print(result_array_last(nums), "(kỳ vọng", expected, ")")
````

````python
# Code nộp LeetCode cho Cách 2 (đổi tên lớp thành Solution khi nộp)
class SolutionLast:
    def resultArray(self, nums: List[int]) -> List[int]:
        arr1, arr2 = [nums[0]], [nums[1]]
        last1, last2 = nums[0], nums[1]
        for x in nums[2:]:
            if last1 > last2:
                arr1.append(x)
                last1 = x
            else:
                arr2.append(x)
                last2 = x
        return arr1 + arr2


print(SolutionLast().resultArray([5, 4, 3, 8]))        # [5, 3, 4, 8]
````

**Độ phức tạp Cách 2:** **Time `O(n)`**, **Space `O(n)`**. Ưu điểm: so sánh hai biến nhanh và ý nghĩa rõ ràng.

### Cách 3 — Gán nhãn từng phần tử, rồi lọc hai lần

**Giải thích.** Bước 1: tạo mảng `labels` (độ dài `n`): `labels[0] = 1`, `labels[1] = 2`, các nhãn còn lại tính theo `last1`, `last2`. Bước 2: kết quả = các phần tử có nhãn `1` (giữ thứ tự) rồi các phần tử có nhãn `2`. Cách này tách rõ "quyết định" và "dựng kết quả".

````python
def result_array_labels(nums: List[int]) -> List[int]:
    n = len(nums)
    labels = [0] * n
    labels[0], labels[1] = 1, 2
    last1, last2 = nums[0], nums[1]
    for i in range(2, n):
        if last1 > last2:
            labels[i] = 1
            last1 = nums[i]
        else:
            labels[i] = 2
            last2 = nums[i]
    first = [nums[i] for i in range(n) if labels[i] == 1]
    second = [nums[i] for i in range(n) if labels[i] == 2]
    return first + second


for nums, expected in examples:
    print(result_array_labels(nums), "(kỳ vọng", expected, ")")
````

**Độ phức tạp Cách 3:** **Time `O(n)`** (ba lượt duyệt vẫn là `O(n)`), **Space `O(n)`** cho `labels` và hai list lọc (nhiều bộ nhớ phụ hơn, nhưng có ích khi cần biết phần tử nào thuộc mảng nào).

### Cách 4 — Dùng `extend` để dựng kết quả một list

**Giải thích.** Thay vì `arr1 + arr2` (tạo list mới), ta cho `arr1.extend(arr2)` rồi trả `arr1`. Hai cách cho cùng kết quả; `extend` tiết kiệm một list tạm. Cách này nhấn mạnh sự khác nhau giữa `+` (không sửa hai list gốc) và `extend` (sửa list bên trái).

````python
def result_array_extend(nums: List[int]) -> List[int]:
    arr1, arr2 = [nums[0]], [nums[1]]
    for x in nums[2:]:
        (arr1 if arr1[-1] > arr2[-1] else arr2).append(x)     # chọn mảng rồi thêm
    arr1.extend(arr2)                                         # nối arr2 vào cuối arr1
    return arr1


for nums, expected in examples:
    print(result_array_extend(nums), "(kỳ vọng", expected, ")")
````

**Độ phức tạp Cách 4:** **Time `O(n)`**, **Space `O(n)`**. Dòng `(arr1 if ... else arr2).append(x)` chọn **đối tượng list** bằng biểu thức điều kiện rồi gọi `append` — gọn nhưng đòi hỏi bạn hiểu rằng `append` tác động lên list được chọn (không phải bản sao).

## 8. TRACE từng bước

Ta trace Cách 2 trên `nums = [5, 4, 3, 8, 2]` và một ví dụ có đổi hướng nhiều lần `nums = [10, 3, 7, 1, 9, 2]`.

````python
def trace_result_array(nums: List[int]) -> List[int]:
    arr1, arr2 = [nums[0]], [nums[1]]
    last1, last2 = nums[0], nums[1]
    print("nums =", nums)
    print(f"khởi tạo: arr1={arr1} arr2={arr2}")
    for i in range(2, len(nums)):
        x = nums[i]
        if last1 > last2:
            where = "arr1"
            arr1.append(x)
            last1 = x
        else:
            where = "arr2"
            arr2.append(x)
            last2 = x
        print(f"i={i} x={x}: so {'last1 > last2' if where == 'arr1' else 'last1 <= last2'} -> {where} | last1={last1} last2={last2} | arr1={arr1} arr2={arr2}")
    result = arr1 + arr2
    print("kết quả:", result)
    return result


assert trace_result_array([5, 4, 3, 8, 2]) == [5, 3, 4, 8, 2]
print()
assert trace_result_array([10, 3, 7, 1, 9, 2]) == result_array_simulate([10, 3, 7, 1, 9, 2])
````

Bảng trace với `nums = [5, 4, 3, 8, 2]` (cột `last1`, `last2` là trước khi quyết định):

| i | x | last1 | last2 | so sánh | vào mảng |
|---|---|---|---|---|---|
| 2 | 3 | 5 | 4 | 5 > 4 | arr1 (last1 = 3) |
| 3 | 8 | 3 | 4 | 3 > 4 sai | arr2 (last2 = 8) |
| 4 | 2 | 3 | 8 | 3 > 8 sai | arr2 (last2 = 2) |

`arr1 = [5, 3]`, `arr2 = [4, 8, 2]` ⇒ kết quả `[5, 3, 4, 8, 2]`.

## 9. Code nộp LeetCode

Bản nộp dùng Cách 1/2 gộp: hai list và đọc `arr[-1]` trực tiếp (rất ngắn, dễ đọc).

````python
from typing import List

class Solution:
    def resultArray(self, nums: List[int]) -> List[int]:
        arr1 = [nums[0]]                       # thao tác 1
        arr2 = [nums[1]]                       # thao tác 2
        for i in range(2, len(nums)):          # thao tác 3 tới n (chỉ số code từ 2)
            if arr1[-1] > arr2[-1]:
                arr1.append(nums[i])
            else:
                arr2.append(nums[i])
        return arr1 + arr2                     # arr1 rồi tới arr2


sol = Solution()
for nums, expected in examples:
    print(nums, "->", sol.resultArray(nums))
````

## 10. Vì sao thuật toán đúng

- **Bám đúng luật:** hai thao tác đầu cố định (`nums[0] → arr1`, `nums[1] → arr2`); mỗi thao tác sau lấy đúng phần tử kế tiếp và quyết định theo phần tử cuối **hiện tại** của hai mảng.
- **An toàn chỉ số:** sau hai thao tác đầu cả hai mảng đều không rỗng, nên `arr[-1]` luôn hợp lệ; `n >= 3` đảm bảo có `nums[1]`.
- **Không có hòa:** các phần tử khác nhau ⇒ hai phần tử cuối khác nhau (chúng là hai phần tử khác vị trí). Do đó nhánh `else` thật sự chỉ xảy ra khi `arr1[-1] < arr2[-1]`, phù hợp đề ("ngược lại").
- **Kết quả:** nối `arr1` rồi `arr2` đúng định nghĩa `result`. Cả bốn cách đều cùng tính nhãn, chỉ khác cách lưu/dựng, nên cho cùng kết quả (kiểm tra ngẫu nhiên ở §12.3).

## 11. Phân tích Time / Space Complexity

Gọi `n = len(nums)`.

| Cách | Time | Space |
|---|---|---|
| 1. Mô phỏng, đọc `arr[-1]` | `O(n)` | `O(n)` |
| 2. Hai biến `last1`, `last2` | `O(n)` | `O(n)` |
| 3. Gán nhãn rồi lọc | `O(n)` | `O(n)` (nhiều hằng số hơn) |

Phải đọc mọi phần tử nên `O(n)` thời gian là tối ưu. Bộ nhớ `O(n)` là bộ nhớ của kết quả.

### 12.2. Phản ví dụ cho các cách làm sai

Hai lỗi phổ biến: (a) **so sánh phần tử đầu** (`arr1[0]`, `arr2[0]`) thay vì phần tử cuối; (b) **nối ngược** `arr2 + arr1`. Ta tạo hàm sai rồi tìm input phân biệt.

## 13. Các lỗi người mới hay gặp

### Lỗi 1: Vòng lặp bắt đầu sai chỗ (đưa lại hai phần tử đầu)

````python
def sai_bat_dau_vong_lap(nums):
    arr1, arr2 = [nums[0]], [nums[1]]
    for x in nums:                      # SAI: nums[0], nums[1] bị thêm lần nữa
        if arr1[-1] > arr2[-1]:
            arr1.append(x)
        else:
            arr2.append(x)
    return arr1 + arr2


print("sai:", sai_bat_dau_vong_lap([2, 1, 3]), "| đúng:", Solution().resultArray([2, 1, 3]))
````

**Vì sao sai:** hai phần tử đầu đã được đặt vào `arr1`, `arr2` rồi; vòng lặp phải bắt đầu từ chỉ số `2` (`nums[2:]` hay `range(2, n)`). Sai ở trên làm kết quả dài hơn `n`. **Cách sửa:** `for i in range(2, len(nums))`.

### Lỗi 2: So sánh phần tử đầu thay vì phần tử cuối

````python
def sai_so_phan_tu_dau(nums):
    arr1, arr2 = [nums[0]], [nums[1]]
    for x in nums[2:]:
        if arr1[0] > arr2[0]:           # SAI
            arr1.append(x)
        else:
            arr2.append(x)
    return arr1 + arr2


print("sai:", sai_so_phan_tu_dau([5, 4, 3, 8]), "| đúng:", Solution().resultArray([5, 4, 3, 8]))
````

**Vì sao sai:** đề so **phần tử cuối** của mỗi mảng; `arr[0]` không đổi nên quyết định sẽ luôn giống nhau sau mỗi lần. **Cách sửa:** dùng `arr[-1]` (hoặc biến `last`).

### Lỗi 3: Nối sai thứ tự hoặc quên nối

````python
def sai_noi_nguoc(nums):
    arr1, arr2 = [nums[0]], [nums[1]]
    for x in nums[2:]:
        (arr1 if arr1[-1] > arr2[-1] else arr2).append(x)
    return arr2 + arr1                  # SAI


print("sai:", sai_noi_nguoc([2, 1, 3]), "| đúng:", Solution().resultArray([2, 1, 3]))
````

**Vì sao sai:** đề yêu cầu `result` là `arr1` **rồi** `arr2`. Nối ngược cho `[1, 2, 3]` thay vì `[2, 3, 1]`. **Cách sửa:** `arr1 + arr2`.

### Lỗi 4: Lệch một vị trí vì đánh số từ 1 của đề

````python
def sai_lech_chi_so(nums):
    arr1, arr2 = [nums[1]], [nums[2]]   # SAI: dùng chỉ số theo đề thay vì chỉ số Python
    for x in nums[3:]:
        if arr1[-1] > arr2[-1]:
            arr1.append(x)
        else:
            arr2.append(x)
    return arr1 + arr2


print("sai:", sai_lech_chi_so([5, 4, 3, 8]), "| đúng:", Solution().resultArray([5, 4, 3, 8]))
````

**Vì sao sai:** đề đánh số từ **1** nên "`nums[1]`" là phần tử đầu tiên, tức `nums[0]` trong Python. Dùng `nums[1]` làm phần tử đầu sẽ bỏ sót phần tử thứ nhất. **Cách sửa:** `arr1 = [nums[0]]`, `arr2 = [nums[1]]`, vòng lặp từ chỉ số `2`.

## 14. Mẹo làm bài nhanh & nhận dạng pattern

- Bài "**mô phỏng**" (simulation): làm đúng từng bước theo đề; chọn cấu trúc dữ liệu đơn giản nhất (hai list).
- Khi quyết định chỉ phụ thuộc vào "phần tử cuối", giữ ngay biến `last1`, `last2` cho rõ.
- **Luôn kiểm tra cách đánh số:** đề từ 1, Python từ 0. Viết ví dụ tay ra rồi đối chiếu với code.
- Dòng `(a if cond else b).append(x)` là mẹo chọn list đích; dùng được nếu bạn hiểu `append` tác động lên chính list được chọn.
- Kết quả là "nối hai list": nhớ `+` tạo list mới, `extend` sửa tại chỗ.

## 15. Biến thể của bài

1. **Distribute Elements Into Two Arrays II (3072):** so sánh theo **số phần tử lớn hơn giá trị mới** (`greaterCount`), cần cấu trúc như Fenwick Tree (BIT) hoặc `SortedList`.
2. **Trả về riêng `(arr1, arr2)`** thay vì nối.
3. **Hai mảng quyết định theo phần tử cuối nhỏ hơn** (đảo dấu so sánh).
4. **Phân phối vào ba mảng**: thêm luật với `arr3`.
5. **Đếm số phần tử vào `arr1`** mà không dựng list (chỉ giữ `last1`, `last2`, đếm).

````python
def split_arrays(nums):
    arr1, arr2 = [nums[0]], [nums[1]]
    for x in nums[2:]:
        (arr1 if arr1[-1] > arr2[-1] else arr2).append(x)
    return arr1, arr2


def reversed_rule(nums):
    """Luật đảo: vào arr1 nếu cuối arr1 NHỎ hơn cuối arr2."""
    arr1, arr2 = [nums[0]], [nums[1]]
    for x in nums[2:]:
        (arr1 if arr1[-1] < arr2[-1] else arr2).append(x)
    return arr1 + arr2


def count_in_arr1(nums):
    last1, last2, cnt = nums[0], nums[1], 1
    for x in nums[2:]:
        if last1 > last2:
            last1, cnt = x, cnt + 1
        else:
            last2 = x
    return cnt


print(split_arrays([5, 4, 3, 8]))       # ([5, 3], [4, 8])
print(reversed_rule([5, 4, 3, 8]))
print(count_in_arr1([5, 4, 3, 8]))      # 2
assert split_arrays([5, 4, 3, 8]) == ([5, 3], [4, 8])
rng = random.Random(5)
for _ in range(300):
    nums = rng.sample(range(1, 101), rng.randint(3, 30))
    assert count_in_arr1(nums) == len(split_arrays(nums)[0])
````
