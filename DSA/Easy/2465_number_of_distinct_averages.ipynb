## 1. Mục tiêu học

Sau notebook này bạn sẽ:

1. Hiểu quy trình đề cho: lặp lại "lấy số nhỏ nhất và số lớn nhất, tính trung bình của chúng, rồi bỏ hai số đó" cho tới khi hết mảng; đếm có bao nhiêu **trung bình khác nhau**.
2. Thấy rằng sau khi **sắp xếp**, cặp (min, max) tại mỗi bước chính là cặp `(nums[i], nums[n - 1 - i])` — bài toán đổi thành **hai con trỏ** từ hai đầu.
3. Dùng `set` để đếm giá trị khác nhau, và mẹo **so sánh tổng thay vì trung bình** (tránh số thực): `(a + b) / 2` khác nhau ⇔ `a + b` khác nhau.
4. Viết 4 cách: mô phỏng bằng `min`/`max`/`remove`, sắp xếp + hai con trỏ, sắp xếp + chỉ số đối xứng `nums[~i]`, và mảng đếm tần suất (không cần sắp xếp); đối chiếu bằng test ngẫu nhiên.
5. Tránh các lỗi: quên sắp xếp, đếm cả cặp trùng (dùng `list` thay `set`), làm tròn trung bình về số nguyên (mất `.5`).

## 2. Đề bài & ví dụ

Cho mảng số nguyên `nums` có **độ dài chẵn**. Lặp lại các bước sau cho tới khi `nums` rỗng:

1. Tìm số **nhỏ nhất** `minNum` và số **lớn nhất** `maxNum` trong `nums` rồi **xoá cả hai** khỏi `nums`.
2. Tính **trung bình** của hai số đó: `(minNum + maxNum) / 2`.

Trả về **số lượng trung bình khác nhau** thu được.

*Lưu ý:* nếu có nhiều số cùng giá trị nhỏ nhất (hoặc lớn nhất), chỉ xoá **một** số trong đó.

| Ví dụ | nums | Output | Giải thích |
|---|---|---|---|
| 1 | `[4,1,4,0,3,5]` | 2 | bước 1: `(0+5)/2 = 2.5`; bước 2: `(1+4)/2 = 2.5`; bước 3: `(3+4)/2 = 3.5` → `{2.5, 3.5}` |
| 2 | `[1,100]` | 1 | chỉ một bước: `50.5` |

Constraints: `2 <= nums.length <= 100`; `nums.length` là **số chẵn**; `0 <= nums[i] <= 100`.

````python
nums = [4, 1, 4, 0, 3, 5]
work = list(nums)
step = 0
while work:
    step += 1
    lo, hi = min(work), max(work)
    work.remove(lo)
    work.remove(hi)
    print(f"bước {step}: min={lo} max={hi} trung bình={(lo + hi) / 2} | còn lại {work}")
````

## 3. Kiến thức nền

### 3.1. `min`, `max` và `list.remove`

`min(a)`, `max(a)` cho giá trị nhỏ nhất/lớn nhất (`O(n)`). `a.remove(x)` xoá **một** phần tử đầu tiên bằng `x` (`O(n)`); nếu có nhiều phần tử trùng, chỉ xoá một — đúng như lưu ý của đề.

````python
a = [4, 1, 4, 0, 3, 5]
a.remove(4)
print(a)
a.remove(max(a))
print(a)
````

### 3.2. `set` — đếm giá trị khác nhau

`set` chỉ giữ mỗi giá trị **một lần**. `len(set(x))` cho số giá trị phân biệt. Thêm phần tử bằng `s.add(v)`; thêm phần tử đã có không làm tăng kích thước.

````python
s = set()
for v in [5, 5, 6, 5, 7]:
    s.add(v)
    print("thêm", v, "->", s, "| kích thước", len(s))
````

### 3.3. Sắp xếp rồi lấy hai đầu: `sorted`, chỉ số `-1` và `~i`

Sau khi sắp xếp tăng dần, `s[0]` nhỏ nhất, `s[-1]` lớn nhất. Chỉ số đối xứng của `i` là `n - 1 - i`, và Python có cách viết gọn là **`~i`** (phủ định bit, `~i == -i - 1`), nên `s[~i] == s[n - 1 - i]`.

````python
s = sorted([4, 1, 4, 0, 3, 5])
n = len(s)
print(s, "n =", n)
for i in range(n // 2):
    print(f"i={i}: s[i]={s[i]}, s[n-1-i]={s[n - 1 - i]}, s[~i]={s[~i]}")
print(~0, ~1, ~2)
````

### 3.4. Hai con trỏ từ hai đầu

`lo = 0`, `hi = n - 1`; lặp trong khi `lo < hi`: xử lý cặp `(s[lo], s[hi])` rồi `lo += 1`, `hi -= 1`. Vì `n` chẵn, hai con trỏ gặp nhau vừa khít sau `n / 2` bước, không bao giờ còn một phần tử lẻ.

````python
s = sorted([4, 1, 4, 0, 3, 5])
lo, hi = 0, len(s) - 1
while lo < hi:
    print(f"cặp ({s[lo]}, {s[hi]}), tổng {s[lo] + s[hi]}")
    lo += 1
    hi -= 1
````

### 3.5. So sánh tổng thay cho trung bình

`(a + b) / 2` là số thực (có thể `.5`). Hai trung bình bằng nhau ⇔ hai tổng `a + b` bằng nhau (nhân đôi hai vế). Nên ta chỉ cần đếm **số tổng khác nhau**: toàn số nguyên, không sợ sai số làm tròn của số thực.

````python
print((0 + 5) / 2 == (1 + 4) / 2, 0 + 5 == 1 + 4)
print((3 + 4) / 2, 3 + 4)
````

### 3.6. Mảng đếm tần suất cho giá trị nhỏ

Vì `0 <= nums[i] <= 100`, ta có thể dùng `cnt = [0] * 101` và `cnt[v] += 1`. Số nhỏ nhất/lớn nhất còn lại được tìm bằng hai chỉ số `lo` (tăng) và `hi` (giảm), bỏ qua các ô có `cnt == 0`. Không cần sắp xếp thật sự — đây là **counting sort ngầm**.

````python
cnt = [0] * 101
for v in [4, 1, 4, 0, 3, 5]:
    cnt[v] += 1
print({v: c for v, c in enumerate(cnt) if c})
````

## 4. Ý tưởng nền tảng / cách nghĩ trực giác

Làm đúng như đề: mỗi bước "bốc" số nhỏ nhất và số lớn nhất ra khỏi dãy, ghi lại trung bình, cho tới hết.

Ví dụ 1 (`[4, 1, 4, 0, 3, 5]`):

```
dãy : 4 1 4 0 3 5
bước 1: min=0, max=5 -> trung bình 2.5 ; còn 4 1 4 3
bước 2: min=1, max=4 -> trung bình 2.5 ; còn 4 3
bước 3: min=3, max=4 -> trung bình 3.5 ; còn rỗng
{2.5, 2.5, 3.5} -> 2 giá trị khác nhau
```

Quan sát: nếu **sắp xếp trước** thì `0 1 3 4 4 5` — bước 1 lấy hai đầu `(0, 5)`, bước 2 lấy hai đầu của phần còn lại `(1, 4)`, bước 3 lấy `(3, 4)`. Nghĩa là các cặp chính là **(đầu, cuối), (đầu thứ hai, cuối thứ hai), ...** của dãy đã sắp xếp. Việc "tìm min/max rồi xoá" biến thành "ghép đầu với cuối".

## 5. Cách 1 — Brute Force / baseline

Mô phỏng **đúng từng chữ** của đề: trong khi `nums` còn phần tử, tìm `min` và `max`, `remove` cả hai, thêm `(min + max)` vào `set`. Ta lưu **tổng** thay vì trung bình (nhân đôi để tránh số thực) mà không đổi kết quả.

Nút thắt: mỗi bước gồm `min`, `max`, hai `remove` đều `O(n)`; có `n/2` bước → `O(n^2)`.

````python
class SolutionBrute:
    def distinctAverages(self, nums):
        work = list(nums)               # không phá mảng gốc
        sums = set()
        while work:
            lo, hi = min(work), max(work)
            work.remove(lo)
            work.remove(hi)
            sums.add(lo + hi)           # lưu tổng: cùng tổng <=> cùng trung bình
        return len(sums)

print(SolutionBrute().distinctAverages([4, 1, 4, 0, 3, 5]))
print(SolutionBrute().distinctAverages([1, 100]))
````

**Độ phức tạp:** `n/2` bước × `O(n)` mỗi bước = `O(n^2)`; Space `O(n)` (bản sao + tập tổng). Với `n <= 100` thì quá nhẹ, nhưng không tận dụng cấu trúc "ghép đầu - cuối".

## 6. Ý tưởng thuật toán / bước đột phá

Sau khi **sắp xếp tăng dần**, tại bước thứ `i` (đếm từ 0) phần tử nhỏ nhất còn lại là `s[i]` và lớn nhất còn lại là `s[n - 1 - i]`. Do đó:

> đáp án = số giá trị khác nhau của `s[i] + s[n - 1 - i]` với `i = 0 .. n/2 - 1`.

**Quy tắc "nếu… thì…":**

- Nếu hai con trỏ `lo < hi` → ghép cặp `(s[lo], s[hi])`, thêm tổng vào `set`, rồi `lo += 1`, `hi -= 1`.
- Nếu `lo >= hi` → dừng (đã dùng hết số).
- Nếu hai cặp có cùng tổng → cùng trung bình → `set` tự gộp.

Invariant: trước mỗi bước, các phần tử `s[lo..hi]` chính là tập còn lại; `s[lo]` là min và `s[hi]` là max của tập đó (vì đã sắp xếp), và các tổng đã ghi tương ứng đúng với các bước trước.

````python
s = sorted([4, 1, 4, 0, 3, 5])
n = len(s)
sums = {s[i] + s[n - 1 - i] for i in range(n // 2)}
print("tổng các cặp:", [s[i] + s[n - 1 - i] for i in range(n // 2)])
print("tập tổng khác nhau:", sums, "->", len(sums))
````

## 7. Các cách tiếp theo

Ta có 4 cách: (1) mô phỏng `min`/`max`/`remove`, (2) sắp xếp + hai con trỏ, (3) sắp xếp + chỉ số đối xứng `~i`, (4) mảng đếm tần suất.

### Cách 2 — Sắp xếp + hai con trỏ

Sắp xếp một lần `O(n log n)`, rồi hai con trỏ `lo`, `hi` tiến vào giữa. Cách "kinh điển" của họ bài hai con trỏ.

````python
class SolutionTwoPointers:
    def distinctAverages(self, nums):
        s = sorted(nums)
        lo, hi = 0, len(s) - 1
        sums = set()
        while lo < hi:
            sums.add(s[lo] + s[hi])
            lo += 1
            hi -= 1
        return len(sums)

print(SolutionTwoPointers().distinctAverages([4, 1, 4, 0, 3, 5]))
print(SolutionTwoPointers().distinctAverages([1, 100]))
````

Độ phức tạp Cách 2: Time `O(n log n)` (sắp xếp) + `O(n)` (hai con trỏ); Space `O(n)` cho `sorted` và `set`.

### Cách 3 — Sắp xếp + chỉ số đối xứng `~i` (một dòng)

Dùng `s[~i]` để lấy phần tử đối xứng và tập hợp (set comprehension). Ngắn gọn, tương đương hai con trỏ.

````python
class SolutionMirror:
    def distinctAverages(self, nums):
        s = sorted(nums)
        return len({s[i] + s[~i] for i in range(len(s) // 2)})

print(SolutionMirror().distinctAverages([4, 1, 4, 0, 3, 5]))
print(SolutionMirror().distinctAverages([1, 100]))
````

Độ phức tạp Cách 3: Time `O(n log n)`, Space `O(n)`.

### Cách 4 — Mảng đếm tần suất (không cần sắp xếp)

Dựng `cnt[0..100]`. Hai chỉ số `lo` (từ 0 tăng) và `hi` (từ 100 giảm): mỗi bước tìm `lo` đầu tiên có `cnt[lo] > 0` và `hi` cuối cùng có `cnt[hi] > 0`, ghi tổng `lo + hi`, giảm `cnt[lo]` và `cnt[hi]` mỗi cái một đơn vị. Lặp `n / 2` lần.

````python
class SolutionCount:
    def distinctAverages(self, nums):
        cnt = [0] * 101
        for v in nums:
            cnt[v] += 1
        lo, hi = 0, 100
        sums = set()
        for _ in range(len(nums) // 2):
            while cnt[lo] == 0:
                lo += 1
            while cnt[hi] == 0:
                hi -= 1
            sums.add(lo + hi)
            cnt[lo] -= 1
            cnt[hi] -= 1
        return len(sums)

print(SolutionCount().distinctAverages([4, 1, 4, 0, 3, 5]))
print(SolutionCount().distinctAverages([1, 100]))
````

Độ phức tạp Cách 4: Time `O(n + V)` với `V = 101` miền giá trị (con trỏ `lo`, `hi` mỗi cái đi tối đa `V` bước tổng cộng), Space `O(V)`. Không có phép sắp xếp `O(n log n)`.

## 8. TRACE từng bước

Trace Cách 2 với `nums = [4, 1, 4, 0, 3, 5]` → sắp xếp `[0, 1, 3, 4, 4, 5]`.

| bước | `lo` | `hi` | `s[lo]` | `s[hi]` | tổng | `sums` |
|---|---|---|---|---|---|---|
| 1 | 0 | 5 | 0 | 5 | 5 | `{5}` |
| 2 | 1 | 4 | 1 | 4 | 5 | `{5}` |
| 3 | 2 | 3 | 3 | 4 | 7 | `{5, 7}` |

Dừng vì `lo = 3`, `hi = 2` (không còn `lo < hi`). Kết quả `len({5, 7}) = 2`.

````python
def trace_two_pointers(nums):
    s = sorted(nums)
    print("sau khi sắp xếp:", s)
    lo, hi = 0, len(s) - 1
    sums = set()
    step = 0
    while lo < hi:
        step += 1
        tong = s[lo] + s[hi]
        sums.add(tong)
        print(f"bước {step}: lo={lo} hi={hi} cặp=({s[lo]}, {s[hi]}) tổng={tong} trung bình={tong / 2} sums={sorted(sums)}")
        lo += 1
        hi -= 1
    print(f"dừng: lo={lo}, hi={hi} -> {len(sums)} trung bình khác nhau")
    return len(sums)

trace_two_pointers([4, 1, 4, 0, 3, 5])
trace_two_pointers([0, 1, 2, 2])
````

## 9. Code nộp LeetCode

````python
from typing import List

class Solution:
    def distinctAverages(self, nums: List[int]) -> int:
        nums = sorted(nums)
        lo, hi = 0, len(nums) - 1
        sums = set()
        while lo < hi:
            sums.add(nums[lo] + nums[hi])
            lo += 1
            hi -= 1
        return len(sums)

print(Solution().distinctAverages([4, 1, 4, 0, 3, 5]), Solution().distinctAverages([1, 100]))
````

## 10. Vì sao thuật toán đúng

- **Cặp tại bước `i` là `(s[i], s[n-1-i])`:** quy nạp theo `i`. Bước 0: min toàn cục là `s[0]` và max toàn cục là `s[n-1]` (mảng đã sắp xếp; nếu có giá trị trùng, chọn bất kỳ phần tử cùng giá trị cho kết quả như nhau). Sau khi xoá hai phần tử này, phần còn lại là `s[1..n-2]`, vẫn đã sắp xếp, nên bước 1 áp dụng cùng lý lẽ, v.v.
- **Giá trị trùng không ảnh hưởng:** đề chỉ xoá **một** số mỗi lần; khi dãy đã sắp xếp, "xoá một số bằng min" tương đương xoá phần tử ở đầu, không đổi tập giá trị còn lại.
- **Tổng ⇔ trung bình:** `(a + b)/2 == (c + d)/2 ⇔ a + b == c + d`. Do đó đếm tổng khác nhau cho đúng bằng số trung bình khác nhau; tránh số thực.
- **Độ dài chẵn:** `n` chẵn nên đúng `n/2` bước và hai con trỏ không bao giờ trùng nhau lẻ phần tử.
- **Cách 4:** `lo` luôn là giá trị nhỏ nhất còn `cnt > 0` và `hi` là giá trị lớn nhất còn `cnt > 0`: hai con trỏ chỉ di chuyển qua các ô đã hết, nên chúng không bỏ sót phần tử nào.

## 11. Phân tích Time / Space Complexity

Gọi `n = len(nums)`, `V = 101` là kích thước miền giá trị.

| Cách | Time | Space |
|---|---|---|
| 1. Mô phỏng `min`/`max`/`remove` | `O(n^2)` | `O(n)` |
| 2. Sắp xếp + hai con trỏ | `O(n log n)` | `O(n)` |
| 3. Sắp xếp + `s[~i]` | `O(n log n)` | `O(n)` |
| 4. Mảng đếm tần suất | `O(n + V)` | `O(V)` |

Sắp xếp là phần tốn nhất ở cách 2, 3 (Timsort worst `O(n log n)`). Với `n <= 100` mọi cách đều tức thì; cách 2 là cân bằng đẹp giữa dễ hiểu và nhanh.

## 12. Bộ test

Test 1: ví dụ đề và edge case (`n = 2` nhỏ nhất, mọi số bằng nhau, chuỗi tăng đều...).

````python
impls = [SolutionBrute(), SolutionTwoPointers(), SolutionMirror(), SolutionCount(), Solution()]
tests = [
    ([4, 1, 4, 0, 3, 5], 2),
    ([1, 100], 1),
    ([0, 0], 1),                          # n = 2, hai số 0
    ([100, 100], 1),                      # biên trên
    ([5, 5, 5, 5], 1),                    # tất cả bằng nhau
    ([1, 2, 3, 4], 1),                    # (1,4) và (2,3) cùng tổng 5 (bắt lỗi dùng list)
    ([0, 1, 2, 2], 2),                    # trung bình 1.0 và 1.5 (bắt lỗi làm tròn //2)
    ([0, 1, 5, 4], 1),                    # chưa sắp xếp: (0,5), (1,4) -> cùng tổng 5 (bắt lỗi quên sort)
    ([0, 0, 100, 100], 1),
    ([0, 1, 2, 3, 4, 5, 6, 7], 1),        # mọi cặp đều có tổng 7
    ([1, 1, 2, 3, 3, 4], 2),              # (1,4)=5, (1,3)=4, (2,3)=5 -> tổng {5, 4}
    ([10, 20, 30, 40, 50, 60], 1),
    ([0, 10, 20, 30, 40, 100], 2),        # (0,100)=100, (10,40)=50, (20,30)=50
]
# oracle độc lập: đúng quy trình của đề trên mảng đã sắp xếp
def oracle(nums):
    work = sorted(nums)
    seen = set()
    while work:
        a, b = work.pop(0), work.pop()
        seen.add((a + b) / 2)
    return len(seen)
for nums, exp in tests:
    assert oracle(nums) == exp, (nums, oracle(nums), exp)
    for im in impls:
        assert im.distinctAverages(list(nums)) == exp, (type(im).__name__, nums, exp)
print("tất cả test cố định OK:", len(tests))
````

Test 2: ngẫu nhiên có seed, đúng constraints (`n` chẵn trong `2..100`, giá trị `0..100`); oracle độc lập tính **trung bình thực** (số thực) bằng đúng quy trình của đề (`pop(0)` và `pop()` trên mảng đã sắp xếp, dùng `fractions` cho chính xác).

````python
import random
from fractions import Fraction
random.seed(2465)

def oracle_exact(nums):
    work = list(nums)
    seen = set()
    while work:
        lo, hi = min(work), max(work)
        work.remove(lo)
        work.remove(hi)
        seen.add(Fraction(lo + hi, 2))
    return len(seen)

for t in range(2500):
    n = 2 * random.randint(1, 50)
    hi_val = random.choice([3, 10, 100])
    nums = [random.randint(0, hi_val) for _ in range(n)]
    assert len(nums) % 2 == 0 and 2 <= len(nums) <= 100
    exp = oracle_exact(nums)
    for im in impls:
        assert im.distinctAverages(list(nums)) == exp, (type(im).__name__, nums)
print("2500 test ngẫu nhiên OK")
````

Test 3: tính chất — đáp án nằm trong `1..n/2`; hoán vị mảng đầu vào không đổi đáp án; và hàm không sửa mảng đầu vào.

## 13. Các lỗi người mới hay gặp

### Lỗi 1: quên sắp xếp trước khi ghép đầu - cuối

````python
def wrong_no_sort(nums):
    n = len(nums)
    return len({nums[i] + nums[n - 1 - i] for i in range(n // 2)})     # SAI: chưa sắp xếp

nums = [0, 1, 5, 4]
print("sai :", wrong_no_sort(nums))
print("đúng:", Solution().distinctAverages(nums))
assert wrong_no_sort(nums) != Solution().distinctAverages(nums)
````

Vì sao sai: "đầu" và "cuối" của mảng **chưa sắp xếp** không phải min/max. Với `[0, 1, 5, 4]` cặp thật là `(0, 5)` và `(1, 4)` (cùng tổng 5), nhưng ghép `(0, 4)`, `(1, 5)` cho hai tổng khác nhau. **Sửa:** `sorted` trước.

### Lỗi 2: dùng `list` (đếm số bước) thay cho `set`

````python
def wrong_list(nums):
    s = sorted(nums)
    lo, hi = 0, len(s) - 1
    avgs = []
    while lo < hi:
        avgs.append(s[lo] + s[hi])           # SAI: không loại bỏ giá trị trùng
        lo += 1
        hi -= 1
    return len(avgs)

nums = [1, 2, 3, 4]
print("sai :", wrong_list(nums))
print("đúng:", Solution().distinctAverages(nums))
assert wrong_list(nums) != Solution().distinctAverages(nums)
````

Vì sao sai: đề hỏi **số trung bình khác nhau**; `(1,4)` và `(2,3)` cùng tổng 5 (cùng trung bình `2.5`) nên chỉ tính một. Dùng `list` thì luôn ra `n / 2`. **Sửa:** `set`.

### Lỗi 3: làm tròn trung bình về số nguyên (`//`)

````python
def wrong_floor_avg(nums):
    s = sorted(nums)
    lo, hi = 0, len(s) - 1
    avgs = set()
    while lo < hi:
        avgs.add((s[lo] + s[hi]) // 2)       # SAI: mất phần .5
        lo += 1
        hi -= 1
    return len(avgs)

nums = [0, 1, 2, 2]
print("sai :", wrong_floor_avg(nums))
print("đúng:", Solution().distinctAverages(nums))
assert wrong_floor_avg(nums) != Solution().distinctAverages(nums)
````

Vì sao sai: các trung bình thật là `1.0` (từ `(0, 2)`) và `1.5` (từ `(1, 2)`) — khác nhau. Nhưng `2 // 2 = 1` và `3 // 2 = 1` gộp chúng làm một. **Sửa:** so sánh **tổng** (số nguyên), hoặc dùng `/ 2` (số thực).

### Lỗi 4 (bonus): xoá phần tử khi đang duyệt, hoặc sửa mảng gốc

Viết `for x in nums: nums.remove(...)` làm thay đổi danh sách trong lúc duyệt → bỏ sót phần tử. Hãy dùng `while nums:` hoặc sao chép (`list(nums)`) và sắp xếp bản sao.

## 14. Mẹo làm bài nhanh & nhận dạng pattern

- "Lặp lại: lấy phần tử nhỏ nhất và lớn nhất rồi xoá" → **sắp xếp một lần**, rồi **hai con trỏ** từ hai đầu.
- "Đếm số giá trị khác nhau" → `set`.
- Trung bình của hai số nguyên → so sánh **tổng** để tránh số thực/làm tròn.
- Giá trị nhỏ (`0..100`) → mảng đếm tần suất cho phép bỏ sắp xếp.
- Độ dài **chẵn** bảo đảm không có phần tử lẻ ở giữa; nếu đề cho lẻ, phải quyết định xử lý phần tử giữa.
- Dùng `s[~i]` là cách viết gọn cho `s[len(s) - 1 - i]`, nhưng chỉ nên dùng khi bạn đã hiểu rõ.

## 15. Biến thể của bài

1. Mảng có độ dài **lẻ** (phần tử giữa không ghép được) — bỏ qua hoặc tính riêng.
2. Trả về **danh sách** các trung bình khác nhau theo thứ tự tăng dần.
3. Đếm **tổng lớn nhất / nhỏ nhất** của các cặp (min, max) (liên quan "Minimize Maximum Pair Sum in Array").
4. Thay "trung bình" bằng "hiệu" (`max - min`): đếm số hiệu khác nhau.
5. Ghép cặp để **tối thiểu hoá tổng lớn nhất** của các cặp — cùng ý tưởng ghép đầu với cuối sau khi sắp xếp.

## 16. So sánh các cách làm

| Cách | Ý tưởng | Time | Space | Khi nào dùng |
|---|---|---|---|---|
| 1. Mô phỏng `min`/`max`/`remove` | làm đúng từng chữ của đề | `O(n^2)` | `O(n)` | hiểu đề, oracle test |
| 2. Sắp xếp + hai con trỏ | ghép đầu - cuối | `O(n log n)` | `O(n)` | **chuẩn**, dễ hiểu |
| 3. Sắp xếp + `s[~i]` | set comprehension một dòng | `O(n log n)` | `O(n)` | code ngắn |
| 4. Mảng đếm tần suất | hai chỉ số trên `cnt[0..100]` | `O(n + V)` | `O(V)` | miền giá trị nhỏ, tránh sắp xếp |

**Mức 1 (nhẹ):** luôn sắp xếp tăng dần; cặp thứ `i` là phần tử thứ `i` từ trái và thứ `i` từ phải.

## 19. Đáp án tham khảo cho bài tập

````python
# Bài 1: sắp xếp [1,1,5,5,9,9] -> các tổng 10, 10, 10 -> 1
assert Solution().distinctAverages([9, 1, 5, 5, 1, 9]) == 1

# Bài 2: [3,8,1,6,2,7] -> [1,2,3,6,7,8] -> (1,8) (2,7) (3,6): tổng 9, trung bình 4.5
s = sorted([3, 8, 1, 6, 2, 7])
pairs = [(s[i], s[-1 - i]) for i in range(len(s) // 2)]
assert pairs == [(1, 8), (2, 7), (3, 6)]
assert {(a + b) / 2 for a, b in pairs} == {4.5}

# Bài 3
def pair_sums(nums):
    s = sorted(nums)
    result = []
    for i in range(len(s) // 2):
        result.append(s[i] + s[len(s) - 1 - i])
    return result
assert pair_sums([4, 1, 4, 0, 3, 5]) == [5, 5, 7]
assert pair_sums([1, 100]) == [101]

# Bài 4
def distinct_averages_list(nums):
    return sorted({t / 2 for t in pair_sums(nums)})
assert distinct_averages_list([4, 1, 4, 0, 3, 5]) == [2.5, 3.5]
assert distinct_averages_list([1, 100]) == [50.5]
assert len(distinct_averages_list([0, 1, 2, 2])) == Solution().distinctAverages([0, 1, 2, 2])

# Bài 5
def min_max_pair_sum(nums):
    return max(pair_sums(nums))
assert min_max_pair_sum([3, 5, 2, 3]) == 7           # [2,3,3,5]: (2,5)=7, (3,3)=6
assert min_max_pair_sum([3, 5, 4, 2, 4, 6]) == 8     # [2,3,4,4,5,6]: 8, 8, 8
print("đáp án bài tập OK")
````
