## 1. Mục tiêu học

Sau notebook này bạn sẽ:

2. Biết định dạng số bằng f-string `f"{x:,}"`, đếm ký tự bằng `str.count`, và đếm chữ số bằng `len(str(x))`.

## 2. Đề bài & ví dụ

Cho số nguyên `n`. Hãy viết **tất cả** các số nguyên từ `1` đến `n` (cả hai đầu) theo cách viết chuẩn: một dấu phẩy được đặt sau mỗi nhóm ba chữ số tính từ **bên phải**, các số có ít hơn 4 chữ số thì không có dấu phẩy. Trả về **tổng số dấu phẩy** đã dùng.

| Ví dụ | n | Output | Giải thích |
|---|---|---|---|
| 1 | `1002` | `3` | `1,000`, `1,001`, `1,002` mỗi số có đúng một dấu phẩy |
| 2 | `998` | `0` | Mọi số từ 1 đến 998 có ít hơn bốn chữ số |

Constraints: `1 <= n <= 10^5`.

````python
for n in [1002, 998, 1000, 999]:
    shown = [f"{x:,}" for x in range(max(1, n - 2), n + 1)]
    print(n, "hiển thị vài số cuối:", shown, "-> số dấu phẩy ở các số này:", sum(s.count(',') for s in shown))
````

## 4. Ý tưởng nền tảng / cách nghĩ trực giác

Số lượng dấu phẩy của một số chỉ phụ thuộc vào **số chữ số** của nó:

| khoảng số | số chữ số | dấu phẩy mỗi số |
|---|---|---|
| 1 .. 999 | 1 - 3 | 0 |
| 1.000 .. 999.999 | 4 - 6 | 1 |
| 1.000.000 .. 999.999.999 | 7 - 9 | 2 |

Với `n <= 10^5`, mọi số nằm trong hai nhóm đầu: nhóm 0 dấu phẩy (`1..999`) và nhóm 1 dấu phẩy (`1000..n`). Vậy tổng số dấu phẩy bằng **số lượng số trong `[1000, n]`**, tức `max(0, n - 999)`.

Kiểm tra: `n = 1002` -> `1002 - 999 = 3` ✓. `n = 998` -> `max(0, -1) = 0` ✓. `n = 1000` -> `1` ✓ (chỉ số `1,000`).

````python
def commas_by_hand(n):
    zero = min(n, 999)
    one = max(0, n - 999)
    print(f"n={n}: {zero} số không có dấu phẩy, {one} số có 1 dấu phẩy -> tổng {one * 1} dấu phẩy")
    return one

for n in (1002, 998, 1000, 100000):
    commas_by_hand(n)
````

## 5. Cách 1 — Brute Force / baseline

Duyệt `x` từ `1` đến `n`, định dạng bằng `f"{x:,}"`, đếm dấu phẩy và cộng vào tổng. Làm đúng theo đề, không cần suy luận.

Nút thắt: mỗi số phải tạo một chuỗi (tốn `O(log x)`), tổng `O(n log n)`. Với `n <= 10^5` vẫn rất nhanh nhưng thao tác trên chuỗi là thừa vì số dấu phẩy chỉ phụ thuộc số chữ số.

````python
class SolutionBrute:
    def countCommas(self, n):
        total = 0
        for x in range(1, n + 1):
            total += f"{x:,}".count(",")
        return total

print(SolutionBrute().countCommas(1002), SolutionBrute().countCommas(998))
````

**Độ phức tạp:** Time `O(n log n)`, Space `O(log n)` cho một chuỗi tạm.

## 6. Ý tưởng thuật toán / bước đột phá

**Bước đột phá:** số dấu phẩy của `x` là `(số chữ số - 1) // 3` — **không cần dựng chuỗi có dấu phẩy**. Hơn nữa, các số có cùng số dấu phẩy nằm trong **khoảng liên tục**:

- có **ít nhất một** dấu phẩy khi `x >= 10^3`,
- có **ít nhất hai** dấu phẩy khi `x >= 10^6`,
- có **ít nhất `k`** dấu phẩy khi `x >= 10^(3k)`.

Tổng số dấu phẩy = tổng, theo từng `k >= 1`, của **số lượng số `x <= n` có ít nhất `k` dấu phẩy** = `max(0, n - 10^(3k) + 1)`. (Mỗi số có đúng `c` dấu phẩy được tính `c` lần, tại `k = 1..c`.)

Với `n <= 10^5`: chỉ `k = 1` đóng góp, đáp án `max(0, n - 999)`.

## 7. Các cách tiếp theo

### Cách 2 — Tính theo số chữ số từng số

Không dựng chuỗi có dấu phẩy; chỉ lấy `len(str(x))` rồi cộng `(d - 1) // 3`.

````python
class SolutionDigits:
    def countCommas(self, n):
        return sum((len(str(x)) - 1) // 3 for x in range(1, n + 1))

print(SolutionDigits().countCommas(1002), SolutionDigits().countCommas(998))
````

**Độ phức tạp:** Time `O(n log n)` (hoặc `O(n)` nếu coi `len(str)` hằng số), Space `O(1)` phụ.

### Cách 3 — Cộng theo các mốc `10^3, 10^6, ...`

`total += max(0, n - p + 1)` cho `p = 1000, 10^6, 10^9, ...` chừng nào `p <= n`. Có tính tổng quát cho `n` lớn hơn.

````python
class SolutionBoundaries:
    def countCommas(self, n):
        total = 0
        p = 1000
        while p <= n:
            total += n - p + 1
            p *= 1000
        return total

print(SolutionBoundaries().countCommas(1002), SolutionBoundaries().countCommas(998), SolutionBoundaries().countCommas(100000))
````

**Độ phức tạp:** Time `O(log n)` (số mốc là `log_1000 n`), Space `O(1)`.

### Cách 4 — Công thức `O(1)` dùng ràng buộc `n <= 10^5`

Vì `n < 10^6`, chỉ có nhóm 0 và nhóm 1 dấu phẩy: `max(0, n - 999)`.

````python
class SolutionFormula:
    def countCommas(self, n):
        return max(0, n - 999)          # đúng vì n <= 10^5 < 10^6

print(SolutionFormula().countCommas(1002), SolutionFormula().countCommas(998), SolutionFormula().countCommas(100000))
````

**Độ phức tạp:** Time `O(1)`, Space `O(1)`.

### Cách 5 — Tự chèn dấu phẩy (không dùng `:,`)

Viết hàm `with_commas(x)`: đổi sang chuỗi và chèn `,` sau mỗi 3 chữ số tính từ phải bằng cắt lát. Hữu ích để hiểu "dấu phẩy sau mỗi nhóm ba chữ số" và để kiểm chứng `:,`.

````python
def with_commas(x):
    s = str(x)
    groups = []
    while len(s) > 3:
        groups.append(s[-3:])
        s = s[:-3]
    groups.append(s)
    return ",".join(reversed(groups))

print(with_commas(7), with_commas(1000), with_commas(12345), with_commas(100000))
assert all(with_commas(x) == f"{x:,}" for x in range(1, 3001))

class SolutionManual:
    def countCommas(self, n):
        return sum(with_commas(x).count(",") for x in range(1, n + 1))

print(SolutionManual().countCommas(1002), SolutionManual().countCommas(998))
````

**Độ phức tạp:** Time `O(n log n)`, Space `O(log n)`.

## 8. TRACE từng bước

Trace Cách 3 với `n = 1002`: mốc `p = 1000 <= 1002` -> cộng `1002 - 1000 + 1 = 3`; `p = 10^6 > 1002` dừng. Tổng `3`.

Trace brute với `n = 1002` ở các số quanh mốc `999 -> 1000`:

| x | `f"{x:,}"` | dấu phẩy | tổng tích lũy (chỉ tính các dòng này) |
|---|---|---|---|
| 999 | `999` | 0 | 0 |
| 1000 | `1,000` | 1 | 1 |
| 1001 | `1,001` | 1 | 2 |
| 1002 | `1,002` | 1 | 3 |

````python
def trace_boundaries(n):
    total = 0
    p = 1000
    step = 0
    while p <= n:
        step += 1
        add = n - p + 1
        total += add
        print(f"bước {step}: p={p} <= n={n}, có {add} số có ít nhất {step} dấu phẩy -> total={total}")
        p *= 1000
    print(f"p={p} > n={n}: dừng. Tổng = {total}")
    return total

assert trace_boundaries(1002) == 3
assert trace_boundaries(998) == 0
assert trace_boundaries(100000) == 99001
````

````python
def trace_brute(n, last=6):
    total = 0
    for x in range(1, n + 1):
        total += f"{x:,}".count(",")
        if x >= n - last + 1 or x in (999, 1000):
            print(f"x={x:6d} -> {x:>9,} tổng={total}")
    return total

assert trace_brute(1002, 3) == 3
````

## 9. Code nộp LeetCode

Dùng Cách 3 (cộng theo mốc): chạy `O(log n)`, đúng cả khi `n` lớn hơn giới hạn đề.

````python
from typing import List, Optional


class Solution:
    def countCommas(self, n: int) -> int:
        total = 0
        p = 1000                       # số đầu tiên có ít nhất 1 dấu phẩy; sau đó 10^6, 10^9, ...
        while p <= n:
            total += n - p + 1
            p *= 1000
        return total


print(Solution().countCommas(1002), Solution().countCommas(998))
````

## 10. Vì sao thuật toán đúng

1. **Quy luật chữ số.** Dấu phẩy được đặt sau mỗi nhóm 3 chữ số kể từ phải, nên số có `d` chữ số có `ceil(d / 3) - 1 = (d - 1) // 3` dấu phẩy.
2. **Phân rã tổng.** Đặt `c(x)` là số dấu phẩy của `x`. Vì `c(x) = số k >= 1 sao cho x >= 10^(3k)` (số có `c` dấu phẩy thỏa đúng với `k = 1, ..., c`), nên `sum c(x) = sum_k #{x <= n : x >= 10^(3k)}`.
3. **Đếm mỗi hạng tử.** `#{x : 10^(3k) <= x <= n} = n - 10^(3k) + 1` khi `10^(3k) <= n`, ngược lại `0` — đúng với vòng `while p <= n`.
4. **Biên.** Tại `n = 999`: `p = 1000 > 999` -> tổng `0`; `n = 1000`: cộng `1`. Đây là lý do công thức `O(1)` là `n - 999` (số lượng số từ `1000` đến `n`) chứ không phải `n - 1000`.
5. Cách 1, 2, 5 đếm trực tiếp từng số nên là "chân lý" để đối chiếu.

## 11. Phân tích Time / Space Complexity

- Cách 1, 5: `O(n log n)` thời gian (mỗi số tạo chuỗi), `O(log n)` bộ nhớ phụ.
- Cách 2: `O(n log n)` (hoặc `O(n)` với `len(str)` coi là hằng số), `O(1)` phụ.
- Cách 3: `O(log n)` thời gian (số mốc `<= log_1000 n`), `O(1)` bộ nhớ.
- Cách 4: `O(1)`, nhưng chỉ đúng khi `n < 10^6`.

Không có best/worst khác nhau đáng kể. Khi `n` tăng tới `10^18`, chỉ cách 3 (và tổng quát hóa của nó) dùng được.

## 12. Bộ test

Đối chiếu mọi cách với ví dụ đề, biên (`999`, `1000`, `100000`) và **toàn bộ** `n` từ 1 đến 3000, cùng 300 giá trị ngẫu nhiên trong `1..10^5`.

````python
import random

fast = [SolutionBoundaries(), SolutionFormula(), Solution()]
slow = [SolutionBrute(), SolutionDigits(), SolutionManual()]

fixed = {1002: 3, 998: 0, 1: 0, 999: 0, 1000: 1, 1001: 2, 5000: 4001, 100000: 99001}
for n, expected in fixed.items():
    for s in fast + [SolutionBrute(), SolutionDigits()]:
        assert s.countCommas(n) == expected, (type(s).__name__, n)

# tất cả n nhỏ: dùng mảng tiền tố của brute làm chân lý
prefix = [0] * 3001
for x in range(1, 3001):
    prefix[x] = prefix[x - 1] + f"{x:,}".count(",")
for n in range(1, 3001):
    for s in fast:
        assert s.countCommas(n) == prefix[n], (type(s).__name__, n)

# ngẫu nhiên có seed, so với tổng theo số chữ số
random.seed(38700)
for _ in range(300):
    n = random.randint(1, 100000)
    ref = sum((len(str(x)) - 1) // 3 for x in range(1, n + 1))
    for s in fast:
        assert s.countCommas(n) == ref, (type(s).__name__, n)
for n in (1002, 998, 77):
    assert SolutionManual().countCommas(n) == SolutionBrute().countCommas(n)
print("tất cả test OK")
````

### Test thuộc tính

(1) Đáp án không giảm theo `n`. (2) Từ `n` đến `n + 1` đáp án tăng thêm đúng `1` nếu `n + 1 >= 1000` (còn nằm trong nhóm 1 dấu phẩy) và `0` nếu `n + 1 <= 999`. (3) Với `n >= 1000`: đáp án = `n - 999`.

````python
prev = Solution().countCommas(1)
for n in range(2, 100001):
    cur = Solution().countCommas(n)
    assert cur - prev == (1 if n >= 1000 else 0)
    prev = cur
for n in random.sample(range(1000, 100001), 100):
    assert Solution().countCommas(n) == n - 999
print("property test OK")
````

## 13. Các lỗi người mới hay gặp

### Lỗi 1: lệch một đơn vị — `n - 1000`

Số lượng số trong `[1000, n]` là `n - 1000 + 1 = n - 999`, không phải `n - 1000`.

````python
def wrong_off_by_one(n):
    return max(0, n - 1000)                                  # SAI: thiếu số 1000

print(wrong_off_by_one(1002), "!=", Solution().countCommas(1002))
assert wrong_off_by_one(1002) == 2 and Solution().countCommas(1002) == 3
assert wrong_off_by_one(1000) == 0 and Solution().countCommas(1000) == 1
````

### Lỗi 2: dùng `n // 1000`

`n // 1000` đếm số bội của 1000, không phải số lượng số có dấu phẩy.

````python
def wrong_floor_div(n):
    return n // 1000                                          # SAI

print(wrong_floor_div(1002), "!=", Solution().countCommas(1002))
assert wrong_floor_div(1002) == 1 and Solution().countCommas(1002) == 3
assert wrong_floor_div(5000) == 5 and Solution().countCommas(5000) == 4001
````

### Lỗi 3: quên số `n` (duyệt `range(1, n)`)

`range(1, n)` dừng ở `n - 1`. Đề tính cả `n`.

````python
def wrong_range(n):
    return sum(f"{x:,}".count(",") for x in range(1, n))     # SAI: thiếu n

print(wrong_range(1002), "!=", Solution().countCommas(1002))
assert wrong_range(1002) == 2 and Solution().countCommas(1002) == 3
assert wrong_range(1000) == 0 and Solution().countCommas(1000) == 1
````

### Lỗi 4: nhầm `len(str(x)) // 3` với `(len(str(x)) - 1) // 3`

Số 3 chữ số như `999` có `len = 3`; `3 // 3 = 1` nên bị đếm nhầm một dấu phẩy.

````python
def wrong_digits(n):
    return sum(len(str(x)) // 3 for x in range(1, n + 1))     # SAI: thiếu -1

print(wrong_digits(999), "!=", Solution().countCommas(999))
assert wrong_digits(999) == 900 and Solution().countCommas(999) == 0
assert wrong_digits(998) == 899 and Solution().countCommas(998) == 0
````

### Lỗi 5: đếm "số lượng số có dấu phẩy" bằng cách đếm chữ `,` trong **một chuỗi nối**

Nếu nối mọi số bằng dấu phẩy phân cách (`"1,2,3,..."`) thì dấu phẩy phân cách bị đếm lẫn với dấu phẩy ngăn nghìn.

````python
def wrong_joined(n):
    joined = ",".join(f"{x:,}" for x in range(1, n + 1))      # SAI: dùng chính dấu phẩy làm phân cách
    return joined.count(",")

print(wrong_joined(1002), "!=", Solution().countCommas(1002))
assert wrong_joined(1002) == 1001 + 3 and Solution().countCommas(1002) == 3
assert wrong_joined(5) == 4 and Solution().countCommas(5) == 0
````

## 18. Gợi ý 3 mức cho bài tập

**Mức 1 (nhẹ):** Bài 1: `2500 - 999` và với `n = 1` không có số nào từ 1000 trở lên. Bài 2: các số từ `1000` đến `12345` đều có đúng một dấu phẩy.

**Mức 3 (gần đáp án):** Bài 1: `1501` và `0`. Bài 2: `12345 - 999 = 11346` dấu phẩy; `999` số không có dấu phẩy. Bài 4: `total_digits(12) = 9*1 + 3*2 = 15`.

## 19. Đáp án tham khảo cho bài tập

````python
# Bài 1, 2
assert Solution().countCommas(2500) == 1501 and Solution().countCommas(1) == 0
assert Solution().countCommas(12345) == 11346

# Bài 3
def count_numbers_with_comma(n):
    return max(0, n - 999)
assert count_numbers_with_comma(12345) == 11346 and count_numbers_with_comma(5) == 0

# Bài 4
def total_digits(n):
    total, lo, d = 0, 1, 1
    while lo <= n:
        hi = min(n, lo * 10 - 1)
        total += (hi - lo + 1) * d
        lo *= 10
        d += 1
    return total
assert total_digits(12) == 15 and total_digits(9) == 9
for n in (1, 99, 100, 5000):
    assert total_digits(n) == sum(len(str(x)) for x in range(1, n + 1))

# Bài 5
def count_commas_range(a, b):
    return Solution().countCommas(b) - Solution().countCommas(a - 1)
assert count_commas_range(1, 1002) == 3 and count_commas_range(1000, 1002) == 3
assert count_commas_range(990, 1001) == 2 and count_commas_range(5, 10) == 0
print("đáp án bài tập OK")
````
