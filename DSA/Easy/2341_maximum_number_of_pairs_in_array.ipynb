## 1. Mục tiêu học

Sau notebook này bạn sẽ:

1. Hiểu bài toán **ghép cặp các phần tử bằng nhau** và tính được số cặp lẫn số phần tử còn thừa.
2. Dùng `collections.Counter`, phép chia `//` và chia dư `%` để suy ra `pairs = count // 2`, `leftover = count % 2`.
3. Biết **sắp xếp rồi quét** (nhóm các giá trị bằng nhau liền kề) và kỹ thuật **bật/tắt (toggle) bằng `set`**: thấy lần lẻ thì thêm, lần chẵn thì bỏ và đếm một cặp.
4. Viết 5 cách: ghép cặp bằng cách xoá phần tử khỏi danh sách (brute force), `Counter`, sắp xếp, `set` toggle, và mảng đếm kích thước 101; đối chiếu bằng test ngẫu nhiên.
5. Tránh các lỗi: đếm `count - 1` thay vì `count // 2`, trả về `[leftover, pairs]` sai thứ tự, tính `leftover` bằng `len(nums) - pairs`, quên rằng một giá trị có thể cho nhiều cặp.

## 2. Đề bài & ví dụ

Cho mảng số nguyên `nums` (chỉ số từ 0). Trong **một thao tác** bạn có thể:

- Chọn **hai số nguyên bằng nhau** trong `nums` (ở hai vị trí khác nhau),
- **Xoá cả hai** khỏi `nums`, tạo thành một **cặp**.

Hãy thực hiện thao tác này **nhiều lần nhất có thể**. Trả về mảng `answer` gồm hai số nguyên `[pairs, leftover]`: `pairs` là **số cặp** đã tạo, `leftover` là **số phần tử còn lại** trong `nums`.

| Ví dụ | nums | Output | Giải thích |
|---|---|---|---|
| 1 | `[1,3,2,1,3,2,2]` | `[3, 1]` | cặp `1,1`, cặp `3,3`, cặp `2,2`; còn một số `2` |
| 2 | `[1,1]` | `[1, 0]` | một cặp `1,1`, không còn gì |
| 3 | `[0]` | `[0, 1]` | không tạo được cặp, còn 1 phần tử |

Constraints: `1 <= len(nums) <= 100`; `0 <= nums[i] <= 100`.

````python
from collections import Counter

examples = [([1, 3, 2, 1, 3, 2, 2], [3, 1]), ([1, 1], [1, 0]), ([0], [0, 1])]
for nums, expected in examples:
    cnt = Counter(nums)
    pairs = sum(c // 2 for c in cnt.values())
    leftover = sum(c % 2 for c in cnt.values())
    print(f"nums={nums}  tần suất={dict(cnt)}  -> [{pairs}, {leftover}] ; mong đợi {expected}")
````

## 3. Kiến thức nền

### 3.1. `collections.Counter`

`Counter(nums)` là từ điển `{giá_trị: số_lần}`. `.values()` là các số lần, `.items()` là các cặp `(giá_trị, số_lần)`.

````python
from collections import Counter
c = Counter([1, 3, 2, 1, 3, 2, 2])
print(c)
print(list(c.values()), c[2], c[99])
````

### 3.2. Chia nguyên `//` và chia dư `%`

Với số lần `c` của một giá trị: `c // 2` là **số cặp** ghép được, `c % 2` là **số thừa** (0 hoặc 1). Ví dụ `c = 5`: hai cặp (4 phần tử) và dư 1.

````python
for c in range(0, 7):
    print(f"c={c}: cặp={c // 2}, dư={c % 2}")
````

### 3.3. `set` — thêm, xoá, kiểm tra thành viên

`s.add(x)`, `s.remove(x)` (lỗi nếu vắng), `s.discard(x)` (không lỗi), `x in s` trung bình `O(1)`. Dùng `set` như công tắc: lần gặp lẻ thì **thêm**, lần gặp chẵn thì **xoá**.

````python
seen = set()
for x in [2, 2, 2]:
    if x in seen:
        seen.remove(x)
        print("ghép cặp", x)
    else:
        seen.add(x)
    print("  seen =", seen)
````

### 3.4. Sắp xếp `sorted` và nhóm các giá trị bằng nhau liền kề

Sau khi sắp xếp, các giá trị bằng nhau đứng **cạnh nhau**. Quét `i` và ghép `nums[i] == nums[i + 1]` thành một cặp rồi nhảy `i += 2`; nếu không bằng thì `i += 1` (phần tử thừa).

````python
nums = sorted([1, 3, 2, 1, 3, 2, 2])
print(nums)
from itertools import groupby
print([(k, len(list(g))) for k, g in groupby(nums)])
````

### 3.5. Mảng đếm (counting array)

Vì `0 <= nums[i] <= 100`, ta có thể dùng mảng `cnt = [0] * 101` với `cnt[v] += 1` — nhanh và không cần thư viện. Các biến thể dùng khi miền giá trị nhỏ.

````python
cnt = [0] * 101
for v in [1, 3, 2, 1, 3, 2, 2]:
    cnt[v] += 1
print(cnt[:5])
print(sum(c // 2 for c in cnt), sum(c % 2 for c in cnt))
````

### 3.6. Phương thức `list.remove(x)` và `list.index(x)`

`lst.remove(x)` xoá phần tử **đầu tiên** bằng `x` (lỗi `ValueError` nếu vắng). Mỗi lần `remove`/`in` quét danh sách nên tốn `O(n)`. Ta dùng nó cho cách brute force dễ hiểu.

````python
lst = [1, 3, 2, 1]
lst.remove(1)
print(lst, 3 in lst, 9 in lst)
try:
    lst.remove(9)
except ValueError as e:
    print("lỗi:", e)
````

## 4. Ý tưởng nền tảng / cách nghĩ trực giác

Giống như **ghép tất chân**: mỗi chiếc tất cùng màu ghép thành một đôi; chiếc lẻ còn lại bỏ sang một bên. Ví dụ 1: `[1,3,2,1,3,2,2]`.

| giá trị | số lần | cặp (`//2`) | dư (`%2`) |
|---|---|---|---|
| 1 | 2 | 1 | 0 |
| 3 | 2 | 1 | 0 |
| 2 | 3 | 1 | 1 |
| **tổng** | 7 | **3** | **1** |

Mỗi giá trị **độc lập** với giá trị khác (chỉ ghép các số bằng nhau). Vậy: tính theo từng giá trị rồi cộng lại.

Kiểm tra: `2 * pairs + leftover = 2 * 3 + 1 = 7 = len(nums)`. Luôn đúng — một cách tự kiểm tra tiện lợi.

````python
nums = [1, 3, 2, 1, 3, 2, 2]
cnt = Counter(nums)
pairs = leftover = 0
for value, c in cnt.items():
    pairs += c // 2
    leftover += c % 2
    print(f"giá trị {value}: số lần={c} -> +{c // 2} cặp, +{c % 2} dư")
print([pairs, leftover], "kiểm tra 2*pairs + leftover == len:", 2 * pairs + leftover == len(nums))
````

## 5. Cách 1 — Brute Force / baseline: mô phỏng đúng thao tác của đề

Lặp: lấy phần tử đầu tiên `x`, tìm **một** phần tử bằng `x` ở phần còn lại; nếu có thì xoá cả hai và tăng `pairs`; nếu không thì phần tử `x` là phần thừa (cũng xoá khỏi danh sách đang xét nhưng tính vào `leftover`). Mỗi bước dùng `in` và `remove` nên chậm hơn.

````python
def pairs_simulate(nums):
    rest = list(nums)                  # sao chép để không phá dữ liệu gốc
    pairs = leftover = 0
    while rest:
        x = rest.pop()                 # lấy một phần tử ra
        if x in rest:                  # còn phần tử bằng x ở lại?
            rest.remove(x)             # xoá nó: ghép thành một cặp
            pairs += 1
        else:
            leftover += 1              # không ghép được
    return [pairs, leftover]

print(pairs_simulate([1, 3, 2, 1, 3, 2, 2]), pairs_simulate([1, 1]), pairs_simulate([0]))
````

````python
from typing import List

class SolutionSimulate:
    def numberOfPairs(self, nums: List[int]) -> List[int]:
        rest = list(nums)
        pairs = leftover = 0
        while rest:
            x = rest.pop()
            if x in rest:
                rest.remove(x)
                pairs += 1
            else:
                leftover += 1
        return [pairs, leftover]

print(SolutionSimulate().numberOfPairs([1, 3, 2, 1, 3, 2, 2]))
print(SolutionSimulate().numberOfPairs([1, 1]))
print(SolutionSimulate().numberOfPairs([0]))
````

**Độ phức tạp:** `n = len(nums)`. Mỗi vòng `in`/`remove` tốn `O(n)`, có `O(n)` vòng → **Time `O(n²)`**, **Space `O(n)`** (bản sao `rest`). Với `n ≤ 100` vẫn nhanh.

**Nút thắt:** việc **tìm phần tử bằng `x`** mỗi lần tốn `O(n)`. Nếu biết trước **mỗi giá trị xuất hiện bao nhiêu lần**, ta tính ngay số cặp mà không cần tìm.

## 6. Ý tưởng thuật toán / bước đột phá

- Chỉ **số lần xuất hiện** của mỗi giá trị quan trọng, không phải vị trí. Một giá trị xuất hiện `c` lần cho `c // 2` cặp và `c % 2` phần tử thừa.
- Các giá trị khác nhau **không tương tác** → cộng độc lập: `pairs = Σ c//2`, `leftover = Σ c%2`.
- Cách "bật/tắt": duyệt mảng một lần với `set`; gặp `x` đã có trong `set` → xoá `x` khỏi `set` và `pairs += 1`; chưa có → thêm. Cuối cùng `leftover = len(set)`.
- Cách sắp xếp: các giá trị bằng nhau đứng cạnh nhau → ghép từng cặp liền kề.
- Bất biến kiểm tra: `2 * pairs + leftover == len(nums)`.

## 7. Các cách tiếp theo

### Cách 2 — `Counter`: `pairs = Σ c//2`, `leftover = Σ c%2` (khuyến nghị khi nộp)

````python
from collections import Counter
from typing import List

class SolutionCounter:
    def numberOfPairs(self, nums: List[int]) -> List[int]:
        counts = Counter(nums)
        pairs = sum(c // 2 for c in counts.values())
        leftover = sum(c % 2 for c in counts.values())
        return [pairs, leftover]

print(SolutionCounter().numberOfPairs([1, 3, 2, 1, 3, 2, 2]))
print(SolutionCounter().numberOfPairs([1, 1]))
print(SolutionCounter().numberOfPairs([0]))
````

**Độ phức tạp:** Time `O(n)`, Space `O(k)` với `k` là số giá trị khác nhau (≤ 101 ở đề này).

### Cách 3 — Sắp xếp rồi quét ghép cặp liền kề

Sắp xếp để các giá trị bằng nhau đứng cạnh nhau. Con trỏ `i`: nếu `nums[i] == nums[i+1]` thì tạo một cặp và `i += 2`; ngược lại `leftover += 1` và `i += 1`.

````python
from typing import List

class SolutionSort:
    def numberOfPairs(self, nums: List[int]) -> List[int]:
        arr = sorted(nums)
        pairs = leftover = 0
        i = 0
        while i < len(arr):
            if i + 1 < len(arr) and arr[i] == arr[i + 1]:
                pairs += 1
                i += 2
            else:
                leftover += 1
                i += 1
        return [pairs, leftover]

print(SolutionSort().numberOfPairs([1, 3, 2, 1, 3, 2, 2]))
print(SolutionSort().numberOfPairs([1, 1]))
print(SolutionSort().numberOfPairs([0]))
````

**Độ phức tạp:** Time `O(n log n)` do sắp xếp, Space `O(n)` cho bản sao đã sắp xếp (hoặc `O(1)` nếu sắp xếp tại chỗ `nums.sort()`).

### Cách 4 — `set` bật/tắt trong một lượt

Mỗi lần gặp `x`: nếu `x` **đang trong `set`** thì có người bạn chờ sẵn → ghép cặp, bỏ `x` khỏi `set`; nếu không thì thêm `x` vào `set` (chờ bạn).

````python
from typing import List

class SolutionToggle:
    def numberOfPairs(self, nums: List[int]) -> List[int]:
        waiting = set()
        pairs = 0
        for x in nums:
            if x in waiting:
                waiting.remove(x)
                pairs += 1
            else:
                waiting.add(x)
        return [pairs, len(waiting)]

print(SolutionToggle().numberOfPairs([1, 3, 2, 1, 3, 2, 2]))
print(SolutionToggle().numberOfPairs([1, 1]))
print(SolutionToggle().numberOfPairs([0]))
````

**Độ phức tạp:** Time `O(n)` (mỗi thao tác `set` trung bình `O(1)`), Space `O(k)` (tối đa số giá trị đang chờ, ≤ 101).

### Cách 5 — Mảng đếm `[0] * 101`

Tận dụng `0 <= nums[i] <= 100`: không cần băm, chỉ cần chỉ số mảng. Nhanh và có thể nhìn thấy rõ vì mọi thứ là số nguyên.

````python
from typing import List

class SolutionArray:
    def numberOfPairs(self, nums: List[int]) -> List[int]:
        cnt = [0] * 101
        for v in nums:
            cnt[v] += 1
        pairs = sum(c // 2 for c in cnt)
        return [pairs, len(nums) - 2 * pairs]

print(SolutionArray().numberOfPairs([1, 3, 2, 1, 3, 2, 2]))
print(SolutionArray().numberOfPairs([1, 1]))
print(SolutionArray().numberOfPairs([0]))
````

**Độ phức tạp:** Time `O(n + 101)`, Space `O(101) = O(1)`. Dòng cuối tính `leftover = len(nums) - 2 * pairs` — đúng vì mỗi cặp bỏ đi 2 phần tử (khác với lỗi `len(nums) - pairs`).

## 8. TRACE từng bước

Trace Cách 4 (`set` bật/tắt) với `nums = [1, 3, 2, 1, 3, 2, 2]`:

| bước | `x` | `x` trong `waiting`? | hành động | `waiting` | `pairs` |
|---|---|---|---|---|---|
| 1 | 1 | không | thêm | `{1}` | 0 |
| 2 | 3 | không | thêm | `{1,3}` | 0 |
| 3 | 2 | không | thêm | `{1,2,3}` | 0 |
| 4 | 1 | **có** | ghép cặp, xoá | `{2,3}` | 1 |
| 5 | 3 | **có** | ghép cặp, xoá | `{2}` | 2 |
| 6 | 2 | **có** | ghép cặp, xoá | `{}` | 3 |
| 7 | 2 | không | thêm | `{2}` | 3 |

Kết quả `[3, len({2})] = [3, 1]`.

````python
def trace_toggle(nums):
    waiting, pairs = set(), 0
    print("nums =", nums)
    for step, x in enumerate(nums, 1):
        if x in waiting:
            waiting.remove(x)
            pairs += 1
            action = "ghép cặp"
        else:
            waiting.add(x)
            action = "chờ bạn"
        print(f"  bước {step}: x={x} {action:9} waiting={sorted(waiting)} pairs={pairs}")
    result = [pairs, len(waiting)]
    print("  kết quả:", result)
    return result

trace_toggle([1, 3, 2, 1, 3, 2, 2])
trace_toggle([0])
````

````python
def trace_sort(nums):
    arr = sorted(nums)
    print("đã sắp xếp:", arr)
    pairs = leftover = i = 0
    while i < len(arr):
        if i + 1 < len(arr) and arr[i] == arr[i + 1]:
            pairs += 1
            print(f"  i={i}: arr[{i}]=arr[{i+1}]={arr[i]} -> cặp #{pairs}, nhảy 2")
            i += 2
        else:
            leftover += 1
            print(f"  i={i}: arr[{i}]={arr[i]} lẻ -> dư #{leftover}, nhảy 1")
            i += 1
    return [pairs, leftover]

print(trace_sort([1, 3, 2, 1, 3, 2, 2]))
````

## 9. Code nộp LeetCode

````python
from collections import Counter
from typing import List

class Solution:
    def numberOfPairs(self, nums: List[int]) -> List[int]:
        counts = Counter(nums)
        pairs = sum(c // 2 for c in counts.values())
        return [pairs, len(nums) - 2 * pairs]

print(Solution().numberOfPairs([1, 3, 2, 1, 3, 2, 2]))
print(Solution().numberOfPairs([1, 1]))
print(Solution().numberOfPairs([0]))
````

## 10. Vì sao thuật toán đúng

- **Các giá trị độc lập:** một cặp chỉ gồm hai số **bằng nhau**, nên các số có giá trị khác nhau không ảnh hưởng đến việc ghép của nhau. Bài toán tách thành từng giá trị.
- **Số cặp tối đa cho một giá trị:** nếu có `c` phần tử bằng `v`, mỗi cặp dùng 2 phần tử nên tối đa `⌊c/2⌋` cặp, và có thể đạt đúng số đó (ghép tuỳ ý theo từng cặp). Sau đó còn `c mod 2` phần tử thừa.
- **Cộng lại:** `pairs = Σ ⌊c/2⌋`; số phần tử còn lại `= len(nums) - 2 * pairs`, vì mỗi cặp loại bỏ đúng 2 phần tử. Bằng `Σ (c mod 2)` do `c = 2·⌊c/2⌋ + (c mod 2)`.
- **Cách `set`:** invariant sau mỗi bước, `waiting` chứa đúng các giá trị đã xuất hiện một số **lẻ** lần; khi gặp lại một giá trị trong `waiting`, nó chuyển sang số lần chẵn (ghép được một cặp). Cuối cùng `len(waiting)` là số giá trị có số lần lẻ = số phần tử thừa.
- **Cách sắp xếp:** sau khi sắp xếp, mỗi giá trị chiếm một đoạn liên tục; ghép từng cặp liền kề trong đoạn cho `⌊c/2⌋` cặp và dư `c mod 2`.

## 11. Phân tích Time / Space Complexity

`n = len(nums) ≤ 100`, `V = 101` là số giá trị khả dĩ.

| Cách | Time | Space |
|---|---|---|
| 1. Mô phỏng `remove` | `O(n²)` | `O(n)` |
| 2. `Counter` | `O(n)` | `O(k)` |
| 3. Sắp xếp rồi quét | `O(n log n)` | `O(n)` (hoặc `O(1)` nếu sắp xếp tại chỗ) |
| 4. `set` bật/tắt | `O(n)` | `O(k)` |
| 5. Mảng đếm 101 | `O(n + V)` | `O(V)` |

`k ≤ min(n, 101)` là số giá trị khác nhau. Cách `Counter`/`set`/mảng là tuyến tính, tối ưu về bậc vì phải đọc mọi phần tử.

## 12. Bộ test

Ví dụ đề, edge case (một phần tử, hai phần tử khác nhau, toàn bộ bằng nhau, tất cả khác nhau, số lần lẻ lớn, biên giá trị `0` và `100`) và test ngẫu nhiên đối chiếu **tất cả cách**, cùng bất biến `2 * pairs + leftover == len(nums)`.

````python
import random

solvers = [SolutionSimulate(), SolutionCounter(), SolutionSort(), SolutionToggle(), SolutionArray(), Solution()]

fixed = [
    ([1, 3, 2, 1, 3, 2, 2], [3, 1]),
    ([1, 1], [1, 0]),
    ([0], [0, 1]),
    ([5, 6], [0, 2]),                         # hai phần tử khác nhau
    ([7] * 100, [50, 0]),                     # 100 phần tử giống nhau
    ([7] * 99, [49, 1]),                      # số lẻ
    (list(range(0, 100)), [0, 100]),          # toàn khác nhau
    ([0, 0, 0], [1, 1]),
    ([100, 100, 100, 100, 100], [2, 1]),      # biên 100
    ([0, 100, 0, 100], [2, 0]),
    ([1, 2, 1, 2, 1, 2, 1], [3, 1]),
    ([4, 4, 4, 4, 3, 3, 9], [3, 1]),
]
for nums, expected in fixed:
    for sol in solvers:
        assert sol.numberOfPairs(list(nums)) == expected, (type(sol).__name__, nums, expected)
print("test cố định: OK")

def reference(nums):
    # tham chiếu độc lập: từ điển thủ công
    d = {}
    for v in nums:
        d[v] = d.get(v, 0) + 1
    pairs = 0
    for c in d.values():
        while c >= 2:
            c -= 2
            pairs += 1
    return [pairs, len(nums) - 2 * pairs]

random.seed(2341)
for _ in range(700):
    n = random.randint(1, 100)
    hi = random.choice([1, 3, 10, 100])
    nums = [random.randint(0, hi) for _ in range(n)]
    expected = reference(nums)
    assert 2 * expected[0] + expected[1] == n
    for sol in solvers:
        got = sol.numberOfPairs(list(nums))
        assert got == expected, (type(sol).__name__, nums, got, expected)
print("700 test ngẫu nhiên: OK")
````

## 13. Các lỗi người mới hay gặp

### Lỗi 1: Dùng `count - 1` hoặc `count` thay vì `count // 2` cho số cặp

Mỗi cặp cần **2** phần tử.

````python
def wrong1(nums):
    cnt = Counter(nums)
    pairs = sum(c - 1 for c in cnt.values())          # sai
    return [pairs, len(nums) - 2 * pairs]

print("sai :", wrong1([1, 3, 2, 1, 3, 2, 2]), "(đáng lẽ [3, 1])")
print("đúng:", Solution().numberOfPairs([1, 3, 2, 1, 3, 2, 2]))
````

### Lỗi 2: Trả về sai thứ tự `[leftover, pairs]`

Đề: `answer[0]` là **số cặp**, `answer[1]` là **số dư**.

````python
def wrong2(nums):
    cnt = Counter(nums)
    pairs = sum(c // 2 for c in cnt.values())
    return [len(nums) - 2 * pairs, pairs]               # đảo thứ tự

print("sai :", wrong2([1, 3, 2, 1, 3, 2, 2]), "(đáng lẽ [3, 1])")
print("đúng:", Solution().numberOfPairs([1, 3, 2, 1, 3, 2, 2]))
````

### Lỗi 3: Tính `leftover = len(nums) - pairs` (quên mỗi cặp bỏ đi 2 phần tử)

````python
def wrong3(nums):
    cnt = Counter(nums)
    pairs = sum(c // 2 for c in cnt.values())
    return [pairs, len(nums) - pairs]                   # phải là len - 2*pairs

print("sai :", wrong3([1, 3, 2, 1, 3, 2, 2]), "(đáng lẽ [3, 1])")
print("đúng:", Solution().numberOfPairs([1, 3, 2, 1, 3, 2, 2]))
````

### Lỗi 4: Dùng `set` thông thường để đếm cặp — chỉ xét "có trùng" một lần

`len(nums) - len(set(nums))` đếm số **phần tử trùng**, không phải số cặp: với ba số `2,2,2` nó cho `2` nhưng chỉ có `1` cặp.

````python
def wrong4(nums):
    pairs = len(nums) - len(set(nums))
    return [pairs, len(nums) - 2 * pairs]

print("sai :", wrong4([2, 2, 2]), "(đáng lẽ [1, 1])")
print("đúng:", Solution().numberOfPairs([2, 2, 2]))
````

### Lỗi 5: `remove` trong lúc đang duyệt chính danh sách đó

Sửa danh sách khi `for x in nums:` đang chạy làm vòng lặp **bỏ sót** phần tử.

````python
def wrong5(nums):
    nums = list(nums)
    pairs = 0
    for x in nums:                      # vừa duyệt vừa xoá
        if nums.count(x) >= 2:
            nums.remove(x)
            nums.remove(x)
            pairs += 1
    return [pairs, len(nums)]

print("sai :", wrong5([1, 1, 1, 1, 1, 1]), "(đáng lẽ [3, 0])")
print("đúng:", Solution().numberOfPairs([1, 1, 1, 1, 1, 1]))
````

## 14. Mẹo làm bài nhanh & nhận dạng pattern

- "Ghép cặp các phần tử **bằng nhau**" → `Counter` rồi `c // 2` (cặp) và `c % 2` (dư).
- Bất biến kiểm tra nhanh: `2 * pairs + leftover == len(nums)`.
- "Đã thấy một số lẻ lần chưa?" → **`set` bật/tắt** (hoặc XOR/bitmask nếu miền nhỏ).
- Miền giá trị nhỏ (`0..100`) → **mảng đếm** thay cho băm.
- Cùng họ: LeetCode 136 Single Number (XOR), 217 Contains Duplicate, 1189, 242 Valid Anagram.
- Đừng sửa danh sách khi đang duyệt nó; hãy dùng bản sao hoặc `Counter`.

## 15. Biến thể của bài

1. Cho phép ghép **hai số hơn kém nhau đúng `k`** (không phải bằng nhau).
2. Ghép bộ **ba** số bằng nhau (`c // 3`).
3. LeetCode 136 Single Number: mọi số xuất hiện hai lần trừ một số → tìm số đó (XOR).
4. LeetCode 2206 Divide Array Into Equal Pairs: có chia được thành toàn các cặp bằng nhau không (`leftover == 0`).
5. Trả về **chính các cặp** (danh sách các cặp giá trị) thay vì chỉ số lượng.

## 16. So sánh các cách làm

| Cách | Ý tưởng | Time | Space | Khi nào dùng |
|---|---|---|---|---|
| 1. Mô phỏng `remove` | làm đúng như đề | `O(n²)` | `O(n)` | hiểu bài, `n` rất nhỏ |
| 2. `Counter` | `Σ c//2`, `Σ c%2` | `O(n)` | `O(k)` | **khuyến nghị khi nộp** |
| 3. Sắp xếp + quét | ghép cặp liền kề | `O(n log n)` | `O(n)` | không muốn dùng băm |
| 4. `set` bật/tắt | một lượt, chờ bạn | `O(n)` | `O(k)` | stream dữ liệu, một lượt |
| 5. Mảng đếm 101 | chỉ số = giá trị | `O(n + V)` | `O(V)` | miền giá trị nhỏ |

## 19. Đáp án tham khảo cho bài tập

Chạy kiểm tra bằng `assert`.

````python
# Bài 1 và 2
assert (7 // 2, 7 % 2, Counter("aab")["a"], Counter("aab")["z"]) == (3, 1, 2, 0)
assert Solution().numberOfPairs([4, 4, 4, 4, 3, 3, 9]) == [3, 1]

# Bài 3
def can_divide_into_pairs(nums):
    return all(c % 2 == 0 for c in Counter(nums).values())

# Bài 4
def list_pairs(nums):
    out = []
    for v, c in sorted(Counter(nums).items()):
        out.extend([v] * (c // 2))
    return out

# Bài 5
def max_triples(nums):
    triples = sum(c // 3 for c in Counter(nums).values())
    return [triples, len(nums) - 3 * triples]

assert can_divide_into_pairs([3, 2, 3, 2, 2, 2]) is True
assert can_divide_into_pairs([1, 2, 3, 4]) is False
assert can_divide_into_pairs([5, 5, 5]) is False
assert list_pairs([1, 3, 2, 1, 3, 2, 2]) == [1, 2, 3]
assert list_pairs([7, 7, 7, 7, 7]) == [7, 7]
assert list_pairs([1, 2, 3]) == []
assert max_triples([1, 1, 1, 2, 2, 2, 2]) == [2, 1]
assert max_triples([1, 1]) == [0, 2]
assert max_triples([9] * 10) == [3, 1]
print("tất cả bài tập OK")
````

from collections import Counter

class Solution:
    def numberOfPairs(self, nums: List[int]) -> List[int]:
