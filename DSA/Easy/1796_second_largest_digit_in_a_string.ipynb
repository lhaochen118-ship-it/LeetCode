## 3. Cách trực giác — gom, sắp xếp, lấy áp chót

1. Gom các chữ số phân biệt vào `set`.
2. Nếu có ít hơn 2 phần tử → `-1`.
3. Ngược lại `sorted(set)[-2]`.

````python
def second_highest_sort(s: str) -> int:
    digits = sorted({int(ch) for ch in s if ch.isdigit()})
    if len(digits) < 2:
        return -1
    return digits[-2]

for s, expected in examples + [("abc", -1), ("a9b9c8", 8)]:
    print(f"{s!r:15} -> {second_highest_sort(s)} | kỳ vọng {expected}")
````

## 4. Nút thắt và insight

Ta không cần biết **toàn bộ** thứ tự — chỉ cần **top 2**. Giữ hai biến:

- `first`: chữ số lớn nhất đã thấy (khởi tạo `-1` = "chưa có").
- `second`: chữ số lớn thứ hai **khác `first`** đã thấy (khởi tạo `-1`).

Khi gặp chữ số `d`:

| Trường hợp | Cập nhật |
|---|---|
| `d > first` | `second = first`, `first = d` (người cũ xuống hạng 2) |
| `first > d > second` | `second = d` |
| `d == first` hoặc `d <= second` | bỏ qua (trùng hoặc quá nhỏ) |

Điểm then chốt: điều kiện `d == first` phải **bỏ qua**, nếu không `"1111"` sẽ cho `second = 1` sai.

`-1` là giá trị khởi tạo rất khéo: mọi chữ số `>= 0 > -1`, và nếu không cập nhật được thì `second` vẫn là `-1` — **đúng luôn đáp án "không tồn tại"**.

````python
# Thử từng bước bằng tay với dãy chữ số 1, 2, 3, 2, 1
first, second = -1, -1
for d in [1, 2, 3, 2, 1]:
    if d > first:
        second, first = first, d
    elif first > d > second:
        second = d
    print(f"d={d}: first={first}, second={second}")
````

`first > d > second` là **so sánh chuỗi (chained comparison)** của Python, nghĩa là `first > d and d > second`.

`second, first = first, d` — gán song song: vế phải tính trước (`first` cũ, `d`), rồi mới gán. Nếu viết hai dòng `first = d; second = first` thì `second` nhận `d` — **sai**!

````python
first, second = 5, 2
d = 7
# Sai thứ tự
f1, s1 = first, second
f1 = d
s1 = f1
print("sai:", f1, s1)
# Đúng: gán song song
f2, s2 = first, second
s2, f2 = f2, d
print("đúng:", f2, s2)
````

## 5. Lời giải chính — một lượt với `first`, `second`

Ghép các đoạn:

1. `first = second = -1`.
2. Duyệt từng `ch` trong `s`; bỏ qua nếu không phải số.
3. `d = int(ch)`, cập nhật theo bảng ở mục 4.
4. Trả `second`.

````python
def second_highest(s: str) -> int:
    first = second = -1
    for ch in s:
        if not ch.isdigit():
            continue
        d = int(ch)
        if d > first:
            second, first = first, d
        elif first > d > second:
            second = d
    return second

for s, expected in examples + [("abc", -1), ("a9b9c8", 8), ("0", -1), ("01", 0)]:
    print(f"{s!r:15} -> {second_highest(s)} | kỳ vọng {expected}")
````

`continue` nhảy ngay sang vòng lặp kế tiếp, bỏ qua phần code còn lại — giúp tránh lồng `if` sâu.

## 6. TRACE từng bước

Trace `s = "dfa12321afd"`:

| ch | là số? | d | Nhánh | first | second |
|---|---|---|---|---|---|
| d,f,a | không | – | bỏ qua | -1 | -1 |
| 1 | có | 1 | `d > first` | 1 | -1 |
| 2 | có | 2 | `d > first` | 2 | 1 |
| 3 | có | 3 | `d > first` | 3 | 2 |
| 2 | có | 2 | `d <= second` → bỏ qua | 3 | 2 |
| 1 | có | 1 | bỏ qua | 3 | 2 |
| a,f,d | không | – | bỏ qua | 3 | 2 |

Kết quả `second = 2`.

````python
def trace_second_highest(s: str) -> int:
    print(f"=== s = {s!r} ===")
    first = second = -1
    for i, ch in enumerate(s):
        if not ch.isdigit():
            continue
        d = int(ch)
        if d > first:
            branch = "d > first: đẩy first xuống second"
            second, first = first, d
        elif first > d > second:
            branch = "first > d > second: cập nhật second"
            second = d
        elif d == first:
            branch = "d == first: trùng max, bỏ qua"
        else:
            branch = "d <= second: bỏ qua"
        print(f"i={i:2} ch={ch} | {branch:38} | first={first:2}, second={second:2}")
    print("-> trả về", second)
    return second

trace_second_highest("dfa12321afd")
print()
trace_second_highest("abc1111")
````

## 7. Các lời giải khác

### Cách 3 — Mảng đánh dấu 10 phần tử, quét từ 9 xuống 0

Vì chỉ có 10 chữ số, dùng list `seen = [False] * 10`. Sau đó quét `d = 9, 8, ..., 0`, đếm các chữ số đã thấy; chữ số thứ hai gặp được chính là đáp án.

````python
def second_highest_mark(s: str) -> int:
    seen = [False] * 10
    for ch in s:
        if ch.isdigit():
            seen[int(ch)] = True
    count = 0
    for d in range(9, -1, -1):
        if seen[d]:
            count += 1
            if count == 2:
                return d
    return -1

print("range(9, -1, -1) =", list(range(9, -1, -1)))
for s, expected in examples:
    print(s, "->", second_highest_mark(s), "| kỳ vọng", expected)
````

`range(9, -1, -1)`: bắt đầu 9, dừng **trước** -1, bước -1 → 9, 8, …, 0.

### Cách 4 — Set rồi bỏ max

Lấy set chữ số, xóa phần tử lớn nhất, phần tử lớn nhất còn lại là đáp án.

````python
def second_highest_remove_max(s: str) -> int:
    digits = {int(ch) for ch in s if ch.isdigit()}
    if len(digits) < 2:
        return -1
    digits.remove(max(digits))
    return max(digits)

for s, expected in examples:
    print(s, "->", second_highest_remove_max(s), "| kỳ vọng", expected)
````

### 7.1. Bảng so sánh

| Cách | Ý tưởng | Time | Space | Ghi chú |
|---|---|---|---|---|
| 1. Set + sort | sorted(set)[-2] | `O(n)` (sort ≤ 10 phần tử) | `O(1)` (≤ 10) | dễ nhất |
| **2. first/second (chính)** | top-2 một lượt | `O(n)` | `O(1)` | tổng quát cho mọi miền giá trị |
| 3. Mảng đánh dấu | quét 9→0 | `O(n + 10)` | `O(10)` | tận dụng miền nhỏ |
| 4. Set bỏ max | max hai lần | `O(n)` | `O(1)` | ngắn |

`n = len(s)`. Với mảng số nguyên tổng quát (không giới hạn 0..9), chỉ cách 2 giữ được `O(n)` thời gian và `O(1)` bộ nhớ.

## 8. Code nộp LeetCode

````python
class Solution:
    def secondHighest(self, s: str) -> int:
        first = second = -1
        for ch in s:
            if ch.isdigit():
                d = int(ch)
                if d > first:
                    second, first = first, d
                elif first > d > second:
                    second = d
        return second

sol = Solution()
for s, expected in examples:
    print(sol.secondHighest(s), "| kỳ vọng", expected)
````

## 9. Vì sao thuật toán đúng?

**Invariant:** sau khi xử lý xong tiền tố `s[0..i]`, gọi `D` là tập chữ số phân biệt trong tiền tố:

- `first = max(D)` (hoặc `-1` nếu `D` rỗng),
- `second = max(D \ {first})` (hoặc `-1` nếu `D` có < 2 phần tử).

Chứng minh giữ invariant khi thêm chữ số `d`:

- `d > first`: max mới là `d`, max của phần còn lại là `first` cũ → gán `second, first = first, d` đúng.
- `d == first`: `D` không đổi → giữ nguyên đúng.
- `second < d < first`: max không đổi, phần tử lớn nhất khác max giờ là `d` → `second = d` đúng.
- `d <= second`: `d` không ảnh hưởng top 2 (nếu `d == second` thì `D` không đổi).

Cuối cùng `second` đúng bằng chữ số lớn thứ hai, hoặc `-1`.

## 12. Lỗi người mới hay gặp

### Lỗi 1: Không loại trùng

Với `"abc1111"` trả `1` thay vì `-1`.

````python
def wrong_no_dedup(s):
    digits = sorted(int(ch) for ch in s if ch.isdigit())
    return digits[-2] if len(digits) >= 2 else -1

print("sai:", wrong_no_dedup("abc1111"), "| đúng:", second_highest("abc1111"))
````

### Lỗi 2: Dùng `>=` thay vì `>` khi so với `first`

`if d >= first: second, first = first, d` → khi gặp lại max, `second` bị gán bằng max.

````python
def wrong_ge(s):
    first = second = -1
    for ch in s:
        if ch.isdigit():
            d = int(ch)
            if d >= first:
                second, first = first, d
            elif d > second:
                second = d
    return second

print("sai:", wrong_ge("abc1111"), "| đúng:", second_highest("abc1111"))
````

### Lỗi 3: Quên nhánh `elif first > d > second`

Chỉ cập nhật khi `d > first` → nếu số lớn nhất đến **trước**, `second` không bao giờ được cập nhật.

````python
def wrong_missing_elif(s):
    first = second = -1
    for ch in s:
        if ch.isdigit():
            d = int(ch)
            if d > first:
                second, first = first, d
    return second

print("sai:", wrong_missing_elif("91"), "| đúng:", second_highest("91"))
````

**Sửa:** luôn `d = int(ch)` trước khi so sánh.

## 13. Mẹo nhận dạng pattern

- "**Lớn thứ hai / nhỏ thứ hai / top-k nhỏ**" → giữ `k` biến và cập nhật dây chuyền (`second, first = first, d`), không cần sort.

## 14. Biến thể

### 14.1. Third Maximum Number (LeetCode 414) — top-3 phân biệt, không có thì trả max

Mở rộng tự nhiên của `first`, `second` thành `first`, `second`, `third`. Dùng `None` làm "chưa có" vì số có thể âm.

````python
def third_max(nums):
    first = second = third = None
    for x in nums:
        if x in (first, second, third):
            continue
        if first is None or x > first:
            first, second, third = x, first, second
        elif second is None or x > second:
            second, third = x, second
        elif third is None or x > third:
            third = x
    return first if third is None else third

print(third_max([3, 2, 1]), third_max([1, 2]), third_max([2, 2, 3, 1]))   # 1 2 1
````

### 14.2. Chữ số lớn thứ `k` trong chuỗi

Dùng mảng đánh dấu và quét từ 9 xuống, dừng ở chữ số thứ `k`.

````python
def kth_largest_digit(s: str, k: int) -> int:
    seen = [False] * 10
    for ch in s:
        if ch.isdigit():
            seen[int(ch)] = True
    for d in range(9, -1, -1):
        if seen[d]:
            k -= 1
            if k == 0:
                return d
    return -1

print(kth_largest_digit("dfa12321afd", 1), kth_largest_digit("dfa12321afd", 3), kth_largest_digit("dfa12321afd", 4))
````

## 17. Đáp án tham khảo

**Bài 1:** `[2, 5, 9]` và `['0', '1']`.

**Bài 2:** `8` → first=8; `8` trùng; `3` → second=3; `9` → first=9, second=8; `3` bỏ qua → đáp án **8**.

````python
print(sorted({int(c) for c in "a5b5c2d9" if c.isdigit()}), [c for c in "z0y1" if c.isdigit()])
assert trace_second_highest("8a8b3c9d3") == 8

def ex3_second_smallest_digit_ans(s):
    first = second = 10
    for ch in s:
        if not ch.isdigit():
            continue
        d = int(ch)
        if d < first:
            second, first = first, d
        elif first < d < second:
            second = d
    return second if second < 10 else -1

def ex4_second_highest_number_ans(nums):
    first = second = None
    for x in nums:
        if x == first:
            continue
        if first is None or x > first:
            second, first = first, x
        elif second is None or x > second:
            second = x
    return second

def ex5_digit_gap_ans(s):
    first = second = -1
    for ch in s:
        if ch.isdigit():
            d = int(ch)
            if d > first:
                second, first = first, d
            elif first > d > second:
                second = d
    return first - second if second != -1 else -1

def ex6_second_highest_integer_ans(s):
    nums, cur = [], ""
    for ch in s + "#":            # ký tự gác '#' để chốt khối cuối
        if ch.isdigit():
            cur += ch
        elif cur:
            nums.append(int(cur))
            cur = ""
    r = ex4_second_highest_number_ans(nums)
    return -1 if r is None else r

assert ex3_second_smallest_digit_ans("dfa12321afd") == 2
assert ex3_second_smallest_digit_ans("abc1111") == -1
assert ex4_second_highest_number_ans([3, 1, 3, 2]) == 2
assert ex4_second_highest_number_ans([-5, -5]) is None
assert ex4_second_highest_number_ans([-1, -7]) == -7
assert ex5_digit_gap_ans("dfa12321afd") == 1
assert ex5_digit_gap_ans("a9b2") == 7
assert ex5_digit_gap_ans("abc") == -1
assert ex6_second_highest_integer_ans("a12b7c12") == 7
assert ex6_second_highest_integer_ans("x100y99z100") == 99
assert ex6_second_highest_integer_ans("abc5") == -1
print("Đáp án tham khảo PASS toàn bộ test!")
````

        first = second = -1
        for ch in s:
            if ch.isdigit():
                d = int(ch)
                if d > first:
                    second, first = first, d      # max cũ xuống hạng 2
