## 2. Đề bài & ví dụ

Một thao tác **lật bit** biến một bit `0` thành `1` hoặc `1` thành `0` trong biểu diễn nhị phân của một số. Ví dụ, lật bit thứ `0` (bit phải nhất) của `7 = 111` cho `110 = 6`.

| Ví dụ | start | goal | Output | Giải thích |
|---|---|---|---|---|
| 1 | `10` | `7` | `3` | `1010 → 0111`: bit 0, 1, 3 khác nhau nên cần lật 3 bit. |
| 2 | `3` | `4` | `3` | `011 → 100`: cả ba bit đều khác nhau. |

Constraints: `0 <= start, goal <= 10^9`.

````python
examples = [(10, 7, 3), (3, 4, 3)]
for start, goal, expected in examples:
    print(f"start={start:>2} = {start:04b} | goal={goal:>2} = {goal:04b} | mong đợi = {expected}")
````

## 3. Kiến thức nền

### 3.1. Biểu diễn nhị phân trong Python

`bin(x)` cho chuỗi dạng `'0b1010'`; `format(x, 'b')` hoặc `f"{x:b}"` cho `'1010'` không có tiền tố. `format(x, '08b')` đệm số `0` bên trái cho đủ 8 ký tự. `int('1010', 2)` đổi ngược chuỗi nhị phân về số.

````python
x = 10
print(bin(x), format(x, "b"), f"{x:08b}")
print(int("1010", 2))
print(bin(0), bin(1))
````

### 3.2. Toán tử bit cơ bản

- `a & b` (AND): bit kết quả là `1` khi **cả hai** bit là `1`.
- `a | b` (OR): `1` khi **ít nhất một** bit là `1`.
- `a ^ b` (XOR): `1` khi hai bit **khác nhau**.
- `x >> 1`: dịch phải một bit (chia nguyên cho 2). `x << 1`: dịch trái (nhân 2).

````python
a, b = 10, 7      # 1010 và 0111
print(f"{a:04b} & {b:04b} = {a & b:04b}")
print(f"{a:04b} | {b:04b} = {a | b:04b}")
print(f"{a:04b} ^ {b:04b} = {a ^ b:04b}")
print(a >> 1, a << 1)
````

### 3.3. XOR là "máy phát hiện khác nhau"

Bảng chân trị XOR: `0^0=0`, `1^1=0`, `0^1=1`, `1^0=1`. Vậy `a ^ b` có bit `1` **đúng ở những vị trí** `a` và `b` khác nhau. Đếm số bit `1` trong `a ^ b` chính là đếm số vị trí khác nhau.

````python
for a, b in [(0, 0), (1, 1), (0, 1), (1, 0)]:
    print(a, "^", b, "=", a ^ b)
print(bin(10 ^ 7), (10 ^ 7).bit_length())
````

### 3.4. Lấy bit cuối: `x & 1`, và đếm bit `1`

`x & 1` là bit cuối (`0` hoặc `1`). Kết hợp `x >>= 1` ta duyệt từng bit từ phải sang trái. Cách khác: `bin(x).count('1')` đếm số ký tự `'1'` trong chuỗi nhị phân; Python 3.10+ có sẵn `x.bit_count()`.

````python
x = 13   # 1101
print(x & 1, (x >> 1) & 1, (x >> 2) & 1, (x >> 3) & 1)
print(bin(x).count("1"), x.bit_count())
````

### 3.5. Mẹo Brian Kernighan: `x & (x - 1)`

Phép `x & (x - 1)` **xóa bit `1` thấp nhất** của `x`. Vì trừ `1` làm bit `1` thấp nhất thành `0` và các bit `0` bên phải nó thành `1`. Lặp cho tới khi `x == 0`, số lần lặp bằng số bit `1`.

````python
x = 0b101100
print(f"{x:06b} -> {x & (x - 1):06b}")
steps = 0
while x:
    x &= x - 1
    steps += 1
    print(f"sau bước {steps}: {x:06b}")
````

### 3.6. Lật một bit cụ thể: `x ^ (1 << p)`

`1 << p` là số chỉ có bit thứ `p` bằng `1` (gọi là **mặt nạ** — mask). XOR với mask lật đúng bit `p` và giữ nguyên các bit khác. Ngoài ra: `x | mask` ép bit `p` thành `1`, `x & ~mask` ép bit `p` thành `0`, `(x >> p) & 1` đọc bit `p`.

Như vậy "một lần lật bit" trong đề chính là phép `x ^= 1 << p` với một `p` nào đó. Điều này giải thích vì sao XOR xuất hiện tự nhiên trong lời giải.

````python
x = 0b1010
for p in range(4):
    mask = 1 << p
    print(f"p={p} mask={mask:04b} | lật: {x ^ mask:04b} | ép 1: {x | mask:04b} | ép 0: {x & ~mask:04b} | đọc bit: {(x >> p) & 1}")
````

## 4. Ý tưởng nền tảng — tính bằng tay

`start = 10 = 1010`, `goal = 7 = 0111`. Căn thẳng hàng từng bit:

| Vị trí bit | 3 | 2 | 1 | 0 |
|---|---|---|---|---|
| start | 1 | 0 | 1 | 0 |
| goal | 0 | 1 | 1 | 1 |
| khác nhau? | có | có | không | có |

Mỗi vị trí **khác nhau** cần đúng **một** lần lật, mỗi vị trí **giống nhau** không cần lật. Vậy đáp án là số vị trí khác nhau: `3`.

Lật một bit chỉ sửa được đúng **một** vị trí, nên không thể làm tốt hơn số vị trí khác nhau. Đó là cận dưới và cũng đạt được ⇒ đáp án chính xác.

````python
start, goal = 10, 7
diff = 0
for pos in range(3, -1, -1):
    bs, bg = (start >> pos) & 1, (goal >> pos) & 1
    print(f"bit {pos}: start={bs} goal={bg} {'khác' if bs != bg else 'giống'}")
    diff += bs != bg
print("số bit khác nhau =", diff)
````

Thử thêm một ví dụ tính tay khác: `start = 29 = 11101`, `goal = 15 = 01111`. Viết thẳng hàng:

```
start = 1 1 1 0 1
goal  = 0 1 1 1 1
khác  = 1 0 0 1 0   -> 2 vị trí khác nhau
```

Chuỗi `khác` chính là `start ^ goal = 10010`. Hai bit `1` nghĩa là cần lật bit thứ `4` (từ `1` thành `0`) và bit thứ `1` (từ `0` thành `1`). Sau khi lật: `11101 ^ 10010 = 01111 = 15`. Đúng như mong muốn: **lật các bit của `start ^ goal` trên `start` cho ra `goal`**, vì `start ^ (start ^ goal) = goal`.

````python
start, goal = 29, 15
x = start ^ goal
print(f"start={start:05b} goal={goal:05b} xor={x:05b}")
print("start ^ xor =", f"{start ^ x:05b}", "== goal?", start ^ x == goal)
````

## 5. Cách 1 — Brute Force: so sánh hai chuỗi nhị phân

Đổi cả hai số sang chuỗi nhị phân, **đệm `0`** cho cùng độ dài, rồi so sánh từng ký tự. Đây là cách thẳng nhất, bám sát cách ta tính tay.

````python
class SolutionBrute:
    def minBitFlips(self, start, goal):
        width = max(start.bit_length(), goal.bit_length(), 1)
        a = format(start, "b").zfill(width)       # đệm 0 bên trái
        b = format(goal, "b").zfill(width)
        flips = 0
        for ca, cb in zip(a, b):
            if ca != cb:
                flips += 1
        return flips

sol = SolutionBrute()
print(sol.minBitFlips(10, 7))
print(sol.minBitFlips(3, 4))
````

**Độ phức tạp:** gọi `B` là số bit (`B <= 30` vì `10^9 < 2^30`). Time `O(B)`, Space `O(B)` cho hai chuỗi.

**Nút thắt:** tạo chuỗi và đệm `0` là thừa. Ta có thể so sánh bit trực tiếp hoặc dùng XOR để gom việc "tìm vị trí khác nhau" vào **một** phép toán.

## 6. Ý tưởng thuật toán — bước đột phá

**Insight:** `start ^ goal` có bit `1` đúng tại các vị trí khác nhau. Bài toán thu về: **đếm số bit `1` của `start ^ goal`** (popcount).

Quy tắc "nếu… thì…":

- **Nếu** hai bit giống nhau **thì** XOR cho `0` ⇒ không cần lật.
- **Nếu** hai bit khác nhau **thì** XOR cho `1` ⇒ cần đúng một lần lật.
- Đáp án = `popcount(start ^ goal)`.

**Invariant (khi đếm bằng Kernighan):** mỗi lần `x &= x - 1` xóa đúng một bit `1`, nên số lần lặp tới khi `x == 0` bằng đúng số bit `1` ban đầu.

## 7. Các cách tiếp theo

### Cách 2 — Duyệt từng bit bằng `& 1` và `>> 1`

Không tạo chuỗi: lặp khi còn bit nào đó khác `0`; so sánh bit cuối của hai số, rồi dịch phải cả hai.

````python
class SolutionLoop:
    def minBitFlips(self, start, goal):
        flips = 0
        while start or goal:                      # còn bit để xét
            if (start & 1) != (goal & 1):
                flips += 1
            start >>= 1
            goal >>= 1
        return flips

sol = SolutionLoop()
print(sol.minBitFlips(10, 7))
print(sol.minBitFlips(3, 4))
````

**Độ phức tạp:** Time `O(B)` với `B` số bit của số lớn hơn, Space `O(1)`.

### Cách 3 — XOR rồi đếm bằng `bin(x).count('1')`

Ngắn gọn nhất ở mức nền tảng: `bin(start ^ goal).count('1')`. Chuỗi `bin(...)` chỉ có `0`, `1` và tiền tố `0b`, nên đếm ký tự `'1'` an toàn.

````python
class SolutionXorBin:
    def minBitFlips(self, start, goal):
        return bin(start ^ goal).count("1")

sol = SolutionXorBin()
print(sol.minBitFlips(10, 7))
print(sol.minBitFlips(3, 4))
````

**Độ phức tạp:** Time `O(B)`, Space `O(B)` cho chuỗi tạm (rất nhỏ).

### Cách 4 — XOR + Brian Kernighan

Thay vì duyệt **mọi** bit, ta chỉ lặp **đúng bằng số bit `1`**: mỗi bước `x &= x - 1` xóa bit `1` thấp nhất. Hiệu quả khi số bit `1` ít.

````python
class SolutionKernighan:
    def minBitFlips(self, start, goal):
        x = start ^ goal
        flips = 0
        while x:
            x &= x - 1          # xóa bit 1 thấp nhất
            flips += 1
        return flips

sol = SolutionKernighan()
print(sol.minBitFlips(10, 7))
print(sol.minBitFlips(3, 4))
````

**Độ phức tạp:** Time `O(k)` với `k` là số bit `1` của `start ^ goal` (`k <= B`), Space `O(1)`.

Ngoài ra Python 3.10+ có `(start ^ goal).bit_count()` — một dòng, chạy bằng C nên rất nhanh; LeetCode hỗ trợ.

## 8. TRACE từng bước

Trace cách 4 với `start = 10`, `goal = 7`:

| Bước | x trước | x - 1 | x & (x - 1) | flips |
|---|---|---|---|---|
| (khởi tạo) | `1010 ^ 0111 = 1101` | | | 0 |
| 1 | `1101` | `1100` | `1100` | 1 |
| 2 | `1100` | `1011` | `1000` | 2 |
| 3 | `1000` | `0111` | `0000` | 3 |

`x = 0` ⇒ dừng, đáp án `3`. Hàm dưới in từng bước, và một hàm khác trace cách duyệt bit.

````python
def trace_kernighan(start, goal, width=8):
    x = start ^ goal
    print(f"start={start:0{width}b} goal={goal:0{width}b} xor={x:0{width}b}")
    flips = 0
    while x:
        nxt = x & (x - 1)
        flips += 1
        print(f"x={x:0{width}b} x-1={x - 1:0{width}b} -> x&(x-1)={nxt:0{width}b} | flips={flips}")
        x = nxt
    return flips

print(trace_kernighan(10, 7, 4))
````

````python
def trace_loop(start, goal):
    flips, pos = 0, 0
    while start or goal:
        bs, bg = start & 1, goal & 1
        if bs != bg:
            flips += 1
        print(f"bit {pos}: start={bs} goal={bg} {'KHÁC' if bs != bg else 'giống'} | flips={flips}")
        start >>= 1; goal >>= 1; pos += 1
    return flips

print(trace_loop(3, 4))
````

## 9. Code nộp LeetCode

Dùng cách 4 (XOR + Kernighan) — không phụ thuộc phiên bản Python, độ phức tạp tốt.

````python
from typing import List

class Solution:
    def minBitFlips(self, start: int, goal: int) -> int:
        x = start ^ goal
        flips = 0
        while x:
            x &= x - 1
            flips += 1
        return flips

print(Solution().minBitFlips(10, 7))
print(Solution().minBitFlips(3, 4))
````

## 10. Vì sao thuật toán đúng?

- **Cận dưới:** một lần lật chỉ đổi đúng **một** bit. Mỗi vị trí mà `start` và `goal` khác nhau bắt buộc phải được lật ít nhất một lần ⇒ cần ít nhất `d` lần, với `d` là số vị trí khác nhau.
- **Cận trên:** lật đúng `d` vị trí đó (mỗi vị trí một lần) biến `start` thành `goal` ⇒ `d` lần là đủ.
- **XOR đúng định nghĩa "khác nhau":** `(a ^ b)` có bit `1` ở vị trí `p` ⇔ bit `p` của `a` và `b` khác nhau. Vậy `d = popcount(start ^ goal)`.
- **Kernighan đúng:** `x - 1` biến bit `1` thấp nhất của `x` thành `0` và mọi bit `0` bên phải nó thành `1`; `&` với `x` giữ nguyên các bit cao và xóa bit đó cùng các bit thấp (vốn đã `0`). Vậy mỗi bước xóa đúng một bit `1`.

## 11. Phân tích Time / Space Complexity

Gọi `B` là số bit của số lớn nhất (`B <= 30`), `k` là số bit `1` của `start ^ goal`.

| Cách | Time | Space |
|---|---|---|
| 1. So sánh chuỗi | `O(B)` | `O(B)` |

Vì `B <= 30`, tất cả đều coi như hằng số; sự khác biệt chỉ quan trọng khi số rất lớn (big int).

### Kiểm chứng độc lập bằng BFS (tìm kiếm theo tầng)

Để chắc chắn rằng "đáp án = số bit khác nhau", ta mô phỏng đúng đề: coi mỗi số là một đỉnh, mỗi lần lật một bit là một cạnh, rồi dùng BFS tìm đường ngắn nhất từ `start` đến `goal` trên các số `6` bit. BFS không dùng công thức XOR nên là một "trọng tài" độc lập.

````python
from collections import deque

def bfs_min_flips(start, goal, width=6):
    dist = {start: 0}
    q = deque([start])
    while q:
        u = q.popleft()
        if u == goal:
            return dist[u]
        for p in range(width):
            v = u ^ (1 << p)                   # một lần lật bit
            if v not in dist:
                dist[v] = dist[u] + 1
                q.append(v)

checked = 0
for a in range(64):
    for b in range(64):
        assert bfs_min_flips(a, b) == Solution().minBitFlips(a, b), (a, b)
        checked += 1
print("BFS khớp công thức trên", checked, "cặp số 6 bit")
````

## 13. Các lỗi người mới hay gặp

### Lỗi 1 — So sánh hai chuỗi nhị phân không đệm cùng độ dài

`bin(3)` = `'11'` còn `bin(4)` = `'100'`. `zip` cắt theo chuỗi ngắn hơn nên bỏ sót bit.

````python
def wrong_no_pad(start, goal):
    a, b = format(start, "b"), format(goal, "b")
    return sum(ca != cb for ca, cb in zip(a, b))     # SAI: zip cắt phần dư

print("sai :", wrong_no_pad(3, 4))     # so '11' với '100' từ TRÁI -> sai vị trí
print("đúng:", Solution().minBitFlips(3, 4))
````

### Lỗi 2 — Đếm nhầm số bit `1` của `start` và `goal` thay vì của XOR

Hai số có cùng số bit `1` vẫn có thể khác nhau hoàn toàn.

````python
def wrong_popcount_diff(start, goal):
    return abs(bin(start).count("1") - bin(goal).count("1"))

print("sai :", wrong_popcount_diff(3, 5))        # 0, vì cả hai đều có hai bit 1
print("đúng:", Solution().minBitFlips(3, 5))     # 011 vs 101 -> 2
````

### Lỗi 3 — Vòng `while` chỉ chờ `start` hết mà quên `goal`

Nếu `goal` có nhiều bit hơn `start`, vòng lặp dừng sớm và bỏ sót các bit cao của `goal`.

````python
def wrong_loop(start, goal):
    flips = 0
    while start:                    # SAI: phải là while start or goal
        if (start & 1) != (goal & 1):
            flips += 1
        start >>= 1
        goal >>= 1
    return flips

print("sai :", wrong_loop(1, 8))    # chỉ xét 1 bit thấp nhất
print("đúng:", Solution().minBitFlips(1, 8))   # 0001 vs 1000 -> 2
````

### Lỗi 4 — Dùng `x - 1` thay cho `x & (x - 1)` khi đếm

Chỉ trừ 1 mỗi lần rồi đếm bước thì số bước bằng **giá trị** của `x`, không phải số bit `1`.

````python
def wrong_minus_one(start, goal):
    x, steps = start ^ goal, 0
    while x:
        x -= 1                      # SAI: thiếu phép AND
        steps += 1
    return steps

print("sai :", wrong_minus_one(10, 7))     # 13 bước
print("đúng:", Solution().minBitFlips(10, 7))   # 3
````

## 14. Mẹo làm bài nhanh & nhận dạng pattern

- "Số thao tác lật bit để biến A thành B" ⇒ **khoảng cách Hamming** = `popcount(A ^ B)`.
- Cụm "đếm số bit 1" ⇒ `bin(x).count('1')`, `x.bit_count()` hoặc Kernighan.
- XOR cho biết "chỗ khác nhau"; AND cho "chỗ chung"; OR cho "hợp".
- Bài họ hàng: Number of 1 Bits (191), Hamming Distance (461), Counting Bits (338), Total Hamming Distance (477), Power of Two (231).
- Nhớ: `x & (x-1)` xóa bit 1 thấp nhất; `x & -x` lấy bit 1 thấp nhất.
- Với số không âm của đề này, không lo biểu diễn bù hai; nhưng nếu gặp số âm trong Python (số nguyên lớn tùy ý, coi như có vô hạn bit `1` ở đầu), cả `x >>= 1` lẫn `x &= x - 1` đều **không bao giờ** về `0` (`x >>= 1` kẹt ở `-1`) — hãy áp mặt nạ `x & 0xFFFFFFFF` trước.
- Khi đếm bit cho nhiều truy vấn lặp lại, có thể tiền tính bảng `popcount` cho 8 hoặc 16 bit rồi tra bảng.

## 15. Biến thể của bài

1. **Hamming Distance (461):** đúng là bài này, chỉ khác tên.
2. **Total Hamming Distance (477):** tổng khoảng cách giữa mọi cặp trong mảng — đếm theo từng bit: `ones * zeros`.
3. **Số lần lật để A OR B = C (1318):** xét từng bit, quyết định lật tối thiểu.
4. **Chỉ được lật bit `1` thành `0`:** cần `start & goal == goal`? và đếm bit.
5. **Lật bit cho nhiều số:** biến cả mảng về cùng một giá trị với tổng lật nhỏ nhất (lấy đa số từng bit).

````python
# Biến thể 2: tổng Hamming distance của mọi cặp, đếm theo từng bit
def total_hamming(nums):
    total = 0
    for bit in range(32):
        ones = sum((x >> bit) & 1 for x in nums)
        total += ones * (len(nums) - ones)
    return total

print(total_hamming([4, 14, 2]))   # 6
````
