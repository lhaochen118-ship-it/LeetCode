## 3. Kiến thức nền

### 3.1. Subarray độ dài 2 và số lượng của chúng

Mảng `n` phần tử có đúng `n - 1` subarray độ dài 2, bắt đầu ở chỉ số `0, 1, ..., n - 2`. Subarray bắt đầu ở `i` là `[nums[i], nums[i + 1]]`.

````python
nums = [10, 20, 30, 40]
for i in range(len(nums) - 1):
    print("bắt đầu ở", i, "->", nums[i:i + 2], "tổng =", sum(nums[i:i + 2]))
````

### 3.2. `zip(nums, nums[1:])` để lấy các cặp liền kề

`zip(a, b)` dừng ở dãy ngắn hơn. Với `nums[1:]` ngắn hơn `nums` đúng 1 phần tử, ta nhận được đúng `n - 1` cặp `(nums[i], nums[i+1])` mà **không cần chỉ số**.

````python
nums = [10, 20, 30, 40]
print(list(zip(nums, nums[1:])))
print([a + b for a, b in zip(nums, nums[1:])])
````

### 3.3. `set` để nhớ "đã gặp"

`set` lưu các giá trị không trùng, kiểm tra `x in s` trung bình `O(1)`. `s.add(x)` thêm vào. Đây là công cụ chuẩn để **phát hiện trùng**.

````python
seen = set()
for v in [6, 6, 7]:
    print(v, "đã gặp?", v in seen)
    seen.add(v)
print(seen)
````

### 3.4. So sánh `len(set(a))` với `len(a)`

Nếu danh sách có phần tử trùng thì tập hợp của nó **ít phần tử hơn**. Đó là cách một dòng để biết "có trùng không".

````python
a = [3, 5, 3]
print(len(a), len(set(a)), len(a) != len(set(a)))
b = [3, 5, 7]
print(len(b), len(set(b)), len(b) != len(set(b)))
````

### 3.5. `sorted` và so sánh hàng xóm

Sau khi sắp xếp, các giá trị bằng nhau **đứng cạnh nhau**. Vì vậy, chỉ cần kiểm tra `s[i] == s[i+1]` là đủ để phát hiện trùng.

````python
s = sorted([9, 3, 5, 3, 1])
print(s)
print(any(s[i] == s[i + 1] for i in range(len(s) - 1)))
````

## 4. Ý tưởng nền tảng / cách nghĩ trực giác

Viết các tổng cặp liền kề ra bằng tay, rồi nhìn xem có số nào xuất hiện hai lần.

```
nums = [4, 2, 4]
tổng: 4+2 = 6, 2+4 = 6  -> 6 xuất hiện 2 lần  -> true

nums = [1, 2, 3, 4, 5]
tổng: 3, 5, 7, 9        -> không lặp          -> false

nums = [0, 0, 0]
tổng: 0, 0              -> 0 xuất hiện 2 lần  -> true
```

Lưu ý quan trọng: "hai subarray khác nhau" nghĩa là **hai vị trí bắt đầu khác nhau**, còn **giá trị** có thể bằng nhau (như `[0,0]` ở ví dụ 3). Nên ta đếm số **lần xuất hiện của tổng** trong danh sách các tổng, không đếm số tổng khác nhau.

Bài toán thực chất là: *cho dãy `sums` có `n - 1` số, hỏi có số nào lặp lại không?*

````python
nums = [0, 0, 0]
sums = [a + b for a, b in zip(nums, nums[1:])]
print(sums)
for i, s in enumerate(sums):
    print(f"chỉ số {i}: tổng {s}")
print("hai vị trí khác nhau cùng tổng 0 -> true")
````

## 5. Cách 1 — Brute Force / baseline: so sánh mọi cặp chỉ số

**Ý tưởng:** có `n - 1` subarray bắt đầu ở `0..n-2`. Với mỗi cặp chỉ số `i < j` trong khoảng đó, so sánh `nums[i] + nums[i+1]` và `nums[j] + nums[j+1]`. Nếu bằng nhau, trả về `True`.

````python
from typing import List

class SolutionBrute:
    def findSubarrays(self, nums: List[int]) -> bool:
        n = len(nums)
        for i in range(n - 1):
            for j in range(i + 1, n - 1):
                if nums[i] + nums[i + 1] == nums[j] + nums[j + 1]:
                    return True
        return False
````

````python
sol = SolutionBrute()
print(sol.findSubarrays([4, 2, 4]))         # True
print(sol.findSubarrays([1, 2, 3, 4, 5]))   # False
print(sol.findSubarrays([0, 0, 0]))         # True
````

**Độ phức tạp:** Time `O(n^2)` (khoảng `n^2 / 2` cặp, với `n = 1000` chỉ ~500.000 phép so sánh), Space `O(1)`. **Nút thắt:** mỗi tổng bị tính lại nhiều lần và ta so sánh từng cặp. Dùng bộ nhớ để nhớ "tổng nào đã gặp" sẽ rút gọn xuống một lượt duyệt.

## 6. Ý tưởng thuật toán / bước đột phá

**Bước đột phá:** đã gặp tổng `s` rồi thì gặp lại lần nữa là xong, nên chỉ cần **một tập hợp các tổng đã thấy** và một lượt duyệt.

- **Invariant:** sau khi xử lý các subarray bắt đầu ở `0..i-1`, `seen` chứa đúng các tổng của chúng; và **chưa có** hai tổng nào trong đó bằng nhau (nếu có thì ta đã trả về `True`).
- **Quy tắc:** với subarray mới có tổng `s`:
  - **nếu** `s in seen` **thì** trả về `True`;
  - **nếu không thì** `seen.add(s)` và tiếp tục.

Thứ tự "**kiểm tra trước, thêm sau**" rất quan trọng: nếu thêm trước, ta luôn tự khớp với chính mình.

## 7. Các cách tiếp theo

### Cách 2 — `set` các tổng đã gặp (một lượt)

````python
from typing import List

class SolutionSet:
    def findSubarrays(self, nums: List[int]) -> bool:
        seen = set()
        for a, b in zip(nums, nums[1:]):
            s = a + b
            if s in seen:
                return True
            seen.add(s)
        return False
````

````python
print(SolutionSet().findSubarrays([4, 2, 4]))         # True
print(SolutionSet().findSubarrays([1, 2, 3, 4, 5]))   # False
print(SolutionSet().findSubarrays([0, 0, 0]))         # True
````

**Độ phức tạp:** Time `O(n)` trung bình (mỗi `in`/`add` trên set là `O(1)` trung bình), Space `O(n)` cho `seen`. Có thể dừng sớm ngay khi thấy trùng.

### Cách 3 — Sắp xếp các tổng rồi so hàng xóm

Tạo danh sách `sums`, `sort`, rồi nếu có hai phần tử kề nhau bằng nhau thì có trùng. Không cần `set` nhưng tốn thời gian sắp xếp.

````python
from typing import List

class SolutionSort:
    def findSubarrays(self, nums: List[int]) -> bool:
        sums = sorted(a + b for a, b in zip(nums, nums[1:]))
        return any(sums[i] == sums[i + 1] for i in range(len(sums) - 1))
````

````python
print(SolutionSort().findSubarrays([4, 2, 4]))         # True
print(SolutionSort().findSubarrays([1, 2, 3, 4, 5]))   # False
print(SolutionSort().findSubarrays([0, 0, 0]))         # True
````

**Độ phức tạp:** Time `O(n log n)` do sắp xếp, Space `O(n)` cho danh sách tổng.

### Cách 4 — So sánh `len(set)` với `len(list)`

Một dòng: nếu số tổng **khác nhau** ít hơn số subarray thì có trùng. Ngắn nhưng luôn tính hết (không dừng sớm).

````python
from typing import List

class SolutionLen:
    def findSubarrays(self, nums: List[int]) -> bool:
        sums = [a + b for a, b in zip(nums, nums[1:])]
        return len(set(sums)) < len(sums)
````

````python
print(SolutionLen().findSubarrays([4, 2, 4]))         # True
print(SolutionLen().findSubarrays([1, 2, 3, 4, 5]))   # False
print(SolutionLen().findSubarrays([0, 0, 0]))         # True
````

**Độ phức tạp:** Time `O(n)` trung bình, Space `O(n)`.

## 8. TRACE từng bước

Bảng trạng thái của cách `set` trên `nums = [4, 2, 4]`:

| Bước | cặp | tổng `s` | `seen` trước | `s in seen`? | Hành động |
|---|---|---|---|---|---|
| 1 | `(4, 2)` | 6 | `{}` | không | thêm 6 |
| 2 | `(2, 4)` | 6 | `{6}` | **có** | trả về `True` |

Và trên `nums = [1, 2, 3, 4, 5]`: tổng `3, 5, 7, 9` đều mới nên duyệt hết, trả về `False`.

````python
def trace_set(nums):
    seen = set()
    for i, (a, b) in enumerate(zip(nums, nums[1:])):
        s = a + b
        print(f"i={i}: cặp=({a}, {b}), tổng={s}, seen={sorted(seen)}")
        if s in seen:
            print(f"   {s} đã có trong seen -> True")
            return True
        seen.add(s)
        print(f"   thêm {s}")
    print("hết mảng -> False")
    return False

print(trace_set([4, 2, 4]))
print(trace_set([1, 2, 3, 4, 5]))
````

## 9. Code nộp LeetCode

Cách `set` một lượt có thể dừng sớm:

````python
from typing import List

class Solution:
    def findSubarrays(self, nums: List[int]) -> bool:
        seen = set()
        for i in range(len(nums) - 1):
            s = nums[i] + nums[i + 1]
            if s in seen:
                return True
            seen.add(s)
        return False
````

````python
print(Solution().findSubarrays([4, 2, 4]))
print(Solution().findSubarrays([1, 2, 3, 4, 5]))
print(Solution().findSubarrays([0, 0, 0]))
````

## 10. Vì sao thuật toán đúng

1. **Quy về trùng lặp.** Hai subarray độ dài 2 bắt đầu ở `i != j` có tổng bằng nhau ⇔ `sums[i] == sums[j]` với `sums[k] = nums[k] + nums[k+1]`. Do đó đáp án là `True` ⇔ dãy `sums` có hai chỉ số khác nhau cùng giá trị.
2. **Invariant của cách `set`.** Trước khi xử lý chỉ số `i`, `seen = {sums[0], ..., sums[i-1]}` và các phần tử này đôi một khác nhau (nếu không, đã trả về `True`).
3. **Bước xử lý.** Nếu `sums[i] in seen` thì tồn tại `j < i` với `sums[j] == sums[i]` ⇒ đáp án `True` đúng. Nếu không thì `sums[i]` khác mọi giá trị trước, thêm vào vẫn giữ invariant.
4. **Kết thúc.** Nếu duyệt hết mà không trả về `True`, mọi `sums` đôi một khác nhau, nên `False`.
5. Các cách khác (brute, sắp xếp, `len(set)`) cùng kiểm tra đúng mệnh đề "có hai chỉ số cùng giá trị trong `sums`" theo các cách khác nhau.

## 11. Phân tích Time / Space Complexity

Đặt `n = len(nums)`; có `n - 1` tổng.

| Cách | Time | Space |
|---|---|---|
| Brute so sánh mọi cặp chỉ số | `O(n^2)` | `O(1)` |
| `set` một lượt (dừng sớm) | `O(n)` trung bình | `O(n)` |
| Sắp xếp các tổng | `O(n log n)` | `O(n)` |
| `len(set)` so với `len(list)` | `O(n)` trung bình | `O(n)` |

Worst case của `set` là `O(n^2)` nếu hàm băm bị va chạm nặng, nhưng với số nguyên Python trong thực tế là `O(n)`. Với `n <= 1000` mọi cách đều chạy rất nhanh.

### 12.2. Đối chiếu ngẫu nhiên (có seed)

## 13. Các lỗi người mới hay gặp

### Lỗi 1: Thêm vào `set` trước khi kiểm tra

````python
from typing import List

def sai_them_truoc(nums: List[int]) -> bool:
    seen = set()
    for a, b in zip(nums, nums[1:]):
        s = a + b
        seen.add(s)               # SAI: thêm trước
        if s in seen:
            return True
    return False

nums = [1, 2, 3, 4, 5]
print("đúng:", Solution().findSubarrays(nums), "| sai:", sai_them_truoc(nums))
assert Solution().findSubarrays(nums) != sai_them_truoc(nums)
````

**Vì sao sai:** sau `seen.add(s)` thì `s in seen` **luôn đúng**, nên hàm luôn trả về `True` ngay ở subarray đầu tiên. Phải kiểm tra trước, thêm sau.

### Lỗi 2: Cho hai chỉ số trùng nhau (cùng một subarray)

````python
def sai_cung_chi_so(nums):
    n = len(nums)
    for i in range(n - 1):
        for j in range(i, n - 1):          # SAI: j bắt đầu từ i
            if nums[i] + nums[i + 1] == nums[j] + nums[j + 1]:
                return True
    return False

nums = [1, 2, 3, 4, 5]
print("đúng:", Solution().findSubarrays(nums), "| sai:", sai_cung_chi_so(nums))
assert Solution().findSubarrays(nums) != sai_cung_chi_so(nums)
````

**Vì sao sai:** khi `j == i`, ta so sánh một subarray với **chính nó**, luôn bằng nhau. Đề yêu cầu hai subarray bắt đầu ở **hai chỉ số khác nhau**, nên `j` phải bắt đầu từ `i + 1`.

### Lỗi 3: So sánh phần tử thay vì tổng

````python
def sai_so_phan_tu(nums):
    seen = set()
    for x in nums:
        if x in seen:                       # SAI: đang tìm phần tử trùng, không phải tổng trùng
            return True
        seen.add(x)
    return False

nums = [1, 5, 2, 4]          # tổng cặp: 6, 7, 6 -> True; nhưng không có phần tử nào trùng
print("đúng:", Solution().findSubarrays(nums), "| sai:", sai_so_phan_tu(nums))
assert Solution().findSubarrays(nums) != sai_so_phan_tu(nums)
````

**Vì sao sai:** `[1, 5, 2, 4]` có hai subarray `[1,5]` và `[2,4]` cùng tổng `6` ⇒ đáp án `True`, nhưng **không** có phần tử nào xuất hiện hai lần nên code sai trả `False`. Ngược lại `[3, 3, 9]` có phần tử trùng mà tổng cặp `6, 12` khác nhau. Bài hỏi về **tổng** của cặp liền kề, không phải về phần tử trùng.

### Lỗi 4: Bỏ sót cặp cuối vì `range(n - 2)`

````python
def sai_thieu_cap_cuoi(nums):
    seen = set()
    for i in range(len(nums) - 2):          # SAI: thiếu cặp (n-2, n-1)
        s = nums[i] + nums[i + 1]
        if s in seen:
            return True
        seen.add(s)
    return False

nums = [4, 2, 4]
print("đúng:", Solution().findSubarrays(nums), "| sai:", sai_thieu_cap_cuoi(nums))
assert Solution().findSubarrays(nums) != sai_thieu_cap_cuoi(nums)
````

**Vì sao sai:** `n` phần tử có `n - 1` cặp liền kề; chỉ số bắt đầu chạy `0..n-2`, tức `range(n - 1)`. Dùng `range(n - 2)` bỏ mất cặp cuối cùng, nên `[4, 2, 4]` chỉ còn một tổng và trả về `False`.

## 14. Mẹo làm bài nhanh & nhận dạng pattern

- Thấy "có hai ... bằng nhau, ở vị trí khác nhau" → **phát hiện trùng lặp** → `set` / `Counter`.
- Biến đổi bài toán trước: tính danh sách tổng cặp liền kề, rồi bài toán chỉ còn là "có phần tử nào xuất hiện hai lần".
- `zip(nums, nums[1:])` là cách ngắn gọn để duyệt cặp liền kề, tránh sai chỉ số.
- Khi `set` lớn hơn nhiều, nhớ thứ tự **kiểm tra rồi mới thêm**.
- Nếu cần biết **mỗi tổng xuất hiện bao nhiêu lần**, dùng `Counter(sums)` rồi `max(c.values()) >= 2`.
- Test nhanh bằng `[0, 0, 0]` (trùng giá trị) và `[1, 1]` (chỉ một subarray) để chắc chắn xử lý đúng biên.

## 15. Biến thể của bài

1. **Subarray độ dài `k` bất kỳ:** dùng tổng cửa sổ trượt (cộng thêm phần tử mới, bớt phần tử cũ) rồi `set`.
2. **Hai subarray không chồng lên nhau:** đòi hỏi `|i - j| >= 2` cho độ dài 2; cần lưu vị trí đầu tiên của mỗi tổng.
3. **Đếm số cặp subarray có tổng bằng nhau:** `Counter` rồi cộng `c * (c - 1) // 2` cho mỗi tổng.
4. **Subarray có tổng bằng `0`** (liên quan tổng tiền tố trùng nhau): tổng tiền tố + `set`.
5. **Tìm tất cả các tổng lặp lại** thay vì chỉ trả `True/False`.
6. **Phiên bản với hai mảng** `nums1`, `nums2`: hỏi có subarray ở mảng này trùng tổng với subarray ở mảng kia không.

## 19. Đáp án tham khảo cho bài tập

Chạy kiểm tra bằng `assert`.

````python
from collections import Counter

# Bài 1: tổng 8, 7, 8 -> True
assert Solution().findSubarrays([7, 1, 6, 2]) is True

# Bài 2: 4 tổng cùng bằng 6 -> C(4, 2) = 6 cặp
def dem_cap_bang_nhau(nums):
    c = Counter(a + b for a, b in zip(nums, nums[1:]))
    return sum(v * (v - 1) // 2 for v in c.values())

assert dem_cap_bang_nhau([3, 3, 3, 3, 3]) == 6

# Bài 3
def co_tong_lap(sums):
    seen = set()
    for s in sums:
        if s in seen:
            return True
        seen.add(s)
    return False

# Bài 5
def findSubarraysK(nums, k):
    n = len(nums)
    cur = sum(nums[:k])
    seen = {cur}
    for i in range(1, n - k + 1):
        cur += nums[i + k - 1] - nums[i - 1]
        if cur in seen:
            return True
        seen.add(cur)
    return False

assert co_tong_lap([6, 6, 7]) is True
assert co_tong_lap([3, 5, 7, 9]) is False
assert co_tong_lap([]) is False
assert dem_cap_bang_nhau([1, 2, 3, 4, 5]) == 0
assert dem_cap_bang_nhau([0, 0, 0]) == 1
assert findSubarraysK([4, 2, 4], 2) is True
assert findSubarraysK([1, 2, 3, 4, 5], 2) is False
assert findSubarraysK([1, 2, 3, 1, 2, 3], 3) is True      # [1,2,3] và [1,2,3] -> tổng 6
assert findSubarraysK([1, 2, 3], 3) is False              # chỉ có một cửa sổ
assert all(findSubarraysK(a, 2) == Solution().findSubarrays(a) for a in ([4, 2, 4], [0, 0, 0], [1, 2, 3, 4], [9, 9]))
print("tất cả bài tập OK")
````
