6. Tránh các lỗi quen thuộc: chọn nhầm thứ tự, quên các tờ `0` che chắn cho `-1`, quên giới hạn số lượng từng loại.

## 3. Kiến thức nền

### 3.1. Nhân list: `[x] * n`

`[1] * 3` tạo list `[1, 1, 1]` (lặp phần tử 3 lần). Nối list bằng `+`. Vậy `[1] * a + [0] * b + [-1] * c` dựng cả cái túi, đã sắp **giảm dần** (1 trước, rồi 0, rồi -1).

````python
items = [1] * 3 + [0] * 2 + [-1] * 4
print(items, "độ dài", len(items))
print([1] * 0, [0] * 0)          # nhân với 0 cho list rỗng
````

### 3.2. Cắt lát `items[:k]` và `sum`

`items[:k]` lấy `k` phần tử đầu (không có lỗi nếu `k = 0`: được list rỗng). `sum(...)` cộng các phần tử, tổng của list rỗng là `0`.

````python
items = [1, 1, 1, 0, 0, -1, -1]
print(items[:4], sum(items[:4]))
print(items[:0], sum(items[:0]))
print(items[:7], sum(items[:7]))
````

### 3.3. `min` và `max` để kẹp số lượng

`min(k, numOnes)` = số tờ `1` thực sự lấy được (không thể nhiều hơn `k` hay `numOnes`). `max(0, x)` đưa số âm về `0` ("không bị thiếu thì phải lấy thêm 0 tờ").

````python
numOnes, k = 3, 2
print(min(k, numOnes))           # 2
numOnes, k = 3, 10
print(min(k, numOnes))           # 3
print(max(0, -5), max(0, 4))     # 0 4
````

### 3.4. Chuỗi `if / elif / else` theo khoảng

Khi cần chia theo khoảng giá trị của `k` (ít, vừa, nhiều) ta dùng chuỗi điều kiện, **kiểm tra từ khoảng nhỏ đến lớn**; nhánh nào đúng đầu tiên sẽ chạy và các nhánh sau bị bỏ qua.

````python
def bucket(x):
    if x <= 3:
        return "ít"
    elif x <= 7:
        return "vừa"
    else:
        return "nhiều"

print([bucket(v) for v in (0, 3, 4, 7, 8)])
````

### 3.5. `itertools.product` / vòng lặp lồng để duyệt mọi cách chọn

Để vét cạn, ta duyệt mọi cặp `(a, b)` (số tờ `1` và số tờ `0` lấy), suy ra `c = k - a - b` là số tờ `-1`. Hai vòng `for` lồng nhau là đủ.

````python
k = 3
for a in range(0, k + 1):
    for b in range(0, k - a + 1):
        c = k - a - b
        print(f"a={a} b={b} c={c}  (a+b+c={a + b + c})")
````

## 4. Ý tưởng nền tảng / cách nghĩ trực giác

Bạn được lấy `k` tờ và muốn tổng **lớn nhất**:

1. Tờ `1` cộng thêm điểm → lấy **càng nhiều càng tốt**.
2. Hết `1` mà còn phải lấy → lấy tờ `0` (không cộng, không trừ).
3. Hết cả `0` mà vẫn còn phải lấy → buộc phải lấy tờ `-1` (mỗi tờ trừ 1).

Ví dụ `numOnes = 2`, `numZeros = 1`, `numNegOnes = 3`, `k = 5`:

```
túi (sắp giảm dần): 1 1 0 -1 -1 -1
lấy 5 tờ đầu       : 1 1 0 -1 -1   -> tổng = 2 + 0 - 2 = 0
```

````python
bag = [1, 1, 0, -1, -1, -1]
k = 5
chosen = bag[:k]
print("lấy:", chosen, "tổng =", sum(chosen))
````

## 5. Cách 1 — Brute Force / baseline

Thử **mọi cách** chọn số lượng `a` tờ `1`, `b` tờ `0`, `c` tờ `-1` với `a + b + c = k` và `a <= numOnes`, `b <= numZeros`, `c <= numNegOnes`. Với mỗi cách hợp lệ, tổng là `a - c`; lấy lớn nhất.

````python
def k_items_bruteforce(numOnes: int, numZeros: int, numNegOnes: int, k: int) -> int:
    best = None
    for a in range(0, numOnes + 1):
        for b in range(0, numZeros + 1):
            c = k - a - b                         # phần còn lại phải lấy từ các tờ -1
            if 0 <= c <= numNegOnes:
                total = a * 1 + b * 0 + c * (-1)
                if best is None or total > best:
                    best = total
    return best
````

````python
print(k_items_bruteforce(3, 2, 0, 2))     # 2
print(k_items_bruteforce(3, 2, 0, 4))     # 3
print(k_items_bruteforce(2, 1, 3, 5))     # 0
````

**Độ phức tạp:** hai vòng lặp `numOnes + 1` và `numZeros + 1` → Time `O(numOnes * numZeros)` (tối đa `51 * 51`); Space `O(1)`.

**Nút thắt:** ta thử cả những cách chọn **chắc chắn không tối ưu** (ví dụ lấy `-1` khi vẫn còn `1` chưa lấy). Quan sát tham lam cho phép đi thẳng tới đáp án.

## 6. Ý tưởng thuật toán / bước đột phá

Giá trị đã sắp xếp: `1 > 0 > -1`. Muốn tổng lớn nhất thì luôn **lấy tờ có giá trị lớn nhất còn lại**. Vì chỉ có ba loại giá trị nên việc "sắp xếp" chỉ là liệt kê `1`, rồi `0`, rồi `-1`; ta lấy `k` tờ đầu.

Chia theo `k`:

- `k <= numOnes` → toàn `1`: tổng `k`.
- `numOnes < k <= numOnes + numZeros` → mọi `1` và vài `0`: tổng `numOnes`.
- `k > numOnes + numZeros` → mọi `1`, mọi `0` và `k - numOnes - numZeros` tờ `-1`: tổng `numOnes - (k - numOnes - numZeros)`.

````python
numOnes, numZeros, numNegOnes, k = 2, 1, 3, 5
if k <= numOnes:
    print("nhánh 1:", k)
elif k <= numOnes + numZeros:
    print("nhánh 2:", numOnes)
else:
    print("nhánh 3:", numOnes - (k - numOnes - numZeros))
````

## 7. Các cách tiếp theo

Ta có thêm 3 cách: **(2)** dựng "túi" thành list rồi lấy `k` phần tử đầu, **(3)** tham lam bằng `if / elif / else` (đã thấy ở mục 6), **(4)** công thức một dòng với `min`/`max`.

### Cách 2 — Dựng list rồi cắt `k` phần tử đầu

Giải thích từng đoạn:

1. `items = [1] * numOnes + [0] * numZeros + [-1] * numNegOnes` — túi đã sắp giảm dần.
2. `items[:k]` — `k` tờ có giá trị lớn nhất.
3. `sum(...)` — tổng.

Dễ nhớ nhất nhưng tạo list dài tới `numOnes + numZeros + numNegOnes` (tối đa 150 phần tử).

````python
def k_items_list(numOnes, numZeros, numNegOnes, k):
    items = [1] * numOnes + [0] * numZeros + [-1] * numNegOnes
    return sum(items[:k])

print(k_items_list(3, 2, 0, 2), k_items_list(3, 2, 0, 4), k_items_list(2, 1, 3, 5))
````

````python
class SolutionList:
    def kItemsWithMaximumSum(self, numOnes: int, numZeros: int, numNegOnes: int, k: int) -> int:
        items = [1] * numOnes + [0] * numZeros + [-1] * numNegOnes
        return sum(items[:k])
````

Độ phức tạp Cách 2: Time `O(numOnes + numZeros + numNegOnes)`; Space `O(numOnes + numZeros + numNegOnes)`.

### Cách 3 — Tham lam bằng `if / elif / else`

Ba nhánh theo `k` (mục 6). Mỗi nhánh là một công thức `O(1)`.

````python
def k_items_greedy(numOnes, numZeros, numNegOnes, k):
    if k <= numOnes:
        return k
    elif k <= numOnes + numZeros:
        return numOnes
    else:
        return numOnes - (k - numOnes - numZeros)

print(k_items_greedy(3, 2, 0, 2), k_items_greedy(3, 2, 0, 4), k_items_greedy(2, 1, 3, 5))
````

````python
class SolutionGreedy:
    def kItemsWithMaximumSum(self, numOnes: int, numZeros: int, numNegOnes: int, k: int) -> int:
        if k <= numOnes:
            return k
        elif k <= numOnes + numZeros:
            return numOnes
        else:
            return numOnes - (k - numOnes - numZeros)
````

Độ phức tạp Cách 3: Time `O(1)`; Space `O(1)`.

### Cách 4 — Công thức một dòng `min` / `max`

Số tờ `1` lấy được: `min(k, numOnes)`. Số tờ `-1` bắt buộc phải lấy: phần `k` vượt quá `numOnes + numZeros`, tức `max(0, k - numOnes - numZeros)`. Các tờ `0` không đóng góp. Vậy:

`đáp án = min(k, numOnes) - max(0, k - numOnes - numZeros)`.

````python
def k_items_formula(numOnes, numZeros, numNegOnes, k):
    return min(k, numOnes) - max(0, k - numOnes - numZeros)

print(k_items_formula(3, 2, 0, 2), k_items_formula(3, 2, 0, 4), k_items_formula(2, 1, 3, 5))
````

````python
class SolutionFormula:
    def kItemsWithMaximumSum(self, numOnes: int, numZeros: int, numNegOnes: int, k: int) -> int:
        return min(k, numOnes) - max(0, k - numOnes - numZeros)
````

Độ phức tạp Cách 4: Time `O(1)`; Space `O(1)`. Chú ý `numNegOnes` **không xuất hiện** trong công thức: vì `k <= tổng số tờ` nên luôn đủ `-1` để lấy khi cần.

## 8. TRACE từng bước

Trace Cách 3 với `numOnes = 2`, `numZeros = 1`, `numNegOnes = 3`, `k = 5`.

| bước | kiểm tra | kết quả |
|---|---|---|
| 1 | `k <= numOnes` ? `5 <= 2` | sai |
| 2 | `k <= numOnes + numZeros` ? `5 <= 3` | sai |
| 3 | nhánh `else`: số `-1` phải lấy = `5 - 2 - 1 = 2` | đáp án `2 - 2 = 0` |

Trace "lấy từng tờ" theo thứ tự giá trị giảm dần:

| lần lấy | tờ lấy | tổng |
|---|---|---|
| 1 | `1` | 1 |
| 2 | `1` | 2 |
| 3 | `0` | 2 |
| 4 | `-1` | 1 |
| 5 | `-1` | 0 |

````python
def trace_pick(numOnes, numZeros, numNegOnes, k):
    ones, zeros, negs = numOnes, numZeros, numNegOnes
    total = 0
    print(f"túi: {ones} tờ 1, {zeros} tờ 0, {negs} tờ -1; phải lấy k = {k}")
    for step in range(1, k + 1):
        if ones > 0:
            ones -= 1
            total += 1
            pick = 1
        elif zeros > 0:
            zeros -= 1
            pick = 0
        else:
            negs -= 1
            total -= 1
            pick = -1
        print(f"lần {step}: lấy {pick:>2} -> tổng = {total} (còn 1:{ones}, 0:{zeros}, -1:{negs})")
    return total

trace_pick(2, 1, 3, 5)
trace_pick(3, 2, 0, 4)
````

````python
def trace_greedy(numOnes, numZeros, numNegOnes, k):
    print(f"numOnes={numOnes}, numZeros={numZeros}, numNegOnes={numNegOnes}, k={k}")
    if k <= numOnes:
        print(f"  k <= numOnes -> toàn tờ 1 -> {k}")
        return k
    if k <= numOnes + numZeros:
        print(f"  k <= numOnes + numZeros = {numOnes + numZeros} -> tất cả 1 rồi thêm 0 -> {numOnes}")
        return numOnes
    need_neg = k - numOnes - numZeros
    print(f"  phải lấy thêm {need_neg} tờ -1 -> {numOnes} - {need_neg} = {numOnes - need_neg}")
    return numOnes - need_neg

trace_greedy(2, 1, 3, 5)
trace_greedy(3, 2, 0, 4)
trace_greedy(3, 2, 0, 2)
````

## 9. Code nộp LeetCode

Nộp Cách 3 (tham lam `if / elif / else`): trực quan và là `O(1)`.

````python
class Solution:
    def kItemsWithMaximumSum(self, numOnes: int, numZeros: int, numNegOnes: int, k: int) -> int:
        if k <= numOnes:
            return k                                   # lấy toàn tờ 1
        if k <= numOnes + numZeros:
            return numOnes                             # lấy hết tờ 1, phần còn lại là tờ 0
        return numOnes - (k - numOnes - numZeros)      # còn phải lấy thêm tờ -1

print(Solution().kItemsWithMaximumSum(3, 2, 0, 2))   # 2
print(Solution().kItemsWithMaximumSum(3, 2, 0, 4))   # 3
````

## 10. Vì sao thuật toán đúng

- **Tham lam (exchange argument):** giả sử một cách chọn tối ưu lấy một tờ có giá trị `v` nhỏ hơn trong khi còn tờ `w > v` chưa lấy. Đổi tờ `v` lấy tờ `w` vẫn lấy đúng `k` tờ và tổng tăng thêm `w - v > 0` — mâu thuẫn với tính tối ưu. Vậy cách tối ưu luôn lấy các tờ theo thứ tự `1`, rồi `0`, rồi `-1`: tức là lấy `k` phần tử đầu của list sắp giảm dần (Cách 2).
- **Ba nhánh:**
  - `k <= numOnes`: `k` tờ đầu đều là `1` → tổng `k`.
  - `numOnes < k <= numOnes + numZeros`: có đúng `numOnes` tờ `1`, còn lại là `0` → tổng `numOnes`.
  - `k > numOnes + numZeros`: có `numOnes` tờ `1`, `numZeros` tờ `0` và `k - numOnes - numZeros` tờ `-1`. Vì ràng buộc `k <= numOnes + numZeros + numNegOnes` nên số tờ `-1` này không vượt `numNegOnes` → tổng `numOnes - (k - numOnes - numZeros)`.
- **Công thức một dòng:** nhánh 1: `min(k, numOnes) = k` và `max(0, k - numOnes - numZeros) = 0` (vì `k <= numOnes`). Nhánh 2: `min = numOnes`, `max = 0`. Nhánh 3: `min = numOnes`, `max = k - numOnes - numZeros > 0`. Khớp ba nhánh.

## 11. Phân tích Time / Space Complexity

Gọi `a = numOnes`, `b = numZeros`, `c = numNegOnes`.

| Cách | Time | Space |
|---|---|---|
| 1. Vét cạn số lượng `(a', b')` | `O(a * b)` | `O(1)` |
| 2. Dựng list + cắt `k` đầu | `O(a + b + c)` | `O(a + b + c)` |
| 3. Tham lam `if / elif` | `O(1)` | `O(1)` |
| 4. Công thức `min`/`max` | `O(1)` | `O(1)` |

Với giới hạn `50` mọi cách đều tức thì; cách 3, 4 không phụ thuộc kích thước túi.

## 12. Bộ test

Test 1: ví dụ đề và edge case: `k = 0`, `k` bằng tổng số tờ, túi chỉ có một loại, các biên giữa các nhánh (`k = numOnes`, `k = numOnes + numZeros`).

````python
impls = [
    ("bruteforce", k_items_bruteforce),
    ("list", k_items_list),
    ("greedy", k_items_greedy),
    ("formula", k_items_formula),
    ("class_list", SolutionList().kItemsWithMaximumSum),
    ("class_greedy", SolutionGreedy().kItemsWithMaximumSum),
    ("class_formula", SolutionFormula().kItemsWithMaximumSum),
    ("submit", Solution().kItemsWithMaximumSum),
]
tests = [
    (3, 2, 0, 2, 2),
    (3, 2, 0, 4, 3),
    (2, 1, 3, 5, 0),
    (5, 5, 5, 0, 0),         # k = 0
    (0, 0, 4, 4, -4),        # chỉ có tờ -1
    (0, 5, 0, 3, 0),         # chỉ có tờ 0
    (4, 0, 0, 4, 4),         # chỉ có tờ 1, lấy hết
    (3, 2, 4, 3, 3),         # k = numOnes (biên nhánh 1 / 2)
    (3, 2, 4, 5, 3),         # k = numOnes + numZeros (biên nhánh 2 / 3)
    (3, 2, 4, 6, 2),         # vượt biên một đơn vị: lấy thêm 1 tờ -1
    (4, 0, 5, 6, 2),         # không có tờ 0: 4 tờ 1 và 2 tờ -1
    (50, 50, 50, 150, 50 - 50),   # lấy hết 150 tờ: tổng = 50 - 50 = 0
    (50, 50, 50, 149, 1),    # 50 tờ 1, 50 tờ 0, 49 tờ -1: 50 - 49 = 1
]
for name, f in impls:
    for a, b, c, k, expected in tests:
        got = f(a, b, c, k)
        assert got == expected, (name, a, b, c, k, got, expected)
    print(name, "PASS", len(tests), "test")
````

Test 2: **toàn bộ** các bộ giá trị nhỏ — mọi `(numOnes, numZeros, numNegOnes)` trong `0..6` và mọi `k` hợp lệ — so với vét cạn bằng **tổ hợp thật** (`itertools.combinations` trên các chỉ số): lấy mọi tập `k` tờ và tính tổng lớn nhất.

````python
from itertools import combinations

def oracle_combinations(a, b, c, k):
    items = [1] * a + [0] * b + [-1] * c
    best = None
    for idx in combinations(range(len(items)), k):
        total = sum(items[i] for i in idx)
        if best is None or total > best:
            best = total
    return best

count = 0
for a in range(0, 7):
    for b in range(0, 7):
        for c in range(0, 7):
            n_items = a + b + c
            for k in range(0, n_items + 1):
                if n_items > 12:
                    continue                    # giữ số tổ hợp nhỏ (<= 924 tập con)
                expected = oracle_combinations(a, b, c, k)
                for name, f in impls:
                    assert f(a, b, c, k) == expected, (name, a, b, c, k)
                count += 1
print("đã so khớp", count, "bộ (a, b, c, k) với oracle tổ hợp")
````

Test 3: ngẫu nhiên có seed, **đúng constraints** (`0..50` mỗi loại, `0 <= k <= tổng`); oracle là "xáo trộn rồi sắp giảm dần, lấy `k` đầu".

````python
import random
random.seed(2600)

for _ in range(3000):
    a, b, c = (random.randint(0, 50) for _ in range(3))
    k = random.randint(0, a + b + c)
    bag = [1] * a + [0] * b + [-1] * c
    random.shuffle(bag)
    expected = sum(sorted(bag, reverse=True)[:k])
    for name, f in impls:
        assert f(a, b, c, k) == expected, (name, a, b, c, k)
print("3000 test ngẫu nhiên: tất cả cách đều khớp oracle (xáo trộn + sắp giảm dần)")
````

## 13. Các lỗi người mới hay gặp

### Lỗi 1: lấy theo thứ tự `-1, 0, 1` (sắp tăng dần) thay vì giảm dần

````python
def wrong_ascending(numOnes, numZeros, numNegOnes, k):
    items = [-1] * numNegOnes + [0] * numZeros + [1] * numOnes     # sắp TĂNG dần
    return sum(items[:k])

print(wrong_ascending(3, 2, 0, 2), "vs đúng", Solution().kItemsWithMaximumSum(3, 2, 0, 2))
print(wrong_ascending(2, 1, 3, 5), "vs đúng", Solution().kItemsWithMaximumSum(2, 1, 3, 5))
````

Vì sao sai: bài hỏi tổng **lớn nhất**, nên phải lấy tờ lớn nhất trước (`1`, rồi `0`, rồi `-1`). Lấy từ nhỏ nhất cho ra tổng **nhỏ nhất** có thể (bài tập 5 chính là bài đó).

### Lỗi 2: quên các tờ `0` — trừ `-1` ngay khi hết tờ `1`

````python
def wrong_no_zeros(numOnes, numZeros, numNegOnes, k):
    if k <= numOnes:
        return k
    return numOnes - (k - numOnes)       # SAI: coi như sau các tờ 1 là tới -1 luôn

print(wrong_no_zeros(3, 2, 0, 4), "vs đúng", Solution().kItemsWithMaximumSum(3, 2, 0, 4))
````

Vì sao sai: sau khi hết tờ `1`, ta còn `numZeros` tờ `0` để "đệm" — lấy chúng không làm tổng giảm. Chỉ khi hết cả `0` mới phải lấy `-1`. Số `-1` phải lấy là `k - numOnes - numZeros`, không phải `k - numOnes`.

### Lỗi 3: quên kẹp số lượng — luôn trả về `k`

````python
def wrong_always_k(numOnes, numZeros, numNegOnes, k):
    return k          # giả sử toàn tờ 1

print(wrong_always_k(1, 2, 3, 4), "vs đúng", Solution().kItemsWithMaximumSum(1, 2, 3, 4))
````

Vì sao sai: chỉ có `numOnes` tờ `1`. Nếu `k > numOnes` thì phải lấy thêm tờ `0` hoặc `-1`; tổng không thể vượt `numOnes`. Công thức dùng `min(k, numOnes)`.

### Lỗi 4: trừ nhầm số tờ `-1` bằng tổng `numNegOnes`

````python
def wrong_all_neg(numOnes, numZeros, numNegOnes, k):
    if k <= numOnes + numZeros:
        return min(k, numOnes)
    return numOnes - numNegOnes          # SAI: chỉ phải lấy k - numOnes - numZeros tờ -1

print(wrong_all_neg(2, 1, 5, 4), "vs đúng", Solution().kItemsWithMaximumSum(2, 1, 5, 4))
````

Vì sao sai: ta chỉ lấy đúng `k` tờ, không phải **tất cả** tờ `-1`. Ví dụ `(2, 1, 5, 4)`: chỉ cần lấy `1` tờ `-1` (tổng `2 - 1 = 1`), không phải `5` tờ.

## 14. Mẹo làm bài nhanh & nhận dạng pattern

- "Chọn đúng `k` phần tử để **tối đa** tổng" → sắp giá trị giảm dần và lấy `k` đầu (**tham lam**). Chỉ có vài loại giá trị → thay sort bằng công thức theo khoảng.
- Gặp bài chỉ có ít loại giá trị (`1, 0, -1`) → chia theo khoảng của `k` bằng `if / elif`.
- Công thức nhớ nhanh: `đáp án = min(k, ones) - max(0, k - ones - zeros)`.
- Kiểm tra biên: `k = 0`; `k` đúng bằng tổng các nhóm đầu; `k` bằng tổng tất cả tờ.
- Khi không chắc thuật toán tham lam, luôn kiểm tra bằng vét cạn tổ hợp trên bộ dữ liệu nhỏ (như mục 12).

## 15. Biến thể của bài

1. **Tổng nhỏ nhất** khi lấy đúng `k` tờ (bài tập 5).
2. Tờ có **giá trị bất kỳ** (không chỉ `1, 0, -1`): sắp xếp giảm dần rồi lấy `k` đầu, hoặc dùng `heapq.nlargest(k, ...)`.
3. Mỗi tờ có thêm **trọng lượng** và chỉ được lấy tối đa trọng lượng `W` (đổi thành bài knapsack).
4. Lấy **không quá** `k` tờ (không bắt buộc đúng `k`): chỉ lấy tờ `1` rồi dừng.
5. Có `m` loại tờ với giá trị khác nhau, mỗi loại có số lượng riêng: sắp theo giá trị, lấy lần lượt.
6. Bài tương tự về chọn `k` phần tử: "Maximum Sum With Exactly K Elements" (2656), "Maximum Score From Removing Substrings", "Kth Largest Element in an Array" (215).

## 16. So sánh các cách làm

| Cách | Ý tưởng | Time | Space | Khi nào dùng |
|---|---|---|---|---|
| 1. Vét cạn số lượng | thử mọi `(a, b, c)` có `a + b + c = k` | `O(a * b)` | `O(1)` | oracle / hiểu đề |
| 2. Dựng list + cắt | `[1]*a + [0]*b + [-1]*c`, `sum(items[:k])` | `O(a + b + c)` | `O(a + b + c)` | dễ nhớ nhất |
| 3. Tham lam `if / elif` | ba nhánh theo `k` | `O(1)` | `O(1)` | cách nộp, dễ giải thích |
| 4. Công thức `min`/`max` | `min(k, a) - max(0, k - a - b)` | `O(1)` | `O(1)` | khi cần một dòng |
