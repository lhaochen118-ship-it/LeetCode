## 2. Nền tảng Python cho bài này

### 2.1. List là **mutable** (sửa được); biến chỉ là "nhãn dán"

Trong Python, biến là một **tên trỏ tới đối tượng**. Khi LeetCode gọi `reverseString(s)`, tham số `s` bên trong hàm và biến `s` của LeetCode **cùng trỏ tới một list**.

- `s[i] = x` hoặc `s[:] = ...` → **sửa nội dung** list đó → bên ngoài thấy thay đổi.
- `s = ...` → chỉ **dán nhãn `s` sang đối tượng khác** bên trong hàm → bên ngoài **không** thấy gì.

Hàm `id(x)` cho biết "địa chỉ" (danh tính) của đối tượng.

````python
def reassign(s):
    s = s[::-1]                 # gán lại tên s cục bộ
    print("  trong hàm (gán lại):", s, "| id =", id(s))

def modify(s):
    s[:] = s[::-1]              # sửa nội dung list gốc
    print("  trong hàm (sửa nội dung):", s, "| id =", id(s))

a = ["a", "b", "c"]
print("id gốc =", id(a))
reassign(a)
print("bên ngoài sau reassign:", a)
modify(a)
print("bên ngoài sau modify  :", a)
````

## 3. Cách trực giác — tạo mảng đảo ngược rồi chép lại

Người mới thường nghĩ: "tạo một list mới chứa các phần tử theo thứ tự ngược, rồi chép vào `s`". Cách này **đúng kết quả** nhưng dùng thêm **O(n) bộ nhớ** — vi phạm yêu cầu O(1).

````python
def reverse_with_copy(s: list[str]) -> None:
    reversed_copy = []
    for i in range(len(s) - 1, -1, -1):     # range(start, stop, step): đi lùi từ len-1 về 0
        reversed_copy.append(s[i])
    for i in range(len(s)):
        s[i] = reversed_copy[i]

s = ["h", "e", "l", "l", "o"]
reverse_with_copy(s)
print(s)
print(list(range(4, -1, -1)))   # giải thích range đi lùi: [4, 3, 2, 1, 0]
````

## 4. Nút thắt

- Bộ nhớ phụ `O(n)` cho `reversed_copy` — đề cấm.
- Đi qua dữ liệu **hai lần** (tạo bản sao, rồi chép lại).

Câu hỏi: có thể đặt mỗi phần tử vào **đúng chỗ** mà không cần mảng phụ không? Nếu ghi đè `s[0] = s[4]` thì mất giá trị cũ của `s[0]`… trừ khi ta **hoán đổi** `s[0]` với `s[4]` — cả hai đều về đúng chỗ cùng lúc!

## 5. Insight — hoán đổi từng cặp đối xứng từ ngoài vào trong

Vị trí `i` và vị trí `n - 1 - i` là một **cặp đối xứng**. Đảo ngược = hoán đổi mọi cặp đối xứng một lần:

```text
h e l l o          cặp (0,4): h <-> o   ->  o e l l h
^       ^
o e l l h          cặp (1,3): e <-> l   ->  o l l e h
  ^   ^
o l l e h          left = right = 2: phần tử giữa đứng yên -> dừng
    ^
```

Hai con trỏ `left` (từ trái) và `right` (từ phải) chính là cặp `(i, n - 1 - i)`. Chỉ cần đi **nửa mảng**.

## 6. Lời giải chính — giảng từng đoạn rồi ghép

**Đoạn 1 — đặt hai con trỏ ở hai đầu:**
```python
left, right = 0, len(s) - 1
```

**Đoạn 2 — lặp khi chưa gặp nhau:**
```python
while left < right:
```
Dùng `<` chứ không phải `<=`: khi `left == right` (phần tử giữa của mảng độ dài lẻ), hoán đổi với chính nó là thừa.

**Đoạn 3 — hoán đổi và tiến vào giữa:**
```python
    s[left], s[right] = s[right], s[left]
    left += 1
    right -= 1
```

**Đoạn 4 — không `return` gì** (mặc định trả `None`).

````python
def reverse_two_pointers(s: list[str]) -> None:
    left, right = 0, len(s) - 1
    while left < right:
        s[left], s[right] = s[right], s[left]
        left += 1
        right -= 1

for data in [list("hello"), list("Hannah"), list("a"), list("ab")]:
    before = data.copy()
    result = reverse_two_pointers(data)
    print(before, "->", data, "| giá trị trả về:", result)
````

## 7. TRACE từng bước

Bảng tay cho `s = ["H","a","n","n","a","h"]` (độ dài chẵn):

| Vòng | left | right | s[left] | s[right] | s sau khi hoán đổi |
|---:|---:|---:|:---:|:---:|---|
| 1 | 0 | 5 | H | h | h a n n a H |
| 2 | 1 | 4 | a | a | h a n n a H (đổi 2 ký tự giống nhau) |
| 3 | 2 | 3 | n | n | h a n n a H |
| — | 3 | 2 | | | dừng vì left > right |

````python
def trace_reverse(s: list[str]) -> None:
    left, right = 0, len(s) - 1
    step = 1
    print("bắt đầu:", "".join(s))
    while left < right:
        a, c = s[left], s[right]
        s[left], s[right] = s[right], s[left]
        print(f"vòng {step}: left={left}, right={right}, đổi {a!r} <-> {c!r} -> {''.join(s)}")
        left += 1
        right -= 1
        step += 1
    print(f"dừng: left={left}, right={right} ({'gặp nhau ở phần tử giữa' if left == right else 'đã vượt qua nhau'})")

trace_reverse(list("Hannah"))
print()
trace_reverse(list("hello"))      # độ dài lẻ: phần tử giữa 'l' đứng yên
````

## 8. Các lời giải khác

### Cách 3 — vòng `for` trên nửa đầu
Chỉ số `i` chạy `0 .. n//2 - 1`, đối xứng là `n - 1 - i`. Với `n` lẻ, `n // 2` tự bỏ qua phần tử giữa.

````python
def reverse_half_for(s: list[str]) -> None:
    n = len(s)
    for i in range(n // 2):
        s[i], s[n - 1 - i] = s[n - 1 - i], s[i]

for n in [5, 6]:
    print(f"n = {n}: n // 2 = {n // 2}, các cặp:", [(i, n - 1 - i) for i in range(n // 2)])
t = list("hello"); reverse_half_for(t); print(t)
````

### Cách 4 — `s.reverse()` (method dựng sẵn, in-place)
`list.reverse()` đảo ngay trên list, trả về `None`. Bên trong nó cũng dùng ý tưởng hoán đổi đối xứng (cài bằng C).

### Cách 5 — `s[:] = s[::-1]` (slice assignment)
`s[::-1]` tạo list mới (O(n) bộ nhớ tạm), rồi `s[:] = ...` chép nội dung vào list gốc. Đúng kết quả, đúng "in-place" về mặt đối tượng, nhưng **không** O(1) bộ nhớ.

````python
t = list("hello")
print(t.reverse(), t)          # None ['o', 'l', 'l', 'e', 'h']

u = list("hello")
original_id = id(u)
u[:] = u[::-1]
print(u, "| vẫn là list cũ?", id(u) == original_id)
````

### Cách 6 — đệ quy
`helper(left, right)`: nếu `left >= right` thì dừng; ngược lại hoán đổi rồi gọi `helper(left + 1, right - 1)`. Dùng `O(n)` stack — với `n = 10^5` sẽ vượt giới hạn đệ quy mặc định của Python (~1000)!

````python
def reverse_recursive(s: list[str]) -> None:
    def helper(left: int, right: int) -> None:
        if left >= right:
            return
        s[left], s[right] = s[right], s[left]
        helper(left + 1, right - 1)
    helper(0, len(s) - 1)

t = list("recursion"); reverse_recursive(t); print("".join(t))
try:
    reverse_recursive(["x"] * 10**5)
except RecursionError as e:
    print("RecursionError với n = 10^5:", str(e)[:50])
````

## 9. Code nộp LeetCode

### Cách chính — hai con trỏ

````python
class Solution:
    def reverseString(self, s: list[str]) -> None:
        """
        Do not return anything, modify s in-place instead.
        """
        left, right = 0, len(s) - 1
        while left < right:
            s[left], s[right] = s[right], s[left]
            left += 1
            right -= 1

s = ["h", "e", "l", "l", "o"]
Solution().reverseString(s)
print(s)
````

### Cách nộp ngắn — `s.reverse()`

````python
class SolutionBuiltin:
    def reverseString(self, s: list[str]) -> None:
        """
        Do not return anything, modify s in-place instead.
        """
        s.reverse()

s = ["H", "a", "n", "n", "a", "h"]
SolutionBuiltin().reverseString(s)
print(s)
````

## 10. Vì sao đúng?

**Invariant (trước mỗi lần kiểm tra `while`):**
- Mọi vị trí `i < left` đã chứa ký tự gốc ở vị trí `n - 1 - i`.
- Mọi vị trí `j > right` đã chứa ký tự gốc ở vị trí `n - 1 - j`.
- Đoạn `s[left..right]` vẫn còn **nguyên như ban đầu**.
- Luôn có `left + right == n - 1` (hai con trỏ đối xứng).

**Ban đầu** `left = 0`, `right = n - 1`: hai vùng "đã xong" rỗng ✔.

**Mỗi vòng** hoán đổi `s[left]` và `s[right]` — đúng cặp đối xứng — rồi thu hẹp cả hai phía: invariant giữ nguyên ✔.

**Khi dừng** `left >= right`: đoạn chưa xử lý có nhiều nhất 1 phần tử (phần tử giữa), mà phần tử giữa đối xứng với chính nó ✔. Vậy mọi vị trí đã đúng.

Mỗi cặp bị hoán đổi **đúng một lần** — không bị đổi lại lần hai (lỗi xảy ra nếu duyệt `for i in range(n)` hết mảng).

## 12. Bộ test — mẫu đề, biên, đối chiếu

Vì hàm trả `None`, ta kiểm tra **chính list sau khi gọi** và kiểm tra giá trị trả về là `None`.

````python
def run(func, data):
    s = list(data)
    ret = func(s)
    assert ret is None, "hàm phải trả về None"
    return s

def check(name, actual, expected):
    assert actual == expected, (name, actual, expected)
    print(f"PASS {name}: {''.join(actual)!r}")

f = Solution().reverseString
check("đề 1", run(f, ["h", "e", "l", "l", "o"]), ["o", "l", "l", "e", "h"])
check("đề 2", run(f, ["H", "a", "n", "n", "a", "h"]), ["h", "a", "n", "n", "a", "H"])
check("1 ký tự", run(f, ["a"]), ["a"])
check("2 ký tự", run(f, ["a", "b"]), ["b", "a"])
check("ký tự đặc biệt", run(f, list("A man, a plan!")), list("!nalp a ,nam A"))
check("palindrome", run(f, list("racecar")), list("racecar"))
big = [chr(33 + i % 94) for i in range(10**5)]
assert run(f, big) == big[::-1]
print("PASS n = 10^5")
````

````python
def wrong_reassign(s):
    s = s[::-1]            # SAI: chỉ đổi biến cục bộ

def wrong_return(s):
    return s[::-1]         # SAI: đề không đọc giá trị trả về, list gốc không đổi

for fn in (wrong_reassign, wrong_return):
    t = list("hello")
    fn(t)
    print(fn.__name__, "->", t)
````

### Lỗi 2: Duyệt hết mảng thay vì nửa mảng
Mỗi cặp bị hoán đổi **hai lần** → trở về như cũ.

### Lỗi 3: Hoán đổi bằng hai lệnh gán tách rời
`s[left] = s[right]; s[right] = s[left]` → giá trị cũ của `s[left]` bị mất, cả hai thành giống nhau.

````python
def wrong_full_loop(s):
    n = len(s)
    for i in range(n):                    # SAI: phải là range(n // 2)
        s[i], s[n - 1 - i] = s[n - 1 - i], s[i]

def wrong_swap(s):
    left, right = 0, len(s) - 1
    while left < right:
        s[left] = s[right]                # SAI: mất giá trị cũ của s[left]
        s[right] = s[left]
        left += 1
        right -= 1

t = list("hello"); wrong_full_loop(t); print("duyệt hết mảng:", "".join(t))
t = list("hello"); wrong_swap(t);      print("hoán đổi tách rời:", "".join(t))
````

### Lỗi 4: Quên cập nhật con trỏ → vòng lặp vô hạn
Thiếu `left += 1` hoặc `right -= 1` thì `left < right` đúng mãi. (Không demo vì sẽ treo kernel — hãy luôn kiểm tra mỗi vòng `while` có làm điều kiện tiến tới sai không.)

### Lỗi 5: `right = len(s)` thay vì `len(s) - 1`
`s[len(s)]` → `IndexError`.

````python
try:
    t = list("abc")
    left, right = 0, len(t)              # SAI
    t[left], t[right] = t[right], t[left]
except IndexError as e:
    print("IndexError:", e)
````

## 15. Biến thể

### Biến thể 1 — đảo một đoạn `[i, j]` của mảng
### Biến thể 2 — Rotate Array (LeetCode 189) bằng 3 lần đảo
Xoay phải `k` bước = đảo cả mảng, đảo `k` phần tử đầu, đảo phần còn lại.
### Biến thể 3 — Reverse Words in a String III (LeetCode 557)

````python
def reverse_range(a: list, i: int, j: int) -> None:
    while i < j:
        a[i], a[j] = a[j], a[i]
        i += 1
        j -= 1

def rotate(nums: list[int], k: int) -> None:
    n = len(nums)
    k %= n
    reverse_range(nums, 0, n - 1)
    reverse_range(nums, 0, k - 1)
    reverse_range(nums, k, n - 1)

def reverse_each_word(sentence: str) -> str:
    chars = list(sentence)
    start = 0
    for end in range(len(chars) + 1):
        if end == len(chars) or chars[end] == " ":
            reverse_range(chars, start, end - 1)
            start = end + 1
    return "".join(chars)

nums = [1, 2, 3, 4, 5, 6, 7]; rotate(nums, 3); print(nums)          # [5, 6, 7, 1, 2, 3, 4]
print(reverse_each_word("Let's take LeetCode contest"))              # s'teL ekat edoCteeL tsetnoc
````

**Bài 2:** `[9, 2, 3] [9, 2, 3] [7, 2, 9]`.

````python
def answer_reverse(s: list[str]) -> None:
    left, right = 0, len(s) - 1
    while left < right:
        s[left], s[right] = s[right], s[left]
        left += 1
        right -= 1

def answer_is_palindrome(s: str) -> bool:
    left, right = 0, len(s) - 1
    while left < right:
        if s[left] != s[right]:
            return False
        left += 1
        right -= 1
    return True

def answer_reverse_str_ii(s: str, k: int) -> str:
    chars = list(s)
    n = len(chars)
    for start in range(0, n, 2 * k):
        reverse_range(chars, start, min(start + k, n) - 1)
    return "".join(chars)

def answer_flip_rows(grid: list[list[int]]) -> None:
    for row in grid:
        reverse_range(row, 0, len(row) - 1)

t = list("python"); answer_reverse(t); assert "".join(t) == "nohtyp"
a = [1, 2, 3]; b_ = a; b_[0] = 9; c = a[::-1]; c[0] = 7
assert (a, b_, c) == ([9, 2, 3], [9, 2, 3], [7, 2, 9])
t = list("hello"); answer_reverse(t); assert t == list("olleh")
assert answer_is_palindrome("racecar") and answer_is_palindrome("a") and answer_is_palindrome("")
assert not answer_is_palindrome("ab")
assert answer_reverse_str_ii("abcdefg", 2) == "bacdfeg"
assert answer_reverse_str_ii("abcd", 2) == "bacd"
assert answer_reverse_str_ii("a", 5) == "a"
g = [[1, 1, 0], [1, 0, 1], [0, 0, 0]]; answer_flip_rows(g)
assert g == [[0, 1, 1], [1, 0, 1], [0, 0, 0]]
print("Đáp án tham khảo PASS toàn bộ test!")
````
