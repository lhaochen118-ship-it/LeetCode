## 2. Đề bài & ví dụ

Cho chuỗi `s` chỉ gồm chữ cái thường và số nguyên `k`. Tạo chuỗi mới gồm **chỉ những ký tự có tần suất xuất hiện trong `s` nhỏ hơn `k`**. Các ký tự được giữ **nguyên thứ tự ban đầu**, và **mọi lần xuất hiện** của ký tự đủ điều kiện đều được giữ. Nếu không ký tự nào đủ điều kiện, trả về chuỗi rỗng.

| Ví dụ | s, k | Output | Giải thích |
|---|---|---|---|
| 1 | `"aadbbcccca"`, `k=3` | `"dbb"` | `a` xuất hiện 3 lần, `b` 2 lần, `c` 4 lần, `d` 1 lần; chỉ `b`, `d` có tần suất `< 3` |
| 2 | `"xyz"`, `k=2` | `"xyz"` | mỗi ký tự xuất hiện 1 lần, `1 < 2` nên giữ hết |

Constraints: `1 <= s.length <= 100`; `s` chỉ gồm chữ cái thường tiếng Anh; `1 <= k <= s.length`.

````python
from collections import Counter
s, k = "aadbbcccca", 3
cnt = Counter(s)
print(cnt)
print("giữ:", [ch for ch in s if cnt[ch] < k])
print("kết quả:", "".join(ch for ch in s if cnt[ch] < k))
````

## 3. Kiến thức nền

### 3.1. Tần suất toàn cục và `Counter`

Tần suất của ký tự là **tổng** số lần nó xuất hiện trong cả chuỗi, không phụ thuộc vị trí. `Counter(s)` trả dictionary ký tự -> số lần. Muốn biết tần suất **trước khi** quyết định giữ/bỏ một ký tự, ta phải đếm xong **toàn bộ** chuỗi rồi mới lọc: đó là hai lượt duyệt.

````python
from collections import Counter
cnt = Counter("hello")
print(cnt)
print(cnt["l"], cnt["z"])          # ký tự không có -> 0
print(sorted(cnt.items()))
````

### 3.2. Mảng đếm 26 ô

Chữ thường có thể ánh xạ sang chỉ số `0..25` bằng `ord(ch) - ord("a")`. Mảng `freq = [0] * 26` thay thế dictionary: nhanh và đơn giản. `freq[ord(ch) - 97] += 1` tăng tần suất.

````python
freq = [0] * 26
for ch in "aadbbcccca":
    freq[ord(ch) - ord("a")] += 1
print(freq[:5])
print({chr(i + 97): v for i, v in enumerate(freq) if v})
````

### 3.3. Dựng chuỗi bằng `join`

Cộng chuỗi trong vòng lặp (`res += ch`) hợp lệ nhưng tạo chuỗi mới nhiều lần. Cách chuẩn: gom ký tự vào list rồi `"".join(list)` (một lần). Với `n <= 100` không khác biệt, nhưng là thói quen tốt.

````python
parts = []
for ch in "abc":
    parts.append(ch.upper())
print("".join(parts))
print("-".join(["a", "b"]), repr("".join([])))
````

### 3.4. `str.translate` để xóa một tập ký tự

`s.translate(table)` biến đổi từng ký tự theo bảng. Bảng `str.maketrans("", "", "abc")` chỉ định xóa các ký tự `a`, `b`, `c`. Cách này xóa **mọi lần xuất hiện** của một tập ký tự trong một lượt.

````python
table = str.maketrans("", "", "ac")
print("aadbbcccca".translate(table))
print("xyz".translate(str.maketrans("", "", "")))
````

## 4. Ý tưởng nền tảng / cách nghĩ trực giác

Làm tay với `s = "aadbbcccca"`, `k = 3`:

```
đếm toàn bộ: a=3 (vị trí 0,1,9), d=1, b=2, c=4
điều kiện giữ: tần suất < 3
a: 3 < 3? không  -> bỏ cả 3 chữ a
d: 1 < 3? có     -> giữ
b: 2 < 3? có     -> giữ cả 2 chữ b
c: 4 < 3? không  -> bỏ
duyệt s theo thứ tự, giữ d, b, b  =>  "dbb"
```

Điểm mấu chốt: quyết định giữ/bỏ một ký tự phụ thuộc vào **tổng** số lần xuất hiện, tức là phải **đếm xong rồi mới lọc**. Không thể quyết định ngay khi gặp ký tự lần đầu.

## 5. Cách 1 — Brute Force / baseline

Với mỗi ký tự `ch` của `s`, gọi `s.count(ch)` để biết tần suất toàn cục; nếu `< k` thì giữ.

Nút thắt: `s.count(ch)` duyệt cả chuỗi **mỗi lần**, nên tổng `O(n^2)` (cùng một ký tự bị đếm lại nhiều lần). Với `n <= 100` không sao, nhưng ý tưởng "đếm một lần" tốt hơn rõ rệt.

````python
class SolutionBrute:
    def filterCharacters(self, s, k):
        res = []
        for ch in s:
            if s.count(ch) < k:
                res.append(ch)
        return "".join(res)

print(SolutionBrute().filterCharacters("aadbbcccca", 3))
print(SolutionBrute().filterCharacters("xyz", 2))
````

**Độ phức tạp:** Time `O(n^2)`, Space `O(n)` cho kết quả.

## 6. Ý tưởng thuật toán / bước đột phá

**Bước đột phá:** tách thành hai lượt, **mỗi lượt `O(n)`**:

1. Lượt 1: đếm tần suất mọi ký tự (bảng `cnt`).
2. Lượt 2: duyệt `s` theo thứ tự, giữ `ch` nếu `cnt[ch] < k`.

**Quy tắc "nếu… thì…":**

- Nếu `cnt[ch] < k` -> giữ **mọi** lần xuất hiện của `ch`.
- Nếu `cnt[ch] >= k` -> bỏ **mọi** lần xuất hiện của `ch`.
- Kết quả giữ thứ tự gốc vì lượt 2 duyệt `s` từ trái sang phải.

Invariant sau lượt 1: `sum(cnt.values()) == len(s)`; mỗi ký tự có tần suất cố định, không đổi ở lượt 2.

````python
def two_pass(s, k):
    cnt = Counter(s)                       # lượt 1
    print("tần suất:", dict(cnt))
    out = []
    for ch in s:                           # lượt 2
        keep = cnt[ch] < k
        out.append(ch if keep else "")
    return "".join(out)

print(two_pass("aadbbcccca", 3))
print(repr(two_pass("aaa", 2)))
````

## 7. Các cách tiếp theo

Tổng cộng 4 cách: (1) `s.count` cho từng ký tự; (2) `Counter` hai lượt; (3) mảng đếm 26 ô; (4) tập ký tự bị loại + `translate`.

### Cách 2 — `Counter` hai lượt

Đúng như mục 6, dùng comprehension trong `join`.

````python
class SolutionCounter:
    def filterCharacters(self, s, k):
        cnt = Counter(s)
        return "".join(ch for ch in s if cnt[ch] < k)

print(SolutionCounter().filterCharacters("aadbbcccca", 3))
print(SolutionCounter().filterCharacters("xyz", 2))
````

Độ phức tạp Cách 2: Time `O(n)`, Space `O(26) = O(1)` cho bảng đếm (cộng kết quả `O(n)`).

### Cách 3 — Mảng đếm 26 ô

Thay `Counter` bằng `freq = [0] * 26`; lượt 2 tra `freq[ord(ch) - 97] < k`. Nhanh nhất về hằng số và không phụ thuộc thư viện.

````python
class SolutionArray:
    def filterCharacters(self, s, k):
        freq = [0] * 26
        for ch in s:
            freq[ord(ch) - 97] += 1
        res = []
        for ch in s:
            if freq[ord(ch) - 97] < k:
                res.append(ch)
        return "".join(res)

print(SolutionArray().filterCharacters("aadbbcccca", 3))
print(SolutionArray().filterCharacters("xyz", 2))
````

Độ phức tạp Cách 3: Time `O(n)`, Space `O(26)` cho bảng.

### Cách 4 — Tập ký tự bị loại + `translate`

Lượt 1: tìm tập `drop = {ch : cnt[ch] >= k}`. Lượt 2: xóa tất cả ký tự của `drop` khỏi `s` bằng `s.translate(str.maketrans("", "", "".join(drop)))`. Tư duy "loại bỏ cả nhóm" thay vì "giữ từng ký tự".

````python
class SolutionTranslate:
    def filterCharacters(self, s, k):
        cnt = Counter(s)
        drop = "".join(ch for ch, c in cnt.items() if c >= k)
        return s.translate(str.maketrans("", "", drop))

print(SolutionTranslate().filterCharacters("aadbbcccca", 3))
print(SolutionTranslate().filterCharacters("xyz", 2))
print(repr(SolutionTranslate().filterCharacters("aaa", 2)))
````

Độ phức tạp Cách 4: Time `O(n)` (`translate` duyệt một lần), Space `O(26 + n)`.

## 8. TRACE từng bước

Trace Cách 2 với `s = "aadbbcccca"`, `k = 3`, sau lượt 1: `a=3, d=1, b=2, c=4`.

| bước | ký tự | `cnt[ch]` | `< 3`? | kết quả đang có |
|---|---|---|---|---|
| 0 | a | 3 | không | `""` |
| 1 | a | 3 | không | `""` |
| 2 | d | 1 | có | `"d"` |
| 3 | b | 2 | có | `"db"` |
| 4 | b | 2 | có | `"dbb"` |
| 5-8 | c c c c | 4 | không | `"dbb"` |
| 9 | a | 3 | không | `"dbb"` |

````python
def trace_filter(s, k):
    cnt = Counter(s)
    print("tần suất:", dict(cnt), "| k =", k)
    res = ""
    for i, ch in enumerate(s):
        keep = cnt[ch] < k
        if keep:
            res += ch
        print(f"i={i} ch={ch} cnt={cnt[ch]} {'< ' if keep else '>='} {k} -> {'giữ' if keep else 'bỏ'} | res={res!r}")
    return res

trace_filter("aadbbcccca", 3)
trace_filter("xyz", 2)
````

## 9. Code nộp LeetCode

````python
from typing import List
from collections import Counter

class Solution:
    def filterCharacters(self, s: str, k: int) -> str:
        cnt = Counter(s)
        return "".join(ch for ch in s if cnt[ch] < k)

print(Solution().filterCharacters("aadbbcccca", 3))
print(Solution().filterCharacters("xyz", 2))
````

## 10. Vì sao thuật toán đúng

- **Đếm đúng tần suất toàn cục:** sau lượt 1, `cnt[ch]` là số lần `ch` xuất hiện trong **cả** `s`, đúng định nghĩa "tần suất" của đề (không phụ thuộc vị trí).
- **Điều kiện giữ nhất quán:** quyết định chỉ phụ thuộc `ch` (qua `cnt[ch]`), nên mọi lần xuất hiện của cùng một ký tự cùng được giữ hoặc cùng bị bỏ — đúng đề "mọi lần xuất hiện".
- **Giữ thứ tự:** lượt 2 duyệt `s` từ trái sang phải và chỉ nối thêm, nên chuỗi kết quả là một **dãy con** của `s` giữ thứ tự gốc.
- **`k` và biên `<`:** đề dùng "nhỏ hơn `k`" (strict). Ký tự có tần suất đúng bằng `k` bị bỏ (ví dụ `a` ở ví dụ 1).
- **Cách 4:** xóa mọi ký tự trong `drop` giữ nguyên các ký tự còn lại theo thứ tự, và `drop` đúng bằng tập ký tự có tần suất `>= k`.
- **Kết quả rỗng:** khi mọi ký tự có tần suất `>= k`, `join` của dãy rỗng trả `""`.

## 11. Phân tích Time / Space Complexity

`n = len(s) <= 100`, bảng chữ cái `26`.

| Cách | Time | Space |
|---|---|---|
| 1. `s.count` từng ký tự | `O(n^2)` | `O(n)` |
| 2. `Counter` hai lượt | `O(n)` | `O(26)` + kết quả |
| 3. Mảng đếm 26 ô | `O(n)` | `O(26)` + kết quả |
| 4. `drop` + `translate` | `O(n)` | `O(26)` + kết quả |

Kết quả có độ dài tối đa `n` nên "Space phụ" ngoài kết quả là `O(1)` (bảng 26). Cách 2 dễ đọc nhất; Cách 3 nhanh nhất.

## 12. Bộ test

Test 1: ví dụ đề và edge case (đều tuân `1 <= len(s) <= 100`, `s` chữ thường, `1 <= k <= len(s)`).

````python
impls = [SolutionBrute(), SolutionCounter(), SolutionArray(), SolutionTranslate(), Solution()]

def oracle(s, k):
    # độc lập: sắp theo vị trí, dùng s.count và danh sách
    out = ""
    for i in range(len(s)):
        c = 0
        for j in range(len(s)):
            if s[j] == s[i]:
                c += 1
        if c < k:
            out += s[i]
    return out

tests = [
    (("aadbbcccca", 3), "dbb"),
    (("xyz", 2), "xyz"),
    (("a", 1), ""),                      # 1 < 1 sai -> rỗng
    (("ab", 2), "ab"),
    (("aa", 2), ""),                     # a=2, không < 2
    (("aa", 1), ""),
    (("abab", 2), ""),                   # a=2,b=2
    (("abab", 3), "abab"),
    (("abcabcd", 2), "d"),
    (("zzzz", 4), ""),                   # bằng k thì bỏ
    (("mississippi", 3), "mpp"),         # m1 i4 s4 p2
    (("mississippi", 5), "mississippi"),
    (("a" * 100, 100), ""),
    (("a" * 99 + "b", 100), "a" * 99 + "b"),
]
for (s, k), exp in tests:
    assert 1 <= len(s) <= 100 and s.islower() and s.isalpha() and 1 <= k <= len(s)
    assert oracle(s, k) == exp, (s, k, oracle(s, k), exp)
    for im in impls:
        assert im.filterCharacters(s, k) == exp, (type(im).__name__, s, k)
print("tất cả test cố định OK:", len(tests))
````

Test 2: ngẫu nhiên có seed, đúng constraints. Dùng bảng chữ cái nhỏ (để có nhiều ký tự lặp) và cả bảng đầy đủ 26 chữ.

````python
import random
import string
random.seed(3662)

empties = 0
for t in range(4000):
    n = random.randint(1, 100)
    alpha = random.choice([2, 3, 5, 26])
    s = "".join(random.choice(string.ascii_lowercase[:alpha]) for _ in range(n))
    k = random.randint(1, n)
    exp = oracle(s, k)
    empties += (exp == "")
    for im in impls:
        assert im.filterCharacters(s, k) == exp, (type(im).__name__, s, k)
print("4000 test ngẫu nhiên OK; số ca kết quả rỗng =", empties)
````

Test 3: tính chất — kết quả là **dãy con** của `s`; `k` lớn hơn thì kết quả **không ngắn hơn** (đơn điệu); `k = len(s)` luôn giữ mọi ký tự trừ khi `s` gồm đúng một loại ký tự; và độ dài kết quả bằng tổng tần suất các ký tự `< k`.

````python
random.seed(662)
def is_subseq(a, b):
    it = iter(b)
    return all(ch in it for ch in a)

for _ in range(1500):
    n = random.randint(1, 100)
    s = "".join(random.choice("abcde") for _ in range(n))
    k = random.randint(1, n)
    res = Solution().filterCharacters(s, k)
    assert is_subseq(res, s)
    if k < n:
        assert len(Solution().filterCharacters(s, k + 1)) >= len(res)
    cnt = Counter(s)
    assert len(res) == sum(c for c in cnt.values() if c < k)
    if len(cnt) > 1:
        assert Solution().filterCharacters(s, n) == s
print("tính chất (dãy con, đơn điệu theo k, độ dài, k = n): OK")
````

## 13. Các lỗi người mới hay gặp

### Lỗi 1: nhầm `<` với `<=`

````python
def wrong_le(s, k):
    cnt = Counter(s)
    return "".join(ch for ch in s if cnt[ch] <= k)          # SAI: đề là "nhỏ hơn k"

print("sai :", wrong_le("aadbbcccca", 3))
print("đúng:", Solution().filterCharacters("aadbbcccca", 3))
assert wrong_le("aadbbcccca", 3) != Solution().filterCharacters("aadbbcccca", 3)
````

Vì sao sai: `a` có tần suất đúng `3 = k` nên phải bị bỏ; `<=` giữ nó lại. **Sửa:** `cnt[ch] < k`.

### Lỗi 2: khử trùng ký tự (chỉ giữ một lần)

````python
def wrong_dedupe(s, k):
    cnt = Counter(s)
    seen, res = set(), ""
    for ch in s:
        if cnt[ch] < k and ch not in seen:                  # SAI: đề giữ MỌI lần xuất hiện
            seen.add(ch)
            res += ch
    return res

print("sai :", wrong_dedupe("aadbbcccca", 3))
print("đúng:", Solution().filterCharacters("aadbbcccca", 3))
assert wrong_dedupe("aadbbcccca", 3) != Solution().filterCharacters("aadbbcccca", 3)
````

Vì sao sai: kết quả đúng là `"dbb"` (giữ cả hai chữ `b`), còn cách sai ra `"db"`. **Sửa:** bỏ phần khử trùng.

### Lỗi 3: dùng tần suất **đang đếm dở** thay vì tổng

````python
def wrong_running(s, k):
    cnt, res = {}, ""
    for ch in s:
        cnt[ch] = cnt.get(ch, 0) + 1
        if cnt[ch] < k:                                      # SAI: tần suất tại thời điểm này, chưa phải tổng
            res += ch
    return res

print("sai :", wrong_running("aadbbcccca", 3))
print("đúng:", Solution().filterCharacters("aadbbcccca", 3))
assert wrong_running("aadbbcccca", 3) != Solution().filterCharacters("aadbbcccca", 3)
````

Vì sao sai: hai chữ `a` đầu có "tần suất đang đếm" `1, 2 < 3` nên bị giữ nhầm, dù tổng của `a` là `3`. **Sửa:** đếm xong toàn bộ rồi mới lọc (hai lượt).

### Lỗi 4: đảo ngược điều kiện (giữ ký tự phổ biến)

````python
def wrong_inverse(s, k):
    cnt = Counter(s)
    return "".join(ch for ch in s if cnt[ch] >= k)          # SAI: giữ ký tự có tần suất >= k

print("sai :", wrong_inverse("aadbbcccca", 3))
print("đúng:", Solution().filterCharacters("aadbbcccca", 3))
assert wrong_inverse("aadbbcccca", 3) != Solution().filterCharacters("aadbbcccca", 3)
````

Vì sao sai: đề giữ ký tự **ít** xuất hiện (`< k`) và bỏ ký tự phổ biến; đảo điều kiện cho đúng phần bù (`"aacccca"` ở ví dụ). **Sửa:** `cnt[ch] < k`.

## 14. Mẹo làm bài nhanh & nhận dạng pattern

- "Lọc phần tử theo **số lần xuất hiện**" -> **đếm trước, lọc sau** (hai lượt).
- Chuỗi chữ thường -> mảng 26 ô hoặc `Counter`; cả hai đều `O(n)`.
- Đọc kỹ biên: "nhỏ hơn `k`" (`<`) khác "ít nhất `k`" (`>=`).
- "Giữ mọi lần xuất hiện" nghĩa là **không khử trùng**; "mỗi ký tự một lần" mới dùng `set`.
- Dựng chuỗi kết quả bằng list rồi `join`.

## 18. Gợi ý 3 mức cho bài tập

**Mức 1 (nhẹ):** đếm tần suất từng ký tự trước, rồi mới xét từng ký tự theo thứ tự trong `s`.

**Mức 2 (vừa):** Bài 3: đổi `<` thành `>=`. Bài 4: dùng `set` `seen` như Lỗi 2 nhưng đúng mục đích. Bài 5: `filterCharacters(s, k) == s` khi mọi tần suất `< k`, tức `k > max(cnt.values())`.

**Mức 3 (gần đáp án):** Bài 1: `a=2, b=2, c=2`, không ai `< 2` -> `""`. Bài 2: `a=2, b=3, c=1` -> giữ `a, a, c` -> `"aac"`. Bài 5: `k = max(cnt.values()) + 1`.

## 19. Đáp án tham khảo cho bài tập

````python
# Bài 1
assert Solution().filterCharacters("abcabc", 2) == ""

# Bài 2
assert dict(Counter("aabbbc")) == {"a": 2, "b": 3, "c": 1}
assert Solution().filterCharacters("aabbbc", 3) == "aac"

# Bài 3
def filter_at_least(s, k):
    cnt = Counter(s)
    return "".join(ch for ch in s if cnt[ch] >= k)
assert filter_at_least("aadbbcccca", 3) == "aacccca"
assert filter_at_least("xyz", 2) == ""

# Bài 4
def unique_filtered(s, k):
    cnt = Counter(s)
    seen, out = set(), []
    for ch in s:
        if cnt[ch] < k and ch not in seen:
            seen.add(ch)
            out.append(ch)
    return "".join(out)
assert unique_filtered("aadbbcccca", 3) == "db"
assert unique_filtered("xyz", 2) == "xyz"

# Bài 5
def smallest_k_for_all(s):
    return max(Counter(s).values()) + 1
for s5 in ["a", "aab", "abcabc", "zzzz"]:
    k5 = smallest_k_for_all(s5)
    assert k5 <= len(s5) + 1
    if k5 <= len(s5):
        assert Solution().filterCharacters(s5, k5) == s5
        assert Solution().filterCharacters(s5, k5 - 1) != s5
print("đáp án bài tập OK")
````
