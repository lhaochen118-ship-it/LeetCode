Máy tính lưu mỗi ký tự bằng một **mã số**. `ord('a')` cho `97`, `ord('b')` cho `98`, ... Các chữ cái liên tiếp có mã liên tiếp nên `ord('c') - ord('a') = 2` chính là "chữ `c` đứng thứ mấy trong bảng chữ cái" (đếm từ `0`). `chr` làm ngược lại.

### 3.3. Đổi chữ số ký tự thành số: `int('3')` và `ord('3') - ord('0')`

### 3.4. Chia lấy dư `%` và tính chẵn/lẻ

`a % 2` bằng `0` nếu `a` **chẵn**, bằng `1` nếu `a` **lẻ**. Python luôn cho kết quả `%` 2 là `0` hoặc `1`, kể cả với số âm (`-3 % 2 == 1`), nên so sánh `a % 2 == b % 2` an toàn với số âm. Lưu ý: `%` được tính **trước** `+` và `-` (giống `*` và `/`).

````python
for a in [0, 1, 2, 7, 8, -3, -4]:
    print(f"{a:>2} % 2 = {a % 2}")
print(2 + 3 % 2)       # 3, vì % làm trước: 2 + (3 % 2) = 2 + 1
print((2 + 3) % 2)     # 1
````

### 3.5. So sánh trả về `bool` và trả thẳng ra khỏi hàm

Biểu thức như `x % 2 == y % 2` **tự nó** là `True`/`False`. Nhờ vậy, thay vì viết `if ...: return True else: return False`, ta có thể `return` thẳng biểu thức so sánh.

````python
def same_parity(x, y):
    return x % 2 == y % 2

print(same_parity(2, 8), same_parity(3, 4), same_parity(-1, 5))   # True False True
````

## 4. Ý tưởng nền tảng / cách nghĩ trực giác (tính tay)

Nhìn bàn cờ: **mỗi bước sang ô kề cạnh thì đổi màu**. Đi sang phải một ô: đổi màu. Đi lên một ô: cũng đổi màu. Đi **chéo** một ô = một bước ngang + một bước dọc = đổi màu **hai lần** = giữ nguyên màu.

Vậy để đi từ ô này sang ô kia, ta đếm tổng số bước ngang + dọc. Nếu tổng đó **chẵn** thì màu đổi một số chẵn lần, tức **cùng màu**; nếu **lẻ** thì **khác màu**.

Tính tay hai ví dụ (cột `a..h` đánh số `1..8`):

- `"a1"` → `"c3"`: cột `1 → 3` (lệch `2`), hàng `1 → 3` (lệch `2`); tổng lệch `4` chẵn ⇒ **cùng màu** ⇒ `True`.
- `"a1"` → `"h3"`: cột `1 → 8` (lệch `7`), hàng `1 → 3` (lệch `2`); tổng lệch `9` lẻ ⇒ **khác màu** ⇒ `False`.

````python
def col_of(coord):
    return ord(coord[0]) - ord("a") + 1      # a..h -> 1..8

def row_of(coord):
    return int(coord[1])                      # '1'..'8' -> 1..8

for c1, c2 in [("a1", "c3"), ("a1", "h3")]:
    dc = abs(col_of(c1) - col_of(c2))
    dr = abs(row_of(c1) - row_of(c2))
    print(f"{c1} -> {c2}: lệch cột = {dc}, lệch hàng = {dr}, tổng = {dc + dr}, chẵn? {(dc + dr) % 2 == 0}")
````

## 5. Cách 1 — Brute Force / baseline: dựng cả bàn cờ rồi tra màu

**Ý tưởng:** làm đúng như đề mô tả. Tô màu cho cả bàn cờ `8 x 8` bằng cách **xen kẽ** (bắt đầu `a1` là đen `"B"`, mỗi ô tiếp theo đổi màu, hết hàng thì đổi thêm một lần vì hàng sau bắt đầu bằng màu ngược lại). Sau đó đổi tọa độ thành chỉ số `(hàng, cột)` và so sánh hai màu.

Cách này **không cần biết toán** nhưng tốn công dựng bảng mỗi lần gọi hàm.

````python
def build_board():
    """board[r][c]: màu ô ở hàng r (0 = hàng 1) và cột c (0 = cột a). 'B' = đen, 'W' = trắng."""
    board = [[None] * 8 for _ in range(8)]
    color = "B"                                   # a1 là ô đen
    for r in range(8):
        for c in range(8):
            board[r][c] = color
            color = "W" if color == "B" else "B"  # sang ô kế bên: đổi màu
        color = "W" if color == "B" else "B"      # hết hàng: đổi thêm lần nữa để hàng sau bắt đầu ngược màu
    return board


def check_board(coordinate1: str, coordinate2: str) -> bool:
    board = build_board()
    def color(coord):
        c = ord(coord[0]) - ord("a")
        r = int(coord[1]) - 1
        return board[r][c]
    return color(coordinate1) == color(coordinate2)


for row in reversed(build_board()):               # in hàng 8 ở trên cùng cho giống bàn cờ thật
    print(" ".join(row))
print(check_board("a1", "c3"), check_board("a1", "h3"))
````

**Độ phức tạp (Cách 1).** Dựng bàn cờ tốn `8 x 8 = 64` thao tác — một hằng số. Vì kích thước bàn cờ cố định nên **Time `O(1)`** và **Space `O(1)`** (`64` ô). Tuy nhiên đây là hằng số **lớn không cần thiết**: ta tô cả 64 ô chỉ để hỏi về hai ô. **Nút thắt:** màu của ô có quy luật rất đơn giản, ta có thể tính thẳng từ tọa độ.

## 6. Ý tưởng thuật toán / bước đột phá

Xem bàn cờ vừa in: ô `(hàng r, cột c)` (đếm từ `0`) đen khi `r + c` **chẵn**, trắng khi `r + c` **lẻ**. Ví dụ `a1` là `(0, 0)` tổng `0` chẵn ⇒ đen; `b1` là `(0, 1)` tổng `1` lẻ ⇒ trắng.

**Màu của ô chỉ là chẵn/lẻ của `cột + hàng`.** Vậy:

> hai ô cùng màu ⟺ `(cột1 + hàng1) % 2 == (cột2 + hàng2) % 2`.

Ta không cần dựng bảng nữa: đổi tọa độ thành hai số rồi so sánh tính chẵn lẻ. Dù đếm cột/hàng từ `0` hay từ `1` thì kết quả so sánh **không đổi**, vì cả hai ô cùng bị cộng thêm `2` (một số chẵn), không đổi tính chẵn lẻ.

````python
# Bảng màu từ tính chẵn lẻ (cột 1..8, hàng 1..8, a1 = (1,1) -> tổng 2 chẵn -> đen)
for r in range(8, 0, -1):
    line = []
    for c in range(1, 9):
        line.append("B" if (r + c) % 2 == 0 else "W")
    print(r, " ".join(line))
print("  " + " ".join("abcdefgh"))
````

## 7. Các cách tiếp theo

### Cách 2 — Công thức `(cột + hàng) % 2` cho từng ô

Giải thích từng đoạn:

- `ord(c[0]) - ord("a") + 1` đổi chữ cái thành cột `1..8`.
- `int(c[1])` đổi ký tự số thành hàng `1..8`.
- `(cột + hàng) % 2` là "màu" của ô: `0` ứng với đen, `1` ứng với trắng.
- So sánh màu hai ô bằng `==`.

````python
def color_parity(coord: str) -> int:
    col = ord(coord[0]) - ord("a") + 1
    row = int(coord[1])
    return (col + row) % 2          # 0: đen, 1: trắng


def check_formula(coordinate1: str, coordinate2: str) -> bool:
    return color_parity(coordinate1) == color_parity(coordinate2)


print(check_formula("a1", "c3"), check_formula("a1", "h3"))      # True False
print(color_parity("a1"), color_parity("b1"), color_parity("h8"))  # 0 1 0
````

````python
# Code nộp LeetCode cho Cách 2 (viết lại thành class Solution)
class Solution2:
    def checkTwoChessboards(self, coordinate1: str, coordinate2: str) -> bool:
        def color(coord: str) -> int:
            return (ord(coord[0]) - ord("a") + 1 + int(coord[1])) % 2
        return color(coordinate1) == color(coordinate2)


print(Solution2().checkTwoChessboards("a1", "c3"), Solution2().checkTwoChessboards("a1", "h3"))
````

**Độ phức tạp Cách 2:** **Time `O(1)`** (vài phép tính), **Space `O(1)`**.

### Cách 3 — Xét hiệu giữa hai ô (cách của lời giải tham khảo)

Không cần tính màu từng ô: chỉ cần **tổng độ lệch cột và hàng** giữa hai ô là chẵn. Lấy `dc = ord(c1[0]) - ord(c2[0])` và `dr = int(c1[1]) - int(c2[1])` (có thể âm, không sao), rồi kiểm tra `(dc + dr) % 2 == 0`. Nhớ ngoặc quanh `dc + dr` vì `%` làm trước `+`.

````python
def check_diff(coordinate1: str, coordinate2: str) -> bool:
    dc = ord(coordinate1[0]) - ord(coordinate2[0])
    dr = int(coordinate1[1]) - int(coordinate2[1])
    return (dc + dr) % 2 == 0


class Solution3:
    def checkTwoChessboards(self, coordinate1: str, coordinate2: str) -> bool:
        return (ord(coordinate1[0]) - ord(coordinate2[0]) + ord(coordinate1[1]) - ord(coordinate2[1])) % 2 == 0


print(check_diff("a1", "c3"), check_diff("a1", "h3"))
print(Solution3().checkTwoChessboards("a1", "c3"), Solution3().checkTwoChessboards("a1", "h3"))
````

Trong `Solution3` ta trừ luôn mã `ord` của hai chữ số `'1'..'8'` (không cần `int`): `ord('3') - ord('1') = 3 - 1 = 2`, bằng đúng hiệu hai chữ số. Biểu thức `a - b + c - d` được tính từ trái sang phải; cả `ord(...)` cũng là số nguyên nên không có gì bất ngờ. Python cho `(âm) % 2` là `0` hoặc `1` nên an toàn.

**Độ phức tạp Cách 3:** **Time `O(1)`**, **Space `O(1)`**.

### Cách 4 — Mẹo `ord` không cần trừ mốc

Mã của `'a'` là `97` (lẻ), mã của `'1'` là `49` (cũng lẻ). Ô `(cột0, hàng0)` (đếm từ `0`) có `ord(chữ) + ord(số) = 97 + cột0 + 49 + hàng0 = 146 + cột0 + hàng0`. Vì `146` chẵn nên **tính chẵn lẻ của `ord(chữ) + ord(số)` đúng bằng tính chẵn lẻ của `cột0 + hàng0`** — không cần trừ mốc, không cần `int`. Ô đen khi tổng này **chẵn**.

````python
def check_ord_trick(coordinate1: str, coordinate2: str) -> bool:
    return (ord(coordinate1[0]) + ord(coordinate1[1])) % 2 == (ord(coordinate2[0]) + ord(coordinate2[1])) % 2


class Solution4:
    def checkTwoChessboards(self, coordinate1: str, coordinate2: str) -> bool:
        return sum(map(ord, coordinate1)) % 2 == sum(map(ord, coordinate2)) % 2


print(ord("a") + ord("1"), (ord("a") + ord("1")) % 2)        # 146 0  -> a1 đen
print(check_ord_trick("a1", "c3"), check_ord_trick("a1", "h3"))
print(Solution4().checkTwoChessboards("a1", "c3"), Solution4().checkTwoChessboards("a1", "h3"))
````

`sum(map(ord, coordinate1))` áp `ord` lên từng ký tự của chuỗi rồi cộng lại — với chuỗi hai ký tự nó chính là `ord(c[0]) + ord(c[1])`.

**Độ phức tạp Cách 4:** **Time `O(1)`**, **Space `O(1)`**.

## 8. TRACE từng bước

Ta trace Cách 2 trên hai cặp `("a1", "c3")` và `("a1", "h3")`, rồi Cách 3 trên `("e4", "d5")` (ví dụ tự dựng).

````python
def trace_formula(c1: str, c2: str) -> bool:
    print(f"coordinate1 = {c1!r}, coordinate2 = {c2!r}")
    parities = []
    for coord in (c1, c2):
        col = ord(coord[0]) - ord("a") + 1
        row = int(coord[1])
        total = col + row
        parities.append(total % 2)
        print(f"  {coord!r}: cột = {col}, hàng = {row}, cột + hàng = {total}, % 2 = {total % 2} -> {'đen' if total % 2 == 0 else 'trắng'}")
    result = parities[0] == parities[1]
    print(f"  so sánh {parities[0]} == {parities[1]} -> {result}")
    return result


assert trace_formula("a1", "c3") is True
assert trace_formula("a1", "h3") is False
````

````python
def trace_diff(c1: str, c2: str) -> bool:
    dc = ord(c1[0]) - ord(c2[0])
    dr = int(c1[1]) - int(c2[1])
    print(f"{c1!r} vs {c2!r}: dc = {ord(c1[0])} - {ord(c2[0])} = {dc}, dr = {c1[1]} - {c2[1]} = {dr}")
    print(f"  dc + dr = {dc + dr}, (dc + dr) % 2 = {(dc + dr) % 2} -> {(dc + dr) % 2 == 0}")
    return (dc + dr) % 2 == 0


assert trace_diff("e4", "d5") is True          # e4 và d5 cùng màu (cùng đường chéo)
assert trace_diff("e4", "e5") is False         # kề cạnh -> khác màu
````

Bảng trace Cách 2:

| tọa độ | cột | hàng | cột + hàng | `% 2` | màu |
|---|---|---|---|---|---|
| `a1` | 1 | 1 | 2 | 0 | đen |
| `c3` | 3 | 3 | 6 | 0 | đen |
| `h3` | 8 | 3 | 11 | 1 | trắng |

- `a1` và `c3`: cùng `0` ⇒ `True`.
- `a1` và `h3`: `0` khác `1` ⇒ `False`.

## 9. Code nộp LeetCode

Bản nộp dùng Cách 2 (tính màu từng ô, dễ đọc nhất cho người mới); Cách 3/4 ngắn hơn nhưng khó nhìn hơn.

````python
class Solution:
    def checkTwoChessboards(self, coordinate1: str, coordinate2: str) -> bool:
        def color(coord: str) -> int:
            col = ord(coord[0]) - ord("a") + 1     # a..h -> 1..8
            row = int(coord[1])                    # '1'..'8' -> 1..8
            return (col + row) % 2                 # 0: đen, 1: trắng
        return color(coordinate1) == color(coordinate2)


sol = Solution()
print(sol.checkTwoChessboards("a1", "c3"), sol.checkTwoChessboards("a1", "h3"))   # True False
````

## 10. Vì sao thuật toán đúng

- **Luân phiên màu:** theo đề, hai ô kề cạnh luôn khác màu. Vậy khi `cột` hoặc `hàng` tăng `1`, màu đổi; khi **cả hai** cùng tăng `1` (đi chéo), màu đổi hai lần nên **giữ nguyên**.

## 13. Các lỗi người mới hay gặp

### Lỗi 1: Cộng chữ số dạng chuỗi với số (`TypeError`)

````python
def sai_quen_int(c1, c2):
    col1 = ord(c1[0]) - ord("a")
    col2 = ord(c2[0]) - ord("a")
    return (col1 + c1[1]) % 2 == (col2 + c2[1]) % 2      # c1[1] là str '1'..'8', chưa phải số

try:
    sai_quen_int("a1", "c3")
except TypeError as e:
    print("TypeError:", e)
````

**Vì sao sai:** `c1[1]` là **chuỗi** `'1'`; cộng `int + str` không hợp lệ. Cần `int(c1[1])` (hoặc dùng mã `ord`).

### Lỗi 2: Quên ngoặc quanh tổng — `%` làm trước `+`

````python
def sai_uu_tien(c1, c2):
    dc = ord(c1[0]) - ord(c2[0])
    dr = int(c1[1]) - int(c2[1])
    return dc + dr % 2 == 0          # = dc + (dr % 2) == 0 -> sai!

print(sai_uu_tien("a1", "c3"), "(đúng là True)")     # dc = -2, dr = -2 -> -2 + 0 == 0 ? False
print(sai_uu_tien("a1", "b2"), "(đúng là True)")     # dc = -1, dr = -1 -> -1 + 1 == 0 ? True (đúng do ăn may)
````

**Vì sao sai:** `dc + dr % 2` được hiểu là `dc + (dr % 2)`, không phải `(dc + dr) % 2`. Kết quả đúng/sai tùy may rủi theo từng cặp. Hãy luôn thêm ngoặc: `(dc + dr) % 2 == 0`.

### Lỗi 3: Chỉ xét một trục (cột hoặc hàng)

````python
def sai_mot_truc(c1, c2):
    return (ord(c1[0]) - ord(c2[0])) % 2 == 0       # chỉ nhìn cột

print(sai_mot_truc("a1", "a2"), "(đúng là False)")   # cùng cột nhưng khác hàng 1 bước -> khác màu
````

**Vì sao sai:** màu phụ thuộc **cả cột lẫn hàng**. `a1` và `a2` cùng cột nhưng kề cạnh nên khác màu. Phải xét `cột + hàng`.

### Lỗi 4: Nhầm "cùng màu" với "cùng hàng hoặc cùng cột"

````python
def sai_cung_duong(c1, c2):
    return c1[0] == c2[0] or c1[1] == c2[1]

print(sai_cung_duong("a1", "a2"), "(đúng là False)")     # cùng cột nhưng khác màu
print(sai_cung_duong("a1", "b2"), "(đúng là True)")      # chéo nhưng cùng màu
````

**Vì sao sai:** cùng hàng/cột chỉ liên quan đến việc xe đi được, không liên quan màu. Chính các ô **cùng đường chéo** mới cùng màu (như tượng/bishop). Cách đúng là một trong các cách ở Mục 7:

**Code đúng (nhắc lại):** `(ord(c1[0]) - ord(c2[0]) + int(c1[1]) - int(c2[1])) % 2 == 0`.

````python
def dung(c1, c2):
    return (ord(c1[0]) - ord(c2[0]) + int(c1[1]) - int(c2[1])) % 2 == 0

for c1, c2, expected in examples + [("a1", "a2", False), ("a1", "b2", True)]:
    assert dung(c1, c2) is expected
print("PASS code đúng")
````

## 14. Mẹo làm bài nhanh & nhận dạng pattern

- "**Màu bàn cờ**", "**xen kẽ**", "**chẵn lẻ của tổng tọa độ**": nghĩ ngay tới `(x + y) % 2`.
- Mỗi lần đề nói "xen kẽ hai trạng thái", hãy nghĩ **tính chẵn lẻ** (parity) thay vì mô phỏng từng ô.
- Đổi chữ cái thành số bằng `ord(ch) - ord('a')`; đổi chữ số thành số bằng `int(ch)`. Nếu chỉ cần so tính chẵn lẻ, mốc `ord('a')`/`ord('1')` đều lẻ nên **không cần trừ**: `(ord(c[0]) + ord(c[1])) % 2`.
- Hai ô **cùng màu** ⟺ **tượng (bishop) có thể đi từ ô này sang ô kia** (nếu không có quân cản). Đây là cách nhớ trực quan.
- Mẹo debug: kiểm tra `"a1"` (đen) và `"h1"` (trắng): `a1` và `h1` cùng hàng nhưng khác màu vì cách nhau `7` cột (lẻ).

## 15. Biến thể của bài

1. **Trả về tên màu** của một ô (`"black"`/`"white"`).
2. **Đếm số ô đen** trong một hình chữ nhật con từ ô `p` đến ô `q`.
3. **Tượng có tới được** ô đích không (cùng màu) và **số nước đi tối thiểu** (`0` nếu trùng, `1` nếu cùng đường chéo, `2` nếu chỉ cùng màu, không thể nếu khác màu).
4. **Mã đi** (knight): mỗi nước đi luôn **đổi màu** ô (vì độ lệch `(2,1)` hoặc `(1,2)` có tổng lẻ).
5. **Bàn cờ `N x M`** tổng quát hoặc **mở rộng** nhận tọa độ có nhiều chữ số.

````python
def color_name(coord):
    return "black" if (ord(coord[0]) + ord(coord[1])) % 2 == 0 else "white"


def bishop_moves(c1, c2):
    """Số nước tượng cần để đi từ c1 sang c2 trên bàn cờ trống; -1 nếu không thể."""
    if c1 == c2:
        return 0
    if color_name(c1) != color_name(c2):
        return -1
    dc = abs(ord(c1[0]) - ord(c2[0]))
    dr = abs(int(c1[1]) - int(c2[1]))
    return 1 if dc == dr else 2


def knight_changes_color(c1, c2):
    dc = abs(ord(c1[0]) - ord(c2[0]))
    dr = abs(int(c1[1]) - int(c2[1]))
    return sorted((dc, dr)) == [1, 2] and color_name(c1) != color_name(c2)


print(color_name("a1"), color_name("h1"))                  # black white
print(bishop_moves("a1", "h8"), bishop_moves("a1", "c1"), bishop_moves("a1", "a2"), bishop_moves("a1", "a1"))   # 1 2 -1 0
print(knight_changes_color("b1", "c3"))                    # True
````
