## 2. Nền tảng Python: tách chữ số

### 2.1. `% 10` và `// 10`

- `x % 10` → **chữ số cuối** (phần dư khi chia cho 10).
- `x // 10` → **bỏ chữ số cuối** (chia lấy phần nguyên).
- Khác với `/` : `38 / 10 = 3.8` là số thực, không dùng được để tách chữ số.

Ví dụ `x = 385`:

| Bước | `x` | `x % 10` | `x // 10` |
|---:|---:|---:|---:|
| 1 | 385 | 5 | 38 |
| 2 | 38 | 8 | 3 |
| 3 | 3 | 3 | 0 → dừng |

````python
x = 385
print("385 % 10 =", 385 % 10, "| 385 // 10 =", 385 // 10, "| 385 / 10 =", 385 / 10)
while x > 0:
    print(f"x = {x:>3}, chữ số cuối = {x % 10}, còn lại = {x // 10}")
    x //= 10        # x //= 10 là viết tắt của x = x // 10
````

### 2.2. Tổng chữ số bằng vòng `while`

Ghép lại: mỗi vòng lấy `x % 10` cộng vào `total`, rồi `x //= 10`. Vòng dừng khi `x == 0`.

Chú ý: với `x = 0`, vòng `while x > 0` **không chạy lần nào** → tổng = 0. Đúng ý: tổng chữ số của 0 là 0.

````python
def digit_sum(x: int) -> int:
    total = 0
    while x > 0:
        total += x % 10
        x //= 10
    return total

for x in [0, 7, 38, 11, 385, 999, 2147483647]:
    print(x, "-> tổng chữ số =", digit_sum(x))
````

### 2.3. Cách "lười": `str`, duyệt ký tự và `int`

- `str(385)` → chuỗi `"385"`.
- `for ch in "385"` lần lượt cho `"3"`, `"8"`, `"5"` (là **ký tự**, chưa phải số).
- `int("8")` → số `8`.
- `sum(...)` cộng tất cả phần tử của một dãy. Biểu thức `int(ch) for ch in s` là **generator expression**: "với mỗi `ch` trong `s`, tạo ra `int(ch)`".

````python
s = str(385)
print(repr(s), type(s))
print([ch for ch in s])                 # list các ký tự
print([int(ch) for ch in s])            # list các số
print(sum(int(ch) for ch in s))         # tổng = 16
print("'8' + '5' =", "8" + "5", "(nối chuỗi, không phải cộng số!)")
````

`a % 9` là số dư khi chia `a` cho 9, luôn nằm trong `0..8`. Ta sẽ cần thêm phép **dịch** `1 + (a - 1) % 9` để đưa kết quả vào `1..9`. Thử trước cho quen tay:

## 3. Cách 1 — Mô phỏng đúng như đề (dùng chuỗi)

Ý tưởng trực giác nhất: **làm đúng lời đề nói**.

```text
trong khi num còn >= 10:
    num = tổng các chữ số của num
trả về num
```

Đây là baseline tốt: ngắn, đúng, và dùng làm "oracle" (đáp án chuẩn) để kiểm tra các cách khéo hơn về sau.

````python
def add_digits_str(num: int) -> int:
    while num >= 10:                       # còn >= 2 chữ số
        num = sum(int(ch) for ch in str(num))
    return num

for num in [38, 0, 9, 10, 99, 12345, 2147483647]:
    print(num, "->", add_digits_str(num))
````

**Vì sao điều kiện là `num >= 10` chứ không phải `num > 10`?** Số `10` có hai chữ số (`1` và `0`), vẫn phải cộng tiếp: `10 -> 1`. Nếu viết `> 10` thì `10` sẽ bị trả nguyên → sai.

**Bao nhiêu vòng?** Số có `d` chữ số thì tổng chữ số `<= 9d`. Với `num <= 2^31 - 1` (10 chữ số), sau 1 vòng còn `<= 90`, sau vòng 2 còn `<= 17`, sau vòng 3 chắc chắn `<= 9`. Vậy số vòng rất ít — mỗi vòng `O(d)`.

````python
def count_rounds(num: int) -> int:
    rounds = 0
    while num >= 10:
        num = sum(int(ch) for ch in str(num))
        rounds += 1
    return rounds

worst = max(range(0, 200000), key=count_rounds)
print("Trong 0..199999, số cần nhiều vòng nhất:", worst, "->", count_rounds(worst), "vòng")
print("2147483647 cần", count_rounds(2147483647), "vòng")
````

## 4. Cách 2 — Mô phỏng bằng toán (`% 10`, `// 10`), không tạo chuỗi

Giống cách 1, chỉ thay việc tính tổng chữ số bằng vòng `while` toán học ở mục 2.2. Lợi ích: không tạo chuỗi tạm, và cách nghĩ này dùng được ở mọi ngôn ngữ.

Ta có **hai vòng lồng nhau**:
- Vòng ngoài: "còn >= 10 thì cộng chữ số tiếp".
- Vòng trong: "tách và cộng từng chữ số".

````python
def add_digits_math(num: int) -> int:
    while num >= 10:
        total = 0
        while num > 0:
            total += num % 10
            num //= 10
        num = total
    return num

for num in [38, 0, 9, 10, 99, 12345, 2147483647]:
    print(num, "->", add_digits_math(num), "| khớp cách 1:", add_digits_math(num) == add_digits_str(num))
````

### 4.1. Biến thể: cộng "dồn" trong một vòng duy nhất

Mẹo nhỏ: mỗi khi `total` vượt quá 9 ta có thể lập tức "cộng chữ số" của nó — `total = total // 10 + total % 10`. Kết quả cuối vẫn giống, vì (như sẽ chứng minh ở mục 5.1) phép cộng chữ số **giữ nguyên số dư khi chia 9**. Đây là bước đệm để hiểu công thức `O(1)`.

````python
def add_digits_one_loop(num: int) -> int:
    total = 0
    while num > 0:
        total += num % 10
        num //= 10
        if total >= 10:
            total = total // 10 + total % 10   # total <= 9 + 9 = 18 nên gộp 1 lần là về <= 9
    return total

print([add_digits_one_loop(x) for x in [38, 0, 9, 10, 99, 12345, 2147483647]])
assert all(add_digits_one_loop(x) == add_digits_str(x) for x in range(5000))
print("Khớp cách 1 trên 0..4999")
````

## 5. Nút thắt, quan sát dữ liệu và insight

Cách 1–2 đã đủ nhanh để Accepted. Nhưng follow-up đòi **không vòng lặp, O(1)**. Muốn vậy phải có **công thức**.

Gợi ý của đề: *"Các kết quả có thể là gì? Chúng xuất hiện tuần hoàn hay ngẫu nhiên?"* → Hãy **in bảng** kết quả cho `num = 0..30` và nhìn.

````python
row_num = []
row_ans = []
for num in range(0, 31):
    row_num.append(f"{num:>3}")
    row_ans.append(f"{add_digits_str(num):>3}")
print("num:", "".join(row_num))
print("ans:", "".join(row_ans))
````

**Quan sát:**
- `0 -> 0` (riêng biệt).
- Từ `1` trở đi kết quả chạy `1, 2, ..., 9, 1, 2, ..., 9, ...` — **chu kỳ 9**.
- Kết quả là `9` đúng khi `num` chia hết cho 9 (`9, 18, 27, ...`).

Phỏng đoán: với `num > 0`, đáp án = "số dư của `num` khi chia 9", nhưng **thay 0 bằng 9**. Viết gọn: `1 + (num - 1) % 9`.

Đừng tin phỏng đoán chỉ vì 30 số đầu đúng — mục 5.1 chứng minh nó.

````python
def guess(num):
    return 0 if num == 0 else 1 + (num - 1) % 9

bad = [n for n in range(0, 100000) if guess(n) != add_digits_str(n)]
print("Số phản ví dụ trong 0..99999:", len(bad))
````

### 5.1. Insight — vì sao lại là chu kỳ 9? (digital root)

**Bổ đề:** `10 ≡ 1 (mod 9)` vì `10 = 9 + 1`. Suy ra `10^k ≡ 1 (mod 9)` với mọi `k`.

Viết `num` theo chữ số: `num = d_k·10^k + ... + d_1·10 + d_0`. Thay mỗi `10^k` bằng `1` (mod 9):

```text
num ≡ d_k + ... + d_1 + d_0   (mod 9)
```

Tức là **num và tổng chữ số của nó có cùng số dư khi chia 9**. Mỗi bước cộng chữ số không làm đổi số dư mod 9 → kết quả cuối cùng `r` (một chữ số) cũng có `r ≡ num (mod 9)`.

Các số một chữ số `1..9` có số dư mod 9 lần lượt `1, 2, ..., 8, 0` — **mỗi số dư ứng đúng một chữ số**. Vậy:
- `num == 0` → `0`.
- `num > 0` và `num % 9 == 0` → `9` (không thể là 0 vì tổng chữ số của số dương luôn dương).
- còn lại → `num % 9`.

Ba trường hợp gộp lại thành **một công thức** `1 + (num - 1) % 9` cho `num > 0`. Kiểm tra số:

````python
num = 38
print("38 % 9 =", 38 % 9, "| 3 + 8 = 11, 11 % 9 =", 11 % 9, "| 1 + 1 = 2, 2 % 9 =", 2 % 9)
print("10^k % 9:", [10 ** k % 9 for k in range(6)])
for num in [9, 18, 27, 99, 38]:
    print(num, "-> num % 9 =", num % 9, "| 1 + (num-1) % 9 =", 1 + (num - 1) % 9)
````

### 5.2. Vì sao `1 + (num - 1) % 9` lại "đổi 0 thành 9"?

Dịch trước khi lấy dư rồi dịch lại: `num - 1` đưa dãy `1..9` về `0..8`; `% 9` giữ nguyên `0..8` (và quấn vòng cho số lớn); `+1` trả về `1..9`.

| num | num - 1 | (num-1) % 9 | +1 |
|---:|---:|---:|---:|
| 1 | 0 | 0 | 1 |
| 8 | 7 | 7 | 8 |
| 9 | 8 | 8 | **9** |
| 10 | 9 | 0 | 1 |
| 18 | 17 | 8 | **9** |

Với `num = 0`: `(0 - 1) % 9` trong **Python** là `8` (dư không âm) → công thức cho `9` — **sai**! Phải tách riêng `num == 0`.

````python
print("(-1) % 9 trong Python =", (-1) % 9)
print("Công thức thô với num = 0:", 1 + (0 - 1) % 9, "(sai, đáp án đúng là 0)")
````

## 6. Lời giải chính — Digital root `O(1)`, giảng từng đoạn

**Đoạn 1 — biên:** `if num == 0: return 0`.

**Đoạn 2 — công thức:** `return 1 + (num - 1) % 9`.

Hai dòng, không vòng lặp, không đệ quy → đáp ứng follow-up.

Một cách viết tương đương hay gặp:
```python
if num == 0: return 0
return 9 if num % 9 == 0 else num % 9
```

````python
def add_digits_formula(num: int) -> int:
    if num == 0:
        return 0
    return 1 + (num - 1) % 9

def add_digits_formula_v2(num: int) -> int:
    if num == 0:
        return 0
    return 9 if num % 9 == 0 else num % 9

for num in [38, 0, 9, 10, 99, 12345, 2147483647]:
    print(num, "->", add_digits_formula(num), add_digits_formula_v2(num))
````

## 7. TRACE từng bước

### 7.1. Trace mô phỏng (cách 2) với `num = 9875`

| Vòng ngoài | num đầu vòng | các chữ số | tổng | num >= 10? |
|---:|---:|---|---:|:---:|
| 1 | 9875 | 5, 7, 8, 9 | 29 | có → tiếp |
| 2 | 29 | 9, 2 | 11 | có → tiếp |
| 3 | 11 | 1, 1 | 2 | không → dừng |

Kết quả: `2`. Đối chiếu công thức: `9875 % 9 = 2` ✔.

````python
def trace_add_digits(num: int) -> int:
    print(f"Bắt đầu num = {num}")
    rnd = 0
    while num >= 10:
        rnd += 1
        total = 0
        x = num
        digits = []
        while x > 0:
            d = x % 10
            digits.append(d)
            total += d
            x //= 10
        print(f"  vòng {rnd}: num = {num:>6} | chữ số (từ cuối) = {digits} | tổng = {total}")
        num = total
    print(f"  dừng vì {num} < 10 -> trả {num}")
    return num

trace_add_digits(9875)
print()
trace_add_digits(38)
print()
trace_add_digits(7)
````

### 7.2. Trace công thức — thấy số dư mod 9 được giữ nguyên

Hàm dưới in `num % 9` ở **mỗi** bước mô phỏng: cột này không đổi. Đó chính là bất biến ở mục 5.1.

````python
def trace_mod9(num: int) -> int:
    print(f"num = {num}, công thức dự đoán = {add_digits_formula(num)}")
    while True:
        print(f"  giá trị = {num:>10} | giá trị % 9 = {num % 9}")
        if num < 10:
            return num
        num = digit_sum(num)

trace_mod9(2147483647)
print()
trace_mod9(999999999)
````

## 8. Code nộp LeetCode

### 8.1. Bản công thức `O(1)` (khuyên dùng)

````python
class Solution:
    def addDigits(self, num: int) -> int:
        if num == 0:
            return 0
        return 1 + (num - 1) % 9

sol = Solution()
print(sol.addDigits(38), sol.addDigits(0))
````

### 8.2. Bản mô phỏng (dễ giải thích, cũng Accepted)

Đặt tên class khác để không ghi đè `Solution` ở trên trong notebook. Khi nộp, đổi lại tên thành `Solution`.

````python
class SolutionSimulation:
    def addDigits(self, num: int) -> int:
        while num >= 10:
            total = 0
            while num > 0:
                total += num % 10
                num //= 10
            num = total
        return num

print(SolutionSimulation().addDigits(38), SolutionSimulation().addDigits(0))
````

## 9. Vì sao đúng? (invariant)

**Mô phỏng:** Bất biến của vòng ngoài: *"đáp án của bài với giá trị `num` hiện tại = đáp án với `num` ban đầu"* — vì ta chỉ thực hiện đúng phép biến đổi mà đề mô tả. Vòng dừng vì với `num >= 10`, tổng chữ số **nhỏ hơn hẳn** `num` (mỗi chữ số `d_k` ở vị trí `k >= 1` đóng góp `d_k·10^k > d_k`), nên `num` giảm nghiêm ngặt và không thể giảm mãi.

**Công thức:** Bất biến *"`num % 9` không đổi qua mỗi lần cộng chữ số"* (mục 5.1). Kết quả cuối là số một chữ số có cùng số dư mod 9 với `num`, và với `num > 0` chỉ có đúng một số trong `1..9` thỏa → chính là `1 + (num - 1) % 9`.

````python
# Kiểm chứng "tổng chữ số < num khi num >= 10" trên một miền lớn
assert all(digit_sum(x) < x for x in range(10, 200000))
# Kiểm chứng bất biến mod 9
assert all(digit_sum(x) % 9 == x % 9 for x in range(0, 200000))
print("Hai bất biến đúng trên 0..199999")
````

## 12. Lỗi người mới hay gặp

### Lỗi 1: Điều kiện dừng `num > 10`

````python
def wrong_stop(num):
    while num > 10:                 # SAI: 10 có 2 chữ số
        num = sum(int(c) for c in str(num))
    return num

print("wrong_stop(10) =", wrong_stop(10), "| đúng =", add_digits_str(10))
print("wrong_stop(19) =", wrong_stop(19), "| đúng =", add_digits_str(19))   # 19 -> 10 -> dừng sai
````

**Sửa:** dùng `while num >= 10` (hoặc `while num > 9`).

### Lỗi 2: Dùng `num % 9` trực tiếp

### Lỗi 5: Dùng `/` thay `//`

`num / 10` cho số thực (`3.8`); lấy `% 10` trên số thực lại ra số thực (`3.8`) chứ không phải một chữ số → tổng sai, và `while num > 0` với số thực chia mãi sẽ chạy rất nhiều vòng vô nghĩa. Luôn dùng `//` cho số nguyên.

## 14. Biến thể

### 14.1. Digital root trong cơ số `b`

Lập luận y hệt: `b ≡ 1 (mod b - 1)`. Vậy digital root cơ số `b` của `num > 0` là `1 + (num - 1) % (b - 1)`.

````python
def digit_sum_base(x, base):
    t = 0
    while x > 0:
        t += x % base
        x //= base
    return t

def droot_sim(x, base):
    while x >= base:
        x = digit_sum_base(x, base)
    return x

def droot_formula(x, base):
    return 0 if x == 0 else 1 + (x - 1) % (base - 1)

for base in [2, 3, 8, 16]:
    assert all(droot_sim(x, base) == droot_formula(x, base) for x in range(3000))
print("Công thức digital root đúng với cơ số 2, 3, 8, 16")
print("Cơ số 16, x = 0xFF ->", droot_formula(0xFF, 16))
````

### 14.2. Kiểm tra chia hết cho 3 và 9 bằng tổng chữ số

Từ bổ đề: `num` chia hết cho 9 ⇔ tổng chữ số chia hết cho 9. Vì `10 ≡ 1 (mod 3)` nên điều tương tự đúng với 3.

````python
for x in [123456, 111111111, 98765, 81]:
    s = digit_sum(x)
    print(f"{x}: tổng chữ số = {s} | chia hết 3? {x % 3 == 0} = {s % 3 == 0} | chia hết 9? {x % 9 == 0} = {s % 9 == 0}")
````

### 14.3. Đếm số vòng mô phỏng ("additive persistence")

Không chỉ hỏi kết quả, mà hỏi **phải cộng bao nhiêu lần** — câu này không có công thức `O(1)`, phải mô phỏng. Hàm `count_rounds` ở mục 3 chính là lời giải.
