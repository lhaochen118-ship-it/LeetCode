## 1. Mục tiêu học

Sau notebook này bạn sẽ:

1. Hiểu đề: cộng các phần tử ở chỉ số **chẵn** và trừ các phần tử ở chỉ số **lẻ** (chỉ số đếm từ `0`), tức `nums[0] - nums[1] + nums[2] - nums[3] + ...`.
2. Dùng được toán tử `%` để phân biệt chỉ số chẵn/lẻ, `enumerate` để lấy cả chỉ số lẫn giá trị, và **slicing có bước** `nums[::2]`, `nums[1::2]`.
3. Biết ba kiểu viết cùng một ý: dùng điều kiện `i % 2`, dùng **biến dấu** đảo qua từng bước, và dùng **tách hai nhóm** rồi trừ tổng.
4. Viết 5 cách khác nhau (điều kiện chẵn/lẻ, biến dấu, slicing, lũy thừa `(-1) ** i`, xử lý theo cặp) và đối chiếu bằng test ngẫu nhiên có `random.seed`.
5. Tránh các lỗi quen thuộc: bắt đầu bằng dấu trừ, đếm chỉ số từ 1, quên phần tử cuối khi độ dài lẻ, và nhầm với tổng thường.

## 2. Đề bài & ví dụ

Cho mảng số nguyên `nums`. **Tổng đan dấu** (alternating sum) của mảng là tổng của các phần tử ở chỉ số chẵn trừ đi tổng của các phần tử ở chỉ số lẻ. Hãy trả về tổng đan dấu đó.

Chỉ số tính từ `0`: phần tử đầu tiên có chỉ số `0` (chẵn) nên được **cộng**.

| Ví dụ | nums | Output | Giải thích |
|---|---|---|---|
| 1 | `[1,3,5,7]` | `-4` | `1 - 3 + 5 - 7 = -4` |
| 2 | `[100]` | `100` | Chỉ có một phần tử ở chỉ số `0`, nên kết quả là chính nó |

Constraints: `1 <= nums.length <= 100`, `1 <= nums[i] <= 100`.

````python
for nums in [[1, 3, 5, 7], [100]]:
    even_sum = sum(v for i, v in enumerate(nums) if i % 2 == 0)
    odd_sum = sum(v for i, v in enumerate(nums) if i % 2 == 1)
    print(f"nums={nums}  chỉ số chẵn cộng: {even_sum}, chỉ số lẻ trừ: {odd_sum}  -> {even_sum - odd_sum}")
````

## 3. Kiến thức nền

### 3.1. Chỉ số bắt đầu từ 0

Trong Python, phần tử đầu tiên của list có chỉ số `0`, phần tử thứ hai có chỉ số `1`, ... Vậy "chỉ số chẵn" là vị trí thứ 1, 3, 5, ... **theo cách đếm của con người**. Đây là chỗ người mới dễ nhầm.

````python
nums = [10, 20, 30, 40]
for i in range(len(nums)):
    print("chỉ số", i, "->", nums[i], "(chẵn)" if i % 2 == 0 else "(lẻ)")
````

### 3.2. Toán tử `%` (phần dư)

`i % 2` là phần dư khi chia `i` cho `2`: bằng `0` nếu `i` chẵn, bằng `1` nếu `i` lẻ. Đây là cách kiểm tra chẵn/lẻ chuẩn nhất.

````python
print([i % 2 for i in range(8)])
print(7 % 2, 10 % 2, 0 % 2)
print([i for i in range(8) if i % 2 == 0])
````

### 3.3. `enumerate`

`enumerate(nums)` sinh từng cặp `(chỉ số, giá trị)` nên không cần viết `range(len(nums))` rồi `nums[i]`.

````python
nums = [5, 6, 7]
for i, v in enumerate(nums):
    print(i, v)
print(list(enumerate(nums)))
print(list(enumerate(nums, start=1)))     # có thể đổi chỉ số bắt đầu
````

### 3.4. Slicing có bước: `a[start:stop:step]`

`nums[::2]` lấy các phần tử ở chỉ số `0, 2, 4, ...`; `nums[1::2]` lấy chỉ số `1, 3, 5, ...`. Hai lát cắt này chính là hai nhóm "cộng" và "trừ" của đề.

````python
nums = [1, 3, 5, 7, 9]
print(nums[::2], nums[1::2])
print(sum(nums[::2]), sum(nums[1::2]))
print(nums[::2] + nums[1::2] == sorted(nums, key=lambda x: nums.index(x) % 2))
````

### 3.5. Lũy thừa của `-1` và biến dấu

`(-1) ** 0 = 1`, `(-1) ** 1 = -1`, `(-1) ** 2 = 1`, ... nên `(-1) ** i` cho dấu `+` ở chỉ số chẵn và `-` ở chỉ số lẻ. Ngoài ra có thể dùng một biến `sign` và đảo dấu mỗi bước bằng `sign = -sign`.

````python
print([(-1) ** i for i in range(6)])
sign = 1
signs = []
for _ in range(6):
    signs.append(sign)
    sign = -sign
print(signs)
````

### 3.6. `sum()` với generator expression

`sum(v for v in nums if cond)` cộng các phần tử thỏa điều kiện mà **không tạo list trung gian**. Biểu thức trong ngoặc là *generator*: sinh từng giá trị khi `sum` cần, nên tốn `O(1)` bộ nhớ. So sánh với list comprehension `[...]` tạo hẳn một list.

````python
nums = [1, 3, 5, 7]
print(sum(v for v in nums if v > 2))                 # generator: không tạo list
print(sum([v for v in nums if v > 2]))               # list comprehension: tạo list rồi mới cộng
print(sum(v for i, v in enumerate(nums) if i % 2 == 0))
gen = (v * 2 for v in nums)
print(type(gen).__name__, list(gen), list(gen))      # generator chỉ dùng được một lần
````

## 4. Ý tưởng nền tảng / cách nghĩ trực giác

Hãy tưởng tượng bạn đi dọc một hàng ô, cứ ô chẵn thì **nhận tiền**, ô lẻ thì **trả tiền**. Cuối hàng, số dư của bạn chính là tổng đan dấu. Vì mỗi ô chỉ ghé một lần nên chỉ cần **một lượt duyệt**.

Tính tay ví dụ 1: `[1, 3, 5, 7]`

- chỉ số 0: nhận `1` -> dư `1`
- chỉ số 1: trả `3` -> dư `-2`
- chỉ số 2: nhận `5` -> dư `3`
- chỉ số 3: trả `7` -> dư `-4`

Kết quả `-4`. Cách nghĩ thứ hai: chia thành hai nhóm: nhóm chẵn `[1, 5]` có tổng `6`, nhóm lẻ `[3, 7]` có tổng `10`, đáp án `6 - 10 = -4`.

### Bảng tính tay

| nums | tổng chỉ số chẵn | tổng chỉ số lẻ | đáp án |
|---|---|---|---|
| `[1,3,5,7]` | 6 | 10 | -4 |
| `[100]` | 100 | 0 | 100 |
| `[2,2,2]` | 4 | 2 | 2 |
| `[9,1,8,2,7]` | 24 | 3 | 21 |

````python
for nums in [[1, 3, 5, 7], [100], [2, 2, 2], [9, 1, 8, 2, 7]]:
    e, o = sum(nums[::2]), sum(nums[1::2])
    print(nums, e, o, e - o)
````

## 5. Cách 1 — Brute Force / baseline

Duyệt từng chỉ số `i`; nếu `i` chẵn thì cộng `nums[i]`, ngược lại thì trừ. Đây là bản dịch **trực tiếp** từ câu chữ của đề.

Nút thắt: bài này gần như không có nút thắt về thời gian (`O(n)` là tối ưu vì phải đọc mọi phần tử). Phần "tối ưu" duy nhất là viết cho gọn và ít lỗi hơn.

````python
class SolutionIndex:
    def alternatingSum(self, nums):
        total = 0
        for i in range(len(nums)):
            if i % 2 == 0:
                total += nums[i]
            else:
                total -= nums[i]
        return total

print(SolutionIndex().alternatingSum([1, 3, 5, 7]), SolutionIndex().alternatingSum([100]))
````

**Độ phức tạp:** Time `O(n)`, Space `O(1)`.

## 6. Ý tưởng thuật toán / bước đột phá

**Bước đột phá:** dấu của phần tử chỉ phụ thuộc vào **tính chẵn lẻ của chỉ số**, không phụ thuộc giá trị. Vì vậy ta có thể:

- đảo dấu sau mỗi phần tử (biến `sign`), hoặc
- tách mảng thành hai nhóm bằng slicing rồi trừ hai tổng, hoặc
- nhân mỗi phần tử với `(-1) ** i`.

**Quy tắc "nếu… thì…":**

- Nếu chỉ số `i` chẵn -> `+nums[i]`; nếu `i` lẻ -> `-nums[i]`.
- Nếu độ dài lẻ thì phần tử cuối (chỉ số chẵn) vẫn được **cộng**.
- Nếu chỉ có một phần tử thì đáp án là chính nó.

**Invariant:** sau khi xử lý `i` phần tử đầu, `total` bằng tổng đan dấu của đoạn `nums[0..i-1]`.

````python
# Kiểm chứng ý tưởng: tổng đan dấu = (tổng chẵn) - (tổng lẻ) = tổng - 2 * (tổng lẻ)
import random
random.seed(3701)
for _ in range(1000):
    n = random.randint(1, 100)
    nums = [random.randint(1, 100) for _ in range(n)]
    alt = sum(nums[::2]) - sum(nums[1::2])
    assert alt == sum(nums) - 2 * sum(nums[1::2])
print("đẳng thức alt = tổng - 2 * tổng_lẻ đúng")
````

## 7. Các cách tiếp theo

### Cách 2 — Biến dấu `sign`

Giữ một biến `sign = 1`; mỗi bước cộng `sign * v` rồi đảo `sign = -sign`. Không cần `%`.

````python
class SolutionSign:
    def alternatingSum(self, nums):
        total, sign = 0, 1
        for v in nums:
            total += sign * v
            sign = -sign
        return total

print(SolutionSign().alternatingSum([1, 3, 5, 7]), SolutionSign().alternatingSum([100]))
````

**Độ phức tạp:** Time `O(n)`, Space `O(1)`.

### Cách 3 — Slicing hai nhóm

`sum(nums[::2]) - sum(nums[1::2])`: một dòng, đọc gần như câu đề. Slicing tạo hai list con nên tốn thêm `O(n)` bộ nhớ.

````python
class SolutionSlice:
    def alternatingSum(self, nums):
        return sum(nums[::2]) - sum(nums[1::2])

print(SolutionSlice().alternatingSum([1, 3, 5, 7]), SolutionSlice().alternatingSum([100]))
````

**Độ phức tạp:** Time `O(n)`, Space `O(n)` (hai list con).

### Cách 4 — `(-1) ** i`

Cho `enumerate`, nhân `v` với `(-1) ** i`. Gọn, nhưng lũy thừa chậm hơn một chút so với đảo dấu.

````python
class SolutionPow:
    def alternatingSum(self, nums):
        return sum(v * (-1) ** i for i, v in enumerate(nums))

print(SolutionPow().alternatingSum([1, 3, 5, 7]), SolutionPow().alternatingSum([100]))
````

**Độ phức tạp:** Time `O(n)`, Space `O(1)` (generator).

### Cách 5 — Xử lý theo cặp

Gom mỗi cặp `(nums[i], nums[i+1])` thành `nums[i] - nums[i+1]`; nếu độ dài lẻ thì cộng thêm phần tử cuối. Cách này làm rõ vì sao phần tử cuối của mảng lẻ được cộng.

````python
class SolutionPairs:
    def alternatingSum(self, nums):
        total = 0
        n = len(nums)
        for i in range(0, n - 1, 2):
            total += nums[i] - nums[i + 1]
        if n % 2 == 1:
            total += nums[-1]
        return total

print(SolutionPairs().alternatingSum([1, 3, 5, 7]), SolutionPairs().alternatingSum([100]), SolutionPairs().alternatingSum([9, 1, 8, 2, 7]))
````

**Độ phức tạp:** Time `O(n)`, Space `O(1)`.

## 8. TRACE từng bước

Trace Cách 2 (biến dấu) với `nums = [9, 1, 8, 2, 7]`:

| i | v | sign trước | total sau | sign sau |
|---|---|---|---|---|
| 0 | 9 | +1 | 9 | -1 |
| 1 | 1 | -1 | 8 | +1 |
| 2 | 8 | +1 | 16 | -1 |
| 3 | 2 | -1 | 14 | +1 |
| 4 | 7 | +1 | 21 | -1 |

Kết quả `21`, khớp bảng tính tay (24 - 3).

````python
def trace_sign(nums):
    total, sign = 0, 1
    print("nums =", nums)
    for i, v in enumerate(nums):
        before = sign
        total += sign * v
        sign = -sign
        print(f"  i={i} v={v:>3} sign_trước={before:+d} -> total={total:>4}  sign_sau={sign:+d}")
    return total

assert trace_sign([9, 1, 8, 2, 7]) == 21
assert trace_sign([1, 3, 5, 7]) == -4
````

````python
def trace_pairs(nums):
    n, total = len(nums), 0
    print("nums =", nums)
    for i in range(0, n - 1, 2):
        total += nums[i] - nums[i + 1]
        print(f"  cặp ({nums[i]}, {nums[i+1]}) -> +{nums[i] - nums[i+1]}  total={total}")
    if n % 2 == 1:
        total += nums[-1]
        print(f"  độ dài lẻ: cộng phần tử cuối {nums[-1]} -> total={total}")
    return total

assert trace_pairs([9, 1, 8, 2, 7]) == 21
````

## 9. Code nộp LeetCode

````python
from typing import List, Optional

class Solution:
    def alternatingSum(self, nums: List[int]) -> int:
        total, sign = 0, 1
        for v in nums:
            total += sign * v
            sign = -sign
        return total
````

## 10. Vì sao thuật toán đúng

- **Đúng theo định nghĩa:** đề định nghĩa tổng đan dấu là `nums[0] - nums[1] + nums[2] - ...`. Biến `sign` bắt đầu bằng `+1` ở chỉ số `0` và đổi dấu sau mỗi phần tử nên ở chỉ số `i` có giá trị `(-1) ** i`.
- **Quy nạp:** với mảng rỗng tổng bằng `0`. Giả sử sau `i` phần tử `total` đúng; phần tử thứ `i+1` (chỉ số `i`) được cộng với dấu `(-1) ** i`, đúng như định nghĩa, nên `total` vẫn đúng.
- **Mảng lẻ:** phần tử cuối có chỉ số chẵn nên dấu `+`, vòng lặp tự xử lý mà không cần nhánh đặc biệt (Cách 5 phải xử lý riêng vì ghép cặp).
- **Cách slicing:** `nums[::2]` đúng là các chỉ số chẵn và `nums[1::2]` là các chỉ số lẻ, hợp hai nhóm này là toàn bộ mảng, không trùng.

## 12. Bộ test

Ví dụ của đề, các biên (một phần tử, hai phần tử, độ dài lẻ/chẵn, toàn số `1`, toàn số `100`, độ dài `100`) và so sánh **tất cả cách** với input ngẫu nhiên (hợp lệ theo constraints).

````python
impls = [SolutionIndex(), SolutionSign(), SolutionSlice(), SolutionPow(), SolutionPairs(), Solution()]

cases = [
    ([1, 3, 5, 7], -4), ([100], 100), ([1], 1), ([5, 3], 2), ([3, 5], -2),
    ([2, 2, 2], 2), ([9, 1, 8, 2, 7], 21), ([1] * 100, 0), ([100] * 100, 0), ([100] * 99, 100),
    ([1, 100], -99), ([100, 1], 99), (list(range(1, 101)), -50),
]
for nums, exp in cases:
    assert 1 <= len(nums) <= 100 and all(1 <= v <= 100 for v in nums)
    for impl in impls:
        got = impl.alternatingSum(list(nums))
        assert got == exp, (type(impl).__name__, nums, got, exp)
print("ví dụ + edge OK:", len(cases), "ca")
````

````python
import random
random.seed(37010)
for _ in range(1000):
    n = random.randint(1, 100)
    nums = [random.randint(1, 100) for _ in range(n)]
    exp = sum(v if i % 2 == 0 else -v for i, v in enumerate(nums))
    outs = [impl.alternatingSum(list(nums)) for impl in impls]
    assert all(o == exp for o in outs), (nums, outs, exp)
    assert outs[0] == sum(nums) - 2 * sum(nums[1::2])
print("1000 test ngẫu nhiên (seed=37010): 6 cách giống nhau")
````

````python
# Tính chất: đảo thứ tự mảng độ dài chẵn thì đổi dấu đáp án; độ dài lẻ thì giữ nguyên
for _ in range(300):
    n = random.randint(1, 100)
    nums = [random.randint(1, 100) for _ in range(n)]
    a, b = Solution().alternatingSum(nums), Solution().alternatingSum(nums[::-1])
    assert (b == -a) if n % 2 == 0 else (b == a)
print("tính chất đảo ngược OK")
````

## 13. Các lỗi người mới hay gặp

### Lỗi 1: bắt đầu bằng dấu trừ

Đề: phần tử chỉ số `0` được **cộng**. Nếu khởi tạo `sign = -1`, kết quả bị đổi dấu.

````python
def wrong_start_minus(nums):
    total, sign = 0, -1                     # SAI: phải bắt đầu bằng +1
    for v in nums:
        total += sign * v
        sign = -sign
    return total

print(wrong_start_minus([1, 3, 5, 7]), "!=", Solution().alternatingSum([1, 3, 5, 7]))
assert wrong_start_minus([1, 3, 5, 7]) == 4 != Solution().alternatingSum([1, 3, 5, 7])
````

### Lỗi 2: đếm chỉ số từ 1

"Vị trí lẻ" theo cách đếm của người (1, 3, 5...) là chỉ số chẵn (0, 2, 4...) của Python. Dùng `enumerate(nums, 1)` rồi kiểm tra `i % 2 == 0` làm dấu bị ngược.

````python
def wrong_one_indexed(nums):
    total = 0
    for i, v in enumerate(nums, 1):          # SAI: chỉ số bắt đầu từ 1
        total += v if i % 2 == 0 else -v
    return total

print(wrong_one_indexed([1, 3, 5, 7]), "!=", Solution().alternatingSum([1, 3, 5, 7]))
assert wrong_one_indexed([1, 3, 5, 7]) == 4 != -4
````

### Lỗi 3: quên phần tử cuối khi ghép cặp với độ dài lẻ

Nếu chỉ ghép cặp `(nums[i], nums[i+1])` mà không cộng phần tử còn lại thì mất phần tử cuối.

````python
def wrong_pairs_forget_last(nums):
    total = 0
    for i in range(0, len(nums) - 1, 2):
        total += nums[i] - nums[i + 1]
    return total                             # SAI: thiếu phần tử cuối khi len lẻ

print(wrong_pairs_forget_last([9, 1, 8, 2, 7]), "!=", Solution().alternatingSum([9, 1, 8, 2, 7]))
assert wrong_pairs_forget_last([9, 1, 8, 2, 7]) == 14 != 21
assert wrong_pairs_forget_last([100]) == 0 != 100
````

### Lỗi 4: trả về tổng thường

Đọc vội đề rồi `return sum(nums)` bỏ qua phần "trừ các chỉ số lẻ".

````python
def wrong_plain_sum(nums):
    return sum(nums)                         # SAI: không đan dấu

print(wrong_plain_sum([1, 3, 5, 7]), "!=", Solution().alternatingSum([1, 3, 5, 7]))
assert wrong_plain_sum([1, 3, 5, 7]) == 16 != -4
````

### Lỗi 5: dùng `abs` hoặc ép kết quả không âm

Đề cho phép đáp án **âm** (ví dụ `-4`). Bọc `abs(...)` làm sai mọi ca nhóm lẻ lớn hơn nhóm chẵn.

````python
def wrong_abs(nums):
    return abs(sum(nums[::2]) - sum(nums[1::2]))     # SAI

print(wrong_abs([1, 3, 5, 7]), "!=", Solution().alternatingSum([1, 3, 5, 7]))
assert wrong_abs([1, 3, 5, 7]) == 4 != -4
````

## 14. Mẹo làm bài nhanh & nhận dạng pattern

- Thấy "chẵn/lẻ chỉ số" -> nghĩ ngay `i % 2`, hoặc slicing `[::2]` và `[1::2]`.
- "Đan dấu" -> biến `sign` đảo mỗi bước, hoặc `(-1) ** i`.
- Một dòng nhanh nhất để nộp: `return sum(nums[::2]) - sum(nums[1::2])`.
- Kiểm tra nhanh bằng ví dụ 2 (`[100] -> 100`): nếu code của bạn trả `-100` là đang bắt đầu sai dấu.
- Độ dài lẻ: phần tử cuối **luôn được cộng**.
- Tính chất để tự kiểm tra: tổng đan dấu `=` tổng `-` 2 lần tổng các phần tử chỉ số lẻ.

### Mở rộng: tổng đan dấu của mọi tiền tố

Nếu đề hỏi tổng đan dấu của **mọi** tiền tố `nums[0..i]` thì `total` sau mỗi bước chính là đáp án của tiền tố đó. Có thể lưu lại để trả lời nhiều truy vấn.

````python
def alternating_prefixes(nums):
    out, total, sign = [], 0, 1
    for v in nums:
        total += sign * v
        sign = -sign
        out.append(total)
    return out

pref = alternating_prefixes([1, 3, 5, 7])
print(pref)                                          # [1, -2, 3, -4]
assert pref == [Solution().alternatingSum([1, 3, 5, 7][:k]) for k in range(1, 5)]
# tổng đan dấu của đoạn [l..r], dấu tính theo chỉ số gốc: pref[r] - pref[l-1] (nhân thêm (-1)**l)
nums = [9, 1, 8, 2, 7]
pref = alternating_prefixes(nums)
l, r = 1, 3
seg = (pref[r] - (pref[l - 1] if l else 0)) * (-1) ** l
assert seg == Solution().alternatingSum(nums[l:r + 1])
print("đoạn", nums[l:r + 1], "-> tổng đan dấu", seg)
````

## 15. Biến thể của bài

1. **Alternating Digit Sum** (LeetCode 2544): đan dấu các **chữ số** của một số nguyên, chữ số đầu luôn dương.
2. **Maximum Alternating Subsequence Sum** (1911): chọn dãy con để tổng đan dấu lớn nhất (quy hoạch động).
3. **Maximum Alternating Subarray Sum** (3196): tối đa hóa tổng đan dấu của một dãy con liên tiếp.
4. **Sum of Even and Odd Index Elements**: trả về cả hai tổng riêng.
5. **Alternating sum theo nhóm k**: dấu đổi sau mỗi `k` phần tử thay vì mỗi phần tử.
6. **Query nhiều đoạn**: dùng mảng tiền tố đan dấu để trả lời tổng đan dấu của `nums[l..r]` trong `O(1)`.

## 19. Đáp án tham khảo cho bài tập

````python
# Bài 1, 2
assert Solution().alternatingSum([4, 4, 4, 4, 4]) == 4
assert sum([10, 20, 30, 40, 50, 60][::2]) == 90 and sum([10, 20, 30, 40, 50, 60][1::2]) == 120
assert Solution().alternatingSum([10, 20, 30, 40, 50, 60]) == -30

# Bài 3
def alternating_digit_sum(n):
    digits = [int(c) for c in str(n)]
    return Solution().alternatingSum(digits)
assert alternating_digit_sum(521) == 4 and alternating_digit_sum(111) == 1 and alternating_digit_sum(886996) == 0

# Bài 4
def even_odd_sums(nums):
    even = odd = 0
    for i, v in enumerate(nums):
        if i % 2 == 0:
            even += v
        else:
            odd += v
    return even, odd
assert even_odd_sums([1, 3, 5, 7]) == (6, 10)
assert even_odd_sums([100]) == (100, 0)

# Bài 5
def alternating_by_block(nums, k):
    return sum(v if (i // k) % 2 == 0 else -v for i, v in enumerate(nums))
assert alternating_by_block([1, 2, 3, 4, 5, 6], 2) == 7
assert alternating_by_block([1, 3, 5, 7], 1) == Solution().alternatingSum([1, 3, 5, 7])
print("đáp án bài tập OK")
````
