## 1. Mục tiêu học

Sau notebook này bạn sẽ:

1. Hiểu đề: tách **mỗi** số trong `nums` thành các chữ số, rồi ghép tất cả các chữ số lại thành **một mảng phẳng**, giữ nguyên thứ tự: các số theo thứ tự xuất hiện, các chữ số của mỗi số từ **trái sang phải**.
2. Phân biệt `list.append` (thêm **một** phần tử, có thể là list con) với `list.extend` (thêm **từng** phần tử của một iterable).
3. Tách chữ số bằng `str(x)` (đúng thứ tự trái → phải) hoặc bằng `divmod(x, 10)` (thứ tự phải → trái, cần **đảo** lại).
4. Viết 5 cách: chuỗi + `extend`, `divmod` + đảo từng nhóm, duyệt ngược toàn mảng + đảo một lần, ghép chuỗi `"".join`, và điền sẵn vào mảng cấp phát trước từ cuối; đối chiếu bằng test ngẫu nhiên.
5. Tránh các lỗi: quên đảo thứ tự chữ số, `append` một list (mảng lồng nhau), nhầm chữ số kiểu chuỗi với kiểu số, đảo cả mảng nhưng không duyệt ngược.

## 2. Đề bài & ví dụ

Cho mảng số nguyên dương `nums`. Trả về một mảng `answer` gồm **các chữ số** của mọi phần tử trong `nums`, giữ **cùng thứ tự** như khi các số được viết ra lần lượt:

- các phần tử theo đúng thứ tự trong `nums`;
- các chữ số của một phần tử theo thứ tự **từ trái sang phải** (chữ số cao nhất trước).

| Ví dụ | nums | Output | Giải thích |
|---|---|---|---|
| 1 | `[13,25,83,77]` | `[1,3,2,5,8,3,7,7]` | tách `13 → 1,3`; `25 → 2,5`; `83 → 8,3`; `77 → 7,7` |
| 2 | `[7,1,3,9]` | `[7,1,3,9]` | mỗi số có một chữ số |

Constraints: `1 <= nums.length <= 1000`; `1 <= nums[i] <= 10^5`.

````python
nums = [13, 25, 83, 77]
answer = []
for x in nums:
    digits = [int(ch) for ch in str(x)]
    print(f"{x} -> {digits}")
    answer.extend(digits)
print("answer =", answer)
````

## 3. Kiến thức nền

### 3.1. `str(x)` và duyệt ký tự — luôn đúng thứ tự trái → phải

`str(13)` là `"13"`, duyệt `for ch in "13"` cho `'1'` rồi `'3'`: đúng thứ tự ta cần. `int(ch)` đổi ký tự `'1'` thành số `1`. Nhớ rằng `'1'` (chuỗi) và `1` (số) là hai kiểu khác nhau: `'1' != 1`.

````python
print([ch for ch in str(13)])
print([int(ch) for ch in str(13)])
print("1" == 1, int("1") == 1)
````

### 3.2. `list.append` khác `list.extend`

`a.append(x)` thêm **một** phần tử `x` (nếu `x` là list thì thành list con → mảng lồng nhau). `a.extend(it)` thêm **từng** phần tử của `it` vào cuối `a`. Dùng `extend` để "làm phẳng" khi ghép nhóm chữ số vào kết quả.

````python
a = [1]
a.append([2, 3])
print(a)                     # [1, [2, 3]]
b = [1]
b.extend([2, 3])
print(b)                     # [1, 2, 3]
c = []
c.extend("13")
print(c)                     # ['1', '3']: extend một chuỗi sinh ra từng KÝ TỰ (kiểu str)
````

### 3.3. `divmod` — tách chữ số cuối, nhưng thứ tự ngược

`x, d = divmod(x, 10)` cho `x // 10` và chữ số cuối `d = x % 10`. Lặp tới khi `x = 0` ta lấy các chữ số từ **phải sang trái** (`13 → 3, 1`), ngược với thứ tự cần. Do đó phải **đảo** nhóm chữ số trước khi nối vào kết quả.

````python
x = 13
digits = []
while x > 0:
    x, d = divmod(x, 10)
    digits.append(d)
print(digits, "-> sau khi đảo:", digits[::-1])
````

### 3.4. Đảo danh sách: `reversed`, `[::-1]`, `.reverse()`

- `lst[::-1]` tạo **bản sao** đảo ngược.
- `lst.reverse()` đảo **tại chỗ** (trả về `None`!).
- `reversed(lst)` trả về một **iterator** đảo ngược (không tạo bản sao ngay); dùng `list(reversed(lst))` hoặc truyền thẳng cho `extend`.

Lỗi hay gặp: `x = lst.reverse()` làm `x` là `None`.

````python
lst = [3, 1]
print(lst[::-1], list(reversed(lst)))
r = lst.reverse()
print(lst, r)                # r là None
````

### 3.5. Ghép chuỗi bằng `"".join(map(str, ...))`

`"".join(map(str, [13, 25]))` cho `"1325"`: nối các số thành **một chuỗi chữ số**; rồi `list(map(int, s))` đổi từng ký tự về số. Đây là cách ngắn nhất cho đúng đề: các chữ số của mọi số đứng liền nhau theo thứ tự.

````python
s = "".join(map(str, [13, 25, 83, 77]))
print(s)
print(list(map(int, s)))
````

### 3.6. Mảng cấp phát trước `[0] * k` và điền từ cuối

Nếu biết trước tổng số chữ số `k`, ta tạo `ans = [0] * k` rồi **điền từ cuối về đầu**: duyệt `nums` ngược, tách chữ số cuối bằng `% 10` và đặt vào `ans[pos]` với `pos` giảm dần. Nhờ điền từ cuối, chữ số tách từ phải sang trái rơi **đúng vị trí** mà không cần đảo.

````python
nums = [13, 25]
k = sum(len(str(x)) for x in nums)
ans = [0] * k
pos = k - 1
for x in reversed(nums):
    while x > 0:
        ans[pos] = x % 10
        x //= 10
        pos -= 1
print(ans)
````

## 4. Ý tưởng nền tảng / cách nghĩ trực giác

Nghĩ như viết các số **liền nhau** trên giấy rồi **bẻ** thành từng chữ số:

```
nums   : 13   25   83   77
viết ra: 1 3  2 5  8 3  7 7
answer : [1, 3, 2, 5, 8, 3, 7, 7]
```

Nên mỗi số cho ra một **nhóm** chữ số theo thứ tự trái → phải; ghép các nhóm theo thứ tự các số. Điểm duy nhất cần cẩn thận là **thứ tự bên trong nhóm** (`13` → `1, 3`, không phải `3, 1`) và việc ghép nhóm bằng `extend` (phẳng) thay vì `append` (lồng).

## 5. Cách 1 — Brute Force / baseline

Với mỗi số `x`: đổi thành chuỗi `str(x)`, mỗi ký tự đổi thành số, rồi `extend` vào kết quả. Gần như viết lại đề thành code.

Nút thắt: không có đáng kể — tổng số chữ số tối đa `6 * 1000 = 6000`. Mỗi chữ số được xử lý một lần nên đã là tuyến tính; các cách sau chỉ khác về cách tách và ghép.

````python
class SolutionString:
    def separateDigits(self, nums):
        answer = []
        for x in nums:
            for ch in str(x):
                answer.append(int(ch))
        return answer

print(SolutionString().separateDigits([13, 25, 83, 77]))
print(SolutionString().separateDigits([7, 1, 3, 9]))
````

**Độ phức tạp:** Time `O(T)` với `T` là tổng số chữ số của mọi phần tử (`T <= 6n`), Space `O(T)` cho kết quả.

## 6. Ý tưởng thuật toán / bước đột phá

Bài toán là "**làm phẳng**" (flatten) một danh sách các nhóm chữ số. Có hai hướng:

1. **Tách trái → phải** bằng chuỗi: thứ tự sẵn đúng, chỉ việc `extend`.
2. **Tách phải → trái** bằng `divmod`: nhanh và không dùng chuỗi, nhưng thứ tự ngược → hoặc **đảo từng nhóm**, hoặc **duyệt `nums` ngược rồi đảo cả kết quả một lần**, hoặc **điền vào mảng cấp phát trước từ cuối**.

**Quy tắc "nếu… thì…":**

- Nếu lấy chữ số bằng `str(x)` → thứ tự **đã đúng**, không đảo.
- Nếu lấy chữ số bằng `% 10` → thứ tự **ngược**, phải đảo nhóm (hoặc điền từ cuối).
- Nếu ghép nhóm → dùng `extend` (hoặc cộng list), không dùng `append(list)`.

Invariant: sau khi xử lý `k` số đầu, `answer` chứa chính xác các chữ số của `k` số đó, theo đúng thứ tự.

````python
nums = [13, 25, 83, 77]
flat = [int(ch) for x in nums for ch in str(x)]      # hai vòng for lồng nhau trong comprehension
print(flat)
print([[int(ch) for ch in str(x)] for x in nums])    # trước khi làm phẳng: list các nhóm
````

## 7. Các cách tiếp theo

Ta có 5 cách: (1) chuỗi + `append`; (2) `divmod` + đảo từng nhóm; (3) duyệt `nums` ngược + đảo một lần; (4) `"".join` rồi `map(int, ...)`; (5) mảng cấp phát trước điền từ cuối.

### Cách 2 — `divmod` + đảo từng nhóm

Với mỗi `x`, lấy chữ số cuối bằng `divmod`, đưa vào danh sách tạm `digits` (thứ tự ngược), rồi `extend(reversed(digits))` vào kết quả. Không dùng chuỗi.

````python
class SolutionDivmod:
    def separateDigits(self, nums):
        answer = []
        for x in nums:
            digits = []
            while x > 0:
                x, d = divmod(x, 10)
                digits.append(d)               # thứ tự phải -> trái
            answer.extend(reversed(digits))    # đảo nhóm về trái -> phải
        return answer

print(SolutionDivmod().separateDigits([13, 25, 83, 77]))
print(SolutionDivmod().separateDigits([7, 1, 3, 9]))
````

Độ phức tạp Cách 2: Time `O(T)`, Space `O(T)` (kết quả) + `O(D)` (nhóm tạm, `D <= 6`).

### Cách 3 — Duyệt `nums` ngược + đảo một lần

Nếu duyệt `nums` **từ cuối về đầu** và với mỗi số lấy chữ số **từ phải sang trái** rồi `append`, thì toàn bộ chữ số được tạo ra theo thứ tự **ngược hoàn toàn** so với đáp án. Chỉ cần `answer.reverse()` **một lần** ở cuối là đúng — tiết kiệm việc đảo từng nhóm.

````python
class SolutionReverseAll:
    def separateDigits(self, nums):
        answer = []
        for x in reversed(nums):               # từ số cuối về số đầu
            while x > 0:
                answer.append(x % 10)          # chữ số từ phải sang trái
                x //= 10
        answer.reverse()                       # đảo cả mảng một lần
        return answer

print(SolutionReverseAll().separateDigits([13, 25, 83, 77]))
print(SolutionReverseAll().separateDigits([7, 1, 3, 9]))
````

Độ phức tạp Cách 3: Time `O(T)`, Space `O(T)`.

### Cách 4 — Ghép chuỗi `"".join` rồi `map(int, ...)`

Nối các số thành một chuỗi duy nhất rồi đổi từng ký tự thành số: đúng bản chất "viết các số liền nhau rồi bẻ chữ số". Một dòng.

````python
class SolutionJoin:
    def separateDigits(self, nums):
        return list(map(int, "".join(map(str, nums))))

print(SolutionJoin().separateDigits([13, 25, 83, 77]))
print(SolutionJoin().separateDigits([7, 1, 3, 9]))
````

Độ phức tạp Cách 4: Time `O(T)`, Space `O(T)` (chuỗi nối + kết quả).

### Cách 5 — Cấp phát trước, điền từ cuối

Đếm tổng số chữ số `k`, tạo `answer = [0] * k` và điền từ vị trí `k - 1` ngược về `0`, duyệt `nums` ngược và lấy chữ số `% 10`. Không đảo và không `append`; hữu ích với ngôn ngữ mảng cố định (C/C++).

````python
class SolutionPrealloc:
    def separateDigits(self, nums):
        total = sum(len(str(x)) for x in nums)
        answer = [0] * total
        pos = total - 1
        for x in reversed(nums):
            while x > 0:
                answer[pos] = x % 10
                x //= 10
                pos -= 1
        return answer

print(SolutionPrealloc().separateDigits([13, 25, 83, 77]))
print(SolutionPrealloc().separateDigits([7, 1, 3, 9]))
````

Độ phức tạp Cách 5: Time `O(T)`, Space `O(T)` (chỉ mảng kết quả; không có nhóm tạm).

## 8. TRACE từng bước

Trace Cách 3 (duyệt ngược + đảo một lần) với `nums = [13, 25, 83, 77]`:

| số đang xử lý (từ cuối) | chữ số lấy được (phải → trái) | `answer` sau bước |
|---|---|---|
| 77 | 7, 7 | `[7, 7]` |
| 83 | 3, 8 | `[7, 7, 3, 8]` |
| 25 | 5, 2 | `[7, 7, 3, 8, 5, 2]` |
| 13 | 3, 1 | `[7, 7, 3, 8, 5, 2, 3, 1]` |

Đảo cả mảng: `[1, 3, 2, 5, 8, 3, 7, 7]` — đúng đáp án.

````python
def trace_reverse_all(nums):
    answer = []
    for x in reversed(nums):
        origin = x
        got = []
        while x > 0:
            got.append(x % 10)
            answer.append(x % 10)
            x //= 10
        print(f"số {origin:>6}: chữ số (phải -> trái) {got} | answer = {answer}")
    answer.reverse()
    print("sau khi đảo:", answer)
    return answer

trace_reverse_all([13, 25, 83, 77])
trace_reverse_all([100000, 7])
````

## 9. Code nộp LeetCode

````python
from typing import List

class Solution:
    def separateDigits(self, nums: List[int]) -> List[int]:
        answer = []
        for x in nums:
            for ch in str(x):
                answer.append(int(ch))
        return answer

print(Solution().separateDigits([13, 25, 83, 77]), Solution().separateDigits([7, 1, 3, 9]))
````

## 10. Vì sao thuật toán đúng

- **Thứ tự các số:** vòng lặp ngoài duyệt `nums` từ đầu đến cuối và `answer` chỉ được thêm vào cuối, nên chữ số của số đứng trước luôn nằm trước chữ số của số đứng sau.
- **Thứ tự trong một số:** `str(x)` liệt kê các chữ số từ chữ số cao nhất (trái) đến thấp nhất (phải), đúng thứ tự đề cần. Mỗi ký tự là một chữ số thập phân hợp lệ nên `int(ch)` cho chính chữ số đó.
- **Cách 2:** `divmod` lấy chữ số từ thấp lên cao; đảo nhóm đưa về thứ tự từ cao xuống thấp.
- **Cách 3:** nếu duyệt toàn bộ ở thứ tự **ngược hoàn toàn** (số cuối trước, chữ số thấp trước) thì dãy sinh ra đúng bằng đáp án **đảo ngược** (đảo ngược một chuỗi chia nhóm cũng đảo thứ tự nhóm lẫn thứ tự trong nhóm), nên đảo một lần cho ra đáp án.
- **Cách 5:** vị trí `pos` bắt đầu ở cuối và giảm một ở mỗi chữ số ghi, vì mỗi số được xử lý từ chữ số thấp (cuối) đến cao nên mỗi chữ số rơi đúng vị trí cuối cùng của nó; `pos` về `-1` khi hết (đã điền đúng `total` ô).
- **Không mất chữ số `0`:** vì `x >= 1`, `x % 10` và `// 10` xử lý đúng các chữ số `0` ở giữa/cuối (ví dụ `100000 → 1,0,0,0,0,0`); vòng `while x > 0` chỉ kết thúc khi đã tách hết chữ số cao nhất (khác `0`).

## 11. Phân tích Time / Space Complexity

Gọi `n = len(nums)` và `T` = tổng số chữ số của mọi phần tử; vì `nums[i] <= 10^5` nên mỗi số có tối đa 6 chữ số, `T <= 6n`.

| Cách | Time | Space |
|---|---|---|
| 1. Chuỗi + `append` | `O(T)` | `O(T)` |
| 2. `divmod` + đảo từng nhóm | `O(T)` | `O(T)` |
| 3. Duyệt ngược + đảo một lần | `O(T)` | `O(T)` |
| 4. `"".join` + `map(int)` | `O(T)` | `O(T)` |
| 5. Cấp phát trước, điền từ cuối | `O(T)` | `O(T)` |

Mọi cách phải tạo ra `T` phần tử đầu ra nên `O(T)` là tối ưu cả về thời gian lẫn bộ nhớ kết quả.

## 12. Bộ test

Test 1: ví dụ đề và các edge case (`n = 1`, số `1`, số `10^5` có nhiều chữ số `0`, số có chữ số `0` ở giữa, các số lặp).

````python
impls = [SolutionString(), SolutionDivmod(), SolutionReverseAll(), SolutionJoin(), SolutionPrealloc(), Solution()]
tests = [
    ([13, 25, 83, 77], [1, 3, 2, 5, 8, 3, 7, 7]),
    ([7, 1, 3, 9], [7, 1, 3, 9]),
    ([1], [1]),                                   # nhỏ nhất
    ([9], [9]),
    ([10], [1, 0]),                               # chữ số 0 ở cuối
    ([100000], [1, 0, 0, 0, 0, 0]),               # biên trên 10^5
    ([101], [1, 0, 1]),                           # chữ số 0 ở giữa
    ([99999], [9, 9, 9, 9, 9]),
    ([123, 45, 6], [1, 2, 3, 4, 5, 6]),           # thứ tự trong và giữa các nhóm
    ([21, 12], [2, 1, 1, 2]),                     # bắt lỗi quên đảo (ra 1,2,2,1)
    ([1000, 20, 3], [1, 0, 0, 0, 2, 0, 3]),
    ([5, 5, 5], [5, 5, 5]),
    ([12, 3, 456, 78910], [1, 2, 3, 4, 5, 6, 7, 8, 9, 1, 0]),
]
for nums, exp in tests:
    assert 1 <= len(nums) <= 1000 and all(1 <= v <= 10**5 for v in nums)
    for im in impls:
        assert im.separateDigits(list(nums)) == exp, (type(im).__name__, nums, exp)
print("tất cả test cố định OK:", len(tests))
````

Test 2: ngẫu nhiên có seed, đúng constraints (`1 <= n <= 1000`, `1 <= nums[i] <= 10^5`); oracle độc lập dùng phép nối chuỗi và so sánh từng ký tự.

````python
import random
random.seed(2553)

def oracle(nums):
    s = "".join(str(x) for x in nums)
    return [ord(c) - ord("0") for c in s]

for t in range(1500):
    n = random.randint(1, 1000) if t % 100 == 0 else random.randint(1, 15)
    hi_val = random.choice([9, 99, 1000, 10**5])
    nums = [random.randint(1, hi_val) for _ in range(n)]
    assert 1 <= len(nums) <= 1000 and all(1 <= v <= 10**5 for v in nums)
    exp = oracle(nums)
    for im in impls:
        assert im.separateDigits(list(nums)) == exp, (type(im).__name__, nums[:10])
print("1500 test ngẫu nhiên OK")
````

Test 3: tính chất — độ dài kết quả bằng tổng số chữ số; mọi phần tử là số nguyên trong `0..9`; ghép lại các chữ số cho đúng chuỗi nối các số; không sửa input; tích luỹ theo số: tách `nums[:k]` là tiền tố của tách `nums`.

## 13. Các lỗi người mới hay gặp

### Lỗi 1: quên đảo thứ tự chữ số khi dùng `% 10`

````python
def wrong_no_reverse(nums):
    answer = []
    for x in nums:
        while x > 0:
            answer.append(x % 10)       # SAI: chữ số lấy từ phải sang trái, chưa đảo nhóm
            x //= 10
    return answer

nums = [13, 25, 83, 77]
print("sai :", wrong_no_reverse(nums))
print("đúng:", Solution().separateDigits(nums))
assert wrong_no_reverse(nums) != Solution().separateDigits(nums)
````

Vì sao sai: `% 10` lấy **chữ số cuối trước**, nên `13` cho `3, 1` thay vì `1, 3`. **Sửa:** đảo từng nhóm (`reversed`), hoặc dùng `str(x)`, hoặc điền mảng từ cuối.

### Lỗi 2: `append` một list thay vì `extend` (mảng lồng nhau)

````python
def wrong_nested(nums):
    answer = []
    for x in nums:
        answer.append([int(ch) for ch in str(x)])     # SAI: append cả một list -> mảng lồng nhau
    return answer

nums = [13, 25, 83, 77]
print("sai :", wrong_nested(nums))
print("đúng:", Solution().separateDigits(nums))
assert wrong_nested(nums) != Solution().separateDigits(nums)
````

Vì sao sai: kết quả là `[[1, 3], [2, 5], ...]` — mỗi phần tử là một list. Đề yêu cầu mảng **phẳng**. **Sửa:** `answer.extend(digits)` hoặc `for d in digits: answer.append(d)`.

### Lỗi 3: quên đổi ký tự thành số (kết quả là chuỗi)

````python
def wrong_strings(nums):
    answer = []
    for x in nums:
        answer.extend(str(x))                # SAI: extend một chuỗi sinh ra các ký tự '1', '3', ...
    return answer

nums = [13, 25, 83, 77]
print("sai :", wrong_strings(nums))
print("đúng:", Solution().separateDigits(nums))
assert wrong_strings(nums) != Solution().separateDigits(nums)
````

Vì sao sai: `['1', '3', ...]` là list các **chuỗi**; `'1' != 1` nên LeetCode báo sai. **Sửa:** `int(ch)` cho từng ký tự (hoặc `map(int, str(x))`).

### Lỗi 4: duyệt `nums` xuôi nhưng đảo cả mảng ở cuối

````python
def wrong_reverse_all_forward(nums):
    answer = []
    for x in nums:                           # SAI: phải duyệt ngược nums nếu định đảo cả mảng ở cuối
        while x > 0:
            answer.append(x % 10)
            x //= 10
    answer.reverse()
    return answer

nums = [13, 25, 83, 77]
print("sai :", wrong_reverse_all_forward(nums))
print("đúng:", Solution().separateDigits(nums))
assert wrong_reverse_all_forward(nums) != Solution().separateDigits(nums)
````

Vì sao sai: đảo cả mảng cũng **đảo thứ tự các số** (`[7,7,8,3,2,5,...]`). Muốn đảo một lần ở cuối thì phải duyệt `nums` **từ cuối về đầu** (Cách 3) để việc đảo cuối cùng trả cả hai thứ tự về đúng.

## 14. Mẹo làm bài nhanh & nhận dạng pattern

- "Tách mỗi số thành các chữ số rồi ghép lại" → `[int(ch) for x in nums for ch in str(x)]` (hai vòng `for` trong comprehension, vòng ngoài viết trước).
- `str(x)` cho chữ số **trái → phải**; `% 10` cho **phải → trái** (phải đảo hoặc điền từ cuối).
- Làm phẳng: dùng `extend` hoặc comprehension hai tầng; `append(list)` tạo mảng lồng.
- Cẩn thận `list.reverse()` trả `None`; dùng `reversed(...)` hoặc `[::-1]` khi cần giá trị.
- Kiểm tra kiểu: kết quả phải là `int`, không phải `str`.
- Số có chữ số `0` (ví dụ `100000`) là test tốt để bắt lỗi tách chữ số (vòng `while x > 0` vẫn giữ các số `0` ở giữa/cuối).

## 15. Biến thể của bài

1. Tách chữ số rồi **đảo** thứ tự (đảo cả mảng kết quả) hoặc đảo từng số (`13 → 31`).
2. Tách theo **cơ số khác** (nhị phân, hệ 16): thay `10` bằng `b`.
3. Trả về **tổng chữ số** hoặc **tần suất các chữ số** (`Counter`) thay vì chính mảng.
4. Cho `nums[i]` có thể bằng `0` (số `0` cần trả `[0]`): vòng `while x > 0` sẽ không tạo ra chữ số nào — phải xử lý riêng.
5. Làm **ngược lại**: từ mảng chữ số ghép thành các số cho trước độ dài mỗi nhóm.

## 16. So sánh các cách làm

| Cách | Ý tưởng | Time | Space | Khi nào dùng |
|---|---|---|---|---|
| 1. Chuỗi + `append` | `str(x)` đúng thứ tự | `O(T)` | `O(T)` | **dễ nhất**, nên nộp |
| 2. `divmod` + đảo nhóm | tách số học, đảo từng nhóm | `O(T)` | `O(T)` | không dùng chuỗi |
| 3. Duyệt ngược + đảo một lần | đảo toàn mảng một lần | `O(T)` | `O(T)` | thủ thuật gọn, ít thao tác đảo |
| 4. `"".join` + `map(int)` | nối số thành một chuỗi | `O(T)` | `O(T)` | một dòng |
| 5. Cấp phát trước | điền từ cuối về đầu | `O(T)` | `O(T)` | ngôn ngữ mảng cố định (C/C++) |

## 18. Gợi ý 3 mức cho bài tập

**Mức 1 (nhẹ):** mỗi số → nhóm chữ số **trái → phải**; nối các nhóm theo thứ tự các số.

## 19. Đáp án tham khảo cho bài tập

````python
# Bài 1: 305 -> 3,0,5 ; 8 -> 8 ; 40 -> 4,0
assert Solution().separateDigits([305, 8, 40]) == [3, 0, 5, 8, 4, 0]

# Bài 2: [1234, 56]
assert Solution().separateDigits([1234, 56]) == [1, 2, 3, 4, 5, 6]
assert wrong_no_reverse([1234, 56]) == [4, 3, 2, 1, 6, 5]

# Bài 3
def digits_of(x):
    digits = []
    while x > 0:
        x, d = divmod(x, 10)
        digits.append(d)
    return digits[::-1]
assert digits_of(13) == [1, 3]
assert digits_of(100000) == [1, 0, 0, 0, 0, 0]
assert digits_of(7) == [7]

# Bài 4
def digit_frequency(nums):
    cnt = [0] * 10
    for d in Solution().separateDigits(nums):
        cnt[d] += 1
    return cnt
assert digit_frequency([13, 25, 83, 77]) == [0, 1, 1, 2, 0, 1, 0, 2, 1, 0]
assert sum(digit_frequency([100000])) == 6
assert digit_frequency([100000])[0] == 5

# Bài 5
def reverse_each_number(nums):
    return [int(str(x)[::-1]) for x in nums]
assert reverse_each_number([13, 100]) == [31, 1]
assert Solution().separateDigits(reverse_each_number([13, 100])) == [3, 1, 1]
print("đáp án bài tập OK")
````
