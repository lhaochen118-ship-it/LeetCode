## 4. Ý tưởng nền tảng / cách nghĩ trực giác (tính tay)

Ví dụ `s = "0101"` (2 bit `1`, 2 bit `0`, độ dài 4). Ta muốn `????` lẻ và lớn nhất.

1. **Lẻ** ⇒ bit cuối phải là `1`. Vậy dùng một bit `1` cho vị trí cuối: `???1`.
2. **Lớn nhất** ⇒ phần còn lại (`???`) phải lớn nhất có thể, từ các bit còn lại: một bit `1` và hai bit `0`. Muốn lớn, đặt `1` ở vị trí cao nhất: `100`.
3. Ghép: `100` + `1` = `"1001"`.

Quy tắc chung: với `c` bit `1` và `z` bit `0`, dành **một** bit `1` cho cuối; `c - 1` bit `1` còn lại dồn **hết lên đầu**, rồi tới `z` bit `0`:

`(c-1 số 1) + (z số 0) + "1"`.

Ví dụ `s = "010"`: `c = 1`, `z = 2` → `"" + "00" + "1" = "001"`.

## 5. Cách 1 — Brute Force / baseline: thử mọi cách sắp xếp

Ý tưởng "vét cạn": tạo **mọi** chuỗi nhị phân có cùng số bit `1` như `s`; giữ lại những chuỗi kết thúc bằng `1` (lẻ); lấy chuỗi lớn nhất. Cách tạo gọn: chọn `c` vị trí đặt bit `1` trong `n` vị trí (`combinations`).

(Cách "ngây thơ" hơn là `set(permutations(s))`, nhưng sinh rất nhiều hoán vị trùng. Dùng `combinations` thì mỗi chuỗi chỉ sinh một lần.)

````python
from itertools import combinations, permutations

def max_odd_bruteforce(s: str) -> str:
    n, c = len(s), s.count("1")
    best = None
    for pos in combinations(range(n), c):       # các vị trí đặt bit 1
        t = ["0"] * n
        for p in pos:
            t[p] = "1"
        cand = "".join(t)
        if cand[-1] == "1" and (best is None or cand > best):
            best = cand
    return best

print(max_odd_bruteforce("010"))       # 001
print(max_odd_bruteforce("0101"))      # 1001

def max_odd_permutations(s: str) -> str:
    """Bản ngây thơ nhất: mọi hoán vị (chỉ dùng cho chuỗi rất ngắn)."""
    cands = {"".join(p) for p in permutations(s)}
    return max(t for t in cands if t[-1] == "1")

print(max_odd_permutations("0101"))    # 1001
````

**Độ phức tạp:** có `C(n, c)` chuỗi, mỗi chuỗi dựng/so sánh `O(n)`. Tổng **Time `O(n · C(n, c))`**, tức tới cỡ `2ⁿ` — với `n = 100` là không thể chạy được (`C(100, 50) ≈ 10²⁹`). **Space `O(n)`** cho chuỗi tạm.

**Nút thắt:** ta thử cả những chuỗi chắc chắn không phải đáp án. Ta cần một quy tắc "cứ làm thẳng như vậy là ra".

## 6. Ý tưởng thuật toán / bước đột phá

Bước đột phá gồm hai mệnh đề đơn giản (tham lam - greedy):

1. **Bit cuối bắt buộc là `1`.** Bỏ nó ra, ta cần xếp `n - 1` bit còn lại (`c - 1` số `1`, `z` số `0`) để được số **lớn nhất**.
2. **Xếp lớn nhất = bit `1` dồn hết sang trái.** Nếu có một cặp `0` đứng trước `1` (`…0…1…`), đổi chỗ chúng thì bit cao hơn tăng từ 0 lên 1, số **tăng ngặt**. Vì vậy ở bố cục tối ưu không còn cặp "0 trước 1": tất cả `1` đứng trước tất cả `0`.

Đáp án chỉ phụ thuộc **hai con số**: số bit `1` (`c`) và số bit `0` (`z`). Không cần sắp xếp thật sự.

Đáp án: `"1" * (c - 1) + "0" * z + "1"`.

## 7. Các cách tiếp theo

### Cách 2 — Sắp xếp giảm dần rồi đổi chỗ

**Giải thích từng đoạn.**

1. `t = sorted(s, reverse=True)` đưa mọi `1` lên đầu, ví dụ `"0101"` → `1100`. Đây là số **lớn nhất** nhưng chưa chắc lẻ.
2. Bit `1` cuối cùng của nhóm số `1` nằm ở chỉ số `c - 1`. Đổi nó với ký tự cuối `t[-1]`: bit `1` đó xuống cuối, ký tự `0` (nếu có) lên thế chỗ: `1100` → `1001`.
3. Nếu `s` toàn là `1` thì `t[c-1]` chính là `t[-1]`, đổi chỗ không thay đổi gì, đúng.

````python
def max_odd_sort(s: str) -> str:
    t = sorted(s, reverse=True)
    c = s.count("1")
    t[c - 1], t[-1] = t[-1], t[c - 1]
    return "".join(t)

print(max_odd_sort("010"))      # 001
print(max_odd_sort("0101"))     # 1001
print(max_odd_sort("111"))      # 111
````

````python
# Code nộp LeetCode cho Cách 2
class SolutionSort:
    def maximumOddBinaryNumber(self, s: str) -> str:
        t = sorted(s, reverse=True)
        c = s.count("1")
        t[c - 1], t[-1] = t[-1], t[c - 1]
        return "".join(t)

print(SolutionSort().maximumOddBinaryNumber("0101"))   # 1001
````

**Độ phức tạp Cách 2:** sắp xếp `O(n log n)`; Space `O(n)` cho danh sách ký tự. (Sắp xếp là thừa vì chỉ có hai loại ký tự; Cách 3 bỏ nó.)

---

### Cách 3 — Đếm rồi nối chuỗi (chính thức, `O(n)`)

**Giải thích.** `c = s.count("1")` và `z = len(s) - c`. Đáp án là `"1" * (c - 1) + "0" * z + "1"`. Không cần sort, không cần list.

````python
def max_odd_count(s: str) -> str:
    c = s.count("1")
    z = len(s) - c
    return "1" * (c - 1) + "0" * z + "1"

print(max_odd_count("010"))     # 001
print(max_odd_count("0101"))    # 1001
print(max_odd_count("1"))       # 1
print(max_odd_count("111"))     # 111
````

````python
# Code nộp LeetCode cho Cách 3
class SolutionCount:
    def maximumOddBinaryNumber(self, s: str) -> str:
        ones = s.count("1")
        zeros = len(s) - ones
        return "1" * (ones - 1) + "0" * zeros + "1"

print(SolutionCount().maximumOddBinaryNumber("0101"))   # 1001
````

**Độ phức tạp Cách 3:** `count` `O(n)`, dựng chuỗi `O(n)`. **Time `O(n)`**, **Space `O(n)`** (chuỗi kết quả bắt buộc phải có).

---

### Cách 4 — Hai con trỏ tại chỗ (partition)

**Giải thích.** Chuyển `s` thành list `a`. Dùng con trỏ `left` là "ô tiếp theo sẽ nhận bit 1". Duyệt `i` từ trái sang phải: gặp `'1'` thì đổi `a[i]` với `a[left]` rồi `left += 1`. Kết quả: mọi `1` ở đầu, mọi `0` ở sau (chỉ dùng hoán đổi, không sort). Cuối cùng đổi `a[left - 1]` (bit `1` cuối nhóm) với `a[-1]`.

````python
def max_odd_two_pointers(s: str) -> str:
    a = list(s)
    left = 0
    for i in range(len(a)):
        if a[i] == "1":
            a[i], a[left] = a[left], a[i]
            left += 1
    a[left - 1], a[-1] = a[-1], a[left - 1]
    return "".join(a)

print(max_odd_two_pointers("0101"))   # 1001
print(max_odd_two_pointers("010"))    # 001
print(max_odd_two_pointers("0011"))   # 1001
````

````python
# Code nộp LeetCode cho Cách 4
class SolutionTwoPointers:
    def maximumOddBinaryNumber(self, s: str) -> str:
        a = list(s)
        left = 0
        for i in range(len(a)):
            if a[i] == "1":
                a[i], a[left] = a[left], a[i]
                left += 1
        a[left - 1], a[-1] = a[-1], a[left - 1]
        return "".join(a)

print(SolutionTwoPointers().maximumOddBinaryNumber("0101"))   # 1001
````

**Độ phức tạp Cách 4:** một lượt duyệt: **Time `O(n)`**, **Space `O(n)`** cho list (không tính kết quả thì phụ trợ `O(1)` nếu thao tác trên mảng cho sẵn).

### Cách 5 — Dựng số nguyên bằng phép dịch bit

**Giải thích.** Gọi `c` số `1`, `z` số `0`. `c - 1` bit `1` đứng cao nhất, nối tiếp `z + 1` bit thấp là `0…01`:

`value = ((1 << (c - 1)) - 1) << (z + 1) | 1`.

- `(1 << (c-1)) - 1` là số có `c - 1` bit `1` (vd `c - 1 = 2` → `0b11`).
- dịch trái `z + 1` bit chừa chỗ cho `z` số `0` và bit cuối.
- `| 1` đặt bit cuối là `1`.

Cuối cùng `format(value, "b").zfill(n)` đổi lại chuỗi đúng độ dài `n` (bổ sung số `0` đứng đầu nếu `c = 1`).

````python
def max_odd_bits(s: str) -> str:
    n = len(s)
    c = s.count("1")
    z = n - c
    value = (((1 << (c - 1)) - 1) << (z + 1)) | 1
    return format(value, "b").zfill(n)

print(max_odd_bits("0101"))    # 1001
print(max_odd_bits("010"))     # 001
print(max_odd_bits("1"))       # 1
````

````python
# Code nộp LeetCode cho Cách 5
class SolutionBits:
    def maximumOddBinaryNumber(self, s: str) -> str:
        n = len(s)
        c = s.count("1")
        z = n - c
        value = (((1 << (c - 1)) - 1) << (z + 1)) | 1
        return format(value, "b").zfill(n)

print(SolutionBits().maximumOddBinaryNumber("0101"))   # 1001
````

**Độ phức tạp Cách 5:** số nguyên có `n` bit nên phép dịch / định dạng `O(n)`. **Time `O(n)`**, **Space `O(n)`**. Cách này chủ yếu để luyện bit; Cách 3 đơn giản hơn.

## 8. TRACE từng bước

Trace Cách 3 trên `s = "0101"`, và Cách 4 trên `s = "0011"`:

| bước | đại lượng | giá trị |
|---|---|---|
| 1 | `c = s.count("1")` | 2 |
| 2 | `z = n - c` | 2 |
| 3 | phần đầu `"1" * (c-1)` | `"1"` |
| 4 | phần giữa `"0" * z` | `"00"` |
| 5 | bit cuối | `"1"` |
| 6 | ghép | `"1001"` |

````python
def trace_count(s):
    c = s.count("1")
    z = len(s) - c
    head, mid, tail = "1" * (c - 1), "0" * z, "1"
    print(f"s={s!r}: c={c}, z={z}")
    print(f"  phần đầu  = '1' * {c - 1} = {head!r}")
    print(f"  phần giữa = '0' * {z} = {mid!r}")
    print(f"  bit cuối  = {tail!r}")
    ans = head + mid + tail
    print("  đáp án =", repr(ans), "| giá trị thập phân =", int(ans, 2))
    return ans

trace_count("0101")
trace_count("010")
trace_count("1")
````

````python
def trace_two_pointers(s):
    a = list(s)
    left = 0
    print("bắt đầu:", a)
    for i in range(len(a)):
        if a[i] == "1":
            a[i], a[left] = a[left], a[i]
            left += 1
            print(f"i={i}: gặp '1', đổi với ô {left - 1} -> {a}, left={left}")
        else:
            print(f"i={i}: gặp '0', bỏ qua -> {a}, left={left}")
    a[left - 1], a[-1] = a[-1], a[left - 1]
    print("đổi bit 1 cuối nhóm xuống cuối:", a)
    return "".join(a)

print("kết quả:", trace_two_pointers("0011"))
````

## 9. Code nộp LeetCode

````python
class Solution:
    def maximumOddBinaryNumber(self, s: str) -> str:
        ones = s.count("1")
        zeros = len(s) - ones
        return "1" * (ones - 1) + "0" * zeros + "1"

print(Solution().maximumOddBinaryNumber("010"))     # 001
print(Solution().maximumOddBinaryNumber("0101"))    # 1001
````

## 10. Vì sao thuật toán đúng

Gọi `c ≥ 1` là số bit `1`, `z` là số bit `0`.

- **Hợp lệ (lẻ):** chuỗi kết thúc bằng `"1"`; bit cuối `1` ⇒ số lẻ. Đáp án dùng đúng `c - 1 + 1 = c` bit `1` và `z` bit `0`, tức là hoán vị của `s`.
- **Tối ưu:** mọi đáp án lẻ phải có bit cuối `1`. Với bit cuối cố định là `1`, ta xếp `c - 1` bit `1` và `z` bit `0` vào `n - 1` vị trí đầu. Giả sử một phương án `P` (cũng lẻ, cũng là hoán vị của `s`) lớn hơn đáp án `A` của ta. Xét vị trí đầu tiên (từ trái) mà `P` và `A` khác nhau. Vì `P > A` nên tại đó `P` có `1`, còn `A` có `0`. Nhưng trong `A`, các bit `0` chỉ xuất hiện sau khi **mọi** bit `1` của phần đầu đã được đặt (và bit cuối `1` thì ở cuối cùng); nên `A` có `0` tại vị trí đó nghĩa là `P` đã dùng hết số bit `1` dành cho phần đầu ở các vị trí trước đó (vì hai chuỗi giống nhau trước vị trí đó) và bit `1` còn lại duy nhất là bit cuối. Do vị trí này không phải vị trí cuối nên `P` không còn bit `1` để đặt ở đây, mâu thuẫn. Vậy không có phương án lớn hơn `A`.
- **Trường hợp `c = 1`:** không còn bit `1` nào dư, phần đầu là `"" ` nên đáp án là toàn `0` rồi `1`.
- **Trường hợp `z = 0`:** toàn bit `1`, đáp án `"1" * (c - 1) + "" + "1"` chính là `s`.

## 11. Phân tích Time / Space Complexity

Gọi `n = len(s)`.

| Cách | Time | Space |
|---|---|---|
| 1. Vét cạn | `O(n · C(n, c))` ~ `O(n · 2ⁿ)` | `O(n)` |
| 2. Sắp xếp + đổi chỗ | `O(n log n)` | `O(n)` |
| 3. Đếm + nối chuỗi | `O(n)` | `O(n)` |
| 4. Hai con trỏ | `O(n)` | `O(n)` |
| 5. Dựng bằng dịch bit | `O(n)` | `O(n)` |

`Space O(n)` ở đây chủ yếu là chuỗi kết quả. Với `n ≤ 100`, mọi cách trừ vét cạn đều chạy tức thì.

### 12.1. Ví dụ của đề và edge case

### 12.2. Vét cạn toàn bộ chuỗi có độ dài ≤ 10

Có `2¹ + … + 2¹⁰ - 10` chuỗi hợp lệ (chứa ít nhất một `1`). Ta kiểm **tất cả**, so với brute force.

````python
from itertools import product
total = 0
for n in range(1, 11):
    for bits in product("01", repeat=n):
        s = "".join(bits)
        if "1" not in s:
            continue
        exp = max_odd_bruteforce(s)
        for sol in sols:
            got = sol.maximumOddBinaryNumber(s)
            assert got == exp, (s, type(sol).__name__, got, exp)
        assert len(exp) == len(s) and sorted(exp) == sorted(s) and exp[-1] == "1"
        total += 1
print("đã kiểm", total, "chuỗi: mọi cách khớp vét cạn")
````

### 12.3. Kiểm tính chất và chuỗi ngẫu nhiên có seed

Với chuỗi dài tới 100 không vét cạn được, nên kiểm các **tính chất** đáp án phải thỏa, và đối chiếu các cách với nhau.

````python
import random
random.seed(2864)
for t in range(500):
    n = random.randint(1, 100)
    s = "".join(random.choice("01") for _ in range(n))
    if "1" not in s:
        s = s[:-1] + "1"                       # đảm bảo có ít nhất một '1'
    ans = Solution().maximumOddBinaryNumber(s)
    assert len(ans) == len(s) and sorted(ans) == sorted(s)      # là hoán vị của s
    assert ans[-1] == "1"                                       # lẻ
    for sol in sols:
        assert sol.maximumOddBinaryNumber(s) == ans, (s, type(sol).__name__)
    # không có hoán vị lẻ nào lớn hơn: thử đổi chỗ hai ký tự bất kỳ
    for _ in range(20):
        i, j = random.randrange(n), random.randrange(n)
        u = list(ans); u[i], u[j] = u[j], u[i]
        u = "".join(u)
        if u[-1] == "1":
            assert u <= ans, (s, ans, u)
print("500 chuỗi ngẫu nhiên: các tính chất đều đúng")
````

## 13. Các lỗi người mới hay gặp

### Lỗi 1: Chỉ sắp xếp giảm dần (quên điều kiện lẻ)

````python
def sai_chi_sort(s):
    return "".join(sorted(s, reverse=True))      # sai: "1100" là số chẵn

print("sai :", sai_chi_sort("0101"))             # 1100
print("đúng:", Solution().maximumOddBinaryNumber("0101"))   # 1001
````

**Vì sao sai:** số lớn nhất chưa chắc lẻ. Bit cuối phải là `1`.

### Lỗi 2: Đặt cả `c` số 1 ở đầu rồi thêm "1" ở cuối

````python
def sai_thua_mot_so_1(s):
    c = s.count("1")
    z = len(s) - c
    return "1" * c + "0" * z + "1"               # sai: dư một bit 1, độ dài n + 1

print("sai :", sai_thua_mot_so_1("0101"), len(sai_thua_mot_so_1("0101")))   # 11001, dài 5
print("đúng:", Solution().maximumOddBinaryNumber("0101"))
````

**Vì sao sai:** bit `1` ở cuối **là một trong** các bit `1` có sẵn, không phải bit mới. Phần đầu chỉ có `c - 1` bit `1`.

### Lỗi 3: Đặt bit `1` bắt buộc ở đầu thay vì ở cuối

````python
def sai_dat_1_o_dau(s):
    c = s.count("1")
    z = len(s) - c
    return "1" + "0" * z + "1" * (c - 1)         # sai: bit cuối có thể là 0 hoặc số nhỏ

print("sai :", sai_dat_1_o_dau("0101"))          # 1001 (trùng ngẫu nhiên)
print("sai :", sai_dat_1_o_dau("01011"))         # 10011 (nhưng đúng phải là 11001)
print("đúng:", Solution().maximumOddBinaryNumber("01011"))
````

**Vì sao sai:** bit cao nhất nên là `1` **nhiều nhất có thể** (các `1` thừa xếp sát nhau ở đầu), còn bit cuối chỉ cần đúng một `1`. Ví dụ với 3 số `1`, đáp án tối ưu `1 1 0 0 1` (hai `1` ở đầu), không phải `1 0 0 1 1`.

### Lỗi 4: Trả về số nguyên làm mất số 0 đứng đầu

````python
def sai_tra_int(s):
    ans = Solution().maximumOddBinaryNumber(s)
    return int(ans, 2)                           # sai kiểu: đề yêu cầu chuỗi

print("sai :", sai_tra_int("010"), type(sai_tra_int("010")))   # 1 <class 'int'>
print("đúng:", repr(Solution().maximumOddBinaryNumber("010"))) # '001'
````

**Vì sao sai:** đề yêu cầu chuỗi (được phép có `0` đứng đầu). `1` và `"001"` khác nhau về kiểu và về độ dài.

## 14. Mẹo làm bài nhanh & nhận dạng pattern

- "**Số nhị phân lẻ**" luôn nghĩ tới **bit cuối = 1**.
- "**Xáo trộn / sắp xếp lại** để được số lớn nhất" thường là **tham lam**: đặt bit lớn ở vị trí cao.
- Khi chỉ có **hai loại ký tự**, đừng sort: **đếm** số lượng rồi dựng đáp án (`O(n)`).
- Cách viết ngắn: `return "1" * (s.count("1") - 1) + "0" * s.count("0") + "1"`.
- Nhớ phép kiểm nhanh: độ dài đáp án = độ dài `s`, số bit `1` bằng nhau, ký tự cuối là `1`.

## 15. Biến thể của bài

1. **Số nhị phân chẵn lớn nhất:** bit cuối phải là `0` (nếu có `0`); `1` còn lại dồn đầu.
2. **Số lẻ nhỏ nhất:** bit cuối `1`; các bit còn lại xếp `0` trước `1` sau cùng (gần cuối).
3. **Số lớn nhất chia hết cho 3 / 5 …** (khó hơn, cần xét dư).
4. **Hoán vị chữ số thập phân** để được số lớn nhất: LeetCode 179 *Largest Number* (sắp xếp với bộ so sánh riêng), LeetCode 2231 *Largest Number After Digit Swaps by Parity*.
5. **Sắp xếp nhị phân theo số bit 1**: LeetCode 1356 *Sort Integers by The Number of 1 Bits*.

## 16. So sánh các cách làm

| Cách | Ý tưởng | Time | Space | Khi nào dùng |
|---|---|---|---|---|
| 1. Vét cạn | thử mọi cách đặt bit 1 | `O(n · 2ⁿ)` | `O(n)` | kiểm chứng chuỗi rất ngắn |
| 2. Sort + đổi chỗ | sắp giảm dần rồi đưa một `1` xuống cuối | `O(n log n)` | `O(n)` | khi quen sort, ít dòng |
| 3. Đếm + nối | `"1"*(c-1) + "0"*z + "1"` | `O(n)` | `O(n)` | **cách nộp bài** |
| 4. Hai con trỏ | partition tại chỗ | `O(n)` | `O(n)` | luyện kỹ thuật con trỏ |
| 5. Dịch bit | dựng số nguyên rồi `format` | `O(n)` | `O(n)` | luyện bit, ít gặp trong thực tế |
