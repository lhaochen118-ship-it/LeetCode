## 1. Mục tiêu học

Sau notebook này bạn sẽ:

1. Hiểu bài toán **tính tổng các số trong `[1, n]` chia hết cho `3`, `5` hoặc `7`** — mỗi số chỉ được cộng **một lần** dù chia hết cho nhiều số cùng lúc (ví dụ `15` chia hết cho cả `3` và `5`).

4. Hiểu **nguyên lý bao hàm – loại trừ (inclusion–exclusion)** để tính tổng trong `O(1)`: cộng các tập đơn, trừ các giao đôi, cộng giao ba.

6. Tránh các lỗi: cộng trùng số chia hết cho nhiều ước (`15`, `21`, `35`, `105`), dùng `range(1, n)` thiếu `n`, dùng `/` thay vì `//`, và dùng `and` thay `or`.

## 2. Đề bài & ví dụ

Cho số nguyên dương `n`. Hãy tìm **tổng của mọi số nguyên trong đoạn `[1, n]` (gồm cả hai đầu) chia hết cho `3`, `5` hoặc `7`**. Trả về tổng đó.

| Ví dụ | n | Output | Giải thích |
|---|---|---|---|
| 1 | `7` | `21` | Các số chia hết cho 3, 5 hoặc 7 trong `[1,7]`: `3, 5, 6, 7` ⇒ `21`. |

| 3 | `9` | `30` | `3, 5, 6, 7, 9` ⇒ `30`. |

Constraints: `1 <= n <= 10^3`.

Lưu ý: "hoặc" ở đây là **ít nhất một** trong ba điều kiện; số chia hết cho nhiều ước chỉ tính **một lần**.

````python
n = 10
chon = [x for x in range(1, n + 1) if x % 3 == 0 or x % 5 == 0 or x % 7 == 0]
print(chon, "-> tổng =", sum(chon))
````

## 3. Kiến thức nền

### 3.1. Chia hết và toán tử `or`

`x % 3 == 0` nghĩa là `x` chia hết cho `3`. `A or B or C` đúng khi **ít nhất một** trong A, B, C đúng. `and` đòi **tất cả** cùng đúng — đây là chỗ hay nhầm.

````python
for x in [3, 5, 6, 7, 8, 15]:
    print(x, "chia hết 3?", x % 3 == 0, "| 5?", x % 5 == 0, "| 7?", x % 7 == 0, "| or:", x % 3 == 0 or x % 5 == 0 or x % 7 == 0)
````

### 3.2. `range` với bước nhảy

`range(a, b, step)` đi từ `a` tới trước `b`, mỗi lần tăng `step`. Ví dụ `range(3, 11, 3)` cho `3, 6, 9` (**không gồm `b`**). Để lấy các bội của `3` trong `[1, n]` ta dùng `range(3, n + 1, 3)` — nhớ `n + 1` vì cận trên bị loại.

````python
n = 10
print(list(range(3, n + 1, 3)))    # [3, 6, 9]
print(list(range(5, n + 1, 5)))    # [5, 10]
print(list(range(7, n + 1, 7)))    # [7]
print(list(range(3, n, 3)))        # [3, 6, 9] (nhưng nếu n = 9 thì range(3, 9, 3) thiếu số 9!)
print(list(range(3, 9, 3)), list(range(3, 9 + 1, 3)))
````

### 3.3. `set` và phép hợp `|`

`set` là tập hợp không trùng. `A | B` là **hợp** (phần tử thuộc A hoặc B, mỗi phần tử một lần). Nhờ đó số `15` (xuất hiện ở cả bội của 3 lẫn bội của 5) chỉ được tính một lần.

````python
a = set(range(3, 16, 3))      # {3, 6, 9, 12, 15}
b = set(range(5, 16, 5))      # {5, 10, 15}
print(a | b)                  # 15 chỉ xuất hiện một lần
print(sum(a) + sum(b), "(cộng trùng 15)")
print(sum(a | b), "(đúng)")
````

### 3.4. Tổng cấp số cộng: `1 + 2 + ... + m`

Tổng `1 + 2 + ... + m = m(m + 1) / 2` (Gauss). Ví dụ `1 + 2 + 3 + 4 = 10 = 4 × 5 / 2`. Mẹo chứng minh: ghép đầu với cuối, `(1 + 4) + (2 + 3) = 5 + 5`, có `m/2` cặp, mỗi cặp tổng `m + 1`.

Với bội của `k`: `k + 2k + ... + mk = k(1 + 2 + ... + m) = k · m(m + 1) / 2`, trong đó `m = n // k` là **số bội** của `k` trong `[1, n]`.

````python
m = 4
print(sum(range(1, m + 1)), m * (m + 1) // 2)       # 10 10
n, k = 10, 3
m = n // k                                           # 3 bội: 3, 6, 9
print(sum(range(k, n + 1, k)), k * m * (m + 1) // 2) # 18 18
````

### 3.5. `//` (chia nguyên) — dùng cho đếm số bội

`n // k` cho **số bội của `k` trong `[1, n]`**. `10 // 3 = 3` (có `3, 6, 9`). Còn `/` cho số thực (`10 / 3 = 3.33...`), không dùng cho đếm. Khi tính `k * m * (m + 1) / 2` hãy dùng `//` ở cuối: tích `m(m+1)` luôn chẵn nên chia 2 chính xác.

````python
print(10 // 3, 10 / 3)                 # 3 3.3333333333333335
m = 3
print(m * (m + 1) // 2, m * (m + 1) / 2)   # 6 6.0 : // cho số nguyên, / cho số thực
````

### 3.6. Bội chung nhỏ nhất (LCM)

Số chia hết **đồng thời** cho `3` và `5` là các bội của `lcm(3, 5) = 15`. Tương tự `lcm(3, 7) = 21`, `lcm(5, 7) = 35`, `lcm(3, 5, 7) = 105`. Đó là các "giao" mà ta sẽ trừ đi ở bao hàm – loại trừ. `math.lcm` (Python 3.9+) tính LCM của nhiều số.

````python
import math
print(math.lcm(3, 5), math.lcm(3, 7), math.lcm(5, 7), math.lcm(3, 5, 7))   # 15 21 35 105
````

## 4. Ý tưởng nền tảng / cách nghĩ trực giác

Cách tự nhiên nhất: **đi qua từng số** từ `1` đến `n`, số nào chia hết cho `3` hoặc `5` hoặc `7` thì cộng vào tổng.

Tính tay `n = 10`:

- `1, 2`: không. `3`: có (chia hết 3) ⇒ tổng `3`.
- `4`: không. `5`: có ⇒ `8`. `6`: có ⇒ `14`. `7`: có ⇒ `21`.
- `8`: không. `9`: có ⇒ `30`. `10`: có ⇒ `40`.

Kết quả `40`, đúng với ví dụ 2.

Điểm cần cẩn thận: nếu thay vì "đi qua từng số" ta cộng **riêng** các bội của `3`, của `5`, của `7`, thì số như `15` (bội của cả 3 và 5) bị **đếm hai lần**.

````python
n = 15
sum3 = sum(range(3, n + 1, 3))
sum5 = sum(range(5, n + 1, 5))
sum7 = sum(range(7, n + 1, 7))
print("cộng riêng (SAI):", sum3 + sum5 + sum7)
print("đáp án thật     :", sum(x for x in range(1, n + 1) if x % 3 == 0 or x % 5 == 0 or x % 7 == 0))
````

## 5. Cách 1 — Brute Force / baseline: duyệt từng số và kiểm tra

**Ý tưởng:** biến `total = 0`; với `x` từ `1` đến `n`, nếu `x % 3 == 0 or x % 5 == 0 or x % 7 == 0` thì `total += x`. Mỗi `x` được xét một lần nên không bị trùng.

````python
class SolutionLoop:
    def sumOfMultiples(self, n: int) -> int:
        total = 0
        for x in range(1, n + 1):          # n + 1 để gồm cả n
            if x % 3 == 0 or x % 5 == 0 or x % 7 == 0:
                total += x
        return total
````

````python
sol = SolutionLoop()
print(sol.sumOfMultiples(7))     # 21
print(sol.sumOfMultiples(10))    # 40
print(sol.sumOfMultiples(9))     # 30
print(sol.sumOfMultiples(1))     # 0
````

**Độ phức tạp:** Time `O(n)` (mỗi số một vài phép `%`); Space `O(1)`.

**Nút thắt:** với `n <= 10^3` quá nhanh. Nhưng nếu `n` tới `10^18`, vòng lặp không thể chạy. Công thức đóng `O(1)` sẽ giải quyết — đó cũng là bài học toán học đáng nhớ.

## 6. Ý tưởng thuật toán / bước đột phá

**Bước đột phá — đếm theo "tập" thay vì theo "số".** Gọi `A₃, A₅, A₇` là tập các số trong `[1, n]` chia hết cho `3`, `5`, `7`. Ta cần tổng các phần tử của `A₃ ∪ A₅ ∪ A₇`.

**Nguyên lý bao hàm – loại trừ:**

`|A ∪ B ∪ C| = |A| + |B| + |C| − |A∩B| − |A∩C| − |B∩C| + |A∩B∩C|`

Áp dụng cho **tổng** (thay `|·|` bằng tổng các phần tử):

`ans = S(3) + S(5) + S(7) − S(15) − S(21) − S(35) + S(105)`

trong đó `S(k)` là tổng các bội của `k` trong `[1, n]`. Vì giao của "chia hết cho `a`" và "chia hết cho `b`" chính là "chia hết cho `lcm(a, b)`" (và `3, 5, 7` đôi một nguyên tố cùng nhau nên `lcm` bằng tích).

Còn `S(k)` tính bằng cấp số cộng: `S(k) = k · m(m + 1) / 2` với `m = n // k`. Toàn bộ cách làm là `O(1)`.

````python
def S(k, n):
    m = n // k
    return k * m * (m + 1) // 2

n = 10
print("S(3) =", S(3, n), "S(5) =", S(5, n), "S(7) =", S(7, n))
print("S(15) =", S(15, n), "S(21) =", S(21, n), "S(35) =", S(35, n), "S(105) =", S(105, n))
print(S(3, n) + S(5, n) + S(7, n) - S(15, n) - S(21, n) - S(35, n) + S(105, n))   # 40
````

## 7. Các cách tiếp theo

### Cách 2 — `sum` với biểu thức sinh có điều kiện

Cùng ý tưởng Cách 1 nhưng gói trong một dòng: `sum(x for x in range(1, n + 1) if ...)`. Biểu thức sinh (generator) không tạo danh sách trung gian.

````python
class SolutionSumGen:
    def sumOfMultiples(self, n: int) -> int:
        return sum(x for x in range(1, n + 1) if x % 3 == 0 or x % 5 == 0 or x % 7 == 0)

print(SolutionSumGen().sumOfMultiples(7))      # 21
print(SolutionSumGen().sumOfMultiples(10))     # 40
````

**Độ phức tạp:** Time `O(n)`, Space `O(1)`.

### Cách 3 — Hợp của ba `range` bằng `set`

Lấy `set(range(3, n + 1, 3)) | set(range(5, n + 1, 5)) | set(range(7, n + 1, 7))` rồi `sum`. `set` tự loại trùng, ta không phải tự nghĩ tới `15`, `21`, `35`, `105`.

````python
class SolutionSet:
    def sumOfMultiples(self, n: int) -> int:
        multiples = set(range(3, n + 1, 3)) | set(range(5, n + 1, 5)) | set(range(7, n + 1, 7))
        return sum(multiples)

print(SolutionSet().sumOfMultiples(7))      # 21
print(SolutionSet().sumOfMultiples(10))     # 40
print(SolutionSet().sumOfMultiples(9))      # 30
````

**Độ phức tạp:** tạo ba `set` tổng kích thước `≈ n(1/3 + 1/5 + 1/7)` ⇒ Time `O(n)`, Space `O(n)`.

### Cách 4 — Bao hàm – loại trừ cố định bảy số hạng, `O(1)`

Dùng công thức ở mục 6. Mỗi `S(k)` có `m = n // k` và cấp số cộng `k · m(m + 1) // 2`.

````python
class SolutionFormula:
    def sumOfMultiples(self, n: int) -> int:
        def S(k: int) -> int:
            m = n // k
            return k * m * (m + 1) // 2
        return S(3) + S(5) + S(7) - S(15) - S(21) - S(35) + S(105)

print(SolutionFormula().sumOfMultiples(7))      # 21
print(SolutionFormula().sumOfMultiples(10))     # 40
print(SolutionFormula().sumOfMultiples(9))      # 30
print(SolutionFormula().sumOfMultiples(10**18)) # chạy tức thì với n khổng lồ
````

**Độ phức tạp:** Time `O(1)` (bảy phép tính), Space `O(1)`. Python có số nguyên lớn tuỳ ý nên `n = 10^18` vẫn đúng.

### Cách 5 — Bao hàm – loại trừ tổng quát (mọi danh sách ước)

Với danh sách ước `ds`, xét mọi tập con khác rỗng: nếu tập con có kích thước **lẻ** thì **cộng** `S(lcm)`, **chẵn** thì **trừ**. Dùng `itertools.combinations` và `math.lcm`. Có thể áp dụng cho bất kỳ `ds` nào, không riêng `3, 5, 7`.

````python
import math
from itertools import combinations

def sum_multiples_general(n: int, divisors) -> int:
    total = 0
    for size in range(1, len(divisors) + 1):
        sign = 1 if size % 2 == 1 else -1
        for group in combinations(divisors, size):
            k = math.lcm(*group)
            m = n // k
            total += sign * k * m * (m + 1) // 2
    return total

class SolutionGeneral:
    def sumOfMultiples(self, n: int) -> int:
        return sum_multiples_general(n, (3, 5, 7))

print(SolutionGeneral().sumOfMultiples(7))       # 21
print(SolutionGeneral().sumOfMultiples(10))      # 40
print(sum_multiples_general(15, (4, 6)))         # bội của 4 hoặc 6 trong [1, 15]: 4,6,8,12 = 30
````

**Độ phức tạp:** với `r` ước: `2^r − 1` tập con ⇒ Time `O(2^r)`; Space `O(1)`. Với `r = 3` chỉ có `7` số hạng, đúng bằng Cách 4.

## 8. TRACE từng bước

Trace Cách 1 trên `n = 10`: cột "dư" là `(x % 3, x % 5, x % 7)`.

| `x` | dư `(%3, %5, %7)` | chia hết cho ít nhất một? | `total` sau |
|---|---|---|---|
| 1 | `(1,1,1)` | không | 0 |
| 2 | `(2,2,2)` | không | 0 |
| 3 | `(0,3,3)` | có | 3 |
| 4 | `(1,4,4)` | không | 3 |
| 5 | `(2,0,5)` | có | 8 |
| 6 | `(0,1,6)` | có | 14 |
| 7 | `(1,2,0)` | có | 21 |
| 8 | `(2,3,1)` | không | 21 |
| 9 | `(0,4,2)` | có | 30 |
| 10 | `(1,0,3)` | có | 40 |

````python
def trace_loop(n):
    total = 0
    for x in range(1, n + 1):
        r = (x % 3, x % 5, x % 7)
        ok = 0 in r
        if ok:
            total += x
        print(f"x={x:2} dư={r} chia hết ít nhất một: {str(ok):5} total={total}")
    print("kết quả:", total)
    return total

trace_loop(10)
````

````python
def trace_formula(n):
    print(f"n = {n}")
    parts = []
    for k, sign in [(3, +1), (5, +1), (7, +1), (15, -1), (21, -1), (35, -1), (105, +1)]:
        m = n // k
        s = k * m * (m + 1) // 2
        parts.append(sign * s)
        print(f"  k={k:3}: m = {n}//{k} = {m}, S = {k}*{m}*{m + 1}//2 = {s}, dấu {'+' if sign > 0 else '-'}")
    print("  tổng =", sum(parts))
    return sum(parts)

trace_formula(10)
trace_formula(105)
````

## 9. Code nộp LeetCode

Với `n <= 10^3`, Cách 2 (một dòng) là đủ và dễ đọc nhất. Nếu muốn lời giải `O(1)`, nộp Cách 4. Ở đây nộp **Cách 4** vì nó tổng quát với mọi `n`.

````python
class Solution:
    def sumOfMultiples(self, n: int) -> int:
        def S(k: int) -> int:
            m = n // k
            return k * m * (m + 1) // 2
        return S(3) + S(5) + S(7) - S(15) - S(21) - S(35) + S(105)
````

````python
print(Solution().sumOfMultiples(7))     # 21
print(Solution().sumOfMultiples(10))    # 40
print(Solution().sumOfMultiples(9))     # 30
````

## 10. Vì sao thuật toán đúng

1. **Cách duyệt (1, 2).** Mỗi `x` trong `[1, n]` được xét đúng một lần và cộng vào tổng đúng khi thoả "chia hết ít nhất một trong `3, 5, 7`" ⇒ tổng đúng theo định nghĩa.
2. **Giao của các tập.** `x` chia hết cho cả `a` và `b` ⟺ `x` chia hết cho `lcm(a, b)`. Vì `3, 5, 7` là các số nguyên tố, `lcm` đôi một bằng tích: `15, 21, 35`; `lcm` cả ba bằng `105`.
3. **Bao hàm – loại trừ.** Phần tử chia hết đúng cho `j` trong ba số (với `j = 1, 2, 3`) được cộng ở các số hạng: `j = 1`: `+1` lần (ở `S(a)`); `j = 2`: `+2 − 1 = 1` lần; `j = 3`: `+3 − 3 + 1 = 1` lần. Mọi phần tử thuộc hợp đều được cộng **đúng một lần**; phần tử không chia hết số nào không được cộng.
4. **Cấp số cộng.** `S(k)` = `k + 2k + ... + mk = k · m(m + 1)/2` với `m = n // k` là số bội nhỏ hơn hoặc bằng `n`. `m(m + 1)` luôn chẵn nên `//` chính xác.
5. **Ví dụ kiểm tra.** `n = 10`: `S(3) = 18`, `S(5) = 15`, `S(7) = 7`, các `S(15) = S(21) = S(35) = S(105) = 0` ⇒ `18 + 15 + 7 = 40`.

## 11. Phân tích Time / Space Complexity

| Cách | Time | Space |
|---|---|---|
| 1. Vòng lặp | `O(n)` | `O(1)` |
| 2. `sum` + generator | `O(n)` | `O(1)` |
| 3. `set` hợp `range` | `O(n)` | `O(n)` |
| 4. Bao hàm – loại trừ 7 số hạng | `O(1)` | `O(1)` |
| 5. Bao hàm – loại trừ tổng quát | `O(2^r)`, `r` = số ước | `O(1)` |

Với `n <= 10^3` mọi cách đều tức thì. Ý nghĩa thực của Cách 4 là cho thấy bài toán có công thức đóng, giúp mở rộng cho `n` rất lớn.

## 12. Bộ test

### 12.1. Ví dụ của đề và edge case

Các ca quan trọng: `n = 1, 2` (chưa có bội); `n = 3, 5, 7` (đúng bằng ước); `n = 15, 21, 35, 105` (số chia hết cho nhiều ước đầu tiên xuất hiện); `n = 1000` (tối đa).

````python
sols = [SolutionLoop(), SolutionSumGen(), SolutionSet(), SolutionFormula(), SolutionGeneral(), Solution()]

def oracle(n):
    # viết bằng danh sách, không dùng công thức
    chon = []
    for x in range(1, n + 1):
        if x % 3 == 0:
            chon.append(x)
        elif x % 5 == 0:
            chon.append(x)
        elif x % 7 == 0:
            chon.append(x)
    return sum(chon)

cases = [
    ("ví dụ 1", 7, 21),
    ("ví dụ 2", 10, 40),
    ("ví dụ 3", 9, 30),
    ("n = 1", 1, 0),
    ("n = 2", 2, 0),
    ("n = 3", 3, 3),
    ("n = 5", 5, 8),
    ("n = 15 (15 xuất hiện ở 3 và 5)", 15, 81),
    ("n = 21 (21 ở 3 và 7)", 21, 140),
    ("n = 35 (35 ở 5 và 7)", 35, 342),
    ("n = 105 (105 ở cả ba)", 105, 3045),
]
for name, n, exp in cases:
    assert exp == oracle(n), (name, exp, oracle(n))
    for s in sols:
        got = s.sumOfMultiples(n)
        assert got == exp, (name, type(s).__name__, got, exp)
    print("PASS", name, "->", exp)
````

Kiểm tra tay `n = 15`: các số là `3, 5, 6, 7, 9, 10, 12, 14, 15`, tổng `81`. Công thức: `S(3) = 3·5·6/2 = 45`, `S(5) = 5·3·4/2 = 30`, `S(7) = 7·2·3/2 = 21`, `S(15) = 15·1·2/2 = 15`, các số hạng còn lại `0` ⇒ `45 + 30 + 21 − 15 = 81`.

## 13. Các lỗi người mới hay gặp

### Lỗi 1: Cộng riêng các bội rồi cộng lại (đếm trùng)

````python
def sai_dem_trung(n: int) -> int:
    return sum(range(3, n + 1, 3)) + sum(range(5, n + 1, 5)) + sum(range(7, n + 1, 7))

print("sai :", sai_dem_trung(15))                 # 96: 15 bị cộng hai lần
print("đúng:", Solution().sumOfMultiples(15))     # 81
````

**Vì sao sai:** `15` là bội của `3` và của `5` nên xuất hiện ở hai tổng và bị cộng **hai lần** (`96 − 81 = 15`). Phải loại trùng bằng `set`, bằng vòng lặp `or`, hoặc bao hàm – loại trừ.

### Lỗi 2: Dùng `range(1, n)` quên mất `n`

````python
def sai_thieu_n(n: int) -> int:
    return sum(x for x in range(1, n) if x % 3 == 0 or x % 5 == 0 or x % 7 == 0)

print("sai :", sai_thieu_n(9), sai_thieu_n(10), sai_thieu_n(7))      # 21 30 15
print("đúng:", Solution().sumOfMultiples(9), Solution().sumOfMultiples(10), Solution().sumOfMultiples(7))   # 30 40 21
````

**Vì sao sai:** `range(1, n)` **không gồm `n`**. Đề viết "trong đoạn `[1, n]` gồm cả hai đầu", nên phải `range(1, n + 1)`. Lỗi này khiến cả ba ví dụ của đề đều sai.

### Lỗi 3: Dùng `and` thay cho `or`

````python
def sai_and(n: int) -> int:
    return sum(x for x in range(1, n + 1) if x % 3 == 0 and x % 5 == 0 and x % 7 == 0)

print("sai :", sai_and(10), sai_and(105))                              # 0 105
print("đúng:", Solution().sumOfMultiples(10), Solution().sumOfMultiples(105))
````

**Vì sao sai:** `and` chỉ giữ các số chia hết cho **cả ba** (tức bội của `105`). Đề cần số chia hết cho **ít nhất một** ⇒ `or`.

### Lỗi 4: Dùng `/` thay vì `//` trong công thức

````python
def sai_chia_thuc(n: int) -> int:
    def S(k):
        m = n // k
        return k * m * (m + 1) / 2          # / trả số thực
    return S(3) + S(5) + S(7) - S(15) - S(21) - S(35) + S(105)

r = sai_chia_thuc(10)
print("sai :", r, type(r).__name__)         # 40.0 float: LeetCode kỳ vọng int
print("đúng:", Solution().sumOfMultiples(10), type(Solution().sumOfMultiples(10)).__name__)
````

**Vì sao sai:** kết quả là `float` (`40.0`) thay vì `int`; với số lớn còn mất độ chính xác. Dùng `//` (chia nguyên): `m(m + 1)` luôn chẵn nên phép chia chính xác.

## 14. Mẹo làm bài nhanh & nhận dạng pattern

- "Tổng/đếm số chia hết cho **a hoặc b**" ⇒ nghĩ tới **bao hàm – loại trừ**: `A + B − (bội chung nhỏ nhất)`.
- "Tổng các bội của `k` đến `n`" ⇒ `k · m(m + 1) / 2` với `m = n // k`. Số bội: `n // k`.
- Khi `n` nhỏ (như `10^3`), cứ viết vòng lặp, đừng sợ; công thức là phần thưởng thêm.
- Mỗi lần viết `range(a, b)` nhớ hỏi: `b` có cần gồm không? (thường phải là `n + 1`).
- Đếm trùng là bẫy kinh điển khi cộng tách nhiều tập: tự hỏi "có phần tử nào thuộc hai tập không?".

````python
import math

# Bài 1
assert 10 // 3 == 3
assert 3 * (3 * 4 // 2) == 18
assert set([3, 6]) | set([5, 10]) == {3, 5, 6, 10}
assert math.lcm(4, 6) == 12

# Bài 2
chon = [x for x in range(1, 16) if x % 3 == 0 or x % 5 == 0 or x % 7 == 0]
assert chon == [3, 5, 6, 7, 9, 10, 12, 14, 15]
assert sum(chon) == 81
assert 45 + 30 + 21 - 15 == 81

# Bài 3
def sum_of_multiples_of(k, n):
    m = n // k
    return k * m * (m + 1) // 2

assert sum_of_multiples_of(3, 10) == 18
assert sum_of_multiples_of(7, 6) == 0
assert sum_of_multiples_of(1, 10) == 55

# Bài 4
def sum_multiples_set(n):
    s = set(range(3, n + 1, 3)) | set(range(5, n + 1, 5)) | set(range(7, n + 1, 7))
    return sum(s)

assert sum_multiples_set(10) == 40
assert sum_multiples_set(7) == 21

# Bài 5
def euler1(limit):
    n = limit - 1
    return sum_of_multiples_of(3, n) + sum_of_multiples_of(5, n) - sum_of_multiples_of(15, n)

assert euler1(10) == 23
assert euler1(1000) == 233168

def count_multiples(n):
    return n // 3 + n // 5 + n // 7 - n // 15 - n // 21 - n // 35 + n // 105

for n in range(1, 300):
    assert count_multiples(n) == sum(1 for x in range(1, n + 1) if x % 3 == 0 or x % 5 == 0 or x % 7 == 0)
print("Tất cả đáp án đúng")
````
