### 3.1. Phép chia nguyên `//` và phép chia dư `%`

- `a // b` là **thương nguyên** (bỏ phần thập phân): `17 // 5 = 3`.
- `a % b` là **số dư**: `17 % 5 = 2`.
- Luôn có `a == (a // b) * b + a % b`.

Trong bài này, `//` giúp ta đếm "có bao nhiêu cụm 5 lít (hoặc 4 lít) đầy đủ".

````python
print(17 // 5, 17 % 5)         # 3 2
print(4 // 5, 5 // 5, 9 // 5)  # 0 1 1
print(8 // 4, 7 // 4, 4 // 4, 3 // 4)  # 2 1 1 0
a, b = 17, 5
print(a == (a // b) * b + a % b)       # True
````

## 5. Cách 1 — Brute Force / baseline: mô phỏng từng lít một

**Ý tưởng:** làm đúng như đề, **mỗi vòng lặp tiêu 1 lít**:

1. Tiêu 1 lít ở bình chính: `mainTank -= 1`, `used += 1`, `km += 10`.
2. Nếu `used == 5` (đã tiêu đủ 5 lít kể từ lần bơm bù trước): đặt lại `used = 0`; nếu `additionalTank > 0` thì chuyển 1 lít: `additionalTank -= 1`, `mainTank += 1`.
3. Lặp tới khi `mainTank == 0`.

Chú ý: ta bơm bù **ngay khi vừa đủ 5 lít**, kể cả khi bình chính vừa về `0` (ví dụ `mainTank = 5`): lúc đó bình chính được cứu và có thêm 1 lít để chạy tiếp.

````python
from typing import List

class SolutionLiter:
    def distanceTraveled(self, mainTank: int, additionalTank: int) -> int:
        km = 0
        used = 0                      # số lít đã tiêu kể từ lần bơm bù gần nhất
        while mainTank > 0:
            mainTank -= 1
            used += 1
            km += 10
            if used == 5:
                used = 0
                if additionalTank > 0:
                    additionalTank -= 1
                    mainTank += 1
        return km
````

````python
sol = SolutionLiter()
print(sol.distanceTraveled(5, 10))   # 60
print(sol.distanceTraveled(1, 2))    # 10
print(sol.distanceTraveled(9, 10))   # 110
````

**Độ phức tạp:** số vòng lặp bằng tổng số lít được tiêu, tối đa khoảng `mainTank + mainTank/4`, tức `O(mainTank)`. Space `O(1)`.

**Nút thắt:** với giới hạn của đề (`<= 100`) cách này nhanh thừa. Nhưng nếu `mainTank` lên tới `10^9`, vòng lặp từng lít là quá chậm. Ta nên nhìn ra cấu trúc "cụm 5 lít" để nhảy một bước.

## 6. Ý tưởng thuật toán / bước đột phá

**Đột phá 1 — gom từng cụm 5 lít.** Thay vì tiêu từng lít, ta tiêu thẳng 5 lít (nếu bình chính còn ≥ 5), cộng 50 km, rồi bơm bù. Khi bình chính còn `< 5` lít thì tiêu nốt và kết thúc, vì không bao giờ đủ 5 lít nữa.

**Đột phá 2 — đếm số lần bơm bù.** Gọi `k` là số lần bơm bù. Lần bơm thứ `j` xảy ra khi tổng số lít đã tiêu đạt đúng `5j`. Để đạt tới `5j` lít, ta phải có đủ nhiên liệu: ban đầu `mainTank` cộng `j − 1` lít đã được bơm trước đó (lần bơm thứ `j` xảy ra *sau* khi tiêu đủ 5 lít, nên lít bơm lần `j` chưa giúp được):

`mainTank + (j − 1) >= 5j  ⟺  mainTank − 1 >= 4j  ⟺  j <= (mainTank − 1) / 4`.

Đồng thời mỗi lần bơm tốn 1 lít bình phụ nên `j <= additionalTank`. Vậy:

`k = min(additionalTank, (mainTank − 1) // 4)`

**Đột phá 3 — công thức cuối:** `km = 10 × (mainTank + k)`.

Kiểm tra: `mainTank = 5, additionalTank = 10` ⇒ `(5−1)//4 = 1`, `k = min(10, 1) = 1` ⇒ `10 × 6 = 60` ✓. `mainTank = 1` ⇒ `0 // 4 = 0` ⇒ `10 × 1 = 10` ✓.

````python
for main, add in [(5, 10), (1, 2), (9, 10), (4, 100), (100, 1), (100, 100)]:
    k = min(add, (main - 1) // 4)
    print(f"main={main:3}, add={add:3} -> số lần bơm k={k:2}, km={10 * (main + k)}")
````

## 7. Các cách tiếp theo

Ta giảng thêm 4 cách nữa, đi từ "mô phỏng cụm" đến công thức.

### Cách 2 — Mô phỏng theo cụm 5 lít

**Từng đoạn:**

- `while mainTank >= 5`: còn đủ 5 lít để hoàn thành một cụm.
- Trong thân vòng lặp: `mainTank -= 5`, `km += 50`; nếu `additionalTank > 0` thì `additionalTank -= 1`, `mainTank += 1`.
- Ra khỏi vòng lặp, `mainTank < 5` lít còn lại sẽ được tiêu hết: `km += mainTank * 10`.

````python
class SolutionChunk:
    def distanceTraveled(self, mainTank: int, additionalTank: int) -> int:
        km = 0
        while mainTank >= 5:
            mainTank -= 5
            km += 50
            if additionalTank > 0:
                additionalTank -= 1
                mainTank += 1
        return km + mainTank * 10
````

````python
sol = SolutionChunk()
print(sol.distanceTraveled(5, 10))   # 60
print(sol.distanceTraveled(1, 2))    # 10
print(sol.distanceTraveled(9, 10))   # 110
````

**Code nộp LeetCode (Cách 2):** chính là class `SolutionChunk` ở trên, đổi tên thành `Solution`.

**Độ phức tạp:** mỗi vòng giảm bình chính ròng ít nhất 4 lít (hoặc 5 nếu bình phụ hết) ⇒ Time `O(mainTank)` (nhỏ hơn cách 1 khoảng 5 lần), Space `O(1)`.

### Cách 3 — Đếm số lần bơm bù bằng điều kiện tích luỹ

**Từng đoạn:** dựa vào đột phá 2. Thử bơm lần thứ `j = 1, 2, ...`; lần `j` hợp lệ khi **cả hai** điều kiện đúng: `j <= additionalTank` và `mainTank + (j − 1) >= 5j`. Gặp lần đầu tiên không hợp lệ thì dừng (các lần sau còn khó hơn).

````python
class SolutionCountPumps:
    def distanceTraveled(self, mainTank: int, additionalTank: int) -> int:
        pumps = 0
        while pumps < additionalTank and mainTank + pumps >= 5 * (pumps + 1):
            pumps += 1
        return 10 * (mainTank + pumps)
````

````python
sol = SolutionCountPumps()
print(sol.distanceTraveled(5, 10))   # 60
print(sol.distanceTraveled(1, 2))    # 10
print(sol.distanceTraveled(9, 10))   # 110
````

**Giải thích điều kiện:** đang có `pumps` lần bơm xong, muốn bơm lần `pumps + 1` thì cần `mainTank + pumps >= 5 * (pumps + 1)` (đủ nhiên liệu để tiêu hết `5 * (pumps + 1)` lít) và còn lít trong bình phụ (`pumps < additionalTank`).

**Độ phức tạp:** Time `O(min(additionalTank, mainTank / 4))`, Space `O(1)`.

### Cách 4 — Công thức O(1) (dùng để nộp)

`k = min(additionalTank, (mainTank − 1) // 4)`, đáp án `10 × (mainTank + k)`.

````python
class SolutionFormula:
    def distanceTraveled(self, mainTank: int, additionalTank: int) -> int:
        pumps = min(additionalTank, (mainTank - 1) // 4)
        return 10 * (mainTank + pumps)
````

````python
sol = SolutionFormula()
print(sol.distanceTraveled(5, 10))   # 60
print(sol.distanceTraveled(1, 2))    # 10
print(sol.distanceTraveled(9, 10))   # 110
````

**Độ phức tạp:** Time `O(1)`, Space `O(1)`.

### Cách 5 — Đệ quy (cho ai thích nghĩ "bài toán nhỏ hơn")

`f(main, add)`: nếu `main < 5` thì đi nốt `10 × main` km; ngược lại tiêu 5 lít (`50` km), nhận bơm bù `b = 1` nếu `add > 0` (ngược lại `0`) rồi gọi `f(main − 5 + b, add − b)`.

````python
class SolutionRecursive:
    def distanceTraveled(self, mainTank: int, additionalTank: int) -> int:
        if mainTank < 5:
            return 10 * mainTank
        b = 1 if additionalTank > 0 else 0
        return 50 + self.distanceTraveled(mainTank - 5 + b, additionalTank - b)
````

````python
sol = SolutionRecursive()
print(sol.distanceTraveled(5, 10))   # 60
print(sol.distanceTraveled(1, 2))    # 10
print(sol.distanceTraveled(9, 10))   # 110
````

**Độ phức tạp:** Time `O(mainTank)`; Space `O(mainTank)` do ngăn xếp đệ quy (tối đa vài chục tầng với đề này). Cách này chủ yếu để luyện tư duy đệ quy; khi nộp nên dùng cách 4.

## 8. TRACE từng bước

Trace Cách 1 (mô phỏng từng lít) trên `mainTank = 5`, `additionalTank = 10`:

| bước | main trước | tiêu 1 lít → main | `used` | km | bơm bù? | main, add sau |
|---|---|---|---|---|---|---|

| 6 | 1 | 0 | 1 | 60 | không | 0, 9 |

Kết thúc vì `main = 0`, đáp án `60`.

````python
def trace_liter(mainTank, additionalTank):
    km = 0
    used = 0
    step = 0
    print(f"bắt đầu: main={mainTank}, add={additionalTank}")
    while mainTank > 0:
        step += 1
        mainTank -= 1
        used += 1
        km += 10
        pump = False
        if used == 5:
            used = 0
            if additionalTank > 0:
                additionalTank -= 1
                mainTank += 1
                pump = True
        print(f"bước {step}: used={used}, km={km}, bơm bù={pump}, main={mainTank}, add={additionalTank}")
    print("kết quả:", km)
    return km

trace_liter(5, 10)
````

````python
def trace_formula(mainTank, additionalTank):
    a = (mainTank - 1) // 4
    pumps = min(additionalTank, a)
    print(f"(main - 1) // 4 = ({mainTank} - 1) // 4 = {a}")
    print(f"pumps = min(add={additionalTank}, {a}) = {pumps}")
    print(f"tổng lít = main + pumps = {mainTank} + {pumps} = {mainTank + pumps}")
    km = 10 * (mainTank + pumps)
    print("km =", km)
    return km

trace_formula(9, 10)
trace_formula(9, 1)
trace_formula(4, 50)
````

## 10. Vì sao thuật toán đúng

**Bước 1 — đáp án = 10 × (tổng lít được tiêu).** Mỗi lít tiêu ở bình chính cho đúng 10 km. Tổng lít tiêu = `mainTank` ban đầu cộng tổng số lít được bơm bù (mỗi lít bơm vào cuối cùng đều được tiêu hết, vì xe chỉ dừng khi bình chính cạn).

**Bước 2 — lần bơm bù thứ `j` xảy ra khi và chỉ khi tổng lít đã tiêu đạt đúng `5j` và bình phụ còn lít.** Ta chứng minh `j` lần bơm đầu đều xảy ra **khi và chỉ khi** `j <= additionalTank` và `mainTank + (j − 1) >= 5j`.

- (Cần) Để tiêu đủ `5j` lít, nhiên liệu bình chính tổng cộng phải đạt ít nhất `5j`. Nhưng trước thời điểm bơm lần `j` chỉ mới có `j − 1` lít được bơm. Nên `mainTank + j − 1 >= 5j`. Còn bình phụ phải còn ít nhất `j` lít ban đầu để bơm `j` lần: `j <= additionalTank`.
- (Đủ) Nếu hai điều kiện đúng cho `j`, thì chúng cũng đúng cho mọi `j' < j` (vế phải nhỏ hơn). Tại thời điểm lần `j'` xảy ra, bình chính vừa tiêu đủ `5j'` lít, và bình phụ còn `additionalTank − (j' − 1) >= 1` lít nên bơm thành công.

**Bước 3 — số lần bơm lớn nhất** là số nguyên `j` lớn nhất thoả `j <= additionalTank` và `4j <= mainTank − 1`, tức `j = min(additionalTank, (mainTank − 1) // 4)` (nếu `mainTank = 1` thì `(0) // 4 = 0` không bơm; `mainTank − 1 >= 0` vì ràng buộc `mainTank >= 1`).

**Bước 4 — đối chiếu.** Cả 5 cách cho cùng kết quả trên mọi input ở mục 12, kể cả ngẫu nhiên, và cách mô phỏng từng lít là bản dịch thẳng của đề nên đóng vai trò **oracle**.

## 11. Phân tích Time / Space Complexity

`m = mainTank`, `a = additionalTank`.

| Cách | Time | Space | Ghi chú |
|---|---|---|---|
| 1. Mô phỏng từng lít | `O(m)` | `O(1)` | số vòng lặp ≈ tổng lít tiêu ≤ `m + m/4` |
| 2. Mô phỏng theo cụm 5 lít | `O(m)` | `O(1)` | số vòng ≤ `m/4` |
| 3. Đếm số lần bơm bù | `O(min(a, m/4))` | `O(1)` | vòng lặp chỉ chạy theo số lần bơm |
| 4. Công thức | `O(1)` | `O(1)` | nhanh nhất |

## 13. Các lỗi người mới hay gặp

### Lỗi 1: Bơm bù mỗi lít thay vì mỗi 5 lít

````python
def sai_bom_moi_lit(mainTank: int, additionalTank: int) -> int:
    km = 0
    while mainTank > 0:
        mainTank -= 1
        km += 10
        if additionalTank > 0:          # sai: không đếm đủ 5 lít
            additionalTank -= 1
            mainTank += 1
    return km

print("sai :", sai_bom_moi_lit(5, 10))                 # 150
print("đúng:", Solution().distanceTraveled(5, 10))     # 60
````

**Vì sao sai:** đề chỉ cho bơm **một lần mỗi 5 lít đã tiêu**. Code sai bơm sau mỗi lít nên bình chính không bao giờ cạn, đi tới 150 km (hết bình phụ) – quá nhiều.

### Lỗi 2: Dùng `mainTank // 5` cho số lần bơm

````python
def sai_chia_5(mainTank: int, additionalTank: int) -> int:
    pumps = min(additionalTank, mainTank // 5)    # sai: lít được bơm lại tiếp tục được tiêu
    return 10 * (mainTank + pumps)

print("sai :", sai_chia_5(9, 10), "| đúng:", Solution().distanceTraveled(9, 10))     # 100 vs 110
print("sai :", sai_chia_5(13, 10), "| đúng:", Solution().distanceTraveled(13, 10))    # 150 vs 160
````

**Vì sao sai:** lít vừa được bơm bù **cũng góp vào** tổng lít tiêu và có thể giúp đạt thêm cụm 5 lít nữa. Với `main = 9` có 2 lần bơm chứ không phải 1. Công thức đúng dùng `(mainTank − 1) // 4`, vì mỗi lần bơm "lời" 1 lít sau khi tiêu ròng 4 lít.

### Lỗi 3: Quên giới hạn bởi `additionalTank`

````python
def sai_quen_min(mainTank: int, additionalTank: int) -> int:
    pumps = (mainTank - 1) // 4        # sai: không chặn bởi bình phụ
    return 10 * (mainTank + pumps)

print("sai :", sai_quen_min(100, 1))                # 1240
print("đúng:", Solution().distanceTraveled(100, 1)) # 1010
````

**Vì sao sai:** mỗi lần bơm lấy 1 lít từ bình phụ; bình phụ chỉ có `additionalTank` lít nên **tối đa** bơm `additionalTank` lần. Phải có `min`.

### Lỗi 4: Quên nhân 10 km/lít hoặc dùng `5 * 10` cho mọi lần

````python
def sai_quen_nhan(mainTank: int, additionalTank: int) -> int:
    pumps = min(additionalTank, (mainTank - 1) // 4)
    return mainTank + pumps          # sai: trả số lít, không phải km

print("sai :", sai_quen_nhan(5, 10))                # 6
print("đúng:", Solution().distanceTraveled(5, 10))  # 60
````

**Vì sao sai:** đề hỏi quãng đường (km), mỗi lít đi 10 km nên cần nhân 10.

### Lỗi 5 (nhỏ): `(mainTank - 1) // 4` với `mainTank = 0`

Ràng buộc đảm bảo `mainTank >= 1`. Nếu tự mở rộng đề cho `mainTank = 0`, `(0 − 1) // 4 = −1` (Python làm tròn xuống), công thức cho kết quả **âm**. Khi ngoài ràng buộc nhớ thêm `max(0, ...)`.

````python
print((0 - 1) // 4)               # -1  (làm tròn xuống, không phải 0)
print(max(0, (0 - 1) // 4))       # 0
````

## 14. Mẹo làm bài nhanh & nhận dạng pattern

- Đề "cứ mỗi `x` đơn vị thì được thêm `y`" thường là **mô phỏng** hoặc **công thức đếm số lần thưởng**. Trước hết viết mô phỏng đúng, sau đó mới tìm công thức.
- Khi phần thưởng cũng **tham gia vào chu kỳ sau** (lít bơm bù lại được tiêu), đừng chia thẳng `main // 5`; hãy hỏi "sau mỗi chu kỳ, mất ròng bao nhiêu?". Ở đây chu kỳ tiêu 5 lít, nhận 1 lít ⇒ **ròng 4 lít**.
- Nhớ cặp `min(nguồn_giới_hạn_1, nguồn_giới_hạn_2)` cho "đếm số lần tối đa khi bị chặn bởi hai thứ".
- Với ràng buộc nhỏ (`<= 100`), mô phỏng luôn an toàn; công thức chỉ cần khi ràng buộc lớn.
- Luôn dùng mô phỏng làm **oracle** để kiểm công thức bằng vét cạn toàn miền.

## 15. Biến thể của bài

1. **Chu kỳ khác:** cứ `c` lít tiêu thì bơm 1 lít (`c >= 2`). Số lần bơm tối đa `min(a, (m − 1) // (c − 1))`.
2. **Mỗi lần bơm `x` lít** (hoặc bơm `min(x, add)` lít): công thức phức tạp hơn, nên mô phỏng theo cụm.
3. **Mức tiêu thụ khác:** `d` km/lít ⇒ đáp án `d × (m + pumps)`.
4. **Bình phụ vô hạn:** số lần bơm `(m − 1) // 4`.
5. **Hỏi ngược:** cần đi ít nhất `D` km, bình chính tối thiểu phải có bao nhiêu lít (khi bình phụ vô hạn)? Có thể tìm bằng tìm kiếm nhị phân hoặc đảo công thức.
6. **Nhiều xe / nhiều bình:** dùng mô phỏng theo sự kiện (event simulation) hoặc heap.

## 16. So sánh các cách làm

| Cách | Ý tưởng | Time | Space | Khi nào dùng |
|---|---|---|---|---|
| 1. Từng lít | dịch thẳng đề | `O(m)` | `O(1)` | làm oracle, ràng buộc nhỏ |
| 2. Cụm 5 lít | nhảy 5 lít một lần | `O(m)` | `O(1)` | muốn mô phỏng nhưng gọn hơn |
| 3. Đếm lần bơm | điều kiện `main + p >= 5(p+1)` | `O(min(a, m/4))` | `O(1)` | khi chưa nhớ công thức |
| 4. Công thức | `min(a, (m−1)//4)` | `O(1)` | `O(1)` | **code nộp**, input lớn |
| 5. Đệ quy | bài toán con `(main − 5 + b, add − b)` | `O(m)` | `O(m)` | luyện tư duy đệ quy |
