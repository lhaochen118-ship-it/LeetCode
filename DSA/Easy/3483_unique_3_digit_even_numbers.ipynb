## 1. Mục tiêu học

Sau notebook này bạn sẽ:

1. Hiểu đề: ghép **đúng 3 chữ số** từ mảng `digits` (mỗi phần tử dùng **tối đa một lần**) thành số có **3 chữ số**, **chẵn**, **không bắt đầu bằng `0`**, rồi đếm số **giá trị khác nhau**.
2. Biết dùng `set` để loại trùng, `itertools.permutations` để sinh hoán vị, và `collections.Counter` để kiểm tra "đủ chữ số để ghép".
3. Nhận ra vì sao đây là bài **liệt kê (enumeration)**: `n <= 10` nên chỉ cần thử mọi cách chọn 3 vị trí.
4. Viết 4 cách: ba vòng lặp chỉ số + `set`, `permutations` + `set`, duyệt mọi số chẵn `100..998` rồi kiểm tra bằng `Counter`, và đếm chữ số với vòng lặp `a, b, c` (không cần `set`).
5. Tránh các lỗi: dùng lại một phần tử nhiều lần, quên điều kiện số đầu khác `0`, quên khử trùng, quên điều kiện chữ số cuối chẵn.

## 2. Đề bài & ví dụ

Cho mảng các chữ số `digits`. Hãy đếm có bao nhiêu **số nguyên khác nhau** có thể tạo thành theo các quy tắc:

1. Số được tạo bằng cách **ghép 3 phần tử** của `digits` theo thứ tự tùy ý; mỗi phần tử (mỗi **vị trí**) chỉ được dùng **tối đa một lần**. Nếu một chữ số xuất hiện hai lần trong `digits` thì được dùng hai lần.
2. Số đó **không có chữ số `0` ở đầu** (nên là số có đúng 3 chữ số).
3. Số đó là số **chẵn**.

Trả về số lượng giá trị phân biệt.

| Ví dụ | digits | Output | Giải thích |
|---|---|---|---|
| 1 | `[1,2,3,4]` | `12` | các số: `124,132,134,142,214,234,312,314,324,342,412,432`; `222` không ghép được vì chỉ có một chữ số `2` |
| 2 | `[0,2,2]` | `2` | chỉ có `202` và `220`; chữ số `2` dùng được hai lần vì xuất hiện hai lần |
| 3 | `[6,6,6]` | `1` | chỉ ghép được `666` |
| 4 | `[1,3,5]` | `0` | không có chữ số chẵn để làm chữ số cuối |

Constraints: `3 <= digits.length <= 10`; `0 <= digits[i] <= 9`.

````python
digits = [0, 2, 2]
found = set()
n = len(digits)
for i in range(n):
    for j in range(n):
        for k in range(n):
            if i != j and j != k and i != k:          # ba vị trí khác nhau
                a, b, c = digits[i], digits[j], digits[k]
                if a != 0 and c % 2 == 0:             # không đầu 0, chữ số cuối chẵn
                    found.add(100 * a + 10 * b + c)
print(sorted(found), "->", len(found), "(mong đợi 2)")
````

## 3. Kiến thức nền

### 3.1. Ghép chữ số thành số

Ba chữ số `a, b, c` (hàng trăm, chục, đơn vị) tạo số `100*a + 10*b + c`. Ví dụ `a=2, b=0, c=2` cho `202`. Số là chẵn khi và chỉ khi **chữ số cuối** `c` chẵn (`c % 2 == 0`). Số có đúng 3 chữ số khi `a != 0`.

````python
a, b, c = 2, 0, 2
print(100 * a + 10 * b + c, (100 * a + 10 * b + c) % 2 == 0)
print(100 * 0 + 10 * 2 + 2)       # a = 0 -> 22 chỉ có 2 chữ số, không hợp lệ
````

### 3.2. `set` để đếm giá trị phân biệt

Nhiều cách chọn vị trí khác nhau có thể cho **cùng một số**: ví dụ `digits = [0, 2, 2]` có hai chữ số `2` ở các vị trí khác nhau, nên `202` ra từ hai hoán vị. `set` tự loại trùng: thêm một giá trị đã có thì không đổi kích thước.

````python
s = set()
for v in [202, 220, 202, 220, 202]:
    s.add(v)
print(s, len(s))
````

### 3.3. `itertools.permutations(iterable, r)`

`permutations(digits, 3)` sinh mọi **bộ ba có thứ tự** từ **vị trí khác nhau** (không lặp vị trí, nhưng giá trị có thể lặp nếu mảng có giá trị lặp). Số bộ ba bằng `n·(n-1)·(n-2)`. Với `n = 3` là `6`, với `n = 10` là `720`.

````python
from itertools import permutations
print(list(permutations([1, 2, 3], 3))[:4], "...")
print(len(list(permutations(range(10), 3))))
print(list(permutations([0, 2, 2], 3)))       # hai số 2 ở hai vị trí khác nhau -> bộ ba lặp
````

### 3.4. `collections.Counter` và kiểm tra "đủ để ghép"

`Counter(digits)` cho biết mỗi chữ số có bao nhiêu bản. Để ghép số `x` cần `Counter(str(x))` **không vượt** `Counter(digits)` ở mọi chữ số. Nhờ đó ta có thể duyệt thẳng các số `100..998` (chẵn) thay vì duyệt hoán vị.

````python
from collections import Counter
have = Counter([0, 2, 2])
for x in [202, 220, 222, 200]:
    need = Counter(str(x))
    ok = all(have[int(ch)] >= cnt for ch, cnt in need.items())
    print(x, dict(need), "ghép được" if ok else "không đủ chữ số")
````

### 3.5. Mảng đếm 10 ô và "mượn–trả" chữ số

`cnt = [0] * 10` với `cnt[d]` là số bản của chữ số `d`. Khi chọn một chữ số ta **mượn** (`cnt[d] -= 1`), thử tiếp các chữ số khác, rồi **trả lại** (`cnt[d] += 1`). Kỹ thuật này là nền của backtracking và đảm bảo mỗi vị trí chỉ dùng một lần mà không cần `set` chỉ số.

````python
cnt = [0] * 10
for d in [0, 2, 2]:
    cnt[d] += 1
print(cnt)
cnt[2] -= 1          # mượn một chữ số 2
print(cnt)
cnt[2] += 1          # trả lại
print(cnt)
````

## 4. Ý tưởng nền tảng / cách nghĩ trực giác

Làm tay với `digits = [0, 2, 2]`. Chữ số cuối phải chẵn: `0` hoặc `2`.

```
chữ số cuối = 0  -> còn lại {2, 2}  -> hàng trăm 2, chục 2 -> 220
chữ số cuối = 2  -> còn lại {0, 2}  -> hàng trăm không được là 0 -> hàng trăm 2, chục 0 -> 202
```

Có hai số: `220` và `202`. Nếu ta cứ thử mọi hoán vị vị trí rồi cho vào `set` thì cũng ra kết quả đó, chỉ là có vài hoán vị cho **cùng** một số.

Nhận xét: số cần xét chỉ có tối đa `10·9·8 = 720` bộ ba vị trí (hoặc `9·10·5 = 450` bộ chữ số), nên brute force là đủ nhanh. Điểm khó của bài **không phải** tốc độ mà là: (i) không dùng một vị trí hai lần, (ii) bỏ số đầu `0`, (iii) chỉ giữ số chẵn, (iv) đếm giá trị **phân biệt**.

## 5. Cách 1 — Brute Force / baseline

Ba vòng lặp lồng nhau theo **chỉ số** `i, j, k` (hàng trăm, chục, đơn vị), bỏ qua khi có hai chỉ số bằng nhau; kiểm tra `digits[i] != 0` và `digits[k]` chẵn; thêm giá trị vào `set`; trả về kích thước `set`.

Vì dùng **chỉ số** chứ không dùng giá trị, quy tắc "mỗi phần tử chỉ dùng một lần" được thỏa tự nhiên. Nút thắt: ta sinh cả các cặp vị trí cho ra cùng giá trị rồi mới khử trùng; không đáng kể với `n <= 10`, nhưng là chỗ ta có thể tối ưu bằng cách đếm chữ số.

````python
class SolutionBrute:
    def totalNumbers(self, digits):
        n = len(digits)
        seen = set()
        for i in range(n):
            for j in range(n):
                for k in range(n):
                    if i == j or j == k or i == k:
                        continue
                    if digits[i] == 0 or digits[k] % 2 != 0:
                        continue
                    seen.add(100 * digits[i] + 10 * digits[j] + digits[k])
        return len(seen)

print(SolutionBrute().totalNumbers([1, 2, 3, 4]))
print(SolutionBrute().totalNumbers([0, 2, 2]))
print(SolutionBrute().totalNumbers([6, 6, 6]))
print(SolutionBrute().totalNumbers([1, 3, 5]))
````

**Độ phức tạp:** `n³` bộ ba chỉ số (thực tế `n(n-1)(n-2)` sau khi bỏ trùng chỉ số) -> `O(n³)` thời gian, `O(n³)` bộ nhớ cho `set` ở mức lý thuyết, nhưng vì chỉ có tối đa `450` số chẵn 3 chữ số nên bộ nhớ thực tế `O(1)` hằng số.

## 6. Ý tưởng thuật toán / bước đột phá

**Bước đột phá:** số cần đếm chỉ có tối đa `900` giá trị 3 chữ số (`100..999`), trong đó `450` số chẵn. Thay vì sinh từ mảng, ta **duyệt các ứng viên** và hỏi: "mảng này có đủ chữ số để ghép số đó không?" — đúng bằng so sánh hai `Counter`.

**Quy tắc "nếu… thì…":**

- Nếu số `x` có chữ số `d` xuất hiện `m` lần mà `digits` chỉ có `< m` bản `d` -> **không** ghép được.
- Nếu mọi chữ số của `x` đều đủ bản -> ghép được, và `x` được tính **đúng một lần** (không cần `set`).
- Số chẵn: chỉ duyệt `range(100, 1000, 2)`.

**Invariant (với cách đếm chữ số):** tại mỗi thời điểm `cnt[d]` = số bản `d` **chưa dùng** ở các hàng đã chọn, nên mỗi vị trí được dùng tối đa một lần.

````python
have = Counter([0, 2, 2])
good = []
for x in range(100, 1000, 2):
    need = Counter(map(int, str(x)))
    if all(have[d] >= m for d, m in need.items()):
        good.append(x)
print(good, len(good))
````

## 7. Các cách tiếp theo

Ta có 4 cách: (1) ba vòng lặp chỉ số + `set`; (2) `permutations` + `set`; (3) duyệt số chẵn `100..998` + `Counter`; (4) vòng lặp chữ số với mảng đếm.

### Cách 2 — `itertools.permutations` + `set`

`permutations(digits, 3)` sinh mọi bộ ba **vị trí khác nhau** (tương đương ba vòng lặp chỉ số của Cách 1). Chỉ giữ các bộ có chữ số đầu khác `0` và chữ số cuối chẵn, ghép thành số rồi đưa vào `set`.

````python
from itertools import permutations

class SolutionPerm:
    def totalNumbers(self, digits):
        seen = set()
        for a, b, c in permutations(digits, 3):
            if a != 0 and c % 2 == 0:
                seen.add(100 * a + 10 * b + c)
        return len(seen)

print(SolutionPerm().totalNumbers([1, 2, 3, 4]))
print(SolutionPerm().totalNumbers([0, 2, 2]))
print(SolutionPerm().totalNumbers([6, 6, 6]))
print(SolutionPerm().totalNumbers([1, 3, 5]))
````

Độ phức tạp Cách 2: `O(n³)` thời gian (`n(n-1)(n-2) <= 720`), bộ nhớ `O(1)` hằng số vì `set` có tối đa `450` phần tử.

### Cách 3 — Duyệt số chẵn `100..998` và so `Counter`

Với mỗi số chẵn `x` từ `100` đến `998`, tách chữ số bằng `str`, đếm bằng `Counter`, và kiểm tra `digits` có đủ bản. Không cần `set` vì mỗi `x` chỉ duyệt một lần. Độ phức tạp **không phụ thuộc** vào nội dung mảng: luôn `450` ứng viên.

````python
class SolutionEnumerate:
    def totalNumbers(self, digits):
        have = Counter(digits)
        total = 0
        for x in range(100, 1000, 2):
            need = Counter(map(int, str(x)))
            if all(have[d] >= m for d, m in need.items()):
                total += 1
        return total

print(SolutionEnumerate().totalNumbers([1, 2, 3, 4]))
print(SolutionEnumerate().totalNumbers([0, 2, 2]))
print(SolutionEnumerate().totalNumbers([6, 6, 6]))
print(SolutionEnumerate().totalNumbers([1, 3, 5]))
````

Độ phức tạp Cách 3: `450` ứng viên × `O(1)` (3 chữ số) = `O(450 + n)` thời gian, tức `O(n)` để dựng `Counter` ban đầu rồi hằng số; bộ nhớ `O(1)`.

### Cách 4 — Vòng lặp chữ số `a, b, c` với mảng đếm "mượn–trả"

Duyệt `a` từ `1..9`, `b` từ `0..9`, `c` trong `0,2,4,6,8` (đã đảm bảo đầu khác `0` và cuối chẵn). Trước khi chọn một chữ số kiểm tra `cnt[d] > 0`, chọn xong thì giảm `cnt[d]`, lúc quay lại thì tăng. Mỗi bộ `(a, b, c)` khác nhau là **một số khác nhau**, nên không cần `set`.

````python
class SolutionDigits:
    def totalNumbers(self, digits):
        cnt = [0] * 10
        for d in digits:
            cnt[d] += 1
        total = 0
        for a in range(1, 10):
            if cnt[a] == 0:
                continue
            cnt[a] -= 1                      # mượn a
            for b in range(10):
                if cnt[b] == 0:
                    continue
                cnt[b] -= 1                  # mượn b
                for c in range(0, 10, 2):
                    if cnt[c] > 0:           # còn bản của c
                        total += 1
                cnt[b] += 1                  # trả b
            cnt[a] += 1                      # trả a
        return total

print(SolutionDigits().totalNumbers([1, 2, 3, 4]))
print(SolutionDigits().totalNumbers([0, 2, 2]))
print(SolutionDigits().totalNumbers([6, 6, 6]))
print(SolutionDigits().totalNumbers([1, 3, 5]))
````

Độ phức tạp Cách 4: `9 · 10 · 5 = 450` lần thử tối đa -> `O(n + 450)` thời gian, bộ nhớ `O(1)` (mảng 10 ô).

## 8. TRACE từng bước

Trace Cách 4 với `digits = [0, 2, 2]` (`cnt = [1,0,2,0,0,0,0,0,0,0]`). Chỉ `a = 2` có `cnt[2] > 0`:

| `a` | `cnt` sau mượn `a` | `b` thử được | `c` thử được | số tìm thấy |
|---|---|---|---|---|
| 2 | `cnt[2]=1`, `cnt[0]=1` | `b=0` (còn 1 bản) | `cnt[2]=1`>0 -> `c=2`; `cnt[0]=0` -> `c=0` không được | `202` |
| 2 | `cnt[2]=1` | `b=2` (còn 1 bản) -> sau mượn `cnt[2]=0`, `cnt[0]=1` | `c=0` được; `c=2` không còn | `220` |

Tổng `2`.

````python
def trace_digits(digits):
    cnt = [0] * 10
    for d in digits:
        cnt[d] += 1
    print("cnt ban đầu:", cnt)
    found = []
    for a in range(1, 10):
        if cnt[a] == 0:
            continue
        cnt[a] -= 1
        for b in range(10):
            if cnt[b] == 0:
                continue
            cnt[b] -= 1
            for c in range(0, 10, 2):
                if cnt[c] > 0:
                    found.append(100 * a + 10 * b + c)
                    print(f"a={a} b={b} c={c} -> {100 * a + 10 * b + c}  (cnt={cnt})")
            cnt[b] += 1
        cnt[a] += 1
    print("cnt cuối    :", cnt, "(đã trả lại hết)")
    print("tổng:", len(found))
    return len(found)

trace_digits([0, 2, 2])
trace_digits([1, 2, 3, 4])
````

## 9. Code nộp LeetCode

````python
from typing import List
from collections import Counter

class Solution:
    def totalNumbers(self, digits: List[int]) -> int:
        cnt = [0] * 10
        for d in digits:
            cnt[d] += 1
        total = 0
        for a in range(1, 10):          # hàng trăm khác 0
            if cnt[a] == 0:
                continue
            cnt[a] -= 1
            for b in range(10):         # hàng chục
                if cnt[b] == 0:
                    continue
                cnt[b] -= 1
                for c in range(0, 10, 2):   # hàng đơn vị chẵn
                    if cnt[c] > 0:
                        total += 1
                cnt[b] += 1
            cnt[a] += 1
        return total

print(Solution().totalNumbers([1, 2, 3, 4]), Solution().totalNumbers([0, 2, 2]), Solution().totalNumbers([6, 6, 6]), Solution().totalNumbers([1, 3, 5]))
````

## 10. Vì sao thuật toán đúng

- **Mỗi số hợp lệ được tính đúng một lần (Cách 3, 4):** số `x` hợp lệ tương ứng **một** bộ `(a, b, c)` các chữ số (hàng trăm, chục, đơn vị). Hai bộ khác nhau cho hai số khác nhau vì biểu diễn thập phân là duy nhất. Nên đếm bộ `(a, b, c)` thỏa điều kiện = đếm số phân biệt.
- **"Ghép được" khi và chỉ khi đủ bản chữ số:** để dựng `x` bằng các **vị trí khác nhau** của `digits`, mỗi chữ số `d` cần `m_d` vị trí mang giá trị `d`, trong đó `m_d` là số lần `d` xuất hiện trong `x`. Các nhu cầu này độc lập nhau nên điều kiện cần và đủ là `m_d <= số bản d trong digits` với mọi `d` — đây chính là phép kiểm tra của Cách 3 và phép "mượn–trả" của Cách 4.
- **Cách 1, 2:** duyệt mọi bộ ba vị trí khác nhau, nên mọi cách ghép đều được xét; `set` khử trùng giá trị.
- **Điều kiện số:** `a != 0` bảo đảm đúng 3 chữ số; `c` chẵn bảo đảm số chẵn (số chẵn khi và chỉ khi chữ số hàng đơn vị chẵn vì `10` chẵn).
- **Bộ test** so sánh cả bốn cách và một oracle backtracking độc lập trên hàng nghìn input ngẫu nhiên.

## 11. Phân tích Time / Space Complexity

Gọi `n = len(digits) <= 10`.

| Cách | Time | Space |
|---|---|---|
| 1. Ba vòng chỉ số + `set` | `O(n³)` (<= 720 bộ ba) | `O(1)` thực tế (<= 450 số) |
| 2. `permutations` + `set` | `O(n³)` | `O(1)` thực tế |
| 3. Duyệt `100..998` + `Counter` | `O(n + 450)` | `O(1)` |
| 4. Vòng chữ số + mảng đếm | `O(n + 450)` | `O(1)` |

Vì `n <= 10`, cả bốn cách chạy trong micro-giây. Cách 3 và 4 có ưu điểm **không đổi theo `n`**, và không cần `set`; Cách 1 và 2 dễ nghĩ nhất, ít bug nhất.

## 12. Bộ test

Trước hết viết một **oracle backtracking độc lập** (không dùng `Counter`, `permutations` hay mảng đếm): chọn lần lượt ba vị trí chưa dùng, giữ lại bộ ba thỏa điều kiện trong một `set`. Oracle này dùng để đối chiếu với năm cài đặt ở trên.

````python
def oracle(digits):
    n = len(digits)
    result = set()
    def go(chosen, used):
        if len(chosen) == 3:
            if chosen[0] != 0 and chosen[2] % 2 == 0:
                result.add(tuple(chosen))
            return
        for idx in range(n):
            if idx not in used:
                used.add(idx)
                chosen.append(digits[idx])
                go(chosen, used)
                chosen.pop()
                used.remove(idx)
    go([], set())
    return len(result)

print(oracle([0, 2, 2]), oracle([1, 2, 3, 4]))
````

Test 1: ví dụ đề và các edge case (`n = 3`, `n = 10`, toàn lẻ, toàn `0`, chỉ một chữ số chẵn, nhiều chữ số trùng). Mọi input tuân thủ `3 <= n <= 10`, `0 <= digits[i] <= 9`.

## 13. Các lỗi người mới hay gặp

### Lỗi 1: cho phép dùng lại một chữ số (không theo dõi số bản)

````python
def wrong_reuse(digits):
    vals = set(digits)                       # SAI: chỉ giữ giá trị, mất thông tin số bản
    seen = set()
    for a in vals:
        for b in vals:
            for c in vals:
                if a != 0 and c % 2 == 0:
                    seen.add(100 * a + 10 * b + c)
    return len(seen)

digits = [1, 2, 3, 4]
print("sai :", wrong_reuse(digits))
print("đúng:", Solution().totalNumbers(digits))
assert wrong_reuse(digits) != Solution().totalNumbers(digits)
````

Vì sao sai: code sai ghép được cả `222`, `444`, `112`... dù mảng chỉ có **một** chữ số `2`; ra `32` thay vì `12`. **Sửa:** dùng chỉ số khác nhau (Cách 1, 2) hoặc kiểm tra/mượn–trả số bản (Cách 3, 4).

### Lỗi 2: quên điều kiện số đầu khác `0`

````python
def wrong_leading_zero(digits):
    seen = set()
    for a, b, c in permutations(digits, 3):
        if c % 2 == 0:                       # SAI: thiếu a != 0
            seen.add((a, b, c))              # dùng tuple để 022 và 22 không bị gộp nhầm
    return len(seen)

digits = [0, 2, 2]
print("sai :", wrong_leading_zero(digits))
print("đúng:", Solution().totalNumbers(digits))
assert wrong_leading_zero(digits) != Solution().totalNumbers(digits)
````

Vì sao sai: `022` là số có **hai** chữ số (giá trị `22`), không phải số có 3 chữ số; code sai đếm thêm nó nên ra `3` thay vì `2`. **Sửa:** thêm `a != 0` (hoặc `a` chạy từ `1`).

### Lỗi 3: không khử trùng (dùng `list`/đếm thay vì `set`)

````python
def wrong_no_dedup(digits):
    count = 0
    for a, b, c in permutations(digits, 3):  # hai số 2 ở hai vị trí khác nhau -> cùng một số bị đếm hai lần
        if a != 0 and c % 2 == 0:
            count += 1                       # SAI: đếm hoán vị, không đếm giá trị phân biệt
    return count

digits = [0, 2, 2]
print("sai :", wrong_no_dedup(digits))
print("đúng:", Solution().totalNumbers(digits))
assert wrong_no_dedup(digits) != Solution().totalNumbers(digits)
````

Vì sao sai: `202` và `220` mỗi số xuất hiện **hai lần** (hai chữ số `2` hoán đổi cho nhau) nên đếm ra `4`. Đề hỏi số giá trị **khác nhau**. **Sửa:** `set` (Cách 1, 2) hoặc cấu trúc đếm mỗi bộ chữ số đúng một lần (Cách 3, 4).

### Lỗi 4: quên điều kiện chữ số cuối chẵn

````python
def wrong_any_parity(digits):
    seen = set()
    for a, b, c in permutations(digits, 3):
        if a != 0:                           # SAI: thiếu c % 2 == 0
            seen.add(100 * a + 10 * b + c)
    return len(seen)

digits = [1, 3, 5]
print("sai :", wrong_any_parity(digits))
print("đúng:", Solution().totalNumbers(digits))
assert wrong_any_parity(digits) != Solution().totalNumbers(digits)
````

Vì sao sai: `[1,3,5]` ghép được `135, 153, 315, 351, 513, 531` (đều lẻ) nên code sai trả `6` trong khi đáp án là `0`. **Sửa:** thêm `c % 2 == 0`.

## 14. Mẹo làm bài nhanh & nhận dạng pattern

- `n <= 10` và ghép 3 phần tử -> **liệt kê** (brute force) là lời giải chuẩn; đừng cố nghĩ công thức.
- "Mỗi phần tử dùng tối đa một lần" -> dùng **chỉ số** (`i != j`) hoặc `permutations` trên **vị trí**, không phải trên giá trị.
- "Các số **khác nhau**" -> `set`.
- Số chẵn -> chữ số cuối chẵn; "không đầu 0" -> chữ số đầu `!= 0`.
- Bài họ hàng: `2094. Finding 3-Digit Even Numbers` (trả về **danh sách** các số, mảng dài tới 100); bài này chỉ đếm và `n <= 10`.
- Khi `n` lớn hơn, chuyển sang cách đếm chữ số (Cách 3, 4) vì không phụ thuộc `n`.

## 15. Biến thể của bài

1. **`2094. Finding 3-Digit Even Numbers`:** trả về mảng các số đó **đã sắp xếp** (cùng ý tưởng, mảng dài đến `100`).
2. Ghép số có `k` chữ số (thay `3` bằng `k`): dùng backtracking với mảng đếm.
3. Đếm số **chia hết cho 5**, hoặc số **nguyên tố**, thay vì chẵn.
4. Cho phép bắt đầu bằng `0` (chuỗi 3 chữ số): bỏ điều kiện `a != 0`, nhưng khi đó "khác nhau" tính theo chuỗi.
5. Đếm số cách (hoán vị) thay vì giá trị phân biệt: bỏ `set` (Lỗi 3 trở thành đáp án đúng).

## 16. So sánh các cách làm

| Cách | Ý tưởng | Time | Space | Khi nào dùng |
|---|---|---|---|---|
| 1. Ba vòng chỉ số + `set` | thử mọi bộ ba vị trí | `O(n³)` | `O(1)` (<= 450) | dễ nghĩ nhất |
| 2. `permutations` + `set` | thư viện sinh hoán vị | `O(n³)` | `O(1)` | gọn, dễ đọc |
| 3. Duyệt `100..998` + `Counter` | hỏi "ghép được số này không?" | `O(n + 450)` | `O(1)` | không phụ thuộc `n` |
