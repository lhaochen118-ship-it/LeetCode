## 3. Cách trực giác — dựng cả bàn cờ rồi tra

Dựng bàn cờ 8×8 theo quy tắc: `a1` đen, ô bên phải hoặc bên trên một ô đen là trắng (và ngược lại). Sau đó tra ô cần hỏi.

Ta dùng chỉ số 0-based: cột `x = 0..7` (a..h), hàng `y = 0..7` (1..8). Ô `(0, 0)` đen. Đi sang phải đổi màu so với ô bên trái; đi lên trên đổi màu so với ô bên dưới.

````python
def build_board():
    # board[y][x] = True nếu trắng
    board = [[False] * 8 for _ in range(8)]
    for y in range(8):
        for x in range(8):
            if x == 0 and y == 0:
                board[y][x] = False                  # a1 đen
            elif x == 0:
                board[y][x] = not board[y - 1][x]    # khác màu ô bên dưới
            else:
                board[y][x] = not board[y][x - 1]    # khác màu ô bên trái
    return board

BOARD = build_board()
for y in range(7, -1, -1):
    print(y + 1, " ".join("W" if BOARD[y][x] else "B" for x in range(8)))
print("  a b c d e f g h")
````

````python
def square_is_white_board(coordinates: str) -> bool:
    x = ord(coordinates[0]) - ord("a")
    y = int(coordinates[1]) - 1
    return BOARD[y][x]

for coord, expected in examples:
    print(coord, "->", square_is_white_board(coord), "| kỳ vọng", expected)
````

### 4.1. Nhìn quy luật trên lưới `(col + row) % 2`

In lưới giá trị `(col + row) % 2` rồi đặt cạnh bàn cờ đã dựng ở mục 3: chỗ nào là `1` thì đúng là ô trắng `W`. Đây là cách **kiểm chứng insight bằng dữ liệu** trước khi tin vào nó.

````python
for y in range(8, 0, -1):
    parity_row = " ".join(str((x + y) % 2) for x in range(1, 9))
    board_row = " ".join("W" if BOARD[y - 1][x - 1] else "B" for x in range(1, 9))
    print(y, parity_row, "   |  ", board_row)
print("  a b c d e f g h        a b c d e f g h")
match = all(((x + y) % 2 == 1) == BOARD[y - 1][x - 1] for x in range(1, 9) for y in range(1, 9))
print("Quy luật khớp toàn bộ bàn cờ?", match)
````

## 5. Lời giải chính — kiểm tra chẵn lẻ của tổng

**Bước 1.** `col = ord(coordinates[0]) - ord('a') + 1`.

**Bước 2.** `row = int(coordinates[1])`.

**Bước 3.** `return (col + row) % 2 == 1`.

````python
def square_is_white(coordinates: str) -> bool:
    col = ord(coordinates[0]) - ord("a") + 1
    row = int(coordinates[1])
    return (col + row) % 2 == 1

for coord, expected in examples:
    print(coord, "->", square_is_white(coord), "| kỳ vọng", expected)
````

## 6. TRACE từng bước

| coordinates | coordinates[0] | ord | col | coordinates[1] | row | col+row | %2 | Kết quả |
|---|---|---|---|---|---|---|---|---|
| `"a1"` | a | 97 | 1 | `'1'` | 1 | 2 | 0 | False |
| `"h3"` | h | 104 | 8 | `'3'` | 3 | 11 | 1 | True |
| `"c7"` | c | 99 | 3 | `'7'` | 7 | 10 | 0 | False |

````python
def trace_square_is_white(coordinates: str) -> bool:
    letter, digit = coordinates
    col = ord(letter) - ord("a") + 1
    row = int(digit)
    total = col + row
    result = total % 2 == 1
    print(f"{coordinates}: ord({letter!r})={ord(letter)} -> col={col}; int({digit!r})={row}; "
          f"col+row={total}; {total}%2={total % 2} -> {'trắng' if result else 'đen'} ({result})")
    return result

for coord, _ in examples:
    trace_square_is_white(coord)
````

## 7. Các lời giải khác

### Cách 3 — Cộng thẳng mã `ord` của hai ký tự

`ord('a') = 97` (lẻ) và `ord('1') = 49` (lẻ). Vì vậy:

- `ord(letter) = 96 + col` → cùng chẵn lẻ với `col` **cộng thêm** một số chẵn (96) → `ord(letter) ≡ col (mod 2)`.
- `ord(digit) = 48 + row` → `ord(digit) ≡ row (mod 2)`.

Do đó `(ord(letter) + ord(digit)) % 2 == (col + row) % 2`.

````python
def square_is_white_ord(coordinates: str) -> bool:
    return (ord(coordinates[0]) + ord(coordinates[1])) % 2 == 1

print("ord('a') - 1 =", ord("a") - 1, "| ord('1') - 1 =", ord("1") - 1)
for coord, expected in examples:
    print(coord, "->", square_is_white_ord(coord), "| kỳ vọng", expected)
````

### Cách 4 — So chẵn lẻ của cột và hàng riêng

Ô trắng ⇔ cột và hàng **khác** tính chẵn lẻ. Cột lẻ là `a, c, e, g`; hàng lẻ là `1, 3, 5, 7`.

````python
def square_is_white_sets(coordinates: str) -> bool:
    col_odd = coordinates[0] in "aceg"
    row_odd = coordinates[1] in "1357"
    return col_odd != row_odd

for coord, expected in examples:
    print(coord, "->", square_is_white_sets(coord), "| kỳ vọng", expected)
````

`col_odd != row_odd` với hai giá trị `bool` chính là phép **XOR** (đúng khi đúng một trong hai đúng).

### 7.1. Bảng so sánh

| Cách | Ý tưởng | Time | Space | Ghi chú |
|---|---|---|---|---|
| 1. Dựng bàn cờ | mô phỏng 8×8 | `O(64)` = `O(1)` | `O(64)` | trực quan, không mở rộng được |
| **2. (col + row) % 2 (chính)** | parity | `O(1)` | `O(1)` | rõ ràng nhất |

## 8. Code nộp LeetCode

````python
class Solution:
    def squareIsWhite(self, coordinates: str) -> bool:
        col = ord(coordinates[0]) - ord("a") + 1
        row = int(coordinates[1])
        return (col + row) % 2 == 1

sol = Solution()
for coord, expected in examples:
    print(sol.squareIsWhite(coord), "| kỳ vọng", expected)
````

## 9. Vì sao thuật toán đúng?

Gọi `f(col, row) = (col + row) % 2`.

- **Cơ sở:** `a1` có `f = (1 + 1) % 2 = 0` và là ô đen (theo bàn cờ chuẩn). ✓
- **Bước chuyển:** hai ô kề nhau theo cạnh (trái-phải hoặc trên-dưới) luôn khác màu trên bàn cờ; đồng thời `col + row` của chúng chênh nhau đúng 1 → `f` khác nhau.
- Mọi ô đều đến được từ `a1` bằng một chuỗi bước kề. Theo quy nạp trên số bước, ô có `f = 0` là đen, `f = 1` là trắng.

Vậy `squareIsWhite` ⇔ `f == 1`. Cách 3 đúng vì `ord(letter) ≡ col` và `ord(digit) ≡ row (mod 2)` như đã chỉ ra.

### Lỗi 4: Dùng 0-based cho một tọa độ và 1-based cho tọa độ kia

`col = ord(c) - ord('a')` (0..7) nhưng `row = int(d)` (1..8) → tổng lệch 1 → **ngược màu toàn bộ**. Hoặc dùng 0-based cho cả hai (`row - 1`), hoặc 1-based cho cả hai — lệch đều cả hai thì parity của tổng không đổi.

````python
def wrong_mixed_base(coordinates):
    col = ord(coordinates[0]) - ord("a")       # 0-based
    row = int(coordinates[1])                  # 1-based
    return (col + row) % 2 == 1

def ok_both_zero_based(coordinates):
    col = ord(coordinates[0]) - ord("a")
    row = int(coordinates[1]) - 1
    return (col + row) % 2 == 1

print("a1 sai:", wrong_mixed_base("a1"), "| 0-based cả hai:", ok_both_zero_based("a1"))
````

### 14.2. Bàn cờ `N × M` với ô góc `(0, 0)` có màu tùy ý

Cho `corner_white` là màu ô `(0, 0)`. Ô `(r, c)` (0-based) cùng màu góc khi `(r + c)` chẵn.

````python
def is_white_general(r: int, c: int, corner_white: bool) -> bool:
    same_as_corner = (r + c) % 2 == 0
    return corner_white if same_as_corner else not corner_white

print(is_white_general(0, 0, False), is_white_general(10**9, 1, False), is_white_general(3, 3, True))
````
