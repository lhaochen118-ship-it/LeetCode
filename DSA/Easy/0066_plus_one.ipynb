## 4. Vì sao cộng từ phải sang trái?

Hãy tính `1299 + 1` bằng tay: 9+1=10 → ghi 0, **nhớ 1** sang trái; 9+1=10 → ghi 0, nhớ 1; 2+1=3 → ghi 3, **hết nhớ**; chữ số 1 không đổi.

Vậy `[1,2,9,9]` biến thành `[1,3,0,0]`.

**Carry** (số nhớ) ở bài cộng 1 chỉ có thể là 0 hoặc 1. Nếu gặp chữ số <9, tăng lên 1 và **dừng ngay**. Nếu là 9, đổi 9 thành 0 và tiếp tục sang trái.

````python
digits=[1,2,9,9]
print("trước",digits)
digits[3]=0; digits[2]=0; digits[1]+=1
print("sau",digits)
````

````python
digits=[1,2,3,4]
for i in range(len(digits)-1,-1,-1):
    print("index",i,"chữ số",digits[i])
````

## 6. Trường hợp thường: chữ số cuối chưa phải 9

Nếu `digits[i] < 9`, tăng nó một đơn vị rồi `return digits` ngay. Tại sao đúng? Vì không có phép nhớ; các chữ số bên trái **giữ nguyên**. Ví dụ `[3,1,2]` → `[3,1,3]`.

````python
digits=[3,1,2]
if digits[-1]<9:
    digits[-1]+=1
print(digits)
````

## 7. Trường hợp dồn nhớ qua nhiều số 9

Nếu chữ số cuối là 9, thay nó bằng 0 và mang 1 sang vị trí kế bên. Cứ tiếp tục cho tới khi gặp một chữ số <9. Trường hợp **tất cả chữ số đều 9**, vòng lặp đã đổi chúng thành 0, phải tạo thêm chữ số 1 ở đầu: `[9,9]` → `[1,0,0]`.

````python
digits=[9,9,9]
for i in range(len(digits)-1,-1,-1):
    if digits[i]==9:
        digits[i]=0
print("Sau khi đi qua tất cả số 9:",digits)
print("Thêm chữ số đầu:",[1]+digits)
````

## 8. Code chính: ngắn nhưng không nhảy tư duy

Code dưới chỉ thực thi quy tắc chúng ta tự rút ra:

- `digits[i] < 9` → tăng và trả ngay.
- `digits[i] == 9` → đổi về 0, chạy tiếp.
- Nếu đi qua **cả mảng** mà chưa `return`, tất cả đều là 9; trả `[1] + digits`.

````python
def plus_one_digits(digits:list[int])->list[int]:
    for i in range(len(digits)-1,-1,-1):
        if digits[i]<9:
            digits[i]+=1
            return digits
        digits[i]=0
    return [1]+digits

for inp in [[1,2,3],[1,2,9,9],[9,9,9]]:
    print(inp,"+ 1 ->",plus_one_digits(inp.copy()))
````

## 9. Bảng trace: `[1,2,9,9]`

| bước | i | digits[i] trước | hành động | digits sau |
|---:|---:|---:|---|---|
| 1 | 3 | 9 | đổi 0, còn carry | `[1,2,9,0]` |
| 2 | 2 | 9 | đổi 0, còn carry | `[1,2,0,0]` |
| 3 | 1 | 2 | tăng 3, hết carry; return | `[1,3,0,0]` |

**Bất biến:** mọi vị trí đã duyệt ở bên phải đều đã được xử lý đúng (nếu chưa dừng thì chúng là các số 9 chuyển thành 0).

````python
def trace_plus_one(original):
    digits=original.copy()
    print("Input:",digits)
    for i in range(len(digits)-1,-1,-1):
        before=digits[i]
        if before<9:
            digits[i]+=1
            print(f"i={i}: {before}<9 -> +1, dừng: {digits}")
            return digits
        digits[i]=0
        print(f"i={i}: {before}=9 -> đặt 0, còn nhớ 1: {digits}")
    print("Tất cả là 9, thêm 1 đầu:",[1]+digits)
    return [1]+digits
assert trace_plus_one([1,2,9,9])==[1,3,0,0]
````

## 10. Trường hợp đặc biệt: một 9 và toàn 9

`[9]` phải thành `[1,0]`: số phần tử tăng thêm 1. `[9,9,9]` phải thành `[1,0,0,0]`, không phải `[0,0,0]` hay `[1,0,0]`. Từng vị trí mới đều có ý nghĩa hàng chục/hàng trăm; nếu thiếu số 1, giá trị sẽ sai hoàn toàn.

````python
assert plus_one_digits([9])==[1,0]
assert plus_one_digits([9,9,9])==[1,0,0,0]
assert plus_one_digits([0])==[1]
print("Các trường hợp biên: PASS")
````

## 11. Vì sao thuật toán đúng?

Ở mỗi bước có đúng hai tình huống:

1. Chữ số đang xét <9: tăng lên 1 không vượt quá 9; không còn nhớ, các vị trí chưa xét không đổi → mảng là số ban đầu +1.
2. Chữ số đang xét =9: 9+1 làm chữ số hiện tại thành 0 và chuyển 1 sang trái → bảo toàn phép cộng.

Nếu tất cả đều bằng 9, sau khi xử lý tất cả, carry 1 vẫn còn ở bên trái nên đặt thêm chữ số 1 là cần thiết và đủ.

## 12. Độ phức tạp và bộ nhớ

`n=len(digits)`.

- **Time O(n)** tệ nhất: khi mọi chữ số đều là 9 phải đi qua n vị trí.
- **Auxiliary Space O(1)** trên nhánh không cần thêm chữ số: ta sửa list cũ ngay tại chỗ.
- Với toàn số 9, biểu thức `[1]+digits` tạo **list kết quả mới O(n)**. Nếu tính cả vùng nhớ dành cho output, nhánh này cần O(n); nhiều lời giải gọi O(1) **bộ nhớ phụ ngoài kết quả**, nhưng Python vẫn cấp phát mảng mới trong nhánh ấy.

## 13. Class Solution để submit

LeetCode tự truyền `digits` vào phương thức `plusOne`. Ta được phép sửa `digits` tại chỗ; nhưng **bắt buộc trả về** list chữ số.

````python
class Solution:
    def plusOne(self, digits:list[int])->list[int]:
        for i in range(len(digits)-1,-1,-1):
            if digits[i]<9:
                digits[i]+=1
                return digits
            digits[i]=0
        return [1]+digits

print(Solution().plusOne([4,3,2,1]))
````
