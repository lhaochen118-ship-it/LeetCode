## 1. Đề bài

Alice có `n` viên kẹo, viên thứ `i` thuộc **loại** `candyType[i]`. Bác sĩ khuyên Alice chỉ ăn **`n / 2`** viên (`n` luôn chẵn). Alice muốn ăn được **càng nhiều loại khác nhau càng tốt**.

Trả về **số loại khác nhau lớn nhất** Alice có thể ăn khi chỉ ăn `n / 2` viên.

| `candyType` | `n/2` | Số loại | Output | Giải thích |
|---|---:|---:|---:|---|
| `[1,1,2,2,3,3]` | 3 | 3 | 3 | ăn mỗi loại một viên |
| `[1,1,2,3]` | 2 | 3 | 2 | chỉ được ăn 2 viên → tối đa 2 loại |
| `[6,6,6,6]` | 2 | 1 | 1 | ăn 2 viên nhưng chỉ có 1 loại |

**Constraints:** `n == candyType.length`, `2 <= n <= 10^4`, `n` chẵn, `-10^5 <= candyType[i] <= 10^5`.

Chú ý: "loại" là một số nguyên bất kỳ (có thể âm) — chỉ dùng để phân biệt, không có ý nghĩa độ lớn.

## 2. Nền tảng Python: `set` (tập hợp)

**Set** là cấu trúc lưu các phần tử **không trùng lặp**, không có thứ tự.

- Tạo: `set()` (rỗng) hoặc `set(list)` (bỏ trùng từ list), hoặc `{1, 2, 3}`.
- Thêm: `s.add(x)` (thêm phần tử đã có thì không đổi gì).
- Kiểm tra: `x in s` — trung bình `O(1)` nhờ **bảng băm** (hash table).
- Kích thước: `len(s)`.

Lưu ý: `{}` là **dict rỗng**, không phải set rỗng.

````python
candy = [1, 1, 2, 2, 3, 3]
kinds = set(candy)
print(kinds, len(kinds))

s = set()
s.add(5)
s.add(5)          # thêm lần nữa không đổi
s.add(-2)
print(s, len(s), 5 in s, 7 in s)
print(type({}), type(set()))
````

### 2.1. Vì sao `x in set` nhanh hơn `x in list`?

- `x in list`: duyệt từng phần tử → `O(n)`.
- `x in set`: tính `hash(x)` để nhảy thẳng tới "ngăn" chứa nó → trung bình `O(1)`.

Đo thử với 10^5 phần tử:

````python
import time

big_list = list(range(100_000))
big_set = set(big_list)

t0 = time.perf_counter()
for _ in range(200):
    99_999 in big_list
t1 = time.perf_counter()
for _ in range(200):
    99_999 in big_set
t2 = time.perf_counter()
print(f"list: {t1 - t0:.4f}s | set: {t2 - t1:.6f}s")
print("set nhanh hơn list:", (t1 - t0) > (t2 - t1))
````

`time.perf_counter()` trả về thời điểm hiện tại (giây) với độ chính xác cao — hiệu hai lần gọi là thời gian chạy. Con số cụ thể tùy máy, nhưng chênh lệch luôn rất lớn.

## 3. Phân tích: hai "trần" của đáp án

Gọi `k` = số loại khác nhau, `half = n // 2`.

1. **Trần do bác sĩ:** chỉ ăn `half` viên → không thể ăn quá `half` loại.
2. **Trần do số loại:** có tất cả `k` loại → không thể ăn quá `k` loại.

⇒ Đáp án **≤ min(k, half)**.

Và luôn **đạt được**:
- Nếu `k ≤ half`: ăn mỗi loại 1 viên (`k` viên), số viên còn lại `half − k` ăn bất kỳ → đủ `k` loại. Có đủ kẹo không? Tổng có `n = 2·half ≥ half` viên ✔.
- Nếu `k > half`: chọn `half` loại khác nhau, mỗi loại 1 viên → `half` loại.

⇒ **Đáp án = min(k, half)**.

````python
for candy in [[1, 1, 2, 2, 3, 3], [1, 1, 2, 3], [6, 6, 6, 6]]:
    k = len(set(candy))
    half = len(candy) // 2
    print(f"{candy}: k={k}, half={half} -> min = {min(k, half)}")
````

## 4. Cách 1 — Brute force: thử mọi cách chọn `n/2` viên

Để **tin** công thức, ta viết brute force cho mảng nhỏ: thử mọi tổ hợp `n/2` vị trí, đếm số loại, lấy max.

`itertools.combinations(iterable, r)` sinh mọi cách chọn `r` phần tử (không quan tâm thứ tự). Số tổ hợp `C(n, n/2)` tăng rất nhanh (n=20 → 184756) nên chỉ dùng cho `n` nhỏ.

````python
from itertools import combinations

print(list(combinations([1, 1, 2, 3], 2)))
````

````python
def distribute_brute(candyType: list[int]) -> int:
    half = len(candyType) // 2
    best = 0
    for chosen in combinations(candyType, half):
        best = max(best, len(set(chosen)))
    return best

print(distribute_brute([1, 1, 2, 2, 3, 3]))
print(distribute_brute([1, 1, 2, 3]))
print(distribute_brute([6, 6, 6, 6]))
````

### Nút thắt

Brute force thử `C(n, n/2)` tổ hợp — với `n = 10^4` là con số khổng lồ. Nhưng phân tích ở mục 3 cho thấy ta **không cần biết chọn viên nào**, chỉ cần **đếm số loại `k`**. Bài toán chuyển thành: **đếm số phần tử khác nhau** trong mảng.

## 5. Cách 2 — Đếm số loại bằng sắp xếp (không dùng hash)

Sau khi sắp xếp, các viên cùng loại đứng **liền nhau**. Số loại = 1 + số vị trí `i` mà `a[i] != a[i-1]`.

````python
def distribute_sort(candyType: list[int]) -> int:
    a = sorted(candyType)
    kinds = 1
    for i in range(1, len(a)):
        if a[i] != a[i - 1]:
            kinds += 1
    return min(kinds, len(a) // 2)

print(distribute_sort([1, 1, 2, 2, 3, 3]), distribute_sort([1, 1, 2, 3]), distribute_sort([6, 6, 6, 6]))
````

````python
def trace_sort(candyType):
    a = sorted(candyType)
    print("sau sắp xếp:", a)
    kinds = 1
    print(f"i=0: a[0]={a[0]} -> loại mới, kinds={kinds}")
    for i in range(1, len(a)):
        new = a[i] != a[i - 1]
        kinds += new
        print(f"i={i}: a[i]={a[i]}, a[i-1]={a[i - 1]} -> {'loại mới' if new else 'trùng'}, kinds={kinds}")
    print(f"kết quả = min({kinds}, {len(a) // 2}) = {min(kinds, len(a) // 2)}")

trace_sort([3, 1, 2, 1, -4, 3])
````

`kinds += new` với `new` là `True/False`: trong Python `True == 1`, `False == 0` nên phép cộng hoạt động.

Độ phức tạp: `O(n log n)` (sắp xếp), bộ nhớ `O(n)` cho `sorted` (hoặc `O(1)` thêm nếu dùng `.sort()` tại chỗ và được phép sửa input).

## 6. Cách 3 — Dùng `set` (lời giải chính)

## 7. Trace Cách 3 (viết tường minh từng bước thêm vào set)

`candyType = [1, 1, 2, 3, 3, 3, 4, 5]` → `n = 8`, `half = 4`.

| i | loại | set trước | Đã có? | set sau | len |
|---:|---:|---|:---:|---|---:|
| 0 | 1 | {} | không | {1} | 1 |
| 1 | 1 | {1} | có | {1} | 1 |
| 2 | 2 | {1} | không | {1,2} | 2 |
| 3 | 3 | {1,2} | không | {1,2,3} | 3 |
| 4–5 | 3 | | có | | 3 |
| 6 | 4 | {1,2,3} | không | {1,2,3,4} | 4 |
| 7 | 5 | | không | {1,2,3,4,5} | 5 |

`min(5, 4) = 4`.

````python
def trace_set(candyType: list[int]) -> int:
    seen = set()
    for i, t in enumerate(candyType):
        before = set(seen)
        existed = t in seen
        seen.add(t)
        print(f"i={i}: loại {t:>2} | trước={sorted(before)} | {'đã có' if existed else 'mới  '} | len={len(seen)}")
    half = len(candyType) // 2
    print(f"kết quả = min({len(seen)}, {half}) = {min(len(seen), half)}")
    return min(len(seen), half)

trace_set([1, 1, 2, 3, 3, 3, 4, 5])
````

## 9. Cách 4 — Set + dừng sớm khi đã đủ `n/2` loại

Đáp án không bao giờ vượt `half`. Khi `seen` đã có `half` loại, đọc thêm cũng vô ích → **dừng sớm**. Hữu ích khi mảng rất dài và đa dạng.

````python
def distribute_early_stop(candyType: list[int]) -> int:
    half = len(candyType) // 2
    seen = set()
    for t in candyType:
        seen.add(t)
        if len(seen) == half:
            return half
    return len(seen)

print(distribute_early_stop([1, 1, 2, 2, 3, 3]), distribute_early_stop([1, 1, 2, 3]), distribute_early_stop([6, 6, 6, 6]))

steps = 0
def count_steps(candyType):
    global steps
    half, seen = len(candyType) // 2, set()
    for t in candyType:
        steps += 1
        seen.add(t)
        if len(seen) == half:
            return half
    return len(seen)

big = list(range(10_000))          # 10^4 loại khác nhau
count_steps(big)
print("Mảng 10^4 loại khác nhau: chỉ đọc", steps, "phần tử trước khi dừng")
````

### Code nộp LeetCode — Cách 4

````python
class Solution:
    def distributeCandies(self, candyType: list[int]) -> int:
        half = len(candyType) // 2
        seen = set()
        for t in candyType:
            seen.add(t)
            if len(seen) == half:
                return half
        return len(seen)

print(Solution().distributeCandies([1, 1, 2, 3]))
````

### Biến thể không dùng hash: mảng đánh dấu (vì loại nằm trong `[-10^5, 10^5]`)

Giá trị bị chặn → dùng list `bool` kích thước `2·10^5 + 1` với **offset** `10^5` để chỉ số không âm. Thời gian `O(n + R)` với `R = 2·10^5 + 1`, không phụ thuộc vào hàm băm.

````python
def distribute_bitmap(candyType: list[int]) -> int:
    OFFSET = 100_000
    seen = [False] * (2 * OFFSET + 1)
    kinds = 0
    for t in candyType:
        if not seen[t + OFFSET]:
            seen[t + OFFSET] = True
            kinds += 1
    return min(kinds, len(candyType) // 2)

print(distribute_bitmap([1, 1, 2, 2, 3, 3]), distribute_bitmap([-100000, 100000, 0, 0]))
````

## 10. Vì sao đúng?

Đã chứng minh ở mục 3:
- **Cận trên:** số loại ăn được ≤ số viên ăn được = `half`, và ≤ tổng số loại = `k`. Vậy ≤ `min(k, half)`.
- **Đạt được:** chọn `min(k, half)` loại khác nhau, mỗi loại một viên; nếu còn thiếu viên (khi `k < half`), bù bằng viên bất kỳ còn lại — luôn đủ vì tổng có `n = 2·half` viên. Số loại ăn được ít nhất `min(k, half)`.

Cách 4: nếu dừng sớm thì `len(seen) == half` ⇒ `k ≥ half` ⇒ `min(k, half) = half` ✔. Nếu không dừng, `len(seen) = k < half` ⇒ đáp án `k` ✔. (`len(seen)` tăng từng 1 nên không thể "nhảy qua" `half`.)

## 11. So sánh và độ phức tạp

Gọi `n = len(candyType)`.

| Cách | Time | Space | Ghi chú |
|---|---|---|---|
| 1. Brute force tổ hợp | `O(C(n, n/2) · n)` | `O(n)` | chỉ để kiểm chứng |
| 2. Sắp xếp + đếm | `O(n log n)` | `O(n)` (`sorted`) | không cần hash |
| 3. `set` | `O(n)` trung bình | `O(n)` | **ngắn nhất, khuyên dùng** |
| 4. `set` + dừng sớm | `O(n)` xấu nhất, thường nhanh hơn | `O(min(n, k))` | tốt khi mảng đa dạng |
| 4b. Mảng đánh dấu | `O(n + R)` | `O(R)` | không phụ thuộc hash |

### Lỗi 5: Đếm loại bằng `list` + `if x not in list`

Đúng nhưng `x not in list` tốn `O(k)` mỗi lần → tổng `O(n·k)`, có thể tới `O(n²)` — chậm với `n = 10^4` khi nhiều loại.

````python
def slow_count_kinds(candy):
    kinds = []
    for t in candy:
        if t not in kinds:          # O(len(kinds)) mỗi lần
            kinds.append(t)
    return len(kinds)

data = list(range(5000)) * 2
t0 = time.perf_counter(); a = slow_count_kinds(data); t1 = time.perf_counter()
b2 = len(set(data)); t2 = time.perf_counter()
print(f"list: {a} loại trong {t1 - t0:.3f}s | set: {b2} loại trong {t2 - t1:.5f}s")
````

## 14. Mẹo nhận dạng pattern và biến thể

### Mẹo
- "Đếm số phần tử **khác nhau**" → `len(set(...))`.
- Đáp án bị chặn bởi **nhiều giới hạn độc lập** → đáp án thường là `min(...)` của các giới hạn; nhớ chứng minh **đạt được**.
- Bài Easy có vẻ phải "chọn" tổ hợp → thử viết brute force nhỏ để tìm công thức đóng.

### Biến thể
1. **LeetCode 217 – Contains Duplicate**: `len(set(nums)) < len(nums)`.
2. **LeetCode 1103 – Distribute Candies to People**: mô phỏng phát kẹo theo vòng (khác hẳn bài này).
3. **LeetCode 349 – Intersection of Two Arrays**: giao hai set.
4. Biến thể tự chế: Alice được ăn `m` viên (không nhất thiết `n/2`) → `min(k, m)`.

````python
# Biến thể 3 demo: giao hai mảng (phần tử chung, không trùng)
print(sorted(set([4, 9, 5]) & set([9, 4, 9, 8, 4])))
# Biến thể 1 demo
print(len(set([1, 2, 3, 1])) < len([1, 2, 3, 1]))
````

`&` giữa hai set là phép **giao**; `|` là hợp; `-` là hiệu.

## 17. Đáp án tham khảo (tự làm trước rồi mới xem)

````python
from collections import Counter

def answer_count_kinds(candy: list[int]) -> int:
    d = {}
    for t in candy:
        d[t] = True
    return len(d)

def answer_distribute_m(candy: list[int], m: int) -> int:
    return min(len(set(candy)), m)

def answer_alice_choice(candy: list[int]) -> list[int]:
    half = len(candy) // 2
    chosen, used_idx, seen = [], set(), set()
    for i, t in enumerate(candy):                 # lượt 1: mỗi loại một viên
        if len(chosen) == half:
            break
        if t not in seen:
            seen.add(t)
            chosen.append(t)
            used_idx.add(i)
    for i, t in enumerate(candy):                 # lượt 2: bù cho đủ half viên
        if len(chosen) == half:
            break
        if i not in used_idx:
            chosen.append(t)
            used_idx.add(i)
    return chosen

def answer_brother_kinds(candy: list[int]) -> int:
    rest = Counter(candy) - Counter(answer_alice_choice(candy))
    return len(rest)

def answer_max_kinds_first_person(candy: list[int], p: int) -> int:
    return min(len(set(candy)), len(candy) // p)

assert answer_count_kinds([1, 1, 2, 3]) == 3
assert answer_count_kinds([6, 6, 6, 6]) == 1
assert answer_distribute_m([1, 1, 2, 3], 3) == 3
assert answer_distribute_m([1, 1, 2, 3], 1) == 1
choice = answer_alice_choice([1, 1, 2, 3])
assert len(choice) == 2 and len(set(choice)) == 2
assert answer_brother_kinds([6, 6, 6, 6]) == 1
assert answer_max_kinds_first_person([1, 2, 3, 4, 5, 6], 3) == 2
assert answer_max_kinds_first_person([1, 1, 1, 2, 2, 2], 2) == 2
for _ in range(300):
    candy = [random.randint(-5, 5) for _ in range(2 * random.randint(1, 20))]
    ch = answer_alice_choice(candy)
    assert len(ch) == len(candy) // 2
    assert len(set(ch)) == distribute_set(candy)
    assert not (Counter(ch) - Counter(candy))      # Alice chỉ ăn kẹo có thật
print("Đáp án chạy PASS toàn bộ test!")
````

`Counter(a) - Counter(b)` trừ số lượng từng key và **bỏ** các key có kết quả ≤ 0. `not (Counter(ch) - Counter(candy))` nghĩa là `ch` không dùng loại nào nhiều hơn số kẹo có thật.
