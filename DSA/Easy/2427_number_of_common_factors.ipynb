| Ví dụ | a | b | Output | Giải thích |

### 3.3. Ước chung lớn nhất `gcd` và quan hệ với ước chung

`math.gcd(a, b)` là ước chung **lớn nhất**. Tính chất quan trọng: **mọi ước chung của `a` và `b` đều là ước của `gcd(a, b)`, và ngược lại**. Nên số ước chung = số ước của `gcd`.

````python
from math import gcd
print(gcd(12, 6), gcd(25, 30), gcd(7, 13), gcd(36, 36))
g = gcd(12, 18)
print("gcd =", g, "| ước của gcd:", [x for x in range(1, g + 1) if g % x == 0])
print("ước chung trực tiếp:", [x for x in range(1, 19) if 12 % x == 0 and 18 % x == 0])
````

### 3.4. Ước đi theo **cặp**

Nếu `i` là ước của `g` thì `g // i` cũng là ước. Cặp `(i, g // i)`: một số `<= sqrt(g)`, một số `>= sqrt(g)`. Vì vậy chỉ cần duyệt `i` đến `sqrt(g)`. Ngoại lệ: khi `i * i == g` thì hai số trong cặp **trùng nhau** và chỉ được đếm **một lần**.

````python
g = 36
for i in range(1, 7):
    if g % i == 0:
        print(f"i={i}: cặp ({i}, {g // i})", "<- trùng nhau" if i == g // i else "")
````

### 3.5. `isqrt`

`math.isqrt(n)` cho phần nguyên căn bậc hai (số nguyên, không sai số thực như `int(n ** 0.5)` với số lớn).

````python
from math import isqrt
print(isqrt(36), isqrt(35), isqrt(1), isqrt(1000))
````

### 3.6. Số ước từ phân tích thừa số nguyên tố

Nếu `g = p1^e1 · p2^e2 · ...` thì số ước là `(e1 + 1)(e2 + 1)...`. Ví dụ `12 = 2^2 · 3^1` có `(2+1)(1+1) = 6` ước (`1,2,3,4,6,12`). Để phân tích: chia liên tiếp cho `2, 3, 4...` và đếm số lần chia hết.

````python
def factorize(n):
    result, p = {}, 2
    while p * p <= n:
        while n % p == 0:
            result[p] = result.get(p, 0) + 1
            n //= p
        p += 1
    if n > 1:
        result[n] = result.get(n, 0) + 1
    return result

print(factorize(12), factorize(36), factorize(97), factorize(1))
````

### 3.7. Thuật toán Euclid tính `gcd` (để hiểu hàm có sẵn làm gì)

Euclid dựa trên `gcd(a, b) = gcd(b, a % b)` và `gcd(a, 0) = a`. Mỗi bước số nhỏ đi ít nhất một nửa sau hai bước nên chỉ cần `O(log)` bước. Bạn có thể tự cài để thấy từng bước, nhưng khi nộp bài hãy dùng `math.gcd`.

````python
def my_gcd(a, b):
    while b:
        print(f"  gcd({a}, {b}) -> gcd({b}, {a % b})")
        a, b = b, a % b
    return a

print("kết quả:", my_gcd(48, 18), "| math.gcd:", gcd(48, 18))
````

## 4. Ý tưởng nền tảng / cách nghĩ trực giác

Muốn tìm ước chung, ta lần lượt thử từng số `x` và hỏi: "`x` có chia hết cả `a` lẫn `b` không?". Ước chung không thể lớn hơn **số nhỏ hơn** trong hai số (vì ước của `n` không vượt quá `n`), nên chỉ cần thử `x` từ `1` đến `min(a, b)`.

Ví dụ `a = 12`, `b = 6` (`min = 6`):

```
x=1: 12%1=0, 6%1=0 ✓
x=2: 12%2=0, 6%2=0 ✓
x=3: 12%3=0, 6%3=0 ✓
x=4: 12%4=0, 6%4=2 ✗
x=5: 12%5=2          ✗
x=6: 12%6=0, 6%6=0 ✓
đếm = 4
```

Ví dụ `25, 30`: chỉ `1` và `5` thoả → `2`. Nhận xét thú vị: các ước chung `1, 2, 3, 6` của `12` và `6` chính là các ước của `gcd(12, 6) = 6`.

## 5. Cách 1 — Brute Force / baseline

Thử mọi `x` từ `1` đến `min(a, b)`, đếm những `x` chia hết cả hai số. Cực kỳ trực tiếp, đúng định nghĩa đề bài.

Nút thắt: số vòng lặp bằng `min(a, b)`; với `a, b <= 1000` thì quá nhẹ, nhưng nếu `a, b` lên `10^12` thì không khả thi.

````python
class SolutionBrute:
    def commonFactors(self, a: int, b: int) -> int:
        count = 0
        for x in range(1, min(a, b) + 1):
            if a % x == 0 and b % x == 0:
                count += 1
        return count

print(SolutionBrute().commonFactors(12, 6))
print(SolutionBrute().commonFactors(25, 30))
````

**Độ phức tạp:** Time `O(min(a, b))`, Space `O(1)`. Nút thắt: kiểm tra chia hết **hai lần** cho mỗi `x`, và duyệt cả những `x` không thể là ước chung (lớn hơn `gcd`).

## 6. Ý tưởng thuật toán / bước đột phá

Mọi ước chung của `a` và `b` đều là ước của `g = gcd(a, b)` (và ngược lại). Vậy:

> **số ước chung = số ước của `g`**.

Chỉ cần đếm ước của **một** số `g`, nhỏ hơn hoặc bằng `min(a, b)`. Sau đó còn có thể tăng tốc thêm:

- Duyệt `x` từ `1` đến `g` (kiểm tra `g % x`) — bớt một phép chia mỗi vòng.
- Duyệt `i` chỉ đến `sqrt(g)` và đếm cặp `(i, g // i)`.
- Hoặc phân tích `g` ra thừa số nguyên tố và nhân `(e + 1)`.

**Quy tắc "nếu… thì…":**

- Nếu `g % i == 0` → có thêm ước `i` và `g // i`.
- Nếu `i * i == g` → chỉ cộng **1** (hai ước trùng nhau).
- Nếu `g == 1` → đáp án `1`.

Invariant: sau khi xét `i` từ `1` đến `k`, biến đếm bằng đúng số ước của `g` mà có ít nhất một thành viên trong cặp `(i, g // i)` với `i <= k`.

````python
from math import gcd
a, b = 12, 6
g = gcd(a, b)
print("g =", g)
print("số ước của g:", sum(1 for x in range(1, g + 1) if g % x == 0))
print("số ước chung trực tiếp:", sum(1 for x in range(1, min(a, b) + 1) if a % x == 0 and b % x == 0))
````

## 7. Các cách tiếp theo

Ta có 4 cách: (1) thử mọi `x` tới `min`, (2) `gcd` + duyệt tới `g`, (3) `gcd` + duyệt tới `sqrt(g)`, (4) `gcd` + thừa số nguyên tố.

### Cách 2 — `gcd` rồi đếm ước của `g`

Đổi bài toán sang một số: `g = gcd(a, b)`, đếm `x` trong `1..g` mà `g % x == 0`.

````python
from math import gcd, isqrt

class SolutionGcd:
    def commonFactors(self, a: int, b: int) -> int:
        g = gcd(a, b)
        return sum(1 for x in range(1, g + 1) if g % x == 0)

print(SolutionGcd().commonFactors(12, 6), SolutionGcd().commonFactors(25, 30))
````

Độ phức tạp Cách 2: Time `O(gcd(a, b))` (nhỏ hơn hoặc bằng `min(a, b)`) cộng chi phí tính `gcd` `O(log min(a, b))`; Space `O(1)`.

### Cách 3 — `gcd` + duyệt tới `sqrt(g)` (đếm theo cặp)

Duyệt `i` từ `1` đến `isqrt(g)`. Mỗi `i` là ước thì cộng `2` (cho `i` và `g // i`), trừ khi `i * i == g` thì cộng `1`.

````python
class SolutionSqrt:
    def commonFactors(self, a: int, b: int) -> int:
        g = gcd(a, b)
        count = 0
        for i in range(1, isqrt(g) + 1):
            if g % i == 0:
                count += 1 if i * i == g else 2
        return count

print(SolutionSqrt().commonFactors(12, 6), SolutionSqrt().commonFactors(25, 30), SolutionSqrt().commonFactors(36, 36))
````

Độ phức tạp Cách 3: Time `O(sqrt(gcd))`, Space `O(1)`. Đây là cách đếm ước chuẩn mà ta sẽ dùng lại ở rất nhiều bài số học.

### Cách 4 — `gcd` + phân tích thừa số nguyên tố

Phân tích `g = p1^e1 · p2^e2 ...` rồi nhân `(e1 + 1)(e2 + 1)...`. Vì mỗi ước ứng với một cách chọn số mũ `0..ei` cho mỗi nguyên tố.

````python
class SolutionFactor:
    def commonFactors(self, a: int, b: int) -> int:
        g = gcd(a, b)
        result = 1
        p = 2
        while p * p <= g:
            e = 0
            while g % p == 0:
                g //= p
                e += 1
            result *= e + 1
            p += 1
        if g > 1:                  # còn lại một số nguyên tố lớn, mũ 1
            result *= 2
        return result

print(SolutionFactor().commonFactors(12, 6), SolutionFactor().commonFactors(25, 30), SolutionFactor().commonFactors(36, 36))
````

Độ phức tạp Cách 4: Time `O(sqrt(gcd))` trong trường hợp xấu nhất (khi `g` là số nguyên tố), thường nhanh hơn nhiều; Space `O(1)`.

## 8. TRACE từng bước

Trace Cách 3 với `a = 36`, `b = 36` (`g = 36`, `isqrt = 6`).

| i | `36 % i` | hành động | count |
|---|---|---|---|
| 1 | 0 | cộng 2 (1 và 36) | 2 |
| 2 | 0 | cộng 2 (2 và 18) | 4 |
| 3 | 0 | cộng 2 (3 và 12) | 6 |
| 4 | 0 | cộng 2 (4 và 9) | 8 |
| 5 | 1 | bỏ qua | 8 |
| 6 | 0 | `6*6 == 36` → cộng 1 | 9 |

````python
def trace_sqrt(a, b):
    g = gcd(a, b)
    print(f"a={a}, b={b}, g={g}, isqrt(g)={isqrt(g)}")
    count = 0
    for i in range(1, isqrt(g) + 1):
        if g % i == 0:
            add = 1 if i * i == g else 2
            count += add
            print(f"  i={i}: {g}%{i}=0 -> ước ({i}, {g // i}), cộng {add}, count={count}")
        else:
            print(f"  i={i}: {g}%{i}={g % i} -> bỏ qua, count={count}")
    print("kết quả:", count)
    return count

trace_sqrt(36, 36)
trace_sqrt(25, 30)
````

## 9. Code nộp LeetCode

````python
from typing import List
from math import gcd, isqrt

class Solution:
    def commonFactors(self, a: int, b: int) -> int:
        g = gcd(a, b)
        count = 0
        for i in range(1, isqrt(g) + 1):
            if g % i == 0:
                count += 1 if i * i == g else 2
        return count

print(Solution().commonFactors(12, 6), Solution().commonFactors(25, 30))
````

## 10. Vì sao thuật toán đúng

- **Ước chung ⇔ ước của gcd:** nếu `x | a` và `x | b` thì `x | gcd(a, b)` (vì `gcd` là tổ hợp tuyến tính `a·s + b·t` theo định lý Bézout, mà `x` chia hết cả hai số hạng). Ngược lại, `gcd | a` và `gcd | b`, nên ước của `gcd` cũng chia hết cả `a` và `b`. Hai tập bằng nhau.
- **Đếm theo cặp:** mỗi ước `d` của `g` ứng với đúng một ước đối xứng `g // d`; một trong hai không vượt quá `sqrt(g)`. Duyệt `i <= sqrt(g)` và đếm cả cặp sẽ phủ mọi ước đúng một lần, ngoại trừ ước `i = sqrt(g)` (khi `g` là số chính phương), nơi `i == g // i` và chỉ được tính 1.
- **Thừa số nguyên tố:** một ước của `g` được xác định duy nhất bởi dãy số mũ `0 <= fi <= ei`, nên có `∏ (ei + 1)` ước.
- **Biên:** `g = 1` cho `isqrt = 1`, `i = 1`, `1 * 1 == 1` → cộng 1, đúng (chỉ ước `1`).

````python
impls = [SolutionBrute(), SolutionGcd(), SolutionSqrt(), SolutionFactor(), Solution()]
tests = [
    (12, 6, 4),
    (25, 30, 2),
    (1, 1, 1),            # nhỏ nhất
    (1000, 1000, 16),     # lớn nhất, 1000 = 2^3 * 5^3 -> 4*4 = 16 ước
    (1, 1000, 1),         # một số là 1
    (7, 13, 1),           # hai số nguyên tố khác nhau
    (7, 7, 2),            # cùng nguyên tố
    (36, 36, 9),          # số chính phương (bắt lỗi đếm đôi)
    (64, 48, 5),          # gcd = 16 = 2^4 -> 5 ước
    (100, 75, 3),         # gcd = 25 -> 1, 5, 25
    (720, 840, 16),       # gcd = 120 = 2^3*3*5 -> 4*2*2 = 16
    (997, 991, 1),        # hai số nguyên tố lớn
]
for a, b, exp in tests:
    for im in impls:
        assert im.commonFactors(a, b) == exp, (type(im).__name__, a, b, exp)
print("tất cả test cố định OK:", len(tests))
````

Test 2: vét cạn mọi cặp `1 <= a, b <= 60` (3600 cặp) và ngẫu nhiên có seed 5000 cặp đúng constraint `1..1000`. Oracle: định nghĩa trực tiếp bằng tập ước.

````python
import random

def oracle(a, b):
    A = {x for x in range(1, a + 1) if a % x == 0}
    B = {x for x in range(1, b + 1) if b % x == 0}
    return len(A & B)

for a in range(1, 61):
    for b in range(1, 61):
        exp = oracle(a, b)
        for im in impls:
            assert im.commonFactors(a, b) == exp, (type(im).__name__, a, b)
print("vét cạn 60 x 60 OK")

random.seed(2427)
for _ in range(5000):
    a, b = random.randint(1, 1000), random.randint(1, 1000)
    exp = oracle(a, b)
    assert all(im.commonFactors(a, b) == exp for im in impls), (a, b)
print("5000 test ngẫu nhiên OK")
````

## 13. Các lỗi người mới hay gặp

### Lỗi 1: `range(1, min(a, b))` bỏ sót chính số nhỏ hơn

````python
def wrong_range(a, b):
    return sum(1 for x in range(1, min(a, b)) if a % x == 0 and b % x == 0)   # SAI: thiếu + 1

print("sai :", wrong_range(12, 6))
print("đúng:", Solution().commonFactors(12, 6))
assert wrong_range(12, 6) != Solution().commonFactors(12, 6)
````

Vì sao sai: `range(1, 6)` chỉ gồm `1..5`, mất ước `6` (chính là `min`). Khi `b` chia hết `a` (như `12` và `6`) thì `min` là một ước chung. **Sửa:** `range(1, min(a, b) + 1)`.

### Lỗi 2: đếm đôi ước khi `g` là số chính phương

````python
def wrong_double(a, b):
    g = gcd(a, b)
    count = 0
    for i in range(1, isqrt(g) + 1):
        if g % i == 0:
            count += 2              # SAI: khi i*i == g thì i và g//i là cùng một số
    return count

print("sai :", wrong_double(36, 36))
print("đúng:", Solution().commonFactors(36, 36))
assert wrong_double(36, 36) != Solution().commonFactors(36, 36)
````

Vì sao sai: với `g = 36`, `i = 6` cho cặp `(6, 6)`: chỉ có **một** ước `6` nhưng đã bị cộng `2`. **Sửa:** `count += 1 if i * i == g else 2`.

### Lỗi 3: nhầm số ước chung với ước chung lớn nhất

````python
def wrong_gcd_only(a, b):
    return gcd(a, b)                # SAI: đây là ước chung LỚN NHẤT, không phải số lượng

print("sai :", wrong_gcd_only(12, 6))
print("đúng:", Solution().commonFactors(12, 6))
assert wrong_gcd_only(12, 6) != Solution().commonFactors(12, 6)
````

Vì sao sai: `gcd(12, 6) = 6` là **giá trị** ước chung lớn nhất, còn đề hỏi **bao nhiêu** ước chung (`1, 2, 3, 6` → `4`). **Sửa:** đếm ước của `gcd`.

### Lỗi 4 (bonus): dùng `int(g ** 0.5)` với số thực

Với số nhỏ như bài này thì ổn, nhưng với số lớn căn bậc hai thực có thể sai số (`int(99999999999999 ** 0.5)` làm tròn lệch). Hãy dùng `math.isqrt` để tránh rủi ro.

## 14. Mẹo làm bài nhanh & nhận dạng pattern

- "Số ước chung / ước chung" → quy về **`gcd`**: ước chung của `a, b` là ước của `gcd(a, b)`.
- "Đếm ước của `n`" → duyệt tới `sqrt(n)` theo cặp; nhớ xử lý số chính phương.
- Muốn đếm ước cho **nhiều** số → sàng/thừa số nguyên tố, công thức `∏ (e + 1)`.
- `a, b <= 1000`: thử mọi `x` cũng qua, nhưng hãy tập viết cách theo `gcd` để dùng lại ở các bài lớn hơn.
- Luôn nhớ `range(1, n + 1)` để bao gồm `n`.
- Kiểm tra các biên: `1`, số nguyên tố, số chính phương, `a` chia hết `b`.
- Khi `gcd == 1` (hai số nguyên tố cùng nhau) đáp án luôn là `1`: chỉ có ước chung `1`.
- Ước chung xuất hiện trong rất nhiều bài khác: rút gọn phân số, chia đều thành nhóm, xếp gạch lát. Hễ đề nói "chia đều cả hai", hãy nghĩ tới `gcd`.
- Nếu quên công thức, vẫn có thể fallback về cách 1 (thử từng `x`) vì ràng buộc nhỏ; quan trọng là không lo lắng và viết đúng trước, tối ưu sau.

## 15. Biến thể của bài

1. Đếm ước chung của **một mảng** số (lấy `gcd` toàn mảng rồi đếm ước) — bài "Find Greatest Common Divisor of Array".
2. Liệt kê (không chỉ đếm) các ước chung theo thứ tự tăng dần.
3. Tính **tổng** các ước chung (`sigma`).
4. Đếm số cặp `(i, j)` có `gcd(nums[i], nums[j]) > 1` hoặc `== 1`.
5. Giới hạn lớn `a, b <= 10^12`: bắt buộc cách `sqrt(g)` hoặc thừa số nguyên tố.

## 16. So sánh các cách làm

| Cách | Ý tưởng | Time | Space | Khi nào dùng |
|---|---|---|---|---|
| 1. Thử mọi `x` tới `min` | đúng định nghĩa | `O(min(a,b))` | `O(1)` | `a, b` nhỏ, làm oracle test |
| 2. `gcd` + duyệt tới `g` | đếm ước của `g` | `O(g)` | `O(1)` | dễ viết, hơi nhanh hơn |
| 3. `gcd` + duyệt tới `sqrt(g)` | đếm theo cặp | `O(sqrt(g))` | `O(1)` | **chuẩn**, mở rộng giới hạn lớn |
| 4. `gcd` + thừa số nguyên tố | `∏ (e + 1)` | `O(sqrt(g))` | `O(1)` | nhiều truy vấn / cần cấu trúc số mũ |

## 18. Gợi ý 3 mức cho bài tập

**Mức 1 (nhẹ):** ước chung = ước của `gcd`. Với Bài 1: `gcd(18, 24) = 6`.

**Mức 2 (vừa):** Bài 3: `if n % i == 0: count += 1 if i * i == n else 2`. Bài 4: tạo list bằng một vòng đến `g`, hoặc duyệt tới căn rồi gom cặp và `sorted`. Bài 5: `functools.reduce(gcd, nums)` cho `g` chung.

## 19. Đáp án tham khảo cho bài tập

````python
from functools import reduce

# Bài 1: gcd(18, 24) = 6 -> ước 1, 2, 3, 6
assert Solution().commonFactors(18, 24) == 4

# Bài 2: gcd(30, 45) = 15 -> 1, 3, 5, 15
assert [x for x in range(1, 31) if 30 % x == 0 and 45 % x == 0] == [1, 3, 5, 15]
assert Solution().commonFactors(30, 45) == 4

# Bài 3
def count_divisors(n):
    count = 0
    for i in range(1, isqrt(n) + 1):
        if n % i == 0:
            count += 1 if i * i == n else 2
    return count
assert count_divisors(36) == 9 and count_divisors(12) == 6 and count_divisors(1) == 1 and count_divisors(97) == 2

# Bài 4
def common_factor_list(a, b):
    g = gcd(a, b)
    return [x for x in range(1, g + 1) if g % x == 0]
assert common_factor_list(12, 6) == [1, 2, 3, 6]
assert common_factor_list(25, 30) == [1, 5]

# Bài 5
def common_factors_array(nums):
    return count_divisors(reduce(gcd, nums))
assert common_factors_array([12, 18, 30]) == 4          # gcd = 6
assert common_factors_array([7]) == 2                     # chỉ một số: số ước của chính nó
assert common_factors_array([8, 9]) == 1
print("đáp án bài tập OK")
````
