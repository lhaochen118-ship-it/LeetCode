## 3. Kiến thức nền

### 3.1. Chỉ số 1-based (đề) và 0-based (Python)

Python đánh số phần tử từ `0`: `nums[0]` là phần tử đầu. Đề đánh số từ `1`. Quy đổi: phần tử số `i` của đề là `nums[i - 1]` của Python.

````python
nums = [2, 7, 1, 19, 18, 3]
for i in range(1, len(nums) + 1):
    print(f"phần tử thứ {i} của đề = nums[{i - 1}] = {nums[i - 1]}")
````

### 3.2. Chia hết và ước: `n % i == 0`

`n % i` là số dư khi chia `n` cho `i`. `n % i == 0` nghĩa là `n` chia hết cho `i`, tức `i` là **ước** của `n`. Ví dụ ước của `6` là `1, 2, 3, 6`; ước của `4` là `1, 2, 4`.

````python
n = 6
print([i for i in range(1, n + 1) if n % i == 0])    # [1, 2, 3, 6]
n = 4
print([i for i in range(1, n + 1) if n % i == 0])    # [1, 2, 4]
n = 7
print([i for i in range(1, n + 1) if n % i == 0])    # [1, 7]  (số nguyên tố)
````

### 3.3. Bình phương và tổng

`x * x` (hoặc `x ** 2`) là bình phương. `sum(...)` cộng mọi phần tử của một iterable.

````python
print(3 * 3, 3 ** 2)                       # 9 9
print(sum([1 ** 2, 2 ** 2, 4 ** 2]))       # 21
print(sum(x * x for x in [2, 7, 1, 3]))    # 63
````

### 3.4. `enumerate(nums, start=1)`

`enumerate(iterable, start=1)` trả về từng cặp `(chỉ_số_từ_1, phần_tử)`. Nhờ vậy ta không cần tự trừ `1`.

````python
nums = [10, 20, 30]
for i, x in enumerate(nums, start=1):
    print(i, x)
# 1 10
# 2 20
# 3 30
````

### 3.5. `math.isqrt` và cặp ước `(i, n // i)`

Ước của `n` đi theo **cặp**: nếu `i` là ước thì `n // i` cũng là ước, và `i * (n // i) = n`. Trong mỗi cặp luôn có một số `<= √n`, nên chỉ cần duyệt `i` từ `1` tới `√n` rồi lấy cả `i` và `n // i`. `math.isqrt(n)` là căn bậc hai nguyên (làm tròn xuống), nhanh và không sai số.

Chú ý: khi `n` là **số chính phương** (như `4`, `9`), có `i = n // i` (ví dụ `n = 4`, `i = 2`); lúc đó chỉ được tính **một lần**.

````python
import math
print(math.isqrt(6), math.isqrt(4), math.isqrt(50))     # 2 2 7
n = 6
pairs = [(i, n // i) for i in range(1, math.isqrt(n) + 1) if n % i == 0]
print(pairs)                  # [(1, 6), (2, 3)]
n = 4
pairs = [(i, n // i) for i in range(1, math.isqrt(n) + 1) if n % i == 0]
print(pairs)                  # [(1, 4), (2, 2)]  <- cặp (2, 2) chỉ tính một lần
````

### 3.6. `set` để loại phần tử trùng

`set` không chứa phần tử trùng, nên `{i, n // i}` cho `1` phần tử khi `i == n // i` và `2` phần tử khi khác nhau.

````python
print({2, 4 // 2})       # {2}
print({1, 6 // 1})       # {1, 6}
````

## 4. Ý tưởng nền tảng / cách nghĩ trực giác

Tính tay ví dụ 2: `nums = [2, 7, 1, 19, 18, 3]`, `n = 6`.

| `i` (đề) | `6 % i` | đặc biệt? | `nums[i - 1]` | bình phương |
|---|---|---|---|---|
| 1 | 0 | có | 2 | 4 |
| 2 | 0 | có | 7 | 49 |
| 3 | 0 | có | 1 | 1 |
| 4 | 2 | không | 19 | — |
| 5 | 1 | không | 18 | — |
| 6 | 0 | có | 3 | 9 |

Tổng `4 + 49 + 1 + 9 = 63`.

**Nhận xét:** phần tử đầu (`i = 1`) và phần tử cuối (`i = n`) **luôn** đặc biệt vì `1` và `n` luôn là ước của `n`. Phần còn lại tuỳ vào `n` có những ước nào. Bài toán quy về: **tìm tập ước của `n`, rồi cộng bình phương các phần tử ở các vị trí đó**.

````python
nums = [2, 7, 1, 19, 18, 3]
n = len(nums)
total = 0
for i in range(1, n + 1):
    if n % i == 0:
        total += nums[i - 1] ** 2
        print(f"i={i}: cộng {nums[i - 1]}^2 = {nums[i - 1] ** 2}, tổng = {total}")
print("kết quả:", total)
````

## 5. Cách 1 — Brute Force / baseline: duyệt mọi `i` từ 1 tới `n`

**Ý tưởng:** làm đúng như đề phát biểu:

1. Với mỗi `i` từ `1` tới `n`.
2. Nếu `n % i == 0` thì cộng `nums[i - 1] ** 2` vào tổng.
3. Trả về tổng.

````python
from typing import List

class SolutionLoop:
    def sumOfSquares(self, nums: List[int]) -> int:
        n = len(nums)
        total = 0
        for i in range(1, n + 1):
            if n % i == 0:
                total += nums[i - 1] * nums[i - 1]
        return total
````

````python
sol = SolutionLoop()
print(sol.sumOfSquares([1, 2, 3, 4]))               # 21
print(sol.sumOfSquares([2, 7, 1, 19, 18, 3]))       # 63
````

**Độ phức tạp:** Time `O(n)` (mỗi `i` một phép `%`), Space `O(1)`.

**Nhận xét:** với `n <= 50` cách này đã quá đủ. Điều đáng học: thay vì thử **mọi** `i`, ta có thể chỉ thử các `i <= √n` nhờ tính chất **ước đi theo cặp**. Điều này trở nên quan trọng khi `n` rất lớn (nếu cho phép duyệt theo công thức mà không cần mảng).

## 6. Ý tưởng thuật toán / bước đột phá

**Đột phá 1 — viết gọn bằng `enumerate`.** `for i, x in enumerate(nums, start=1)` cho cả chỉ số kiểu đề và giá trị, không cần `nums[i - 1]`. Kết hợp với `sum` và biểu thức sinh: `sum(x * x for i, x in enumerate(nums, 1) if n % i == 0)`.

**Đột phá 2 — ước đi theo cặp.** Nếu `i | n` (i là ước của `n`) thì `n // i` cũng là ước. Do đó chỉ cần duyệt `i` từ `1` tới `√n`, mỗi ước `i` tìm được sinh ra **hai** chỉ số: `i` và `n // i` (nhớ tránh tính hai lần khi `i == n // i`). Số phép `%` giảm từ `n` xuống `√n`.

**Đột phá 3 — tập ước tính trước.** Tách bài toán làm hai bước: (1) tìm tập ước `D` của `n`; (2) cộng bình phương các `nums[d − 1]` với `d ∈ D`. Cách tách này sạch về mặt tư duy và dễ thay "special" bằng điều kiện khác.

**Đột phá 4 — lọc bằng mặt nạ (mask).** Dựng mặt nạ boolean `[n % i == 0 for i in 1..n]`, rồi chọn phần tử ứng với `True`. `itertools.compress(nums, mask)` làm đúng việc đó.

````python
nums = [2, 7, 1, 19, 18, 3]
n = len(nums)
print(sum(x * x for i, x in enumerate(nums, start=1) if n % i == 0))   # 63
````

## 7. Các cách tiếp theo

### Cách 2 — `enumerate(nums, start=1)` + biểu thức sinh (code nộp)

**Từng đoạn:**

- `n = len(nums)`.
- `enumerate(nums, start=1)` cho `(i, x)` với `i` từ `1`.
- Điều kiện `n % i == 0` giữ lại các phần tử đặc biệt.
- `x * x` bình phương; `sum` cộng lại.

````python
class SolutionEnum:
    def sumOfSquares(self, nums: List[int]) -> int:
        n = len(nums)
        return sum(x * x for i, x in enumerate(nums, start=1) if n % i == 0)
````

````python
sol = SolutionEnum()
print(sol.sumOfSquares([1, 2, 3, 4]))               # 21
print(sol.sumOfSquares([2, 7, 1, 19, 18, 3]))       # 63
````

**Độ phức tạp:** Time `O(n)`, Space `O(1)` (biểu thức sinh không tạo danh sách).

### Cách 3 — Duyệt tới `√n`, ghép cặp ước `(i, n // i)`

**Từng đoạn:**

- `for i in range(1, isqrt(n) + 1)`: mỗi `i` có thể là ước nhỏ của một cặp.
- Nếu `n % i == 0`: cộng `nums[i − 1]²`; nếu `i != n // i` thì cộng thêm `nums[n // i − 1]²` (ước lớn của cặp).
- Điều kiện `i != n // i` tránh tính một ước hai lần khi `n` là số chính phương.

````python
import math

class SolutionSqrt:
    def sumOfSquares(self, nums: List[int]) -> int:
        n = len(nums)
        total = 0
        for i in range(1, math.isqrt(n) + 1):
            if n % i == 0:
                total += nums[i - 1] ** 2
                j = n // i
                if j != i:
                    total += nums[j - 1] ** 2
        return total
````

````python
sol = SolutionSqrt()
print(sol.sumOfSquares([1, 2, 3, 4]))               # 21  (n=4: cặp (1,4) và (2,2))
print(sol.sumOfSquares([2, 7, 1, 19, 18, 3]))       # 63  (n=6: cặp (1,6) và (2,3))
````

**Độ phức tạp:** Time `O(√n)` (số phép `%`; mỗi ước tìm được tốn thêm `O(1)` để đọc mảng), Space `O(1)`.

### Cách 4 — Tập ước tính trước, rồi cộng bình phương

**Từng đoạn:**

- Bước 1: `divisors = set()`; với `i` từ `1` tới `isqrt(n)`, nếu `n % i == 0` thì thêm `i` và `n // i` vào tập (tập tự loại trùng).
- Bước 2: `sum(nums[d - 1] ** 2 for d in divisors)`.

````python
class SolutionDivisorSet:
    def sumOfSquares(self, nums: List[int]) -> int:
        n = len(nums)
        divisors = set()
        for i in range(1, math.isqrt(n) + 1):
            if n % i == 0:
                divisors.add(i)
                divisors.add(n // i)
        return sum(nums[d - 1] ** 2 for d in divisors)
````

````python
sol = SolutionDivisorSet()
print(sol.sumOfSquares([1, 2, 3, 4]))               # 21
print(sol.sumOfSquares([2, 7, 1, 19, 18, 3]))       # 63
n = 12
d = set()
for i in range(1, math.isqrt(n) + 1):
    if n % i == 0:
        d.update({i, n // i})
print(sorted(d))      # [1, 2, 3, 4, 6, 12]
````

**Độ phức tạp:** Time `O(√n)` (tìm ước) `+ O(d(n))` (cộng; `d(n)` là số ước), Space `O(d(n))`. Hai bước tách rời giúp dễ đọc và dễ mở rộng.

### Cách 5 — Mặt nạ boolean + `itertools.compress`

**Từng đoạn:** `mask[i − 1] = (n % i == 0)` cho `i = 1..n`. `compress(nums, mask)` trả về các phần tử của `nums` ở những vị trí mask `True`. Rồi bình phương và cộng.

````python
from itertools import compress

class SolutionCompress:
    def sumOfSquares(self, nums: List[int]) -> int:
        n = len(nums)
        mask = [n % i == 0 for i in range(1, n + 1)]
        return sum(x * x for x in compress(nums, mask))
````

````python
print(list(compress("ABCDEF", [1, 0, 1, 0, 1, 1])))        # ['A', 'C', 'E', 'F']
sol = SolutionCompress()
print(sol.sumOfSquares([1, 2, 3, 4]))               # 21
print(sol.sumOfSquares([2, 7, 1, 19, 18, 3]))       # 63
````

**Độ phức tạp:** Time `O(n)`, Space `O(n)` (mặt nạ). Cách này tách "chọn" khỏi "tính" và gọn khi nhiều bước lọc.

## 8. TRACE từng bước

Trace Cách 1 trên `nums = [2, 7, 1, 19, 18, 3]` (`n = 6`):

|---|---|---|---|---|---|
| 1 | 0 | có | 2 | 4 | 4 |
| 2 | 0 | có | 7 | 49 | 53 |
| 3 | 0 | có | 1 | 1 | 54 |
| 4 | 2 | không | 19 | 0 | 54 |

Trace Cách 3 (`n = 4`, `nums = [1, 2, 3, 4]`): `isqrt(4) = 2`. `i = 1`: ước, cộng `nums[0]² = 1`, `j = 4 ≠ 1` nên cộng `nums[3]² = 16`. `i = 2`: ước, cộng `nums[1]² = 4`, `j = 2 == i` nên **không** cộng lần nữa. Tổng `21`.

````python
def trace_loop(nums):
    n = len(nums)
    total = 0
    print("n =", n)
    for i in range(1, n + 1):
        special = n % i == 0
        add = nums[i - 1] ** 2 if special else 0
        total += add
        print(f"i={i}, n%i={n % i}, special={special}, nums[{i-1}]={nums[i-1]}, cộng={add}, total={total}")
    return total

trace_loop([2, 7, 1, 19, 18, 3])
````

````python
def trace_sqrt(nums):
    n = len(nums)
    total = 0
    r = math.isqrt(n)
    print(f"n={n}, isqrt(n)={r}")
    for i in range(1, r + 1):
        if n % i != 0:
            print(f"i={i}: {n} % {i} = {n % i} -> không phải ước")
            continue
        j = n // i
        total += nums[i - 1] ** 2
        msg = f"i={i}: ước, cộng nums[{i - 1}]^2={nums[i - 1] ** 2}"
        if j != i:
            total += nums[j - 1] ** 2
            msg += f"; ước đôi j={j}, cộng nums[{j - 1}]^2={nums[j - 1] ** 2}"
        else:
            msg += f"; j={j} trùng i, không cộng lần hai"
        print(msg, "-> total =", total)
    return total

trace_sqrt([1, 2, 3, 4])
trace_sqrt([2, 7, 1, 19, 18, 3])
````

## 9. Code nộp LeetCode

Dùng Cách 2 (`enumerate` + `sum`), ngắn gọn, đúng, `O(n)`.

````python
from typing import List

class Solution:
    def sumOfSquares(self, nums: List[int]) -> int:
        n = len(nums)
        # i chạy từ 1 (theo cách đánh số của đề); phần tử đặc biệt khi i là ước của n
        return sum(x * x for i, x in enumerate(nums, start=1) if n % i == 0)
````

````python
print(Solution().sumOfSquares([1, 2, 3, 4]))               # 21
print(Solution().sumOfSquares([2, 7, 1, 19, 18, 3]))       # 63
````

## 10. Vì sao thuật toán đúng

**Cách 1 / Cách 2.** Phần tử đặc biệt của đề là `nums[i]` (1-based) với `n % i == 0`. Phần tử đó là `nums[i − 1]` trong Python, và `enumerate(nums, start=1)` gán `i` đúng chỉ số 1-based cho từng phần tử. Mỗi phần tử được xét đúng một lần, nên tổng `Σ nums[i]² trên i | n` bằng đúng đại lượng đề yêu cầu.

**Cách 3 (ghép cặp ước).** Tập ước của `n` là hợp của các cặp `{i, n // i}` với `i` chạy trên các ước thoả `i <= √n`. Thật vậy:

- Nếu `d` là ước của `n` và `d <= √n` thì `d` được duyệt trực tiếp.
- Nếu `d` là ước và `d > √n` thì `e = n // d` là ước và `e = n / d < √n` (vì `d > √n`), do đó `e` được duyệt và sinh ra `d = n // e`.
- Mỗi ước được sinh đúng một lần nếu ta loại trường hợp `i = n // i`. Khi `i = n // i` thì `i² = n`, hai ước của cặp trùng nhau và chỉ tính một lần.

Do đó tập ước được cộng đầy đủ, không trùng, và kết quả bằng Cách 1.

**Cách 4, 5.** Cách 4 xây tập ước giống Cách 3 (dùng `set` để loại trùng) rồi cộng; Cách 5 lọc theo định nghĩa. Đều tương đương Cách 1; test ở mục 12 kiểm chứng trên mọi `n` từ `1` tới `50`.

## 11. Phân tích Time / Space Complexity

`n = len(nums)`, `d(n)` là số ước của `n` (rất nhỏ: với `n <= 50`, tối đa `6`).

| Cách | Time | Space | Ghi chú |
|---|---|---|---|
| 1. Vòng `for` | `O(n)` | `O(1)` | đọc dễ nhất |
| 2. `enumerate` + `sum` | `O(n)` | `O(1)` | **code nộp** |
| 3. Ghép cặp ước đến `√n` | `O(√n)` phép chia dư | `O(1)` | tốt khi `n` rất lớn và chỉ cần xem vài chỉ số |

## 13. Các lỗi người mới hay gặp

### Lỗi 1: Nhầm chỉ số 0-based và 1-based

````python
def sai_chi_so(nums: List[int]) -> int:
    n = len(nums)
    total = 0
    for i in range(n):                  # i chạy 0..n-1
        if n % (i + 1) == 0:            # đúng điều kiện nhưng...
            total += nums[i + 1] ** 2    # sai: lấy phần tử kế tiếp (có thể IndexError)
    return total

try:
    print("sai :", sai_chi_so([1, 2, 3, 4]))
except IndexError as e:
    print("sai : IndexError:", e)
print("đúng:", Solution().sumOfSquares([1, 2, 3, 4]))      # 21
````

**Vì sao sai:** phần tử thứ `i` của đề là `nums[i - 1]` trong Python. Khi `i = n` (phần tử cuối, luôn đặc biệt), `nums[n]` ra ngoài mảng ⇒ `IndexError`. Hoặc dùng `enumerate(nums, start=1)` để khỏi tự trừ.

### Lỗi 2: `n % i` với `i` bắt đầu từ 0

````python
def sai_chia_cho_0(nums: List[int]) -> int:
    n = len(nums)
    total = 0
    for i in range(n):                 # i = 0 ngay vòng đầu
        if n % i == 0:                 # ZeroDivisionError khi i = 0
            total += nums[i] ** 2
    return total

try:
    sai_chia_cho_0([1, 2, 3, 4])
except ZeroDivisionError as e:
    print("lỗi:", e)
````

**Vì sao sai:** chia (dư) cho `0` không xác định ⇒ `ZeroDivisionError`. Chỉ số 1-based đảm bảo `i >= 1`.

### Lỗi 3: Quên bình phương

````python
def sai_quen_binh_phuong(nums: List[int]) -> int:
    n = len(nums)
    return sum(x for i, x in enumerate(nums, 1) if n % i == 0)

print("sai :", sai_quen_binh_phuong([1, 2, 3, 4]))         # 7 (1 + 2 + 4)
print("đúng:", Solution().sumOfSquares([1, 2, 3, 4]))      # 21
````

**Vì sao sai:** đề yêu cầu **tổng bình phương**, không phải tổng.

### Lỗi 4: Ghép cặp ước nhưng đếm trùng khi `n` là số chính phương

````python
def sai_dem_trung(nums: List[int]) -> int:
    n = len(nums)
    total = 0
    for i in range(1, math.isqrt(n) + 1):
        if n % i == 0:
            total += nums[i - 1] ** 2
            total += nums[n // i - 1] ** 2        # sai: không loại trường hợp i == n // i
    return total

print("sai :", sai_dem_trung([1, 2, 3, 4]))               # 25 (đếm 2^2 hai lần)
print("đúng:", Solution().sumOfSquares([1, 2, 3, 4]))     # 21
````

**Vì sao sai:** với `n = 4`, `i = 2` có `n // i = 2`; ước `2` bị cộng hai lần. Cần `if j != i`, hoặc dùng `set`.

### Lỗi 5: Đảo `n % i` thành `i % n`

````python
def sai_dao_phep_du(nums: List[int]) -> int:
    n = len(nums)
    return sum(x * x for i, x in enumerate(nums, 1) if i % n == 0)    # sai: i chia hết n?

print("sai :", sai_dao_phep_du([1, 2, 3, 4]))              # 16 (chỉ i = 4)
print("đúng:", Solution().sumOfSquares([1, 2, 3, 4]))      # 21
````

**Vì sao sai:** "`i` chia hết `n`" nghĩa là `n % i == 0` (`n` là số bị chia). `i % n == 0` chỉ đúng khi `i` là bội của `n` (ở đây chỉ `i = n`).

## 14. Mẹo làm bài nhanh & nhận dạng pattern

- Đề nói "1-indexed" mà bạn dùng Python: **luôn** dùng `enumerate(nums, start=1)` hoặc `nums[i - 1]`.
- "Phần tử ở vị trí `i` với `i` chia hết `n`" = "vị trí là **ước** của `n`". Phần tử đầu và cuối luôn nằm trong số đó.
- Ước đi **theo cặp** `(i, n // i)`; chỉ duyệt `i <= √n`; **nhớ trường hợp `i == n // i`**.
- `sum(biểu thức for ... if điều kiện)` là mẫu quen thuộc cho "tổng các phần tử thoả điều kiện".
- Với `n <= 50`, đừng tối ưu quá mức: ưu tiên code dễ đọc và đúng.
- Nếu đề cho `n` rất lớn nhưng chỉ hỏi về các vị trí ước, ghép cặp ước là cách duy nhất khả thi.

## 15. Biến thể của bài

1. **Tổng lập phương** hay tổng `x` thay vì `x²`: đổi biểu thức trong `sum`.
2. **Chỉ số 0-based:** `i` đặc biệt nếu `n % (i + 1) == 0` (cùng kết quả) hoặc nếu `i` chia hết `n` (sai khác).
3. **Special nếu `i²` chia hết `n`** hoặc `i` là **số nguyên tố**: đổi điều kiện.
4. **Tích** các phần tử đặc biệt, hoặc **số** phần tử đặc biệt (số ước của `n`).
5. **`n` lên tới `10^12`** và `nums` được cho như một hàm `f(i)`: dùng ghép cặp ước `O(√n)`.
6. **Chỉ số `i` là bội của `k`** (thay vì ước của `n`): `range(k, n + 1, k)`.

## 16. So sánh các cách làm

| Cách | Ý tưởng | Time | Space | Khi nào dùng |
|---|---|---|---|---|
| 1. Vòng `for` | thử mọi `i` từ 1 tới `n` | `O(n)` | `O(1)` | dễ đọc nhất |
| 2. `enumerate` + `sum` | cùng ý, viết gọn | `O(n)` | `O(1)` | **code nộp** |
| 3. Ghép cặp ước | `i <= √n` và `n // i` | `O(√n)` | `O(1)` | `n` rất lớn |
| 4. Tập ước | tìm ước trước, cộng sau | `O(√n + d(n))` | `O(d(n))` | tách bước, dễ đổi điều kiện |
| 5. Mặt nạ `compress` | mask boolean rồi chọn | `O(n)` | `O(n)` | nhiều lớp lọc |
