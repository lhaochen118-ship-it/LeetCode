## 1. Đọc đề trước khi nghĩ code

Cho chuỗi `s` gồm các từ (word) và dấu cách `' '`. Hãy trả **số ký tự của từ cuối cùng**; không tính các dấu cách sau từ đó.

| s | kết quả | vì sao |
|---|---:|---|
| `"Hello World"` | 5 | từ cuối là World |
| `"   fly me   to   the moon  "` | 4 | từ cuối là moon |
| `"luffy is still joyboy"` | 6 | từ cuối là joyboy |

Trong đề gốc, có ít nhất một word; các word gồm chữ cái tiếng Anh. Không được giả sử chuỗi kết thúc bằng chữ: nó có thể có **trailing spaces** (dấu cách cuối).

````python
s="Hello World"
print("ký tự cuối:",repr(s[-1]))
print("độ dài toàn chuỗi:",len(s))
````

## 2. Python cơ bản: len, chỉ số âm, ký tự khoảng trắng

`len(s)` là số ký tự của **cả chuỗi**, gồm cả space. `s[-1]` lấy ký tự **cuối**; `s[-2]` lấy ký tự gần cuối. Trong code thực hành, nhớ dùng `repr(ch)` để **thấy được** dấu cách rõ ràng.

````python
s="hi  "
print("len:",len(s))
for i,ch in enumerate(s):
    print("index",i,"ký tự",repr(ch))
print("s[-1] =",repr(s[-1]))
````

## 3. Lối tắt đầu tiên: `.split()`

`split()` **không có tham số** tự tách theo các đoạn whitespace liên tiếp, bỏ khoảng trắng đầu và cuối. Ví dụ `" a  bb ".split()` → `['a','bb']`. Lấy phần tử cuối `[-1]` rồi `len` cho đáp án.

Cách này rất dễ hiểu, nhưng tạo ra **list** các word và các chuỗi con → dùng thêm bộ nhớ O(n).

````python
s="   fly me   to   the moon  "
words=s.split()
print("Danh sách từ:",words)
print("Từ cuối:",repr(words[-1]),"độ dài:",len(words[-1]))
````

## 4. Code cách đơn giản

Nếu đầu vào bảo đảm có ít nhất một word như đề thì không cần xử lý list rỗng. Hàm demo dưới đây vẫn hỗ trợ chuỗi toàn dấu cách như **phần mở rộng** để luyện code.

````python
def last_word_split(s:str)->int:
    words=s.split()
    return len(words[-1]) if words else 0

for s in ["Hello World", " a  bb  ", "single", "   "]:
    print(repr(s),"->",last_word_split(s))
````

## 5. Insight: chỉ cần đi từ PHẢI qua TRÁI

Ta cần **mỗi từ cuối**. Vì vậy thay vì tách tất cả các từ, bắt đầu tại ký tự cuối:

1. **Bỏ qua** các space ở cuối.
2. **Đếm** liên tiếp các ký tự không phải space.
3. Gặp space ngăn cách trước từ cuối hoặc hết chuỗi → **dừng**.

Không cần tạo danh sách từ!

````python
s="Hi Bob  "; i=len(s)-1
while i>=0 and s[i]==" ":
    i-=1
print("Vị trí ký tự cuối của word:",i,"ký tự:",repr(s[i]))
````

## 6. Cú pháp `while`, `and` và điều kiện biên

`while condition:` lặp chừng nào điều kiện còn True. `and` có short-circuit: nếu `i>=0` sai, Python **không đọc** `s[i]` ở vế sau. Thứ tự rất quan trọng.

`i -= 1` là viết tắt của `i = i - 1`. Bắt đầu `i=len(s)-1`, lùi dần tới tối thiểu -1.

````python
s="a  ";i=len(s)-1
while i>=0 and s[i]==" ":
    print("Bỏ qua khoảng trắng tại",i)
    i-=1
print("Đã tới",i,"chữ",repr(s[i]))
````

## 7. Code chính O(1) bộ nhớ

Dùng biến `length=0` đếm từng chữ của từ cuối. Khi gặp space ngăn từ hoặc hết chuỗi, kết thúc. Không cần `split` hay `strip`.

````python
def length_last_word_backward(s:str)->int:
    i=len(s)-1
    while i>=0 and s[i]==" ":
        i-=1
    length=0
    while i>=0 and s[i]!=" ":
        length+=1
        i-=1
    return length

print(length_last_word_backward("Hello World"))
print(length_last_word_backward("   fly me   to   the moon  "))
````

## 8. Bảng trace bằng tay

Ví dụ `s="Go home  "`. Các index: `0 G, 1 o, 2 ' ', 3 h, 4 o, 5 m, 6 e, 7 ' ', 8 ' '`.

| bước | i trước | s[i] | hành động | length sau |
|---:|---:|---|---|---:|
| 1 | 8 | space | bỏ qua | 0 |
| 2 | 7 | space | bỏ qua | 0 |
| 3 | 6 | e | đếm | 1 |
| 4 | 5 | m | đếm | 2 |
| 5 | 4 | o | đếm | 3 |
| 6 | 3 | h | đếm | 4 |
| 7 | 2 | space | dừng | 4 |

Mấu chốt: **không** đếm các space ở cuối; khi đã bắt đầu đếm một từ, gặp space mới dừng.

````python
def trace_last_word(s):
    i=len(s)-1;length=0
    print("Chuỗi:",repr(s))
    while i>=0 and s[i]==" ":
        print(f"i={i}: space cuối -> bỏ qua, length={length}")
        i-=1
    while i>=0 and s[i]!=" ":
        length+=1
        print(f"i={i}: gặp {s[i]!r} -> length={length}")
        i-=1
    print("Dừng; đáp án",length)
    return length
assert trace_last_word("Go home  ")==4
````

## 9. Vì sao cách duyệt ngược đúng?

Sau vòng 1, `i` đứng ở ký tự cuối của từ cuối (hoặc -1). Vòng 2 đi lùi qua từng ký tự của chính từ ấy, tăng `length` **một lần cho mỗi chữ**. Vì gặp space là dừng, nó không thể đếm nhầm từ trước. Từ cuối được đếm đủ đúng số ký tự.

## 10. Phân tích độ phức tạp

Gọi `n=len(s)`.

- `split()` cần duyệt chuỗi O(n), có thể tạo bộ nhớ phụ O(n).
- Quét ngược: mỗi ký tự bị đi qua **tối đa một lần**, nên **Time O(n)** tệ nhất (ví dụ chỉ có một từ rất dài), **Space O(1)** (chỉ dùng `i` và `length`).
- Trường hợp dừng sớm (từ cuối ngắn và không có nhiều space cuối), số bước thực tế nhỏ hơn n; nhưng Big-O worst-case vẫn O(n).

````python
s="code"*500
assert length_last_word_backward(s)==2000
print("Một từ dài 2000 ký tự: PASS")
````

## 11. Code chuẩn nộp LeetCode

Tên method trong đề: `lengthOfLastWord` (chữ O, L, W viết hoa), nhận `s: str`, trả số nguyên. Không cần import và không nên `print` trong code submit.

````python
class Solution:
    def lengthOfLastWord(self,s:str)->int:
        i=len(s)-1
        while i>=0 and s[i]==" ":
            i-=1
        count=0
        while i>=0 and s[i]!=" ":
            count+=1
            i-=1
        return count

assert Solution().lengthOfLastWord("Hello World")==5
````
