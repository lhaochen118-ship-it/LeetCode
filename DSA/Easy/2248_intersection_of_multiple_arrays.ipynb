## 1. Mục tiêu học

Sau notebook này bạn sẽ:

1. Hiểu **giao** (intersection) của nhiều tập: các phần tử xuất hiện trong **mọi** mảng con.
2. Làm việc với **mảng hai chiều** (`List[List[int]]`): duyệt hàng, duyệt từng phần tử.
3. Viết 4 cách: brute force (kiểm tra từng ứng viên), giao tập hợp `set`, **đếm tần suất** (mảng/`Counter`), và hai con trỏ trên mảng đã sắp xếp.
4. Rút ra mẹo: vì các giá trị trong **mỗi mảng con đều khác nhau**, "xuất hiện trong mọi mảng" ⇔ "đếm được đúng `len(nums)` lần".
5. Tránh lỗi hay gặp: quên sắp xếp kết quả, giao sai khi dùng `set` rỗng khởi tạo, đếm trùng khi mảng con có giá trị lặp.

## 2. Đề bài & ví dụ

Cho mảng hai chiều `nums` gồm các mảng số nguyên **đôi một khác nhau trong từng mảng con**. Hãy trả về danh sách các số nguyên xuất hiện trong **mọi** mảng con, **sắp xếp tăng dần**.

| Ví dụ | nums | Output | Giải thích |
|---|---|---|---|
| 1 | `[[3,1,2,4,5],[1,2,3,4],[3,4,5,6]]` | `[3,4]` | `3` và `4` có mặt trong cả ba mảng. |
| 2 | `[[1,2,3],[4,5,6]]` | `[]` | Hai mảng không có phần tử chung. |

Constraints:

- `1 <= nums.length <= 1000`
- `1 <= sum(nums[i].length) <= 1000`
- `1 <= nums[i][j] <= 1000`
- Mọi giá trị trong **mỗi** `nums[i]` đều khác nhau.

Chú ý: các mảng con **không** nhất thiết đã sắp xếp (ví dụ `[3,1,2,4,5]`).

````python
examples = [([[3, 1, 2, 4, 5], [1, 2, 3, 4], [3, 4, 5, 6]], [3, 4]), ([[1, 2, 3], [4, 5, 6]], [])]
for nums, expected in examples:
    common = sorted(set(nums[0]).intersection(*nums[1:]))
    print("nums =", nums, "| mong đợi =", expected, "| tự tính nhanh:", common)
````

## 3. Kiến thức nền

### 3.1. Mảng hai chiều (list của list)

`nums[i]` là mảng con thứ `i`; `nums[i][j]` là phần tử thứ `j` của mảng con đó. `len(nums)` là số mảng con, `len(nums[i])` là độ dài mảng con thứ `i`.

````python
nums = [[3, 1, 2, 4, 5], [1, 2, 3, 4], [3, 4, 5, 6]]
print(len(nums), len(nums[0]), nums[2][1])
for row in nums:
    print(row)
print(sum(len(row) for row in nums))        # tổng số phần tử
````

### 3.2. `set` và phép giao `&`

`set` là tập hợp không trùng. Giao hai tập: `a & b` hoặc `a.intersection(b)` — gồm phần tử có trong **cả hai**. Với nhiều tập: `a.intersection(b, c)` hoặc cập nhật dần `a &= b`.

````python
a, b, c = {3, 1, 2, 4, 5}, {1, 2, 3, 4}, {3, 4, 5, 6}
print(a & b)
print(a & b & c)
print(a.intersection(b, c))
print(set([1, 2, 3]) & set([4, 5, 6]))
````

### 3.3. `sorted` và chuyển `set` ↔ `list`

`set` **không có thứ tự**; đề đòi kết quả tăng dần nên phải `sorted(...)`. `sorted` trả về **list** mới.

````python
s = {4, 3}
print(s, list(s))                # thứ tự không đảm bảo
print(sorted(s))                 # [3, 4] chắc chắn tăng dần
````

### 3.4. `collections.Counter` và mảng đếm

`Counter(iterable)` đếm số lần mỗi phần tử xuất hiện. Vì giá trị tối đa `1000`, ta cũng có thể dùng mảng `cnt = [0] * 1001` và cộng `cnt[v] += 1` — nhanh và đơn giản.

````python
from collections import Counter

c = Counter([3, 1, 2, 4, 5] + [1, 2, 3, 4] + [3, 4, 5, 6])
print(c)
print(sorted(v for v, k in c.items() if k == 3))
cnt = [0] * 7
for row in [[3, 1, 2], [1, 2, 3], [3, 4]]:
    for v in row:
        cnt[v] += 1
print(cnt)
````

### 3.5. `functools.reduce` (tùy chọn)

`reduce(f, [x1, x2, x3])` tính `f(f(x1, x2), x3)`. Dùng để giao nhiều tập: `reduce(lambda a, b: a & b, danh_sách_set)`.

````python
from functools import reduce

sets = [set([3, 1, 2, 4, 5]), set([1, 2, 3, 4]), set([3, 4, 5, 6])]
print(reduce(lambda a, b: a & b, sets))
````

1. Lấy ứng viên từ mảng đầu: `3, 1, 2, 4, 5`.
2. Kiểm tra từng ứng viên có trong `B` và `C` không:

| Ứng viên | trong `B`? | trong `C`? | giữ? |

````python
A, B, C = [3, 1, 2, 4, 5], [1, 2, 3, 4], [3, 4, 5, 6]
for v in A:
    print(v, "trong B:", v in B, "| trong C:", v in C)
print(sorted(v for v in A if v in B and v in C))
````

## 5. Cách 1 — Brute Force: kiểm tra từng ứng viên trong mọi mảng

Lấy mỗi `v` của mảng đầu tiên làm **ứng viên**. Với mỗi ứng viên, duyệt mọi mảng con và kiểm tra `v in mảng` (tìm tuần tự). Nếu có mặt trong tất cả thì giữ. Cuối cùng sắp xếp.

````python
class SolutionBrute:
    def intersection(self, nums):
        res = []
        for v in nums[0]:
            ok = True
            for row in nums:
                if v not in row:             # tìm tuần tự O(len(row))
                    ok = False
                    break
            if ok:
                res.append(v)
        return sorted(res)

sol = SolutionBrute()
print(sol.intersection([[3, 1, 2, 4, 5], [1, 2, 3, 4], [3, 4, 5, 6]]))
print(sol.intersection([[1, 2, 3], [4, 5, 6]]))
````

**Độ phức tạp:** ứng viên tối đa `L0 = len(nums[0])`, mỗi ứng viên quét tối đa mọi phần tử (`N = tổng số phần tử`) ⇒ Time `O(L0 · N)` ≤ `O(N²)`, Space `O(1)` ngoài kết quả. Với `N <= 1000` vẫn nhanh.

**Nút thắt:** `v in row` trên list là tìm tuần tự. Đổi sang `set` hoặc **đếm một lượt** sẽ tránh việc quét lặp.

## 6. Ý tưởng thuật toán — bước đột phá

**Insight:** "có mặt trong mọi mảng" là điều kiện về **số mảng chứa `v`**. Vì mỗi mảng không có giá trị lặp, số mảng chứa `v` **bằng** số lần `v` xuất hiện trong toàn bộ dữ liệu. Vậy:

- Đếm tần suất mỗi giá trị trong **tất cả** mảng (một lượt, `O(N)`).
- Giá trị có tần suất **đúng bằng `len(nums)`** là giá trị chung.

Hai hướng song song:

1. **`set`:** giao dần các tập `set(row)`.
2. **Đếm:** `cnt[v] == len(nums)`.

Quy tắc "nếu… thì…": **nếu** `cnt[v] == số_mảng` **thì** `v` thuộc giao; ngược lại thì không.

**Invariant (cách giao `set`):** sau khi xử lý `k` mảng đầu, tập `common` đúng bằng giao của `k` mảng đó.

## 7. Các cách tiếp theo

### Cách 2 — Giao tập hợp bằng `set`

Khởi tạo `common = set(nums[0])` (**không** phải tập rỗng, kẻo giao luôn rỗng). Với mỗi mảng còn lại, `common &= set(row)`. Cuối cùng `sorted(common)`.

````python
class SolutionSet:
    def intersection(self, nums):
        common = set(nums[0])
        for row in nums[1:]:
            common &= set(row)
        return sorted(common)

sol = SolutionSet()
print(sol.intersection([[3, 1, 2, 4, 5], [1, 2, 3, 4], [3, 4, 5, 6]]))
print(sol.intersection([[1, 2, 3], [4, 5, 6]]))
````

**Độ phức tạp:** Time `O(N + k log k)` với `N` là tổng số phần tử, `k` là kích thước kết quả (cho sắp xếp); Space `O(N)`.

### Cách 3 — Đếm tần suất bằng mảng 1001 ô

Cộng `cnt[v] += 1` cho mọi phần tử của mọi mảng. Rồi quét `v` từ `1` đến `1000`: nếu `cnt[v] == len(nums)` thì thêm vào kết quả — **kết quả tự động tăng dần**, không cần `sorted`.

````python
class SolutionCount:
    def intersection(self, nums):
        cnt = [0] * 1001
        for row in nums:
            for v in row:
                cnt[v] += 1
        m = len(nums)
        return [v for v in range(1, 1001) if cnt[v] == m]

sol = SolutionCount()
print(sol.intersection([[3, 1, 2, 4, 5], [1, 2, 3, 4], [3, 4, 5, 6]]))
print(sol.intersection([[1, 2, 3], [4, 5, 6]]))
````

**Độ phức tạp:** Time `O(N + V)` với `V = 1000` là miền giá trị, Space `O(V)`. Không cần sắp xếp, không cần hash.

### Cách 4 — Hai con trỏ trên mảng đã sắp xếp

Sắp xếp từng mảng, rồi giao **hai mảng** bằng hai con trỏ `i`, `j`: nếu `a[i] == b[j]` thì ghi nhận, tăng cả hai; nếu `a[i] < b[j]` thì tăng `i`; ngược lại tăng `j`. Giao dần: `kq = giao(kq, mảng_kế_tiếp)`.

````python
class SolutionTwoPointer:
    @staticmethod
    def _inter(a, b):
        i = j = 0
        out = []
        while i < len(a) and j < len(b):
            if a[i] == b[j]:
                out.append(a[i]); i += 1; j += 1
            elif a[i] < b[j]:
                i += 1
            else:
                j += 1
        return out

    def intersection(self, nums):
        cur = sorted(nums[0])
        for row in nums[1:]:
            cur = self._inter(cur, sorted(row))
            if not cur:
                break                       # rỗng thì dừng sớm
        return cur

sol = SolutionTwoPointer()
print(sol.intersection([[3, 1, 2, 4, 5], [1, 2, 3, 4], [3, 4, 5, 6]]))
print(sol.intersection([[1, 2, 3], [4, 5, 6]]))
````

**Độ phức tạp:** Time `O(N log N)` do sắp xếp từng mảng (giao hai mảng đã sắp xếp là tuyến tính), Space `O(N)`. Ưu điểm: không cần giả định "giá trị nhỏ" (hữu ích khi giá trị rất lớn).

## 8. TRACE từng bước

Trace cách 3 (mảng đếm) với `nums = [[3,1,2,4,5],[1,2,3,4],[3,4,5,6]]`, `m = 3`.

| Sau mảng | `cnt[1]` | `cnt[2]` | `cnt[3]` | `cnt[4]` | `cnt[5]` | `cnt[6]` |
|---|---|---|---|---|---|---|
| 1 `[3,1,2,4,5]` | 1 | 1 | 1 | 1 | 1 | 0 |
| 2 `[1,2,3,4]` | 2 | 2 | 2 | 2 | 1 | 0 |
| 3 `[3,4,5,6]` | 2 | 2 | **3** | **3** | 2 | 1 |

Giá trị có `cnt == 3` là `3` và `4` ⇒ đáp án `[3, 4]`.

````python
def trace_count(nums):
    cnt = [0] * 8
    m = len(nums)
    for r, row in enumerate(nums, 1):
        for v in row:
            cnt[v] += 1
        print(f"sau mảng {r} {row}: cnt[1..6] = {cnt[1:7]}")
    ans = [v for v in range(1, 8) if cnt[v] == m]
    print("cnt == m =", m, "tại:", ans)
    return ans

trace_count([[3, 1, 2, 4, 5], [1, 2, 3, 4], [3, 4, 5, 6]])
````

````python
def trace_set(nums):
    common = set(nums[0])
    print("khởi tạo:", sorted(common))
    for r, row in enumerate(nums[1:], 2):
        common &= set(row)
        print(f"giao với mảng {r} {row}: -> {sorted(common)}")
    return sorted(common)

print(trace_set([[3, 1, 2, 4, 5], [1, 2, 3, 4], [3, 4, 5, 6]]))
````

## 9. Code nộp LeetCode

Dùng cách 3 (mảng đếm): `O(N + V)`, kết quả tự động tăng dần.

````python
from typing import List

class Solution:
    def intersection(self, nums: List[List[int]]) -> List[int]:
        cnt = [0] * 1001
        for row in nums:
            for v in row:
                cnt[v] += 1
        m = len(nums)
        return [v for v in range(1, 1001) if cnt[v] == m]

print(Solution().intersection([[3, 1, 2, 4, 5], [1, 2, 3, 4], [3, 4, 5, 6]]))
print(Solution().intersection([[1, 2, 3], [4, 5, 6]]))
````

## 10. Vì sao thuật toán đúng?

- **Cách đếm:** Gọi `m = len(nums)`. Vì mỗi mảng con có các giá trị khác nhau, mỗi mảng đóng góp **tối đa `1`** vào `cnt[v]`, và đóng góp đúng `1` khi và chỉ khi `v` thuộc mảng đó. Do đó `cnt[v]` = số mảng chứa `v` ≤ `m`, và `cnt[v] == m` ⇔ `v` thuộc **mọi** mảng. Quét `v` tăng dần nên kết quả tăng dần.
- **Cách `set`:** quy nạp theo số mảng: `common_k = common_{k-1} ∩ set(row_k)` đúng bằng giao `k` mảng đầu; `common_1 = set(nums[0])` hiển nhiên đúng.
- **Cách hai con trỏ:** với hai mảng tăng dần, nếu `a[i] < b[j]` thì `a[i]` nhỏ hơn mọi phần tử còn lại của `b` (`b[j..]` đều `>= b[j]`) nên không thể ở trong giao ⇒ bỏ qua an toàn; đối xứng cho trường hợp ngược lại.
- **Brute force:** kiểm tra trực tiếp định nghĩa; ứng viên chỉ cần lấy từ mảng đầu vì giao ⊆ mảng đầu.

Điều kiện "giá trị trong mỗi mảng khác nhau" là then chốt cho cách đếm; nếu không, ta phải dùng `set(row)` khi đếm (xem mục 13).

## 11. Phân tích Time / Space Complexity

Gọi `m = len(nums)`, `N = tổng số phần tử` (`N <= 1000`), `V = 1000` miền giá trị, `L0 = len(nums[0])`.

| Cách | Time | Space |
|---|---|---|
| 1. Brute force | `O(L0 · N)` | `O(1)` |
| 2. Giao `set` | `O(N + k log k)` | `O(N)` |
| 3. Mảng đếm | `O(N + V)` | `O(V)` |
| 4. Hai con trỏ | `O(N log N)` | `O(N)` |

Mọi cách đều rất nhanh với `N <= 1000`. Cách đếm thể hiện tốt nhất mẹo "đếm đủ `m` lần".

````python
solutions = [SolutionBrute(), SolutionSet(), SolutionCount(), SolutionTwoPointer(), Solution()]

for s in solutions:
    assert s.intersection([[3, 1, 2, 4, 5], [1, 2, 3, 4], [3, 4, 5, 6]]) == [3, 4]
    assert s.intersection([[1, 2, 3], [4, 5, 6]]) == []
print("ví dụ đề: OK")
````

## 13. Các lỗi người mới hay gặp

### Lỗi 1 — Quên sắp xếp kết quả

`set` không có thứ tự. Đề đòi tăng dần; quên `sorted` có thể cho thứ tự lung tung (hoặc kiểu `set` thay vì `list`).

````python
def wrong_no_sort(nums):
    common = set(nums[0])
    for row in nums[1:]:
        common &= set(row)
    return list(common)          # thứ tự không đảm bảo

res = wrong_no_sort([[9, 8, 7], [7, 8, 9, 1]])
print("kiểu:", type(res).__name__, "| không đảm bảo tăng dần:", res)
print("đúng:", Solution().intersection([[9, 8, 7], [7, 8, 9, 1]]))
````

### Lỗi 2 — Khởi tạo `common = set()` rồi giao

Giao với tập rỗng luôn rỗng, nên kết quả luôn là `[]`.

````python
def wrong_empty_init(nums):
    common = set()               # SAI: phải là set(nums[0])
    for row in nums:
        common &= set(row)
    return sorted(common)

print("sai :", wrong_empty_init([[1, 2], [2, 3]]))   # []
print("đúng:", Solution().intersection([[1, 2], [2, 3]]))
````

### Lỗi 3 — Đếm trùng khi một mảng con có giá trị lặp

Đề bảo đảm giá trị trong mỗi mảng khác nhau. Nhưng nếu bạn áp dụng cách đếm cho dữ liệu **có lặp** (ví dụ bài tương tự không có điều kiện này), `cnt[v]` có thể vượt `m` hoặc đạt `m` oan.

````python
def count_no_dedupe(nums):
    cnt = {}
    for row in nums:
        for v in row:
            cnt[v] = cnt.get(v, 0) + 1
    return sorted(v for v, c in cnt.items() if c == len(nums))

def count_with_dedupe(nums):
    cnt = {}
    for row in nums:
        for v in set(row):                 # mỗi mảng chỉ đóng góp 1
            cnt[v] = cnt.get(v, 0) + 1
    return sorted(v for v, c in cnt.items() if c == len(nums))

data = [[1, 1], [2, 3]]                    # vi phạm giả thiết "khác nhau trong mảng"
print("không khử trùng:", count_no_dedupe(data))   # [1] — sai, 1 không có trong mảng thứ hai
print("khử trùng      :", count_with_dedupe(data)) # []
````

### Lỗi 4 — Sửa mảng gốc khi giao

Nếu viết `common = nums[0]` rồi `common.remove(...)`, ta làm hỏng đầu vào (list là đối tượng có thể thay đổi). Hãy tạo bản sao hoặc dùng `set(nums[0])`.

````python
def wrong_mutate(nums):
    common = nums[0]                      # KHÔNG sao chép: cùng một đối tượng
    for v in list(common):
        if any(v not in row for row in nums[1:]):
            common.remove(v)
    return sorted(common)

data = [[1, 2, 3], [2, 3, 4]]
print("kết quả:", wrong_mutate(data), "| đầu vào bị sửa thành:", data[0])
````

## 14. Mẹo làm bài nhanh & nhận dạng pattern

- "Có mặt trong **mọi** tập" ⇒ giao `set`, hoặc đếm tần suất `== số tập`.
- Khi mỗi tập không có phần tử lặp, **đếm tổng** là đủ (không cần `set` từng hàng).
- Miền giá trị nhỏ (`<= 1000`) ⇒ **mảng đếm** + quét `1..V` cho kết quả tự sắp xếp.
- `set.intersection(*list_of_sets)` là một dòng gọn: `sorted(set.intersection(*map(set, nums)))`.
- Họ hàng: Intersection of Two Arrays (349, 350), Find Common Characters (1002), Intersection of Three Sorted Arrays (1213), Common Elements in Two Arrays.

## 15. Biến thể của bài

1. **Giao của hai mảng (LC 349):** `set(a) & set(b)`.
2. **Giao có lặp (LC 350):** dùng `Counter` và `&` giữa hai `Counter` (lấy min).
3. **Hợp nhiều mảng:** `set().union(*nums)`.
4. **Phần tử xuất hiện trong ít nhất `k` mảng:** đếm và so `cnt[v] >= k`.
5. **Giao ba mảng đã sắp xếp (LC 1213):** ba con trỏ.

````python
# Biến thể 4: phần tử xuất hiện trong ít nhất k mảng
def at_least_k_arrays(nums, k):
    cnt = {}
    for row in nums:
        for v in set(row):
            cnt[v] = cnt.get(v, 0) + 1
    return sorted(v for v, c in cnt.items() if c >= k)

print(at_least_k_arrays([[3, 1, 2, 4, 5], [1, 2, 3, 4], [3, 4, 5, 6]], 2))
````

## 19. Đáp án tham khảo cho bài tập

Mỗi đáp án được kiểm tra bằng `assert`.

````python
# Bài 1: 2 và 6 có trong cả ba mảng
assert Solution().intersection([[2, 4, 6], [6, 2], [1, 2, 6, 9]]) == [2, 6]

# Bài 2: cnt[1]=1, cnt[2]=3, cnt[3]=1, cnt[5]=1 -> giao [2]
cnt = {}
for row in [[1, 2], [2, 3], [2, 5]]:
    for v in row:
        cnt[v] = cnt.get(v, 0) + 1
assert cnt == {1: 1, 2: 3, 3: 1, 5: 1}
assert Solution().intersection([[1, 2], [2, 3], [2, 5]]) == [2]

# Bài 3
def union_all_ans(nums):
    u = set()
    for row in nums:
        u.update(row)
    return sorted(u)
assert union_all_ans([[1, 2], [2, 3], [2, 5]]) == [1, 2, 3, 5]

# Bài 4
def intersect_two_sorted_ans(a, b):
    i = j = 0
    out = []
    while i < len(a) and j < len(b):
        if a[i] == b[j]:
            out.append(a[i]); i += 1; j += 1
        elif a[i] < b[j]:
            i += 1
        else:
            j += 1
    return out
assert intersect_two_sorted_ans([1, 3, 5, 7], [3, 4, 5, 8]) == [3, 5]
assert intersect_two_sorted_ans([1, 2], [3, 4]) == []

# Bài 5
def common_in_at_least_half_ans(nums):
    need = (len(nums) + 1) // 2
    cnt = {}
    for row in nums:
        for v in set(row):
            cnt[v] = cnt.get(v, 0) + 1
    return sorted(v for v, c in cnt.items() if c >= need)
assert common_in_at_least_half_ans([[1, 2], [2, 3], [2, 5]]) == [2]
assert common_in_at_least_half_ans([[1, 2], [3, 4]]) == [1, 2, 3, 4]
print("Tất cả đáp án bài tập: OK")
````

from typing import List

class Solution:
    def intersection(self, nums: List[List[int]]) -> List[int]:
        cnt = [0] * 1001
        for row in nums:

                cnt[v] += 1                      # mỗi mảng đóng góp tối đa 1
        m = len(nums)
        return [v for v in range(1, 1001) if cnt[v] == m]
