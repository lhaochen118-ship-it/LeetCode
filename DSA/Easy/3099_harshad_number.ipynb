## 2. Đề bài & ví dụ

Một số nguyên được gọi là **số Harshad** nếu nó **chia hết cho tổng các chữ số** của chính nó.

Cho một số nguyên `x`. Trả về **tổng các chữ số của `x`** nếu `x` là số Harshad; ngược lại trả về `-1`.

**Ví dụ 1**

- Input: `x = 18`
- Output: `9`
- Giải thích: tổng chữ số của `18` là `1 + 8 = 9`; `18` chia hết cho `9` nên `18` là số Harshad. Trả về `9`.

**Ví dụ 2**

- Input: `x = 23`
- Output: `-1`
- Giải thích: tổng chữ số của `23` là `2 + 3 = 5`; `23` không chia hết cho `5` nên không phải số Harshad. Trả về `-1`.

**Ví dụ 3 (tự dựng)**

- Input: `x = 100`
- Output: `1`
- Giải thích: tổng chữ số `1 + 0 + 0 = 1`; `100` chia hết cho `1`. Trả về `1`.

**Ràng buộc:** `1 <= x <= 100`.

````python
import random

examples = [
    (18, 9),
    (23, -1),
    (100, 1),
]
for x, expected in examples:
    print("x =", x, "-> kỳ vọng", expected)
````

## 3. Kiến thức nền

### 3.1. Chữ số của một số: dùng chuỗi `str(x)`

`str(18)` cho chuỗi `"18"`. Duyệt chuỗi ta được từng **ký tự chữ số** `'1'`, `'8'`; đổi mỗi ký tự về số bằng `int(c)`. Khi đó `sum(int(c) for c in str(x))` là tổng chữ số. Lưu ý `'8'` (ký tự) khác `8` (số).

````python
x = 18
s = str(x)
print(s, list(s))                              # 18 ['1', '8']
print([int(c) for c in s])                     # [1, 8]
print(sum(int(c) for c in s))                  # 9
print("8" == 8, int("8") == 8)                 # False True
````

### 3.2. Chữ số bằng toán: `% 10` và `// 10`

- `x % 10`: **chữ số cuối** (phần dư khi chia `10`).
- `x // 10`: `x` **bỏ chữ số cuối** (chia nguyên, làm tròn xuống).

Lặp "lấy chữ số cuối rồi bỏ đi" cho tới khi `x` thành `0` ta duyệt hết các chữ số.

````python
x = 123
print(x % 10, x // 10)       # 3 12
print(12 % 10, 12 // 10)     # 2 1
print(1 % 10, 1 // 10)       # 1 0
````

### 3.3. `divmod(a, b)`: thương và dư cùng lúc

`divmod(a, b)` trả về cặp `(a // b, a % b)`. Với `divmod(123, 10)` ta được `(12, 3)`: một lệnh cho cả "bỏ chữ số cuối" và "lấy chữ số cuối".

````python
print(divmod(123, 10))       # (12, 3)
q, r = divmod(123, 10)
print(q, r)                  # 12 3
````

### 3.4. Ba phép chia và phép chia hết

- `/`: chia **thực**, kết quả luôn là số thực (`18 / 9 = 2.0`).
- `//`: chia **nguyên**, bỏ phần thập phân (`23 // 5 = 4`).
- `%`: **phần dư** (`23 % 5 = 3`).

`a` **chia hết cho** `b` khi và chỉ khi `a % b == 0`. Thứ tự rất quan trọng: `18 % 9 == 0` nhưng `9 % 18 == 9`.

````python
print(18 / 9, 18 // 9, 18 % 9)       # 2.0 2 0
print(23 / 5, 23 // 5, 23 % 5)       # 4.6 4 3
print(18 % 9 == 0, 9 % 18 == 0)      # True False  (thứ tự quan trọng)
````

### 3.5. Biểu thức điều kiện một dòng

`A if điều_kiện else B` cho `A` khi điều kiện đúng, ngược lại cho `B`. Ta sẽ viết "trả về tổng chữ số nếu chia hết, ngược lại `-1`" gọn bằng cách này.

````python
s = 9
print(s if 18 % s == 0 else -1)      # 9
s = 5
print(s if 23 % s == 0 else -1)      # -1
````

## 4. Ý tưởng nền tảng / cách nghĩ trực giác (tính tay)

Với `x = 18`:

1. Chữ số: `1` và `8`; tổng `s = 1 + 8 = 9`.
2. Kiểm tra `18 % 9`: `18 = 2 × 9` nên dư `0` ⇒ chia hết ⇒ Harshad ⇒ trả `9`.

Với `x = 23`:

2. `23 % 5 = 3 ≠ 0` ⇒ không chia hết ⇒ trả `-1`.

Bài toán gồm hai bước đơn giản: **(1) tính tổng chữ số**, **(2) kiểm tra chia hết**.

````python
for x in [18, 23, 7, 100, 11]:
    s = sum(int(c) for c in str(x))
    r = x % s
    print(f"x={x}: tổng chữ số={s}, x % s = {r} -> {'Harshad, trả ' + str(s) if r == 0 else 'không, trả -1'}")
````

## 5. Cách 1 — Brute Force / baseline: tổng chữ số bằng chuỗi

**Ý tưởng.** Đổi `x` thành chuỗi, cộng từng chữ số (`int(c)`), rồi kiểm tra `x % s == 0`. Ngắn, dễ đọc, bám sát lời đề.

````python
def harshad_string(x: int) -> int:
    s = sum(int(c) for c in str(x))       # tổng các chữ số
    if x % s == 0:                         # x chia hết cho tổng chữ số
        return s
    return -1


for x, expected in examples:
    print(x, harshad_string(x), "(kỳ vọng", expected, ")")
````

**Độ phức tạp (Cách 1).** Gọi `D` là số chữ số của `x` (`D = O(log x)`). Chuyển chuỗi và cộng mất `O(D)` ⇒ **Time `O(D)`**, **Space `O(D)`** (chuỗi). Với `x <= 100`, `D <= 3`.

**Nút thắt:** hoàn toàn không có nút thắt về tốc độ; nhưng việc chuyển sang chuỗi chỉ để cộng chữ số hơi "vòng". Cách toán học không cần chuỗi và là kỹ năng nền cho nhiều bài khác.

## 6. Ý tưởng thuật toán / bước đột phá

Ta có thể lấy từng chữ số **không cần chuỗi**: lặp `digit = x % 10` (chữ số cuối), cộng vào tổng, rồi `x //= 10` (bỏ chữ số cuối), cho tới khi `x` bằng `0`. Chi tiết quan trọng: vòng lặp **gọt** `x` về `0`, nên ta phải **lưu bản sao của `x`** (`n = x`) để vòng lặp gọt `n`, còn `x` gốc giữ nguyên cho phép kiểm tra chia hết.

Về toán: `x` Harshad ⇔ `x % S(x) == 0` với `S(x)` là tổng chữ số.

````python
# Gọt chữ số: ví dụ x = 345
x = 345
n = x                        # bản sao để gọt (x gốc còn nguyên)
total = 0
while n > 0:
    digit = n % 10
    total += digit
    print(f"n={n}: chữ số cuối={digit}, tổng={total}")
    n //= 10
print("tổng chữ số =", total, "| x vẫn là", x)
````

## 7. Các cách tiếp theo

### Cách 2 — Vòng lặp toán học `% 10`, `// 10`

**Giải thích từng đoạn.**

1. `n = x` (bản sao), `total = 0`.
2. `while n > 0`: `total += n % 10`; `n //= 10`.
3. So sánh `x % total == 0` (dùng `x` gốc).
4. Trả `total` hoặc `-1`.

````python
def harshad_math(x: int) -> int:
    n = x                          # bản sao để gọt
    total = 0
    while n > 0:
        total += n % 10            # cộng chữ số cuối
        n //= 10                   # bỏ chữ số cuối
    return total if x % total == 0 else -1


for x, expected in examples:
    print(x, harshad_math(x), "(kỳ vọng", expected, ")")
````

````python
# Code nộp LeetCode cho Cách 2 (đổi tên lớp thành Solution khi nộp)
class SolutionMath:
    def sumOfTheDigitsOfHarshadNumber(self, x: int) -> int:
        n, total = x, 0
        while n > 0:
            total += n % 10
            n //= 10
        return total if x % total == 0 else -1


print(SolutionMath().sumOfTheDigitsOfHarshadNumber(18))      # 9
````

**Độ phức tạp Cách 2:** **Time `O(D)`** với `D = ⌊log₁₀ x⌋ + 1`, **Space `O(1)`**.

### Cách 3 — `divmod` (và đệ quy) cho tổng chữ số

**Giải thích.** `divmod(n, 10)` cho `(n // 10, n % 10)` trong một lệnh, nên vòng lặp gọn hơn. Ta cũng có thể viết đệ quy: `digit_sum(n) = n % 10 + digit_sum(n // 10)`, với trường hợp cơ sở `digit_sum(0) = 0`. Đệ quy tốn `O(D)` bộ nhớ ngăn xếp, nhưng với `D <= 3` thì vô hại.

````python
def digit_sum_divmod(n: int) -> int:
    total = 0
    while n:                         # n != 0
        n, digit = divmod(n, 10)     # n //= 10 và digit = n % 10 cùng lúc
        total += digit
    return total


def digit_sum_recursive(n: int) -> int:
    if n == 0:                       # trường hợp cơ sở
        return 0
    return n % 10 + digit_sum_recursive(n // 10)


def harshad_divmod(x: int) -> int:
    s = digit_sum_divmod(x)
    return s if x % s == 0 else -1


def harshad_recursive(x: int) -> int:
    s = digit_sum_recursive(x)
    return s if x % s == 0 else -1


for x, expected in examples:
    print(x, harshad_divmod(x), harshad_recursive(x), "(kỳ vọng", expected, ")")
````

**Độ phức tạp Cách 3:** vòng lặp `divmod`: **Time `O(D)`**, **Space `O(1)`**; đệ quy: **Time `O(D)`**, **Space `O(D)`** do ngăn xếp.

### Cách 4 — Bảng tra: tính trước cho mọi `x` từ `1` đến `100`

**Giải thích.** `x <= 100` nên chỉ có 100 đầu vào. Ta tính một lần `table[x]` rồi tra `O(1)`. Đây là mẫu "tính trước" (precompute) và còn là **bài kiểm chứng chéo**: bảng sinh bằng định nghĩa thô sẽ được dùng để so với các cách còn lại.

````python
def harshad_by_definition(x: int) -> int:
    """Bản 'đọc đề': dùng chuỗi và so sánh thương nguyên."""
    s = sum(int(c) for c in str(x))
    return s if (x // s) * s == x else -1       # x = thương * s  <=>  chia hết


table = {x: harshad_by_definition(x) for x in range(1, 101)}


def harshad_table(x: int) -> int:
    return table[x]


for x, expected in examples:
    print(x, harshad_table(x), "(kỳ vọng", expected, ")")
print("các số Harshad từ 1 đến 100:", [x for x in range(1, 101) if table[x] != -1])
````

**Độ phức tạp Cách 4:** tiền xử lý `O(V · D)` với `V = 100`; mỗi truy vấn **`O(1)`**; **Space `O(V)`**. Với một lần gọi duy nhất thì không đáng, nhưng nếu hỏi hàng nghìn lần thì có lợi.

## 8. TRACE từng bước

Ta trace Cách 2 (vòng lặp toán học) trên `x = 18` (Harshad) và `x = 23` (không).

````python
def trace_harshad(x: int) -> int:
    n = x
    total = 0
    print(f"x = {x}")
    while n > 0:
        digit = n % 10
        total += digit
        print(f"  n={n}: chữ số cuối={digit}, tổng={total}, n // 10 = {n // 10}")
        n //= 10
    r = x % total
    print(f"  tổng chữ số = {total}; x % tổng = {x} % {total} = {r}")
    result = total if r == 0 else -1
    print("  ->", result)
    return result


assert trace_harshad(18) == 9
print()
assert trace_harshad(23) == -1
````

Bảng trace Cách 2 với `x = 18`:

| bước | n | `n % 10` | tổng | `n // 10` |
|---|---|---|---|---|
| 1 | 18 | 8 | 8 | 1 |
| 2 | 1 | 1 | 9 | 0 |

Hết chữ số (`n = 0`). `x % tổng = 18 % 9 = 0` ⇒ Harshad ⇒ trả `9`.

Với `x = 23`: `3` rồi `2` ⇒ tổng `5`; `23 % 5 = 3` ⇒ trả `-1`.

## 9. Code nộp LeetCode

Bản nộp dùng Cách 2 (vòng lặp toán học): không phụ thuộc chuỗi, bộ nhớ `O(1)`, ngắn gọn.

````python
class Solution:
    def sumOfTheDigitsOfHarshadNumber(self, x: int) -> int:
        n = x                                   # bản sao để "gọt" chữ số (giữ nguyên x)
        total = 0
        while n > 0:
            total += n % 10                     # cộng chữ số cuối
            n //= 10                            # bỏ chữ số cuối
        return total if x % total == 0 else -1  # Harshad: x chia hết cho tổng chữ số


sol = Solution()
for x, expected in examples:
    print(x, "->", sol.sumOfTheDigitsOfHarshadNumber(x))
````

## 10. Vì sao thuật toán đúng

- **Tổng chữ số đúng:** mỗi vòng lặp cộng chữ số cuối của `n` rồi bỏ nó; `n` giảm nghiêm ngặt về `0` sau đúng `D` vòng, nên `total` là tổng đúng `D` chữ số. Vì `x >= 1` nên `D >= 1` và `total >= 1`.
- **Không chia cho `0`:** `x >= 1` có ít nhất một chữ số khác `0` (chữ số đầu), nên `total >= 1` và `x % total` luôn hợp lệ.
- **Kiểm tra Harshad đúng:** theo định nghĩa, `x` là Harshad ⇔ `total` chia hết `x` ⇔ `x % total == 0`. Ta dùng `x` **gốc** (không bị gọt) cho phép chia này.
- **Trả đúng giá trị:** nếu Harshad thì đề yêu cầu trả **tổng chữ số** (`total`), ngược lại `-1`; các cách khác nhau chỉ khác cách tính `total`.

## 11. Phân tích Time / Space Complexity

Gọi `D` là số chữ số của `x` (`D = O(log x)`, ở đây `D <= 3`).

| Cách | Time | Space |
|---|---|---|
| 1. Chuỗi | `O(D)` | `O(D)` |
| 2. Vòng lặp `% 10`, `// 10` | `O(D)` | `O(1)` |
| 3. `divmod` / đệ quy | `O(D)` | `O(1)` / `O(D)` |
| 4. Bảng tra | `O(1)` mỗi truy vấn | `O(100)` |

Với `x <= 100`, tất cả đều chạy tức thì.

````python
sol = Solution()
impls = {
    "string": harshad_string,
    "math": harshad_math,
    "divmod": harshad_divmod,
    "recursive": harshad_recursive,
    "table": harshad_table,
    "leetcode": sol.sumOfTheDigitsOfHarshadNumber,
}


def check_all(x, expected, label):
    for key, fn in impls.items():
        actual = fn(x)
        assert actual == expected, (label, key, x, actual, expected)
    print("PASS", label)


for idx, (x, expected) in enumerate(examples, 1):
    check_all(x, expected, f"ví dụ {idx}")
````

### 12.2. Phản ví dụ cho các cách làm sai

Hai lỗi phổ biến: (a) **kiểm tra chia hết sau khi `x` đã bị gọt** (nên `x == 0`); (b) **đảo thứ tự phép chia** (`s % x` thay vì `x % s`).

````python
def wrong_lost_x(x):
    total = 0
    while x > 0:
        total += x % 10
        x //= 10                       # x bị gọt về 0
    return total if x % total == 0 else -1     # SAI: x = 0 nên 0 % total == 0 luôn đúng


def wrong_swapped(x):
    s = sum(int(c) for c in str(x))
    return s if s % x == 0 else -1     # SAI: đảo thứ tự


print(wrong_lost_x(23), "!=", harshad_string(23))
print(wrong_swapped(18), "!=", harshad_string(18))
assert wrong_lost_x(23) != harshad_string(23)
assert wrong_swapped(18) != harshad_string(18)
````

## 13. Các lỗi người mới hay gặp

### Lỗi 1: Làm mất `x` khi vòng `while x > 0` gọt hết chữ số

````python
def sai_mat_x(x):
    total = 0
    while x > 0:
        total += x % 10
        x //= 10                              # x bị gọt về 0
    return total if x % total == 0 else -1    # SAI: x lúc này bằng 0


print("sai:", sai_mat_x(23), "| đúng:", Solution().sumOfTheDigitsOfHarshadNumber(23))
````

**Vì sao sai:** sau vòng lặp `x = 0`, và `0 % total == 0` **luôn đúng**, nên mọi số đều bị coi là Harshad (ở trên `23` trả `5` thay vì `-1`). **Cách sửa:** dùng bản sao (`n = x`) để gọt, giữ `x` nguyên.

### Lỗi 2: Đảo thứ tự `x % s` thành `s % x`

````python
def sai_dao_thu_tu(x):
    s = sum(int(c) for c in str(x))
    return s if s % x == 0 else -1            # SAI: s thường nhỏ hơn x


print("sai:", sai_dao_thu_tu(18), "| đúng:", Solution().sumOfTheDigitsOfHarshadNumber(18))
````

**Vì sao sai:** đề hỏi `x` **chia hết cho** tổng chữ số, tức `x % s == 0`. `s % x` với `s < x` là `s` (khác `0`) nên luôn "không chia hết". **Cách sửa:** `x % s`.

### Lỗi 3: Dùng `/` (chia thực) thay cho `//` khi tách chữ số

````python
def sai_chia_thuc(x):
    total = 0
    n = x
    count = 0
    while n >= 1 and count < 10:              # count chỉ để tránh vòng lặp dài khi demo
        total += n % 10                       # với số thực, % 10 không còn là chữ số
        n = n / 10                            # SAI: 18 -> 1.8 -> 0.18 ...
        count += 1
    return total


print("sai:", sai_chia_thuc(18), "| đúng tổng chữ số:", 1 + 8)
print("18 / 10 =", 18 / 10, "| 18 % 10 =", 18 % 10, "| (18 / 10) % 10 =", (18 / 10) % 10)
````

**Vì sao sai:** `/` cho số **thực** (`1.8`), nên `n % 10` ở bước sau là `1.8` thay vì chữ số `1`. Tổng chữ số tính ra là một số thực lẻ. **Cách sửa:** luôn dùng `//` khi bỏ chữ số cuối.

### Lỗi 4: Trả `True/False` hoặc trả `x` thay vì tổng chữ số

````python
def sai_tra_bool(x):
    s = sum(int(c) for c in str(x))
    return x % s == 0                          # SAI: đề yêu cầu trả tổng chữ số hoặc -1


print("sai:", sai_tra_bool(18), "| đúng:", Solution().sumOfTheDigitsOfHarshadNumber(18))
print("sai:", sai_tra_bool(23), "| đúng:", Solution().sumOfTheDigitsOfHarshadNumber(23))
````

**Vì sao sai:** hàm trả `True`/`False` (và `True == 1`, `False == 0`), không phải `9` / `-1`. Đây là lỗi đọc đề. **Cách sửa:** `return s if x % s == 0 else -1`.

5. **Happy Number (202)**: lặp tổng bình phương chữ số.
