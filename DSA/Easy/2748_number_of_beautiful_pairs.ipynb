## 1. Mục tiêu học

Sau notebook này bạn sẽ:

1. Hiểu đề: đếm các cặp `(i, j)` với `i < j` mà **chữ số đầu** của `nums[i]` và **chữ số cuối** của `nums[j]` **nguyên tố cùng nhau** (`gcd = 1`).
2. Biết tách **chữ số cuối** bằng `x % 10`, **chữ số đầu** bằng chuỗi `str(x)[0]` hoặc vòng `while x >= 10: x //= 10`, và dùng `math.gcd`.
3. Hiểu **nguyên tố cùng nhau (coprime)** là gì và vì sao `gcd(a, b) == 1` kiểm tra được điều đó.
4. Đi từ **Brute Force O(n²)** → **đếm theo chữ số đầu đã gặp (chỉ 9 giá trị)** → **bảng coprime tính trước** → **bảng đếm cặp chữ số 9×9** → **đếm từ phải sang trái**, và nhận ra bài này là "đếm theo nhóm nhỏ" thay vì so từng cặp.
5. Tránh các lỗi: so sánh chuỗi với số trong `gcd`, lấy chữ số đầu bằng `x // 10`, đảo vai trò `i`/`j`, và đếm nhầm một số với chính nó.

## 2. Đề bài & ví dụ

Cho mảng số nguyên `nums` (chỉ số bắt đầu từ `0`). Một cặp chỉ số `(i, j)` với `0 <= i < j < nums.length` được gọi là **đẹp (beautiful)** nếu **chữ số đầu tiên** của `nums[i]` và **chữ số cuối cùng** của `nums[j]` là **nguyên tố cùng nhau**.

Hai số nguyên `x` và `y` gọi là nguyên tố cùng nhau nếu **ước chung lớn nhất** của chúng bằng `1`, tức `gcd(x, y) == 1`.

Trả về **tổng số cặp đẹp** trong `nums`.

| Ví dụ | nums | Output | Giải thích |
|---|---|---|---|
| 1 | `[2,5,1,4]` | `5` | Các cặp đẹp: `(0,1)`: 2 và 5; `(0,2)`: 2 và 1; `(1,2)`: 5 và 1; `(1,3)`: 5 và 4; `(2,3)`: 1 và 4. Cặp `(0,3)` không đẹp vì `gcd(2, 4) = 2`. |
| 2 | `[11,21,12]` | `2` | Cặp `(0,1)`: đầu của 11 là 1, cuối của 21 là 1 ⇒ gcd 1. Cặp `(0,2)`: 1 và 2. Cặp `(1,2)`: đầu của 21 là 2, cuối của 12 là 2 ⇒ gcd 2 ⇒ không đẹp. |

Constraints:

- `2 <= nums.length <= 100`
- `1 <= nums[i] <= 9999`
- `nums[i] % 10 != 0` (chữ số cuối khác `0`)

````python
import math
nums = [2, 5, 1, 4]
n = len(nums)
for i in range(n):
    for j in range(i + 1, n):
        first = int(str(nums[i])[0])
        last = nums[j] % 10
        print((i, j), "đầu", first, "cuối", last, "gcd", math.gcd(first, last), "đẹp" if math.gcd(first, last) == 1 else "")
````

### 3.1. Chữ số cuối bằng `x % 10`

`%` là phép chia dư. Chia cho `10`, số dư là **chữ số hàng đơn vị**: `1234 % 10 = 4`.

````python
print(1234 % 10)     # 4
print(7 % 10)        # 7
print(21 % 10)       # 1
````

### 3.2. Chữ số đầu tiên: hai cách

**Cách A — dùng chuỗi:** `str(x)[0]` là ký tự đầu (kiểu `str`), bọc `int(...)` để thành số.

**Cách B — dùng toán:** chia nguyên `x //= 10` liên tục cho tới khi `x < 10`; khi đó `x` chính là chữ số đầu.

````python
x = 1234
print(str(x)[0], type(str(x)[0]))          # 1 <class 'str'>
print(int(str(x)[0]))                      # 1

y = 1234
while y >= 10:
    y //= 10
    print("y =", y)                        # 123, 12, 1
print("chữ số đầu:", y)                    # 1
````

### 3.3. Ước chung lớn nhất `math.gcd` và "nguyên tố cùng nhau"

`math.gcd(a, b)` trả ước chung lớn nhất của `a` và `b`. Hai số **nguyên tố cùng nhau (coprime)** khi `gcd(a, b) == 1`, tức chúng không có ước chung nào lớn hơn 1.

- `gcd(2, 5) = 1` ⇒ coprime.
- `gcd(2, 4) = 2` ⇒ không coprime.
- `gcd(1, x) = 1` với mọi `x` ⇒ chữ số `1` coprime với mọi chữ số.

````python
import math
print(math.gcd(2, 5), math.gcd(2, 4), math.gcd(1, 9), math.gcd(6, 9))   # 1 2 1 3
print(math.gcd(7, 7))      # 7  (một số chỉ coprime với chính nó nếu bằng 1)
print(math.gcd(1, 1))      # 1
````

### 3.4. Danh sách đếm theo chữ số (mảng tần suất)

Vì chữ số chỉ thuộc `0..9`, ta có thể dùng danh sách độ dài 10: `cnt[d]` = số lần chữ số `d` đã xuất hiện. Đây là **counting array** (mảng đếm), thay cho dict.

````python
cnt = [0] * 10
for d in [2, 5, 5, 1]:
    cnt[d] += 1
print(cnt)             # [0, 1, 1, 0, 0, 2, 0, 0, 0, 0]
print(cnt[5])          # 2
````

### 3.5. Duyệt cặp `i < j` và cách "đếm trước, cập nhật sau"

Khi duyệt `j` từ trái sang phải, mọi `i < j` chính là những phần tử **đã duyệt**. Nếu ta gom thông tin của chúng vào một cấu trúc (ví dụ `cnt`), thì tại `j` chỉ cần tra cấu trúc đó thay vì quay lại từng `i`. Quan trọng: **đếm cho `j` trước, rồi mới thêm `nums[j]` vào cấu trúc** để không ghép `j` với chính nó.

````python
cnt = [0] * 10
nums = [2, 5, 1, 4]
for j, x in enumerate(nums):
    print(f"j={j}: các chữ số đầu đã gặp (i<j) = {cnt}")
    cnt[int(str(x)[0])] += 1       # thêm SAU khi đã dùng
````

## 4. Ý tưởng nền tảng / cách nghĩ trực giác

Hãy tách bài toán thành hai "nửa":

- Nửa **trái** của cặp: chỉ quan tâm **chữ số đầu** của `nums[i]`.
- Nửa **phải** của cặp: chỉ quan tâm **chữ số cuối** của `nums[j]`.

Tính tay ví dụ 2: `nums = [11, 21, 12]`.

| số | chữ số đầu | chữ số cuối |
|---|---|---|
| 11 | 1 | 1 |
| 21 | 2 | 1 |
| 12 | 1 | 2 |

Cặp `(i, j)` đẹp ⇔ `gcd(đầu[i], cuối[j]) = 1`:

- `(0,1)`: `gcd(1, 1) = 1` ✓
- `(0,2)`: `gcd(1, 2) = 1` ✓
- `(1,2)`: `gcd(2, 2) = 2` ✗

Tổng `2`. Nhận xét: **chỉ có 9 chữ số đầu khả dĩ (1–9)** và **9 chữ số cuối khả dĩ (1–9)**, nên số "loại" cặp chỉ là `9 × 9 = 81`, rất nhỏ so với `n²` khi `n` lớn.

````python
nums = [11, 21, 12]
first = [int(str(x)[0]) for x in nums]
last = [x % 10 for x in nums]
print("đầu :", first)    # [1, 2, 1]
print("cuối:", last)     # [1, 1, 2]
````

## 5. Cách 1 — Brute Force / baseline: thử mọi cặp (i, j)

**Ý tưởng:** hai vòng lồng nhau `i < j`; với mỗi cặp tính chữ số đầu của `nums[i]`, chữ số cuối của `nums[j]`, rồi kiểm tra `gcd == 1`.

````python
import math
from typing import List

class SolutionBrute:
    def countBeautifulPairs(self, nums: List[int]) -> int:
        count = 0
        n = len(nums)
        for i in range(n):
            first = int(str(nums[i])[0])
            for j in range(i + 1, n):
                last = nums[j] % 10
                if math.gcd(first, last) == 1:
                    count += 1
        return count
````

````python
sol = SolutionBrute()
print(sol.countBeautifulPairs([2, 5, 1, 4]))      # 5
print(sol.countBeautifulPairs([11, 21, 12]))      # 2
````

**Độ phức tạp:** `n(n − 1)/2` cặp, mỗi cặp một `gcd` (rất nhanh với số một chữ số) ⇒ Time `O(n²)`, Space `O(1)`.

**Nút thắt:** với `n <= 100` thì chạy tức thì. Nhưng ta lặp lại việc "tính chữ số đầu / kiểm tra coprime" cho rất nhiều cặp có **cùng** cặp chữ số. Chỉ có 81 cặp chữ số khác nhau; ta nên **đếm theo chữ số**, không so từng cặp.

## 6. Ý tưởng thuật toán / bước đột phá

**Đột phá 1 — đếm theo chữ số đầu đã gặp.** Duyệt `j` từ trái sang phải. Giữ `cnt[d]` = số phần tử `i < j` có chữ số đầu bằng `d`. Tại `j` (chữ số cuối `l = nums[j] % 10`), số cặp đẹp kết thúc ở `j` là:

`sum(cnt[d] for d in 1..9 if gcd(d, l) == 1)`

Sau đó cập nhật `cnt[chữ số đầu của nums[j]] += 1`. Mỗi phần tử tốn tối đa 9 phép kiểm tra ⇒ `O(9n)`, tức tuyến tính theo `n`.

**Đột phá 2 — tính trước bảng coprime.** `gcd` chỉ cần tính cho 81 cặp chữ số một lần. Lưu `coprime[a][b] = (gcd(a, b) == 1)`, hoặc danh sách `co[l] = [d for d in 1..9 if gcd(d, l) == 1]`.

**Đột phá 3 — tách "đếm" khỏi "kiểm tra coprime".** Ta có thể đếm `pair[a][b]` = số cặp `(i, j)` với đầu `a`, cuối `b` trước, rồi cuối cùng cộng các ô mà `gcd(a, b) == 1`. Cách này tách rõ hai việc.

**Đột phá 4 — đối xứng.** Đếm từ phải sang trái với `cntLast[l]` = số phần tử `j > i` có chữ số cuối `l` cũng đúng.

````python
import math
# Ví dụ bước 1 trên [2, 5, 1, 4]
cnt = [0] * 10
total = 0
for j, x in enumerate([2, 5, 1, 4]):
    last = x % 10
    add = sum(cnt[d] for d in range(1, 10) if math.gcd(d, last) == 1)
    total += add
    print(f"j={j}: cuối={last}, cặp mới={add}, tổng={total}")
    cnt[int(str(x)[0])] += 1
````

## 7. Các cách tiếp theo

### Cách 2 — Mảng đếm chữ số đầu + kiểm tra `gcd` trực tiếp

**Từng đoạn:**

- `cnt = [0] * 10`, `ans = 0`.
- Với mỗi `x`: `last = x % 10`; cộng `cnt[d]` cho mọi `d` trong `1..9` thoả `gcd(d, last) == 1`.
- Sau đó thêm chữ số đầu của `x` vào `cnt`. Chữ số đầu tính bằng vòng `while first >= 10: first //= 10`.

````python
class SolutionCount:
    def countBeautifulPairs(self, nums: List[int]) -> int:
        cnt = [0] * 10
        ans = 0
        for x in nums:
            last = x % 10
            for d in range(1, 10):
                if cnt[d] and math.gcd(d, last) == 1:
                    ans += cnt[d]
            first = x
            while first >= 10:
                first //= 10
            cnt[first] += 1
        return ans
````

````python
sol = SolutionCount()
print(sol.countBeautifulPairs([2, 5, 1, 4]))      # 5
print(sol.countBeautifulPairs([11, 21, 12]))      # 2
````

**Độ phức tạp:** Time `O(9n)` (và `O(n log x)` cho tìm chữ số đầu với `x <= 9999` tối đa 3 lần chia), Space `O(1)` (mảng 10 phần tử).

### Cách 3 — Bảng coprime tính trước (code nộp)

**Từng đoạn:** trước khi duyệt, tính `co[l]` = danh sách các chữ số đầu `d` coprime với chữ số cuối `l`. Trong vòng lặp chỉ cần `sum(cnt[d] for d in co[last])`.

````python
class SolutionTable:
    def countBeautifulPairs(self, nums: List[int]) -> int:
        co = [[d for d in range(1, 10) if math.gcd(d, l) == 1] for l in range(10)]
        cnt = [0] * 10
        ans = 0
        for x in nums:
            ans += sum(cnt[d] for d in co[x % 10])
            cnt[int(str(x)[0])] += 1
        return ans
````

````python
print([ (l, [d for d in range(1, 10) if math.gcd(d, l) == 1]) for l in range(1, 10)])
sol = SolutionTable()
print(sol.countBeautifulPairs([2, 5, 1, 4]))      # 5
print(sol.countBeautifulPairs([11, 21, 12]))      # 2
````

**Độ phức tạp:** Time `O(n × 9)` thực tế nhỏ hơn vì mỗi `co[l]` ngắn; Space `O(1)`.

### Cách 4 — Bảng đếm cặp chữ số 9×9, cộng coprime ở cuối

**Từng đoạn:**

- `cnt[d]`: số phần tử đã gặp có chữ số đầu `d`.
- `pair[a][b]`: số cặp `(i, j)` có đầu `a` (ở `i`) và cuối `b` (ở `j`). Tại `j` với cuối `b`: với mọi `a`, `pair[a][b] += cnt[a]`.
- Cuối cùng cộng `pair[a][b]` với mọi `(a, b)` có `gcd(a, b) == 1`.

````python
class SolutionPairTable:
    def countBeautifulPairs(self, nums: List[int]) -> int:
        cnt = [0] * 10
        pair = [[0] * 10 for _ in range(10)]
        for x in nums:
            b = x % 10
            for a in range(1, 10):
                pair[a][b] += cnt[a]
            cnt[int(str(x)[0])] += 1
        return sum(pair[a][b] for a in range(1, 10) for b in range(10) if math.gcd(a, b) == 1)
````

````python
sol = SolutionPairTable()
print(sol.countBeautifulPairs([2, 5, 1, 4]))      # 5
print(sol.countBeautifulPairs([11, 21, 12]))      # 2
````

**Độ phức tạp:** Time `O(9n + 81)`, Space `O(100)`. Tiện khi cần thay điều kiện "coprime" bằng điều kiện khác (chỉ sửa phần cộng cuối).

### Cách 5 — Đếm từ phải sang trái (chữ số cuối phía sau)

Đảo vai trò: duyệt `i` từ `n − 1` về `0`. `cntLast[l]` = số phần tử `j > i` có chữ số cuối `l`. Tại `i` với chữ số đầu `f`: cộng `cntLast[l]` cho mọi `l` có `gcd(f, l) == 1`; rồi thêm `nums[i] % 10` vào `cntLast`.

````python
class SolutionRightToLeft:
    def countBeautifulPairs(self, nums: List[int]) -> int:
        cnt_last = [0] * 10
        ans = 0
        for i in range(len(nums) - 1, -1, -1):
            f = int(str(nums[i])[0])
            for l in range(10):
                if cnt_last[l] and math.gcd(f, l) == 1:
                    ans += cnt_last[l]
            cnt_last[nums[i] % 10] += 1
        return ans
````

````python
sol = SolutionRightToLeft()
print(sol.countBeautifulPairs([2, 5, 1, 4]))      # 5
print(sol.countBeautifulPairs([11, 21, 12]))      # 2
````

**Độ phức tạp:** Time `O(10n)`, Space `O(1)`.

**Lưu ý chữ số `0`:** nhờ ràng buộc `nums[i] % 10 != 0`, chữ số cuối không bao giờ là `0`. (Nếu có `0`, `gcd(f, 0) = f` nên chỉ `f = 1` thoả; các cách dùng `math.gcd` vẫn đúng.)

## 8. TRACE từng bước

Trace Cách 3 trên `nums = [2, 5, 1, 4]`:

| j | `nums[j]` | cuối | `co[cuối]` | `cnt` trước (đầu đã gặp) | cặp mới | tổng | `cnt` sau |
|---|---|---|---|---|---|---|---|
| 0 | 2 | 2 | `[1,3,5,7,9]` | `{}` | 0 | 0 | `{2:1}` |
| 1 | 5 | 5 | `[1,2,3,4,6,7,8,9]` | `{2:1}` | 1 (đầu 2) | 1 | `{2:1,5:1}` |
| 2 | 1 | 1 | `[1..9]` | `{2:1,5:1}` | 2 | 3 | `{2:1,5:1,1:1}` |
| 3 | 4 | 4 | `[1,3,5,7,9]` | `{2:1,5:1,1:1}` | 2 (đầu 5, 1) | 5 | `{...,4:1}` |

Ở `j = 3`: chữ số đầu đã gặp là 2, 5, 1. `gcd(2, 4) = 2` ⇒ không; `gcd(5, 4) = 1` ✓; `gcd(1, 4) = 1` ✓ ⇒ 2 cặp. Tổng `5`, khớp đáp án.

````python
def trace_table(nums):
    co = [[d for d in range(1, 10) if math.gcd(d, l) == 1] for l in range(10)]
    cnt = [0] * 10
    ans = 0
    for j, x in enumerate(nums):
        last = x % 10
        first = int(str(x)[0])
        gained = [(d, cnt[d]) for d in co[last] if cnt[d]]
        add = sum(c for _, c in gained)
        ans += add
        shown = {d: c for d, c in enumerate(cnt) if c}
        print(f"j={j}, x={x}, cuối={last}, cnt_trước={shown}, cặp mới={add} {gained}, tổng={ans}")
        cnt[first] += 1
    print("kết quả:", ans)
    return ans

trace_table([2, 5, 1, 4])
trace_table([11, 21, 12])
````

````python
def trace_brute(nums):
    n = len(nums)
    ans = 0
    for i in range(n):
        for j in range(i + 1, n):
            f = int(str(nums[i])[0])
            l = nums[j] % 10
            g = math.gcd(f, l)
            if g == 1:
                ans += 1
            print(f"({i},{j}): đầu({nums[i]})={f}, cuối({nums[j]})={l}, gcd={g} -> {'đẹp' if g == 1 else 'không'}")
    print("kết quả:", ans)
    return ans

trace_brute([11, 21, 12])
````

## 9. Code nộp LeetCode

Dùng Cách 3 (bảng coprime tính trước + mảng đếm).

````python
import math
from typing import List

class Solution:
    def countBeautifulPairs(self, nums: List[int]) -> int:
        # co[l] = các chữ số đầu d (1..9) nguyên tố cùng nhau với chữ số cuối l
        co = [[d for d in range(1, 10) if math.gcd(d, l) == 1] for l in range(10)]
        cnt = [0] * 10            # cnt[d] = số phần tử đã gặp có chữ số đầu = d
        ans = 0
        for x in nums:
            ans += sum(cnt[d] for d in co[x % 10])   # ghép x làm "j" với mọi "i" đã gặp
            cnt[int(str(x)[0])] += 1                  # rồi mới thêm x làm "i" cho các j sau
        return ans
````

````python
print(Solution().countBeautifulPairs([2, 5, 1, 4]))      # 5
print(Solution().countBeautifulPairs([11, 21, 12]))      # 2
````

## 10. Vì sao thuật toán đúng

**Bất biến vòng lặp.** Ngay trước khi xử lý `nums[j]` (đếm `j = 0, 1, ...`), `cnt[d]` bằng **số chỉ số `i < j`** có chữ số đầu của `nums[i]` bằng `d`.

- Ban đầu `cnt` toàn `0` (chưa có `i < 0`).
- Sau khi xử lý `nums[j]`, ta tăng `cnt[đầu(nums[j])]` nên bất biến đúng cho `j + 1`.

**Tại bước `j`.** Số cặp đẹp có chỉ số phải bằng `j` là số `i < j` thoả `gcd(đầu(nums[i]), cuối(nums[j])) = 1`. Nhóm các `i` theo chữ số đầu `d`: có `cnt[d]` chỉ số với mỗi `d`, và điều kiện chỉ phụ thuộc `d` (và `cuối(nums[j])`). Vậy số cặp là `sum(cnt[d] for d if gcd(d, cuối) == 1)`, đúng bằng phần cộng vào `ans`.

**Không bỏ sót, không đếm trùng.** Mỗi cặp `(i, j)`, `i < j`, được đếm đúng một lần, tại bước `j`. Việc **cộng trước, tăng `cnt` sau** đảm bảo `i` luôn nhỏ hơn `j` thật sự (không ghép `j` với `j`).

**Chữ số đầu thuộc `1..9`.** Vì `nums[i] >= 1` nên chữ số đầu khác `0`; do đó `cnt[0]` luôn bằng `0` và bảng `co` chỉ cần `d` từ `1` tới `9`.

Cách 1 (vét cạn) là bản dịch thẳng của đề; test ở mục 12 chứng minh bằng thực nghiệm các cách còn lại khớp với nó.

## 13. Các lỗi người mới hay gặp

### Lỗi 1: Đưa chuỗi vào `math.gcd`

````python
nums = [11, 21, 12]
try:
    print(math.gcd(str(nums[0])[0], nums[1] % 10))
except TypeError as e:
    print("lỗi:", e)
print("đúng:", math.gcd(int(str(nums[0])[0]), nums[1] % 10))
````

**Vì sao sai:** `str(x)[0]` là **ký tự** `'1'`, không phải số `1`. `math.gcd` cần số nguyên ⇒ phải bọc `int(...)`.

### Lỗi 2: Lấy chữ số đầu bằng `x // 10`

````python
def sai_dau(x: int) -> int:
    return x // 10          # sai: chỉ bỏ chữ số cuối

print(sai_dau(1234), "(sai, muốn 1)")      # 123
print(sai_dau(7), "(sai, muốn 7)")         # 0
print(int(str(1234)[0]), int(str(7)[0]))   # 1 7  (đúng)
````

**Vì sao sai:** `x // 10` **bỏ chữ số cuối**, không cho chữ số đầu. Muốn chữ số đầu: `int(str(x)[0])` hoặc lặp `while x >= 10: x //= 10`.

### Lỗi 3: Đảo vai trò `i` và `j`

````python
def sai_dao_vai_tro(nums: List[int]) -> int:
    n, count = len(nums), 0
    for i in range(n):
        for j in range(i + 1, n):
            # sai: lấy CUỐI của nums[i] và ĐẦU của nums[j]
            if math.gcd(nums[i] % 10, int(str(nums[j])[0])) == 1:
                count += 1
    return count

print("sai :", sai_dao_vai_tro([11, 21, 12]))               # 3
print("đúng:", Solution().countBeautifulPairs([11, 21, 12]))  # 2
````

**Vì sao sai:** đề quy định **đầu của `nums[i]`** (phần tử đứng trước) và **cuối của `nums[j]`** (phần tử đứng sau). Với số có một chữ số hai vai trò trùng nhau nên lỗi này khó thấy ở ví dụ 1 nhưng lộ ra ở ví dụ 2.

### Lỗi 4: Thêm vào `cnt` trước khi đếm (ghép một số với chính nó)

````python
def sai_them_truoc(nums: List[int]) -> int:
    cnt = [0] * 10
    ans = 0
    for x in nums:
        cnt[int(str(x)[0])] += 1             # sai: thêm trước
        ans += sum(cnt[d] for d in range(1, 10) if math.gcd(d, x % 10) == 1)
    return ans

print("sai :", sai_them_truoc([11, 21, 12]))                  # 5
print("đúng:", Solution().countBeautifulPairs([11, 21, 12]))  # 2
````

**Vì sao sai:** sau khi thêm `x` vào `cnt`, bước đếm tính luôn `i = j` (ghép `x` với chính nó). Phải **đếm trước, thêm sau**.

### Lỗi 5: Dùng `gcd > 1` thay vì `gcd == 1`

````python
def sai_dieu_kien(nums: List[int]) -> int:
    n, count = len(nums), 0
    for i in range(n):
        for j in range(i + 1, n):
            if math.gcd(int(str(nums[i])[0]), nums[j] % 10) > 1:     # sai: đếm cặp KHÔNG coprime
                count += 1
    return count

print("sai :", sai_dieu_kien([2, 5, 1, 4]))                   # 1
print("đúng:", Solution().countBeautifulPairs([2, 5, 1, 4]))  # 5
````

**Vì sao sai:** "nguyên tố cùng nhau" nghĩa là `gcd == 1`. Đếm `gcd > 1` cho đúng phần bù (`6 − 5 = 1` ở ví dụ 1).

## 14. Mẹo làm bài nhanh & nhận dạng pattern

- Đề nói "đếm cặp `(i, j)` với `i < j` thoả điều kiện chỉ phụ thuộc **một tính chất nhỏ** của mỗi phần tử" → nghĩ tới **mảng đếm theo tính chất** thay vì hai vòng lặp.
- Khi giá trị tính chất chỉ có ít loại (ở đây 9), `O(n × số_loại)` rất tốt.
- Công thức xương sống: **duyệt `j`; cộng `Σ cnt[loại hợp lệ]`; sau đó `cnt[loại của j] += 1`.** (Giống Two Sum, Count Good Pairs...)
- Chữ số đầu: `int(str(x)[0])` (ngắn gọn); chữ số cuối: `x % 10`.
- `gcd(1, x) = 1`: chữ số `1` ở đầu luôn "đẹp" với mọi chữ số cuối.
- Hãy kiểm tra các số một chữ số, và các số mà đầu = cuối (như `11`).

## 15. Biến thể của bài

1. **Đếm cặp mà đầu của `nums[i]` bằng cuối của `nums[j]`** (điều kiện khác: bằng nhau): dùng `cnt[d]` và cộng `cnt[last]`.
2. **`gcd(đầu, cuối) = g` cho trước** hoặc **`gcd > 1`**: thay điều kiện ở bảng coprime.
3. **`n` lên tới `10^5` và `nums[i]` lên `10^9`:** cùng một thuật toán (chỉ có 9 chữ số đầu, 10 chữ số cuối).
4. **Cho phép chữ số cuối `0`:** `gcd(f, 0) = f`, chỉ `f = 1` là coprime với `0`; thuật toán không đổi.
5. **Cặp không có thứ tự** (`i != j`, mỗi cặp đếm một lần theo mọi hướng): đếm theo cả hai hướng.
6. **Đếm các cặp `(i, j)` mà tổng hai chữ số / tích hai chữ số có tính chất nào đó:** cùng khung "đếm theo chữ số".

## 16. So sánh các cách làm

| Cách | Ý tưởng | Time | Space | Khi nào dùng |
|---|---|---|---|---|
| 1. Brute force | mọi cặp `i < j`, `gcd` từng cặp | `O(n²)` | `O(1)` | `n` nhỏ, làm oracle |
| 2. Đếm + `gcd` | `cnt` đầu đã gặp, `gcd` mỗi lần | `O(9n)` | `O(1)` | khi chưa tạo bảng |
| 3. Bảng coprime | `co[l]` tính trước | `O(n)` | `O(1)` | **code nộp**, gọn |
| 4. Bảng cặp 9×9 | đếm cặp, lọc coprime cuối | `O(9n + 81)` | `O(100)` | điều kiện có thể đổi |
| 5. Phải sang trái | đếm cuối phía sau | `O(10n)` | `O(1)` | đối xứng, kiểm tra chéo |

class Solution:

        return ans
