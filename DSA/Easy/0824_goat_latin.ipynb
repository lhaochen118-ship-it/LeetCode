## 1. Đề bài và ví dụ

## 3. Biến từng quy tắc thành một hàm nhỏ

Bài mô phỏng: chia nhỏ để dễ kiểm tra. Viết hàm `convert_word(w, idx)` cho **một** từ.

````python
VOWELS = set("aeiouAEIOU")

def convert_word(w: str, idx: int) -> str:
    if w[0] in VOWELS:            # quy tắc 1
        core = w
    else:                          # quy tắc 2
        core = w[1:] + w[0]
    return core + "ma" + "a" * idx # quy tắc 3

for idx, w in enumerate(["I", "speak", "Goat", "Latin", "apple"], start=1):
    print(f"{idx}: {w:>6} -> {convert_word(w, idx)}")
````

## 4. Cách 1 — Cộng chuỗi trực tiếp (cách tự nhiên)

Người mới thường viết: tạo `result = ""`, mỗi từ cộng thêm `từ_mới + " "`, cuối cùng cắt dấu cách thừa bằng `strip()` hoặc `result[:-1]`.

````python
def goat_concat(sentence: str) -> str:
    result = ""
    for idx, w in enumerate(sentence.split(), start=1):
        result += convert_word(w, idx) + " "
    return result[:-1]   # bỏ dấu cách cuối

print(goat_concat("I speak Goat Latin"))
````

### Nút thắt của cách 1

Chuỗi Python là **bất biến (immutable)**: `result += x` thường phải **tạo chuỗi mới** và sao chép toàn bộ nội dung cũ. Nếu kết quả cuối dài `L`, tổng chi phí sao chép có thể lên tới `O(L²)` trong trường hợp xấu (CPython đôi khi tối ưu được, nhưng không nên dựa vào).

Với bài này `len(sentence) <= 150` nên cách 1 vẫn qua. Nhưng thói quen tốt là: **gom vào list rồi `join` một lần**.

````python
s = "abc"
print(id(s))
s += "d"
print(id(s), "<- thường là đối tượng mới (chuỗi bất biến)")
try:
    s[0] = "X"
except TypeError as e:
    print("Không sửa tại chỗ được:", e)
````

## 5. Insight — Kích thước output và cách ghép hiệu quả

Gọi `n = len(sentence)`, `m` = số từ. Từ thứ `i` thêm `i` chữ `a`, nên tổng số `a` là `1 + 2 + ... + m = m(m+1)/2`. **Output** đã dài cỡ `O(n + m²)` — không thuật toán nào nhanh hơn kích thước output được.

Vì vậy mục tiêu: mỗi ký tự output được tạo **một lần**. `list.append` + `" ".join(...)` làm đúng điều đó.

````python
for m in [1, 2, 4, 10, 75]:
    print(f"m={m:>3} từ -> tổng số 'a' thêm vào = {m * (m + 1) // 2}")
````

## 6. Lời giải chính — list + join (từng đoạn rồi ghép)

**Đoạn 1:** cắt câu `words = sentence.split()`.

**Đoạn 2:** duyệt từng từ với `idx` từ 1; áp dụng quy tắc 1/2 để có `core`.

**Đoạn 3:** thêm `"ma"` và `"a" * idx`, `append` vào list `out`.

**Đoạn 4:** `return " ".join(out)`.

````python
def goat_join(sentence: str) -> str:
    vowels = set("aeiouAEIOU")
    out = []
    for idx, w in enumerate(sentence.split(), start=1):
        if w[0] in vowels:
            core = w
        else:
            core = w[1:] + w[0]
        out.append(core + "ma" + "a" * idx)
    return " ".join(out)

print(goat_join("I speak Goat Latin"))
print(goat_join("The quick brown fox jumped over the lazy dog"))
````

### 7.1. Bảng trace cho `"Each word apple"`

| idx | w | w[0] | nguyên âm? | core | + "ma" | + "a"*idx | out sau append |
|---|---|---|---|---|---|---|---|
| 1 | Each | E | có | Each | Eachma | Eachmaa | [Eachmaa] |
| 2 | word | w | không | ordw | ordwma | ordwmaaa | [.., ordwmaaa] |
| 3 | apple | a | có | apple | applema | applemaaaa | [.., applemaaaa] |

Kết quả: `"Eachmaa ordwmaaa applemaaaa"`. Chú ý `E` **hoa** vẫn là nguyên âm.

````python
def trace_goat(sentence: str) -> str:
    vowels = set("aeiouAEIOU")
    out = []
    for idx, w in enumerate(sentence.split(), start=1):
        is_vowel = w[0] in vowels
        core = w if is_vowel else w[1:] + w[0]
        new_w = core + "ma" + "a" * idx
        out.append(new_w)
        print(f"idx={idx} w={w!r:8} w[0]={w[0]!r} nguyên_âm={is_vowel!s:5} core={core!r:8} -> {new_w!r}")
    res = " ".join(out)
    print("Kết quả:", repr(res))
    return res

trace_goat("Each word apple");
````

## 8. Code nộp LeetCode

````python
class Solution:
    def toGoatLatin(self, sentence: str) -> str:
        vowels = set("aeiouAEIOU")
        out = []
        for idx, w in enumerate(sentence.split(), start=1):
            if w[0] in vowels:
                core = w
            else:
                core = w[1:] + w[0]
            out.append(core + "ma" + "a" * idx)
        return " ".join(out)

sol = Solution()
assert sol.toGoatLatin("I speak Goat Latin") == "Imaa peaksmaaa oatGmaaaa atinLmaaaaa"
assert sol.toGoatLatin("The quick brown fox jumped over the lazy dog") == \
    "heTmaa uickqmaaa rownbmaaaa oxfmaaaaa umpedjmaaaaaa overmaaaaaaa hetmaaaaaaaa azylmaaaaaaaaa ogdmaaaaaaaaaa"
print("Solution PASS 2 ví dụ đề")
````

## 9. Vì sao đúng? Độ phức tạp

**Đúng:** `split()` cho đúng danh sách từ theo thứ tự. Với mỗi từ, `if/else` áp dụng **đúng một** trong hai quy tắc 1/2 (hai trường hợp loại trừ nhau và phủ hết: chữ đầu hoặc là nguyên âm, hoặc không). Sau đó luôn thêm `"ma"` và đúng `idx` chữ `a` với `idx` là thứ tự từ (bắt đầu 1). `" ".join` ghép lại với đúng một dấu cách giữa các từ — khớp định dạng đề.

**Invariant:** sau khi xử lý từ thứ `idx`, `out` chứa đúng `idx` từ đã chuyển đổi của `idx` từ đầu tiên, theo thứ tự.

**Độ phức tạp:** gọi `n = len(sentence)`, `m` = số từ.
- Time: `O(n + m²)` — mỗi ký tự gốc được xử lý hằng số lần, cộng `m(m+1)/2` chữ `a`. Vì đây là kích thước output nên là tối ưu.
- Space: `O(n + m²)` cho output (list `out` và chuỗi kết quả).

````python
sentence = "a b c d"
res = Solution().toGoatLatin(sentence)
m = len(sentence.split())
print(res, "| độ dài:", len(res), "| số 'a' thêm:", m * (m + 1) // 2)
````

## 10. Cách 3 — Hậu tố tăng dần (không nhân chuỗi)

Thay vì tính `"a" * idx` mỗi lần, giữ biến `suffix` và mỗi từ **thêm một** `a`. Ý tưởng "trạng thái tăng dần" này hay gặp ở nhiều bài khác.

````python
def goat_suffix(sentence: str) -> str:
    vowels = set("aeiouAEIOU")
    out = []
    suffix = "ma"
    for w in sentence.split():
        suffix += "a"                       # từ thứ k: "ma" + k chữ a
        core = w if w[0] in vowels else w[1:] + w[0]
        out.append(core + suffix)
    return " ".join(out)

print(goat_suffix("I speak Goat Latin"))
````

## 11. Cách 4 — List comprehension một dòng

Khi đã hiểu rõ, có thể viết gọn bằng list comprehension và biểu thức điều kiện `A if cond else B`. Đọc từ trong ra ngoài: với mỗi `(i, w)`, tính `core`, ghép `"ma"` và `"a" * i`.

````python
x = 5
print("chẵn" if x % 2 == 0 else "lẻ")    # biểu thức điều kiện

def goat_oneliner(sentence: str) -> str:
    return " ".join(
        (w if w[0].lower() in "aeiou" else w[1:] + w[0]) + "ma" + "a" * i
        for i, w in enumerate(sentence.split(), 1)
    )

print(goat_oneliner("I speak Goat Latin"))
````

### Code nộp LeetCode — bản một dòng

````python
class SolutionOneLiner:
    def toGoatLatin(self, sentence: str) -> str:
        return " ".join(
            (w if w[0].lower() in "aeiou" else w[1:] + w[0]) + "ma" + "a" * i
            for i, w in enumerate(sentence.split(), 1)
        )

print(SolutionOneLiner().toGoatLatin("The quick brown fox jumped over the lazy dog"))
````

### So sánh các cách

| Cách | Ý tưởng | Time | Ghi chú |
|---|---|---|---|
| 1. Cộng chuỗi | `result += ...` | có thể `O(L²)` tệ nhất (`L` = độ dài output) | Dễ viết, phải bỏ dấu cách cuối |
| 2. List + join | `append` rồi `" ".join` | `O(n + m²)` | **Khuyến nghị** |
| 3. Hậu tố tăng dần | `suffix += "a"` | `O(n + m²)` | Minh họa trạng thái tăng dần |
| 4. One-liner | comprehension | `O(n + m²)` | Gọn, cần quen cú pháp |

## 13. Lỗi người mới hay gặp

### Lỗi 1: quên nguyên âm chữ hoa

```python
if w[0] in "aeiou":   # thiếu AEIOU
```

Vì sao sai: `"I"` (đề 1!) bị coi là phụ âm → `"I"[1:] + "I"` = `"I"` → `"Ima..."` tình cờ đúng với từ 1 chữ, nhưng `"Apple"` thì sai hẳn.

````python
def wrong_case(sentence):
    out = []
    for idx, w in enumerate(sentence.split(), start=1):
        core = w if w[0] in "aeiou" else w[1:] + w[0]
        out.append(core + "ma" + "a" * idx)
    return " ".join(out)

print("sai :", wrong_case("Apple"))
print("đúng:", Solution().toGoatLatin("Apple"))
````

### Lỗi 2: đếm chữ `a` bắt đầu từ 0

`enumerate(words)` mặc định bắt đầu 0 → từ đầu tiên không có `a` nào.

````python
def wrong_index(sentence):
    out = []
    for idx, w in enumerate(sentence.split()):          # thiếu start=1
        core = w if w[0] in "aeiouAEIOU" else w[1:] + w[0]
        out.append(core + "ma" + "a" * idx)
    return " ".join(out)

print("sai :", wrong_index("I speak"))
print("đúng:", Solution().toGoatLatin("I speak"))
````

### Lỗi 3: thừa dấu cách cuối khi cộng chuỗi

`result += new_w + " "` rồi `return result` → có `" "` ở cuối, LeetCode báo Wrong Answer dù nhìn "giống".

````python
def wrong_space(sentence):
    result = ""
    for idx, w in enumerate(sentence.split(), start=1):
        core = w if w[0] in "aeiouAEIOU" else w[1:] + w[0]
        result += core + "ma" + "a" * idx + " "
    return result

print(repr(wrong_space("I speak")))
print(repr(Solution().toGoatLatin("I speak")))
````

### Lỗi 4: đưa chữ đầu ra cuối **sau** khi đã thêm `"ma"`

Thứ tự đúng: xoay chữ đầu trước, rồi mới thêm `"ma"`. Làm ngược sẽ ra `"oatmag"` thay vì `"oatgma"`.

````python
w = "goat"
tmp = w + "ma"
print("sai :", tmp[1:] + tmp[0])
print("đúng:", w[1:] + w[0] + "ma")
````

## 14. Mẹo nhận dạng pattern

- Đề liệt kê **các quy tắc biến đổi** từng từ/ký tự → bài **mô phỏng**: viết hàm cho một đơn vị, kiểm thử riêng, rồi lặp.
- "Xử lý từng từ trong câu" → `split()` → biến đổi → `" ".join(...)`. Áp dụng cho *Reverse Words in a String III (557)*, *Truncate Sentence (1816)*, *Capitalize the Title (2129)*.
- Phân loại theo tập ký tự (nguyên âm, chữ số…) → dựng `set` một lần, tra `in` nhiều lần.
- Lượng thêm vào phụ thuộc **vị trí** → `enumerate(..., start=1)` hoặc biến trạng thái tăng dần.

## 15. Biến thể

### 15.1. Pig Latin cổ điển: chuyển **cụm phụ âm đầu** ra cuối + `"ay"`

Ví dụ `"string" → "ingstray"`, `"apple" → "appleay"`. Cần tìm vị trí nguyên âm đầu tiên.

````python
def pig_latin_word(w: str) -> str:
    vowels = set("aeiouAEIOU")
    k = 0
    while k < len(w) and w[k] not in vowels:
        k += 1
    if k == 0:
        return w + "ay"
    return w[k:] + w[:k] + "ay"

print([pig_latin_word(w) for w in ["string", "apple", "rhythm", "Goat"]])
````

### 15.2. Giải mã Goat Latin về câu gốc

Bỏ `idx` chữ `a` cuối và bỏ `"ma"` thì thu được `core`. Nhưng có vấn đề: nhìn `core` ta **không biết** nó đã bị xoay (từ gốc bắt đầu bằng phụ âm) hay giữ nguyên (từ gốc bắt đầu bằng nguyên âm). Thử tìm hai từ gốc khác nhau cho cùng kết quả:

````python
def goat_word(w):
    return (w if w[0] in "aeiouAEIOU" else w[1:] + w[0]) + "ma"

print(goat_word("ea"), goat_word("ae"), "<- 'ea' -> 'eama'; 'ae' -> 'aema' (khác nhau)")
print(goat_word("bea"), goat_word("eab"), "<- 'bea' và 'eab' cho cùng 'eabma'? ", goat_word("bea") == goat_word("eab"))
````

Kết quả trên cho thấy `"bea"` (xoay `b` ra cuối → `"eab"`) và `"eab"` (bắt đầu nguyên âm, giữ nguyên) cùng thành `"eabma"` → không giải mã duy nhất được.

### 15.3. Đếm độ dài kết quả mà không cần xây chuỗi

Mỗi từ dài thêm `2 + idx`; cộng thêm `m - 1` dấu cách.

````python
def goat_length(sentence: str) -> int:
    words = sentence.split()
    m = len(words)
    return sum(len(w) + 2 + i for i, w in enumerate(words, 1)) + (m - 1)

s = "The quick brown fox jumped over the lazy dog"
print(goat_length(s), len(Solution().toGoatLatin(s)))
````

## 18. Đáp án tham khảo và tổng kết

**Bài 2:** `"Hi"` → `"iHmaa"`; `"you"` → `"ouymaaa"`; `"Ok"` → `"Okmaaaa"` ⇒ `"iHmaa ouymaaa Okmaaaa"`.

````python
print("hello"[1:] + "hello"[0], "ab".split(" "), "-".join(["x"]), "a" * 0 == "")

def ans3_goat(sentence: str) -> str:
    vowels = set("aeiouAEIOU")
    out = []
    for idx, w in enumerate(sentence.split(), start=1):
        core = w if w[0] in vowels else w[1:] + w[0]
        out.append(core + "ma" + "a" * idx)
    return " ".join(out)

def ans4_reverse_each_word(s: str) -> str:
    return " ".join(w[::-1] for w in s.split())

def ans5_goat_custom(sentence: str, tail: str, ch: str) -> str:
    vowels = set("aeiouAEIOU")
    out = []
    for idx, w in enumerate(sentence.split(), start=1):
        core = w if w[0] in vowels else w[1:] + w[0]
        out.append(core + tail + ch * idx)
    return " ".join(out)

def ans6_pig_latin_sentence(sentence: str) -> str:
    vowels = set("aeiouAEIOU")
    def pig(w):
        k = 0
        while k < len(w) and w[k] not in vowels:
            k += 1
        return w[k:] + w[:k] + "ay"      # k == len(w) -> w + "ay"
    return " ".join(pig(w) for w in sentence.split())

assert ans3_goat("Hi you Ok") == "iHmaa ouymaaa Okmaaaa"
assert ans3_goat("I speak Goat Latin") == "Imaa peaksmaaa oatGmaaaa atinLmaaaaa"
assert ans4_reverse_each_word("Let's take LeetCode contest") == "s'teL ekat edoCteeL tsetnoc"
assert ans5_goat_custom("I speak", "xy", "z") == "Ixyz peaksxyzz"
assert ans6_pig_latin_sentence("string apple rhythm") == "ingstray appleay rhythmay"
print("Đáp án bài 1-6 PASS")
````
