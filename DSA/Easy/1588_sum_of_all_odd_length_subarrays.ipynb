## Mục tiêu học
1. Hiểu **subarray (mảng con liên tiếp)**, cách liệt kê bằng hai chỉ số `i`, `j` và slicing `arr[i:j]`.
2. Đi qua 4 lời giải từ dễ đến khéo: `O(n³)` → `O(n²)` (cộng dồn) → `O(n²)` (prefix sum) → `O(n)` (đếm đóng góp).
3. Học kỹ thuật **prefix sum** — tính tổng đoạn bất kỳ trong `O(1)`.
4. Học tư duy **contribution technique**: thay vì cộng từng mảng con, hỏi "mỗi phần tử được cộng bao nhiêu lần?".
5. Trace, chứng minh công thức `((i+1)*(n-i)+1)//2`, và luyện biến thể (độ dài chẵn, mọi độ dài, độ dài chia hết cho k).

## 1. Đề bài

Cho mảng số nguyên **dương** `arr`. Trả về **tổng của tất cả các mảng con có độ dài lẻ**.

**Mảng con (subarray)** là một dãy **liên tiếp** của mảng.

Ví dụ 1: `arr = [1,4,2,5,3]` → `58`

| Độ dài | Các mảng con | Tổng |
|:---:|---|---:|
| 1 | `[1]`, `[4]`, `[2]`, `[5]`, `[3]` | 15 |
| 3 | `[1,4,2]`=7, `[4,2,5]`=11, `[2,5,3]`=10 | 28 |
| 5 | `[1,4,2,5,3]`=15 | 15 |
| | **Tổng cộng** | **58** |

Ví dụ 2: `arr = [1,2]` → `3` (chỉ có `[1]` và `[2]`).
Ví dụ 3: `arr = [10,11,12]` → `66` (`10+11+12 + 33`).

**Constraints:** `1 <= arr.length <= 100`; `1 <= arr[i] <= 1000`.
**Follow up:** giải được trong `O(n)` không?

````python
examples = [([1, 4, 2, 5, 3], 58), ([1, 2], 3), ([10, 11, 12], 66)]
for arr, expected in examples:
    print(arr, "-> kỳ vọng", expected)
print("Kiểm tra tay ví dụ 3:", 10 + 11 + 12 + (10 + 11 + 12))
````

## 2. Nền tảng Python

### 2.1. Slicing `arr[i:j]` — lấy đoạn liên tiếp

`arr[i:j]` lấy các phần tử từ chỉ số `i` **đến trước** `j` (không lấy `arr[j]`). Độ dài đoạn là `j - i`.

Mảng con bắt đầu ở `i`, kết thúc ở `j` (tính cả `j`) là `arr[i:j+1]`, độ dài `j - i + 1`.

````python
arr = [1, 4, 2, 5, 3]
print("arr[0:3] =", arr[0:3], "| độ dài", 3 - 0)
print("arr[1:4] =", arr[1:4])
i, j = 2, 4
print(f"mảng con từ i={i} tới j={j} (tính cả j):", arr[i:j + 1], "| độ dài", j - i + 1)
````

### 2.2. `sum()` và `range(start, stop, step)`

- `sum(list)` cộng mọi phần tử.
- `range(1, n + 1, 2)` sinh `1, 3, 5, ...` — các **độ dài lẻ** từ 1 tới n. `step = 2` là bước nhảy.

````python
print(sum([1, 4, 2]))
n = 5
print("độ dài lẻ khi n=5:", list(range(1, n + 1, 2)))
n = 6
print("độ dài lẻ khi n=6:", list(range(1, n + 1, 2)))
````

### 2.3. Liệt kê mọi mảng con bằng hai chỉ số

`i` là điểm bắt đầu, `j` là điểm kết thúc với `i <= j`. Có `n(n+1)/2` mảng con. Độ dài lẻ ⟺ `(j - i + 1) % 2 == 1`.

`%` là phép **chia lấy dư**: `7 % 2 = 1` (lẻ), `8 % 2 = 0` (chẵn).

````python
arr = [10, 11, 12]
n = len(arr)
for i in range(n):
    for j in range(i, n):
        length = j - i + 1
        tag = "LẺ" if length % 2 == 1 else "chẵn"
        print(f"i={i}, j={j}, đoạn={arr[i:j+1]}, độ dài={length} ({tag})")
print("Tổng số mảng con:", n * (n + 1) // 2)
````

## 3. Cách 1 — Brute Force `O(n³)`: liệt kê, lọc lẻ, cộng từng đoạn

Đúng như gợi ý của LeetCode: thử mọi cặp `(i, j)`, nếu độ dài lẻ thì cộng tổng đoạn.

Mỗi lần `sum(arr[i:j+1])` tốn tới `O(n)` → tổng `O(n³)`. Với `n ≤ 100` vẫn chạy được (khoảng vài trăm nghìn phép cộng).

````python
def sum_odd_brute(arr: list[int]) -> int:
    n = len(arr)
    total = 0
    for i in range(n):
        for j in range(i, n):
            if (j - i + 1) % 2 == 1:
                total += sum(arr[i:j + 1])
    return total

for arr, expected in examples:
    print(arr, "->", sum_odd_brute(arr), "| kỳ vọng", expected)
````

Một cách viết khác của brute force: duyệt theo **độ dài** `L = 1, 3, 5, ...` rồi theo điểm bắt đầu `i` (từ `0` tới `n - L`).

````python
def sum_odd_by_length(arr: list[int]) -> int:
    n = len(arr)
    total = 0
    for L in range(1, n + 1, 2):
        for i in range(n - L + 1):
            total += sum(arr[i:i + L])
    return total

print([sum_odd_by_length(a) for a, _ in examples])
````

## 4. Cách 2 — `O(n²)`: cộng dồn khi mở rộng `j`

**Chỗ lặp việc của cách 1:** tổng `arr[i..j]` và `arr[i..j+1]` chỉ khác nhau đúng `arr[j+1]`. Vậy khi cố định `i` và tăng `j`, giữ biến `running` và cộng thêm một phần tử mỗi bước.

```text
i = 0: running = 1 -> 5 -> 7 -> 12 -> 15
             (j=0)  (j=1) (j=2) (j=3) (j=4)
lẻ ở j = 0, 2, 4 -> cộng 1 + 7 + 15
```

````python
def sum_odd_running(arr: list[int]) -> int:
    n = len(arr)
    total = 0
    for i in range(n):
        running = 0
        for j in range(i, n):
            running += arr[j]              # tổng arr[i..j]
            if (j - i + 1) % 2 == 1:
                total += running
    return total

print([sum_odd_running(a) for a, _ in examples])
````

## 5. Cách 3 — Prefix sum `O(n²)`: tổng đoạn bất kỳ trong `O(1)`

**Prefix sum** `prefix[k]` = tổng `k` phần tử đầu: `prefix[0] = 0`, `prefix[k+1] = prefix[k] + arr[k]`.

Khi đó tổng đoạn `arr[i..j]` = `prefix[j+1] - prefix[i]`.

```text
arr    =    [1, 4, 2, 5, 3]
prefix = [0, 1, 5, 7, 12, 15]
tổng arr[1..3] = prefix[4] - prefix[1] = 12 - 1 = 11  (4+2+5)
```

Ta có thể dựng prefix bằng `itertools.accumulate` (thư viện chuẩn) — nó trả về các tổng cộng dồn.

````python
from itertools import accumulate
arr = [1, 4, 2, 5, 3]
prefix = [0]
for x in arr:
    prefix.append(prefix[-1] + x)
print("prefix (tự dựng)  =", prefix)
print("prefix (accumulate)=", [0] + list(accumulate(arr)))
print("tổng arr[1..3] =", prefix[4] - prefix[1], "| kiểm tra:", sum(arr[1:4]))
````

````python
def sum_odd_prefix(arr: list[int]) -> int:
    n = len(arr)
    prefix = [0] * (n + 1)
    for k in range(n):
        prefix[k + 1] = prefix[k] + arr[k]
    total = 0
    for i in range(n):
        for j in range(i, n, 2):           # j - i chẵn  <=>  độ dài lẻ
            total += prefix[j + 1] - prefix[i]
    return total

print([sum_odd_prefix(a) for a, _ in examples])
````

Mẹo nhỏ ở trên: `range(i, n, 2)` cho `j = i, i+2, i+4, ...` nên `j - i` luôn chẵn, độ dài `j - i + 1` luôn lẻ — không cần `if` nữa.

## 6. Nút thắt và insight — Đếm **đóng góp** của từng phần tử

Cả 3 cách trên đều **liệt kê mảng con**, mà số mảng con là `~n²/2` → không thể xuống `O(n)` theo hướng này.

Đổi góc nhìn: đáp án là tổng của nhiều đoạn; mỗi đoạn là tổng của các phần tử. Gom lại theo phần tử:

$$\text{đáp án} = \sum_i arr[i] \times (\text{số mảng con độ dài lẻ chứa } i)$$

Vậy chỉ cần biết **phần tử thứ `i` nằm trong bao nhiêu mảng con lẻ**. Đó gọi là **contribution technique** (kỹ thuật đếm đóng góp).

````python
# Đếm thô: mỗi chỉ số xuất hiện trong bao nhiêu mảng con lẻ?
def count_by_brute(n):
    cnt = [0] * n
    for i in range(n):
        for j in range(i, n):
            if (j - i + 1) % 2 == 1:
                for k in range(i, j + 1):
                    cnt[k] += 1
    return cnt

c = count_by_brute(5)
print("số lần xuất hiện (n=5):", c)
arr = [1, 4, 2, 5, 3]
print("tổng đóng góp:", sum(a * k for a, k in zip(arr, c)))   # 58
````

### 6.1. Bước 1: số mảng con (mọi độ dài) chứa chỉ số `i`

Một mảng con chứa `i` phải bắt đầu ở `start ∈ {0, 1, ..., i}` (**`i + 1` cách**) và kết thúc ở `end ∈ {i, ..., n-1}` (**`n - i` cách**). Chọn độc lập → `(i + 1) * (n - i)` mảng con.

Ví dụ `n = 5`, `i = 2`: `3 * 3 = 9` mảng con chứa vị trí 2.

````python
n = 5
for i in range(n):
    print(f"i={i}: bắt đầu có {i+1} cách, kết thúc có {n-i} cách -> {(i+1)*(n-i)} mảng con chứa i")
````

### 6.2. Bước 2: trong số đó, bao nhiêu mảng con có độ dài lẻ?

Đặt `a = i - start` (số phần tử bên trái `i`, từ `0` tới `i`) và `b = end - i` (bên phải, từ `0` tới `n-1-i`). Độ dài `= a + b + 1` lẻ ⟺ `a + b` **chẵn** ⟺ `a`, `b` **cùng tính chẵn lẻ**.

- Số `a` chẵn trong `0..i`: `i // 2 + 1`; số `a` lẻ: `(i + 1) // 2`.
- Số `b` chẵn trong `0..n-1-i`: `(n - 1 - i) // 2 + 1`; số `b` lẻ: `(n - i) // 2`.

Số mảng con lẻ chứa `i` = `chẵn_trái * chẵn_phải + lẻ_trái * lẻ_phải`.

Rút gọn được (sẽ kiểm chứng bằng code): **`((i + 1) * (n - i) + 1) // 2`** — tức "tổng số mảng con chứa `i`, chia 2, làm tròn lên".

````python
def odd_count_parity(i, n):
    even_left, odd_left = i // 2 + 1, (i + 1) // 2
    even_right, odd_right = (n - 1 - i) // 2 + 1, (n - i) // 2
    return even_left * even_right + odd_left * odd_right

def odd_count_formula(i, n):
    return ((i + 1) * (n - i) + 1) // 2

for n in range(1, 40):
    brute = count_by_brute(n) if n <= 12 else None
    for i in range(n):
        assert odd_count_parity(i, n) == odd_count_formula(i, n)
        if brute is not None:
            assert brute[i] == odd_count_formula(i, n)
print("n=5:", [odd_count_formula(i, 5) for i in range(5)])
print("Công thức khớp đếm thô (n<=12) và khớp đếm chẵn/lẻ (n<40)")
````

**Vì sao "chia 2 làm tròn lên"?** Đặt `L = i + 1` (số giá trị của `a`) và `R = n - i` (số giá trị của `b`). Ta đếm số cặp `(a, b)` cùng tính chẵn lẻ.

- Nếu `R` **chẵn**: với mỗi `a` cố định, các `b = 0..R-1` có đúng `R/2` số cùng tính chẵn lẻ với `a` → tổng `L·R/2` (đúng một nửa). Tương tự nếu `L` chẵn.
- Nếu `L`, `R` đều **lẻ**: số chẵn nhiều hơn số lẻ đúng 1 ở mỗi bên: chẵn `(L+1)/2`, lẻ `(L-1)/2`. Số cặp cùng tính = `[(L+1)(R+1) + (L-1)(R-1)] / 4 = (2LR + 2)/4 = (LR + 1)/2`.

Cả hai trường hợp gộp lại đúng là `ceil(L·R / 2)`.

Trong Python, `ceil(x / 2)` với số nguyên dương có thể viết `(x + 1) // 2` (không cần `math.ceil`, tránh số thực).

````python
import math
for x in [1, 4, 5, 8, 9]:
    print(x, "-> (x+1)//2 =", (x + 1) // 2, "| math.ceil(x/2) =", math.ceil(x / 2))
````

## 7. Lời giải chính `O(n)` — ghép lại

````python
def sum_odd_contribution(arr: list[int]) -> int:
    n = len(arr)
    total = 0
    for i, x in enumerate(arr):
        times = ((i + 1) * (n - i) + 1) // 2   # số mảng con lẻ chứa vị trí i
        total += x * times
    return total

for arr, expected in examples:
    print(arr, "->", sum_odd_contribution(arr), "| kỳ vọng", expected)
````

## 8. TRACE từng bước

Ví dụ `arr = [1, 4, 2, 5, 3]`, `n = 5`:

| i | arr[i] | i+1 (trái) | n-i (phải) | tổng mảng con chứa i | số mảng con lẻ | đóng góp | total |
|:-:|:-:|:-:|:-:|:-:|:-:|:-:|:-:|
| 0 | 1 | 1 | 5 | 5 | 3 | 3 | 3 |
| 1 | 4 | 2 | 4 | 8 | 4 | 16 | 19 |
| 2 | 2 | 3 | 3 | 9 | 5 | 10 | 29 |
| 3 | 5 | 4 | 2 | 8 | 4 | 20 | 49 |
| 4 | 3 | 5 | 1 | 5 | 3 | 9 | **58** |

Kiểm tra cột "số mảng con lẻ" với i = 0: `[1]`, `[1,4,2]`, `[1,4,2,5,3]` → đúng 3.

````python
def trace_sum_odd(arr):
    n = len(arr)
    total = 0
    print(" i | x  | trái | phải | chứa i | lẻ | đóng góp | total")
    for i, x in enumerate(arr):
        left, right = i + 1, n - i
        times = (left * right + 1) // 2
        total += x * times
        print(f"{i:>2} | {x:<2} | {left:>4} | {right:>4} | {left*right:>6} | {times:>2} | {x*times:>8} | {total}")
    return total

print("Kết quả:", trace_sum_odd([1, 4, 2, 5, 3]))
````

Trace thứ hai với `n` chẵn để thấy trường hợp `L·R` chẵn (không có "+1 lệch"): `arr = [1, 2]` → mỗi phần tử chỉ nằm trong 1 mảng con lẻ (chính nó).

````python
print("Kết quả:", trace_sum_odd([1, 2]))
print()
print("Kết quả:", trace_sum_odd([10, 11, 12]))
````

Đối chiếu với cách liệt kê (cách 2) bằng trace: in từng mảng con lẻ được cộng.

````python
def trace_running(arr):
    n = len(arr)
    total = 0
    for i in range(n):
        running = 0
        for j in range(i, n):
            running += arr[j]
            if (j - i + 1) % 2 == 1:
                total += running
                print(f"  cộng {arr[i:j+1]} = {running:<3} -> total = {total}")
    return total

print("Kết quả:", trace_running([1, 4, 2, 5, 3]))
````

## 9. Code nộp LeetCode

### 9.1. Cách chính — đếm đóng góp `O(n)` (giải được Follow up)

````python
class Solution:
    def sumOddLengthSubarrays(self, arr: list[int]) -> int:
        n = len(arr)
        total = 0
        for i, x in enumerate(arr):
            total += x * (((i + 1) * (n - i) + 1) // 2)
        return total

sol = Solution()
print([sol.sumOddLengthSubarrays(a) for a, _ in examples])   # [58, 3, 66]
````

### 9.2. Cách cộng dồn `O(n²)` — dễ nghĩ ra trong phòng thi

````python
class SolutionRunning:
    def sumOddLengthSubarrays(self, arr: list[int]) -> int:
        n = len(arr)
        total = 0
        for i in range(n):
            running = 0
            for j in range(i, n):
                running += arr[j]
                if (j - i) % 2 == 0:
                    total += running
        return total

print([SolutionRunning().sumOddLengthSubarrays(a) for a, _ in examples])
````

### 9.3. Cách prefix sum `O(n²)`

````python
class SolutionPrefix:
    def sumOddLengthSubarrays(self, arr: list[int]) -> int:
        prefix = [0]
        for x in arr:
            prefix.append(prefix[-1] + x)
        n = len(arr)
        return sum(prefix[j + 1] - prefix[i] for i in range(n) for j in range(i, n, 2))

print([SolutionPrefix().sumOddLengthSubarrays(a) for a, _ in examples])
````

### 9.4. Cách Brute Force `O(n³)` — một dòng

````python
class SolutionBrute:
    def sumOddLengthSubarrays(self, arr: list[int]) -> int:
        n = len(arr)
        return sum(sum(arr[i:i + L]) for L in range(1, n + 1, 2) for i in range(n - L + 1))

print([SolutionBrute().sumOddLengthSubarrays(a) for a, _ in examples])
````

## 10. Vì sao đúng? Độ phức tạp và so sánh

**Đúng (cách đóng góp):** đáp án là tổng trên các mảng con lẻ `S` của `Σ_{k∈S} arr[k]`. Đổi thứ tự cộng (cộng theo phần tử trước, theo mảng con sau) không đổi kết quả: `Σ_k arr[k] · #{S lẻ : k ∈ S}`. Mục 6 đã đếm `#{S lẻ : k ∈ S} = ((k+1)(n-k)+1)//2` bằng cách chọn độc lập điểm bắt đầu/kết thúc và xét tính chẵn lẻ, nên mỗi mảng con lẻ được cộng **đúng một lần** cho mỗi phần tử của nó.

**Đúng (cách cộng dồn):** invariant — sau khi xử lý `j`, `running == sum(arr[i..j])`.

Gọi `n = len(arr)`.

| Cách | Ý tưởng | Time | Space phụ |
|---|---|---|---|
| 1. Brute Force | liệt kê + `sum` từng đoạn | `O(n³)` | `O(n)` (slice tạm) |
| 2. Cộng dồn | giữ `running` khi tăng `j` | `O(n²)` | `O(1)` |
| 3. Prefix sum | tổng đoạn = hiệu prefix | `O(n²)` | `O(n)` |
| 4. Đóng góp | đếm số lần mỗi phần tử được cộng | **`O(n)`** | **`O(1)`** |

Với `n ≤ 100` cả 4 cách đều qua; cách 4 là đáp án cho Follow up và là kỹ thuật đáng nhớ nhất.

## 12. Lỗi người mới hay gặp

### Lỗi 1: Slice lệch một — `arr[i:j]` thay vì `arr[i:j+1]`

`arr[i:j]` **không** lấy `arr[j]`. Nếu `j` là chỉ số cuối (tính cả), phải dùng `arr[i:j+1]`.

````python
def wrong_slice(arr):
    n = len(arr)
    return sum(sum(arr[i:j]) for i in range(n) for j in range(i, n) if (j - i + 1) % 2 == 1)

print("sai:", wrong_slice([1, 4, 2, 5, 3]), "| đúng:", sum_odd_brute([1, 4, 2, 5, 3]))
````

### Lỗi 2: Quên làm tròn lên — dùng `(i+1)*(n-i)//2`

Khi `(i+1)*(n-i)` lẻ, số mảng con lẻ nhiều hơn một nửa. Làm tròn xuống làm mất phần đó. Ví dụ `[7]`: `1*1//2 = 0` → đáp án sai `0`.

````python
def wrong_floor(arr):
    n = len(arr)
    return sum(x * ((i + 1) * (n - i) // 2) for i, x in enumerate(arr))

print("sai:", wrong_floor([7]), wrong_floor([1, 4, 2, 5, 3]), "| đúng:", sum_odd_contribution([7]), sum_odd_contribution([1, 4, 2, 5, 3]))
````

### Lỗi 3: `range(1, n, 2)` thiếu độ dài `n`

Khi `n` lẻ, độ dài lớn nhất cũng lẻ (= n) và phải được tính. `range(1, n, 2)` dừng trước `n` nên bỏ mất cả mảng. **Sửa:** `range(1, n + 1, 2)`.

````python
def wrong_range(arr):
    n = len(arr)
    return sum(sum(arr[i:i + L]) for L in range(1, n, 2) for i in range(n - L + 1))

print("sai:", wrong_range([1, 4, 2, 5, 3]), "| đúng: 58  (thiếu đúng 15 = tổng cả mảng)")
````

### Lỗi 4: Reset `running` sai chỗ

`running = 0` phải đặt **bên trong** vòng `i` (mỗi điểm bắt đầu mới, tổng làm lại từ 0). Đặt ngoài vòng `i` sẽ cộng lẫn đoạn cũ.

````python
def wrong_reset(arr):
    n = len(arr)
    total = running = 0
    for i in range(n):
        for j in range(i, n):
            running += arr[j]
            if (j - i) % 2 == 0:
                total += running
    return total

print("sai:", wrong_reset([1, 4, 2, 5, 3]), "| đúng:", sum_odd_running([1, 4, 2, 5, 3]))
````

## 13. Mẹo nhận dạng pattern

- Đề hỏi "**tổng trên mọi mảng con** (có điều kiện)" → thử **contribution technique**: mỗi phần tử được cộng bao nhiêu lần?
- Số mảng con chứa vị trí `i`: `(i + 1) * (n - i)` — công thức nền của rất nhiều bài.
- Cần **tổng đoạn** nhiều lần → **prefix sum**: `sum(i..j) = prefix[j+1] - prefix[i]`.
- Điều kiện chẵn/lẻ trên độ dài → tách theo tính chẵn lẻ của số phần tử trái/phải.
- Bài liên quan: 303 Range Sum Query, 907 Sum of Subarray Minimums (đóng góp + stack), 2104 Sum of Subarray Ranges, 828 Count Unique Characters of All Substrings, 2063 Vowels of All Substrings.

## 14. Biến thể

### 14.1. Tổng của **mọi** mảng con (không quan tâm độ dài)

Mỗi phần tử được cộng `(i+1)*(n-i)` lần.

````python
def sum_all_subarrays(arr):
    n = len(arr)
    return sum(x * (i + 1) * (n - i) for i, x in enumerate(arr))

def sum_all_brute(arr):
    n = len(arr)
    return sum(sum(arr[i:j + 1]) for i in range(n) for j in range(i, n))

print(sum_all_subarrays([1, 4, 2, 5, 3]), sum_all_brute([1, 4, 2, 5, 3]))
````

### 14.2. Tổng các mảng con độ dài **chẵn**

Chẵn = Tất cả − Lẻ. Số mảng con chẵn chứa `i` là `(i+1)*(n-i) // 2` (làm tròn xuống).

````python
def sum_even_subarrays(arr):
    n = len(arr)
    return sum(x * ((i + 1) * (n - i) // 2) for i, x in enumerate(arr))

a = [1, 4, 2, 5, 3]
print(sum_even_subarrays(a), "=", sum_all_subarrays(a), "-", sum_odd_contribution(a))
````

### 14.3. Độ dài chia hết cho `k` — quay về prefix sum

Khi điều kiện độ dài phức tạp hơn chẵn/lẻ, công thức đóng góp khó hơn. Prefix sum + duyệt `j = i + k - 1, i + 2k - 1, ...` vẫn cho `O(n²/k)` gọn gàng.

````python
def sum_len_multiple_of_k(arr, k):
    prefix = [0]
    for x in arr:
        prefix.append(prefix[-1] + x)
    n = len(arr)
    return sum(prefix[j + 1] - prefix[i] for i in range(n) for j in range(i + k - 1, n, k))

print(sum_len_multiple_of_k([1, 4, 2, 5, 3], 2), "(phải bằng tổng chẵn:", sum_even_subarrays([1, 4, 2, 5, 3]), ")")
print(sum_len_multiple_of_k([1, 4, 2, 5, 3], 5))   # chỉ cả mảng -> 15
````

## 17. Đáp án tham khảo

**Bài 1:** `[2, 3, 3, 2]` (= ceil của `[4, 6, 6, 4] / 2`). Kiểm tra: tổng = 10 = tổng độ dài các mảng con lẻ khi n=4 (`4·1 + 2·3`).

**Bài 2:** `[2, 5, 8]` và `[6, 7]`.

````python
def answer_running(arr):
    n, total = len(arr), 0
    for i in range(n):
        running = 0
        for j in range(i, n):
            running += arr[j]
            if (j - i) % 2 == 0:
                total += running
    return total

def answer_contribution(arr):
    n = len(arr)
    return sum(x * (((i + 1) * (n - i) + 1) // 2) for i, x in enumerate(arr))

def answer_even_prefix(arr):
    prefix = [0]
    for x in arr:
        prefix.append(prefix[-1] + x)
    n = len(arr)
    return sum(prefix[j + 1] - prefix[i] for i in range(n) for j in range(i + 1, n, 2))

def answer_odd_len_odd_sum_n2(arr):
    prefix = [0]
    for x in arr:
        prefix.append(prefix[-1] + x)
    n = len(arr)
    return sum(1 for i in range(n) for j in range(i, n, 2) if (prefix[j + 1] - prefix[i]) % 2 == 1)

def answer_odd_len_odd_sum(arr):
    # cnt[p][v]: số prefix có chỉ số k với k%2 == p và prefix[k]%2 == v
    cnt = [[0, 0], [0, 0]]
    cnt[0][0] = 1          # prefix[0] = 0 tại chỉ số 0
    s, ans = 0, 0
    for k, x in enumerate(arr, start=1):
        s += x
        ans += cnt[1 - k % 2][1 - s % 2]   # chỉ số khác chẵn lẻ, giá trị khác chẵn lẻ
        cnt[k % 2][s % 2] += 1
    return ans

assert [((i + 1) * (4 - i) + 1) // 2 for i in range(4)] == [2, 3, 3, 2]
assert list(range(2, 9, 3)) == [2, 5, 8] and [5, 6, 7][1:3] == [6, 7]
assert answer_running([1, 4, 2, 5, 3]) == 58
assert answer_contribution([10, 11, 12]) == 66
assert answer_even_prefix([1, 4, 2, 5, 3]) == 52 == sum_even_subarrays([1, 4, 2, 5, 3])
assert answer_odd_len_odd_sum_n2([1, 4, 2, 5, 3]) == 6 == answer_odd_len_odd_sum([1, 4, 2, 5, 3])
rng = random.Random(42)
for _ in range(300):
    arr = [rng.randint(1, 20) for _ in range(rng.randint(1, 15))]
    assert answer_running(arr) == answer_contribution(arr) == sum_odd_brute(arr)
    assert answer_even_prefix(arr) == sum_even_subarrays(arr)
    assert answer_odd_len_odd_sum_n2(arr) == answer_odd_len_odd_sum(arr)
print("Đáp án tham khảo: PASS toàn bộ!")
````
