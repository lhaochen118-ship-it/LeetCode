## 3. Cách trực giác: thử cả hai bước đầu tiên

Nếu còn `n` bậc thì ta có hai nhánh: **bước 1**, còn `n-1`; hoặc **bước 2**, còn `n-2`. Tổng số cách bằng tổng số cách của hai nhánh.

Đệ quy (recursion) là khi một hàm **tự gọi lại** trên một bài toán nhỏ hơn. Cần có **điểm dừng (base case)**:

- `n=0`: có đúng **1 cách** để không bước thêm nữa — chuỗi bước rỗng.
- `n<0`: không thể hoàn tất đúng đích, có **0 cách**.

`ways(0)=1` là công cụ đếm toán học, không có nghĩa đề yêu cầu kiểm thử n=0.

```python
def climb_recursive(n: int) -> int:
    if n == 0:
        return 1
    if n < 0:
        return 0
    return climb_recursive(n-1) + climb_recursive(n-2)

for n in range(1,7):
    print("n=",n,"số cách=",climb_recursive(n))
```

## 4. Xem cây lời gọi: tại sao đệ quy chậm?

Khi tính `ways(5)`, ta gọi `ways(4)` và `ways(3)`. Nhưng khi tính `ways(4)`, ta lại gọi `ways(3)` một lần nữa. Cùng **bài toán con bị lặp** nhiều lần.

Ta in lịch sử gọi hàm với `n=4` để nhận ra sự trùng lặp. Không dùng cách in này khi `n` lớn.

```python
def trace_recursive(n, depth=0):
    print("  "*depth + f"ways({n})")
    if n==0: return 1
    if n<0: return 0
    return trace_recursive(n-1,depth+1)+trace_recursive(n-2,depth+1)

print("Kết quả:",trace_recursive(4))
```

## 5. Insight: nhóm các cách theo BƯỚC CUỐI

Muốn tới bậc `n`, bước **cuối** chỉ có hai khả năng:

- Từ bậc `n-1`, bước thêm 1 bậc: `ways(n-1)` cách.
- Từ bậc `n-2`, bước thêm 2 bậc: `ways(n-2)` cách.

Hai nhóm **không trùng nhau** và bao phủ toàn bộ cách. Do đó:

**`ways(n) = ways(n-1) + ways(n-2)`**, với `ways(0)=1`, `ways(1)=1`.

Ví dụ: `ways(4)=ways(3)+ways(2)=3+2=5`. Đây là ý tưởng của **Dynamic Programming (DP)**: lưu lời giải của bài toán con để không tính lại.

```python
ways = [1, 1]   # ways[0]=1, ways[1]=1
for n in range(2,8):
    next_count = ways[n-1] + ways[n-2]
    ways.append(next_count)
    print(f"ways({n}) = ways({n-1}) + ways({n-2}) = {ways[n-1]} + {ways[n-2]} = {next_count}")
```

## 6. Python cơ sở: list `.append`, chỉ số và `range`

`dp[i]` là giá trị lưu tại vị trí `i`. `append(x)` **thêm phần tử ở cuối** list. `range(2,n+1)` duyệt các số 2,3,...,n; `n+1` **không được tính**.

Khi đã biết `dp[i-1]` và `dp[i-2]`, ta mới tính `dp[i]`. Đó là lý do phải đi từ **bài toán nhỏ lên lớn**.

```python
dp = [1,1]
for i in range(2,5):
    dp.append(dp[i-1]+dp[i-2])
    print("sau i=",i,"->",dp)
print("dp[4] =",dp[4])
```

## 7. DP dùng mảng: rõ ràng nhất cho người mới

`dp[k]` biểu thị **số cách đến bậc k**. Sau khi xây `dp[0]` và `dp[1]`, mỗi lượt tính thêm một kết quả. Ta chỉ tính mỗi `k` đúng một lần.

```python
def climb_dp_array(n: int) -> int:
    dp = [1] * (n + 1)  # tạo n+1 ô, từ 0 đến n
    if n <= 1:
        return 1
    for i in range(2, n + 1):
        dp[i] = dp[i-1] + dp[i-2]
    return dp[n]

for n in range(1,8):
    print(n, "->", climb_dp_array(n))
```

## 8. Tại sao dùng 2 biến là đủ?

Ở bậc `k`, công thức chỉ đọc 2 số liền trước. Không cần giữ toàn bộ mảng `dp`.

Đặt:
- `prev2` = ways(k-2)
- `prev1` = ways(k-1)
- `current = prev1 + prev2` = ways(k)

Sau khi tính xong, **dịch cửa sổ**: `prev2 = prev1`, `prev1 = current`.

Python cho phép gán đồng thời `prev2, prev1 = prev1, prev2 + prev1`: các vế phải được tính từ **giá trị cũ** trước khi gán.

```python
prev2, prev1 = 1, 1
print("trước: ways(0)=",prev2,", ways(1)=",prev1)
prev2, prev1 = prev1, prev2 + prev1
print("sau  : ways(1)=",prev2,", ways(2)=",prev1)
prev2, prev1 = prev1, prev2 + prev1
print("tiếp : ways(2)=",prev2,", ways(3)=",prev1)
```

## 9. Code chính: DP tối ưu bộ nhớ

Từ `ways(0)=1`, `ways(1)=1`, duyệt `k=2` đến `n`. Cuối cùng `prev1` đúng bằng `ways(n)`.

Đề LeetCode cho n dương; cách cài đặt này còn định nghĩa được trường hợp mở rộng n=0.

```python
def climb_dp_rolling(n: int) -> int:
    prev2, prev1 = 1, 1
    for k in range(2, n + 1):
        prev2, prev1 = prev1, prev2 + prev1
    return prev1

for n in range(1,8):
    print(n, "->", climb_dp_rolling(n))
```

## 10. Trace bằng tay cho `n=5`

| k | prev2 trước = ways(k-2) | prev1 trước = ways(k-1) | current | prev2/prev1 sau |
|---:|---:|---:|---:|---|
| 2 | 1 | 1 | 2 | 1,2 |
| 3 | 1 | 2 | 3 | 2,3 |
| 4 | 2 | 3 | 5 | 3,5 |
| 5 | 3 | 5 | 8 | 5,8 |

Do đó `ways(5)=8`. Khi trace phải nhìn thấy **trước** và **sau**, không chỉ có kết quả cuối.

```python
def trace_climb(n):
    prev2, prev1 = 1, 1
    print("Khởi đầu: ways(0)=1, ways(1)=1")
    for k in range(2,n+1):
        current = prev2 + prev1
        print(f"k={k}: trước={prev2},{prev1} -> tính={current}")
        prev2,prev1 = prev1,current
        print(f"     sau  ={prev2},{prev1}")
    print("TRẢ VỀ",prev1)
    return prev1

trace_climb(5)
```

## 11. Vì sao DP đúng? (invariant)

Trước khi tính bậc k, hai biến luôn là **số cách chính xác** đến `k-2` và `k-1`. Theo cách phân nhóm theo bước cuối, tổng của chúng chính xác là số cách đến `k`. Gán đồng thời duy trì bất biến cho bậc tiếp theo.

Khi k chạy hết đến n, `prev1=ways(n)`, nên kết quả đúng. Không nhầm `ways(0)=1` với “có một cách di chuyển một bước”: ở bậc 0, cách hợp lệ là **không di chuyển**.

## 12. Cách nộp LeetCode

Chỉ dán lớp `Solution`, tên hàm phải là `climbStairs`, input là số nguyên `n`, output là số nguyên (không trả danh sách các đường đi).

```python
class Solution:
    def climbStairs(self, n: int) -> int:
        prev2, prev1 = 1, 1
        for k in range(2, n+1):
            prev2, prev1 = prev1, prev2 + prev1
        return prev1
```
