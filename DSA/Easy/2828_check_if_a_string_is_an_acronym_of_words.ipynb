## 1. Mục tiêu học

Sau notebook này bạn sẽ:

1. Hiểu đề: cho danh sách từ `words` và chuỗi `s`; hỏi `s` có phải là **từ viết tắt (acronym)** của `words` không, tức `s` bằng cách nối **chữ cái đầu** của từng từ **theo đúng thứ tự**.
2. Biết **tự tính tay**: lấy chữ cái đầu của từng từ, ghép lại, rồi so với `s`.
3. Học các công cụ Python: chỉ số `w[0]`, nối chuỗi bằng `+` và `"".join(...)`, biểu thức sinh (generator expression), `zip`, `enumerate`, `all`, so sánh chuỗi `==`.
4. Đi từ **Brute Force: xây chuỗi rồi so sánh** → **`"".join` một dòng** → **so sánh từng vị trí, thoát sớm (không tạo chuỗi phụ)** → **`zip` + `all` (nhớ kiểm độ dài!)**.
5. Chứng minh vì sao phải **kiểm độ dài trước** và tránh các lỗi kinh điển: `zip` âm thầm cắt cụt, so sánh `in` thay vì `==`, quên thứ tự, nhầm chữ cái đầu với chữ cái cuối.

## 2. Đề bài & ví dụ

Cho một mảng chuỗi `words` và một chuỗi `s`. Hãy xác định `s` có phải là **từ viết tắt** (acronym) của `words` hay không.

Chuỗi `s` được coi là từ viết tắt của `words` nếu nó có thể được tạo thành bằng cách **nối chữ cái đầu tiên của mỗi chuỗi trong `words` theo đúng thứ tự**. Ví dụ `"ab"` là từ viết tắt của `["apple", "banana"]`, nhưng `"ba"` thì không.

Trả về `true` nếu `s` là từ viết tắt của `words`, ngược lại trả về `false`.

| Ví dụ | `words`, `s` | Output | Giải thích |
|---|---|---|---|
| 1 | `["alice","bob","charlie"]`, `"abc"` | `true` | Chữ cái đầu: `a`, `b`, `c`; nối lại `"abc"` bằng `s`. |
| 2 | `["an","apple"]`, `"a"` | `false` | Chữ cái đầu: `a`, `a`; nối lại `"aa"`, khác `"a"`. |
| 3 | `["never","gonna","give","up","on","you"]`, `"ngguoy"` | `true` | Chữ cái đầu: `n,g,g,u,o,y` ⇒ `"ngguoy"`. |

Constraints:

- `1 <= words.length <= 100`
- `1 <= words[i].length <= 10`
- `1 <= s.length <= 100`
- `words[i]` và `s` chỉ gồm chữ cái thường tiếng Anh.

````python
# Chạy thử "bằng tay" ví dụ 1
words = ["alice", "bob", "charlie"]
for w in words:
    print(w, "-> chữ cái đầu:", w[0])
print("ghép lại:", "".join(w[0] for w in words))
````

## 3. Kiến thức nền

### 3.1. Chỉ số chuỗi `w[0]`

Chuỗi là dãy ký tự đánh chỉ số từ `0`. `w[0]` là **ký tự đầu**, `w[-1]` là **ký tự cuối**. Với đề này mọi từ có độ dài `>= 1` nên `w[0]` luôn tồn tại (nếu chuỗi rỗng thì `""[0]` gây `IndexError`).

````python
w = "alice"
print(w[0], w[1], w[-1])      # a l e
print(len(w))                 # 5
try:
    print(""[0])
except IndexError as e:
    print("lỗi với chuỗi rỗng:", e)
````

### 3.2. Nối chuỗi: `+` và `"".join(...)`

`"a" + "b"` nối hai chuỗi. Để nối **nhiều** chuỗi, `"".join(danh_sách)` hiệu quả hơn nhiều so với cộng dồn trong vòng lặp (mỗi lần `+` có thể sao chép chuỗi). `"-".join(["a", "b"])` cho `"a-b"` — chuỗi đứng trước `.join` là **dấu nối** giữa các phần tử.

````python
print("a" + "b")                       # ab
print("".join(["a", "b", "c"]))        # abc
print("-".join(["a", "b", "c"]))       # a-b-c
````

### 3.3. Biểu thức sinh (generator expression)

`(w[0] for w in words)` là cách viết ngắn "với mỗi `w` trong `words`, lấy `w[0]`". Nó tạo các giá trị **lười** (không tạo list ngay). Đưa nó thẳng vào `"".join(...)` hoặc `all(...)` rất tiện. Dạng với dấu `[...]` tạo list thật.

````python
words = ["alice", "bob"]
print([w[0] for w in words])           # ['a', 'b']  (list comprehension)
gen = (w[0] for w in words)
print(list(gen))                       # ['a', 'b']
print("".join(w[0] for w in words))    # ab
````

### 3.4. `zip` và `enumerate`

- `zip(a, b)` ghép từng cặp phần tử cùng vị trí; **dừng khi một trong hai hết** (cắt cụt theo cái ngắn hơn!).
- `enumerate(a)` cho từng cặp `(chỉ_số, phần_tử)`.

````python
print(list(zip(["x", "y", "z"], "ab")))       # [('x', 'a'), ('y', 'b')] : 'z' bị bỏ rơi!
for i, w in enumerate(["alice", "bob"]):
    print(i, w)
````

### 3.5. `all(...)` và `any(...)`

`all(iterable)` trả `True` nếu **mọi** phần tử đều đúng (và trả `True` cho dãy rỗng), `any` trả `True` nếu **có ít nhất một** phần tử đúng. Cả hai **dừng sớm** khi biết kết quả.

````python
print(all([True, True]), all([True, False]), all([]))     # True False True
print(any([False, True]), any([False, False]), any([]))   # True False False
````

### 3.6. So sánh chuỗi `==` và toán tử `in`

`a == b` đúng khi hai chuỗi **giống hệt** (cùng độ dài, cùng từng ký tự). Toán tử `x in s` kiểm tra `x` có là **chuỗi con** của `s` — **khác hẳn** `==`. Bài này cần `==`.

````python
print("abc" == "abc", "abc" == "ab", "abc" == "abd")   # True False False
print("ab" in "abc", "ab" == "abc")                     # True False
````

## 4. Ý tưởng nền tảng / cách nghĩ trực giác

Tính tay ví dụ 3 `words = ["never","gonna","give","up","on","you"]`, `s = "ngguoy"`.

| vị trí `i` | từ `words[i]` | chữ cái đầu | ký tự `s[i]` | khớp? |
|---|---|---|---|---|
| 0 | never | n | n | có |
| 1 | gonna | g | g | có |
| 2 | give | g | g | có |
| 3 | up | u | u | có |
| 4 | on | o | o | có |
| 5 | you | y | y | có |

Có **6 từ** và `s` có **6 ký tự**: độ dài khớp, mọi vị trí khớp ⇒ `true`.

Với ví dụ 2 `["an", "apple"]`, `s = "a"`:

| vị trí | từ | chữ cái đầu | `s[i]` | khớp? |
|---|---|---|---|---|
| 0 | an | a | a | có |
| 1 | apple | a | — (`s` hết rồi) | **không** |

`s` chỉ có 1 ký tự nhưng có 2 từ ⇒ `false`. Đây là lý do phải xét cả **độ dài**: số từ phải bằng độ dài `s`.

**Nhận xét:** điều kiện đúng ⇔ (số từ == `len(s)`) **và** (với mọi `i`, `words[i][0] == s[i]`).

````python
words, s = ["an", "apple"], "a"
print("số từ:", len(words), "| độ dài s:", len(s))
print("khớp độ dài?", len(words) == len(s))
````

## 5. Cách 1 — Brute Force / baseline: xây chuỗi viết tắt rồi so sánh

**Ý tưởng:** làm đúng như đề. Khởi tạo `acronym = ""`; với mỗi từ `w` trong `words`, nối `w[0]` vào `acronym`; cuối cùng `return acronym == s`.

````python
from typing import List

def is_acronym_build(words: List[str], s: str) -> bool:
    acronym = ""
    for w in words:
        acronym = acronym + w[0]      # nối chữ cái đầu
    return acronym == s
````

````python
print(is_acronym_build(["alice", "bob", "charlie"], "abc"))                       # True
print(is_acronym_build(["an", "apple"], "a"))                                      # False
print(is_acronym_build(["never", "gonna", "give", "up", "on", "you"], "ngguoy"))   # True
````

**Độ phức tạp:** gọi `n = len(words)`. Mỗi lần nối `+` có thể sao chép chuỗi hiện tại (độ dài tới `n`) ⇒ tệ nhất `O(n²)` theo mô hình sao chép, dù CPython thường tối ưu phép `+=`; so sánh cuối `O(n)`. Space `O(n)` cho chuỗi `acronym`.

**Nút thắt:** ta tạo cả một chuỗi trung gian chỉ để so sánh; và nối bằng `+` trong vòng lặp là thói quen kém hiệu quả. Nếu chuỗi `s` dài khác `n`, ta biết ngay là `false` mà vẫn xây hết chuỗi.

## 6. Ý tưởng thuật toán / bước đột phá

**Đột phá 1 — `"".join` thay cho nối từng bước.** `"".join(w[0] for w in words)` ghép một lần, `O(n)`.

**Đột phá 2 — kiểm độ dài trước.** Nếu `len(words) != len(s)` thì chắc chắn `false`: trả về ngay, không cần động tới ký tự nào.

**Đột phá 3 — so sánh từng vị trí, thoát sớm, không tạo chuỗi phụ.** Sau khi biết độ dài bằng nhau, duyệt `i` từ `0` tới `n − 1`: nếu `words[i][0] != s[i]` thì trả `false` ngay; duyệt hết mà không lệch thì `true`. Không tạo chuỗi nào ⇒ Space `O(1)`.

Đây là mô hình "so sánh hai dãy song song" rất hay gặp.

````python
words = ["an", "apple"]
s = "a"
if len(words) != len(s):
    print("độ dài khác nhau -> False ngay, không cần so ký tự")
````

````python
words = ["alice", "bob", "carol"]
s = "abd"
for i in range(len(words)):
    print(i, words[i][0], s[i], "khớp" if words[i][0] == s[i] else "LỆCH -> dừng, False")
    if words[i][0] != s[i]:
        break
````

## 7. Các cách tiếp theo

Ta giảng thêm 3 cách nữa.

### Cách 2 — `"".join` một dòng

**Từng đoạn:**

- `(w[0] for w in words)`: biểu thức sinh các chữ cái đầu.
- `"".join(...)`: ghép chúng thành một chuỗi.
- `== s`: so sánh với `s`.

Ngắn và đọc dễ; nhưng vẫn tạo chuỗi phụ độ dài `n`.

````python
def is_acronym_join(words: List[str], s: str) -> bool:
    return "".join(w[0] for w in words) == s
````

````python
print(is_acronym_join(["alice", "bob", "charlie"], "abc"))     # True
print(is_acronym_join(["an", "apple"], "a"))                    # False
````

**Code nộp LeetCode (Cách 2):**

```python

class Solution:
    def isAcronym(self, words: List[str], s: str) -> bool:
        return "".join(w[0] for w in words) == s
```

**Độ phức tạp:** Time `O(n)`, Space `O(n)`.

### Cách 3 — So sánh từng vị trí, thoát sớm (không chuỗi phụ)

**Từng đoạn:**

- `if len(words) != len(s): return False` — độ dài khác nhau thì chắc chắn sai.
- `for i in range(len(words))`: duyệt song song.
- `if words[i][0] != s[i]: return False` — gặp chỗ lệch thì dừng.
- Hết vòng lặp ⇒ `True`.

````python
def is_acronym_index(words: List[str], s: str) -> bool:
    if len(words) != len(s):
        return False
    for i in range(len(words)):
        if words[i][0] != s[i]:
            return False
    return True
````

````python
print(is_acronym_index(["alice", "bob", "charlie"], "abc"))     # True
print(is_acronym_index(["an", "apple"], "a"))                    # False
print(is_acronym_index(["ab", "cd"], "ad"))                      # False (lệch ở vị trí 1)
````

**Code nộp LeetCode (Cách 3):**

```python
from typing import List

class Solution:
    def isAcronym(self, words: List[str], s: str) -> bool:
        if len(words) != len(s):
            return False
        for i in range(len(words)):
            if words[i][0] != s[i]:
                return False

```

**Độ phức tạp:** Time `O(n)` (tốt nhất `O(1)` nếu lệch độ dài hoặc lệch ngay vị trí đầu), Space `O(1)`.

### Cách 4 — `zip` + `all` (nhớ kiểm độ dài!)

**Từng đoạn:**

- `zip(words, s)` ghép `words[i]` với `s[i]` cùng vị trí.
- `all(w[0] == c for w, c in zip(words, s))` kiểm mọi cặp khớp.
- Nhưng `zip` **cắt cụt** theo dãy ngắn hơn ⇒ **bắt buộc** thêm `len(words) == len(s)`; nếu không `["an", "apple"]` với `"a"` bị coi là `True` sai.

````python
def is_acronym_zip(words: List[str], s: str) -> bool:
    return len(words) == len(s) and all(w[0] == c for w, c in zip(words, s))
````

````python
print(is_acronym_zip(["alice", "bob", "charlie"], "abc"))     # True
print(is_acronym_zip(["an", "apple"], "a"))                    # False
````

**Code nộp LeetCode (Cách 4):**

```python
from typing import List

    def isAcronym(self, words: List[str], s: str) -> bool:

```

**Độ phức tạp:** Time `O(n)` (`and` và `all` đều dừng sớm), Space `O(1)`.

## 8. TRACE từng bước

Trace Cách 3 (so sánh từng vị trí) trên hai ví dụ: một `true` (`["never","gonna","give","up","on","you"]`, `"ngguoy"`) và một `false` ở giữa (`["alice","bob","carol"]`, `"abd"`).

| `i` | `words[i]` | `words[i][0]` | `s[i]` | kết luận |
|---|---|---|---|---|
| 0 | alice | a | a | khớp, tiếp |
| 1 | bob | b | b | khớp, tiếp |
| 2 | carol | c | d | **lệch ⇒ trả `False`** |

````python
def trace_acronym(words, s):
    print(f"số từ = {len(words)}, len(s) = {len(s)}")
    if len(words) != len(s):
        print("độ dài khác nhau -> False")
        return False
    for i in range(len(words)):
        a, b = words[i][0], s[i]
        trang_thai = "khớp" if a == b else "LỆCH -> False"
        print(f"i={i}: words[i]={words[i]!r:10} chữ cái đầu={a!r}, s[i]={b!r} -> {trang_thai}")
        if a != b:
            return False
    print("mọi vị trí khớp -> True")
    return True

trace_acronym(["never", "gonna", "give", "up", "on", "you"], "ngguoy")
````

````python
trace_acronym(["alice", "bob", "carol"], "abd")
print("---")
trace_acronym(["an", "apple"], "a")
````

## 9. Code nộp LeetCode

Dùng Cách 3 (so sánh từng vị trí): rõ ràng, thoát sớm, không chuỗi phụ.

````python
from typing import List

class Solution:
    def isAcronym(self, words: List[str], s: str) -> bool:
        if len(words) != len(s):          # số từ phải bằng độ dài s
            return False
        for i in range(len(words)):
            if words[i][0] != s[i]:       # chữ cái đầu của từ thứ i phải là s[i]
                return False
        return True
````

````python
print(Solution().isAcronym(["alice", "bob", "charlie"], "abc"))     # True
print(Solution().isAcronym(["an", "apple"], "a"))                    # False
````

## 10. Vì sao thuật toán đúng

Gọi `A` là chuỗi nối chữ cái đầu của `words` theo thứ tự: `A = words[0][0] + words[1][0] + ... + words[n−1][0]`, trong đó `n = len(words)`. Đề hỏi `A == s`.

**Hai chuỗi bằng nhau ⇔ cùng độ dài và mọi vị trí giống nhau.** `A` có đúng `n` ký tự (mỗi từ cho một chữ cái đầu, vì `words[i].length >= 1`). Vậy `A == s` ⇔ `len(s) == n` **và** với mọi `0 <= i < n`: `A[i] == s[i]`, mà `A[i] = words[i][0]`.

- **Kiểm `len(words) != len(s)` ⇒ `False`:** nếu độ dài khác nhau thì `A != s`, đúng với phát biểu trên.
- **Vòng lặp `for` với `words[i][0] != s[i] ⇒ False`:** nếu một vị trí lệch thì `A != s`.
- **Chạy hết không lệch ⇒ `True`:** độ dài bằng nhau và mọi vị trí khớp ⇒ `A == s`.

**Vì sao `s[i]` không vượt biên:** sau khi đã chắc `len(s) == n`, các chỉ số `i = 0..n−1` đều hợp lệ cho `s`. Nếu bỏ kiểm độ dài, `s[i]` có thể gây `IndexError` (khi `s` ngắn hơn) hoặc bỏ sót phần dư (khi `s` dài hơn).

**Vì sao `words[i][0]` luôn hợp lệ:** constraints cam kết `words[i].length >= 1`.

## 11. Phân tích Time / Space Complexity

`n = len(words)`; xét độ dài `s` cũng `<= 100`.

| Cách | Time | Space | Ghi chú |
|---|---|---|---|
| 1. Xây chuỗi bằng `+` rồi so sánh | `O(n)`–`O(n²)` (tuỳ cách `+` sao chép) | `O(n)` | dễ hiểu |
| 2. `"".join(...) == s` | `O(n)` | `O(n)` | ngắn gọn |
| 3. So sánh từng vị trí, thoát sớm | `O(n)` (tốt nhất `O(1)`) | `O(1)` | **code nộp** |
| 4. `zip` + `all` | `O(n)` | `O(1)` | ngắn, nhớ kiểm độ dài |

Với `n <= 100`, mọi cách tức thì; khác biệt chỉ ở độ gọn và tư duy "thoát sớm".

## 12. Bộ test

### 12.1. Ví dụ của đề và edge case

Các ca quan trọng: đúng; `s` ngắn hơn số từ; `s` dài hơn số từ; một từ một chữ; cùng chữ cái đầu lặp lại; sai thứ tự; sai ở cuối; từ dài nhưng chữ đầu khớp.

````python
impls = {
    "build": is_acronym_build,
    "join": is_acronym_join,
    "index": is_acronym_index,
    "zip": is_acronym_zip,
    "Solution": Solution().isAcronym,
}
cases = [
    ("ví dụ 1", ["alice", "bob", "charlie"], "abc", True),
    ("ví dụ 2", ["an", "apple"], "a", False),
    ("ví dụ 3", ["never", "gonna", "give", "up", "on", "you"], "ngguoy", True),
    ("s dài hơn số từ", ["a"], "ab", False),
    ("s ngắn hơn số từ", ["a", "b", "c"], "ab", False),
    ("một từ một chữ", ["z"], "z", True),
    ("một từ, sai chữ", ["z"], "y", False),
    ("sai thứ tự", ["apple", "banana"], "ba", False),
    ("sai ở cuối", ["ab", "bc", "cd"], "abd", False),
    ("từ dài chữ đầu lặp", ["aaaaaaaaaa", "aaaaaaaaaa"], "aa", True),
    ("chữ cái cuối không phải chữ đầu", ["ab", "cd"], "bd", False),
]
for name, words, s, exp in cases:
    for k, f in impls.items():
        got = f(list(words), s)
        assert got == exp, (name, k, got, exp)
    print("PASS", name, "->", exp)
````

## 13. Các lỗi người mới hay gặp

### Lỗi 1: Dùng `zip` mà quên kiểm độ dài

````python
def sai_zip_khong_kiem(words, s):
    return all(w[0] == c for w, c in zip(words, s))     # sai: zip cắt cụt

print("sai :", sai_zip_khong_kiem(["an", "apple"], "a"))        # True (sai)
print("đúng:", Solution().isAcronym(["an", "apple"], "a"))       # False
print("sai :", sai_zip_khong_kiem(["a"], "abc"), "| đúng:", Solution().isAcronym(["a"], "abc"))
````

**Vì sao sai:** `zip` dừng khi một dãy hết nên phần dư bị bỏ qua. Ví dụ 2 của đề chính là bẫy này. Luôn so sánh `len(words) == len(s)`.

### Lỗi 2: Dùng `in` thay `==`

````python
def sai_dung_in(words, s):
    acronym = "".join(w[0] for w in words)
    return s in acronym             # sai: chỉ kiểm chuỗi con

print("sai :", sai_dung_in(["alice", "bob", "charlie"], "ab"))      # True (sai)
print("đúng:", Solution().isAcronym(["alice", "bob", "charlie"], "ab"))   # False
````

**Vì sao sai:** `"ab" in "abc"` là `True`, nhưng `s` phải **bằng đúng** chuỗi viết tắt. Dùng `==`.

### Lỗi 3: Nhầm chữ cái cuối với chữ cái đầu

````python
def sai_chu_cuoi(words, s):
    return "".join(w[-1] for w in words) == s      # sai: w[-1] là chữ cuối

print("sai :", sai_chu_cuoi(["alice", "bob", "charlie"], "abc"))     # False (sai)
print("đúng:", Solution().isAcronym(["alice", "bob", "charlie"], "abc"))   # True
````

**Vì sao sai:** `w[-1]` là ký tự **cuối**; chữ cái **đầu** là `w[0]`. Chỉ số âm đếm từ cuối.

### Lỗi 4: Sắp xếp lại hoặc bỏ qua thứ tự

````python
def sai_sap_xep(words, s):
    return sorted(w[0] for w in words) == sorted(s)   # sai: bỏ qua thứ tự

print("sai :", sai_sap_xep(["apple", "banana"], "ba"))             # True (sai)
print("đúng:", Solution().isAcronym(["apple", "banana"], "ba"))     # False
````

**Vì sao sai:** đề yêu cầu đúng **thứ tự**; `"ab"` là viết tắt của `["apple","banana"]` nhưng `"ba"` thì không.

LeetCode Python trả kiểu `bool` của Python: `True` / `False` (viết hoa chữ cái đầu). Chuỗi `"true"` hoặc `"True"` bị coi là kết quả sai kiểu.

## 14. Mẹo làm bài nhanh & nhận dạng pattern

- Đề "ghép chữ cái đầu / cuối / chữ thứ `k` của từng từ rồi so với chuỗi cho trước" ⇒ `"".join(w[k] for w in words) == s`, hoặc so từng vị trí.
- Khi so hai dãy **song song**, nhớ kiểm **độ dài bằng nhau** trước; `zip` không cảnh báo độ dài khác.
- Mẫu "thoát sớm": `for ... if sai: return False` rồi `return True` ở cuối. Rất phổ biến trong bài kiểm tra điều kiện.
- `w[0]` = đầu, `w[-1]` = cuối; chỉ số âm đếm ngược từ cuối.
- Dùng `==` cho "bằng nhau", `in` cho "chứa".

## 15. Biến thể của bài

1. **Viết tắt bằng chữ cái cuối** hoặc chữ cái thứ `k` (nếu đủ dài).
2. **Không phân biệt hoa / thường** (`lower()` trước khi so sánh).
3. **Chuỗi cho dạng một câu:** `sentence.split()` rồi lấy chữ cái đầu.
4. **Hỏi ngược:** cho `acronym`, kiểm một danh sách từ có phải nguồn của nó không (đúng như bài này).
5. **Cho phép bỏ qua một số từ** (như "of", "the") khi tạo viết tắt: tập `stop_words`.
6. **Viết tắt là tiền tố của chuỗi** (không phải bằng): `s.startswith(acronym)` hay so sánh đoạn đầu.

## 16. So sánh các cách làm

| Cách | Ý tưởng | Time | Space | Khi nào dùng |

| 1. Xây chuỗi bằng `+` | dịch thẳng đề | `O(n)`–`O(n²)` | `O(n)` | người mới, oracle |
| 2. `"".join(...) == s` | một dòng | `O(n)` | `O(n)` | code ngắn, dễ đọc |

class Solution:
    def isAcronym(self, words: List[str], s: str) -> bool:
