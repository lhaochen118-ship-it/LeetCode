## 1. Mục tiêu học

Sau notebook này bạn sẽ:

1. Hiểu đề: **nối** (concatenate) hai số bằng cách viết chúng liền nhau (`15` nối `49` ra `1549`); lặp lại việc ghép **số đầu** với **số cuối** của mảng và cộng giá trị nối vào kết quả; nếu còn **một** số thì cộng chính nó.
2. Dùng **hai con trỏ** `lo`, `hi` từ hai đầu mảng, và xử lý riêng trường hợp **còn đúng một phần tử** (mảng có độ dài lẻ).
3. Nối số bằng chuỗi (`int(str(a) + str(b))`) và bằng **toán học** (`a * 10^(số chữ số của b) + b`).
4. Viết 4 cách: mô phỏng `pop(0)`/`pop()`, hai con trỏ + chuỗi, hai con trỏ + toán học (không chuỗi), và `deque`; đối chiếu bằng test ngẫu nhiên.
5. Tránh các lỗi: bỏ sót phần tử giữa khi `n` lẻ, **cộng** thay vì **nối**, nhân cố định `* 10`, và nối phần tử giữa với chính nó.

## 2. Đề bài & ví dụ

Cho mảng số nguyên `nums` (chỉ số từ 0). **Nối** hai số là viết chúng liền nhau theo thứ tự: ví dụ nối `15` và `49` được `1549`.

**Giá trị nối của mảng (concatenation value)** ban đầu là `0`. Lặp lại cho tới khi `nums` rỗng:

- Nếu `nums` có **nhiều hơn một** phần tử: lấy số **đầu tiên** và số **cuối cùng**, nối chúng (đầu trước, cuối sau), **cộng** giá trị nối đó vào kết quả, rồi **xoá** cả hai khỏi `nums`.
- Nếu `nums` có **đúng một** phần tử: cộng giá trị của nó vào kết quả và xoá nó.

Trả về giá trị nối cuối cùng.

| Ví dụ | nums | Output | Giải thích |
|---|---|---|---|
| 1 | `[7,52,2,4]` | 596 | `7` nối `4` = `74`; `52` nối `2` = `522`; `74 + 522 = 596` |
| 2 | `[5,14,13,8,12]` | 673 | `5` nối `12` = `512`; `14` nối `8` = `148`; còn `13`; `512 + 148 + 13 = 673` |

Constraints: `1 <= nums.length <= 1000`; `1 <= nums[i] <= 10^4`.

````python
nums = [5, 14, 13, 8, 12]
work = list(nums)
total = 0
while work:
    if len(work) > 1:
        a, b = work.pop(0), work.pop()
        value = int(str(a) + str(b))
        print(f"nối {a} và {b} -> {value}")
    else:
        value = work.pop()
        print(f"còn một số {value}")
    total += value
print("tổng =", total)
````

## 3. Kiến thức nền

### 3.1. Nối số bằng chuỗi: `str` rồi `int`

`str(15)` là `"15"`; phép `+` giữa hai chuỗi là **nối chuỗi**: `"15" + "49" = "1549"`; `int("1549")` đổi lại thành số `1549`. Lưu ý: `15 + 49` (cộng số) là `64`, hoàn toàn khác.

````python
a, b = 15, 49
print(str(a) + str(b), int(str(a) + str(b)))
print(a + b)
print(int(str(7) + str(4)), int(str(52) + str(2)))
````

### 3.2. Nối số bằng toán học: `a * 10^k + b`

Nối `a` với `b` (có `k` chữ số) nghĩa là **đẩy** `a` sang trái `k` hàng rồi cộng `b`: `a * 10**k + b`. Ví dụ `15` nối `49` (`k = 2`): `15 * 100 + 49 = 1549`. Số chữ số `k` có thể lấy bằng `len(str(b))`, hoặc tính bằng vòng lặp nhân `10` cho tới khi vượt `b` (không dùng chuỗi).

````python
a, b = 15, 49
k = len(str(b))
print(k, a * 10 ** k + b)
p = 1
while p <= b:          # tìm 10^k: luỹ thừa 10 nhỏ nhất lớn hơn b
    p *= 10
print(p, a * p + b)
````

### 3.3. Hai con trỏ từ hai đầu

`lo = 0`, `hi = len(nums) - 1`. Lặp trong khi `lo < hi`: xử lý cặp `(nums[lo], nums[hi])`, rồi `lo += 1`, `hi -= 1`. Khi vòng lặp dừng: nếu `lo == hi` thì còn **đúng một** phần tử ở giữa (mảng độ dài lẻ); nếu `lo > hi` thì đã hết (độ dài chẵn).

````python
for n in [4, 5]:
    lo, hi = 0, n - 1
    pairs = []
    while lo < hi:
        pairs.append((lo, hi))
        lo += 1
        hi -= 1
    print(f"n={n}: các cặp {pairs}; dừng với lo={lo}, hi={hi}; còn giữa? {lo == hi}")
````

### 3.4. `list.pop(0)` và `list.pop()`; `collections.deque`

`lst.pop()` bỏ và trả về phần tử **cuối** (`O(1)`). `lst.pop(0)` bỏ phần tử **đầu** nhưng phải dịch các phần tử còn lại nên tốn `O(n)`. `collections.deque` hỗ trợ cả `popleft()` và `pop()` đều `O(1)`, nên phù hợp khi cần lấy từ hai đầu.

````python
from collections import deque
lst = [1, 2, 3, 4]
print(lst.pop(0), lst.pop(), lst)
dq = deque([1, 2, 3, 4])
print(dq.popleft(), dq.pop(), list(dq))
````

### 3.5. Độ dài chẵn / lẻ và `n // 2`, `n % 2`

Với `n` phần tử, ta ghép được `n // 2` cặp `(đầu, cuối)`; nếu `n % 2 == 1` còn **một** phần tử ở chỉ số `n // 2`. Ví dụ `n = 5`: 2 cặp `(0,4)`, `(1,3)` và phần tử giữa ở chỉ số `2`. Biết điều này cho phép viết vòng `for i in range(n // 2)` và một câu `if n % 2: ...` thay vì dùng `while`.

````python
for n in [1, 4, 5, 6]:
    print(f"n={n}: số cặp = {n // 2}, còn phần tử giữa? {n % 2 == 1}", end="")
    print(f", chỉ số giữa = {n // 2}" if n % 2 else "")
````

### 3.6. Tổng giá trị lớn và kiểu số nguyên

`nums[i] <= 10^4` nên mỗi số có tối đa 5 chữ số, và giá trị nối tối đa `10^4` nối `10^4` ≈ `10^9` (10 chữ số); với `n = 1000` có tối đa 500 cặp nên tổng nhỏ hơn `5 * 10^11`. Vượt `int` 32 bit (`~2,1 * 10^9`) → ở C++/Java cần `long long`. Python không tràn.

````python
big = int(str(10**4) + str(10**4))
print(big, 500 * big, 500 * big > 2**31 - 1)
````

## 4. Ý tưởng nền tảng / cách nghĩ trực giác

Làm đúng như đề: mỗi vòng lấy hai đầu, nối, cộng, bỏ đi; còn một số thì cộng luôn.

Ví dụ 1 `[7, 52, 2, 4]`:

```
vòng 1: đầu = 7, cuối = 4 -> "74" = 74      còn [52, 2]    tổng 74
vòng 2: đầu = 52, cuối = 2 -> "522" = 522    còn []         tổng 596
```

Ví dụ 2 `[5, 14, 13, 8, 12]`:

```
vòng 1: 5 và 12 -> 512   còn [14, 13, 8]
vòng 2: 14 và 8 -> 148   còn [13]
vòng 3: còn một số 13 -> cộng 13      tổng 512 + 148 + 13 = 673
```

Nhận xét: mảng thực ra **không bị thay đổi thật sự** nếu ta dùng hai chỉ số `lo`, `hi` thay cho việc xoá phần tử — đó là hai con trỏ.

## 5. Cách 1 — Brute Force / baseline

Mô phỏng nguyên văn: tạo bản sao `work`, trong khi `work` còn phần tử: nếu còn nhiều hơn một thì `pop(0)` và `pop()`, nối bằng chuỗi, cộng; nếu còn một thì `pop()` và cộng.

Nút thắt: `pop(0)` tốn `O(n)` (dịch toàn bộ phần tử còn lại) và lặp `n / 2` lần → `O(n^2)` trong trường hợp xấu; ngoài ra phải tạo bản sao để không phá `nums`.

````python
class SolutionPop:
    def findTheArrayConcVal(self, nums):
        work = list(nums)
        total = 0
        while work:
            if len(work) > 1:
                a = work.pop(0)
                b = work.pop()
                total += int(str(a) + str(b))
            else:
                total += work.pop()
        return total

print(SolutionPop().findTheArrayConcVal([7, 52, 2, 4]))
print(SolutionPop().findTheArrayConcVal([5, 14, 13, 8, 12]))
````

**Độ phức tạp:** Time `O(n^2)` do `pop(0)` (cộng thêm thời gian nối chuỗi `O(1)` vì tối đa 5 chữ số mỗi số), Space `O(n)` cho bản sao.

## 6. Ý tưởng thuật toán / bước đột phá

**Bước đột phá:** không cần xoá phần tử. Dùng hai chỉ số `lo` (đầu) và `hi` (cuối) di chuyển vào trong; "xoá" cả hai nghĩa là `lo += 1`, `hi -= 1`. Mảng gốc giữ nguyên và mỗi bước chỉ `O(1)`.

**Quy tắc "nếu… thì…":**

- Nếu `lo < hi` → nối `nums[lo]` với `nums[hi]`, cộng vào tổng, rồi `lo += 1`, `hi -= 1`.
- Nếu `lo == hi` → còn đúng một số: cộng `nums[lo]` (vì đề: còn một thì cộng chính nó, **không nối**).
- Nếu `lo > hi` → xong.

Invariant: tại đầu mỗi vòng, các phần tử `nums[lo..hi]` chính là phần còn lại của mảng; `total` bằng tổng các giá trị nối của các cặp đã xử lý.

````python
nums = [5, 14, 13, 8, 12]
lo, hi, total = 0, len(nums) - 1, 0
while lo < hi:
    total += int(str(nums[lo]) + str(nums[hi]))
    lo += 1
    hi -= 1
if lo == hi:
    total += nums[lo]
print("tổng =", total)
````

## 7. Các cách tiếp theo

Ta có 4 cách: (1) mô phỏng `pop(0)`/`pop()`; (2) hai con trỏ + nối chuỗi; (3) hai con trỏ + toán học (`a * 10^k + b`, không chuỗi); (4) `deque` với `popleft()`/`pop()`.

### Cách 2 — Hai con trỏ + nối chuỗi

Dùng `lo`, `hi` như mục 6 và nối bằng `int(str(a) + str(b))`. Không sửa mảng, `O(n)`.

````python
class SolutionTwoPointers:
    def findTheArrayConcVal(self, nums):
        lo, hi = 0, len(nums) - 1
        total = 0
        while lo < hi:
            total += int(str(nums[lo]) + str(nums[hi]))
            lo += 1
            hi -= 1
        if lo == hi:                      # còn đúng một phần tử ở giữa
            total += nums[lo]
        return total

print(SolutionTwoPointers().findTheArrayConcVal([7, 52, 2, 4]))
print(SolutionTwoPointers().findTheArrayConcVal([5, 14, 13, 8, 12]))
````

Độ phức tạp Cách 2: Time `O(n)` (nối chuỗi trên số tối đa 5 chữ số là `O(1)`), Space `O(1)` thêm.

### Cách 3 — Hai con trỏ + toán học (không dùng chuỗi)

Nối `a` với `b` bằng `a * p + b`, trong đó `p` là luỹ thừa `10` nhỏ nhất **lớn hơn** `b` (`p = 10` nếu `b` một chữ số, `100` nếu hai chữ số, ...). Tính `p` bằng vòng lặp nhân `10`. Cách này hữu ích khi không muốn đổi qua chuỗi.

````python
class SolutionMath:
    def findTheArrayConcVal(self, nums):
        def concat(a, b):
            p = 1
            while p <= b:
                p *= 10
            return a * p + b

        total = 0
        for i in range(len(nums) // 2):
            total += concat(nums[i], nums[len(nums) - 1 - i])
        if len(nums) % 2 == 1:
            total += nums[len(nums) // 2]
        return total

print(SolutionMath().findTheArrayConcVal([7, 52, 2, 4]))
print(SolutionMath().findTheArrayConcVal([5, 14, 13, 8, 12]))
````

Độ phức tạp Cách 3: Time `O(n)` (mỗi `concat` tối đa 5 lần nhân), Space `O(1)`.

### Cách 4 — `deque` với `popleft()` / `pop()`

Giống cách 1 (mô phỏng xoá hai đầu) nhưng dùng `deque` nên lấy hai đầu đều `O(1)`: `O(n)` tổng. Gần với mô tả của đề nhất mà vẫn nhanh.

````python
class SolutionDeque:
    def findTheArrayConcVal(self, nums):
        dq = deque(nums)
        total = 0
        while dq:
            if len(dq) > 1:
                a = dq.popleft()
                b = dq.pop()
                total += int(str(a) + str(b))
            else:
                total += dq.pop()
        return total

print(SolutionDeque().findTheArrayConcVal([7, 52, 2, 4]))
print(SolutionDeque().findTheArrayConcVal([5, 14, 13, 8, 12]))
````

Độ phức tạp Cách 4: Time `O(n)`, Space `O(n)` (deque là một bản sao của mảng).

## 8. TRACE từng bước

Trace Cách 2 với `nums = [5, 14, 13, 8, 12]`:

| bước | `lo` | `hi` | `nums[lo]` | `nums[hi]` | giá trị nối | `total` |
|---|---|---|---|---|---|---|
| 1 | 0 | 4 | 5 | 12 | 512 | 512 |
| 2 | 1 | 3 | 14 | 8 | 148 | 660 |
| 3 | 2 | 2 | 13 | (trùng) | cộng `13` | 673 |

Sau bước 2: `lo = 2`, `hi = 2` → `lo == hi`, còn một số ở giữa; cộng `13`. Kết quả `673`.

````python
def trace_concat(nums):
    lo, hi, total = 0, len(nums) - 1, 0
    step = 0
    while lo < hi:
        step += 1
        value = int(str(nums[lo]) + str(nums[hi]))
        total += value
        print(f"bước {step}: lo={lo} hi={hi} nối {nums[lo]} và {nums[hi]} = {value} | total={total}")
        lo += 1
        hi -= 1
    if lo == hi:
        step += 1
        total += nums[lo]
        print(f"bước {step}: lo=hi={lo}, còn một số {nums[lo]} -> cộng nguyên giá trị | total={total}")
    else:
        print(f"dừng: lo={lo} > hi={hi}, không còn phần tử giữa")
    return total

trace_concat([5, 14, 13, 8, 12])
trace_concat([7, 52, 2, 4])
````

## 9. Code nộp LeetCode

````python
from typing import List

class Solution:
    def findTheArrayConcVal(self, nums: List[int]) -> int:
        lo, hi = 0, len(nums) - 1
        total = 0
        while lo < hi:
            total += int(str(nums[lo]) + str(nums[hi]))
            lo += 1
            hi -= 1
        if lo == hi:
            total += nums[lo]
        return total

print(Solution().findTheArrayConcVal([7, 52, 2, 4]), Solution().findTheArrayConcVal([5, 14, 13, 8, 12]))
````

## 10. Vì sao thuật toán đúng

- **Hai con trỏ mô phỏng đúng việc xoá hai đầu:** tại mỗi bước, phần còn lại của mảng là `nums[lo..hi]`; phần tử đầu là `nums[lo]` và cuối là `nums[hi]`. Sau khi ghép và "xoá" cả hai, phần còn lại là `nums[lo+1..hi-1]`, tương ứng `lo += 1`, `hi -= 1`.
- **Điều kiện dừng đúng:** nếu `n` chẵn thì sau `n / 2` bước `lo = hi + 1` (không còn gì); nếu `n` lẻ thì sau `(n-1)/2` bước `lo = hi` và còn đúng một số; đề nói số đơn lẻ được **cộng chính nó** (không nối với ai) nên `total += nums[lo]`.
- **Nối số:** viết `a` liền `b` trong hệ thập phân là `a * 10^k + b` với `k` là số chữ số của `b`: các chữ số của `a` dịch sang trái `k` hàng, `b` chiếm `k` hàng thấp. `int(str(a) + str(b))` cho cùng giá trị. (Không có số `0` đứng đầu vì `nums[i] >= 1`, nên `k = len(str(b))` chính xác.)
- **Tổng:** kết quả là tổng giá trị nối của từng cặp cộng với phần tử giữa (nếu có), mỗi cặp/phần tử được xử lý đúng một lần.
- **Cách 4 (deque):** là mô phỏng trực tiếp của đề với hai thao tác lấy phần tử đầu/cuối `O(1)`.

## 11. Phân tích Time / Space Complexity

Gọi `n = len(nums)`; mỗi số có tối đa 5 chữ số nên nối là `O(1)`.

| Cách | Time | Space |
|---|---|---|
| 1. `pop(0)` / `pop()` trên list | `O(n^2)` | `O(n)` |
| 2. Hai con trỏ + nối chuỗi | `O(n)` | `O(1)` |
| 3. Hai con trỏ + toán học | `O(n)` | `O(1)` |
| 4. `deque` | `O(n)` | `O(n)` |

Cách 2 và 3 là tối ưu (`O(n)` thời gian, `O(1)` bộ nhớ thêm). Cách 1 chậm hơn vì `pop(0)` nhưng với `n <= 1000` vẫn tức thì.

## 12. Bộ test

Test 1: ví dụ đề và các edge case (`n = 1`, `n = 2`, `n` lẻ/chẵn, số `10^4` nhiều chữ số, các số một chữ số).

````python
impls = [SolutionPop(), SolutionTwoPointers(), SolutionMath(), SolutionDeque(), Solution()]
tests = [
    ([7, 52, 2, 4], 596),
    ([5, 14, 13, 8, 12], 673),
    ([1], 1),                              # n = 1: chỉ còn một số
    ([10000], 10000),                      # n = 1, biên trên
    ([1, 2], 12),                          # n = 2: 1 nối 2 (bắt lỗi nối sai thứ tự, cộng)
    ([7, 52], 752),                        # b có 2 chữ số (bắt lỗi nhân cố định * 10)
    ([10000, 10000], 1000010000),          # biên trên, 10 chữ số
    ([1, 2, 3], 15),                       # (1 nối 3) = 13 + 2 = 15 (bắt lỗi bỏ phần tử giữa)
    ([13], 13),                            # bắt lỗi nối phần tử giữa với chính nó (1313)
    ([1, 1, 1, 1], 22),                    # 11 + 11
    ([9, 99, 999, 9999, 9], 99 + 999999 + 999),         # 9||9, 99||9999, giữa 999
    ([100, 20, 3], 1003 + 20),
    ([2, 3, 4, 5, 6, 7], 27 + 36 + 45),
]
def oracle(nums):
    total = 0
    work = [str(x) for x in nums]
    while work:
        if len(work) > 1:
            total += int(work[0] + work[-1])
            work = work[1:-1]
        else:
            total += int(work[0])
            work = []
    return total
for nums, exp in tests:
    assert 1 <= len(nums) <= 1000 and all(1 <= v <= 10**4 for v in nums)
    assert oracle(nums) == exp, (nums, oracle(nums), exp)
    for im in impls:
        assert im.findTheArrayConcVal(list(nums)) == exp, (type(im).__name__, nums, exp)
print("tất cả test cố định OK:", len(tests))
````

Test 2: ngẫu nhiên có seed, đúng constraints (`1 <= n <= 1000`, `1 <= nums[i] <= 10^4`); oracle độc lập cắt dần hai đầu của danh sách chuỗi.

````python
import random
random.seed(2562)

for t in range(1500):
    n = random.randint(1, 1000) if t % 100 == 0 else random.randint(1, 12)
    hi_val = random.choice([9, 99, 999, 10**4])
    nums = [random.randint(1, hi_val) for _ in range(n)]
    assert 1 <= len(nums) <= 1000 and all(1 <= v <= 10**4 for v in nums)
    exp = oracle(nums)
    for im in impls:
        assert im.findTheArrayConcVal(list(nums)) == exp, (type(im).__name__, nums[:10])
print("1500 test ngẫu nhiên OK")
````

## 13. Các lỗi người mới hay gặp

### Lỗi 1: bỏ sót phần tử giữa khi `n` lẻ

````python
def wrong_no_middle(nums):
    lo, hi, total = 0, len(nums) - 1, 0
    while lo < hi:
        total += int(str(nums[lo]) + str(nums[hi]))
        lo += 1
        hi -= 1
    return total                              # SAI: thiếu "nếu lo == hi thì cộng nums[lo]"

nums = [5, 14, 13, 8, 12]
print("sai :", wrong_no_middle(nums))
print("đúng:", Solution().findTheArrayConcVal(nums))
assert wrong_no_middle(nums) != Solution().findTheArrayConcVal(nums)
````

Vì sao sai: khi `n` lẻ, sau khi hai con trỏ gặp nhau vẫn còn một số (`13`) mà đề bắt buộc cộng vào kết quả. Kết quả sai `660` thiếu đúng `13`. **Sửa:** sau vòng lặp, `if lo == hi: total += nums[lo]`.

### Lỗi 2: **cộng** hai số thay vì **nối**

````python
def wrong_add(nums):
    lo, hi, total = 0, len(nums) - 1, 0
    while lo < hi:
        total += nums[lo] + nums[hi]              # SAI: cộng chứ không nối
        lo += 1
        hi -= 1
    if lo == hi:
        total += nums[lo]
    return total

nums = [7, 52, 2, 4]
print("sai :", wrong_add(nums))
print("đúng:", Solution().findTheArrayConcVal(nums))
assert wrong_add(nums) != Solution().findTheArrayConcVal(nums)
````

Vì sao sai: `7 + 4 = 11` nhưng nối `7` và `4` cho `74`. Nối là viết hai số **liền nhau**. **Sửa:** `int(str(a) + str(b))` hoặc `a * 10 ** len(str(b)) + b`.

### Lỗi 3: nhân cố định `* 10` thay vì theo số chữ số của số cuối

````python
def wrong_times_ten(nums):
    lo, hi, total = 0, len(nums) - 1, 0
    while lo < hi:
        total += nums[lo] * 10 + nums[hi]         # SAI: chỉ đúng khi nums[hi] có đúng một chữ số
        lo += 1
        hi -= 1
    if lo == hi:
        total += nums[lo]
    return total

nums = [7, 52]
print("sai :", wrong_times_ten(nums))
print("đúng:", Solution().findTheArrayConcVal(nums))
assert wrong_times_ten(nums) != Solution().findTheArrayConcVal(nums)
````

Vì sao sai: nối `7` với `52` là `752 = 7 * 100 + 52`; nhân `10` cho `122`. Hệ số nhân phải là `10^k` với `k` là **số chữ số của số đứng sau** (`52` có 2 chữ số → `100`).

### Lỗi 4: nối phần tử giữa với chính nó (dùng `<=` thay `<`)

````python
def wrong_le(nums):
    lo, hi, total = 0, len(nums) - 1, 0
    while lo <= hi:                                # SAI: khi lo == hi lại nối số đó với chính nó
        total += int(str(nums[lo]) + str(nums[hi]))
        lo += 1
        hi -= 1
    return total

nums = [13]
print("sai :", wrong_le(nums))
print("đúng:", Solution().findTheArrayConcVal(nums))
assert wrong_le(nums) != Solution().findTheArrayConcVal(nums)
````

Vì sao sai: khi chỉ còn một số, đề cộng **chính giá trị của nó** (`13`), không nối nó với chính nó (`1313`). Dùng `while lo < hi` rồi xử lý `lo == hi` riêng.

## 14. Mẹo làm bài nhanh & nhận dạng pattern

- "Ghép phần tử đầu với phần tử cuối, rồi thu hẹp vào trong" → **hai con trỏ** `lo`, `hi`.
- Luôn nghĩ tới **mảng độ dài lẻ**: sau khi `lo == hi` còn một phần tử cần xử lý riêng.
- Nối số: `int(str(a) + str(b))` (dễ) hoặc `a * 10 ** len(str(b)) + b` (không chuỗi); nhớ nối khác cộng.
- Lấy hai đầu liên tục trên **danh sách**: `pop(0)` chậm (`O(n)`); dùng `deque` hoặc hai chỉ số.
- Không sửa mảng đầu vào nếu không cần; dùng chỉ số thay vì xoá phần tử.
- Kiểm tra các biên: `n = 1`, `n = 2`, `n = 3` và số có nhiều chữ số (`10^4`).

## 15. Biến thể của bài

1. Nối theo thứ tự **ngược** (cuối trước, đầu sau).
2. Trả về **danh sách** các giá trị nối của từng cặp thay vì tổng.
3. Số có thể bằng `0` (nối với số có chữ số `0` đứng đầu cần cẩn thận khi dùng `a * 10^k + b`): dùng chuỗi để chắc chắn.
4. Kết quả lấy **modulo** `10^9 + 7` khi `n`, giá trị rất lớn.
5. Ghép **ba** phần tử một lúc (đầu, giữa, cuối) hoặc ghép **hai đầu rồi hai đầu của phần còn lại** theo kiểu cây (tạo "concat tournament").

## 16. So sánh các cách làm

| Cách | Ý tưởng | Time | Space | Khi nào dùng |
|---|---|---|---|---|
| 1. `pop(0)` / `pop()` | mô phỏng nguyên văn trên list | `O(n^2)` | `O(n)` | hiểu đề, oracle |
| 2. Hai con trỏ + chuỗi | `lo`, `hi` + `int(str(a) + str(b))` | `O(n)` | `O(1)` | **chuẩn, nên nộp** |
| 3. Hai con trỏ + toán học | `a * 10^k + b` không chuỗi | `O(n)` | `O(1)` | không muốn đổi chuỗi |
| 4. `deque` | `popleft()` và `pop()` đều `O(1)` | `O(n)` | `O(n)` | giống đề, vẫn nhanh |

## 18. Gợi ý 3 mức cho bài tập

**Mức 1 (nhẹ):** lấy phần tử đầu với phần tử cuối nối lại; còn một số thì cộng nguyên giá trị.

**Mức 2 (vừa):** Bài 3: `p = 1; while p <= b: p *= 10`, rồi `a * p + b`. Bài 4: dùng `lo`, `hi`; khi `lo == hi` thêm `nums[lo]` vào danh sách. Bài 5: tính `total` rồi lấy `% mod` cuối cùng (hoặc lấy modulo từng bước).

**Mức 3 (gần đáp án):** Bài 1: `6` nối `4` = `64`; `3` nối `10` = `310` → `374`. Bài 2: `2` nối `9` = `29`; `11` nối `4` = `114`; còn `30` → `29 + 114 + 30 = 173`.

## 19. Đáp án tham khảo cho bài tập

````python
# Bài 1: 6||4 = 64 ; 3||10 = 310 -> 374
assert Solution().findTheArrayConcVal([6, 3, 10, 4]) == 374

# Bài 2: 2||9 = 29 ; 11||4 = 114 ; giữa 30 -> 173
assert Solution().findTheArrayConcVal([2, 11, 30, 4, 9]) == 29 + 114 + 30 == 173

# Bài 3
def concat(a, b):
    p = 1
    while p <= b:
        p *= 10
    return a * p + b
assert concat(15, 49) == 1549
assert concat(7, 52) == 752
assert concat(10000, 10000) == 1000010000
assert concat(1, 9) == 19

# Bài 4
def concat_pairs(nums):
    lo, hi, out = 0, len(nums) - 1, []
    while lo < hi:
        out.append(concat(nums[lo], nums[hi]))
        lo += 1
        hi -= 1
    if lo == hi:
        out.append(nums[lo])
    return out
assert concat_pairs([7, 52, 2, 4]) == [74, 522]
assert concat_pairs([5, 14, 13, 8, 12]) == [512, 148, 13]
assert sum(concat_pairs([5, 14, 13, 8, 12])) == Solution().findTheArrayConcVal([5, 14, 13, 8, 12])

# Bài 5
def conc_val_mod(nums, mod):
    total = 0
    for v in concat_pairs(nums):
        total = (total + v) % mod
    return total
MOD = 10**9 + 7
for nums in ([7, 52, 2, 4], [5, 14, 13, 8, 12], [10000] * 1000):
    assert conc_val_mod(nums, MOD) == Solution().findTheArrayConcVal(nums) % MOD
print("đáp án bài tập OK")
````
