## 1. Mục tiêu học

Sau notebook này bạn sẽ:

1. Hiểu đề: mảng `nums` là **"tốt" (good)** nếu nó là một **hoán vị** của mảng `base[n] = [1, 2, ..., n − 1, n, n]` (độ dài `n + 1`, số `n` xuất hiện **hai lần**) với một số `n` nào đó.
2. Hiểu **hoán vị (permutation)** theo nghĩa "cùng các phần tử với cùng số lần xuất hiện, thứ tự tuỳ ý" và biết kiểm tra bằng **sắp xếp** hoặc **đếm tần suất**.
3. Biết dùng `sorted`, `max`, `list.count`, `set`, `collections.Counter` và so sánh hai danh sách.
4. Đi từ **Brute Force thử mọi `n`** → **suy ra `n = max(nums)` rồi sắp xếp và so sánh** → **`Counter`** → **mảng đếm (không dùng băm)** → **`set` + `count`**.
5. Rút ra **các điều kiện cần và đủ**: `len = n + 1`, `max = n`, `n` xuất hiện đúng `2` lần, mỗi số `1..n−1` đúng `1` lần.
6. Tránh các lỗi: quên kiểm tra độ dài, chỉ kiểm tra "có `n` hai lần", chọn sai `n`, và nhầm `range(1, n)` với `range(1, n + 1)`.

## 2. Đề bài & ví dụ

Cho mảng số nguyên `nums`. Ta nói `nums` là **tốt (good)** nếu nó là một **hoán vị** của mảng `base[n]`, trong đó `base[n]` có độ dài `n + 1` và bằng:

`base[n] = [1, 2, ..., n − 1, n, n]`

tức là chứa mỗi số từ `1` tới `n − 1` **đúng một lần**, và số `n` **đúng hai lần**. Ví dụ `base[1] = [1, 1]`, `base[3] = [1, 2, 3, 3]`.

Trả về `true` nếu `nums` là tốt, ngược lại trả `false`.

| Ví dụ | nums | Output | Giải thích |
|---|---|---|---|
| 1 | `[2,1,3]` | `false` | Số lớn nhất là `3` nên chỉ có thể là `base[3] = [1,2,3,3]` (độ dài 4), nhưng `nums` có độ dài 3. |
| 2 | `[1,3,3,2]` | `true` | Số lớn nhất là `3`; `nums` là một hoán vị của `base[3] = [1,2,3,3]`. |
| 3 | `[1,1]` | `true` | Số lớn nhất là `1`; `nums` là hoán vị của `base[1] = [1,1]`. |
| 4 | `[3,4,4,1,2,1]` | `false` | Số lớn nhất là `4` nên chỉ có thể là `base[4] = [1,2,3,4,4]` (độ dài 5), nhưng `nums` có độ dài 6. |

Constraints:

- `1 <= nums.length <= 100`
- `1 <= nums[i] <= 200`

````python
def base(n):
    return list(range(1, n)) + [n, n]

print(base(1))   # [1, 1]
print(base(3))   # [1, 2, 3, 3]
print(base(4))   # [1, 2, 3, 4, 4]
print(sorted([1, 3, 3, 2]) == base(3))   # True
````

## 3. Kiến thức nền

### 3.1. Hoán vị (permutation) và đa tập (multiset)

Hai danh sách là **hoán vị của nhau** nếu có **cùng các phần tử với cùng số lần xuất hiện**, chỉ khác thứ tự. Ví dụ `[1, 3, 3, 2]` là hoán vị của `[1, 2, 3, 3]`, nhưng `[1, 3, 2, 2]` thì không (số lần của `2` và `3` khác nhau).

Cách đơn giản kiểm tra: **sắp xếp cả hai** rồi so sánh (`sorted(a) == sorted(b)`), vì sau khi sắp xếp, các hoán vị của nhau thành **cùng một danh sách**.

````python
a = [1, 3, 3, 2]
b = [1, 2, 3, 3]
c = [1, 3, 2, 2]
print(sorted(a) == sorted(b))   # True
print(sorted(a) == sorted(c))   # False
print(a == b)                    # False (so sánh trực tiếp phụ thuộc thứ tự)
````

### 3.2. `sorted`, `max`, `len` và nối danh sách bằng `+`

- `sorted(a)` trả về danh sách **mới** đã sắp xếp tăng dần (không sửa `a`).
- `max(a)` trả về phần tử lớn nhất; `len(a)` là độ dài.
- `a + b` nối hai danh sách; `list(range(1, n))` là `[1, ..., n − 1]`.

````python
a = [3, 1, 2]
print(sorted(a), a)                  # [1, 2, 3] [3, 1, 2]
print(max(a), len(a))                # 3 3
n = 4
print(list(range(1, n)) + [n, n])    # [1, 2, 3, 4, 4]
````

### 3.3. `list.count` và `collections.Counter`

- `a.count(x)`: số lần `x` xuất hiện trong `a` (duyệt cả danh sách: `O(n)`).
- `Counter(a)`: **từ điển đếm** tất cả phần tử một lần (`O(n)`), tra cứu `cnt[x]` nhanh, và trả `0` cho khoá không có (không gây `KeyError`).

````python
from collections import Counter
a = [1, 3, 3, 2]
print(a.count(3))             # 2
cnt = Counter(a)
print(cnt)                    # Counter({3: 2, 1: 1, 2: 1})
print(cnt[3], cnt[2], cnt[9]) # 2 1 0
````

### 3.4. Mảng đếm (counting array)

Khi giá trị nằm trong khoảng nhỏ `1..V`, ta dùng danh sách `cnt = [0] * (V + 1)` và `cnt[x] += 1`. Đây là cách đếm **không dùng băm**, rất nhanh và hay dùng khi biết trước miền giá trị.

````python
a = [1, 3, 3, 2]
cnt = [0] * (3 + 1)
for x in a:
    cnt[x] += 1
print(cnt)          # [0, 1, 1, 2]  (cnt[1]=1, cnt[2]=1, cnt[3]=2)
````

### 3.5. `set` và `all(...)`

- `set(a)` là tập các phần tử khác nhau; `len(set(a))` là số giá trị phân biệt.
- `all(điều_kiện for x in ...)` đúng khi **mọi** phần tử thoả điều kiện (`True` nếu rỗng).

````python
a = [1, 3, 3, 2]
print(set(a), len(set(a)))                              # {1, 2, 3} 3
print(all(a.count(v) == 1 for v in range(1, 3)))        # True: 1 và 2 mỗi số đúng 1 lần
print(all(x < 0 for x in []))                            # True (rỗng)
````

## 4. Ý tưởng nền tảng / cách nghĩ trực giác

Nhìn kỹ `base[n] = [1, 2, ..., n − 1, n, n]`:

- **Số lớn nhất** là `n`.
- **Độ dài** là `n + 1`.
- Số `n` xuất hiện **hai lần**, các số nhỏ hơn mỗi số **một lần**.

Vậy nếu `nums` là tốt thì `n` **bị xác định ngay** bởi `nums`: `n = max(nums)` (và cũng `n = len(nums) − 1`). Không cần thử nhiều `n`.

Tính tay ví dụ `[3, 4, 4, 1, 2, 1]`: số lớn nhất `4` ⇒ `n = 4` ⇒ cần độ dài `5`, mà `nums` dài `6` ⇒ **không tốt**. Chưa cần nhìn các số khác.

Tính tay `[1, 3, 3, 2]`: `n = 3`, độ dài `4 = n + 1` ✓; đếm: `1` một lần, `2` một lần, `3` hai lần ✓ ⇒ **tốt**.

````python
for nums in ([2, 1, 3], [1, 3, 3, 2], [1, 1], [3, 4, 4, 1, 2, 1]):
    n = max(nums)
    print(nums, "n =", n, "| cần độ dài", n + 1, "| có", len(nums), "| khớp độ dài?", len(nums) == n + 1)
````

## 5. Cách 1 — Brute Force / baseline: thử mọi `n` rồi so sánh với `base[n]`

**Ý tưởng:** chưa dùng nhận xét nào. Với mỗi `n` có thể (từ `1` tới một cận đủ lớn như `len(nums)`), dựng `base[n] = list(range(1, n)) + [n, n]`; nếu `sorted(nums) == base[n]` thì `nums` là hoán vị của `base[n]`, trả `True`. Hết các `n` mà không khớp thì `False`.

Độ dài `base[n]` là `n + 1` nên chỉ cần thử `n` đến `len(nums)` (nếu `n + 1 > len(nums)` thì chắc chắn không khớp độ dài).

````python
from typing import List

class SolutionTryAll:
    def isGood(self, nums: List[int]) -> bool:
        s = sorted(nums)
        for n in range(1, len(nums) + 1):
            base = list(range(1, n)) + [n, n]
            if s == base:
                return True
        return False
````

````python
sol = SolutionTryAll()
print(sol.isGood([2, 1, 3]))              # False
print(sol.isGood([1, 3, 3, 2]))           # True
print(sol.isGood([1, 1]))                 # True
print(sol.isGood([3, 4, 4, 1, 2, 1]))     # False
````

**Độ phức tạp:** sắp xếp `O(m log m)` với `m = len(nums)`; vòng thử `n` có `m` lần, mỗi lần dựng `base` `O(n)` và so sánh `O(n)` ⇒ Time `O(m²)`, Space `O(m)`.

**Nút thắt:** ta thử rất nhiều `n` vô ích, trong khi `n` thực ra bị **xác định** bởi `nums` (`n = max(nums)`).

## 6. Ý tưởng thuật toán / bước đột phá

**Đột phá 1 — `n` xác định duy nhất.** Nếu `nums` tốt thì `max(nums) = n`. Vậy chỉ cần thử đúng **một** `n = max(nums)`. Thậm chí độ dài cũng cho `n = len(nums) − 1`; hai giá trị này phải bằng nhau.

**Đột phá 2 — các điều kiện cần và đủ.** `nums` tốt **khi và chỉ khi** với `n = max(nums)`:

1. `len(nums) == n + 1`;
2. số `n` xuất hiện **đúng 2 lần**;
3. mỗi số `1, 2, ..., n − 1` xuất hiện **đúng 1 lần**.

(Từ 1, 2, 3 suy ra không còn chỗ cho phần tử nào khác: `2 + (n − 1) = n + 1`.)

**Đột phá 3 — chọn công cụ đếm.** Có ba cách kiểm tra 2 và 3:

- sắp xếp rồi so với `base[n]`;
- `Counter` (băm);
- mảng đếm (không băm) với miền `1..n`.

**Đột phá 4 — dừng sớm.** Kiểm tra `max` và độ dài trước, vì chúng rất rẻ và loại phần lớn trường hợp sai.

````python
nums = [1, 3, 3, 2]
n = max(nums)
cnt = [0] * (n + 1)
for x in nums:
    cnt[x] += 1
print("n =", n, "cnt =", cnt)
print("n xuất hiện đúng 2 lần?", cnt[n] == 2)
print("1..n-1 mỗi số đúng 1 lần?", all(cnt[v] == 1 for v in range(1, n)))
````

## 7. Các cách tiếp theo

### Cách 2 — Suy ra `n = max(nums)`, sắp xếp và so sánh với `base[n]`

**Từng đoạn:**

- `n = max(nums)`.
- `base = list(range(1, n)) + [n, n]`.
- `sorted(nums) == base` (so sánh danh sách đã sắp xếp; hai danh sách bằng nhau cả độ dài lẫn từng phần tử, nên điều kiện độ dài được kiểm tra luôn).

````python
class SolutionSort:
    def isGood(self, nums: List[int]) -> bool:
        n = max(nums)
        base = list(range(1, n)) + [n, n]
        return sorted(nums) == base
````

````python
sol = SolutionSort()
print(sol.isGood([2, 1, 3]))              # False
print(sol.isGood([1, 3, 3, 2]))           # True
print(sol.isGood([1, 1]))                 # True
print(sol.isGood([3, 4, 4, 1, 2, 1]))     # False
````

**Độ phức tạp:** Time `O(m log m)` (sắp xếp) — dựng `base` `O(n)`, `n <= m`; Space `O(m)`.

### Cách 3 — `Counter` và kiểm tra các điều kiện (code nộp)

**Từng đoạn:**

- `n = len(nums) − 1`; nếu `max(nums) != n` thì `False`. (Điều kiện này gộp "độ dài `= n + 1`" và "`n` là phần tử lớn nhất".)
- `cnt = Counter(nums)`.
- `cnt[n] == 2` và mọi `v` trong `1..n − 1` có `cnt[v] == 1`.

Vì `len(nums) = n + 1` và số phần tử đã tính là `2 + (n − 1) = n + 1`, không còn chỗ cho phần tử nào khác (như `0` hoặc giá trị `> n`); đồng thời `max(nums) == n` đảm bảo không có giá trị lớn hơn `n`.

````python
from collections import Counter

class SolutionCounter:
    def isGood(self, nums: List[int]) -> bool:
        n = len(nums) - 1
        if max(nums) != n:
            return False
        cnt = Counter(nums)
        return cnt[n] == 2 and all(cnt[v] == 1 for v in range(1, n))
````

````python
sol = SolutionCounter()
print(sol.isGood([2, 1, 3]))              # False
print(sol.isGood([1, 3, 3, 2]))           # True
print(sol.isGood([1, 1]))                 # True
print(sol.isGood([3, 4, 4, 1, 2, 1]))     # False
````

**Độ phức tạp:** Time `O(m)`, Space `O(m)`.

### Cách 4 — Mảng đếm (counting array), không băm

**Từng đoạn:**

- Kiểm tra `max(nums) == len(nums) − 1 = n`.
- `cnt = [0] * (n + 1)`; duyệt `nums` và tăng `cnt[x]`. Giá trị `x` nằm trong `1..n` nhờ `max = n` và `nums[i] >= 1` nên chỉ số hợp lệ.
- Nếu một số vượt quá số lần cho phép thì trả `False` ngay: `cnt[x] > 1` với `x < n`, hoặc `cnt[n] > 2`.
- Cuối cùng chỉ cần kiểm tra không thiếu: `cnt[n] == 2` và mọi `cnt[v] == 1`.

````python
class SolutionArray:
    def isGood(self, nums: List[int]) -> bool:
        n = len(nums) - 1
        if max(nums) != n:
            return False
        cnt = [0] * (n + 1)
        for x in nums:
            cnt[x] += 1
            limit = 2 if x == n else 1
            if cnt[x] > limit:
                return False          # xuất hiện quá nhiều lần
        return cnt[n] == 2 and all(cnt[v] == 1 for v in range(1, n))
````

````python
sol = SolutionArray()
print(sol.isGood([2, 1, 3]))              # False
print(sol.isGood([1, 3, 3, 2]))           # True
print(sol.isGood([1, 1]))                 # True
print(sol.isGood([3, 4, 4, 1, 2, 1]))     # False
````

**Độ phức tạp:** Time `O(m)`, Space `O(n)` (mảng đếm độ dài `n + 1`). Không băm nên hằng số nhỏ.

**Lưu ý chỉ số:** `cnt[x]` với `x <= max(nums) = n` an toàn vì ta đã kiểm `max(nums) == n` **trước** khi tạo/truy cập mảng đếm (nếu quên kiểm, `x > n` gây `IndexError`).

### Cách 5 — `set` + `count`

**Từng đoạn (dựa trên đếm số phần tử phân biệt):**

- `n = len(nums) − 1`; kiểm `max(nums) == n`.
- `set(nums)` phải là `{1, 2, ..., n}` (đủ `n` giá trị phân biệt, không thiếu). Khi đó `len(nums) = n + 1` nghĩa là còn đúng **một phần tử dư** so với `n` giá trị phân biệt, tức một giá trị lặp thêm đúng một lần.
- Phần tử lặp đó phải là `n`: `nums.count(n) == 2`.

````python
class SolutionSetCount:
    def isGood(self, nums: List[int]) -> bool:
        n = len(nums) - 1
        if max(nums) != n:
            return False
        return set(nums) == set(range(1, n + 1)) and nums.count(n) == 2
````

````python
sol = SolutionSetCount()
print(sol.isGood([2, 1, 3]))              # False
print(sol.isGood([1, 3, 3, 2]))           # True
print(sol.isGood([1, 1]))                 # True
print(sol.isGood([3, 4, 4, 1, 2, 1]))     # False
````

**Độ phức tạp:** Time `O(m)` (`set`, `count`), Space `O(m)`. Ngắn gọn nhất, nhưng cần nhớ lý do `count(n) == 2` đủ (xem mục 10).

## 8. TRACE từng bước

Trace Cách 3 trên `nums = [1, 3, 3, 2]`:

| bước | giá trị | kết quả |
|---|---|---|
| `n = len − 1` | 3 | |
| `max(nums)` | 3 | khớp `n` ✓ |
| `cnt` | `{1:1, 3:2, 2:1}` | |
| `cnt[3]` | 2 | đúng 2 lần ✓ |
| `cnt[1]`, `cnt[2]` | 1, 1 | mỗi số 1 lần ✓ |
| kết quả | | `True` |

Trace `[3, 4, 4, 1, 2, 1]`: `n = 6 − 1 = 5`, `max = 4 ≠ 5` ⇒ `False` ngay.

Trace `[1, 1, 3, 3]` (không tốt): `n = 3`, `max = 3` ✓; `cnt[3] = 2` ✓; nhưng `cnt[1] = 2 ≠ 1` ⇒ `False`. (Số `2` bị thiếu.)

````python
def trace_counter(nums):
    n = len(nums) - 1
    print(f"nums={nums}: n = len - 1 = {n}, max = {max(nums)}")
    if max(nums) != n:
        print("  max != n  -> False")
        return False
    cnt = Counter(nums)
    print("  cnt =", dict(sorted(cnt.items())))
    print(f"  cnt[{n}] == 2 ? {cnt[n] == 2}")
    for v in range(1, n):
        print(f"  cnt[{v}] == 1 ? {cnt[v] == 1}")
    ans = cnt[n] == 2 and all(cnt[v] == 1 for v in range(1, n))
    print("  kết quả:", ans)
    return ans

trace_counter([1, 3, 3, 2])
trace_counter([3, 4, 4, 1, 2, 1])
trace_counter([1, 1, 3, 3])
````

````python
def trace_array(nums):
    n = len(nums) - 1
    if max(nums) != n:
        print(nums, "-> max != n, False")
        return False
    cnt = [0] * (n + 1)
    for i, x in enumerate(nums):
        cnt[x] += 1
        limit = 2 if x == n else 1
        print(f"i={i}, x={x}, cnt={cnt}, giới hạn của {x} là {limit}")
        if cnt[x] > limit:
            print("  vượt giới hạn -> False")
            return False
    ans = cnt[n] == 2 and all(cnt[v] == 1 for v in range(1, n))
    print("kết quả:", ans)
    return ans

trace_array([1, 3, 3, 2])
trace_array([1, 2, 2, 3])
````

## 9. Code nộp LeetCode

Dùng Cách 3 (`Counter`): rõ ràng, `O(m)`.

````python
from collections import Counter
from typing import List

class Solution:
    def isGood(self, nums: List[int]) -> bool:
        n = len(nums) - 1                 # base[n] có độ dài n + 1
        if max(nums) != n:                # n phải là phần tử lớn nhất
            return False
        cnt = Counter(nums)
        # n xuất hiện đúng 2 lần, mỗi số 1..n-1 đúng 1 lần
        return cnt[n] == 2 and all(cnt[v] == 1 for v in range(1, n))
````

````python
print(Solution().isGood([2, 1, 3]))              # False
print(Solution().isGood([1, 3, 3, 2]))           # True
print(Solution().isGood([1, 1]))                 # True
print(Solution().isGood([3, 4, 4, 1, 2, 1]))     # False
````

## 10. Vì sao thuật toán đúng

**Chiều thuận (tốt ⇒ qua kiểm tra).** Nếu `nums` là hoán vị của `base[n]` thì độ dài là `n + 1` (nên `n = len − 1`), phần tử lớn nhất là `n`, số `n` xuất hiện đúng 2 lần và mỗi số `1..n − 1` đúng 1 lần. Mọi điều kiện của thuật toán thoả.

**Chiều ngược (qua kiểm tra ⇒ tốt).** Giả sử `n = len − 1`, `max(nums) = n`, `cnt[n] = 2`, `cnt[v] = 1` với mọi `v ∈ [1, n − 1]`. Số phần tử đã giải thích là `2 + (n − 1) = n + 1 = len(nums)`; vậy **không còn phần tử nào khác** (không có `0`, không có giá trị `> n`; `max = n` còn bảo vệ điều đó, và đề bảo đảm `nums[i] >= 1`). Do đó `nums` có đúng nội dung của `base[n]`, tức là hoán vị của `base[n]`. ✓

**Với `n = 1`:** `range(1, 1)` rỗng nên chỉ cần `cnt[1] == 2`, `max = 1`, `len = 2`: đúng là `[1, 1]` ✓.

**Về Cách 5.** Với `max = n = len − 1`, `set(nums) = {1..n}` nghĩa là có đúng `n` giá trị phân biệt, mỗi giá trị xuất hiện ít nhất một lần. Tổng số phần tử `n + 1` nên chỉ còn **một** lần lặp thừa; điều kiện `count(n) == 2` bảo đảm phần thừa đó thuộc về `n`. Hệ quả: các giá trị khác xuất hiện đúng một lần. ✓

**Về Cách 1 và 2.** Dựng `base[n]` và so sánh sau khi sắp xếp là định nghĩa của "hoán vị của `base[n]`", nên đúng hiển nhiên; Cách 2 chỉ bỏ các `n` không thể (vì `max(nums) ≠ n`).

## 11. Phân tích Time / Space Complexity

`m = len(nums)`; `n = m − 1`.

| Cách | Time | Space | Ghi chú |
|---|---|---|---|
| 1. Thử mọi `n` | `O(m²)` | `O(m)` | baseline chậm nhất |
| 2. `max` + sắp xếp + so sánh | `O(m log m)` | `O(m)` | rất dễ đúng |
| 3. `Counter` | `O(m)` | `O(m)` | **code nộp** |
| 4. Mảng đếm | `O(m)` | `O(m)` | không băm, dừng sớm |
| 5. `set` + `count` | `O(m)` | `O(m)` | ngắn nhất |

### 12.3. Đối chiếu ngẫu nhiên (hợp lệ theo đề)

Ta sinh mảng tốt (xáo trộn `base[n]`) và mảng "gần tốt" (đổi một phần tử, đổi độ dài, thêm trùng...) với `1 <= len <= 100`, `1 <= nums[i] <= 200`.

````python
import random
random.seed(2784)

def gen_good():
    n = random.randint(1, 99)
    nums = list(range(1, n)) + [n, n]
    random.shuffle(nums)
    return nums

def gen_near():
    nums = gen_good()
    mode = random.choice(["change", "extend", "drop", "swap_dup", "random"])
    if mode == "change":
        nums[random.randrange(len(nums))] = random.randint(1, 200)
    elif mode == "extend" and len(nums) < 100:
        nums.append(random.randint(1, 100))
    elif mode == "drop" and len(nums) > 1:
        nums.pop(random.randrange(len(nums)))
    elif mode == "swap_dup":
        i = random.randrange(len(nums))
        nums[i] = random.choice(nums)
    else:
        nums = [random.randint(1, 8) for _ in range(random.randint(1, 12))]
    return nums

n_good = n_bad = 0
for _ in range(800):
    for nums in (gen_good(), gen_near()):
        assert 1 <= len(nums) <= 100 and all(1 <= v <= 200 for v in nums)
        exp = SolutionTryAll().isGood(nums[:])
        n_good += exp
        n_bad += (not exp)
        for s in sols:
            assert s.isGood(nums[:]) == exp, (type(s).__name__, nums)
        snap = nums[:]
        Solution().isGood(nums)
        assert nums == snap            # không sửa dữ liệu vào
print("1600 ca ngẫu nhiên; True:", n_good, ", False:", n_bad, "- 6 cách khớp nhau")
````

## 13. Các lỗi người mới hay gặp

### Lỗi 1: Quên kiểm tra độ dài (chỉ kiểm "có `n` hai lần")

````python
def sai_quen_do_dai(nums: List[int]) -> bool:
    n = max(nums)
    return nums.count(n) == 2            # sai: quá lỏng

print("sai :", sai_quen_do_dai([1, 1, 3, 3]))           # True
print("đúng:", Solution().isGood([1, 1, 3, 3]))         # False
````

**Vì sao sai:** `[1, 1, 3, 3]` có `3` hai lần nhưng **thiếu `2`** và thừa `1`. Phải kiểm cả độ dài (`n + 1`) **và** các số nhỏ xuất hiện đúng một lần.

### Lỗi 2: Chọn `n = len(nums)` thay vì `len(nums) − 1`

````python
def sai_chon_n(nums: List[int]) -> bool:
    n = len(nums)                        # sai: base[n] có độ dài n + 1
    return sorted(nums) == list(range(1, n)) + [n, n]

print("sai :", sai_chon_n([1, 3, 3, 2]), sai_chon_n([1, 1]))      # False False
print("đúng:", Solution().isGood([1, 3, 3, 2]), Solution().isGood([1, 1]))     # True True
````

**Vì sao sai:** `base[n]` có `n + 1` phần tử nên `n = len(nums) − 1`. Chọn `n = len(nums)` dựng sai `base` ⇒ luôn khớp sai.

### Lỗi 3: Dựng `base` sai bằng `range(1, n + 1)`

````python
def sai_dung_base(n):
    return list(range(1, n + 1)) + [n, n]      # sai: n xuất hiện 3 lần, độ dài n + 2

print(sai_dung_base(3))       # [1, 2, 3, 3, 3]
print(list(range(1, 3)) + [3, 3])   # đúng: [1, 2, 3, 3]
````

**Vì sao sai:** `range(1, n + 1)` đã chứa `n`; cộng thêm `[n, n]` làm `n` có **ba** lần. Chỉ cần `range(1, n)` (từ `1` tới `n − 1`) rồi thêm `[n, n]`.

### Lỗi 4: Cho rằng mọi số phải khác nhau

````python
def sai_phai_khac_nhau(nums: List[int]) -> bool:
    return len(set(nums)) == len(nums) and sorted(nums) == list(range(1, len(nums) + 1))

print("sai :", sai_phai_khac_nhau([1, 3, 3, 2]))          # False
print("đúng:", Solution().isGood([1, 3, 3, 2]))           # True
````

**Vì sao sai:** mảng tốt **luôn có** một số lặp (số `n` xuất hiện hai lần). Kiểm tra "là hoán vị của `1..len`" cho bài khác (mảng không lặp).

### Lỗi 5: `IndexError` với mảng đếm khi chưa kiểm `max`

````python
def sai_mang_dem(nums: List[int]) -> bool:
    n = len(nums) - 1
    cnt = [0] * (n + 1)                 # quên kiểm max(nums) == n trước
    for x in nums:
        cnt[x] += 1                      # x > n gây IndexError
    return cnt[n] == 2 and all(cnt[v] == 1 for v in range(1, n))

try:
    print(sai_mang_dem([5, 1, 1]))
except IndexError as e:
    print("lỗi:", e)
print("đúng:", Solution().isGood([5, 1, 1]))       # False
````

**Vì sao sai:** giá trị `5` vượt quá chỉ số tối đa `n = 2` của mảng đếm. Kiểm `max(nums) == n` **trước** khi dùng mảng đếm (hoặc dùng `Counter`).

## 14. Mẹo làm bài nhanh & nhận dạng pattern

- Đề "mảng có phải **hoán vị** của một mảng chuẩn" → nghĩ tới **sắp xếp so sánh** hoặc **đếm tần suất** (`Counter` / mảng đếm).
- Khi cấu trúc chuẩn có **tham số ẩn** (`n`), hãy tìm đại lượng xác định nó (ở đây `max` và `len`).
- Viết điều kiện **cần và đủ** dưới dạng danh sách ngắn, kiểm từng điều kiện rẻ trước.
- Mảng đếm `[0] * (V + 1)` rất hợp khi giá trị nhỏ; đừng quên kiểm tra chỉ số hợp lệ.
- Test kinh điển: độ dài 1, độ dài 2, `[1, 1]`, mảng không lặp, `n` ba lần, số thiếu được thay bằng số lặp.
- Kiểm chứng bằng **vét cạn mọi mảng nhỏ** là cách tốt nhất để tin vào điều kiện cần và đủ.

## 15. Biến thể của bài

1. **Mảng là hoán vị của `1..n` (không lặp)** (LeetCode kiểu "Check if permutation"): `sorted(nums) == list(range(1, len + 1))`.
2. **Tìm số lặp** (LeetCode 287 / 442) hoặc **số thiếu** (268 / 448): dùng cùng kỹ thuật đếm.
3. **`base[n]` với số `k` xuất hiện hai lần** (không nhất thiết `n`): điều kiện đếm đổi tương ứng.
4. **Đếm số cách** sắp xếp một mảng tốt (hoán vị lặp): `(n + 1)! / 2!`.
5. **Hai mảng là hoán vị của nhau** (LeetCode 242 anagram, 1941...): đếm hoặc sắp xếp.
6. **Cho `nums` rất lớn** (`10^6` phần tử): dùng mảng đếm hoặc `Counter`, tránh sắp xếp `O(m log m)` nếu cần tối ưu.

## 16. So sánh các cách làm

| Cách | Ý tưởng | Time | Space | Khi nào dùng |
|---|---|---|---|---|
| 1. Thử mọi `n` | dựng `base[n]` rồi so sánh sắp xếp | `O(m²)` | `O(m)` | baseline, làm oracle |
| 2. `max` + sắp xếp | một `n` duy nhất | `O(m log m)` | `O(m)` | rất dễ đọc |
| 3. `Counter` | đếm tần suất, kiểm điều kiện | `O(m)` | `O(m)` | **code nộp** |
| 4. Mảng đếm | đếm không băm, dừng sớm | `O(m)` | `O(m)` | giá trị nhỏ, cần nhanh |
| 5. `set` + `count` | số phân biệt và `count(n)` | `O(m)` | `O(m)` | ngắn gọn, cần hiểu bất biến |
