## 1. Mục tiêu học

Sau notebook này bạn sẽ:

1. Hiểu đề: tìm **chỉ số `i` nhỏ nhất** sao cho **tổng các chữ số** của `nums[i]` đúng bằng `i`; không có thì trả `-1`.
2. Biết tính tổng chữ số của một số nguyên bằng vòng lặp `% 10` / `// 10`, bằng chuyển sang chuỗi, và bằng bảng tính sẵn (quy hoạch động nhỏ).
3. Biết dùng `enumerate`, **thoát sớm** (early return) khi tìm đúng chỉ số nhỏ nhất, và `next(generator, default)`.
4. Viết 4 cách: vòng lặp chia 10, chuỗi + `sum(map(int, ...))`, bảng tổng chữ số dựng sẵn, và cách **chặn cận**: tổng chữ số tối đa `27` nên chỉ cần xét `i <= 27`.
5. Tránh các lỗi: so `nums[i]` với `i` thay vì **tổng chữ số**, lấy chỉ số **lớn nhất**, trả `0` thay `-1`, và tính tổng chữ số sai (chỉ lấy chữ số cuối).

## 2. Đề bài & ví dụ

Cho mảng số nguyên `nums`. Trả về **chỉ số nhỏ nhất** `i` (đánh số từ `0`) sao cho **tổng các chữ số** của `nums[i]` bằng `i`. Nếu không có chỉ số nào thỏa thì trả về `-1`.

| Ví dụ | nums | Output | Giải thích |
|---|---|---|---|
| 1 | `[1,3,2]` | `2` | tại `i = 2`, `nums[2] = 2` có tổng chữ số `2`, bằng `i` |
| 2 | `[1,10,11]` | `1` | tại `i = 1`, `nums[1] = 10` có tổng chữ số `1 + 0 = 1`; tại `i = 2` thì `11` cũng thỏa (`1 + 1 = 2`) nhưng `1` nhỏ hơn |
| 3 | `[1,2,3]` | `-1` | không chỉ số nào thỏa |

Constraints: `1 <= nums.length <= 100`; `0 <= nums[i] <= 1000`.

Chú ý: chỉ số bắt đầu từ `0`; tổng chữ số của `0` là `0`, nên `nums[0] = 0` thỏa ngay ở `i = 0`.

````python
def digit_sum(x):
    total = 0
    while x > 0:
        total += x % 10
        x //= 10
    return total

for nums, expected in [([1, 3, 2], 2), ([1, 10, 11], 1), ([1, 2, 3], -1)]:
    rows = [(i, v, digit_sum(v)) for i, v in enumerate(nums)]
    ans = next((i for i, v, ds in rows if ds == i), -1)
    print(f"nums={nums}: (i, giá trị, tổng chữ số) = {rows} -> {ans} (mong đợi {expected})")
````

## 3. Kiến thức nền

### 3.1. Tổng các chữ số bằng `% 10` và `// 10`

Với số nguyên không âm `x`: `x % 10` là chữ số **cuối**, `x // 10` là `x` sau khi **bỏ** chữ số cuối. Lặp lại đến khi `x == 0` thì cộng hết các chữ số. Ví dụ `x = 1234`: `4 + 3 + 2 + 1 = 10`. Với `x = 0` vòng lặp không chạy nên tổng là `0` (đúng).

````python
x = 1234
total = 0
while x > 0:
    print("chữ số cuối:", x % 10, "| phần còn lại:", x // 10)
    total += x % 10
    x //= 10
print("tổng =", total)
````

### 3.2. Tổng chữ số qua chuỗi

`str(x)` đổi số thành chuỗi các chữ số; `map(int, str(x))` đổi từng ký tự về số; `sum(...)` cộng lại. Ngắn gọn nhưng chậm hơn phép chia một chút (không đáng kể ở đây).

````python
x = 1234
print(str(x), list(map(int, str(x))), sum(map(int, str(x))))
print(sum(map(int, str(0))))             # '0' -> [0] -> 0
````

### 3.3. `enumerate` và thoát sớm

`for i, v in enumerate(nums)` cho cả chỉ số lẫn giá trị. Vì ta muốn chỉ số **nhỏ nhất**, duyệt từ `0` tăng dần và `return i` ngay khi gặp chỉ số đầu tiên thỏa — các chỉ số sau không thể nhỏ hơn.

````python
def first_even_index(nums):
    for i, v in enumerate(nums):
        if v % 2 == 0:
            return i              # thoát sớm: chỉ số đầu tiên chính là nhỏ nhất
    return -1

print(first_even_index([3, 5, 8, 6]), first_even_index([1, 3]))
````

### 3.4. `next(generator, default)`

`next(g, d)` lấy phần tử **đầu tiên** của generator `g`, hoặc `d` nếu `g` rỗng. Cú pháp gọn cho "tìm phần tử đầu tiên thỏa điều kiện hoặc trả mặc định".

````python
nums = [1, 10, 11]
print(next((i for i, v in enumerate(nums) if sum(map(int, str(v))) == i), -1))
print(next((i for i in range(5) if i > 10), -1))
````

### 3.5. Bảng tính sẵn bằng quy hoạch động nhỏ

Vì `nums[i] <= 1000`, ta có thể tính trước `ds[x]` = tổng chữ số của `x` cho `x = 0..1000` bằng công thức `ds[x] = ds[x // 10] + x % 10` (tổng chữ số của `x` = tổng chữ số của phần còn lại cộng chữ số cuối). Sau đó mỗi lần xét chỉ tra bảng `O(1)`.

````python
ds = [0] * 1001
for x in range(1, 1001):
    ds[x] = ds[x // 10] + x % 10
print(ds[0], ds[10], ds[999], ds[1000])
print(max(ds))                       # tổng chữ số lớn nhất khi x <= 1000
````

## 4. Ý tưởng nền tảng / cách nghĩ trực giác

Cứ làm đúng như đề với `nums = [1, 10, 11]`:

```
i = 0: nums[0] = 1  -> tổng chữ số 1  != 0
i = 1: nums[1] = 10 -> tổng chữ số 1  == 1  => dừng, đáp án 1
```

Ta không cần xét `i = 2` vì đã tìm được `i = 1` nhỏ hơn. Với `nums = [1, 2, 3]`: tổng chữ số là `1, 2, 3` còn chỉ số là `0, 1, 2` — **không** khớp ở đâu -> `-1`.

Nhận xét thêm: mỗi giá trị `<= 1000` có tối đa **4 chữ số**, nên tổng chữ số rất nhỏ (tối đa `9 + 9 + 9 = 27`, đạt ở `999`). Do đó chỉ số `i` mà tổng chữ số có thể bằng chỉ nằm trong `0..27`. Đó là bước tối ưu ở Cách 4.

## 5. Cách 1 — Brute Force / baseline

Với mỗi chỉ số `i` từ `0`, tính tổng chữ số của `nums[i]` bằng vòng lặp `% 10` / `// 10`; nếu bằng `i` thì trả `i` ngay; hết mảng trả `-1`.

````python
class SolutionLoop:
    def smallestIndex(self, nums):
        for i in range(len(nums)):
            x = nums[i]
            total = 0
            while x > 0:
                total += x % 10
                x //= 10
            if total == i:
                return i
        return -1

print(SolutionLoop().smallestIndex([1, 3, 2]))
print(SolutionLoop().smallestIndex([1, 10, 11]))
print(SolutionLoop().smallestIndex([1, 2, 3]))
````

**Độ phức tạp:** gọi `n = len(nums)`. Mỗi số có tối đa 4 chữ số nên tính tổng chữ số `O(1)` (`O(log V)` với `V` là giá trị lớn nhất). Tổng `O(n)` thời gian, `O(1)` bộ nhớ.

## 6. Ý tưởng thuật toán / bước đột phá

Bài đã `O(n)` — ta không thể nhanh hơn bậc, nhưng có thể làm gọn và nhanh hơn một hằng số:

- **Thoát sớm:** duyệt từ trái sang phải và trả ngay lần đầu thỏa (chỉ số nhỏ nhất).
- **Bảng tính sẵn:** `ds[0..1000]` bằng quy hoạch động, rồi tra `O(1)`.
- **Chặn cận:** `ds[x] <= 27` với `x <= 1000`. Nếu `i > 27` thì `ds[nums[i]] == i` là **không thể**. Vậy chỉ cần xét `i` trong `0..min(n, 28) - 1`.

**Quy tắc "nếu… thì…":**

- Nếu `digit_sum(nums[i]) == i` thì `i` là ứng viên; duyệt tăng dần nên ứng viên đầu tiên là nhỏ nhất.
- Nếu `i >= 28` thì dừng, vì không còn khả năng.
- Nếu không thấy ứng viên nào thì `-1` (chứ không phải `0`, vì `0` là chỉ số hợp lệ!).

````python
ds = [0] * 1001
for x in range(1, 1001):
    ds[x] = ds[x // 10] + x % 10
print("tổng chữ số lớn nhất:", max(ds), "đạt tại", [x for x in range(1001) if ds[x] == max(ds)])
````

## 7. Các cách tiếp theo

Ta có 4 cách: (1) vòng lặp `% 10`; (2) chuỗi + `sum(map(int, ...))`; (3) bảng `ds` dựng sẵn; (4) chặn cận `i <= 27` với `next`.

### Cách 2 — Chuỗi và `sum(map(int, str(v)))`

Gọn nhất: `sum(map(int, str(v)))` là tổng chữ số. Dùng `enumerate` để lấy `(i, v)` và `return i` khi thỏa.

````python
class SolutionStr:
    def smallestIndex(self, nums):
        for i, v in enumerate(nums):
            if sum(map(int, str(v))) == i:
                return i
        return -1

print(SolutionStr().smallestIndex([1, 3, 2]))
print(SolutionStr().smallestIndex([1, 10, 11]))
print(SolutionStr().smallestIndex([1, 2, 3]))
````

Độ phức tạp Cách 2: `O(n)` thời gian (mỗi số tối đa 4 chữ số), `O(1)` bộ nhớ thêm.

### Cách 3 — Bảng tổng chữ số dựng sẵn

Dựng `ds` một lần với `ds[x] = ds[x // 10] + x % 10`; mỗi lần xét chỉ tra `ds[nums[i]]`. Lợi ích rõ khi có **nhiều lần gọi** hoặc nhiều mảng (bảng dùng lại), hoặc khi giá trị lớn hơn một chút.

````python
class SolutionTable:
    def smallestIndex(self, nums):
        ds = [0] * 1001
        for x in range(1, 1001):
            ds[x] = ds[x // 10] + x % 10
        for i, v in enumerate(nums):
            if ds[v] == i:
                return i
        return -1

print(SolutionTable().smallestIndex([1, 3, 2]))
print(SolutionTable().smallestIndex([1, 10, 11]))
print(SolutionTable().smallestIndex([1, 2, 3]))
````

Độ phức tạp Cách 3: dựng bảng `O(V)` với `V = 1000` (hằng số), tra mỗi phần tử `O(1)` -> `O(V + n)` thời gian, `O(V)` bộ nhớ.

### Cách 4 — Chặn cận `i <= 27` với `next`

Vì `digit_sum(v) <= 27` khi `v <= 1000`, chỉ các chỉ số `0..27` có thể thỏa. Ta chỉ cần duyệt `min(n, 28)` phần tử đầu, dùng `next(generator, -1)` để trả chỉ số đầu tiên thỏa hoặc `-1`.

````python
class SolutionBound:
    def smallestIndex(self, nums):
        limit = min(len(nums), 28)                 # i = 0..27 (tổng chữ số tối đa 27)
        return next((i for i in range(limit) if sum(map(int, str(nums[i]))) == i), -1)

print(SolutionBound().smallestIndex([1, 3, 2]))
print(SolutionBound().smallestIndex([1, 10, 11]))
print(SolutionBound().smallestIndex([1, 2, 3]))
````

Độ phức tạp Cách 4: duyệt tối đa `28` phần tử -> `O(min(n, 28))` = `O(1)` thời gian với `n <= 100`, `O(1)` bộ nhớ. Đây chỉ là tối ưu hằng số, không đổi bậc của lời giải tổng quát `O(n)`.

## 8. TRACE từng bước

Trace Cách 1 với `nums = [1, 10, 11]`:

| bước | `i` | `nums[i]` | các chữ số | tổng | `tổng == i`? | hành động |
|---|---|---|---|---|---|---|
| 1 | 0 | 1 | `1` | 1 | `1 == 0` sai | tiếp tục |
| 2 | 1 | 10 | `0, 1` | 1 | `1 == 1` **đúng** | trả `1` |

Không cần xét `i = 2` (giá trị `11`, tổng `2`, cũng thỏa nhưng lớn hơn).

````python
def trace_smallest_index(nums):
    for i in range(len(nums)):
        x = nums[i]
        digits = []
        total = 0
        while x > 0:
            digits.append(x % 10)
            total += x % 10
            x //= 10
        print(f"i={i}: nums[i]={nums[i]} chữ số (từ cuối)={digits} tổng={total} so với i -> {'KHỚP' if total == i else 'không'}")
        if total == i:
            return i
    return -1

print("kết quả:", trace_smallest_index([1, 10, 11]))
print("kết quả:", trace_smallest_index([1, 2, 3]))
print("kết quả:", trace_smallest_index([0, 5]))
````

## 9. Code nộp LeetCode

````python
from typing import List

class Solution:
    def smallestIndex(self, nums: List[int]) -> int:
        for i, v in enumerate(nums):
            total = 0
            while v > 0:
                total += v % 10
                v //= 10
            if total == i:
                return i
        return -1

print(Solution().smallestIndex([1, 3, 2]), Solution().smallestIndex([1, 10, 11]), Solution().smallestIndex([1, 2, 3]))
````

## 10. Vì sao thuật toán đúng

- **Tổng chữ số:** vòng lặp `% 10` / `// 10` cộng đúng từng chữ số của `v` (mỗi lần lấy chữ số cuối rồi bỏ nó); với `v = 0` tổng là `0`, đúng định nghĩa.
- **Nhỏ nhất:** ta duyệt `i = 0, 1, 2, ...` tăng dần và trả ngay lần đầu tiên điều kiện đúng; mọi chỉ số nhỏ hơn đã bị loại, nên chỉ số trả về là nhỏ nhất.
- **Không có đáp án:** nếu duyệt hết không gặp, tức mọi `i` đều không thỏa, trả `-1`.
- **Cách 3:** `ds[x] = ds[x // 10] + x % 10` đúng vì chữ số của `x` gồm chữ số của `x // 10` và chữ số cuối `x % 10`.
- **Cách 4 (chặn cận):** với `0 <= v <= 1000`, tổng chữ số tối đa là `27` (ở `v = 999`; còn `1000` có tổng `1`). Nếu `i >= 28` thì tổng không thể bằng `i`, nên chỉ cần xét `i < 28`. Cả bộ test đối chiếu Cách 4 với các cách còn lại.

## 11. Phân tích Time / Space Complexity

Gọi `n = len(nums)`, `V = 1000` là giá trị tối đa.

| Cách | Time | Space |
|---|---|---|
| 1. Vòng lặp `% 10` | `O(n)` (mỗi số `O(log V)`) | `O(1)` |
| 2. Chuỗi + `sum(map(int, ...))` | `O(n)` | `O(1)` |
| 3. Bảng `ds` dựng sẵn | `O(V + n)` | `O(V)` |
| 4. Chặn cận `i <= 27` | `O(min(n, 28))` | `O(1)` |

Với `n <= 100` mọi cách đều tức thì. Trong phỏng vấn, Cách 1 hoặc 2 với thoát sớm là đủ; nêu thêm nhận xét `27` để thể hiện hiểu bài.

## 12. Bộ test

Test 1: ví dụ đề và các edge case: `nums[0] = 0` (chỉ số `0` hợp lệ), `n = 1`, không có đáp án, nhiều đáp án (chọn nhỏ nhất), giá trị biên `1000` và `999`, chỉ số lớn hơn `27`. Mọi input tuân thủ `1 <= n <= 100`, `0 <= nums[i] <= 1000`.

````python
impls = [SolutionLoop(), SolutionStr(), SolutionTable(), SolutionBound(), Solution()]
tests = [
    ([1, 3, 2], 2),
    ([1, 10, 11], 1),
    ([1, 2, 3], -1),
    ([0], 0),                                   # n = 1, nums[0] = 0 -> tổng chữ số 0 == chỉ số 0
    ([5], -1),                                  # n = 1, không khớp
    ([0, 1, 2, 3], 0),                          # mọi chỉ số đều khớp -> chọn nhỏ nhất 0
    ([9, 9, 2], 2),                             # chỉ i = 2 khớp
    ([1000, 1], 1),                             # 1000 có tổng chữ số 1; i = 0 không khớp, i = 1 (nums[1] = 1) khớp
    ([999] + [0] * 99, -1),                     # nums[0]=999 tổng 27 != 0; các số 0 có tổng 0 != i (i >= 1)
    ([999] * 28, 27),                           # i = 27: 999 có tổng 27 (chỉ số lớn nhất có thể khớp)
    ([999] * 27 + [998] + [999], -1),           # đủ 29 phần tử nhưng không chỉ số nào khớp (tổng tối đa 27)
    ([19, 28, 37, 46], -1),                     # tổng chữ số đều 10
    ([10, 20, 30, 3], 3),                       # tổng 1, 2, 3, 3 -> chỉ i = 3 khớp
]
for nums, exp in tests:
    assert 1 <= len(nums) <= 100 and all(0 <= v <= 1000 for v in nums)
    for im in impls:
        got = im.smallestIndex(list(nums))
        assert got == exp, (type(im).__name__, nums, got, exp)
print("tất cả test cố định OK:", len(tests))
````

Test 2: ngẫu nhiên có seed. Dùng dải giá trị nhỏ (`0..30`, nhiều ca có đáp án) và dải đầy đủ (`0..1000`, nhiều ca `-1`), cộng ca **cài sẵn đáp án**: chọn chỉ số `i` rồi đặt `nums[i]` là một số có tổng chữ số `i` (tìm bằng duyệt `0..1000`). Oracle độc lập: gom mọi chỉ số thỏa vào danh sách rồi lấy `min`.

````python
import random
random.seed(3550)

def ds_str(v):
    return sum(int(ch) for ch in str(v))

def oracle(nums):
    good = [i for i in range(len(nums)) if ds_str(nums[i]) == i]
    return min(good) if good else -1

by_sum = {}
for v in range(1001):
    by_sum.setdefault(ds_str(v), []).append(v)

hits = misses = 0
for t in range(3000):
    n = random.randint(1, 100)
    hi_val = random.choice([30, 100, 1000])
    nums = [random.randint(0, hi_val) for _ in range(n)]
    if t % 3 == 0:                                       # cài sẵn một đáp án
        i = random.randint(0, min(n - 1, 27))
        nums[i] = random.choice(by_sum[i])
    assert 1 <= len(nums) <= 100 and all(0 <= v <= 1000 for v in nums)
    ref = oracle(nums)
    hits += ref != -1
    misses += ref == -1
    for im in impls:
        assert im.smallestIndex(list(nums)) == ref, (type(im).__name__, nums)
print("3000 test ngẫu nhiên OK; có đáp án:", hits, "| -1:", misses)
assert hits > 500 and misses > 100
````

Test 3: tính chất — đáp án (nếu khác `-1`) luôn `<= 27` và `< n`; nối thêm phần tử vào **cuối** mảng không làm đổi đáp án nếu đã có đáp án; không sửa input.

````python
random.seed(50)
for _ in range(2000):
    n = random.randint(1, 100)
    nums = [random.randint(0, 40) for _ in range(n)]
    snapshot = list(nums)
    ans = Solution().smallestIndex(nums)
    assert nums == snapshot
    assert ans == -1 or (0 <= ans <= 27 and ans < n)
    if ans != -1 and n < 100:
        assert Solution().smallestIndex(nums + [random.randint(0, 1000)]) == ans
print("tính chất (<= 27, < n, thêm cuối không đổi, không sửa input): OK")
````

## 13. Các lỗi người mới hay gặp

### Lỗi 1: so `nums[i]` với `i` thay vì **tổng chữ số** của `nums[i]`

````python
def wrong_compare_value(nums):
    for i, v in enumerate(nums):
        if v == i:                                # SAI: phải so tổng chữ số của v với i
            return i
    return -1

nums = [1, 10, 11]
print("sai :", wrong_compare_value(nums))
print("đúng:", Solution().smallestIndex(nums))
assert wrong_compare_value(nums) != Solution().smallestIndex(nums)
````

Vì sao sai: `[1,10,11]` không có `nums[i] == i` nên code sai trả `-1`, trong khi `10` có tổng chữ số `1 == i`. **Sửa:** tính tổng chữ số trước khi so.

### Lỗi 2: trả về chỉ số **lớn nhất** (duyệt ngược)

````python
def wrong_largest_index(nums):
    for i in range(len(nums) - 1, -1, -1):        # SAI: duyệt từ cuối -> ra chỉ số lớn nhất
        if sum(map(int, str(nums[i]))) == i:
            return i
    return -1

nums = [1, 10, 11]
print("sai :", wrong_largest_index(nums))
print("đúng:", Solution().smallestIndex(nums))
assert wrong_largest_index(nums) != Solution().smallestIndex(nums)
````

Vì sao sai: cả `i = 1` và `i = 2` đều thỏa; đề hỏi chỉ số **nhỏ nhất** (`1`) còn code sai trả `2`. **Sửa:** duyệt từ `0` tăng dần và thoát ngay lần đầu.

### Lỗi 3: trả `0` thay vì `-1` khi không có đáp án

````python
def wrong_default_zero(nums):
    ans = 0                                       # SAI: 0 cũng là một chỉ số hợp lệ
    for i, v in enumerate(nums):
        if sum(map(int, str(v))) == i:
            ans = i
            break
    return ans

nums = [1, 2, 3]
print("sai :", wrong_default_zero(nums))
print("đúng:", Solution().smallestIndex(nums))
assert wrong_default_zero(nums) != Solution().smallestIndex(nums)
````

Vì sao sai: không có đáp án thì phải trả `-1`; trả `0` bị nhầm với "đáp án là chỉ số 0" (ví dụ `nums = [0]` có đáp án đúng `0`). **Sửa:** `return -1` sau vòng lặp.

### Lỗi 4: tính tổng chữ số sai (chỉ lấy chữ số cuối)

````python
def wrong_last_digit(nums):
    for i, v in enumerate(nums):
        if v % 10 == i:                           # SAI: v % 10 chỉ là chữ số cuối, không phải tổng
            return i
    return -1

nums = [1, 10, 11]
print("sai :", wrong_last_digit(nums))
print("đúng:", Solution().smallestIndex(nums))
assert wrong_last_digit(nums) != Solution().smallestIndex(nums)
````

Vì sao sai: `10 % 10 = 0`, `11 % 10 = 1`, `1 % 10 = 1` nên không thấy khớp (`i = 0` cần `0`, `i = 1` cần `1` nhưng `10 % 10 = 0`, `i = 2` cần `2`) -> trả `-1`. Tổng chữ số phải cộng **mọi** chữ số: `1 + 0 = 1`. **Sửa:** vòng lặp `% 10` / `// 10` cộng dồn.

## 14. Mẹo làm bài nhanh & nhận dạng pattern

- "Chỉ số **nhỏ nhất** thỏa điều kiện" -> duyệt từ `0`, **thoát sớm** ở lần đầu tiên.
- "Tổng chữ số" -> `while x: s += x % 10; x //= 10`, hoặc `sum(map(int, str(x)))`.
- Không có đáp án -> `-1`; đừng nhầm với chỉ số `0` hợp lệ.
- Giá trị nhỏ (`<= 1000`) -> tổng chữ số rất nhỏ (`<= 27`): có thể chặn cận để biết đáp án chỉ nằm ở đầu mảng.
- Kiểm tra `0`: tổng chữ số của `0` là `0`, nên `nums[0] = 0` thỏa ngay.
- Bài họ hàng: `2535. Difference Between Element Sum and Digit Sum of an Array`, `258. Add Digits`, `1356. Sort Integers by The Number of 1 Bits`.

## 15. Biến thể của bài

1. **Chỉ số lớn nhất** thỏa điều kiện (duyệt ngược hoặc lấy `max`).
2. **Đếm** số chỉ số thỏa `digit_sum(nums[i]) == i`.
3. Điều kiện khác: `digit_sum(nums[i]) == nums[i] % k`, hoặc tổng chữ số của **chỉ số** bằng `nums[i]`.
4. **Chỉ số đánh từ 1** thay vì từ 0: so `digit_sum(nums[i]) == i + 1`.
5. Giá trị đến `10^9`: tổng chữ số tối đa `81`; chặn cận tương tự: chỉ xét `i <= 81`.

## 16. So sánh các cách làm

|---|---|---|---|---|
| 1. Vòng lặp `% 10` | tính tổng chữ số thủ công | `O(n)` | `O(1)` | **chuẩn, nên nộp** |
| 2. Chuỗi + `sum(map(...))` | đổi sang chuỗi | `O(n)` | `O(1)` | ngắn gọn nhất |
| 3. Bảng `ds` dựng sẵn | `ds[x] = ds[x//10] + x%10` | `O(V + n)` | `O(V)` | nhiều lần gọi / nhiều mảng |
| 4. Chặn cận `i <= 27` | chỉ xét 28 phần tử đầu | `O(min(n, 28))` | `O(1)` | hiểu cấu trúc, tối ưu hằng số |

## 19. Đáp án tham khảo cho bài tập

````python
# Bài 1
assert Solution().smallestIndex([5, 4, 3, 2]) == -1

# Bài 2: 20 -> 2 (i=0), 6 -> 6 (i=1), 11 -> 2 (i=2 đúng), 120 -> 3 (i=3 cũng đúng)
assert Solution().smallestIndex([20, 6, 11, 120]) == 2

# Bài 3
def digit_sum_ex(x):
    total = 0
    while x > 0:
        total += x % 10
        x //= 10
    return total
assert digit_sum_ex(1234) == 10 and digit_sum_ex(0) == 0 and digit_sum_ex(999) == 27

# Bài 4
def count_matching(nums):
    count = 0
    for i, v in enumerate(nums):
        if digit_sum_ex(v) == i:
            count += 1
    return count
assert count_matching([20, 6, 11, 120]) == 2
assert count_matching([0, 1, 2, 3]) == 4
assert count_matching([5, 4, 3, 2]) == 0

# Bài 5
def largest_index_digit_sum(nums):
    for i in range(len(nums) - 1, -1, -1):
        if digit_sum_ex(nums[i]) == i:
            return i
    return -1
assert largest_index_digit_sum([20, 6, 11, 120]) == 3
assert largest_index_digit_sum([1, 10, 11]) == 2
assert largest_index_digit_sum([1, 2, 3]) == -1
print("đáp án bài tập OK")
````
