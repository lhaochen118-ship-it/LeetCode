## 1. Mục tiêu học

Sau notebook này bạn sẽ:

1. Hiểu đề: xóa **đúng một** ký tự bằng `digit` khỏi chuỗi `number` sao cho kết quả (xét như số) là **lớn nhất**.
2. Dùng **cắt chuỗi** (slicing) `s[:i] + s[i+1:]` để xóa một ký tự, và biết so sánh chuỗi cùng độ dài tương đương so sánh số.
3. Viết 4 cách: liệt kê mọi ứng viên (brute force), `max` trên chuỗi, **tham lam một lượt** (xóa lần xuất hiện đầu tiên mà chữ số sau nó lớn hơn), và tham lam bằng `find`/`rfind`.

## 2. Đề bài & ví dụ

Cho chuỗi `number` biểu diễn một số nguyên dương và một ký tự chữ số `digit`. Hãy **xóa đúng một lần xuất hiện** của `digit` khỏi `number` sao cho giá trị số của chuỗi còn lại là **lớn nhất** có thể. Trả về chuỗi kết quả. Đề đảm bảo `digit` xuất hiện ít nhất một lần trong `number`.

| Ví dụ | number | digit | Output | Giải thích |
|---|---|---|---|---|
| 1 | `"123"` | `"3"` | `"12"` | Chỉ có một số `3`; xóa nó. |
| 2 | `"1231"` | `"1"` | `"231"` | Xóa `1` đầu → `"231"`; xóa `1` cuối → `"123"`. Lớn nhất là `"231"`. |
| 3 | `"551"` | `"5"` | `"51"` | Xóa `5` nào cũng cho `"51"`. |

Constraints:

- `2 <= number.length <= 100`
- `number` gồm các chữ số từ `'1'` đến `'9'` (không có `'0'`).
- `digit` là một chữ số từ `'1'` đến `'9'`.
- `digit` xuất hiện ít nhất một lần trong `number`.

````python
examples = [("123", "3", "12"), ("1231", "1", "231"), ("551", "5", "51")]
for number, digit, expected in examples:
    candidates = [number[:i] + number[i + 1:] for i, c in enumerate(number) if c == digit]
    print(f"number={number!r}, digit={digit!r} | ứng viên: {candidates} | mong đợi = {expected!r}")
````

## 4. Ý tưởng nền tảng — tính bằng tay

| Xóa vị trí | Chuỗi còn lại |
|---|---|
| `0` | `"231"` |
| `3` | `"123"` |

## 5. Cách 1 — Brute Force: thử xóa từng lần xuất hiện

Duyệt mọi vị trí `i` có `number[i] == digit`, tạo chuỗi sau khi xóa, rồi chọn chuỗi lớn nhất. Vì độ dài mọi ứng viên bằng nhau (`n - 1`) và không có ký tự `0` đứng đầu, so sánh theo **giá trị số** `int(...)` an toàn (và cũng tương đương so sánh chuỗi).

````python
class SolutionBrute:
    def removeDigit(self, number, digit):
        best_val, best_str = -1, ""
        for i in range(len(number)):
            if number[i] == digit:
                cand = number[:i] + number[i + 1:]
                if int(cand) > best_val:
                    best_val, best_str = int(cand), cand
        return best_str

sol = SolutionBrute()
print(sol.removeDigit("123", "3"))
print(sol.removeDigit("1231", "1"))
print(sol.removeDigit("551", "5"))
````

**Độ phức tạp:** có tối đa `n` vị trí; mỗi lần tạo chuỗi và đổi sang `int` mất `O(n)` ⇒ Time `O(n²)`, Space `O(n)`. Với `n <= 100` rất nhẹ.

**Nút thắt:** ta tạo `n` chuỗi dài `n` rồi so sánh. Có thể **quyết định ngay trong một lượt** nhờ quy luật về chữ số đứng sau.

## 6. Ý tưởng thuật toán — bước đột phá

So sánh kết quả khi xóa hai vị trí `i < j` đều có `digit`:

- `r_i` = xóa `i`, `r_j` = xóa `j`. Tại mỗi vị trí `k` trong `[i, j - 1]`: `r_i[k] = number[k + 1]`, còn `r_j[k] = number[k]`. Ngoài đoạn đó hai chuỗi giống nhau.
- Vậy hai chuỗi khác nhau **lần đầu** ở vị trí `k` nhỏ nhất (từ `i`) mà `number[k] != number[k + 1]`; tại đó `r_i` lớn hơn nếu `number[k + 1] > number[k]`.

Rút ra quy tắc "nếu… thì…":

- **Nếu** tồn tại lần xuất hiện `i` của `digit` mà `number[i + 1] > digit` **thì** xóa **lần đầu tiên như vậy** (kết quả lớn nhất).
- **Nếu không** có lần nào như thế (sau mọi `digit` là chữ số `<= digit` hoặc hết chuỗi) **thì** xóa lần xuất hiện **cuối cùng**.

**Invariant khi quét trái sang phải:** nếu chưa gặp `digit` nào có chữ số sau lớn hơn nó, thì lần xóa tốt nhất cho tới giờ là **lần xuất hiện gần nhất** (`last`).

## 7. Các cách tiếp theo

### Cách 2 — `max` trên chuỗi ứng viên (ngắn nhất)

Sinh các ứng viên bằng generator rồi `max`. Đúng vì ứng viên cùng độ dài nên so sánh chuỗi = so sánh số.

````python
class SolutionMax:
    def removeDigit(self, number, digit):
        return max(number[:i] + number[i + 1:] for i, c in enumerate(number) if c == digit)

sol = SolutionMax()
print(sol.removeDigit("123", "3"))
print(sol.removeDigit("1231", "1"))
print(sol.removeDigit("551", "5"))
````

**Độ phức tạp:** Time `O(n²)` (tạo tối đa `n` chuỗi, mỗi chuỗi `O(n)`), Space `O(n)`. Rất ngắn và dễ đúng; dùng được khi `n` nhỏ như ở đây.

### Cách 3 — Tham lam một lượt

Quét trái sang phải, nhớ `last` là vị trí `digit` gần nhất. Gặp `digit` tại `i`: cập nhật `last = i`; nếu `number[i + 1] > digit` thì xóa ngay `i` và trả về. Hết vòng mà chưa trả về thì xóa `last`.

````python
class SolutionGreedy:
    def removeDigit(self, number, digit):
        n = len(number)
        last = -1
        for i in range(n):
            if number[i] == digit:
                last = i
                if i + 1 < n and number[i + 1] > digit:
                    return number[:i] + number[i + 1:]     # xóa lần đầu có chữ số sau lớn hơn
        return number[:last] + number[last + 1:]            # không có -> xóa lần cuối

sol = SolutionGreedy()
print(sol.removeDigit("123", "3"))
print(sol.removeDigit("1231", "1"))
print(sol.removeDigit("551", "5"))
````

**Độ phức tạp:** Time `O(n)` (một lượt + một lần cắt chuỗi), Space `O(n)` cho chuỗi kết quả.

### Cách 4 — Tham lam bằng `find` / `rfind`

Cùng quy tắc nhưng dùng hàm tìm kiếm có sẵn: lần lượt tìm từng vị trí `digit` bằng `find`; nếu chữ số sau lớn hơn thì xóa; nếu không có vị trí nào như vậy thì xóa vị trí `rfind`.

````python
class SolutionFind:
    def removeDigit(self, number, digit):
        n = len(number)
        i = number.find(digit)
        while i != -1:
            if i + 1 < n and number[i + 1] > digit:
                return number[:i] + number[i + 1:]
            i = number.find(digit, i + 1)                   # lần xuất hiện kế tiếp
        j = number.rfind(digit)
        return number[:j] + number[j + 1:]

sol = SolutionFind()
print(sol.removeDigit("123", "3"))
print(sol.removeDigit("1231", "1"))
print(sol.removeDigit("551", "5"))
````

**Độ phức tạp:** Time `O(n)` (mỗi `find` tiếp tục từ vị trí sau, tổng quét `O(n)`), Space `O(n)`.

## 8. TRACE từng bước

Trace cách 3 với `number = "52535"`, `digit = "5"`:

| i | number[i] | là digit? | number[i+1] | `> digit`? | last | hành động |
|---|---|---|---|---|---|---|
| 0 | `5` | có | `2` | không | 0 | tiếp tục |
| 1 | `2` | không | | | 0 | bỏ qua |
| 2 | `5` | có | `3` | không | 2 | tiếp tục |
| 3 | `3` | không | | | 2 | bỏ qua |
| 4 | `5` | có | (hết) | không | 4 | tiếp tục |

Hết vòng ⇒ xóa `last = 4` ⇒ `"5253"`. Kiểm chứng: xóa vị trí `0` → `"2535"`, vị trí `2` → `"5235"`, vị trí `4` → `"5253"`. Lớn nhất là `"5253"`. Hàm dưới in từng bước.

````python
def trace_greedy(number, digit):
    n, last = len(number), -1
    for i in range(n):
        c = number[i]
        if c == digit:
            last = i
            nxt = number[i + 1] if i + 1 < n else None
            print(f"i={i} c={c} là digit | số sau={nxt} | last={last}", end="")
            if nxt is not None and nxt > digit:
                res = number[:i] + number[i + 1:]
                print(f" -> sau lớn hơn, xóa i, kết quả {res}")
                return res
            print(" -> chưa có lý do xóa, tiếp tục")
        else:
            print(f"i={i} c={c} không phải digit")
    res = number[:last] + number[last + 1:]
    print(f"hết chuỗi -> xóa last={last}, kết quả {res}")
    return res

print(trace_greedy("52535", "5"))
````

## 9. Code nộp LeetCode

Dùng cách 3 (tham lam một lượt).

````python
from typing import List

class Solution:
    def removeDigit(self, number: str, digit: str) -> str:
        n = len(number)
        last = -1
        for i in range(n):
            if number[i] == digit:
                last = i
                if i + 1 < n and number[i + 1] > digit:
                    return number[:i] + number[i + 1:]
        return number[:last] + number[last + 1:]

print(Solution().removeDigit("123", "3"))
print(Solution().removeDigit("1231", "1"))
print(Solution().removeDigit("551", "5"))
````

## 10. Vì sao thuật toán đúng?

Xét hai lần xuất hiện `i < j` của `digit`. Hai chuỗi `r_i`, `r_j` chỉ có thể khác nhau ở các vị trí `k ∈ [i, j-1]`, nơi `r_i[k] = number[k+1]` và `r_j[k] = number[k]`.

1. **Nếu `number[i..j]` toàn ký tự giống nhau** thì `r_i = r_j` (xóa chữ nào trong dãy bằng nhau cũng như nhau).
2. **Ngược lại**, gọi `k` là vị trí đầu tiên trong `[i, j-1]` mà `number[k] != number[k+1]`. Hai chuỗi khác nhau lần đầu ở `k`: `r_i` hơn `r_j` ⇔ `number[k+1] > number[k]`.

**Nếu có `i` đầu tiên mà `number[i+1] > digit`:**

- Với mọi `j > i` có `digit`: `number[i] = digit` và `number[i+1] > digit` nên vị trí khác biệt đầu tiên là `k = i`, tại đó `r_i[k] = number[i+1] > digit = r_j[k]` ⇒ `r_i > r_j`.
- Với các lần xuất hiện **trước** `i` (gọi chúng lần lượt `o_1 < o_2 < ... < i`), mỗi cái có chữ số sau `<= digit`. So hai lần xuất hiện **liền kề** `o_s` và `o_{s+1}`: nếu chữ số sau `o_s` bằng `digit` thì `o_{s+1} = o_s + 1` và hai kết quả **giống hệt**; nếu nhỏ hơn `digit` thì tại `k = o_s` ta có `r_{o_s}[k] < digit = r_{o_{s+1}}[k]` ⇒ `r_{o_s} < r_{o_{s+1}}`. Vậy dãy kết quả **không giảm** khi đi từ lần xuất hiện đầu tới `i`, nên `r_i` lớn nhất trong nhóm này.

**Nếu không có `i` nào như vậy:** mọi lần xuất hiện đều có chữ số sau `<= digit` (hoặc là ký tự cuối). Lập luận chuỗi ở trên áp dụng cho **mọi** cặp liền kề ⇒ dãy kết quả không giảm đến lần xuất hiện **cuối cùng**, nên lần cuối là tối ưu.

Để chắc chắn, mục 12 so sánh cách tham lam với brute force trên **mọi** chuỗi ngắn.

## 13. Các lỗi người mới hay gặp

### Lỗi 1 — Xóa **mọi** chữ số bằng `digit` (dùng `replace`)

Đề chỉ cho xóa **đúng một** lần xuất hiện. `number.replace(digit, "")` xóa hết.

### Lỗi 2 — Luôn xóa lần xuất hiện đầu tiên (hoặc luôn xóa lần cuối)

Đúng với một số ví dụ nhưng sai tổng quát.

````python
def wrong_first(number, digit):
    i = number.find(digit)
    return number[:i] + number[i + 1:]

def wrong_last(number, digit):
    i = number.rfind(digit)
    return number[:i] + number[i + 1:]

print("luôn xóa đầu, 1231:", wrong_first("1231", "1"), "(đúng là 231 - trùng hợp)")
print("luôn xóa đầu, 52535:", wrong_first("52535", "5"), "(đúng là 5253 - SAI)")
print("luôn xóa cuối, 1231:", wrong_last("1231", "1"), "(đúng là 231 - SAI)")
print("đúng:", Solution().removeDigit("52535", "5"), Solution().removeDigit("1231", "1"))
````

### Lỗi 3 — So sánh chuỗi khác độ dài

Mọi ứng viên có cùng độ dài nên an toàn. Nhưng nếu bạn tự "xóa" kiểu khác (ví dụ cắt bỏ phần đầu), độ dài thay đổi và so sánh chuỗi sai.

````python
print("so sánh chuỗi khác độ dài:", "9" > "100", "| so sánh số:", int("9") > int("100"))
cands_same_len = ["231", "123"]
print("cùng độ dài, an toàn:", max(cands_same_len), max(cands_same_len, key=int))
````

### Lỗi 4 — Truy cập `number[i + 1]` vượt biên

Khi `digit` ở **cuối** chuỗi, `number[i + 1]` gây `IndexError`. Phải kiểm tra `i + 1 < n` trước.

````python
def wrong_index(number, digit):
    for i in range(len(number)):
        if number[i] == digit and number[i + 1] > digit:   # thiếu kiểm tra biên
            return number[:i] + number[i + 1:]
    j = number.rfind(digit)
    return number[:j] + number[j + 1:]

try:
    print(wrong_index("123", "3"))
except IndexError as e:
    print("IndexError:", e)
print("đúng:", Solution().removeDigit("123", "3"))
````

## 14. Mẹo làm bài nhanh & nhận dạng pattern

- "Xóa một phần tử để được số **lớn nhất**" ⇒ nhìn **chữ số đứng sau**: nếu lớn hơn thì xóa ngay (kéo chữ số lớn lên hàng cao).
- Nếu không tìm được vị trí nào "đáng xóa", xóa **vị trí cuối cùng** (chữ số bị bỏ ở hàng thấp nhất, ảnh hưởng ít nhất).
- Chuỗi chữ số **cùng độ dài, không có số 0 đầu** ⇒ so sánh chuỗi = so sánh số.
- Với `n` nhỏ (`<= 100`), brute force + `max` là phương án an toàn để đối chiếu.
- Họ hàng: Remove K Digits (402), Largest Number (179), Maximum Swap (670), Create Maximum Number (321), Remove Duplicate Letters (316).

## 15. Biến thể của bài

1. **Số nhỏ nhất:** xóa một lần xuất hiện để kết quả nhỏ nhất (đảo ngược quy tắc: xóa lần đầu có chữ số sau **nhỏ hơn**, ngược lại xóa lần cuối).
2. **Xóa `k` chữ số bất kỳ (LC 402):** dùng ngăn xếp đơn điệu.
3. **Cho phép xóa nhiều lần cùng `digit`:** chọn tập con tối ưu.
4. **`number` có thể chứa `0`:** cần xử lý số 0 đứng đầu.
5. **Hoán đổi hai chữ số để lớn nhất (LC 670):** cùng tư duy "so sánh vị trí".

````python
# Biến thể 1: số nhỏ nhất sau khi xóa một lần xuất hiện của digit
def remove_digit_min(number, digit):
    return min(number[:i] + number[i + 1:] for i, c in enumerate(number) if c == digit)

print(remove_digit_min("1231", "1"), remove_digit_min("52535", "5"))
````

## 16. So sánh các cách làm

| Cách | Ý tưởng | Time | Space | Khi nào dùng |
|---|---|---|---|---|
| 1. Brute force (`int`) | Thử mọi vị trí, so sánh số | `O(n²)` | `O(n)` | Hiểu đề, dùng làm chuẩn đối chiếu |
| 2. `max` trên chuỗi | Generator ứng viên + `max` | `O(n²)` | `O(n)` | Viết nhanh khi `n` nhỏ |
| 3. Tham lam một lượt | Xóa lần đầu có chữ số sau lớn hơn, không thì lần cuối | `O(n)` | `O(n)` | **Nộp bài**, tối ưu |
| 4. Tham lam `find`/`rfind` | Cùng quy tắc, dùng hàm tìm có sẵn | `O(n)` | `O(n)` | Khi thích hàm chuỗi |

## 19. Đáp án tham khảo cho bài tập

Mỗi đáp án được kiểm tra bằng `assert`.

````python
# Bài 1: xóa 3 đầu -> "123"; xóa 3 cuối -> "312"; lớn nhất "312"?  (so sánh "123" và "312")
assert Solution().removeDigit("3123", "3") == "312"

# Bài 2: "73279", digit 7 -> xóa vị trí 0: "3279"; xóa vị trí 3: "7329" -> lớn nhất "7329"
assert "73279"[:0] + "73279"[1:] == "3279"
assert "73279"[:3] + "73279"[4:] == "7329"
assert Solution().removeDigit("73279", "7") == "7329"

# Bài 3
def remove_at_ans(s, i):
    return s[:i] + s[i + 1:]
assert remove_at_ans("1231", 0) == "231"
assert remove_at_ans("1231", 3) == "123"

# Bài 4
def remove_digit_min_greedy_ans(number, digit):
    n, last = len(number), -1
    for i in range(n):
        if number[i] == digit:
            last = i
            if i + 1 < n and number[i + 1] < digit:
                return number[:i] + number[i + 1:]
    return number[:last] + number[last + 1:]
assert remove_digit_min_greedy_ans("1231", "1") == "123"
assert remove_digit_min_greedy_ans("52535", "5") == "2535"
for number, digit in [("1231", "1"), ("52535", "5"), ("3123", "3"), ("73279", "7")]:
    assert remove_digit_min_greedy_ans(number, digit) == min(number[:i] + number[i + 1:] for i, c in enumerate(number) if c == digit)

# Bài 5
def count_best_positions_ans(number, digit):
    cands = [number[:i] + number[i + 1:] for i, c in enumerate(number) if c == digit]
    best = max(cands)
    return sum(1 for c in cands if c == best)
assert count_best_positions_ans("551", "5") == 2
assert count_best_positions_ans("1231", "1") == 1
assert count_best_positions_ans("7777", "7") == 4
print("Tất cả đáp án bài tập: OK")
````

class Solution:
    def removeDigit(self, number: str, digit: str) -> str:
        n, last = len(number), -1
        for i in range(n):
            if number[i] == digit:
                last = i
                if i + 1 < n and number[i + 1] > digit:   # chữ số sau lớn hơn -> xóa ngay
