### 3.6. Ghi nhớ kết quả: `functools.lru_cache`

`@lru_cache(None)` đặt trên một hàm đệ quy sẽ **nhớ** kết quả theo tham số, nên mỗi tổ hợp tham số chỉ tính một lần. Ta dùng cho lời giải vét cạn (quy hoạch động) ở mục 5.

````python
from functools import lru_cache

calls = 0
@lru_cache(None)
def fib(n):
    global calls
    calls += 1
    return n if n < 2 else fib(n - 1) + fib(n - 2)

print(fib(30), "chỉ cần", calls, "lần tính (không có cache sẽ hàng triệu lần)")
````

## 4. Ý tưởng nền tảng / cách nghĩ trực giác

Ví dụ 1: `money = 20`, `children = 3`.

1. Phát trước mỗi đứa 1 đô → dùng 3 đô, còn **17**.
2. Muốn nâng một đứa lên 8 đô cần thêm 7. Nâng đứa thứ nhất: còn `17 - 7 = 10`. Nâng đứa thứ hai: còn `10 - 7 = 3`. Nâng đứa thứ ba cần 7 mà chỉ còn 3 → không đủ.
3. Vậy tối đa **2** đứa lên 8 đô, còn dư **3** đô cho đứa thứ ba: đứa đó nhận `1 + 3 = 4` đô — **bị cấm**!
4. Sửa: chỉ cho **1** đứa nhận 8; hai đứa còn lại chia `20 - 8 = 12` đô, ví dụ `9` và `3` (hoặc `6, 6`) — hợp lệ. Đáp án `1`.

Các tình huống "vướng luật" chính là chỗ bài này khó nhằn.

````python
money, children = 20, 3
extra = money - children               # 17
print("extra =", extra)
for kids_with_8 in range(0, children + 1):
    need = 7 * kids_with_8
    if need <= extra:
        print(f"nâng {kids_with_8} đứa lên 8: cần {need}, dư {extra - need}")
````

## 5. Cách 1 — Brute Force / baseline (vét cạn bằng quy hoạch động)

Ta **thử** `k` từ `children` giảm dần tới `0`. Với mỗi `k`, đặt `k` đứa nhận đúng 8 đô; còn `rem = money - 8k` đô phải chia cho `c = children - k` đứa còn lại sao cho mỗi đứa nhận **ít nhất 1**, **không 4 và không 8** (vì ta muốn *đúng* `k` đứa nhận 8). Việc "chia được hay không" kiểm tra bằng hàm đệ quy có nhớ `can_split(c, m)`: thử đứa đầu tiên nhận `v` đô (`v` hợp lệ), rồi hỏi `can_split(c - 1, m - v)`.

````python
from functools import lru_cache

@lru_cache(None)
def can_split(c, m):
    """Có chia được m đô cho c đứa, mỗi đứa >= 1, không đứa nào nhận 4 hoặc 8 không?"""
    if c == 0:
        return m == 0
    for v in range(1, m - (c - 1) + 1):          # các đứa còn lại cần ít nhất 1 mỗi đứa
        if v == 4 or v == 8:
            continue
        if can_split(c - 1, m - v):
            return True
    return False

def dist_bruteforce(money: int, children: int) -> int:
    for k in range(children, -1, -1):
        rem = money - 8 * k
        if rem < 0:
            continue
        if can_split(children - k, rem):
            return k
    return -1
````

````python
print(dist_bruteforce(20, 3))     # 1
print(dist_bruteforce(16, 2))     # 2
print(dist_bruteforce(3, 4))      # -1  (không đủ cho mỗi đứa 1 đô)
````

**Độ phức tạp:** có `O(children * money)` trạng thái `(c, m)` (tối đa `30 * 200 = 6000`), mỗi trạng thái thử `O(money)` giá trị `v` → Time `O(children * money^2)`; Space `O(children * money)` cho bộ nhớ đệm. Với giới hạn đề bài vẫn chạy được, nhưng nặng và dễ sai khi viết; dùng làm **oracle** để kiểm chứng công thức.

**Nút thắt:** ta đang thử mọi cách chia thay vì dùng quan sát "chỉ cần chú ý vài tình huống đặc biệt".

## 6. Ý tưởng thuật toán / bước đột phá

Phát 1 đô cho mỗi đứa trước (`money -= children`; nếu không đủ → `-1`). Sau đó mỗi đứa muốn lên 8 đô cần thêm 7:

- `k = min(money // 7, children)` là số đứa tối đa lên được 8 đô (nếu chỉ xét số tiền), `rem = money - 7 * k` là tiền dư.

Có đúng **hai** tình huống làm cách chia ở trên không hợp lệ, buộc giảm `k` đi 1:

1. `k == children` và `rem > 0`: **ai cũng đã 8 đô** mà vẫn còn dư tiền. Phải đưa phần dư cho ai đó, làm đứa đó nhận `> 8` đô, nên chỉ còn `children - 1` đứa nhận đúng 8.
2. `k == children - 1` và `rem == 3`: còn đúng **một** đứa chưa đủ 8, nó nhận `1 + 3 = 4` đô — bị cấm. Phải hạ một đứa xuống (ví dụ `8, 4` thành `7, 5`) nên `k` giảm 1.

Nếu còn **từ hai** đứa trở lên chưa đủ 8, ta chia phần dư thoải mái (ví dụ dư 3 thì chia `2` và `3` đô tổng, hai đứa nhận `1 + 1 = 2` và `1 + 2 = 3` — không ai nhận 4).

````python
def dist_formula_demo(money, children):
    if money < children:
        return -1
    money -= children
    k = min(money // 7, children)
    rem = money - 7 * k
    print(f"   sau phát 1 đô: extra={money}, k={k}, rem={rem}")
    if (k == children and rem > 0) or (k == children - 1 and rem == 3):
        k -= 1
        print("   -> rơi vào tình huống đặc biệt, giảm k còn", k)
    return k

for m, c in [(20, 3), (16, 2), (17, 2), (12, 2), (3, 4)]:
    print(f"money={m}, children={c}:")
    print("   đáp án:", dist_formula_demo(m, c))
````

## 7. Các cách tiếp theo

Ta có thêm 3 cách: **(2)** duyệt `k` từ lớn xuống với điều kiện khả thi viết gọn, **(3)** công thức tham lam `O(1)`, **(4)** dựng cụ thể cách chia kèm bộ kiểm tra hợp lệ.

### Cách 2 — Duyệt `k` giảm dần + điều kiện khả thi

Thay vì quy hoạch động, ta viết thẳng điều kiện "chia được `rem` đô cho `c` đứa (không 4, không 8, mỗi đứa >= 1)":

- `c == 0`: phải `rem == 0`.
- `rem < c`: không đủ mỗi đứa 1 đô → không.
- `c == 1`: một đứa duy nhất nhận cả `rem` đô → được nếu `rem` khác 4 và khác 8.
- `c >= 2` (và `rem >= c`): luôn chia được (tránh 4 và 8 bằng cách chia lệch).

Mệnh đề cuối là điểm mấu chốt; ta sẽ **kiểm chứng bằng oracle** ở mục 12.

````python
def feasible(c, rem):
    if c == 0:
        return rem == 0
    if rem < c:
        return False
    if c == 1:
        return rem != 4 and rem != 8
    return True

def dist_scan(money, children):
    for k in range(children, -1, -1):
        if money - 8 * k >= 0 and feasible(children - k, money - 8 * k):
            return k
    return -1

print(dist_scan(20, 3), dist_scan(16, 2), dist_scan(3, 4))
````

````python
class SolutionScan:
    def distMoney(self, money: int, children: int) -> int:
        def feasible(c, rem):
            if c == 0:
                return rem == 0
            if rem < c:
                return False
            if c == 1:
                return rem != 4 and rem != 8
            return True

        for k in range(children, -1, -1):
            if money - 8 * k >= 0 and feasible(children - k, money - 8 * k):
                return k
        return -1
````

Độ phức tạp Cách 2: tối đa `children + 1` lần thử, mỗi lần `O(1)` → Time `O(children)`; Space `O(1)`.

### Cách 3 — Công thức tham lam `O(1)`

Đây là cách ở mục 6: phát 1 đô, `k = min(extra // 7, children)`, `rem = extra - 7k`, rồi hai điều kiện giảm `k`.

````python
def dist_greedy(money, children):
    if money < children:
        return -1
    money -= children                        # mỗi đứa 1 đô
    k = min(money // 7, children)            # tối đa bao nhiêu đứa lên 8 (cần thêm 7 mỗi đứa)
    rem = money - 7 * k
    if (k == children and rem > 0) or (k == children - 1 and rem == 3):
        k -= 1
    return k

print(dist_greedy(20, 3), dist_greedy(16, 2), dist_greedy(3, 4))
````

````python
class SolutionGreedy:
    def distMoney(self, money: int, children: int) -> int:
        if money < children:
            return -1
        money -= children
        k = min(money // 7, children)
        rem = money - 7 * k
        if (k == children and rem > 0) or (k == children - 1 and rem == 3):
            k -= 1
        return k
````

Độ phức tạp Cách 3: Time `O(1)`; Space `O(1)`.

### Cách 4 — Dựng cụ thể một cách chia, rồi đếm số đứa nhận 8

Thay vì chỉ trả về số, ta **dựng danh sách `amounts`** (số tiền từng đứa) và chỉnh lại nếu vi phạm luật. Cách này vừa trả lời bài toán, vừa cho ta "bằng chứng" (một cách chia hợp lệ) để kiểm tra bằng code.

Giải thích từng đoạn:

1. Mỗi đứa nhận 1 đô; `left` là phần còn lại.
2. Lần lượt nâng từng đứa: `give = min(7, left)` (nếu đủ 7 thì lên 8, nếu không thì đưa hết phần còn lại).
3. Nếu hết một lượt mà `left > 0`: ai cũng đã 8, đưa phần dư cho đứa cuối (nó thành `> 8`).
4. Nếu có đứa nhận đúng 4: nếu đó không phải đứa cuối thì chuyển 1 đô cho đứa kế tiếp (đang có 1 đô) → `3` và `2`; nếu là đứa cuối thì lấy 1 đô từ đứa trước (đang là 8) → `7` và `5`.

````python
def distribute_construct(money, children):
    """Trả về (đáp án, danh sách số tiền từng đứa). Không hợp lệ -> (-1, None)."""
    if money < children:
        return -1, None
    amounts = [1] * children
    left = money - children
    for i in range(children):
        give = min(7, left)
        amounts[i] += give
        left -= give
    if left > 0:                              # ai cũng 8 mà còn dư
        amounts[-1] += left
    for i in range(children):
        if amounts[i] == 4:
            if i < children - 1:              # đứa sau đang có 1 đô
                amounts[i] -= 1
                amounts[i + 1] += 1
            else:                             # đứa cuối: lấy 1 đô từ đứa trước (đang là 8)
                amounts[i] += 1
                amounts[i - 1] -= 1
    return amounts.count(8), amounts

for m, c in [(20, 3), (16, 2), (12, 2), (3, 4)]:
    print(m, c, "->", distribute_construct(m, c))
````

````python
class SolutionConstruct:
    def distMoney(self, money: int, children: int) -> int:
        if money < children:
            return -1
        amounts = [1] * children
        left = money - children
        for i in range(children):
            give = min(7, left)
            amounts[i] += give
            left -= give
        if left > 0:
            amounts[-1] += left
        for i in range(children):
            if amounts[i] == 4:
                if i < children - 1:
                    amounts[i] -= 1
                    amounts[i + 1] += 1
                else:
                    amounts[i] += 1
                    amounts[i - 1] -= 1
        return amounts.count(8)
````

Độ phức tạp Cách 4: Time `O(children)`; Space `O(children)` cho `amounts`.

## 8. TRACE từng bước

Trace Cách 3 với `money = 20`, `children = 3`:

| bước | giá trị |
|---|---|
| `money >= children`? | `20 >= 3` → tiếp tục |
| phát 1 đô mỗi đứa: `money -= 3` | `17` |
| `k = min(17 // 7, 3)` | `min(2, 3) = 2` |
| `rem = 17 - 7 * 2` | `3` |
| `k == children`? | `2 == 3` → không |
| `k == children - 1 and rem == 3`? | `2 == 2` và `3 == 3` → có |
| giảm `k` | `k = 1` |

Trace Cách 4 (dựng cụ thể) với `money = 12`, `children = 2` (cho thấy trường hợp cuối cùng):

| bước | `amounts` | `left` |
|---|---|---|
| bắt đầu | `[1, 1]` | 10 |
| đứa 0: `give = min(7, 10) = 7` | `[8, 1]` | 3 |
| đứa 1: `give = min(7, 3) = 3` | `[8, 4]` | 0 |
| đứa 1 đúng 4 và là đứa cuối → lấy 1 từ đứa trước | `[7, 5]` | 0 |

Kết quả `amounts.count(8) = 0`.

````python
def trace_greedy(money, children):
    print(f"money={money}, children={children}")
    if money < children:
        print("  money < children -> -1")
        return -1
    money -= children
    print("  sau khi phát 1 đô mỗi đứa, extra =", money)
    k = min(money // 7, children)
    rem = money - 7 * k
    print(f"  k = min({money} // 7, {children}) = {k}; rem = {money} - 7*{k} = {rem}")
    if k == children and rem > 0:
        print("  ai cũng 8 mà còn dư -> k -= 1")
        k -= 1
    elif k == children - 1 and rem == 3:
        print("  đứa còn lại sẽ nhận 1 + 3 = 4 (bị cấm) -> k -= 1")
        k -= 1
    else:
        print("  không có tình huống đặc biệt")
    print("  đáp án:", k)
    return k

trace_greedy(20, 3)
trace_greedy(17, 2)
trace_greedy(16, 2)
````

````python
def trace_construct(money, children):
    print(f"money={money}, children={children}")
    if money < children:
        print("  -> -1")
        return
    amounts = [1] * children
    left = money - children
    print("  bắt đầu:", amounts, "left =", left)
    for i in range(children):
        give = min(7, left)
        amounts[i] += give
        left -= give
        print(f"  đứa {i}: give={give} -> {amounts}, left={left}")
    if left > 0:
        amounts[-1] += left
        print("  dư tiền, đưa cho đứa cuối:", amounts)
    print("  cách chia sau cùng (trước bước chống số 4):", amounts)

trace_construct(12, 2)
trace_construct(20, 3)
````

## 9. Code nộp LeetCode

Nộp Cách 3 (công thức tham lam): ngắn, `O(1)`, nhưng nhớ **hai** điều kiện đặc biệt.

````python
class Solution:
    def distMoney(self, money: int, children: int) -> int:
        if money < children:
            return -1
        money -= children                     # ai cũng có 1 đô
        k = min(money // 7, children)         # số đứa lên được 8 đô
        rem = money - 7 * k                   # tiền còn dư
        if (k == children and rem > 0) or (k == children - 1 and rem == 3):
            k -= 1
        return k

print(Solution().distMoney(20, 3))   # 1
print(Solution().distMoney(16, 2))   # 2
````

## 10. Vì sao thuật toán đúng

Đặt `extra = money - children` (sau khi phát 1 đô mỗi đứa) — bắt buộc `extra >= 0`, nếu không **không có** cách chia nào nên `-1`.

- **Cận trên của `k`:** muốn `k` đứa nhận đúng 8 cần ít nhất `7k` đô từ `extra` (mỗi đứa thêm 7), và `k <= children`. Vậy `k <= min(extra // 7, children)`.
- **Khi nào đạt được cận trên?** Gọi `k0 = min(extra // 7, children)`, `rem = extra - 7 * k0` (tiền dư còn phải chia). Có `children - k0` đứa còn lại, mỗi đứa đã có 1 đô và phải nhận thêm đúng `rem` đô tổng cộng, không ai thành 4 hoặc 8.
  - `k0 == children`: không còn đứa nào để nhận `rem`. `rem == 0` thì xong (ai cũng 8). `rem > 0` thì phải đưa cho một đứa nào đó → nó nhận `> 8`, nên **chỉ `children - 1` đứa đạt đúng 8** (và cách chia `8, ..., 8, 8 + rem` hợp lệ vì `8 + rem >= 9`).
  - `k0 == children - 1`: còn **một** đứa duy nhất, nhận `1 + rem` với `rem < 7` (vì nếu `rem >= 7` thì `k0` còn nâng được thêm một đứa, mâu thuẫn với việc `k0` là min hoặc đã chạm `children`). Hợp lệ ⇔ `1 + rem != 4` ⇔ `rem != 3` (và `1 + rem <= 7` nên không thành 8). Nếu `rem == 3` thì phải hạ `k` xuống `children - 2`.
  - `k0 <= children - 2`: có ít nhất **hai** đứa còn lại chia `rem < 7` đô thêm. Cần một cách chia không tạo `1 + x = 4` (tức `x = 3`) và `1 + x = 8` (tức `x = 7`; không xảy ra vì `rem < 7`). Nếu `rem == 3` chia `1` và `2`; các giá trị khác chia tuỳ ý (đưa hết cho một đứa hoặc chia nhỏ tránh 3). Luôn có cách.
- **Hạ `k0` xuống `k0 - 1` có luôn khả thi trong hai tình huống đặc biệt không?** Tình huống 1 (`k0 == children`, `rem > 0`): cách chia `8, ..., 8, 8 + rem` hợp lệ và có `children - 1` đứa nhận 8. Tình huống 2 (`k0 == children - 1`, `rem == 3`): hạ một đứa từ 8 xuống 7 và đứa còn lại nhận `1 + 3 + 1 = 5` (cách chia `8, ..., 8, 7, 5`) — không ai nhận 4, và có `children - 2` đứa nhận 8.
- **Kiểm chứng bằng máy:** vì miền đầu vào nhỏ (`200 x 29` cặp), mục 12 so sánh công thức với oracle vét cạn trên **mọi** cặp hợp lệ, nên mọi lập luận trên đều được xác nhận.

## 11. Phân tích Time / Space Complexity

Gọi `m = money`, `c = children`.

| Cách | Time | Space |
|---|---|---|
| 1. Vét cạn quy hoạch động | `O(c * m^2)` | `O(c * m)` |
| 2. Duyệt `k` giảm dần + điều kiện khả thi | `O(c)` | `O(1)` |
| 3. Công thức tham lam | `O(1)` | `O(1)` |
| 4. Dựng cụ thể cách chia | `O(c)` | `O(c)` |

Với `m <= 200`, `c <= 30` mọi cách đều chạy tức thì ngoại trừ có thể cảm nhận được độ nặng của Cách 1.

## 12. Bộ test

````python
impls = [
    ("bruteforce", dist_bruteforce),
    ("scan", dist_scan),
    ("greedy", dist_greedy),
    ("construct", lambda m, c: distribute_construct(m, c)[0]),
    ("class_scan", SolutionScan().distMoney),
    ("class_greedy", SolutionGreedy().distMoney),
    ("class_construct", SolutionConstruct().distMoney),
    ("submit", Solution().distMoney),
]
tests = [
    (20, 3, 1),
    (16, 2, 2),
    (3, 4, -1),         # money < children
    (2, 2, 0),          # mỗi đứa đúng 1 đô
    (4, 2, 0),          # 1 + 3 hoặc 2 + 2 (không ai 4)
    (24, 3, 3),         # 8, 8, 8
    (25, 3, 2),         # ai cũng 8 mà dư 1 -> 8, 8, 9
    (17, 2, 1),         # 8, 9
    (12, 2, 0),         # 8 + 4 bị cấm
    (13, 2, 1),         # 8 + 5
    (11, 2, 1),         # 8 + 3
    (200, 30, 24),      # biên trên: 24 đứa nhận 8 (192 đô), 6 đứa còn lại chia 8 đô (1,1,1,1,2,2)
    (200, 2, 1),        # 8 + 192
]
for name, f in impls:
    for money, children, expected in tests:
        got = f(money, children)
        assert got == expected, (name, money, children, got, expected)
    print(name, "PASS", len(tests), "test")
````

Test 3: kiểm chứng **bằng chứng cụ thể**: với mọi cặp hợp lệ mà đáp án khác `-1`, cách chia do `distribute_construct` dựng ra phải hợp lệ (mọi đứa >= 1, không ai nhận 4, tổng bằng `money`) và có đúng `đáp án` đứa nhận 8.

## 13. Các lỗi người mới hay gặp

### Lỗi 1: chia thẳng `money // 8` (quên mỗi đứa phải có ít nhất 1 đô)

````python
def wrong_div8(money, children):
    return min(money // 8, children)

print(wrong_div8(16, 3), "vs đúng", Solution().distMoney(16, 3))
print(wrong_div8(17, 2), "vs đúng", Solution().distMoney(17, 2))
````

Vì sao sai: `money = 16`, `children = 3` — `16 // 8 = 2` nhưng nếu hai đứa nhận 8 thì đứa thứ ba nhận `0` đô (vi phạm "ít nhất 1 đô"). Phải phát 1 đô cho mỗi đứa **trước** rồi mới xét việc nâng lên 8 (cần thêm 7).

### Lỗi 2: quên luật "không ai nhận 4"

````python
def wrong_no_four(money, children):
    if money < children:
        return -1
    money -= children
    k = min(money // 7, children)
    rem = money - 7 * k
    if k == children and rem > 0:           # có xử lý trường hợp ai cũng 8, nhưng THIẾU luật số 4
        k -= 1
    return k

print(wrong_no_four(20, 3), "vs đúng", Solution().distMoney(20, 3))
print(wrong_no_four(12, 2), "vs đúng", Solution().distMoney(12, 2))
````

Vì sao sai: `money = 20`, `children = 3` cho `k = 2` với `rem = 3` — đứa còn lại nhận `1 + 3 = 4` đô, bị cấm. Cần thêm điều kiện `k == children - 1 and rem == 3`.

### Lỗi 3: quên trường hợp "ai cũng 8 mà còn dư tiền"

````python
def wrong_all_eight(money, children):
    if money < children:
        return -1
    money -= children
    k = min(money // 7, children)
    rem = money - 7 * k
    if k == children - 1 and rem == 3:      # chỉ có luật số 4, THIẾU trường hợp k == children
        k -= 1
    return k

print(wrong_all_eight(25, 3), "vs đúng", Solution().distMoney(25, 3))
print(wrong_all_eight(200, 2), "vs đúng", Solution().distMoney(200, 2))
````

Vì sao sai: `money = 25`, `children = 3` — sau khi mỗi đứa có 1 đô còn 22, `k = min(3, 3) = 3`, dư `1`. Số dư này phải đưa cho ai đó, làm đứa đó nhận 9 đô nên chỉ **2** đứa nhận đúng 8. Khi `k == children` và `rem > 0` phải `k -= 1`.

### Lỗi 4: quên trả `-1` khi không đủ tiền cho mỗi đứa 1 đô

````python
def wrong_no_minus_one(money, children):
    money -= children                      # money có thể thành số âm
    k = min(money // 7, children)          # -3 // 7 = -1 (chia nguyên làm tròn xuống)
    rem = money - 7 * k
    if (k == children and rem > 0) or (k == children - 1 and rem == 3):
        k -= 1
    return k

print(wrong_no_minus_one(3, 4), "vs đúng", Solution().distMoney(3, 4), "(trùng hợp)")
print(wrong_no_minus_one(1, 30), "vs đúng", Solution().distMoney(1, 30))
````

Vì sao sai: khi `money < children` không thể cho mỗi đứa ít nhất 1 đô và đề yêu cầu trả `-1`. Với `(3, 4)` đoạn code trên ra `-1` chỉ do trùng hợp phép chia làm tròn xuống (`-1 // 7 = -1`); với `(1, 30)` nó trả `-5`. Hãy xử lý `money < children` **ngay đầu** hàm.

## 14. Mẹo làm bài nhanh & nhận dạng pattern

- "Chia hết, mỗi phần ít nhất X, tối đa hoá số phần đạt giá trị Y" → **phát phần bắt buộc trước**, rồi tham lam nâng từng phần lên Y (cần thêm `Y - X` mỗi phần).
- Khi bài có **luật cấm** (không được nhận 4), hãy tìm **các trường hợp biên** nơi luật làm lời giải tham lam bị phá: ở đây chỉ có 2 (dư làm ai đó thành 4, và ai cũng 8 mà dư).
- Miền input nhỏ (`200 * 30`) → luôn có thể **kiểm tra toàn bộ** bằng vét cạn, thay vì chỉ thử vài test tay.
- Viết thêm hàm `valid_distribution(...)` để kiểm tra cách chia mà thuật toán tạo ra — "chứng minh bằng ví dụ" mỗi test.
- Nhớ `//` làm tròn **xuống** (kể cả với số âm); luôn xử lý điều kiện vô nghiệm (`-1`) trước.

## 15. Biến thể của bài

1. Thay "đúng 8" bằng số `X` bất kỳ và "không ai nhận 4" bằng một **tập số cấm** `{4, 5, ...}` (xem bài tập 5).
2. Hỏi **số cách** chia (đếm) thay vì số đứa tối đa — thành bài quy hoạch động (đếm tổ hợp có ràng buộc).
3. Hỏi **số đứa tối thiểu** nhận đúng 8 (bài ngược lại).
4. Mỗi đứa nhận **tối đa** `M` đô (thêm cận trên) — phải thêm trường hợp "dư quá nhiều" thì `-1`.
5. Các đứa có "độ tuổi" khác nhau nên số tiền tối thiểu mỗi đứa khác nhau.
6. Bài tương tự về tham lam + trường hợp biên: "Candy" (135), "Assign Cookies" (455), "Maximum Number of Coins You Can Get" (1561).

## 16. So sánh các cách làm

| Cách | Ý tưởng | Time | Space | Khi nào dùng |
|---|---|---|---|---|
| 1. Vét cạn quy hoạch động | thử `k`, kiểm tra chia được bằng DP có nhớ | `O(c * m^2)` | `O(c * m)` | oracle kiểm chứng |
| 2. Duyệt `k` + điều kiện khả thi | điều kiện rút gọn `feasible(c, rem)` | `O(c)` | `O(1)` | dễ suy luận, ít sai |
| 3. Công thức tham lam | phát 1 đô, `k = min(extra // 7, c)`, 2 điều kiện | `O(1)` | `O(1)` | cách nộp |
| 4. Dựng cụ thể | tạo `amounts`, chỉnh số 4 | `O(c)` | `O(c)` | cần chứng minh bằng cách chia |

````python
# Bài 1
assert Solution().distMoney(23, 3) == 2 == dist_bruteforce(23, 3)

# Bài 2
assert Solution().distMoney(12, 2) == 0
assert valid_distribution([6, 6], 12) and [6, 6].count(8) == 0
assert not valid_distribution([8, 4], 12)         # 8 + 4 bị cấm vì có số 4

# Bài 3
def valid_distribution_ex_answer(amounts, money):
    return all(a >= 1 for a in amounts) and 4 not in amounts and sum(amounts) == money

assert valid_distribution_ex_answer([8, 9, 3], 20)
assert not valid_distribution_ex_answer([8, 4, 8], 20)
assert not valid_distribution_ex_answer([0, 20], 20)
assert not valid_distribution_ex_answer([8, 9, 3], 21)

# Bài 4
def dist_scan_again(money, children):
    def feasible(c, rem):
        if c == 0:
            return rem == 0
        if rem < c:
            return False
        if c == 1:
            return rem not in (4, 8)
        return True
    for k in range(children, -1, -1):
        rem = money - 8 * k
        if rem >= 0 and feasible(children - k, rem):
            return k
    return -1

for money in range(1, 201):
    for children in range(2, 31):
        assert dist_scan_again(money, children) == Solution().distMoney(money, children)

# Bài 5
def dist_money_forbidden(money, children, forbidden):
    forbidden = set(forbidden)
    assert 8 not in forbidden

    @lru_cache(None)
    def can_split_general(c, m):
        if c == 0:
            return m == 0
        for v in range(1, m - (c - 1) + 1):
            if v in forbidden or v == 8:
                continue
            if can_split_general(c - 1, m - v):
                return True
        return False

    for k in range(children, -1, -1):
        rem = money - 8 * k
        if rem >= 0 and can_split_general(children - k, rem):
            return k
    return -1

# với forbidden = {4} phải trùng bài gốc
for money in range(1, 120, 3):
    for children in range(2, 12):
        assert dist_money_forbidden(money, children, {4}) == Solution().distMoney(money, children)
# thêm luật cấm 5: (8 + 5) không còn dùng được; kiểm tra vài ca tay
assert dist_money_forbidden(13, 2, {4, 5}) == 0     # 8 + 5 bị cấm; 6 + 7 hợp lệ nhưng không ai nhận 8
assert dist_money_forbidden(16, 2, {4, 5}) == 2
print("tất cả đáp án bài tập: PASS")
````
