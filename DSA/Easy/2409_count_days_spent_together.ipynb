## 1. Mục tiêu học

Sau notebook này bạn sẽ:

1. Biến bài toán lịch thành bài **giao của hai đoạn** trên trục số: `[arriveA, leaveA]` và `[arriveB, leaveB]`.

3. Dùng công thức độ dài giao `max(0, min(leaveA, leaveB) - max(arriveA, arriveB) + 1)`, nhớ `+1` vì hai đầu **đều tính**.

## 2. Đề bài & ví dụ

Alice và Bob sắp đến thăm Rome, mỗi người ở một khoảng ngày liên tiếp trong **cùng một năm không nhuận** (365 ngày). Cho bốn chuỗi định dạng `"MM-DD"`:

- `arriveAlice`, `leaveAlice`: ngày Alice đến và đi;
- `arriveBob`, `leaveBob`: ngày Bob đến và đi.

Ngày đến và ngày đi đều **được tính** (inclusive). Hãy trả về **tổng số ngày cả hai cùng ở Rome**.

Số ngày mỗi tháng (năm không nhuận): `[31, 28, 31, 30, 31, 30, 31, 31, 30, 31, 30, 31]`.

| Ví dụ | Alice | Bob | Output |
|---|---|---|---|
| 1 | `"08-15"` → `"08-18"` | `"08-16"` → `"08-19"` | 3 |
| 2 | `"10-01"` → `"10-31"` | `"11-01"` → `"12-31"` | 0 |

Giải thích ví dụ 1: cùng ở ngày 16, 17, 18 tháng 8 → 3 ngày. Ví dụ 2: Alice đi ngày 31/10, Bob đến ngày 1/11 → không giao nhau.

Constraints: tất cả ngày đều có dạng `"MM-DD"` hợp lệ trong năm không nhuận; `arriveAlice <= leaveAlice`; `arriveBob <= leaveBob`.

````python
days_in_month = [31, 28, 31, 30, 31, 30, 31, 31, 30, 31, 30, 31]
print("tổng số ngày trong năm:", sum(days_in_month))
a1, a2, b1, b2 = "08-15", "08-18", "08-16", "08-19"
print("tách chuỗi:", a1.split('-'), "->", [int(x) for x in a1.split('-')])
````

## 3. Kiến thức nền

### 3.1. Tách chuỗi `"MM-DD"`

`"08-15".split('-')` cho `['08', '15']`. Dùng `int` để đổi sang số (`int('08') == 8`, không gặp rắc rối với số 0 đứng đầu). Có thể dùng `map(int, ...)` hoặc gán nhiều biến cùng lúc.

````python
mm, dd = map(int, "08-15".split('-'))
print(mm, dd, type(mm))
print(int("08"), int("015"))
````

### 3.2. Danh sách số ngày và tổng tiền tố

Số ngày **trước** tháng `m` là `sum(days_in_month[:m - 1])` (`m` bắt đầu từ 1, lát cắt bỏ tháng hiện tại). Ví dụ tháng 8: `sum(days[:7])`.

````python
days = [31, 28, 31, 30, 31, 30, 31, 31, 30, 31, 30, 31]
print("trước tháng 8:", sum(days[:7]))
print("trước tháng 1:", sum(days[:0]))
from itertools import accumulate
start = [0] + list(accumulate(days))
print("start =", start)
````

### 3.3. Số thứ tự ngày trong năm

`day_of_year("MM-DD") = (số ngày trước tháng MM) + DD`. Ngày 1/1 là 1, ngày 31/12 là 365. Hai ngày được so sánh dễ dàng bằng số này.

````python
def day_of_year(date):
    m, d = map(int, date.split('-'))
    return sum(days[:m - 1]) + d

for dt in ("01-01", "02-28", "03-01", "08-15", "12-31"):
    print(dt, "->", day_of_year(dt))
````

### 3.4. Giao của hai đoạn `[a, b]` và `[c, d]` (số nguyên, hai đầu gồm cả)

Phần giao bắt đầu ở `max(a, c)` và kết thúc ở `min(b, d)`. Số nguyên trong đoạn `[lo, hi]` là `hi - lo + 1` nếu `lo <= hi`, ngược lại `0`. Công thức: `max(0, min(b, d) - max(a, c) + 1)`.

````python
def overlap(a, b, c, d):
    return max(0, min(b, d) - max(a, c) + 1)

print(overlap(1, 5, 3, 9), overlap(1, 5, 5, 9), overlap(1, 5, 6, 9), overlap(1, 10, 3, 4))
````

## 4. Ý tưởng nền tảng / cách nghĩ trực giác

Hãy tưởng tượng một thước 365 vạch (mỗi vạch là một ngày). Alice tô một đoạn màu đỏ từ ngày đến tới ngày đi, Bob tô một đoạn màu xanh. Số vạch **được tô cả hai màu** là đáp án.

Ví dụ 1 (`08-15..08-18` và `08-16..08-19`): đổi sang số thứ tự ngày (tháng 8 bắt đầu sau 212 ngày): Alice `227..230`, Bob `228..231`.

```
ngày : 227 228 229 230 231
Alice:  █   █   █   █
Bob  :      █   █   █   █
chung:      ✓   ✓   ✓          -> 3 ngày
```

Phần giao bắt đầu ở `max(227, 228) = 228` và kết thúc ở `min(230, 231) = 230` → `230 - 228 + 1 = 3`.

Ví dụ 2: Alice kết thúc ở `10-31 = 304`, Bob bắt đầu ở `11-01 = 305` → `min(leave) = 304`, `max(arrive) = 305` → `304 - 305 + 1 = 0` (vừa chạm, không có ngày chung).

## 5. Cách 1 — Brute Force / baseline

Duyệt **từng ngày** `d = 1..365`; ngày đó tính nếu cả Alice **và** Bob đang ở Rome (`arrive <= d <= leave`). Đếm.

Đây là cách "mô phỏng đúng đề": cực kỳ khó sai. Nút thắt: 365 vòng lặp (hằng số nhỏ) nhưng không cần thiết, vì đáp án có thể tính trực tiếp từ bốn mốc.

````python
def day_of_year(date_str):
    m, d = map(int, date_str.split('-'))
    return sum([31, 28, 31, 30, 31, 30, 31, 31, 30, 31, 30, 31][:m - 1]) + d

class SolutionBrute:
    def countDaysTogether(self, arriveAlice, leaveAlice, arriveBob, leaveBob):
        a1, a2 = day_of_year(arriveAlice), day_of_year(leaveAlice)
        b1, b2 = day_of_year(arriveBob), day_of_year(leaveBob)
        count = 0
        for d in range(1, 366):
            if a1 <= d <= a2 and b1 <= d <= b2:
                count += 1
        return count

print(SolutionBrute().countDaysTogether("08-15", "08-18", "08-16", "08-19"))
print(SolutionBrute().countDaysTogether("10-01", "10-31", "11-01", "12-31"))
````

**Độ phức tạp:** vòng lặp cố định 365 lần → `O(365) = O(1)` (hằng số). Space `O(1)`. Nút thắt: mô phỏng từng ngày dù công thức `O(1)` đủ dùng.

## 6. Ý tưởng thuật toán / bước đột phá

Hai đoạn `[a1, a2]` và `[b1, b2]` có giao là `[max(a1, b1), min(a2, b2)]`. Độ dài giao:

- Nếu `max(a1, b1) <= min(a2, b2)` → `min(a2, b2) - max(a1, b1) + 1`.
- Ngược lại → `0`.

**Quy tắc "nếu… thì…":**

- Nếu một người đi **trước khi** người kia đến (`a2 < b1` hoặc `b2 < a1`) → không giao → `0`.
- Nếu hai đoạn chạm đúng một ngày (`a2 == b1`) → giao `1` ngày (vì hai đầu tính).
- Nếu một đoạn nằm hoàn toàn trong đoạn kia → độ dài của đoạn nhỏ hơn.

Invariant: `lo = max(arrive)` luôn là ngày đầu tiên cả hai cùng có mặt; `hi = min(leave)` luôn là ngày cuối cùng cả hai còn ở lại; mọi ngày chung nằm trong `[lo, hi]`.

````python
a1, a2, b1, b2 = map(day_of_year, ("08-15", "08-18", "08-16", "08-19"))
lo, hi = max(a1, b1), min(a2, b2)
print("lo =", lo, "hi =", hi, "-> số ngày =", max(0, hi - lo + 1))
````

## 7. Các cách tiếp theo

Ta có 4 cách: (1) duyệt từng ngày, (2) giao hai tập hợp, (3) công thức + tiền tố số ngày, (4) `datetime`.

### Cách 2 — Giao hai tập hợp

Tạo `set` các ngày của Alice và của Bob, lấy giao, đếm. Dễ đọc, dựa thẳng vào khái niệm "ngày chung" của đề.

````python
class SolutionSet:
    def countDaysTogether(self, arriveAlice, leaveAlice, arriveBob, leaveBob):
        alice = set(range(day_of_year(arriveAlice), day_of_year(leaveAlice) + 1))
        bob = set(range(day_of_year(arriveBob), day_of_year(leaveBob) + 1))
        return len(alice & bob)

print(SolutionSet().countDaysTogether("08-15", "08-18", "08-16", "08-19"))
print(SolutionSet().countDaysTogether("10-01", "10-31", "11-01", "12-31"))
````

Độ phức tạp Cách 2: Time `O(365)` (tạo hai tập tối đa 365 phần tử), Space `O(365)`. Cả hai đều là hằng số, nhưng bộ nhớ lớn hơn Cách 1.

### Cách 3 — Công thức với tiền tố số ngày (cách chuẩn)

Dựng mảng `start[m]` = số ngày trước tháng `m` một lần. Đổi `MM-DD` sang số bằng `start[mm - 1] + dd`, rồi áp công thức giao `O(1)`.

Mẹo: chuỗi `"MM-DD"` được **đệm số 0**, nên so sánh/`max`/`min` bằng chuỗi cũng cho kết quả đúng thứ tự thời gian; nhưng để tính độ dài vẫn phải đổi sang số.

````python
from itertools import accumulate

class SolutionFormula:
    DAYS = [31, 28, 31, 30, 31, 30, 31, 31, 30, 31, 30, 31]
    START = [0] + list(accumulate(DAYS))          # START[m-1] = số ngày trước tháng m

    def toDay(self, s):
        mm, dd = int(s[:2]), int(s[3:])
        return self.START[mm - 1] + dd

    def countDaysTogether(self, arriveAlice, leaveAlice, arriveBob, leaveBob):
        lo = max(self.toDay(arriveAlice), self.toDay(arriveBob))
        hi = min(self.toDay(leaveAlice), self.toDay(leaveBob))
        return max(0, hi - lo + 1)

print(SolutionFormula().countDaysTogether("08-15", "08-18", "08-16", "08-19"))
print(SolutionFormula().countDaysTogether("10-01", "10-31", "11-01", "12-31"))
````

Độ phức tạp Cách 3: Time `O(1)` (hằng số: tách chuỗi và vài phép so sánh), Space `O(1)` (mảng 13 phần tử cố định).

### Cách 4 — Dùng `datetime.date`

Thư viện chuẩn lo sẵn số ngày của từng tháng. Chọn **năm 2021** (không nhuận) để khớp đề. `date.toordinal()` cho số thứ tự liên tục nên trừ trực tiếp.

````python
from datetime import date

class SolutionDatetime:
    def countDaysTogether(self, arriveAlice, leaveAlice, arriveBob, leaveBob):
        def to_date(s):
            m, d = map(int, s.split('-'))
            return date(2021, m, d).toordinal()          # 2021: năm không nhuận
        lo = max(to_date(arriveAlice), to_date(arriveBob))
        hi = min(to_date(leaveAlice), to_date(leaveBob))
        return max(0, hi - lo + 1)

print(SolutionDatetime().countDaysTogether("08-15", "08-18", "08-16", "08-19"))
print(SolutionDatetime().countDaysTogether("10-01", "10-31", "11-01", "12-31"))
````

Độ phức tạp Cách 4: Time `O(1)`, Space `O(1)`. Chú ý: **không** dùng năm nhuận như 2020 vì `date(2020, 2, 29)` hợp lệ còn `02-29` không xuất hiện trong đề; và `date(2021, 2, 29)` sẽ lỗi — nhưng đề đảm bảo ngày hợp lệ.

## 8. TRACE từng bước

Trace Cách 3 với `("08-15", "08-18", "08-16", "08-19")`:

| bước | giá trị |
|---|---|
| `toDay("08-15")` | `START[7] + 15 = 212 + 15 = 227` |
| `toDay("08-18")` | `212 + 18 = 230` |
| `toDay("08-16")` | `228` |
| `toDay("08-19")` | `231` |
| `lo = max(227, 228)` | `228` |
| `hi = min(230, 231)` | `230` |
| `hi - lo + 1` | `3` |

````python
def trace_formula(a1s, a2s, b1s, b2s):
    DAYS = [31, 28, 31, 30, 31, 30, 31, 31, 30, 31, 30, 31]
    START = [0]
    for dm in DAYS:
        START.append(START[-1] + dm)
    print("START =", START)
    def to_day(s):
        mm, dd = int(s[:2]), int(s[3:])
        v = START[mm - 1] + dd
        print(f"  {s} -> START[{mm - 1}] + {dd} = {START[mm - 1]} + {dd} = {v}")
        return v
    a1, a2, b1, b2 = to_day(a1s), to_day(a2s), to_day(b1s), to_day(b2s)
    lo, hi = max(a1, b1), min(a2, b2)
    print(f"lo = max({a1}, {b1}) = {lo}; hi = min({a2}, {b2}) = {hi}")
    ans = max(0, hi - lo + 1)
    print("đáp án = max(0, hi - lo + 1) =", ans)
    return ans

trace_formula("08-15", "08-18", "08-16", "08-19")
trace_formula("10-01", "10-31", "11-01", "12-31")
````

## 9. Code nộp LeetCode

````python
from typing import List, Optional
from itertools import accumulate

class Solution:
    def countDaysTogether(self, arriveAlice: str, leaveAlice: str, arriveBob: str, leaveBob: str) -> int:
        days = [31, 28, 31, 30, 31, 30, 31, 31, 30, 31, 30, 31]
        start = [0] + list(accumulate(days))

        def to_day(s: str) -> int:
            return start[int(s[:2]) - 1] + int(s[3:])

        lo = max(to_day(arriveAlice), to_day(arriveBob))
        hi = min(to_day(leaveAlice), to_day(leaveBob))
        return max(0, hi - lo + 1)

print(Solution().countDaysTogether("08-15", "08-18", "08-16", "08-19"),
      Solution().countDaysTogether("10-01", "10-31", "11-01", "12-31"))
````

## 10. Vì sao thuật toán đúng

- **Đổi ngày sang số là đơn điệu:** ngày sau trong năm luôn có số thứ tự lớn hơn, nên so sánh số tương đương so sánh ngày.
- **Giao hai đoạn:** một ngày `d` thuộc cả hai đoạn ⇔ `d >= a1`, `d >= b1`, `d <= a2`, `d <= b2` ⇔ `max(a1, b1) <= d <= min(a2, b2)`. Vậy tập ngày chung chính là đoạn `[lo, hi]`.
- **Đếm số nguyên trong đoạn:** `[lo, hi]` có `hi - lo + 1` số nguyên nếu `lo <= hi`; nếu `lo > hi` đoạn rỗng và công thức cho giá trị `<= 0`, nên `max(0, ...)` sửa thành `0`.
- **Đầu mút tính cả:** đề nói ngày đến và ngày đi đều tính, nên ta dùng `+ 1`.
- **Khớp với brute:** Cách 1 chính là đếm trực tiếp tập `{d : a1<=d<=a2, b1<=d<=b2}`, và Cách 3 tính cùng tập đó bằng công thức.

## 11. Phân tích Time / Space Complexity

Gọi số ngày trong năm là `D = 365` (hằng số).

| Cách | Time | Space |
|---|---|---|
| 1. Duyệt từng ngày | `O(D)` | `O(1)` |
| 2. Giao hai tập | `O(D)` | `O(D)` |
| 3. Công thức + tiền tố | `O(1)` | `O(1)` |
| 4. `datetime` | `O(1)` | `O(1)` |

Vì `D` cố định, mọi cách đều là hằng số; điều quan trọng là **công thức gọn và ít chỗ sai**. Nếu đề mở rộng sang nhiều năm, Cách 1/2 trở thành `O(độ dài khoảng)` còn cách 3/4 vẫn `O(1)`. Không có worst/average khác nhau.

## 12. Bộ test

Test 1: ví dụ đề, biên của tháng/năm và các trường hợp chạm đầu mút.

````python
impls = [SolutionBrute(), SolutionSet(), SolutionFormula(), SolutionDatetime(), Solution()]
tests = [
    (("08-15", "08-18", "08-16", "08-19"), 3),
    (("10-01", "10-31", "11-01", "12-31"), 0),
    (("01-01", "12-31", "01-01", "12-31"), 365),     # cả năm
    (("01-01", "01-01", "01-01", "01-01"), 1),       # một ngày trùng
    (("02-28", "03-01", "02-28", "03-01"), 2),       # qua ranh giới tháng 2 (28 ngày)
    (("12-31", "12-31", "12-31", "12-31"), 1),       # cuối năm
    (("03-10", "03-20", "03-20", "03-30"), 1),       # chạm đúng 1 đầu mút
    (("03-10", "03-20", "03-21", "03-30"), 0),       # sát nhau nhưng không chạm
    (("05-01", "09-30", "06-15", "06-20"), 6),       # đoạn này lồng trong đoạn kia
    (("01-01", "01-10", "03-01", "03-05"), 0),       # xa nhau (bắt lỗi thiếu max(0, ...))
    (("11-30", "12-02", "12-01", "12-05"), 2),
]
for args, exp in tests:
    for im in impls:
        got = im.countDaysTogether(*args)
        assert got == exp, (type(im).__name__, args, got, exp)
print("tất cả test cố định OK:", len(tests))
````

Test 2: ngẫu nhiên có seed. Ngày được tạo hợp lệ trong năm không nhuận bằng cách chọn số thứ tự `1..365` rồi đổi lại `MM-DD` bằng `datetime`, bảo đảm `arrive <= leave`. Oracle độc lập: đếm từng ngày trong khoảng.

````python
import random
from datetime import date, timedelta
random.seed(2409)

def to_mmdd(k):                                   # k trong 1..365
    d = date(2021, 1, 1) + timedelta(days=k - 1)
    return f"{d.month:02d}-{d.day:02d}"

def oracle(a1, a2, b1, b2):
    return sum(1 for k in range(1, 366) if a1 <= k <= a2 and b1 <= k <= b2)

zero = 0
for _ in range(5000):
    a1, a2 = sorted(random.sample(range(1, 366), 2)) if random.random() < 0.9 else (random.randint(1, 365),) * 2
    b1, b2 = sorted(random.sample(range(1, 366), 2)) if random.random() < 0.9 else (random.randint(1, 365),) * 2
    args = (to_mmdd(a1), to_mmdd(a2), to_mmdd(b1), to_mmdd(b2))
    assert args[0] <= args[1] and args[2] <= args[3]          # ràng buộc arrive <= leave (so sánh chuỗi)
    exp = oracle(a1, a2, b1, b2)
    zero += exp == 0
    for im in impls:
        assert im.countDaysTogether(*args) == exp, (type(im).__name__, args, exp)
print("5000 test ngẫu nhiên OK; số test đáp án 0:", zero)
assert 0 < zero < 5000
````

Test 3: kiểm tra tính chất — đối xứng (đổi vai Alice và Bob không đổi đáp án) và khi hai đoạn giống nhau thì đáp án bằng độ dài đoạn.

````python
random.seed(11)
for _ in range(500):
    a1, a2 = sorted(random.sample(range(1, 366), 2))
    b1, b2 = sorted(random.sample(range(1, 366), 2))
    A = (to_mmdd(a1), to_mmdd(a2)); B = (to_mmdd(b1), to_mmdd(b2))
    assert Solution().countDaysTogether(*A, *B) == Solution().countDaysTogether(*B, *A)
    assert Solution().countDaysTogether(*A, *A) == a2 - a1 + 1
print("tính đối xứng và tự giao: OK")
````

## 13. Các lỗi người mới hay gặp

### Lỗi 1: coi mọi tháng có 30 ngày

````python
def wrong_30(arriveAlice, leaveAlice, arriveBob, leaveBob):
    def f(s):
        m, d = map(int, s.split('-'))
        return m * 30 + d                      # SAI: tháng không đều 30 ngày
    lo = max(f(arriveAlice), f(arriveBob))
    hi = min(f(leaveAlice), f(leaveBob))
    return max(0, hi - lo + 1)

args = ("02-28", "03-01", "02-28", "03-01")
print("sai :", wrong_30(*args))
print("đúng:", Solution().countDaysTogether(*args))
assert wrong_30(*args) != Solution().countDaysTogether(*args)
````

Vì sao sai: tháng 2 chỉ có 28 ngày; từ `02-28` sang `03-01` chỉ có 2 ngày (28/2 và 1/3), còn công thức `m*30 + d` cho `88 → 91` tức 4 ngày. **Sửa:** dùng mảng số ngày thực của từng tháng.

### Lỗi 2: quên `+ 1` (ngày đến và ngày đi đều tính)

````python
def wrong_no_plus1(arriveAlice, leaveAlice, arriveBob, leaveBob):
    lo = max(day_of_year(arriveAlice), day_of_year(arriveBob))
    hi = min(day_of_year(leaveAlice), day_of_year(leaveBob))
    return max(0, hi - lo)                     # SAI: thiếu + 1

args = ("08-15", "08-18", "08-16", "08-19")
print("sai :", wrong_no_plus1(*args))
print("đúng:", Solution().countDaysTogether(*args))
assert wrong_no_plus1(*args) != Solution().countDaysTogether(*args)
````

Vì sao sai: `hi - lo` là **khoảng cách** giữa hai mốc, còn số ngày gồm cả hai đầu là `hi - lo + 1` (giống đếm cọc và khoảng hàng rào). Hệ quả: hai người chỉ trùng một ngày bị tính `0`.

### Lỗi 3: quên `max(0, ...)` khi hai đoạn không giao nhau

````python
def wrong_negative(arriveAlice, leaveAlice, arriveBob, leaveBob):
    lo = max(day_of_year(arriveAlice), day_of_year(arriveBob))
    hi = min(day_of_year(leaveAlice), day_of_year(leaveBob))
    return hi - lo + 1                         # SAI: có thể âm

args = ("01-01", "01-10", "03-01", "03-05")
print("sai :", wrong_negative(*args))
print("đúng:", Solution().countDaysTogether(*args))
assert wrong_negative(*args) != Solution().countDaysTogether(*args)
````

Vì sao sai: khi `hi < lo` công thức cho số âm (ở đây `-49`), nhưng số ngày chung không thể âm. **Sửa:** bọc `max(0, ...)`.

### Lỗi 4 (bonus): so sánh chuỗi khi chưa đệm số 0

`"08-15"` và `"8-15"` không còn so sánh đúng bằng chuỗi (`"8-15" > "08-15"` đúng nhưng `"8-15" > "10-01"` lại sai theo thời gian). Đề đảm bảo `MM-DD` đệm số 0 nên so sánh chuỗi dùng được, nhưng vẫn nên đổi sang số để tính độ dài.

## 14. Mẹo làm bài nhanh & nhận dạng pattern

- "Số ngày cả hai cùng ở lại / số đơn vị chung" → **giao hai đoạn**: `max(0, min(đuôi) - max(đầu) + 1)`.
- Bài lịch không xuất hiện năm nhuận thì dùng sẵn mảng 12 số ngày và `accumulate` để có tiền tố; chạy nhanh và không phụ thuộc thư viện.
- `datetime` tiện nhưng nhớ chọn **năm không nhuận** (ví dụ 2021) để đúng đề.
- Với đoạn đóng `[a, b]` luôn nhớ `b - a + 1` (đếm cả hai đầu); với đoạn nửa mở `[a, b)` mới là `b - a`.
- Thử ngay hai trường hợp biên: chạm một ngày (`a2 == b1`) và sát nhau (`a2 + 1 == b1`).
- Có thể kiểm tra nhanh bằng brute 365 ngày (oracle) trước khi nộp.

## 15. Biến thể của bài

1. Cho **năm nhuận** (có `02-29`): chỉ cần đổi `28 → 29`.
2. Khoảng thời gian trải qua **nhiều năm**: dùng `date.toordinal()` với năm đầy đủ.
3. Nhiều người (3 người trở lên): ngày chung = giao của tất cả → `max` đầu và `min` đuôi.
4. Tính số ngày **chỉ một trong hai** có mặt (đối xứng hiệu): `(len A) + (len B) - 2 * giao`.
5. Hợp nhất lịch/Meeting rooms: kiểm tra giao của hai khoảng (xem bài 2446).

````python
# Bài 1: Alice 06-01..06-30, Bob 06-15..07-15 -> giao 06-15..06-30 = 16 ngày
assert Solution().countDaysTogether("06-01", "06-30", "06-15", "07-15") == 16

# Bài 2: 03-25 = 59+25 = 84 ; 04-05 = 90+5 = 95 ; 04-01 = 91 ; 04-20 = 110
assert (day_of_year("03-25"), day_of_year("04-05"), day_of_year("04-01"), day_of_year("04-20")) == (84, 95, 91, 110)
assert Solution().countDaysTogether("03-25", "04-05", "04-01", "04-20") == 5

# Bài 3
def interval_overlap(a, b, c, d):
    return max(0, min(b, d) - max(a, c) + 1)
assert interval_overlap(1, 5, 3, 9) == 3
assert interval_overlap(1, 5, 6, 9) == 0

# Bài 4
def days_between(d1, d2):
    return day_of_year(d2) - day_of_year(d1)
assert days_between("01-01", "12-31") == 364
assert days_between("08-15", "08-15") == 0

# Bài 5
def days_all_three(*intervals):
    lo = max(day_of_year(a) for a, _ in intervals)
    hi = min(day_of_year(l) for _, l in intervals)
    return max(0, hi - lo + 1)
assert days_all_three(("01-01", "01-31"), ("01-10", "02-10"), ("01-20", "03-01")) == 12
assert days_all_three(("01-01", "01-05"), ("01-03", "01-10")) == 3
assert days_all_three(("01-01", "01-05"), ("01-03", "01-10"), ("01-08", "01-09")) == 0
print("đáp án bài tập OK")
````
