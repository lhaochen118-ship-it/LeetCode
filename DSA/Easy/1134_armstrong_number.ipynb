## 1. Đề bài — `n` có phải số Armstrong?

Cho số nguyên `n`. Trả về `True` nếu và chỉ nếu `n` là **số Armstrong**.

Số nguyên `k` chữ số `n` là số Armstrong khi **tổng lũy thừa bậc `k` của từng chữ số bằng `n`**.

| n | k | Tính | Kết quả |
|---:|---:|---|:---:|
| 153 | 3 | `1³ + 5³ + 3³ = 1 + 125 + 27 = 153` | `True` |
| 123 | 3 | `1³ + 2³ + 3³ = 1 + 8 + 27 = 36` | `False` |
| 9474 | 4 | `9⁴ + 4⁴ + 7⁴ + 4⁴ = 6561 + 256 + 2401 + 256 = 9474` | `True` |
| 7 | 1 | `7¹ = 7` | `True` |

**Constraints (theo đề gốc):** `1 <= n <= 10^8`.

Chú ý: số mũ là **số chữ số của `n`**, không cố định bằng 3.

````python
print("153 :", 1**3 + 5**3 + 3**3)
print("123 :", 1**3 + 2**3 + 3**3)
print("9474:", 9**4 + 4**4 + 7**4 + 4**4)
````

### 2.3. Tách chữ số bằng toán: `% 10` và `// 10`

- `n % 10` → chữ số **cuối** (hàng đơn vị).
- `n // 10` → bỏ chữ số cuối.

Lặp đến khi `n == 0` sẽ lấy được mọi chữ số (theo thứ tự **từ phải sang trái**).

````python
n = 153
while n > 0:
    print(f"n={n:>3}: chữ số cuối = n % 10 = {n % 10}, bỏ đi -> n // 10 = {n // 10}")
    n //= 10
````

### 2.4. Đếm số chữ số — 3 cách

1. `len(str(n))` — ngắn nhất.
2. Vòng `while`: đếm số lần chia 10 tới khi về 0.
3. `math.log10(n)`: số chữ số của `n >= 1` là `floor(log10(n)) + 1`. Đây là **số thực** — với `n` rất lớn có thể sai do làm tròn, và `log10(0)` lỗi. Trong miền `n <= 10^8` thì an toàn, nhưng nên biết cạm bẫy.

````python
import math

def digits_while(n):
    k = 0
    while n > 0:
        k += 1
        n //= 10
    return k

for n in [1, 9, 10, 153, 9474, 10**8]:
    print(f"n={n:>9}: len(str)={len(str(n))}, while={digits_while(n)}, log10={math.floor(math.log10(n)) + 1}")

big = 10**15 - 1             # 999999999999999 có 15 chữ số
print("cạm bẫy số thực:", len(str(big)), "vs", math.floor(math.log10(big)) + 1)
````

`10**15 - 1` có 15 chữ số, nhưng `log10` của nó (dạng số thực) bị làm tròn thành `15.0` nên công thức cho **16** — sai. Bài học: với số nguyên, ưu tiên cách đếm bằng số nguyên.

## 3. Cách 1 — Dùng chuỗi (trực giác)

Theo đúng định nghĩa: tách chữ số, lấy mũ `k`, cộng, so sánh.

````python
def is_armstrong_string(n: int) -> bool:
    s = str(n)
    k = len(s)
    return sum(int(ch) ** k for ch in s) == n

for n in [153, 123, 9474, 7, 10, 370, 371, 407]:
    print(n, "->", is_armstrong_string(n))
````

**Độ phức tạp:** gọi `d` = số chữ số (`d <= 9`). Time `O(d)` lần mũ (mỗi lần mũ số nhỏ coi như hằng số), Space `O(d)` cho chuỗi. Với `n <= 10^8` thì gần như tức thì.

Đây là bài **không có nút thắt hiệu năng**; điểm học là **xử lý chữ số không qua chuỗi** (kỹ năng cần cho nhiều bài khác) và **tư duy toán** về giới hạn của bài.

## 4. Insight — tách chữ số bằng toán, và cần `k` **trước khi** cộng

Trở ngại khi làm thuần toán: số mũ `k` phụ thuộc **số chữ số**, mà ta chỉ biết khi đã tách hết chữ số. Hai lựa chọn:
1. **Hai lượt:** lượt 1 đếm `k`, lượt 2 tính tổng.
2. **Một lượt:** tách chữ số vào list, `k = len(list)`, rồi tính tổng.

Đồng thời phải **giữ lại `n` gốc** vì vòng lặp sẽ biến `n` về 0.

## 5. Cách 2 — Thuần toán, hai lượt (lời giải chính)

````python
def is_armstrong_math(n: int) -> bool:
    # Lượt 1: đếm số chữ số
    k, x = 0, n
    while x > 0:
        k += 1
        x //= 10
    # Lượt 2: cộng lũy thừa bậc k của từng chữ số
    total, x = 0, n
    while x > 0:
        total += (x % 10) ** k
        x //= 10
    return total == n

for n in [153, 123, 9474, 7, 10, 370, 371, 407]:
    print(n, "->", is_armstrong_math(n))
````

Biến `x` là **bản sao làm việc**: ta chia nhỏ `x`, còn `n` giữ nguyên để so sánh ở cuối.

### 5.1. Dừng sớm (tối ưu nhỏ)

Nếu `total` đã **vượt** `n` thì chắc chắn không bằng → trả `False` ngay. Không đổi Big-O nhưng là thói quen tốt.

````python
def is_armstrong_early_exit(n: int) -> bool:
    k = len(str(n))
    total, x = 0, n
    while x > 0:
        total += (x % 10) ** k
        if total > n:
            return False
        x //= 10
    return total == n

print(is_armstrong_early_exit(9474), is_armstrong_early_exit(9999))
````

## 6. TRACE từng bước

`n = 9474`, `k = 4`. Chữ số được lấy từ phải sang trái:

| Bước | x | chữ số `x % 10` | chữ số^4 | total | x sau `//= 10` |
|---:|---:|---:|---:|---:|---:|
| 1 | 9474 | 4 | 256 | 256 | 947 |
| 2 | 947 | 7 | 2401 | 2657 | 94 |
| 3 | 94 | 4 | 256 | 2913 | 9 |
| 4 | 9 | 9 | 6561 | 9474 | 0 |

`total == n` → `True`.

````python
def trace_armstrong(n: int) -> bool:
    k = digits_while(n)
    print(f"n = {n}, số chữ số k = {k}")
    total, x, step = 0, n, 1
    while x > 0:
        d = x % 10
        total += d ** k
        print(f"Bước {step}: x={x:>5}, chữ số={d}, {d}^{k}={d ** k:>5}, total={total:>5}, x sau={x // 10}")
        x //= 10
        step += 1
    print(f"So sánh: total={total} == n={n}? {total == n}")
    return total == n

_ = trace_armstrong(9474)
print()
_ = trace_armstrong(123)
````

## 7. Code nộp LeetCode

Bản thuần toán:

````python
class Solution:
    def isArmstrong(self, n: int) -> bool:
        k, x = 0, n
        while x:
            k += 1
            x //= 10
        total, x = 0, n
        while x:
            total += (x % 10) ** k
            x //= 10
        return total == n

sol = Solution()
print(sol.isArmstrong(153), sol.isArmstrong(123))
````

`while x:` tương đương `while x != 0:` vì số `0` được coi là **sai** (falsy), số khác 0 là **đúng** (truthy).

Bản chuỗi (một dòng):

````python
class SolutionString:
    def isArmstrong(self, n: int) -> bool:
        s = str(n)
        return sum(int(c) ** len(s) for c in s) == n

print(SolutionString().isArmstrong(153), SolutionString().isArmstrong(123))
````

## 8. Cách 3 — Góc nhìn toán: chỉ có hữu hạn số Armstrong → tính trước

Với số `k` chữ số: tổng lớn nhất có thể là `k · 9^k`, còn số nhỏ nhất `k` chữ số là `10^(k-1)`. Khi `k` đủ lớn, `k · 9^k < 10^(k-1)` → **không** số `k` chữ số nào có thể là Armstrong. (Điều này xảy ra từ `k = 61`, nên số Armstrong trong hệ thập phân là **hữu hạn** — thực tế số lớn nhất chỉ có 39 chữ số.)

Trong miền đề bài (`n <= 10^8`, tối đa 9 chữ số), ta có thể **liệt kê trước** mọi số Armstrong rồi chỉ việc tra `n in set`. Liệt kê bằng vét cạn `10^8` số thì quá chậm trong Python, nhưng có mẹo: tổng `Σ d^k` **chỉ phụ thuộc vào tập (multiset) chữ số**, không phụ thuộc thứ tự. Ta duyệt mọi multiset chữ số bằng `itertools.combinations_with_replacement` (với `k = 9` chỉ có `C(18, 9) = 48620` multiset), tính tổng, rồi kiểm tra tổng đó có đúng là số tạo bởi chính multiset đó không.

````python
for k in [1, 3, 9, 20, 60, 61]:
    print(f"k={k:>2}: k*9^k = {k * 9**k:.3e}  vs  10^(k-1) = {10**(k - 1):.3e}  -> còn khả năng? {k * 9**k >= 10**(k - 1)}")
````

````python
from itertools import combinations_with_replacement

def all_armstrong_up_to_digits(max_digits: int) -> list:
    found = set()
    for k in range(1, max_digits + 1):
        for combo in combinations_with_replacement(range(10), k):
            total = sum(d ** k for d in combo)
            # total phải có đúng k chữ số và multiset chữ số của total phải bằng combo
            if len(str(total)) == k and sorted(int(c) for c in str(total)) == list(combo):
                found.add(total)
    found.discard(0)                 # đề cho n >= 1
    return sorted(found)

ARMSTRONG = all_armstrong_up_to_digits(9)
print(len(ARMSTRONG), "số Armstrong có tối đa 9 chữ số:")
print(ARMSTRONG)
````

`combinations_with_replacement(range(10), 3)` sinh các bộ không giảm như `(0,0,0), (0,0,1), …, (1,3,5), …` — mỗi multiset đúng một lần. Ví dụ bộ `(1,3,5)` cho tổng `1+27+125 = 153`, và chữ số của 153 sắp xếp lại là `[1,3,5]` → khớp → 153 là số Armstrong.

````python
ARMSTRONG_SET = set(x for x in ARMSTRONG if x <= 10**8)

class SolutionPrecomputed:
    def isArmstrong(self, n: int) -> bool:
        return n in ARMSTRONG_SET

print(SolutionPrecomputed().isArmstrong(153), SolutionPrecomputed().isArmstrong(123))
print("Kiểm tra chéo:", all(is_armstrong_math(x) for x in ARMSTRONG))
````

Trên LeetCode, có thể dán thẳng list hằng số này vào code. Đây là ví dụ của kỹ thuật **tiền xử lý (precompute) khi miền đáp án nhỏ**. Tuy vậy, người phỏng vấn thường muốn thấy cách 2.

## 9. Vì sao đúng? — Độ phức tạp — So sánh

**Tính đúng (cách 2):** lượt 1 đếm đúng `k` vì mỗi phép `//= 10` bỏ đúng một chữ số và dừng khi hết chữ số (`n >= 1` nên `k >= 1`). Invariant lượt 2: sau mỗi vòng, `total` = tổng `d^k` của các chữ số **đã bóc** (các chữ số cuối của `n`), còn `x` chứa các chữ số **chưa bóc**. Khi `x = 0`, mọi chữ số đã được cộng đúng một lần → `total` chính là tổng Armstrong của `n`. So sánh với `n` gốc (không bị sửa) cho đáp án.

**Tính đúng (cách 3):** mọi số Armstrong `m` có `k` chữ số đều tạo ra từ đúng multiset chữ số của nó, nên sẽ được tìm thấy khi duyệt multiset đó; ngược lại điều kiện kiểm tra bảo đảm chỉ thêm số thật sự là Armstrong.

Gọi `d` = số chữ số của `n` (`d <= 9`).

| Cách | Time | Space |
|---|---|---|
| 1. Chuỗi | `O(d)` | `O(d)` |
| 2. Toán, 2 lượt | `O(d)` | `O(1)` |
| 3. Tính trước + set | `O(1)` mỗi truy vấn (tiền xử lý một lần ~`Σ C(k+9, k)` multiset) | `O(1)` (31 số tối đa 9 chữ số) |

## 11. Lỗi người mới hay gặp

### Lỗi 1: Luôn dùng mũ 3

````python
def wrong_cube(n):
    return sum(int(c) ** 3 for c in str(n)) == n     # SAI: mũ phải là số chữ số

for n in [9474, 1, 153]:
    print(n, "sai:", wrong_cube(n), "| đúng:", is_armstrong_math(n))
````

**Vì sao sai:** chỉ trùng hợp đúng với số 3 chữ số. `9474` (4 chữ số) phải dùng mũ 4; `1` (1 chữ số) mũ 1 — mà `1³ = 1` nên tình cờ đúng, rất dễ đánh lừa khi test sơ sài.

### Lỗi 2: Dùng `^` thay cho `**`

````python
def wrong_xor(n):
    k = len(str(n))
    return sum(int(c) ^ k for c in str(n)) == n      # SAI: ^ là XOR

print("153 sai:", wrong_xor(153), "| 4^3 =", 4 ^ 3, "| 4**3 =", 4 ** 3)
````

### Lỗi 3: Phá hủy `n` rồi so sánh với chính nó

````python
def wrong_destroy_n(n):
    k = len(str(n))
    total = 0
    while n > 0:
        total += (n % 10) ** k
        n //= 10
    return total == n                                # SAI: n đã thành 0

print("153 sai:", wrong_destroy_n(153), "| đúng:", is_armstrong_math(153))
````

**Sửa:** làm việc trên bản sao `x = n`, hoặc lưu `original = n` trước vòng lặp.

### Lỗi 4: Đếm chữ số sau khi đã chia `n` về 0

````python
def wrong_count_after(n):
    digits = []
    while n > 0:
        digits.append(n % 10)
        n //= 10
    k = len(str(n))                                  # SAI: n giờ là 0 -> k = 1
    return sum(d ** k for d in digits)

print("tổng của 153 (sai):", wrong_count_after(153), "| đúng phải là 153")
````

**Sửa:** `k = len(digits)` (đếm số phần tử đã tách) hoặc tính `k` trước vòng lặp.

## 13. Biến thể

### 13.1. #1281 Subtract the Product and Sum of Digits

````python
def subtract_product_and_sum(n: int) -> int:
    prod, total = 1, 0
    while n:
        d = n % 10
        prod *= d
        total += d
        n //= 10
    return prod - total

print(subtract_product_and_sum(234), subtract_product_and_sum(4421))   # 15 21
````

### 13.2. #202 Happy Number — lặp phép "tổng bình phương chữ số"

Một số "vui" nếu lặp phép biến đổi `n → Σ d²` cuối cùng về `1`. Nếu không, dãy sẽ rơi vào **chu trình** — phát hiện bằng `set` các giá trị đã gặp. (Lập luận chặn trên giống mục 8 cho thấy dãy luôn bị giới hạn nên chắc chắn lặp lại.)

````python
def digit_square_sum(n: int) -> int:
    total = 0
    while n:
        total += (n % 10) ** 2
        n //= 10
    return total

def is_happy(n: int) -> bool:
    seen = set()
    while n != 1 and n not in seen:
        seen.add(n)
        n = digit_square_sum(n)
    return n == 1

print(is_happy(19), is_happy(2))      # True False
````

### 13.3. Liệt kê số Armstrong trong một khoảng

````python
def armstrong_in_range(lo: int, hi: int) -> list:
    return [x for x in range(lo, hi + 1) if is_armstrong_math(x)]

print(armstrong_in_range(100, 999))   # [153, 370, 371, 407]
````
