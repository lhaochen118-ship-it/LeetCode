## 4. Ý tưởng nền tảng / cách nghĩ trực giác

Đặt hai ngón tay: một ở đầu chuỗi (`i`), một ở cuối (`n - 1 - i`) rồi cho cả hai tiến vào giữa. Ở mỗi bước, so hai ký tự: nếu bằng nhau thì `i` là đáp án (vì ta đi từ ngoài vào, `i` tăng dần nên gặp đầu tiên là nhỏ nhất).

Tính tay `"abcacbd"` (`n = 7`):

| i | s[i] | n-1-i | s[n-1-i] | bằng nhau? |
|---|---|---|---|---|
| 0 | a | 6 | d | không |
| 1 | b | 5 | b | **có** -> đáp án 1 |

Tính tay `"abcdab"` (`n = 6`): `(0,5)`: a-b, `(1,4)`: b-a, `(2,3)`: c-d, rồi `(3,2)`, `(4,1)`, `(5,0)` là lặp lại các cặp đã xét (đối xứng) nên đều không bằng nhau -> `-1`.

Ý quan trọng: nếu `s[i] == s[n-1-i]` thì cũng có `s[j] == s[n-1-j]` với `j = n-1-i`. Do đó cặp khớp luôn xuất hiện **hai lần** (ở `i` và `n-1-i`) trừ vị trí giữa; chỉ số nhỏ hơn nằm ở **nửa đầu**.

````python
def explain(s):
    n = len(s)
    for i in range(n):
        j = n - 1 - i
        eq = s[i] == s[j]
        print(f"  i={i} j={j}: {s[i]!r} vs {s[j]!r} -> {eq}")
        if eq:
            return i
    return -1

print(explain("abcacbd"))
print(explain("abc"))
print(explain("abcdab"))
````

## 5. Cách 1 — Brute Force / baseline

Duyệt `i` từ `0` đến `n - 1`, so `s[i]` với `s[n - i - 1]`; gặp bằng nhau đầu tiên thì trả `i`. Hết vòng lặp trả `-1`.

Cách này đúng và chạy `O(n)`. "Nút thắt" duy nhất: ta duyệt cả **nửa sau** vô ích (các cặp đã xét lặp lại), nên số bước có thể giảm một nửa; đó là Cách 3.

````python
class SolutionBrute:
    def firstMatchingIndex(self, s):
        n = len(s)
        for i in range(n):
            if s[i] == s[n - i - 1]:
                return i
        return -1

print(SolutionBrute().firstMatchingIndex("abcacbd"), SolutionBrute().firstMatchingIndex("abc"), SolutionBrute().firstMatchingIndex("abcdab"))
````

**Độ phức tạp:** Time `O(n)`, Space `O(1)`.

## 6. Ý tưởng thuật toán / bước đột phá

**Bước đột phá:** tính đối xứng. Nếu `s[i] == s[n-1-i]` với `i > (n-1)/2` thì đặt `j = n-1-i < (n-1)/2`: ta cũng có `s[j] == s[n-1-j]`, và `j < i`. Vậy **chỉ số nhỏ nhất luôn nằm trong nửa đầu** (gồm cả ký tự giữa khi `n` lẻ), tức `i <= (n - 1) // 2`... cụ thể `i in range((n + 1) // 2)`. Không cần kiểm tra nửa sau.

**Invariant** của vòng lặp hai con trỏ `l = 0, r = n - 1`: mọi cặp `(k, n-1-k)` với `k < l` đã được xét và **không khớp**. Khi `l == r` (tâm của chuỗi lẻ) thì khớp chắc chắn. Khi `l > r` thì đã xét hết, trả `-1`.

Quy tắc: nếu `s[l] == s[r]` trả `l`; ngược lại `l += 1`, `r -= 1`.

Hệ quả: nếu `n` **lẻ** thì đáp án luôn tồn tại (nhiều nhất là `(n - 1) / 2`, ký tự giữa); chỉ khi `n` **chẵn** mới có thể trả `-1`.

## 7. Các cách tiếp theo

### Cách 2 — So với chuỗi đảo ngược bằng `zip`

Ghép `s` với `s[::-1]`, tìm vị trí đầu tiên hai ký tự bằng nhau.

````python
class SolutionZip:
    def firstMatchingIndex(self, s):
        for i, (a, b) in enumerate(zip(s, s[::-1])):
            if a == b:
                return i
        return -1

print(SolutionZip().firstMatchingIndex("abcacbd"), SolutionZip().firstMatchingIndex("abc"), SolutionZip().firstMatchingIndex("abcdab"))
````

**Độ phức tạp:** Time `O(n)`, Space `O(n)` cho chuỗi đảo.

### Cách 3 — Chỉ duyệt nửa đầu

Vòng lặp `for i in range((n + 1) // 2)` (nửa đầu cộng ký tự giữa). Nếu chạy hết mà chưa thấy thì `-1`. Số bước bằng một nửa.

````python
class SolutionHalf:
    def firstMatchingIndex(self, s):
        n = len(s)
        for i in range((n + 1) // 2):
            if s[i] == s[n - 1 - i]:
                return i
        return -1

print(SolutionHalf().firstMatchingIndex("abcacbd"), SolutionHalf().firstMatchingIndex("abc"), SolutionHalf().firstMatchingIndex("abcdab"))
````

**Độ phức tạp:** Time `O(n)` (hằng số nhỏ hơn), Space `O(1)`.

### Cách 4 — Hai con trỏ từ hai đầu

`l = 0`, `r = n - 1`; `while l <= r`: nếu `s[l] == s[r]` trả `l`, ngược lại dịch vào giữa. Điều kiện `l <= r` (không phải `<`) để xét cả tâm khi `n` lẻ.

````python
class SolutionTwoPointers:
    def firstMatchingIndex(self, s):
        l, r = 0, len(s) - 1
        while l <= r:
            if s[l] == s[r]:
                return l
            l += 1
            r -= 1
        return -1

print(SolutionTwoPointers().firstMatchingIndex("abcacbd"), SolutionTwoPointers().firstMatchingIndex("abc"), SolutionTwoPointers().firstMatchingIndex("abcdab"))
````

**Độ phức tạp:** Time `O(n)`, Space `O(1)`.

### Cách 5 — Danh sách cờ boolean rồi `index`

Tạo `flags = [a == b for a, b in zip(s, reversed(s))]` rồi lấy `flags.index(True)` nếu có. Phong cách "tạo mảng đặc trưng rồi tra cứu", tiện khi cần dùng lại mảng đó.

````python
class SolutionFlags:
    def firstMatchingIndex(self, s):
        flags = [a == b for a, b in zip(s, reversed(s))]
        return flags.index(True) if True in flags else -1

print(SolutionFlags().firstMatchingIndex("abcacbd"), SolutionFlags().firstMatchingIndex("abc"), SolutionFlags().firstMatchingIndex("abcdab"))
````

**Độ phức tạp:** Time `O(n)`, Space `O(n)`.

## 8. TRACE từng bước

Trace Cách 4 (hai con trỏ) với `"abcacbd"` và `"abcdab"`:

| s | l | r | s[l] | s[r] | kết luận |
|---|---|---|---|---|---|
| `abcacbd` | 0 | 6 | a | d | khác, dịch |
| `abcacbd` | 1 | 5 | b | b | bằng -> trả 1 |
| `abcdab` | 0 | 5 | a | b | khác |
| `abcdab` | 1 | 4 | b | a | khác |
| `abcdab` | 2 | 3 | c | d | khác |
| `abcdab` | 3 | 2 | | | `l > r`, dừng -> -1 |

````python
def trace_two_pointers(s):
    l, r = 0, len(s) - 1
    print(f"s={s!r}")
    while l <= r:
        eq = s[l] == s[r]
        print(f"  l={l} r={r}: {s[l]!r} vs {s[r]!r} -> {'bằng, trả ' + str(l) if eq else 'khác, dịch vào giữa'}")
        if eq:
            return l
        l += 1
        r -= 1
    print(f"  l={l} > r={r}: dừng")
    return -1

assert trace_two_pointers("abcacbd") == 1
assert trace_two_pointers("abcdab") == -1
assert trace_two_pointers("abc") == 1
assert trace_two_pointers("a") == 0
````

````python
def trace_half(s):
    n = len(s)
    print(f"n={n}, nửa đầu: range({(n + 1) // 2}) = {list(range((n + 1) // 2))}")
    for i in range((n + 1) // 2):
        print(f"  i={i}: s[{i}]={s[i]!r}, s[{n - 1 - i}]={s[n - 1 - i]!r}")
        if s[i] == s[n - 1 - i]:
            return i
    return -1

assert trace_half("xyzzy") == 2
assert trace_half("xyyx") == 0
````

## 9. Code nộp LeetCode

Dùng Cách 4 (hai con trỏ) — một vòng `while`, không cần bộ nhớ phụ.

````python
from typing import List, Optional


class Solution:
    def firstMatchingIndex(self, s: str) -> int:
        l, r = 0, len(s) - 1
        while l <= r:                # l <= r để xét cả ký tự giữa khi n lẻ
            if s[l] == s[r]:
                return l
            l += 1
            r -= 1
        return -1


print(Solution().firstMatchingIndex("abcacbd"), Solution().firstMatchingIndex("abc"), Solution().firstMatchingIndex("abcdab"))
````

## 10. Vì sao thuật toán đúng

1. **Ánh xạ đối xứng.** Với `l` tăng và `r = n - 1 - l` giảm cùng nhịp, mỗi lần lặp xét đúng cặp `(l, n - 1 - l)`. Điều kiện `s[l] == s[n - l - 1]` chính là điều kiện của đề tại `i = l`.
2. **Chỉ số tăng dần.** `l` tăng từ `0` nên lần đầu thỏa điều kiện chính là chỉ số **nhỏ nhất**.
3. **Không bỏ sót.** Vòng lặp chạy khi `l <= r`, tức `l <= (n - 1) / 2` (nửa đầu và ký tự giữa). Với `l` ở nửa sau, cặp `(l, n-1-l)` là bản sao đối xứng của cặp `(n-1-l, l)` đã xét (khi `n-1-l < l`), nên nếu nó khớp thì cặp tương ứng ở nửa đầu cũng đã khớp và đã trả trước đó.
4. **Ký tự giữa (n lẻ).** Khi `l == r`, `s[l] == s[r]` luôn đúng nên ta trả `l`; vì vậy `n` lẻ không bao giờ trả `-1`.
5. **n chẵn không có khớp.** Vòng lặp kết thúc với `l > r` và trả `-1`: mọi cặp đã xét không khớp.

## 11. Phân tích Time / Space Complexity

Gọi `n = len(s)`.

- Cách 1: `O(n)` thời gian, `O(1)` bộ nhớ.
- Cách 2: `O(n)` thời gian, `O(n)` bộ nhớ (chuỗi đảo).
- Cách 3, 4: `O(n/2) = O(n)` thời gian, `O(1)` bộ nhớ.
- Cách 5: `O(n)` thời gian, `O(n)` bộ nhớ (list cờ).

Best case: `s[0] == s[n - 1]` -> trả `0` sau một bước. Worst case: không khớp (chỉ với `n` chẵn) hoặc khớp ở tâm: duyệt `n / 2` bước.

## 12. Bộ test

Đối chiếu 5 cách + code nộp với ví dụ đề, biên (độ dài 1, 2, 3) và 500 chuỗi ngẫu nhiên đúng constraints. Kèm kiểm tra tính chất: `n` lẻ luôn có đáp án.

````python
import random, string

solutions = [SolutionBrute(), SolutionZip(), SolutionHalf(), SolutionTwoPointers(), SolutionFlags(), Solution()]

fixed = [
    ("abcacbd", 1),
    ("abc", 1),
    ("abcdab", -1),
    ("a", 0),
    ("aa", 0),
    ("ab", -1),
    ("aba", 0),
    ("abb", 1),
    ("abcd", -1),
    ("abca", 0),
    ("xyzzy", 2),
    ("xyzwvuy", 3),
]
for s_, expected in fixed:
    for sol in solutions:
        assert sol.firstMatchingIndex(s_) == expected, (type(sol).__name__, s_)

random.seed(38840)
for _ in range(500):
    n = random.randint(1, 100)
    alphabet = random.choice(["ab", "abc", "abcdef", string.ascii_lowercase])
    s_ = ''.join(random.choice(alphabet) for _ in range(n))
    outs = [sol.firstMatchingIndex(s_) for sol in solutions]
    assert all(o == outs[0] for o in outs), (s_, outs)
    if n % 2 == 1:
        assert outs[0] != -1 and outs[0] <= n // 2
    if outs[0] != -1:
        i = outs[0]
        assert s_[i] == s_[n - 1 - i] and all(s_[k] != s_[n - 1 - k] for k in range(i))
print("tất cả test OK")
````

### Test thuộc tính

(1) Đáp án (nếu có) luôn `<= (n - 1) // 2`. (2) Đảo ngược chuỗi không đổi đáp án (cặp đối xứng không phụ thuộc chiều). (3) Palindrome luôn có đáp án `0`.

````python
random.seed(3884)
for _ in range(300):
    n = random.randint(1, 100)
    s_ = ''.join(random.choice("abcd") for _ in range(n))
    ans = Solution().firstMatchingIndex(s_)
    if ans != -1:
        assert ans <= (n - 1) // 2
    assert Solution().firstMatchingIndex(s_[::-1]) == ans
    half = s_[: n // 2]
    pal = half + (random.choice("abcd") if n % 2 else "") + half[::-1]
    assert Solution().firstMatchingIndex(pal) == 0
print("property test OK")
````

## 13. Các lỗi người mới hay gặp

### Lỗi 1: bỏ sót ký tự giữa khi `n` lẻ (dùng `range(n // 2)`)

Với `n = 3`, `n // 2 = 1` nên chỉ xét `i = 0`; ký tự giữa (`i = 1`) bị bỏ qua và ví dụ `"abc"` cho `-1` thay vì `1`.

````python
def wrong_half_floor(s):
    n = len(s)
    for i in range(n // 2):                                       # SAI: phải (n + 1) // 2
        if s[i] == s[n - 1 - i]:
            return i
    return -1

print(wrong_half_floor("abc"), "!=", Solution().firstMatchingIndex("abc"))
assert wrong_half_floor("abc") == -1 and Solution().firstMatchingIndex("abc") == 1
assert wrong_half_floor("a") == -1 and Solution().firstMatchingIndex("a") == 0
````

### Lỗi 2: dùng `while l < r` thay vì `l <= r`

Cũng bỏ sót ký tự giữa của chuỗi lẻ. Khi `l == r` hai con trỏ cùng chỉ vào tâm.

````python
def wrong_strict(s):
    l, r = 0, len(s) - 1
    while l < r:                                                  # SAI: phải l <= r
        if s[l] == s[r]:
            return l
        l += 1
        r -= 1
    return -1

print(wrong_strict("abc"), "!=", Solution().firstMatchingIndex("abc"))
assert wrong_strict("abc") == -1 and Solution().firstMatchingIndex("abc") == 1
assert wrong_strict("abcacbd") == 1
````

### Lỗi 3: lệch chỉ số đối xứng — dùng `n - i` thay vì `n - i - 1`

Ở `i = 0`, `s[n - 0] = s[n]` ngoài biên (`IndexError`). Một số bạn dùng `s[-i]`, nhưng `s[-0]` lại là `s[0]` nên luôn khớp chính nó.

````python
def wrong_negative_index(s):
    for i in range(len(s)):
        if s[i] == s[-i]:                                         # SAI: s[-i], i=0 -> s[0]
            return i
    return -1

print(wrong_negative_index("abcdab"), "!=", Solution().firstMatchingIndex("abcdab"))
assert wrong_negative_index("abcdab") == 0 and Solution().firstMatchingIndex("abcdab") == -1

def wrong_n_minus_i(s):
    n = len(s)
    for i in range(n):
        if s[i] == s[n - i]:                                      # SAI: ngoài biên khi i = 0
            return i
    return -1

try:
    wrong_n_minus_i("abc")
except IndexError as e:
    print("IndexError:", e)
````

### Lỗi 4: nhầm với "hai ký tự giống nhau liền kề"

Đề so sánh các ký tự **đối xứng**, không phải hai ký tự đứng cạnh nhau.

````python
def wrong_adjacent(s):
    for i in range(len(s) - 1):
        if s[i] == s[i + 1]:                                      # SAI: so hai ký tự kề nhau
            return i
    return -1

print(wrong_adjacent("abcacbd"), "!=", Solution().firstMatchingIndex("abcacbd"))
assert wrong_adjacent("abcacbd") == -1 and Solution().firstMatchingIndex("abcacbd") == 1
assert wrong_adjacent("aab") == 0 and Solution().firstMatchingIndex("aab") == 1
````

### Lỗi 5: trả về chỉ số phía bên phải (`n - 1 - i`) hoặc ký tự

Đề hỏi **chỉ số nhỏ nhất** `i`, không phải chỉ số đối xứng, và không phải ký tự.

````python
def wrong_return_right(s):
    n = len(s)
    for i in range(n):
        if s[i] == s[n - 1 - i]:
            return n - 1 - i                                      # SAI: trả chỉ số phía phải
    return -1

print(wrong_return_right("abcacbd"), "!=", Solution().firstMatchingIndex("abcacbd"))
assert wrong_return_right("abcacbd") == 5 and Solution().firstMatchingIndex("abcacbd") == 1
````

## 18. Gợi ý 3 mức cho bài tập

**Mức 1 (nhẹ):** Bài 1: `"racecar"` là palindrome; `"abcd"` có `n` chẵn. Bài 2: `"leetcode"` có `n = 8`, các cặp `(0,7), (1,6), (2,5), (3,4)`.

## 19. Đáp án tham khảo cho bài tập

````python
# Bài 1, 2
assert Solution().firstMatchingIndex("racecar") == 0 and Solution().firstMatchingIndex("abcd") == -1
assert Solution().firstMatchingIndex("leetcode") == -1

# Bài 3
def count_matching_positions(s):
    n = len(s)
    return sum(1 for i in range(n) if s[i] == s[n - 1 - i])
assert count_matching_positions("abcacbd") == 5          # i = 1, 2, 3, 4, 5
assert count_matching_positions("abc") == 1 and count_matching_positions("abcd") == 0

# Bài 4
def last_matching_index(s):
    n = len(s)
    for i in range(n - 1, -1, -1):
        if s[i] == s[n - 1 - i]:
            return i
    return -1
assert last_matching_index("abcacbd") == 5 and last_matching_index("abcdab") == -1 and last_matching_index("a") == 0

# Bài 5
def is_palindrome_by_matches(s):
    return count_matching_positions(s) == len(s)
assert is_palindrome_by_matches("racecar") and is_palindrome_by_matches("abba")
assert not is_palindrome_by_matches("abca") and not is_palindrome_by_matches("ab")
random.seed(5)
for _ in range(100):
    t = ''.join(random.choice("ab") for _ in range(random.randint(1, 10)))
    assert is_palindrome_by_matches(t) == (t == t[::-1])
print("đáp án bài tập OK")
````

## 20. Template nhớ nhanh + Checklist tự kiểm tra trước khi submit

**Template "so sánh đối xứng hai đầu":**

```python
l, r = 0, len(s) - 1
while l <= r:                 # l <= r để xét cả tâm khi n lẻ
    if s[l] == s[r]:
        return l
    l += 1
    r -= 1
return -1
```
