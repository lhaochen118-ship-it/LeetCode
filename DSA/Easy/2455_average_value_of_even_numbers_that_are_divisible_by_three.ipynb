## 3. Kiến thức nền

### 3.1. Số chẵn và chia hết cho 3 bằng `%`

`x % 2 == 0` nghĩa là `x` chẵn; `x % 3 == 0` nghĩa là `x` chia hết cho 3. Phép `%` trả phần dư; phần dư `0` nghĩa là chia hết.

````python
print([x % 2 for x in [1, 2, 3, 4]], [x % 3 for x in [1, 2, 3, 4, 6]])
````

### 3.2. `and` và `or`

`A and B` đúng khi **cả hai** đúng; `A or B` đúng khi **ít nhất một** đúng. Bài này cần **và** — "vừa chẵn, vừa chia hết cho 3". Nếu dùng `or` bạn sẽ lọc luôn cả số chẵn không chia hết 3 hoặc số lẻ chia hết cho 3.

````python
for x in [2, 3, 6, 10, 15]:
    print(x, "and:", x % 2 == 0 and x % 3 == 0, "| or:", x % 2 == 0 or x % 3 == 0)
````

### 3.3. Chẵn **và** chia hết 3 ⇔ chia hết cho 6

Số chia hết cho cả 2 lẫn 3 (hai số nguyên tố cùng nhau) thì chia hết cho tích `2 × 3 = 6`; ngược lại số chia hết cho 6 thì chia hết cho 2 và 3. Do đó `x % 6 == 0` là điều kiện gọn.

````python
print(all((x % 2 == 0 and x % 3 == 0) == (x % 6 == 0) for x in range(1, 1001)))
print([x for x in range(1, 40) if x % 6 == 0])
````

### 3.4. Chia nguyên `//` và chia thực `/`

`7 / 2 == 3.5` (số thực). `7 // 2 == 3` (chia nguyên, làm tròn **xuống**). Với số **dương** thì `//` đúng bằng "làm tròn xuống" mà đề yêu cầu. LeetCode kiểm tra kết quả là `int`, nên dùng `//`.

````python
print(7 / 2, 7 // 2, 66 / 4, 66 // 4, 30 // 3)
print(type(66 / 4), type(66 // 4))
````

### 3.5. `sum`, `len` và danh sách rỗng

`sum([])` là `0`; `len([])` là `0`. Chia cho `len([])` gây `ZeroDivisionError`. Bài này **phải** kiểm tra danh sách rỗng trước khi chia.

````python
lst = []
print(sum(lst), len(lst))
try:
    print(sum(lst) // len(lst))
except ZeroDivisionError as e:
    print("ZeroDivisionError:", e)
````

### 3.6. List comprehension có điều kiện

`[x for x in nums if x % 6 == 0]` tạo list mới chỉ gồm phần tử thoả điều kiện. Có thể thay `[...]` bằng `(...)` để thành biểu thức sinh (không lưu danh sách).

````python
nums = [1, 3, 6, 10, 12, 15]
print([x for x in nums if x % 6 == 0])
print(sum(x for x in nums if x % 6 == 0))
````

## 4. Ý tưởng nền tảng / cách nghĩ trực giác

Hãy làm như tính điểm trung bình môn: **chọn** các điểm đủ điều kiện, **cộng**, **đếm**, rồi **chia**.

Ví dụ 1, `nums = [1, 3, 6, 10, 12, 15]`:

```
1  : lẻ                      -> bỏ
3  : lẻ                      -> bỏ
6  : chẵn, chia hết 3        -> giữ
10 : chẵn nhưng 10 % 3 = 1   -> bỏ
12 : chẵn, chia hết 3        -> giữ
15 : lẻ                      -> bỏ
giữ: 6, 12 -> tổng 18, đếm 2 -> 18 // 2 = 9
```

Ví dụ 2, `[1, 2, 4, 7, 10]`: không số nào chia hết cho 6 → đếm = 0 → trả `0` (không được chia).

Hai chỗ cần cẩn thận: (1) **mẫu số là số phần tử được giữ**, không phải `len(nums)`; (2) nếu không giữ phần tử nào thì trả `0`.

## 5. Cách 1 — Brute Force / baseline

Một vòng `for`, kiểm tra **hai điều kiện riêng** `x % 2 == 0` và `x % 3 == 0`, cộng vào `total` và tăng `count`. Cuối cùng, nếu `count == 0` trả `0`, ngược lại trả `total // count`.

Rõ ràng và bám sát đề, hầu như không thể sai. Nút thắt: không có gì đáng kể về tốc độ; chỉ là hơi dài dòng.

````python
class SolutionBrute:
    def averageValue(self, nums):
        total = 0
        count = 0
        for x in nums:
            if x % 2 == 0 and x % 3 == 0:
                total += x
                count += 1
        if count == 0:
            return 0
        return total // count

print(SolutionBrute().averageValue([1, 3, 6, 10, 12, 15]))
print(SolutionBrute().averageValue([1, 2, 4, 7, 10]))
````

**Độ phức tạp:** Time `O(n)` (một lượt duyệt), Space `O(1)`. Đây đã là tối ưu về bậc (phải đọc mọi phần tử ít nhất một lần).

## 6. Ý tưởng thuật toán / bước đột phá

Đơn giản hoá điều kiện: **chẵn và chia hết cho 3 ⇔ `x % 6 == 0`**. Từ đó:

- Lọc một điều kiện duy nhất `x % 6 == 0` (ít phép tính hơn, ít chỗ sai hơn).
- Dùng `sum` và `len` của danh sách đã lọc.

**Quy tắc "nếu… thì…":**

- Nếu danh sách lọc **rỗng** → `0`.
- Ngược lại → `sum(lọc) // len(lọc)` (chia nguyên làm tròn xuống vì mọi số dương).

Ngoài ra còn hai biến thể có ích: dùng **mảng tần suất** (đếm số lần xuất hiện của mỗi giá trị `1..1000`) và **một lượt `reduce`** giữ cặp `(total, count)`.

Invariant: tại mọi thời điểm, `total` là tổng và `count` là số lượng các phần tử đã duyệt mà chia hết cho 6; cuối cùng `total // count` là đáp án nếu `count > 0`.

````python
nums = [1, 3, 6, 10, 12, 15]
chosen = [x for x in nums if x % 6 == 0]
print("chosen:", chosen, "| tổng:", sum(chosen), "| số lượng:", len(chosen))
print("đáp án:", sum(chosen) // len(chosen) if chosen else 0)
````

## 7. Các cách tiếp theo

Ta có 4 cách: (1) vòng `for` hai điều kiện, (2) lọc `% 6` bằng list comprehension, (3) mảng tần suất, (4) `reduce` một lượt.

### Cách 2 — List comprehension với `% 6`

Lọc rồi `sum // len`. Ngắn gọn nhất. Nhớ xử lý danh sách rỗng.

````python
class SolutionFilter:
    def averageValue(self, nums):
        chosen = [x for x in nums if x % 6 == 0]
        return sum(chosen) // len(chosen) if chosen else 0

print(SolutionFilter().averageValue([1, 3, 6, 10, 12, 15]))
print(SolutionFilter().averageValue([1, 2, 4, 7, 10]))
````

Độ phức tạp Cách 2: Time `O(n)`, Space `O(k)` với `k` là số phần tử được chọn (tối đa `n`).

### Cách 3 — Mảng tần suất theo giá trị

Vì giá trị chỉ trong `1..1000`, ta đếm `freq[v]` = số lần `v` xuất hiện. Rồi chỉ cần duyệt các **bội của 6** trong `6, 12, ..., 996`: `total += v * freq[v]`, `count += freq[v]`. Cách này hữu ích khi mảng rất dài nhưng miền giá trị nhỏ.

````python
class SolutionFreq:
    def averageValue(self, nums):
        freq = [0] * 1001
        for x in nums:
            freq[x] += 1
        total = count = 0
        for v in range(6, 1001, 6):
            total += v * freq[v]
            count += freq[v]
        return total // count if count else 0

print(SolutionFreq().averageValue([1, 3, 6, 10, 12, 15]))
print(SolutionFreq().averageValue([1, 2, 4, 7, 10]))
````

Độ phức tạp Cách 3: Time `O(n + V)` với `V = 1000` là miền giá trị, Space `O(V)`. Với bài này `V` cố định nên coi là `O(n)`.

### Cách 4 — `functools.reduce` một lượt

Giữ cặp `(total, count)` làm "trạng thái" và cập nhật qua từng phần tử. Đây là kiểu **fold** (gấp) trong lập trình hàm; không lưu danh sách trung gian.

````python
from functools import reduce

class SolutionReduce:
    def averageValue(self, nums):
        total, count = reduce(
            lambda acc, x: (acc[0] + x, acc[1] + 1) if x % 6 == 0 else acc,
            nums,
            (0, 0),
        )
        return total // count if count else 0

print(SolutionReduce().averageValue([1, 3, 6, 10, 12, 15]))
print(SolutionReduce().averageValue([1, 2, 4, 7, 10]))
````

Độ phức tạp Cách 4: Time `O(n)`, Space `O(1)`.

## 8. TRACE từng bước

Trace Cách 1 với `nums = [1, 3, 6, 10, 12, 15]`.

| x | `x % 2` | `x % 3` | giữ? | total | count |
|---|---|---|---|---|---|
| 1 | 1 | 1 | không | 0 | 0 |
| 3 | 1 | 0 | không | 0 | 0 |
| 6 | 0 | 0 | có | 6 | 1 |
| 10 | 0 | 1 | không | 6 | 1 |
| 12 | 0 | 0 | có | 18 | 2 |
| 15 | 1 | 0 | không | 18 | 2 |

Kết quả `18 // 2 = 9`.

````python
def trace_average(nums):
    total = count = 0
    for x in nums:
        keep = (x % 2 == 0 and x % 3 == 0)
        if keep:
            total += x
            count += 1
        print(f"x={x:3d} x%2={x % 2} x%3={x % 3} giữ={'có' if keep else 'không':5s} total={total:3d} count={count}")
    result = total // count if count else 0
    print(f"kết quả: {total} // {count} = {result}" if count else "không có số nào -> 0")
    return result

trace_average([1, 3, 6, 10, 12, 15])
trace_average([1, 2, 4, 7, 10])
````

## 9. Code nộp LeetCode

````python
from typing import List

class Solution:
    def averageValue(self, nums: List[int]) -> int:
        total = 0
        count = 0
        for x in nums:
            if x % 6 == 0:
                total += x
                count += 1
        return total // count if count else 0

print(Solution().averageValue([1, 3, 6, 10, 12, 15]), Solution().averageValue([1, 2, 4, 7, 10]))
````

## 10. Vì sao thuật toán đúng

- **Đúng tập số:** `x` chẵn và chia hết cho 3 ⇔ `2 | x` và `3 | x`. Vì `gcd(2, 3) = 1`, điều này tương đương `6 | x` (nếu `x = 2a = 3b` thì `3 | 2a` ⇒ `3 | a`, tức `x = 6c`). Nên lọc `x % 6 == 0` giữ chính xác các số cần xét.
- **Đúng trung bình:** theo đề trung bình = `tổng / số phần tử` được chọn (không phải toàn mảng), rồi **làm tròn xuống**. Với `tổng` và `count` nguyên dương, `tổng // count` chính là phần nguyên của phép chia — đúng "làm tròn xuống".
- **Trường hợp rỗng:** khi `count = 0` thì đề quy định `0`; ta kiểm tra riêng nên không chia cho 0.
- **Các cách khác:** Cách 3 duyệt theo giá trị; vì mỗi giá trị `v` xuất hiện `freq[v]` lần, `v * freq[v]` đóng góp đúng tổng của các lần đó; các bội của 6 trong `1..1000` là `6, 12, ..., 996` nên không sót. Cách 4 cập nhật cặp `(total, count)` theo đúng quy tắc ở mục 6.

## 11. Phân tích Time / Space Complexity

Gọi `n = len(nums)`, `V = 1000` là miền giá trị.

| Cách | Time | Space |
|---|---|---|
| 1. `for` hai điều kiện | `O(n)` | `O(1)` |
| 2. List comprehension `% 6` | `O(n)` | `O(k)`, `k <= n` |
| 3. Mảng tần suất | `O(n + V)` | `O(V)` |
| 4. `reduce` một lượt | `O(n)` | `O(1)` |

Cả bốn cách đều tuyến tính; không có sự khác biệt average/worst. Về tốc độ thực tế, list comprehension thường nhanh nhất trong Python.

## 12. Bộ test

Test 1: ví dụ đề và các edge case.

````python
impls = [SolutionBrute(), SolutionFilter(), SolutionFreq(), SolutionReduce(), Solution()]
tests = [
    ([1, 3, 6, 10, 12, 15], 9),
    ([1, 2, 4, 7, 10], 0),
    ([6], 6),                          # n = 1, thoả
    ([1], 0),                          # n = 1, không thoả
    ([1000], 0),                       # 1000 chẵn nhưng 1000 % 3 = 1
    ([996], 996),                      # bội 6 lớn nhất <= 1000
    ([6, 12, 18, 30], 16),             # 66 / 4 = 16.5 -> làm tròn xuống 16 (bắt lỗi dùng /)
    ([6, 12, 1, 1], 9),                # mẫu số phải là số phần tử được chọn (bắt lỗi chia len(nums))
    ([10, 6], 6),                      # bắt lỗi dùng or
    ([3, 9, 15], 0),                   # chia hết 3 nhưng lẻ
    ([2, 4, 8], 0),                    # chẵn nhưng không chia hết 3
    ([6, 6, 6, 6], 6),                 # lặp
    ([6, 12], 9),
    ([6, 18, 18], 14),                 # 42 // 3 = 14
]
for nums, exp in tests:
    for im in impls:
        got = im.averageValue(list(nums))
        assert got == exp and isinstance(got, int), (type(im).__name__, nums, got, exp)
print("tất cả test cố định OK:", len(tests))
````

Test 2: ngẫu nhiên có seed, đúng constraints (`1 <= n <= 1000`, `1 <= nums[i] <= 1000`). Oracle độc lập dựa trên định nghĩa (lọc `chẵn` rồi lọc `chia hết 3`, dùng `divmod`).

````python
import random
random.seed(2455)

def oracle(nums):
    evens = [x for x in nums if x % 2 == 0]
    both = [x for x in evens if x % 3 == 0]
    if not both:
        return 0
    q, r = divmod(sum(both), len(both))
    return q

zeros = 0
for t in range(3000):
    n = random.randint(1, 1000) if t % 10 == 0 else random.randint(1, 15)
    hi = random.choice([10, 50, 1000])
    nums = [random.randint(1, hi) for _ in range(n)]
    exp = oracle(nums)
    zeros += exp == 0
    for im in impls:
        assert im.averageValue(list(nums)) == exp, (type(im).__name__, nums[:10])
print("3000 test ngẫu nhiên OK; số test đáp án 0:", zeros)
assert 0 < zeros < 3000
````

Test 3: tính chất — đáp án luôn nằm giữa số nhỏ nhất và số lớn nhất của các bội 6 trong mảng, và là bội của `1` hiển nhiên; thêm vào mảng một số không thoả điều kiện không làm đổi đáp án.

````python
random.seed(8)
for _ in range(500):
    nums = [random.randint(1, 1000) for _ in range(random.randint(1, 40))]
    mult6 = [x for x in nums if x % 6 == 0]
    ans = Solution().averageValue(nums)
    if mult6:
        assert min(mult6) <= ans <= max(mult6)
    else:
        assert ans == 0
    noise = random.choice([x for x in range(1, 1001) if x % 6 != 0])
    assert Solution().averageValue(nums + [noise]) == ans
print("tính chất biên + thêm số nhiễu: OK")
````

## 13. Các lỗi người mới hay gặp

### Lỗi 1: dùng `or` thay cho `and`

````python
def wrong_or(nums):
    chosen = [x for x in nums if x % 2 == 0 or x % 3 == 0]      # SAI: phải là "và"
    return sum(chosen) // len(chosen) if chosen else 0

nums = [10, 6]
print("sai :", wrong_or(nums))
print("đúng:", Solution().averageValue(nums))
assert wrong_or(nums) != Solution().averageValue(nums)
````

Vì sao sai: `10` chẵn nhưng không chia hết 3, vẫn bị `or` chọn; trung bình `(10 + 6) // 2 = 8` thay vì `6`. **Sửa:** `and` (hoặc `x % 6 == 0`).

### Lỗi 2: dùng `/` thay vì `//` (ra số thực)

````python
def wrong_float(nums):
    chosen = [x for x in nums if x % 6 == 0]
    return sum(chosen) / len(chosen) if chosen else 0           # SAI: trả số thực

nums = [6, 12, 18, 30]
print("sai :", wrong_float(nums), type(wrong_float(nums)).__name__)
print("đúng:", Solution().averageValue(nums), type(Solution().averageValue(nums)).__name__)
assert wrong_float(nums) != Solution().averageValue(nums)
````

Vì sao sai: `66 / 4 = 16.5` (số thực) trong khi đề cần số nguyên **làm tròn xuống** `16`. **Sửa:** `//` (hoặc `int(...)` với số dương).

### Lỗi 3: chia cho `len(nums)` thay vì số phần tử được chọn

````python
def wrong_denominator(nums):
    return sum(x for x in nums if x % 6 == 0) // len(nums)       # SAI: mẫu số phải là count

nums = [6, 12, 1, 1]
print("sai :", wrong_denominator(nums))
print("đúng:", Solution().averageValue(nums))
assert wrong_denominator(nums) != Solution().averageValue(nums)
````

Vì sao sai: trung bình chỉ tính trên các số **thoả điều kiện**; chia cho cả mảng làm giá trị tụt xuống (`18 // 4 = 4` thay vì `9`).

### Lỗi 4: không xử lý trường hợp không có số nào (chia cho 0)

````python
def wrong_zero_div(nums):
    chosen = [x for x in nums if x % 6 == 0]
    return sum(chosen) // len(chosen)                             # SAI: chosen có thể rỗng

try:
    wrong_zero_div([1, 2, 4, 7, 10])
except ZeroDivisionError as e:
    print("sai : ZeroDivisionError ->", e)
print("đúng:", Solution().averageValue([1, 2, 4, 7, 10]))
````

Vì sao sai: ví dụ 2 của đề chính là trường hợp này; chia cho `len([]) = 0` gây lỗi. **Sửa:** kiểm tra `if count == 0: return 0` (hoặc `if chosen else 0`).

## 14. Mẹo làm bài nhanh & nhận dạng pattern

- "Chia hết cho **cả a lẫn b**" → nghĩ tới `lcm(a, b)`: ở đây `lcm(2, 3) = 6`, nên `x % 6 == 0`.
- "Trung bình làm tròn xuống" → `tổng // số_lượng` (số dương); nhớ mẫu số là số phần tử **được chọn**.
- Mọi phép chia cần kiểm tra mẫu số bằng 0 → đề đã gợi ý bằng câu "nếu không có số nào, trả 0".
- Với miền giá trị nhỏ (`<= 1000`) có thể dùng mảng tần suất; với bài này một vòng `for` là đủ.
- LeetCode so sánh kiểu dữ liệu: trả `9.0` thay vì `9` thường vẫn bị coi là sai (hoặc không khớp kiểu trong Python typed). Dùng `//`.
- Viết điều kiện lọc **một lần** (`x % 6 == 0`) để tránh nhầm giữa `and`/`or`.

## 15. Biến thể của bài

1. Trung bình **làm tròn lên** (`-(-total // count)` hoặc `math.ceil`).
2. Trung bình của các số chia hết cho **k** bất kỳ (khái quát hoá bằng `x % k == 0`).
3. Chẵn **hoặc** chia hết cho 3 (đúng với `or`, thay đổi tập số).
4. Trả về **trung vị** thay vì trung bình của các số thoả điều kiện.
5. Trung bình có trọng số hoặc trung bình các số nằm ở **chỉ số chẵn** của mảng.

## 16. So sánh các cách làm

| Cách | Ý tưởng | Time | Space | Khi nào dùng |
|---|---|---|---|---|
| 1. `for` hai điều kiện | duyệt, kiểm tra `%2` và `%3` | `O(n)` | `O(1)` | học cơ bản, bám sát đề |
| 2. List comprehension `% 6` | lọc rồi `sum // len` | `O(n)` | `O(k)` | **ngắn gọn**, dễ đọc |
| 3. Mảng tần suất | đếm theo giá trị, duyệt bội của 6 | `O(n + V)` | `O(V)` | miền giá trị nhỏ, nhiều lần truy vấn |
| 4. `reduce` | gấp `(total, count)` | `O(n)` | `O(1)` | phong cách hàm, dữ liệu dạng stream |

## 18. Gợi ý 3 mức cho bài tập

**Mức 1 (nhẹ):** lọc theo `x % 6 == 0`, tính tổng và đếm, nhớ trường hợp rỗng.

**Mức 2 (vừa):** Bài 3: giống đề nhưng thay `6` bằng tham số `k`. Bài 4: làm tròn lên với số dương: `(total + count - 1) // count`. Bài 5: duyệt một lần, cập nhật `total` và `count`, rồi tính trung bình ở cuối.

**Mức 3 (gần đáp án):** Bài 3: `chosen = [x for x in nums if x % k == 0]; return sum(chosen) // len(chosen) if chosen else 0`. Bài 4: `return (total + count - 1) // count if count else 0`. Bài 2: chọn `6, 12, 18` → tổng 36, đếm 3 → 12.

## 19. Đáp án tham khảo cho bài tập

````python
# Bài 1: 12, 18, 24 -> 54 // 3 = 18
assert Solution().averageValue([12, 18, 24, 5, 7]) == 18

# Bài 2: [3, 6, 9, 12, 15, 18] -> chọn 6, 12, 18 -> tổng 36, đếm 3 -> 12
chosen = [x for x in [3, 6, 9, 12, 15, 18] if x % 6 == 0]
assert chosen == [6, 12, 18] and sum(chosen) == 36 and len(chosen) == 3
assert Solution().averageValue([3, 6, 9, 12, 15, 18]) == 12

# Bài 3
def average_divisible_by(nums, k):
    chosen = [x for x in nums if x % k == 0]
    return sum(chosen) // len(chosen) if chosen else 0
assert average_divisible_by([1, 3, 6, 10, 12, 15], 6) == 9
assert average_divisible_by([1, 3, 6, 10, 12, 15], 5) == 12      # 10 và 15 -> 25 // 2
assert average_divisible_by([1, 2, 4], 7) == 0

# Bài 4
def average_ceil(nums):
    chosen = [x for x in nums if x % 6 == 0]
    return (sum(chosen) + len(chosen) - 1) // len(chosen) if chosen else 0
assert average_ceil([6, 12, 18, 30]) == 17          # 66 / 4 = 16.5 -> 17
assert average_ceil([6, 12]) == 9
assert average_ceil([1]) == 0

# Bài 5
def count_and_average(nums):
    total = count = 0
    for x in nums:
        if x % 2 == 0 and x % 3 == 0:
            total += x
            count += 1
    return count, (total // count if count else 0)
assert count_and_average([1, 3, 6, 10, 12, 15]) == (2, 9)
assert count_and_average([1, 2, 4, 7, 10]) == (0, 0)
print("đáp án bài tập OK")
````
