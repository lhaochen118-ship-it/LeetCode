## 1. Bài toán yêu cầu gì?

Một **câu (sentence)** là danh sách các từ cách nhau bởi **đúng một** dấu cách, không có dấu cách ở đầu hay cuối. Mỗi từ chỉ gồm chữ cái tiếng Anh hoa/thường.

Cho câu `s` và số nguyên `k`. Hãy **cắt** câu chỉ giữ lại `k` từ đầu tiên và trả về câu đó.

| `s` | `k` | Kết quả |
|---|:---:|---|
| `"Hello how are you Contestant"` | 4 | `"Hello how are you"` |
| `"What is the solution to this problem"` | 4 | `"What is the solution"` |
| `"chopper is not a tanuki"` | 5 | `"chopper is not a tanuki"` |
| `"Hi"` | 1 | `"Hi"` |

**Ràng buộc:** `1 <= len(s) <= 500`, `1 <= k <=` số từ trong `s`, chỉ gồm chữ cái và dấu cách, các từ cách nhau đúng một dấu cách, không có dấu cách đầu/cuối.

````python
examples = [
    ("Hello how are you Contestant", 4, "Hello how are you"),
    ("What is the solution to this problem", 4, "What is the solution"),
    ("chopper is not a tanuki", 5, "chopper is not a tanuki"),
]
for s, k, expected in examples:
    print(f"k={k}: {s!r}\n  -> {expected!r}")
````

## 2. Nền tảng Python cần biết

### 2.1. `str.split` — tách chuỗi thành list

- `s.split()` (không đối số): tách theo mọi chuỗi khoảng trắng, bỏ mảnh rỗng.
- `s.split(" ")`: tách tại **từng** dấu cách.

Với câu đúng chuẩn của đề (một dấu cách giữa các từ), hai cách cho cùng kết quả.

````python
s = "Hello how are you Contestant"
words = s.split()
print(words)
print("số từ:", len(words))
print('"a  b".split()   ->', "a  b".split())
print('"a  b".split(" ") ->', "a  b".split(" "))
````

### 2.2. Slicing list: `words[:k]`

`words[:k]` lấy `k` phần tử đầu (chỉ số `0..k-1`). Nếu `k` lớn hơn độ dài list thì lấy hết — **không lỗi**.

````python
words = ["Hello", "how", "are", "you", "Contestant"]
print(words[:4])
print(words[:1])
print(words[:100], "<- k lớn hơn len vẫn không lỗi")
````

### 2.3. `" ".join(list)` — ghép list chuỗi lại

`sep.join(items)` nối các phần tử của `items`, chèn `sep` vào **giữa** chúng (không thêm ở đầu/cuối).

````python
print(" ".join(["Hello", "how", "are", "you"]))
print("-".join(["a", "b", "c"]))
print(repr(" ".join(["solo"])), "<- một phần tử: không có dấu cách")
````

### 2.4. Slicing chuỗi `s[:i]` và `enumerate`

`s[:i]` lấy `i` ký tự đầu (chỉ số `0..i-1`). Nếu `s[i]` là dấu cách thì `s[:i]` dừng **ngay trước** dấu cách đó.

````python
s = "Hello how are you"
for i, ch in enumerate(s):
    if ch == " ":
        print(f"dấu cách tại i={i} -> s[:i] = {s[:i]!r}")
````

## 3. Cách trực giác — tách, lấy k từ, ghép lại

Đúng như gợi ý của đề: *"giải trên mảng các từ sẽ dễ hơn"*.

1. `words = s.split()`
2. `words[:k]`
3. `" ".join(...)`

````python
def truncate_sentence_split(s: str, k: int) -> str:
    words = s.split()
    return " ".join(words[:k])

for s, k, expected in examples:
    result = truncate_sentence_split(s, k)
    print(repr(result), "| đúng?", result == expected)
````

### 3.1. Đánh giá

- Rất gọn, `O(n)` thời gian với `n = len(s)`.
- Nhưng luôn tách **toàn bộ** câu, kể cả phần sau từ thứ `k` mà ta sẽ bỏ đi; tạo list `words` (`O(n)` bộ nhớ) rồi lại tạo chuỗi mới.
- Nếu câu rất dài (hàng triệu từ) và `k` nhỏ, ta làm thừa rất nhiều.

Câu hỏi: *Có cách nào biết ngay **vị trí cắt** trong chuỗi gốc?*

## 4. Nút thắt và insight

Vì các từ cách nhau **đúng một** dấu cách và không có dấu cách đầu/cuối:

- Từ thứ 1 kết thúc ngay trước dấu cách thứ 1.
- Từ thứ `k` kết thúc ngay trước dấu cách thứ `k`.

> **Insight:** Đếm dấu cách từ trái sang. Khi gặp dấu cách **thứ `k`** tại chỉ số `i`, đáp án là `s[:i]`. Nếu không có đủ `k` dấu cách (tức câu có đúng `k` từ), đáp án là cả `s`.

````python
s = "Hello how are you Contestant"
positions = [i for i, ch in enumerate(s) if ch == " "]
print("vị trí các dấu cách:", positions)
k = 4
print(f"dấu cách thứ {k} ở chỉ số {positions[k - 1]} -> s[:{positions[k - 1]}] = {s[:positions[k - 1]]!r}")
````

Lưu ý `positions[k - 1]`: list đánh số từ 0, nên dấu cách **thứ k** nằm ở chỉ số `k - 1` của list.

## 5. Lời giải chính — đếm dấu cách, cắt tại dấu cách thứ k

**Bước 1.** `count = 0`.

**Bước 2.** Duyệt `i, ch` trong `enumerate(s)`:
- Nếu `ch == ' '`: `count += 1`; nếu `count == k` → `return s[:i]`.

**Bước 3.** Hết vòng lặp mà chưa trả → câu có đúng `k` từ → `return s`.

````python
def truncate_sentence(s: str, k: int) -> str:
    count = 0
    for i, ch in enumerate(s):
        if ch == " ":
            count += 1
            if count == k:
                return s[:i]
    return s

for s, k, expected in examples:
    result = truncate_sentence(s, k)
    print(repr(result), "| đúng?", result == expected)
````

## 6. TRACE từng bước

Trace `s = "What is the solution to this problem"`, `k = 4`:

| i | ch | count sau | count == k? | Hành động |
|---|---|---|---|---|
| 4 | ' ' | 1 | không | đi tiếp |
| 7 | ' ' | 2 | không | đi tiếp |
| 11 | ' ' | 3 | không | đi tiếp |
| 20 | ' ' | 4 | **có** | return `s[:20]` = `"What is the solution"` |

Các ký tự chữ cái không thay đổi gì nên không ghi trong bảng. Phần `" to this problem"` **không bao giờ được đọc**.

````python
def trace_truncate_sentence(s: str, k: int) -> str:
    print(f"=== s={s!r}, k={k} ===")
    count = 0
    for i, ch in enumerate(s):
        if ch == " ":
            count += 1
            print(f"i={i:2}: gặp dấu cách -> count={count}", end="")
            if count == k:
                print(f" == k -> return s[:{i}] = {s[:i]!r}")
                return s[:i]
            print()
    print(f"hết chuỗi, chỉ có {count} dấu cách (< k) -> return s nguyên vẹn")
    return s

trace_truncate_sentence("What is the solution to this problem", 4)
print()
trace_truncate_sentence("chopper is not a tanuki", 5)
````

## 7. Các lời giải khác

### Cách 3 — `split` với `maxsplit`

`s.split(" ", k)` tách **tối đa `k` lần** → tối đa `k + 1` mảnh; mảnh cuối chứa "phần còn lại". Lấy `k` mảnh đầu rồi ghép. Ít việc hơn `split()` toàn bộ.

````python
s = "Hello how are you Contestant"
print(s.split(" ", 2))
print(s.split(" ", 4))

def truncate_sentence_maxsplit(s: str, k: int) -> str:
    return " ".join(s.split(" ", k)[:k])

for s, k, expected in examples:
    print(repr(truncate_sentence_maxsplit(s, k)), "| đúng?", truncate_sentence_maxsplit(s, k) == expected)
````

### Cách 4 — Dùng `str.find` nhảy từ dấu cách này sang dấu cách kế

`s.find(" ", start)` trả chỉ số dấu cách đầu tiên từ vị trí `start`, hoặc `-1` nếu không có. Nhảy `k` lần.

````python
print('"a b c".find(" ") =', "a b c".find(" "), '| find(" ", 2) =', "a b c".find(" ", 2), '| find(" ", 4) =', "a b c".find(" ", 4))

def truncate_sentence_find(s: str, k: int) -> str:
    pos = -1
    for _ in range(k):
        pos = s.find(" ", pos + 1)
        if pos == -1:
            return s
    return s[:pos]

for s, k, expected in examples:
    print(repr(truncate_sentence_find(s, k)), "| đúng?", truncate_sentence_find(s, k) == expected)
````

`for _ in range(k)`: dấu `_` là quy ước đặt tên cho biến **không dùng tới** — ta chỉ cần lặp `k` lần.

### 7.1. Bảng so sánh

| Cách | Ý tưởng | Time | Space phụ | Ghi chú |
|---|---|---|---|---|
| 1. `split` + `join` | list từ | `O(n)` | `O(n)` | dễ nhất, gợi ý của đề |
| **2. Đếm dấu cách (chính)** | cắt tại dấu cách thứ k | `O(p)` | `O(1)` ngoài kết quả | dừng sớm |
| 3. `split(" ", k)` | maxsplit | `O(n)` | `O(n)` | gọn, ít tách hơn |
| 4. `find` lặp | nhảy dấu cách | `O(p)` | `O(1)` ngoài kết quả | dùng hàm có sẵn (chạy bằng C, nhanh) |

`n = len(s)`, `p` = vị trí dấu cách thứ `k` (≤ `n`). Mọi cách đều phải tạo chuỗi kết quả độ dài ≤ `n`.

## 8. Code nộp LeetCode

````python
class Solution:
    def truncateSentence(self, s: str, k: int) -> str:
        count = 0
        for i, ch in enumerate(s):
            if ch == " ":
                count += 1
                if count == k:
                    return s[:i]
        return s

sol = Solution()
for s, k, expected in examples:
    print(repr(sol.truncateSentence(s, k)), "| đúng?", sol.truncateSentence(s, k) == expected)
````

Bản một dòng (dễ nhớ, cũng được chấp nhận):

````python
class SolutionSplit:
    def truncateSentence(self, s: str, k: int) -> str:
        return " ".join(s.split()[:k])

print(SolutionSplit().truncateSentence("Hello how are you Contestant", 4))
````

## 9. Vì sao thuật toán đúng?

Theo constraints, `s = w1 + " " + w2 + " " + ... + " " + wm` với `m >= k` từ, mỗi từ không chứa dấu cách.

- Dấu cách thứ `j` (đếm từ 1) nằm ngay sau từ `wj`. Do đó tiền tố `s[:i]` với `i` = chỉ số dấu cách thứ `k` là đúng `w1 + " " + ... + " " + wk` — chính là câu gồm `k` từ đầu.
- Nếu `m == k`, chuỗi chỉ có `k - 1` dấu cách → vòng lặp không bao giờ đạt `count == k` → trả `s`, cũng chính là `k` từ đầu.

**Invariant:** trước khi xét chỉ số `i`, `count` = số dấu cách trong `s[0..i-1]` = số từ **đã kết thúc hoàn toàn** trước vị trí `i`.

## 10. Độ phức tạp

`n = len(s)`.

- **Time:** `O(n)` trong trường hợp xấu nhất (`k` = số từ), thường dừng sớm hơn.
- **Space:** `O(1)` cho biến đếm; chuỗi kết quả `s[:i]` tốn `O(i)` — bộ nhớ bắt buộc cho đầu ra, thường không tính là bộ nhớ phụ.

### Lỗi 1: Trả về list thay vì chuỗi

### Lỗi 2: Cắt tại dấu cách thứ `k - 1` (lệch một / off-by-one)

Nhầm "từ thứ k kết thúc ở dấu cách thứ k-1" → thiếu một từ.

````python
def wrong_off_by_one(s, k):
    count = 0
    for i, ch in enumerate(s):
        if ch == " ":
            count += 1
            if count == k - 1:
                return s[:i]
    return s

print("sai:", repr(wrong_off_by_one("Hello how are you Contestant", 4)), "| đúng:", repr(truncate_sentence("Hello how are you Contestant", 4)))
````

### Lỗi 3: Trả `s[:i + 1]` — dính dấu cách ở cuối

`s[:i + 1]` gồm cả ký tự `s[i]` là dấu cách → `"Hello how are you "` — LeetCode chấm sai vì thừa dấu cách.

````python
def wrong_trailing_space(s, k):
    count = 0
    for i, ch in enumerate(s):
        if ch == " ":
            count += 1
            if count == k:
                return s[:i + 1]
    return s

print("sai:", repr(wrong_trailing_space("Hello how are you Contestant", 4)))
````

### Lỗi 4: Quên trường hợp `k` bằng số từ

Nếu sau vòng lặp không `return s` (hàm trả `None`), ví dụ 3 của đề sẽ sai.

````python
def wrong_forget_tail(s, k):
    count = 0
    for i, ch in enumerate(s):
        if ch == " ":
            count += 1
            if count == k:
                return s[:i]

print("sai:", wrong_forget_tail("chopper is not a tanuki", 5))
````

## 14. Biến thể

### 14.1. Lấy `k` từ **cuối** câu

Đếm dấu cách từ phải sang, hoặc `" ".join(s.split()[-k:])`.

````python
def last_k_words(s: str, k: int) -> str:
    count = 0
    for i in range(len(s) - 1, -1, -1):
        if s[i] == " ":
            count += 1
            if count == k:
                return s[i + 1:]
    return s

print(repr(last_k_words("Hello how are you Contestant", 2)), repr(" ".join("Hello how are you Contestant".split()[-2:])))
````

### 14.2. Câu "bẩn": nhiều dấu cách liên tiếp, có dấu cách đầu/cuối

Khi đó cách đếm dấu cách **sai**; phải dùng `split()` (không đối số) để chuẩn hóa.

````python
dirty = "  Hello   how are   you  "
print(repr(truncate_sentence(dirty, 2)), "<- cách đếm dấu cách sai với câu bẩn")
print(repr(" ".join(dirty.split()[:2])), "<- split() chuẩn hóa được")
````

| 1 | maxsplit = 2 → 3 mảnh | `[:10]` lấy hết | `['a', 'b', 'c d']`, `'a b c d'` |
| 2 | đếm vị trí từng dấu cách | 1, 6, 13, 18 | `s[:13]` = `"I love Python"` |
| 3 | giống lời giải chính | trả `i` thay vì `s[:i]` | cuối hàm `return len(s)` |
| 4 | dấu cách thứ k | `s[i + 1:]` | không có thì `""` |
| 5 | duyệt từ, cộng độ dài | độ dài mới = cũ + 1 + len(từ) (trừ từ đầu) | vượt `limit` thì dừng |
| 6 | tách dòng | gọi lại `truncate_sentence` với `k` hợp lệ | `min(k, số từ)` |

## 17. Đáp án tham khảo

**Bài 1:** `['a', 'b', 'c d']` và `'a b c d'`.

**Bài 2:** dấu cách tại 1, 6, 13, 18; thứ 3 ở chỉ số 13 → `"I love Python"`.

````python
print("a b c d".split(" ", 2), repr(" ".join("a b c d".split()[:10])))
assert trace_truncate_sentence("I love Python very much", 3) == "I love Python"

def ex3_kth_space_index_ans(s, k):
    count = 0
    for i, ch in enumerate(s):
        if ch == " ":
            count += 1
            if count == k:
                return i
    return len(s)

def ex4_drop_first_k_ans(s, k):
    i = ex3_kth_space_index_ans(s, k)
    return s[i + 1:]

def ex5_truncate_chars_ans(s, limit):
    result, length = [], 0
    for w in s.split():
        new_length = len(w) if not result else length + 1 + len(w)
        if new_length > limit:
            break
        result.append(w)
        length = new_length
    return " ".join(result)

def ex6_truncate_each_line_ans(text, k):
    lines = text.split("\n")
    return "\n".join(truncate_sentence(line, min(k, len(line.split()))) for line in lines)

assert ex3_kth_space_index_ans("Hello how are you Contestant", 4) == 17
assert ex3_kth_space_index_ans("a b", 5) == 3
assert ex4_drop_first_k_ans("Hello how are you Contestant", 4) == "Contestant"
assert ex4_drop_first_k_ans("a b c", 3) == ""
assert ex5_truncate_chars_ans("Hello how are you", 9) == "Hello how"
assert ex5_truncate_chars_ans("Hello how are you", 8) == "Hello"
assert ex5_truncate_chars_ans("Hello", 3) == ""
assert ex6_truncate_each_line_ans("a b c\nd e\nf", 2) == "a b\nd e\nf"
print("Đáp án tham khảo PASS toàn bộ test!")
````

```python
class Solution:
    def truncateSentence(self, s: str, k: int) -> str:
        count = 0
        for i, ch in enumerate(s):
            if ch == " ":
                count += 1
                if count == k:
                    return s[:i]      # dừng NGAY TRƯỚC dấu cách thứ k
        return s                      # câu có đúng k từ
        # ngắn: return " ".join(s.split()[:k])
```
