Skeleton LeetCode:

```python
class Solution:
    def countMatches(self, items: list[list[str]], ruleKey: str, ruleValue: str) -> int:

```

Mục tiêu học:

1. Đọc dữ liệu dạng **list của list** (mỗi item là một "bản ghi" 3 trường) và truy cập trường bằng chỉ số.
2. Chuyển một **tên trường** (`"type"`, `"color"`, `"name"`) thành **chỉ số cột** (`0`, `1`, `2`) bằng `if/elif`, `dict`, hoặc `list.index`.
3. Nhận ra nút thắt: **tra chỉ số một lần** ngoài vòng lặp thay vì so `ruleKey` cho từng item.
4. Viết mẫu **đếm có điều kiện** bằng vòng `for` và bằng `sum(1 for ... if ...)`.
5. Làm quen `zip(*items)` để lấy **cả một cột**, và `list.count`.
6. Tự giải các biến thể: nhiều luật cùng lúc, đếm theo nhóm, lọc trả về bản ghi…

## 1. Bài toán yêu cầu gì?

Cho mảng `items`, mỗi `items[i] = [type_i, color_i, name_i]` mô tả loại, màu và tên của một món đồ. Cho thêm một **luật** gồm `ruleKey` và `ruleValue`.

Món thứ `i` **khớp luật** nếu:

- `ruleKey == "type"` và `ruleValue == type_i`, hoặc
- `ruleKey == "color"` và `ruleValue == color_i`, hoặc
- `ruleKey == "name"` và `ruleValue == name_i`.

Trả về **số món khớp luật**.

| items | ruleKey | ruleValue | Kết quả |
|---|---|---|---:|
| `[["phone","blue","pixel"],["computer","silver","lenovo"],["phone","gold","iphone"]]` | `"color"` | `"silver"` | 1 |
| `[["phone","blue","pixel"],["computer","silver","phone"],["phone","gold","iphone"]]` | `"type"` | `"phone"` | 2 |

Constraints:

- `1 <= items.length <= 10^4`
- `1 <= type_i.length, color_i.length, name_i.length, ruleValue.length <= 10`
- `ruleKey` là một trong `"type"`, `"color"`, `"name"`; mọi chuỗi chỉ gồm chữ thường.

Chú ý ví dụ 2: món `["computer","silver","phone"]` có **tên** là `"phone"` nhưng luật hỏi **loại** → không khớp.

### Nhìn `items` như một bảng

```text
           cột 0 (type)  cột 1 (color)  cột 2 (name)
item 0     phone         blue           pixel
item 1     computer      silver         phone
item 2     phone         gold           iphone
```

`ruleKey` chọn **cột**, `ruleValue` là giá trị cần tìm trong cột đó.

````python
items = [["phone", "blue", "pixel"], ["computer", "silver", "phone"], ["phone", "gold", "iphone"]]
print(f"{'':<8}{'type':<10}{'color':<8}{'name'}")
for i, (t, c, n) in enumerate(items):
    print(f"item {i}: {t:<10}{c:<8}{n}")
````

`for i, (t, c, n) in enumerate(items)`: mỗi phần tử là list 3 chuỗi, ta **mở gói** (unpack) ngay thành 3 biến `t, c, n`.

## 2. Nền tảng Python cần biết

### 2.1. List lồng list và chỉ số hai cấp

`items[i]` là một món (list 3 chuỗi); `items[i][k]` là trường thứ `k` của món đó.

````python
items = [["phone", "blue", "pixel"], ["computer", "silver", "lenovo"]]
print(items[1])
print(items[1][0], items[1][1], items[1][2])
````

### 2.2. Mở gói (unpacking)

Gán một list 3 phần tử cho 3 biến cùng lúc. Số biến phải bằng số phần tử, nếu không sẽ `ValueError`.

````python
t, c, n = ["phone", "gold", "iphone"]
print("type =", t, "| color =", c, "| name =", n)
````

### 2.3. Từ tên trường sang chỉ số: ba cách

1. `if/elif`: so từng trường hợp.
2. `dict`: `{"type": 0, "color": 1, "name": 2}[ruleKey]`.
3. `list.index`: `["type", "color", "name"].index(ruleKey)` — trả vị trí đầu tiên của phần tử.

````python
ruleKey = "color"

if ruleKey == "type":
    k1 = 0
elif ruleKey == "color":
    k1 = 1
else:
    k1 = 2

k2 = {"type": 0, "color": 1, "name": 2}[ruleKey]
k3 = ["type", "color", "name"].index(ruleKey)
print(k1, k2, k3)
````

### 2.4. Đếm có điều kiện

Mẫu cơ bản: biến đếm `count = 0`, gặp phần tử thỏa thì `count += 1`. Viết gọn: `sum(1 for x in data if cond(x))`.

````python
data = ["phone", "computer", "phone"]
count = 0
for x in data:
    if x == "phone":
        count += 1
print(count, sum(1 for x in data if x == "phone"), data.count("phone"))
````

### 2.5. `zip(*items)` — chuyển hàng thành cột

`*items` "rải" từng hàng thành đối số riêng của `zip`; `zip` ghép phần tử thứ nhất của mọi hàng, rồi thứ hai, … → ta được **các cột**.

````python
items = [["phone", "blue", "pixel"], ["computer", "silver", "phone"], ["phone", "gold", "iphone"]]
cols = list(zip(*items))
for name, col in zip(["type", "color", "name"], cols):
    print(name, "->", col)
````

## 3. Cách trực giác — với mỗi món, kiểm tra đúng như đề viết

Chép nguyên văn ba điều kiện của đề vào một `if` với `or`:

````python
def count_matches_literal(items, ruleKey, ruleValue):
    count = 0
    for t, c, n in items:
        if (ruleKey == "type" and ruleValue == t) or \
           (ruleKey == "color" and ruleValue == c) or \
           (ruleKey == "name" and ruleValue == n):
            count += 1
    return count

ex1 = [["phone", "blue", "pixel"], ["computer", "silver", "lenovo"], ["phone", "gold", "iphone"]]
ex2 = [["phone", "blue", "pixel"], ["computer", "silver", "phone"], ["phone", "gold", "iphone"]]
print(count_matches_literal(ex1, "color", "silver"))   # 1
print(count_matches_literal(ex2, "type", "phone"))     # 2
````

Dấu `\` ở cuối dòng cho phép viết tiếp biểu thức ở dòng sau.

### 3.1. Phiên bản `if/elif` trong vòng lặp

Một cách viết hay gặp khác: trong vòng lặp, hỏi `ruleKey` là gì rồi so trường tương ứng.

````python
def count_matches_branch(items, ruleKey, ruleValue):
    count = 0
    for item in items:
        if ruleKey == "type":
            value = item[0]
        elif ruleKey == "color":
            value = item[1]
        else:
            value = item[2]
        if value == ruleValue:
            count += 1
    return count

print(count_matches_branch(ex1, "color", "silver"), count_matches_branch(ex2, "type", "phone"))
````

## 4. Nút thắt và insight

**Nút thắt:** `ruleKey` **không đổi** suốt vòng lặp, nhưng hai cách trên lại so sánh `ruleKey` với `"type"`, `"color"`, `"name"` **cho từng món** — tới 3 phép so chuỗi thừa mỗi món, `10^4` món là `3·10^4` phép thừa. Code cũng dài và dễ gõ sai tên trường.

**Insight:** tách bài thành 2 bước độc lập:

1. **Một lần duy nhất:** đổi `ruleKey` → chỉ số cột `k` (0, 1 hoặc 2).
2. **Cho từng món:** chỉ so **một** trường `item[k] == ruleValue`.

Đây là kỹ thuật **"đưa phần bất biến ra khỏi vòng lặp"** (loop-invariant hoisting) — rất phổ biến khi tối ưu code.

````python
# Đếm số phép so sánh chuỗi: cách literal so ruleKey + ruleValue nhiều lần
def comparisons_literal(n_items):
    return n_items * 6          # tối đa: 3 lần so ruleKey + 3 lần so ruleValue

def comparisons_hoisted(n_items):
    return 3 + n_items          # 3 lần tìm chỉ số + 1 lần so mỗi món

for n in [3, 100, 10000]:
    print(f"n={n:>5}: literal tối đa ~{comparisons_literal(n):>6} | tách bước ~{comparisons_hoisted(n):>6}")
````

## 5. Lời giải chính — tra chỉ số một lần, rồi đếm

### 5.1. Bảng tra tên trường → chỉ số

````python
KEY_INDEX = {"type": 0, "color": 1, "name": 2}
print(KEY_INDEX["type"], KEY_INDEX["color"], KEY_INDEX["name"])
````

Tên viết HOA (`KEY_INDEX`) là quy ước Python cho **hằng số** — giá trị không định thay đổi.

### 5.2. Đếm theo một cột

````python
k = KEY_INDEX["type"]
count = 0
for item in ex2:
    if item[k] == "phone":
        count += 1
print("số món có type == 'phone':", count)
````

### 5.3. Ghép thành hàm

````python
def count_matches(items: list[list[str]], ruleKey: str, ruleValue: str) -> int:
    k = KEY_INDEX[ruleKey]                  # bước 1: một lần duy nhất
    count = 0
    for item in items:                      # bước 2: mỗi món một phép so
        if item[k] == ruleValue:
            count += 1
    return count

print(count_matches(ex1, "color", "silver"), count_matches(ex2, "type", "phone"), count_matches(ex2, "name", "phone"))
````

## 6. TRACE từng bước

Trace ví dụ 2: `ruleKey = "type"` → `k = 0`, `ruleValue = "phone"`.

| i | item | item[k] | == "phone"? | count |
|---:|---|---|---|---:|
| 0 | `["phone","blue","pixel"]` | `phone` | Có | 1 |
| 1 | `["computer","silver","phone"]` | `computer` | Không | 1 |
| 2 | `["phone","gold","iphone"]` | `phone` | Có | 2 |

Món 1 có `"phone"` ở cột **name** nhưng ta chỉ nhìn cột **type** → không đếm.

````python
def trace_count_matches(items, ruleKey, ruleValue):
    k = KEY_INDEX[ruleKey]
    print(f"ruleKey={ruleKey!r} -> cột k={k}; ruleValue={ruleValue!r}")
    count = 0
    for i, item in enumerate(items):
        ok = item[k] == ruleValue
        if ok:
            count += 1
        print(f"  i={i}: {str(item):<34} item[{k}]={item[k]!r:<11} khớp? {str(ok):<5} count={count}")
    print("=> trả về", count)
    return count

trace_count_matches(ex2, "type", "phone")
````

### 6.1. Cùng dữ liệu, đổi luật sang `name`

Giờ món 1 lại khớp (vì tên là `"phone"`) còn món 0, 2 thì không.

````python
trace_count_matches(ex2, "name", "phone")
````

## 7. Code nộp LeetCode

````python
class Solution:
    def countMatches(self, items: list[list[str]], ruleKey: str, ruleValue: str) -> int:
        k = {"type": 0, "color": 1, "name": 2}[ruleKey]
        count = 0
        for item in items:
            if item[k] == ruleValue:
                count += 1
        return count

sol = Solution()
print(sol.countMatches(ex1, "color", "silver"), sol.countMatches(ex2, "type", "phone"))
````

## 8. Các lời giải khác

### 8.1. `list.index` + `sum` — một dòng

````python
class SolutionIndexSum:
    def countMatches(self, items: list[list[str]], ruleKey: str, ruleValue: str) -> int:
        k = ["type", "color", "name"].index(ruleKey)
        return sum(1 for item in items if item[k] == ruleValue)

print(SolutionIndexSum().countMatches(ex1, "color", "silver"), SolutionIndexSum().countMatches(ex2, "type", "phone"))
````

### 8.2. Lấy cả cột bằng `zip(*items)` rồi `count`

`list(zip(*items))[k]` là **tuple** chứa toàn bộ cột `k`; `tuple.count(ruleValue)` đếm (chạy ở tốc độ C).

````python
class SolutionColumn:
    def countMatches(self, items: list[list[str]], ruleKey: str, ruleValue: str) -> int:
        k = ["type", "color", "name"].index(ruleKey)
        column = list(zip(*items))[k]
        return column.count(ruleValue)

print(SolutionColumn().countMatches(ex1, "color", "silver"), SolutionColumn().countMatches(ex2, "type", "phone"))
````

Nhược điểm: `zip(*items)` dựng **cả 3 cột** dù chỉ cần một → tốn `O(3n)` bộ nhớ phụ. Có thể lấy riêng một cột: `[item[k] for item in items]`.

### 8.3. `if/elif` trong vòng lặp — bản nộp (cách trực giác)

````python
class SolutionBranch:
    def countMatches(self, items: list[list[str]], ruleKey: str, ruleValue: str) -> int:
        count = 0
        for t, c, n in items:
            if ruleKey == "type" and t == ruleValue:
                count += 1
            elif ruleKey == "color" and c == ruleValue:
                count += 1
            elif ruleKey == "name" and n == ruleValue:
                count += 1
        return count

print(SolutionBranch().countMatches(ex2, "type", "phone"))
````

### 8.4. `operator.itemgetter` (tham khảo)

`itemgetter(k)` tạo một hàm `f` sao cho `f(item) == item[k]`. Kết hợp `map` để lấy cột.

````python
from operator import itemgetter

get_color = itemgetter(1)
print(get_color(["phone", "blue", "pixel"]))
print(list(map(itemgetter(1), ex1)).count("silver"))
````

### 8.5. Bảng so sánh

Gọi `n = len(items)`.

| Cách | Ý tưởng | Time | Space phụ | Ghi chú |
|---|---|---:|---:|---|
| Chép nguyên điều kiện đề | `or` 3 nhánh | `O(n)` | `O(1)` | Đúng nhưng dài, nhiều phép so thừa |
| `if/elif` trong vòng | hỏi `ruleKey` mỗi món | `O(n)` | `O(1)` | Như trên |
| Tra chỉ số 1 lần + đếm | `dict` → `k` | `O(n)` | `O(1)` | **Lời giải chính** |
| `index` + `sum` | một dòng | `O(n)` | `O(1)` | Gọn |
| `zip(*items)` + `count` | lấy cột | `O(n)` | `O(n)` | Nhanh thực tế, tốn bộ nhớ |

(Mỗi phép so chuỗi tốn `O(L)` với `L ≤ 10` là độ dài chuỗi — hằng số nhỏ, nên ta ghi `O(n)`.)

## 9. Vì sao lời giải đúng?

Ba điều kiện của đề **loại trừ nhau** theo `ruleKey`: với mỗi giá trị của `ruleKey`, chỉ đúng một nhánh có thể đúng:

- `ruleKey == "type"` ⇒ món khớp ⇔ `item[0] == ruleValue`.
- `ruleKey == "color"` ⇒ món khớp ⇔ `item[1] == ruleValue`.
- `ruleKey == "name"` ⇒ món khớp ⇔ `item[2] == ruleValue`.

Gộp lại: món khớp ⇔ `item[KEY_INDEX[ruleKey]] == ruleValue`. Vòng lặp kiểm tra điều kiện này cho **mỗi món đúng một lần** và cộng 1 khi đúng → `count` = số món khớp.

Ta đối chiếu bằng máy giữa cách "chép nguyên đề" (định nghĩa) và lời giải chính trên dữ liệu ngẫu nhiên.

````python
import random

rng = random.Random(1773)
TYPES = ["phone", "computer", "tablet"]
COLORS = ["blue", "silver", "gold", "phone"]      # cố tình cho "phone" xuất hiện ở nhiều cột
NAMES = ["pixel", "lenovo", "iphone", "phone"]
for _ in range(500):
    items = [[rng.choice(TYPES), rng.choice(COLORS), rng.choice(NAMES)] for _ in range(rng.randint(1, 30))]
    key = rng.choice(["type", "color", "name"])
    val = rng.choice(TYPES + COLORS + NAMES)
    assert count_matches(items, key, val) == count_matches_literal(items, key, val)
print("500 bộ dữ liệu: lời giải chính == định nghĩa của đề")
````

## 10. Độ phức tạp

Gọi `n = len(items)`, `L` = độ dài tối đa của chuỗi (≤ 10).

- **Time `O(n · L)`**, coi `L` là hằng → **`O(n)`**: tra chỉ số `O(1)`, mỗi món một phép so chuỗi.
- **Space `O(1)`** phụ: chỉ `k` và `count` (bảng tra 3 phần tử là hằng số).
- Cách `zip(*items)`: Space `O(n)` cho các cột.

Phải xem mọi món ít nhất một lần → `O(n)` là tối ưu.

## 12. Lỗi người mới hay gặp

### Lỗi 1: Tìm `ruleValue` ở **bất kỳ cột nào**

Viết `if ruleValue in item` — quên rằng luật chỉ định **một cột**.

````python
def wrong_any_column(items, ruleKey, ruleValue):
    return sum(1 for item in items if ruleValue in item)

print("sai:", wrong_any_column(ex2, "type", "phone"), "| đúng:", count_matches(ex2, "type", "phone"))
````

**Sửa:** chỉ so `item[k]` với `k` lấy từ `ruleKey`.

### Lỗi 2: Sai thứ tự cột

Nhầm `"color"` là cột 2, `"name"` là cột 1. Đề quy định `[type, color, name]` = `[0, 1, 2]`.

````python
WRONG_INDEX = {"type": 0, "color": 2, "name": 1}
def wrong_order(items, ruleKey, ruleValue):
    k = WRONG_INDEX[ruleKey]
    return sum(1 for it in items if it[k] == ruleValue)

print("sai:", wrong_order(ex1, "color", "silver"), "| đúng:", count_matches(ex1, "color", "silver"))
````

### Lỗi 3: Dùng `in` cho so sánh chuỗi con

`ruleValue in item[k]` kiểm tra **chuỗi con**, không phải bằng nhau: `"phone" in "iphone"` là `True`!

````python
print('"phone" in "iphone" ->', "phone" in "iphone")
def wrong_substring(items, ruleKey, ruleValue):
    k = KEY_INDEX[ruleKey]
    return sum(1 for it in items if ruleValue in it[k])

print("sai:", wrong_substring(ex1, "name", "phone"), "| đúng:", count_matches(ex1, "name", "phone"))
````

**Sửa:** dùng `==`.

### Lỗi 4: Viết sai chính tả tên trường trong `dict`

`{"types": 0, ...}` → `KeyError` khi `ruleKey == "type"`. Lỗi này lộ ngay khi chạy test, nên **luôn chạy đủ ba loại `ruleKey`**.

````python
TYPO = {"types": 0, "color": 1, "name": 2}
try:
    TYPO["type"]
except KeyError as e:
    print("KeyError:", e)
````

### Lỗi 5: `return` nằm trong vòng lặp

Thụt lề sai làm hàm trả về sau món đầu tiên.

````python
def wrong_return_inside(items, ruleKey, ruleValue):
    k = KEY_INDEX[ruleKey]
    count = 0
    for it in items:
        if it[k] == ruleValue:
            count += 1
        return count          # thụt lề sai!

print("sai:", wrong_return_inside(ex2, "type", "phone"), "| đúng:", count_matches(ex2, "type", "phone"))
````

## 13. Mẹo nhận dạng pattern

- Dữ liệu dạng **bảng** (list các bản ghi cùng cấu trúc) + "lọc / đếm theo một trường" → đổi **tên trường → chỉ số cột** một lần, rồi duyệt.
- Phần nào **không đổi** trong vòng lặp → **đưa ra ngoài** (hoisting).
- **Đếm có điều kiện** → `sum(1 for x in ... if ...)` hoặc `list.count`.
- So bằng nhau dùng `==`, đừng dùng `in` với chuỗi.
- Gặp lại ở thực tế: lọc dữ liệu kiểu SQL `SELECT COUNT(*) WHERE color = 'silver'`.

## 14. Biến thể

### 14.1. Nhiều luật cùng lúc (AND)

Món phải khớp **mọi** luật trong danh sách `rules = [(key, value), ...]`.

````python
def count_matches_all(items, rules):
    idx = [(KEY_INDEX[k], v) for k, v in rules]     # tra chỉ số một lần
    return sum(1 for it in items if all(it[k] == v for k, v in idx))

print(count_matches_all(ex2, [("type", "phone"), ("color", "gold")]))   # 1
print(count_matches_all(ex2, [("type", "phone")]))                       # 2
````

### 14.2. Đếm theo nhóm của một trường (GROUP BY)

````python
from collections import Counter

def group_count(items, key):
    k = KEY_INDEX[key]
    return Counter(it[k] for it in items)

print(group_count(ex2, "type"))
print(group_count(ex2, "color"))
````

### 14.3. Trả về danh sách **tên** các món khớp

````python
def matching_names(items, ruleKey, ruleValue):
    k = KEY_INDEX[ruleKey]
    return [it[2] for it in items if it[k] == ruleValue]

print(matching_names(ex2, "type", "phone"))   # ['pixel', 'iphone']
````
