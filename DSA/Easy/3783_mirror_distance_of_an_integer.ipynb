Sau notebook này bạn sẽ:

1. Hiểu đề: **khoảng cách gương** của `n` là `abs(n - reverse(n))`, trong đó `reverse(n)` là số có các chữ số của `n` đọc ngược (bỏ số `0` đứng đầu sau khi đảo).
2. Thành thạo hai cách đảo số: bằng **chuỗi** (`str`, `[::-1]`, `int`) và bằng **số học** (`% 10`, `// 10`, `rev * 10 + d`).

## 2. Đề bài & ví dụ

Cho số nguyên `n`. **Khoảng cách gương** (mirror distance) của `n` được định nghĩa là `abs(n - reverse(n))`, với `reverse(n)` là số nguyên thu được bằng cách **viết các chữ số của `n` theo thứ tự ngược lại**. Hãy trả về khoảng cách gương của `n`.

| Ví dụ | n | Output | Giải thích |
|---|---|---|---|
| 1 | `25` | `27` | `reverse(25) = 52`, `abs(25 - 52) = 27` |
| 2 | `10` | `9` | `reverse(10) = 01 = 1`, `abs(10 - 1) = 9` |
| 3 | `7` | `0` | `reverse(7) = 7`, `abs(7 - 7) = 0` |

Constraints: `1 <= n <= 10^9`.

````python
def mirror(n):
    return abs(n - int(str(n)[::-1]))

for n, exp in [(25, 27), (10, 9), (7, 0)]:
    print(f"n={n:>3} reverse={int(str(n)[::-1]):>3} abs(n - reverse)={mirror(n)} (đề: {exp})")
    assert mirror(n) == exp
````

````python
s = str(1200)
print(s, s[::-1], int(s[::-1]))        # '1200' -> '0021' -> 21
print(int('01'), int('0021'))
````

### 3.2. `% 10` và `// 10`: bóc chữ số cuối

`n % 10` là chữ số hàng đơn vị, `n // 10` là phần còn lại sau khi bỏ chữ số đó. Lặp `while n:` ta bóc được từng chữ số từ **phải sang trái**.

````python
n = 1200
digits = []
while n:
    digits.append(n % 10)
    n //= 10
print(digits)          # [0, 0, 2, 1]
````

````python
rev = 0
for d in [0, 0, 2, 1]:              # các chữ số của 1200 bóc từ phải sang trái
    rev = rev * 10 + d
    print("rev =", rev)
````

## 4. Ý tưởng nền tảng / cách nghĩ trực giác

Bài này gần như chỉ là "dịch nguyên văn" định nghĩa sang code: tìm `reverse(n)` rồi lấy `abs` hiệu. Thử vài ví dụ bằng tay:

```
n = 25   -> 52     |25 - 52| = 27
n = 10   -> 01 = 1 |10 - 1|  = 9
n = 7    -> 7      |7 - 7|   = 0
n = 121  -> 121    |121 - 121| = 0   (palindrome)
n = 1200 -> 0021 = 21  |1200 - 21| = 1179
```

Hai điểm cần lưu ý:

1. **Số `0` ở cuối `n` biến mất khi đảo** (vì thành `0` đứng đầu). `reverse(1200) = 21`, không phải `0021` hay `210`.
2. **Palindrome** (đọc xuôi ngược như nhau) có khoảng cách gương `0`.

Phần chứng minh vì sao `reverse` bằng vòng lặp số học cho đúng kết quả nằm ở mục 10.

### Bảng tính tay

| n | chữ số | đảo | `reverse(n)` | `abs(n - reverse)` |
|---|---|---|---|---|
| `25` | 2, 5 | 5, 2 | 52 | 27 |
| `10` | 1, 0 | 0, 1 | 1 | 9 |
| `7` | 7 | 7 | 7 | 0 |
| `121` | 1, 2, 1 | 1, 2, 1 | 121 | 0 |
| `1200` | 1, 2, 0, 0 | 0, 0, 2, 1 | 21 | 1179 |
| `1000000000` | 1 và chín số 0 | chín số 0 rồi 1 | 1 | 999999999 |

````python
for n in [25, 10, 7, 121, 1200, 10 ** 9]:
    r = int(str(n)[::-1])
    print(f"n={n:>10} reverse={r:>10} khoảng cách={abs(n - r):>10}")
````

### Quan sát thêm: tác dụng của số `0` ở cuối

Nếu `n = m * 10^t` với `m` không chia hết cho `10` thì `reverse(n) = reverse(m)`: các số `0` ở cuối biến mất. Vì vậy `n` và `n * 10` có **cùng số đảo**, nhưng khoảng cách gương khác nhau.

````python
for n in [12, 120, 1200, 12000]:
    r = int(str(n)[::-1])
    print(f"n={n:>6} reverse={r:>3} khoảng cách={abs(n - r):>6}")
assert all(int(str(m * 10 ** t)[::-1]) == int(str(m)[::-1]) for m in range(1, 200) if m % 10 for t in range(5))
````

## 5. Cách 1 — Brute Force / baseline

Làm đúng như định nghĩa, dùng chuỗi: `str(n)`, đảo bằng `[::-1]`, `int(...)`, rồi `abs`.

Nút thắt: chưa có gì "chậm" (số có tối đa 10 chữ số), nhưng cách này phụ thuộc việc đổi qua lại giữa số và chuỗi. Ta sẽ xem cách làm **thuần số học**, hữu ích trong những bài cấm chuyển sang chuỗi.

````python
class SolutionString:
    def mirrorDistance(self, n):
        rev = int(str(n)[::-1])
        return abs(n - rev)

print(SolutionString().mirrorDistance(25), SolutionString().mirrorDistance(10), SolutionString().mirrorDistance(7))
````

**Độ phức tạp:** Time `O(d)` với `d = ⌊log10 n⌋ + 1` là số chữ số, Space `O(d)` cho chuỗi.

## 6. Ý tưởng thuật toán / bước đột phá

**Bước đột phá (cách số học):** đảo số bằng cách **bóc chữ số cuối** của `n` và **gắn vào cuối** `rev`:

```
rev = 0
while x > 0:
    rev = rev * 10 + x % 10
    x //= 10
```

Biến `x` là bản sao của `n` để còn giữ `n` gốc cho phép trừ cuối cùng.

**Bước đột phá (cách ghép cặp):** viết `n = Σ d_i * 10^i` (với `d_0` là chữ số hàng đơn vị) và `reverse(n) = Σ d_i * 10^(L-1-i)`. Hiệu `n - reverse(n) = Σ d_i * (10^i - 10^(L-1-i))`. Hai vị trí đối xứng `i` và `L-1-i` gộp thành một số hạng `(d_i - d_(L-1-i)) * (10^i - 10^(L-1-i))`; chữ số giữa (nếu `L` lẻ) cho `0`.

**Quy tắc "nếu… thì…":**

- Nếu `n` là palindrome -> đáp án `0`.
- Nếu `n` có `0` ở cuối -> `reverse` ngắn hơn, cứ để `int`/vòng lặp tự xử lý.
- Luôn bọc kết quả bằng `abs`.

**Invariant của vòng lặp số học:** sau mỗi vòng, `rev` là số tạo bởi các chữ số đã bóc (theo thứ tự đã bóc) và `x` là phần còn lại chưa xử lý; `n` gốc = (phần đã bóc đọc xuôi) kết hợp với `x`.

````python
# Kiểm chứng: hiệu luôn chia hết cho 9, và palindrome cho 0
for n in range(1, 20000):
    r = int(str(n)[::-1])
    assert (n - r) % 9 == 0
    if str(n) == str(n)[::-1]:
        assert abs(n - r) == 0
print("hiệu n - reverse(n) chia hết cho 9; palindrome cho 0")
````

## 7. Các cách tiếp theo

### Cách 2 — Đảo bằng vòng lặp số học

Dùng biến `x = n`; lặp `while x:` bóc `d = x % 10`, `rev = rev * 10 + d`, `x //= 10`. Không dùng chuỗi, `O(1)` bộ nhớ.

````python
class SolutionMath:
    def mirrorDistance(self, n):
        x, rev = n, 0
        while x:
            rev = rev * 10 + x % 10
            x //= 10
        return abs(n - rev)

print(SolutionMath().mirrorDistance(25), SolutionMath().mirrorDistance(10), SolutionMath().mirrorDistance(7))
````

**Độ phức tạp:** Time `O(d)`, Space `O(1)`.

### Cách 3 — Dựng từ danh sách chữ số

Bóc các chữ số vào danh sách `digits` (từ phải sang trái). Khi đó **chính `digits` đã là thứ tự của số đảo** đọc từ trái sang phải; nên `rev = Σ digits[i] * 10^(d-1-i)` hay dùng `reduce`. Cách này tách hai việc "bóc" và "ghép", dễ gỡ lỗi hơn.

````python
from functools import reduce

class SolutionDigits:
    def mirrorDistance(self, n):
        digits, x = [], n
        while x:
            digits.append(x % 10)          # từ hàng đơn vị lên: chính là thứ tự của số đảo
            x //= 10
        rev = reduce(lambda acc, d: acc * 10 + d, digits, 0)
        return abs(n - rev)

print(SolutionDigits().mirrorDistance(25), SolutionDigits().mirrorDistance(10), SolutionDigits().mirrorDistance(7))
````

**Độ phức tạp:** Time `O(d)`, Space `O(d)`.

### Cách 4 — Công thức ghép cặp chữ số đối xứng

Dùng hai con trỏ trên chuỗi chữ số `s` (độ dài `L`, chỉ số từ trái bắt đầu `0`). Chữ số `a = s[i]` đứng cách đầu `i` bước, nên trong `n` nó có trọng số `10^(L-1-i)`; trong số đảo nó chuyển sang vị trí đối xứng nên trọng số là `10^i`. Chữ số `b = s[L-1-i]` thì ngược lại. Vậy cặp `(i, L-1-i)` đóng góp `a * (10^(L-1-i) - 10^i) + b * (10^i - 10^(L-1-i)) = (a - b) * (10^(L-1-i) - 10^i)` vào `n - reverse(n)`. Cộng mọi cặp (với `i < L-1-i`) rồi lấy `abs`. Cách này nhất quán với chứng minh ở mục 10 và không cần tạo số đảo.

````python
class SolutionPairs:
    def mirrorDistance(self, n):
        s = str(n)
        L = len(s)
        diff = 0
        i, j = 0, L - 1
        while i < j:
            a, b = int(s[i]), int(s[j])
            # s[i] có trọng số 10^(L-1-i) trong n và 10^i trong số đảo; s[j] ngược lại
            diff += (a - b) * (10 ** (L - 1 - i) - 10 ** i)
            i += 1
            j -= 1
        return abs(diff)

print(SolutionPairs().mirrorDistance(25), SolutionPairs().mirrorDistance(10), SolutionPairs().mirrorDistance(7))
````

**Độ phức tạp:** Time `O(d)` (chỉ `d/2` vòng), Space `O(d)` cho chuỗi.

## 8. TRACE từng bước

Trace Cách 2 với `n = 1200`:

| vòng | `x` trước | `d = x % 10` | `rev` sau | `x` sau |
|---|---|---|---|---|
| 1 | 1200 | 0 | 0 | 120 |
| 2 | 120 | 0 | 0 | 12 |
| 3 | 12 | 2 | 2 | 1 |
| 4 | 1 | 1 | 21 | 0 |

`rev = 21`, đáp án `abs(1200 - 21) = 1179`.

````python
def trace_math(n):
    x, rev, step = n, 0, 0
    print("n =", n)
    while x:
        step += 1
        d = x % 10
        rev = rev * 10 + d
        print(f"  vòng {step}: x={x:>5} d={d} -> rev={rev:>5}, x tiếp theo={x // 10}")
        x //= 10
    print(f"  reverse = {rev}, khoảng cách = {abs(n - rev)}")
    return abs(n - rev)

assert trace_math(1200) == 1179
assert trace_math(25) == 27
````

````python
def trace_pairs(n):
    s = str(n)
    L = len(s)
    print("s =", s)
    diff, i, j = 0, 0, L - 1
    while i < j:
        a, b = int(s[i]), int(s[j])
        term = (a - b) * (10 ** (L - 1 - i) - 10 ** i)
        diff += term
        print(f"  cặp (i={i}, j={j}): a={a} b={b} số hạng={term} tích lũy={diff}")
        i += 1
        j -= 1
    print("  |tổng| =", abs(diff))
    return abs(diff)

assert trace_pairs(1200) == 1179
assert trace_pairs(7) == 0
````

## 9. Code nộp LeetCode

````python
from typing import List, Optional

class Solution:
    def mirrorDistance(self, n: int) -> int:
        x, rev = n, 0
        while x:
            rev = rev * 10 + x % 10      # gắn chữ số cuối của x vào cuối rev
            x //= 10
        return abs(n - rev)
````

## 10. Vì sao thuật toán đúng

- **Đảo bằng vòng lặp:** quy nạp theo số vòng. Trước vòng đầu `rev = 0` (số tạo bởi 0 chữ số). Giả sử sau `k` vòng, `rev` đúng bằng số tạo bởi `k` chữ số cuối của `n` đã bóc theo thứ tự bóc. Vòng tiếp theo bóc chữ số `d` kế tiếp: `rev * 10 + d` đẩy các chữ số cũ sang trái một hàng rồi gắn `d` vào hàng đơn vị, nên `rev` là số tạo bởi `k + 1` chữ số theo thứ tự bóc. Khi `x = 0` đã bóc hết nên `rev = reverse(n)`.
- **Số `0` cuối:** khi bóc các chữ số `0` đầu tiên, `rev = 0 * 10 + 0 = 0` nên chúng không đóng góp: chính là hiện tượng "bỏ `0` đầu sau khi đảo".
- **`abs`:** định nghĩa của đề, bảo đảm kết quả không âm dù `n` nhỏ hơn hay lớn hơn số đảo.
- **Công thức ghép cặp:** với `d_i` là chữ số hàng `10^i` (từ phải), `n - reverse(n) = Σ_i d_i (10^i - 10^(L-1-i))`; hai số hạng `i` và `L-1-i` gộp thành `(d_i - d_(L-1-i)) * (10^i - 10^(L-1-i))`, số hạng ở giữa (`i = L-1-i`) bằng `0`. Đây là cùng một giá trị với cách đảo trực tiếp, chỉ khác cách cộng; mục 12 kiểm chứng bằng vét cạn.
- **Chia hết cho 9:** `n ≡ reverse(n) (mod 9)` do cùng tổng chữ số nên đáp án luôn chia hết cho 9; test ở mục 12 dùng tính chất này như một kiểm tra chéo.

## 12. Bộ test

Ví dụ của đề, các biên (`n = 1`, `n = 10^9`, số có `0` ở cuối, palindrome) và so sánh **tất cả cách** trên nhiều input; thêm kiểm tra chia hết cho `9`.

````python
impls = [SolutionString(), SolutionMath(), SolutionDigits(), SolutionPairs(), Solution()]

cases = [
    (25, 27), (10, 9), (7, 0), (1, 0), (121, 0), (1200, 1179), (100, 99), (120, 99),
    (10 ** 9, 10 ** 9 - 1), (987654321, 987654321 - 123456789), (1001, 0), (12, 9), (21, 9),
    (999999999, 0), (123, 198),
]
for n, exp in cases:
    assert 1 <= n <= 10 ** 9
    for impl in impls:
        got = impl.mirrorDistance(n)
        assert got == exp, (type(impl).__name__, n, got, exp)
print("ví dụ + edge OK:", len(cases), "ca")
````

## 13. Các lỗi người mới hay gặp

### Lỗi 1: quên `abs`

`n - reverse(n)` có thể âm (`25 - 52 = -27`).

````python
def wrong_no_abs(n):
    return n - int(str(n)[::-1])                 # SAI: thiếu abs

print(wrong_no_abs(25), "!=", Solution().mirrorDistance(25))
assert wrong_no_abs(25) == -27 and Solution().mirrorDistance(25) == 27
````

### Lỗi 2: dùng `n` đã bị "ăn mất" trong vòng lặp

Nếu vừa dùng `n` làm biến bóc chữ số (`n //= 10`) vừa dùng nó trong phép trừ cuối, `n` đã bằng `0` nên kết quả chỉ là `rev`.

````python
def wrong_consumed(n):
    rev = 0
    while n:
        rev = rev * 10 + n % 10
        n //= 10                                # n bị tiêu hao
    return abs(n - rev)                         # SAI: n lúc này là 0

print(wrong_consumed(25), "!=", Solution().mirrorDistance(25))
assert wrong_consumed(25) == 52 and Solution().mirrorDistance(25) == 27
````

### Lỗi 3: điều kiện vòng lặp dừng sớm

`while x > 10` bỏ sót chữ số đầu cùng của số đang xử lý (khi `x` còn một chữ số).

````python
def wrong_loop(n):
    x, rev = n, 0
    while x > 10:                                # SAI: phải là while x > 0 (hay while x)
        rev = rev * 10 + x % 10
        x //= 10
    return abs(n - rev)

print(wrong_loop(25), "!=", Solution().mirrorDistance(25))
assert wrong_loop(25) == 20 and Solution().mirrorDistance(25) == 27
````

### Lỗi 4: nhầm "đảo" với "sắp xếp giảm dần các chữ số"

Hai thao tác này chỉ trùng nhau khi các chữ số vốn đã tăng dần (như `1234`). Với `1324`, đảo cho `4231` còn sắp giảm cho `4321`.

````python
def wrong_sorted_desc(n):
    rev = int("".join(sorted(str(n), reverse=True)))      # SAI: sắp xếp thay vì đảo
    return abs(n - rev)

print(wrong_sorted_desc(1324), "!=", Solution().mirrorDistance(1324))
print(wrong_sorted_desc(1324), "vs", Solution().mirrorDistance(1324))        # 4321 vs 4231
assert wrong_sorted_desc(1324) != Solution().mirrorDistance(1324)
````

## 14. Mẹo làm bài nhanh & nhận dạng pattern

- "Đảo số" -> hoặc `int(str(n)[::-1])`, hoặc vòng `rev = rev * 10 + x % 10`; nhớ giữ lại `n` gốc.
- Nếu đề nói "bỏ số `0` đầu", `int()` làm việc đó miễn phí; vòng số học cũng vậy.
- Gặp `abs(a - b)` với `a`, `b` không biết lớn nhỏ -> luôn dùng `abs`.
- Tự kiểm tra: kết quả phải chia hết cho `9`; palindrome phải ra `0`.
- Với `n <= 10^9` đảo ra tối đa 10 chữ số, không lo tràn (Python dùng số nguyên lớn tùy ý; ở ngôn ngữ khác cần `long`).
- Nếu bị cấm chuyển sang chuỗi, dùng cách số học; nếu muốn ngắn gọn, dùng chuỗi.

## 15. Biến thể của bài

1. **Reverse Integer** (LeetCode 7): đảo số có dấu và kiểm tra tràn 32 bit.
2. **Palindrome Number** (9): `n == reverse(n)`.
3. **Sum of Number and Its Reverse** (2443): tìm `x` sao cho `x + reverse(x) = num`.
4. **Count Number of Distinct Integers After Reverse Operations** (2442): thêm số đảo của từng phần tử rồi đếm số phân biệt.
5. **Mirror Distance** cho nhiều số: tính khoảng cách gương lớn nhất/nhỏ nhất trong một mảng.
6. Khoảng cách gương **trong cơ số khác** (cơ số 2): đảo chuỗi nhị phân của `n`.

## 18. Gợi ý 3 mức cho bài tập

**Mức 1 (nhẹ):** Bài 1: `340` đảo là `043`. Bài 2: chữ số của `5071` là `5, 0, 7, 1`.

**Mức 3 (gần đáp án):** Bài 1: `reverse(340) = 43` nên `abs(340 - 43) = 297`. Bài 2: `reverse = 1705`, khoảng cách `|5071 - 1705| = 3366`. Bài 5: `max(Solution().mirrorDistance(x) for x in nums)`.

## 19. Đáp án tham khảo cho bài tập

````python
# Bài 1, 2
assert Solution().mirrorDistance(340) == 297
assert Solution().mirrorDistance(5071) == 3366 and int(str(5071)[::-1]) == 1705

# Bài 3
def reverse_number(n):
    rev = 0
    while n:
        rev = rev * 10 + n % 10
        n //= 10
    return rev
assert reverse_number(25) == 52 and reverse_number(1200) == 21 and reverse_number(7) == 7
assert all(reverse_number(n) == int(str(n)[::-1]) for n in range(1, 5000))

# Bài 4
def is_palindrome_number(n):
    return n == reverse_number(n)
assert is_palindrome_number(121) and is_palindrome_number(7) and not is_palindrome_number(10) and not is_palindrome_number(1200)
assert all(is_palindrome_number(n) == (str(n) == str(n)[::-1]) for n in range(1, 5000))

# Bài 5
def max_mirror_distance(nums):
    return max(Solution().mirrorDistance(x) for x in nums)
assert max_mirror_distance([25, 10, 7]) == 27
assert max_mirror_distance([7, 121, 1]) == 0
assert max_mirror_distance([10 ** 9, 5]) == 10 ** 9 - 1
print("đáp án bài tập OK")
````
