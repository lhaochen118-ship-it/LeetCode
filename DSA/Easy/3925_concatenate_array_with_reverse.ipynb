## 2. Đề bài & ví dụ

Cho mảng số nguyên `nums` độ dài `n`. Hãy tạo mảng `ans` độ dài `2 * n` sao cho:

- `ans[i] = nums[i]` với `0 <= i < n` (nửa đầu giống `nums`),
- `ans[i + n] = nums[n - i - 1]` với `0 <= i < n` (nửa sau là `nums` viết ngược).

Trả về `ans`.

| Ví dụ | nums | Output | Giải thích |
|---|---|---|---|
| 1 | `[1,2,3]` | `[1,2,3,3,2,1]` | Ba phần tử đầu là bản sao `nums`; `ans[3] = nums[2] = 3`, `ans[4] = nums[1] = 2`, `ans[5] = nums[0] = 1` |
| 2 | `[1]` | `[1,1]` | Mảng một phần tử đảo ngược vẫn là chính nó |

Constraints: `1 <= nums.length <= 100`; `1 <= nums[i] <= 100`.

````python
nums = [1, 2, 3]
n = len(nums)
for i in range(n):
    print(f"ans[{i}] = nums[{i}] = {nums[i]};  ans[{i + n}] = nums[{n - i - 1}] = {nums[n - i - 1]}")
````

## 3. Kiến thức nền

### 3.1. Cắt lát đảo ngược `nums[::-1]`

`nums[::-1]` tạo **list mới** gồm các phần tử theo thứ tự ngược, **không** thay đổi `nums`. Cú pháp `[start:stop:step]` với `step = -1`.

````python
nums = [1, 2, 3]
r = nums[::-1]
print(r, nums, r is nums)
````

### 3.2. Nối list bằng `+` và lặp bằng `*`

`a + b` tạo list mới gồm `a` rồi `b`. `a * 2` là `a + a` (lặp **cùng thứ tự**, không đảo). Phân biệt rõ để không nhầm `nums * 2` với bài này.

````python
a = [1, 2, 3]
print(a + a[::-1], a * 2)
````

### 3.3. `list.reverse()` khác `reversed()` khác `[::-1]`

- `lst.reverse()` đảo **tại chỗ** và trả về `None`.
- `reversed(lst)` trả về **iterator** (không phải list).
- `lst[::-1]` trả về **list mới**.

````python
a = [1, 2, 3]
print(a[::-1], list(reversed(a)))
b = [1, 2, 3]
res = b.reverse()
print(res, b)                    # None [3, 2, 1]
````

### 3.4. `extend` và `append`

`lst.extend(iterable)` thêm **từng phần tử** của iterable vào cuối. `lst.append(x)` thêm **một** phần tử `x` (kể cả `x` là list -> list lồng). Cả hai sửa tại chỗ và trả về `None`.

````python
a = [1, 2]
a.extend([3, 4])
print(a)
b = [1, 2]
b.append([3, 4])
print(b)
````

### 3.5. Sao chép list: `list(nums)`, `nums[:]`, `nums.copy()`

`ans = nums` **không** sao chép: cả hai tên cùng trỏ một list (aliasing). Muốn bản sao độc lập dùng `list(nums)` hoặc `nums[:]`. Sửa `ans` mà chưa sao chép sẽ làm hỏng `nums`.

````python
nums = [1, 2, 3]
alias = nums
alias.append(9)
print(nums)                       # nums bị đổi theo!
nums = [1, 2, 3]
copy = list(nums)
copy.append(9)
print(nums, copy)
````

## 4. Ý tưởng nền tảng / cách nghĩ trực giác

Hãy hình dung "lật" dãy như soi gương: nửa đầu là `nums`, nửa sau là hình phản chiếu của nó. Phần tử cuối của nửa đầu (`nums[n-1]`) đứng cạnh phần tử đầu của nửa sau (cũng là `nums[n-1]`), nên `ans` luôn là một **palindrome** (đọc xuôi = đọc ngược) độ dài chẵn.

Tính tay `[4, 5, 6]`: nửa đầu `4 5 6`; nửa sau đảo `6 5 4`. Kết quả `[4,5,6,6,5,4]`.

| i | `ans[i]` | `ans[i + 3]` |
|---|---|---|
| 0 | `nums[0] = 4` | `nums[2] = 6` |
| 1 | `nums[1] = 5` | `nums[1] = 5` |
| 2 | `nums[2] = 6` | `nums[0] = 4` |

````python
nums = [4, 5, 6]
ans = nums + nums[::-1]
print(ans, ans == ans[::-1], len(ans) == 2 * len(nums))
````

## 5. Cách 1 — Brute Force / baseline

Làm đúng công thức của đề: tạo `ans = [0] * (2 * n)`, rồi với mỗi `i` gán `ans[i] = nums[i]` và `ans[i + n] = nums[n - i - 1]`.

Cách này đúng và `O(n)`; "nút thắt" duy nhất là sự dài dòng và dễ lệch chỉ số. Python cho cách ngắn hơn nhiều.

````python
class SolutionBrute:
    def concatWithReverse(self, nums):
        n = len(nums)
        ans = [0] * (2 * n)
        for i in range(n):
            ans[i] = nums[i]
            ans[i + n] = nums[n - i - 1]
        return ans

print(SolutionBrute().concatWithReverse([1, 2, 3]), SolutionBrute().concatWithReverse([1]))
````

**Độ phức tạp:** Time `O(n)`, Space `O(n)` cho `ans`.

## 6. Ý tưởng thuật toán / bước đột phá

**Bước đột phá:** "nửa đầu giống `nums`, nửa sau là `nums` đảo ngược" chính là **nối** hai list: `nums + nums[::-1]`. Không cần chỉ số.

Điều cần cẩn thận không phải thuật toán mà là **ngữ nghĩa của các thao tác list** trong Python:

- dùng thao tác **tạo list mới** (`+`, `[::-1]`) để không làm hỏng `nums`;
- tránh các hàm sửa tại chỗ trả về `None` (`reverse`, `extend`, `sort`) trong biểu thức.

**Invariant:** `len(ans) == 2n`; `ans[:n] == nums`; `ans[n:] == nums[::-1]`; do đó `ans[i + n] = nums[n - 1 - i]`.

## 7. Các cách tiếp theo

### Cách 2 — Nối với lát cắt đảo ngược

Một dòng: `nums + nums[::-1]`.

````python
class SolutionSlice:
    def concatWithReverse(self, nums):
        return nums + nums[::-1]

print(SolutionSlice().concatWithReverse([1, 2, 3]), SolutionSlice().concatWithReverse([1]))
````

**Độ phức tạp:** Time `O(n)`, Space `O(n)` (tạo hai list: lát cắt đảo và kết quả).

### Cách 3 — Sao chép rồi `extend(reversed(...))`

`ans = list(nums)` (bản sao độc lập), rồi `ans.extend(reversed(nums))`. Chú ý `reversed(nums)` là iterator của `nums` (chưa bị sửa), nên an toàn.

````python
class SolutionExtend:
    def concatWithReverse(self, nums):
        ans = list(nums)               # bản sao độc lập, không alias
        ans.extend(reversed(nums))
        return ans

print(SolutionExtend().concatWithReverse([1, 2, 3]), SolutionExtend().concatWithReverse([1]))
````

**Độ phức tạp:** Time `O(n)`, Space `O(n)` (không tạo thêm list lát cắt phụ).

### Cách 4 — List comprehension theo chỉ số của `ans`

Với chỉ số `k` của `ans` (`0 <= k < 2n`): nếu `k < n` thì lấy `nums[k]`, ngược lại lấy `nums[2n - 1 - k]` (đối xứng qua tâm).

````python
class SolutionComprehension:
    def concatWithReverse(self, nums):
        n = len(nums)
        return [nums[k] if k < n else nums[2 * n - 1 - k] for k in range(2 * n)]

print(SolutionComprehension().concatWithReverse([1, 2, 3]), SolutionComprehension().concatWithReverse([1]))
````

**Độ phức tạp:** Time `O(n)`, Space `O(n)`.

### Cách 5 — `itertools.chain` trải dài hai iterable

`list(chain(nums, reversed(nums)))`: `chain` nối hai iterable mà không tạo list trung gian.

````python
from itertools import chain

class SolutionChain:
    def concatWithReverse(self, nums):
        return list(chain(nums, reversed(nums)))

print(SolutionChain().concatWithReverse([1, 2, 3]), SolutionChain().concatWithReverse([1]))
````

**Độ phức tạp:** Time `O(n)`, Space `O(n)` (chỉ kết quả).

## 8. TRACE từng bước

Trace Cách 1 với `nums = [4, 5, 6]` (`n = 3`, `ans` khởi tạo `[0, 0, 0, 0, 0, 0]`):

| i | `ans[i] = nums[i]` | `ans[i+3] = nums[2-i]` | `ans` sau bước |
|---|---|---|---|
| 0 | `ans[0] = 4` | `ans[3] = 6` | `[4,0,0,6,0,0]` |
| 1 | `ans[1] = 5` | `ans[4] = 5` | `[4,5,0,6,5,0]` |
| 2 | `ans[2] = 6` | `ans[5] = 4` | `[4,5,6,6,5,4]` |

````python
def trace_brute(nums):
    n = len(nums)
    ans = [0] * (2 * n)
    for i in range(n):
        ans[i] = nums[i]
        ans[i + n] = nums[n - i - 1]
        print(f"i={i}: ans[{i}]={nums[i]}, ans[{i + n}]=nums[{n - i - 1}]={nums[n - i - 1]} -> {ans}")
    return ans

assert trace_brute([4, 5, 6]) == [4, 5, 6, 6, 5, 4]
assert trace_brute([1]) == [1, 1]
````

````python
def trace_aliasing():
    nums = [1, 2, 3]
    ans_bad = nums                     # alias
    ans_bad.extend(reversed(nums))
    print("alias: nums bị sửa ->", nums, "| ans_bad cùng đối tượng:", ans_bad is nums)
    nums = [1, 2, 3]
    ans_good = list(nums)              # bản sao độc lập
    ans_good.extend(reversed(nums))
    print("copy : nums giữ nguyên ->", nums, "| ans_good:", ans_good)

trace_aliasing()
````

## 9. Code nộp LeetCode

Dùng Cách 2: ngắn nhất và không sửa mảng đầu vào.

````python
from typing import List, Optional


class Solution:
    def concatWithReverse(self, nums: List[int]) -> List[int]:
        return nums + nums[::-1]


print(Solution().concatWithReverse([1, 2, 3]), Solution().concatWithReverse([1]))
````

## 10. Vì sao thuật toán đúng

1. **Độ dài.** `len(nums + nums[::-1]) = n + n = 2n`.
2. **Nửa đầu.** Các vị trí `0..n-1` của kết quả là các phần tử của `nums` theo thứ tự: `ans[i] = nums[i]`.
3. **Nửa sau.** Các vị trí `n..2n-1` là `nums[::-1]`, phần tử thứ `i` của nó là `nums[n - 1 - i]`; do đó `ans[i + n] = nums[n - 1 - i]`, đúng công thức của đề.
4. **Không sửa đầu vào.** `+` và `[::-1]` đều tạo list mới, không thay đổi `nums`.
5. **Tương đương giữa các cách.** Cách 1 gán theo công thức, Cách 4 dùng đối xứng `2n - 1 - k`, các cách còn lại nối hai dãy; tất cả mô tả cùng một dãy `ans`.

## 11. Phân tích Time / Space Complexity

Gọi `n = len(nums)`.

- Cách 1: `O(n)` thời gian, `O(n)` bộ nhớ.
- Cách 2: `O(n)` thời gian, `O(n)` bộ nhớ (lát cắt đảo ngược `O(n)` cộng kết quả `O(2n)`).
- Cách 3, 5: `O(n)` thời gian, `O(n)` bộ nhớ (không list phụ).
- Cách 4: `O(n)` thời gian, `O(n)` bộ nhớ.

Không thể nhanh hơn `O(n)` vì phải tạo `2n` phần tử đầu ra. Không có best/worst khác nhau.

## 12. Bộ test

Đối chiếu 5 cách + code nộp với ví dụ đề, biên (độ dài 1, 2, 100) và 400 input ngẫu nhiên đúng constraints, đồng thời kiểm tra các tính chất: độ dài `2n`, palindrome, **mảng đầu vào không bị thay đổi**.

````python
import random

solutions = [SolutionBrute(), SolutionSlice(), SolutionExtend(), SolutionComprehension(), SolutionChain(), Solution()]

fixed = [
    ([1, 2, 3], [1, 2, 3, 3, 2, 1]),
    ([1], [1, 1]),
    ([4, 5, 6], [4, 5, 6, 6, 5, 4]),
    ([1, 2], [1, 2, 2, 1]),
    ([7, 7, 7], [7, 7, 7, 7, 7, 7]),
    ([100, 1], [100, 1, 1, 100]),
    (list(range(1, 6)), [1, 2, 3, 4, 5, 5, 4, 3, 2, 1]),
]
for nums, expected in fixed:
    for s in solutions:
        arg = list(nums)
        assert s.concatWithReverse(arg) == expected, (type(s).__name__, nums)
        assert arg == nums, "đầu vào bị thay đổi: " + type(s).__name__

random.seed(39250)
for _ in range(400):
    n = random.randint(1, 100)
    nums = [random.randint(1, 100) for _ in range(n)]
    outs = []
    for s in solutions:
        arg = list(nums)
        outs.append(s.concatWithReverse(arg))
        assert arg == nums
    assert all(o == outs[0] for o in outs)
    ans = outs[0]
    assert len(ans) == 2 * n and ans == ans[::-1]
    assert ans[:n] == nums and all(ans[i + n] == nums[n - i - 1] for i in range(n))
print("tất cả test OK")
````

### Test thuộc tính

(1) Kết quả luôn là palindrome độ dài chẵn. (2) Áp dụng cho `ans` lần nữa: `ans + ans[::-1] == ans + ans` vì `ans` là palindrome. (3) Tổng các phần tử của `ans` bằng `2 * sum(nums)`; số lần xuất hiện của mỗi giá trị nhân đôi.

````python
from collections import Counter
random.seed(3925)
for _ in range(300):
    nums = [random.randint(1, 100) for _ in range(random.randint(1, 100))]
    ans = Solution().concatWithReverse(nums)
    assert ans == ans[::-1] and len(ans) % 2 == 0
    assert Solution().concatWithReverse(ans) == ans + ans
    assert sum(ans) == 2 * sum(nums)
    assert Counter(ans) == Counter({k: 2 * v for k, v in Counter(nums).items()})
print("property test OK")
````

## 13. Các lỗi người mới hay gặp

### Lỗi 1: dùng `nums.reverse()` trong biểu thức (trả về `None`)

`list.reverse()` đảo tại chỗ và trả `None`; `nums + nums.reverse()` gây `TypeError`, và còn làm hỏng `nums`.

````python
def wrong_reverse(nums):
    return nums + nums.reverse()                                     # SAI: reverse() trả về None

try:
    wrong_reverse([1, 2, 3])
except TypeError as e:
    print("TypeError:", e)
assert Solution().concatWithReverse([1, 2, 3]) == [1, 2, 3, 3, 2, 1]
````

### Lỗi 2: dùng `nums * 2` (lặp, không đảo)

`[1,2,3] * 2` là `[1,2,3,1,2,3]`: cùng thứ tự, khác đề.

````python
def wrong_repeat(nums):
    return nums * 2                                                  # SAI: lặp, không đảo

print(wrong_repeat([1, 2, 3]), "!=", Solution().concatWithReverse([1, 2, 3]))
assert wrong_repeat([1, 2, 3]) == [1, 2, 3, 1, 2, 3] and Solution().concatWithReverse([1, 2, 3]) == [1, 2, 3, 3, 2, 1]
assert wrong_repeat([1]) == Solution().concatWithReverse([1])       # trùng khi n = 1 nên dễ lọt ví dụ 2
````

### Lỗi 3: `append` thay vì `extend` (list lồng)

`ans.append(nums[::-1])` thêm **một** phần tử là list con, ra `[1, 2, 3, [3, 2, 1]]` (độ dài `n + 1`).

````python
def wrong_append(nums):
    ans = list(nums)
    ans.append(nums[::-1])                                           # SAI: phải extend
    return ans

print(wrong_append([1, 2, 3]), "!=", Solution().concatWithReverse([1, 2, 3]))
assert wrong_append([1, 2, 3]) == [1, 2, 3, [3, 2, 1]] and len(wrong_append([1, 2, 3])) == 4
````

### Lỗi 4: aliasing — sửa trực tiếp mảng đầu vào

`ans = nums` rồi `ans.extend(...)` làm `nums` dài ra; hàm trả về chính đối tượng đầu vào đã bị sửa. Trên LeetCode điều này thường vẫn qua test nhưng là thói quen xấu và gây lỗi khi `nums` được dùng lại.

````python
def wrong_inplace(nums):
    ans = nums                                                       # SAI: alias, không sao chép
    ans.extend(reversed(nums))
    return ans

original = [1, 2, 3]
out = wrong_inplace(original)
print(out, "| đầu vào sau khi gọi:", original)
assert original == [1, 2, 3, 3, 2, 1] and out is original
safe = [1, 2, 3]
assert Solution().concatWithReverse(safe) == [1, 2, 3, 3, 2, 1] and safe == [1, 2, 3]
````

### Lỗi 5: lệch chỉ số khi viết theo công thức

Viết `ans[i + n] = nums[n - i]` (thiếu `- 1`) gây `IndexError` tại `i = 0` (`nums[n]` ngoài biên).

````python
def wrong_index(nums):
    n = len(nums)
    ans = [0] * (2 * n)
    for i in range(n):
        ans[i] = nums[i]
        ans[i + n] = nums[n - i]                                     # SAI: phải nums[n - i - 1]
    return ans

try:
    wrong_index([1, 2, 3])
except IndexError as e:
    print("IndexError:", e)
assert Solution().concatWithReverse([1, 2, 3]) == [1, 2, 3, 3, 2, 1]
````

## 14. Mẹo làm bài nhanh & nhận dạng pattern

- "Nối một dãy với bản đảo của nó" -> `a + a[::-1]`.
- Nhớ ba thứ: `a[::-1]` (list mới), `reversed(a)` (iterator), `a.reverse()` (tại chỗ, trả `None`).
- Muốn **không sửa đầu vào**, dùng phép tạo list mới (`+`, lát cắt) hoặc `list(a)` trước khi `extend`.
- Công thức đối xứng: phần tử `k` của dãy dài `2n` ở nửa sau tương ứng `2n - 1 - k`.
- Kiểm tra bằng tính chất: kết quả là **palindrome** độ dài chẵn.

## 15. Biến thể của bài

1. **Concatenation of Array** (1929): `nums + nums` (không đảo).
2. **Reverse String** (344), **Rotate Array** (189): đảo/xoay mảng.
3. **Palindrome Linked List** (234), **Valid Palindrome** (125): kiểm tra đối xứng.
4. **Shuffle the Array** (1470): ghép xen kẽ hai nửa.
5. Biến thể: nối `nums` với bản đảo **bỏ phần tử đầu** (để không lặp phần tử giữa), hay nối `k` lần luân phiên xuôi/ngược ("zigzag").
6. Biến thể: yêu cầu thực hiện **tại chỗ** với `O(1)` bộ nhớ phụ nếu mảng đã được cấp phát dung lượng `2n`.

| 3. `list` + `extend(reversed)` | sao chép rồi nối | `O(n)` | `O(n)` | Khi muốn tránh list phụ |
| 4. Comprehension | `k < n` hay `2n-1-k` | `O(n)` | `O(n)` | Làm rõ phép đối xứng |
| 5. `chain` | nối iterable | `O(n)` | `O(n)` | Khi dữ liệu là iterator/lớn |

## 19. Đáp án tham khảo cho bài tập

````python
# Bài 1, 2
assert Solution().concatWithReverse([2, 0, 9]) == [2, 0, 9, 9, 0, 2] and Solution().concatWithReverse([5, 5]) == [5, 5, 5, 5]
a = Solution().concatWithReverse([1, 2, 3, 4])
assert a == [1, 2, 3, 4, 4, 3, 2, 1] and a[4] == 4 and a[7] == 1

# Bài 3
def concat_with_reverse_skip_mid(nums):
    return nums + nums[::-1][1:]
assert concat_with_reverse_skip_mid([1, 2, 3]) == [1, 2, 3, 2, 1]
assert concat_with_reverse_skip_mid([1]) == [1] and concat_with_reverse_skip_mid([4, 5]) == [4, 5, 4]

# Bài 4
def concat_zigzag(nums, k):
    out = []
    for t in range(k):
        out += nums if t % 2 == 0 else nums[::-1]
    return out
assert concat_zigzag([1, 2], 3) == [1, 2, 2, 1, 1, 2]
assert concat_zigzag([1, 2, 3], 2) == Solution().concatWithReverse([1, 2, 3]) and concat_zigzag([1], 0) == []

# Bài 5
def is_valid_concat_with_reverse(ans):
    return len(ans) % 2 == 0 and ans == ans[::-1]
assert is_valid_concat_with_reverse([1, 2, 2, 1]) and not is_valid_concat_with_reverse([1, 2, 1])
assert not is_valid_concat_with_reverse([1, 2, 3, 4])
random.seed(23)
for _ in range(100):
    nums = [random.randint(1, 100) for _ in range(random.randint(1, 30))]
    assert is_valid_concat_with_reverse(Solution().concatWithReverse(nums))
print("đáp án bài tập OK")
````
