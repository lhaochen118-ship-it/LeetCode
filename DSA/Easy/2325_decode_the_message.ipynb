## 1. Mục tiêu học

Sau notebook này bạn sẽ:

1. Hiểu **bảng thay thế (substitution cipher)**: mỗi chữ cái được thay bằng đúng một chữ cái khác theo một bảng cố định.
2. Biết lập bảng thay thế từ `key` theo **thứ tự xuất hiện đầu tiên** của từng chữ cái, bằng `dict` và biến đếm.
3. Biết dùng `dict.get`, `str.join`, `str.maketrans` / `str.translate`, `string.ascii_lowercase`, `ord`/`chr` và `dict.fromkeys` (giữ thứ tự chèn).
4. Viết 5 cách: dict + vòng lặp, `translate`, mảng 26 ô với `ord`, `dict.fromkeys` + `zip`, và duyệt `key` bằng danh sách đã thấy.
5. Trace, test ngẫu nhiên đối chiếu, và tránh lỗi: quên giữ khoảng trắng, ghi đè mapping khi chữ cái lặp lại, gán theo thứ tự `a, b, c...` thay vì theo thứ tự trong `key`.

## 2. Đề bài & ví dụ

Cho hai chuỗi `key` và `message`. `key` là **khóa**, `message` là **tin nhắn mã hóa**. Giải mã `message` như sau:

1. Duyệt `key`, lấy **lần xuất hiện đầu tiên** của mỗi chữ cái thường theo thứ tự. Xếp các chữ cái đó thành một dãy 26 chữ cái (đề đảm bảo `key` chứa đủ 26 chữ cái).
2. Gán chữ cái thứ nhất trong dãy cho `'a'`, thứ hai cho `'b'`, ..., thứ 26 cho `'z'`: tức bảng `thứ i trong dãy -> chữ thứ i trong bảng chữ cái`.
3. Thay mỗi chữ cái trong `message` theo bảng; **khoảng trắng giữ nguyên**.

Trả về `message` đã giải mã.

| Ví dụ | key | message | Output |
|---|---|---|---|
| 1 | `"the quick brown fox jumps over the lazy dog"` | `"vkbs bs t suepuv"` | `"this is a secret"` |
| 2 | `"eljuxhpwnyrdgtqkviszcfmabo"` | `"zwx hnfx lqantp mnoeius ycgk vcnjrdb"` | `"the five boxing wizards jump quickly"` |

Constraints: `26 <= len(key) <= 2000`; `key` chỉ gồm chữ thường và khoảng trắng, **chứa đủ 26 chữ cái**; `1 <= len(message) <= 2000`; `message` chỉ gồm chữ thường và khoảng trắng.

````python
key = "the quick brown fox jumps over the lazy dog"
seen = []
for ch in key:
    if ch != " " and ch not in seen:
        seen.append(ch)
print("thứ tự xuất hiện đầu tiên:", "".join(seen), len(seen))
table = dict(zip(seen, "abcdefghijklmnopqrstuvwxyz"))
print(table)
print("".join(table.get(c, c) for c in "vkbs bs t suepuv"))
````

````python
d = {"t": "a", "h": "b"}
print(d["t"], d.get("h"), d.get("z"), d.get("z", "?"))
print("t" in d, "z" in d)
d["e"] = "c"
print(d, len(d))
````

````python
import string

print(string.ascii_lowercase)
print(ord("a"), ord("z"), chr(97), chr(97 + 25))
print([chr(ord("a") + i) for i in range(5)])
````

````python
table = str.maketrans("abc", "xyz")
print("a b c d".translate(table))
print("hello world".translate(str.maketrans("lo", "01")))
````

## 4. Ý tưởng nền tảng / cách nghĩ trực giác

Lấy `key = "the quick brown fox jumps over the lazy dog"`. Đọc từ trái sang phải, mỗi lần gặp **chữ cái mới** ta gán chữ tiếp theo của bảng chữ cái:

- `t` là chữ mới đầu tiên -> `t` giải mã thành `a`.
- `h` mới -> `b`; `e` mới -> `c`; khoảng trắng bỏ qua; `q` -> `d`; `u` -> `e`; `i` -> `f`; `c` -> `g`; `k` -> `h`; ...
- Chữ đã gặp rồi (vd `e` trong `over`) **không** gán lại.

Giải mã `"vkbs"`: `v -> t`, `k -> h`, `b -> i`, `s -> s` ra `"this"`. Khoảng trắng đứng nguyên nên `"vkbs bs t suepuv"` ra `"this is a secret"`.

Vậy bài gồm hai bước: (1) **lập bảng** theo thứ tự xuất hiện đầu tiên, (2) **thay thế** từng ký tự của `message`.

````python
key = "the quick brown fox jumps over the lazy dog"
table, nxt = {}, 0
for ch in key:
    if ch != " " and ch not in table:
        table[ch] = chr(ord("a") + nxt)
        nxt += 1
print("bảng (5 cặp đầu):", list(table.items())[:5], "... tổng", len(table))
print("".join(table.get(c, c) for c in "vkbs bs t suepuv"))
````

## 5. Cách 1 — Brute Force / baseline: dict + biến đếm

**Ý tưởng:** duyệt `key`, mỗi chữ cái chưa có trong `table` thì gán `chr(ord('a') + nxt)` và `nxt += 1`. Rồi dịch `message` từng ký tự bằng `table.get(c, c)`.

**Độ phức tạp:** `O(|key| + |message|)` thời gian, `O(26)` bộ nhớ cho bảng. Đây đã tối ưu; "nút thắt" duy nhất là viết dài; các cách sau cho code gọn hơn.

````python
class SolutionDict:
    def decodeMessage(self, key: str, message: str) -> str:
        table = {}
        nxt = 0                                         # chữ cái đúng kế tiếp sẽ gán
        for ch in key:
            if ch != " " and ch not in table:
                table[ch] = chr(ord("a") + nxt)
                nxt += 1
        out = []
        for ch in message:
            out.append(table.get(ch, ch))               # khoảng trắng không có trong table nên giữ nguyên
        return "".join(out)

print(SolutionDict().decodeMessage("the quick brown fox jumps over the lazy dog", "vkbs bs t suepuv"))
print(SolutionDict().decodeMessage("eljuxhpwnyrdgtqkviszcfmabo", "zwx hnfx lqantp mnoeius ycgk vcnjrdb"))
````

## 6. Ý tưởng thuật toán / bước đột phá

**Insight:** bảng thay thế chính là một **phép ghép** giữa "dãy chữ cái theo thứ tự xuất hiện lần đầu trong `key`" và "bảng chữ cái `a..z`". Có sẵn công cụ cho cả hai việc:

- `dict.fromkeys(key.replace(' ', ''))` lấy dãy chữ cái duy nhất theo thứ tự xuất hiện (đủ 26 chữ);
- `zip(dãy_đó, ascii_lowercase)` ghép theo thứ tự;
- `str.maketrans` + `translate` thay nguyên chuỗi và giữ khoảng trắng.

**Invariant:** sau bước lập bảng, `table[c]` là chữ cái thứ `k` trong bảng chữ cái, với `k` là số thứ tự xuất hiện lần đầu của `c` trong `key` (đếm từ 0); các chữ cái khác nhau có ảnh khác nhau.

**Quy tắc:** bỏ khoảng trắng, loại trùng giữ thứ tự, ghép với `a..z`, thay thế.

## 7. Các cách tiếp theo

### Cách 2 — `dict.fromkeys` + `zip` + `get`

Gọn: lập bảng trong một dòng rồi dịch bằng list comprehension.

````python
import string

class SolutionZip:
    def decodeMessage(self, key: str, message: str) -> str:
        order = dict.fromkeys(key.replace(" ", ""))               # chữ cái duy nhất theo thứ tự xuất hiện
        table = dict(zip(order, string.ascii_lowercase))
        return "".join(table.get(ch, ch) for ch in message)

print(SolutionZip().decodeMessage("the quick brown fox jumps over the lazy dog", "vkbs bs t suepuv"))
print(SolutionZip().decodeMessage("eljuxhpwnyrdgtqkviszcfmabo", "zwx hnfx lqantp mnoeius ycgk vcnjrdb"))
````

**Độ phức tạp Cách 2:** Time `O(|key| + |message|)`, Space `O(26 + |message|)`.

### Cách 3 — `str.maketrans` và `str.translate`

Tạo `from_chars` là dãy chữ cái theo thứ tự xuất hiện (chuỗi 26 ký tự), `to_chars = ascii_lowercase`. Một lệnh `translate` thay cả `message`; ký tự không có trong bảng (khoảng trắng) tự giữ.

````python
import string

class SolutionTranslate:
    def decodeMessage(self, key: str, message: str) -> str:
        order = "".join(dict.fromkeys(key.replace(" ", "")))      # 26 chữ cái theo thứ tự xuất hiện
        return message.translate(str.maketrans(order, string.ascii_lowercase))

print(SolutionTranslate().decodeMessage("the quick brown fox jumps over the lazy dog", "vkbs bs t suepuv"))
print(SolutionTranslate().decodeMessage("eljuxhpwnyrdgtqkviszcfmabo", "zwx hnfx lqantp mnoeius ycgk vcnjrdb"))
````

**Độ phức tạp Cách 3:** Time `O(|key| + |message|)`, Space `O(26 + |message|)`. `translate` chạy bằng C nên nhanh nhất thực tế.

### Cách 4 — Mảng 26 ô với `ord`

Không dùng `dict`: mảng `mp = [''] * 26` (hoặc `[None] * 26`) chỉ số `ord(c) - 97`. Gán khi ô còn trống.

````python
class SolutionArray:
    def decodeMessage(self, key: str, message: str) -> str:
        mp = [""] * 26
        nxt = 0
        for ch in key:
            if ch == " ":
                continue
            idx = ord(ch) - 97
            if mp[idx] == "":                      # lần đầu thấy chữ này
                mp[idx] = chr(97 + nxt)
                nxt += 1
        return "".join(" " if ch == " " else mp[ord(ch) - 97] for ch in message)

print(SolutionArray().decodeMessage("the quick brown fox jumps over the lazy dog", "vkbs bs t suepuv"))
print(SolutionArray().decodeMessage("eljuxhpwnyrdgtqkviszcfmabo", "zwx hnfx lqantp mnoeius ycgk vcnjrdb"))
````

**Độ phức tạp Cách 4:** Time `O(|key| + |message|)`, Space `O(26 + |message|)`. Không băm, chỉ truy cập mảng; cực nhanh với bảng chữ cái nhỏ.

### Cách 5 — Dừng sớm khi đủ 26 chữ

Vì `key` luôn đủ 26 chữ cái nên khi bảng đã có 26 phần tử thì dừng duyệt `key` (không cần đọc phần còn lại). Khi `key` dài tới 2000 ký tự điều này tiết kiệm công việc.

````python
import string

class SolutionEarlyStop:
    def decodeMessage(self, key: str, message: str) -> str:
        table = {}
        for ch in key:
            if ch != " " and ch not in table:
                table[ch] = string.ascii_lowercase[len(table)]   # chữ cái thứ len(table)
                if len(table) == 26:
                    break                                         # đủ rồi, khỏi duyệt thêm
        return "".join(table.get(ch, ch) for ch in message)

print(SolutionEarlyStop().decodeMessage("the quick brown fox jumps over the lazy dog", "vkbs bs t suepuv"))
print(SolutionEarlyStop().decodeMessage("eljuxhpwnyrdgtqkviszcfmabo" + " abc" * 100, "zwx hnfx"))
````

**Độ phức tạp Cách 5:** Time `O(|key|)` trong trường hợp xấu nhất nhưng thường dừng sớm, cộng `O(|message|)`; Space `O(26 + |message|)`. Dùng `len(table)` thay biến đếm riêng.

## 8. TRACE từng bước

Trace lập bảng với `key = "the quick brown fox jumps over the lazy dog"` (chỉ 10 chữ cái mới đầu tiên):

| Ký tự `key` | Chữ mới? | `nxt` | `table` thêm |
|---|---|---|---|
| `t` | có | 0 | `t -> a` |
| `h` | có | 1 | `h -> b` |
| `e` | có | 2 | `e -> c` |
| ` ` | bỏ qua | | |
| `q` | có | 3 | `q -> d` |
| `u` | có | 4 | `u -> e` |
| `i` | có | 5 | `i -> f` |
| `c` | có | 6 | `c -> g` |
| `k` | có | 7 | `k -> h` |
| ` ` | bỏ qua | | |
| `b` | có | 8 | `b -> i` |

Giải mã: `v -> t`, `k -> h`, `b -> i`, `s -> s`: `"vkbs"` = `"this"`.

````python
def trace_table(key, upto=12):
    table, nxt = {}, 0
    for i, ch in enumerate(key):
        if ch == " ":
            print(f"  i={i:2}: ' ' bỏ qua")
            continue
        if ch in table:
            print(f"  i={i:2}: {ch!r} đã có -> {table[ch]!r} (không gán lại)")
            continue
        table[ch] = chr(97 + nxt)
        nxt += 1
        print(f"  i={i:2}: {ch!r} mới -> {table[ch]!r}")
        if nxt >= upto:
            print("  ... (dừng in)")
            break
    return table

t = trace_table("the quick brown fox jumps over the lazy dog")
````

````python
def trace_decode(table_key, message):
    order = "".join(dict.fromkeys(table_key.replace(" ", "")))
    table = dict(zip(order, "abcdefghijklmnopqrstuvwxyz"))
    out = []
    for ch in message:
        out.append(table.get(ch, ch))
        print(f"  {ch!r} -> {out[-1]!r}")
    return "".join(out)

print("kết quả:", trace_decode("the quick brown fox jumps over the lazy dog", "vkbs bs t"))
````

## 9. Code nộp LeetCode

````python
import string

class Solution:
    def decodeMessage(self, key: str, message: str) -> str:
        order = "".join(dict.fromkeys(key.replace(" ", "")))
        return message.translate(str.maketrans(order, string.ascii_lowercase))

print(Solution().decodeMessage("the quick brown fox jumps over the lazy dog", "vkbs bs t suepuv"))
print(Solution().decodeMessage("eljuxhpwnyrdgtqkviszcfmabo", "zwx hnfx lqantp mnoeius ycgk vcnjrdb"))
````

- **Thứ tự lần đầu.** Ta chỉ gán khi gặp chữ lần đầu (`not in table`), nên chữ xuất hiện sớm nhất nhận chữ cái nhỏ nhất còn lại — đúng yêu cầu.
- **Giữ khoảng trắng.** Khoảng trắng không nằm trong bảng (đã loại khỏi `key`), nên `get(c, c)` / `translate` giữ nguyên.
- **Giải mã từng ký tự độc lập.** Phép thay thế đơn ký tự không phụ thuộc vị trí hay ngữ cảnh, nên dịch từng ký tự rồi nối là đúng.

## 13. Các lỗi người mới hay gặp

### Lỗi 1: Ghi đè bảng khi chữ cái lặp lại trong `key`

Nếu không kiểm tra `ch not in table` thì chữ lặp lại sẽ bị gán lại chữ cái đích khác, làm hỏng bảng.

````python
def sai_ghi_de(key, message):
    table, nxt = {}, 0
    for ch in key:
        if ch == " ":
            continue
        table[ch] = chr(97 + nxt)        # SAI: gán lại cả khi đã có
        nxt += 1
    return "".join(table.get(c, c) for c in message)

k = "the quick brown fox jumps over the lazy dog"
print("sai :", sai_ghi_de(k, "vkbs bs t suepuv"))
print("đúng:", Solution().decodeMessage(k, "vkbs bs t suepuv"))
````

### Lỗi 2: Quên khoảng trắng trong `key` khi đếm chữ cái mới

Nếu cũng gán cho `' '` thì khoảng trắng chiếm một chữ cái đích và mọi chữ sau lệch 1 (đến cuối vượt quá `z`). Phải `continue` khi `ch == ' '`.

````python
def sai_gan_khoang_trang(key, message):
    table, nxt = {}, 0
    for ch in key:
        if ch not in table:               # SAI: khoảng trắng cũng được gán
            table[ch] = chr(97 + nxt)
            nxt += 1
    return "".join(table.get(c, c) for c in message)

k = "the quick brown fox jumps over the lazy dog"
print("sai :", sai_gan_khoang_trang(k, "vkbs bs t suepuv"))
print("đúng:", Solution().decodeMessage(k, "vkbs bs t suepuv"))
````

### Lỗi 3: Gán theo thứ tự bảng chữ cái thay vì thứ tự trong `key`

Một số người sắp xếp `set(key)` rồi ghép với `a..z`; thứ tự `set` không phải thứ tự xuất hiện, kết quả sai.

````python
def sai_sorted_set(key, message):
    letters = sorted(set(key.replace(" ", "")))       # SAI: sắp xếp theo chữ cái, không theo xuất hiện
    table = dict(zip(letters, "abcdefghijklmnopqrstuvwxyz"))
    return "".join(table.get(c, c) for c in message)

k = "the quick brown fox jumps over the lazy dog"
print("sai :", sai_sorted_set(k, "vkbs bs t suepuv"))
print("đúng:", Solution().decodeMessage(k, "vkbs bs t suepuv"))
````

### Lỗi 4: Tra `table[ch]` thay vì `table.get(ch, ch)` khi `message` có khoảng trắng

Khoảng trắng không có trong bảng nên `table[' ']` gây `KeyError`.

````python
table = {"a": "x"}
try:
    print("".join(table[c] for c in "a a"))
except KeyError as e:
    print("KeyError:", e)
print("".join(table.get(c, c) for c in "a a"))
````

## 14. Mẹo làm bài nhanh & nhận dạng pattern

- Đề "thay mỗi chữ cái bằng chữ khác theo bảng" -> **`str.translate`** hoặc `dict.get`.
- "Thứ tự xuất hiện lần đầu, bỏ trùng" -> `dict.fromkeys(...)` (Python 3.7+) hoặc vòng `if ch not in table`.
- Bảng chữ cái cố định `a..z` -> mảng 26 ô hoặc `string.ascii_lowercase`.
- Ký tự lạ (khoảng trắng) cần giữ nguyên -> `dict.get(c, c)` hoặc `translate` (tự bỏ qua ký tự không có trong bảng).
- Pattern gặp lại ở: Isomorphic Strings, Word Pattern, Verifying an Alien Dictionary.

## 15. Biến thể của bài

- **Mã hóa** (chiều ngược): bảng `a -> chữ đầu tiên trong key`, ... (nghịch đảo bảng).
- **Cho phép chữ hoa:** giữ hoa/thường bằng cách ánh xạ cả hai.
- **`key` thiếu chữ cái:** chữ thiếu giữ nguyên hoặc báo lỗi.
- **Nhiều khóa (hai lớp mã):** áp bảng thứ nhất rồi bảng thứ hai (hợp thành hoán vị).
- **Phá mã:** không biết `key`, đoán bảng bằng thống kê tần suất chữ cái.

## 16. So sánh các cách làm

| Cách | Ý tưởng | Time | Space | Khi nào dùng |
|---|---|---|---|---|
| 1. `dict` + biến đếm | gán khi chữ mới | `O(k+m)` | `O(26)` | dễ hiểu nhất, người mới |
| 2. `fromkeys` + `zip` | ghép dãy duy nhất với `a..z` | `O(k+m)` | `O(26)` | code ngắn |
| 3. `translate` | `maketrans` rồi dịch cả chuỗi | `O(k+m)` | `O(26)` | nộp bài, nhanh |
| 4. Mảng 26 ô | `ord(c) - 97` | `O(k+m)` | `O(26)` | không cần băm |
| 5. Dừng sớm | `break` khi đủ 26 | `O(k'+m)` | `O(26)` | `key` rất dài |

## 18. Gợi ý 3 mức cho bài tập

- **Mức 1 (nhẹ):** Bài 1: trong `key` đảo ngược, `z` đứng đầu nên `z -> a`, ..., còn `a` đứng cuối nên `a -> z`, `b -> y`, `c -> x`; vậy `"abc"` thành `"zyx"`. Bài 2: `c -> a`, `a -> b`, `b -> c` nên `"cabcab"` thành `"abcabc"`.
- **Mức 2 (vừa):** Bài 3: dùng `dict` và `ch not in table`. Bài 4: đảo `table`: `{v: k for k, v in table.items()}`.
- **Mức 3 (sâu):** Bài 5: `len(set(key.replace(' ', ''))) == 26` và kiểm tra mọi ký tự là chữ thường hoặc khoảng trắng.

## 19. Đáp án tham khảo cho bài tập

Chạy kiểm tra bằng `assert`.

````python
import string

# Bài 1 và 2
assert Solution().decodeMessage("zyxwvutsrqponmlkjihgfedcba", "abc") == "zyx"
assert Solution().decodeMessage("cabdefghijklmnopqrstuvwxyz", "cabcab") == "abcabc"

# Bài 3
def build_table(key):
    table = {}
    for ch in key:
        if ch != " " and ch not in table:
            table[ch] = string.ascii_lowercase[len(table)]
    return table

# Bài 4
def encode_message(key, plain):
    inv = {v: k for k, v in build_table(key).items()}
    return "".join(inv.get(c, c) for c in plain)

# Bài 5
def is_valid_key(key):
    letters = key.replace(" ", "")
    return all("a" <= c <= "z" for c in letters) and len(set(letters)) == 26

assert build_table("bca" + "defghijklmnopqrstuvwxyz")["b"] == "a"
assert len(build_table("the quick brown fox jumps over the lazy dog")) == 26
k = "the quick brown fox jumps over the lazy dog"
assert encode_message(k, "this is a secret") == "vkbs bs t suepuv"
assert Solution().decodeMessage(k, encode_message(k, "hello world")) == "hello world"
assert is_valid_key(k) and not is_valid_key("abc") and not is_valid_key("a" * 26)
print("tất cả bài tập OK")
````

import string

class Solution:
    def decodeMessage(self, key: str, message: str) -> str:
        order = "".join(dict.fromkeys(key.replace(" ", "")))
        return message.translate(str.maketrans(order, string.ascii_lowercase))
