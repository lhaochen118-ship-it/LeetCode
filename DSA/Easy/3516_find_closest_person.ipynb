## 1. Mục tiêu học

Sau notebook này bạn sẽ:

1. Hiểu đề: hai người đi cùng tốc độ về phía người thứ ba đứng yên; ai tới trước thì thắng, tới cùng lúc thì hòa — tất cả quy về **so sánh hai khoảng cách**.
2. Biết tính khoảng cách trên trục số bằng **giá trị tuyệt đối** `abs(a - b)`.
3. Biết viết so sánh ba trường hợp (`<`, `>`, `==`) thành `if/elif/else`, thành tra bảng bằng chỉ số âm, hoặc bằng `min` trên các cặp `(khoảng cách, nhãn)`.
4. Viết 4 cách: so sánh trực tiếp hai khoảng cách, **mô phỏng từng giây**, mẹo `(d > 0) - (d < 0)` kết hợp tra bảng, và dùng `sorted` các cặp `(khoảng cách, id)` kèm kiểm tra hòa.
5. Tránh các lỗi: quên `abs`, trả sai khi hòa, so sánh **vị trí** thay vì **khoảng cách**, và trả về khoảng cách thay vì mã người.

## 2. Đề bài & ví dụ

Ba người đứng trên **trục số** tại các vị trí nguyên: Người 1 ở `x`, Người 2 ở `y`, Người 3 ở `z`. Người 3 **đứng yên**. Người 1 và Người 2 **cùng bắt đầu** đi về phía Người 3 với **cùng tốc độ** (mỗi giây đi `1` đơn vị).

Trả về:

- `1` nếu Người 1 tới chỗ Người 3 **trước**;
- `2` nếu Người 2 tới trước;
- `0` nếu hai người tới **cùng lúc**.

| Ví dụ | x, y, z | Output | Giải thích |
|---|---|---|---|
| 1 | `2, 7, 4` | `1` | Người 1 cách Người 3 `2`, Người 2 cách `3` -> Người 1 tới trước |
| 2 | `2, 5, 6` | `2` | Người 1 cách `4`, Người 2 cách `1` -> Người 2 tới trước |
| 3 | `1, 5, 3` | `0` | cả hai cách `2` -> cùng lúc |

Constraints: `1 <= x, y, z <= 100`.

Chú ý: hai người **không** nhất thiết ở hai phía khác nhau của Người 3; họ có thể cùng ở bên trái hoặc bên phải. Kết quả chỉ phụ thuộc vào **khoảng cách**, không phụ thuộc phía.

````python
examples = [((2, 7, 4), 1), ((2, 5, 6), 2), ((1, 5, 3), 0)]
for (x, y, z), expected in examples:
    d1, d2 = abs(x - z), abs(y - z)
    result = 1 if d1 < d2 else (2 if d2 < d1 else 0)
    print(f"x={x}, y={y}, z={z}: khoảng cách {d1} và {d2} -> {result} (mong đợi {expected})")
````

## 3. Kiến thức nền

### 3.1. Khoảng cách trên trục số và giá trị tuyệt đối

Khoảng cách giữa hai điểm `a` và `b` trên trục số là `|a - b|` (không âm). Python viết `abs(a - b)`. Ví dụ điểm `2` và `7`: `abs(2 - 7) = 5`, và `abs(7 - 2) = 5` (đối xứng).

````python
print(abs(2 - 7), abs(7 - 2), abs(4 - 4))
print(2 - 4, abs(2 - 4))        # hiệu có thể âm, khoảng cách thì không
````

### 3.2. Vận tốc bằng nhau => thời gian tỉ lệ với khoảng cách

Nếu đi cùng tốc độ `v`, thời gian đến nơi là `khoảng_cách / v`. Vì `v` chung nên **khoảng cách nhỏ hơn <=> tới sớm hơn**. Ta không cần nhân hay chia gì cả, chỉ so sánh hai khoảng cách. Ở đây `v = 1` đơn vị/giây nên thời gian **bằng đúng** khoảng cách (tính bằng giây).

````python
v = 1
for (x, y, z) in [(2, 7, 4), (2, 5, 6)]:
    t1, t2 = abs(x - z) / v, abs(y - z) / v
    print((x, y, z), "thời gian:", t1, t2)
````

### 3.3. `if / elif / else` cho ba trường hợp

Có đúng ba khả năng loại trừ nhau: `d1 < d2`, `d1 > d2`, `d1 == d2`. Hai nhánh đầu cho `1` và `2`, nhánh `else` (bằng nhau) cho `0`. Thứ tự các nhánh không quan trọng, miễn là phủ hết ba khả năng.

````python
def compare(d1, d2):
    if d1 < d2:
        return 1
    elif d1 > d2:
        return 2
    else:
        return 0

print(compare(2, 3), compare(4, 1), compare(2, 2))
````

### 3.4. Chỉ số âm của danh sách và `True`/`False` là `1`/`0`

`lst[-1]` là phần tử **cuối**. Trong Python `True == 1` và `False == 0`, nên `(d > 0) - (d < 0)` cho `1` nếu `d > 0`, `-1` nếu `d < 0`, `0` nếu `d == 0` (hàm dấu `sign`). Kết hợp với `[0, 2, 1][sign]`: chỉ số `0 -> 0`, `1 -> 2`, `-1 -> 1` (phần tử cuối).

````python
table = [0, 2, 1]
for d in (-5, 0, 7):
    sign = (d > 0) - (d < 0)
    print(f"d={d:>2} sign={sign:>2} -> table[{sign}] = {table[sign]}")
````

### 3.5. `sorted` / `min` trên danh sách tuple

Python so sánh tuple **từng phần tử từ trái sang phải**. `min([(3, 1), (2, 2)])` là `(2, 2)`: phần tử đầu nhỏ hơn. Ta dùng `(khoảng_cách, id_người)` để lấy người gần nhất, và so hai khoảng cách đầu tiên để phát hiện hòa.

````python
cands = sorted([(abs(2 - 4), 1), (abs(7 - 4), 2)])
print(cands)
print(cands[0][0] == cands[1][0])         # có hòa không?
````

## 4. Ý tưởng nền tảng / cách nghĩ trực giác

Vẽ trục số cho ví dụ 1 (`x = 2`, `y = 7`, `z = 4`):

```
vị trí:  1   2   3   4   5   6   7   8
              x       z               y
               --2-->             <--3--
```

Người 1 cần `2` bước, Người 2 cần `3` bước. Sau `2` giây Người 1 đã tới, Người 2 còn cách `1`. Vậy Người 1 thắng.

Với ví dụ 3 (`x = 1`, `y = 5`, `z = 3`): hai người ở hai phía, cùng cách `2` -> tới **cùng một giây** (`t = 2`) -> `0`.

Khi hai người ở **cùng phía** (ví dụ `x = 2`, `y = 5`, `z = 6`) ta vẫn chỉ cần so `abs(x - z) = 4` với `abs(y - z) = 1`. Không cần quan tâm họ có cản nhau hay không: đề không có luật va chạm.

## 5. Cách 1 — Brute Force / baseline

Cách "ngu ngốc nhất" nhưng đúng nhất về mặt hình dung: **mô phỏng từng giây**. Mỗi giây hai người cùng tiến một bước về phía `z`. Ngay khi một (hoặc cả hai) người tới `z` thì dừng và trả kết quả. Dùng làm chuẩn đối chiếu (oracle) cho cách trực tiếp.

Dùng cách tiến bước `p += (z > p) - (z < p)`: nếu `z` ở bên phải `p` thì `+1`, bên trái thì `-1`, bằng nhau thì đứng yên.

````python
class SolutionSimulate:
    def findClosest(self, x, y, z):
        p1, p2 = x, y
        while True:
            a, b = (p1 == z), (p2 == z)          # ai đã tới nơi?
            if a and b:
                return 0
            if a:
                return 1
            if b:
                return 2
            p1 += (z > p1) - (z < p1)             # tiến một bước về phía z
            p2 += (z > p2) - (z < p2)

print(SolutionSimulate().findClosest(2, 7, 4))
print(SolutionSimulate().findClosest(2, 5, 6))
print(SolutionSimulate().findClosest(1, 5, 3))
````

**Độ phức tạp:** số giây tối đa bằng khoảng cách lớn nhất `<= 99` (vì `1 <= x, y, z <= 100`) -> `O(max(|x-z|, |y-z|))` thời gian, `O(1)` bộ nhớ. Nút thắt: ta lặp đi lặp lại những bước mà có thể suy ra ngay bằng phép trừ.

## 6. Ý tưởng thuật toán / bước đột phá

**Bước đột phá:** vì cả hai đi cùng tốc độ, người ở **gần** `z` hơn thì tới trước. Mô phỏng chỉ để đếm "ai hết quãng đường trước", và quãng đường đó chính là `abs(x - z)`, `abs(y - z)`. Vậy:

```
d1 = abs(x - z)
d2 = abs(y - z)
d1 < d2  ->  1
d1 > d2  ->  2
d1 == d2 ->  0
```

**Quy tắc "nếu… thì…":**

- Nếu `d1 < d2` thì Người 1 tới trước (`1`).
- Nếu `d2 < d1` thì Người 2 tới trước (`2`).
- Nếu `d1 == d2` thì hòa (`0`).
- Phía của `x` hay `y` so với `z` **không** ảnh hưởng.

**Invariant của mô phỏng:** sau `t` giây, khoảng cách của mỗi người còn `max(0, d - t)`; người đầu tiên có khoảng cách `0` là người có `d` nhỏ hơn.

````python
for (x, y, z) in [(2, 7, 4), (2, 5, 6), (1, 5, 3)]:
    d1, d2 = abs(x - z), abs(y - z)
    print((x, y, z), "d1 =", d1, "d2 =", d2, "->", 1 if d1 < d2 else 2 if d2 < d1 else 0)
````

## 7. Các cách tiếp theo

Ta có 4 cách: (1) mô phỏng từng giây; (2) so sánh trực tiếp hai khoảng cách; (3) mẹo dấu của hiệu + tra bảng chỉ số âm; (4) `sorted` các cặp `(khoảng_cách, id)` kèm kiểm tra hòa.

### Cách 2 — So sánh trực tiếp hai khoảng cách

Chính là bước đột phá ở mục 6: `abs` hai lần, ba nhánh `if`. Đây là lời giải chuẩn.

````python
class SolutionDirect:
    def findClosest(self, x, y, z):
        d1 = abs(x - z)
        d2 = abs(y - z)
        if d1 < d2:
            return 1
        if d2 < d1:
            return 2
        return 0

print(SolutionDirect().findClosest(2, 7, 4))
print(SolutionDirect().findClosest(2, 5, 6))
print(SolutionDirect().findClosest(1, 5, 3))
````

Độ phức tạp Cách 2: `O(1)` thời gian, `O(1)` bộ nhớ.

### Cách 3 — Dấu của hiệu + tra bảng (chỉ số âm)

Đặt `d = d1 - d2`. Dấu của `d`: `0` (hòa) -> `0`; `d > 0` (người 1 xa hơn) -> `2`; `d < 0` (người 1 gần hơn) -> `1`. Tính dấu bằng `(d > 0) - (d < 0)` rồi tra `[0, 2, 1][sign]` (chỉ số `-1` lấy phần tử cuối `1`). Không có `if`; hơi "mẹo", dùng để luyện tư duy.

````python
class SolutionSign:
    def findClosest(self, x, y, z):
        d = abs(x - z) - abs(y - z)
        sign = (d > 0) - (d < 0)            # 1, -1 hoặc 0
        return [0, 2, 1][sign]

print(SolutionSign().findClosest(2, 7, 4))
print(SolutionSign().findClosest(2, 5, 6))
print(SolutionSign().findClosest(1, 5, 3))
````

Độ phức tạp Cách 3: `O(1)` thời gian, `O(1)` bộ nhớ.

### Cách 4 — `sorted` các cặp `(khoảng_cách, id)` + kiểm tra hòa

Tạo `cands = sorted([(d1, 1), (d2, 2)])`. Nếu hai khoảng cách đầu bằng nhau thì hòa (`0`); ngược lại phần tử đầu là người gần nhất, trả `cands[0][1]`. Cách này mở rộng tự nhiên khi có **nhiều hơn hai** người.

````python
class SolutionSorted:
    def findClosest(self, x, y, z):
        cands = sorted([(abs(x - z), 1), (abs(y - z), 2)])
        if cands[0][0] == cands[1][0]:
            return 0
        return cands[0][1]

print(SolutionSorted().findClosest(2, 7, 4))
print(SolutionSorted().findClosest(2, 5, 6))
print(SolutionSorted().findClosest(1, 5, 3))
````

Độ phức tạp Cách 4: sắp xếp danh sách hai phần tử -> `O(1)` thời gian và bộ nhớ (với `k` người thì `O(k log k)`).

## 8. TRACE từng bước

Trace Cách 1 (mô phỏng) với `x = 2`, `y = 7`, `z = 4`:

| giây | `p1` | `p2` | `p1 == z`? | `p2 == z`? | hành động |
|---|---|---|---|---|---|
| 0 | 2 | 7 | không | không | cả hai tiến một bước |
| 1 | 3 | 6 | không | không | tiến tiếp |
| 2 | 4 | 5 | **có** | không | Người 1 tới -> trả `1` |

Kết quả `1`: khớp với `abs(2-4) = 2 < abs(7-4) = 3`.

````python
def trace_simulate(x, y, z):
    p1, p2 = x, y
    t = 0
    while True:
        a, b = (p1 == z), (p2 == z)
        print(f"giây {t}: p1={p1} p2={p2} | tới nơi? người 1: {a}, người 2: {b}")
        if a and b:
            return 0
        if a:
            return 1
        if b:
            return 2
        p1 += (z > p1) - (z < p1)
        p2 += (z > p2) - (z < p2)
        t += 1

print("kết quả:", trace_simulate(2, 7, 4))
print("kết quả:", trace_simulate(1, 5, 3))
print("kết quả:", trace_simulate(2, 5, 6))
````

## 9. Code nộp LeetCode

````python
class Solution:
    def findClosest(self, x: int, y: int, z: int) -> int:
        d1 = abs(x - z)
        d2 = abs(y - z)
        if d1 < d2:
            return 1
        if d2 < d1:
            return 2
        return 0

print(Solution().findClosest(2, 7, 4), Solution().findClosest(2, 5, 6), Solution().findClosest(1, 5, 3))
````

## 10. Vì sao thuật toán đúng

- **Thời gian tới nơi = khoảng cách:** mỗi giây một người đi đúng `1` đơn vị về phía `z`, mà không đi quá `z` (dừng khi tới). Sau `t` giây người ban đầu cách `d` đơn vị cách `max(0, d - t)`. Người đó tới `z` ở giây `t = d`.
- **So sánh thời gian = so sánh khoảng cách:** Người 1 tới ở giây `d1`, Người 2 ở giây `d2`. Ai có số nhỏ hơn tới trước; bằng nhau thì tới cùng giây. Đó chính là ba nhánh của code.
- **Cách 1 (mô phỏng) trùng khớp:** vòng lặp dừng ở giây nhỏ nhất `t = min(d1, d2)` và kiểm tra ai (hoặc cả hai) đã ở `z`; nên kết quả bằng nhánh tương ứng.
- **Cách 3:** `sign(d1 - d2)` tương ứng `d1 > d2 -> 2`, `d1 < d2 -> 1`, bằng nhau -> `0`.
- **Cách 4:** `sorted` đưa khoảng cách nhỏ nhất lên đầu; hòa khi hai khoảng cách bằng nhau.
- **Test:** ta duyệt **toàn bộ** các bộ `x, y, z` trong `1..20` cho mọi cài đặt và thêm ngẫu nhiên trong `1..100` để bao phủ hết miền.

## 11. Phân tích Time / Space Complexity

Gọi `D = max(abs(x - z), abs(y - z)) <= 99`.

| Cách | Time | Space |
|---|---|---|
| 1. Mô phỏng từng giây | `O(D)` | `O(1)` |
| 2. So sánh trực tiếp | `O(1)` | `O(1)` |
| 3. Dấu của hiệu + bảng | `O(1)` | `O(1)` |
| 4. `sorted` các cặp | `O(1)` (2 phần tử) | `O(1)` |

Với `x, y, z <= 100` mọi cách đều tức thì; Cách 2 là chuẩn để nộp. Phân biệt: Cách 1 là `O(D)` chỉ vì ta cố tình mô phỏng.

## 12. Bộ test

Test 1: ví dụ đề và các edge case (một người đã đứng ngay tại `z`, hai người cùng một chỗ, hai người cùng phía, biên `1` và `100`). Mọi input tuân thủ `1 <= x, y, z <= 100`.

````python
impls = [SolutionSimulate(), SolutionDirect(), SolutionSign(), SolutionSorted(), Solution()]
tests = [
    ((2, 7, 4), 1),
    ((2, 5, 6), 2),
    ((1, 5, 3), 0),
    ((5, 9, 5), 1),             # Người 1 đã đứng tại z: tới ở giây 0
    ((9, 5, 5), 2),             # Người 2 đã đứng tại z
    ((5, 5, 5), 0),             # cả ba cùng một chỗ
    ((3, 3, 8), 0),             # hai người cùng chỗ -> luôn hòa
    ((1, 100, 50), 1),          # cách 49 và 50: người 1 gần hơn một chút
    ((100, 1, 50), 2),
    ((1, 100, 1), 1),           # biên: người 1 tại z = 1
    ((100, 100, 1), 0),         # hai người cùng biên trên
    ((1, 2, 100), 2),           # cùng phía, người 2 gần z hơn
    ((10, 20, 15), 0),          # hai phía đối xứng quanh z
]
for (x, y, z), exp in tests:
    assert 1 <= x <= 100 and 1 <= y <= 100 and 1 <= z <= 100
    for im in impls:
        got = im.findClosest(x, y, z)
        assert got == exp, (type(im).__name__, (x, y, z), got, exp)
print("tất cả test cố định OK:", len(tests))
````

Test 2: **vét cạn** mọi bộ `(x, y, z)` với `1 <= x, y, z <= 20` (8000 bộ) cho cả năm cài đặt; chúng phải trùng nhau tuyệt đối, và kết quả chỉ nhận giá trị `0, 1, 2`.

````python
count = {0: 0, 1: 0, 2: 0}
for x in range(1, 21):
    for y in range(1, 21):
        for z in range(1, 21):
            ref = impls[0].findClosest(x, y, z)         # mô phỏng làm chuẩn
            assert ref in (0, 1, 2)
            count[ref] += 1
            for im in impls[1:]:
                assert im.findClosest(x, y, z) == ref, (type(im).__name__, (x, y, z))
print("vét cạn 20^3 = 8000 bộ OK; phân bố kết quả:", count)
````

Test 3: ngẫu nhiên có seed trên toàn miền `1..100`, cộng các **tính chất**: hoán đổi `x` và `y` thì `1 <-> 2` (và `0` giữ nguyên); dịch cả ba số cùng một đoạn (nếu vẫn trong `1..100`) không đổi kết quả; phản chiếu qua `101 - v` không đổi kết quả.

````python
import random
random.seed(3516)
swap = {0: 0, 1: 2, 2: 1}
for _ in range(5000):
    x, y, z = (random.randint(1, 100) for _ in range(3))
    ref = Solution().findClosest(x, y, z)
    for im in impls:
        assert im.findClosest(x, y, z) == ref, (type(im).__name__, (x, y, z))
    assert Solution().findClosest(y, x, z) == swap[ref]                 # đổi chỗ hai người
    assert Solution().findClosest(101 - x, 101 - y, 101 - z) == ref     # phản chiếu qua trục
    s = random.randint(-min(x, y, z) + 1, 100 - max(x, y, z))           # dịch giữ trong 1..100
    assert Solution().findClosest(x + s, y + s, z + s) == ref
print("5000 test ngẫu nhiên + tính chất (đổi chỗ, phản chiếu, tịnh tiến) OK")
````

## 13. Các lỗi người mới hay gặp

### Lỗi 1: quên giá trị tuyệt đối

````python
def wrong_no_abs(x, y, z):
    d1, d2 = x - z, y - z                         # SAI: hiệu có thể âm, không phải khoảng cách
    if d1 < d2:
        return 1
    if d2 < d1:
        return 2
    return 0

args = (2, 3, 8)
print("sai :", wrong_no_abs(*args))
print("đúng:", Solution().findClosest(*args))
assert wrong_no_abs(*args) != Solution().findClosest(*args)
````

Vì sao sai: `x - z = -6` và `y - z = -5`; so sánh `-6 < -5` cho `1`, nhưng khoảng cách thật là `6` và `5`, Người 2 gần hơn. **Sửa:** `abs(x - z)`, `abs(y - z)`.

### Lỗi 2: trả sai khi hòa (gộp trường hợp bằng vào một nhánh)

````python
def wrong_tie(x, y, z):
    d1, d2 = abs(x - z), abs(y - z)
    return 1 if d1 <= d2 else 2                   # SAI: hòa phải trả 0

args = (1, 5, 3)
print("sai :", wrong_tie(*args))
print("đúng:", Solution().findClosest(*args))
assert wrong_tie(*args) != Solution().findClosest(*args)
````

Vì sao sai: khi `d1 == d2` đề yêu cầu `0`, không phải `1`. Dùng `<=` gộp luôn trường hợp hòa vào nhánh "người 1". **Sửa:** kiểm tra `==` riêng (hoặc để nhánh `else` trả `0`).

### Lỗi 3: so sánh **vị trí** thay vì **khoảng cách**

````python
def wrong_compare_positions(x, y, z):
    if x < y:                                      # SAI: ai đứng bên trái chưa chắc gần z hơn
        return 1
    if y < x:
        return 2
    return 0

args = (2, 5, 6)
print("sai :", wrong_compare_positions(*args))
print("đúng:", Solution().findClosest(*args))
assert wrong_compare_positions(*args) != Solution().findClosest(*args)
````

Vì sao sai: `x = 2 < y = 5` nhưng `z = 6` ở bên phải cả hai, nên người đứng ở `5` mới gần hơn. Chỉ khi nào ta biết `z` ở bên nào ta mới dùng vị trí thay cho khoảng cách. **Sửa:** luôn đo `abs(· - z)`.

### Lỗi 4: trả về **khoảng cách** thay vì mã người

````python
def wrong_return_distance(x, y, z):
    return min(abs(x - z), abs(y - z))             # SAI: đề hỏi AI tới trước (1/2/0)

args = (2, 7, 4)
print("sai :", wrong_return_distance(*args))
print("đúng:", Solution().findClosest(*args))
assert wrong_return_distance(*args) != Solution().findClosest(*args)
````

Vì sao sai: `min(2, 3) = 2` là khoảng cách nhỏ nhất, nhưng đáp án là **mã người** `1`. Dễ nhầm vì kết quả đôi khi trùng (ví dụ khoảng cách nhỏ nhất cũng bằng `1`). **Sửa:** trả `1`, `2` hoặc `0` theo ba nhánh.

## 14. Mẹo làm bài nhanh & nhận dạng pattern

- "Hai đối tượng đi **cùng tốc độ** về cùng một đích" -> chỉ cần so **khoảng cách**.
- Khoảng cách trên trục số = `abs(a - b)`; quên `abs` là lỗi số một.
- Ba kết quả loại trừ (`<`, `>`, `==`) -> viết đủ ba nhánh, nhánh `else` cho trường hợp còn lại.
- Hòa là trường hợp đặc biệt: kiểm tra riêng hoặc để nhánh `else`.
- Kiểm tra bằng ví dụ: một người đứng sẵn tại `z`; hai người đối xứng quanh `z`; hai người cùng phía.
- Bài `O(1)` Easy như thế này nên tự test **vét cạn** miền nhỏ để chắc chắn (mục 12).

## 15. Biến thể của bài

1. **Tốc độ khác nhau** `v1, v2`: so sánh **thời gian** `d1 / v1` và `d2 / v2` (cẩn thận số thực; dùng nhân chéo `d1 * v2` với `d2 * v1` để tránh chia).
2. **Nhiều hơn hai người:** trả về người gần nhất (và cách xử lý hòa), mở rộng Cách 4.
3. **Hai chiều (mặt phẳng):** khoảng cách Manhattan `|dx| + |dy|` hoặc Euclid.
4. **Hòa trả về chỉ số nhỏ nhất** thay vì `0`.
5. Người 3 **cũng di chuyển** (ví dụ chạy trốn): bài toán đuổi bắt, cần xét tốc độ và hướng.

## 18. Gợi ý 3 mức cho bài tập

**Mức 1 (nhẹ):** ai gần `z` hơn thì tới trước; đo khoảng cách bằng `abs`.

**Mức 2 (vừa):** Bài 3: `return abs(a - b)`. Bài 4: tính ba khoảng cách, lấy `m = min(...)`, đếm số người có khoảng cách `m`; nếu `> 1` trả `0`, ngược lại trả chỉ số người đó. Bài 5: thời gian `d / v`, nên `d1 / v1 < d2 / v2 <=> d1 * v2 < d2 * v1` (vì `v1, v2 > 0`).

## 19. Đáp án tham khảo cho bài tập

````python
# Bài 1
assert Solution().findClosest(10, 3, 7) == 1

# Bài 2: 2 và 2 -> hòa
assert Solution().findClosest(6, 2, 4) == 0

# Bài 3
def distance(a, b):
    return abs(a - b)
assert distance(2, 7) == 5 and distance(7, 2) == 5 and distance(4, 4) == 0

# Bài 4
def closest_of_three(a, b, c, z):
    ds = [distance(a, z), distance(b, z), distance(c, z)]
    m = min(ds)
    if ds.count(m) > 1:
        return 0
    return ds.index(m) + 1
assert closest_of_three(1, 5, 9, 4) == 2          # khoảng cách 3, 1, 5
assert closest_of_three(1, 7, 9, 4) == 0          # 3 và 3 hòa nhau ở mức nhỏ nhất
assert closest_of_three(4, 4, 9, 4) == 0          # hai người cùng đứng tại z
assert closest_of_three(2, 8, 4, 4) == 3

# Bài 5
def find_closest_speed(x, y, z, v1, v2):
    t1 = distance(x, z) * v2                       # tương ứng d1 / v1 sau khi nhân v1 * v2
    t2 = distance(y, z) * v1
    if t1 < t2:
        return 1
    if t2 < t1:
        return 2
    return 0
assert find_closest_speed(2, 7, 4, 1, 1) == Solution().findClosest(2, 7, 4)
assert find_closest_speed(1, 9, 5, 1, 2) == 2     # d = 4 và 4, nhưng người 2 nhanh gấp đôi
assert find_closest_speed(1, 9, 5, 2, 2) == 0
print("đáp án bài tập OK")
````
