## 1. Mục tiêu học

Sau notebook này bạn sẽ:

1. Hiểu đề: mảng chuỗi `words` được xem là **vòng tròn** (phần tử cuối nối với phần tử đầu); từ vị trí `startIndex`, mỗi bước đi sang trái hoặc sang phải một ô; tìm số bước **ít nhất** để đứng ở một ô có giá trị bằng `target`, hoặc `-1` nếu không có.
2. Nắm khái niệm **khoảng cách trên vòng tròn**: giữa hai chỉ số `i` và `s` là `min(|i - s|, n - |i - s|)`.
3. Dùng phép `%` (modulo) của Python để "cuộn vòng" chỉ số, kể cả chỉ số âm (`(-1) % n == n - 1`).
4. Viết 4 cách: mở rộng từng bước hai phía, duyệt mọi chỉ số + công thức vòng, đi riêng từng hướng, và xoay mảng; đối chiếu bằng test ngẫu nhiên.
5. Tránh các lỗi: quên tính đường vòng, so sánh bằng `in` trên chuỗi (khớp chuỗi con), chỉ đi một hướng, quên trả `-1` khi không tìm thấy.

## 2. Đề bài & ví dụ

Cho mảng chuỗi `words` **đánh chỉ số từ 0, dạng vòng tròn**: phần tử kế tiếp của `words[i]` là `words[(i + 1) % n]`, phần tử trước của `words[i]` là `words[(i - 1 + n) % n]`, với `n = len(words)`.

Cho chuỗi `target` và số nguyên `startIndex`. Từ `startIndex`, mỗi lần đi **một bước** sang phải (`+1`) hoặc sang trái (`-1`) theo vòng. Trả về **số bước ngắn nhất** để tới một vị trí có `words[i] == target`. Nếu `target` không có trong `words`, trả về `-1`.

| Ví dụ | words | target | startIndex | Output | Giải thích |
|---|---|---|---|---|---|
| 1 | `["hello","i","am","leetcode","hello"]` | `"hello"` | 1 | 1 | từ chỉ số 1 lùi một bước tới chỉ số 0 (`"hello"`); tiến thì cần 3 bước tới chỉ số 4 |
| 2 | `["a","b","leetcode"]` | `"leetcode"` | 0 | 1 | từ chỉ số 0 lùi một bước (vòng) tới chỉ số 2 |
| 3 | `["i","eat","leetcode"]` | `"ate"` | 0 | -1 | không có `"ate"` |

Constraints: `1 <= words.length <= 100`; `1 <= words[i].length <= 100`; `words[i]` và `target` chỉ gồm chữ cái thường tiếng Anh; `0 <= startIndex < words.length`.

````python
words = ["hello", "i", "am", "leetcode", "hello"]
target, start = "hello", 1
n = len(words)
for i, w in enumerate(words):
    if w == target:
        d = abs(i - start)
        print(f"chỉ số {i}: |i - start| = {d}, đi vòng = {n - d}, khoảng cách vòng tròn = {min(d, n - d)}")
````

## 3. Kiến thức nền

### 3.1. Mảng vòng tròn và toán tử `%`

Để "cuộn vòng", dùng `% n`: `(i + 1) % n` là ô kế tiếp (từ `n - 1` quay về `0`), `(i - 1) % n` là ô trước (từ `0` quay về `n - 1`). Khác nhiều ngôn ngữ khác, `%` của Python với số âm vẫn cho kết quả **không âm**: `(-1) % 5 == 4`. Vì vậy không cần viết `+ n` thủ công.

````python
n = 5
print([(i + 1) % n for i in range(n)])       # kế tiếp
print([(i - 1) % n for i in range(n)])       # ô trước
print((-1) % 5, (-6) % 5)
````

### 3.2. Hai đường đi giữa hai ô trên vòng tròn

Giữa hai chỉ số `i` và `s` có hai đường: đi thẳng `d = |i - s|` bước, hoặc đi vòng qua đầu/cuối `n - d` bước. Khoảng cách ngắn nhất là `min(d, n - d)`. Nếu `i == s` thì `d = 0` và kết quả là `0` (đứng yên).

````python
def ring_dist(i, s, n):
    d = abs(i - s)
    return min(d, n - d)

n = 5
print([ring_dist(i, 1, n) for i in range(n)])    # khoảng cách từ ô 1 tới mọi ô
print(ring_dist(0, 0, n), ring_dist(0, 4, n))
````

### 3.3. `enumerate` và so sánh chuỗi bằng `==`

`for i, w in enumerate(words)` cho cả chỉ số và giá trị. So sánh chuỗi `w == target` đòi hỏi **khớp hoàn toàn**; còn `target in w` (toán tử `in` trên chuỗi) kiểm tra **chuỗi con**: `"leet" in "leetcode"` là `True` nhưng `"leet" == "leetcode"` là `False`. Đề cần khớp hoàn toàn.

````python
print("leet" in "leetcode", "leet" == "leetcode")
print("leetcode" in ["leetcode", "a"], "leet" in ["leetcode", "a"])      # in trên LIST là so sánh ==
````

### 3.4. `list.index` và xử lý khi không tìm thấy

`lst.index(x)` trả chỉ số **đầu tiên** bằng `x`; nếu không có sẽ ném `ValueError`. Dùng `if x in lst` trước, hoặc `try/except`. Để tìm vị trí **cuối cùng**, đảo mảng: `n - 1 - lst[::-1].index(x)`.

````python
lst = ["hello", "i", "am", "leetcode", "hello"]
print(lst.index("hello"), len(lst) - 1 - lst[::-1].index("hello"))
try:
    lst.index("ate")
except ValueError as e:
    print("ValueError:", e)
````

### 3.5. Cắt lát để xoay mảng

`words[s:] + words[:s]` là mảng **xoay trái** `s` bước, sao cho phần tử ở `startIndex` về đầu. Sau khi xoay, vị trí `p` trong mảng mới nghĩa là đi **sang phải** `p` bước (hoặc sang trái `n - p` bước, với `p > 0`) từ `startIndex`.

````python
words = ["hello", "i", "am", "leetcode", "hello"]
s = 1
rot = words[s:] + words[:s]
print(rot)
print([(p, min(p, len(rot) - p)) for p, w in enumerate(rot) if w == "hello"])
````

### 3.6. `min` với giá trị khởi tạo "vô cực"

Khi tìm minimum qua nhiều ứng viên, ta cần giá trị khởi tạo lớn hơn mọi đáp án: `float("inf")` hoặc `n` (khoảng cách vòng tròn tối đa là `n // 2`). Cuối cùng nếu vẫn là giá trị khởi tạo nghĩa là không tìm thấy → trả `-1`.

````python
best = float("inf")
for d in [3, 1, 4]:
    best = min(best, d)
print(best)
print(-1 if float("inf") == float("inf") else 0)
````

## 4. Ý tưởng nền tảng / cách nghĩ trực giác

Hãy tưởng tượng các từ xếp quanh một vòng tròn, bạn đứng tại `startIndex`. Bạn có thể bước sang trái hoặc phải; ô ở bước thứ `k` sang phải là `(s + k) % n`, sang trái là `(s - k) % n`. **Ô gần nhất** cho `target` là điều ta cần.

Ví dụ 1 (`s = 1`, `n = 5`):

```
chỉ số : 0       1    2    3         4
từ     : hello   i    am   leetcode  hello
bước k : trái 1  đứng phải 1 phải 2   phải 3 (cũng là trái 2)
          ^ target ở bước 1 (trái)
```

Thử `k = 0, 1, 2, ...`: tại `k = 0` ô `"i"` không khớp; tại `k = 1` ô bên trái là `words[0] = "hello"` khớp → đáp án `1`.

Vì đường vòng tròn chỉ dài `n`, đi quá `n // 2` bước về một hướng là không cần thiết (đi hướng ngược lại sẽ ngắn hơn); nhưng thử tới `n - 1` bước cũng không sai.

## 5. Cách 1 — Brute Force / baseline

Với `k = 0, 1, ..., n - 1`: kiểm tra ô `(s + k) % n` và ô `(s - k) % n`. Gặp `target` lần đầu tại bước `k` thì trả `k` (vì ta thử `k` tăng dần nên `k` đầu tiên là nhỏ nhất). Nếu hết vòng không thấy → `-1`.

Nút thắt: mỗi bước kiểm tra hai ô, tối đa `n` bước → `O(n)` phép so sánh chuỗi, mỗi lần so sánh chuỗi tốn `O(L)` → `O(n * L)`. Đây đã là tuyến tính; nhược điểm chủ yếu là phải nghĩ kỹ về modulo.

````python
class SolutionExpand:
    def closetTarget(self, words, target, startIndex):
        n = len(words)
        for k in range(n):
            if words[(startIndex + k) % n] == target or words[(startIndex - k) % n] == target:
                return k
        return -1

print(SolutionExpand().closetTarget(["hello", "i", "am", "leetcode", "hello"], "hello", 1))
print(SolutionExpand().closetTarget(["a", "b", "leetcode"], "leetcode", 0))
print(SolutionExpand().closetTarget(["i", "eat", "leetcode"], "ate", 0))
````

**Độ phức tạp:** Time `O(n * L)` (`n` ô, so sánh chuỗi tối đa `L` ký tự), Space `O(1)`.

## 6. Ý tưởng thuật toán / bước đột phá

**Bước đột phá:** không cần "đi" từng bước. Mỗi ô `i` chứa `target` có **khoảng cách vòng tròn** tới `startIndex` là `min(|i - s|, n - |i - s|)`. Đáp án là **nhỏ nhất** của các khoảng cách đó trên mọi ô khớp.

**Quy tắc "nếu… thì…":**

- Nếu `words[i] == target` → ứng viên `min(|i - s|, n - |i - s|)`.
- Nếu không có ô nào khớp → `-1`.
- Nếu `i == s` và khớp → khoảng cách `0` (tốt nhất có thể).

Invariant: sau khi duyệt các chỉ số `0..i`, `best` là khoảng cách vòng tròn nhỏ nhất tới một ô khớp trong `0..i` (hoặc vô cực nếu chưa có).

````python
words, target, s = ["hello", "i", "am", "leetcode", "hello"], "hello", 1
n = len(words)
cands = [min(abs(i - s), n - abs(i - s)) for i, w in enumerate(words) if w == target]
print("các ứng viên:", cands, "->", min(cands) if cands else -1)
````

## 7. Các cách tiếp theo

Ta có 4 cách: (1) mở rộng từng bước hai phía, (2) duyệt mọi chỉ số + công thức vòng tròn, (3) đi riêng từng hướng, (4) xoay mảng rồi dùng `index`.

### Cách 2 — Duyệt mọi chỉ số + công thức vòng tròn

Đúng ý mục 6: một lượt `for i, w in enumerate(words)`, với mỗi ô khớp tính `d = abs(i - s)` rồi `min(d, n - d)`, giữ `best` nhỏ nhất.

````python
class SolutionFormula:
    def closetTarget(self, words, target, startIndex):
        n = len(words)
        best = n                       # khoảng cách vòng tròn tối đa là n // 2 < n
        found = False
        for i, w in enumerate(words):
            if w == target:
                found = True
                d = abs(i - startIndex)
                best = min(best, d, n - d)
        return best if found else -1

print(SolutionFormula().closetTarget(["hello", "i", "am", "leetcode", "hello"], "hello", 1))
print(SolutionFormula().closetTarget(["a", "b", "leetcode"], "leetcode", 0))
print(SolutionFormula().closetTarget(["i", "eat", "leetcode"], "ate", 0))
````

Độ phức tạp Cách 2: Time `O(n * L)`, Space `O(1)`.

### Cách 3 — Đi riêng từng hướng

Đi **sang phải** từ `startIndex` tối đa `n` bước để đếm số bước tới ô khớp đầu tiên (`right`), rồi đi **sang trái** tương tự (`left`). Đáp án là `min(right, left)` nếu có. Cách này gần với mô tả "đi bước một" của đề nhất.

````python
class SolutionDirections:
    def closetTarget(self, words, target, startIndex):
        n = len(words)
        right = left = None
        for k in range(n):
            if words[(startIndex + k) % n] == target:
                right = k
                break
        for k in range(n):
            if words[(startIndex - k) % n] == target:
                left = k
                break
        if right is None:                 # nếu một hướng không thấy thì hướng kia cũng không thấy
            return -1
        return min(right, left)

print(SolutionDirections().closetTarget(["hello", "i", "am", "leetcode", "hello"], "hello", 1))
print(SolutionDirections().closetTarget(["a", "b", "leetcode"], "leetcode", 0))
print(SolutionDirections().closetTarget(["i", "eat", "leetcode"], "ate", 0))
````

Độ phức tạp Cách 3: Time `O(n * L)` (hai lượt), Space `O(1)`.

### Cách 4 — Xoay mảng + `index`

Xoay mảng để `startIndex` về `0`: `rot = words[s:] + words[:s]`. Khi đó vị trí khớp đầu tiên `f = rot.index(target)` là số bước đi phải; vị trí khớp cuối cùng `l` cho số bước đi trái là `n - l` (nếu `l > 0`). Nếu `f == 0` thì đáp án `0`.

````python
class SolutionRotate:
    def closetTarget(self, words, target, startIndex):
        n = len(words)
        if target not in words:
            return -1
        rot = words[startIndex:] + words[:startIndex]
        first = rot.index(target)
        if first == 0:
            return 0
        last = n - 1 - rot[::-1].index(target)
        return min(first, n - last)

print(SolutionRotate().closetTarget(["hello", "i", "am", "leetcode", "hello"], "hello", 1))
print(SolutionRotate().closetTarget(["a", "b", "leetcode"], "leetcode", 0))
print(SolutionRotate().closetTarget(["i", "eat", "leetcode"], "ate", 0))
````

Độ phức tạp Cách 4: Time `O(n * L)`, Space `O(n)` (mảng xoay và bản đảo).

## 8. TRACE từng bước

Trace Cách 1 (mở rộng hai phía) với `words = ["hello","i","am","leetcode","hello"]`, `target = "hello"`, `startIndex = 1`, `n = 5`:

| `k` | ô phải `(1+k)%5` | `words[...]` | ô trái `(1-k)%5` | `words[...]` | khớp? |
|---|---|---|---|---|---|
| 0 | 1 | `i` | 1 | `i` | không |
| 1 | 2 | `am` | 0 | `hello` | **có** (ô trái) → trả `1` |

Với ví dụ 3 (`["i","eat","leetcode"]`, `"ate"`, `0`): các `k = 0, 1, 2` đều không khớp → `-1`.

````python
def trace_expand(words, target, start):
    n = len(words)
    print(f"words={words}, target={target!r}, start={start}, n={n}")
    for k in range(n):
        r, l = (start + k) % n, (start - k) % n
        print(f"k={k}: phải ô {r} -> {words[r]!r}, trái ô {l} -> {words[l]!r}")
        if words[r] == target or words[l] == target:
            print("-> tìm thấy ở bước", k)
            return k
    print("-> không có target")
    return -1

trace_expand(["hello", "i", "am", "leetcode", "hello"], "hello", 1)
trace_expand(["i", "eat", "leetcode"], "ate", 0)
````

## 9. Code nộp LeetCode

````python
from typing import List

class Solution:
    def closetTarget(self, words: List[str], target: str, startIndex: int) -> int:
        n = len(words)
        best = n
        found = False
        for i, w in enumerate(words):
            if w == target:
                found = True
                d = abs(i - startIndex)
                best = min(best, d, n - d)
        return best if found else -1

print(Solution().closetTarget(["hello", "i", "am", "leetcode", "hello"], "hello", 1),
      Solution().closetTarget(["a", "b", "leetcode"], "leetcode", 0),
      Solution().closetTarget(["i", "eat", "leetcode"], "ate", 0))
````

## 10. Vì sao thuật toán đúng

- **Khoảng cách vòng tròn:** giữa ô `s` và ô `i` chỉ có hai đường đi đơn giản trên một vòng: đi theo chiều tăng chỉ số hoặc theo chiều giảm chỉ số. Một đường dài `|i - s|` (không qua cạnh nối cuối - đầu) và đường kia dài `n - |i - s|` (qua cạnh nối). Mọi đường đi khác (đi quanh nhiều vòng hoặc đổi chiều) chỉ dài hơn hoặc bằng. Vậy khoảng cách ngắn nhất là `min(|i - s|, n - |i - s|)`.
- **Đáp án là min trên các ô khớp:** để chạm một ô khớp cần đi tới ô khớp gần nhất; ta xét mọi ô khớp nên không bỏ sót; lấy min cho ra đáp án tối ưu. Không có ô khớp → `-1`.
- **Cách 1 (mở rộng):** ở bước `k`, hai ô cách `start` đúng `k` bước (về mỗi hướng) được kiểm tra; `k` tăng dần nên `k` đầu tiên có ô khớp chính là khoảng cách nhỏ nhất (mọi ô cách `< k` bước đều đã bị loại).
- **Cách 3:** `right` là khoảng cách theo hướng phải tới ô khớp gần nhất, `left` theo hướng trái; mọi ô khớp được tiếp cận theo một trong hai hướng nên `min` bao quát mọi trường hợp. Nếu hướng phải duyệt hết `n` ô mà không thấy thì target không tồn tại (đã duyệt cả vòng).
- **Cách 4:** sau xoay, chỉ số `p` là số bước phải; số bước trái tới cùng ô là `n - p` (với `p > 0`). Ô khớp đầu tiên cho bước phải nhỏ nhất, ô khớp cuối cùng cho bước trái nhỏ nhất.

## 11. Phân tích Time / Space Complexity

Gọi `n = len(words)`, `L` = độ dài chuỗi lớn nhất (so sánh hai chuỗi tốn `O(L)`).

| Cách | Time | Space |
|---|---|---|
| 1. Mở rộng hai phía | `O(n * L)` | `O(1)` |
| 2. Mọi chỉ số + công thức vòng | `O(n * L)` | `O(1)` |
| 3. Đi riêng từng hướng | `O(n * L)` | `O(1)` |
| 4. Xoay mảng + `index` | `O(n * L)` | `O(n)` |

Mọi cách tuyến tính theo kích thước đầu vào, đây là tối ưu vì phải xem các ô ít nhất một lần trong trường hợp xấu (target không có). Với `n, L <= 100` rất nhanh.

## 12. Bộ test

Test 1: ví dụ đề và các edge case (`n = 1`, target ở chính vị trí bắt đầu, target nhiều lần, đường vòng ngắn hơn, không có target).

````python
impls = [SolutionExpand(), SolutionFormula(), SolutionDirections(), SolutionRotate(), Solution()]
tests = [
    (["hello", "i", "am", "leetcode", "hello"], "hello", 1, 1),
    (["a", "b", "leetcode"], "leetcode", 0, 1),
    (["i", "eat", "leetcode"], "ate", 0, -1),
    (["a"], "a", 0, 0),                         # n = 1, có target tại chỗ
    (["a"], "b", 0, -1),                        # n = 1, không có
    (["a", "b", "c", "d"], "a", 0, 0),          # đứng ngay tại target
    (["a", "b", "c", "d"], "c", 0, 2),          # n = 4, cách đúng n/2 bước cả hai phía
    (["a", "b", "c", "d", "e"], "e", 0, 1),     # đường vòng (lùi 1) ngắn hơn đi thẳng (4)
    (["a", "b", "c", "d", "e"], "b", 4, 2),     # đi vòng phải: 4 -> 0 -> 1
    (["x", "y", "x", "y", "x", "y"], "y", 0, 1),
    (["leetcode", "a"], "leet", 1, -1),         # bắt lỗi khớp chuỗi con
    (["ab", "abc", "ab"], "ab", 1, 1),          # target xuất hiện hai phía
    (["a", "b", "c", "d", "e", "f"], "d", 0, 3),
    (["a", "b", "c", "d", "e", "f"], "f", 1, 2),
]
for words, target, start, exp in tests:
    assert 1 <= len(words) <= 100 and 0 <= start < len(words)
    assert all(w.isalpha() and w.islower() and 1 <= len(w) <= 100 for w in words + [target])
    for im in impls:
        assert im.closetTarget(list(words), target, start) == exp, (type(im).__name__, words, target, start, exp)
print("tất cả test cố định OK:", len(tests))
````

Test 2: ngẫu nhiên có seed, đúng constraints (`1 <= n <= 100`, chỉ chữ cái thường, `0 <= startIndex < n`); oracle độc lập mô phỏng **BFS trên vòng tròn**.

````python
import random
from collections import deque
random.seed(2515)

def oracle_bfs(words, target, start):
    n = len(words)
    dist = {start: 0}
    q = deque([start])
    while q:
        u = q.popleft()
        if words[u] == target:
            return dist[u]
        for v in ((u + 1) % n, (u - 1) % n):
            if v not in dist:
                dist[v] = dist[u] + 1
                q.append(v)
    return -1

vocab = ["a", "b", "ab", "abc", "z", "leet", "leetcode"]
for t in range(2000):
    n = random.randint(1, 100) if t % 20 == 0 else random.randint(1, 12)
    k = random.choice([1, 2, 3, len(vocab)])
    words = [random.choice(vocab[:k]) for _ in range(n)]
    target = random.choice(vocab)
    start = random.randrange(n)
    exp = oracle_bfs(words, target, start)
    for im in impls:
        assert im.closetTarget(list(words), target, start) == exp, (type(im).__name__, words, target, start)
print("2000 test ngẫu nhiên OK")
````

Test 3: tính chất — đáp án `-1` hoặc nằm trong `0..n // 2`; xoay vòng mảng cùng với `startIndex` không đổi đáp án; đáp án `0` ⇔ `words[startIndex] == target`.

````python
random.seed(21)
for _ in range(500):
    n = random.randint(1, 30)
    words = [random.choice(["a", "b", "c"]) for _ in range(n)]
    target = random.choice(["a", "b", "c", "d"])
    start = random.randrange(n)
    ans = Solution().closetTarget(words, target, start)
    assert ans == -1 or 0 <= ans <= n // 2
    assert (ans == 0) == (words[start] == target)
    shift = random.randrange(n)
    rotated = words[shift:] + words[:shift]
    new_start = (start - shift) % n
    assert Solution().closetTarget(rotated, target, new_start) == ans
print("tính chất (khoảng giá trị, bất biến khi xoay vòng, đáp án 0): OK")
````

## 13. Các lỗi người mới hay gặp

### Lỗi 1: quên rằng mảng là vòng tròn (chỉ dùng `abs(i - start)`)

````python
def wrong_no_wrap(words, target, start):
    ds = [abs(i - start) for i, w in enumerate(words) if w == target]    # SAI: thiếu đường vòng n - d
    return min(ds) if ds else -1

words, target, start = ["a", "b", "leetcode"], "leetcode", 0
print("sai :", wrong_no_wrap(words, target, start))
print("đúng:", Solution().closetTarget(words, target, start))
assert wrong_no_wrap(words, target, start) != Solution().closetTarget(words, target, start)
````

Vì sao sai: từ ô `0` tới ô `2` đi thẳng mất `2` bước, nhưng lùi một bước (từ `0` quay về `2` theo vòng) chỉ mất `1`. Khoảng cách vòng tròn là `min(d, n - d)`.

### Lỗi 2: khớp chuỗi con bằng `in` thay vì `==`

````python
def wrong_substring(words, target, start):
    n = len(words)
    ds = []
    for i, w in enumerate(words):
        if target in w:                       # SAI: "leet" in "leetcode" là True, nhưng "leet" != "leetcode"
            d = abs(i - start)
            ds.append(min(d, n - d))
    return min(ds) if ds else -1

words, target, start = ["leetcode", "a"], "leet", 1
print("sai :", wrong_substring(words, target, start))
print("đúng:", Solution().closetTarget(words, target, start))
assert wrong_substring(words, target, start) != Solution().closetTarget(words, target, start)
````

Vì sao sai: đề yêu cầu ô có giá trị **bằng** `target`, không phải chứa `target`. **Sửa:** `w == target`.

### Lỗi 3: chỉ đi một hướng

````python
def wrong_one_direction(words, target, start):
    n = len(words)
    for k in range(n):
        if words[(start + k) % n] == target:       # SAI: chỉ xét hướng phải
            return k
    return -1

words, target, start = ["hello", "i", "am", "leetcode", "hello"], "hello", 1
print("sai :", wrong_one_direction(words, target, start))
print("đúng:", Solution().closetTarget(words, target, start))
assert wrong_one_direction(words, target, start) != Solution().closetTarget(words, target, start)
````

Vì sao sai: hướng phải cần `3` bước tới chỉ số 4, nhưng hướng trái chỉ cần `1`. Phải xét **cả hai hướng** (hoặc dùng `min(d, n - d)`).

### Lỗi 4: không trả `-1` khi không tìm thấy

````python
def wrong_no_minus_one(words, target, start):
    n = len(words)
    best = n                                  # SAI: quên phân biệt "không tìm thấy"
    for i, w in enumerate(words):
        if w == target:
            d = abs(i - start)
            best = min(best, d, n - d)
    return best

words, target, start = ["i", "eat", "leetcode"], "ate", 0
print("sai :", wrong_no_minus_one(words, target, start))
print("đúng:", Solution().closetTarget(words, target, start))
assert wrong_no_minus_one(words, target, start) != Solution().closetTarget(words, target, start)
````

Vì sao sai: nếu `target` không có, `best` giữ nguyên giá trị khởi tạo `n` và hàm trả `3` thay vì `-1`. Cần cờ `found` (hoặc kiểm tra `best == inf`) để trả `-1`.

## 14. Mẹo làm bài nhanh & nhận dạng pattern

- "Mảng vòng tròn / đi lại hai chiều" → khoảng cách giữa hai chỉ số là `min(d, n - d)` với `d = abs(i - j)`.
- Cuộn chỉ số bằng `% n`; trong Python, `(-1) % n == n - 1` nên không cần `+ n`.
- Tìm "gần nhất" → xét **mọi ứng viên** và lấy `min`, hoặc mở rộng ra hai phía theo từng bước `k`.
- So sánh chuỗi: dùng `==` khi cần khớp hoàn toàn; `in` trên chuỗi là kiểm tra chuỗi con — hai thứ khác nhau.
- Luôn xử lý riêng "không tìm thấy" (`-1`) bằng cờ hoặc giá trị vô cực.
- Tên phương thức đề cho là `closetTarget` (sai chính tả của LeetCode); chép đúng từng chữ khi nộp.

## 15. Biến thể của bài

1. Chỉ được đi **một chiều** (chỉ sang phải) — khi đó là `(i - start) % n`.
2. Mỗi bước sang trái tốn `1`, sang phải tốn `2` (chi phí khác nhau) — chọn hướng rẻ hơn.
3. Trả về **chỉ số** của ô gần nhất (nếu hoà, ô nhỏ hơn).
4. Có **nhiều truy vấn** `(target, start)` cho cùng mảng: dựng sẵn `dict target -> danh sách chỉ số` và dùng tìm nhị phân trên danh sách đã sắp xếp.
5. Mảng vòng tròn hai chiều (lưới có các cạnh nối vòng) — khoảng cách theo từng trục `min(d, n - d)`.

## 16. So sánh các cách làm

| Cách | Ý tưởng | Time | Space | Khi nào dùng |
|---|---|---|---|---|
| 1. Mở rộng hai phía | thử `k = 0, 1, ...` hai ô mỗi bước | `O(n * L)` | `O(1)` | trực quan nhất, đúng mô tả đề |
| 2. Mọi chỉ số + công thức | `min(d, n - d)` cho từng ô khớp | `O(n * L)` | `O(1)` | **gọn, nên nộp** |
| 3. Đi riêng từng hướng | tìm ô khớp đầu tiên mỗi hướng | `O(n * L)` | `O(1)` | khi hai hướng có luật khác nhau |
| 4. Xoay mảng + `index` | đưa `start` về `0` rồi `index` | `O(n * L)` | `O(n)` | thích dùng hàm có sẵn |

## 18. Gợi ý 3 mức cho bài tập

**Mức 1 (nhẹ):** giữa hai ô trên vòng có hai đường: đi thẳng và đi vòng; lấy đường ngắn hơn.

**Mức 2 (vừa):** Bài 3: `d = abs(i - j)` rồi `min(d, n - d)`. Bài 4: duyệt `enumerate(words)`, so sánh `(khoảng cách, chỉ số)` bằng tuple. Bài 5: số bước đi phải tới chỉ số `i` là `(i - start) % n`.

## 19. Đáp án tham khảo cho bài tập

````python
# Bài 1: "e" ở chỉ số 4, start 1: d=3, vòng=2 -> 2
assert Solution().closetTarget(["a", "b", "c", "d", "e"], "e", 1) == 2

# Bài 2: "y" ở chỉ số 1 và 4; từ chỉ số 2 -> 1 hoặc min(2, 3) = 2
words2 = ["x", "y", "z", "x", "y"]
assert [min(abs(i - 2), 5 - abs(i - 2)) for i, w in enumerate(words2) if w == "y"] == [1, 2]
assert Solution().closetTarget(words2, "y", 2) == 1

# Bài 3
def ring_distance(i, j, n):
    d = abs(i - j)
    return min(d, n - d)
assert ring_distance(0, 4, 5) == 1
assert ring_distance(1, 4, 5) == 2
assert ring_distance(3, 3, 5) == 0

# Bài 4
def closest_index(words, target, start):
    n = len(words)
    best = None
    for i, w in enumerate(words):
        if w == target:
            cand = (ring_distance(i, start, n), i)
            if best is None or cand < best:
                best = cand
    return -1 if best is None else best[1]
assert closest_index(["hello", "i", "am", "leetcode", "hello"], "hello", 1) == 0
assert closest_index(["a", "b", "a"], "a", 1) == 0          # hoà: chọn chỉ số nhỏ hơn
assert closest_index(["i", "eat"], "ate", 0) == -1

# Bài 5
def one_way_distance(words, target, start):
    n = len(words)
    for k in range(n):
        if words[(start + k) % n] == target:
            return k
    return -1
assert one_way_distance(["hello", "i", "am", "leetcode", "hello"], "hello", 1) == 3
assert one_way_distance(["a", "b", "c"], "a", 0) == 0
assert one_way_distance(["a", "b", "c"], "z", 0) == -1
print("đáp án bài tập OK")
````
