## 1. Mục tiêu học

Sau notebook này bạn sẽ:

2. Nắm cơ bản **danh sách liên kết đơn**: lớp `ListNode`, duyệt bằng `while cur`, và cách **nhảy hai nút một lần** (`cur = cur.next.next`).
3. Biết tự **dựng danh sách từ list Python** và **in danh sách ra** để thử nghiệm ngoài LeetCode.
4. Đi từ cách **đổ danh sách thành mảng rồi duyệt theo chỉ số** → **duyệt cặp một lượt (hai con trỏ)** → **một bộ đếm hiệu số `balance`** → **duyệt từng nút theo chẵn/lẻ** → **đệ quy**.
5. Tránh các lỗi kinh điển: nhảy `cur = cur.next` (thay vì `.next.next`), hiểu nhầm đội nào ứng với chỉ số nào, viết sai chữ hoa/thường của kết quả, và `AttributeError` do `cur.next` là `None`.

## 2. Đề bài & ví dụ

Cho `head` là đầu của một danh sách liên kết đơn có **độ dài chẵn**, chứa các số nguyên. Các nút ở **chỉ số chẵn** (0, 2, 4, ...) chứa **số chẵn**; các nút ở **chỉ số lẻ** (1, 3, 5, ...) chứa **số lẻ**.

Ta gọi mỗi nút ở chỉ số chẵn cùng nút ngay sau nó là một **cặp**: các nút chỉ số `(0, 1)` là một cặp, `(2, 3)` là một cặp, v.v. Với mỗi cặp ta so sánh giá trị hai nút:

- Nếu nút **chỉ số lẻ** có giá trị **lớn hơn**, đội **`"Odd"`** được 1 điểm.
- Nếu nút **chỉ số chẵn** có giá trị **lớn hơn**, đội **`"Even"`** được 1 điểm.

Trả về tên đội có **nhiều điểm hơn**: `"Even"` hoặc `"Odd"`. Nếu bằng điểm, trả về `"Tie"`.

**Ví dụ 1**

- Input: `head = [2, 1]`
- Output: `"Even"`
- Giải thích: có một cặp `(2, 1)`; nút chỉ số chẵn (`2`) lớn hơn nên `Even` được 1 điểm, `Odd` được 0 điểm.

**Ví dụ 2**

- Input: `head = [2, 5, 4, 7, 20, 5]`
- Output: `"Odd"`
- Giải thích: ba cặp `(2, 5)`, `(4, 7)`, `(20, 5)`. Hai cặp đầu nút lẻ lớn hơn (`Odd` được 2 điểm); cặp cuối nút chẵn lớn hơn (`Even` được 1 điểm). `Odd` thắng.

**Ví dụ 3**

- Input: `head = [4, 5, 2, 1]`
- Output: `"Tie"`
- Giải thích: cặp `(4, 5)` cho `Odd` 1 điểm, cặp `(2, 1)` cho `Even` 1 điểm. Hòa.

**Ràng buộc** (theo trí nhớ, nên đối chiếu đề gốc): số nút nằm trong `[2, 100]` và là số **chẵn**; `1 <= Node.val <= 100`; nút chỉ số chẵn có giá trị chẵn, nút chỉ số lẻ có giá trị lẻ (nên hai nút trong một cặp **không bao giờ bằng nhau**).

````python
from typing import List, Optional
import random


class ListNode:
    def __init__(self, val=0, next=None):
        self.val = val
        self.next = next


def build_list(values: List[int]) -> Optional[ListNode]:
    """Dựng danh sách liên kết từ list Python, trả về nút đầu (head)."""
    dummy = ListNode()
    tail = dummy
    for v in values:
        tail.next = ListNode(v)
        tail = tail.next
    return dummy.next


def list_to_array(head: Optional[ListNode]) -> List[int]:
    """Đọc danh sách liên kết thành list Python."""
    out = []
    cur = head
    while cur:
        out.append(cur.val)
        cur = cur.next
    return out


def show_list(head: Optional[ListNode]) -> str:
    """In danh sách dạng 2 -> 1 -> None."""
    return " -> ".join(str(v) for v in list_to_array(head)) + " -> None"


examples = [
    ([2, 1], "Even"),
    ([2, 5, 4, 7, 20, 5], "Odd"),
    ([4, 5, 2, 1], "Tie"),
]
for values, expected in examples:
    print(show_list(build_list(values)), "-> kỳ vọng", expected)
````

## 3. Kiến thức nền

### 3.1. Danh sách liên kết đơn và lớp `ListNode`

Một **danh sách liên kết đơn** là chuỗi các **nút**. Mỗi nút giữ hai thứ: giá trị `val` và con trỏ `next` trỏ tới nút kế tiếp (nút cuối có `next = None`). Ta chỉ giữ **đầu danh sách** `head`; muốn tới nút thứ 3 phải đi qua nút 1, rồi nút 2. Khác list Python, **không thể truy cập `lst[i]` trực tiếp**.

````python
a = ListNode(2)
b = ListNode(1)
a.next = b                       # nối: 2 -> 1 -> None
print(a.val, a.next.val, a.next.next)   # 2 1 None
print(show_list(a))
````

### 3.2. Duyệt danh sách bằng `while cur`

`cur` bắt đầu ở `head`. Khi `cur` còn là một nút (không phải `None`), ta xử lý rồi **đi tiếp** bằng `cur = cur.next`. Khi `cur` thành `None` nghĩa là đã đi hết. Điều kiện `while cur:` đủ để dừng vì nút là "đúng" còn `None` là "sai".

````python
head = build_list([2, 5, 4, 7])
cur = head
idx = 0
while cur:
    print("chỉ số", idx, "giá trị", cur.val)
    cur = cur.next
    idx += 1
````

### 3.3. Nhảy hai nút một lần: `cur = cur.next.next`

Vì mỗi cặp gồm **hai nút liền nhau**, sau khi xử lý cặp `(cur, cur.next)` ta nhảy tới cặp tiếp theo bằng `cur = cur.next.next`. Điều này **an toàn** vì đề đảm bảo độ dài chẵn: nếu `cur` là một nút thì `cur.next` luôn tồn tại, còn `cur.next.next` có thể là `None` (hết danh sách) — và `while cur:` sẽ dừng.

````python
head = build_list([2, 5, 4, 7, 20, 5])
cur = head
while cur:
    print("cặp:", (cur.val, cur.next.val))
    cur = cur.next.next        # nhảy qua cả cặp
````

### 3.4. So sánh để chấm điểm bằng biểu thức boolean

Trong Python `True` cộng như `1`, `False` như `0`. Có thể viết `even_score += a > b` để cộng 1 khi điều kiện đúng. Ta cũng có thể dùng một **bộ đếm hiệu số**: `balance += 1` khi `Even` thắng cặp, `balance -= 1` khi `Odd` thắng. Cuối cùng chỉ cần nhìn dấu của `balance`.

````python
a, b = 2, 5
print(a > b, b > a)                 # False True
even_score = 0
even_score += a > b
print(even_score)                   # 0
balance = 0
balance += 1 if a > b else -1       # Odd thắng cặp này
print(balance)                      # -1
````

### 3.5. Chuỗi trả về: đúng chữ hoa/thường

Đề yêu cầu **đúng** ba chuỗi `"Even"`, `"Odd"`, `"Tie"` (chữ cái đầu viết hoa). LeetCode so sánh chuỗi chính xác nên `"even"` hay `"EVEN"` đều sai.

````python
def name_of(balance: int) -> str:
    if balance > 0:
        return "Even"
    if balance < 0:
        return "Odd"
    return "Tie"


print(name_of(1), name_of(-2), name_of(0))      # Even Odd Tie
````

## 4. Ý tưởng nền tảng / cách nghĩ trực giác (tính tay)

Lấy `head = [2, 5, 4, 7, 20, 5]`. Viết các nút kèm chỉ số:

| chỉ số | 0 | 1 | 2 | 3 | 4 | 5 |
|---|---|---|---|---|---|---|
| giá trị | 2 | 5 | 4 | 7 | 20 | 5 |

Các cặp (chỉ số chẵn, chỉ số lẻ): `(2, 5)`, `(4, 7)`, `(20, 5)`.

| cặp | so sánh | ai được điểm | Even | Odd |
|---|---|---|---|---|
| (2, 5) | 5 > 2 | Odd | 0 | 1 |
| (4, 7) | 7 > 4 | Odd | 0 | 2 |
| (20, 5) | 20 > 5 | Even | 1 | 2 |

`Odd = 2 > Even = 1` nên đáp án `"Odd"`. Bài toán chỉ là: **đi hết danh sách theo từng cặp, so sánh hai số, cộng điểm, rồi so tổng điểm.**

````python
values = [2, 5, 4, 7, 20, 5]
even_score = odd_score = 0
for i in range(0, len(values), 2):          # i = 0, 2, 4: đầu mỗi cặp
    left, right = values[i], values[i + 1]
    if left > right:
        even_score += 1
    elif right > left:
        odd_score += 1
    print(f"cặp ({left}, {right}) -> Even={even_score}, Odd={odd_score}")
print("kết quả:", name_of(even_score - odd_score))
````

## 5. Cách 1 — Brute Force / baseline: đổ danh sách thành mảng rồi duyệt theo chỉ số

**Ý tưởng.** Danh sách liên kết khó truy cập theo chỉ số, nên bước đầu: **đọc hết giá trị vào một list Python** (`list_to_array`). Sau đó dùng `range(0, n, 2)` để lấy từng cặp `values[i]`, `values[i + 1]` như ở §4.

Cách này dễ nhất để kiểm tra logic chấm điểm; nó tốn thêm bộ nhớ cho mảng phụ.

````python
def game_result_array(head: Optional[ListNode]) -> str:
    values = list_to_array(head)                 # bước 1: đổ ra mảng
    even_score = odd_score = 0
    for i in range(0, len(values), 2):           # bước 2: duyệt từng cặp
        if values[i] > values[i + 1]:
            even_score += 1
        elif values[i + 1] > values[i]:
            odd_score += 1
    if even_score > odd_score:
        return "Even"
    if odd_score > even_score:
        return "Odd"
    return "Tie"


for values, expected in examples:
    print(game_result_array(build_list(values)), "(kỳ vọng", expected, ")")
````

**Độ phức tạp (Cách 1).** Gọi `n` là số nút. Đọc vào mảng `O(n)`, duyệt cặp `O(n)` ⇒ **Time `O(n)`**, **Space `O(n)`** cho mảng phụ.

**Nút thắt:** mảng phụ không cần thiết — ta chỉ cần nhìn mỗi nút **một lần** khi đi qua. Có thể chấm điểm ngay trong lúc duyệt danh sách, bỏ mảng phụ.

## 6. Ý tưởng thuật toán / bước đột phá

Điều quan trọng: để biết đội nào thắng ta **không cần giữ lại các giá trị**, chỉ cần **hai con số** (điểm của mỗi đội) hoặc thậm chí **một số** (hiệu số điểm `balance = even_score - odd_score`). Khi đi qua mỗi cặp, ta chỉ cần:

- nếu nút chẵn-chỉ-số lớn hơn: `balance += 1`;
- ngược lại (nút lẻ-chỉ-số lớn hơn): `balance -= 1`.

Cuối cùng dấu của `balance` cho đáp án (`> 0` là `Even`, `< 0` là `Odd`, `== 0` là `Tie`). Chạy trực tiếp trên danh sách bằng cách nhảy `cur = cur.next.next`: **Time `O(n)`**, **Space `O(1)`**.

````python
# balance trên ví dụ 2 bằng tay: [2, 5, 4, 7, 20, 5]
balance = 0
for left, right in [(2, 5), (4, 7), (20, 5)]:
    balance += 1 if left > right else -1
    print(f"cặp ({left}, {right}): balance = {balance}")
print(name_of(balance))      # Odd
````

## 7. Các cách tiếp theo

### Cách 2 — Duyệt cặp một lượt bằng `cur` và `cur.next` (hai điểm số)

**Giải thích từng đoạn.**

- `cur` đứng ở nút đầu mỗi cặp (chỉ số chẵn); `cur.next` là nút thứ hai của cặp (chỉ số lẻ).
- So sánh `cur.val` với `cur.next.val`, cộng điểm cho `even_score` hoặc `odd_score`.
- `cur = cur.next.next` để nhảy sang cặp sau.
- Hết vòng thì so hai điểm số.

````python
def game_result_pairs(head: Optional[ListNode]) -> str:
    even_score = odd_score = 0
    cur = head
    while cur:
        if cur.val > cur.next.val:        # nút chỉ số chẵn thắng cặp
            even_score += 1
        elif cur.next.val > cur.val:      # nút chỉ số lẻ thắng cặp
            odd_score += 1
        cur = cur.next.next               # sang cặp tiếp theo
    if even_score > odd_score:
        return "Even"
    if odd_score > even_score:
        return "Odd"
    return "Tie"


for values, expected in examples:
    print(game_result_pairs(build_list(values)), "(kỳ vọng", expected, ")")
````

````python
# Code nộp LeetCode cho Cách 2 (đổi tên lớp thành Solution khi nộp)
class SolutionPairs:
    def gameResult(self, head: Optional[ListNode]) -> str:
        even_score = odd_score = 0
        cur = head
        while cur:
            if cur.val > cur.next.val:
                even_score += 1
            elif cur.next.val > cur.val:
                odd_score += 1
            cur = cur.next.next
        if even_score > odd_score:
            return "Even"
        if odd_score > even_score:
            return "Odd"
        return "Tie"


print(SolutionPairs().gameResult(build_list([4, 5, 2, 1])))      # Tie
````

**Độ phức tạp Cách 2:** **Time `O(n)`** (mỗi nút thăm một lần), **Space `O(1)`**.

### Cách 3 — Một bộ đếm hiệu số `balance`

**Giải thích.** Thay hai biến điểm bằng một biến `balance`. Cặp nào `Even` thắng thì `+1`, cặp nào `Odd` thắng thì `-1`. Vì giá trị chẵn/lẻ khác tính chẵn lẻ nên hai nút trong cặp không bao giờ bằng nhau; ta chỉ cần một câu `if ... else`. (Nếu muốn an toàn cho cả trường hợp bằng nhau, thêm nhánh `elif`; xem §13.)

````python
def game_result_balance(head: Optional[ListNode]) -> str:
    balance = 0
    cur = head
    while cur:
        balance += 1 if cur.val > cur.next.val else -1
        cur = cur.next.next
    return "Even" if balance > 0 else "Odd" if balance < 0 else "Tie"


for values, expected in examples:
    print(game_result_balance(build_list(values)), "(kỳ vọng", expected, ")")
````

**Độ phức tạp Cách 3:** **Time `O(n)`**, **Space `O(1)`**. Mã ngắn nhất, nhưng dựa vào đảm bảo "không bằng nhau" của đề (ràng buộc chẵn/lẻ). Nếu tự mở rộng đề cho giá trị bất kỳ thì Cách 2 an toàn hơn.

### Cách 4 — Duyệt từng nút một, đổi vai theo chỉ số (không nhìn trước `cur.next`)

**Giải thích.** Có thể đi qua **mọi nút** bằng `cur = cur.next` bình thường và nhớ lại **nút chỉ số chẵn trước đó** trong biến `prev`. Khi `idx` chẵn, ta lưu `prev = cur.val`; khi `idx` lẻ, ta so `prev` với `cur.val` rồi chấm điểm. Cách này không bao giờ đọc `cur.next.val`, nên không lo `None` dù độ dài lẻ (an toàn hơn nếu tự mở rộng đề).

````python
def game_result_stream(head: Optional[ListNode]) -> str:
    balance = 0
    prev = 0
    idx = 0
    cur = head
    while cur:
        if idx % 2 == 0:                    # nút đầu của cặp: chỉ ghi nhớ
            prev = cur.val
        else:                               # nút thứ hai của cặp: so với prev
            if prev > cur.val:
                balance += 1                # Even thắng
            elif cur.val > prev:
                balance -= 1                # Odd thắng
        idx += 1
        cur = cur.next
    return "Even" if balance > 0 else "Odd" if balance < 0 else "Tie"


for values, expected in examples:
    print(game_result_stream(build_list(values)), "(kỳ vọng", expected, ")")
````

**Độ phức tạp Cách 4:** **Time `O(n)`**, **Space `O(1)`**.

### Cách 5 — Đệ quy

**Giải thích.** Định nghĩa `diff(node)` = hiệu số điểm (Even trừ Odd) của **phần danh sách bắt đầu từ `node`**. Trường hợp cơ sở: `node` là `None` ⇒ `0`. Ngược lại: điểm của cặp đầu (`+1`, `-1` hoặc `0`) cộng `diff(node.next.next)`. Đệ quy hay, nhưng dùng ngăn xếp gọi hàm `O(n)`.

````python
def game_result_recursive(head: Optional[ListNode]) -> str:
    def diff(node: Optional[ListNode]) -> int:
        if node is None:
            return 0
        gain = (node.val > node.next.val) - (node.next.val > node.val)   # +1, -1 hoặc 0
        return gain + diff(node.next.next)

    d = diff(head)
    return "Even" if d > 0 else "Odd" if d < 0 else "Tie"


for values, expected in examples:
    print(game_result_recursive(build_list(values)), "(kỳ vọng", expected, ")")
````

**Độ phức tạp Cách 5:** **Time `O(n)`**, **Space `O(n)`** do ngăn xếp đệ quy (tối đa `n / 2 = 50` tầng nên không lo tràn).

## 8. TRACE từng bước

Ta trace Cách 2 (hai điểm số) và Cách 4 (từng nút một) trên `[2, 5, 4, 7, 20, 5]`.

````python
def trace_pairs(values: List[int]) -> str:
    head = build_list(values)
    print("danh sách:", show_list(head))
    even_score = odd_score = 0
    cur = head
    step = 0
    while cur:
        a, b = cur.val, cur.next.val
        if a > b:
            even_score += 1
            note = "Even thắng cặp"
        elif b > a:
            odd_score += 1
            note = "Odd thắng cặp"
        else:
            note = "hòa cặp"
        print(f"cặp {step}: ({a}, {b}) -> {note} | Even={even_score} Odd={odd_score}")
        cur = cur.next.next
        step += 1
    result = "Even" if even_score > odd_score else "Odd" if odd_score > even_score else "Tie"
    print("kết quả:", result)
    return result


assert trace_pairs([2, 5, 4, 7, 20, 5]) == "Odd"
````

````python
def trace_stream(values: List[int]) -> str:
    head = build_list(values)
    balance, prev, idx = 0, 0, 0
    cur = head
    while cur:
        if idx % 2 == 0:
            prev = cur.val
            print(f"idx={idx} val={cur.val}: chỉ số chẵn -> nhớ prev={prev}")
        else:
            if prev > cur.val:
                balance += 1
                note = "Even +1"
            elif cur.val > prev:
                balance -= 1
                note = "Odd +1 (balance -1)"
            else:
                note = "hòa"
            print(f"idx={idx} val={cur.val}: so với prev={prev} -> {note}, balance={balance}")
        idx += 1
        cur = cur.next
    result = "Even" if balance > 0 else "Odd" if balance < 0 else "Tie"
    print("kết quả:", result)
    return result


assert trace_stream([2, 5, 4, 7, 20, 5]) == "Odd"
````

Bảng trace Cách 2 với `[2, 5, 4, 7, 20, 5]`:

| cặp | (a, b) | điều kiện | Even | Odd |
|---|---|---|---|---|
| 0 | (2, 5) | 5 > 2 | 0 | 1 |
| 1 | (4, 7) | 7 > 4 | 0 | 2 |
| 2 | (20, 5) | 20 > 5 | 1 | 2 |

Hết danh sách (`cur` thành `None`), `Odd (2) > Even (1)` nên trả `"Odd"`.

## 9. Code nộp LeetCode

Bản nộp dùng Cách 2 (hai điểm số): an toàn cả khi hai giá trị bằng nhau, vẫn `O(n)` thời gian và `O(1)` bộ nhớ. Trên LeetCode, lớp `ListNode` có sẵn nên bạn **chỉ nộp phần `class Solution`**.

````python
class Solution:
    def gameResult(self, head: Optional[ListNode]) -> str:
        even_score = odd_score = 0
        cur = head
        while cur:                                   # cur luôn là nút ở chỉ số chẵn
            if cur.val > cur.next.val:
                even_score += 1                      # đội Even thắng cặp này
            elif cur.next.val > cur.val:
                odd_score += 1                       # đội Odd thắng cặp này
            cur = cur.next.next                      # sang cặp tiếp theo
        if even_score > odd_score:
            return "Even"
        if odd_score > even_score:
            return "Odd"
        return "Tie"


sol = Solution()
for values, expected in examples:
    print(values, "->", sol.gameResult(build_list(values)))
````

## 10. Vì sao thuật toán đúng

- **Phân hoạch thành cặp:** vì độ dài chẵn, các nút chia thành các cặp `(0,1), (2,3), ...` không sót, không trùng. `cur` luôn đứng ở chỉ số chẵn nhờ nhảy hai bước từ chỉ số `0`.
- **Không đọc `None`:** khi `cur` là một nút ở chỉ số chẵn, chỉ số kế tiếp `cur.next` tồn tại (độ dài chẵn). Còn `cur.next.next` có thể là `None`, và ta chỉ gán nó cho `cur` (không đọc `.val`), vòng lặp dừng đúng.

## 13. Các lỗi người mới hay gặp

### Lỗi 1: Nhảy một nút thay vì hai nút

````python
def sai_nhay_mot_nut(head):
    balance = 0
    cur = head
    while cur and cur.next:
        balance += 1 if cur.val > cur.next.val else -1
        cur = cur.next                  # SAI: nhảy 1 nút, nên ghép cả (idx 1, idx 2), (idx 2, idx 3)...
    return "Even" if balance > 0 else "Odd" if balance < 0 else "Tie"


print("sai:", sai_nhay_mot_nut(build_list([2, 1, 4, 5, 6, 3])), "| đúng:", Solution().gameResult(build_list([2, 1, 4, 5, 6, 3])))
````

**Vì sao sai:** `cur = cur.next` làm `cur` lần lượt đứng ở **mọi** chỉ số, nên ta tính cả các cặp `(1, 2)`, `(3, 4)` — những cặp không có trong đề. **Cách sửa:** `cur = cur.next.next`.

### Lỗi 2: Gán nhầm đội theo chỉ số

````python
def sai_dao_doi(head):
    even_score = odd_score = 0
    cur = head
    while cur:
        if cur.val > cur.next.val:
            odd_score += 1              # SAI: nút chỉ số chẵn lớn hơn thì Even được điểm
        elif cur.next.val > cur.val:
            even_score += 1
        cur = cur.next.next
    return "Even" if even_score > odd_score else "Odd" if odd_score > even_score else "Tie"


print("sai:", sai_dao_doi(build_list([2, 1])), "| đúng:", Solution().gameResult(build_list([2, 1])))
````

**Vì sao sai:** "Even" là đội của nút ở **chỉ số chẵn** (không phải của số "chẵn" trong tên, dù đề đảm bảo hai điều trùng nhau). Nút chỉ số chẵn lớn hơn ⇒ **Even** được điểm. **Cách sửa:** `cur.val > cur.next.val` ⇒ `even_score += 1`.

### Lỗi 3: Chữ hoa/thường hoặc tên đội sai

````python
def sai_ten_doi(head):
    result = Solution().gameResult(head)
    return result.lower()               # SAI: LeetCode yêu cầu "Even", "Odd", "Tie"


print("sai:", sai_ten_doi(build_list([2, 1])), "| đúng:", Solution().gameResult(build_list([2, 1])))
````

### Lỗi 4: Đọc `cur.next.val` khi `cur.next` là `None` (nếu tự thử danh sách lẻ)

````python
def game_pairs_unsafe(head):
    balance = 0
    cur = head
    while cur:
        balance += 1 if cur.val > cur.next.val else -1     # lỗi nếu cur là nút cuối (độ dài lẻ)
        cur = cur.next.next
    return balance


try:
    game_pairs_unsafe(build_list([2, 1, 4]))              # độ dài lẻ: vi phạm đề
except AttributeError as e:
    print("AttributeError:", e)
print("Cách 4 (stream) vẫn chạy được:", game_result_stream(build_list([2, 1, 4])))
````

## 14. Mẹo làm bài nhanh & nhận dạng pattern

- **Hai nút một lần** trên danh sách liên kết ⇒ `cur = cur.next.next`. Điều kiện dừng `while cur:` (độ dài chẵn) hoặc `while cur and cur.next:` (độ dài bất kỳ).
- Khi chỉ cần **tổng kết** (ai thắng) thì giữ **một số** (`balance`) thay vì nhiều biến, tiết kiệm code.
- Danh sách liên kết hiếm khi cần mảng phụ; đổ ra mảng là cách **debug** tốt, nhưng bản nộp thường là một lượt `O(1)` bộ nhớ.
- `(a > b) - (a < b)` cho `+1`, `0`, `-1` (hàm "dấu") — mẹo gọn khi so sánh hai số.
- Khi tự thử trên máy, nhớ **tự dựng `ListNode`** và hàm `build_list` / `show_list`.

## 15. Biến thể của bài

1. **Danh sách độ dài bất kỳ** (nút cuối lẻ bị bỏ): dùng `while cur and cur.next`.
2. **Trả về điểm số** `(even_score, odd_score)` thay vì tên đội.
3. **Ghép cặp theo cách khác:** `(1, 2), (3, 4), ...` (bắt đầu từ chỉ số 1).
4. **Đội thắng là đội có tổng giá trị lớn hơn** (không phải số cặp thắng).
5. **Swap Nodes in Pairs (24)**, **Odd Even Linked List (328)**: cũng thao tác theo cặp / chỉ số chẵn lẻ trên danh sách liên kết.

````python
def scores(head):
    even_score = odd_score = 0
    cur = head
    while cur and cur.next:                 # an toàn cho độ dài bất kỳ
        if cur.val > cur.next.val:
            even_score += 1
        elif cur.next.val > cur.val:
            odd_score += 1
        cur = cur.next.next
    return even_score, odd_score


def total_sum_winner(head):
    values = list_to_array(head)
    e = sum(values[0::2])
    o = sum(values[1::2])
    return "Even" if e > o else "Odd" if o > e else "Tie"


print(scores(build_list([2, 5, 4, 7, 20, 5])))            # (1, 2)
print(scores(build_list([2, 1, 4])))                      # (1, 0): nút cuối bị bỏ
print(total_sum_winner(build_list([2, 5, 4, 7, 20, 5])))  # Even: 26 vs 17
assert scores(build_list([2, 5, 4, 7, 20, 5])) == (1, 2)
assert scores(build_list([2, 1, 4])) == (1, 0)
assert total_sum_winner(build_list([2, 5, 4, 7, 20, 5])) == "Even"
````
