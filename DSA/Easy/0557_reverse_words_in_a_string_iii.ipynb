## 3. Cách 1 — Tách, đảo từng từ, ghép lại

Ba bước tự nhiên:
1. Tách câu thành các từ: `s.split(" ")`.
2. Đảo từng từ: `w[::-1]`.
3. Ghép lại với một dấu cách: `" ".join(...)`.

````python
def reverse_words_split(s: str) -> str:
    words = s.split(" ")
    reversed_words = [w[::-1] for w in words]
    return " ".join(reversed_words)

print(reverse_words_split("Let's take LeetCode contest"))
print(reverse_words_split("Mr Ding"))
````

````python
def trace_split(s: str) -> str:
    words = s.split(" ")
    print("Bước 1 - split:", words)
    rev = []
    for i, w in enumerate(words):
        rev.append(w[::-1])
        print(f"Bước 2.{i} - đảo {w!r:>12} -> {w[::-1]!r}")
    ans = " ".join(rev)
    print("Bước 3 - join:", repr(ans))
    return ans

trace_split("Let's take LeetCode contest")
````

### Code nộp LeetCode — Cách 1

Có thể viết một dòng với **generator expression** `(w[::-1] for w in ...)` — giống list comprehension nhưng dùng ngoặc tròn, không tạo list trung gian.

````python
class Solution:
    def reverseWords(self, s: str) -> str:
        return " ".join(w[::-1] for w in s.split(" "))

print(Solution().reverseWords("Let's take LeetCode contest"))
````

## 4. Nút thắt: tự tìm ranh giới của mỗi từ

Nếu không có `split`, ta phải tự biết **mỗi từ bắt đầu và kết thúc ở đâu**.

Quan sát với `s = "Mr Ding"`:

```text
vị trí: 0 1 2 3 4 5 6
ký tự : M r _ D i n g
```

- Từ 1: `[0, 1]`, kết thúc vì `s[2]` là dấu cách.
- Từ 2: `[3, 6]`, kết thúc vì hết chuỗi.

Quy tắc: dùng biến `start` = vị trí đầu từ hiện tại. Duyệt `i`; khi `i == n` (hết chuỗi) **hoặc** `s[i] == " "` thì từ hiện tại là `s[start:i]` → xử lý, rồi đặt `start = i + 1`.

Mẹo: cho `i` chạy tới `n` (tức `range(n + 1)`) để từ cuối cùng cũng được xử lý trong cùng vòng lặp, không cần code riêng sau vòng.

````python
s = "Mr Ding"
start = 0
for i in range(len(s) + 1):
    if i == len(s) or s[i] == " ":
        print(f"tìm thấy từ s[{start}:{i}] = {s[start:i]!r}")
        start = i + 1
````

Thứ tự điều kiện `i == len(s) or s[i] == " "` rất quan trọng: nếu `i == len(s)` đúng thì Python **không** đánh giá `s[i]` nữa (gọi là **short-circuit**), nhờ đó không bị `IndexError`.

## 5. Cách 2 — Tự quét từ, đảo bằng vòng lặp ngược

Với mỗi từ `s[start:i]`, ta tự đảo bằng cách đi từ `i - 1` lùi về `start`: `range(i - 1, start - 1, -1)`.

````python
print(list(range(6, 2, -1)))   # 6 5 4 3 -> duyệt lùi từ 6 về 3
````

````python
def reverse_words_scan(s: str) -> str:
    out = []
    start = 0
    n = len(s)
    for i in range(n + 1):
        if i == n or s[i] == " ":
            for j in range(i - 1, start - 1, -1):   # chép từ theo thứ tự ngược
                out.append(s[j])
            if i < n:
                out.append(" ")                    # giữ dấu cách
            start = i + 1
    return "".join(out)

print(reverse_words_scan("Let's take LeetCode contest"))
print(reverse_words_scan("Mr Ding"))
print(reverse_words_scan("a"))
````

## 6. Cách 3 — Hai con trỏ trên list (đảo tại chỗ)

Ý tưởng giống C++/Java: đổi `s` thành list ký tự, rồi với mỗi từ `[start, end]` dùng hai con trỏ `left = start`, `right = end` đổi chỗ và tiến vào giữa.

Phần đảo một đoạn tách thành hàm nhỏ `reverse_range(chars, left, right)` cho dễ đọc.

````python
def reverse_range(chars: list, left: int, right: int) -> None:
    while left < right:
        chars[left], chars[right] = chars[right], chars[left]
        left += 1
        right -= 1

demo = list("LeetCode")
reverse_range(demo, 0, len(demo) - 1)
print("".join(demo))
````

````python
def reverse_words_two_pointers(s: str) -> str:
    chars = list(s)
    n = len(chars)
    start = 0
    for i in range(n + 1):
        if i == n or chars[i] == " ":
            reverse_range(chars, start, i - 1)
            start = i + 1
    return "".join(chars)

print(reverse_words_two_pointers("Let's take LeetCode contest"))
print(reverse_words_two_pointers("Mr Ding"))
````

## 7. Trace Cách 3

`s = "Mr Ding"`:

| i | `chars[i]` | Ranh giới? | Đoạn đảo `[start, i-1]` | chars sau | start mới |
|---:|---|:---:|---|---|---:|
| 0 | M | không | | `Mr Ding` | 0 |
| 1 | r | không | | `Mr Ding` | 0 |
| 2 | ` ` | **có** | `[0, 1]` | `rM Ding` | 3 |
| 3..6 | D,i,n,g | không | | | 3 |
| 7 | (hết) | **có** | `[3, 6]` | `rM gniD` | 8 |

````python
def trace_two_pointers(s: str) -> str:
    chars = list(s)
    n = len(chars)
    start = 0
    for i in range(n + 1):
        boundary = i == n or chars[i] == " "
        label = "(hết)" if i == n else repr(chars[i])
        if boundary:
            before = "".join(chars)
            reverse_range(chars, start, i - 1)
            print(f"i={i} {label:>6} ranh giới -> đảo [{start}, {i - 1}] : {before!r} -> {''.join(chars)!r}")
            start = i + 1
        else:
            print(f"i={i} {label:>6}")
    return "".join(chars)

trace_two_pointers("Mr Ding")
````

### Code nộp LeetCode — Hai con trỏ

````python
class Solution:
    def reverseWords(self, s: str) -> str:
        chars = list(s)
        n = len(chars)
        start = 0
        for i in range(n + 1):
            if i == n or chars[i] == " ":
                left, right = start, i - 1
                while left < right:
                    chars[left], chars[right] = chars[right], chars[left]
                    left += 1
                    right -= 1
                start = i + 1
        return "".join(chars)

print(Solution().reverseWords("Let's take LeetCode contest"))
````

## 8. Cách 4 — Mẹo "đảo hai lần" (sáng tạo)

Quan sát:
- Đảo **cả câu**: `"Mr Ding"` → `"gniD rM"` — ký tự trong mỗi từ đã bị đảo, nhưng **thứ tự từ** cũng bị đảo.
- Muốn trả thứ tự từ về như cũ: tách theo dấu cách rồi **đảo thứ tự list từ**.

```text
"Mr Ding" --đảo cả chuỗi--> "gniD rM" --split--> ["gniD", "rM"] --đảo list--> ["rM", "gniD"] --join--> "rM gniD"
```

Mẹo đảo hai lần này rất hay gặp (ví dụ "xoay mảng" bằng 3 lần đảo, LeetCode 189).

````python
def reverse_words_double(s: str) -> str:
    return " ".join(s[::-1].split(" ")[::-1])

s = "Mr Ding"
print(repr(s[::-1]))
print(s[::-1].split(" "))
print(s[::-1].split(" ")[::-1])
print(reverse_words_double(s))
print(reverse_words_double("Let's take LeetCode contest"))
````

## 9. Vì sao đúng?

**Cách 1:** vì các từ cách nhau đúng một dấu cách và không có dấu cách đầu/cuối, `" ".join(s.split(" "))` == `s`. Thay mỗi từ bằng bản đảo của nó không đổi độ dài từ, nên kết quả đúng định nghĩa: mỗi từ bị đảo, thứ tự từ và dấu cách giữ nguyên.

**Cách 3 — invariant:** trước khi xét vị trí `i`,
- mọi từ kết thúc trước `start - 1` đã được đảo đúng trong `chars`;
- `chars[start:i]` vẫn là nguyên bản của từ đang đọc dở (chưa bị đụng tới).

Khi gặp ranh giới, đảo đoạn `[start, i-1]` hoàn tất từ đó; dấu cách tại `i` không bao giờ bị di chuyển. Khi `i = n`, từ cuối cũng được đảo ⇒ toàn bộ đúng.

**Cách 4:** đảo cả chuỗi biến `w1 _ w2 _ ... _ wk` thành `rev(wk) _ ... _ rev(w1)`. Đảo thứ tự list từ cho `rev(w1) _ ... _ rev(wk)` ✔.

## 10. So sánh và độ phức tạp

Gọi `n = len(s)`.

| Cách | Ý tưởng | Time | Space thêm | Ghi chú |
|---|---|---|---|---|
| 1. split + `[::-1]` + join | dùng thư viện | `O(n)` | `O(n)` | **ngắn, khuyên dùng** |
| 2. Tự quét + chép ngược | tự tìm ranh giới | `O(n)` | `O(n)` | luyện chỉ số |
| 3. Hai con trỏ trên list | đảo tại chỗ từng đoạn | `O(n)` | `O(n)` cho list (chuỗi Python bất biến) | giống C++/Java, `O(1)` nếu input là mảng |
| 4. Đảo hai lần | đảo câu + đảo list từ | `O(n)` | `O(n)` | mẹo hay, áp dụng cho bài khác |

Mỗi ký tự chỉ được đọc/ghi một số lần hằng số → `O(n)`.

### Lỗi 3: Quên xử lý từ cuối cùng khi tự quét

Nếu chỉ xử lý khi gặp dấu cách (`range(n)` thay vì `range(n + 1)`), từ cuối không bao giờ được đảo.

````python
def wrong_forget_last(s: str) -> str:
    chars = list(s)
    start = 0
    for i in range(len(chars)):              # SAI: không có i == n
        if chars[i] == " ":
            reverse_range(chars, start, i - 1)
            start = i + 1
    return "".join(chars)

print("sai :", wrong_forget_last("Mr Ding"), "| đúng:", reverse_words_two_pointers("Mr Ding"))
````

### Lỗi 4: Đảo đoạn `[start, i]` thay vì `[start, i - 1]`

`i` là vị trí **dấu cách** (hoặc `n`), không thuộc từ. Đảo tới `i` sẽ kéo dấu cách vào đầu từ, hoặc `IndexError` khi `i == n`.

````python
def wrong_off_by_one(s: str) -> str:
    chars = list(s)
    start = 0
    for i in range(len(chars) + 1):
        if i == len(chars) or chars[i] == " ":
            reverse_range(chars, start, min(i, len(chars) - 1))   # SAI: tới i
            start = i + 1
    return "".join(chars)

print("sai :", repr(wrong_off_by_one("Mr Ding")))
````

````python
s = "ab"
i = 2
try:
    print(s[i] == " " or i == len(s))
except IndexError as e:
    print("IndexError:", e)
print(i == len(s) or s[i] == " ")   # an toàn nhờ short-circuit
````

## 13. Mẹo nhận dạng pattern

- Đề "xử lý **từng từ** riêng biệt" → `split` + xử lý + `join`, hoặc quét với `start`/ranh giới.
- Cần đảo một đoạn tại chỗ → **hai con trỏ** `left`, `right`.
- Xử lý theo đoạn mà đoạn cuối không có "dấu kết thúc" → cho vòng lặp chạy tới `n` (sentinel) hoặc xử lý riêng sau vòng.
- Muốn đảo thứ tự **khối** mà giữ nội dung khối → mẹo **đảo hai lần**.

````python
# Biến thể 1 demo: đảo thứ tự từ, bỏ khoảng trắng thừa
def reverse_word_order(s: str) -> str:
    return " ".join(reversed(s.split()))

print(repr(reverse_word_order("  the sky   is blue  ")))

# Biến thể 4 demo: xoay mảng sang phải k bước bằng 3 lần đảo
def rotate(nums: list, k: int) -> list:
    nums = nums[:]
    n = len(nums)
    k %= n
    reverse_range(nums, 0, n - 1)
    reverse_range(nums, 0, k - 1)
    reverse_range(nums, k, n - 1)
    return nums

print(rotate([1, 2, 3, 4, 5, 6, 7], 3))   # [5, 6, 7, 1, 2, 3, 4]
````

`reversed(list)` trả về iterator duyệt ngược; `" ".join` nhận iterator được. `nums[:]` tạo **bản sao nông** của list để không sửa list gốc.

## 17. Đáp án tham khảo (tự làm trước rồi mới xem)

````python
def answer_count_words(s: str) -> int:
    return s.count(" ") + 1

def answer_reverse_long_words(s: str, m: int) -> str:
    return " ".join(w[::-1] if len(w) >= m else w for w in s.split(" "))

def answer_reverse_words_ex(s: str) -> str:
    chars = list(s)
    n = len(chars)
    start = 0
    for i in range(n + 1):
        if i == n or chars[i] == " ":
            left, right = start, i - 1
            while left < right:
                chars[left], chars[right] = chars[right], chars[left]
                left += 1
                right -= 1
            start = i + 1
    return "".join(chars)

def answer_reverse_words_keep_spaces(s: str) -> str:
    # split(" ") giữ các phần rỗng giữa những dấu cách liền nhau -> join lại khôi phục đúng dấu cách
    return " ".join(w[::-1] for w in s.split(" "))

def answer_reverse_word_order_inplace(chars: list) -> list:
    n = len(chars)
    reverse_range(chars, 0, n - 1)          # 1) đảo cả list
    start = 0
    for i in range(n + 1):                  # 2) đảo lại từng từ
        if i == n or chars[i] == " ":
            reverse_range(chars, start, i - 1)
            start = i + 1
    return chars

assert answer_count_words("Let's take LeetCode contest") == 4
assert answer_count_words("a") == 1
assert answer_reverse_long_words("Mr Ding is here", 3) == "Mr gniD is ereh"
assert answer_reverse_words_ex("Let's take LeetCode contest") == "s'teL ekat edoCteeL tsetnoc"
assert answer_reverse_words_ex("a") == "a"
assert answer_reverse_words_keep_spaces("  ab  cd ") == "  ba  dc "
assert answer_reverse_words_keep_spaces("   ") == "   "
assert "".join(answer_reverse_word_order_inplace(list("the sky is blue"))) == "blue is sky the"
assert "".join(answer_reverse_word_order_inplace(list("one"))) == "one"
print("Đáp án chạy PASS toàn bộ test!")
````
