## 3. Cách 1 — Qua chuỗi (dễ nghĩ nhất)

1. Lấy list chữ số bằng `str` + `int`.
2. Tích bằng vòng `for` (khởi tạo `1`), tổng bằng `sum`.
3. Trả `tích − tổng`.

````python
def subtract_string(n: int) -> int:
    digits = [int(ch) for ch in str(n)]
    product = 1                   # phần tử trung hòa của phép nhân
    for d in digits:
        product *= d
    return product - sum(digits)

print(subtract_string(234))     # 15
print(subtract_string(4421))    # 21
print(subtract_string(10))      # -1
````

**Độ phức tạp:** gọi `d` = số chữ số (`d = ⌊log₁₀ n⌋ + 1`, với `n ≤ 10^5` thì `d ≤ 6`). Thời gian `O(d)`, bộ nhớ `O(d)` cho chuỗi và list chữ số.

### Phiên bản thư viện một dòng

````python
import math

def subtract_builtin(n: int) -> int:
    digits = list(map(int, str(n)))
    return math.prod(digits) - sum(digits)

print(subtract_builtin(234), subtract_builtin(4421))   # 15 21
````

Chú ý: phải `list(...)` trước khi dùng hai lần. Một đối tượng `map` chỉ duyệt được **một lần** — lần thứ hai nó rỗng!

````python
m = map(int, "234")
print(math.prod(m))   # 24
print(sum(m))         # 0  <- map đã bị "tiêu thụ" hết
````

## 4. Nút thắt

Cách 1 đúng và gọn, nhưng:
- Tạo thêm chuỗi và list (bộ nhớ `O(d)`).
- Duyệt dữ liệu **hai lần** (một lần cho tích, một lần cho tổng).
- Phỏng vấn thường yêu cầu **không dùng chuỗi**, để kiểm tra kỹ năng `% 10` / `// 10`.

## 5. Insight — một lượt bóc chữ số, cập nhật cả tích lẫn tổng

Trong mỗi vòng `while`, ta có đúng một chữ số `d = n % 10`. Cập nhật luôn cả hai:
```text
product = product * d
total   = total + d
```
Cuối cùng trả `product - total`. Không cần lưu chữ số nào.

## 6. Cách 2 — Toán học một lượt (lời giải chính)

````python
def subtract_math(n: int) -> int:
    product, total = 1, 0
    while n > 0:
        d = n % 10
        product *= d
        total += d
        n //= 10
    return product - total

print(subtract_math(234))     # 15
print(subtract_math(4421))    # 21
print(subtract_math(10))      # -1
````

### Viết gọn với `divmod`

`n, d = divmod(n, 10)` vừa bỏ chữ số cuối (`n // 10`) vừa lấy nó (`n % 10`) trong một lệnh.
`product, total = 1, 0` là **gán nhiều biến** cùng lúc (tuple unpacking).

````python
def subtract_divmod(n: int) -> int:
    product, total = 1, 0
    while n:                       # n != 0
        n, d = divmod(n, 10)
        product *= d
        total += d
    return product - total

print(subtract_divmod(234), subtract_divmod(4421))   # 15 21
````

`while n:` tương đương `while n != 0:` vì số `0` được coi là **False** khi đặt trong điều kiện ("falsy"), mọi số khác `0` là **True**.

## 7. Cách 3 — `functools.reduce` (góc nhìn lập trình hàm)

`reduce(f, [a, b, c], init)` tính `f(f(f(init, a), b), c)` — "gấp" list thành một giá trị. Có thể gấp **một lần** ra cặp `(tích, tổng)`.

````python
from functools import reduce
import operator

print(reduce(operator.mul, [2, 3, 4], 1))     # ((1*2)*3)*4 = 24
print(reduce(operator.add, [2, 3, 4], 0))     # 9

def subtract_reduce(n: int) -> int:
    p, s = reduce(lambda acc, d: (acc[0] * d, acc[1] + d), map(int, str(n)), (1, 0))
    return p - s

print(subtract_reduce(234), subtract_reduce(4421))   # 15 21
````

- `operator.mul`, `operator.add` là các hàm tương ứng với `*`, `+`.
- `lambda acc, d: (...)` là hàm vô danh nhận "tích lũy" `acc = (tích, tổng)` và chữ số `d`, trả về cặp mới.

Cách này hay để mở rộng tầm nhìn, nhưng khó đọc hơn Cách 2 với người mới.

## 8. Cách 4 — Đệ quy

Định nghĩa: chữ số của `n` = chữ số của `n // 10` + chữ số cuối `n % 10`. Ta viết hàm đệ quy trả về `(tích, tổng)`.

````python
def prod_sum_rec(n: int) -> tuple[int, int]:
    if n < 10:                          # trường hợp cơ sở: 1 chữ số
        return n, n
    p, s = prod_sum_rec(n // 10)        # bài toán con nhỏ hơn
    d = n % 10
    return p * d, s + d


def subtract_recursive(n: int) -> int:
    p, s = prod_sum_rec(n)
    return p - s

print(prod_sum_rec(234))                 # (24, 9)
print(subtract_recursive(234), subtract_recursive(4421))   # 15 21
````

Đệ quy dùng `O(d)` bộ nhớ cho **ngăn xếp lời gọi** (call stack). Với `d ≤ 6` không vấn đề gì.

## 9. Trace từng bước (Cách 2)

`n = 4421`:

| vòng | `n` trước | `d = n % 10` | `product` sau | `total` sau | `n` sau (`// 10`) |
|---:|---:|---:|---:|---:|---:|
| 1 | 4421 | 1 | 1 | 1 | 442 |
| 2 | 442 | 2 | 2 | 3 | 44 |
| 3 | 44 | 4 | 8 | 7 | 4 |
| 4 | 4 | 4 | 32 | 11 | 0 |

Dừng vì `n = 0`. Kết quả `32 − 11 = 21`.

````python
def trace_subtract(n: int) -> int:
    product, total = 1, 0
    step = 1
    print(f"bắt đầu: n={n}, product={product}, total={total}")
    while n > 0:
        d = n % 10
        product *= d
        total += d
        print(f"vòng {step}: d={d} -> product={product:>3}, total={total:>2}, n -> {n // 10}")
        n //= 10
        step += 1
    print(f"kết quả = {product} - {total} = {product - total}")
    return product - total

trace_subtract(4421)
print()
trace_subtract(105)
````

Ví dụ `105`: vừa gặp chữ số `0`, tích về `0` và **ở lại 0 mãi** — đúng về toán học. (Có thể dừng sớm phần tích, nhưng vẫn phải tính tổng đầy đủ.)

## 10. Code nộp LeetCode

````python
class Solution:
    def subtractProductAndSum(self, n: int) -> int:
        product, total = 1, 0
        while n > 0:
            n, d = divmod(n, 10)
            product *= d
            total += d
        return product - total

print(Solution().subtractProductAndSum(234))    # 15
print(Solution().subtractProductAndSum(4421))   # 21
````

## 11. Vì sao đúng?

**Invariant của vòng `while`:** gọi `n0` là giá trị ban đầu. Sau mỗi vòng, `product` = tích các chữ số **đã bóc** (các chữ số cuối của `n0`), `total` = tổng của chúng, và `n` chứa đúng các chữ số **chưa bóc**.

- Ban đầu: chưa bóc gì → `product = 1` (tích rỗng), `total = 0` (tổng rỗng), `n = n0`. Đúng.
- Mỗi vòng: chữ số cuối của `n` được chuyển từ "chưa bóc" sang "đã bóc", cập nhật đúng vào tích và tổng.
- Kết thúc khi `n = 0`: không còn chữ số nào chưa bóc → `product`, `total` là tích và tổng **mọi** chữ số.
- Vòng lặp dừng vì `n` giảm thực sự sau mỗi vòng (`n // 10 < n` khi `n > 0`).

Vì `n >= 1`, vòng lặp chạy ít nhất một lần, không có trường hợp "số 0 không có chữ số".

## 12. Độ phức tạp và so sánh

Gọi `d` = số chữ số của `n` (≈ `log₁₀ n`).

| Cách | Thời gian | Bộ nhớ phụ | Ghi chú |
|---|---|---|---|
| 1. Chuỗi + list | `O(d)` | `O(d)` | dễ hiểu nhất |
| 1'. `map` + `math.prod` | `O(d)` | `O(d)` | ngắn nhất |
| 2. `% 10` / `divmod` một lượt | `O(d)` | `O(1)` | lời giải chính |
| 3. `reduce` | `O(d)` | `O(d)` (chuỗi) | lập trình hàm |
| 4. Đệ quy | `O(d)` | `O(d)` (call stack) | luyện tư duy đệ quy |

### Thí nghiệm nhỏ: đo thời gian bằng `timeit`

Big-O giống nhau (`O(d)`) không có nghĩa là tốc độ thực tế giống nhau. Module `timeit` chạy một hàm nhiều lần và đo tổng thời gian. Con số cụ thể **phụ thuộc máy**, nên ta chỉ in tỉ lệ để so sánh tương đối, không assert gì về tốc độ.

Điều đáng học: với số rất nhỏ (≤ 6 chữ số), chênh lệch giữa các cách là vài trăm nano giây — **không đáng để hy sinh độ dễ đọc**. Chọn cách bạn giải thích được rõ ràng nhất.

````python
import timeit

funcs = [subtract_string, subtract_builtin, subtract_math, subtract_divmod, subtract_reduce, subtract_recursive]
times = {f.__name__: timeit.timeit(lambda f=f: f(98765), number=20000) for f in funcs}
base = times["subtract_math"]
for name, t in sorted(times.items(), key=lambda kv: kv[1]):
    print(f"{name:<20} ~ {t / base:4.2f} x (so với subtract_math)")
````

Giải thích `lambda f=f: f(98765)`: trong dict comprehension, nếu viết `lambda: f(98765)` thì mọi lambda sẽ "nhìn" cùng biến `f` và đến lúc chạy chỉ thấy giá trị **cuối cùng** của nó (late binding). Gán `f=f` làm **tham số mặc định** sẽ "chụp" giá trị `f` tại thời điểm tạo lambda. Đây là một bẫy Python kinh điển, cell dưới minh họa.

````python
fs_bad = [lambda: i for i in range(3)]
fs_good = [lambda i=i: i for i in range(3)]
print([g() for g in fs_bad])    # [2, 2, 2]  <- cùng thấy i cuối cùng
print([g() for g in fs_good])   # [0, 1, 2]
````

## 14. Lỗi người mới hay gặp

### Lỗi 1: Khởi tạo tích bằng `0`

`0 * bất_kỳ = 0` → tích luôn bằng 0.

````python
def wrong_product_zero(n):
    product, total = 0, 0                 # SAI
    while n > 0:
        d = n % 10
        product *= d
        total += d
        n //= 10
    return product - total

print("sai :", wrong_product_zero(234), "| đúng:", Solution().subtractProductAndSum(234))
````

### Lỗi 2: Dùng `/` thay vì `//`

`n / 10` ra số thực (`23.4`), `% 10` của số thực ra số thực lẻ (`3.4`)… và vòng lặp chạy rất nhiều lần cho đến khi số thực **tràn về 0** (underflow). Kết quả sai hoàn toàn.

````python
def wrong_true_division(n):
    product, total = 1, 0
    steps = 0
    while n > 0 and steps < 5:            # giới hạn 5 vòng để minh họa
        d = n % 10
        product *= d
        total += d
        n = n / 10                        # SAI
        steps += 1
        print(f"  vòng {steps}: d={d}, n={n}")
    return product - total

print("sai :", wrong_true_division(234))
````

### Lỗi 3: Đảo thứ tự phép trừ

Đề yêu cầu **tích − tổng**. Viết `total - product` cho kết quả **đổi dấu**.

````python
def wrong_order(n):
    digits = [int(ch) for ch in str(n)]
    return sum(digits) - math.prod(digits)        # SAI

print("sai :", wrong_order(234), "| đúng:", Solution().subtractProductAndSum(234))
````

## 15. Mẹo nhận dạng pattern và biến thể

**Dấu hiệu "bài xử lý chữ số":** đề nhắc tới *digits*, *sum of digits*, *reverse integer*, *palindrome number*, *count digits*…
→ Khuôn mẫu:
```python
while n > 0:
    d = n % 10      # chữ số cuối
    ...             # xử lý d
    n //= 10        # bỏ chữ số cuối
```
hoặc nhanh gọn: `for ch in str(n): d = int(ch)`.

**Bài liên quan:**
- LeetCode 258 *Add Digits*, 202 *Happy Number* (tổng bình phương chữ số).
- LeetCode 9 *Palindrome Number*, 7 *Reverse Integer*.
- LeetCode 2544 *Alternating Digit Sum*, 1295 *Find Numbers with Even Number of Digits*.
- LeetCode 1742 *Maximum Number of Balls in a Box* (nhóm theo tổng chữ số).

````python
# Mini: Happy Number (202) dùng đúng khuôn bóc chữ số
def digit_square_sum(n):
    s = 0
    while n > 0:
        n, d = divmod(n, 10)
        s += d * d
    return s

def is_happy(n):
    seen = set()
    while n != 1 and n not in seen:
        seen.add(n)
        n = digit_square_sum(n)
    return n == 1

print(digit_square_sum(19), is_happy(19), is_happy(2))   # 82 True False
````

## 18. Đáp án tham khảo

````python
ans_ex1 = 0 - (1 + 2 + 0 + 3)


def ans_count_digits(n):
    cnt = 0
    while n > 0:
        n //= 10
        cnt += 1
    return cnt


def ans_alternate_digit_sum(n):
    return sum(int(c) if i % 2 == 0 else -int(c) for i, c in enumerate(str(n)))


def ans_product_nonzero_minus_sum(n):
    product, total = 1, 0
    while n > 0:
        n, d = divmod(n, 10)
        if d:
            product *= d
        total += d
    return product - total


def ans_add_digits_loop(num):
    while num >= 10:
        s = 0
        while num > 0:
            num, d = divmod(num, 10)
            s += d
        num = s
    return num


def ans_add_digits_formula(num):
    return 0 if num == 0 else 1 + (num - 1) % 9


assert ans_ex1 == Solution().subtractProductAndSum(1203) == -6
assert ans_count_digits(7) == 1
assert ans_count_digits(100000) == 6
assert ans_alternate_digit_sum(521) == 4
assert ans_alternate_digit_sum(111) == 1
assert ans_alternate_digit_sum(886996) == 0
assert ans_product_nonzero_minus_sum(105) == -1
assert ans_product_nonzero_minus_sum(234) == 15
assert ans_add_digits_loop(38) == 2 and ans_add_digits_formula(38) == 2
assert ans_add_digits_loop(0) == 0 and ans_add_digits_formula(0) == 0
for v in range(0, 20000):
    assert ans_add_digits_loop(v) == ans_add_digits_formula(v)
    if v:
        assert ans_count_digits(v) == len(str(v))
print("Đáp án bài tập: tất cả assert PASS")
````
