## 1. Mục tiêu học

Sau notebook này bạn sẽ:

1. Hiểu đề: thực hiện **đúng thứ tự** ba bước — đổi mỗi số **chẵn** thành `0`, đổi mỗi số **lẻ** thành `1`, rồi **sắp xếp** mảng theo thứ tự không giảm.
2. Dùng phép `% 2` (hoặc `& 1`) để xét tính chẵn/lẻ, và dùng `sorted`, list comprehension một cách thành thạo.
3. Nhận ra bước đột phá: sau khi đổi, mảng chỉ chứa hai giá trị `0` và `1`, nên **không cần sort** — chỉ cần **đếm** số chẵn rồi viết `0` lặp đó lần, `1` lặp phần còn lại.
4. Viết 4 cách: ánh xạ rồi `sorted`, đếm số chẵn, điền hai đầu bằng hai con trỏ, và dùng phép bit `x & 1`; đối chiếu chúng bằng test ngẫu nhiên có `random.seed`.
5. Tránh các lỗi quen thuộc: sắp xếp **trước** khi ánh xạ, quên ánh xạ, đảo thứ tự `0`/`1`, và nhầm `0` với `1` khi kiểm tra chẵn/lẻ.

## 2. Đề bài & ví dụ

Cho mảng số nguyên `nums`. Hãy biến đổi `nums` bằng các thao tác sau, **theo đúng thứ tự**:

1. Thay mỗi số **chẵn** bằng `0`.
2. Thay mỗi số **lẻ** bằng `1`.
3. **Sắp xếp** mảng đã biến đổi theo thứ tự **không giảm** (từ nhỏ đến lớn).

Trả về mảng sau khi thực hiện xong.

| Ví dụ | nums | Output | Giải thích |
|---|---|---|---|
| 1 | `[4,3,2,1]` | `[0,0,1,1]` | chẵn `4,2` -> `0,0`; lẻ `3,1` -> `1,1`; mảng `[0,1,0,1]` sort thành `[0,0,1,1]` |
| 2 | `[1,5,1,4,2]` | `[0,0,1,1,1]` | lẻ `1,5,1` -> `1,1,1`; chẵn `4,2` -> `0,0`; sort được `[0,0,1,1,1]` |

Constraints: `1 <= nums.length <= 100`; `1 <= nums[i] <= 1000`.

````python
nums = [4, 3, 2, 1]
mapped = []
for x in nums:
    if x % 2 == 0:
        mapped.append(0)
    else:
        mapped.append(1)
print("sau khi đổi :", mapped)
mapped.sort()
print("sau khi sort:", mapped)
````

## 3. Kiến thức nền

### 3.1. Chẵn / lẻ với toán tử `%`

`x % 2` là **phần dư** khi chia `x` cho `2`: bằng `0` nếu `x` chẵn, bằng `1` nếu `x` lẻ (với số nguyên dương). Vì kết quả chỉ là `0` hoặc `1`, nên `x % 2` **chính là** giá trị mà đề muốn gán cho `x`: chẵn -> `0`, lẻ -> `1`. Đây là điểm mấu chốt của cả bài.

````python
for x in [1, 2, 3, 4, 5, 1000]:
    print(f"{x} % 2 = {x % 2} -> {'lẻ' if x % 2 else 'chẵn'}")
````

### 3.2. Toán tử bit `x & 1`

Ở dạng nhị phân, chữ số **cuối** của một số cho biết nó chẵn hay lẻ: số chẵn kết thúc bằng bit `0`, số lẻ bằng bit `1`. Phép `x & 1` giữ lại đúng bit cuối đó. Với số nguyên dương, `x & 1 == x % 2`.

````python
for x in [4, 3, 2, 1, 7]:
    print(f"{x:>2} = {x:04b} (nhị phân) -> x & 1 = {x & 1}, x % 2 = {x % 2}")
````

### 3.3. `sorted()` và `list.sort()`

`sorted(a)` trả về **danh sách mới** đã sắp xếp (không sửa `a`); `a.sort()` sắp xếp **tại chỗ** và trả về `None`. Mặc định là thứ tự **không giảm** (tăng dần, cho phép bằng nhau). Cả hai tốn `O(n log n)` thời gian.

````python
a = [0, 1, 0, 1]
b = sorted(a)
print(a, b)
c = a.copy()
result = c.sort()
print(c, result)   # sort() tại chỗ trả về None - lỗi hay gặp: gán return = c.sort()
````

### 3.4. List comprehension và nhân danh sách

`[expr for x in nums]` tạo danh sách mới từ mỗi phần tử; `[0] * 3` tạo `[0, 0, 0]`; phép `+` nối hai danh sách. Ba công cụ này cho phép viết lời giải ở dạng một dòng: `[0] * số_chẵn + [1] * số_lẻ`.

````python
nums = [1, 5, 1, 4, 2]
parity = [x % 2 for x in nums]
print(parity)
print([0] * 2 + [1] * 3)
evens = sum(1 for x in nums if x % 2 == 0)
print("số chẵn =", evens, "| số lẻ =", len(nums) - evens)
````

### 3.5. Đếm bằng `sum` với biểu thức boolean

Trong Python `True` bằng `1` và `False` bằng `0`, nên `sum(x % 2 == 0 for x in nums)` đếm số phần tử chẵn. Cũng vì vậy `sum(x % 2 for x in nums)` đếm số phần tử **lẻ** (mỗi số lẻ cho `1`, mỗi số chẵn cho `0`).

````python
nums = [4, 3, 2, 1]
print(sum(x % 2 == 0 for x in nums), sum(x % 2 for x in nums), sum(x & 1 for x in nums))
````

## 4. Ý tưởng nền tảng / cách nghĩ trực giác

Làm đúng từng bước của đề, trên ví dụ `nums = [1, 5, 1, 4, 2]`:

```
bước 1+2 (đổi theo chẵn/lẻ): 1->1, 5->1, 1->1, 4->0, 2->0   => [1, 1, 1, 0, 0]
bước 3   (sắp xếp tăng dần): [0, 0, 1, 1, 1]
```

Nhận xét: sau bước đổi, mảng **chỉ có hai loại giá trị** `0` và `1`. Khi sắp xếp, mọi số `0` đứng trước mọi số `1`. Vậy kết quả hoàn toàn được xác định bởi **hai con số**: có bao nhiêu `0` (= số phần tử chẵn) và bao nhiêu `1` (= số phần tử lẻ). Thứ tự ban đầu của các phần tử **không còn quan trọng**.

## 5. Cách 1 — Brute Force / baseline

Làm nguyên văn đề: tạo mảng `mapped` bằng cách đổi từng phần tử, rồi gọi `sorted`. Rất dễ viết, ít lỗi, dùng làm **chuẩn đối chiếu** cho các cách khác.

Nút thắt: bước sort mất `O(n log n)` dù mảng chỉ có hai giá trị khác nhau — một thuật toán sắp xếp tổng quát không biết điều đó.

````python
class SolutionSort:
    def transformArray(self, nums):
        mapped = [0 if x % 2 == 0 else 1 for x in nums]
        return sorted(mapped)

print(SolutionSort().transformArray([4, 3, 2, 1]))
print(SolutionSort().transformArray([1, 5, 1, 4, 2]))
````

**Độ phức tạp:** Time `O(n log n)` (do `sorted`), Space `O(n)` cho mảng mới.

## 6. Ý tưởng thuật toán / bước đột phá

**Bước đột phá:** vì chỉ có hai giá trị `0` và `1`, sắp xếp = **gom hết số `0` về trước**. Không cần so sánh gì cả: chỉ cần đếm `e` = số phần tử chẵn, kết quả là `e` số `0` rồi `n - e` số `1`.

**Quy tắc "nếu… thì…":**

- Nếu `x` chẵn -> nó góp thêm một số `0` ở **đầu** kết quả.
- Nếu `x` lẻ -> nó góp thêm một số `1` ở **cuối** kết quả.
- Độ dài kết quả luôn bằng `n`; vị trí của `0`/`1` chỉ phụ thuộc vào **số lượng**, không phụ thuộc thứ tự xuất hiện.

Invariant (với cách đếm): sau khi duyệt xong, `e + o = n` và kết quả là `[0] * e + [1] * o`.

````python
nums = [1, 5, 1, 4, 2]
evens = 0
for x in nums:
    if x % 2 == 0:
        evens += 1
odds = len(nums) - evens
print("evens =", evens, "odds =", odds)
print([0] * evens + [1] * odds)
````

## 7. Các cách tiếp theo

Ta có 4 cách: (1) ánh xạ rồi `sorted`; (2) đếm số chẵn rồi dựng mảng; (3) hai con trỏ điền từ hai đầu; (4) dùng bit `x & 1` và `sum`.

### Cách 2 — Đếm số chẵn

Đếm `e` số chẵn, trả về `[0] * e + [1] * (n - e)`. Không sort, chỉ một lần duyệt: `O(n)`.

````python
class SolutionCount:
    def transformArray(self, nums):
        evens = 0
        for x in nums:
            if x % 2 == 0:
                evens += 1
        return [0] * evens + [1] * (len(nums) - evens)

print(SolutionCount().transformArray([4, 3, 2, 1]))
print(SolutionCount().transformArray([1, 5, 1, 4, 2]))
````

Độ phức tạp Cách 2: Time `O(n)`, Space `O(n)` cho mảng kết quả (không tính được là `O(1)` vì phải trả về mảng mới).

### Cách 3 — Hai con trỏ điền từ hai đầu

Tạo `res = [0] * n`. Dùng `lo = 0` (vị trí `0` kế tiếp) và `hi = n - 1` (vị trí `1` kế tiếp tính từ cuối). Với mỗi `x`: nếu chẵn thì **giữ** `res[lo] = 0` rồi `lo += 1`; nếu lẻ thì `res[hi] = 1` rồi `hi -= 1`. Tại mọi lúc, đoạn `res[0..lo)` toàn `0`, đoạn `res(hi..n-1]` toàn `1`. Cách này gợi cho ta kỹ thuật **phân hoạch** (partition) hay gặp trong quicksort / "Dutch flag".

````python
class SolutionTwoPointers:
    def transformArray(self, nums):
        n = len(nums)
        res = [0] * n
        lo, hi = 0, n - 1
        for x in nums:
            if x % 2 == 0:
                res[lo] = 0
                lo += 1
            else:
                res[hi] = 1
                hi -= 1
        return res

print(SolutionTwoPointers().transformArray([4, 3, 2, 1]))
print(SolutionTwoPointers().transformArray([1, 5, 1, 4, 2]))
````

Độ phức tạp Cách 3: Time `O(n)`, Space `O(n)` (mảng kết quả). Sau khi duyệt hết, `lo == hi + 1` vì `lo` đếm số chẵn và `n - 1 - hi` đếm số lẻ.

### Cách 4 — Bit `x & 1` và `sum`

Số lẻ <-> bit cuối bằng `1`. Vậy `ones = sum(x & 1 for x in nums)` là số phần tử lẻ, kết quả là `[0] * (n - ones) + [1] * ones`. Cách này gọn nhất, chỉ dùng một biểu thức đếm.

````python
class SolutionBit:
    def transformArray(self, nums):
        ones = sum(x & 1 for x in nums)
        return [0] * (len(nums) - ones) + [1] * ones

print(SolutionBit().transformArray([4, 3, 2, 1]))
print(SolutionBit().transformArray([1, 5, 1, 4, 2]))
````

Độ phức tạp Cách 4: Time `O(n)`, Space `O(n)` cho mảng kết quả.

## 8. TRACE từng bước

Trace Cách 3 (hai con trỏ) với `nums = [1, 5, 1, 4, 2]` (`n = 5`):

| bước | `x` | chẵn/lẻ | `lo` | `hi` | `res` sau bước |
|---|---|---|---|---|---|
| bắt đầu | | | 0 | 4 | `[0,0,0,0,0]` |
| 1 | 1 | lẻ | 0 | 3 | `[0,0,0,0,1]` |
| 2 | 5 | lẻ | 0 | 2 | `[0,0,0,1,1]` |
| 3 | 1 | lẻ | 0 | 1 | `[0,0,1,1,1]` |
| 4 | 4 | chẵn | 1 | 1 | `[0,0,1,1,1]` |
| 5 | 2 | chẵn | 2 | 1 | `[0,0,1,1,1]` |

Kết thúc: `lo = 2` = số chẵn, `hi = 1`, `lo == hi + 1`. Kết quả `[0, 0, 1, 1, 1]`.

````python
def trace_parity(nums):
    n = len(nums)
    res = [0] * n
    lo, hi = 0, n - 1
    print(f"bắt đầu: lo={lo} hi={hi} res={res}")
    for step, x in enumerate(nums, start=1):
        if x % 2 == 0:
            res[lo] = 0
            lo += 1
            kind = "chẵn"
        else:
            res[hi] = 1
            hi -= 1
            kind = "lẻ"
        print(f"bước {step}: x={x} ({kind}) lo={lo} hi={hi} res={res}")
    print("lo == hi + 1 ?", lo == hi + 1)
    return res

trace_parity([1, 5, 1, 4, 2])
trace_parity([4, 3, 2, 1])
````

## 9. Code nộp LeetCode

````python
from typing import List

class Solution:
    def transformArray(self, nums: List[int]) -> List[int]:
        evens = 0
        for x in nums:
            if x % 2 == 0:
                evens += 1
        return [0] * evens + [1] * (len(nums) - evens)

print(Solution().transformArray([4, 3, 2, 1]), Solution().transformArray([1, 5, 1, 4, 2]))
````

## 10. Vì sao thuật toán đúng

- **Giá trị mỗi phần tử:** theo đề, phần tử chẵn thành `0` và phần tử lẻ thành `1`; sau đó mảng chỉ chứa `0` và `1`.
- **Sắp xếp không giảm = `0` trước `1`:** với hai giá trị `0 < 1`, mảng không giảm nghĩa là mọi `0` đứng trước mọi `1` (nếu có `1` đứng trước `0` thì cặp đó giảm). Vì các số bằng nhau không phân biệt được, **chỉ có một** mảng không giảm với đúng `e` số `0` và `o` số `1`, đó là `[0]*e + [1]*o`.
- **Số lượng:** số `0` bằng số phần tử chẵn `e`, số `1` bằng số phần tử lẻ `o = n - e` (mỗi số nguyên dương hoặc chẵn hoặc lẻ, không có trường hợp thứ ba).
- **Cách 3:** `lo` luôn bằng số chẵn đã thấy, `n - 1 - hi` luôn bằng số lẻ đã thấy; các ô được ghi chưa bao giờ chồng nhau vì `lo + (n - 1 - hi) <= n`.
- **Cách 4:** `x & 1` bằng `x % 2` với số nguyên dương nên cùng số `1` được đếm.

## 11. Phân tích Time / Space Complexity

Gọi `n = len(nums)`.

| Cách | Time | Space |
|---|---|---|
| 1. Ánh xạ + `sorted` | `O(n log n)` | `O(n)` |
| 2. Đếm số chẵn | `O(n)` | `O(n)` (mảng trả về) |
| 3. Hai con trỏ điền hai đầu | `O(n)` | `O(n)` |
| 4. Bit `x & 1` + `sum` | `O(n)` | `O(n)` |

Bộ nhớ thêm ngoài mảng kết quả của Cách 2, 3, 4 là `O(1)`. Với `n <= 100` mọi cách chạy tức thì; sự khác biệt chỉ có ý nghĩa về cách **nghĩ**: đếm thay vì sort khi giá trị chỉ có vài loại.

## 12. Bộ test

Test 1: ví dụ đề và các edge case (`n = 1`, toàn chẵn, toàn lẻ, giá trị biên `1` và `1000`, đã xen kẽ, đã sắp xếp).

````python
impls = [SolutionSort(), SolutionCount(), SolutionTwoPointers(), SolutionBit(), Solution()]
tests = [
    ([4, 3, 2, 1], [0, 0, 1, 1]),
    ([1, 5, 1, 4, 2], [0, 0, 1, 1, 1]),
    ([1], [1]),                                   # n = 1, lẻ, giá trị nhỏ nhất
    ([2], [0]),                                   # n = 1, chẵn
    ([1000], [0]),                                # biên trên, chẵn
    ([999], [1]),                                 # biên trên, lẻ
    ([2, 4, 6, 8], [0, 0, 0, 0]),                 # toàn chẵn
    ([1, 3, 5, 7, 9], [1, 1, 1, 1, 1]),           # toàn lẻ
    ([1, 2, 3, 4, 5, 6], [0, 0, 0, 1, 1, 1]),     # xen kẽ
    ([6, 5, 4, 3, 2, 1], [0, 0, 0, 1, 1, 1]),     # xen kẽ ngược
    ([1, 1, 2, 2], [0, 0, 1, 1]),                 # lặp giá trị
    ([1000, 999], [0, 1]),                        # hai biên
]
for nums, exp in tests:
    assert 1 <= len(nums) <= 100 and all(1 <= v <= 1000 for v in nums)
    for im in impls:
        got = im.transformArray(list(nums))
        assert got == exp, (type(im).__name__, nums, got, exp)
print("tất cả test cố định OK:", len(tests))
````

Test 2: ngẫu nhiên có seed, đúng constraints (`1 <= n <= 100`, `1 <= nums[i] <= 1000`). Oracle độc lập không dùng `sorted` của mảng ánh xạ mà kiểm tra **tính chất**: độ dài, chỉ gồm `0/1`, không giảm, số `1` bằng số phần tử lẻ.

````python
import random
random.seed(3467)

def check_properties(nums, out):
    assert len(out) == len(nums)
    assert set(out) <= {0, 1}
    assert all(out[i] <= out[i + 1] for i in range(len(out) - 1))
    assert sum(out) == len([x for x in nums if x % 2 == 1])

for t in range(1500):
    n = random.randint(1, 100)
    hi_val = random.choice([2, 10, 1000])
    nums = [random.randint(1, hi_val) for _ in range(n)]
    assert 1 <= len(nums) <= 100 and all(1 <= v <= 1000 for v in nums)
    ref = impls[0].transformArray(list(nums))
    check_properties(nums, ref)
    for im in impls:
        assert im.transformArray(list(nums)) == ref, (type(im).__name__, nums)
print("1500 test ngẫu nhiên OK")
````

Test 3: tính chất — không sửa input; kết quả **không phụ thuộc thứ tự** của `nums` (xáo trộn không đổi); cộng thêm `2` vào mọi phần tử (giữ chẵn/lẻ, nhưng vẫn ≤ 1000 nếu ta chặn lại) cho cùng kết quả.

````python
random.seed(67)
for _ in range(500):
    n = random.randint(1, 100)
    nums = [random.randint(1, 998) for _ in range(n)]
    snapshot = list(nums)
    base = Solution().transformArray(nums)
    assert nums == snapshot
    shuffled = list(nums)
    random.shuffle(shuffled)
    assert Solution().transformArray(shuffled) == base
    shifted = [x + 2 for x in nums]
    assert all(1 <= v <= 1000 for v in shifted)
    assert Solution().transformArray(shifted) == base
print("tính chất (không sửa input, xáo trộn, +2): OK")
````

## 13. Các lỗi người mới hay gặp

### Lỗi 1: sắp xếp `nums` **trước** khi ánh xạ

````python
def wrong_sort_first(nums):
    nums = sorted(nums)                       # SAI: sort số gốc rồi mới đổi 0/1
    return [x % 2 for x in nums]

nums = [4, 3, 2, 1]
print("sai :", wrong_sort_first(nums))
print("đúng:", Solution().transformArray(nums))
assert wrong_sort_first(nums) != Solution().transformArray(nums)
````

Vì sao sai: sort `[4,3,2,1]` thành `[1,2,3,4]` rồi đổi ra `[1,0,1,0]`, **không** có dạng `0` trước `1`. Đề yêu cầu thứ tự: đổi **trước**, sort **sau**. **Sửa:** ánh xạ xong mới `sorted`, hoặc dùng cách đếm.

### Lỗi 2: quên ánh xạ, chỉ sắp xếp

````python
def wrong_only_sort(nums):
    return sorted(nums)                       # SAI: chưa thay chẵn/lẻ bằng 0/1

nums = [1, 5, 1, 4, 2]
print("sai :", wrong_only_sort(nums))
print("đúng:", Solution().transformArray(nums))
assert wrong_only_sort(nums) != Solution().transformArray(nums)
````

Vì sao sai: kết quả `[1,1,2,4,5]` vẫn chứa các giá trị gốc, trong khi đề chỉ cho phép `0` và `1`. **Sửa:** thay mỗi `x` bằng `x % 2` trước.

### Lỗi 3: đảo vai trò `0` và `1` (nhầm chẵn = `1`)

````python
def wrong_swapped(nums):
    mapped = [1 if x % 2 == 0 else 0 for x in nums]   # SAI: chẵn phải là 0, lẻ phải là 1
    return sorted(mapped)

nums = [1, 5, 1, 4, 2]
print("sai :", wrong_swapped(nums))
print("đúng:", Solution().transformArray(nums))
assert wrong_swapped(nums) != Solution().transformArray(nums)
````

Vì sao sai: đề nói chẵn -> `0`, lẻ -> `1`. Nhầm ngược sẽ ra `[0,0,0,1,1]` thay vì `[0,0,1,1,1]` (ba số lẻ nhưng chỉ ra hai số `1`). **Sửa:** dùng chính `x % 2` (chẵn cho `0`, lẻ cho `1`).

### Lỗi 4: sắp xếp giảm dần (`reverse=True`)

````python
def wrong_descending(nums):
    mapped = [x % 2 for x in nums]
    return sorted(mapped, reverse=True)       # SAI: đề yêu cầu không giảm (tăng dần)

nums = [4, 3, 2, 1]
print("sai :", wrong_descending(nums))
print("đúng:", Solution().transformArray(nums))
assert wrong_descending(nums) != Solution().transformArray(nums)
````

Vì sao sai: `[1,1,0,0]` là thứ tự **giảm**. "Không giảm" = từ nhỏ đến lớn, tức `0` đứng trước `1`. **Sửa:** bỏ `reverse=True`, hoặc dựng `[0] * e + [1] * o` (số `0` trước).

## 14. Mẹo làm bài nhanh & nhận dạng pattern

- Bài "đổi giá trị rồi sắp xếp" mà sau khi đổi chỉ còn **vài giá trị khác nhau** (ở đây 2) -> thay sort bằng **đếm** (counting sort): `O(n)`.
- Chẵn/lẻ -> `x % 2` hoặc `x & 1`; chính phần dư cũng là giá trị cần gán, khỏi `if/else`.
- Luôn đọc **thứ tự thao tác** trong đề: "đổi, rồi sort" khác "sort, rồi đổi" (Lỗi 1).
- Dựng kết quả bằng `[0] * a + [1] * b`; cẩn thận `a + b = n`.
- Đừng nhầm với bài `905. Sort Array By Parity`: ở bài đó các số **giữ nguyên giá trị** và chỉ đưa số chẵn lên trước; ở bài này giá trị bị **đổi** thành `0/1`.
- Kiểm tra biên: `n = 1`, toàn chẵn, toàn lẻ.

## 15. Biến thể của bài

1. **Giữ nguyên giá trị** (`905. Sort Array By Parity`): chỉ đưa số chẵn lên trước, số lẻ về sau (hai con trỏ in-place).
2. **Ba lớp** (`x % 3`): đổi theo dư khi chia `3` rồi sắp xếp -> đếm ba loại (Dutch national flag, `75. Sort Colors`).
3. **Sắp xếp giảm dần** hoặc để số lẻ trước số chẵn.
4. Yêu cầu **sửa tại chỗ** (`O(1)` bộ nhớ thêm): ghi `0` vào `nums[0..e)` và `1` vào `nums[e..n)` sau khi đếm.
5. Giá trị có thể bằng `0` hoặc **âm**: trong Python `-3 % 2 == 1` (đúng), nhưng ở C++/Java `-3 % 2 == -1`, nên dùng `x & 1` hoặc `x % 2 != 0` để kiểm tra lẻ.

## 16. So sánh các cách làm

| Cách | Ý tưởng | Time | Space | Khi nào dùng |
|---|---|---|---|---|
| 1. Ánh xạ + `sorted` | làm nguyên văn đề | `O(n log n)` | `O(n)` | viết nhanh, oracle |
| 2. Đếm số chẵn | `[0]*e + [1]*(n-e)` | `O(n)` | `O(n)` | **chuẩn, nên nộp** |
| 3. Hai con trỏ hai đầu | điền `0` từ trái, `1` từ phải | `O(n)` | `O(n)` | học kỹ thuật partition |
| 4. Bit `x & 1` + `sum` | đếm số lẻ bằng bit | `O(n)` | `O(n)` | gọn, một dòng |

## 18. Gợi ý 3 mức cho bài tập

## 19. Đáp án tham khảo cho bài tập

````python
# Bài 1: chẵn 10,8,6 -> 3 số 0 ; lẻ 7,3 -> 2 số 1
assert Solution().transformArray([10, 7, 8, 3, 6]) == [0, 0, 0, 1, 1]

# Bài 2: chẵn 2,2,4 ; lẻ 9,7,1
assert Solution().transformArray([9, 2, 2, 7, 4, 1]) == [0, 0, 0, 1, 1, 1]

# Bài 3
def count_evens(nums):
    c = 0
    for x in nums:
        if x % 2 == 0:
            c += 1
    return c
assert count_evens([9, 2, 2, 7, 4, 1]) == 3
assert count_evens([1, 3]) == 0
assert count_evens([2]) == 1

# Bài 4
def transform_in_place(nums):
    e = count_evens(nums)
    for i in range(len(nums)):
        nums[i] = 0 if i < e else 1
    return nums
arr = [4, 3, 2, 1]
assert transform_in_place(arr) == [0, 0, 1, 1] and arr == [0, 0, 1, 1]
assert transform_in_place([1, 5, 1, 4, 2]) == [0, 0, 1, 1, 1]

# Bài 5
def transform_mod3(nums):
    cnt = [0, 0, 0]
    for x in nums:
        cnt[x % 3] += 1
    return [0] * cnt[0] + [1] * cnt[1] + [2] * cnt[2]
assert transform_mod3([4, 3, 2, 1]) == sorted(x % 3 for x in [4, 3, 2, 1])
assert transform_mod3([5, 6, 7, 8, 9]) == [0, 0, 1, 2, 2]
print("đáp án bài tập OK")
````
