## 1. Mục tiêu học

Sau notebook này bạn sẽ:

1. Hiểu đề: mỗi chuỗi có một **giá trị** — nếu chỉ gồm chữ số thì là số nguyên mà nó biểu diễn (bỏ số `0` đứng đầu), ngược lại là **độ dài** của chuỗi; cần trả về giá trị **lớn nhất** trong mảng.
2. Phân biệt một chuỗi "toàn chữ số" bằng `str.isdigit()`, kiểm tra từng ký tự, hoặc bắt ngoại lệ `ValueError` của `int()`.
3. Nắm hành vi của `int("00012") == 12` và `len("00012") == 5`, và vì sao `"01"` có giá trị `1` chứ không phải `2`.
4. Viết 5 cách (vòng lặp thủ công, `isdigit`, `try/except`, `max` với generator, và sắp xếp theo hàm giá trị); đối chiếu bằng test ngẫu nhiên.

## 2. Đề bài & ví dụ

**Giá trị** của một chuỗi chữ-số `s` (gồm chữ cái thường và chữ số) được định nghĩa:

- nếu `s` **chỉ gồm chữ số** → giá trị là số nguyên cơ số 10 của nó (ví dụ `"00123"` → `123`);
- ngược lại (có ít nhất một chữ cái) → giá trị là **độ dài** `len(s)`.

Cho mảng `strs`, trả về **giá trị lớn nhất** của các chuỗi trong mảng.

| Ví dụ | strs | Output | Giải thích |
|---|---|---|---|
| 1 | `["alic3","bob","3","4","00000"]` | 5 | `"alic3"` → 5, `"bob"` → 3, `"3"` → 3, `"4"` → 4, `"00000"` → 0 |
| 2 | `["1","01","001","0001"]` | 1 | cả bốn chuỗi đều có giá trị `1` |

Constraints: `1 <= strs.length <= 100`; `1 <= strs[i].length <= 9`; `strs[i]` chỉ gồm chữ cái thường tiếng Anh và chữ số.

````python
strs = ["alic3", "bob", "3", "4", "00000"]
for s in strs:
    if s.isdigit():
        print(f"{s!r}: toàn chữ số -> int = {int(s)}")
    else:
        print(f"{s!r}: có chữ cái -> len = {len(s)}")
````

## 3. Kiến thức nền

### 3.1. `str.isdigit()` — chuỗi toàn chữ số?

`s.isdigit()` trả `True` khi **mọi** ký tự của `s` là chữ số và `s` không rỗng. Với chuỗi chứa chữ cái thường và chữ số (đúng như đề) nó đủ chính xác. (Với ký tự Unicode lạ như `"²"` thì `isdigit()` cũng trả `True` nhưng `int()` không chuyển được; điều này không xảy ra với ràng buộc của đề.)

````python
for s in ["123", "007", "a1", "abc", "9", ""]:
    print(repr(s), s.isdigit())
````

### 3.2. `int(s)` và số `0` đứng đầu

`int("00012")` cho `12`: các số `0` ở đầu bị bỏ qua. Nhưng `len("00012")` vẫn là `5`. Hai khái niệm "giá trị số" và "độ dài chuỗi" tách biệt hoàn toàn. Nếu `s` có chữ cái, `int(s)` ném lỗi `ValueError`.

````python
print(int("00012"), len("00012"))
print(int("000"), int("0"))
try:
    int("alic3")
except ValueError as e:
    print("ValueError:", e)
````

### 3.3. `try / except ValueError` — "xin lỗi sau" thay vì "hỏi trước"

Thay vì kiểm tra trước (`isdigit`), ta có thể thử chuyển rồi bắt lỗi. Cách này hay dùng trong Python (EAFP). Lưu ý `int()` còn chấp nhận khoảng trắng, dấu `+/-` và dấu `_` (`int("1_0") == 10`), nhưng đề chỉ có chữ thường và chữ số nên không gặp.

````python
def value_try(s):
    try:
        return int(s)
    except ValueError:
        return len(s)

print([value_try(s) for s in ["alic3", "bob", "3", "4", "00000"]])
````

### 3.4. `max` với generator expression và `key`

`max(iterable)` trả phần tử lớn nhất. Truyền một **generator** `max(f(x) for x in arr)` để tính giá trị cho mỗi phần tử rồi lấy lớn nhất mà không tạo list trung gian. `max(arr, key=f)` trả **phần tử** có `f` lớn nhất (không phải giá trị `f`).

````python
words = ["alic3", "bob", "3", "4", "00000"]
f = lambda s: int(s) if s.isdigit() else len(s)
print(max(f(s) for s in words))          # giá trị lớn nhất
print(max(words, key=f))                 # chuỗi có giá trị lớn nhất
````

### 3.5. Kiểm tra từng ký tự bằng `all`

`all(c in "0123456789" for c in s)` là cách tự viết của `isdigit()`; hữu ích khi bạn muốn hiểu hoặc kiểm soát chính xác "thế nào là chữ số". Cũng có thể so sánh `"0" <= c <= "9"`.

````python
digits = "0123456789"
for s in ["alic3", "00000", "4"]:
    print(s, all(c in digits for c in s), all("0" <= c <= "9" for c in s))
````

### 3.6. Thứ tự từ điển của chuỗi khác thứ tự số

`"bob" > "4"` là `True` vì so sánh ký tự `'b'` (mã 98) với `'4'` (mã 52). `max(strs)` chọn chuỗi lớn nhất theo **từ điển**, không phải theo "giá trị" theo đề. Đây là nguồn của một lỗi phổ biến (xem mục 13).

````python
print("bob" > "4", "9" > "10", 9 > 10)
print(max(["alic3", "bob", "3", "4", "00000"]))
````

## 4. Ý tưởng nền tảng / cách nghĩ trực giác

Chỉ cần làm đúng như đề: với mỗi chuỗi, quyết định nó thuộc loại nào rồi tính giá trị; giữ lại giá trị lớn nhất.

Ví dụ 1 tính tay:

```
"alic3"  : có chữ 'a','l','i','c' -> độ dài 5
"bob"    : có chữ cái            -> độ dài 3
"3"      : toàn chữ số           -> 3
"4"      : toàn chữ số           -> 4
"00000"  : toàn chữ số           -> int = 0   (không phải 5!)
max(5, 3, 3, 4, 0) = 5
```

Ví dụ 2: `"1"`, `"01"`, `"001"`, `"0001"` đều toàn chữ số, `int` đều bằng `1` → đáp án `1`, **không** phải `4`.

Nhận xét: mỗi chuỗi độc lập với nhau; bài toán chỉ là "ánh xạ rồi lấy lớn nhất" (map → max).

## 5. Cách 1 — Brute Force / baseline

Duyệt từng chuỗi, **tự kiểm tra từng ký tự** có phải chữ số không (cờ `all_digits`); nếu có chữ cái dùng `len`, nếu toàn chữ số thì tính số thủ công `value = value * 10 + digit` (không dùng `int`). Cách này chỉ dùng phép lặp cơ bản và hoàn toàn khớp với định nghĩa của đề.

Nút thắt: chẳng có nút thắt lớn: mỗi chuỗi tốn `O(L)` với `L <= 9`. Nhưng code dài, dễ nhầm ở chỗ khởi tạo `value` và tính số.

````python
class SolutionBrute:
    def maximumValue(self, strs):
        best = 0
        for s in strs:
            all_digits = True
            value = 0
            for ch in s:
                if "0" <= ch <= "9":
                    value = value * 10 + (ord(ch) - ord("0"))   # tự dựng số
                else:
                    all_digits = False
            current = value if all_digits else len(s)
            if current > best:
                best = current
        return best

print(SolutionBrute().maximumValue(["alic3", "bob", "3", "4", "00000"]))
print(SolutionBrute().maximumValue(["1", "01", "001", "0001"]))
````

**Độ phức tạp:** Time `O(n * L)` với `n` số chuỗi, `L` độ dài tối đa (`<= 9`); Space `O(1)`.

## 6. Ý tưởng thuật toán / bước đột phá

Tách bài toán thành hai bước nhỏ:

1. Viết một **hàm giá trị** `value(s)`: `int(s)` nếu `s.isdigit()`, ngược lại `len(s)`.
2. Đáp án là `max(value(s) for s in strs)`.

**Quy tắc "nếu… thì…":**

- Nếu `s.isdigit()` → giá trị = `int(s)` (tự bỏ `0` đầu).
- Nếu không (có chữ cái) → giá trị = `len(s)`.
- Giá trị luôn `>= 0`; vì `strs` không rỗng và mỗi chuỗi dài `>= 1`, đáp án là giá trị của một chuỗi cụ thể.

Invariant khi duyệt: `best` luôn bằng giá trị lớn nhất của các chuỗi đã duyệt.

````python
def value(s):
    return int(s) if s.isdigit() else len(s)

strs = ["alic3", "bob", "3", "4", "00000"]
print([value(s) for s in strs], "->", max(value(s) for s in strs))
````

## 7. Các cách tiếp theo

Ta có 5 cách: (1) kiểm tra từng ký tự, (2) `isdigit` + vòng lặp, (3) `try/except ValueError`, (4) `max` với generator một dòng, (5) sắp xếp theo hàm giá trị rồi lấy phần tử cuối.

### Cách 2 — `isdigit` với vòng lặp

Dùng hàm dựng sẵn `isdigit` thay cho kiểm tra thủ công; cập nhật `best` khi gặp giá trị lớn hơn.

````python
class SolutionIsdigit:
    def maximumValue(self, strs):
        best = 0
        for s in strs:
            v = int(s) if s.isdigit() else len(s)
            best = max(best, v)
        return best

print(SolutionIsdigit().maximumValue(["alic3", "bob", "3", "4", "00000"]))
print(SolutionIsdigit().maximumValue(["1", "01", "001", "0001"]))
````

Độ phức tạp Cách 2: Time `O(n * L)`, Space `O(1)`.

### Cách 3 — `try / except ValueError`

Thử `int(s)`; nếu thất bại (có chữ cái) thì dùng `len(s)`. Ngắn, đúng với ràng buộc của đề. Nhược điểm: ngoại lệ chậm hơn một chút nếu rất nhiều chuỗi có chữ.

````python
class SolutionTry:
    def maximumValue(self, strs):
        best = 0
        for s in strs:
            try:
                v = int(s)
            except ValueError:
                v = len(s)
            best = max(best, v)
        return best

print(SolutionTry().maximumValue(["alic3", "bob", "3", "4", "00000"]))
print(SolutionTry().maximumValue(["1", "01", "001", "0001"]))
````

Độ phức tạp Cách 3: Time `O(n * L)`, Space `O(1)`.

### Cách 4 — `max` với generator (một dòng)

Biểu thức `max(int(s) if s.isdigit() else len(s) for s in strs)` — phong cách Python gọn nhất, dễ đọc khi đã quen.

````python
class SolutionOneLine:
    def maximumValue(self, strs):
        return max(int(s) if s.isdigit() else len(s) for s in strs)

print(SolutionOneLine().maximumValue(["alic3", "bob", "3", "4", "00000"]))
print(SolutionOneLine().maximumValue(["1", "01", "001", "0001"]))
````

Độ phức tạp Cách 4: Time `O(n * L)`, Space `O(1)` (generator không tạo list).

### Cách 5 — Sắp xếp theo hàm giá trị

Sắp xếp `strs` theo khoá `value(s)` rồi lấy phần tử cuối, tính lại giá trị của nó. Không tối ưu (tốn `O(n log n)` cho việc chỉ cần tìm max), nhưng cho thấy `key=` hoạt động ra sao và dễ mở rộng sang "trả về chuỗi có giá trị lớn nhất".

````python
class SolutionSort:
    def maximumValue(self, strs):
        def value(s):
            return int(s) if s.isdigit() else len(s)
        ordered = sorted(strs, key=value)
        return value(ordered[-1])

print(SolutionSort().maximumValue(["alic3", "bob", "3", "4", "00000"]))
print(SolutionSort().maximumValue(["1", "01", "001", "0001"]))
````

Độ phức tạp Cách 5: Time `O(n log n * L)` (so sánh khoá), Space `O(n)`.

## 8. TRACE từng bước

Trace Cách 2 với `["alic3", "bob", "3", "4", "00000"]`:

| bước | `s` | `isdigit` | giá trị `v` | `best` sau bước |
|---|---|---|---|---|
| 1 | `"alic3"` | `False` | `len = 5` | 5 |
| 2 | `"bob"` | `False` | `len = 3` | 5 |
| 3 | `"3"` | `True` | `int = 3` | 5 |
| 4 | `"4"` | `True` | `int = 4` | 5 |
| 5 | `"00000"` | `True` | `int = 0` | 5 |

Kết quả `5`.

````python
def trace_max_value(strs):
    best = 0
    for step, s in enumerate(strs, start=1):
        if s.isdigit():
            v, why = int(s), f"toàn chữ số -> int={int(s)}"
        else:
            v, why = len(s), f"có chữ cái -> len={len(s)}"
        best = max(best, v)
        print(f"bước {step}: s={s!r:10} {why:28} best={best}")
    return best

print("kết quả:", trace_max_value(["alic3", "bob", "3", "4", "00000"]))
print("kết quả:", trace_max_value(["1", "01", "001", "0001"]))
````

## 9. Code nộp LeetCode

````python
from typing import List

class Solution:
    def maximumValue(self, strs: List[str]) -> int:
        return max(int(s) if s.isdigit() else len(s) for s in strs)

print(Solution().maximumValue(["alic3", "bob", "3", "4", "00000"]), Solution().maximumValue(["1", "01", "001", "0001"]))
````

## 10. Vì sao thuật toán đúng

- **Mỗi chuỗi có đúng một giá trị:** hoặc toàn chữ số (rẽ nhánh `isdigit()` đúng), hoặc có ít nhất một chữ cái (nhánh `len`). Hai trường hợp loại trừ lẫn nhau và bao phủ tất cả chuỗi hợp lệ (chuỗi không rỗng).
- **`int(s)` đúng giá trị:** chuỗi chữ số chuyển thành số nguyên cơ số 10, bỏ `0` đứng đầu — khớp chính xác định nghĩa "biểu diễn cơ số 10" của đề.
- **`max` trên các giá trị:** đáp án là giá trị lớn nhất trong tất cả; hàm `max` duyệt mọi phần tử nên không bỏ sót. Invariant: sau khi xử lý `k` chuỗi, `best` bằng giá trị lớn nhất của `k` chuỗi đầu.
- **Khởi tạo `best = 0` an toàn:** mọi giá trị `>= 0`, nên `0` không bao giờ lớn hơn đáp án thực.
- **`try/except`:** vì chuỗi chỉ gồm chữ thường và chữ số, `int(s)` chỉ thất bại khi có chữ cái, đúng trường hợp cần `len`.

## 11. Phân tích Time / Space Complexity

Gọi `n = len(strs)`, `L` = độ dài chuỗi lớn nhất (`<= 9`).

| Cách | Time | Space |
|---|---|---|
| 1. Kiểm tra từng ký tự thủ công | `O(n * L)` | `O(1)` |
| 2. `isdigit` + vòng lặp | `O(n * L)` | `O(1)` |
| 3. `try/except` | `O(n * L)` | `O(1)` |
| 4. `max` + generator | `O(n * L)` | `O(1)` |
| 5. Sắp xếp theo khoá | `O(n log n * L)` | `O(n)` |

Mọi cách phải đọc từng ký tự ít nhất một lần nên `O(n * L)` là tối ưu; với `n <= 100`, `L <= 9` thì tức thì.

## 12. Bộ test

Test 1: ví dụ đề và edge case (một chuỗi, chuỗi `"0"`, số `0` đứng đầu, chuỗi toàn chữ cái dài, chuỗi chữ số dài 9 ký tự).

````python
impls = [SolutionBrute(), SolutionIsdigit(), SolutionTry(), SolutionOneLine(), SolutionSort(), Solution()]
tests = [
    (["alic3", "bob", "3", "4", "00000"], 5),
    (["1", "01", "001", "0001"], 1),
    (["0"], 0),                              # giá trị 0, chuỗi dài 1 (bắt lỗi khởi tạo best = 1)
    (["00", "0", "000"], 0),                 # toàn số 0
    (["a"], 1),                              # một chữ cái
    (["abcdefghi"], 9),                      # dài 9, toàn chữ cái
    (["999999999"], 999999999),              # chữ số dài 9: giá trị lớn
    (["999999999", "abcdefghi"], 999999999),
    (["a1"], 2),                             # lẫn chữ và số -> len
    (["123", "abcd"], 123),                  # số lớn hơn độ dài
    (["12", "abcdef"], 12),                  # giá trị số lớn hơn len(chuỗi chữ) là 6
    (["4", "abcdef"], 6),                    # độ dài lớn hơn giá trị số
    (["bob", "alic3"], 5),                   # max theo từ điển là "bob" (bắt lỗi max(strs))
    (["007"], 7),                            # số 0 đầu bị bỏ
]
for strs, exp in tests:
    assert 1 <= len(strs) <= 100 and all(1 <= len(s) <= 9 for s in strs)
    assert all(c.isdigit() or ("a" <= c <= "z") for s in strs for c in s)
    for im in impls:
        assert im.maximumValue(list(strs)) == exp, (type(im).__name__, strs, exp)
print("tất cả test cố định OK:", len(tests))
````

Test 2: ngẫu nhiên có seed, đúng constraints (`1..100` chuỗi, độ dài `1..9`, ký tự chữ thường/chữ số); oracle độc lập dùng `re.fullmatch` thay cho `isdigit`.

````python
import random, re, string
random.seed(2496)

def oracle(strs):
    best = -1
    for s in strs:
        v = int(s) if re.fullmatch(r"[0-9]+", s) else len(s)
        best = max(best, v)
    return best

for t in range(1500):
    n = random.randint(1, 100) if t % 10 == 0 else random.randint(1, 8)
    alphabet = random.choice([string.digits, "0" + "1", string.ascii_lowercase, string.ascii_lowercase + string.digits])
    strs = ["".join(random.choice(alphabet) for _ in range(random.randint(1, 9))) for _ in range(n)]
    assert 1 <= len(strs) <= 100 and all(1 <= len(s) <= 9 for s in strs)
    exp = oracle(strs)
    for im in impls:
        assert im.maximumValue(list(strs)) == exp, (type(im).__name__, strs)
print("1500 test ngẫu nhiên OK")
````

Test 3: tính chất — đáp án `>= 0`; không phụ thuộc thứ tự mảng; không sửa mảng; thêm một chuỗi toàn chữ cái dài 9 thì đáp án `>= 9`.

````python
random.seed(5)
for _ in range(300):
    n = random.randint(1, 20)
    strs = ["".join(random.choice(string.ascii_lowercase + string.digits) for _ in range(random.randint(1, 9))) for _ in range(n)]
    base = Solution().maximumValue(strs)
    assert base >= 0
    shuffled = list(strs)
    random.shuffle(shuffled)
    assert Solution().maximumValue(shuffled) == base
    snapshot = list(strs)
    Solution().maximumValue(strs)
    assert strs == snapshot
    assert Solution().maximumValue(strs + ["abcdefghi"]) >= 9
print("tính chất (không âm, bất biến theo hoán vị, không sửa input, thêm chuỗi chữ dài 9): OK")
````

## 13. Các lỗi người mới hay gặp

### Lỗi 1: luôn dùng độ dài (quên ý "chữ số → giá trị số")

````python
def wrong_always_len(strs):
    return max(len(s) for s in strs)        # SAI: chuỗi toàn chữ số phải dùng giá trị int

strs = ["1", "01", "001", "0001"]
print("sai :", wrong_always_len(strs))
print("đúng:", Solution().maximumValue(strs))
assert wrong_always_len(strs) != Solution().maximumValue(strs)
````

Vì sao sai: chuỗi toàn chữ số có giá trị là số nguyên (`"0001"` → `1`), **không** phải độ dài `4`. **Sửa:** rẽ nhánh `isdigit()`.

### Lỗi 2: so sánh chuỗi theo thứ tự từ điển (`max(strs)`)

````python
def wrong_lexicographic(strs):
    return max(strs)                         # SAI: lấy chuỗi lớn nhất theo từ điển, không phải giá trị lớn nhất

strs = ["alic3", "bob", "3", "4", "00000"]
print("sai :", wrong_lexicographic(strs))
print("đúng:", Solution().maximumValue(strs))
assert wrong_lexicographic(strs) != Solution().maximumValue(strs)
````

Vì sao sai: `max` trên chuỗi so sánh theo mã ký tự, ra chuỗi `"bob"` (cũng là một **chuỗi**, không phải số). Đề cần **giá trị** đã tính. **Sửa:** tính giá trị của từng chuỗi **rồi** lấy `max` của các giá trị.

### Lỗi 3: khởi tạo đáp án quá lớn

````python
def wrong_init(strs):
    best = 1                                  # SAI: giá trị nhỏ nhất có thể là 0 (chuỗi "0")
    for s in strs:
        best = max(best, int(s) if s.isdigit() else len(s))
    return best

strs = ["0"]
print("sai :", wrong_init(strs))
print("đúng:", Solution().maximumValue(strs))
assert wrong_init(strs) != Solution().maximumValue(strs)
````

Vì sao sai: giá trị có thể bằng `0` (với `"0"`, `"000"`), nhưng `best = 1` làm đáp án luôn `>= 1`. **Sửa:** khởi tạo `best = 0` (hoặc `-1`), hoặc dùng `max(generator)` không cần khởi tạo.

### Lỗi 4: phân loại bằng `isalpha()` rồi gọi `int` lên chuỗi lẫn chữ và số

````python
def wrong_isalpha(strs):
    best = 0
    for s in strs:
        v = len(s) if s.isalpha() else int(s)     # SAI: "alic3" không isalpha, nhưng cũng không phải số
        best = max(best, v)
    return best

try:
    wrong_isalpha(["alic3"])
    crashed = False
except ValueError:
    crashed = True
print("sai : ValueError?", crashed)
print("đúng:", Solution().maximumValue(["alic3"]))
assert crashed
````

Vì sao sai: chuỗi **lẫn** chữ cái và chữ số (như `"alic3"`) không phải "toàn chữ cái" nên rơi vào nhánh `int` và gây `ValueError`. Điều kiện đúng là "**toàn chữ số**" (`isdigit`) — mọi trường hợp còn lại dùng `len`.

## 14. Mẹo làm bài nhanh & nhận dạng pattern

- Bài "định nghĩa giá trị cho từng phần tử rồi lấy lớn nhất" → tách hàm `value(x)` và viết `max(value(x) for x in arr)`.
- Chuỗi "toàn chữ số" → `s.isdigit()`; mọi trường hợp còn lại là nhánh còn lại — tránh kiểm tra hai lần.
- Số `0` đứng đầu: `int()` bỏ chúng; `len()` thì không. Hai thao tác cho kết quả khác nhau.
- `max(strs)` luôn là so sánh từ điển; nếu cần so sánh theo giá trị phải đưa qua `key=` hoặc tính giá trị trước.
- Khởi tạo `best` bằng giá trị **nhỏ nhất có thể** (ở đây `0`), hoặc dùng `max` của generator để khỏi phải khởi tạo.
- Ràng buộc chữ thường + chữ số làm `isdigit` và `try int` an toàn; nếu có ký tự lạ (dấu, khoảng trắng) hãy kiểm tra thủ công.

## 15. Biến thể của bài

1. Trả về **chuỗi** có giá trị lớn nhất thay vì giá trị (dùng `max(strs, key=value)`).
2. Chuỗi có thể chứa ký tự hoa, dấu hoặc khoảng trắng — cần định nghĩa lại "toàn chữ số".
3. Trả về **chỉ số** đầu tiên đạt giá trị lớn nhất.
4. Đếm **bao nhiêu chuỗi** đạt giá trị lớn nhất.
5. Cơ số khác: chuỗi toàn chữ số hex (`int(s, 16)`) thì giá trị theo hệ 16.

## 16. So sánh các cách làm

| Cách | Ý tưởng | Time | Space | Khi nào dùng |
|---|---|---|---|---|
| 1. Thủ công từng ký tự | tự dựng số, cờ `all_digits` | `O(n * L)` | `O(1)` | học nguyên lý, ngôn ngữ không có `isdigit` |
| 2. `isdigit` + vòng lặp | rẽ nhánh rồi `max` | `O(n * L)` | `O(1)` | rõ ràng, dễ debug |
| 3. `try/except` | thử `int`, lỗi thì `len` | `O(n * L)` | `O(1)` | phong cách EAFP |
| 4. `max` + generator | một dòng | `O(n * L)` | `O(1)` | **gọn nhất**, nên dùng khi nộp |
| 5. Sắp xếp theo khoá | `sorted(key=value)` | `O(n log n * L)` | `O(n)` | khi cần cả thứ hạng |

## 18. Gợi ý 3 mức cho bài tập

**Mức 1 (nhẹ):** mỗi chuỗi có đúng một giá trị: toàn chữ số → `int`, còn lại → `len`.

**Mức 2 (vừa):** Bài 3: `return int(s) if s.isdigit() else len(s)`. Bài 4: `max(strs, key=value_of)` (trả phần tử **đầu tiên** đạt max khi có hoà). Bài 5: tính `m = max(value_of(s) for s in strs)` rồi đếm `value_of(s) == m`.

**Mức 3 (gần đáp án):** Bài 1: `"12"`→12, `"ab"`→2, `"007"`→7, `"abcdefg"`→7 → max `12`. Bài 2: `"x9"`→2, `"0420"`→420, `"z"`→1, `"99"`→99 → max `420`.

## 19. Đáp án tham khảo cho bài tập

````python
# Bài 1: 12, 2, 7, 7 -> 12
assert Solution().maximumValue(["12", "ab", "007", "abcdefg"]) == 12

# Bài 2: "x9"->2, "0420"->420, "z"->1, "99"->99
vals = [int(s) if s.isdigit() else len(s) for s in ["x9", "0420", "z", "99"]]
assert vals == [2, 420, 1, 99]
assert max(vals) == 420 == Solution().maximumValue(["x9", "0420", "z", "99"])

# Bài 3
def value_of(s):
    return int(s) if s.isdigit() else len(s)
assert value_of("00123") == 123
assert value_of("alic3") == 5
assert value_of("0") == 0

# Bài 4
def best_string(strs):
    return max(strs, key=value_of)
assert best_string(["alic3", "bob", "3", "4", "00000"]) == "alic3"
assert best_string(["1", "01", "001"]) == "1"          # hoà -> phần tử đầu tiên

# Bài 5
def count_max(strs):
    m = max(value_of(s) for s in strs)
    return sum(1 for s in strs if value_of(s) == m)
assert count_max(["1", "01", "001", "0001"]) == 4
assert count_max(["alic3", "bob", "3", "4", "00000"]) == 1
assert count_max(["ab", "5", "cd"]) == 1                # "5" có giá trị 5 > 2
print("đáp án bài tập OK")
````
