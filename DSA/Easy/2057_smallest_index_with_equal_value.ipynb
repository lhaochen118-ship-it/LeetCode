## 1. Đề bài yêu cầu gì?

Cho mảng số nguyên `nums` **đánh số từ 0**. Trả về **chỉ số nhỏ nhất** `i` sao cho `i mod 10 == nums[i]`; nếu không có, trả `-1`.

`x mod y` là **số dư** khi chia `x` cho `y`.

| Ví dụ | nums | Output | Giải thích |
|---|---|---|---|
| 1 | `[0,1,2]` | `0` | mọi i đều thỏa, nhỏ nhất là 0 |
| 2 | `[4,3,2,1]` | `2` | chỉ i = 2: `2 mod 10 = 2 == nums[2]` |
| 3 | `[1,2,3,4,5,6,7,8,9,0]` | `-1` | không i nào thỏa |

Constraints: `1 <= nums.length <= 100`, `0 <= nums[i] <= 9`.

Để ý: `nums[i]` luôn là **một chữ số 0..9**, và `i mod 10` cũng luôn là **một chữ số 0..9** — đó là lý do đề dùng `mod 10`: so sánh "chữ số hàng đơn vị của chỉ số" với giá trị.

````python
nums = [4, 3, 2, 1]
for i in range(len(nums)):
    print(f"i={i}: i % 10 = {i % 10}, nums[i] = {nums[i]}, bằng nhau? {i % 10 == nums[i]}")
````

### 2.6. `return` trong vòng lặp = dừng sớm

Khi gặp `return`, hàm kết thúc **ngay lập tức**, vòng lặp không chạy tiếp. Đây là chìa khóa của "tìm phần tử **đầu tiên**".

````python
def first_even(nums):
    for i, x in enumerate(nums):
        print("  đang xét i =", i)
        if x % 2 == 0:
            return i
    return -1

print("kết quả:", first_even([3, 5, 8, 10, 12]))
````

### 2.7. `=` (gán) khác `==` (so sánh)

````python
i, nums = 12, [0] * 12 + [2]
print(i % 10 == nums[i])          # (12 % 10) == 2 -> True
print((i % 10) == nums[i])        # giống hệt
try:
    compile("if i % 10 = nums[i]: pass", "<thử>", "exec")
except SyntaxError as e:
    print("SyntaxError:", e.msg)
````

## 3. Cách 1 — Trực giác: thu thập mọi chỉ số thỏa rồi lấy min

Người mới hay nghĩ theo đúng chữ trong đề: "tìm **tất cả** i thỏa điều kiện, rồi lấy **nhỏ nhất**".

````python
def smallest_equal_collect(nums):
    good = [i for i in range(len(nums)) if i % 10 == nums[i]]
    return min(good) if good else -1          # list rỗng thì min() báo lỗi -> phải kiểm tra

print(smallest_equal_collect([0, 1, 2]))                       # 0
print(smallest_equal_collect([4, 3, 2, 1]))                    # 2
print(smallest_equal_collect([1, 2, 3, 4, 5, 6, 7, 8, 9, 0]))  # -1
````

`if good` — một list **rỗng** được coi là `False`, list có phần tử là `True` ("truthiness"). Nếu bỏ kiểm tra, `min([])` gây `ValueError`.

````python
print(bool([]), bool([3]))
try:
    min([])
except ValueError as e:
    print("ValueError:", e)
````

## 4. Nút thắt và insight

Cách 1 đúng, `O(n)`, nhưng:

- Luôn duyệt **toàn bộ** mảng, kể cả khi đáp án ở ngay `i = 0`.
- Tạo thêm list `good` (`O(n)` bộ nhớ), rồi còn duyệt lần nữa để tìm `min`.

**Insight:** ta duyệt `i` theo thứ tự **tăng dần**. Chỉ số **đầu tiên** thỏa điều kiện mà ta gặp **chính là chỉ số nhỏ nhất** — không cần gom, không cần `min`. Trả về ngay khi gặp.

Đây đúng là gợi ý của đề: *"Starting with i=0, check the condition for each index. The first one you find to be true is the smallest index."*

## 5. Cách 2 (lời giải chính) — Duyệt từ trái, trả ngay khi gặp

Ba dòng cốt lõi:

```python
for i, x in enumerate(nums):
    if i % 10 == x:
        return i
return -1
```

`return -1` đặt **sau** vòng lặp (không thụt vào trong `for`): chỉ chạy khi duyệt hết mà không tìm thấy.

````python
def smallest_equal(nums):
    for i, x in enumerate(nums):
        if i % 10 == x:
            return i
    return -1

print(smallest_equal([0, 1, 2]))
print(smallest_equal([4, 3, 2, 1]))
print(smallest_equal([1, 2, 3, 4, 5, 6, 7, 8, 9, 0]))
````

## 6. TRACE từng bước

Ví dụ 2: `nums = [4,3,2,1]`.

| i | nums[i] | i % 10 | bằng? | hành động |
|---|---|---|---|---|
| 0 | 4 | 0 | ✗ | tiếp |
| 1 | 3 | 1 | ✗ | tiếp |
| 2 | 2 | 2 | ✓ | **return 2** |
| 3 | 1 | 3 | – | không bao giờ xét |

Ví dụ với mảng dài hơn 10 để thấy chu kỳ: ta cần một mảng mà **12 chỉ số đầu đều không thỏa**, rồi chỉ số 12 thỏa (`12 % 10 = 2 == nums[12]`).

Cẩn thận khi tự tạo dữ liệu test! Một mảng như `[9] * 12 + [2]` (12 số 9 rồi số 2) trông có vẻ ổn, nhưng `nums[9] = 9` và `9 % 10 = 9` → đáp án lại là **9**, không phải 12. Đây là lỗi tạo test rất dễ mắc. Cách an toàn: đặt `nums[i] = (i + 1) % 10` — luôn lệch 1 so với `i % 10` nên không bao giờ thỏa.

````python
def trace_smallest_equal(nums):
    for i, x in enumerate(nums):
        ok = i % 10 == x
        print(f"i={i:>2}, nums[i]={x}, i%10={i % 10}, bằng? {ok}")
        if ok:
            print("-> return", i)
            return i
    print("-> duyệt hết, return -1")
    return -1

trace_smallest_equal([4, 3, 2, 1])
````

````python
LONG = [(i + 1) % 10 for i in range(12)] + [2]
print("LONG =", LONG)
print("Mảng 'trông có vẻ ổn' [9]*12+[2] cho đáp án:", smallest_equal([9] * 12 + [2]), "(do nums[9] = 9)")
trace_smallest_equal(LONG)
````

`[(i + 1) % 10 for i in range(12)]` tạo `[1, 2, ..., 9, 0, 1, 2]` (12 phần tử); `+ [2]` nối thêm phần tử 2 ở cuối → mảng 13 phần tử. `[9] * 12` là phép nhân list: lặp list `[9]` 12 lần.

## 7. Cách 3 — `next()` với generator và giá trị mặc định

`next(iterator, default)` lấy **phần tử đầu tiên** của iterator; nếu iterator rỗng thì trả `default`. Kết hợp với generator expression `(i for ... if ...)`, ta có "tìm đầu tiên, dừng sớm" trong một dòng — generator **lười** (lazy), chỉ tính đến khi tìm thấy.

````python
gen = (x * x for x in range(5))
print(next(gen), next(gen), next(gen))   # 0 1 4 — mỗi lần next lấy tiếp một phần tử
print(next((i for i in range(3) if i > 10), "không có"))
````

````python
def smallest_equal_next(nums):
    return next((i for i, x in enumerate(nums) if i % 10 == x), -1)

print(smallest_equal_next([0, 1, 2]))
print(smallest_equal_next([4, 3, 2, 1]))
print(smallest_equal_next([1, 2, 3, 4, 5, 6, 7, 8, 9, 0]))
````

## 8. Cách 4 — Nhìn theo "chữ số hàng đơn vị" (minh họa tư duy theo nhóm dư)

Một góc nhìn khác: chia các chỉ số thành 10 nhóm theo `i % 10 = r`. Trong nhóm `r`, ta cần `nums[i] == r`. Với mỗi `r`, tìm chỉ số nhỏ nhất trong nhóm thỏa, rồi lấy `min` giữa 10 nhóm.

Cách này **không nhanh hơn** (vẫn `O(n)`), nhưng luyện kỹ năng duyệt có bước nhảy `range(start, stop, step)` — rất hay dùng ở bài chu kỳ.

````python
print(list(range(2, 25, 10)))    # các chỉ số có i % 10 == 2: [2, 12, 22]

def smallest_equal_by_residue(nums):
    best = -1
    for r in range(10):
        for i in range(r, len(nums), 10):     # mọi i có i % 10 == r, tăng dần
            if nums[i] == r:
                if best == -1 or i < best:
                    best = i
                break                         # trong nhóm r, chỉ số đầu tiên là nhỏ nhất
    return best

print(smallest_equal_by_residue([4, 3, 2, 1]))
print(smallest_equal_by_residue(LONG))
print(smallest_equal_by_residue([1, 2, 3, 4, 5, 6, 7, 8, 9, 0]))
````

`break` thoát khỏi vòng lặp **gần nhất** (vòng `for i`), rồi tiếp tục vòng `for r` bên ngoài.

## 9. Code nộp LeetCode

````python
from typing import List

class Solution:
    def smallestEqual(self, nums: List[int]) -> int:
        for i, x in enumerate(nums):
            if i % 10 == x:
                return i
        return -1

sol = Solution()
print(sol.smallestEqual([4, 3, 2, 1]))
````

## 10. Vì sao đúng?

**Invariant:** trước khi xét chỉ số `i`, **không có** chỉ số `j < i` nào thỏa `j % 10 == nums[j]` (nếu có, hàm đã `return` ở `j`).

- Khi gặp `i` đầu tiên thỏa điều kiện: mọi chỉ số nhỏ hơn đều không thỏa (invariant) → `i` là **nhỏ nhất** → trả `i` đúng.
- Nếu duyệt hết không thấy: không chỉ số nào thỏa → `-1` đúng.

Cách 1 đúng vì `min` của tập các chỉ số thỏa chính là định nghĩa đề; cách 2 cho cùng kết quả nhưng dừng sớm.

## 11. Độ phức tạp

Gọi `n = len(nums)` (≤ 100).

| Cách | Time (tệ nhất) | Time (tốt nhất) | Space |
|---|---|---|---|
| 1. Gom list + min | `O(n)` | `O(n)` (luôn duyệt hết) | `O(n)` |
| 2. Duyệt dừng sớm | `O(n)` | `O(1)` (đáp án ở i = 0) | `O(1)` |
| 3. `next` + generator | `O(n)` | `O(1)` | `O(1)` |
| 4. Theo nhóm dư | `O(n)` (mỗi chỉ số xét tối đa 1 lần) + 10 nhóm | `O(10)` | `O(1)` |

Không thể tốt hơn `O(n)` trong trường hợp tệ nhất: nếu chưa nhìn một chỉ số nào đó, nó có thể là chỉ số duy nhất thỏa.

## 13. Lỗi người mới hay gặp

### Lỗi 1: Đặt `return -1` bên trong vòng lặp

````python
def wrong_return_inside(nums):
    for i, x in enumerate(nums):
        if i % 10 == x:
            return i
        return -1              # thụt lề sai: trả -1 ngay ở i = 0 nếu i=0 không thỏa

print("Sai :", wrong_return_inside([4, 3, 2, 1]))   # -1
print("Đúng:", smallest_equal([4, 3, 2, 1]))        # 2
````

### Lỗi 2: Quên `% 10`, so sánh `i == nums[i]`

Với mảng ≤ 10 phần tử thì không khác, nhưng với `i ≥ 10` sẽ sai (vì `nums[i] ≤ 9` không bao giờ bằng `i ≥ 10`).

````python
def wrong_no_mod(nums):
    for i, x in enumerate(nums):
        if i == x:
            return i
    return -1

print("Sai :", wrong_no_mod(LONG))   # -1
print("Đúng:", smallest_equal(LONG))  # 12
````

### Lỗi 3: Đảo vai trò — `nums[i] % 10 == i`

Đề là `i mod 10 == nums[i]`, không phải `nums[i] mod 10 == i`. Vì `nums[i] ≤ 9`, `nums[i] % 10 = nums[i]`, nên vế này lại thành `nums[i] == i` — sai tương tự lỗi 2.

### Lỗi 4: Trả về **giá trị** thay vì **chỉ số**

`return x` thay vì `return i`. Với ví dụ 2 thì tình cờ đúng (`nums[2] = 2`), nhưng với mảng `LONG` (đáp án ở chỉ số 12, `nums[12] = 2`) trả `2` thay vì `12`.

````python
def wrong_return_value(nums):
    for i, x in enumerate(nums):
        if i % 10 == x:
            return x
    return -1

print("Ví dụ 2 — sai mà vẫn trùng:", wrong_return_value([4, 3, 2, 1]))
print("Mảng dài — lộ lỗi:", wrong_return_value(LONG), "thay vì", smallest_equal(LONG))
````

Bài học: **ví dụ của đề có thể che giấu lỗi**. Luôn thêm test có chỉ số ≥ 10.

### Lỗi 5: Duyệt ngược để "tìm nhanh hơn"

Duyệt từ cuối về đầu sẽ tìm được chỉ số **lớn nhất** trước. Muốn nhỏ nhất bằng cách duyệt ngược thì phải duyệt hết và cập nhật liên tục — mất ưu điểm dừng sớm.

## 14. Mẹo nhận dạng pattern

- "Chỉ số **nhỏ nhất** / phần tử **đầu tiên** thỏa ..." → duyệt **từ trái**, `return` ngay khi gặp; sau vòng lặp trả giá trị "không tìm thấy" (`-1`).
- "Chỉ số **lớn nhất** / **cuối cùng**" → duyệt **từ phải** (`range(n-1, -1, -1)`), `return` ngay.
- Điều kiện có `i % m` → nghĩ về **chu kỳ m**; test với mảng dài hơn `m`.
- Python: `next((... for ... if ...), default)` là "find first" một dòng.
- Bài cùng kiểu: 1351 Count Negative Numbers, 2089 Find Target Indices After Sorting, 1572, 27/35 (tìm vị trí), 2614.

## 15. Biến thể

1. **Chỉ số lớn nhất** thỏa điều kiện.
2. **Tất cả** chỉ số thỏa (giờ thì cách 1 lại là đúng công cụ).
3. **Mod m tổng quát:** `i % m == nums[i]`.
4. **Đếm** số chỉ số thỏa.

````python
def largest_equal(nums):
    for i in range(len(nums) - 1, -1, -1):     # n-1, n-2, ..., 0
        if i % 10 == nums[i]:
            return i
    return -1

def all_equal_indices(nums, m=10):
    return [i for i, x in enumerate(nums) if i % m == x]

print(list(range(4, -1, -1)))
print(largest_equal([0, 1, 2]), all_equal_indices([0, 1, 2]), all_equal_indices([0, 1, 0, 1], m=2))
````

`range(start, stop, step)` với `step = -1` đếm lùi; `stop = -1` để **bao gồm** 0 (vì `stop` không được lấy).

## 16. Bài tập tự luyện

````python
def exercise_smallest_equal_answer(nums):
    for i in range(len(nums)):
        if i % 10 == nums[i]:
            return i
    return -1

def count_equal_answer(nums):
    return sum(i % 10 == x for i, x in enumerate(nums))

def smallest_equal_mod_answer(nums, m):
    for i, x in enumerate(nums):
        if i % m == x:
            return i
    return -1

def smallest_equal_queries_answer(nums, queries):
    n = len(nums)
    nxt = [-1] * (n + 1)
    for i in range(n - 1, -1, -1):
        nxt[i] = i if i % 10 == nums[i] else nxt[i + 1]
    out = []
    for l, r in queries:
        j = nxt[l]
        out.append(j if j != -1 and j <= r else -1)
    return out

assert exercise_smallest_equal_answer([4, 3, 2, 1]) == 2
assert count_equal_answer([0, 1, 2]) == 3
assert smallest_equal_mod_answer([1, 1, 0], 2) == 1
assert smallest_equal_queries_answer([4, 3, 2, 1, 4], [(0, 4), (3, 4), (0, 1)]) == [2, 4, -1]
random.seed(1)
for _ in range(300):
    nums = [random.randint(0, 9) for _ in range(random.randint(1, 30))]
    qs = [tuple(sorted(random.sample(range(len(nums)), 2))) if len(nums) > 1 else (0, 0) for _ in range(5)]
    # oracle: duyệt trực tiếp trên CHỈ SỐ GỐC i trong [l, r] (không cắt mảng)
    brute = [next((i for i in range(l, r + 1) if i % 10 == nums[i]), -1) for l, r in qs]
    assert smallest_equal_queries_answer(nums, qs) == brute
    assert exercise_smallest_equal_answer(nums) == smallest_equal(nums)
print("Tất cả đáp án bài tập — PASS")
````

Để ý vì sao oracle trong cell trên duyệt `range(l, r + 1)` chứ không cắt mảng: nếu cắt `nums[l:r+1]` thì chỉ số trong đoạn cắt bắt đầu lại từ 0, làm `i % 10` sai lệch. Đây là một lỗi rất thật khi làm bài truy vấn đoạn — luôn giữ **chỉ số gốc**.
