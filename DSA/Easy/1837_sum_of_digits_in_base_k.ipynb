## 2. Nền tảng: hệ cơ số là gì?

Trong cơ số 10, `345 = 3·10² + 4·10¹ + 5·10⁰`. Mỗi vị trí "nặng" gấp 10 lần vị trí bên phải.

Trong cơ số `k`, mỗi vị trí nặng gấp `k` lần. Ví dụ cơ số 6: `54₆ = 5·6¹ + 4·6⁰ = 30 + 4 = 34`.

Chữ số trong cơ số `k` chỉ được từ `0` đến `k-1` (cơ số 2 chỉ có 0 và 1).

````python
print("345 =", 3 * 10**2 + 4 * 10**1 + 5 * 10**0)
print("54 (base 6) =", 5 * 6**1 + 4 * 6**0)
print("1011 (base 2) =", 1 * 2**3 + 0 * 2**2 + 1 * 2**1 + 1 * 2**0)
````

`a ** b` là lũy thừa `a^b` trong Python (`6**2 = 36`). Đừng nhầm với `^` — trong Python `^` là **XOR**, không phải mũ.

````python
print(6 ** 2, "| 6 ^ 2 =", 6 ^ 2, "(XOR, không phải mũ!)")
````

### 2.1. `%` và `//` — tách chữ số cuối

Với cơ số 10 ta đã quen: `345 % 10 = 5` (chữ số cuối), `345 // 10 = 34` (bỏ chữ số cuối).

Điều tương tự đúng với **mọi cơ số** `k`:

- `n % k` = chữ số **cuối cùng** của `n` khi viết ở cơ số `k`.
- `n // k` = `n` sau khi **bỏ** chữ số cuối (ở cơ số `k`).

Vì sao? Viết `n = (...)·k + d` với `0 <= d < k`: phần `(...)·k` chia hết cho `k`, phần dư chính là `d` — chữ số hàng đơn vị.

````python
n, k = 34, 6
print("34 % 6  =", n % k, "  <- chữ số cuối (4)")
print("34 // 6 =", n // k, "  <- phần còn lại (5)")
print("kiểm tra: 5*6 + 4 =", 5 * 6 + 4)
print("345 % 10 =", 345 % 10, "| 345 // 10 =", 345 // 10)
````

### 2.2. `divmod(a, b)` — lấy cả thương và dư một lần

`divmod(a, b)` trả về **tuple** `(a // b, a % b)`. Có thể "mở gói" (unpack) thành hai biến: `q, r = divmod(a, b)`.

````python
q, r = divmod(34, 6)
print("divmod(34, 6) =", divmod(34, 6), "-> q =", q, ", r =", r)
````

### 2.3. Vòng `while n > 0`

Lặp "bóc chữ số cuối" cho tới khi `n` thành `0`. Mỗi vòng `n` nhỏ đi khoảng `k` lần nên vòng lặp chắc chắn dừng.

````python
n, k = 34, 6
while n > 0:
    print(f"n={n:>3}: chữ số cuối = {n % k}, còn lại = {n // k}")
    n //= k
print("n cuối =", n)
````

Chú ý: chữ số được bóc ra theo thứ tự **từ phải sang trái** (4 rồi 5). Muốn viết đúng số `54` phải **đảo ngược**; nhưng với bài này ta chỉ cần **tổng**, mà phép cộng không phụ thuộc thứ tự — không cần đảo!

### 2.4. Một số, nhiều cách viết

Cùng một **giá trị** 34 nhưng mỗi cơ số cho một chuỗi chữ số khác nhau, và **tổng chữ số cũng khác nhau**. Bảng dưới giúp thấy đề bài phụ thuộc vào `k` thế nào. (Cách tách chữ số sẽ được giảng kỹ ở mục 3; cell này chỉ để quan sát.)

````python
def _digits(n, k):
    out = []
    while n > 0:
        out.append(n % k)
        n //= k
    return out[::-1] or [0]

print(" k | 34 viết ở cơ số k | tổng chữ số")
for k in range(2, 11):
    d = _digits(34, k)
    print(f"{k:>2} | {''.join(map(str, d)):>17} | {sum(d)}")
````

Quan sát: `k` càng nhỏ thì càng nhiều chữ số nhưng mỗi chữ số càng nhỏ. Không có quy luật đơn điệu cho tổng — phải tính thật.

`out[::-1] or [0]`: nếu list rỗng (khi `n = 0`) thì `[]` được coi là `False`, biểu thức `or` trả về `[0]`.

## 3. Cách 1 — đổi sang list chữ số rồi cộng (trực giác)

Làm đúng như đề mô tả: (1) đổi `n` sang cơ số `k` thành list chữ số, (2) cộng các chữ số.

````python
def to_base_digits(n: int, k: int) -> list:
    if n == 0:
        return [0]
    digits = []
    while n > 0:
        digits.append(n % k)
        n //= k
    digits.reverse()          # đảo để chữ số cao nhất đứng đầu
    return digits

for n, k in [(34, 6), (10, 10), (5, 2), (100, 3)]:
    d = to_base_digits(n, k)
    print(f"{n} (base 10) = {''.join(map(str, d))} (base {k})")
````

- `list.append(x)` thêm vào cuối; `list.reverse()` đảo **tại chỗ**.
- `map(str, d)` đổi từng chữ số thành chuỗi; `''.join(...)` nối chúng lại.

````python
def sum_base_via_list(n: int, k: int) -> int:
    return sum(to_base_digits(n, k))

for (n, k), expected in examples:
    print(n, k, "->", sum_base_via_list(n, k), "| expected", expected)
    assert sum_base_via_list(n, k) == expected
````

````python
# Kiểm tra hàm đổi cơ số bằng int(s, base)
for n in range(1, 101):
    for k in range(2, 11):
        s = "".join(map(str, to_base_digits(n, k)))
        assert int(s, k) == n
print("to_base_digits đúng với mọi n 1..100, k 2..10")
````

### 3.1. Có gì thừa?

Cách 1 đúng, nhưng tạo list, đảo list, rồi lại duyệt list để cộng. Với **tổng**, ta không cần biết thứ tự, cũng không cần lưu chữ số. Chỉ cần **cộng dồn ngay khi bóc** ra.

Đây không phải nút thắt về tốc độ (n rất nhỏ) mà là bài học về **bỏ bớt trạng thái không cần thiết** — viết code gọn, ít chỗ sai.

### 3.2. Insight

> Tổng chữ số = cộng dồn `n % k` qua mỗi vòng `n //= k`.

Thứ tự bóc (phải → trái) không ảnh hưởng tổng vì phép cộng giao hoán. Không cần list, không cần đảo.

````python
n, k = 34, 6
total = 0
total += n % k; n //= k     # cộng 4, n = 5
total += n % k; n //= k     # cộng 5, n = 0
print("total =", total, "| n =", n)
````

Dấu `;` cho phép viết hai lệnh trên một dòng — chỉ dùng trong minh họa, code thật nên tách dòng.

## 4. Cách 2 — lời giải chính: cộng dồn trong vòng lặp

Đoạn 1: `total = 0`.

Đoạn 2: `while n > 0:` — còn chữ số thì còn bóc.

Đoạn 3: `total += n % k` — cộng chữ số cuối.

Đoạn 4: `n //= k` — bỏ chữ số cuối.

````python
def sum_base_loop(n: int, k: int) -> int:
    total = 0
    while n > 0:
        total += n % k
        n //= k
    return total

for (n, k), expected in examples:
    assert sum_base_loop(n, k) == expected
print(sum_base_loop(34, 6), sum_base_loop(10, 10))
````

## 5. TRACE từng bước

Trace `n = 100, k = 3`. Ta biết `100 = 81 + 18 + 1 = 1·81 + 0·27 + 2·9 + 0·3 + 1·1` ⇒ `10201₃`, tổng = 1+0+2+0+1 = **4**.

| vòng | n (trước) | n % k (chữ số) | total (sau) | n // k |
|---:|---:|---:|---:|---:|
| 1 | 100 | 1 | 1 | 33 |
| 2 | 33 | 0 | 1 | 11 |
| 3 | 11 | 2 | 3 | 3 |
| 4 | 3 | 0 | 3 | 1 |
| 5 | 1 | 1 | 4 | 0 |

`n = 0` ⇒ dừng, trả `4`.

````python
def trace_sum_base(n: int, k: int) -> int:
    total = 0
    step = 1
    print(f"Đổi {n} sang cơ số {k}")
    while n > 0:
        digit = n % k
        total += digit
        print(f"vòng {step}: n={n:>3} -> chữ số {digit}, total={total:>2}, n mới = {n // k}")
        n //= k
        step += 1
    print("Dừng vì n = 0. Kết quả:", total)
    return total

assert trace_sum_base(100, 3) == 4
````

Trace thứ hai — trường hợp đặc biệt `k = 10` và `n = k`: kết quả luôn là 1 (`10₁₀`, `10₆ = 6`, …). Cũng thử cơ số 2 để thấy tổng chữ số = **số bit 1**.

````python
assert trace_sum_base(10, 10) == 1
print()
assert trace_sum_base(13, 2) == bin(13).count("1")
print("13 =", bin(13), "-> số bit 1 =", bin(13).count("1"))
````

## 6. Cách 3 — `divmod`

Gộp `%` và `//` thành một lần gọi: `n, d = divmod(n, k)`. Dòng này **gán đồng thời**: vế phải tính xong hết (dùng `n` cũ) rồi mới gán cho `n` và `d`.

````python
def sum_base_divmod(n: int, k: int) -> int:
    total = 0
    while n:
        n, d = divmod(n, k)
        total += d
    return total

print(sum_base_divmod(34, 6), sum_base_divmod(100, 3))
````

`while n:` tương đương `while n != 0:` — số `0` được coi là `False`, số khác 0 là `True` (truthiness).

````python
for v in [0, 1, 7]:
    print(v, "->", bool(v))
````

## 7. Cách 4 — đệ quy

Định nghĩa toán học: `S(0) = 0`, `S(n) = (n % k) + S(n // k)`. "Tổng chữ số = chữ số cuối + tổng chữ số của phần còn lại."

Hàm **đệ quy** là hàm gọi chính nó với input nhỏ hơn, có **điều kiện dừng** (base case) để không gọi mãi.

````python
def sum_base_recursive(n: int, k: int) -> int:
    if n == 0:                 # base case
        return 0
    return n % k + sum_base_recursive(n // k, k)

print(sum_base_recursive(34, 6), sum_base_recursive(100, 3))
````

````python
def trace_recursive(n, k, depth=0):
    pad = "  " * depth
    if n == 0:
        print(pad + "S(0) = 0")
        return 0
    print(pad + f"S({n}) = {n % k} + S({n // k})")
    res = n % k + trace_recursive(n // k, k, depth + 1)
    print(pad + f"=> S({n}) = {res}")
    return res

trace_recursive(34, 6)
````

Mỗi tầng thụt vào thể hiện một lần gọi trên **call stack**. Kết quả được "trả ngược" từ tầng sâu nhất lên.

## 8. Code nộp LeetCode

### Cách A — vòng lặp cộng dồn (khuyên dùng)

````python
class Solution:
    def sumBase(self, n: int, k: int) -> int:
        total = 0
        while n > 0:
            total += n % k
            n //= k
        return total

print(Solution().sumBase(34, 6), Solution().sumBase(10, 10))
````

### Cách B — `divmod`

````python
class SolutionDivmod:
    def sumBase(self, n: int, k: int) -> int:
        total = 0
        while n:
            n, d = divmod(n, k)
            total += d
        return total

print(SolutionDivmod().sumBase(34, 6))
````

### Cách C — đệ quy

````python
class SolutionRecursive:
    def sumBase(self, n: int, k: int) -> int:
        if n == 0:
            return 0
        return n % k + self.sumBase(n // k, k)

print(SolutionRecursive().sumBase(34, 6))
````

## 9. Vì sao đúng?

Viết `n = d_m·k^m + ... + d_1·k + d_0` (biểu diễn cơ số `k`, `0 <= d_i < k`).

- `n % k = d_0` vì mọi số hạng khác chia hết cho `k`, và `d_0 < k`.
- `n // k = d_m·k^(m-1) + ... + d_1` — chính là số có các chữ số `d_m ... d_1`.

**Invariant:** ở đầu mỗi vòng, `total + (tổng chữ số của n hiện tại ở cơ số k) = đáp án`.

- Ban đầu: `total = 0`, `n` là input ⇒ đúng.
- Mỗi vòng: chuyển `d_0` từ "phần còn lại" sang `total`, và `n` mất đúng chữ số `d_0` ⇒ vẫn đúng.
- Khi `n = 0`: tổng chữ số của `0` là `0` ⇒ `total = đáp án`.

**Dừng:** `k >= 2` nên `n // k < n` với `n > 0` — `n` giảm nghiêm ngặt về 0.

## 10. Độ phức tạp

Số chữ số của `n` ở cơ số `k` là `⌊log_k n⌋ + 1`. Gọi `d` là số đó.

| Cách | Time | Space |
|---|---|---|
| List chữ số | `O(d)` = `O(log_k n)` | `O(d)` (list) |
| Cộng dồn | `O(log_k n)` | `O(1)` |
| `divmod` | `O(log_k n)` | `O(1)` |
| Đệ quy | `O(log_k n)` | `O(log_k n)` (call stack) |

Với `n <= 100`, `k = 2` là xấu nhất: `d = 7` vòng.

````python
import math
for k in [2, 3, 6, 10]:
    d = len(to_base_digits(100, k))
    print(f"k={k:>2}: 100 có {d} chữ số, log_k(100)+1 ≈ {math.log(100, k) + 1:.2f}")
````

Nhóm test: mẫu đề; `n < k` (một chữ số ⇒ tổng = `n`); `n = k` (⇒ 1); lũy thừa của `k` (⇒ 1); cơ số 2 (⇒ số bit 1); vét toàn bộ miền `n 1..100`, `k 2..10` so với oracle dùng chuỗi.

````python
def check(name, actual, expected):
    assert actual == expected, (name, actual, expected)
    print(f"{name:<22} PASS ({actual})")

sol = Solution()
check("mẫu 1", sol.sumBase(34, 6), 9)
check("mẫu 2", sol.sumBase(10, 10), 1)
check("n < k", sol.sumBase(7, 9), 7)
check("n = k", sol.sumBase(5, 5), 1)
check("lũy thừa 2**6", sol.sumBase(64, 2), 1)
check("n = 1", sol.sumBase(1, 2), 1)
check("100 base 2", sol.sumBase(100, 2), bin(100).count("1"))
check("100 base 9", sol.sumBase(100, 9), 1 + 2 + 1)   # 100 = 121 (base 9)
````

````python
def oracle(n, k):
    # nguồn độc lập: format() có sẵn cho cơ số 2/8/10; cơ số khác dùng list rồi tự kiểm bằng int(s, k)
    if k == 2:
        s = format(n, "b")
    elif k == 8:
        s = format(n, "o")
    elif k == 10:
        s = str(n)
    else:
        s = "".join(map(str, to_base_digits(n, k)))
    assert int(s, k) == n
    return sum(int(c) for c in s)

impls = [sum_base_via_list, sum_base_loop, sum_base_divmod, sum_base_recursive,
         Solution().sumBase, SolutionDivmod().sumBase, SolutionRecursive().sumBase]
for n in range(1, 101):
    for k in range(2, 11):
        exp = oracle(n, k)
        assert all(f(n, k) == exp for f in impls), (n, k)
print("Vét cạn 900 cặp (n, k) x 7 cài đặt PASS")
````

`format(n, "b")` cho chuỗi nhị phân không có tiền tố `0b`; `"o"` cho bát phân. Đây là nguồn đối chiếu **độc lập** với code ta viết cho `k = 2, 8, 10`.

## 12. Lỗi người mới hay gặp

### Lỗi 1: dùng `/` thay vì `//`

````python
def wrong_true_division(n, k):
    total = 0
    steps = 0
    while n > 0 and steps < 50:     # chặn vòng lặp để demo không treo
        total += n % k
        n = n / k                   # ra số thực!
        steps += 1
    return total

print("sai :", wrong_true_division(34, 6), "| đúng:", sum_base_loop(34, 6))
````

`n / k` cho số thực (`5.666…`), `n` không bao giờ đúng bằng 0 nhanh, và `%` trên số thực cho phần dư lẻ. Sửa: `n //= k`.

### Lỗi 2: đổi sang chuỗi cơ số 10 rồi cộng chữ số

````python
def wrong_base10_digits(n, k):
    return sum(int(c) for c in str(n))     # quên đổi cơ số!

print("sai :", wrong_base10_digits(34, 6), "| đúng:", sum_base_loop(34, 6))
````

`str(n)` luôn là cơ số 10. Sửa: tách chữ số bằng `% k`.

### Lỗi 3: thứ tự `n //= k` trước `total += n % k`

````python
def wrong_order(n, k):
    total = 0
    while n > 0:
        n //= k                   # bỏ chữ số cuối TRƯỚC khi cộng
        total += n % k
    return total

print("sai :", wrong_order(34, 6), "| đúng:", sum_base_loop(34, 6))
````

Bỏ mất chữ số hàng đơn vị. Sửa: cộng trước, chia sau (hoặc dùng `divmod`).

### Lỗi 4: dùng `^` làm lũy thừa khi tự kiểm tra

````python
print("5*6^1 + 4 =", 5 * 6 ^ 1 + 4, "(sai, ^ là XOR)")
print("5*6**1 + 4 =", 5 * 6 ** 1 + 4, "(đúng)")
````

## 13. Mẹo nhận dạng pattern

- Thấy "chữ số", "cơ số", "đổi hệ", "reverse số" ⇒ nghĩ ngay cặp **`n % k` / `n // k`** trong `while n > 0`.
- Chỉ cần **tổng / đếm / max** của chữ số ⇒ không cần lưu list, không cần đảo.
- Cần **biểu diễn đầy đủ** (chuỗi) ⇒ append rồi `reverse()` ở cuối.
- Muốn kiểm thử ⇒ dùng `int(s, base)` hoặc `format(n, 'b')`.
- Cơ số 2 ⇒ tổng chữ số = số bit 1 (`bin(n).count('1')`, hoặc `n.bit_count()` từ Python 3.10).

### 13.1. Hình dung `n // k` và `n % k` bằng cách gom nhóm

`n // k` là "có bao nhiêu nhóm `k` đầy đủ trong `n`", còn `n % k` là "phần lẻ không đủ một nhóm". Đổi cơ số chính là **gom nhóm lặp lại**: gom `n` viên bi thành nhóm `k`, rồi gom các nhóm thành "siêu nhóm" `k`, … Số dư ở mỗi tầng gom là một chữ số.

````python
n, k = 34, 6
level = 0
while n > 0:
    groups, left = divmod(n, k)
    print(f"tầng {level}: {n} vật -> {groups} nhóm {k} + dư {left}  (chữ số hàng k^{level} = {left})")
    n = groups
    level += 1
````

## 14. Biến thể

1. **Biểu diễn chuỗi** của `n` ở cơ số `k` (bài 504 Base 7, 405 hex).
2. **Tích chữ số** hoặc **chữ số lớn nhất** ở cơ số `k`.
3. **Kiểm tra palindrome ở cơ số k** (bài 2396 Strictly Palindromic Number).
4. **Cơ số > 10** ⇒ chữ số ≥ 10 dùng chữ cái `A, B, ...`.

````python
DIGITS = "0123456789ABCDEFGHIJKLMNOPQRSTUVWXYZ"

def to_base_string(n, k):
    if n == 0:
        return "0"
    out = []
    while n > 0:
        out.append(DIGITS[n % k])
        n //= k
    return "".join(reversed(out))

print(to_base_string(34, 6), to_base_string(255, 16), to_base_string(100, 7))
assert int(to_base_string(255, 16), 16) == 255
print("n=255 có", (255).bit_count() if hasattr(int, "bit_count") else bin(255).count("1"), "bit 1")
````

`reversed(out)` trả về một **iterator** duyệt ngược (không sửa list), dùng được trong `join`.

````python
def answer_sum_base(n, k):
    total = 0
    while n > 0:
        total += n % k
        n //= k
    return total


def answer_to_base(n, k):
    if n == 0:
        return "0"
    out = []
    while n > 0:
        out.append(str(n % k))
        n //= k
    return "".join(reversed(out))


def answer_is_palindrome_base(n, k):
    s = answer_to_base(n, k)
    return s == s[::-1]


def answer_best_base(n):
    best_sum, best_k = min((answer_sum_base(n, k), k) for k in range(2, 11))
    return (best_k, best_sum)


assert answer_sum_base(34, 6) == 9
assert answer_sum_base(10, 10) == 1
assert answer_to_base(34, 6) == "54"
assert answer_to_base(0, 2) == "0"
assert answer_to_base(5, 2) == "101"
assert answer_is_palindrome_base(5, 2) is True
assert answer_is_palindrome_base(6, 2) is False
assert answer_best_base(64) == (2, 1)
assert answer_best_base(10) == (10, 1)
for n in range(0, 200):
    for k in range(2, 11):
        assert int(answer_to_base(n, k), k) == n
print("Đáp án tham khảo PASS toàn bộ test!")
````
