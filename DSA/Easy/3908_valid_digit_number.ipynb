## 1. Mục tiêu học

Sau notebook này bạn sẽ:

1. Hiểu đề: số `n` là **hợp lệ** nếu **chứa ít nhất một chữ số `x`** nhưng **không bắt đầu bằng chữ số `x`**.
2. Biết tách chữ số của một số nguyên bằng chuyển sang chuỗi `str(n)` hoặc bằng `divmod(n, 10)`, và lấy **chữ số đầu** của số.
3. Biết ba cách diễn đạt tương đương: `x in digits[1:]` và `digits[0] != x`, hoặc `s.find(x) > 0`.
4. Viết 5 cách khác nhau và đối chiếu bằng cách kiểm tra **toàn bộ** `n` từ `0` đến `1000` với mọi `x`, cùng test ngẫu nhiên có `random.seed`.
5. Tránh các lỗi: so sánh chuỗi với số nguyên (`'2' != 2` luôn đúng), lấy nhầm chữ số cuối thay vì đầu, quên rằng `n = 0` vẫn có chữ số `0`, bỏ sót điều kiện "chứa".

## 2. Đề bài & ví dụ

Cho số nguyên `n` và một chữ số `x`. Số `n` được gọi là **hợp lệ** nếu:

1. biểu diễn thập phân của `n` **chứa** chữ số `x` ít nhất một lần; và
2. chữ số **đầu tiên** (ngoài cùng bên trái) của `n` **khác** `x`.

Trả về `true` nếu `n` hợp lệ, ngược lại `false`.

| Ví dụ | n | x | Output | Giải thích |
|---|---|---|---|---|
| 1 | `101` | `0` | `true` | Có chữ số `0` ở vị trí giữa và chữ số đầu là `1` (khác `0`) |
| 2 | `232` | `2` | `false` | Số bắt đầu bằng `2`, vi phạm điều kiện 2 |
| 3 | `5` | `1` | `false` | Không chứa chữ số `1` |

Constraints: `0 <= n <= 10^5`; `0 <= x <= 9`.

````python
for n, x in [(101, 0), (232, 2), (5, 1)]:
    s = str(n)
    print(f"n={n} x={x}: chữ số = {list(s)}, chứa x? {str(x) in s}, bắt đầu bằng x? {s[0] == str(x)}")
````

## 3. Kiến thức nền

### 3.1. Đổi số sang chuỗi và duyệt chữ số

`str(n)` cho biểu diễn thập phân (không có số `0` thừa ở đầu, trừ chính số `0` là `"0"`). Mỗi ký tự là một chữ số. Dùng `int(ch)` để đổi ngược một ký tự chữ số thành số.

````python
n = 232
s = str(n)
print(s, list(s), [int(c) for c in s], str(0), len(str(100000)))
````

### 3.2. Phân biệt `'2'` (chuỗi) và `2` (số)

`'2' == 2` là `False` trong Python (khác kiểu). Khi so sánh chữ số của chuỗi với `x` (số nguyên) phải đổi kiểu: `s[0] == str(x)` hoặc `int(s[0]) == x`.

````python
print('2' == 2, '2' == str(2), int('2') == 2)
s = "232"
print(s[0] != 2, s[0] != str(2))
````

### 3.3. Toán tử `in` và `str.find`

`'2' in "232"` kiểm tra có chứa hay không. `"232".find('2')` trả **chỉ số đầu tiên** của ký tự (`0`), hoặc `-1` nếu không có. Vì vậy `s.find(c) > 0` nghĩa là "có xuất hiện và lần xuất hiện đầu tiên không ở vị trí 0".

````python
for s, c in [("101", "0"), ("232", "2"), ("5", "1")]:
    print(s, c, c in s, s.find(c), s.find(c) > 0)
````

### 3.4. Tách chữ số bằng `divmod`

`divmod(n, 10)` trả `(n // 10, n % 10)`: phần còn lại và **chữ số cuối**. Lặp đến khi `n == 0` ta lấy được các chữ số từ **cuối lên đầu**; chữ số cuối cùng lấy được (ở lần lặp cuối) chính là **chữ số đầu** của số.

````python
n = 1234
digits = []
while n > 0:
    n, d = divmod(n, 10)
    digits.append(d)
print(digits, "chữ số đầu =", digits[-1])
````

### 3.5. Trường hợp đặc biệt `n = 0`

Vòng `while n > 0` không chạy với `n = 0`, nhưng `0` có đúng một chữ số `0`. Với cách chuỗi `str(0) = "0"` không có vấn đề; với cách chia cần xử lý riêng (hoặc dùng `do-while`).

````python
print(str(0), list(str(0)))
n = 0
digits = []
while True:
    n, d = divmod(n, 10)
    digits.append(d)
    if n == 0:
        break
print(digits)
````

## 4. Ý tưởng nền tảng / cách nghĩ trực giác

Hai điều kiện độc lập:

- **Có `x`** trong các chữ số;
- **Chữ số đầu không phải `x`**.

Ghép lại: "`x` xuất hiện ở **ít nhất một vị trí không phải vị trí đầu**, và chữ số đầu khác `x`." Nói gọn: **chữ số đầu khác `x` và `x` có mặt ở phần còn lại**.

Tính tay:

| n | x | chữ số | đầu | chứa `x`? | đầu `!= x`? | kết quả |
|---|---|---|---|---|---|---|
| 101 | 0 | 1,0,1 | 1 | có | có | `true` |
| 232 | 2 | 2,3,2 | 2 | có | không | `false` |
| 5 | 1 | 5 | 5 | không | có | `false` |
| 0 | 0 | 0 | 0 | có | không | `false` |
| 0 | 3 | 0 | 0 | không | có | `false` |
| 120 | 0 | 1,2,0 | 1 | có | có | `true` |

````python
def explain(n, x):
    s = str(n)
    has = str(x) in s
    first_ok = s[0] != str(x)
    print(f"n={n:>3} x={x}: chứa={has!s:5} đầu!=x={first_ok!s:5} -> {has and first_ok}")
    return has and first_ok

for n, x in [(101, 0), (232, 2), (5, 1), (0, 0), (0, 3), (120, 0)]:
    explain(n, x)
````

## 5. Cách 1 — Brute Force / baseline

Duyệt qua **từng chữ số** của `str(n)`: đếm xem `x` xuất hiện bao nhiêu lần (`found`), và kiểm tra chữ số đầu (`str(n)[0]`). Trả `found and chữ_số_đầu != x`.

Rất thẳng thắn, đúng nghĩa đề. "Nút thắt": viết dài hơn mức cần thiết; ta có thể gộp thành một biểu thức hoặc dùng `find`.

````python
class SolutionBrute:
    def validDigit(self, n, x):
        s = str(n)
        found = False
        for ch in s:
            if int(ch) == x:
                found = True
        first = int(s[0])
        return found and first != x

print(SolutionBrute().validDigit(101, 0), SolutionBrute().validDigit(232, 2), SolutionBrute().validDigit(5, 1))
````

**Độ phức tạp:** Time `O(d)` với `d` là số chữ số của `n` (`d <= 6`), Space `O(d)` cho chuỗi.

## 6. Ý tưởng thuật toán / bước đột phá

**Bước đột phá:** gom hai điều kiện thành một câu hỏi: "**`x` có xuất hiện ở một vị trí không phải đầu không, và đầu có khác `x` không?**" Ta không cần đếm số lần xuất hiện, chỉ cần biết **có/không**.

- Bằng chuỗi: `d = str(x)`, `s = str(n)`: `s[0] != d and d in s`.
- Bằng `find`: `s.find(d) > 0` (vị trí xuất hiện đầu tiên khác `0` và không phải `-1`). Điều kiện này **tự đã bao hàm** cả "chứa" lẫn "không bắt đầu bằng": nếu lần xuất hiện đầu tiên ở vị trí `> 0` thì chữ số đầu khác `x` (vì nếu chữ số đầu là `x` thì lần xuất hiện đầu tiên là `0`).

**Invariant:** `s.find(d)` là chỉ số nhỏ nhất của `d`. `> 0` nghĩa là `d` có mặt và vị trí 0 không phải `d`.

## 7. Các cách tiếp theo

### Cách 2 — Biểu thức chuỗi `s[0] != d and d in s`

Đúng hai điều kiện của đề, một dòng.

````python
class SolutionString:
    def validDigit(self, n, x):
        s, d = str(n), str(x)
        return s[0] != d and d in s

print(SolutionString().validDigit(101, 0), SolutionString().validDigit(232, 2), SolutionString().validDigit(5, 1))
````

**Độ phức tạp:** Time `O(d)`, Space `O(d)`.

### Cách 3 — `str.find(d) > 0`

Chỉ một phép kiểm tra: lần xuất hiện đầu tiên phải ở vị trí dương.

````python
class SolutionFind:
    def validDigit(self, n, x):
        return str(n).find(str(x)) > 0

print(SolutionFind().validDigit(101, 0), SolutionFind().validDigit(232, 2), SolutionFind().validDigit(5, 1))
````

**Độ phức tạp:** Time `O(d)`, Space `O(d)`.

### Cách 4 — Phép toán số học (không dùng chuỗi)

Dùng `divmod` để lấy các chữ số từ cuối lên đầu, ghi nhận `seen` (đã gặp `x` ở vị trí không phải đầu) và lấy chữ số đầu ở lần lặp cuối. Dùng cấu trúc `while True ... break` để xử lý `n = 0`.

````python
class SolutionMath:
    def validDigit(self, n, x):
        seen = False                 # đã thấy x ở vị trí không phải đầu?
        while True:
            rest, d = divmod(n, 10)
            if rest == 0:            # d là chữ số đầu
                return seen and d != x
            if d == x:
                seen = True
            n = rest

print(SolutionMath().validDigit(101, 0), SolutionMath().validDigit(232, 2), SolutionMath().validDigit(5, 1), SolutionMath().validDigit(0, 0))
````

**Độ phức tạp:** Time `O(d)`, Space `O(1)`.

### Cách 5 — Danh sách chữ số và `any` trên phần đuôi

`digits = [int(c) for c in str(n)]`, trả `digits[0] != x and any(d == x for d in digits[1:])`. Rõ ràng hóa cấu trúc "đầu" và "phần còn lại".

````python
class SolutionDigits:
    def validDigit(self, n, x):
        digits = [int(c) for c in str(n)]
        head, tail = digits[0], digits[1:]
        return head != x and x in tail

print(SolutionDigits().validDigit(101, 0), SolutionDigits().validDigit(232, 2), SolutionDigits().validDigit(5, 1), SolutionDigits().validDigit(0, 0))
````

**Độ phức tạp:** Time `O(d)`, Space `O(d)`.

## 8. TRACE từng bước

Trace Cách 4 (số học) với `n = 120`, `x = 0` (lấy chữ số từ cuối lên đầu):

| bước | n | rest | d | `rest == 0`? | `d == x`? | seen |
|---|---|---|---|---|---|---|
| 1 | 120 | 12 | 0 | không | có | True |
| 2 | 12 | 1 | 2 | không | không | True |
| 3 | 1 | 0 | 1 | có (d=1 là chữ số đầu) | | trả `True and 1 != 0` = True |

````python
def trace_math(n, x):
    seen = False
    step = 0
    while True:
        step += 1
        rest, d = divmod(n, 10)
        print(f"bước {step}: n={n} rest={rest} d={d}", end="")
        if rest == 0:
            ans = seen and d != x
            print(f" -> d là chữ số đầu; seen={seen}, d != x: {d != x} => {ans}")
            return ans
        if d == x:
            seen = True
            print(f" -> d == x, seen=True")
        else:
            print()
        n = rest

assert trace_math(120, 0) is True
assert trace_math(232, 2) is False
assert trace_math(5, 1) is False
assert trace_math(0, 0) is False
````

````python
def trace_find(n, x):
    s, d = str(n), str(x)
    idx = s.find(d)
    print(f"n={n}, x={x}: s={s!r}, find({d!r}) = {idx} -> {idx > 0}")
    return idx > 0

for args in [(101, 0), (232, 2), (5, 1), (0, 0), (100000, 0)]:
    trace_find(*args)
````

## 9. Code nộp LeetCode

Dùng Cách 2 (hai điều kiện rõ ràng trên chuỗi): dễ đọc và dễ kiểm tra từng vế.

````python
from typing import List, Optional


class Solution:
    def validDigit(self, n: int, x: int) -> bool:
        s, d = str(n), str(x)
        return s[0] != d and d in s


print(Solution().validDigit(101, 0), Solution().validDigit(232, 2), Solution().validDigit(5, 1))
````

## 10. Vì sao thuật toán đúng

1. **Biểu diễn đúng.** `str(n)` là biểu diễn thập phân chuẩn: `s[0]` là chữ số đầu (khác `0` trừ khi `n = 0`), các ký tự còn lại là các chữ số tiếp theo.
2. **Đổi `x` sang cùng kiểu.** `d = str(x)` là ký tự chữ số tương ứng, nên `s[0] != d` đúng nghĩa "chữ số đầu khác `x`", và `d in s` đúng nghĩa "chứa `x`".
3. **Hai điều kiện của đề.** Trả về `True` khi cả hai đúng, chính là định nghĩa "hợp lệ".
4. **Tương đương với `find > 0`.** Nếu chữ số đầu khác `x` và `x` có mặt thì lần xuất hiện đầu tiên ở vị trí `>= 1`; ngược lại nếu `find > 0` thì `x` có mặt và vị trí `0` không phải `x`. Hai điều kiện tương đương.
5. **Biên.** `n = 0`: `s = "0"`; với `x = 0` thì `s[0] == d` nên `False`; với `x != 0` thì `d not in s` nên `False`. Số một chữ số luôn `False` (chữ số duy nhất vừa là đầu vừa là toàn bộ).

## 11. Phân tích Time / Space Complexity

Gọi `d` là số chữ số của `n` (`d <= 6` vì `n <= 10^5`).

- Cách 1, 2, 3, 5: `O(d)` thời gian, `O(d)` bộ nhớ cho chuỗi.
- Cách 4: `O(d)` thời gian, `O(1)` bộ nhớ.

Vì `d` tối đa 6 nên mọi cách đều chạy hằng số. Không có sự khác biệt best / worst đáng kể.

## 12. Bộ test

Vì `n <= 10^5` và `x <= 9`, ta có thể kiểm tra **toàn bộ** `n` từ `0` đến `1000` với mọi `x` từ `0` đến `9` (10 010 cặp) cho cả 5 cách, so với một **bộ chuẩn độc lập** (liệt kê các chữ số theo `divmod`), cộng test ngẫu nhiên có seed trên `0..10^5`.

````python
import random

solutions = [SolutionBrute(), SolutionString(), SolutionFind(), SolutionMath(), SolutionDigits(), Solution()]

def reference(n, x):
    digits = []
    m = n
    while True:
        m, d = divmod(m, 10)
        digits.append(d)
        if m == 0:
            break
    digits.reverse()                  # digits[0] là chữ số đầu
    return (x in digits) and digits[0] != x

fixed = [((101, 0), True), ((232, 2), False), ((5, 1), False), ((0, 0), False), ((0, 5), False),
         ((120, 0), True), ((100, 0), True), ((10, 0), True), ((10, 1), False), ((100000, 0), True), ((100000, 1), False),
         ((99999, 9), False), ((12321, 3), True)]
for (n, x), expected in fixed:
    for s in solutions:
        assert s.validDigit(n, x) is expected, (type(s).__name__, n, x)

for n in range(0, 1001):
    for x in range(10):
        ref = reference(n, x)
        for s in solutions:
            assert s.validDigit(n, x) is ref, (type(s).__name__, n, x)

random.seed(39080)
for _ in range(500):
    n = random.randint(0, 10**5)
    x = random.randint(0, 9)
    ref = reference(n, x)
    assert all(s.validDigit(n, x) is ref for s in solutions)
print("tất cả test OK")
````

### Test thuộc tính

(1) Số có một chữ số luôn `False`. (2) Nếu `n` hợp lệ với `x` thì thêm một chữ số **bất kỳ** vào **cuối** `n` vẫn hợp lệ (chữ số đầu không đổi, vẫn chứa `x`), miễn `n * 10 + c <= 10^5`. (3) Với `x` cố định, đếm số `n` hợp lệ trong `0..99` bằng công thức.

````python
random.seed(3908)
for n in range(10):
    for x in range(10):
        assert Solution().validDigit(n, x) is False
for _ in range(500):
    n = random.randint(1, 9999)
    x = random.randint(0, 9)
    if Solution().validDigit(n, x):
        c = random.randint(0, 9)
        assert Solution().validDigit(n * 10 + c, x)
# hai chữ số "ab": hợp lệ khi b == x và a != x  => với x cố định có 9 - (x != 0) ... đếm trực tiếp
for x in range(10):
    cnt = sum(Solution().validDigit(n, x) for n in range(10, 100))
    expected = sum(1 for a in range(1, 10) for b in range(10) if b == x and a != x)
    assert cnt == expected
print("property test OK")
````

## 13. Các lỗi người mới hay gặp

### Lỗi 1: so sánh chuỗi với số nguyên

`'2' != 2` luôn `True` vì khác kiểu; điều kiện "không bắt đầu bằng `x`" lúc nào cũng đúng nên `232, x = 2` trả `True` thay vì `False`.

````python
def wrong_type(n, x):
    s = str(n)
    return s[0] != x and str(x) in s                              # SAI: s[0] là chuỗi, x là số

print(wrong_type(232, 2), "!=", Solution().validDigit(232, 2))
assert wrong_type(232, 2) is True and Solution().validDigit(232, 2) is False
assert ('2' != 2) is True
````

### Lỗi 2: lấy chữ số **cuối** thay vì chữ số **đầu**

`n % 10` là chữ số cuối. Dùng nó làm "chữ số đầu" cho kết quả sai, ví dụ `101, x = 0`: chữ số cuối là `1` còn chữ số đầu cũng `1`, nhưng `100, x = 0`: chữ số cuối `0` làm ta loại nhầm.

````python
def wrong_last_digit(n, x):
    return n % 10 != x and str(x) in str(n)                       # SAI: n % 10 là chữ số cuối

print(wrong_last_digit(100, 0), "!=", Solution().validDigit(100, 0))
assert wrong_last_digit(100, 0) is False and Solution().validDigit(100, 0) is True
assert wrong_last_digit(210, 2) is True and Solution().validDigit(210, 2) is False
````

### Lỗi 3: chỉ kiểm tra "chứa `x`" (bỏ điều kiện không bắt đầu bằng `x`)

`232, x = 2` chứa `2` nhưng bắt đầu bằng `2` nên không hợp lệ.

````python
def wrong_only_contains(n, x):
    return str(x) in str(n)                                       # SAI: thiếu điều kiện chữ số đầu

print(wrong_only_contains(232, 2), "!=", Solution().validDigit(232, 2))
assert wrong_only_contains(232, 2) is True and Solution().validDigit(232, 2) is False
````

### Lỗi 4: chỉ kiểm tra "không bắt đầu bằng `x`" (bỏ điều kiện chứa)

`5, x = 1` không bắt đầu bằng `1` nhưng cũng không chứa `1` nên không hợp lệ.

````python
def wrong_only_not_start(n, x):
    return str(n)[0] != str(x)                                    # SAI: thiếu điều kiện chứa x

print(wrong_only_not_start(5, 1), "!=", Solution().validDigit(5, 1))
assert wrong_only_not_start(5, 1) is True and Solution().validDigit(5, 1) is False
````

### Lỗi 5: số học bỏ sót `n = 0`

Vòng `while n > 0` không chạy với `n = 0` nên không có chữ số nào được xét; hàm trả giá trị mặc định có thể sai (ở đây `False` đúng tình cờ với `x = 0`, nhưng cách tư duy "bỏ qua số 0" dễ gây lỗi ở bài khác). Hãy xử lý `n = 0` bằng `while True ... break` hoặc chuỗi.

````python
def wrong_digit_count_zero(n):
    count = 0
    while n > 0:                                                  # SAI: n = 0 có 1 chữ số nhưng vòng không chạy
        n //= 10
        count += 1
    return count

print(wrong_digit_count_zero(0), len(str(0)))
assert wrong_digit_count_zero(0) == 0 and len(str(0)) == 1
assert Solution().validDigit(0, 0) is False and Solution().validDigit(0, 4) is False
````

## 14. Mẹo làm bài nhanh & nhận dạng pattern

- Bài về chữ số của số nguyên -> **đổi sang chuỗi** cho nhanh; cần `O(1)` bộ nhớ thì dùng `divmod`.
- Luôn chú ý **kiểu dữ liệu**: ký tự `'2'` khác số `2`. Đổi `x` thành `str(x)` rồi mới so sánh với ký tự.
- `s.find(c) > 0` là mẹo gọn cho "có `c` và lần đầu không ở vị trí 0".
- Test biên: `n = 0`, số một chữ số, `x = 0` (chữ số `0` không thể đứng đầu trừ chính số `0`), số có nhiều chữ số `x`.
- Khi `n` nhỏ có thể kiểm tra **toàn bộ** miền với bộ chuẩn độc lập.

````python
# Bài 1, 2
assert Solution().validDigit(7007, 0) is True and Solution().validDigit(7007, 7) is False
assert sum(Solution().validDigit(n, 5) for n in range(10, 100)) == 8

# Bài 3
def count_digit(n, x):
    return str(n).count(str(x))
assert count_digit(7007, 0) == 2 and count_digit(0, 0) == 1 and count_digit(123, 9) == 0

# Bài 4
def valid_digit_exactly_once(n, x):
    s, d = str(n), str(x)
    return s[0] != d and s.count(d) == 1
assert valid_digit_exactly_once(101, 0) is True and valid_digit_exactly_once(1001, 0) is False
assert valid_digit_exactly_once(232, 2) is False

# Bài 5
def count_valid(limit, x):
    return sum(1 for n in range(limit + 1) if Solution().validDigit(n, x))
assert count_valid(99, 5) == 8 and count_valid(9, 3) == 0
assert count_valid(1000, 0) == sum(1 for n in range(1001) if "0" in str(n)[1:])
print("đáp án bài tập OK")
````
