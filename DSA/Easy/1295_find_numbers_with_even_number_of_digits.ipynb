**Constraints:** `1 <= nums.length <= 500`, `1 <= nums[i] <= 10^5`.

Chú ý: đề hỏi **số chữ số** chẵn, không phải **giá trị** chẵn. `12` có 2 chữ số (chẵn) — đúng; `2` là số chẵn nhưng có **1** chữ số (lẻ) — không tính.

````python
nums = [12, 345, 2, 6, 7896]
for x in nums:
    d = len(str(x))
    print(f"{x:>5}: {d} chữ số -> {'CHẴN' if d % 2 == 0 else 'lẻ'}")
````

### 2.4. Đếm chữ số bằng `// 10`

Mỗi lần `x //= 10` bỏ đi một chữ số cuối. Đếm số lần cho tới khi `x == 0`.

````python
x = 7896
cnt = 0
while x > 0:
    x //= 10
    cnt += 1
    print("x =", x, "| đã bỏ", cnt, "chữ số")
print("số chữ số:", cnt)
````

## 3. Cách 1 — Chuỗi (dễ nghĩ nhất)

````python
def find_numbers_str(nums: list[int]) -> int:
    count = 0
    for x in nums:
        if len(str(x)) % 2 == 0:
            count += 1
    return count

print(find_numbers_str([12, 345, 2, 6, 7896]))     # 2
print(find_numbers_str([555, 901, 482, 1771]))     # 1
````

**Độ phức tạp:** gọi `n = len(nums)`, `D` = số chữ số tối đa (`D ≤ 6`). Thời gian `O(n · D)`, bộ nhớ `O(D)` cho chuỗi tạm.

### Viết gọn một dòng

````python
def find_numbers_str_short(nums: list[int]) -> int:
    return sum(len(str(x)) % 2 == 0 for x in nums)

print(find_numbers_str_short([12, 345, 2, 6, 7896]))   # 2
````

## 4. Cách 2 — Đếm chữ số bằng toán học (không dùng chuỗi)

Đúng như hint của đề: "chia cho 10 nhiều lần để biết số chữ số". Tách thành hàm phụ `count_digits` để dễ đọc và test riêng.

````python
def count_digits(x: int) -> int:
    cnt = 0
    while x > 0:
        x //= 10
        cnt += 1
    return cnt


def find_numbers_math(nums: list[int]) -> int:
    return sum(1 for x in nums if count_digits(x) % 2 == 0)

print([count_digits(x) for x in [1, 10, 99, 100, 100000]])   # [1, 2, 2, 3, 6]
print(find_numbers_math([12, 345, 2, 6, 7896]))              # 2
````

**Lưu ý:** `count_digits(0)` trả `0` (vòng `while` không chạy) — sai về mặt toán (0 có 1 chữ số) nhưng đề đảm bảo `nums[i] >= 1` nên không gặp. Nếu tái sử dụng cho bài khác, nhớ xử lý `x == 0`.

## 5. Cách 3 — Logarit `log10`

````python
import math

def find_numbers_log(nums: list[int]) -> int:
    return sum(1 for x in nums if (math.floor(math.log10(x)) + 1) % 2 == 0)

print(find_numbers_log([12, 345, 2, 6, 7896]))   # 2
print(find_numbers_log([555, 901, 482, 1771]))   # 1
````

### Cảnh báo sai số số thực

`log10` trả về `float`. Với số **rất lớn** sát lũy thừa của 10, sai số làm tròn có thể khiến `floor` lệch 1. Trong miền của đề (`≤ 10^5`) thì an toàn (ta sẽ vét cạn kiểm chứng ở mục test), nhưng với số khổng lồ thì **không** nên tin `log10`.

````python
big = 10**16 - 1                       # 9999999999999999: 16 chữ số
print(len(str(big)))                   # 16
print(math.log10(big))                 # 16.0 (!) do float không phân biệt nổi
print(math.floor(math.log10(big)) + 1) # 17 -> SAI
````

## 6. Nút thắt — có cần "đếm" chữ số không?

Ba cách trên đều tính **chính xác** số chữ số rồi mới xét chẵn lẻ. Nhưng constraints nói `1 <= x <= 10^5`, tức số chữ số chỉ là 1..6. Các khoảng giá trị:

| số chữ số | khoảng giá trị | chẵn? |
|---:|---|---|
| 1 | `1 .. 9` | |
| 2 | `10 .. 99` | ✔ |
| 3 | `100 .. 999` | |
| 4 | `1000 .. 9999` | ✔ |
| 5 | `10000 .. 99999` | |
| 6 | `100000` | ✔ |

## 7. Insight — so sánh khoảng giá trị

`x` có số chữ số chẵn ⇔ `10 <= x <= 99` **hoặc** `1000 <= x <= 9999` **hoặc** `x == 100000`. Chỉ vài phép so sánh, không vòng lặp con, không chuỗi.

Python cho viết **so sánh nối chuỗi** `10 <= x <= 99` (nghĩa là `10 <= x and x <= 99`).

````python
x = 57
print(10 <= x <= 99)          # True
print(10 <= x and x <= 99)    # cách viết dài tương đương
print(1000 <= 999 <= 9999)    # False
````

## 8. Cách 4 — So sánh khoảng (lời giải chính, tận dụng constraints)

````python
def has_even_digits_range(x: int) -> bool:
    return 10 <= x <= 99 or 1000 <= x <= 9999 or x == 100000


def find_numbers_range(nums: list[int]) -> int:
    return sum(1 for x in nums if has_even_digits_range(x))

print(find_numbers_range([12, 345, 2, 6, 7896]))   # 2
print(find_numbers_range([555, 901, 482, 1771]))   # 1
````

### Tổng quát hóa (không phụ thuộc giới hạn `10^5`)

Ý tưởng khoảng vẫn tổng quát được: số chữ số chẵn `2k` ⇔ `10^(2k-1) <= x < 10^(2k)`. Ta có thể "nhảy" theo **lũy thừa của 100**: chia `x` cho 100 cho tới khi `x < 100`, rồi xem phần còn lại có 2 chữ số (`>= 10`) hay 1 chữ số.

````python
def has_even_digits_div100(x: int) -> bool:
    while x >= 100:
        x //= 100             # bỏ 2 chữ số một lần: tính chẵn lẻ của số chữ số không đổi
    return x >= 10            # còn 2 chữ số -> chẵn; còn 1 chữ số -> lẻ

print([has_even_digits_div100(x) for x in [5, 12, 345, 7896, 100000, 10**15]])
# [False, True, False, True, True, True]
print(len(str(10**15)), has_even_digits_div100(10**15))   # 16 True: 10**15 = số 1 và 15 số 0
````

Chú ý `10**15` có **16** chữ số (số 1 và 15 số 0) → chẵn → `True`. Người mới hay đếm nhầm thành 15 — luôn kiểm tra bằng chạy thật.

Bỏ 2 chữ số mỗi lần không làm đổi tính chẵn lẻ của số chữ số — đó là **bất biến** giúp lời giải đúng. Số vòng chỉ còn một nửa.

## 9. Trace từng bước

`nums = [12, 345, 2, 6, 7896, 100000]` với Cách 4 và Cách 2:

| x | `count_digits` | chẵn? | thuộc khoảng nào | đếm |
|---:|---:|---|---|---:|
| 12 | 2 | ✔ | `10..99` | 1 |
| 345 | 3 | | — | 1 |
| 2 | 1 | | — | 1 |
| 6 | 1 | | — | 1 |
| 7896 | 4 | ✔ | `1000..9999` | 2 |
| 100000 | 6 | ✔ | `== 100000` | 3 |

````python
def trace_find_numbers(nums):
    count = 0
    for x in nums:
        d = count_digits(x)
        ok = has_even_digits_range(x)
        assert ok == (d % 2 == 0)
        if ok:
            count += 1
        print(f"x={x:>6} | digits={d} | even? {ok!s:<5} | count={count}")
    return count

trace_find_numbers([12, 345, 2, 6, 7896, 100000])
````

### Giải thích định dạng f-string dùng trong trace

- `{x:>6}`: căn **phải** trong ô rộng 6 ký tự — các số thẳng cột dù dài ngắn khác nhau.
- `{ok!s:<5}`: `!s` đổi giá trị sang chuỗi bằng `str()`, `:<5` căn **trái** rộng 5 (vì `True` 4 ký tự, `False` 5 ký tự).
- `assert ok == (d % 2 == 0)` bên trong trace là một **kiểm tra chéo**: hai cách (khoảng giá trị và đếm chữ số) phải cho cùng kết luận ở mọi bước.

````python
for v in [7, 1234, 100000]:
    print(f"[{v:>6}] [{v:<6}] [{v:^8}]")
print(f"[{True!s:<5}] [{False!s:<5}]")
````

### Trace hàm `count_digits` từng vòng

````python
def trace_count_digits(x):
    cnt = 0
    print(f"bắt đầu x={x}")
    while x > 0:
        x //= 10
        cnt += 1
        print(f"  vòng {cnt}: x -> {x}")
    print("  => số chữ số =", cnt)
    return cnt

trace_count_digits(7896)
````

## 10. Code nộp LeetCode

````python
class Solution:
    def findNumbers(self, nums: list[int]) -> int:
        count = 0
        for x in nums:
            if len(str(x)) % 2 == 0:
                count += 1
        return count

print(Solution().findNumbers([12, 345, 2, 6, 7896]))   # 2
print(Solution().findNumbers([555, 901, 482, 1771]))   # 1
````

Bản tận dụng constraints (không chuỗi, không vòng con):

```python
class Solution:
    def findNumbers(self, nums: list[int]) -> int:
        return sum(10 <= x <= 99 or 1000 <= x <= 9999 or x == 100000 for x in nums)
```

## 11. Vì sao đúng?

- **Cách 1/2/3:** mỗi cách tính đúng số chữ số `d` của `x >= 1` (chuỗi thập phân không có số 0 thừa ở đầu; `// 10` bỏ đúng 1 chữ số mỗi vòng; `⌊log₁₀ x⌋ + 1` đúng với số thực chính xác, và đã kiểm chứng trong miền đề). Biến `count` tăng đúng khi `d` chẵn.
- **Cách 4:** bảng ở mục 6 liệt kê **toàn bộ** miền `[1, 10^5]` thành 6 khoảng rời nhau theo số chữ số; ta chọn đúng các khoảng có số chữ số chẵn.
- **Invariant vòng ngoài:** sau khi xét `i` phần tử đầu, `count` = số phần tử trong `nums[0..i-1]` có số chữ số chẵn.

## 12. Độ phức tạp và so sánh

Gọi `n = len(nums)`, `D` = số chữ số tối đa (≤ 6).

| Cách | Thời gian | Bộ nhớ phụ | Ghi chú |
|---|---|---|---|
| 1. `len(str(x))` | `O(n · D)` | `O(D)` | dễ nhất |
| 2. `// 10` lặp | `O(n · D)` | `O(1)` | không dùng chuỗi |
| 3. `log10` | `O(n)` | `O(1)` | cẩn thận sai số float với số lớn |
| 4. So sánh khoảng | `O(n)` | `O(1)` | dựa vào constraints |
| 4'. Chia 100 lặp | `O(n · D/2)` | `O(1)` | tổng quát, nhanh gấp đôi Cách 2 |

Vì `D ≤ 6` là hằng số, cả 4 cách thực chất đều `O(n)`.

### Lỗi 4: Đếm chữ số bằng `/ 10` (chia thực)

`x / 10` không bao giờ về đúng `0` mà giảm dần như `0.7896 → 0.07896 → …` — vòng lặp chạy hàng trăm lần cho tới khi số thực **tràn dưới** về `0.0`.

````python
def wrong_true_division_count(x):
    cnt = 0
    while x > 0:
        x = x / 10                   # SAI
        cnt += 1
    return cnt

print("sai :", wrong_true_division_count(7896), "| đúng:", count_digits(7896))
````

## 15. Mẹo nhận dạng pattern và biến thể

**Dấu hiệu:**
- "Số chữ số" → `len(str(x))` hoặc vòng `// 10`.
- Constraints nhỏ, cố định → cân nhắc **liệt kê khoảng** / **bảng tra** thay vì tính toán.
- "Đếm phần tử thỏa điều kiện" → `sum(cond for x in nums)`.

**Bài liên quan:**
- LeetCode 1281 *Subtract the Product and Sum of Digits* (bóc chữ số).
- LeetCode 2520 *Count the Digits That Divide a Number*.
- LeetCode 2843 *Count Symmetric Integers* (số chữ số chẵn, so nửa đầu với nửa sau).
- LeetCode 7 *Reverse Integer*, 9 *Palindrome Number*.

````python
# Mini: Count Symmetric Integers (2843) — dùng lại ý "số chữ số chẵn"
def count_symmetric(low, high):
    res = 0
    for x in range(low, high + 1):
        s = str(x)
        if len(s) % 2 == 0:
            h = len(s) // 2
            if sum(map(int, s[:h])) == sum(map(int, s[h:])):
                res += 1
    return res

print(count_symmetric(1, 100), count_symmetric(1200, 1230))   # 9 4
````

## 17. Gợi ý 3 mức

**Bài 1:** (1) Đếm chữ số từng số. (2) 1, 2, 3, 4, 5, 6. (3) Các số chẵn trong đó.

**Bài 2:** (1) Dùng `count_digits`. (2) Điều kiện `% 2 == 1`. (3) Hoặc `len(nums) - số_chẵn`.

**Bài 3:** (1) Số âm: lấy `abs`. (2) `0` có 1 chữ số — xử lý riêng. (3) Dùng vòng `// 10`.

**Bài 4:** (1) `dict` + `setdefault` hoặc `defaultdict(list)`. (2) Key = `len(str(x))`. (3) `groups.setdefault(d, []).append(x)`.

**Bài 5:** (1) Số có `k` chữ số nằm trong `[10^(k-1), 10^k - 1]`. (2) Giao hai đoạn `[a, b]` và `[c, d]` là `[max(a,c), min(b,d)]`, độ dài `max(0, min - max + 1)`. (3) Cộng cho mọi `k` chẵn từ 2 tới số chữ số của `hi`.

## 18. Đáp án tham khảo

````python
ans_ex1 = 3     # 70 (2), 7000 (4), 100000 (6)


def ans_find_numbers_odd(nums):
    return sum(1 for x in nums if count_digits(x) % 2 == 1)


def ans_count_digits_any(x):
    x = abs(x)
    if x == 0:
        return 1
    cnt = 0
    while x > 0:
        x //= 10
        cnt += 1
    return cnt


def ans_group_by_digits(nums):
    groups = {}
    for x in nums:
        groups.setdefault(len(str(x)), []).append(x)
    return groups


def ans_count_even_digit_in_range(lo, hi):
    total = 0
    k = 2
    while 10 ** (k - 1) <= hi:
        a, b_ = max(lo, 10 ** (k - 1)), min(hi, 10 ** k - 1)
        total += max(0, b_ - a + 1)
        k += 2
    return total


assert ans_ex1 == sum(len(str(x)) % 2 == 0 for x in [7, 70, 700, 7000, 70000, 100000])
assert ans_find_numbers_odd([12, 345, 2, 6, 7896]) == 3
assert ans_count_digits_any(0) == 1
assert ans_count_digits_any(-345) == 3
assert ans_count_digits_any(100000) == 6
assert ans_group_by_digits([12, 345, 2, 6, 7896]) == {2: [12], 3: [345], 1: [2, 6], 4: [7896]}
assert ans_count_even_digit_in_range(1, 100) == 90
assert ans_count_even_digit_in_range(1, 10**5) == 90 + 9000 + 1
random.seed(3)
for _ in range(300):
    lo = random.randint(1, 20000)
    hi = random.randint(lo, 30000)
    assert ans_count_even_digit_in_range(lo, hi) == sum(len(str(x)) % 2 == 0 for x in range(lo, hi + 1))
for x in range(-2000, 2001):
    assert ans_count_digits_any(x) == len(str(abs(x)))
print("số có số chữ số chẵn trong [1, 10^18]:", ans_count_even_digit_in_range(1, 10**18))
print("Đáp án bài tập: tất cả assert PASS")
````

`groups.setdefault(key, [])` trả về list đã có của `key`, hoặc tạo list rỗng mới nếu chưa có — rồi ta `.append(x)` vào đó.
