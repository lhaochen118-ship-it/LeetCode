## 2. Nền tảng Python

### 2.1. Toán tử `in` với chuỗi

`ch in "aeiou"` trả `True` nếu `ch` là một **chuỗi con** của `"aeiou"`. Với `ch` là một ký tự, đó chính là câu hỏi "có phải nguyên âm không?". Python phải dò từng ký tự của `"aeiou"` → tối đa 5 phép so sánh.

````python
VOWELS_STR = "aeiou"
for ch in ["a", "b", "u", "y"]:
    print(repr(ch), "là nguyên âm?", ch in VOWELS_STR)
print("Cẩn thận: 'ae' in 'aeiou' =", "ae" in VOWELS_STR, "(chuỗi con, không phải 1 ký tự)")
````

### 2.2. `set` — tập hợp, tra cứu trung bình `O(1)`

`set` là tập các phần tử **không trùng**, không có thứ tự. Nó dùng **bảng băm** (hash table): `x in my_set` tính băm của `x` rồi nhảy thẳng tới vị trí → trung bình `O(1)`, không phụ thuộc kích thước tập.

Tạo set: `{"a", "e"}` hoặc `set("aeiou")` (tách chuỗi thành từng ký tự). Chú ý `{}` là **dict rỗng**, set rỗng phải viết `set()`.

````python
VOWELS = set("aeiou")
print(VOWELS, type(VOWELS).__name__, "| số phần tử:", len(VOWELS))
print("'e' in VOWELS:", "e" in VOWELS, "| 'z' in VOWELS:", "z" in VOWELS)
print("set('banana') =", set("banana"), "(tự loại trùng)")
print("type({}) =", type({}).__name__, "| type(set()) =", type(set()).__name__)
````

### 2.3. List comprehension và generator expression

- `[ch for ch in s if ch not in VOWELS]` → tạo **list** các ký tự không phải nguyên âm.
- `(ch for ch in s if ...)` → **generator**: sinh từng phần tử khi cần, không tạo list trung gian.

`"".join(...)` nhận được cả hai.

````python
s = "hello"
lst = [ch for ch in s if ch not in VOWELS]
print("list:", lst, "->", "".join(lst))
gen = (ch for ch in s if ch not in VOWELS)
print("generator:", gen)
print("join generator:", "".join(gen))
````

Đọc list comprehension như câu tiếng Việt: "**lấy** `ch` **với mỗi** `ch` **trong** `s` **nếu** `ch` không phải nguyên âm". Nó tương đương:

```python
lst = []
for ch in s:
    if ch not in VOWELS:
        lst.append(ch)
```

## 3. Cách 1 — Duyệt và cộng dồn chuỗi (trực giác)

````python
def remove_vowels_concat(s: str) -> str:
    result = ""
    for ch in s:
        if ch not in "aeiou":
            result += ch
    return result

print(remove_vowels_concat("leetcodeisacommunityforcoders"))
print(repr(remove_vowels_concat("aeiou")))
````

**Nút thắt:** chuỗi Python bất biến, nên `result += ch` về nguyên tắc tạo chuỗi mới và sao chép cả `result` → trường hợp xấu `O(n²)` (CPython thường tối ưu được, nhưng không đảm bảo). Với `n <= 1000` vẫn nhanh, nhưng ta nên học thói quen chuẩn: **gom vào list, `join` một lần**.

Hãy đo thử trên chuỗi lớn để thấy khác biệt (con số cụ thể phụ thuộc máy):

````python
import timeit
big = "leetcodeisacommunityforcoders" * 3000      # ~87 000 ký tự

def by_join(s):
    return "".join([ch for ch in s if ch not in "aeiou"])

t1 = timeit.timeit(lambda: remove_vowels_concat(big), number=5)
t2 = timeit.timeit(lambda: by_join(big), number=5)
print(f"+= : {t1:.4f}s | join: {t2:.4f}s (5 lần chạy)")
assert remove_vowels_concat(big) == by_join(big)
````

`timeit.timeit(hàm, number=k)` chạy hàm `k` lần và trả tổng thời gian (giây). `lambda: ...` là **hàm vô danh** một dòng, ở đây dùng để "gói" lời gọi có tham số.

## 4. Insight — đây là bài **lọc**, và điều kiện lọc nên tra bằng `set`

- Kết quả = các ký tự của `s` **giữ nguyên thứ tự**, bỏ đi những ký tự thỏa "là nguyên âm".
- Mọi ngôn ngữ đều có mẫu "filter": duyệt một lần, quyết định giữ/bỏ từng phần tử → `O(n)`.
- Điều kiện "là nguyên âm" kiểm bằng `set` cho tra cứu trung bình `O(1)` — với 5 nguyên âm thì `in "aeiou"` cũng nhanh, nhưng khi tập điều kiện lớn (ví dụ 1000 từ cấm), `set` khác biệt rõ.

### 4.1. Thí nghiệm: `in` trên list lớn và trên set lớn

Với 5 nguyên âm, khác biệt không đáng kể. Nhưng hãy tưởng tượng điều kiện lọc là "thuộc danh sách 10 000 mã bị cấm". `x in list` phải dò tuần tự (`O(m)`), còn `x in set` băm rồi nhảy thẳng (`O(1)` trung bình). Thí nghiệm dưới đây cho thấy điều đó (số đo phụ thuộc máy, nhưng chênh lệch rất rõ).

````python
import timeit
banned_list = list(range(10_000))
banned_set = set(banned_list)
queries = list(range(9_000, 11_000))      # nửa có, nửa không

t_list = timeit.timeit(lambda: sum(q in banned_list for q in queries), number=3)
t_set = timeit.timeit(lambda: sum(q in banned_set for q in queries), number=3)
print(f"in list: {t_list:.4f}s | in set: {t_set:.5f}s")
print("Kết quả giống nhau:", sum(q in banned_list for q in queries) == sum(q in banned_set for q in queries))
````

Bài học: **dựng `set` một lần ở ngoài vòng lặp**, rồi tra cứu bên trong. Đừng dựng lại `set(...)` ở mỗi vòng — chi phí dựng là `O(m)` mỗi lần.

## 5. Cách 2 — List comprehension + `join` (lời giải chính)

Ghép các mảnh:
1. `VOWELS = set("aeiou")` — tập nguyên âm.
2. Lọc: `ch for ch in s if ch not in VOWELS`.
3. Nối: `"".join(...)`.

````python
def remove_vowels(s: str) -> str:
    vowels = set("aeiou")
    return "".join(ch for ch in s if ch not in vowels)

print(remove_vowels("leetcodeisacommunityforcoders"))   # ltcdscmmntyfrcdrs
print(repr(remove_vowels("aeiou")))                     # ''
print(remove_vowels("xyz"))                             # xyz
````

## 6. TRACE từng bước

Ví dụ `s = "education"`:

| i | ch | nguyên âm? | quyết định | kết quả tạm |
|---:|:---:|:---:|---|---|
| 0 | e | Có | bỏ | `""` |
| 1 | d | Không | giữ | `"d"` |
| 2 | u | Có | bỏ | `"d"` |
| 3 | c | Không | giữ | `"dc"` |
| 4 | a | Có | bỏ | `"dc"` |
| 5 | t | Không | giữ | `"dct"` |
| 6 | i | Có | bỏ | `"dct"` |
| 7 | o | Có | bỏ | `"dct"` |
| 8 | n | Không | giữ | `"dctn"` |

Kết quả `"dctn"`.

````python
def trace_remove_vowels(s: str) -> str:
    vowels = set("aeiou")
    kept = []
    for i, ch in enumerate(s):
        is_v = ch in vowels
        if not is_v:
            kept.append(ch)
        print(f"i={i}, ch={ch!r}, nguyên âm? {is_v!s:<5} -> {'bỏ ' if is_v else 'giữ'} | tạm: {''.join(kept)!r}")
    result = "".join(kept)
    print("Kết quả:", repr(result))
    return result

_ = trace_remove_vowels("education")
````

## 7. Code nộp LeetCode

## 8. Cách 3 — Hàm `filter`

`filter(hàm, dãy)` giữ lại các phần tử mà `hàm(phần tử)` trả `True`. Kết quả là một **iterator** (sinh dần), có thể đưa thẳng vào `join`.

````python
print(list(filter(lambda x: x % 2 == 0, [1, 2, 3, 4, 5])))   # giữ số chẵn

def remove_vowels_filter(s: str) -> str:
    return "".join(filter(lambda ch: ch not in "aeiou", s))

print(remove_vowels_filter("leetcodeisacommunityforcoders"))
````

## 9. Cách 4 — `str.translate` (xóa ký tự bằng bảng dịch)

`str.maketrans(x, y, z)` tạo bảng dịch: ký tự trong `x` đổi thành ký tự tương ứng trong `y`, ký tự trong `z` bị **xóa**. Ta chỉ cần tham số thứ 3: `str.maketrans("", "", "aeiou")`.

`s.translate(table)` áp bảng lên toàn chuỗi trong một lượt — được cài đặt bằng C nên thường **nhanh nhất** cho thao tác ký tự đơn.

````python
table = str.maketrans("", "", "aeiou")
print(table)                                    # dict: mã Unicode -> None (xóa)
print("ord('a') =", ord("a"))
print("hello world".translate(table))

def remove_vowels_translate(s: str) -> str:
    return s.translate(str.maketrans("", "", "aeiou"))

print(remove_vowels_translate("leetcodeisacommunityforcoders"))
````

`ord(ch)` trả về **mã Unicode** (số nguyên) của ký tự; bảng dịch dùng mã này làm key. Giá trị `None` nghĩa là "xóa ký tự".

## 10. Cách 5 — Biểu thức chính quy `re.sub`

Trong regex, `[aeiou]` là **lớp ký tự** (character class): khớp **một** ký tự bất kỳ trong ngoặc vuông. `re.sub(r"[aeiou]", "", s)` thay mọi nguyên âm bằng chuỗi rỗng, tức là xóa.

````python
import re

print(re.findall(r"[aeiou]", "education"))     # các đoạn khớp
def remove_vowels_regex(s: str) -> str:
    return re.sub(r"[aeiou]", "", s)

print(remove_vowels_regex("leetcodeisacommunityforcoders"))
print("Không phân biệt hoa thường:", re.sub(r"[aeiou]", "", "EducAtion", flags=re.IGNORECASE))
````

## 11. Vì sao đúng? — Độ phức tạp — So sánh

**Tính đúng:** invariant của vòng lọc — sau khi xét `k` ký tự đầu, `kept` chứa đúng các ký tự **không phải nguyên âm** của `s[:k]`, theo đúng thứ tự xuất hiện. Bước `k+1` chỉ thêm `s[k]` khi nó không phải nguyên âm, nên invariant được giữ; `k = n` cho đáp án. Không ký tự phụ âm nào bị mất (mọi ký tự được xét đúng một lần) và không nguyên âm nào lọt qua (mỗi lần thêm đều qua kiểm tra).

Gọi `n = len(s)`, `V = 5` số nguyên âm.

| Cách | Time | Space phụ | Nhận xét |
|---|---|---|---|
| 1. `+=` | `O(n)` thường, `O(n²)` xấu nhất | `O(n)` | Dễ nghĩ, không nên cho chuỗi lớn |
| 2. comprehension + `join` + `set` | `O(n)` | `O(n)` | **Chuẩn, dễ đọc** |
| 3. `filter` | `O(n·V)` với `in "aeiou"` (V hằng số → `O(n)`) | `O(n)` | Phong cách hàm |
| 4. `translate` | `O(n)` | `O(n)` | Nhanh nhất thực tế |
| 5. `re.sub` | `O(n)` | `O(n)` | Linh hoạt cho mẫu phức tạp |

`O(n)` bộ nhớ là cho chuỗi kết quả (bắt buộc vì chuỗi bất biến).

## 13. Lỗi người mới hay gặp

### Lỗi 1: Xóa phần tử khỏi list trong khi đang duyệt

````python
def wrong_remove_while_iterating(s):
    chars = list(s)
    for ch in chars:
        if ch in "aeiou":
            chars.remove(ch)          # SAI: list co lại, vòng lặp nhảy cóc
    return "".join(chars)

print("sai :", wrong_remove_while_iterating("aeiou"), "| đúng:", repr(remove_vowels("aeiou")))
print("sai :", wrong_remove_while_iterating("baab"), "| đúng:", remove_vowels("baab"))
````

**Vì sao sai:** khi xóa phần tử ở vị trí `i`, mọi phần tử sau dời lên một ô; vòng `for` tiếp tục ở vị trí `i+1` nên **bỏ qua** phần tử vừa dời vào `i`. Thêm nữa `list.remove(x)` xóa lần xuất hiện **đầu tiên** của `x`, không chắc là phần tử đang xét. **Sửa:** xây list mới (lọc), không sửa list đang duyệt.

### Lỗi 2: Dùng `replace` nhưng quên gán lại

````python
def wrong_replace(s):
    for v in "aeiou":
        s.replace(v, "")             # SAI: kết quả bị bỏ đi
    return s

def right_replace(s):
    for v in "aeiou":
        s = s.replace(v, "")         # gán lại -> đúng (5 lượt quét, O(5n))
    return s

print("sai :", wrong_replace("education"), "| đúng:", right_replace("education"))
````

## 15. Biến thể

### 15.1. Hỗ trợ cả chữ hoa

````python
def remove_vowels_any_case(s: str) -> str:
    vowels = set("aeiouAEIOU")
    return "".join(ch for ch in s if ch not in vowels)

print(remove_vowels_any_case("LeetCode Is Awesome"))
````

### 15.2. Reverse Vowels (#345) — đảo thứ tự các nguyên âm, giữ phụ âm tại chỗ

Hai con trỏ từ hai đầu: tiến `i` tới nguyên âm kế tiếp, lùi `j` tới nguyên âm trước đó, đổi chỗ. Chuỗi bất biến nên chuyển sang list trước.

````python
def reverse_vowels(s: str) -> str:
    vowels = set("aeiouAEIOU")
    chars = list(s)
    i, j = 0, len(chars) - 1
    while i < j:
        if chars[i] not in vowels:
            i += 1
        elif chars[j] not in vowels:
            j -= 1
        else:
            chars[i], chars[j] = chars[j], chars[i]    # hoán đổi
            i += 1
            j -= 1
    return "".join(chars)

print(reverse_vowels("hello"), reverse_vowels("leetcode"))   # holle leotcede
````

`a, b = b, a` là cách **hoán đổi** hai biến trong Python (đóng gói thành tuple rồi tách ra), không cần biến tạm.

## 17. Đáp án tham khảo

- **Bài 1:** `"q"` và `"prorain"`.
- **Bài 2:** `"aurora"` → bỏ a, u, o, a → `"rr"`.

````python
def answer_remove_vowels(s):
    vowels = set("aeiou"); kept = []
    for ch in s:
        if ch not in vowels:
            kept.append(ch)
    return "".join(kept)

def answer_count_vowels(s):
    d = {}
    for ch in s:
        if ch in "aeiou":
            d[ch] = d.get(ch, 0) + 1
    return d

def answer_keep_first_vowel(sentence):
    words = []
    for w in sentence.split(" "):
        seen_vowel = False
        out = []
        for ch in w:
            if ch in "aeiou":
                if not seen_vowel:
                    out.append(ch)
                    seen_vowel = True
            else:
                out.append(ch)
        words.append("".join(out))
    return " ".join(words)

def answer_remove_vowels_inplace(chars):
    write = 0
    for read in range(len(chars)):
        if chars[read] not in "aeiou":
            chars[write] = chars[read]
            write += 1
    return write

assert "".join(ch for ch in "queue" if ch not in "aeiou") == "q"
assert "programming".translate(str.maketrans("", "", "gm")) == "prorain"
assert answer_remove_vowels("aurora") == "rr"
assert answer_remove_vowels("leetcodeisacommunityforcoders") == "ltcdscmmntyfrcdrs"
assert answer_remove_vowels("aeiou") == ""
assert answer_count_vowels("education") == {"e": 1, "u": 1, "a": 1, "i": 1, "o": 1}
assert answer_count_vowels("xyz") == {}
assert answer_keep_first_vowel("hello world") == "hell world"
assert answer_keep_first_vowel("aeiou banana") == "a bann"
for _ in range(1000):
    s = "".join(rng.choice("abcdeiou") for _ in range(rng.randint(1, 20)))
    arr = list(s)
    k = answer_remove_vowels_inplace(arr)
    assert "".join(arr[:k]) == remove_vowels(s) == answer_remove_vowels(s)
print("Đáp án tham khảo: PASS toàn bộ")
````
