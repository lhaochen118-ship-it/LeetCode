## 1. Mục tiêu học

Sau notebook này bạn sẽ:

1. Hiểu đề: biến chuỗi `s` thành **palindrome (chuỗi đối xứng)** bằng **ít thao tác nhất**; nếu có nhiều đáp án thì chọn chuỗi **nhỏ nhất theo thứ tự từ điển**.

4. Hiểu **tham lam (greedy)**: mỗi cặp độc lập với các cặp khác ⇒ chọn kí tự nhỏ hơn ở từng cặp.
5. Viết 4 cách: thử cả 26 chữ cho mỗi cặp (baseline), hai con trỏ + `min`, dựng nửa trái rồi lật, và một dòng `zip(s, reversed(s))`.
6. Chứng minh vì sao vừa ít thao tác nhất vừa nhỏ nhất, kiểm bằng **vét cạn mọi palindrome** trên chuỗi ngắn.
7. Tránh các lỗi quen thuộc: luôn chép nửa trái, đổi cả hai kí tự, quên kí tự giữa khi `n` lẻ, gán trực tiếp `s[i] = ...`.

## 2. Đề bài & ví dụ

Cho chuỗi `s` gồm **chữ thường** tiếng Anh. Trong **một thao tác** bạn có thể **thay một kí tự bất kỳ** của `s` bằng **một chữ thường khác**.

Nhiệm vụ: làm cho `s` trở thành **palindrome** bằng **số thao tác ít nhất**. Nếu có nhiều palindrome đạt số thao tác ít nhất, trả về chuỗi **nhỏ nhất theo thứ tự từ điển**.

Chuỗi `a` nhỏ hơn chuỗi `b` (cùng độ dài) nếu ở vị trí đầu tiên mà hai chuỗi khác nhau, kí tự của `a` đứng **trước** trong bảng chữ cái.

| Ví dụ | s | Output | Giải thích |
|---|---|---|---|
| 1 | `"egcfe"` | `"efcfe"` | chỉ cần 1 thao tác: đổi `g` thành `f` (hoặc `f` thành `g`); chọn `efcfe` vì nhỏ hơn `egcge` |
| 2 | `"abcd"` | `"abba"` | cần 2 thao tác: đổi `d` → `a` và `c` → `b` |
| 3 | `"seven"` | `"neven"` | 1 thao tác: đổi `s` → `n` (vì `n < s`) |

Constraints: `1 <= s.length <= 1000`; `s` chỉ gồm chữ thường tiếng Anh.

````python
for s in ["egcfe", "abcd", "seven"]:
    print(s, "-> đọc ngược:", s[::-1], "| palindrome?", s == s[::-1])
print("neven" == "neven"[::-1])     # True: đáp án ví dụ 3 là palindrome
print("efcfe" < "egcge")            # True: so sánh từ điển
````

## 3. Kiến thức nền

### 3.1. Palindrome và các cặp đối xứng

Chuỗi độ dài `n` là palindrome nếu `s[i] == s[n - 1 - i]` với mọi `i`. Ta **ghép cặp** vị trí đầu–cuối: `(0, n-1)`, `(1, n-2)`, ... Có `n // 2` cặp; nếu `n` lẻ thì còn **một kí tự ở giữa** (vị trí `n // 2`) không có cặp, và nó **luôn tự đối xứng**.

Ví dụ `n = 5`: cặp `(0, 4)`, `(1, 3)`; kí tự giữa ở chỉ số `2`.

````python
s = "egcfe"
n = len(s)
for i in range(n // 2):
    print(f"cặp ({i}, {n - 1 - i}): {s[i]!r} và {s[n - 1 - i]!r}")
print("kí tự giữa:", s[n // 2] if n % 2 == 1 else "(không có)")
````

### 3.2. Chỉ số âm, `s[::-1]`, `reversed`, `zip`

- `s[-1]` là kí tự cuối, `s[n - 1 - i]` là phản chiếu của `s[i]`.
- `s[::-1]` là chuỗi đảo ngược; `reversed(s)` là iterator đi ngược.
- `zip(a, b)` ghép từng cặp phần tử cùng vị trí: `zip("abc", "xyz")` cho `('a','x'), ('b','y'), ('c','z')`. Vậy `zip(s, reversed(s))` cho đúng các cặp phản chiếu **(kể cả hai chiều)**.

````python
s = "abcd"
print(s[-1], s[3], s[len(s) - 1 - 0])      # d d d
print(s[::-1], list(reversed(s)))          # dcba ['d', 'c', 'b', 'a']
print(list(zip(s, reversed(s))))           # [('a','d'), ('b','c'), ('c','b'), ('d','a')]
````

### 3.3. So sánh kí tự và chuỗi: `min`, `<`

Kí tự được so sánh theo mã (bảng chữ cái đã sắp theo thứ tự): `'a' < 'b' < ... < 'z'`. `min('s', 'n')` là `'n'`. So sánh chuỗi so từng vị trí từ trái sang phải (thứ tự từ điển).

````python
print(min("s", "n"), max("s", "n"))           # n s
print("a" < "b", ord("a"), ord("z"))          # True 97 122
print("abba" < "abcb", "efcfe" < "egcge")     # True True
````

### 3.4. Chuỗi bất biến: không `s[i] = 'x'`; dùng `list` rồi `"".join`

Chuỗi Python **không sửa tại chỗ** được. Cách sửa: đổi sang `list(s)`, sửa các phần tử, rồi ghép lại bằng `"".join(...)`.

````python
s = "seven"
try:
    s[0] = "n"
except TypeError as e:
    print("Lỗi:", e)

arr = list(s)       # ['s', 'e', 'v', 'e', 'n']
arr[0] = "n"
print("".join(arr))     # neven
````

### 3.5. Hai con trỏ và `while i < j`

Hai con trỏ `i = 0`, `j = n - 1` đi vào giữa: mỗi bước xét cặp `(s[i], s[j])` rồi `i += 1; j -= 1`. Dừng khi `i >= j` (đã gặp nhau, hoặc chỉ còn kí tự giữa).

````python
s = "egcfe"
i, j = 0, len(s) - 1
while i < j:
    print(f"i={i}, j={j}: {s[i]} vs {s[j]}")
    i += 1
    j -= 1
print("dừng với i =", i, "j =", j)
````

## 4. Ý tưởng nền tảng / cách nghĩ trực giác

Palindrome nghĩa là **mỗi cặp đối xứng phải bằng nhau**. Hai cặp khác nhau chạm vào các vị trí khác nhau nên **không ảnh hưởng nhau**. Do đó bài toán tách thành nhiều bài toán nhỏ, mỗi bài nhỏ là một cặp `(a, b)`:

- Nếu `a == b`: đã đối xứng, **không cần làm gì** (0 thao tác).
- Nếu `a != b`: cần **ít nhất 1** thao tác (phải đổi ít nhất một trong hai). Một thao tác là đủ: đổi một kí tự thành kí tự kia.

Tính tay `s = "abcd"`:

| cặp | kí tự | cần ít nhất | chọn chữ chung |
|---|---|---|---|
| (0, 3) | a, d | 1 thao tác | `a` hoặc `d` ⇒ nhỏ hơn: `a` |
| (1, 2) | b, c | 1 thao tác | `b` hoặc `c` ⇒ nhỏ hơn: `b` |

Ghép lại: vị trí 0 và 3 là `a`, vị trí 1 và 2 là `b` ⇒ `abba`, tổng 2 thao tác. Đúng ví dụ 2.

````python
s = "abcd"
res = list(s)
ops = 0
for i in range(len(s) // 2):
    a, b = s[i], s[len(s) - 1 - i]
    if a != b:
        ops += 1
        c = min(a, b)
        res[i] = res[len(s) - 1 - i] = c
print("".join(res), "số thao tác:", ops)    # abba 2
````

## 5. Cách 1 — Brute Force / baseline: thử cả 26 chữ cho mỗi cặp

**Ý tưởng:** với mỗi cặp `(a, b)`, **thử từng chữ cái** `c` từ `'a'` đến `'z'` làm chữ chung. Chi phí: `cost(c) = (a != c) + (b != c)` (số kí tự phải đổi). Chọn `c` có `cost` nhỏ nhất, hoà thì chọn `c` nhỏ hơn. Cách này **không giả định** "chỉ chọn một trong hai chữ" nên rất trung thực với đề.

Nếu `n` lẻ thì kí tự giữa giữ nguyên.

````python
import string

def smallest_palindrome_try26(s: str) -> str:
    n = len(s)
    res = list(s)
    for i in range(n // 2):
        a, b = s[i], s[n - 1 - i]
        best_c, best_cost = None, 3
        for c in string.ascii_lowercase:          # thử 26 chữ, theo thứ tự a -> z
            cost = (a != c) + (b != c)
            if cost < best_cost:                  # chỉ thay khi nghiêm ngặt tốt hơn -> hoà giữ chữ nhỏ hơn
                best_c, best_cost = c, cost
        res[i] = res[n - 1 - i] = best_c
    return "".join(res)

print(smallest_palindrome_try26("egcfe"), smallest_palindrome_try26("abcd"), smallest_palindrome_try26("seven"))
````

**Độ phức tạp:** `n / 2` cặp × 26 chữ ⇒ Time `O(26 · n) = O(n)` (hằng số 26 lớn), Space `O(n)` cho `list` kết quả.

**Nút thắt:** vòng 26 chữ thật ra thừa — chữ tốt nhất luôn là **một trong hai chữ của cặp** (mục 10). Nhận ra điều đó ta bỏ vòng 26 chữ và chỉ cần `min(a, b)`.

## 6. Ý tưởng thuật toán / bước đột phá

Với một cặp `(a, b)`:

- Nếu chọn chữ chung là **một trong hai** chữ ⇒ tốn `0` thao tác (nếu `a == b`) hoặc đúng `1` thao tác (nếu khác nhau).
- Nếu chọn chữ **khác cả `a` và `b`** ⇒ tốn `2` thao tác nên **không phải tối ưu** khi `a != b` (và tốn 2 khi `a == b` thay vì 0).

Vậy chữ chung phải thuộc `{a, b}` và trong hai chữ đó, chọn chữ **nhỏ hơn** để chuỗi nhỏ nhất: **`min(a, b)`**.

Công thức toàn chuỗi gọn đến mức bất ngờ: kí tự kết quả ở vị trí `i` là `min(s[i], s[n - 1 - i])`. (Với vị trí giữa `i = n-1-i`, `min(x, x) = x` nên không cần xử lý riêng!)

````python
s = "seven"
n = len(s)
print([min(s[i], s[n - 1 - i]) for i in range(n)])           # ['n', 'e', 'v', 'e', 'n']
print("".join(min(s[i], s[n - 1 - i]) for i in range(n)))     # neven
````

## 7. Các cách tiếp theo

Ngoài Cách 1 (thử 26 chữ), ta xem 3 cách nữa: (2) hai con trỏ + `min`, (3) dựng nửa trái rồi lật, (4) một dòng với `zip`.

### Cách 2 — Hai con trỏ + `min` (sửa trên `list`)

Giải thích từng đoạn:

- `arr = list(s)`: chuỗi bất biến nên sửa trên list.
- `i, j = 0, len(s) - 1`: hai đầu.
- `while i < j`: còn cặp để xét (khi `i == j` là kí tự giữa, không cần xét).
- `c = min(arr[i], arr[j])`: chữ nhỏ hơn của cặp.
- `arr[i] = arr[j] = c`: gán cho cả hai đầu (gán nếu bằng nhau thì vô hại).
- `i += 1; j -= 1`: vào trong.
- `"".join(arr)`: ghép lại.

````python
def smallest_palindrome_two_pointers(s: str) -> str:
    arr = list(s)
    i, j = 0, len(arr) - 1
    while i < j:
        c = min(arr[i], arr[j])
        arr[i] = arr[j] = c
        i += 1
        j -= 1
    return "".join(arr)

print(smallest_palindrome_two_pointers("egcfe"), smallest_palindrome_two_pointers("abcd"), smallest_palindrome_two_pointers("seven"))
````

````python
class SolutionTwoPointers:
    def makeSmallestPalindrome(self, s: str) -> str:
        arr = list(s)
        i, j = 0, len(arr) - 1
        while i < j:
            c = min(arr[i], arr[j])
            arr[i] = arr[j] = c
            i += 1
            j -= 1
        return "".join(arr)

print(SolutionTwoPointers().makeSmallestPalindrome("abcd"))
````

**Độ phức tạp:** Time `O(n)` (mỗi cặp một bước), Space `O(n)` (list kí tự).

### Cách 3 — Dựng nửa trái rồi lật làm nửa phải

Vì kết quả là palindrome, chỉ cần quyết định **nửa trái** `half`, rồi ghép: `half + (kí tự giữa nếu n lẻ) + half[::-1]`.

Giải thích từng đoạn:

- `n = len(s)`; `half = [min(s[i], s[n-1-i]) for i in range(n // 2)]` — chữ chung của từng cặp.
- `mid = s[n // 2] if n % 2 else ""` — nếu `n` lẻ lấy kí tự giữa, nếu chẵn thì chuỗi rỗng.
- `"".join(half) + mid + "".join(half)[::-1]`: nửa trái, giữa, nửa phải là nửa trái đảo ngược.

Ưu điểm: **không phải sửa tại chỗ**, và thể hiện rõ cấu trúc "palindrome = nửa + giữa + nửa đảo".

````python
def smallest_palindrome_half(s: str) -> str:
    n = len(s)
    left = "".join(min(s[i], s[n - 1 - i]) for i in range(n // 2))
    mid = s[n // 2] if n % 2 == 1 else ""
    return left + mid + left[::-1]

print(smallest_palindrome_half("egcfe"), smallest_palindrome_half("abcd"), smallest_palindrome_half("seven"))
print(repr(smallest_palindrome_half("a")), repr(smallest_palindrome_half("ba")))
````

````python
class SolutionHalf:
    def makeSmallestPalindrome(self, s: str) -> str:
        n = len(s)
        left = "".join(min(s[i], s[n - 1 - i]) for i in range(n // 2))
        mid = s[n // 2] if n % 2 == 1 else ""
        return left + mid + left[::-1]

print(SolutionHalf().makeSmallestPalindrome("seven"))
````

**Độ phức tạp:** Time `O(n)`; Space `O(n)` (chuỗi `left`, kết quả).

### Cách 4 — Một dòng: `min` trên từng cặp `zip(s, reversed(s))`

Công thức mục 6: `res[i] = min(s[i], s[n-1-i])` với **mọi** `i`. Cặp `(s[i], s[n-1-i])` chính là phần tử thứ `i` của `zip(s, reversed(s))`, nên:

`"".join(min(a, b) for a, b in zip(s, reversed(s)))`

Hay: `"".join(map(min, s, reversed(s)))`. Đây là cách ngắn nhất, nhưng bạn phải hiểu vì sao đúng (xem mục 10).

````python
def smallest_palindrome_zip(s: str) -> str:
    return "".join(min(a, b) for a, b in zip(s, reversed(s)))

print(smallest_palindrome_zip("egcfe"), smallest_palindrome_zip("abcd"), smallest_palindrome_zip("seven"))
print("".join(map(min, "seven", reversed("seven"))))     # neven, cách viết bằng map
````

````python
class SolutionZip:
    def makeSmallestPalindrome(self, s: str) -> str:
        return "".join(min(a, b) for a, b in zip(s, reversed(s)))

print(SolutionZip().makeSmallestPalindrome("egcfe"))
````

**Độ phức tạp:** Time `O(n)`, Space `O(n)` (chuỗi kết quả). Duyệt `n` cặp (mỗi cặp phản chiếu xuất hiện hai lần, ta tính hai lần `min` giống nhau — hằng số 2, không đổi độ phức tạp).

## 8. TRACE từng bước

Trace Cách 2 với `s = "egcfe"`:

| bước | `i` | `j` | `arr[i]` | `arr[j]` | `min` | `arr` sau |
|---|---|---|---|---|---|---|
| 1 | 0 | 4 | `e` | `e` | `e` | `[e,g,c,f,e]` |
| 2 | 1 | 3 | `g` | `f` | `f` | `[e,f,c,f,e]` |
| — | 2 | 2 | dừng vì `i < j` sai | | | `efcfe` |

Tổng số thao tác = số cặp mà `arr[i] != arr[j]` = **1** ⇒ trùng ví dụ 1.

````python
def trace_two_pointers(s):
    arr = list(s)
    i, j = 0, len(arr) - 1
    ops = 0
    while i < j:
        a, b = arr[i], arr[j]
        c = min(a, b)
        if a != b:
            ops += 1
        arr[i] = arr[j] = c
        print(f"i={i}, j={j}: {a} vs {b} -> {c}   arr={''.join(arr)}  ops={ops}")
        i += 1
        j -= 1
    print(f"dừng: i={i}, j={j}; kết quả {''.join(arr)!r}, {ops} thao tác")
    return "".join(arr)

trace_two_pointers("egcfe")
````

````python
# n chẵn, nhiều cặp khác nhau: "abcd"; rồi n lẻ khác: "seven"
trace_two_pointers("abcd")
trace_two_pointers("seven")
````

## 10. Vì sao thuật toán đúng

Gọi `t` là palindrome ta tạo ra. Cần chứng minh: (a) `t` đạt số thao tác **ít nhất**; (b) trong các palindrome đạt số thao tác ít nhất, `t` **nhỏ nhất**.

**(a) Số thao tác ít nhất.** Số thao tác = số vị trí `i` mà `t[i] != s[i]`. Palindrome `t` đòi hỏi `t[i] = t[n-1-i]`. Xét một cặp `(i, j)` với `s[i] = a`, `s[j] = b`:

- `a == b`: tốn tối thiểu `0` (giữ nguyên); chọn chữ chung khác `a` tốn `2`.
- `a != b`: tốn tối thiểu `1`, đạt được khi chữ chung là `a` hoặc `b` (đổi đúng một kí tự); chọn chữ khác cả `a, b` tốn `2`.

Các cặp **rời nhau** (chạm vào các vị trí khác nhau) nên tổng thao tác = tổng chi phí từng cặp, và tổng nhỏ nhất đạt được khi **từng cặp** đạt chi phí nhỏ nhất. Kí tự giữa (n lẻ) không cần đổi. Cách của ta giữ chi phí nhỏ nhất ở **mọi** cặp ⇒ tổng nhỏ nhất.

**(b) Nhỏ nhất theo từ điển.** Palindrome đạt tổng nhỏ nhất bắt buộc, ở mỗi cặp **khác nhau**, chữ chung thuộc `{a, b}`; còn cặp **bằng nhau** thì chữ chung phải đúng bằng `a` (chọn khác thì chi phí tăng). Vậy chỗ còn lựa chọn duy nhất là mỗi cặp khác nhau chọn `a` hay `b`. Chuỗi nhỏ nhất theo từ điển được tạo bằng cách tại **vị trí đầu tiên** còn lựa chọn ta chọn chữ nhỏ nhất, rồi tiếp tục (vì các cặp độc lập, lựa chọn cặp trước không làm hạn chế cặp sau). Với cặp `(i, j)`, `i < j` nên vị trí `i` đứng trước `j` trong chuỗi: chọn `min(a, b)` làm `t[i]` rồi `t[j] = t[i]` vì phải đối xứng. Do đó `t[i] = min(s[i], s[n-1-i])` cho nửa trái, nửa phải sao lại đối xứng ⇒ chuỗi nhỏ nhất.

Ta cũng **vét cạn** mọi palindrome cùng độ dài trên chuỗi ngắn ở mục 12.4 để kiểm chứng bằng thực nghiệm.

### 12.3. Phản ví dụ cho cách làm sai

````python
def sai_chep_nua_trai(s):
    n = len(s)
    left = s[: n // 2]
    mid = s[n // 2] if n % 2 else ""
    return left + mid + left[::-1]

def dem_thao_tac(s, t):
    return sum(1 for x, y in zip(s, t) if x != y)

print(sai_chep_nua_trai("dcba"), "!=", Solution().makeSmallestPalindrome("dcba"))
assert sai_chep_nua_trai("dcba") != Solution().makeSmallestPalindrome("dcba")
print("số thao tác 'toàn a' =", dem_thao_tac("abcd", "aaaa"), "> tối thiểu", dem_thao_tac("abcd", "abba"))
````

### 12.4. Đối chiếu vét cạn mọi palindrome + ngẫu nhiên (có seed)

**Oracle vét cạn:** với `n ≤ 6`, thử **mọi** palindrome độ dài `n` trên bảng 26 chữ (tối đa `26^3 = 17576` palindrome). Với mỗi palindrome tính (số thao tác, chuỗi) rồi lấy nhỏ nhất theo cặp đó. Cách này **không dựa vào bất kỳ giả định nào** về lời giải.

Sau đó ta kiểm tra thêm với chuỗi dài (tới 1000): mọi cách khớp nhau, kết quả là palindrome, và số thao tác đúng bằng số cặp đối xứng khác nhau.

````python
import random, string, itertools

def oracle_bruteforce(s):
    n = len(s)
    half_len = (n + 1) // 2
    best = None
    for half in itertools.product(string.ascii_lowercase, repeat=half_len):
        h = "".join(half)
        t = h + h[: n // 2][::-1]           # palindrome xác định bởi nửa đầu (gồm cả kí tự giữa)
        key = (sum(1 for x, y in zip(s, t) if x != y), t)
        if best is None or key < best:
            best = key
    return best[1]

random.seed(2697)
for _ in range(40):
    n = random.randint(1, 5)               # 26^3 = 17576 palindrome nhiều nhất cho mỗi test
    s = "".join(random.choice("abcdxyz") for _ in range(n))
    want = oracle_bruteforce(s)
    for f in funcs:
        assert f(s) == want, (s, want, f(s))
print("OK: 40 chuỗi ngắn khớp oracle vét cạn")
````

## 13. Các lỗi người mới hay gặp

### Lỗi 1: Luôn chép nửa trái (hoặc nửa phải)

Nghĩ "đối xứng thì cứ lấy nửa trái rồi lật là xong" — thoả đối xứng và số thao tác ít nhất, nhưng **không phải nhỏ nhất**.

````python
def sai_chep_trai(s: str) -> str:
    n = len(s)
    return s[: (n + 1) // 2] + s[: n // 2][::-1]

print("sai :", sai_chep_trai("dcba"))                    # dccd  (đối xứng, 2 thao tác, nhưng không nhỏ nhất)
print("đúng:", Solution().makeSmallestPalindrome("dcba"))   # abba
assert sai_chep_trai("dcba") != Solution().makeSmallestPalindrome("dcba")
````

**Vì sao sai:** đề yêu cầu **nhỏ nhất theo từ điển** trong các đáp án ít thao tác nhất. Mỗi cặp khác nhau có hai lựa chọn (đổi bên phải theo bên trái, hoặc ngược lại); phải chọn chữ **nhỏ hơn** ở từng cặp, tức `min(a, b)`.

### Lỗi 2: Đổi cả hai kí tự (tốn 2 thao tác thay vì 1)

````python
def sai_doi_ca_hai(s: str) -> str:
    n = len(s)
    arr = list(s)
    for i in range(n // 2):
        if arr[i] != arr[n - 1 - i]:
            arr[i] = arr[n - 1 - i] = "a"        # SAI: luôn thay cả hai bằng 'a'
    return "".join(arr)

def dem_thao_tac(s, t):
    return sum(1 for x, y in zip(s, t) if x != y)

print("sai :", sai_doi_ca_hai("seven"), dem_thao_tac("seven", sai_doi_ca_hai("seven")), "thao tác")
print("đúng:", Solution().makeSmallestPalindrome("seven"), dem_thao_tac("seven", Solution().makeSmallestPalindrome("seven")), "thao tác")
assert dem_thao_tac("seven", sai_doi_ca_hai("seven")) > dem_thao_tac("seven", Solution().makeSmallestPalindrome("seven"))
````

**Vì sao sai:** đề ưu tiên **số thao tác ít nhất** trước, rồi mới đến từ điển. Chọn một chữ ngoài `{s[i], s[j]}` (ở đây `a`) làm cả hai kí tự tốn tới 2 thao tác cho cặp đó. Chỉ được chọn trong `{s[i], s[j]}`.

### Lỗi 3: Quên kí tự giữa khi `n` lẻ

````python
def sai_quen_giua(s: str) -> str:
    n = len(s)
    left = "".join(min(s[i], s[n - 1 - i]) for i in range(n // 2))
    return left + left[::-1]                      # SAI: n lẻ thì mất kí tự giữa

print("sai :", repr(sai_quen_giua("seven")), len(sai_quen_giua("seven")))      # độ dài 4 thay vì 5
print("đúng:", repr(Solution().makeSmallestPalindrome("seven")))
assert len(sai_quen_giua("seven")) != len("seven")
````

**Vì sao sai:** với `n = 5`, `n // 2 = 2` cặp chỉ tạo 4 kí tự; kí tự ở giữa (`v`) bị rơi mất. Cách sửa: `mid = s[n // 2] if n % 2 else ""` (Cách 3), hoặc dùng công thức `min` trên **mọi** vị trí (Cách 4), trong đó vị trí giữa tự cho `min(x, x) = x`.

### Lỗi 4: Gán trực tiếp vào chuỗi

````python
def sai_gan_chuoi(s: str) -> str:
    i, j = 0, len(s) - 1
    while i < j:
        s[i] = s[j] = min(s[i], s[j])        # SAI: chuỗi bất biến
        i += 1
        j -= 1
    return s

try:
    sai_gan_chuoi("abcd")
except TypeError as e:
    print("TypeError:", e)
print("đúng:", Solution().makeSmallestPalindrome("abcd"))
````

**Vì sao sai:** `str` không hỗ trợ gán theo chỉ số. Hãy đổi sang `list(s)` rồi `"".join(...)`, hoặc dựng chuỗi mới (Cách 3, 4).

## 14. Mẹo làm bài nhanh & nhận dạng pattern

- Từ khoá **palindrome** ⇒ nghĩ ngay đến **hai con trỏ** từ hai đầu vào giữa, hoặc **cặp đối xứng** `(i, n-1-i)`.
- "**Ít thao tác nhất**" + các vị trí độc lập ⇒ tham lam từng vị trí/cặp; kiểm tra tính độc lập trước.
- "Nhỏ nhất theo từ điển" ⇒ ở **vị trí đầu tiên còn lựa chọn**, chọn chữ nhỏ nhất. Hoà nhau thì xét vị trí tiếp theo.
- Với số vị trí giữa, nhớ trường hợp `n` lẻ (kí tự giữa tự đối xứng); công thức `min(s[i], s[n-1-i])` cho mọi `i` xử lý luôn mà không cần `if`.
- Chuỗi Python bất biến: nghĩ sớm cách xây kết quả (`list` + `join`, hoặc xây chuỗi mới bằng `join` trên generator).
- Hằng số 26 chữ là nhỏ nhưng nên **bỏ** khi biết đáp án nằm trong `{a, b}`.

## 15. Biến thể của bài

1. **Số thao tác tối thiểu** để thành palindrome: `sum(a != b for a, b in zip(s[: n // 2], s[::-1]))` — chính là số cặp khác nhau.
2. **Palindrome lớn nhất theo từ điển** (cùng số thao tác ít nhất): thay `min` bằng `max`.
3. **LeetCode 1328 Break a Palindrome**: đổi đúng **một** kí tự để chuỗi **không còn** là palindrome và nhỏ nhất.
4. **LeetCode 680 Valid Palindrome II**: có thể **xoá tối đa một** kí tự — hai con trỏ + thử bỏ một bên.
5. **LeetCode 214 Shortest Palindrome**: thêm kí tự **vào đầu** để ngắn nhất — cần KMP/hash.
6. **Cho phép thao tác đổi chữ có chi phí khác nhau** (đổi `x` thành `y` tốn `|x - y|`): chi phí cặp là `|a - b|`, vẫn tham lam từng cặp.

````python
def min_ops(s):
    return sum(1 for a, b in zip(s[: len(s) // 2], reversed(s)) if a != b)

def largest_palindrome(s):
    return "".join(max(a, b) for a, b in zip(s, reversed(s)))

print(min_ops("abcd"), min_ops("egcfe"), min_ops("abba"))      # 2 1 0
print(largest_palindrome("abcd"), largest_palindrome("seven"))  # dccd seves
````

````python
# Bài 1
assert min("q", "e") == "e"
assert list(zip("ab", "xy")) == [("a", "x"), ("b", "y")]
assert "abc"[::-1] == "cba"
assert "abba" == "abba"[::-1]

# Bài 2: leetcode -> cặp (l,e)->e, (e,d)->d, (e,o)->e, (t,c)->c => "edeccede", 4 thao tác
def small(s):
    return "".join(min(a, b) for a, b in zip(s, reversed(s)))

ans2 = small("leetcode")
assert ans2 == ans2[::-1]
ops2 = sum(1 for x, y in zip("leetcode", ans2) if x != y)
assert (ans2, ops2) == ("edeccede", 4), (ans2, ops2)

# Bài 3
def make_palindrome_todo_ans(s):
    arr = list(s)
    i, j = 0, len(arr) - 1
    while i < j:
        c = min(arr[i], arr[j])
        arr[i] = arr[j] = c
        i += 1
        j -= 1
    return "".join(arr)

assert make_palindrome_todo_ans("egcfe") == "efcfe"
assert make_palindrome_todo_ans("abcd") == "abba"

# Bài 4
def make_palindrome_half(s):
    n = len(s)
    left = "".join(min(s[i], s[n - 1 - i]) for i in range(n // 2))
    mid = s[n // 2] if n % 2 else ""
    return left + mid + left[::-1]

assert make_palindrome_half("egcfe") == "efcfe"
assert make_palindrome_half("abcd") == "abba"
assert make_palindrome_half("seven") == "neven"

# Bài 5
def min_cost_palindrome(s):
    n = len(s)
    return sum(abs(ord(s[i]) - ord(s[n - 1 - i])) for i in range(n // 2))

assert min_cost_palindrome("abcd") == 3 + 1
assert min_cost_palindrome("abba") == 0

def break_palindrome(s):
    n = len(s)
    if n == 1:
        return ""
    for i in range(n // 2):
        if s[i] != "a":
            return s[:i] + "a" + s[i + 1:]
    return s[:-1] + "b"

assert break_palindrome("abccba") == "aaccba"
assert break_palindrome("a") == ""
assert break_palindrome("aa") == "ab"
print("Tất cả đáp án đúng")
````
