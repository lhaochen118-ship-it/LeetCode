## 1. Mục tiêu học

Sau notebook này bạn sẽ:

1. Hiểu đề: đếm các bit `1` của `n` đang nằm ở vị trí **chẵn** (`0, 2, 4, ...`) và vị trí **lẻ** (`1, 3, 5, ...`), **đánh số từ phải sang trái** (bit thấp nhất là vị trí 0).
2. Đọc biểu diễn nhị phân bằng `bin(n)`, đảo chuỗi bằng `[::-1]` và duyệt bằng `enumerate`.
3. Dùng các toán tử bit cơ bản: `n & 1` (lấy bit thấp nhất), `n >> 1` (bỏ bit thấp nhất), `^` (XOR) để đổi chẵn/lẻ.
4. Biết **mặt nạ bit (bit mask)** `0x55555555` (chọn các vị trí chẵn) và `0xAAAAAAAA` (các vị trí lẻ), rồi đếm số bit 1 (popcount).
5. Viết 4 cách và đối chiếu **toàn bộ** miền `1..1000` bằng một oracle viết bằng phép chia thay vì phép bit.
6. Tránh các lỗi quen thuộc: đánh số từ trái sang phải, nhầm vị trí chẵn/lẻ (bắt đầu từ 1), đếm mọi bit thay vì chỉ bit 1, chọn sai mặt nạ.

## 2. Đề bài & ví dụ

Cho số nguyên dương `n`.

- Gọi `even` là số **bit 1 ở các chỉ số chẵn** trong biểu diễn nhị phân của `n`.
- Gọi `odd` là số **bit 1 ở các chỉ số lẻ** trong biểu diễn nhị phân của `n`.

Lưu ý: các bit được đánh chỉ số **từ phải sang trái** (bit ngoài cùng bên phải có chỉ số 0).

Trả về mảng `[even, odd]`.

| Ví dụ | n | Output | Giải thích |
|---|---|---|---|
| 1 | `50` | `[1, 2]` | `50` viết nhị phân là `110010`; các bit 1 ở chỉ số `1, 4, 5` → chỉ số chẵn: `4` (1 bit), chỉ số lẻ: `1, 5` (2 bit) |
| 2 | `2` | `[0, 1]` | `2` là `10`; bit 1 duy nhất ở chỉ số `1` (lẻ) |

Constraints: `1 <= n <= 1000`.

````python
for n in (50, 2):
    b = bin(n)[2:]
    print(f"n={n} nhị phân={b}")
    # đánh chỉ số từ PHẢI sang TRÁI
    for idx in range(len(b)):
        bit = b[len(b) - 1 - idx]
        print(f"   chỉ số {idx}: bit {bit}")
````

## 3. Kiến thức nền

### 3.1. Biểu diễn nhị phân và `bin`

Mỗi số nguyên dương viết được dưới dạng tổng các lũy thừa của 2: `50 = 32 + 16 + 2 = 2^5 + 2^4 + 2^1`. Các bit `1` nằm ở **chỉ số** `5, 4, 1` (số mũ). Hàm `bin(n)` trả chuỗi dạng `'0b110010'` (tiền tố `0b` + các bit); ta cắt `[2:]` để bỏ tiền tố.

````python
print(bin(50), bin(50)[2:], 2 ** 5 + 2 ** 4 + 2 ** 1)
print(bin(2), bin(1), bin(8))
````

### 3.2. Đảo chuỗi `[::-1]` và `enumerate`

Chỉ số bit đếm **từ phải**, nhưng chuỗi nhị phân đọc **từ trái**. Đảo chuỗi bằng `s[::-1]` (bước `-1`) thì ký tự đầu của chuỗi mới chính là bit chỉ số 0. `enumerate(s)` cho từng cặp `(chỉ số, ký tự)`.

````python
b = bin(50)[2:]
r = b[::-1]
print(b, "->", r)
for i, ch in enumerate(r):
    print(i, ch, "(chẵn)" if i % 2 == 0 else "(lẻ)")
````

### 3.3. Toán tử bit `&` và `>>`

- `n & 1` lấy **bit thấp nhất** (`1` nếu `n` lẻ, `0` nếu `n` chẵn).
- `n >> 1` **dịch phải** 1 bit: bỏ bit thấp nhất, tương đương `n // 2`.
- Lặp "lấy bit thấp nhất, rồi dịch phải" cho ta duyệt các bit từ chỉ số 0 trở lên.

````python
n = 50          # 110010
while n:
    print(f"n={n:>3} ({bin(n)[2:]:>6}) bit thấp nhất = {n & 1}")
    n >>= 1
````

### 3.4. Đổi chẵn/lẻ: `i % 2`, `i & 1` và `i ^= 1`

Chỉ số `i` là chẵn ⇔ `i % 2 == 0`. Phép **XOR** `i ^= 1` đổi `0 -> 1` và `1 -> 0`; dùng để "bật qua bật lại" giữa chẵn và lẻ mà không cần `if`.

````python
i = 0
seq = []
for _ in range(6):
    seq.append(i)
    i ^= 1
print(seq)          # [0, 1, 0, 1, 0, 1]
print([k % 2 for k in range(6)], [k & 1 for k in range(6)])
````

### 3.5. Mặt nạ bit (bit mask) và số hệ 16

`x & mask` giữ lại các bit của `x` ở những vị trí mà `mask` có bit `1`, và đặt các vị trí khác về `0`. Hệ 16 (hexa) cho viết mặt nạ ngắn gọn vì một chữ số hexa là **4 bit**: `5` ↔ `0101`, `A` ↔ `1010`.

- `0x5 = 0101`: bit chỉ số `0` và `2` bật → mặt nạ cho vị trí **chẵn**.
- `0xA = 1010`: bit chỉ số `1` và `3` bật → mặt nạ cho vị trí **lẻ**.
- Lặp lại: `0x55555555` chọn mọi chỉ số chẵn trong 32 bit; `0xAAAAAAAA` chọn mọi chỉ số lẻ.

````python
print(bin(0x5), bin(0xA))
print(bin(0x55555555)[2:].zfill(32))
print(bin(0xAAAAAAAA)[2:].zfill(32))
n = 50
print("n & 0x55555555 =", bin(n & 0x55555555), "-> bit 1 ở chỉ số chẵn")
print("n & 0xAAAAAAAA =", bin(n & 0xAAAAAAAA), "-> bit 1 ở chỉ số lẻ")
````

### 3.6. Đếm số bit 1 (popcount)

`bin(x).count('1')` đếm số ký tự `'1'` trong chuỗi nhị phân; từ Python 3.10 còn có `x.bit_count()` làm việc tương tự nhanh hơn. Với `x = 0`, `bin(0) = '0b0'` nên số bit 1 là `0`.

````python
x = 0b110010
print(bin(x).count("1"))      # 3
print(bin(0), bin(0).count("1"))      # 0 không có bit 1 nào
````

## 4. Ý tưởng nền tảng / cách nghĩ trực giác

Viết nhị phân của `n` rồi **đánh số từ phải**; chỉ nhìn các chỗ có `1`, phân loại vị trí chẵn / lẻ.

`n = 50 = 110010`:

```
vị trí (từ phải):  5 4 3 2 1 0
bit             :  1 1 0 0 1 0
                   ^ ^     ^
                   | |     +-- vị trí 1: lẻ
                   | +-------- vị trí 4: chẵn
                   +---------- vị trí 5: lẻ
even = 1 (vị trí 4), odd = 2 (vị trí 1 và 5)  ->  [1, 2]
```

````python
n = 50
b = bin(n)[2:]
even = odd = 0
for pos in range(len(b)):                 # pos = chỉ số tính từ phải
    bit = b[len(b) - 1 - pos]
    if bit == "1":
        if pos % 2 == 0:
            even += 1
        else:
            odd += 1
print([even, odd])
````

## 5. Cách 1 — Brute Force / baseline

Đổi `n` sang chuỗi nhị phân, **đảo chuỗi** để chỉ số bit khớp chỉ số chuỗi, rồi duyệt `enumerate`: gặp `'1'` thì tăng `even` hoặc `odd` theo tính chẵn lẻ của chỉ số.

````python
from typing import List

def even_odd_string(n: int) -> List[int]:
    bits = bin(n)[2:][::-1]              # đảo để bits[i] là bit chỉ số i
    even = odd = 0
    for i, ch in enumerate(bits):
        if ch == "1":
            if i % 2 == 0:
                even += 1
            else:
                odd += 1
    return [even, odd]
````

````python
print(even_odd_string(50))    # [1, 2]
print(even_odd_string(2))     # [0, 1]
print(even_odd_string(1))     # [1, 0]
````

**Độ phức tạp:** gọi `b = log2(n)` là số bit. Đổi sang chuỗi, đảo, duyệt đều `O(b)` → Time `O(log n)`; Space `O(log n)` cho chuỗi.

**Nút thắt:** ta tạo chuỗi trung gian chỉ để đọc từng bit, trong khi có thể lấy bit trực tiếp từ số bằng `n & 1` và `n >> 1`. Hơn nữa việc đếm có thể làm **song song** cho mọi bit bằng một mặt nạ.

## 6. Ý tưởng thuật toán / bước đột phá

Hai quan sát:

1. **Không cần chuỗi:** lặp `while n:` lấy `n & 1` rồi `n >>= 1`; chỉ số bit hiện tại chính là số lần đã dịch.
2. **Không cần phân nhánh chẵn/lẻ:** giữ mảng `cnt = [even, odd]` và một chỉ số `i` luân phiên `0, 1, 0, 1, ...` (`i ^= 1`); mỗi bước `cnt[i] += n & 1`.

Và một bước nhảy xa hơn: **mặt nạ**. `n & 0x55555555` giữ lại các bit 1 ở chỉ số chẵn; đếm số bit 1 của kết quả cho `even`. Tương tự `0xAAAAAAAA` cho `odd`.

Quy tắc "nếu… thì…": bit thấp nhất bằng `1` → cộng vào `cnt[i]`; luôn đổi `i` sau mỗi bit.

````python
cnt = [0, 0]                 # cnt[0] = even, cnt[1] = odd
n, i = 50, 0
while n:
    cnt[i] += n & 1
    i ^= 1
    n >>= 1
print(cnt)
````

## 7. Các cách tiếp theo

Ta có thêm 3 cách: **(2)** dịch bit và dùng `i % 2`, **(3)** mảng `cnt` + `i ^= 1` không `if`, **(4)** mặt nạ + popcount.

### Cách 2 — Dịch bit, dùng `i % 2`

Giải thích từng đoạn:

1. `i` là chỉ số bit hiện tại, bắt đầu từ `0`.
2. `while n:` — khi `n` còn bit 1 nào đó phía trên thì còn phải xử lý.
3. Nếu `n & 1` (bit thấp nhất là 1): tăng `even` nếu `i` chẵn, ngược lại `odd`.
4. `n >>= 1` bỏ bit vừa xử lý và `i += 1`.

````python
def even_odd_shift(n):
    even = odd = 0
    i = 0
    while n:
        if n & 1:
            if i % 2 == 0:
                even += 1
            else:
                odd += 1
        n >>= 1
        i += 1
    return [even, odd]

print(even_odd_shift(50), even_odd_shift(2))
````

````python
from typing import List

class SolutionShift:
    def evenOddBit(self, n: int) -> List[int]:
        even = odd = 0
        i = 0
        while n:
            if n & 1:
                if i % 2 == 0:
                    even += 1
                else:
                    odd += 1
            n >>= 1
            i += 1
        return [even, odd]
````

Độ phức tạp Cách 2: Time `O(log n)` (số lần dịch bằng độ dài nhị phân); Space `O(1)`.

### Cách 3 — Mảng `cnt` và `i ^= 1` (không `if`)

`cnt[i] += n & 1` cộng `0` hoặc `1` vào ô đúng chẵn/lẻ; `i ^= 1` đổi qua lại `0 <-> 1`. Nhờ vậy vòng lặp chỉ có 3 dòng.

````python
def even_odd_toggle(n):
    cnt = [0, 0]
    i = 0
    while n:
        cnt[i] += n & 1
        i ^= 1
        n >>= 1
    return cnt

print(even_odd_toggle(50), even_odd_toggle(2))
````

````python
from typing import List

class SolutionToggle:
    def evenOddBit(self, n: int) -> List[int]:
        cnt = [0, 0]
        i = 0
        while n:
            cnt[i] += n & 1
            i ^= 1
            n >>= 1
        return cnt
````

Độ phức tạp Cách 3: Time `O(log n)`; Space `O(1)`.

### Cách 4 — Mặt nạ bit + popcount

`n & 0x55555555` chỉ giữ các bit 1 ở chỉ số chẵn; số bit 1 còn lại chính là `even`. `n & 0xAAAAAAAA` tương tự cho `odd`. Vì `n <= 1000 < 2^10` nên 32 bit là dư; với `n` nhỏ hơn `2^32` công thức đúng. (Nếu `n` có thể lớn hơn thì dùng mặt nạ dài hơn.)

````python
def even_odd_mask(n):
    even = bin(n & 0x55555555).count("1")
    odd = bin(n & 0xAAAAAAAA).count("1")
    return [even, odd]

print(even_odd_mask(50), even_odd_mask(2))
print("bit_count:", (50 & 0x55555555).bit_count(), (50 & 0xAAAAAAAA).bit_count())
````

````python
from typing import List

class SolutionMask:
    def evenOddBit(self, n: int) -> List[int]:
        return [bin(n & 0x55555555).count("1"), bin(n & 0xAAAAAAAA).count("1")]
````

Độ phức tạp Cách 4: Time `O(1)` theo độ rộng cố định 32 bit (hoặc `O(log n)` nếu tính cả việc đếm bit 1); Space `O(1)`. (Ghi chú: `int.bit_count()` có từ Python 3.10, nên khi nộp ta dùng `bin(...).count("1")` cho chắc.)

## 8. TRACE từng bước

Trace Cách 3 (mảng `cnt` + `i ^= 1`) với `n = 50 = 110010`.

| bước | `n` (nhị phân) | `i` | `n & 1` | `cnt` sau | `n` sau `>>= 1` |
|---|---|---|---|---|---|
| 1 | `110010` | 0 | 0 | `[0, 0]` | `11001` |
| 2 | `11001` | 1 | 1 | `[0, 1]` | `1100` |
| 3 | `1100` | 0 | 0 | `[0, 1]` | `110` |
| 4 | `110` | 1 | 0 | `[0, 1]` | `11` |
| 5 | `11` | 0 | 1 | `[1, 1]` | `1` |
| 6 | `1` | 1 | 1 | `[1, 2]` | `0` |

Hết vòng vì `n = 0`. Kết quả `[1, 2]`. Chú ý: bước 2 (chỉ số 1) và bước 6 (chỉ số 5) đều ở chỉ số lẻ.

````python
def trace_toggle(n):
    cnt = [0, 0]
    i = 0
    step = 0
    print("n ban đầu:", n, bin(n)[2:])
    while n:
        step += 1
        low = n & 1
        cnt[i] += low
        print(f"bước {step}: n={bin(n)[2:]:>6} chỉ số bit={step - 1} (i={i}) bit thấp nhất={low} -> cnt={cnt}")
        i ^= 1
        n >>= 1
    print("kết quả:", cnt)
    return cnt

trace_toggle(50)
````

````python
trace_toggle(2)      # bit 1 duy nhất nằm ở chỉ số lẻ
trace_toggle(1000)   # biên trên của đề
````

````python
# Trace cách mặt nạ: nhìn trực tiếp bit nào được giữ lại
n = 50
print("n            :", bin(n)[2:].zfill(8))
print("mặt nạ chẵn  :", bin(0x55)[2:].zfill(8))
print("n & mặt nạ   :", bin(n & 0x55)[2:].zfill(8), "-> even =", bin(n & 0x55).count("1"))
print("mặt nạ lẻ    :", bin(0xAA)[2:].zfill(8))
print("n & mặt nạ   :", bin(n & 0xAA)[2:].zfill(8), "-> odd  =", bin(n & 0xAA).count("1"))
````

## 9. Code nộp LeetCode

Nộp Cách 3 (mảng `cnt` + `i ^= 1`): ngắn, `O(log n)` thời gian, `O(1)` bộ nhớ.

````python
from typing import List

class Solution:
    def evenOddBit(self, n: int) -> List[int]:
        cnt = [0, 0]          # cnt[0] = even, cnt[1] = odd
        i = 0
        while n:
            cnt[i] += n & 1
            i ^= 1
            n >>= 1
        return cnt

print(Solution().evenOddBit(50))   # [1, 2]
print(Solution().evenOddBit(2))    # [0, 1]
````

## 10. Vì sao thuật toán đúng

- **Invariant:** ở đầu mỗi vòng lặp, `i` bằng chỉ số bit (tính từ phải) của bit thấp nhất hiện có của `n`, và `cnt` đã đếm đúng mọi bit 1 ở các chỉ số `< i`. Ban đầu `i = 0`, `cnt = [0, 0]` và chưa có bit nào được xử lý.
- **Bước lặp:** `n & 1` là bit chỉ số `i` của số gốc (vì `n` đã bị dịch phải `i` lần). `cnt[i] += n & 1` — ở đây `i` chỉ nhận giá trị `0` hoặc `1`, **đúng bằng `(chỉ số bit) mod 2`**, vì `i` được đổi qua lại sau mỗi bit — nên bit 1 chẵn vào `cnt[0]`, bit 1 lẻ vào `cnt[1]`.
- **Kết thúc:** khi `n = 0`, mọi bit còn lại đều là `0` nên không còn bit 1 nào chưa đếm. Đồng thời vòng lặp luôn dừng vì `n` giảm ít nhất một nửa mỗi bước (`n >>= 1`) và `n >= 0`.
- **Cách mặt nạ:** `0x55555555` có bit 1 đúng ở các chỉ số chẵn `0, 2, ..., 30`, `0xAAAAAAAA` đúng ở các chỉ số lẻ `1, 3, ..., 31`. Với `n < 2^32`, mọi bit 1 của `n` nằm đúng ở một trong hai mặt nạ nên `popcount(n & mask)` đếm đúng.
- **Cách chuỗi:** `bin(n)[2:][::-1]` đưa bit chỉ số 0 lên đầu, nên chỉ số chuỗi bằng chỉ số bit; ta đếm đúng định nghĩa.

## 11. Phân tích Time / Space Complexity

Gọi `b = floor(log2 n) + 1` là số bit của `n` (`b <= 10` với `n <= 1000`).

| Cách | Time | Space |
|---|---|---|
| 1. Chuỗi nhị phân đảo + `enumerate` | `O(b)` | `O(b)` |
| 2. Dịch bit + `i % 2` | `O(b)` | `O(1)` |
| 3. Mảng `cnt` + `i ^= 1` | `O(b)` | `O(1)` |
| 4. Mặt nạ + popcount | `O(1)` theo 32 bit cố định | `O(1)` |

Với `n <= 1000` mọi cách xong trong vài micro-giây. Phần học chính là cách **nghĩ theo bit**.

## 12. Bộ test

Test 1: ví dụ đề và các edge case: `n = 1` (nhỏ nhất, chỉ bit 0), lũy thừa của 2, số có toàn bit 1 (`2^k - 1`), và biên trên `1000`.

````python
impls = [
    ("string", even_odd_string),
    ("shift", even_odd_shift),
    ("toggle", even_odd_toggle),
    ("mask", even_odd_mask),
    ("class_shift", SolutionShift().evenOddBit),
    ("class_toggle", SolutionToggle().evenOddBit),
    ("class_mask", SolutionMask().evenOddBit),
    ("submit", Solution().evenOddBit),
]
tests = [
    (50, [1, 2]),
    (2, [0, 1]),
    (1, [1, 0]),            # chỉ bit 0
    (4, [1, 0]),            # 100: bit chỉ số 2
    (8, [0, 1]),            # 1000: bit chỉ số 3
    (7, [2, 1]),            # 111: chỉ số 0, 1, 2
    (15, [2, 2]),           # 1111
    (255, [4, 4]),          # 11111111
    (1000, None),           # biên trên: 1111101000, đối chiếu bằng oracle ở Test 2
]
for name, f in impls:
    for n, expected in tests:
        got = f(n)
        if expected is not None:
            assert got == expected, (name, n, got, expected)
        assert len(got) == 2 and got[0] + got[1] == bin(n).count("1")
    print(name, "PASS", len(tests), "test")
````

Test 2: **kiểm chứng toàn bộ** miền input `1..1000` với oracle độc lập, viết bằng **phép chia / lấy dư** (không dùng phép bit, không dùng chuỗi nhị phân): chữ số nhị phân thứ `k` của `n` là `(n // 2**k) % 2`.

````python
def oracle(n):
    even = odd = 0
    for k in range(0, 12):                       # 2^11 = 2048 > 1000: đủ để phủ mọi bit
        bit = (n // (2 ** k)) % 2
        if bit == 1:
            if k % 2 == 0:
                even += 1
            else:
                odd += 1
    return [even, odd]

for n in range(1, 1001):
    expected = oracle(n)
    for name, f in impls:
        assert f(n) == expected, (name, n, f(n), expected)
print("đã so khớp toàn bộ n trong [1, 1000]: tất cả cách đều đúng")
````

Test 3: tính chất — `even + odd = số bit 1 của n`; nhân đôi `n` (dịch trái 1 bit) **hoán đổi** `even` và `odd` (mọi chỉ số tăng 1 nên chẵn thành lẻ); với `n <= 500` thì `2n <= 1000` hợp lệ.

````python
for n in range(1, 501):
    e, o = Solution().evenOddBit(n)
    e2, o2 = Solution().evenOddBit(2 * n)
    assert (e2, o2) == (o, e), (n, [e, o], [e2, o2])
    assert e + o == bin(n).count("1")
print("tính chất: dịch trái 1 bit làm hoán đổi even và odd: PASS")
````

## 13. Các lỗi người mới hay gặp

### Lỗi 1: đánh chỉ số từ **trái** sang phải

````python
def wrong_left_to_right(n):
    bits = bin(n)[2:]                  # KHÔNG đảo chuỗi
    even = odd = 0
    for i, ch in enumerate(bits):
        if ch == "1":
            if i % 2 == 0:
                even += 1
            else:
                odd += 1
    return [even, odd]

print(wrong_left_to_right(50), "vs đúng", Solution().evenOddBit(50))
````

Vì sao sai: `50 = 110010`; đếm từ trái, các bit 1 ở chỉ số `0, 1, 4` → `even = 2`, `odd = 1` (`[2, 1]`), ngược với đúng (`[1, 2]`). Đề nói rõ chỉ số tính **từ phải sang trái**: phải đảo chuỗi (`[::-1]`) hoặc duyệt bằng `n & 1, n >>= 1`.

### Lỗi 2: bắt đầu chỉ số từ `1` (hoán đổi chẵn / lẻ)

````python
def wrong_start_one(n):
    even = odd = 0
    i = 1                              # SAI: bit thấp nhất có chỉ số 0, không phải 1
    while n:
        if n & 1:
            if i % 2 == 0:
                even += 1
            else:
                odd += 1
        n >>= 1
        i += 1
    return [even, odd]

print(wrong_start_one(50), "vs đúng", Solution().evenOddBit(50))
````

Vì sao sai: lập trình viên thường nghĩ "bit đầu tiên là bit số 1". Đề đánh chỉ số từ `0` nên bit thấp nhất là chỉ số **chẵn**. Khởi tạo `i = 0`.

### Lỗi 3: đếm mọi bit thay vì chỉ bit `1`

````python
def wrong_count_all(n):
    even = odd = 0
    i = 0
    while n:
        if i % 2 == 0:                 # thiếu kiểm tra n & 1
            even += 1
        else:
            odd += 1
        n >>= 1
        i += 1
    return [even, odd]

print(wrong_count_all(50), "vs đúng", Solution().evenOddBit(50))
````

Vì sao sai: đề chỉ đếm các bit **bằng 1**. Code trên đếm số chỉ số chẵn/lẻ (độ dài nhị phân) kể cả bit 0. Phải thêm điều kiện `if n & 1` hoặc cộng `n & 1` thay vì `1`.

### Lỗi 4: chọn sai mặt nạ (đảo `0x55...` và `0xAA...`)

````python
def wrong_mask(n):
    even = bin(n & 0xAAAAAAAA).count("1")      # SAI: 0xAAAAAAAA là mặt nạ vị trí LẺ
    odd = bin(n & 0x55555555).count("1")
    return [even, odd]

print(wrong_mask(50), "vs đúng", Solution().evenOddBit(50))
print("0x5 =", bin(0x5), "-> chỉ số 0 và 2 (chẵn);  0xA =", bin(0xA), "-> chỉ số 1 và 3 (lẻ)")
````

Vì sao sai: `0x5 = 0101` bật các chỉ số `0, 2` (chẵn); `0xA = 1010` bật các chỉ số `1, 3` (lẻ). Nhớ: **5 ↔ chẵn, A ↔ lẻ**; hoặc tự viết ra `bin(0x5)` để kiểm tra.

## 14. Mẹo làm bài nhanh & nhận dạng pattern

- "Đếm bit 1" → `bin(x).count("1")` hoặc `x.bit_count()` hoặc vòng `x & 1`, `x >>= 1`.
- "Vị trí chẵn / lẻ của bit" → mặt nạ `0x55555555` / `0xAAAAAAAA`, hoặc biến `i` luân phiên (`i ^= 1`).
- Nhớ **chỉ số từ phải sang trái, bắt đầu từ 0**. Khi nghi ngờ, viết ví dụ nhỏ ra giấy (`2 = 10`).
- Muốn debug nhanh: `print(bin(n))` hoặc `print(f"{n:b}")`; để thấy rõ độ dài cố định dùng `f"{n:08b}"`.
- Dịch trái `n << 1` = nhân 2 = mọi chỉ số tăng 1 → chẵn và lẻ hoán đổi (đã kiểm ở mục 12).
- Họ bài bit cơ bản: Number of 1 Bits (191), Counting Bits (338), Power of Two (231), Binary Number with Alternating Bits (693), Hamming Distance (461).

## 15. Biến thể của bài

1. Đếm bit 1 ở vị trí **chia 3 dư 0, 1, 2** (tổng quát hoá `mod 2` thành `mod m`) — xem bài tập 5.
2. Đếm bit **0** (trong độ dài nhị phân thực tế của `n`) ở vị trí chẵn / lẻ.
3. Trả về **các chỉ số** của bit 1 thay vì chỉ số lượng (bài tập 3).
4. `n` rất lớn (`10^18`): cần mặt nạ 64 bit `0x5555555555555555`, vẫn `O(1)`.
5. Hỏi "`n` có các bit 1 **xen kẽ** hay không" (Binary Number with Alternating Bits, 693).
6. Đếm theo **byte** (nhóm 8 bit) hoặc theo nibble (4 bit): dùng mặt nạ khác nhau.

## 19. Đáp án tham khảo cho bài tập

````python
# Bài 1: 11 = 1011 -> bit 1 ở chỉ số 0, 1, 3
assert Solution().evenOddBit(11) == [1, 2]

# Bài 2: 100 = 1100100 -> chỉ số 2, 5, 6
assert bin(100)[2:] == "1100100"
assert Solution().evenOddBit(100) == [2, 1]

# Bài 3
def one_bit_positions_answer(n):
    positions = []
    i = 0
    while n:
        if n & 1:
            positions.append(i)
        n >>= 1
        i += 1
    return positions

assert one_bit_positions_answer(50) == [1, 4, 5]
assert one_bit_positions_answer(2) == [1]
assert one_bit_positions_answer(11) == [0, 1, 3]

# Bài 4
def even_odd_by_mask(n):
    return [bin(n & 0x55555555).count("1"), bin(n & 0xAAAAAAAA).count("1")]

for n in range(1, 1001):
    assert even_odd_by_mask(n) == Solution().evenOddBit(n)

# Bài 5
def count_by_mod(n, m):
    res = [0] * m
    for i in one_bit_positions_answer(n):
        res[i % m] += 1
    return res

assert count_by_mod(50, 2) == [1, 2]
# 50 = bit ở chỉ số 1, 4, 5 -> mod 3 = 1, 1, 2 -> [0, 2, 1]
assert count_by_mod(50, 3) == [0, 2, 1]
for n in range(1, 1001):
    assert count_by_mod(n, 2) == Solution().evenOddBit(n)
print("tất cả đáp án bài tập: PASS")
````
