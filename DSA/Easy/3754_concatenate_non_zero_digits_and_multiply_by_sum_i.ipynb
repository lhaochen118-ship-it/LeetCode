## 1. Mục tiêu học

Sau notebook này bạn sẽ:

1. Hiểu đề: từ số `n`, nối các chữ số **khác 0** (giữ thứ tự) thành số `x`, tính `sum` là tổng các chữ số của `x`, rồi trả về `x * sum`.
2. Thành thạo hai cách xử lý chữ số: qua **chuỗi** (`str`, `int`) và qua **số học** (`//`, `%`, `divmod`).
3. Biết cách dựng số từ chữ số theo cả hai hướng: `x = x * 10 + d` (trái sang phải) và `d * place` (phải sang trái).
4. Viết 5 cách khác nhau (chuỗi, lọc + `sum`, `divmod` + `place`, một lượt trái sang phải, đệ quy) và đối chiếu bằng test ngẫu nhiên có `random.seed`.

## 2. Đề bài & ví dụ

Cho số nguyên `n`. Hãy tạo số nguyên `x` bằng cách **nối các chữ số khác 0** của `n` theo **đúng thứ tự xuất hiện** từ trái sang phải. Nếu `n` không có chữ số khác 0 thì `x = 0`. Gọi `sum` là **tổng các chữ số của `x`**. Trả về `x * sum`.

| Ví dụ | n | Output | Giải thích |
|---|---|---|---|
| 1 | `10203004` | `12340` | Chữ số khác 0: `1, 2, 3, 4` nên `x = 1234`; `sum = 1+2+3+4 = 10`; `1234 * 10 = 12340` |
| 2 | `1000` | `1` | Chỉ có chữ số `1` nên `x = 1`, `sum = 1`, kết quả `1` |

Constraints: `0 <= n <= 10^9`. Trường hợp `n = 0` được phép: không có chữ số khác 0, nên `x = 0` và kết quả `0`.

````python
for n in [10203004, 1000, 0, 7, 909]:
    s = str(n)
    kept = [c for c in s if c != '0']
    x = int(''.join(kept)) if kept else 0
    total = sum(int(c) for c in kept)
    print(f"n={n}  giữ lại={kept}  x={x}  sum={total}  -> {x * total}")
````

## 4. Ý tưởng nền tảng / cách nghĩ trực giác

Bài có hai bước rõ ràng:

## 5. Cách 1 — Brute Force / baseline

Dùng chuỗi: `str(n)`, lọc ký tự khác `'0'`, nối lại thành `x` (nếu rỗng thì `0`), tính tổng chữ số bằng `sum(int(c) ...)`, trả về tích. Đây là cách dịch thẳng từng bước của đề.

Nút thắt: không có nút thắt về thời gian (tối đa 10 chữ số); điều cần chú ý là **xử lý `n = 0`** (chuỗi sau khi lọc rỗng nên không thể `int('')`).

````python
class SolutionString:
    def sumAndMultiply(self, n):
        kept = [c for c in str(n) if c != '0']
        if not kept:                              # n = 0 (hoặc không có chữ số khác 0)
            return 0
        x = int(''.join(kept))
        total = sum(int(c) for c in str(x))
        return x * total

print(SolutionString().sumAndMultiply(10203004), SolutionString().sumAndMultiply(1000), SolutionString().sumAndMultiply(0))
````

**Độ phức tạp:** Time `O(d)`, Space `O(d)` với `d` là số chữ số của `n` (tối đa 10).

## 6. Ý tưởng thuật toán / bước đột phá

**Bước đột phá:** gộp hai bước thành **một lượt duyệt chữ số**: vừa dựng `x` vừa cộng `sum`, vì chữ số 0 bị bỏ nên không ảnh hưởng đến cả hai.

**Quy tắc "nếu… thì…":**

- Nếu chữ số `d == 0` -> bỏ qua (không đổi `x`, không đổi `sum`).
- Nếu `d != 0` và duyệt từ trái sang phải -> `x = x * 10 + d; total += d`.
- Nếu duyệt từ phải sang trái -> `x += d * place; place *= 10; total += d`.
- Nếu `n = 0` -> `x = 0`, `total = 0`, kết quả `0` (vòng lặp không chạy nên tự đúng).

**Invariant:** sau khi xử lý một số chữ số, `x` là số tạo bởi các chữ số khác 0 đã xử lý và `total` là tổng của chúng.

````python
# Kiểm chứng: tổng chữ số của x = tổng chữ số của n
import random
random.seed(3754)
for _ in range(2000):
    n = random.randint(0, 10 ** 9)
    kept = [c for c in str(n) if c != '0']
    x = int(''.join(kept)) if kept else 0
    assert sum(int(c) for c in str(x)) == sum(int(c) for c in str(n))
print("tổng chữ số của x bằng tổng chữ số của n: đúng")
````

## 7. Các cách tiếp theo

### Cách 2 — Lọc chữ số rồi nối bằng công thức

Tách `digits = [int(c) for c in str(n) if c != '0']`, dựng `x` bằng `reduce`-kiểu `x * 10 + d`, và `total = sum(digits)`. Gọn hơn vì không đổi qua lại giữa chuỗi và số.

````python
class SolutionDigits:
    def sumAndMultiply(self, n):
        digits = [int(c) for c in str(n) if c != '0']
        x = 0
        for d in digits:
            x = x * 10 + d
        return x * sum(digits)

print(SolutionDigits().sumAndMultiply(10203004), SolutionDigits().sumAndMultiply(1000), SolutionDigits().sumAndMultiply(0))
````

**Độ phức tạp:** Time `O(d)`, Space `O(d)`.

### Cách 3 — Số học thuần: `divmod` + `place`

Không dùng chuỗi: lấy chữ số hàng đơn vị bằng `divmod`; nếu khác 0 thì `x += d * place; place *= 10; total += d`. `n = 0` làm vòng lặp không chạy nên cho `0 * 0 = 0`.

````python
class SolutionDivmod:
    def sumAndMultiply(self, n):
        x, place, total = 0, 1, 0
        while n > 0:
            n, d = divmod(n, 10)
            if d != 0:
                x += d * place
                place *= 10
                total += d
        return x * total

print(SolutionDivmod().sumAndMultiply(10203004), SolutionDivmod().sumAndMultiply(1000), SolutionDivmod().sumAndMultiply(0))
````

**Độ phức tạp:** Time `O(d)`, Space `O(1)`.

### Cách 4 — Một lượt trái sang phải trên chuỗi

Duyệt `for c in str(n)`; mỗi chữ số khác 0 cập nhật cả `x = x * 10 + d` và `total += d` cùng lúc.

````python
class SolutionOnePass:
    def sumAndMultiply(self, n):
        x = total = 0
        for c in str(n):
            d = ord(c) - ord('0')          # đổi ký tự số thành giá trị số
            if d:
                x = x * 10 + d
                total += d
        return x * total

print(SolutionOnePass().sumAndMultiply(10203004), SolutionOnePass().sumAndMultiply(1000), SolutionOnePass().sumAndMultiply(0))
````

**Độ phức tạp:** Time `O(d)`, Space `O(d)` cho chuỗi (hoặc `O(1)` nếu bỏ qua chi phí `str(n)`).

### Cách 5 — Đệ quy trên `n // 10`

`f(n)` trả cặp `(x, total)` cho số `n`: gọi `f(n // 10)` để lấy cặp của phần đầu, rồi xét chữ số cuối `d = n % 10`. Nếu `d != 0` thì `x = x * 10 + d` và `total += d`. Cách này thể hiện cấu trúc bài toán con "bỏ chữ số cuối".

````python
class SolutionRecursive:
    def _f(self, n):
        if n == 0:
            return 0, 0
        x, total = self._f(n // 10)
        d = n % 10
        if d:
            x, total = x * 10 + d, total + d
        return x, total

    def sumAndMultiply(self, n):
        x, total = self._f(n)
        return x * total

print(SolutionRecursive().sumAndMultiply(10203004), SolutionRecursive().sumAndMultiply(1000), SolutionRecursive().sumAndMultiply(0))
````

**Độ phức tạp:** Time `O(d)`, Space `O(d)` (ngăn xếp đệ quy tối đa 10 mức).

## 8. TRACE từng bước

Trace Cách 4 (một lượt trái sang phải) với `n = 10203004`:

| vị trí | chữ số | hành động | `x` | `total` |
|---|---|---|---|---|
| 0 | 1 | `x = 0*10+1` | 1 | 1 |
| 1 | 0 | bỏ | 1 | 1 |
| 2 | 2 | `x = 1*10+2` | 12 | 3 |
| 3 | 0 | bỏ | 12 | 3 |
| 4 | 3 | `x = 12*10+3` | 123 | 6 |
| 5 | 0 | bỏ | 123 | 6 |
| 6 | 0 | bỏ | 123 | 6 |
| 7 | 4 | `x = 123*10+4` | 1234 | 10 |

Kết quả `1234 * 10 = 12340`.

````python
def trace_one_pass(n):
    x = total = 0
    print("n =", n)
    for i, c in enumerate(str(n)):
        d = int(c)
        if d:
            x, total = x * 10 + d, total + d
            act = f"x = x*10+{d}"
        else:
            act = "bỏ"
        print(f"  vị trí {i}: chữ số {d} -> {act:<10} x={x:<5} total={total}")
    print(f"  kết quả = {x} * {total} = {x * total}")
    return x * total

assert trace_one_pass(10203004) == 12340
assert trace_one_pass(1000) == 1
````

````python
def trace_divmod(n):
    x, place, total, step = 0, 1, 0, 0
    print("n =", n)
    while n > 0:
        step += 1
        n, d = divmod(n, 10)
        if d:
            x += d * place
            place *= 10
            total += d
        print(f"  bước {step}: d={d} -> x={x} place={place} total={total}")
    return x * total

assert trace_divmod(10203004) == 12340
````

## 9. Code nộp LeetCode

````python
from typing import List, Optional

class Solution:
    def sumAndMultiply(self, n: int) -> int:
        x = total = 0
        for c in str(n):
            d = int(c)
            if d:
                x = x * 10 + d
                total += d
        return x * total
````

## 10. Vì sao thuật toán đúng

- **`x` đúng thứ tự:** duyệt chuỗi từ trái sang phải, mỗi chữ số khác 0 được nối vào cuối bằng `x * 10 + d`, đúng như "nối các chữ số khác 0 theo thứ tự xuất hiện".
- **`total` đúng:** `sum` là tổng chữ số của `x`; chữ số của `x` chính là các chữ số khác 0 của `n`, nên `total` (cộng dồn các `d != 0`) bằng đúng tổng đó. (Chữ số 0 vốn cộng thêm 0 nên không ảnh hưởng.)
- **Biên `n = 0`:** chuỗi `'0'` chỉ có chữ số `0`, bị bỏ, nên `x = 0`, `total = 0`, kết quả `0` đúng theo đề.
- **Không tràn:** `n <= 10^9` có tối đa 10 chữ số nên `x < 10^10` và `total <= 90`, tích `< 10^12`, `int` của Python không tràn dù sao.
- **Kết quả không đổi khi dùng hai hướng:** cách `divmod` đặt chữ số vào đúng hàng `place`, cho cùng `x`.

## 13. Các lỗi người mới hay gặp

### Lỗi 1: dựng `x` bị ngược thứ tự

Lấy chữ số từ phải sang trái nhưng dựng `x` bằng `x * 10 + d` sẽ đảo ngược thứ tự: `1234` thành `4321`.

````python
def wrong_reversed(n):
    x = total = 0
    while n > 0:
        n, d = divmod(n, 10)
        if d:
            x = x * 10 + d               # SAI: đi từ phải nhưng dựng kiểu trái
            total += d
    return x * total

print(wrong_reversed(10203004), "!=", Solution().sumAndMultiply(10203004))
assert wrong_reversed(10203004) == 4321 * 10 != 12340
````

### Lỗi 2: quên xử lý `n = 0` khi dùng chuỗi

Sau khi lọc `'0'` ra khỏi `'0'` thì chuỗi rỗng; `int('')` gây `ValueError`. Đề **cho phép** `n = 0`.

````python
def wrong_no_zero_case(n):
    x = int(''.join(c for c in str(n) if c != '0'))       # SAI: lỗi khi n = 0
    return x * sum(int(c) for c in str(x))

try:
    wrong_no_zero_case(0)
except ValueError as e:
    print("ValueError:", e)
assert Solution().sumAndMultiply(0) == 0
````

### Lỗi 3: không loại chữ số 0 khi nối

Nếu dùng luôn `n` làm `x`, kết quả nhân với tổng chữ số của `n`, quá lớn so với `1234 * 10`.

````python
def wrong_keep_zeros(n):
    return n * sum(int(c) for c in str(n))               # SAI: x phải bỏ số 0

print(wrong_keep_zeros(10203004), "!=", Solution().sumAndMultiply(10203004))
assert wrong_keep_zeros(10203004) == 10203004 * 10 != 12340
````

### Lỗi 4: trả về `x + sum` hoặc `x` thay vì `x * sum`

Đọc vội đề: phép tính cuối cùng là **nhân**, không phải cộng.

````python
def wrong_add(n):
    kept = [int(c) for c in str(n) if c != '0']
    x = 0
    for d in kept:
        x = x * 10 + d
    return x + sum(kept)                                  # SAI: phải nhân

print(wrong_add(10203004), "!=", Solution().sumAndMultiply(10203004))
assert wrong_add(10203004) == 1244 != 12340
````

### Lỗi 5: tăng `place` cả khi gặp chữ số 0

Ở cách `divmod`, nếu `place *= 10` chạy với mọi chữ số (kể cả 0) thì các chữ số 0 vẫn "chiếm chỗ" và `x` không bị nén.

````python
def wrong_place_always(n):
    x, place, total = 0, 1, 0
    while n > 0:
        n, d = divmod(n, 10)
        x += d * place
        total += d
        place *= 10                                       # SAI: tăng cả khi d == 0
    return x * total

print(wrong_place_always(10203004), "!=", Solution().sumAndMultiply(10203004))
assert wrong_place_always(10203004) == 10203004 * 10 != 12340
````

## 14. Mẹo làm bài nhanh & nhận dạng pattern

- "Nối/lọc chữ số" -> chuỗi `str(n)` + lọc; hoặc `divmod` nếu muốn tránh chuỗi.
- Tổng chữ số của `x` (sau khi bỏ 0) bằng tổng chữ số của `n`, nên có thể tính `total` trong cùng vòng lặp.
- Luôn thử `n = 0`: nhiều lời giải chuỗi bị `int('')`.
- Muốn dựng số: trái sang phải -> `x * 10 + d`; phải sang trái -> `d * place`. **Không trộn hai hướng.**
- Số chữ số rất nhỏ (<= 10) nên không lo hiệu năng; ưu tiên đúng và dễ đọc.
- Phiên bản "II" của bài này hỏi **nhiều truy vấn** trên đoạn chữ số, cần mảng tiền tố.

## 19. Đáp án tham khảo cho bài tập

````python
# Bài 1, 2
assert Solution().sumAndMultiply(30405) == 4140 and Solution().sumAndMultiply(99) == 1782
assert Solution().sumAndMultiply(2002003) == 223 * 7 == 1561

# Bài 3
def digit_sum(n):
    total = 0
    while n > 0:
        n, d = divmod(n, 10)
        total += d
    return total
assert digit_sum(10203004) == 10 and digit_sum(0) == 0 and digit_sum(999) == 27

# Bài 4
def concat_nonzero(n):
    x = 0
    for c in str(n):
        if c != '0':
            x = x * 10 + int(c)
    return x
assert concat_nonzero(10203004) == 1234 and concat_nonzero(0) == 0 and concat_nonzero(1000) == 1
assert all(concat_nonzero(v) * digit_sum(v) == Solution().sumAndMultiply(v) for v in (10203004, 1000, 0, 909, 12))

# Bài 5
def sum_and_multiply_range(n, l, r):
    return Solution().sumAndMultiply(int(str(n)[l:r + 1])) if str(n)[l:r + 1] else 0
assert sum_and_multiply_range(10203004, 2, 5) == 115
assert sum_and_multiply_range(10203004, 0, 7) == Solution().sumAndMultiply(10203004)
print("đáp án bài tập OK")
````
