## 1. Đề bài và ví dụ

| `n` | Kết quả | Vì sao |
|---:|:---:|---|
| `1` | `True` | `2**0 = 1` |
| `16` | `True` | `2**4 = 16` |

| `0` | `False` | `2**x` luôn > 0 |
| `-8` | `False` | lũy thừa của 2 luôn dương |

**Constraints:** `-2^31 <= n <= 2^31 - 1`.

## 2. Nền tảng: `%`, `//`, `**`

- `a % b`: phần dư khi chia `a` cho `b`. `n % 2 == 0` nghĩa là `n` chẵn.
- `a // b`: chia lấy phần nguyên (floor division).
- `a ** b`: lũy thừa.

Một số là lũy thừa của 2 ⇔ chia liên tiếp cho 2 mà **luôn chia hết**, cuối cùng còn lại đúng `1`.

````python
n = 40
print(n % 2, n // 2)   # 0 20
n = 7
print(n % 2, n // 2)   # 1 3
print(2 ** 5)          # 32
````

## 3. Cách 1 — Chia liên tiếp cho 2 (baseline)

1. Nếu `n <= 0` → `False` (không có lũy thừa của 2 nào ≤ 0).
2. Trong khi `n` chẵn: `n //= 2`.
3. Cuối cùng `n == 1` thì là lũy thừa của 2.

````python
def is_power_of_two_loop(n: int) -> bool:
    if n <= 0:
        return False
    while n % 2 == 0:
        n //= 2
    return n == 1

for v in [1, 16, 3, 0, -8, 6, 1024]:
    print(v, "->", is_power_of_two_loop(v))
````

### Trace cách chia

Bảng cho `n = 24`:

| Bước | `n` | `n % 2` | Hành động |
|---|---|---|---|
| 0 | 24 | 0 | chia → 12 |
| 1 | 12 | 0 | chia → 6 |
| 2 | 6 | 0 | chia → 3 |
| 3 | 3 | 1 | dừng; `3 != 1` → `False` |

````python
def trace_loop(n: int) -> bool:
    print(f"Bắt đầu n = {n}")
    if n <= 0:
        print("  n <= 0 -> False")
        return False
    step = 0
    while n % 2 == 0:
        print(f"  bước {step}: n={n} chẵn -> n //= 2 -> {n // 2}")
        n //= 2
        step += 1
    print(f"  dừng: n={n} lẻ; n == 1? {n == 1}")
    return n == 1

trace_loop(24)
trace_loop(32)
````

### Nút thắt

- Số vòng lặp = số lần chia cho 2 ≈ `log2(n)` → tối đa ~31 vòng. Đã nhanh, nhưng Follow-up muốn **không vòng lặp**.
- Để làm được, ta cần nhìn số dưới dạng **nhị phân**.

## 4. Cách 2 — Đệ quy (cùng ý tưởng)

Đệ quy = hàm gọi lại chính nó với bài toán nhỏ hơn, có **điều kiện dừng** (base case).

- Base: `n == 1` → `True`; `n <= 0` hoặc `n` lẻ → `False`.
- Bước đệ quy: `n` chẵn → kết quả của `n // 2`.

````python
def is_power_of_two_rec(n: int) -> bool:
    if n == 1:
        return True
    if n <= 0 or n % 2 == 1:
        return False
    return is_power_of_two_rec(n // 2)

for v in [1, 16, 3, 0, -8, 6, 1024]:
    print(v, "->", is_power_of_two_rec(v))
````

Độ sâu đệ quy tối đa ~31 nên an toàn. Space `O(log n)` do call stack.

## 5. Nền tảng nhị phân — chìa khóa để bỏ vòng lặp

`bin(n)` trả chuỗi nhị phân với tiền tố `0b`. Hãy so sánh lũy thừa của 2 với các số khác:

````python
for v in [1, 2, 4, 8, 16, 3, 6, 12, 15]:
    print(f"{v:>3} = {bin(v):>8}   số bit 1 = {bin(v).count('1')}")
````

**Quan sát:** lũy thừa của 2 có **đúng một bit 1** (`1`, `10`, `100`, `1000`, ...). Các số khác có ≥ 2 bit 1.

Vậy câu hỏi trở thành: *`n > 0` và có đúng một bit 1 không?*

### Cách 3 — Đếm bit 1

````python
def is_power_of_two_bincount(n: int) -> bool:
    return n > 0 and bin(n).count("1") == 1

for v in [1, 16, 3, 0, -8, 6, 1024]:
    print(v, "->", is_power_of_two_bincount(v))
````

Chú ý `bin(-8)` là `'-0b1000'` — cũng có một chữ `1`! Nên điều kiện `n > 0` là **bắt buộc**.

````python
print(bin(-8), bin(-8).count("1"))
````

## 6. Toán tử bit cần biết

| Toán tử | Ý nghĩa | Ví dụ |
|---|---|---|
| `a & b` | AND từng bit: bit kết quả = 1 khi cả hai bit là 1 | `12 & 10` = `1100 & 1010` = `1000` = 8 |
| `a \| b` | OR từng bit | `12 \| 10` = `1110` = 14 |
| `a >> k` | dịch phải k bit (≈ `// 2**k`) | `12 >> 2` = 3 |
| `-a` | số đối (biểu diễn bù 2) | dùng trong mẹo `n & -n` |

````python
a, b = 12, 10
print(f"{a:04b} & {b:04b} = {a & b:04b} ({a & b})")
print(f"{a:04b} | {b:04b} = {a | b:04b} ({a | b})")
print(f"{a:04b} >> 2  = {a >> 2:04b} ({a >> 2})")
````

`f"{a:04b}"` định dạng số ở hệ nhị phân (`b`), độ rộng 4, đệm số 0 ở trái.

## 7. Insight — `n - 1` lật bit 1 thấp nhất và mọi bit 0 phía sau

Lấy `n = 12 = 1100`. Trừ 1: `11 = 1011`. Bit 1 thấp nhất (vị trí 2) thành 0, các bit 0 sau nó thành 1, phần trước giữ nguyên.

Do đó `n & (n - 1)` = `n` **bỏ đi bit 1 thấp nhất**:

```text
n       = 1100
n - 1   = 1011
n&(n-1) = 1000
```

Nếu `n` chỉ có **một** bit 1, bỏ nó đi thì còn `0`. Nếu có ≥ 2 bit 1 thì vẫn còn bit khác → khác 0.

````python
for v in [12, 16, 6, 1, 7, 64]:
    print(f"n={v:>2} = {v:07b} | n-1 = {v-1:07b} | n&(n-1) = {v & (v-1):07b} -> {v & (v-1)}")
````

Với `16 = 0010000`, `n & (n-1) = 0` → lũy thừa của 2. Với `12`, kết quả `8 != 0` → không phải.

## 8. Lời giải chính — `n > 0 and n & (n - 1) == 0`

Ghép lại:

1. `n > 0`: loại 0 và số âm (với `n = 0`, `0 & -1 = 0` sẽ cho kết quả sai nếu thiếu điều kiện này).
2. `n & (n - 1) == 0`: đúng một bit 1.

**Cẩn thận độ ưu tiên toán tử:** thứ tự trong Python là `-` (trừ) > `&` > `==`. Nên `n & n - 1 == 0` được hiểu là `(n & (n - 1)) == 0`. Tuy vậy, ở C/C++/Java thì `==` lại cao hơn `&`, cùng biểu thức sẽ thành `n & ((n - 1) == 0)` — sai. Thói quen tốt: **luôn đặt ngoặc** `(n & (n - 1)) == 0`.

Ô dưới cũng cho thấy vì sao cần `n > 0`: với `n = 0`, `0 & -1 = 0` nên phép bit một mình trả `True` sai.

````python
n = 12
print("Không ngoặc :", n & n - 1 == 0)        # Python hiểu là (n & (n-1)) == 0
print("Có ngoặc    :", (n & (n - 1)) == 0)
n = 0
print("n=0, thiếu n>0:", (n & (n - 1)) == 0, "<- sai!")
````

````python
def is_power_of_two_bit(n: int) -> bool:
    return n > 0 and (n & (n - 1)) == 0

for v in [1, 16, 3, 0, -8, 6, 1024, 2**30, 2**31 - 1]:
    print(v, "->", is_power_of_two_bit(v))
````

## 9. Trace mẹo bit

````python
def trace_bit(n: int) -> bool:
    print(f"n = {n}")
    if n <= 0:
        print("  n <= 0 -> False ngay")
        return False
    w = max(n.bit_length(), 1)
    print(f"  n      = {n:0{w}b}")
    print(f"  n - 1  = {n-1:0{w}b}")
    r = n & (n - 1)
    print(f"  n&(n-1)= {r:0{w}b}  -> {'bằng 0: True' if r == 0 else 'khác 0: False'}")
    return r == 0

trace_bit(40)
trace_bit(32)
trace_bit(0)
trace_bit(1)
````

`n.bit_length()` cho số bit cần để biểu diễn `n` (ví dụ `40 -> 6`). `f"{x:0{w}b}"` lồng biến `w` vào định dạng.

## 10. Cách 5 — `n & -n` (cô lập bit thấp nhất) và cách "chia số lớn nhất"

**`n & -n`** cho giá trị của bit 1 thấp nhất. Nếu bit đó chính là `n` thì `n` chỉ có một bit 1.

````python
for v in [12, 16, 40, 1]:
    print(f"n={v:>2}: n & -n = {v & -v}  -> power of two? {v & -v == v}")

def is_power_of_two_lowbit(n: int) -> bool:
    return n > 0 and (n & -n) == n

print([is_power_of_two_lowbit(v) for v in [1, 16, 3, 0, -8]])
````

### Vì sao `n & -n` cô lập bit thấp nhất? — biểu diễn bù 2

Máy tính lưu số âm theo **bù 2 (two's complement)**: `-n = (~n) + 1`, trong đó `~n` lật mọi bit.

```text
n      = 0010 1000   (40)
~n     = 1101 0111
~n + 1 = 1101 1000   (-40)
n & -n = 0000 1000   (8)  <- chỉ còn bit 1 thấp nhất
```

Lật bit rồi cộng 1 làm "dòng nhớ" chạy qua các bit 0 ở cuối cho tới đúng bit 1 thấp nhất. Phía trên bit đó, `n` và `-n` ngược nhau hoàn toàn nên AND ra 0. Ta kiểm chứng bằng 8 bit thấp (`& 0xFF` để cắt bớt các bit cao vô hạn của Python):

````python
n = 40
print(f"n      = {n & 0xFF:08b}")
print(f"~n     = {~n & 0xFF:08b}")
print(f"-n     = {-n & 0xFF:08b}")
print(f"n & -n = {n & -n:08b} = {n & -n}")
print("~n + 1 == -n ?", ~n + 1 == -n)
````

**Chia số lớn nhất:** trong miền int32, lũy thừa của 2 lớn nhất là `2**30`. Mọi ước dương của `2**30` đều là lũy thừa của 2 (vì phân tích thừa số nguyên tố chỉ có số 2). Vậy `n > 0 and 2**30 % n == 0`.

````python
def is_power_of_two_divisor(n: int) -> bool:
    return n > 0 and (1 << 30) % n == 0

print(1 << 30 == 2**30)          # 1 << k = 2**k
print([is_power_of_two_divisor(v) for v in [1, 16, 3, 0, -8, 2**30]])
````

`1 << 30` là dịch bit 1 sang trái 30 vị trí = `2**30`. Cách này chỉ đúng **vì constraints giới hạn int32**; nếu `n` có thể lớn hơn `2**30` (như `2**40`) thì cách này sai.

## 11. Code nộp LeetCode

````python
class Solution:
    def isPowerOfTwo(self, n: int) -> bool:
        return n > 0 and (n & (n - 1)) == 0

sol = Solution()
print(sol.isPowerOfTwo(1), sol.isPowerOfTwo(16), sol.isPowerOfTwo(3))
````

```python
        return n == 1
```

## 12. Vì sao đúng? + Độ phức tạp

**Mệnh đề:** với `n > 0`, `n` là lũy thừa của 2 ⇔ `n` có đúng một bit 1 ⇔ `n & (n-1) == 0`.

- Chiều 1: `2**x` trong nhị phân là `1` theo sau `x` số 0 → đúng một bit 1.
- Chiều 2: số có đúng một bit 1 ở vị trí `x` có giá trị `2**x`.
- Chiều 3: `n - 1` giữ nguyên các bit cao hơn bit 1 thấp nhất, nên `n & (n-1)` = `n` bỏ bit thấp nhất. Kết quả = 0 ⇔ không còn bit nào khác ⇔ đúng một bit.

**Cách chia vòng lặp — invariant:** "`n` ban đầu là lũy thừa của 2 ⇔ `n` hiện tại là lũy thừa của 2", vì chia một số chẵn cho 2 bảo toàn tính chất này; dừng khi `n` lẻ, và số lẻ duy nhất là lũy thừa của 2 là `1`.

| Cách | Time | Space | Ghi chú |
|---|---|---|---|
| Chia liên tiếp | `O(log n)` | `O(1)` | baseline dễ nghĩ |
| Đệ quy | `O(log n)` | `O(log n)` stack | minh họa đệ quy |
| `bin(n).count('1')` | `O(log n)` (tạo chuỗi) | `O(log n)` | ngắn, cần `n > 0` |
| `n & (n-1)` | `O(1)` | `O(1)` | **đáp án Follow-up** |
| `n & -n` / `2**30 % n` | `O(1)` | `O(1)` | cách thứ hai dựa vào int32 |

(`log n` ở đây ≤ 31, nên thực tế mọi cách đều rất nhanh.)

## 15. Mẹo nhận dạng pattern

- "Có phải lũy thừa của 2?", "đúng một bit 1", "số bit 1" → nghĩ tới **`n & (n - 1)`**.
- Đếm bit 1 (LeetCode 191): lặp `n &= n - 1` cho đến khi `n == 0`, số lần lặp = số bit 1.
- "Lũy thừa của 3/5/..." không có mẹo bit, nhưng có mẹo **ước của lũy thừa lớn nhất** (`3**19 % n == 0`, vì 3 là số nguyên tố).
- "Lũy thừa của 4" = lũy thừa của 2 **và** bit 1 nằm ở vị trí chẵn.

## 16. Biến thể

````python
def count_ones(n: int) -> int:
    """Đếm bit 1 bằng cách xóa dần bit thấp nhất (Brian Kernighan)."""
    cnt = 0
    while n:
        n &= n - 1
        cnt += 1
    return cnt

def is_power_of_three(n: int) -> bool:
    return n > 0 and 3**19 % n == 0       # 3**19 là lũy thừa của 3 lớn nhất trong int32

print([count_ones(v) for v in [0, 1, 7, 8, 255]])
print([is_power_of_three(v) for v in [1, 3, 9, 27, 6, 0, 45]])
````

| 5 | Số bit của `n-1` | `1 << k` | `1 << (n - 1).bit_length()` |

### Đáp án tham khảo

Bài 1: `bin(20) = '0b10100'`, `20 & 19 = 16`, `20 & -20 = 4`, `1 << 5 = 32`.

````python
def answer_power_two_loop(n):
    if n <= 0:
        return False
    while n % 2 == 0:
        n //= 2
    return n == 1

def answer_power_two_bit(n):
    return n > 0 and (n & (n - 1)) == 0

def answer_power_of_four(n):
    return n > 0 and (n & (n - 1)) == 0 and (n & 0x55555555) != 0

def answer_next_power_of_two(n):
    return 1 << (n - 1).bit_length()

def answer_powers_with_exponent(nums):
    return [(v, v.bit_length() - 1) for v in nums if v > 0 and (v & (v - 1)) == 0]

print(bin(20), 20 & 19, 20 & -20, 1 << 5)
assert answer_power_two_loop(16) and not answer_power_two_loop(0)
assert answer_power_two_bit(1) and not answer_power_two_bit(6)
assert answer_power_of_four(16) and answer_power_of_four(1) and not answer_power_of_four(8)
assert [v for v in range(1, 2000) if answer_power_of_four(v)] == [1, 4, 16, 64, 256, 1024]
assert answer_next_power_of_two(5) == 8 and answer_next_power_of_two(8) == 8 and answer_next_power_of_two(1) == 1
assert all(answer_next_power_of_two(k) == min(p for p in power_set if p >= k) for k in range(1, 3000))
assert answer_powers_with_exponent([3, 8, 1, 0, 64]) == [(8, 3), (1, 0), (64, 6)]
print("Đáp án tham khảo: tất cả test PASS")
````
