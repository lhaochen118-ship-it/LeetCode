## 2. Đề bài & ví dụ

Cho chuỗi `s` và **một ký tự** `letter`. Trả về **phần trăm** các ký tự trong `s` bằng `letter`, **làm tròn xuống** thành số nguyên gần nhất (tức lấy phần nguyên, bỏ phần thập phân).

| Ví dụ | s | letter | Output | Giải thích |
|---|---|---|---|---|
| 1 | `"foobar"` | `"o"` | `33` | có 2 chữ `o` trên 6 ký tự: `2/6 = 33.33...%` → `33` |
| 2 | `"jjjj"` | `"k"` | `0` | không có `k` nào: `0%` |

Constraints: `1 <= len(s) <= 100`; `s` chỉ gồm chữ cái tiếng Anh thường; `letter` là một chữ cái thường.

Công thức: `đáp án = floor(count(letter) * 100 / len(s))`.

````python
examples = [("foobar", "o", 33), ("jjjj", "k", 0)]
for s, letter, expected in examples:
    count = 0
    for ch in s:
        if ch == letter:
            count += 1
    print(f"s={s!r:9} letter={letter!r} count={count} n={len(s)} -> {count}/{len(s)} = {count / len(s) * 100:.2f}% ; mong đợi {expected}")
````

## 3. Kiến thức nền

### 3.1. Duyệt chuỗi và so sánh ký tự

Chuỗi là dãy ký tự; `for ch in s` lấy từng ký tự. Một ký tự cũng là chuỗi độ dài 1, so sánh bằng `==`.

````python
s = "foobar"
for i, ch in enumerate(s):
    print(i, ch, ch == "o")
````

### 3.2. Phương thức `str.count`

`s.count(sub)` đếm số lần `sub` xuất hiện **không chồng lấp**. Với `sub` là một ký tự, đó chính là số ký tự bằng nó. Độ phức tạp `O(n)` nhưng chạy bằng C nên rất nhanh.

````python
print("foobar".count("o"), "jjjj".count("k"), "banana".count("a"))
print("aaaa".count("aa"))     # không chồng lấp: chỉ 2, không phải 3
````

### 3.3. `sum` với biểu thức sinh (generator expression)

`True` bằng `1`, `False` bằng `0` khi cộng. Vì vậy `sum(ch == letter for ch in s)` đếm số ký tự thỏa điều kiện mà không cần danh sách trung gian.

````python
print(True + True, False + 1)
print(sum(ch == "o" for ch in "foobar"))
print(sum(1 for ch in "foobar" if ch == "o"))
````

### 3.4. `collections.Counter`

`Counter(s)` tạo từ điển đếm số lần mỗi ký tự. Tra khóa vắng mặt trả `0` thay vì lỗi, rất hợp bài này.

````python
from collections import Counter
c = Counter("foobar")
print(c)
print(c["o"], c["k"])       # 'k' vắng mặt -> 0
````

### 3.5. Chia nguyên `//` và chia thực `/`

- `7 / 2 == 3.5` (số thực).
- `7 // 2 == 3` (làm tròn **xuống**, kết quả nguyên).
- Với số **không âm**, `a // b` đúng bằng `floor(a / b)` và hoàn toàn chính xác (không có sai số thực).

Ta muốn `floor(count * 100 / n)`. Nhân `100` **trước** rồi chia nguyên: `count * 100 // n`.

````python
print(7 / 2, 7 // 2, 200 // 6, 2 * 100 // 6)
print(int(33.99), int(-0.5), round(33.6))     # int cắt phần thập phân, round làm tròn
````

### 3.6. Vì sao tránh số thực?

Số thực nhị phân không biểu diễn chính xác mọi số thập phân. `0.29 * 100` cho `28.999999999999996`; `int()` của nó là `28`, **sai** (đúng là `29`). Dùng số nguyên thì không bao giờ gặp vấn đề này.

````python
print(0.29 * 100, int(0.29 * 100))
print(29 / 100 * 100, int(29 / 100 * 100))
print(29 * 100 // 100)
````

## 4. Ý tưởng nền tảng / cách nghĩ trực giác

Tính tay ví dụ 1: `s = "foobar"`, `letter = "o"`.

| bước | ký tự | `== 'o'`? | `count` |
|---|---|---|---|
| 1 | `f` | không | 0 |
| 2 | `o` | có | 1 |
| 3 | `o` | có | 2 |
| 4 | `b` | không | 2 |

`count = 2`, `n = 6`. `2 * 100 = 200`, `200 // 6 = 33` (vì `6 * 33 = 198 <= 200 < 204 = 6 * 34`). Vậy đáp án `33`.

Ba bước: **đếm → nhân 100 → chia nguyên cho độ dài**.

````python
s, letter = "foobar", "o"
count = 0
for ch in s:
    if ch == letter:
        count += 1
    print(f"ký tự {ch!r}: count={count}")
print("count*100 =", count * 100, "; // len =", count * 100 // len(s))
````

## 5. Cách 1 — Brute Force / baseline: vòng `for` đếm thủ công

Duyệt từng ký tự, tăng biến đếm mỗi khi bằng `letter`, rồi tính phần trăm bằng số nguyên.

````python
def percentage_loop(s, letter):
    count = 0
    for ch in s:
        if ch == letter:
            count += 1
    return count * 100 // len(s)

print(percentage_loop("foobar", "o"), percentage_loop("jjjj", "k"))
````

````python
from typing import List

class SolutionLoop:
    def percentageLetter(self, s: str, letter: str) -> int:
        count = 0
        for ch in s:
            if ch == letter:
                count += 1
        return count * 100 // len(s)

print(SolutionLoop().percentageLetter("foobar", "o"))
print(SolutionLoop().percentageLetter("jjjj", "k"))
````

**Độ phức tạp:** `n = len(s)`. Một lượt duyệt → **Time `O(n)`**, chỉ vài biến → **Space `O(1)`**.

**Nút thắt:** thuật toán đã tối ưu về bậc; chỉ còn cách làm **ngắn gọn hơn** (dùng hàm có sẵn) và **đúng hơn** (tránh số thực).

## 6. Ý tưởng thuật toán / bước đột phá

Hai insight:

- **Đếm** là việc cốt lõi; mọi cách chỉ khác nhau ở *công cụ đếm* (`for`, `count`, `sum`, `Counter`, lọc).
- **Tính phần trăm bằng số nguyên:** `floor(c * 100 / n) = (c * 100) // n` với `c, n` nguyên không âm. Không cần `float`, không cần `math.floor`.

Quy tắc: nếu `letter` không xuất hiện → `c = 0` → kết quả `0`. Nếu mọi ký tự bằng `letter` → `c = n` → kết quả `100`. Khoảng giá trị luôn là `0..100`.

## 7. Các cách tiếp theo

### Cách 2 — `str.count` (ngắn nhất, chạy bằng C)

````python
from typing import List

class SolutionCount:
    def percentageLetter(self, s: str, letter: str) -> int:
        return s.count(letter) * 100 // len(s)

print(SolutionCount().percentageLetter("foobar", "o"))
print(SolutionCount().percentageLetter("jjjj", "k"))
````

**Độ phức tạp:** Time `O(n)`, Space `O(1)`. Đây là cách thông dụng nhất khi nộp.

### Cách 3 — `sum` với biểu thức sinh

Không tạo danh sách, mỗi ký tự khớp đóng góp `1`. Cách này mở rộng dễ cho điều kiện phức tạp hơn (ví dụ "ký tự là nguyên âm").

````python
from typing import List

class SolutionSum:
    def percentageLetter(self, s: str, letter: str) -> int:
        count = sum(1 for ch in s if ch == letter)
        return count * 100 // len(s)

print(SolutionSum().percentageLetter("foobar", "o"))
print(SolutionSum().percentageLetter("jjjj", "k"))
````

**Độ phức tạp:** Time `O(n)`, Space `O(1)`.

### Cách 4 — `collections.Counter`

Dựng bảng đếm một lần rồi tra `letter`. Hữu ích khi cần hỏi **nhiều** chữ cái trên cùng một chuỗi: dựng `Counter` một lần, mỗi truy vấn `O(1)`.

````python
from collections import Counter
from typing import List

class SolutionCounter:
    def percentageLetter(self, s: str, letter: str) -> int:
        freq = Counter(s)
        return freq[letter] * 100 // len(s)

print(SolutionCounter().percentageLetter("foobar", "o"))
print(SolutionCounter().percentageLetter("jjjj", "k"))
````

**Độ phức tạp:** Time `O(n)` để dựng, tra `O(1)`; Space `O(k)` với `k ≤ 26` chữ cái khác nhau, tức hằng số.

### Cách 5 — Lọc ra danh sách rồi dùng `len`

Lọc các ký tự bằng `letter` thành một danh sách, lấy `len`. Tốn bộ nhớ hơn nhưng rất dễ đọc: "những ký tự khớp" / "tất cả ký tự".

````python
from typing import List

class SolutionFilter:
    def percentageLetter(self, s: str, letter: str) -> int:
        matches = [ch for ch in s if ch == letter]
        return len(matches) * 100 // len(s)

print(SolutionFilter().percentageLetter("foobar", "o"))
print(SolutionFilter().percentageLetter("jjjj", "k"))
````

**Độ phức tạp:** Time `O(n)`, Space `O(c)` với `c` là số ký tự khớp (tối đa `n`).

## 8. TRACE từng bước

Trace Cách 1 với `s = "banana"`, `letter = "a"` (đáp án `3 * 100 // 6 = 50`):

| bước | `ch` | khớp? | `count` |
|---|---|---|---|
| 1 | `b` | không | 0 |
| 2 | `a` | có | 1 |
| 3 | `n` | không | 1 |
| 4 | `a` | có | 2 |
| 5 | `n` | không | 2 |
| 6 | `a` | có | 3 |

Sau vòng lặp: `count * 100 = 300`, `300 // 6 = 50`.

````python
def trace_percentage(s, letter):
    print(f"s={s!r}, letter={letter!r}, n={len(s)}")
    count = 0
    for step, ch in enumerate(s, 1):
        match = ch == letter
        if match:
            count += 1
        print(f"  bước {step}: ch={ch!r} khớp={match!s:5} count={count}")
    result = count * 100 // len(s)
    print(f"  count*100 = {count * 100}; // {len(s)} = {result}  (chính xác {count * 100 / len(s):.3f})")
    return result

print("kết quả:", trace_percentage("banana", "a"))
print("kết quả:", trace_percentage("foobar", "o"))
````

## 9. Code nộp LeetCode

````python
from typing import List

class Solution:
    def percentageLetter(self, s: str, letter: str) -> int:
        return s.count(letter) * 100 // len(s)

print(Solution().percentageLetter("foobar", "o"))
print(Solution().percentageLetter("jjjj", "k"))
````

## 10. Vì sao thuật toán đúng

- **Đếm đúng:** `count` bằng số vị trí `i` có `s[i] == letter`. Mỗi cách (vòng lặp, `count`, `sum`, `Counter`, lọc) đều đếm đúng tập vị trí đó; `str.count` với một ký tự không có vấn đề chồng lấp.
- **Phần trăm đúng:** tỉ lệ là `count / n`, tức `count * 100 / n` phần trăm. Với số nguyên không âm `a` và `b > 0`, `a // b` là số nguyên lớn nhất `q` thỏa `q * b <= a`, đó chính xác là `floor(a / b)`. Ở đây `a = count * 100`, `b = n`.
- **Không chia cho 0:** đề đảm bảo `len(s) >= 1`.
- **Không sai số thực:** mọi phép tính đều trên số nguyên, nên không có lỗi kiểu `28.999999999999996`.
- **Biên:** `count = 0` cho `0`; `count = n` cho `100 * n // n = 100`.

Đặt `n = len(s)` (`≤ 100`).

| Cách | Time | Space |

## 12. Bộ test

Ví dụ đề, edge case (một ký tự, toàn bộ giống nhau, không có, đúng tỉ lệ chẵn) và test ngẫu nhiên đối chiếu **tất cả cách** với phép tính bằng `fractions.Fraction` (chính xác tuyệt đối).

````python
import random
from fractions import Fraction
from math import floor

solvers = [SolutionLoop(), SolutionCount(), SolutionSum(), SolutionCounter(), SolutionFilter(), Solution()]

fixed = [
    ("foobar", "o", 33),
    ("jjjj", "k", 0),
    ("a", "a", 100),                 # một ký tự, khớp
    ("a", "b", 0),                   # một ký tự, không khớp
    ("aaaa", "a", 100),              # toàn bộ khớp
    ("abab", "a", 50),               # chẵn
    ("abc", "a", 33),                # 33.33 -> cắt xuống
    ("abc", "c", 33),
    ("aabbcc", "b", 33),
    ("a" * 29 + "b" * 71, "a", 29),  # 29% : chỗ số thực dễ sai
    ("a" * 57 + "b" * 43, "a", 57),
    ("zzzzzzzzzy", "y", 10),
    ("abcdefghijklmnopqrstuvwxyz", "m", 3),   # 100/26 = 3.84 -> 3 (không làm tròn lên)
]
for s, letter, expected in fixed:
    for sol in solvers:
        assert sol.percentageLetter(s, letter) == expected, (type(sol).__name__, s, letter, expected)
print("test cố định: OK")

def reference(s, letter):
    return floor(Fraction(s.count(letter) * 100, len(s)))

random.seed(2278)
for _ in range(700):
    n = random.randint(1, 100)
    alphabet = random.choice(["ab", "abc", "abcdefghijklmnopqrstuvwxyz"])
    s = "".join(random.choice(alphabet) for _ in range(n))
    letter = random.choice("abcdefghijklmnopqrstuvwxyz")
    expected = reference(s, letter)
    for sol in solvers:
        assert sol.percentageLetter(s, letter) == expected, (type(sol).__name__, s, letter)
print("700 test ngẫu nhiên: OK")

# thử với mọi letter có thể trên một chuỗi cố định: tổng các phần trăm không vượt 100 (mỗi cái đã cắt xuống)
text = "thequickbrownfoxjumpsoverthelazydog"
parts = [Solution().percentageLetter(text, ch) for ch in "abcdefghijklmnopqrstuvwxyz"]
assert sum(parts) <= 100 and sum(parts) >= 100 - 26
print("tổng phần trăm các chữ cái của", text, "=", sum(parts), "(<= 100 vì mỗi số bị cắt xuống)")

# quét mọi (count, n): đếm bao nhiêu cặp mà cách dùng số thực int(c / n * 100) sai
bad = [(c, n) for n in range(1, 101) for c in range(0, n + 1) if int(c / n * 100) != c * 100 // n]
print("số cặp (count, n) mà int(c / n * 100) sai:", len(bad), "ví dụ:", bad[:5])
````

## 13. Các lỗi người mới hay gặp

### Lỗi 1: Chia thực rồi `int` — dính sai số thực

`int(count / n * 100)` đôi khi thấp hơn đáp án đúng 1 đơn vị.

````python
def wrong1(s, letter):
    return int(s.count(letter) / len(s) * 100)

s = "a" * 29 + "b" * 71
print("sai :", wrong1(s, "a"), "(đáng lẽ 29)")
print("đúng:", Solution().percentageLetter(s, "a"))
print("các cặp (count, n) bị sai đầu tiên:", bad[:6])
````

### Lỗi 2: Chia nguyên trước khi nhân 100

`count // n * 100` — `count // n` chỉ là `0` hoặc `1`, mất hết phần thập phân.

````python
def wrong2(s, letter):
    return s.count(letter) // len(s) * 100

print("sai :", wrong2("foobar", "o"), "(đáng lẽ 33)")
print("đúng:", Solution().percentageLetter("foobar", "o"))
````

### Lỗi 3: Dùng `round` thay vì làm tròn xuống

Đề yêu cầu **làm tròn xuống** (cắt phần thập phân), không phải làm tròn gần nhất.

````python
def wrong3(s, letter):
    return round(s.count(letter) * 100 / len(s))

print("sai :", wrong3("aab", "a"), "(66.67 bị làm tròn lên 67; đáng lẽ 66)")
print("đúng:", Solution().percentageLetter("aab", "a"))
````

### Lỗi 4: Quên nhân 100 hoặc trả về phân số

Đề cần **số nguyên phần trăm**, không phải tỉ lệ `0.33`.

````python
def wrong4(s, letter):
    return s.count(letter) / len(s)

print("sai :", wrong4("foobar", "o"), "(đây là tỉ lệ, không phải phần trăm nguyên)")
print("đúng:", Solution().percentageLetter("foobar", "o"))
````

### Lỗi 5: Chia cho số ký tự khác `letter` hoặc cho độ dài sai

Mẫu số phải là **toàn bộ độ dài** `len(s)`, không phải số ký tự còn lại.

````python
def wrong5(s, letter):
    count = s.count(letter)
    others = len(s) - count
    return count * 100 // others if others else 100

print("sai :", wrong5("foobar", "o"), "(2*100//4 = 50; đáng lẽ 33)")
print("đúng:", Solution().percentageLetter("foobar", "o"))
````

## 14. Mẹo làm bài nhanh & nhận dạng pattern

- "Phần trăm / tỉ lệ làm tròn xuống" → `a * 100 // b` bằng **số nguyên**, nhân trước chia sau.
- "Đếm số lần một giá trị xuất hiện" → `s.count(x)` (chuỗi/danh sách) hoặc `Counter`.
- Điều kiện đếm phức tạp → `sum(1 for x in data if điều_kiện)`.
- Luôn tự hỏi "mẫu số có thể bằng `0` không?" — ở đây `len(s) >= 1` nên an toàn.
- Nhớ `True == 1` để cộng boolean trực tiếp.

## 15. Biến thể của bài

1. Phần trăm **làm tròn lên** (`ceil`): `(c * 100 + n - 1) // n`.
2. Phần trăm của **nhiều** chữ cái / tập ký tự (`sum(ch in letters for ch in s)`).
3. Trả về phần trăm với **hai chữ số thập phân** (dùng số nguyên `c * 10000 // n`, rồi định dạng).
4. Tìm chữ cái có phần trăm **cao nhất** (`Counter.most_common(1)`).
5. Các bài đếm ký tự khác như 387 First Unique Character in a String, 242 Valid Anagram.

## 19. Đáp án tham khảo cho bài tập

Chạy kiểm tra bằng `assert`.

````python
from collections import Counter

# Bài 1 và 2
assert (7 // 2, 7 / 2, int(5.99), 150 // 4) == (3, 3.5, 5, 37)
assert Solution().percentageLetter("mississippi", "s") == 36      # 4*100//11 = 36

# Bài 3
def percentage_ceil(s, letter):
    count = s.count(letter)
    n = len(s)
    return (count * 100 + n - 1) // n

# Bài 4
def percentage_vowels(s):
    return sum(ch in "aeiou" for ch in s) * 100 // len(s)

# Bài 5
def top_letter(s):
    freq = Counter(s)
    best = min(freq, key=lambda ch: (-freq[ch], ch))
    return best, freq[best] * 100 // len(s)

assert percentage_ceil("foobar", "o") == 34
assert percentage_ceil("abab", "a") == 50
assert percentage_ceil("jjjj", "k") == 0
assert percentage_vowels("education") == 55          # 5/9 = 55.5
assert percentage_vowels("xyz") == 0
assert top_letter("banana") == ("a", 50)
assert top_letter("abab") == ("a", 50)               # hòa: chọn 'a'
assert top_letter("z") == ("z", 100)
print("tất cả bài tập OK")
````
