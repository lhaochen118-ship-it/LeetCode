## 1. Bài toán yêu cầu gì?

Cho chuỗi `s` chỉ gồm `'0'` và `'1'`. Mỗi thao tác: đổi một ký tự `'0'` thành `'1'` hoặc ngược lại.

Chuỗi **xen kẽ** nếu không có hai ký tự kề nhau giống nhau. Ví dụ `"010"` xen kẽ, `"0100"` thì không.

Trả về **số thao tác ít nhất** để `s` thành chuỗi xen kẽ.

| s | Kết quả | Giải thích |
|---|---:|---|
| `"0100"` | 1 | đổi ký tự cuối → `"0101"` |
| `"10"` | 0 | đã xen kẽ |
| `"1111"` | 2 | thành `"0101"` hoặc `"1010"` |

Constraints: `1 <= s.length <= 10^4`, `s[i]` là `'0'` hoặc `'1'`.

### Đáp án cuối cùng trông như thế nào?

Đây là gợi ý quan trọng nhất của đề: hãy tưởng tượng chuỗi **sau khi** đã sửa xong.

- Nếu ký tự đầu là `'0'` thì ký tự thứ hai buộc là `'1'`, thứ ba buộc là `'0'`… → `"0101…"`.
- Nếu ký tự đầu là `'1'` → `"1010…"`.

Chỉ có **2 ứng viên** cho chuỗi đích, cùng độ dài với `s`.

````python
s = "0100"
n = len(s)
target0 = "".join("01"[i % 2] for i in range(n))
target1 = "".join("10"[i % 2] for i in range(n))
print("s       =", s)
print("dạng 01 =", target0)
print("dạng 10 =", target1)
````

## 3. Cách trực giác — dựng cả hai chuỗi đích, đếm khác biệt

1. Dựng `target0 = "0101…"`, `target1 = "1010…"` (độ dài `n`).
2. `cost0` = số vị trí `s` khác `target0`; `cost1` tương tự.
3. Trả `min(cost0, cost1)`.

Mỗi vị trí khác nhau cần đúng **một** thao tác (đổi ký tự đó), và vị trí giống nhau thì không cần đụng tới.

````python
def min_operations_targets(s: str) -> int:
    n = len(s)
    target0 = ("01" * (n // 2 + 1))[:n]
    target1 = ("10" * (n // 2 + 1))[:n]
    cost0 = sum(a != b for a, b in zip(s, target0))
    cost1 = sum(a != b for a, b in zip(s, target1))
    return min(cost0, cost1)

print(min_operations_targets("0100"), min_operations_targets("10"), min_operations_targets("1111"))
````

### 3.1. Baseline "vét cạn thật sự" (chỉ để kiểm chứng với `n` nhỏ)

Thử **mọi** chuỗi nhị phân cùng độ dài (`2^n` chuỗi), giữ những chuỗi xen kẽ, tính số thao tác, lấy min. Rất chậm (`n = 20` đã hơn 1 triệu chuỗi) nhưng đúng theo định nghĩa → dùng làm **oracle** để kiểm tra các cách nhanh.

````python
from itertools import product

def is_alternating(t: str) -> bool:
    return all(t[i] != t[i + 1] for i in range(len(t) - 1))

def min_operations_exhaustive(s: str) -> int:
    best = None
    for bits in product("01", repeat=len(s)):
        t = "".join(bits)
        if is_alternating(t):
            cost = sum(a != b for a, b in zip(s, t))
            best = cost if best is None else min(best, cost)
    return best

print(min_operations_exhaustive("0100"), min_operations_exhaustive("1111"))
print("số chuỗi xen kẽ độ dài 4:", sum(is_alternating("".join(b)) for b in product("01", repeat=4)))
````

Kết quả cuối cho thấy: trong 16 chuỗi độ dài 4 chỉ có **2** chuỗi xen kẽ — đúng như nhận xét "chỉ có 2 dạng".

## 4. Nút thắt và insight

**Nút thắt 1:** vét cạn `2^n` vô vọng với `n = 10^4`. Nhận xét "chỉ 2 dạng" đưa về `O(n)`.

**Nút thắt 2:** dựng hai chuỗi đích tốn thêm `O(n)` bộ nhớ, và ta duyệt `s` hai lần.

**Insight 1 (chẵn lẻ):** ký tự mong muốn của dạng `"0101…"` tại vị trí `i` là `str(i % 2)` — tính ngay, không cần dựng chuỗi.

**Insight 2 (bù):** tại mỗi vị trí, `s[i]` **hoặc** khớp dạng 01 **hoặc** khớp dạng 10 (hai dạng luôn khác nhau ở mọi vị trí). Vậy:

```text
cost0 + cost1 = n    →    cost1 = n - cost0
```

Chỉ cần đếm `cost0` trong **một lần duyệt**, đáp án là `min(cost0, n - cost0)`.

````python
for s in ["0100", "10", "1111", "110010"]:
    c0 = sum(ch != str(i % 2) for i, ch in enumerate(s))
    c1 = sum(ch != str(1 - i % 2) for i, ch in enumerate(s))
    print(f"s={s:<7} cost0={c0} cost1={c1} cost0+cost1={c0 + c1} n={len(s)}")
````

## 5. Lời giải chính — một lần duyệt, đếm theo chẵn lẻ, lấy bù

### 5.1. Đếm `cost0` (so với dạng `"0101…"`)

````python
s = "0100"
cost0 = 0
for i, ch in enumerate(s):
    if ch != str(i % 2):
        cost0 += 1
print("cost0 =", cost0)
````

### 5.2. Lấy bù và chọn min

````python
n = len(s)
print("cost1 = n - cost0 =", n - cost0)
print("đáp án =", min(cost0, n - cost0))
````

### 5.3. Ghép lại thành hàm

````python
def min_operations(s: str) -> int:
    cost0 = 0                              # số thao tác để thành "0101..."
    for i, ch in enumerate(s):
        if ch != str(i % 2):
            cost0 += 1
    return min(cost0, len(s) - cost0)      # dạng "1010..." tốn n - cost0

print(min_operations("0100"), min_operations("10"), min_operations("1111"))
````

## 6. TRACE từng bước

Trace cho `s = "110010"` (n = 6):

| i | s[i] | mong muốn dạng 01 | khác? | cost0 |
|---:|:---:|:---:|---|---:|
| 0 | 1 | 0 | Có | 1 |
| 1 | 1 | 1 | Không | 1 |
| 2 | 0 | 0 | Không | 1 |
| 3 | 0 | 1 | Có | 2 |
| 4 | 1 | 0 | Có | 3 |
| 5 | 0 | 1 | Có | 4 |

`cost0 = 4`, `cost1 = 6 - 4 = 2` → đáp án `2` (đổi `s` thành `"101010"`).

````python
def trace_min_operations(s: str) -> int:
    cost0 = 0
    print(f"s = {s!r}, n = {len(s)}")
    print(" i | s[i] | muốn(01) | muốn(10) | khác dạng 01? | cost0")
    for i, ch in enumerate(s):
        want0 = str(i % 2)
        want1 = str(1 - i % 2)
        diff = ch != want0
        if diff:
            cost0 += 1
        print(f"{i:>2} |  {ch}   |    {want0}     |    {want1}     | {str(diff):<13} | {cost0}")
    cost1 = len(s) - cost0
    print(f"cost0 = {cost0}, cost1 = n - cost0 = {cost1} -> min = {min(cost0, cost1)}")
    return min(cost0, cost1)

trace_min_operations("110010")
````

## 7. Code nộp LeetCode

````python
class Solution:
    def minOperations(self, s: str) -> int:
        cost0 = 0
        for i, ch in enumerate(s):
            if ch != str(i % 2):
                cost0 += 1
        return min(cost0, len(s) - cost0)

sol = Solution()
print(sol.minOperations("0100"), sol.minOperations("10"), sol.minOperations("1111"))
````

## 8. Các lời giải khác

### 8.1. Dựng hai chuỗi đích — bản nộp

Dễ hiểu nhất, bộ nhớ `O(n)`.

````python
class SolutionTargets:
    def minOperations(self, s: str) -> int:
        n = len(s)
        t0 = ("01" * (n // 2 + 1))[:n]
        t1 = ("10" * (n // 2 + 1))[:n]
        return min(sum(a != b for a, b in zip(s, t0)),
                   sum(a != b for a, b in zip(s, t1)))

print(SolutionTargets().minOperations("0100"))
````

### 8.2. Một dòng với số nguyên: `int(ch) ^ (i & 1)`

- `int(ch)` là `0` hoặc `1`.
- `i & 1` là bit cuối của `i` — giống `i % 2` nhưng dùng phép AND bit.
- `a ^ b` (XOR) bằng `1` khi `a != b`. Vậy `int(ch) ^ (i & 1)` = 1 đúng khi ký tự lệch khỏi dạng 01.

````python
for a in (0, 1):
    for b in (0, 1):
        print(f"{a} ^ {b} = {a ^ b}")
print("6 & 1 =", 6 & 1, "| 7 & 1 =", 7 & 1)
````

````python
class SolutionXor:
    def minOperations(self, s: str) -> int:
        cost0 = sum(int(ch) ^ (i & 1) for i, ch in enumerate(s))
        return min(cost0, len(s) - cost0)

print(SolutionXor().minOperations("110010"), SolutionXor().minOperations("1111"))
````

### 8.3. Đếm theo cặp (chẵn/lẻ) bằng slicing

`s[0::2]` lấy các ký tự ở chỉ số chẵn, `s[1::2]` lấy chỉ số lẻ. Dạng 01 muốn chỉ số chẵn là `'0'` và chỉ số lẻ là `'1'`:

`cost0 = (số '1' ở vị trí chẵn) + (số '0' ở vị trí lẻ)`. Hàm `str.count` làm việc đếm ở tốc độ C.

````python
s = "110010"
print("chẵn:", s[0::2], "| lẻ:", s[1::2])

class SolutionSlice:
    def minOperations(self, s: str) -> int:
        cost0 = s[0::2].count('1') + s[1::2].count('0')
        return min(cost0, len(s) - cost0)

print(SolutionSlice().minOperations("110010"), SolutionSlice().minOperations("0100"))
````

### 8.4. Bảng so sánh

Gọi `n = len(s)`.

| Cách | Ý tưởng | Time | Space | Ghi chú |
|---|---|---:|---:|---|
| Vét cạn `2^n` chuỗi | thử mọi chuỗi | `O(2^n · n)` | `O(n)` | Chỉ làm oracle |
| Dựng 2 chuỗi đích | đếm khác biệt 2 lần | `O(n)` | `O(n)` | Dễ hiểu |

## 9. Vì sao lời giải đúng?

1. **Chỉ có 2 chuỗi xen kẽ độ dài n.** Ký tự đầu quyết định toàn bộ: mỗi ký tự sau buộc phải khác ký tự trước nó, mà chỉ có 2 lựa chọn `'0'/'1'`.
2. **Chi phí để biến `s` thành một chuỗi đích `t` cố định = số vị trí `s[i] != t[i]`.** Mỗi thao tác sửa đúng một vị trí; vị trí khác nhau phải bị sửa ít nhất một lần; sửa vị trí giống nhau là thừa.
3. **Đáp án = min chi phí trên 2 chuỗi đích** — vì chuỗi kết quả phải là một trong hai.
4. **`cost1 = n - cost0`:** hai dạng đích khác nhau ở **mọi** vị trí, nên với mỗi `i`, `s[i]` khớp đúng một trong hai → mỗi vị trí đóng góp đúng 1 vào `cost0 + cost1`.

Kiểm chứng bằng oracle vét cạn trên **mọi** chuỗi độ dài 1..10:

````python
checked = 0
for n in range(1, 11):
    for bits in product("01", repeat=n):
        t = "".join(bits)
        assert min_operations(t) == min_operations_exhaustive(t), t
        checked += 1
print("Khớp oracle vét cạn trên", checked, "chuỗi")
````

## 12. Lỗi người mới hay gặp

### Lỗi 1: Tham lam "thấy hai ký tự kề bằng nhau thì sửa ký tự sau"

Nghe hợp lý nhưng **không tối ưu**: nó luôn giữ ký tự đầu, tức là luôn chọn dạng đích bắt đầu bằng `s[0]`.

````python
def wrong_greedy(s):
    t = list(s)
    ops = 0
    for i in range(1, len(t)):
        if t[i] == t[i - 1]:
            t[i] = '1' if t[i] == '0' else '0'
            ops += 1
    return ops

for s in ["0100", "0010", "1101"]:
    print(f"{s}: tham lam = {wrong_greedy(s)} | tối ưu = {min_operations(s)}")
````

Với `"0010"`: tham lam giữ `'0'` đầu → đích `"0101"`, tốn 3. Nhưng đích `"1010"` chỉ tốn 1 (đổi luôn ký tự đầu). **Sửa:** so sánh cả hai dạng đích.

### Lỗi 2: Chỉ tính một dạng đích

Quên dạng `"1010…"` → `"10"` ra `2` thay vì `0`.

````python
def wrong_one_target(s):
    return sum(ch != str(i % 2) for i, ch in enumerate(s))

print("sai:", wrong_one_target("10"), "| đúng:", min_operations("10"))
````

### Lỗi 3: So sánh ký tự với số

`ch != i % 2` luôn `True` vì `'0' != 0` (chuỗi khác số nguyên trong Python) → `cost0 = n`, `n - cost0 = 0`, hàm **luôn trả 0**. **Sửa:** `ch != str(i % 2)` hoặc `int(ch) != i % 2`.

````python
print("'0' == 0 ?", '0' == 0)
def wrong_type(s):
    c0 = sum(ch != i % 2 for i, ch in enumerate(s))
    return min(c0, len(s) - c0)
print("sai:", wrong_type("1111"), "| đúng:", min_operations("1111"))
````

### Lỗi 4: Đếm số cặp kề nhau bằng nhau

`"1111"` có 3 cặp bằng nhau nhưng chỉ cần 2 thao tác (một thao tác có thể sửa 2 cặp cùng lúc). Số cặp bằng nhau **không** phải đáp án.

````python
def wrong_count_pairs(s):
    return sum(s[i] == s[i + 1] for i in range(len(s) - 1))

print("sai:", wrong_count_pairs("1111"), "| đúng:", min_operations("1111"))
````

## 14. Biến thể

### 14.1. Trả về luôn chuỗi xen kẽ gần nhất (ưu tiên dạng 01 nếu hòa)

````python
def nearest_alternating(s):
    n = len(s)
    c0 = sum(ch != str(i % 2) for i, ch in enumerate(s))
    start = "01" if c0 <= n - c0 else "10"
    return "".join(start[i % 2] for i in range(n))

print(nearest_alternating("0100"), nearest_alternating("110010"), nearest_alternating("1111"))
````

### 14.2. Mẫu tuần hoàn chu kỳ 3 trên bảng chữ `"abc"`

Chuỗi đích là một phép xoay của `"abcabc…"` — có 3 ứng viên. Khung "thử mọi đích, lấy min" vẫn dùng được.

````python
def min_changes_period3(s):
    n = len(s)
    best = n
    for shift in range(3):
        pat = "abc"[shift:] + "abc"[:shift]
        cost = sum(ch != pat[i % 3] for i, ch in enumerate(s))
        best = min(best, cost)
    return best

print(min_changes_period3("abcabc"), min_changes_period3("aaaaaa"), min_changes_period3("bcabca"))
````

### 14.3. Xen kẽ theo **khối** độ dài 2: `"0011…"` hoặc `"1100…"`

Ký tự mong muốn tại `i` là `str((i // 2) % 2)` — chỉ đổi công thức vị trí.

````python
def min_changes_block2(s):
    c0 = sum(ch != str((i // 2) % 2) for i, ch in enumerate(s))
    return min(c0, len(s) - c0)

print(min_changes_block2("0011"), min_changes_block2("0101"), min_changes_block2("1100"))
````

## 17. Đáp án tham khảo

**Bài 1:** `cost0 = 5`, `cost1 = 2` → đáp án **2**.

**Bài 2:** `'0'`, `"01010"`, `2`.

````python
trace_min_operations("1011001")
print("10"[3 % 2], ("01" * 3)[:5], sum(a != b for a, b in zip("0110", "0101")))
````

````python
def answer_min_operations(s):
    cost0 = 0
    for i, ch in enumerate(s):
        if ch != str(i % 2):
            cost0 += 1
    return min(cost0, len(s) - cost0)


def answer_min_operations_targets(s):
    n = len(s)
    t0 = ("01" * (n // 2 + 1))[:n]
    t1 = ("10" * (n // 2 + 1))[:n]
    return min(sum(x != y for x, y in zip(s, t0)), sum(x != y for x, y in zip(s, t1)))


def answer_equal_adjacent_pairs(s):
    return sum(s[i] == s[i + 1] for i in range(len(s) - 1))


def answer_min_cost(s, a, b):
    best = None
    for start in ("01", "10"):
        cost = 0
        for i, ch in enumerate(s):
            want = start[i % 2]
            if ch != want:
                cost += a if ch == '0' else b
        best = cost if best is None else min(best, cost)
    return best


assert answer_min_operations("0100") == 1
assert answer_min_operations("1111") == 2
assert answer_min_operations_targets("110010") == 2
assert answer_equal_adjacent_pairs("1111") == 3
assert answer_equal_adjacent_pairs("0101") == 0
assert answer_min_cost("1111", 1, 5) == 10
assert answer_min_cost("0000", 3, 100) == 6
assert answer_min_cost("0110", 1, 2) == 3
# chi phí đều 1 đồng -> phải trùng bài gốc
rng = random.Random(58)
for _ in range(300):
    s = "".join(rng.choice("01") for _ in range(rng.randint(1, 20)))
    assert answer_min_cost(s, 1, 1) == answer_min_operations(s) == answer_min_operations_targets(s)
print("Đáp án tham khảo PASS toàn bộ test!")
````
