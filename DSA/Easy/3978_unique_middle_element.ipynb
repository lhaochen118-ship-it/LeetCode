2. Tìm phần tử giữa bằng chỉ số `len(nums) // 2` (chia nguyên) và nhớ vì sao độ dài lẻ cho một phần tử giữa duy nhất.
3. Biết các cách đếm lần xuất hiện: `list.count`, `collections.Counter`, duyệt thủ công, so sánh `index` / vị trí xuất hiện đầu và cuối.

## 2. Đề bài & ví dụ

Cho mảng số nguyên `nums` có độ dài **lẻ** `n`. Hãy trả về `True` nếu **phần tử ở giữa** của `nums` xuất hiện **đúng một lần** trong toàn mảng, ngược lại `False`.

Phần tử giữa là `nums[n // 2]` (vì `n` lẻ nên có đúng một phần tử có cùng số phần tử bên trái và bên phải).

| Ví dụ | `nums` | Output | Giải thích |
|---|---|---|---|
| 1 | `[1,2,3]` | `True` | Phần tử giữa là `2`, xuất hiện đúng 1 lần |
| 2 | `[1,2,2]` | `False` | Phần tử giữa là `2`, xuất hiện 2 lần |

Constraints: `1 <= n == len(nums) <= 100`; `n` là số lẻ; `1 <= nums[i] <= 100`.

Lưu ý: chỉ quan tâm **đúng một phần tử giữa**, không cần các phần tử khác có trùng nhau hay không.

````python
for nums in ([1, 2, 3], [1, 2, 2], [5], [4, 4, 4]):
    n = len(nums)
    mid = nums[n // 2]
    print(f"nums={nums}: n={n}, chỉ số giữa={n // 2}, phần tử giữa={mid}, số lần xuất hiện={nums.count(mid)}")
````

## 3. Kiến thức nền

### 3.1. Chỉ số phần tử giữa và `//`

Với `n` lẻ, chỉ số giữa là `n // 2` (chia nguyên, bỏ phần thập phân): `n = 3 -> 1`, `n = 5 -> 2`, `n = 1 -> 0`. Toán tử `/` trả về **số thực** (`3 / 2 = 1.5`), **không** dùng làm chỉ số được.

````python
for n in (1, 3, 5, 7):
    print(n, n // 2, n / 2)
try:
    [10, 20, 30][3 / 2]
except TypeError as e:
    print("TypeError:", e)
````

### 3.2. `list.count(x)`

`nums.count(x)` trả về số lần `x` xuất hiện trong `nums`. Duyệt toàn bộ danh sách: `O(n)`.

````python
print([1, 2, 2].count(2), [1, 2, 3].count(5))
````

### 3.3. `collections.Counter`

`Counter(nums)` dựng từ điển `giá_trị -> số lần`. Tra `cnt[x]` là `O(1)`, khóa không có cho `0`. Phù hợp khi cần tra **nhiều** giá trị.

````python
from collections import Counter
cnt = Counter([1, 2, 2])
print(cnt, cnt[2], cnt[7])
````

### 3.4. `index` và `rindex` kiểu thủ công

`nums.index(x)` cho vị trí **đầu tiên** của `x`. Vị trí **cuối cùng** có thể tính bằng `len(nums) - 1 - nums[::-1].index(x)`. Phần tử xuất hiện đúng một lần **khi và chỉ khi** vị trí đầu bằng vị trí cuối.

````python
nums = [3, 1, 3, 9]
first = nums.index(3)
last = len(nums) - 1 - nums[::-1].index(3)
print(first, last, first == last)
````

### 3.5. Thoát sớm khi đếm

Khi chỉ cần biết "có đúng một lần hay không", ta có thể dừng ngay khi thấy bản sao thứ hai (đếm `> 1`), không cần đếm hết mảng. Đây là tối ưu nhỏ nhưng là thói quen tốt.

````python
def appears_once(nums, target):
    seen = 0
    for x in nums:
        if x == target:
            seen += 1
            if seen > 1:
                return False
    return seen == 1

print(appears_once([1, 2, 3], 2), appears_once([1, 2, 2], 2), appears_once([1, 2, 2], 9))
````

## 4. Ý tưởng nền tảng / cách nghĩ trực giác

Hai việc tách bạch:

1. **Xác định phần tử giữa** `mid = nums[len(nums) // 2]`.
2. **Đếm** `mid` xuất hiện bao nhiêu lần; `True` nếu đúng `1`.

Tính tay `[7, 3, 5, 3, 9]`: `n = 5`, chỉ số giữa `2`, `mid = 5`; `5` xuất hiện 1 lần -> `True`.
Tính tay `[7, 3, 3, 3, 9]`: `mid = 3`, xuất hiện 3 lần -> `False`.
Tính tay `[3, 1, 3, 1, 1]`: `mid = 3` (chỉ số 2), `3` còn xuất hiện ở chỉ số 0 -> 2 lần -> `False` (dù hai hàng xóm đều là `1`).

| `nums` | `mid` | số lần | kết quả |
|---|---|---|---|
| `[7,3,5,3,9]` | 5 | 1 | `True` |
| `[7,3,3,3,9]` | 3 | 3 | `False` |
| `[3,1,3,1,1]` | 3 | 2 | `False` |
| `[4]` | 4 | 1 | `True` |

````python
for nums in ([7, 3, 5, 3, 9], [7, 3, 3, 3, 9], [3, 1, 3, 1, 1], [4]):
    mid = nums[len(nums) // 2]
    print(nums, "mid =", mid, "count =", nums.count(mid), "->", nums.count(mid) == 1)
````

## 5. Cách 1 — Brute Force / baseline: đếm thủ công bằng vòng lặp

1. `mid = nums[len(nums) // 2]`.
2. Duyệt mọi phần tử, đếm số phần tử bằng `mid`.
3. Trả `count == 1`.

Đây là cách trực tiếp nhất, `O(n)`. Chưa có "nút thắt" thực sự (mọi cách đều phải đọc cả mảng), nhưng ta sẽ làm ngắn gọn và thoát sớm hơn.

````python
class SolutionLoop:
    def isMiddleElementUnique(self, nums):
        mid = nums[len(nums) // 2]
        count = 0
        for x in nums:
            if x == mid:
                count += 1
        return count == 1

print(SolutionLoop().isMiddleElementUnique([1, 2, 3]), SolutionLoop().isMiddleElementUnique([1, 2, 2]))
````

**Độ phức tạp:** Time `O(n)`, Space `O(1)`.

## 6. Ý tưởng thuật toán / bước đột phá

**Bước đột phá:** không cần dựng bảng đếm cho **mọi** giá trị. Chỉ **một** giá trị (phần tử giữa) được hỏi, nên một lần đếm trên giá trị đó là đủ.

- `nums.count(mid) == 1` làm đúng việc đó trong một dòng;
- Nếu muốn thoát sớm: đếm tới `2` là dừng;
- Nếu muốn một lượt không đếm: so **vị trí xuất hiện đầu** và **cuối** của `mid` (hai vị trí bằng nhau khi duy nhất; chắc chắn bằng vị trí giữa `n // 2`).

**Quy tắc "nếu... thì...":**

- nếu tồn tại chỉ số `i != n // 2` mà `nums[i] == mid` thì `False`;
- nếu không có chỉ số nào như vậy thì `True`.

**Invariant:** `mid` luôn xuất hiện ít nhất một lần (tại chỉ số giữa), nên "đúng một lần" tương đương "**không** có bản sao nào ở chỉ số khác".

## 7. Các cách tiếp theo

### Cách 2 — `list.count`

Một dòng: `nums.count(nums[len(nums) // 2]) == 1`.

````python
class SolutionCount:
    def isMiddleElementUnique(self, nums):
        return nums.count(nums[len(nums) // 2]) == 1

print(SolutionCount().isMiddleElementUnique([1, 2, 3]), SolutionCount().isMiddleElementUnique([1, 2, 2]))
````

**Độ phức tạp:** Time `O(n)`, Space `O(1)`.

### Cách 3 — `Counter`

Dựng bảng đếm rồi tra phần tử giữa. Thích hợp khi bài mở rộng ra nhiều truy vấn.

````python
from collections import Counter

class SolutionCounter:
    def isMiddleElementUnique(self, nums):
        cnt = Counter(nums)
        return cnt[nums[len(nums) // 2]] == 1

print(SolutionCounter().isMiddleElementUnique([1, 2, 3]), SolutionCounter().isMiddleElementUnique([1, 2, 2]))
````

**Độ phức tạp:** Time `O(n)`, Space `O(d)` với `d` là số giá trị khác nhau.

### Cách 4 — Tìm bản sao ở chỉ số khác (thoát sớm)

Duyệt mọi chỉ số `i != n // 2`; nếu `nums[i] == mid` thì trả `False` ngay. Hết mà không thấy thì `True`.

````python
class SolutionEarly:
    def isMiddleElementUnique(self, nums):
        mid_idx = len(nums) // 2
        mid = nums[mid_idx]
        for i, x in enumerate(nums):
            if i != mid_idx and x == mid:
                return False
        return True

print(SolutionEarly().isMiddleElementUnique([1, 2, 3]), SolutionEarly().isMiddleElementUnique([1, 2, 2]))
````

**Độ phức tạp:** Time `O(n)` (best `O(1)`: bản sao ở chỉ số 0), Space `O(1)`.

### Cách 5 — Hai nửa riêng biệt

Phần tử giữa chỉ trùng nếu xuất hiện ở **nửa trái** hoặc **nửa phải**: `mid in nums[:h] or mid in nums[h + 1:]` với `h = n // 2`. Trực quan hóa cấu trúc bài toán: mảng = nửa trái + giữa + nửa phải.

````python
class SolutionHalves:
    def isMiddleElementUnique(self, nums):
        h = len(nums) // 2
        mid = nums[h]
        return mid not in nums[:h] and mid not in nums[h + 1:]

print(SolutionHalves().isMiddleElementUnique([1, 2, 3]), SolutionHalves().isMiddleElementUnique([1, 2, 2]))
````

**Độ phức tạp:** Time `O(n)`, Space `O(n)` cho hai lát cắt (có thể tránh bằng chỉ số).

## 8. TRACE từng bước

Trace Cách 4 trên `nums = [3, 1, 3, 1, 1]` (`n = 5`, chỉ số giữa `2`, `mid = 3`):

| `i` | `nums[i]` | `i != 2`? | `== mid`? | hành động |
|---|---|---|---|---|
| 0 | 3 | có | có | **return False** |

Trace trên `[1, 2, 3]` (`mid = 2`, chỉ số giữa `1`): `i = 0` (1 != 2), `i = 1` (là chỉ số giữa, bỏ qua), `i = 2` (3 != 2) -> không có bản sao -> `True`.

````python
def trace_early(nums):
    mid_idx = len(nums) // 2
    mid = nums[mid_idx]
    print(f"nums={nums}, chỉ số giữa={mid_idx}, mid={mid}")
    for i, x in enumerate(nums):
        if i == mid_idx:
            print(f"  i={i}: chính phần tử giữa, bỏ qua")
            continue
        same = x == mid
        print(f"  i={i}: nums[i]={x}, bằng mid? {same}")
        if same:
            print("  -> tìm thấy bản sao, return False")
            return False
    print("  -> không có bản sao, return True")
    return True

assert trace_early([3, 1, 3, 1, 1]) is False
assert trace_early([1, 2, 3]) is True
assert trace_early([5]) is True
````

````python
def trace_counter(nums):
    cnt = {}
    for x in nums:
        cnt[x] = cnt.get(x, 0) + 1
    mid = nums[len(nums) // 2]
    print("bảng đếm:", cnt, "| mid =", mid, "| số lần =", cnt[mid])
    return cnt[mid] == 1

assert trace_counter([1, 2, 2]) is False and trace_counter([1, 2, 3]) is True
````

## 9. Code nộp LeetCode

Dùng Cách 2 (`list.count`): ngắn và đủ nhanh.

````python
from typing import List, Optional


class Solution:
    def isMiddleElementUnique(self, nums: List[int]) -> bool:
        return nums.count(nums[len(nums) // 2]) == 1


print(Solution().isMiddleElementUnique([1, 2, 3]), Solution().isMiddleElementUnique([1, 2, 2]))
````

## 10. Vì sao thuật toán đúng

1. **Phần tử giữa duy nhất.** Vì `n` lẻ, `n = 2h + 1`, chỉ số giữa `h = n // 2` có đúng `h` phần tử bên trái và `h` bên phải. Đó là phần tử giữa của đề.
2. **`mid` luôn xuất hiện ít nhất một lần** (tại chỉ số `h`), nên `count(mid) >= 1`.
3. **Đúng một lần** nghĩa là `count(mid) == 1`, hay tương đương không có chỉ số `i != h` nào có `nums[i] == mid`. `list.count` đếm mọi chỉ số, trong đó chỉ số `h` đóng góp `1`; vậy `count == 1` khi và chỉ khi không có chỉ số nào khác.
4. **Các cách khác** (Counter, tìm bản sao, hai nửa) đều kiểm tra cùng mệnh đề "không tồn tại `i != h` với `nums[i] = mid`".
5. **Ca `n = 1`:** chỉ có phần tử giữa, không có bản sao -> `True`.

## 11. Phân tích Time / Space Complexity

Gọi `n = len(nums) <= 100`.

- Cách 1, 2, 4: Time `O(n)`, Space `O(1)` (Cách 4 thoát sớm được).
- Cách 3 (`Counter`): Time `O(n)`, Space `O(d)`.
- Cách 5 (hai nửa): Time `O(n)`, Space `O(n)` do lát cắt.

Không thể dưới `O(n)` ở trường hợp xấu nhất (kết quả `True` buộc phải xem mọi phần tử để chắc không có bản sao).

## 12. Bộ test

### Cách kiểm thử trong notebook này

Chúng ta dùng ba lớp kiểm thử, từ dễ đến khó:

1. **Test cố định**: ví dụ của đề + các trường hợp biên tự nghĩ ra. Mỗi test là một `assert hàm(input) == kết_quả_mong_đợi`. Nếu sai, Python ném `AssertionError` và dừng ngay, kèm thông tin ta in ra để dễ lần lại.
2. **Test ngẫu nhiên có `random.seed`**: gọi `random.seed(số)` trước vòng lặp để **mỗi lần chạy lại sinh đúng cùng dữ liệu** (tái lập được). Input luôn được sinh **trong đúng constraints** của đề.
3. **Đối chiếu nhiều cách (oracle)**: một cách rất đơn giản, hiển nhiên đúng (brute force) đóng vai "trọng tài"; mọi cách nhanh hơn phải cho cùng kết quả với trọng tài trên mọi input ngẫu nhiên.

````python
import random

solutions = [SolutionLoop(), SolutionCount(), SolutionCounter(), SolutionEarly(), SolutionHalves(), Solution()]

fixed = [
    ([1, 2, 3], True), ([1, 2, 2], False), ([5], True), ([4, 4, 4], False),
    ([7, 3, 5, 3, 9], True), ([7, 3, 3, 3, 9], False), ([3, 1, 3, 1, 1], False),
    ([2, 1, 2], True), ([1, 1, 2], False), ([100] * 99, False),
    (list(range(1, 100, 1)), True), ([1, 2, 3, 4, 5, 4, 7], False), ([9, 8, 7, 6, 5], True),
]
for nums, expected in fixed:
    for sol in solutions:
        assert sol.isMiddleElementUnique(list(nums)) == expected, (type(sol).__name__, nums)
print("test cố định OK:", len(fixed), "ca")

random.seed(3978)
cnt = {True: 0, False: 0}
for _ in range(800):
    n = random.choice(range(1, 100, 2))
    hi = random.choice([2, 5, 30, 100])
    nums = [random.randint(1, hi) for _ in range(n)]
    expected = sum(1 for x in nums if x == nums[n // 2]) == 1      # trọng tài viết tay
    cnt[expected] += 1
    for sol in solutions:
        assert sol.isMiddleElementUnique(list(nums)) == expected, (type(sol).__name__, nums)
print("test ngẫu nhiên OK, phân bố:", cnt)
assert cnt[True] > 50 and cnt[False] > 50
````

### Test thuộc tính (property test)

(1) Đảo ngược mảng không đổi kết quả (phần tử giữa không đổi). (2) Hoán vị các phần tử **bên trái** (hay bên phải) không đổi kết quả. (3) Đổi tên giá trị theo song ánh (ví dụ `x -> 101 - x`) không đổi kết quả. (4) Nếu thêm một bản sao của phần tử giữa vào cuối, đáp án trở thành `False`... (nhưng độ dài thành chẵn nên chỉ dùng như kiểm tra logic `count`).

````python
random.seed(39780)
for _ in range(500):
    n = random.choice(range(1, 60, 2))
    nums = [random.randint(1, 10) for _ in range(n)]
    r = Solution().isMiddleElementUnique(nums)
    assert Solution().isMiddleElementUnique(nums[::-1]) == r
    h = n // 2
    left = nums[:h]; random.shuffle(left)
    right = nums[h + 1:]; random.shuffle(right)
    assert Solution().isMiddleElementUnique(left + [nums[h]] + right) == r
    assert Solution().isMiddleElementUnique([101 - x for x in nums]) == r
    assert (nums + [nums[h]]).count(nums[h]) == nums.count(nums[h]) + 1
print("property test OK")
````

## 13. Các lỗi người mới hay gặp

### Lỗi 1: dùng `n / 2` làm chỉ số (số thực)

`3 / 2 = 1.5` -> `TypeError: list indices must be integers`. Dùng `//`.

````python
def wrong_float_index(nums):
    return nums.count(nums[len(nums) / 2]) == 1               # SAI: / cho số thực

try:
    wrong_float_index([1, 2, 3])
except TypeError as e:
    print("TypeError:", e)
assert Solution().isMiddleElementUnique([1, 2, 3]) is True
````

### Lỗi 2: lấy sai phần tử giữa (`n // 2 - 1` hoặc `n // 2 + 1`)

Lệch một chỉ số nghĩa là hỏi về phần tử khác. Với `[1,2,2]` đúng là chỉ số `1` (giá trị `2`, hai lần -> `False`); lệch sang chỉ số `0` (giá trị `1`, một lần) cho `True`.

````python
def wrong_off_by_one(nums):
    return nums.count(nums[len(nums) // 2 - 1]) == 1          # SAI: lệch trái một vị trí

print(wrong_off_by_one([1, 2, 2]), "!=", Solution().isMiddleElementUnique([1, 2, 2]))
assert wrong_off_by_one([1, 2, 2]) is True and Solution().isMiddleElementUnique([1, 2, 2]) is False
````

### Lỗi 3: chỉ so sánh phần tử giữa với **hai hàng xóm**

Bản sao có thể nằm **xa** phần tử giữa. `[3,1,3,1,1]`: hai hàng xóm của `3` (chỉ số 2) đều là `1`, nhưng `3` còn xuất hiện ở chỉ số 0.

````python
def wrong_neighbors_only(nums):
    h = len(nums) // 2
    left = nums[h - 1] if h > 0 else None
    right = nums[h + 1] if h + 1 < len(nums) else None
    return nums[h] != left and nums[h] != right                 # SAI: bỏ sót bản sao ở xa

print(wrong_neighbors_only([3, 1, 3, 1, 1]), "!=", Solution().isMiddleElementUnique([3, 1, 3, 1, 1]))
assert wrong_neighbors_only([3, 1, 3, 1, 1]) is True and Solution().isMiddleElementUnique([3, 1, 3, 1, 1]) is False
````

### Lỗi 4: kiểm tra mọi phần tử đều khác nhau (`len(set(nums)) == len(nums)`)

Đề chỉ quan tâm **phần tử giữa**; các phần tử khác được phép trùng nhau.

````python
def wrong_all_unique(nums):
    return len(set(nums)) == len(nums)                          # SAI: yêu cầu cả mảng đôi một khác nhau

print(wrong_all_unique([1, 5, 1]), "!=", Solution().isMiddleElementUnique([1, 5, 1]))
assert wrong_all_unique([1, 5, 1]) is False and Solution().isMiddleElementUnique([1, 5, 1]) is True
````

### Lỗi 5: dùng `nums.index(mid) == nums.rindex(mid)` (không có `rindex` với list)

List của Python **không có** `rindex`. Muốn vị trí cuối phải tự tính. Ví dụ này cho thấy lỗi `AttributeError`.

````python
def wrong_rindex(nums):
    mid = nums[len(nums) // 2]
    return nums.index(mid) == nums.rindex(mid)                  # SAI: list không có rindex

try:
    wrong_rindex([1, 2, 3])
except AttributeError as e:
    print("AttributeError:", e)
mid = 2
nums = [1, 2, 3]
assert nums.index(mid) == len(nums) - 1 - nums[::-1].index(mid)
````

## 18. Gợi ý 3 mức cho bài tập

**Mức 1 (nhẹ):** Bài 1: tìm chỉ số giữa rồi nhìn xem giá trị đó còn xuất hiện ở đâu. Bài 2: `n = 5`, chỉ số giữa `2`.

**Mức 2 (vừa):** Bài 3: `nums.count(nums[len(nums) // 2])`. Bài 4: duyệt `enumerate`, bỏ qua `i == len(nums) // 2`. Bài 5: với `n` chẵn hai phần tử giữa ở `n // 2 - 1` và `n // 2`.

**Mức 3 (gần đáp án):** Bài 1: `[9,9,4,9,9] -> True` (giữa là `4`, duy nhất); `[4,6,4] -> True`. Bài 2: `mid = 2`, các chỉ số `0, 2, 4`. Bài 4: `[2,5,2,5,2] -> 0`. Bài 5: `[1,2,3,4] -> True`, `[1,2,2,4] -> False`.

## 19. Đáp án tham khảo cho bài tập

````python
# Bài 1, 2
assert Solution().isMiddleElementUnique([9, 9, 4, 9, 9]) is True and Solution().isMiddleElementUnique([4, 6, 4]) is True
nums2 = [2, 5, 2, 5, 2]
assert len(nums2) // 2 == 2 and nums2[2] == 2 and [i for i, x in enumerate(nums2) if x == 2] == [0, 2, 4]

# Bài 3
def middle_count(nums):
    return nums.count(nums[len(nums) // 2])

assert middle_count([2, 5, 2, 5, 2]) == 3 and middle_count([1, 2, 3]) == 1

# Bài 4
def first_duplicate_index_of_middle(nums):
    h = len(nums) // 2
    for i, x in enumerate(nums):
        if i != h and x == nums[h]:
            return i
    return -1

assert first_duplicate_index_of_middle([2, 5, 2, 5, 2]) == 0 and first_duplicate_index_of_middle([1, 2, 3]) == -1

# Bài 5
def are_both_middles_unique(nums):
    n = len(nums)
    a, b = nums[n // 2 - 1], nums[n // 2]
    return nums.count(a) == 1 and nums.count(b) == 1

assert are_both_middles_unique([1, 2, 3, 4]) is True and are_both_middles_unique([1, 2, 2, 4]) is False
print("đáp án bài tập OK")
````
