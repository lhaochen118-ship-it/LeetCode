## 1. Mục tiêu học

Sau notebook này bạn sẽ:

2. Nắm các công cụ Python: `sum` với biểu thức sinh có điều kiện (`sum(x for x in nums if x < 10)`), `len(str(x))` để đếm chữ số, so sánh `>` / `!=`, `dict`/`defaultdict` gom nhóm theo số chữ số, và biến tích lũy `diff`.

5. Tránh các lỗi kinh điển: chỉ xét một lựa chọn (`a > b`), dùng `>=` (tính cả hòa là thắng), phân loại sai ranh giới (`x <= 10`), và đếm **số phần tử** thay vì **tổng giá trị**.

## 2. Đề bài & ví dụ

Cho mảng các số nguyên **dương** `nums`.

Alice và Bob chơi một trò chơi. Alice có thể chọn **hoặc** tất cả các số có **một chữ số**, **hoặc** tất cả các số có **hai chữ số** trong `nums`; những số còn lại thuộc về Bob. Alice thắng nếu **tổng các số của cô lớn hơn hẳn** tổng các số của Bob.

Trả về `true` nếu Alice có thể thắng, ngược lại trả về `false`.

**Ví dụ 1**

- Input: `nums = [1,2,3,4,10]`
- Output: `false`
- Giải thích: Alice không thắng được dù chọn một chữ số (`1+2+3+4 = 10` so với Bob `10`: hòa) hay hai chữ số (`10` so với Bob `10`: hòa).

**Ví dụ 2**

- Input: `nums = [1,2,3,4,5,14]`
- Output: `true`
- Giải thích: Alice chọn các số một chữ số có tổng `15`, còn Bob có `14`; `15 > 14`.

**Ví dụ 3**

- Input: `nums = [5,5,5,25]`
- Output: `true`
- Giải thích: Alice chọn các số hai chữ số có tổng `25`; Bob có `5+5+5 = 15`.

**Ràng buộc:** `1 <= nums.length <= 100`; `1 <= nums[i] <= 99`.

````python
examples = [
    ([1, 2, 3, 4, 10], False),
    ([1, 2, 3, 4, 5, 14], True),
    ([5, 5, 5, 25], True),
]
for nums, expected in examples:
    print(nums, "-> kỳ vọng", expected)
````

## 3. Kiến thức nền

### 3.1. Biểu thức sinh có điều kiện trong `sum`

`sum(x for x in nums if điều_kiện)` cộng chỉ những phần tử thỏa điều kiện. Phần `for x in nums` duyệt từng phần tử, phần `if ...` lọc, và phần `x` ở đầu là giá trị được cộng. Nếu không phần tử nào thỏa, `sum` của dãy rỗng bằng `0`.

````python
nums = [1, 2, 3, 4, 5, 14]
print(sum(x for x in nums if x < 10))      # 15 (các số một chữ số)
print(sum(x for x in nums if x >= 10))     # 14 (các số hai chữ số)
print(sum(x for x in [5, 7] if x >= 10))   # 0  (rỗng)
````

### 3.2. Một chữ số hay hai chữ số: `x < 10` và `len(str(x))`

Vì `1 <= x <= 99`, số có **một** chữ số là `1..9` (tức `x < 10`, hay `x <= 9`), số có **hai** chữ số là `10..99` (tức `x >= 10`). Một cách khác, đọc được ý nghĩa hơn: `len(str(x))` đếm số ký tự của biểu diễn thập phân. Lưu ý `10` là số **hai** chữ số (ranh giới hay bị nhầm).

````python
for x in [1, 9, 10, 99]:
    print(x, "x < 10:", x < 10, "| len(str(x)):", len(str(x)))
````

### 3.3. "Lớn hơn hẳn" (strictly greater): `>` khác `>=`

`a > b` đúng khi `a` **lớn hơn** `b`; khi `a == b` nó **sai**. Đề nói "strictly greater" nên **hòa không phải thắng**. Phép `a != b` đúng khi hai số khác nhau (hoặc `a > b`, hoặc `a < b`).

````python
a, b = 10, 10
print(a > b, a >= b, a != b)     # False True False
a, b = 15, 14
print(a > b, a >= b, a != b)     # True True True
````

### 3.4. Tổng của hai phần bù nhau

Nếu chia mọi phần tử của `nums` thành **đúng hai nhóm** (một chữ số / hai chữ số) thì `tổng_nhóm_1 + tổng_nhóm_2 = sum(nums)`. Biết một nhóm và `sum(nums)` ta suy ra nhóm kia: `b = total − a`. Đây là ý tưởng "hai phần bù nhau" mà ta dùng ở Cách 4.

````python
nums = [1, 2, 3, 4, 5, 14]
a = sum(x for x in nums if x < 10)
b = sum(x for x in nums if x >= 10)
print(a, b, a + b, sum(nums), "b == total - a:", b == sum(nums) - a)
````

### 3.5. Biến tích lũy có dấu và `defaultdict` gom nhóm

- **Biến tích lũy có dấu:** `diff = 0`, rồi `diff += x` cho nhóm này và `diff -= x` cho nhóm kia: `diff = a − b`.
- **`defaultdict(int)`**: từ điển mà khóa chưa có tự động bằng `0`, nên `groups[len(str(x))] += x` cộng dồn theo từng nhóm số chữ số mà không cần kiểm tra khóa có chưa.

````python
from collections import defaultdict

nums = [5, 5, 5, 25]
diff = 0
for x in nums:
    diff += x if x < 10 else -x
print("diff = a - b =", diff)                 # 15 - 25 = -10

groups = defaultdict(int)
for x in nums:
    groups[len(str(x))] += x
print(dict(groups))                           # {1: 15, 2: 25}
````

## 4. Ý tưởng nền tảng / cách nghĩ trực giác (tính tay)

Ví dụ 2: `nums = [1,2,3,4,5,14]`. Chia thành hai nhóm:

| nhóm | các số | tổng |
|---|---|---|
| một chữ số (`a`) | 1, 2, 3, 4, 5 | 15 |
| hai chữ số (`b`) | 14 | 14 |

- Alice chọn **nhóm 1**: cô có `15`, Bob có `14` ⇒ `15 > 14` ⇒ thắng.
- Alice chọn **nhóm 2**: cô có `14`, Bob có `15` ⇒ `14 > 15` sai.

Ví dụ 1: `[1,2,3,4,10]`: `a = 10`, `b = 10`. Chọn nhóm 1: `10 > 10` sai. Chọn nhóm 2: `10 > 10` sai. ⇒ Alice **không thể** thắng.

Nhận xét: Alice chọn nhóm nào thì Bob giữ nhóm còn lại; so sánh luôn là "tổng nhóm chọn" với "tổng nhóm kia". Nếu hai tổng khác nhau thì **chọn nhóm có tổng lớn hơn** là thắng; nếu bằng nhau thì không cách nào thắng.

````python
nums = [1, 2, 3, 4, 5, 14]
a = sum(x for x in nums if x < 10)
b = sum(x for x in nums if x >= 10)
print(f"chọn nhóm 1: Alice {a} vs Bob {b} -> thắng? {a > b}")
print(f"chọn nhóm 2: Alice {b} vs Bob {a} -> thắng? {b > a}")
````

## 5. Cách 1 — Brute Force / baseline: thử cả hai lựa chọn của Alice

**Ý tưởng.** Bám sát đề: với mỗi lựa chọn, chia `nums` thành phần của Alice và phần của Bob rồi so sánh tổng.

- Lựa chọn "một chữ số": Alice có mọi `x < 10`, Bob có mọi `x >= 10`.
- Lựa chọn "hai chữ số": ngược lại.
- Nếu **một trong hai** lựa chọn cho tổng Alice **lớn hơn hẳn** tổng Bob thì trả `True`.

Không cần nhận xét thông minh gì; chỉ làm đúng theo đề, kể cả việc dựng danh sách của Bob.

````python
def can_alice_win_try_both(nums: list[int]) -> bool:
    for pick_single in (True, False):                 # hai lựa chọn của Alice
        alice, bob = [], []
        for x in nums:
            is_single = x < 10
            if is_single == pick_single:
                alice.append(x)                       # số thuộc nhóm Alice chọn
            else:
                bob.append(x)                         # còn lại thuộc Bob
        if sum(alice) > sum(bob):                     # strictly greater
            return True
    return False


for nums, expected in examples:
    print(nums, can_alice_win_try_both(nums), "(kỳ vọng", expected, ")")
````

**Độ phức tạp (Cách 1).** Hai lựa chọn, mỗi lựa chọn duyệt `n` phần tử ⇒ **Time `O(n)`**, **Space `O(n)`** vì dựng hai danh sách phụ.

**Nút thắt:** ta dựng cả hai danh sách chỉ để cộng chúng; thực ra **hai lựa chọn dùng lại cùng hai tổng** `a` và `b` (chỉ đổi vai). Không cần danh sách.

## 6. Ý tưởng thuật toán / bước đột phá

Gọi `a` = tổng các số một chữ số, `b` = tổng các số hai chữ số.

- Alice chọn nhóm 1 ⇒ cô có `a`, Bob có `b` ⇒ thắng ⇔ `a > b`.
- Alice chọn nhóm 2 ⇒ cô có `b`, Bob có `a` ⇒ thắng ⇔ `b > a`.

Cô thắng nếu **ít nhất một** điều kiện đúng: `a > b` **hoặc** `b > a`. Hai điều đó cùng nhau chính là **`a != b`** (vì với hai số thực thì hoặc bằng nhau, hoặc một số lớn hơn). Vậy:

> **Alice thắng ⇔ `a != b`**.

Chỉ cần hai tổng, không cần danh sách; thậm chí chỉ cần hiệu `a − b` (Cách 3).

````python
for nums, expected in examples:
    a = sum(x for x in nums if x < 10)
    b = sum(x for x in nums if x >= 10)
    print(f"a={a}, b={b}, a != b -> {a != b} (kỳ vọng {expected})")
````

## 7. Các cách tiếp theo

### Cách 2 — Hai tổng bằng `sum` có điều kiện rồi `a != b`

**Giải thích từng đoạn.**

1. `a = sum(x for x in nums if x < 10)`: tổng các số một chữ số.
2. `b = sum(x for x in nums if x > 9)`: tổng các số hai chữ số.
3. `return a != b`: Alice thắng khi hai tổng khác nhau.

````python
def can_alice_win_two_sums(nums: list[int]) -> bool:
    a = sum(x for x in nums if x < 10)        # tổng các số một chữ số
    b = sum(x for x in nums if x > 9)         # tổng các số hai chữ số
    return a != b


for nums, expected in examples:
    print(nums, can_alice_win_two_sums(nums), "(kỳ vọng", expected, ")")
````

````python
# Code nộp LeetCode cho Cách 2 (đổi tên lớp thành Solution khi nộp)
class SolutionTwoSums:
    def canAliceWin(self, nums: list[int]) -> bool:
        a = sum(x for x in nums if x < 10)
        b = sum(x for x in nums if x > 9)
        return a != b


print(SolutionTwoSums().canAliceWin([1, 2, 3, 4, 10]), SolutionTwoSums().canAliceWin([5, 5, 5, 25]))     # False True
````

**Độ phức tạp Cách 2:** **Time `O(n)`** (hai lần duyệt), **Space `O(1)`**.

### Cách 3 — Một vòng lặp, tích lũy hiệu `diff = a − b`

**Giải thích.** Thay hai tổng bằng một biến `diff`: số một chữ số **cộng** vào, số hai chữ số **trừ** đi. Cuối cùng `diff = a − b` và Alice thắng ⇔ `diff != 0`. Chỉ duyệt mảng **một lần**.

````python
def can_alice_win_diff(nums: list[int]) -> bool:
    diff = 0
    for x in nums:
        if x < 10:
            diff += x          # góp cho nhóm một chữ số
        else:
            diff -= x          # góp (âm) cho nhóm hai chữ số
    return diff != 0           # diff = a - b


for nums, expected in examples:
    print(nums, can_alice_win_diff(nums), "(kỳ vọng", expected, ")")
````

**Độ phức tạp Cách 3:** **Time `O(n)`** một lượt, **Space `O(1)`**.

### Cách 4 — Mẹo "hai phần bù nhau": `2a != tổng`

**Giải thích.** `a + b = total` nên `a = b ⇔ 2a = total`. Do đó chỉ cần **một** tổng `a` và `total = sum(nums)`: Alice thắng ⇔ `2 * a != total`. Lợi ích: hai phép `sum` gọn, không phải lọc nhóm hai chữ số. Nhược điểm: khó đọc hơn nếu chưa nắm ý tưởng bù nhau.

````python
def can_alice_win_total(nums: list[int]) -> bool:
    a = sum(x for x in nums if x < 10)      # chỉ cần tổng nhóm một chữ số
    return 2 * a != sum(nums)               # a = b  <=>  2a = a + b = tổng


for nums, expected in examples:
    print(nums, can_alice_win_total(nums), "(kỳ vọng", expected, ")")
````

````python
# Code nộp LeetCode cho Cách 4
class SolutionTotal:
    def canAliceWin(self, nums: list[int]) -> bool:
        a = sum(x for x in nums if x < 10)
        return 2 * a != sum(nums)


print(SolutionTotal().canAliceWin([1, 2, 3, 4, 5, 14]), SolutionTotal().canAliceWin([1, 2, 3, 4, 10]))   # True False
````

**Độ phức tạp Cách 4:** **Time `O(n)`**, **Space `O(1)`**.

### Cách 5 — Gom nhóm theo số chữ số bằng `defaultdict`

**Giải thích.** Phân loại bằng `len(str(x))` thay cho `x < 10`, gom tổng vào `groups[số_chữ_số]`. Cách này **tổng quát hóa** dễ khi `nums[i]` lớn hơn (ba chữ số, bốn chữ số, ...): Alice thắng nếu có nhóm nào mà `2 * tổng_nhóm > tổng_tất_cả`. Với đề này chỉ có hai nhóm nên kết quả tương đương `a != b`.

````python
def can_alice_win_groups(nums: list[int]) -> bool:
    groups = defaultdict(int)
    for x in nums:
        groups[len(str(x))] += x            # gom tổng theo số chữ số
    total = sum(groups.values())
    return any(2 * g > total for g in groups.values())     # có nhóm nào "hơn hẳn phần còn lại"?


for nums, expected in examples:
    print(nums, can_alice_win_groups(nums), "(kỳ vọng", expected, ")")
````

**Độ phức tạp Cách 5:** **Time `O(n)`**, **Space `O(d)`** với `d` là số nhóm số chữ số (ở đề này `d <= 2`, coi như `O(1)`).

**Vì sao `2 * g > total` đúng nghĩa "hơn hẳn":** nhóm `g` hơn phần còn lại `total − g` ⇔ `g > total − g` ⇔ `2g > total`.

## 8. TRACE từng bước

Ta trace Cách 3 (hiệu `diff`) trên `nums = [5, 5, 5, 25]`, Cách 2 trên `nums = [1, 2, 3, 4, 10]` (hòa), và một ví dụ tự dựng `nums = [9, 10, 1, 20, 3]` đủ "khó" vì trộn hai nhóm.

````python
def trace_diff(nums: list[int]) -> bool:
    print("nums =", nums)
    diff = 0
    for i, x in enumerate(nums):
        if x < 10:
            diff += x
            print(f"  i={i} x={x:>2} (một chữ số)  -> diff += {x}  => diff = {diff}")
        else:
            diff -= x
            print(f"  i={i} x={x:>2} (hai chữ số)  -> diff -= {x}  => diff = {diff}")
    print(f"  diff = a - b = {diff} -> Alice thắng? {diff != 0}")
    return diff != 0


assert trace_diff([5, 5, 5, 25]) is True
assert trace_diff([1, 2, 3, 4, 10]) is False
````

````python
def trace_two_sums(nums: list[int]) -> bool:
    print("nums =", nums)
    a = b = 0
    for x in nums:
        if x < 10:
            a += x
            print(f"  x={x:>2}: nhóm 1 -> a = {a}, b = {b}")
        else:
            b += x
            print(f"  x={x:>2}: nhóm 2 -> a = {a}, b = {b}")
    print(f"  chọn nhóm 1: {a} > {b}? {a > b}; chọn nhóm 2: {b} > {a}? {b > a}")
    return a != b


assert trace_two_sums([9, 10, 1, 20, 3]) is True      # a = 13, b = 30
assert trace_two_sums([4, 6, 10]) is False            # a = 10, b = 10
````

Bảng trace Cách 3 với `nums = [5, 5, 5, 25]`:

| `i` | `x` | nhóm | `diff` sau |
|---|---|---|---|
| 0 | 5 | một chữ số (`+`) | 5 |
| 1 | 5 | một chữ số (`+`) | 10 |
| 2 | 5 | một chữ số (`+`) | 15 |
| 3 | 25 | hai chữ số (`−`) | −10 |

`diff = −10 != 0` ⇒ Alice thắng (chọn nhóm 2 vì `b = 25 > a = 15`). Với `[1,2,3,4,10]`: `diff = 10 − 10 = 0` ⇒ không thắng được.

## 9. Code nộp LeetCode

Bản nộp dùng Cách 2 (hai tổng, `a != b`): ngắn, dễ đọc, đúng ý nghĩa đề.

````python
class Solution:
    def canAliceWin(self, nums: list[int]) -> bool:
        a = sum(x for x in nums if x < 10)       # tổng các số một chữ số
        b = sum(x for x in nums if x > 9)        # tổng các số hai chữ số
        return a != b                            # khác nhau thì chọn nhóm lớn hơn là thắng


sol = Solution()
for nums, expected in examples:
    print(nums, "->", sol.canAliceWin(nums))
````

## 10. Vì sao thuật toán đúng

- **Phân hoạch.** Vì `1 <= nums[i] <= 99`, mỗi số thuộc **đúng một** trong hai nhóm: một chữ số (`1..9`) hoặc hai chữ số (`10..99`). Nên `a + b = tổng`, và nếu Alice lấy nhóm này thì Bob lấy đúng nhóm kia.
- **Hai lựa chọn của Alice.** Chọn nhóm một chữ số: thắng ⇔ `a > b`. Chọn nhóm hai chữ số: thắng ⇔ `b > a`.
- **Hợp lại.** Có lựa chọn thắng ⇔ `a > b` hoặc `b > a` ⇔ `a != b`. (Trong tập số nguyên, nếu `a != b` thì chắc chắn một trong hai bất đẳng thức đúng.)
- **Trường hợp một nhóm rỗng.** Ví dụ không có số hai chữ số ⇒ `b = 0`, còn `a >= 1` (vì `nums` không rỗng và các số dương) ⇒ `a != b` ⇒ Alice thắng (chọn nhóm 1, Bob có `0`). Công thức vẫn đúng không cần xử lý riêng.
- **Hòa không thắng.** `a == b` nghĩa là không cách nào có "lớn hơn hẳn"; vì vậy phải dùng `>`/`!=`, không dùng `>=`.
- **Đối chiếu độc lập.** Cách 1 (thử cả hai lựa chọn đúng theo đề) được so với công thức trên hàng nghìn mảng ngẫu nhiên và vét cạn các mảng nhỏ ở mục 12.

## 11. Time / Space Complexity

| Cách | Time | Space |
|---|---|---|
| 1. Thử cả hai lựa chọn (dựng danh sách) | `O(n)` | `O(n)` |
| 2. Hai tổng `a`, `b` | `O(n)` | `O(1)` |
| 3. Một vòng, hiệu `diff` | `O(n)` | `O(1)` |
| 4. `2a != tổng` | `O(n)` | `O(1)` |
| 5. `defaultdict` theo số chữ số | `O(n)` | `O(d)`, `d <= 2` |

Với `n <= 100` mọi cách đều tức thì. Không thể nhanh hơn `O(n)` vì phải đọc mọi phần tử.

## 13. Các lỗi người mới hay gặp

### Lỗi 1: Chỉ xét lựa chọn "một chữ số"

````python
def sai_chi_nhom_1(nums):
    a = sum(x for x in nums if x < 10)
    b = sum(x for x in nums if x >= 10)
    return a > b                                 # SAI: bỏ qua việc Alice có thể chọn nhóm hai chữ số


print("sai:", sai_chi_nhom_1([5, 5, 5, 25]), "| đúng:", Solution().canAliceWin([5, 5, 5, 25]))
````

**Vì sao sai:** Ví dụ 3 có `a = 15`, `b = 25`; chọn nhóm hai chữ số thắng nhưng `a > b` sai. **Cách sửa:** `a != b` (hoặc `a > b or b > a`).

### Lỗi 2: Dùng `>=` (coi hòa là thắng)

````python
def sai_dau_bang2(nums):
    a = sum(x for x in nums if x < 10)
    b = sum(x for x in nums if x >= 10)
    return a >= b or b >= a                      # SAI: luôn True, vì một trong hai luôn đúng


print("sai:", sai_dau_bang2([1, 2, 3, 4, 10]), "| đúng:", Solution().canAliceWin([1, 2, 3, 4, 10]))
````

**Vì sao sai:** `a >= b or b >= a` luôn đúng với mọi `a, b`. Đề yêu cầu **lớn hơn hẳn**, nên hòa (`a == b`) là thua. **Cách sửa:** dùng `>`/`!=`.

### Lỗi 3: Phân loại sai ranh giới (`x <= 10` là "một chữ số")

````python
def sai_ranh_gioi(nums):
    a = sum(x for x in nums if x <= 10)          # SAI: 10 có HAI chữ số
    b = sum(x for x in nums if x > 10)
    return a != b


for nums in ([10], [1, 2, 3, 4, 10]):
    print(nums, "sai:", sai_ranh_gioi(nums), "| đúng:", Solution().canAliceWin(nums))
assert sai_ranh_gioi([1, 2, 3, 4, 10]) != Solution().canAliceWin([1, 2, 3, 4, 10])
````

**Vì sao sai:** `10` có hai chữ số (`'1'`, `'0'`). Đưa `10` vào nhóm một chữ số làm `a = 20`, `b = 0` ⇒ trả `True` cho `[1,2,3,4,10]`, trong khi đáp án là `False`. **Cách sửa:** một chữ số là `x < 10` (hoặc `x <= 9`), hai chữ số là `x >= 10`.

### Lỗi 4: Đếm số phần tử thay vì cộng giá trị

````python
def sai_dem_phan_tu(nums):
    a = len([x for x in nums if x < 10])          # SAI: đếm số lượng, không phải tổng
    b = len([x for x in nums if x >= 10])
    return a != b


for nums in ([1, 2, 3, 4, 10], [5, 5, 5, 25], [1, 99]):
    print(nums, "sai:", sai_dem_phan_tu(nums), "| đúng:", Solution().canAliceWin(nums))
assert sai_dem_phan_tu([1, 2, 3, 4, 10]) != Solution().canAliceWin([1, 2, 3, 4, 10])
assert sai_dem_phan_tu([1, 99]) != Solution().canAliceWin([1, 99])      # đếm 1 vs 1 (False), tổng 1 vs 99 (True)
````

**Vì sao sai:** đề so sánh **tổng giá trị**, không phải số lượng. Với `[1,2,3,4,10]`: `4` số một chữ số và `1` số hai chữ số (khác nhau ⇒ `True`), nhưng tổng đều là `10` (⇒ `False`). **Cách sửa:** dùng `sum(x ...)`, không `len(...)`.

## 14. Mẹo làm bài nhanh & nhận dạng pattern

- "**Chọn một trong hai nhóm bù nhau, thắng nếu nhóm mình hơn hẳn phần còn lại**" ⇒ chỉ cần so sánh hai tổng; thắng ⇔ **hai tổng khác nhau** (chọn nhóm lớn hơn).
- Khi các nhóm **chia hết toàn bộ** mảng, luôn có `tổng_nhóm_này + tổng_nhóm_kia = tổng`; dùng quan hệ này để rút gọn code (`2a != tổng`).
- Phân loại theo số chữ số: `x < 10`, hoặc `len(str(x))` nếu muốn tổng quát; nhớ ranh giới `9 / 10`.
- Đọc kỹ từ khóa **strictly** (lớn hơn hẳn) ⇒ hòa là không thắng.
- Trước khi code, kiểm tra các biên: mảng một phần tử, một nhóm rỗng, hai tổng bằng nhau.

## 15. Biến thể của bài

1. **Ba nhóm** (một, hai, ba chữ số) hoặc `d` nhóm theo số chữ số: Alice chọn **một** nhóm, thắng nếu `2 * tổng_nhóm > tổng_tất_cả` (Cách 5).
2. **Hiệu thắng lớn nhất:** Alice muốn tối đa hóa `tổng_của_cô − tổng_của_Bob`; đáp án `abs(a − b)`.
3. **Đếm số cách thắng:** có bao nhiêu trong hai lựa chọn đưa tới thắng (`0`, `1` hoặc `2` — chỉ `0` hoặc `1` ở bài này).
4. **Cho phép chọn nhiều nhóm** (một tập con các nhóm): thắng nếu tồn tại tập con có tổng `> tổng/2`.
5. Bài liên quan: **Stone Game (877)**, **Predict the Winner (486)**, **Sum of Digits in the Base K (1837)**, **Find Numbers with Even Number of Digits (1295)**.

````python
def best_margin(nums):
    a = sum(x for x in nums if x < 10)
    b = sum(x for x in nums if x >= 10)
    return abs(a - b)


def winning_choices(nums):
    a = sum(x for x in nums if x < 10)
    b = sum(x for x in nums if x >= 10)
    return int(a > b) + int(b > a)


def can_win_any_group(nums):
    groups = defaultdict(int)
    for x in nums:
        groups[len(str(x))] += x
    total = sum(groups.values())
    return any(2 * g > total for g in groups.values())


print(best_margin([1, 2, 3, 4, 5, 14]))          # 1
print(winning_choices([1, 2, 3, 4, 10]), winning_choices([5, 5, 5, 25]))   # 0 1
print(can_win_any_group([5, 500, 20, 20, 20]))   # nhóm 3 chữ số: 500 > 45 -> True
print(can_win_any_group([100, 100, 10, 10, 1]))  # nhóm 3 chữ số 200 vs 21 -> True
print(can_win_any_group([100, 50, 50, 1]))       # các nhóm: 100 / 100 / 1, tổng 201: 200 > 201 sai -> False
assert best_margin([1, 2, 3, 4, 5, 14]) == 1
assert winning_choices([1, 2, 3, 4, 10]) == 0 and winning_choices([5, 5, 5, 25]) == 1
assert can_win_any_group([5, 500, 20, 20, 20]) is True
assert can_win_any_group([100, 50, 50, 1]) is False
````
