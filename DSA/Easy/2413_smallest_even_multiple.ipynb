## 3. Kiến thức nền

### 3.1. Bội và chia hết

`m` là **bội** của `n` nếu `m = n * k` với `k` nguyên dương, tương đương `m % n == 0`. Bội của `n` là `n, 2n, 3n, ...`.

````python
n = 7
print([n * k for k in range(1, 6)])
print(14 % 7, 15 % 7, 21 % 7)
````

### 3.2. Tính chất chẵn/lẻ khi nhân

| a | b | a × b |
|---|---|---|
| chẵn | chẵn | chẵn |
| chẵn | lẻ | chẵn |
| lẻ | chẵn | chẵn |
| lẻ | lẻ | **lẻ** |

Vì vậy bội `n * k` là chẵn nếu `n` chẵn **hoặc** `k` chẵn. Với `n` lẻ, `k` phải chẵn; `k` chẵn nhỏ nhất là `2`.

````python
for n in (3, 4):
    for k in (1, 2, 3):
        m = n * k
        print(f"n={n} k={k} -> {m} {'chẵn' if m % 2 == 0 else 'lẻ'}")
````

### 3.3. Ước chung lớn nhất và thuật toán Euclid

`gcd(a, b)` là ước chung lớn nhất. Euclid: `gcd(a, b) = gcd(b, a % b)`, dừng khi `b == 0` (kết quả là `a`). Thư viện `math.gcd` làm sẵn việc này.

````python
import math

def my_gcd(a, b):
    while b:
        a, b = b, a % b
    return a

print(my_gcd(12, 8), math.gcd(12, 8))
print(my_gcd(5, 2), math.gcd(5, 2))
print(my_gcd(6, 2), math.gcd(6, 2))
````

### 3.4. Bội chung nhỏ nhất (LCM)

`lcm(a, b) = a * b // gcd(a, b)`. Python 3.9+ có `math.lcm(a, b)`. Số chẵn nhỏ nhất là bội của `n` chính là `lcm(2, n)`, vì "chẵn" nghĩa là "bội của 2".

````python
import math
print(2 * 5 // math.gcd(2, 5))      # lcm(2, 5) = 10
print(2 * 6 // math.gcd(2, 6))      # lcm(2, 6) = 6
print(math.lcm(2, 5), math.lcm(2, 6))
````

### 3.5. Toán tử bit: `n & 1` và `n << 1`

`n & 1` lấy bit thấp nhất: `0` nếu `n` chẵn, `1` nếu lẻ. `n << k` dịch trái `k` bit, tức nhân `2**k`. Vậy `n << (n & 1)` là `n` khi chẵn, `2n` khi lẻ.

````python
for n in (5, 6, 7, 8):
    print(n, bin(n), "n & 1 =", n & 1, "-> n << (n & 1) =", n << (n & 1))
````

## 4. Ý tưởng nền tảng / cách nghĩ trực giác

Liệt kê bội của `n` từ nhỏ đến lớn và dừng ở số chẵn đầu tiên.

- `n = 5`: `5` (lẻ), `10` (chẵn) ⇒ dừng ở `10`.
- `n = 6`: `6` (chẵn) ⇒ dừng ngay ở `6`.
- `n = 7`: `7` (lẻ), `14` (chẵn) ⇒ `14`.
- `n = 1`: `1` (lẻ), `2` (chẵn) ⇒ `2`.

Quan sát: nếu `n` **chẵn**, chính `n` là đáp án. Nếu `n` **lẻ**, bội thứ nhất `n` lẻ, bội thứ hai `2n` chẵn ⇒ đáp án là `2n`. Không bao giờ phải tới bội thứ ba!

````python
for n in (1, 2, 3, 4, 5, 6, 7, 10, 11, 150):
    m = n
    while m % 2 != 0:
        m += n
    print(f"n={n:3d} -> {m:4d}  (= {m // n} * n)")
````

### 4.1. Bảng quy luật cho `n = 1..12`

Nhìn bảng sẽ thấy ngay quy luật: cột "đáp án" chỉ bằng `n` (hàng chẵn) hoặc `2n` (hàng lẻ), và cột `k` (đáp án chia `n`) chỉ nhận giá trị `1` hoặc `2`.

````python
print(f"{'n':>3} {'chẵn/lẻ':>8} {'đáp án':>7} {'k = đáp án / n':>15}")
for n in range(1, 13):
    ans = n if n % 2 == 0 else 2 * n
    print(f"{n:>3} {'chẵn' if n % 2 == 0 else 'lẻ':>8} {ans:>7} {ans // n:>15}")
````

Một cách nhớ: "muốn chẵn mà đang lẻ thì chỉ cần nhân với một số chẵn nhỏ nhất, tức `2`; đã chẵn rồi thì nhân với `1` (giữ nguyên)". Hệ số nhân `k` chỉ là `1` hoặc `2`; đó là lý do đáp án không bao giờ lớn hơn `2n`, tức tối đa `300` với `n <= 150`.

## 5. Cách 1 — Brute Force / baseline: duyệt từng bội của `n`

**Ý tưởng:** bắt đầu từ `m = n`, nếu `m` chưa chẵn thì tăng thêm `n` (sang bội tiếp theo) cho đến khi chẵn.

````python
class SolutionBrute:
    def smallestEvenMultiple(self, n: int) -> int:
        m = n
        while m % 2 != 0:
            m += n
        return m
````

````python
sol = SolutionBrute()
print(sol.smallestEvenMultiple(5))   # 10
print(sol.smallestEvenMultiple(6))   # 6
print(sol.smallestEvenMultiple(1))   # 2
````

**Độ phức tạp:** số vòng lặp tối đa là 2 (vì bội thứ hai luôn chẵn), nên Time `O(1)` trong thực tế; về lý thuyết viết là `O(k)` với `k <= 2`. Space `O(1)`. **Nút thắt:** không có, nhưng vòng lặp che mất quy luật; ta có thể viết thẳng kết quả.

## 6. Ý tưởng thuật toán / bước đột phá

**Bước đột phá:** đáp án chỉ có thể là `n` hoặc `2n`.

- **Invariant / quy tắc:**
  - **nếu** `n` chẵn **thì** đáp án là `n` (là bội của `n` nhỏ nhất có thể và đã chẵn);
  - **nếu** `n` lẻ **thì** đáp án là `2n` (`n` lẻ nên bị loại; `2n` là bội của `n`, chẵn, và không có bội nào của `n` nằm giữa `n` và `2n`).
- Như vậy: `answer = n if n % 2 == 0 else 2 * n`.

Cách nhìn khác: đáp án = `lcm(2, n)`. Nếu `n` chẵn, `gcd(2, n) = 2` nên `lcm = n`. Nếu `n` lẻ, `gcd(2, n) = 1` nên `lcm = 2n`.

## 7. Các cách tiếp theo

### Cách 2 — Thử hai ứng viên `n` và `2n`

Dựa trên nhận xét "chỉ cần xét `n` và `2n`": chọn ứng viên chẵn nhỏ nhất.

````python
class SolutionCandidates:
    def smallestEvenMultiple(self, n: int) -> int:
        candidates = [n, 2 * n]
        return min(c for c in candidates if c % 2 == 0)
````

````python
print(SolutionCandidates().smallestEvenMultiple(5))   # 10
print(SolutionCandidates().smallestEvenMultiple(6))   # 6
print(SolutionCandidates().smallestEvenMultiple(1))   # 2
````

**Độ phức tạp:** Time `O(1)`, Space `O(1)`.

### Cách 3 — Rẽ nhánh chẵn/lẻ

Cách ngắn gọn nhất dùng điều kiện.

````python
class SolutionBranch:
    def smallestEvenMultiple(self, n: int) -> int:
        if n % 2 == 0:
            return n
        return 2 * n
````

````python
print(SolutionBranch().smallestEvenMultiple(5))   # 10
print(SolutionBranch().smallestEvenMultiple(6))   # 6
print(SolutionBranch().smallestEvenMultiple(1))   # 2
````

**Độ phức tạp:** Time `O(1)`, Space `O(1)`.

### Cách 4 — Công thức LCM

Dùng định nghĩa: đáp án là `lcm(2, n)`.

````python
import math

class SolutionLcm:
    def smallestEvenMultiple(self, n: int) -> int:
        return 2 * n // math.gcd(2, n)
````

````python
print(SolutionLcm().smallestEvenMultiple(5))   # 10
print(SolutionLcm().smallestEvenMultiple(6))   # 6
print(SolutionLcm().smallestEvenMultiple(1))   # 2
````

**Độ phức tạp:** `math.gcd` trên số nhỏ chạy `O(log n)` theo Euclid (thực tế là hằng số vì `gcd(2, n)` kết thúc sau 1–2 bước). Space `O(1)`.

### Cách 5 — Mẹo bit

`n & 1` là `1` khi `n` lẻ, `0` khi chẵn; `n << 1` nhân đôi. Kết hợp: `n << (n & 1)`.

````python
class SolutionBit:
    def smallestEvenMultiple(self, n: int) -> int:
        return n << (n & 1)
````

````python
print(SolutionBit().smallestEvenMultiple(5))   # 10
print(SolutionBit().smallestEvenMultiple(6))   # 6
print(SolutionBit().smallestEvenMultiple(1))   # 2
````

**Độ phức tạp:** Time `O(1)`, Space `O(1)`. Cách này ngắn nhưng khó đọc; trong bài thật nên ưu tiên cách rẽ nhánh cho dễ hiểu.

## 8. TRACE từng bước

Trace cách **brute** cho `n = 5` và `n = 6`, và trace thuật toán Euclid cho `gcd(2, n)` (cách LCM).

| n | m ban đầu | `m % 2` | thao tác | m sau |
|---|---|---|---|---|
| 5 | 5 | 1 (lẻ) | `m += 5` | 10 |
| 5 | 10 | 0 (chẵn) | dừng | 10 |
| 6 | 6 | 0 (chẵn) | dừng | 6 |

````python
def trace_brute(n):
    m = n
    step = 0
    while True:
        print(f"n={n}, bước {step}: m={m}, m%2={m % 2}")
        if m % 2 == 0:
            print("   m chẵn -> dừng, kết quả:", m)
            return m
        m += n
        step += 1
        print(f"   m lẻ -> chuyển sang bội kế tiếp, m={m}")

trace_brute(5)
trace_brute(6)
trace_brute(1)
````

````python
def trace_euclid(a, b):
    print(f"gcd({a}, {b})")
    while b:
        print(f"  a={a}, b={b} -> (a, b) = ({b}, {a % b})")
        a, b = b, a % b
    print("  kết quả gcd =", a)
    return a

g = trace_euclid(2, 5)
print("lcm(2, 5) =", 2 * 5 // g)
g = trace_euclid(2, 6)
print("lcm(2, 6) =", 2 * 6 // g)
````

## 9. Code nộp LeetCode

Dùng cách rẽ nhánh (rõ ràng nhất):

````python
class Solution:
    def smallestEvenMultiple(self, n: int) -> int:
        if n % 2 == 0:
            return n
        return 2 * n
````

````python
print(Solution().smallestEvenMultiple(5))
print(Solution().smallestEvenMultiple(6))
````

## 10. Vì sao thuật toán đúng

Cần chứng minh: đáp án là `n` khi `n` chẵn, `2n` khi `n` lẻ.

1. **`n` chẵn.** Mọi bội dương của `n` là `n, 2n, ...`, nhỏ nhất là `n`, và `n` chẵn nên thỏa cả hai điều kiện. Không có số dương nào nhỏ hơn `n` mà là bội của `n`, nên `n` là nhỏ nhất.
2. **`n` lẻ.** Bội nhỏ nhất `n` lẻ nên bị loại. Các bội của `n` nhỏ hơn `2n` chỉ có `n`; bội kế tiếp là `2n`, chẵn (vì có thừa số 2) và là bội của `n`. Vậy `2n` thỏa mãn và là nhỏ nhất.
3. **Liên hệ với LCM.** Điều kiện "là bội của 2 và của `n`" chính là "bội chung của `2` và `n`", nên đáp án là `lcm(2, n) = 2n / gcd(2, n)`. `gcd(2, n) ∈ {1, 2}`: bằng `2` khi `n` chẵn, `1` khi `n` lẻ. Cả hai công thức cho cùng kết quả.
4. **Mẹo bit.** `n & 1 = 1` ⇔ `n` lẻ, khi đó `n << 1 = 2n`; còn `n & 1 = 0` thì `n << 0 = n`.
5. **Vòng lặp brute dừng sau tối đa 2 lần**, bởi bội thứ hai `2n` luôn chẵn.

### 12.2. Đối chiếu toàn bộ miền và test ngẫu nhiên (có seed)

Miền hợp lệ chỉ có `150` giá trị, nên ta kiểm tra **toàn bộ**, đối chiếu với định nghĩa gốc: số chẵn nhỏ nhất `m` có `m % 2 == 0` và `m % n == 0`.

````python
def by_definition(n):
    m = 1
    while not (m % 2 == 0 and m % n == 0):
        m += 1
    return m

for n in range(1, 151):
    want = by_definition(n)
    for s in sols:
        assert s.smallestEvenMultiple(n) == want, (type(s).__name__, n)
print("đúng với cả 150 giá trị n trong [1, 150], 6 cách")
````

## 13. Các lỗi người mới hay gặp

### Lỗi 1: Luôn trả `2 * n`

````python
def sai_luon_nhan_doi(n: int) -> int:
    return 2 * n                      # SAI: n = 6 đã chẵn rồi

n = 6
print("đúng:", Solution().smallestEvenMultiple(n), "| sai:", sai_luon_nhan_doi(n))
assert Solution().smallestEvenMultiple(n) != sai_luon_nhan_doi(n)
````

**Vì sao sai:** `12` vẫn là bội chẵn của `6` nhưng không phải **nhỏ nhất**; `6` đã thỏa. Chỉ nhân đôi khi `n` lẻ.

### Lỗi 2: Đảo điều kiện chẵn/lẻ

````python
def sai_dao_dieu_kien(n: int) -> int:
    if n % 2 == 1:                    # SAI: lẻ lại trả n
        return n
    return 2 * n

for n in (5, 6):
    print(f"n={n}: đúng={Solution().smallestEvenMultiple(n)}, sai={sai_dao_dieu_kien(n)}")
    assert Solution().smallestEvenMultiple(n) != sai_dao_dieu_kien(n)
````

**Vì sao sai:** nếu `n` lẻ thì `n` không phải số chẵn; phải nhân đôi. Đảo điều kiện cho ra `5` thay vì `10` (không chẵn) và `12` thay vì `6`.

### Lỗi 3: Trả `n + 1` cho số lẻ (nhầm "chẵn gần nhất" với "bội")

````python
def sai_cong_1(n: int) -> int:
    if n % 2 == 0:
        return n
    return n + 1                      # SAI: n+1 chẵn nhưng không là bội của n (trừ n = 1)

for n in (5, 7):
    print(f"n={n}: đúng={Solution().smallestEvenMultiple(n)}, sai={sai_cong_1(n)}, sai % n = {sai_cong_1(n) % n}")
    assert Solution().smallestEvenMultiple(n) != sai_cong_1(n)
````

**Vì sao sai:** đề yêu cầu **bội của `n`**. `6` không chia hết cho `5`. Số chẵn nhỏ nhất lớn hơn `n` không phải bội của `n`.

### Lỗi 4: Duyệt bội nhưng bắt đầu từ `1` thay vì từ `n`

````python
def sai_bat_dau_tu_1(n: int) -> int:
    m = 1
    while m % 2 != 0:                 # SAI: chỉ kiểm tra chẵn, quên kiểm tra là bội của n
        m += 1
    return m

for n in (5, 6, 9):
    print(f"n={n}: đúng={Solution().smallestEvenMultiple(n)}, sai={sai_bat_dau_tu_1(n)}")
assert Solution().smallestEvenMultiple(5) != sai_bat_dau_tu_1(5)
````

**Vì sao sai:** hàm này luôn trả `2` (số chẵn đầu tiên), bỏ qua điều kiện "là bội của `n`". Điều kiện dừng phải là `m % 2 == 0 and m % n == 0`, hoặc đi theo các bội `n, 2n, ...`.

## 14. Mẹo làm bài nhanh & nhận dạng pattern

- "Số nhỏ nhất vừa chia hết cho `a` vừa chia hết cho `b`" ⇒ **LCM**: `a * b // gcd(a, b)` hoặc `math.lcm`.
- Với `2` thì chỉ có hai khả năng: `n` chẵn ⇒ `n`; `n` lẻ ⇒ `2n`. Không cần vòng lặp.
- Nhớ tính chất chẵn/lẻ khi nhân: chỉ lẻ × lẻ mới ra lẻ.
- Mẹo bit `n & 1` thay `n % 2` khi bạn đã quen; nhưng ưu tiên dễ đọc.
- Brute-force kiểm tra nhanh: nếu miền nhỏ (ở đây 150 giá trị), hãy **kiểm tra toàn bộ miền** bằng định nghĩa.
- Luôn thử `n = 1` và `n = 2` vì dễ lỗi ở biên.
- Khi bí, hãy viết ra 3–4 bội đầu của `n` như ở mục 4; quy luật gần như hiện ra tự nhiên, và đó là cách học tốt nhất cho các bài toán số học nhỏ.
- Đừng viết vòng `while True` không có điều kiện dừng rõ ràng; với bài này vòng lặp luôn dừng sau tối đa 2 lượt, nhưng hãy tự chứng minh điều đó trước khi dùng.

## 15. Biến thể của bài

1. **Bội chung nhỏ nhất của `a` và `b` bất kỳ:** `lcm(a, b)`.
2. **Bội nhỏ nhất của `n` chia hết cho `k`** (thay `2` bằng `k`): `lcm(n, k)`.
3. **Bội của `n` nhỏ nhất lớn hơn hoặc bằng `L`:** `ceil(L / n) * n`.
4. **LCM của cả dãy** `1..n` (bài kinh điển "smallest multiple"): lặp `lcm` dần.
5. **Số chẵn nhỏ nhất không bé hơn `n` và chia hết cho `d`.**
6. **Đếm bội chẵn của `n` trong khoảng `[1, M]`:** đếm bội của `lcm(2, n)`.

### 15.1. Minh họa vài biến thể bằng code

Dưới đây là ba biến thể nhỏ để bạn thấy cùng một ý tưởng (LCM) dùng lại nhiều lần.

````python
import math
from functools import reduce

def lcm_two(a, b):
    return a // math.gcd(a, b) * b

def lcm_range(n):                       # LCM của 1..n
    return reduce(lcm_two, range(1, n + 1), 1)

def smallest_multiple_at_least(n, L):    # bội nhỏ nhất của n mà >= L
    return -(-L // n) * n

print("lcm(1..10) =", lcm_range(10))               # 2520
print("lcm(4, 6) =", lcm_two(4, 6))                # 12
print("bội của 7 nhỏ nhất >= 30:", smallest_multiple_at_least(7, 30))   # 35
assert lcm_range(10) == 2520
assert smallest_multiple_at_least(7, 30) == 35
assert smallest_multiple_at_least(5, 10) == 10
````
