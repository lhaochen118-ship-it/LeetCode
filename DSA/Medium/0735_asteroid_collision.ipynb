## 3. Kiến thức nền

### 3.1. Dấu là hướng, trị tuyệt đối là kích thước

`abs(x)` cho trị tuyệt đối. Tiểu hành tinh `x > 0` bay sang phải, `x < 0` bay sang trái; kích thước `abs(x)`. Một cặp **va chạm được** chỉ khi tiểu hành tinh **bên trái bay sang phải** (`> 0`) và tiểu hành tinh **bên phải bay sang trái** (`< 0`), tức cặp `(+, −)` kề nhau. Cặp `(−, +)` thì **bay xa nhau**, cặp cùng dấu thì **cùng chiều** nên không gặp.

````python
def can_collide(left, right):
    return left > 0 and right < 0           # chỉ (+, -) mới gặp nhau

for pair in [(5, -3), (-5, 3), (5, 3), (-5, -3)]:
    print(pair, "-> va chạm?", can_collide(*pair), "| kích thước", abs(pair[0]), abs(pair[1]))
````

### 3.2. Một vụ va chạm: ai sống sót

Với cặp `(a, b)` có `a > 0 > b`: so `a` với `abs(b) = -b`.

- `a > -b`: `b` nổ, **`a` sống**.
- `a < -b`: `a` nổ, **`b` sống**.
- `a == -b`: **cả hai** nổ.

Ta viết thành hàm `fight` trả list những cái còn sống:

````python
def fight(a, b):
    if a > -b:
        return [a]
    if a < -b:
        return [b]
    return []

print(fight(10, -5), fight(2, -5), fight(8, -8))
````

### 3.3. List dùng như ngăn xếp (stack)

**Ngăn xếp** là cấu trúc "vào sau, ra trước". Với list Python: `stk.append(x)` đẩy vào đỉnh; `stk.pop()` lấy và bỏ đỉnh; `stk[-1]` xem đỉnh **mà không bỏ**; `if stk:` kiểm tra không rỗng (list rỗng là "giả"). Gọi `stk[-1]` khi rỗng sẽ báo `IndexError`, nên luôn kiểm tra `stk and ...` trước.

````python
stk = []
stk.append(3); stk.append(5)
print(stk, stk[-1])        # [3, 5] 5
print(stk.pop(), stk)      # 5 [3]
print(bool([]), bool([1]))
try:
    [][-1]
except IndexError as e:
    print("IndexError:", e)
````

### 3.4. Vòng `while` với điều kiện ghép và thứ tự đánh giá

Điều kiện ghép `A and B and C` được đánh giá **từ trái sang phải** và **dừng ngay** khi gặp vế sai (*short-circuit*). Vì thế ta viết `while stk and stk[-1] > 0 and stk[-1] < -x:` — nếu `stk` rỗng thì vế sau **không bao giờ được chạy** (tránh `IndexError`). Hãy luôn đặt phép kiểm tra "tồn tại" **trước** phép đọc phần tử.

````python
stk, x = [], -5
print(stk and stk[-1] > 0)          # [] (giả): dừng sớm, không đọc stk[-1]
stk = [3, 4]
n = 0
while stk and stk[-1] > 0 and stk[-1] < -x:
    stk.pop()
    n += 1
print("đã pop", n, "lần, còn", stk)
````

### 3.5. Lặp tới khi không còn thay đổi (cố định điểm)

Một cách mô phỏng tự nhiên: **quét cả mảng, xử lý mọi va chạm tìm được, rồi lặp lại** cho tới khi **một lượt quét không đổi gì**. Dùng một biến cờ `changed` và vòng `while changed:`. Mỗi lượt có ít nhất một va chạm (làm mảng ngắn đi), nên vòng lặp luôn dừng sau tối đa `n` lượt.

````python
nums = [1, 2, 3]
changed = True
rounds = 0
while changed:
    changed = False
    if len(nums) > 1:
        nums.pop()           # "một thay đổi"
        changed = True
    rounds += 1
print(rounds, nums)
````

### 3.6. Danh sách liên kết cài bằng mảng chỉ số

Để xoá phần tử giữa list trong `O(1)` mà không dịch chuyển, ta lưu **hai mảng chỉ số**: `prv[i]` (chỉ số phần tử đứng **trước** `i`, hoặc `-1`) và `nxt[i]` (đứng **sau** `i`, hoặc `n`). Xoá `d` nghĩa là nối `prv[d]` với `nxt[d]`: `nxt[prv[d]] = nxt[d]` và `prv[nxt[d]] = prv[d]`. Ta cũng giữ mảng `alive[i]` để đánh dấu đã xoá.

````python
vals = [10, 20, 30, 40]
n = len(vals)
prv = list(range(-1, n - 1))        # [-1, 0, 1, 2]
nxt = list(range(1, n + 1))         # [1, 2, 3, 4]  (n = "không có")
alive = [True] * n

def remove(d):
    alive[d] = False
    p, q = prv[d], nxt[d]
    if p >= 0:
        nxt[p] = q
    if q < n:
        prv[q] = p

remove(1)
remove(2)
i, out = 0, []
while i < n:                          # duyệt qua các nút còn sống theo liên kết
    out.append(vals[i])
    i = nxt[i]
print(out, alive)
````

### 3.7. `random.Random(seed)` và kiểm thử ngẫu nhiên có thể lặp lại

`rng = random.Random(123)` tạo một bộ sinh số ngẫu nhiên **riêng** với hạt giống cố định, nên mỗi lần chạy lại notebook ta nhận **cùng một** dãy số (kết quả test lặp lại được). `rng.randint(a, b)` lấy số nguyên trong `[a, b]`; `rng.choice(list)` chọn một phần tử; `rng.choice([-1, 1])` chọn dấu. Để sinh tiểu hành tinh **khác 0**: lấy kích thước `1..K` rồi nhân với dấu ngẫu nhiên.

````python
def random_asteroids(rng, n, k):
    return [rng.randint(1, k) * rng.choice([-1, 1]) for _ in range(n)]

rng = random.Random(123)
print(random_asteroids(rng, 8, 6))
rng = random.Random(123)
print(random_asteroids(rng, 8, 6))      # cùng seed -> cùng kết quả
````

## 4. Ý tưởng nền tảng / cách nghĩ trực giác (tính tay)

Xét `asteroids = [3, 5, -6, 2, -1, 4]`. Vẽ hướng bay:

```
  3 →   5 →   ← 6   2 →   ← 1   4 →
```

Ai gặp ai? Chỉ cặp `(+, −)` kề nhau: `5` và `-6`; `2` và `-1`.

1. `5` gặp `-6`: `5 < 6` nên `5` nổ. Giờ `3` thành hàng xóm bên trái của `-6`: `3` gặp `-6`: `3 < 6` nên `3` nổ. `-6` không còn ai bên trái nên bay mãi sang trái: **sống**.
2. `2` gặp `-1`: `2 > 1` nên `-1` nổ. `2` và `4` cùng bay sang phải.

Kết quả: `[-6, 2, 4]`.

Nhận xét quan trọng:

- Một tiểu hành tinh **âm** có thể **phá liên tiếp nhiều** tiểu hành tinh dương bên trái nó (như `-6` phá `5` rồi `3`), cho đến khi gặp cái **lớn hơn** (nó nổ), **bằng** (cả hai nổ) hoặc hết dương (sống tiếp).
- Một tiểu hành tinh **dương** thì **chưa chắc** an toàn: số phận của nó phụ thuộc các âm **ở bên phải**. Nhưng ta có thể **xét lần lượt từ trái sang phải**: khi gặp một tiểu hành tinh mới, mọi thứ ở bên trái đã "ổn định" (không còn va chạm nội bộ), và tiểu hành tinh mới chỉ có thể va chạm với **những cái dương đứng ngay bên trái** (từ gần ra xa).

Cấu trúc "phần đã ổn định, chỉ quan tâm cái gần nhất ở bên phải của nó" chính là **ngăn xếp**.

````python
# Mô phỏng tay bằng code trên [3, 5, -6, 2, -1, 4]
a = [3, 5, -6, 2, -1, 4]
stack = []
for x in a:
    if x > 0:
        stack.append(x)
        print(f"x={x:>2}: dương, đẩy vào        -> {stack}")
    else:
        popped = []
        while stack and stack[-1] > 0 and stack[-1] < -x:
            popped.append(stack.pop())
        if stack and stack[-1] == -x:
            stack.pop()
            print(f"x={x:>2}: phá {popped}, rồi bằng nhau, cả hai nổ -> {stack}")
        elif not stack or stack[-1] < 0:
            stack.append(x)
            print(f"x={x:>2}: phá {popped}, sống sót, đẩy vào -> {stack}")
        else:
            print(f"x={x:>2}: phá {popped}, nhưng gặp cái lớn hơn -> x nổ -> {stack}")
print("kết quả:", stack)
````

## 5. Cách 1 — Brute Force / baseline: mô phỏng nhiều lượt quét

**Ý tưởng "dịch nguyên văn đề bài":**

1. Quét mảng từ trái sang phải. Gặp cặp kề `(+, −)` thì xử lý va chạm (giữ lại cái lớn hơn, hoặc không giữ gì nếu bằng nhau), rồi **nhảy qua cả hai** phần tử.
2. Các phần tử còn lại chép sang mảng mới.
3. Nếu lượt quét có ít nhất một va chạm → lặp lại với mảng mới (vì có thể xuất hiện cặp `(+, −)` mới kề nhau). Nếu không → dừng.

Dễ đúng vì mô phỏng đúng quá trình vật lý; ta dùng nó làm **oracle**.

````python
def collide_passes(asteroids):
    a = list(asteroids)
    changed = True
    while changed:
        changed = False
        out, i = [], 0
        while i < len(a):
            if i + 1 < len(a) and a[i] > 0 and a[i + 1] < 0:
                out.extend(fight(a[i], a[i + 1]))
                changed = True
                i += 2
            else:
                out.append(a[i])
                i += 1
        a = out
    return a

for a, expected in examples:
    got = collide_passes(a)
    print(a, "->", got)
    assert got == expected
````

**Độ phức tạp.** Mỗi lượt quét `O(n)`. Số lượt có thể lên tới `~n/2`: ví dụ `[1, 2, 3, ..., k, -100, -100, ...]`: mỗi lượt chỉ xử lý được **một** cặp vì sau mỗi vụ va chạm, tiểu hành tinh dương kế tiếp mới trở thành hàng xóm của `-100`.

- **Time:** `O(n²)` xấu nhất.
- **Space:** `O(n)` cho mảng mới của mỗi lượt.

**Nút thắt:** sau mỗi vụ va chạm ta **quét lại từ đầu** dù chỉ có vị trí **ngay bên cạnh** vừa va chạm mới có thể xuất hiện cặp `(+, −)` mới. Ta cần một cấu trúc nhớ "cái dương gần nhất bên trái".

## 6. Ý tưởng thuật toán / bước đột phá

**Bước đột phá:** duyệt **một lượt từ trái sang phải** và giữ một **ngăn xếp** các tiểu hành tinh **đang sống** ở bên trái. Khi xét tiểu hành tinh mới `x`:

- `x > 0` (bay phải): bên trái nó **không** có gì để va chạm (những cái bên trái hoặc bay trái — đã bay xa, hoặc bay phải — cùng chiều). Đẩy `x` vào ngăn xếp.
- `x < 0` (bay trái): nó chỉ có thể va chạm với các tiểu hành tinh **dương trên đỉnh ngăn xếp** (từ gần ra xa). Lặp:
  - Đỉnh **dương nhỏ hơn** `|x|` → đỉnh nổ (`pop`), tiếp tục với đỉnh mới.
  - Đỉnh **dương bằng** `|x|` → cả hai nổ (`pop` đỉnh, **không** đẩy `x`), xong.
  - Đỉnh **dương lớn hơn** `|x|` → `x` nổ (không đẩy `x`), xong.
  - Ngăn xếp **hết** hoặc đỉnh là **số âm** → `x` sống, **đẩy `x`** vào.

Mỗi phần tử bị đẩy vào tối đa một lần và bị `pop` tối đa một lần, nên tổng thời gian `O(n)`.

Thử bước đột phá tách riêng bằng một hàm xử lý **một tiểu hành tinh âm** trên ngăn xếp:

````python
def hit(stk, x):
    """x < 0 đánh vào ngăn xếp stk (đã ổn định). Trả về True nếu x sống sót (đã được đẩy vào)."""
    while stk and stk[-1] > 0 and stk[-1] < -x:
        stk.pop()                          # đỉnh dương nhỏ hơn: nổ
    if stk and stk[-1] == -x:
        stk.pop()                          # bằng nhau: cả hai nổ
        return False
    if not stk or stk[-1] < 0:
        stk.append(x)                      # hết dương: x sống
        return True
    return False                           # đỉnh dương lớn hơn: x nổ

s = [3, 5]
print(hit(s, -6), s)       # True  [-6]
s = [10, 2]
print(hit(s, -5), s)       # False [10]
s = [4, 6]
print(hit(s, -6), s)       # False [4]
````

## 7. Các cách tiếp theo

### Cách 2 — Ngăn xếp, một lượt (`O(n)`, cách chuẩn)

**Giải thích từng đoạn:**

1. `stk = []` giữ các tiểu hành tinh đang sống bên trái (đã ổn định).
2. Với mỗi `x` trong `asteroids`:
   - `x > 0`: `stk.append(x)`.
   - `x < 0`: `while stk and stk[-1] > 0 and stk[-1] < -x: stk.pop()`. Sau vòng `while`:
     - `if stk and stk[-1] == -x: stk.pop()` (cả hai nổ),
     - `elif not stk or stk[-1] < 0: stk.append(x)` (`x` sống),
     - còn lại: đỉnh dương lớn hơn → `x` nổ, không làm gì.
3. Trả `stk`.

````python
def collide_stack(asteroids):
    stk = []
    for x in asteroids:
        if x > 0:
            stk.append(x)
        else:
            while stk and stk[-1] > 0 and stk[-1] < -x:
                stk.pop()
            if stk and stk[-1] == -x:
                stk.pop()
            elif not stk or stk[-1] < 0:
                stk.append(x)
    return stk

for a, expected in examples:
    got = collide_stack(a)
    print(a, "->", got)
    assert got == expected
````

- **Space:** `O(n)` cho ngăn xếp (kết quả).

### Cách 3 — Danh sách liên kết + hàng việc các cặp `(+, −)` kề nhau (`O(n)`)

**Ý tưởng khác hẳn:** thay vì duyệt trái sang phải, ta mô phỏng **theo sự kiện**. Mọi va chạm đều xảy ra ở một cặp `(+, −)` **kề nhau trong danh sách đang sống**. Ta:

1. Đưa mọi cặp kề `(i, i+1)` có `a[i] > 0 > a[i+1]` vào **hàng việc** `work`.
2. Lấy một cặp `(i, j)` ra. Nếu một trong hai đã nổ, hoặc `i` và `j` không còn kề nhau (`nxt[i] != j`) → **bỏ qua** (cặp đã cũ).
3. Giải quyết va chạm: đánh dấu và **xoá** (nối `prv`, `nxt`) tiểu hành tinh nổ.
4. **Phát sinh cặp mới:** nếu `i` sống thì hàng xóm phải mới của `i` có thể là âm → đưa `(i, nxt[i])` vào; nếu `j` sống thì hàng xóm trái mới của `j` có thể là dương → `(prv[j], j)`; nếu cả hai nổ thì hai hàng xóm mới `(prv[i], nxt[j])` trở nên kề nhau, kiểm tra `(+, −)`.
5. Khi `work` rỗng, các tiểu hành tinh còn `alive` (theo thứ tự chỉ số) là kết quả.

Cách này cũng cho thấy một sự thật quan trọng: **kết quả không phụ thuộc thứ tự xử lý các va chạm** (ta lấy cặp từ cuối hàng việc chứ không từ trái sang phải).

````python
def collide_linked(asteroids):
    a = list(asteroids)
    n = len(a)
    prv = list(range(-1, n - 1))
    nxt = list(range(1, n + 1))
    alive = [True] * n
    work = [(i, i + 1) for i in range(n - 1) if a[i] > 0 > a[i + 1]]

    def remove(d):
        alive[d] = False
        p, q = prv[d], nxt[d]
        if p >= 0:
            nxt[p] = q
        if q < n:
            prv[q] = p

    while work:
        i, j = work.pop()
        if not (alive[i] and alive[j]) or nxt[i] != j:
            continue                                   # cặp đã cũ
        if a[i] > -a[j]:
            remove(j)
            k = nxt[i]                                 # hàng xóm phải mới của i
            if k < n and a[k] < 0:
                work.append((i, k))
        elif a[i] < -a[j]:
            remove(i)
            k = prv[j]                                 # hàng xóm trái mới của j
            if k >= 0 and a[k] > 0:
                work.append((k, j))
        else:
            p, q = prv[i], nxt[j]
            remove(i)
            remove(j)
            if p >= 0 and q < n and a[p] > 0 and a[q] < 0:
                work.append((p, q))
    return [a[i] for i in range(n) if alive[i]]

for a, expected in examples:
    got = collide_linked(a)
    print(a, "->", got)
    assert got == expected
````

**Độ phức tạp (cách 3):**

- **Time:** `O(n)` — mỗi va chạm hợp lệ làm ít nhất một tiểu hành tinh nổ (tối đa `n` va chạm), mỗi va chạm sinh tối đa một cặp mới; mỗi cặp cũ bị bỏ qua tốn `O(1)`.
- **Space:** `O(n)` cho `prv`, `nxt`, `alive`, `work`.

### Cách 4 — Duyệt từ **phải sang trái** với ngăn xếp đối xứng (`O(n)`)

Ý tưởng đối xứng: thay vì cho **âm** đánh các **dương** bên trái, ta duyệt **từ phải sang trái** và cho **dương** `x` đánh các **âm** đang nằm bên phải nó.

**Giải thích từng đoạn:**

1. Ngăn xếp `stk` giữ các tiểu hành tinh **đã ổn định ở bên phải**; **đỉnh** là cái **ngoài cùng bên trái** (vì ta thêm dần sang trái).
2. Với mỗi `x` (từ cuối về đầu):
   - `x < 0`: bay trái, không va chạm với cái nào đã ở bên phải → `append`.
   - `x > 0`: đánh các âm trên đỉnh: `while stk and stk[-1] < 0`: nếu `-stk[-1] < x` thì đỉnh nổ (`pop`), tiếp tục; nếu bằng nhau thì **cả hai nổ** (`pop` rồi `x` mất); nếu `-stk[-1] > x` thì `x` nổ. Nếu hết âm trên đỉnh (ngăn xếp rỗng hoặc đỉnh dương) mà `x` còn sống → `append(x)`.
3. Kết quả là `stk` **đảo ngược** (`stk[::-1]`) để về đúng thứ tự trái → phải.

````python
def collide_right_to_left(asteroids):
    stk = []
    for x in reversed(asteroids):
        if x < 0:
            stk.append(x)
            continue
        alive = True
        while stk and stk[-1] < 0:
            if -stk[-1] < x:
                stk.pop()                  # âm nhỏ hơn x: nổ
            elif -stk[-1] == x:
                stk.pop()                  # bằng nhau: cả hai nổ
                alive = False
                break
            else:
                alive = False              # âm lớn hơn x: x nổ
                break
        if alive:
            stk.append(x)
    return stk[::-1]

for a, expected in examples:
    got = collide_right_to_left(a)
    print(a, "->", got)
    assert got == expected
````

````python
class Solution:
    def asteroidCollision(self, asteroids: List[int]) -> List[int]:
        stk = []
        for x in reversed(asteroids):
            if x < 0:
                stk.append(x)
                continue
            alive = True
            while stk and stk[-1] < 0:
                if -stk[-1] < x:
                    stk.pop()
                elif -stk[-1] == x:
                    stk.pop()
                    alive = False
                    break
                else:
                    alive = False
                    break
            if alive:
                stk.append(x)
        return stk[::-1]

print([Solution().asteroidCollision(a) == e for a, e in examples])
````

**Độ phức tạp (cách 4):**

- **Time:** `O(n)` (mỗi phần tử vào/ra ngăn xếp tối đa một lần).
- **Space:** `O(n)`.

Cách 4 cho thấy bài toán **đối xứng gương**: đổi chiều duyệt và vai trò dương/âm cho cùng kết quả — một dấu hiệu nữa rằng đáp án không phụ thuộc thứ tự xử lý.

Ghi chú: còn một biến thể của cách 2 dùng **chính mảng đầu vào làm ngăn xếp** (con trỏ `top`) để chỉ tốn `O(1)` bộ nhớ phụ — xem bài tập 5.

### 8.1. Trace cách 2 (ngăn xếp) trên `[3, 5, -6, 2, -1, 4]`

| Bước | `x` | Ngăn xếp trước | Hành động | Ngăn xếp sau |
|---|---|---|---|---|
| 1 | `3` | `[]` | dương → đẩy | `[3]` |
| 2 | `5` | `[3]` | dương → đẩy | `[3, 5]` |
| 3 | `-6` | `[3, 5]` | pop `5` (`5 < 6`), pop `3` (`3 < 6`), ngăn xếp rỗng → đẩy `-6` | `[-6]` |
| 4 | `2` | `[-6]` | dương → đẩy | `[-6, 2]` |
| 5 | `-1` | `[-6, 2]` | đỉnh `2 > 1` (`2` không nhỏ hơn `1`) và `2 ≠ 1` → `-1` nổ | `[-6, 2]` |
| 6 | `4` | `[-6, 2]` | dương → đẩy | `[-6, 2, 4]` |

Kết quả `[-6, 2, 4]`.

````python
def trace_stack(asteroids):
    stk = []
    for step, x in enumerate(asteroids, 1):
        before = list(stk)
        note = ""
        if x > 0:
            stk.append(x)
            note = "dương -> đẩy"
        else:
            popped = []
            while stk and stk[-1] > 0 and stk[-1] < -x:
                popped.append(stk.pop())
            if stk and stk[-1] == -x:
                stk.pop()
                note = f"pop nhỏ hơn {popped}; đỉnh bằng {-x} -> cả hai nổ"
            elif not stk or stk[-1] < 0:
                stk.append(x)
                note = f"pop nhỏ hơn {popped}; hết dương -> {x} sống, đẩy"
            else:
                note = f"pop nhỏ hơn {popped}; đỉnh {stk[-1]} lớn hơn {-x} -> {x} nổ"
        print(f"bước {step}: x={x:>3} | trước {str(before):<14} | {note:<58} | sau {stk}")
    return stk

print("kết quả:", trace_stack([3, 5, -6, 2, -1, 4]))
print()
print("kết quả:", trace_stack([5, 5, -5, -5]))
````

### 8.2. Trace cách 3 (danh sách liên kết + hàng việc) trên `[3, 5, -6, 2, -1, 4]`

Hàng việc ban đầu: cặp `(i, j)` với `a[i] > 0 > a[j]` kề nhau: `(1, 2)` (tức `5` và `-6`) và `(3, 4)` (tức `2` và `-1`).

| Bước | Lấy ra | Kết quả | Cặp mới được thêm |
|---|---|---|---|
| 1 | `(3, 4)` (`2`, `-1`) | `2 > 1` → `-1` nổ | hàng xóm phải mới của `2` là `4` (dương) → không có |
| 2 | `(1, 2)` (`5`, `-6`) | `5 < 6` → `5` nổ | hàng xóm trái mới của `-6` là `3` (dương) → thêm `(0, 2)` |
| 3 | `(0, 2)` (`3`, `-6`) | `3 < 6` → `3` nổ | hàng xóm trái mới của `-6`: không còn → không có |

Còn sống: `-6`, `2`, `4`. Lưu ý thứ tự xử lý ở đây **khác** cách 2 (bắt đầu từ cặp bên phải), nhưng kết quả **giống hệt**.

````python
def trace_linked(asteroids):
    a = list(asteroids)
    n = len(a)
    prv = list(range(-1, n - 1))
    nxt = list(range(1, n + 1))
    alive = [True] * n
    work = [(i, i + 1) for i in range(n - 1) if a[i] > 0 > a[i + 1]]
    print("hàng việc ban đầu:", work)

    def remove(d):
        alive[d] = False
        p, q = prv[d], nxt[d]
        if p >= 0:
            nxt[p] = q
        if q < n:
            prv[q] = p

    step = 0
    while work:
        i, j = work.pop()
        step += 1
        if not (alive[i] and alive[j]) or nxt[i] != j:
            print(f"bước {step}: ({i},{j}) đã cũ -> bỏ qua")
            continue
        if a[i] > -a[j]:
            remove(j)
            msg = f"{a[i]} thắng {a[j]}"
            k = nxt[i]
            if k < n and a[k] < 0:
                work.append((i, k)); msg += f" | thêm cặp ({i},{k})"
        elif a[i] < -a[j]:
            remove(i)
            msg = f"{a[j]} thắng {a[i]}"
            k = prv[j]
            if k >= 0 and a[k] > 0:
                work.append((k, j)); msg += f" | thêm cặp ({k},{j})"
        else:
            p, q = prv[i], nxt[j]
            remove(i); remove(j)
            msg = f"{a[i]} và {a[j]} cùng nổ"
            if p >= 0 and q < n and a[p] > 0 and a[q] < 0:
                work.append((p, q)); msg += f" | thêm cặp ({p},{q})"
        print(f"bước {step}: lấy ({i},{j}) = ({a[i]},{a[j]}) -> {msg}")
    return [a[i] for i in range(n) if alive[i]]

print("kết quả:", trace_linked([3, 5, -6, 2, -1, 4]))
print("kết quả:", trace_linked([5, 5, -5, -5]))
````

### 8.3. Trace cách 4 (phải sang trái) trên `[10, 2, -5]`

| Bước | `x` | Ngăn xếp trước | Hành động | Ngăn xếp sau |
|---|---|---|---|---|
| 1 | `-5` | `[]` | âm → đẩy | `[-5]` |
| 2 | `2` | `[-5]` | đỉnh `-5`: `5 > 2` → `2` nổ | `[-5]` |
| 3 | `10` | `[-5]` | đỉnh `-5`: `5 < 10` → pop; ngăn xếp rỗng → đẩy `10` | `[10]` |

Đảo ngược `[10]` → `[10]`.

````python
def trace_right_to_left(asteroids):
    stk = []
    for step, x in enumerate(reversed(asteroids), 1):
        before = list(stk)
        if x < 0:
            stk.append(x)
            note = "âm -> đẩy"
        else:
            alive, popped = True, []
            while stk and stk[-1] < 0:
                if -stk[-1] < x:
                    popped.append(stk.pop())
                elif -stk[-1] == x:
                    popped.append(stk.pop()); alive = False
                    break
                else:
                    alive = False
                    break
            if alive:
                stk.append(x)
            note = f"đánh các âm {popped}; " + ("sống, đẩy" if alive else "nổ")
        print(f"bước {step}: x={x:>3} | trước {str(before):<10} | {note:<32} | sau {stk}")
    return stk[::-1]

print("kết quả:", trace_right_to_left([10, 2, -5]))
````

## 10. Vì sao đúng

**Bất biến (cách 2).** Sau khi xử lý `a[0..i]`, ngăn xếp `stk` chứa đúng trạng thái **cuối cùng** (mọi va chạm đã xong) của riêng các tiểu hành tinh `a[0..i]` nếu ta coi chúng là cả hệ thống. Hơn nữa `stk` luôn có dạng **"một dãy âm rồi một dãy dương"** (không có cặp `(+, −)` kề nhau), vì cặp đó sẽ va chạm.

**Chứng minh quy nạp.**

- Cơ sở: rỗng.
- Bước: thêm `x`.
  - `x > 0`: cái duy nhất có thể gặp `x` là cái **âm bên phải** `x` — hiện chưa có. Những cái bên trái `x`: âm thì bay xa, dương thì cùng chiều. Vậy trạng thái cuối của `a[0..i]` là `stk + [x]`; dạng "âm rồi dương" giữ nguyên.
  - `x < 0`: `x` chỉ có thể gặp **cái dương gần nhất bên trái** (đỉnh nếu đỉnh dương). Nếu đỉnh nhỏ hơn `|x|`, đỉnh nổ và `x` tiếp tục gặp cái dương kế tiếp; nếu bằng, cả hai nổ; nếu lớn hơn, `x` nổ. Nếu hết dương (đỉnh là âm hoặc rỗng), `x` bay mãi sang trái, không cản trở; thêm `x` vào đỉnh vẫn giữ dạng "âm rồi dương" (vì không còn dương nào dưới nó).
  - Vì mọi va chạm xảy ra giữa `x` và các dương liên tiếp ở đỉnh, và kết quả của mỗi va chạm được quyết định bởi kích thước, đây đúng là quá trình mô tả trong đề.

**Kết quả không phụ thuộc thứ tự va chạm (confluence).** Ta thấy ở cách 3 và 4 các thứ tự xử lý khác nhau cho cùng kết quả. Lý do: khi hai va chạm khác nhau cùng "chờ" (hai cặp `(+, −)` rời nhau) chúng **độc lập**; khi chúng chia sẻ một tiểu hành tinh (ví dụ `+ + −`: `−` có thể va chạm với dương gần nó trước), thứ tự duy nhất có thể xảy ra là gần-trước-xa-sau vì tiểu hành tinh xa bị che bởi cái gần.

Ta kiểm chứng bằng code: các cách cho cùng kết quả với **oracle ngẫu nhiên**: mỗi lần chọn **một cặp `(+, −)` kề nhau ngẫu nhiên** để va chạm, lặp đến khi hết cặp, với nhiều seed khác nhau.

````python
def collide_random_order(asteroids, rng):
    """Oracle: mỗi bước chọn NGẪU NHIÊN một cặp (+, -) kề nhau để va chạm, cho tới khi hết cặp."""
    a = list(asteroids)
    while True:
        idx = [k for k in range(len(a) - 1) if a[k] > 0 > a[k + 1]]
        if not idx:
            return a
        k = rng.choice(idx)
        a[k:k + 2] = fight(a[k], a[k + 1])

rng = random.Random(735)
count = 0
for _ in range(500):
    a = random_asteroids(rng, rng.randint(2, 14), rng.choice([3, 6, 12]))
    ref = collide_stack(a)
    for _ in range(5):                                  # 5 thứ tự ngẫu nhiên khác nhau
        assert collide_random_order(a, rng) == ref, a
    count += 1
print("đã kiểm chứng tính 'không phụ thuộc thứ tự' trên", count, "mảng ngẫu nhiên (mỗi mảng 5 thứ tự)")
````

| 3. Danh sách liên kết + hàng việc | `O(n)` | `O(n)` | mô phỏng theo sự kiện, thứ tự tuỳ ý |
| 4. Phải sang trái | `O(n)` | `O(n)` | đối xứng gương của cách 2 |

Với cách 2: vòng `for` chạy `n` lần; vòng `while` bên trong mỗi lần `pop` đúng một phần tử, mà mỗi phần tử chỉ được `append` một lần nên tổng số `pop` `≤ n` → tổng `O(n)`.

Với `n = 10^4` cả bốn cách trừ cách 1 đều chạy trong chớp mắt; cách 1 có thể mất hàng giây ở trường hợp xấu.

## 12. Bộ test

Gồm: ví dụ đề, edge case (một cặp, bằng nhau, chuỗi va chạm liên hoàn, tất cả cùng dấu, âm trước dương sau, số `±1000`), phản ví dụ cho các cách làm sai, **hàm kiểm tra tính hợp lệ** `is_valid_result` (kết quả là dãy con của đầu vào theo đúng thứ tự và **không còn cặp `(+, −)` kề nhau**), đối chiếu 4 cách với nhau và với **oracle ngẫu nhiên thứ tự** trên nhiều input có seed, và test lớn `n = 10^4`.

````python
METHODS = {
    "passes": collide_passes,
    "stack": collide_stack,
    "linked": collide_linked,
    "right_to_left": collide_right_to_left,
}

def is_subsequence(small, big):
    it = iter(big)
    return all(any(x == y for y in it) for x in small)

def is_valid_result(asteroids, res):
    """Điều kiện cần: dãy con của đầu vào, không còn cặp (+, -) kề nhau."""
    return is_subsequence(res, asteroids) and not any(res[i] > 0 > res[i + 1] for i in range(len(res) - 1))

def check_all(a, expected=None):
    ref = None
    for name, fn in METHODS.items():
        got = fn(list(a))
        assert is_valid_result(a, got), (name, a, got)
        if expected is not None:
            assert got == expected, (name, a, got, expected)
        if ref is None:
            ref = got
        assert got == ref, (name, a, got, ref)
    return ref

for a, expected in examples:
    check_all(a, expected)

# edge case
check_all([1, -1], [])
check_all([-1, 1], [-1, 1])                          # bay xa nhau
check_all([1, 1, -1], [1])                           # -1 cùng nổ với một số 1
check_all([5, 5, -5, -5], [])
check_all([1, 2, 3, -5], [-5])                       # một âm phá liên tiếp nhiều dương
check_all([1, 2, 3, -3], [1, 2])                     # phá 2 cái rồi bằng cái thứ ba
check_all([5, -3], [5])                              # âm nhỏ hơn: âm nổ
check_all([3, -5], [-5])
check_all([-5, -3, 2], [-5, -3, 2])
check_all([1, 2, 3, 4], [1, 2, 3, 4])                # toàn dương
check_all([-4, -3, -2, -1], [-4, -3, -2, -1])        # toàn âm
check_all([1, -2, 3, -4], [-2, -4])                  # xen kẽ
check_all([1000, -1000], [])
check_all([1000, 999, -1000], [])                    # -1000 phá 999 rồi cả hai nổ với 1000
check_all([-1000, 1000, -999, 999], [-1000, 1000, 999])
check_all([2, 1, -1, -2], [])                        # 1 và -1 nổ; rồi 2 và -2 nổ
check_all([4, -2, -3, 6, -6, 1, -7], [-7])
print("edge case: OK")
````

## 13. Các lỗi người mới hay gặp

### Lỗi 1: Dùng `if` thay `while` (chỉ xử lý một vụ va chạm)

**Code sai:**

```python
if stk and stk[-1] > 0 and stk[-1] < -x:       # chỉ pop MỘT lần
    stk.pop()
```

**Vì sao sai:** một tiểu hành tinh âm lớn có thể phá **nhiều** tiểu hành tinh dương liên tiếp. Phải dùng `while` để tiếp tục cho tới khi gặp cái lớn hơn hoặc bằng, hoặc hết dương.

````python
def wrong_if(asteroids):
    stk = []
    for x in asteroids:
        if x > 0:
            stk.append(x)
        else:
            if stk and stk[-1] > 0 and stk[-1] < -x:       # SAI: if, không phải while
                stk.pop()
            if stk and stk[-1] == -x:
                stk.pop()
            elif not stk or stk[-1] < 0:
                stk.append(x)
    return stk

print("sai :", wrong_if([1, 2, 3, -5]))       # [1, 2]: sai
print("đúng:", collide_stack([1, 2, 3, -5]))  # [-5]
````

### Lỗi 2: Quên trường hợp **bằng nhau** (cả hai cùng nổ)

**Code sai:**

```python
while stk and stk[-1] > 0 and stk[-1] <= -x:   # <=: pop cả khi bằng nhau, nhưng x vẫn sống
    stk.pop()
...
stk.append(x)                                    # x bị đẩy vào dù đã nổ cùng cái kia
```

**Vì sao sai:** khi hai tiểu hành tinh **bằng kích thước**, **cả hai** nổ. Nếu chỉ `pop` đỉnh mà vẫn đẩy `x` vào thì còn lại một tiểu hành tinh "ma".

````python
def wrong_equal(asteroids):
    stk = []
    for x in asteroids:
        if x > 0:
            stk.append(x)
        else:
            while stk and stk[-1] > 0 and stk[-1] <= -x:     # SAI: <= rồi vẫn đẩy x
                stk.pop()
            if not stk or stk[-1] < 0:
                stk.append(x)
    return stk

print("sai :", wrong_equal([8, -8]))          # [-8]
print("đúng:", collide_stack([8, -8]))        # []
````

### Lỗi 3: Coi **âm gặp âm** (hoặc âm đứng trước dương) là va chạm

**Code sai:**

```python
while stk and stk[-1] < -x:        # thiếu điều kiện stk[-1] > 0
    stk.pop()
```

**Vì sao sai:** đỉnh **âm** bay sang trái, cùng chiều với `x` âm nên **không** va chạm; thiếu `stk[-1] > 0` thì ta pop nhầm cả các âm nhỏ hơn.

````python
def wrong_neg_neg(asteroids):
    stk = []
    for x in asteroids:
        if x > 0:
            stk.append(x)
        else:
            while stk and abs(stk[-1]) < -x:                 # SAI: thiếu stk[-1] > 0
                stk.pop()
            if stk and stk[-1] == -x:
                stk.pop()
            elif not stk or stk[-1] < 0:
                stk.append(x)
    return stk

print("sai :", wrong_neg_neg([-2, -3]))        # [-3]: sai, -2 và -3 không hề va chạm
print("đúng:", collide_stack([-2, -3]))        # [-2, -3]
````

### Lỗi 4: Đẩy tiểu hành tinh **đã nổ** vào ngăn xếp

**Code sai:**

```python
while stk and stk[-1] > 0 and stk[-1] < -x:
    stk.pop()
stk.append(x)                      # luôn đẩy, kể cả khi đỉnh dương lớn hơn
```

**Vì sao sai:** nếu đỉnh dương **lớn hơn** `|x|` thì `x` **nổ**; đẩy nó vào là sai. Chỉ đẩy khi ngăn xếp hết dương (rỗng hoặc đỉnh âm).

````python
def wrong_push_always(asteroids):
    stk = []
    for x in asteroids:
        if x > 0:
            stk.append(x)
        else:
            while stk and stk[-1] > 0 and stk[-1] < -x:
                stk.pop()
            if stk and stk[-1] == -x:
                stk.pop()
            else:
                stk.append(x)                                # SAI: đẩy cả khi đỉnh dương lớn hơn
    return stk

print("sai :", wrong_push_always([5, -3]))       # [5, -3]: -3 đã nổ nhưng vẫn còn
print("đúng:", collide_stack([5, -3]))           # [5]
````

### Lỗi 5: Coi **mọi cặp trái dấu** là va chạm (không để ý thứ tự)

**Code sai:**

```python
while stk and stk[-1] * x < 0 and abs(stk[-1]) < abs(x):     # trái dấu là va chạm
    stk.pop()
```

**Vì sao sai:** chỉ cặp `(+, −)` (dương **bên trái**, âm **bên phải**) mới gặp nhau. Cặp `(−, +)` (âm bên trái, dương bên phải) **bay xa nhau**. Điều kiện `stk[-1] * x < 0` không phân biệt hai trường hợp nên `[-2, 3]` bị coi là va chạm.

````python
def wrong_any_opposite(asteroids):
    stk = []
    for x in asteroids:
        while stk and stk[-1] * x < 0 and abs(stk[-1]) < abs(x):     # SAI: trái dấu bất kể thứ tự
            stk.pop()
        if stk and stk[-1] * x < 0 and abs(stk[-1]) == abs(x):
            stk.pop()
        elif not stk or stk[-1] * x > 0 or abs(stk[-1]) < abs(x):
            stk.append(x)
    return stk

print("sai :", wrong_any_opposite([-2, 3]))        # [3]: sai, -2 và 3 bay xa nhau
print("đúng:", collide_stack([-2, 3]))             # [-2, 3]
````

## 14. Mẹo & nhận dạng pattern

- **"Mỗi phần tử mới tương tác với những phần tử gần nhất ở bên trái, rồi có thể xoá nhiều cái liên tiếp":** dấu hiệu của **ngăn xếp**. Cùng họ: *Remove All Adjacent Duplicates*, *Valid Parentheses*, *Daily Temperatures*, *Next Greater Element*, *Largest Rectangle in Histogram*, *Remove K Digits*.
- **Mô phỏng + chỉ cặp kề nhau mới tương tác:** thử mô phỏng ngây thơ (lặp quét) làm oracle, rồi tối ưu bằng ngăn xếp (nếu có "bên trái ổn định") hoặc danh sách liên kết + hàng việc (nếu thứ tự tuỳ ý).
- **Kiểm tra tính chất "không phụ thuộc thứ tự":** viết oracle ngẫu nhiên chọn cặp va chạm ngẫu nhiên; nếu mọi thứ tự cho cùng kết quả (confluence) thì mọi cách cài đặt hợp lệ đều đúng.
- **Dạng kết quả:** ngăn xếp cuối cùng có dạng "âm... rồi dương..."; một **dấu hiệu** để tự kiểm.
- **Điều kiện ghép trong `while`:** đặt kiểm tra tồn tại (`stk`) trước khi đọc `stk[-1]`.
- **Ba nhánh sau vòng `while`:** bằng nhau → cả hai nổ; hết dương → sống; còn lại → nổ. Đừng bỏ nhánh nào.
- **Đối xứng gương:** duyệt từ phải sang trái và đổi vai dương/âm cho cùng kết quả; hữu ích để tự kiểm.

## 15. Biến thể

1. **Ghi lại nhật ký va chạm:** trả danh sách các vụ va chạm (cái nào thắng, cái nào nổ) theo thứ tự xảy ra.
2. **Đếm số tiểu hành tinh nổ / số vụ va chạm.**
3. **Robot Collisions (2751):** mỗi robot có "máu" và hướng; va chạm trừ máu của bên thắng đi 1 — cần **thay đổi giá trị** của phần tử sống sót trong ngăn xếp.
4. **Tốc độ khác nhau:** khó hơn nhiều; cần sắp xếp theo thời điểm va chạm (hàng đợi ưu tiên).
5. **Hướng/kích thước trên vòng tròn:** hàng kề nhau quấn vòng; phải xử lý đoạn đầu và cuối.
6. **Remove All Adjacent Duplicates In String (1047):** cùng khuôn ngăn xếp, "va chạm" khi hai chữ cái liền nhau giống nhau.

Dưới đây là biến thể 1 và 2 (nhật ký va chạm, đếm số vụ nổ):

````python
def collision_log(asteroids):
    """Trả (nhật ký, kết quả). Mỗi sự kiện là (dương, âm, 'dương thắng' | 'âm thắng' | 'cùng nổ')."""
    log, stk = [], []
    for x in asteroids:
        if x > 0:
            stk.append(x)
            continue
        alive = True
        while alive and stk and stk[-1] > 0:
            top = stk[-1]
            if top < -x:
                log.append((top, x, "âm thắng"))
                stk.pop()
            elif top == -x:
                log.append((top, x, "cùng nổ"))
                stk.pop()
                alive = False
            else:
                log.append((top, x, "dương thắng"))
                alive = False
        if alive:
            stk.append(x)
    return log, stk

log, res = collision_log([3, 5, -6, 2, -1, 4])
for event in log:
    print(event)
print("kết quả:", res)
assert res == [-6, 2, 4]

# số tiểu hành tinh nổ = n - số sống; mỗi sự kiện làm nổ 1 (hoặc 2 nếu cùng nổ)
rng = random.Random(15)
for _ in range(300):
    a = random_asteroids(rng, rng.randint(2, 14), 6)
    log, res = collision_log(a)
    assert res == collide_stack(a)
    exploded = sum(2 if t == "cùng nổ" else 1 for _, _, t in log)
    assert exploded == len(a) - len(res), a
print("nhật ký khớp: số nổ = n - số sống")
````

## 18. Gợi ý (3 mức)

**Bài 1.**
- Mức 1: `fight(a, b)` so `a` với `-b`. `pop()` lấy bỏ phần tử cuối, `stk[-1]` xem phần tử cuối.
- Mức 2: `[-2, 2]`: âm bên trái, dương bên phải: bay xa nhau, không va chạm. `[6, -4, -3]`: `-4` nổ vì `6 > 4`; `-3` cũng nổ.
- Mức 3: kết quả theo thứ tự: `7 [6] []`, `3 [3]`, `[5, 10]`, `[-2, 2]`, `[6]`.

**Bài 2.**
- Mức 1: `4` đẩy vào; `-2` và `-3` gặp đỉnh `4` lớn hơn nên nổ.
- Mức 2: `6` đẩy vào `[4, 6]`; `-6` bằng `6` nên cả hai nổ → `[4]`; `1` đẩy vào `[4, 1]`.
- Mức 3: `-7` phá `1`, rồi phá `4`, rồi ngăn xếp rỗng nên đẩy `-7`. Các trạng thái: `[4], [4], [4], [4, 6], [4], [4, 1], [-7]`; đáp án `[-7]`.

**Bài 3.**
- Mức 1: `fight_pair`: so `a` với `-b`; ba nhánh.
- Mức 2: `survives_stack`: `while stk and stk[-1] > 0 and stk[-1] < -x: stk.pop()`.
- Mức 3: sau `while`: `if stk and stk[-1] == -x: stk.pop(); return False`; `if stk and stk[-1] > 0: return False`; `return True`.

**Bài 4.**
- Mức 1: duyệt `for x in asteroids`; `x > 0` thì `append`.
- Mức 2: `x < 0`: `while` pop dương nhỏ hơn; rồi xử lý bằng nhau / hết dương / đỉnh lớn hơn.
- Mức 3: dùng đúng ba nhánh: `if stk and stk[-1] == -x: stk.pop()`, `elif not stk or stk[-1] < 0: stk.append(x)`.

**Bài 5.**
- Mức 1: `top` là số phần tử của ngăn xếp; đỉnh là `arr[top - 1]`.
- Mức 2: "đẩy" = `arr[top] = x; top += 1`; "pop" = `top -= 1`.
- Mức 3: duyệt `for x in asteroids` (đọc từ đầu vào gốc, ghi vào `arr`), vì `top ≤` chỉ số đang đọc nên không ghi đè phần chưa đọc.

````python
# Bài 1
assert (abs(-7), fight(6, -4), fight(3, -3)) == (7, [6], [])
stk = [3, 5]; stk.pop()
assert (stk[-1], stk) == (3, [3])
assert collide_stack([5, 10, -5]) == [5, 10]
assert collide_stack([-2, 2]) == [-2, 2]
assert collide_stack([6, -4, -3]) == [6]
print("Bài 1: 7 [6] [] | 3 [3] | [5, 10] | [-2, 2] | [6]")

# Bài 2
my_states = [[4], [4], [4], [4, 6], [4], [4, 1], [-7]]
my_answer = [-7]
a = [4, -2, -3, 6, -6, 1, -7]
truth = [collide_stack(a[:i]) for i in range(1, len(a) + 1)]
assert my_states == truth and my_answer == collide_stack(a)
print("Bài 2: OK", my_states)

# Bài 3
def fight_pair(a, b):
    if a > -b:
        return [a]
    if a < -b:
        return [b]
    return []

def survives_stack(stk, x):
    while stk and stk[-1] > 0 and stk[-1] < -x:
        stk.pop()
    if stk and stk[-1] == -x:
        stk.pop()
        return False
    if stk and stk[-1] > 0:
        return False
    return True

assert fight_pair(10, -5) == [10] and fight_pair(2, -5) == [-5] and fight_pair(8, -8) == []
s = [3, 5]; assert survives_stack(s, -6) is True and s == []
s = [10, 2]; assert survives_stack(s, -5) is False and s == [10]
s = [4, 6]; assert survives_stack(s, -6) is False and s == [4]
s = [-1, 3]; assert survives_stack(s, -3) is False and s == [-1]
print("Bài 3: OK")
````

````python
# Bài 4
def survivors(asteroids):
    stk = []
    for x in asteroids:
        if x > 0:
            stk.append(x)
        else:
            while stk and stk[-1] > 0 and stk[-1] < -x:
                stk.pop()
            if stk and stk[-1] == -x:
                stk.pop()
            elif not stk or stk[-1] < 0:
                stk.append(x)
    return stk

for a, expected in examples:
    assert survivors(a) == expected
rng = random.Random(4)
for _ in range(300):
    a = random_asteroids(rng, rng.randint(2, 12), 6)
    assert survivors(a) == collide_passes(a), a
print("Bài 4: OK")

# Bài 5
def collide_in_place(asteroids):
    arr = list(asteroids)
    top = 0
    for x in asteroids:
        if x > 0:
            arr[top] = x
            top += 1
        else:
            while top > 0 and arr[top - 1] > 0 and arr[top - 1] < -x:
                top -= 1
            if top > 0 and arr[top - 1] == -x:
                top -= 1
            elif top == 0 or arr[top - 1] < 0:
                arr[top] = x
                top += 1
    return arr[:top]

for a, expected in examples:
    assert collide_in_place(a) == expected
rng = random.Random(5)
for _ in range(500):
    a = random_asteroids(rng, rng.randint(2, 12), 6)
    b = list(a)
    assert collide_in_place(a) == collide_stack(a) and a == b
print("Bài 5: OK")
````

## 20. Template nhớ nhanh + Checklist trước khi submit

**Template ngăn xếp cho va chạm `(+, −)`:**

```python
class Solution:
    def asteroidCollision(self, asteroids: List[int]) -> List[int]:
        stk = []
        for x in asteroids:
            if x > 0:
                stk.append(x)                               # bay phải: không va chạm với bên trái
            else:
                while stk and stk[-1] > 0 and stk[-1] < -x:
                    stk.pop()                               # dương nhỏ hơn: nổ
                if stk and stk[-1] == -x:
                    stk.pop()                               # bằng nhau: cả hai nổ
                elif not stk or stk[-1] < 0:
                    stk.append(x)                           # hết dương: x sống
                # đỉnh dương lớn hơn: x nổ (không làm gì)
        return stk
```

**Ghi nhớ một câu:** *"Dương thì đẩy; âm thì phá các dương nhỏ hơn trên đỉnh, gặp bằng thì cùng nổ, hết dương thì sống, gặp dương lớn hơn thì nổ."*
