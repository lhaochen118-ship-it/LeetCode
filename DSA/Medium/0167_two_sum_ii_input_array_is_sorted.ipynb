# LeetCode #167 — Two Sum II (Input Array Is Sorted): hai con trỏ ở hai đầu

**Mức độ:** Medium · **Chủ đề:** Array, Two Pointers, Binary Search

Skeleton trên LeetCode:

```python
class Solution:
    def twoSum(self, numbers: List[int], target: int) -> List[int]:
```

Mục tiêu của notebook:

1. So sánh bốn cách: thử mọi cặp `O(n²)`, tìm nhị phân `O(n log n)`, bảng băm `O(n)` thời gian + `O(n)` bộ nhớ, **hai con trỏ** `O(n)` + `O(1)` bộ nhớ.
2. Hiểu vì sao **mảng đã sắp xếp** cho phép bỏ hẳn một đầu sau mỗi phép so sánh.
3. Chứng minh hai con trỏ **không bỏ sót** cặp đáp án.
4. Chú ý chỉ số **bắt đầu từ 1** theo đề.

# 1. Bài toán là gì?

Cho mảng số nguyên `numbers` đánh số **từ 1**, đã **sắp xếp không giảm**. Tìm hai số có tổng bằng `target`: trả về `[index1, index2]` với `1 <= index1 < index2 <= numbers.length`.

Đề đảm bảo có **đúng một** đáp án; không được dùng một phần tử hai lần. Lời giải chỉ được dùng **bộ nhớ phụ hằng số**.

**Constraints (theo đề):**

- `2 <= numbers.length <= 3 * 10^4`
- `-1000 <= numbers[i] <= 1000`
- `numbers` sắp xếp không giảm.
- `-1000 <= target <= 1000`
- Có đúng một đáp án.

---

## Ví dụ

```text
numbers = [2, 7, 11, 15], target = 9   ->  [1, 2]   (2 + 7)
numbers = [2, 3, 4],      target = 6   ->  [1, 3]   (2 + 4)
numbers = [-1, 0],        target = -1  ->  [1, 2]
```

# 2. Kiến thức nền cần nắm trước

Để hiểu trọn lời giải bài này, bạn cần nắm chắc các kiến thức dưới đây. Mỗi phần trình bày: khái niệm là gì, hình minh họa, code mẫu **chạy được**, độ phức tạp và lỗi hay gặp. Phần nào đã vững, bạn có thể đọc lướt.

1. Hai con trỏ (two pointers)
2. Tìm kiếm nhị phân (binary search)

Cuối mục có phần **2.3 Áp dụng vào bài này** nối các kiến thức trên với lời giải.

## 2.1 Hai con trỏ (two pointers)

### Ý tưởng

Dùng **hai chỉ số** di chuyển trên mảng/chuỗi theo quy tắc, để thay vòng lặp lồng `O(n²)` bằng một lượt `O(n)`. Hai kiểu chính:

**1. Hai đầu tiến vào nhau** (thường trên mảng đã sắp xếp):

```text
[1, 2, 4, 7, 11, 15], tìm hai số có tổng 15
 L                 R     1 + 15 = 16 > 15 → R lùi (cần nhỏ hơn)
 L             R         1 + 11 = 12 < 15 → L tiến (cần lớn hơn)
    L          R         2 + 11 = 13 < 15 → L tiến
       L       R         4 + 11 = 15 ✓
```

Vì sao được phép bỏ? Khi `a[L] + a[R] > target`, mọi cặp `(L', R)` với `L' > L` còn lớn hơn → `R` không thể là đáp án, bỏ `R`.

**2. Cùng chiều — con trỏ đọc và con trỏ ghi** (sửa mảng tại chỗ):

```text
xóa mọi số 0, giữ thứ tự: [0, 1, 0, 3, 12]
read đi qua từng phần tử; gặp số khác 0 thì chép vào vị trí write, write += 1
→ [1, 3, 12, ...]
```

````python
def tong_hai_so_da_sap_xep(a, target):
    L, R = 0, len(a) - 1
    while L < R:
        s = a[L] + a[R]
        print(f"   L={L} R={R}: {a[L]} + {a[R]} = {s}")
        if s == target:
            return L, R
        if s < target:
            L += 1
        else:
            R -= 1
    return None

print("→", tong_hai_so_da_sap_xep([1, 2, 4, 7, 11, 15], 15))

def bo_so_khong(a):
    write = 0
    for read in range(len(a)):
        if a[read] != 0:
            a[write] = a[read]
            write += 1
    return a[:write]

print("bỏ số 0:", bo_so_khong([0, 1, 0, 3, 12]))

def la_palindrome(s):
    i, j = 0, len(s) - 1
    while i < j:
        if s[i] != s[j]:
            return False
        i += 1; j -= 1
    return True
print("'racecar' palindrome?", la_palindrome("racecar"), "| 'abca'?", la_palindrome("abca"))
````

### Đảo ngược một đoạn tại chỗ

Hai con trỏ ở hai đầu, đổi chỗ rồi tiến vào trong — dùng rất nhiều (đảo mảng, xoay mảng, đảo từ).

````python
def dao_doan(a, i, j):
    while i < j:
        a[i], a[j] = a[j], a[i]
        i += 1; j -= 1

arr_tp = [1, 2, 3, 4, 5, 6]
dao_doan(arr_tp, 1, 4)
print("đảo đoạn [1..4]:", arr_tp)
````

### Lỗi hay gặp

- Quên **sắp xếp** trước khi dùng hai đầu cho bài tổng.
- Điều kiện `L <= R` thay vì `L < R` → dùng một phần tử hai lần.
- Không chứng minh được vì sao bỏ một đầu là an toàn → dễ bỏ mất đáp án.
- Con trỏ ghi vượt con trỏ đọc (không thể xảy ra nếu `write` chỉ tăng khi ghi).

## 2.2 Tìm kiếm nhị phân (binary search)

### Ý tưởng

Trên mảng **đã sắp xếp**, so phần tử cần tìm với phần tử **ở giữa**: nếu nhỏ hơn thì chắc chắn nằm ở nửa trái, lớn hơn thì ở nửa phải. Mỗi bước loại **một nửa** → chỉ `O(log n)` bước (mảng 1 triệu phần tử: ~20 bước).

```text
tìm 23 trong [2, 5, 8, 12, 16, 23, 38, 56, 72, 91]
lo=0 hi=9 mid=4 → 16 < 23 → lo = 5
lo=5 hi=9 mid=7 → 56 > 23 → hi = 6
lo=5 hi=6 mid=5 → 23 = 23 → tìm thấy ở chỉ số 5
```

````python
def tim_nhi_phan(a, x):
    lo, hi = 0, len(a) - 1                # khoảng tìm đóng [lo, hi]
    while lo <= hi:
        mid = (lo + hi) // 2
        print(f"   lo={lo} hi={hi} mid={mid} a[mid]={a[mid]}")
        if a[mid] == x:
            return mid
        if a[mid] < x:
            lo = mid + 1
        else:
            hi = mid - 1
    return -1

a_bs = [2, 5, 8, 12, 16, 23, 38, 56, 72, 91]
print("→", tim_nhi_phan(a_bs, 23))
print("→", tim_nhi_phan(a_bs, 7), "(không có)")
````

### Dạng tổng quát: tìm vị trí **đầu tiên** thỏa điều kiện

Rất nhiều bài không hỏi "có `x` không" mà hỏi "vị trí đầu tiên mà điều kiện `P(i)` đúng", với `P` **đơn điệu**: `False False … False True True … True`. Khung:

```text
lo, hi = 0, n            # đáp án nằm trong [lo, hi]; n nghĩa là "không có"
while lo < hi:
    mid = (lo + hi) // 2
    if P(mid): hi = mid          # mid có thể là đáp án → giữ lại
    else:      lo = mid + 1      # mid chắc chắn không phải
return lo
```

Hai trường hợp quan trọng:

- **lower_bound**: vị trí đầu tiên `a[i] >= x` (`bisect_left`).
- **upper_bound**: vị trí đầu tiên `a[i] > x` (`bisect_right`).

Số lần xuất hiện của `x` = `upper_bound − lower_bound`.

````python
from bisect import bisect_left, bisect_right

def dau_tien(n, P):
    lo, hi = 0, n
    while lo < hi:
        mid = (lo + hi) // 2
        if P(mid):
            hi = mid
        else:
            lo = mid + 1
    return lo

b_bs = [1, 2, 2, 2, 5, 7]
lb = dau_tien(len(b_bs), lambda i: b_bs[i] >= 2)
ub = dau_tien(len(b_bs), lambda i: b_bs[i] > 2)
print("lower_bound(2) =", lb, "| upper_bound(2) =", ub, "| số lần xuất hiện:", ub - lb)
print("bisect_left / bisect_right:", bisect_left(b_bs, 2), bisect_right(b_bs, 2))
print("chèn 6 vào vị trí:", bisect_left(b_bs, 6))
print("số nguyên nhỏ nhất k với k*k >= 50:", dau_tien(100, lambda k: k * k >= 50))
````

### Lỗi hay gặp (nguồn lỗi số 1 là "lệch 1")

- Trộn hai kiểu khoảng: `while lo <= hi` đi với `hi = mid - 1`; `while lo < hi` đi với `hi = mid`. Trộn lẫn → **lặp vô hạn** hoặc bỏ sót.
- `lo = mid` (thay vì `mid + 1`) với `mid = (lo+hi)//2` → khi `hi = lo + 1` thì `mid = lo`, không tiến → lặp vô hạn.
- Áp dụng trên mảng **chưa sắp xếp** / điều kiện không đơn điệu.
- (Ngôn ngữ khác) `(lo + hi) / 2` tràn số → dùng `lo + (hi - lo) // 2`; Python không bị.

````python
def sai_lo_bang_mid(a, x, gioi_han=20):
    lo, hi, buoc = 0, len(a) - 1, 0
    while lo < hi and buoc < gioi_han:
        mid = (lo + hi) // 2
        if a[mid] < x:
            lo = mid            # SAI: không loại mid
        else:
            hi = mid
        buoc += 1
    return buoc
print("lo = mid: số vòng trước khi bị chặn:", sai_lo_bang_mid([1, 3], 3))
````

## 2.3 Áp dụng vào bài này

Mảng **đã sắp xếp** → **hai con trỏ ở hai đầu**: tổng nhỏ quá thì tăng trái, lớn quá thì giảm phải; mỗi bước loại được cả một hàng/cột các cặp — `O(n)` thời gian, `O(1)` bộ nhớ (đề yêu cầu bộ nhớ hằng). Cách thứ hai: với mỗi phần tử, **tìm nhị phân** phần bù trong phần còn lại — `O(n log n)`.

# 4. Ba cách chưa tối ưu

**Cách A — thử mọi cặp:** `O(n²)`, với `n = 3·10^4` là `4.5·10^8` cặp → quá chậm.

**Cách B — tìm nhị phân phần bù:** với mỗi `i`, tìm `target − numbers[i]` trong `numbers[i+1:]` bằng `bisect` → `O(n log n)`, `O(1)` bộ nhớ.

**Cách C — bảng băm (như Two Sum #1):** `O(n)` thời gian nhưng `O(n)` bộ nhớ — **vi phạm** yêu cầu bộ nhớ hằng số, và không tận dụng việc mảng đã sắp xếp.

````python
def two_sum_brute(numbers, target):
    for i in range(len(numbers)):
        for j in range(i + 1, len(numbers)):
            if numbers[i] + numbers[j] == target:
                return [i + 1, j + 1]

def two_sum_binary(numbers, target):
    for i, x in enumerate(numbers):
        j = bisect_left(numbers, target - x, i + 1)       # chỉ tìm bên phải i
        if j < len(numbers) and numbers[j] == target - x:
            return [i + 1, j + 1]

def two_sum_hash(numbers, target):
    seen = {}
    for j, x in enumerate(numbers):
        if target - x in seen:
            return [seen[target - x] + 1, j + 1]
        seen.setdefault(x, j)

for f in [two_sum_brute, two_sum_binary, two_sum_hash]:
    print(f"{f.__name__:<15}", f([2, 7, 11, 15], 9), f([2, 3, 4], 6), f([-1, 0], -1))
````

# 5. Ý tưởng chính: hai con trỏ ở hai đầu

Đặt `L = 0`, `R = n − 1`, xét `s = numbers[L] + numbers[R]`:

- `s == target` → xong.
- `s < target` → cần tổng **lớn hơn**. Giữ `L` và giảm `R` chỉ làm tổng nhỏ đi; vậy **mọi** cặp `(L, R')` với `R' <= R` đều `<= s < target` → `L` không thể là một nửa của đáp án → bỏ `L`: `L += 1`.
- `s > target` → đối xứng: mọi cặp `(L', R)` với `L' >= L` đều `>= s > target` → bỏ `R`: `R -= 1`.

```text
[2, 7, 11, 15], target 9
 L         R    2 + 15 = 17 > 9  → R lùi
 L     R        2 + 11 = 13 > 9  → R lùi
 L  R           2 + 7  = 9       ✓ → [1, 2]
```

Mỗi bước loại một phần tử → tối đa `n − 1` bước.

````python
def two_sum_pointers(numbers: List[int], target: int) -> List[int]:
    L, R = 0, len(numbers) - 1
    while L < R:
        s = numbers[L] + numbers[R]
        if s == target:
            return [L + 1, R + 1]
        if s < target:
            L += 1
        else:
            R -= 1
    return []                                   # đề đảm bảo không xảy ra

print(two_sum_pointers([2, 7, 11, 15], 9), two_sum_pointers([2, 3, 4], 6), two_sum_pointers([-1, 0], -1))
````

# 6. Mô phỏng

````python
def trace_two_sum(numbers, target):
    L, R, step = 0, len(numbers) - 1, 0
    while L < R:
        s = numbers[L] + numbers[R]
        step += 1
        if s == target:
            print(f"bước {step}: L={L} R={R}: {numbers[L]} + {numbers[R]} = {s} = target ✓")
            return [L + 1, R + 1]
        act = "nhỏ hơn → L += 1" if s < target else "lớn hơn → R -= 1"
        print(f"bước {step}: L={L} R={R}: {numbers[L]} + {numbers[R]} = {s:>3} {act}")
        if s < target:
            L += 1
        else:
            R -= 1

print("→", trace_two_sum([1, 3, 4, 6, 8, 11, 14], 10))
````

# 7. Phiên bản đúng format LeetCode

````python
class Solution:
    def twoSum(self, numbers: List[int], target: int) -> List[int]:
        L, R = 0, len(numbers) - 1
        while L < R:
            s = numbers[L] + numbers[R]
            if s == target:
                return [L + 1, R + 1]
            if s < target:
                L += 1
            else:
                R -= 1
        return []
````

````python
def check(name, actual, expected):
    assert actual == expected, (name, actual, expected)
    print(f"{name:<14} PASS  {actual}")


sol = Solution()
check("Ví dụ 1", sol.twoSum([2, 7, 11, 15], 9), [1, 2])
check("Ví dụ 2", sol.twoSum([2, 3, 4], 6), [1, 3])
check("Ví dụ 3", sol.twoSum([-1, 0], -1), [1, 2])
check("Trùng giá trị", sol.twoSum([1, 2, 2, 7], 4), [2, 3])
check("Hai số giống", sol.twoSum([0, 0, 3, 4], 0), [1, 2])

def random_unique_answer(n):
    """Sinh mảng sắp xếp có ĐÚNG MỘT cặp chỉ số tổng bằng target."""
    while True:
        a = sorted(random.randint(-30, 30) for _ in range(n))
        sums = {}
        for i in range(n):
            for j in range(i + 1, n):
                sums.setdefault(a[i] + a[j], []).append((i, j))
        uniq = [t for t, ps in sums.items() if len(ps) == 1]
        if uniq:
            t = random.choice(uniq)
            return a, t, [sums[t][0][0] + 1, sums[t][0][1] + 1]

random.seed(167)
for _ in range(2000):
    a, t, exp = random_unique_answer(random.randint(2, 10))
    assert sol.twoSum(a, t) == exp == two_sum_brute(a, t) == two_sum_binary(a, t), (a, t)
print("2000 mảng ngẫu nhiên (đáp án duy nhất): khớp brute force")

n = 3 * 10**4
big = sorted(random.sample(range(-10**6, 10**6), n))
t_ = big[n // 2 - 1] + big[n // 2]
for name, f in [("hai con trỏ", sol.twoSum), ("nhị phân", two_sum_binary)]:
    t0 = time.perf_counter(); r = f(big, t_); t1 = time.perf_counter()
    print(f"n = 3·10^4, {name:<11}: {(t1 - t0) * 1000:.2f} ms → {r}")
small = big[::10]
t0 = time.perf_counter(); two_sum_brute(small, small[1498] + small[1501]); t1 = time.perf_counter()
print(f"thử mọi cặp, chỉ với n = 3000: {(t1 - t0) * 1000:.0f} ms (độ phức tạp O(n²): n gấp 10 thì trường hợp xấu lâu gấp ~100 lần)")
````

# 8. Vì sao đúng? Độ phức tạp?

Gọi đáp án là `(i*, j*)` với `i* < j*`. **Bất biến:** luôn có `L <= i*` và `j* <= R`.

- Ban đầu `L = 0`, `R = n − 1` → đúng.
- Giả sử `L < i*` và ta tăng `L` vì `s < target`. Nếu `L` là một nửa của đáp án thì `L = i*`, mâu thuẫn. Còn nếu `L = i*` thì `numbers[i*] + numbers[R] < target = numbers[i*] + numbers[j*]` → `numbers[R] < numbers[j*]`, nhưng `R >= j*` và mảng tăng nên `numbers[R] >= numbers[j*]` — mâu thuẫn. Vậy khi tăng `L` thì chắc chắn `L < i*`, bất biến giữ.
- Đối xứng cho việc giảm `R`.

`R − L` giảm 1 mỗi bước; bất biến đảm bảo vòng lặp dừng tại đúng `(i*, j*)`.

```text
                 Time          Bộ nhớ phụ
Thử mọi cặp      O(n²)         O(1)
Nhị phân         O(n log n)    O(1)
Bảng băm         O(n)          O(n)
Hai con trỏ      O(n)          O(1)   ← đáp ứng đề
```

# 9. Những lỗi dễ mắc

## Lỗi 1: Trả về chỉ số 0-based

````python
def wrong_zero_based(numbers, target):
    L, R = 0, len(numbers) - 1
    while L < R:
        s = numbers[L] + numbers[R]
        if s == target:
            return [L, R]                          # SAI: đề dùng chỉ số từ 1
        if s < target: L += 1
        else: R -= 1

print("Sai :", wrong_zero_based([2, 7, 11, 15], 9), "| Đúng:", sol.twoSum([2, 7, 11, 15], 9))
````

## Lỗi 2: Điều kiện `L <= R` (dùng một phần tử hai lần)

````python
def wrong_le(numbers, target):
    L, R = 0, len(numbers) - 1
    while L <= R:                                  # SAI: cho phép L == R
        s = numbers[L] + numbers[R]
        if s == target:
            return [L + 1, R + 1]
        if s < target: L += 1
        else: R -= 1
    return []

print("numbers = [1, 3, 5], target = 6 (đáp án [1, 3]) → sai:", wrong_le([1, 3, 5], 6), "| đúng:", sol.twoSum([1, 3, 5], 6))
print("numbers = [1, 2, 3], target = 6 → sai:", wrong_le([1, 2, 3], 6), "(dùng số 3 hai lần), trong khi không có cặp hợp lệ")
````

Với đầu vào hợp lệ của đề (luôn có đáp án), hai con trỏ gặp đáp án trước khi `L == R` nên lỗi bị che; nhưng với đầu vào không có đáp án, `L <= R` trả về cặp `[3, 3]` vô nghĩa.

## Lỗi 3: Dời sai con trỏ

````python
def wrong_direction(numbers, target, limit=50):
    L, R, steps = 0, len(numbers) - 1, 0
    while L < R and steps < limit:
        s = numbers[L] + numbers[R]
        if s == target:
            return [L + 1, R + 1]
        if s < target:
            R -= 1                                 # SAI: tổng nhỏ mà lại làm nhỏ thêm
        else:
            L += 1
        steps += 1
    return []

print("Sai :", wrong_direction([2, 7, 11, 15], 9), "| Đúng:", sol.twoSum([2, 7, 11, 15], 9))
````

`2 + 15 = 17 > 9`, bản sai lại tăng `L` (bỏ số `2` — một nửa của đáp án) → không bao giờ tìm thấy.

## Lỗi 4: Dùng bảng băm khi đề yêu cầu `O(1)` bộ nhớ

Kết quả đúng nhưng **không** đạt yêu cầu bộ nhớ; trong phỏng vấn đây là điểm trừ vì bỏ qua thông tin "đã sắp xếp".

# 10. Mẹo nhận diện pattern

- **Mảng đã sắp xếp + tìm cặp thỏa điều kiện về tổng/hiệu** → hai con trỏ ở hai đầu.
- **Chưa sắp xếp** → sắp xếp trước (`O(n log n)`) rồi hai con trỏ, hoặc bảng băm.

## Đáp án Bài 3

````python
def answer_3_count_pairs_at_most(numbers: List[int], target: int) -> int:
    L, R, cnt = 0, len(numbers) - 1, 0
    while L < R:
        if numbers[L] + numbers[R] <= target:
            cnt += R - L
            L += 1
        else:
            R -= 1
    return cnt

assert answer_3_count_pairs_at_most([1, 2, 3, 4], 5) == 4
assert answer_3_count_pairs_at_most([1, 1, 1], 2) == 3
assert answer_3_count_pairs_at_most([5, 6], 3) == 0
for _ in range(3000):
    a = sorted(random.randint(-10, 10) for _ in range(random.randint(0, 12)))
    t = random.randint(-20, 20)
    assert answer_3_count_pairs_at_most(a, t) == sum(a[i] + a[j] <= t for i in range(len(a)) for j in range(i + 1, len(a)))
print("Bài 3 PASS")
````

## Đáp án Bài 4

````python
def answer_4_closest_pair_sum(numbers: List[int], target: int) -> int:
    L, R = 0, len(numbers) - 1
    best = numbers[L] + numbers[R]
    while L < R:
        s = numbers[L] + numbers[R]
        if (abs(s - target), s) < (abs(best - target), best):
            best = s
        if s == target:
            return s
        if s < target:
            L += 1
        else:
            R -= 1
    return best

assert answer_4_closest_pair_sum([1, 4, 6, 9], 12) == 13
assert answer_4_closest_pair_sum([1, 4, 6, 9], 11) == 10
assert answer_4_closest_pair_sum([-3, -1, 2], 0) == -1
for _ in range(3000):
    a = sorted(random.randint(-10, 10) for _ in range(random.randint(2, 10)))
    t = random.randint(-25, 25)
    exp = min((abs(a[i] + a[j] - t), a[i] + a[j]) for i in range(len(a)) for j in range(i + 1, len(a)))[1]
    assert answer_4_closest_pair_sum(a, t) == exp, (a, t)
print("Bài 4 PASS")
````

## Đáp án Bài 5

````python
def answer_5_find_target(root: Optional[TreeNode], k: int) -> bool:
    vals = []
    def inorder(node):
        if node:
            inorder(node.left); vals.append(node.val); inorder(node.right)
    inorder(root)
    L, R = 0, len(vals) - 1
    while L < R:
        s = vals[L] + vals[R]
        if s == k:
            return True
        if s < k:
            L += 1
        else:
            R -= 1
    return False

assert answer_5_find_target(bst_from_sorted([2, 3, 4, 5, 6, 7]), 9) is True
assert answer_5_find_target(bst_from_sorted([2, 3, 4, 5, 6, 7]), 28) is False
assert answer_5_find_target(bst_from_sorted([1]), 2) is False
for _ in range(2000):
    a = sorted(random.sample(range(-20, 20), random.randint(1, 10)))
    k = random.randint(-40, 40)
    exp = any(a[i] + a[j] == k for i in range(len(a)) for j in range(i + 1, len(a)))
    assert answer_5_find_target(bst_from_sorted(a), k) == exp
print("Bài 5 PASS")
````

# 14. Tổng kết

> Mảng đã sắp xếp → hai con trỏ ở hai đầu: tổng nhỏ hơn thì `L += 1`, lớn hơn thì `R -= 1`, mỗi bước loại chắc chắn một phần tử không thể là đáp án. `O(n)` thời gian, `O(1)` bộ nhớ. Nhớ trả chỉ số **từ 1** và dùng `L < R`.
