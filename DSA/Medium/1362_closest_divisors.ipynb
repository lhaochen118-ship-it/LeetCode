## 1. Mục tiêu học

1. Biết rằng ước số luôn đi theo **cặp** `(i, x/i)` và chỉ cần duyệt `i` tới `√x`.
2. Hiểu vì sao duyệt `i` **từ `√x` đi xuống**, ước đầu tiên gặp được cho **cặp gần nhau nhất** (hiệu `x/i − i` giảm khi `i` tăng).
3. Áp dụng cho **hai** số `num + 1` và `num + 2`, so sánh đúng cách; tối ưu bằng **một vòng lặp chung**.
4. Dùng `math.isqrt` (căn nguyên chính xác) thay cho `int(math.sqrt(x))`.
5. Viết test cho bài **nhiều đáp án đúng** (thứ tự tùy ý) bằng hàm kiểm tra tính hợp lệ.

## 2. Đề bài & ví dụ

Cho số nguyên `num`. Tìm hai số nguyên có **hiệu tuyệt đối nhỏ nhất** sao cho **tích** của chúng bằng `num + 1` **hoặc** `num + 2`. Trả về hai số theo thứ tự bất kỳ.

**Ví dụ 1:** `num = 8` → `[3, 3]` (với 9: cặp gần nhất 3 & 3; với 10: 2 & 5; chọn 3 & 3).

**Ví dụ 2:** `num = 123` → `[5, 25]`.

**Ví dụ 3:** `num = 999` → `[40, 25]`.

**Ràng buộc:** `1 <= num <= 10^9`.

Lưu ý: đáp án có thể có nhiều dạng (`[5, 25]` hay `[25, 5]`), ta kiểm tra bằng **hàm hợp lệ**: tích thuộc `{num+1, num+2}` và hiệu bằng hiệu tối ưu.

````python
import math
import random
import time

examples = [
    (8, [3, 3]),
    (123, [5, 25]),
    (999, [40, 25]),
]
for num, want in examples:
    a, b = want
    print(f"num={num}: mong đợi {want} | tích = {a * b} (num+1={num + 1}, num+2={num + 2}) | hiệu = {abs(a - b)}")
````

## 3. Kiến thức nền

### 3.1 Ước số đi theo cặp

Nếu `i` chia hết `x` thì `x // i` cũng là ước, và `i · (x//i) = x`. Trong mỗi cặp, số nhỏ `≤ √x ≤` số lớn. Vậy chỉ cần duyệt `i = 1..⌊√x⌋` là liệt kê đủ mọi cặp.

````python
x = 36
pairs = [(i, x // i) for i in range(1, math.isqrt(x) + 1) if x % i == 0]
print("các cặp ước của 36:", pairs)
````

### 3.2 `math.isqrt` và vì sao không nên dùng `int(math.sqrt(x))`

`math.isqrt(x)` trả về `⌊√x⌋` **chính xác** cho số nguyên bất kỳ. `math.sqrt` dùng số thực 64-bit — với số rất lớn có thể làm tròn sai (ở bài này `x ≤ 10^9 + 2` thì vẫn đúng, nhưng thói quen dùng `isqrt` an toàn hơn).

````python
for x in [9, 10, 10**9 + 2, 10**30 - 1]:
    print(f"x={x}: isqrt={math.isqrt(x)}, int(sqrt)={int(math.sqrt(x))}")
print("(dòng cuối: int(sqrt) sai vì làm tròn số thực — ngoài phạm vi bài nhưng là bài học)")
````

### 3.3 Hiệu của cặp ước giảm khi số nhỏ tăng

Với `i ≤ √x`, hiệu `d(i) = x/i − i` là hàm **giảm** theo `i` (khi `i` tăng, `x/i` giảm và `−i` cũng giảm). Do đó ước `i ≤ √x` **lớn nhất** cho cặp gần nhất.

````python
x = 360
for i in range(1, math.isqrt(x) + 1):
    if x % i == 0:
        print(f"i={i:2d}, x/i={x // i:3d}, hiệu={x // i - i}")
````

### 3.4 Vòng lặp ngược với `range(start, stop, -1)` và `for ... else`

`range(k, 0, -1)` cho `k, k−1, …, 1`. Ta `return` ngay khi gặp ước đầu tiên.

````python
print(list(range(5, 0, -1)))
def first_divisor_from_top(x):
    for i in range(math.isqrt(x), 0, -1):
        if x % i == 0:
            return i
print(first_divisor_from_top(10), first_divisor_from_top(13), first_divisor_from_top(49))
````

### 3.5 Chọn phần tử tốt nhất bằng `min(..., key=...)`

`min(list_cặp, key=lambda p: abs(p[0] − p[1]))` chọn cặp có hiệu nhỏ nhất.

````python
cands = [(3, 3), (2, 5)]
print(min(cands, key=lambda p: abs(p[0] - p[1])))
````

## 4. Ý tưởng nền tảng / trực giác

Tính tay **Ví dụ 2** (`num = 123` ⇒ xét 124 và 125):

| x | ⌊√x⌋ | duyệt i giảm dần | ước đầu tiên | cặp | hiệu |
|---|---|---|---|---|---|
| 124 | 11 | 11, 10, 9, 8, 7, 6, 5, **4** | 4 | (4, 31) | 27 |
| 125 | 11 | 11, 10, 9, 8, 7, 6, **5** | 5 | (5, 25) | 20 |

Chọn hiệu nhỏ hơn ⇒ `[5, 25]` ✔.

**Ví dụ 3** (`num = 999` ⇒ 1000 và 1001): `√1000 ≈ 31.6`; duyệt xuống 31, 30, …, **25** (1000 = 25·40, hiệu 15). Với 1001: duyệt xuống tới **13** (1001 = 7·11·13 = 13·77, hiệu 64). Chọn `[25, 40]` ✔.

````python
def closest_pair_of(x):
    for i in range(math.isqrt(x), 0, -1):
        if x % i == 0:
            return i, x // i

for num, want in examples:
    p1, p2 = closest_pair_of(num + 1), closest_pair_of(num + 2)
    print(f"num={num}: {num + 1} -> {p1} (hiệu {p1[1] - p1[0]}), {num + 2} -> {p2} (hiệu {p2[1] - p2[0]})")
````

## 5. Cách 1 — Brute Force / baseline: duyệt mọi ước

**Ý tưởng:** với mỗi `x ∈ {num+1, num+2}`, duyệt **mọi** `i` từ 1 tới `x`; nếu `i` chia hết `x` thì xét cặp `(i, x//i)`, giữ cặp có hiệu nhỏ nhất.

````python
def closest_brute(num):
    best = None
    for x in (num + 1, num + 2):
        for i in range(1, x + 1):
            if x % i == 0:
                j = x // i
                if best is None or abs(i - j) < abs(best[0] - best[1]):
                    best = [i, j]
    return best

for num, want in examples:
    print(f"num={num} -> {closest_brute(num)} (mong đợi {want} hoặc đảo thứ tự)")
````

**Độ phức tạp:** `O(num)` — với `num = 10^9` là hai tỷ phép chia, không thể chạy kịp.

**Nút thắt:** (1) mỗi cặp bị xét hai lần (`(i, j)` và `(j, i)`), (2) duyệt quá `√x` là thừa, (3) duyệt từ nhỏ lên phải đi hết mới biết cặp nào tốt nhất.

## 6. Ý tưởng thuật toán / bước đột phá

**Insight 1:** chỉ cần `i ≤ √x` ⇒ `O(√x)` thay vì `O(x)`.

**Insight 2:** hiệu `x/i − i` giảm khi `i` tăng (với `i ≤ √x`) ⇒ duyệt `i` **từ `⌊√x⌋` xuống**, ước **đầu tiên** gặp là tốt nhất ⇒ dừng sớm.

**Insight 3 (một vòng chung):** duyệt `i` từ `⌊√(num+2)⌋` xuống; tại mỗi `i`, kiểm tra `num+1` trước rồi `num+2`; ước đầu tiên tìm được (cho bất kỳ số nào) là đáp án tối ưu.
- Cùng một `i`: `(num+1)/i − i < (num+2)/i − i` ⇒ ưu tiên `num+1`.
- `i` lớn hơn luôn thắng `i` nhỏ hơn (chứng minh ở mục 10).

**Quy tắc nếu… thì…:**
- Nếu `(num+1) % i == 0` thì trả `[i, (num+1)//i]`.
- Ngược lại nếu `(num+2) % i == 0` thì trả `[i, (num+2)//i]`.
- Vòng lặp chắc chắn dừng vì `i = 1` chia hết mọi số.

## 7. Các cách tiếp theo

### 7.1 Cách 2 — Duyệt `i` từ 1 tới `√x`, giữ cặp tốt nhất

**Giải thích:** áp dụng Insight 1 nhưng vẫn duyệt từ nhỏ lên: cặp tốt nhất là cặp cuối cùng tìm được (vì hiệu giảm dần), nhưng phải duyệt hết tới `√x`.

````python
def closest_sqrt_up(num):
    best = None
    for x in (num + 1, num + 2):
        for i in range(1, math.isqrt(x) + 1):
            if x % i == 0:
                j = x // i
                if best is None or j - i < best[1] - best[0]:
                    best = [i, j]
    return best

for num, want in examples:
    print(f"num={num} -> {closest_sqrt_up(num)}")
````

**Độ phức tạp:** `O(√num)` ≈ `2 × 31623` phép chia — nhanh.

### 7.2 Cách 3 — Từ `√x` đi xuống cho từng số, rồi so sánh

**Giải thích từng đoạn:**
1. Hàm `closest_pair_of(x)`: `for i in range(isqrt(x), 0, -1)`, gặp ước đầu tiên thì trả `(i, x//i)`.
2. Gọi cho `num+1` và `num+2`, chọn cặp có hiệu nhỏ hơn.

````python
def closest_down_each(num):
    a = closest_pair_of(num + 1)
    b = closest_pair_of(num + 2)
    return list(a) if a[1] - a[0] <= b[1] - b[0] else list(b)

for num, want in examples:
    print(f"num={num} -> {closest_down_each(num)}")
````

Code nộp LeetCode (Cách 3):

````python
class Solution:
    def closestDivisors(self, num: int) -> list[int]:
        def best(x):
            for i in range(math.isqrt(x), 0, -1):
                if x % i == 0:
                    return [i, x // i]
        a, b = best(num + 1), best(num + 2)
        return a if a[1] - a[0] <= b[1] - b[0] else b

print(Solution().closestDivisors(999))
````

**Độ phức tạp:** worst `O(√num)` (khi số nguyên tố — phải đi tới `i = 1`), thường dừng rất sớm.

### 7.3 Cách 4 — Một vòng lặp chung cho cả hai số

**Giải thích:** `i` chạy từ `isqrt(num + 2)` xuống 1; ở mỗi bước kiểm tra `num+1` rồi `num+2`. Ước đầu tiên gặp là đáp án. Gọn và dừng sớm nhất có thể.

````python
def closest_joint(num):
    for i in range(math.isqrt(num + 2), 0, -1):
        if (num + 1) % i == 0:
            return [i, (num + 1) // i]
        if (num + 2) % i == 0:
            return [i, (num + 2) // i]

for num, want in examples:
    print(f"num={num} -> {closest_joint(num)}")
````

Code nộp LeetCode (Cách 4):

````python
class Solution:
    def closestDivisors(self, num: int) -> list[int]:
        for i in range(math.isqrt(num + 2), 0, -1):
            if (num + 1) % i == 0:
                return [i, (num + 1) // i]
            if (num + 2) % i == 0:
                return [i, (num + 2) // i]

print(Solution().closestDivisors(8), Solution().closestDivisors(123))
````

### 7.4 Cách 5 — Phân tích thừa số nguyên tố rồi sinh mọi ước

**Giải thích:** phân tích `x = p1^e1 · p2^e2 · …` (chia thử tới `√x`), sinh **toàn bộ** ước từ các lũy thừa, chọn ước `d ≤ √x` lớn nhất. Không nhanh hơn về độ phức tạp worst-case, nhưng cho thấy liên hệ với chủ đề *Prime Factorization* và hữu ích khi cần **tất cả** ước.

````python
def factorize(x):
    f, p = {}, 2
    while p * p <= x:
        while x % p == 0:
            f[p] = f.get(p, 0) + 1
            x //= p
        p += 1 if p == 2 else 2
    if x > 1:
        f[x] = f.get(x, 0) + 1
    return f

def all_divisors(x):
    divs = [1]
    for p, e in factorize(x).items():
        divs = [d * p ** k for d in divs for k in range(e + 1)]
    return sorted(divs)

def closest_factorize(num):
    best = None
    for x in (num + 1, num + 2):
        r = math.isqrt(x)
        d = max(v for v in all_divisors(x) if v <= r)
        if best is None or x // d - d < best[1] - best[0]:
            best = [d, x // d]
    return best

print("1000 =", factorize(1000), "| ước:", all_divisors(1000))
for num, want in examples:
    print(f"num={num} -> {closest_factorize(num)}")
````

## 8. TRACE từng bước

Trace Cách 4 trên `num = 999` (xét 1000 và 1001, bắt đầu từ `isqrt(1001) = 31`):

| i | 1000 % i | 1001 % i | Kết quả |
|---|---|---|---|
| 31 | 8 | 9 | |
| 30 | 10 | 11 | |
| 29 | 14 | 15 | |
| 28 | 20 | 21 | |
| 27 | 1 | 2 | |
| 26 | 12 | 13 | |
| 25 | **0** | | trả `[25, 40]` |

````python
def trace_joint(num):
    a, b = num + 1, num + 2
    print(f"num={num}: xét {a} và {b}, bắt đầu từ i = isqrt({b}) = {math.isqrt(b)}")
    for i in range(math.isqrt(b), 0, -1):
        ra, rb = a % i, b % i
        print(f"  i={i:3d}: {a} % i = {ra:3d}, {b} % i = {rb:3d}")
        if ra == 0:
            print(f"  -> {a} = {i} × {a // i}")
            return [i, a // i]
        if rb == 0:
            print(f"  -> {b} = {i} × {b // i}")
            return [i, b // i]

trace_joint(999)
print()
trace_joint(8)
````

Trace Cách 3 trên `num = 123` (mỗi số một vòng riêng):

````python
def trace_each(num):
    res = []
    for x in (num + 1, num + 2):
        tried = []
        for i in range(math.isqrt(x), 0, -1):
            tried.append(i)
            if x % i == 0:
                print(f"x={x}: thử {tried} -> ước {i}, cặp ({i}, {x // i}), hiệu {x // i - i}")
                res.append([i, x // i])
                break
    ans = min(res, key=lambda p: p[1] - p[0])
    print("chọn:", ans)
    return ans

trace_each(123)
````

## 9. Code nộp LeetCode

````python
import math

class Solution:
    def closestDivisors(self, num: int) -> list[int]:
        # i giảm dần từ căn của num+2: ước đầu tiên gặp cho cặp gần nhau nhất
        for i in range(math.isqrt(num + 2), 0, -1):
            if (num + 1) % i == 0:              # cùng i thì num+1 cho hiệu nhỏ hơn
                return [i, (num + 1) // i]
            if (num + 2) % i == 0:
                return [i, (num + 2) // i]

def is_valid_answer(num, ans):
    """Kiểm tra đáp án (thứ tự tùy ý): tích đúng và hiệu tối ưu."""
    a, b = ans
    if a * b not in (num + 1, num + 2) or a < 1 or b < 1:
        return False
    best = min(closest_pair_of(num + 1)[1] - closest_pair_of(num + 1)[0],
               closest_pair_of(num + 2)[1] - closest_pair_of(num + 2)[0])
    return abs(a - b) == best

sol = Solution()
for num, want in examples:
    got = sol.closestDivisors(num)
    assert is_valid_answer(num, got) and sorted(got) == sorted(want)
    print(f"closestDivisors({num}) = {got}")
````

## 10. Vì sao thuật toán đúng

**Bổ đề 1 (một số):** với `x` cố định và ước `i ≤ √x`, hiệu `d(i) = x/i − i` giảm ngặt theo `i`. Mọi cặp `(a, b)` với `a·b = x`, `a ≤ b` có `a ≤ √x`. Vậy ước lớn nhất `≤ ⌊√x⌋` cho hiệu nhỏ nhất — chính là ước đầu tiên khi duyệt từ `⌊√x⌋` xuống. ⇒ Cách 3 đúng.

**Bổ đề 2 (vòng chung, Cách 4):** đặt `A = num+1`, `B = num+2 = A+1`.
- Nếu `B` là số chính phương `r²` thì `i = r = isqrt(B)` chia hết `B` ⇒ hiệu 0, tối ưu tuyệt đối. (Lúc đó `A = r² − 1` chỉ chia hết `r` khi `r = 1`, trường hợp `num = 0` ngoài ràng buộc.) Còn lại `isqrt(B) = isqrt(A)`, nên mọi `i` được duyệt đều `≤ √A` — là "số nhỏ" hợp lệ của cả hai.
- Cùng `i`: `A/i − i < B/i − i` ⇒ kiểm tra `A` trước là đúng.
- Khác `i`: giả sử tìm thấy ở `i1` (cho `X ∈ {A, B}`), còn một cặp khác ở `i2 < i1` (cho `Y ∈ {A, B}`, `Y ≥ X − 1`). Khi đó
  `Y/i2 − i2 ≥ (X−1)/(i1−1) − (i1−1)`, và `(X−1)/(i1−1) − (i1−1) − (X/i1 − i1) = (X − i1)/(i1(i1−1)) + 1 > 0`.
  ⇒ cặp ở `i1` luôn tốt hơn mọi cặp ở `i` nhỏ hơn. Vậy ước đầu tiên gặp là tối ưu.
- Vòng lặp luôn dừng vì `i = 1` chia hết `A`.

## 11. Phân tích Time / Space

Gọi `S = √(num + 2)` (≤ ~31623).

| Cách | Time worst | Time thường gặp | Space |
|---|---|---|---|
| 1. Duyệt mọi ước | `O(num)` | `O(num)` | `O(1)` |
| 2. Lên tới √x | `O(S)` | `O(S)` | `O(1)` |
| 3. Từ √x xuống, từng số | `O(S)` | dừng sớm | `O(1)` |
| 4. Vòng chung | `O(S)` | dừng sớm nhất | `O(1)` |
| 5. Phân tích thừa số | `O(S + τ(x))` | `O(S)` | `O(τ(x))` (số ước) |

Worst case của Cách 4: cả `num+1`, `num+2` có ước "nhỏ" duy nhất rất nhỏ (ví dụ một số nguyên tố và gấp đôi một số nguyên tố) ⇒ duyệt gần hết `S` bước.

## 12. Bộ test

### 12.1 Ví dụ đề + edge case (kiểm tra bằng hàm hợp lệ)

````python
METHODS = {
    "brute": closest_brute,
    "sqrt_up": closest_sqrt_up,
    "down_each": closest_down_each,
    "joint": closest_joint,
    "factorize": closest_factorize,
    "submit": Solution().closestDivisors,
}
small_cases = [1, 2, 3, 7, 8, 14, 15, 123, 999, 1000]
for num in small_cases:
    for name, f in METHODS.items():
        got = f(num)
        assert is_valid_answer(num, got), (name, num, got)
print("Một số giá trị:", {num: Solution().closestDivisors(num) for num in [1, 2, 3, 7, 14, 15]})
print(f"{len(small_cases)} giá trị × {len(METHODS)} cách: đều hợp lệ")
````

### 12.2 Đối chiếu toàn bộ `num = 1..3000` và ngẫu nhiên cỡ lớn (seed cố định)

Với `num` nhỏ, so hiệu với brute force; ngoài ra vì hai cặp có cùng hiệu tối ưu có thể khác nhau, ta chỉ so **hiệu** và kiểm tra **tích**.

````python
def best_diff_brute(num):
    p = closest_brute(num)
    return abs(p[0] - p[1])

for num in range(1, 3001):
    bd = best_diff_brute(num)
    for name, f in METHODS.items():
        if name == "brute":
            continue
        a, b = f(num)
        assert a * b in (num + 1, num + 2) and abs(a - b) == bd, (name, num)
print("num = 1..3000: mọi cách cho hiệu tối ưu")

rng = random.Random(1362)
for _ in range(200):
    num = rng.randint(1, 10**9)
    outs = {name: METHODS[name](num) for name in ["sqrt_up", "down_each", "joint", "submit"]}
    diffs = {abs(a - b) for a, b in outs.values()}
    assert len(diffs) == 1 and all(a * b in (num + 1, num + 2) for a, b in outs.values()), num
print("200 num ngẫu nhiên tới 1e9: các cách nhanh thống nhất hiệu tối ưu")
````

### 12.3 Trường hợp xấu & thời gian

### Lỗi 1 — Chỉ xét `num + 1` (hoặc chỉ `num + 2`)

````python
def wrong_only_plus1(num):
    return list(closest_pair_of(num + 1))

def wrong_only_plus2(num):
    return list(closest_pair_of(num + 2))

for num in [7, 8, 14]:
    print(f"num={num}: chỉ +1 -> {wrong_only_plus1(num)}, chỉ +2 -> {wrong_only_plus2(num)}, đúng -> {closest_joint(num)}")
````

### Lỗi 2 — Duyệt từ 1 đi lên rồi dừng ở ước đầu tiên

````python
def wrong_up_break(num):
    best = None
    for x in (num + 1, num + 2):
        for i in range(1, math.isqrt(x) + 1):
            if x % i == 0:
                cand = [i, x // i]
                break                          # dừng ở i = 1 luôn!
        if best is None or cand[1] - cand[0] < best[1] - best[0]:
            best = cand
    return best

for num, want in examples:
    print(f"num={num}: sai cho {wrong_up_break(num)}, đúng là {want}")
````

**Vì sao sai:** `i = 1` luôn là ước ⇒ luôn trả `[1, x]` — cặp **xa** nhất. **Sửa:** duyệt **từ `√x` xuống** rồi mới dừng sớm.

### Lỗi 3 — `range(..., 1, -1)` bỏ sót `i = 1`

````python
def wrong_stop_at_2(num):
    for i in range(math.isqrt(num + 2), 1, -1):    # không bao giờ thử i = 1
        if (num + 1) % i == 0:
            return [i, (num + 1) // i]
        if (num + 2) % i == 0:
            return [i, (num + 2) // i]
    return None

print("num=1 (2 và 3 đều nguyên tố): sai cho", wrong_stop_at_2(1), "| đúng là", closest_joint(1))
````

**Vì sao sai:** khi `num+1` và `num+2` đều nguyên tố (chỉ xảy ra với `num = 1`: 2 và 3), ước duy nhất là 1. **Sửa:** `range(isqrt(...), 0, -1)`.

### Lỗi 4 — Vòng chung nhưng kiểm tra `num + 2` trước `num + 1`

````python
def wrong_order(num):
    for i in range(math.isqrt(num + 2), 0, -1):
        if (num + 2) % i == 0:
            return [i, (num + 2) // i]
        if (num + 1) % i == 0:
            return [i, (num + 1) // i]

bad = [num for num in range(1, 2000) if abs(wrong_order(num)[0] - wrong_order(num)[1]) != best_diff_brute(num)]
print("số giá trị num ≤ 2000 bị sai:", len(bad), "| ví dụ:", bad[:5])
num = bad[0]
print(f"num={num}: sai cho {wrong_order(num)}, đúng là {closest_joint(num)}")
````

**Vì sao sai:** cùng một `i`, `num+1` cho hiệu `(num+1)/i − i` **nhỏ hơn** `num+2`. Vì `gcd(num+1, num+2) = 1`, hai số chỉ cùng chia hết cho `i = 1` — tức trường hợp cả hai đều nguyên tố (`num = 1`). Lỗi rất hiếm nên dễ lọt qua test ví dụ, nhưng vẫn là sai (bộ test ẩn có `num = 1`). **Sửa:** kiểm tra `num+1` trước (hoặc so sánh hiệu).

### Lỗi 5 — So sánh "cặp tốt hơn" bằng tổng thay vì hiệu

````python
def wrong_compare_sum(num):
    a, b = closest_pair_of(num + 1), closest_pair_of(num + 2)
    return list(a) if a[0] + a[1] <= b[0] + b[1] else list(b)

bad = [num for num in range(1, 2000) if abs(wrong_compare_sum(num)[0] - wrong_compare_sum(num)[1]) != best_diff_brute(num)]
print("số giá trị bị sai:", len(bad), "| ví dụ num =", bad[0], "->", wrong_compare_sum(bad[0]), "đúng:", closest_joint(bad[0]))
````

**Vì sao sai:** tổng nhỏ ≠ hiệu nhỏ khi hai tích khác nhau (`num+1` vs `num+2`). **Sửa:** so sánh đúng tiêu chí đề: `|a − b|`.

## 14. Mẹo làm bài nhanh & nhận dạng pattern

- **Ước số / cặp nhân tử** ⇒ duyệt tới `isqrt(x)`; muốn "gần nhau nhất" thì duyệt **ngược từ √x**.
- Luôn dùng `math.isqrt` cho căn nguyên.
- Bài "chọn trong nhiều ứng viên" (ở đây 2 số) ⇒ hoặc tính riêng rồi so sánh, hoặc gộp vào một vòng với **thứ tự ưu tiên** đúng.
- Bài trả về "thứ tự tùy ý / nhiều đáp án" ⇒ test bằng hàm hợp lệ, không so list trực tiếp.
- Pattern liên quan: **LeetCode 492 — Construct the Rectangle** (cùng ý tưởng cho một số).

## 15. Biến thể của bài

1. **LeetCode 492 — Construct the Rectangle:** chỉ một số `area`, trả `[L, W]` với `L ≥ W`, hiệu nhỏ nhất.
2. **Tích thuộc đoạn `[num+1, num+k]`** — duyệt chung, kiểm tra `k` số ở mỗi `i` (Bài 5).
3. **Ba thừa số gần nhau nhất** (`a·b·c = x`, tối thiểu `max − min`) — duyệt `a` tới `∛x`, rồi giải bài hai thừa số cho `x/a`.
4. **Đếm số ước** / **tổng ước** — cùng vòng `√x`.
5. **Số lớn tới `10^18`** — duyệt `√x` quá chậm (10^9 bước) ⇒ cần phân tích thừa số nhanh (Pollard rho) rồi sinh ước.

````python
def construct_rectangle(area):
    """Biến thể 1 (LC 492)."""
    for w in range(math.isqrt(area), 0, -1):
        if area % w == 0:
            return [area // w, w]

print(construct_rectangle(4), construct_rectangle(37), construct_rectangle(122122))
````

````python
# Bài 4
def closest_joint_count_sol(num):
    loops = 0
    for i in range(math.isqrt(num + 2), 0, -1):
        loops += 1
        if (num + 1) % i == 0:
            return [(num + 1) // i, i], loops
        if (num + 2) % i == 0:
            return [(num + 2) // i, i], loops

pair, loops = closest_joint_count_sol(999)
print("Bài 4:", pair, "sau", loops, "vòng")
assert pair == [40, 25] and loops == 31 - 25 + 1
for num in range(1, 500):
    p, l = closest_joint_count_sol(num)
    assert p[0] >= p[1] and sorted(p) == sorted(closest_joint(num)) and l == math.isqrt(num + 2) - p[1] + 1
print("Bài 4: đúng với num = 1..499")
````

````python
# Bài 5
def closest_in_range_sol(num, k):
    best = None                                   # (hiệu, tích, cặp)
    for x in range(num + 1, num + k + 1):
        a, b = closest_single_sol(x)
        cand = (b - a, x, [a, b])
        if best is None or cand[:2] < best[:2]:
            best = cand
    return best[2]

def closest_in_range_brute(num, k):
    best = None
    for x in range(num + 1, num + k + 1):
        for a in range(1, x + 1):
            if x % a == 0:
                b = x // a
                cand = (abs(a - b), x, sorted([a, b]))
                if best is None or cand[:2] < best[:2]:
                    best = cand
    return best[2]

print("Bài 5:", closest_in_range_sol(8, 2), closest_in_range_sol(10, 5), closest_in_range_sol(1, 1))
rng = random.Random(5)
for _ in range(300):
    num, k = rng.randint(1, 300), rng.randint(1, 6)
    assert closest_in_range_sol(num, k) == closest_in_range_brute(num, k), (num, k)
assert sorted(closest_in_range_sol(123, 2)) == [5, 25]
print("Bài 5: khớp brute force trên 300 bộ ngẫu nhiên")
````
