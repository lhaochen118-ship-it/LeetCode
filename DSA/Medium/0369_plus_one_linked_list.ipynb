## 1. Mục tiêu học

Sau notebook này bạn sẽ:

1. Hiểu đề: một số nguyên **không âm** được lưu thành **danh sách liên kết đơn** (mỗi nút giữ **một chữ số**, chữ số **cao nhất ở đầu** danh sách); ta phải **cộng thêm 1** và trả về danh sách mới biểu diễn kết quả.
2. Nắm các công cụ: lớp **`ListNode`** (`val`, `next`), cách **duyệt** danh sách bằng `while cur:`, **nút giả (`dummy`)** đứng trước `head`, phép **`divmod` / `// 10` / `% 10`** cho cộng có nhớ, **đảo danh sách**, và **đệ quy** trả về giá trị "nhớ".
3. Đi từ cách **đổi cả danh sách thành số nguyên rồi cộng 1** → **đưa các nút vào mảng rồi cộng từ phải sang trái** → **đảo danh sách – cộng – đảo lại** → **đệ quy** → **một lượt duyệt tìm chữ số khác 9 cuối cùng** (tối ưu nhất, `O(1)` bộ nhớ phụ).
4. Hiểu mấu chốt: cộng 1 chỉ ảnh hưởng tới **đuôi toàn chữ số 9** và **chữ số đứng ngay trước đuôi đó**; chữ số này tăng thêm 1, các chữ số 9 phía sau thành 0, nếu **tất cả là 9** thì cần **thêm một nút `1` mới ở đầu**.
5. Tránh các lỗi kinh điển: quên **nút mới ở đầu** khi toàn số 9 (`999 -> 1000`), **cộng 1 vào nút cuối mà quên nhớ** (`9 -> 10` trong một nút), quên **đặt các số 9 phía sau về 0**, và **trả về số nguyên** thay vì danh sách.

## 2. Đề bài & ví dụ

Cho `head` là nút đầu của một danh sách liên kết đơn biểu diễn một số nguyên **không âm**: mỗi nút chứa **một chữ số** (`0..9`) và **chữ số có nghĩa lớn nhất nằm ở đầu danh sách** (giống cách ta viết số trên giấy, từ trái sang phải).

Hãy **cộng thêm 1** vào số đó và trả về nút đầu của danh sách biểu diễn kết quả.

**Ví dụ 1**

- Input: `head = [1,2,3]`
- Output: `[1,2,4]`
- Giải thích: danh sách biểu diễn số `123`; `123 + 1 = 124`.

**Ví dụ 2**

- Input: `head = [0]`
- Output: `[1]`
- Giải thích: số `0`, cộng 1 được `1`.

**Ví dụ tự dựng (đối chiếu lại đề gốc nếu có Premium)**

- Input: `head = [9,9,9]` → Output: `[1,0,0,0]` (`999 + 1 = 1000`, danh sách dài thêm một nút).
- Input: `head = [1,9,9]` → Output: `[2,0,0]` (`199 + 1 = 200`).

**Ràng buộc (theo hiểu biết về đề)**

- Số nút trong danh sách nằm trong khoảng `[1, 100]`.
- `0 <= Node.val <= 9`.
- Số được biểu diễn **không có chữ số 0 thừa ở đầu** (trừ chính số `0`).

Vì trong notebook không có sẵn `ListNode`, ta tự định nghĩa nó cùng ba hàm tiện ích: `build` (từ list Python thành danh sách liên kết), `to_list` (ngược lại) và `show` (in dạng `1 -> 2 -> 3`).

````python
from typing import List, Optional
import random


class ListNode:
    def __init__(self, val=0, next=None):
        self.val = val
        self.next = next


def build(digits):
    """Dựng danh sách liên kết từ list Python, trả về nút đầu (None nếu list rỗng)."""
    head = None
    for d in reversed(digits):          # dựng từ cuối lên để mỗi nút mới trỏ vào phần đã dựng
        head = ListNode(d, head)
    return head


def to_list(head, limit=10_000):
    """Đọc danh sách liên kết thành list Python (có giới hạn để không bị lặp vô hạn nếu có chu trình)."""
    out = []
    while head is not None:
        out.append(head.val)
        head = head.next
        assert len(out) <= limit, "danh sách quá dài hoặc có chu trình"
    return out


def show(head):
    vals = to_list(head)
    return " -> ".join(map(str, vals)) if vals else "(rỗng)"


examples = [
    ([1, 2, 3], [1, 2, 4]),
    ([0], [1]),
    ([9, 9, 9], [1, 0, 0, 0]),
    ([1, 9, 9], [2, 0, 0]),
]
for digits, expected in examples:
    print(digits, "->", show(build(digits)), "| kỳ vọng", expected)
````

## 3. Kiến thức nền

### 3.1. `ListNode` và duyệt danh sách

Một danh sách liên kết đơn là chuỗi các **nút** (node). Mỗi nút có hai ngăn: **`val`** (giá trị, ở đây là một chữ số) và **`next`** (trỏ tới nút kế tiếp, hoặc `None` nếu là nút cuối). Ta chỉ giữ **`head`** (nút đầu) để đi hết danh sách: bắt đầu từ `head`, rồi liên tục `cur = cur.next` cho tới khi `cur` là `None`.

Khác với list Python, **không thể truy cập `lst[i]` trực tiếp**: muốn tới nút thứ `i` phải đi qua `i` bước. Đặc biệt, với danh sách **đơn** ta **không đi lùi được** — đây là khó khăn chính của bài: cộng có nhớ cần đi từ **phải sang trái** (từ chữ số cuối), nhưng danh sách chỉ đi được từ **trái sang phải**.

````python
head = build([1, 2, 3])
cur = head
step = 0
while cur is not None:                  # duyệt: dừng khi vượt qua nút cuối
    print(f"bước {step}: val={cur.val}, next={'None' if cur.next is None else cur.next.val}")
    cur = cur.next
    step += 1
print(show(head))
````

### 3.2. Cộng có nhớ: `total % 10` và `total // 10`

Khi cộng hai chữ số cộng với nhớ, ta được `total` từ `0` đến `19`. **Chữ số viết xuống** là `total % 10` (chia lấy dư), **số nhớ sang bên trái** là `total // 10` (chia lấy phần nguyên). `divmod(total, 10)` trả về cả hai cùng lúc.

Ở bài này ta chỉ cộng `1`: nếu chữ số hiện tại là `9` thì `9 + 1 = 10` → viết `0`, nhớ `1`; nếu là chữ số khác `9` thì **không còn nhớ** và dừng ngay.

````python
for digit in [3, 8, 9]:
    carry_in = 1
    total = digit + carry_in
    print(f"{digit} + {carry_in} = {total} -> viết {total % 10}, nhớ {total // 10}  | divmod = {divmod(total, 10)}")
````

### 3.3. Nút giả (`dummy`)

Khi kết quả có thể **dài hơn** đầu vào (`999 -> 1000`), ta cần chèn một nút mới **ở đầu**. Mẹo quen thuộc: tạo sẵn `dummy = ListNode(0, head)` đứng **trước** `head`. Nếu cuối cùng `dummy.val` trở thành `1` thì `dummy` chính là đầu danh sách mới; nếu vẫn `0` thì trả về `dummy.next` (tức `head`). Nhờ vậy ta không phải viết riêng trường hợp "thêm nút ở đầu".

````python
head = build([9, 9])
dummy = ListNode(0, head)
print("dummy ->", show(dummy))          # 0 -> 9 -> 9
dummy.val += 1
print("sau khi dummy.val += 1:", show(dummy))   # 1 -> 9 -> 9 (chưa đặt 9 về 0, chỉ minh họa)
print("nếu dummy.val == 0 thì trả dummy.next:", show(ListNode(0, head).next))
````

### 3.4. Đảo danh sách liên kết

Đảo danh sách (reverse) là kỹ thuật nền: duyệt từng nút, **đổi hướng mũi tên `next`** cho nó trỏ về nút trước. Cần ba con trỏ: `prev` (nút trước, ban đầu `None`), `cur` (nút đang xét) và `nxt` (cất `cur.next` **trước khi** đổi hướng, kẻo mất phần còn lại).

Sau khi đảo, nút cuối cũ (chữ số hàng đơn vị) thành đầu danh sách, nên ta có thể cộng có nhớ **từ trái sang phải** như bình thường.

````python
def reverse_list(head):
    prev, cur = None, head
    while cur is not None:
        nxt = cur.next          # 1) cất phần còn lại
        cur.next = prev         # 2) đổi hướng mũi tên
        prev = cur              # 3) tiến prev
        cur = nxt               # 4) tiến cur
    return prev                 # prev là đầu danh sách mới


h = build([1, 2, 3])
print(show(h), "->", show(reverse_list(h)))
````

### 3.5. Đệ quy trả về giá trị "nhớ"

Đệ quy cho ta "đi lùi miễn phí": hàm đi sâu tới nút cuối rồi **quay lui** từ nút cuối về đầu, đúng thứ tự ta cần cho phép cộng có nhớ. Ta cho hàm `dfs(node)` **trả về số nhớ** sau khi xử lý `node` và mọi nút phía sau. Với `node = None` (vượt qua nút cuối) ta trả về `1` — chính là số `1` cần cộng thêm. Mỗi nút nhận số nhớ từ nút sau, tính `total`, ghi `total % 10` và trả `total // 10` cho nút trước.

Ví dụ nhỏ trên list Python (chưa dùng danh sách liên kết) để thấy "đi sâu rồi quay lui":

````python
def demo(i, arr, depth=0):
    pad = "  " * depth
    if i == len(arr):
        print(f"{pad}hết mảng, trả nhớ 1")
        return 1
    carry = demo(i + 1, arr, depth + 1)
    total = arr[i] + carry
    arr[i] = total % 10
    print(f"{pad}arr[{i}]: {total - carry}+{carry}={total} -> ghi {total % 10}, trả nhớ {total // 10}")
    return total // 10


arr = [1, 9, 9]
print("nhớ cuối cùng:", demo(0, arr), "| arr =", arr)
````

### 3.6. Đổi qua lại giữa danh sách chữ số và số nguyên

Python có số nguyên **không giới hạn**, nên `int("123") + 1` luôn đúng dù số rất dài. `str(124)` cho chuỗi `"124"`, rồi `[int(ch) for ch in "124"]` cho các chữ số. Điều này cho ta **cách giải "gian lận" nhưng đúng** trong Python (cách 1). Lưu ý: Python 3.11+ giới hạn đổi chuỗi-số dài hơn **4300 chữ số** (`sys.set_int_max_str_digits`), nhưng đề chỉ có tối đa `100` chữ số nên không vướng.

````python
digits = [1, 2, 3]
number = int("".join(str(d) for d in digits))
print(number, number + 1, [int(ch) for ch in str(number + 1)])
print(int("999") + 1, [int(ch) for ch in str(int("999") + 1)])
````

## 4. Ý tưởng nền tảng / cách nghĩ trực giác (tính tay)

Hãy làm đúng như cộng tay trên giấy. Cộng 1 vào số `1 9 9 9`:

```
  1 9 9 9
+       1
---------
```

- Hàng đơn vị: `9 + 1 = 10` → viết `0`, nhớ `1`.
- Hàng chục: `9 + 1 = 10` → viết `0`, nhớ `1`.
- Hàng trăm: `9 + 1 = 10` → viết `0`, nhớ `1`.
- Hàng nghìn: `1 + 1 = 2` → viết `2`, **hết nhớ**, dừng.

Kết quả `2 0 0 0`. Quan sát quan trọng: **chuỗi các số 9 ở cuối biến thành 0**, **chữ số khác 9 đứng ngay trước chúng tăng thêm 1**, mọi chữ số còn lại **giữ nguyên**.

Nếu **toàn bộ là 9** (`9 9 9`), chuỗi nhớ chạy hết ra ngoài và ta cần **thêm một chữ số `1`** ở đầu: `1 0 0 0`.

| Số | Đuôi toàn 9 | Chữ số khác 9 đứng trước đuôi | Kết quả |
|---|---|---|---|
| `123` | không có (`3` khác 9) | `3` | `124` |
| `1999` | `999` | `1` | `2000` |
| `8 9` | `9` | `8` | `90` |
| `999` | `999` | không có | `1000` |

````python
# Tính tay bằng code: tìm vị trí chữ số khác 9 cuối cùng của list chữ số
for digits in ([1, 2, 3], [1, 9, 9, 9], [8, 9], [9, 9, 9], [0]):
    last = -1
    for i, d in enumerate(digits):
        if d != 9:
            last = i
    print(digits, "-> vị trí chữ số khác 9 cuối cùng:", last)
````

## 5. Cách 1 — Brute Force / baseline: đổi thành số nguyên, cộng 1, dựng lại danh sách

**Ý tưởng.** Duyệt danh sách, ghép các chữ số thành chuỗi, `int(...)`, cộng `1`, đổi lại chuỗi rồi dựng danh sách mới. Rất ngắn, bám sát đề nhưng **né** hoàn toàn bài toán danh sách liên kết: trong ngôn ngữ như C++/Java số `100` chữ số **không** vừa kiểu `int/long`, nên cách này không dùng được ở đó (còn bài kiểm tra "bản chất" thì chính là thao tác trên nút).

````python
def plus_one_int(head):
    chars = []
    cur = head
    while cur is not None:
        chars.append(str(cur.val))
        cur = cur.next
    number = int("".join(chars)) + 1                 # số nguyên lớn tùy ý của Python
    return build([int(ch) for ch in str(number)])    # dựng danh sách mới


for digits, expected in examples:
    print(digits, "->", to_list(plus_one_int(build(digits))), "(kỳ vọng", expected, ")")
````

**Độ phức tạp (Cách 1).** Gọi `n` là số nút. Duyệt `n` nút, ghép chuỗi `O(n)`, `int(...)`/`str(...)` `O(n)` đến `O(n²)` tùy cài đặt với số rất lớn (với `n <= 100` là tức thì), dựng danh sách `O(n)`. Tóm lại **Time `O(n)`** (coi chuyển đổi tuyến tính), **Space `O(n)`** (chuỗi + danh sách mới).

**Nút thắt:** ta bỏ qua cấu trúc nút, tốn `O(n)` bộ nhớ, và không dùng được khi số vượt giới hạn kiểu số nguyên. Ta muốn **sửa trực tiếp các nút** theo luật cộng có nhớ.

## 6. Ý tưởng thuật toán / bước đột phá

**Bước đột phá: nhớ chỉ chạy ngược qua đuôi toàn số 9.** Khi cộng `1`, số nhớ dừng ngay tại chữ số khác 9 đầu tiên (tính từ phải). Vì vậy kết quả luôn có dạng:

```
 [ phần đầu giữ nguyên ] [ x ] [ 9 9 ... 9 ]        trước
 [ phần đầu giữ nguyên ] [x+1] [ 0 0 ... 0 ]        sau
```

Ta chỉ cần **tìm nút `x` = nút khác 9 cuối cùng** (một lượt duyệt từ trái sang phải, cập nhật mỗi khi gặp nút khác 9), rồi **tăng `x` thêm 1** và **đặt mọi nút phía sau về 0**. Nếu **không có** nút nào khác 9, `x` "nằm" ở nút `dummy` đứng trước đầu danh sách, và `dummy.val` thành `1` — đó là nút mới ở đầu.

Điều này **không cần đi lùi**, **không cần đệ quy**, **không cần mảng phụ**: chỉ `O(n)` thời gian và `O(1)` bộ nhớ phụ. Các cách còn lại (mảng, đảo, đệ quy) là những cách khác để "đi lùi" trên danh sách đơn.

````python
# Một lượt duyệt: ghi nhớ nút khác 9 cuối cùng
head = build([1, 9, 9])
dummy = ListNode(0, head)
not_nine = dummy                      # mặc định: nút giả (dùng khi toàn số 9)
cur = head
while cur is not None:
    if cur.val != 9:
        not_nine = cur
    print(f"duyệt nút {cur.val}: nút khác 9 cuối cùng hiện là {not_nine.val}")
    cur = cur.next
print("kết luận: tăng nút có val =", not_nine.val)
````

## 7. Các cách tiếp theo

Ngoài Cách 1 (đổi sang số nguyên), ta có thêm **4 cách khác nhau về ý tưởng**: Cách 2 (đưa các nút vào mảng), Cách 3 (đảo danh sách), Cách 4 (đệ quy), Cách 5 (nút giả + tìm chữ số khác 9 cuối cùng).

### Cách 2 — Đưa các nút vào mảng rồi cộng từ phải sang trái

**Giải thích từng đoạn.**

1. Duyệt một lượt, `nodes.append(cur)`: mảng `nodes` giữ **tham chiếu tới các nút** (không phải giá trị), nên sửa `nodes[i].val` là sửa luôn danh sách.
2. Mảng cho phép **đi lùi** bằng chỉ số `i = len(nodes) - 1` giảm dần.
3. Cộng có nhớ với `carry = 1`: `total = nodes[i].val + carry`, ghi `total % 10`, `carry = total // 10`. Dừng khi `carry == 0` (sớm) hoặc hết nút.
4. Nếu hết nút mà `carry` vẫn `1` ⇒ tất cả là 9 ⇒ trả về `ListNode(1, head)`.

````python
def plus_one_array(head):
    nodes = []
    cur = head
    while cur is not None:
        nodes.append(cur)                  # lưu NÚT, không lưu giá trị
        cur = cur.next
    carry = 1
    i = len(nodes) - 1
    while i >= 0 and carry:
        total = nodes[i].val + carry
        nodes[i].val = total % 10
        carry = total // 10
        i -= 1
    if carry:                              # nhớ chạy hết ra ngoài: 999 -> 1000
        return ListNode(1, head)
    return head


for digits, expected in examples:
    print(digits, "->", to_list(plus_one_array(build(digits))), "(kỳ vọng", expected, ")")
````

````python
# Code nộp LeetCode cho Cách 2 (đổi tên lớp thành Solution khi nộp)
class SolutionArray:
    def plusOne(self, head: Optional[ListNode]) -> Optional[ListNode]:
        nodes = []
        cur = head
        while cur:
            nodes.append(cur)
            cur = cur.next
        carry = 1
        i = len(nodes) - 1
        while i >= 0 and carry:
            total = nodes[i].val + carry
            nodes[i].val = total % 10
            carry = total // 10
            i -= 1
        return ListNode(1, head) if carry else head


print(to_list(SolutionArray().plusOne(build([1, 9, 9]))))        # [2, 0, 0]
````

**Độ phức tạp Cách 2:** **Time `O(n)`**, **Space `O(n)`** (mảng `n` tham chiếu).

### Cách 3 — Đảo danh sách, cộng từ đầu, rồi đảo lại

**Giải thích từng đoạn.**

1. `head = reverse_list(head)`: chữ số hàng đơn vị lên đầu.
2. Đi từ đầu (mới) với `carry = 1`: `total = cur.val + carry`, ghi `% 10`, `carry = // 10`, tiến `cur`. Nhớ lại `last` là nút cuối đã xử lý.
3. Dừng khi hết nhớ hoặc hết nút. Nếu **còn nhớ** thì `last.next = ListNode(1)`: thêm chữ số cao nhất ở **cuối** danh sách đã đảo (vì ngược lại sẽ là đầu).
4. Đảo lại để được thứ tự ban đầu.

Ưu điểm: **không tốn mảng**, chỉ `O(1)` bộ nhớ phụ. Nhược điểm: đảo **hai lần**, và **thay đổi tạm** danh sách (phải nhớ đảo lại).

````python
def plus_one_reverse(head):
    head = reverse_list(head)              # 1) đảo
    cur, last, carry = head, None, 1
    while cur is not None and carry:       # 2) cộng từ hàng đơn vị
        total = cur.val + carry
        cur.val = total % 10
        carry = total // 10
        last = cur
        cur = cur.next
    if carry:                              # 3) còn nhớ: thêm chữ số cao nhất (ở cuối danh sách đảo)
        last.next = ListNode(carry)
    return reverse_list(head)              # 4) đảo lại


for digits, expected in examples:
    print(digits, "->", to_list(plus_one_reverse(build(digits))), "(kỳ vọng", expected, ")")
````

````python
# Code nộp LeetCode cho Cách 3
class SolutionReverse:
    def plusOne(self, head: Optional[ListNode]) -> Optional[ListNode]:
        def reverse(node):
            prev = None
            while node:
                nxt = node.next
                node.next = prev
                prev = node
                node = nxt
            return prev

        head = reverse(head)
        cur, last, carry = head, None, 1
        while cur and carry:
            total = cur.val + carry
            cur.val = total % 10
            carry = total // 10
            last = cur
            cur = cur.next
        if carry:
            last.next = ListNode(carry)
        return reverse(head)


print(to_list(SolutionReverse().plusOne(build([9, 9, 9]))))      # [1, 0, 0, 0]
````

**Độ phức tạp Cách 3:** **Time `O(n)`** (ba lượt tuyến tính), **Space `O(1)`** phụ.

### Cách 4 — Đệ quy: đi sâu tới cuối, quay lui mang theo số nhớ

**Giải thích từng đoạn.**

1. `dfs(node)` trả về **số nhớ** sau khi xử lý `node` và phần phía sau.
2. **Điều kiện cơ sở:** `node is None` ⇒ trả về `1` (đây chính là số `1` ta cộng thêm).
3. **Quay lui:** `carry = dfs(node.next)`; `total = node.val + carry`; `node.val = total % 10`; trả `total // 10`.
4. Sau khi gọi `dfs(head)`: nếu còn nhớ `1` ⇒ trả `ListNode(1, head)`, ngược lại trả `head`.

Độ sâu đệ quy bằng số nút (`<= 100`), an toàn với giới hạn đệ quy mặc định của Python (`1000`). Nếu `n` lớn hơn nhiều, nên dùng cách lặp.

````python
def plus_one_recursive(head):
    def dfs(node):
        if node is None:
            return 1                       # số 1 cần cộng thêm, "đặt" ở sau nút cuối
        carry = dfs(node.next)             # xử lý phần sau trước
        total = node.val + carry
        node.val = total % 10
        return total // 10                 # số nhớ trả cho nút đứng trước

    carry = dfs(head)
    return ListNode(1, head) if carry else head


for digits, expected in examples:
    print(digits, "->", to_list(plus_one_recursive(build(digits))), "(kỳ vọng", expected, ")")
````

````python
# Code nộp LeetCode cho Cách 4
class SolutionRecursive:
    def plusOne(self, head: Optional[ListNode]) -> Optional[ListNode]:
        def dfs(node):
            if not node:
                return 1
            carry = dfs(node.next)
            total = node.val + carry
            node.val = total % 10
            return total // 10

        return ListNode(1, head) if dfs(head) else head


print(to_list(SolutionRecursive().plusOne(build([8, 9]))))       # [9, 0]
````

**Độ phức tạp Cách 4:** **Time `O(n)`**, **Space `O(n)`** (ngăn xếp đệ quy sâu `n`).

### Cách 5 — Nút giả + nút khác 9 cuối cùng (tối ưu)

**Giải thích từng đoạn.**

1. `dummy = ListNode(0, head)`; `not_nine = dummy`.
2. Một lượt duyệt: mỗi khi `cur.val != 9` thì `not_nine = cur`. Kết thúc, `not_nine` là nút khác 9 **cuối cùng** (hoặc `dummy` nếu toàn 9).
3. `not_nine.val += 1`.
4. Mọi nút sau `not_nine` (toàn số 9) đặt về `0`.
5. `dummy.val` bằng `1` ⇒ `dummy` là đầu mới; bằng `0` ⇒ trả `dummy.next`.

Đây là cách **gọn nhất**: một lượt duyệt tìm nút, một lượt đặt về 0 (tổng cộng không quá `2n` bước), **không đảo, không đệ quy, không mảng**.

````python
def plus_one_dummy(head):
    dummy = ListNode(0, head)
    not_nine = dummy
    cur = head
    while cur is not None:                 # lượt 1: tìm nút khác 9 cuối cùng
        if cur.val != 9:
            not_nine = cur
        cur = cur.next
    not_nine.val += 1                      # tăng nó thêm 1
    cur = not_nine.next
    while cur is not None:                 # lượt 2: các số 9 phía sau thành 0
        cur.val = 0
        cur = cur.next
    return dummy if dummy.val == 1 else dummy.next


for digits, expected in examples:
    print(digits, "->", to_list(plus_one_dummy(build(digits))), "(kỳ vọng", expected, ")")
````

````python
# Code nộp LeetCode cho Cách 5
class SolutionDummy:
    def plusOne(self, head: Optional[ListNode]) -> Optional[ListNode]:
        dummy = ListNode(0, head)
        not_nine = dummy
        cur = head
        while cur:
            if cur.val != 9:
                not_nine = cur
            cur = cur.next
        not_nine.val += 1
        cur = not_nine.next
        while cur:
            cur.val = 0
            cur = cur.next
        return dummy if dummy.val else head


print(to_list(SolutionDummy().plusOne(build([1, 2, 9, 9]))))     # [1, 3, 0, 0]
````

**Độ phức tạp Cách 5:** **Time `O(n)`**, **Space `O(1)`** phụ (chỉ vài con trỏ và một nút giả).

## 8. TRACE từng bước

Ta trace **Cách 5** trên `[1, 9, 9]` (có đuôi 9) và `[9, 9]` (toàn 9), rồi **Cách 3** (đảo) trên `[8, 9, 9]` để thấy cách số nhớ chạy.

Bảng trace Cách 5 với `[1, 9, 9]` (`dummy = 0`):

| Bước | `cur.val` | Khác 9? | `not_nine.val` sau bước |
|---|---|---|---|
| 1 | 1 | có | 1 |
| 2 | 9 | không | 1 |
| 3 | 9 | không | 1 |

Sau duyệt: `not_nine` là nút `1` → thành `2`; hai nút sau thành `0` ⇒ `[2, 0, 0]`. Dummy vẫn `0` nên trả `dummy.next`.

````python
def trace_dummy(digits):
    head = build(digits)
    dummy = ListNode(0, head)
    not_nine = dummy
    print("bắt đầu:", show(dummy), "(dummy ở đầu)")
    cur, step = head, 1
    while cur is not None:
        if cur.val != 9:
            not_nine = cur
        print(f"  bước {step}: cur.val={cur.val}, khác 9? {cur.val != 9}, not_nine.val={not_nine.val}")
        cur = cur.next
        step += 1
    not_nine.val += 1
    print("  tăng not_nine ->", show(dummy))
    cur = not_nine.next
    while cur is not None:
        print(f"  đặt nút 9 phía sau về 0 (hiện val={cur.val})")
        cur.val = 0
        cur = cur.next
    result = dummy if dummy.val == 1 else dummy.next
    print("  kết quả:", show(result), "| dummy.val =", dummy.val)
    return to_list(result)


assert trace_dummy([1, 9, 9]) == [2, 0, 0]
print()
assert trace_dummy([9, 9]) == [1, 0, 0]
````

````python
def trace_reverse(digits):
    head = reverse_list(build(digits))
    print("sau khi đảo:", show(head))
    cur, last, carry, step = head, None, 1, 1
    while cur is not None and carry:
        total = cur.val + carry
        print(f"  bước {step}: val={cur.val} + nhớ {carry} = {total} -> ghi {total % 10}, nhớ mới {total // 10}")
        cur.val, carry = total % 10, total // 10
        last, cur = cur, cur.next
        step += 1
    if carry:
        print("  còn nhớ sau nút cuối: thêm nút", carry)
        last.next = ListNode(carry)
    result = reverse_list(head)
    print("đảo lại:", show(result))
    return to_list(result)


assert trace_reverse([8, 9, 9]) == [9, 0, 0]
print()
assert trace_reverse([9, 9]) == [1, 0, 0]
````

## 9. Code nộp LeetCode

Bản nộp dùng **Cách 5** (nút giả + nút khác 9 cuối cùng): `O(n)` thời gian, `O(1)` bộ nhớ phụ, không đệ quy. Trên LeetCode `ListNode` đã được định nghĩa sẵn; ở đây ta dùng `ListNode` tự dựng ở mục 2.

````python
class Solution:
    def plusOne(self, head: Optional[ListNode]) -> Optional[ListNode]:
        dummy = ListNode(0, head)               # nút giả: phòng trường hợp toàn số 9
        not_nine = dummy
        cur = head
        while cur:                              # tìm nút khác 9 cuối cùng
            if cur.val != 9:
                not_nine = cur
            cur = cur.next
        not_nine.val += 1                       # tăng nó 1
        cur = not_nine.next
        while cur:                              # các số 9 phía sau thành 0
            cur.val = 0
            cur = cur.next
        return dummy if dummy.val else head


sol = Solution()
for digits, expected in examples:
    print(digits, "->", show(sol.plusOne(build(digits))))
````

## 10. Vì sao thuật toán đúng

- **Cộng 1 chỉ làm nhớ lan qua các số 9.** Tại chữ số hàng đơn vị ta cộng `1`. Nếu chữ số đó khác 9 thì kết quả là chữ số + 1 và **không có nhớ** nên mọi chữ số khác giữ nguyên. Nếu là 9 thì viết `0`, nhớ `1`, và bài toán lặp lại ở chữ số bên trái. Do đó nhớ dừng ở chữ số **khác 9 đầu tiên tính từ phải**, tức **nút khác 9 cuối cùng** trong danh sách.
- **Các nút sau `not_nine` đều là 9.** Nút `not_nine` là nút khác 9 cuối cùng, nên mọi nút phía sau nó bắt buộc bằng 9 và đều biến thành `0`.
- **`not_nine + 1` không tràn.** Vì `not_nine.val <= 8`, tăng 1 được tối đa `9`, không sinh nhớ mới.
- **Toàn số 9.** Không có nút khác 9, `not_nine` giữ giá trị `dummy`; `dummy.val` từ `0` thành `1`, và các nút sau (toàn bộ danh sách) về `0`: `1 0 0 ... 0`, đúng với `99..9 + 1 = 10..0`.
- **Không có số 0 thừa ở đầu.** Đầu vào không có 0 thừa; kết quả chỉ có thể dài hơn một nút bằng đúng chữ số `1`, nên cũng không có 0 thừa. Riêng `[0]` được tăng thành `[1]`.

### 12.2. Phản ví dụ cho các cách làm sai

Hai lỗi hay gặp: (a) **chỉ tăng nút cuối** mà không xử lý nhớ (`9 + 1 = 10` nằm trong một nút); (b) **tăng nút khác 9 cuối cùng nhưng quên đặt các số 9 phía sau về 0**. Ta chứng minh hai hàm sai này **khác** oracle ở một số input.

````python
def wrong_last_only(head):
    cur = head
    while cur.next is not None:
        cur = cur.next
    cur.val += 1                    # SAI: nếu cur.val == 9 sẽ thành 10 (không phải chữ số)
    return head


def wrong_no_zeroing(head):
    dummy = ListNode(0, head)
    not_nine, cur = dummy, head
    while cur:
        if cur.val != 9:
            not_nine = cur
        cur = cur.next
    not_nine.val += 1               # SAI: quên đặt các số 9 phía sau về 0
    return dummy if dummy.val else head


print("wrong_last_only [1,9]:", run(wrong_last_only, [1, 9]), "!=", oracle([1, 9]))
print("wrong_no_zeroing [1,9,9]:", run(wrong_no_zeroing, [1, 9, 9]), "!=", oracle([1, 9, 9]))
assert run(wrong_last_only, [1, 9]) != oracle([1, 9])
assert run(wrong_no_zeroing, [1, 9, 9]) != oracle([1, 9, 9])
````

### 12.3. Đối chiếu ngẫu nhiên (có seed) và vét cạn

Đầu vào hợp lệ theo đề: độ dài `1..100`, chữ số `0..9`, **không có 0 thừa ở đầu** (trừ số `0`). Ta kiểm tra: (1) **ngẫu nhiên**, thiên về chữ số 9 để thử nhiều đuôi 9; (2) **vét cạn mọi số từ `0` đến `99999`**.

````python
rng = random.Random(369)


def random_digits():
    n = rng.randint(1, 100)
    nine_bias = rng.choice([0.0, 0.3, 0.8, 0.95])         # xác suất chữ số là 9
    digits = [9 if rng.random() < nine_bias else rng.randint(0, 9) for _ in range(n)]
    if n > 1 and digits[0] == 0:
        digits[0] = rng.randint(1, 9)                      # không có 0 thừa ở đầu
    if n == 1 and rng.random() < 0.2:
        digits = [0]
    return digits


for trial in range(3000):
    digits = random_digits()
    assert 1 <= len(digits) <= 100 and all(0 <= d <= 9 for d in digits)
    assert len(digits) == 1 or digits[0] != 0
    ref = oracle(digits)
    for key, fn in impls.items():
        assert run(fn, list(digits)) == ref, (key, digits)
print("PASS 3000 danh sách ngẫu nhiên đúng ràng buộc")

for number in range(100_000):
    digits = [int(ch) for ch in str(number)]
    ref = [int(ch) for ch in str(number + 1)]
    for key, fn in impls.items():
        assert run(fn, list(digits)) == ref, (key, number)
print("PASS vét cạn 100000 số từ 0 đến 99999 với 6 cách")
````

## 13. Các lỗi người mới hay gặp

### Lỗi 1: Quên thêm nút mới ở đầu khi toàn số 9

````python
def loi1_quen_nut_dau(head):
    dummy = ListNode(0, head)
    not_nine, cur = dummy, head
    while cur:
        if cur.val != 9:
            not_nine = cur
        cur = cur.next
    not_nine.val += 1
    cur = not_nine.next
    while cur:
        cur.val = 0
        cur = cur.next
    return head                    # SAI: luôn trả head, bỏ qua dummy


print("sai:", run(loi1_quen_nut_dau, [9, 9, 9]), "| đúng:", run(Solution().plusOne, [9, 9, 9]))
````

**Vì sao sai:** với `999`, nút `1` mới nằm ở `dummy`, nhưng hàm vẫn trả về `head` nên **mất chữ số `1` ở đầu** (kết quả `000`). **Cách sửa:** trả về `dummy` nếu `dummy.val == 1`, ngược lại `head`.

### Lỗi 2: Chỉ cộng 1 vào nút cuối (không xử lý nhớ)

````python
def loi2_chi_nut_cuoi(head):
    cur = head
    while cur.next:
        cur = cur.next
    cur.val += 1                   # SAI: 9 + 1 = 10 không phải một chữ số
    return head


print("sai:", run(loi2_chi_nut_cuoi, [1, 2, 9]), "| đúng:", run(Solution().plusOne, [1, 2, 9]))
````

**Vì sao sai:** `9 + 1 = 10` không nằm vừa một nút (`val` chỉ là **một chữ số**). Phải **viết `0` và nhớ `1` sang nút bên trái**. **Cách sửa:** dùng `% 10` / `// 10` với biến `carry`, hoặc tìm nút khác 9 cuối cùng như Cách 5.

### Lỗi 3: Quên đặt các số 9 phía sau về 0

````python
def loi3_quen_dat_0(head):
    dummy = ListNode(0, head)
    not_nine, cur = dummy, head
    while cur:
        if cur.val != 9:
            not_nine = cur
        cur = cur.next
    not_nine.val += 1              # SAI: các nút 9 phía sau vẫn là 9
    return dummy if dummy.val else head


print("sai:", run(loi3_quen_dat_0, [1, 9, 9]), "| đúng:", run(Solution().plusOne, [1, 9, 9]))
````

**Vì sao sai:** `199 + 1` phải là `200`, nhưng hàm cho `299`. Mỗi chữ số 9 phía sau sinh `0` và nhớ, nên **phải đặt chúng về 0**. **Cách sửa:** thêm vòng `while cur: cur.val = 0` từ `not_nine.next`.

### Lỗi 4: Trả về số nguyên (hoặc list) thay vì nút đầu của danh sách

````python
def loi4_tra_so(head):
    number = 0
    while head:
        number = number * 10 + head.val
        head = head.next
    return number + 1              # SAI: đề yêu cầu trả về danh sách liên kết (nút đầu)


result = loi4_tra_so(build([1, 2, 3]))
print("sai:", result, type(result).__name__, "| đúng:", show(Solution().plusOne(build([1, 2, 3]))))
````

**Vì sao sai:** kiểu trả về là `Optional[ListNode]`, không phải `int`. Trình chấm sẽ cố duyệt `.next` trên số nguyên và báo lỗi. **Cách sửa:** dựng lại danh sách (Cách 1) hoặc sửa trực tiếp các nút rồi trả về nút đầu.

### Lỗi 5: Sửa danh sách mà đánh mất `head`

````python
def loi5_mat_head(head):
    carry = 1
    while head and carry:          # SAI: dùng chính head để duyệt, rồi trả head đã bị đẩy tới cuối
        total = head.val + carry
        head.val, carry = total % 10, total // 10
        head = head.next
    return head


print("sai:", show(loi5_mat_head(build([1, 2, 3]))), "| đúng:", show(Solution().plusOne(build([1, 2, 3]))))
````

**Vì sao sai:** biến `head` bị đẩy `head = head.next` nên khi return ta đã **mất đầu danh sách** (ở đây chỉ còn phần đuôi `2 -> 3`). **Cách sửa:** dùng biến phụ `cur = head` để duyệt, giữ nguyên `head` (hoặc `dummy`) để trả về. Ngoài ra cộng từ **trái** sang như thế cũng sai hướng: nhớ phải đi từ **phải** sang trái.

## 14. Mẹo làm bài nhanh & nhận dạng pattern

- "**Cộng/trừ số lớn lưu trong danh sách liên kết**" ⇒ nghĩ ngay tới **cộng có nhớ**, và câu hỏi "làm sao đi từ phải sang trái": **đảo danh sách**, **ngăn xếp / mảng**, **đệ quy**, hoặc **tìm nút khác 9 cuối cùng** (chỉ khi cộng 1).
- Mỗi khi kết quả có thể **dài thêm một nút ở đầu** hoặc **xóa nút đầu**, hãy dùng **nút giả `dummy`** để không phải viết nhánh riêng.
- Khi duyệt, luôn dùng biến `cur` thay vì sửa `head`; giữ `head`/`dummy` nguyên để trả về.
- `carry` + `total % 10` + `total // 10` là bộ ba chuẩn của mọi bài cộng có nhớ (cả **Add Two Numbers**, **Add Binary**, **Plus One** trên mảng).
- Số nhớ khi **cộng 1** chỉ chạy qua đuôi toàn số 9: kỹ thuật "nút khác 9 cuối cùng" chỉ áp dụng cho **cộng 1** (cộng số tổng quát cần carry đầy đủ).
- Test luôn phải có: `[0]`, `[9]`, `[9,9,9]` và một đuôi 9 sau chữ số khác (`[1,9,9]`).

## 15. Biến thể của bài

1. **Plus One (66)** — cùng bài nhưng đầu vào là **mảng** chữ số.
2. **Add Two Numbers (2)** — hai danh sách, chữ số **thấp** ở đầu (đảo ngược), cộng có nhớ.
3. **Add Two Numbers II (445)** — hai danh sách, chữ số **cao** ở đầu (như bài này): dùng hai ngăn xếp hoặc đảo.
4. **Cộng `k` bất kỳ** (`0 <= k <= 9`, hoặc `k` nhiều chữ số) vào danh sách.
5. **Trừ 1** khỏi số lưu trong danh sách (mượn thay vì nhớ; chú ý số 0 thừa ở đầu).
6. **Nhân đôi số** trong danh sách (Double a Number Represented as a Linked List, 2816).

````python
def add_small(head, k):
    """Cộng k (0 <= k <= 9) vào số trong danh sách: đảo, cộng nhớ, đảo lại."""
    head = reverse_list(head)
    cur, last, carry = head, None, k
    while cur is not None and carry:
        total = cur.val + carry
        cur.val, carry = total % 10, total // 10
        last, cur = cur, cur.next
    if carry:
        last.next = ListNode(carry)
    return reverse_list(head)


def minus_one_list(digits):
    """Biến thể trừ 1 (đầu vào là số >= 1) bằng số nguyên, để so sánh."""
    return [int(ch) for ch in str(int("".join(map(str, digits))) - 1)]


print(to_list(add_small(build([9, 9]), 5)), int("99") + 5)   # [1, 0, 4] 104
print(minus_one_list([1, 0, 0]), minus_one_list([1]))
for number in range(0, 300):
    for k in range(10):
        digits = [int(c) for c in str(number)]
        assert run(lambda h: add_small(h, k), digits) == [int(c) for c in str(number + k)]
print("add_small đúng với mọi số 0..299 và k = 0..9")
````

| Cách | Ý tưởng | Time | Space | Khi nào dùng |
|---|---|---|---|---|
| 1. Đổi sang `int` | ghép chữ số → số → cộng 1 → dựng lại | `O(n)` | `O(n)` | Python, kiểm tra nhanh; làm **oracle** |

| 4. Đệ quy | quay lui mang theo số nhớ | `O(n)` | `O(n)` | code ngắn, `n` nhỏ |
| 5. Nút giả + nút khác 9 cuối | nhớ chỉ lan qua đuôi 9 | `O(n)` | `O(1)` | **nên nộp**: gọn, tối ưu, riêng cho cộng 1 |
