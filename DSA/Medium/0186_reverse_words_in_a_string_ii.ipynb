## 2.1 Hai con trỏ (two pointers)

### Ý tưởng

Dùng **hai chỉ số** di chuyển trên mảng/chuỗi theo quy tắc, để thay vòng lặp lồng `O(n²)` bằng một lượt `O(n)`. Hai kiểu chính:

**1. Hai đầu tiến vào nhau** (thường trên mảng đã sắp xếp):

```text
[1, 2, 4, 7, 11, 15], tìm hai số có tổng 15
 L                 R     1 + 15 = 16 > 15 → R lùi (cần nhỏ hơn)
 L             R         1 + 11 = 12 < 15 → L tiến (cần lớn hơn)
    L          R         2 + 11 = 13 < 15 → L tiến
       L       R         4 + 11 = 15 ✓
```

Vì sao được phép bỏ? Khi `a[L] + a[R] > target`, mọi cặp `(L', R)` với `L' > L` còn lớn hơn → `R` không thể là đáp án, bỏ `R`.

**2. Cùng chiều — con trỏ đọc và con trỏ ghi** (sửa mảng tại chỗ):

```text
xóa mọi số 0, giữ thứ tự: [0, 1, 0, 3, 12]
read đi qua từng phần tử; gặp số khác 0 thì chép vào vị trí write, write += 1
→ [1, 3, 12, ...]
```

````python
def tong_hai_so_da_sap_xep(a, target):
    L, R = 0, len(a) - 1
    while L < R:
        s = a[L] + a[R]
        print(f"   L={L} R={R}: {a[L]} + {a[R]} = {s}")
        if s == target:
            return L, R
        if s < target:
            L += 1
        else:
            R -= 1
    return None

print("→", tong_hai_so_da_sap_xep([1, 2, 4, 7, 11, 15], 15))

def bo_so_khong(a):
    write = 0
    for read in range(len(a)):
        if a[read] != 0:
            a[write] = a[read]
            write += 1
    return a[:write]

print("bỏ số 0:", bo_so_khong([0, 1, 0, 3, 12]))

def la_palindrome(s):
    i, j = 0, len(s) - 1
    while i < j:
        if s[i] != s[j]:
            return False
        i += 1; j -= 1
    return True
print("'racecar' palindrome?", la_palindrome("racecar"), "| 'abca'?", la_palindrome("abca"))
````

### Đảo ngược một đoạn tại chỗ

Hai con trỏ ở hai đầu, đổi chỗ rồi tiến vào trong — dùng rất nhiều (đảo mảng, xoay mảng, đảo từ).

````python
def dao_doan(a, i, j):
    while i < j:
        a[i], a[j] = a[j], a[i]
        i += 1; j -= 1

arr_tp = [1, 2, 3, 4, 5, 6]
dao_doan(arr_tp, 1, 4)
print("đảo đoạn [1..4]:", arr_tp)
````

### Lỗi hay gặp

- Quên **sắp xếp** trước khi dùng hai đầu cho bài tổng.
- Điều kiện `L <= R` thay vì `L < R` → dùng một phần tử hai lần.
- Không chứng minh được vì sao bỏ một đầu là an toàn → dễ bỏ mất đáp án.
- Con trỏ ghi vượt con trỏ đọc (không thể xảy ra nếu `write` chỉ tăng khi ghi).

### Gom nhóm ký tự liên tiếp

`itertools.groupby` gom các phần tử **liên tiếp bằng nhau** — tiện cho mã hóa độ dài loạt (run-length): `"aaabcc" → 3a 1b 2c`. Tự viết bằng hai con trỏ: `i` ở đầu nhóm, `j` chạy tới khi ký tự khác.

````python
from itertools import groupby
print([(k, len(list(g))) for k, g in groupby("aaabcc")])

def run_length(s):
    out, i = [], 0
    while i < len(s):
        j = i
        while j < len(s) and s[j] == s[i]:
            j += 1
        out.append(f"{j - i}{s[i]}")
        i = j
    return "".join(out)
print("run_length('aaabcc') =", run_length("aaabcc"))
````

Lỗi hay gặp: `split(" ")` sinh phần rỗng khi có nhiều dấu cách; so sánh chuỗi số `"10" < "9"` là **True** (so từ điển, không phải số); nối chuỗi trong vòng lặp lớn; quên chuỗi rỗng.

## 2.3 Áp dụng vào bài này

Đảo thứ tự các từ **tại chỗ** bằng hai lần đảo: đảo **toàn bộ** mảng ký tự (thứ tự từ đúng nhưng mỗi từ bị ngược), rồi đảo lại **từng từ**. Mỗi lần đảo một đoạn là kỹ thuật **hai con trỏ** đổi chỗ từ hai đầu vào giữa — `O(n)` thời gian, `O(1)` bộ nhớ thêm.

# 3. Nền tảng

## Đảo một đoạn tại chỗ

Hai con trỏ ở hai đầu đoạn, đổi chỗ rồi tiến vào trong. `O(độ dài đoạn)`, `O(1)` bộ nhớ.

````python
from typing import List
import random, time

def reverse_range(a: List[str], i: int, j: int) -> None:
    while i < j:
        a[i], a[j] = a[j], a[i]
        i += 1
        j -= 1

a = list("abcdef")
reverse_range(a, 1, 4)
print("".join(a))
````

Chuỗi Python bất biến, nên đề đưa vào `list` ký tự. `"".join(s)` để xem kết quả, `list("...")` để tạo đầu vào.

# 4. Baseline: tách từ, đảo, ghi lại (không tại chỗ)

````python
def reverse_words_copy(s: List[str]) -> None:
    words = "".join(s).split(" ")
    s[:] = list(" ".join(reversed(words)))     # gán lại nội dung, nhưng đã tạo chuỗi/list tạm O(n)

s1 = list("the sky is blue")
reverse_words_copy(s1)
print("".join(s1))
````

Đúng, nhưng dùng `O(n)` bộ nhớ phụ (chuỗi, list các từ). Đề muốn `O(1)`.

# 5. Ý tưởng chính: đảo cả mảng, rồi đảo từng từ

```text
ban đầu:          the sky is blue
đảo cả mảng:      eulb si yks eht       ← thứ tự từ đã đúng, nhưng mỗi từ bị ngược
đảo từng từ:      blue is sky the       ✓
```

Vì sao? Viết chuỗi là `w1 _ w2 _ … _ wk`. Đảo cả chuỗi cho `rev(wk) _ … _ rev(w2) _ rev(w1)` (tính chất `rev(XY) = rev(Y) rev(X)`). Đảo lại từng khối `rev(wi)` cho `wi`.

Tìm từng từ: duyệt `k` từ 0 tới `n`; khi gặp dấu cách (hoặc hết mảng) thì đoạn `[start, k − 1]` là một từ.

````python
def reverse_words_inplace(s: List[str]) -> None:
    n = len(s)
    reverse_range(s, 0, n - 1)
    start = 0
    for k in range(n + 1):
        if k == n or s[k] == " ":
            reverse_range(s, start, k - 1)
            start = k + 1

s2 = list("the sky is blue")
reverse_words_inplace(s2)
print("".join(s2))
````

# 6. Mô phỏng

````python
def trace_reverse_words(text):
    s = list(text)
    print(f"ban đầu     : {''.join(s)!r}")
    reverse_range(s, 0, len(s) - 1)
    print(f"đảo cả mảng : {''.join(s)!r}")
    start = 0
    for k in range(len(s) + 1):
        if k == len(s) or s[k] == " ":
            reverse_range(s, start, k - 1)
            print(f"đảo s[{start}:{k}]  : {''.join(s)!r}")
            start = k + 1
    return "".join(s)

_ = trace_reverse_words("the sky is blue")
````

# 7. Phiên bản đúng format LeetCode

````python
class Solution:
    def reverseWords(self, s: List[str]) -> None:
        def rev(i: int, j: int) -> None:
            while i < j:
                s[i], s[j] = s[j], s[i]
                i += 1
                j -= 1

        n = len(s)
        rev(0, n - 1)
        start = 0
        for k in range(n + 1):
            if k == n or s[k] == " ":
                rev(start, k - 1)
                start = k + 1
````

````python
def check(name, text, expected):
    s = list(text)
    ident = id(s)
    out = Solution().reverseWords(s)
    assert out is None and id(s) == ident, "phải sửa tại chỗ, không trả về"
    assert "".join(s) == expected, (name, "".join(s), expected)
    print(f"{name:<12} PASS  {''.join(s)!r}")


check("Ví dụ 1", "the sky is blue", "blue is sky the")
check("Ví dụ 2", "a", "a")
check("Hai từ", "hello world", "world hello")
check("Một từ dài", "abcdef", "abcdef")
check("Chữ số", "a1 B2 c3", "c3 B2 a1")

random.seed(186)
for _ in range(3000):
    words = ["".join(random.choice("abXY12") for _ in range(random.randint(1, 4))) for _ in range(random.randint(1, 6))]
    text = " ".join(words)
    s = list(text)
    Solution().reverseWords(s)
    assert "".join(s) == " ".join(reversed(words)), text
print("3000 chuỗi ngẫu nhiên: khớp ' '.join(reversed(words))")

big = list(" ".join("w" * random.randint(1, 8) for _ in range(20000)))[:10**5]
while big[-1] == " ":
    big.pop()
t0 = time.time(); Solution().reverseWords(big); t1 = time.time()
print(f"n = {len(big)}: {t1 - t0:.3f}s")
````

# 8. Vì sao đúng? Độ phức tạp?

- Đảo cả mảng: `rev(w1 _ w2 _ … _ wk) = rev(wk) _ … _ rev(w1)` (dấu cách là một ký tự, đảo vẫn là dấu cách và nằm đúng giữa các khối).
- Lượt thứ hai tìm đúng các khối giữa các dấu cách (đó là `rev(wk), …, rev(w1)`) và đảo mỗi khối → `wk _ … _ w1`.

Mỗi ký tự bị đổi chỗ tối đa hai lần.

```text
Time:  O(n)
Space: O(1) bộ nhớ phụ
```

## Lỗi 2: Quên đảo **từ cuối cùng** (vòng lặp chỉ tới `n − 1`)

````python
def wrong_last_word(s):
    n = len(s)
    reverse_range(s, 0, n - 1)
    start = 0
    for k in range(n):                             # SAI: không có lượt k == n cho từ cuối
        if s[k] == " ":
            reverse_range(s, start, k - 1)
            start = k + 1

s_w2 = list("the sky is blue")
wrong_last_word(s_w2)
print("Sai :", repr("".join(s_w2)), "| Đúng:", repr("blue is sky the"))
````

Từ cuối cùng không có dấu cách theo sau nên không bao giờ được đảo — thêm lượt `k == n` (hoặc đảo đoạn cuối sau vòng lặp).

## Lỗi 3: Gán `s = …` thay vì sửa nội dung

````python
def wrong_rebind(s):
    s = list(" ".join(reversed("".join(s).split(" "))))    # SAI: chỉ đổi biến cục bộ

s_w3 = list("the sky is blue")
wrong_rebind(s_w3)
print("Sai :", repr("".join(s_w3)), "← mảng gốc không đổi")
````

Gán lại tên `s` trong hàm **không** ảnh hưởng mảng bên ngoài. Muốn thay toàn bộ nội dung phải dùng `s[:] = …` (nhưng như vậy vẫn tốn bộ nhớ phụ) hoặc đổi chỗ tại chỗ.

## Lỗi 4: Đổi chỗ bằng `s[i] = s[j]; s[j] = s[i]`

Ghi đè `s[i]` trước khi lưu → cả hai thành `s[j]`. Python dùng gán bộ `s[i], s[j] = s[j], s[i]`.

**Bài 5**
- Mức 1: đếm số từ `m`, `k %= m`; nếu `k == 0` thì giữ nguyên.
- Mức 2: sau khi đảo cả mảng, `rev(B)` chiếm `len(B)` ký tự đầu — `len(B)` = vị trí của dấu cách thứ `k` (đếm từ đầu mảng đã đảo).

````python
def answer_3_reverse_each_word(s: List[str]) -> None:
    start = 0
    for k in range(len(s) + 1):
        if k == len(s) or s[k] == " ":
            reverse_range(s, start, k - 1)
            start = k + 1

e3 = list("Let's take LeetCode contest"); answer_3_reverse_each_word(e3); assert "".join(e3) == "s'teL ekat edoCteeL tsetnoc"
e3b = list("ab"); answer_3_reverse_each_word(e3b); assert "".join(e3b) == "ba"
print("Bài 3 PASS")
````

## Đáp án Bài 4

````python
def answer_4_rotate_left(s: List[str], k: int) -> None:
    n = len(s)
    k %= n
    reverse_range(s, 0, k - 1)
    reverse_range(s, k, n - 1)
    reverse_range(s, 0, n - 1)

e4 = list("abcdefg"); answer_4_rotate_left(e4, 2); assert "".join(e4) == "cdefgab"
e4b = list("abc"); answer_4_rotate_left(e4b, 4); assert "".join(e4b) == "bca"
for _ in range(2000):
    t = "".join(random.choice("abcd") for _ in range(random.randint(1, 8)))
    k = random.randint(0, 20)
    a = list(t); answer_4_rotate_left(a, k)
    assert "".join(a) == t[k % len(t):] + t[:k % len(t)]
print("Bài 4 PASS")
````

## Đáp án Bài 5

````python
def answer_5_rotate_words(s: List[str], k: int) -> None:
    n = len(s)
    m = s.count(" ") + 1                       # số từ
    k %= m
    if k == 0:
        return
    reverse_range(s, 0, n - 1)                 # rev(B) _ rev(A)
    spaces, p = 0, 0
    while True:                                # vị trí dấu cách thứ k trong mảng đã đảo
        if s[p] == " ":
            spaces += 1
            if spaces == k:
                break
        p += 1
    reverse_range(s, 0, p - 1)                 # rev(B) → B
    reverse_range(s, p + 1, n - 1)             # rev(A) → A

e5 = list("a bb ccc dddd"); answer_5_rotate_words(e5, 1); assert "".join(e5) == "dddd a bb ccc"
e5b = list("a bb ccc dddd"); answer_5_rotate_words(e5b, 2); assert "".join(e5b) == "ccc dddd a bb"
e5c = list("x y"); answer_5_rotate_words(e5c, 2); assert "".join(e5c) == "x y"
for _ in range(2000):
    words = ["".join(random.choice("ab") for _ in range(random.randint(1, 3))) for _ in range(random.randint(1, 5))]
    k = random.randint(0, 7)
    a = list(" ".join(words)); answer_5_rotate_words(a, k)
    kk = k % len(words)
    assert "".join(a) == " ".join(words[len(words) - kk:] + words[:len(words) - kk])
print("Bài 5 PASS")
````
