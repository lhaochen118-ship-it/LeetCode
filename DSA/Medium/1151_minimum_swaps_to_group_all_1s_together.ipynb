## 4. Chuyển bài toán swap thành bài toán chọn đoạn

Giả sử mảng có `k` số `1`. Khi đã gom xong, **khối số `1` phải dài đúng `k`**. Vậy ta có thể chọn một đoạn liên tiếp dài `k` làm nơi nhận số `1`.

Nếu đoạn này đang có `ones` số `1`, thì nó có `k - ones` số `0`.

Vì toàn bộ mảng cũng chỉ có `k` số `1`, **bên ngoài** đoạn sẽ còn đúng `k - ones` số `1`. Ta có thể đổi từng số `0` trong đoạn với một số `1` bên ngoài, mỗi cặp mất một swap. Không thể ít hơn, bởi mỗi `0` nằm trong đoạn đích phải bị thay đi.

Vì thế:

\[\boxed{\text{swap tối thiểu} = k - \max_{\text{cửa sổ độ dài }k}(\text{số lượng 1 trong cửa sổ})}\]

**Ví dụ:** `[1,0,1,0,1]` có `k=3`; các đoạn `[1,0,1]`, `[0,1,0]`, `[1,0,1]` chứa lần lượt `2,1,2` số `1` → cần `1,2,1` swap → đáp án `1`.

```python
example = [1, 0, 1, 0, 1]
k = example.count(1)
for left in range(len(example) - k + 1):
    segment = example[left:left + k]
    ones = sum(segment)
    print(f'left={left}, đoạn={segment}, ones={ones}, swap={k-ones}')
```

## 5. Baseline (cách dễ nghĩ): xét lại mọi đoạn

Ta liệt kê mọi vị trí bắt đầu `left`, tạo đoạn độ dài `k`, đếm số `1` và lấy số swap nhỏ nhất.

Nhược điểm: mỗi lần `sum(data[left:left+k])` quét tối đa `k` phần tử, trong khi hai đoạn kế tiếp gần như trùng nhau.

- **Thời gian:** `O(nk)` trong trường hợp xấu (ví dụ `k ≈ n/2`).
- **Bộ nhớ phụ:** `O(k)` vì phép cắt `data[left:left+k]` tạo một list mới. Nếu đếm bằng vòng lặp trực tiếp không cắt mảng thì dùng `O(1)` bộ nhớ.

```python
def min_swaps_bruteforce(data):
    k = data.count(1)
    if k <= 1:
        return 0
    answer = k
    for left in range(len(data) - k + 1):
        window = data[left:left + k]
        answer = min(answer, k - sum(window))
    return answer

for arr in [[1,0,1,0,1], [0,0,0,1,0], [1,1,1,1], [0,0,0,0]]:
    print(arr, '→', min_swaps_bruteforce(arr))
```

## 6. Mấu chốt Sliding Window: chỉ cập nhật phần thay đổi

Cửa sổ trước dài `k` là `data[left : left+k]`. Khi dịch sang phải một bước:

- Phần tử **ra khỏi** cửa sổ: `data[left]`.
- Phần tử **đi vào** cửa sổ: `data[left+k]`.
- `ones_moi = ones_cu - data[left] + data[left+k]`.

Như vậy thay vì tính lại tổng `k` số, mỗi lần dịch chỉ làm **hai phép cộng/trừ**.

Đây là bất biến (*invariant*): tại mỗi bước, `ones` luôn bằng tổng các phần tử của cửa sổ hiện hành.

```python
arr = [1, 0, 1, 0, 1]
k = 3
ones = sum(arr[:k])
print('Cửa sổ đầu:', arr[:k], '| ones =', ones)
old = arr[0]
new = arr[k]
ones = ones - old + new
print('Bỏ', old, ', nhận', new, '→ ones mới =', ones)
print('Đoạn mới:', arr[1:k+1])
```

## 7. Lời giải `O(n)` dạng hàm dễ đọc

Khởi tạo `k` là tổng số `1`, đếm số `1` ở cửa sổ đầu. Duyệt `right` từ `k` đến `n-1`: `right` là phần tử mới sắp đi vào. Vị trí cần loại đi là `right-k`.

Giữ `best` là số `1` lớn nhất trong **tất cả** cửa sổ đã nhìn thấy. Cuối cùng trả `k-best`.

Với `k=0` hoặc `k=1`, đáp án chắc chắn bằng 0.

```python
def min_swaps_sliding_window(data):
    k = data.count(1)
    if k <= 1:
        return 0
    ones = sum(data[:k])
    best = ones
    for right in range(k, len(data)):
        ones += data[right] - data[right - k]
        best = max(best, ones)
    return k - best

print('Ví dụ 1:', min_swaps_sliding_window([1, 0, 1, 0, 1]))
print('Ví dụ 3:', min_swaps_sliding_window([1,0,1,0,1,0,0,1,1,0,1]))
```

## 8. Trace bằng tay: cửa sổ đi qua những đoạn nào?

Với `data = [1,0,1,0,1]`, `k=3`:

| `left` | `right` (bao gồm) | Cửa sổ | `ones` | `best` | swap nếu chọn đoạn này |
|---:|---:|---|---:|---:|---:|
| 0 | 2 | `[1,0,1]` | 2 | 2 | 1 |
| 1 | 3 | `[0,1,0]` | 1 | 2 | 2 |
| 2 | 4 | `[1,0,1]` | 2 | 2 | 1 |

Ta không cần biết `1` nào ở ngoài sẽ swap với `0` nào ở trong để **đếm** số swap tối thiểu.

```python
def trace_min_swaps(data):
    k = data.count(1)
    if k <= 1:
        print('k <= 1: đã gom xong, trả 0')
        return 0
    ones = sum(data[:k]); best = ones
    print(f'left=0, window={data[:k]}, ones={ones}, best={best}, swaps={k-ones}')
    for right in range(k, len(data)):
        gone = data[right-k]
        incoming = data[right]
        ones += incoming - gone
        best = max(best, ones)
        left = right - k + 1
        print(f'left={left}, ra={gone}, vào={incoming}, window={data[left:right+1]}, ones={ones}, best={best}, swaps={k-ones}')
    return k - best

assert trace_min_swaps([1, 0, 1, 0, 1]) == 1
```

## 9. Code nộp LeetCode

LeetCode gọi phương thức `minSwaps` của lớp `Solution`; không cần `print`, không thay đổi mảng đầu vào. `List[int]` chỉ là chú thích kiểu; ta import `List` từ `typing`.

```python
from typing import List

class Solution:
    def minSwaps(self, data: List[int]) -> int:
        k = data.count(1)
        if k <= 1:
            return 0
        ones = sum(data[:k])
        max_ones = ones
        for right in range(k, len(data)):
            ones += data[right] - data[right - k]
            if ones > max_ones:
                max_ones = ones
        return k - max_ones

sol = Solution()
assert sol.minSwaps([1,0,1,0,1]) == 1
assert sol.minSwaps([0,0,0,1,0]) == 0
assert sol.minSwaps([1,0,1,0,1,0,0,1,1,0,1]) == 3
print('Đã đúng 3 ví dụ trong đề.')
```

## 10. Vì sao thuật toán đúng? — chứng minh ngắn

1. **Tất cả đáp án hợp lệ** đều chọn một đoạn dài `k`, vì có đúng `k` số `1`.
2. **Với đoạn cố định:** nếu chứa `z` số `0`, cần ít nhất `z` lần đổi, bởi mỗi số `0` phải bị thay; ngoài đoạn có đúng `z` số `1`, nên `z` lần đổi là đủ.
3. **Sliding Window không bỏ sót đoạn nào:** bắt đầu từ trái nhất, mỗi vòng dịch phải một vị trí và xét cả đoạn cuối cùng.
4. `ones` được cập nhật bằng `+ phần tử vào - phần tử ra`, nên luôn đếm đúng; `max_ones` ghi nhận đoạn tốt nhất.

Vậy `k-max_ones` chính là **số swap tối thiểu trên tất cả các lựa chọn**.

## 13. Kiểm thử độc lập: mô phỏng thật các phép swap bằng BFS

Để kiểm tra công thức không chỉ “tình cờ đúng”, ta thử **BFS trên các trạng thái mảng ngắn**:

- Mỗi đỉnh là một mảng nhị phân.
- Một cạnh là một lần swap hai vị trí bất kỳ có giá trị khác nhau.
- BFS (*Breadth-First Search*) duyệt theo số bước tăng dần nên lần đầu đạt khối `1` liên tiếp cho số swap ít nhất.

BFS khá tốn kém nên chỉ dùng làm **bộ kiểm thử độc lập**, không nộp LeetCode với `n` lớn.

```python
from collections import deque
from itertools import combinations

def min_swaps_by_bfs(data):
    start = tuple(data)
    k = sum(start)
    if k <= 1:
        return 0
    q = deque([(start, 0)])
    visited = {start}
    while q:
        state, steps = q.popleft()
        if any(all(x == 1 for x in state[i:i+k]) for i in range(len(state)-k+1)):
            return steps
        for i, j in combinations(range(len(state)), 2):
            if state[i] == state[j]:
                continue
            nxt = list(state)
            nxt[i], nxt[j] = nxt[j], nxt[i]
            nxt = tuple(nxt)
            if nxt not in visited:
                visited.add(nxt)
                q.append((nxt, steps+1))
    raise AssertionError('Không thể đạt đích')

assert min_swaps_by_bfs([1,0,1,0,1]) == 1
print('BFS kiểm chứng ví dụ cơ bản:', min_swaps_by_bfs([1,0,1,0,1]))
```

## 17. Mẹo nhận diện Sliding Window

Hãy tự hỏi:
1. Đề có nói **một đoạn liên tiếp** không?
2. Độ dài đoạn có được xác định trước không (`k`, tổng số `1`, số ngày liên tiếp...)?
3. Khi dịch đoạn đi một bước, có thể **bỏ một phần tử cũ và thêm một phần tử mới** thay vì tính lại không?

Nếu có, hãy thử Sliding Window cố định. Bài tương tự: *Maximum Average Subarray I*, *Grumpy Bookstore Owner* (một số kỹ thuật có thể thay đổi).

```python
def max_sum_window(arr, window_size):
    if not 1 <= window_size <= len(arr):
        raise ValueError('window_size không hợp lệ')
    current = sum(arr[:window_size]); best = current
    for right in range(window_size, len(arr)):
        current += arr[right] - arr[right - window_size]
        best = max(best, current)
    return best

assert max_sum_window([3,-1,4,0,5], 2) == 5
print('Tổng đoạn dài 2 lớn nhất:', max_sum_window([3,-1,4,0,5], 2))
```

- **Mức 2 — công cụ:** `range(len(data)-k+1)` liệt kê vị trí bắt đầu; `sum(data[left:left+k])` là số `1` trong đoạn; `data.count(1)` là `k`.
- **Mức 3 — gần code:** với cửa sổ kế tiếp, `ones += data[right]-data[right-k]`; lưu `best=max(best,ones)`; kết quả `k-best`.

Đáp án bài 1: `data` dài 7, có `k=4` số `1`, vì vậy có đúng `7-4+1=4` cửa sổ. Số `1` trong từng cửa sổ lần lượt là **`[2,2,2,3]`**; cửa sổ cuối chứa nhiều số `1` nhất, nên đáp án `4-3=1`.

Việc tự tính số cửa sổ `n-k+1` là một bài học quan trọng để tránh lỗi *off-by-one*.
