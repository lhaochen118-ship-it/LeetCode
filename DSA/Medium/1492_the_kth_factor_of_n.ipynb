## 1. Mục tiêu học

1. Hiểu khái niệm **ước số** và tính chất **ước đi theo cặp** `(i, n // i)` quanh `√n`.
2. Đi từ duyệt `1..n` (`O(n)`) lên duyệt tới `√n` (`O(√n)`) — trả lời câu follow-up "nhanh hơn `O(n)`".
3. Xử lý đúng trường hợp **số chính phương** (ước `√n` chỉ được đếm một lần).
4. Biết dùng `math.isqrt` thay vì `int(math.sqrt(...))` để tránh sai số dấu phẩy động.
5. Làm quen cách sinh mọi ước từ **phân tích thừa số nguyên tố**, và đếm/sắp xếp ước.

## 2. Đề bài & ví dụ

Cho hai số nguyên dương `n` và `k`. Một **ước** (factor) của `n` là số nguyên `i` sao cho `n % i == 0`. Xét danh sách mọi ước của `n` sắp **tăng dần**; trả về ước thứ `k` (đếm từ 1), hoặc `-1` nếu `n` có ít hơn `k` ước.

**Ví dụ 1:** `n = 12, k = 3` → `3` (các ước: `[1, 2, 3, 4, 6, 12]`).

**Ví dụ 2:** `n = 7, k = 2` → `7` (các ước: `[1, 7]`).

**Ví dụ 3:** `n = 4, k = 4` → `-1` (chỉ có 3 ước `[1, 2, 4]`).

**Ràng buộc:** `1 <= k <= n <= 1000`.

**Follow-up:** giải được với độ phức tạp nhỏ hơn `O(n)` không?

````python
import math
import random
import time
from typing import List

examples = [((12, 3), 3), ((7, 2), 7), ((4, 4), -1)]
for (n, k), want in examples:
    divs = [i for i in range(1, n + 1) if n % i == 0]
    print(f"n={n}, k={k} -> ước: {divs} -> đáp án {want}")
````

## 3. Kiến thức nền

### 3.1 Phép chia lấy dư và ước số

`n % i == 0` ⇔ `i` chia hết `n` ⇔ `i` là ước của `n`. Luôn có `1` và `n` là ước.

````python
n = 12
for i in range(1, n + 1):
    print(i, "là ước" if n % i == 0 else "-", end=" | ")
print()
````

### 3.2 Ước đi theo cặp

Nếu `i` là ước thì `n // i` cũng là ước, và `i * (n // i) == n`. Trong mỗi cặp, **một số `≤ √n`**, số kia `≥ √n` (vì nếu cả hai `> √n` thì tích `> n`). Riêng khi `i * i == n` thì cặp "trùng" — chỉ có **một** ước.

````python
for n in [12, 36]:
    pairs = [(i, n // i) for i in range(1, math.isqrt(n) + 1) if n % i == 0]
    print(f"n={n}: cặp ước {pairs}")
````

### 3.3 `math.isqrt` vs `int(math.sqrt(x))`

`math.isqrt(x)` trả **phần nguyên chính xác** của căn bậc hai (số học nguyên). `math.sqrt` dùng số thực — với số rất lớn có thể làm tròn lên sai. Với `n ≤ 1000` thì không sao, nhưng nên tập thói quen dùng `isqrt`.

````python
x = 94906267 ** 2 - 1            # căn đúng phải là 94906266.99... -> phần nguyên 94906266
print("int(math.sqrt(x)) =", int(math.sqrt(x)))
print("math.isqrt(x)     =", math.isqrt(x))
print("isqrt(12) =", math.isqrt(12), "| isqrt(16) =", math.isqrt(16))
````

### 3.4 Duyệt ngược với `range(a, b, -1)` và `reversed`

`range(r, 0, -1)` chạy `r, r-1, …, 1`. `reversed(list)` duyệt list từ cuối. Dùng để liệt kê các ước lớn theo thứ tự tăng dần.

````python
small = [1, 2, 3]
n = 12
print("ước lớn (tăng dần):", [n // d for d in reversed(small)])
print(list(range(3, 0, -1)))
````

### 3.5 Phân tích thừa số nguyên tố

`12 = 2² · 3`. Mọi ước có dạng `2^a · 3^b` với `0 ≤ a ≤ 2`, `0 ≤ b ≤ 1` ⇒ số ước = `(2+1)(1+1) = 6`. Phân tích bằng cách chia thử tới `√n`.

````python
def factorize(n):
    f, p = {}, 2
    while p * p <= n:
        while n % p == 0:
            f[p] = f.get(p, 0) + 1
            n //= p
        p += 1
    if n > 1:
        f[n] = f.get(n, 0) + 1
    return f

for n in [12, 360, 997, 1000]:
    f = factorize(n)
    cnt = 1
    for e in f.values():
        cnt *= e + 1
    print(n, f, "số ước =", cnt)
````

## 5. Cách 1 — Brute Force / baseline: liệt kê mọi ước

**Ý tưởng:** duyệt `i = 1..n`, gom các ước vào list (tự nhiên đã tăng dần), trả `divs[k-1]` nếu đủ, ngược lại `-1`.

````python
def kth_factor_list(n, k):
    divs = [i for i in range(1, n + 1) if n % i == 0]
    return divs[k - 1] if k <= len(divs) else -1

for (n, k), want in examples:
    print(n, k, "->", kth_factor_list(n, k), "| mong đợi", want)
````

**Độ phức tạp:** `O(n)` thời gian, `O(d(n))` bộ nhớ (`d(n)` = số ước, với `n ≤ 1000` thì `d(n) ≤ 32`).

**Nút thắt:** duyệt cả những `i > √n` — nhưng các ước đó đã "lộ diện" qua cặp `(n // i, i)` của một `i ≤ √n`. Có thể chỉ duyệt tới `√n`.

## 6. Ý tưởng thuật toán / bước đột phá

**Insight:** ước đi theo cặp `(i, n // i)` với `i ≤ √n`. Duyệt `i = 1..⌊√n⌋`:
- các `i` chia hết cho ta **nửa nhỏ** (tăng dần);
- các `n // i` tương ứng, đọc theo `i` **giảm dần**, cho **nửa lớn** (tăng dần).

**Quy tắc "nếu… thì…":**
- Nếu `k ≤ số ước nhỏ` thì đáp án là ước nhỏ thứ `k`.
- Ngược lại, đếm tiếp trong nửa lớn: duyệt `i` từ `⌊√n⌋` về 1; **nếu `i * i == n` thì bỏ qua** (ước `√n` đã đếm ở nửa nhỏ).
- Hết cả hai nửa mà chưa đủ `k` ⇒ `-1`.

**Invariant khi đếm ngược `k`:** sau khi xử lý một ước, `k` = số ước **còn phải bỏ qua** nữa để tới đáp án; khi `k` về 0 thì ước vừa xét là đáp án.

Phiên bản không cần list: duyệt hai lượt và giảm `k` ⇒ `O(√n)` thời gian, `O(1)` bộ nhớ.

## 7. Các cách tiếp theo

### Cách 2 — Duyệt `1..n` và đếm ngược `k` (không lưu list)

Bớt bộ nhớ của Cách 1: mỗi khi gặp ước thì `k -= 1`; `k == 0` ⇒ trả ngay. Dừng sớm khi tìm thấy.

````python
def kth_factor_count(n, k):
    for i in range(1, n + 1):
        if n % i == 0:
            k -= 1
            if k == 0:
                return i
    return -1

class SolutionCount:
    def kthFactor(self, n: int, k: int) -> int:
        for i in range(1, n + 1):
            if n % i == 0:
                k -= 1
                if k == 0:
                    return i
        return -1

for (n, k), want in examples:
    print(kth_factor_count(n, k), SolutionCount().kthFactor(n, k), want)
````

**Độ phức tạp:** `O(n)` thời gian tệ nhất (khi `k` lớn hoặc đáp án `-1`), `O(1)` bộ nhớ.

### Cách 3 — Duyệt tới `√n`, hai nửa bằng list

Gom `small` (ước `≤ √n`) và `large` (ước `> √n`, tăng dần) rồi chọn.

````python
def kth_factor_sqrt_list(n, k):
    r = math.isqrt(n)
    small = [i for i in range(1, r + 1) if n % i == 0]
    large = [n // i for i in reversed(small) if i * i != n]
    divs = small + large
    return divs[k - 1] if k <= len(divs) else -1

class SolutionSqrtList:
    def kthFactor(self, n: int, k: int) -> int:
        r = math.isqrt(n)
        small = [i for i in range(1, r + 1) if n % i == 0]
        large = [n // i for i in reversed(small) if i * i != n]
        divs = small + large
        return divs[k - 1] if k <= len(divs) else -1

for (n, k), want in examples:
    print(kth_factor_sqrt_list(n, k), SolutionSqrtList().kthFactor(n, k), want)
````

**Độ phức tạp:** `O(√n)` thời gian, `O(d(n))` bộ nhớ.

### Cách 4 — Duyệt tới `√n`, hai lượt, `O(1)` bộ nhớ

1. **Lượt tăng:** `i = 1..r` (`r = isqrt(n)`); gặp ước thì `k -= 1`, về 0 thì trả `i`.
2. **Lượt giảm:** `i = r..1`; gặp ước và `i * i != n` thì `k -= 1`, về 0 thì trả `n // i`.
3. Không tìm thấy ⇒ `-1`.

````python
def kth_factor_sqrt(n, k):
    r = math.isqrt(n)
    for i in range(1, r + 1):
        if n % i == 0:
            k -= 1
            if k == 0:
                return i
    for i in range(r, 0, -1):
        if n % i == 0 and i * i != n:
            k -= 1
            if k == 0:
                return n // i
    return -1

for (n, k), want in examples:
    print(kth_factor_sqrt(n, k), want)
````

### Cách 5 — Sinh ước từ phân tích thừa số nguyên tố

Phân tích `n = p1^e1 · p2^e2 · …` (chia thử tới `√n`), rồi sinh mọi ước bằng cách nhân dần: bắt đầu `[1]`, với mỗi `p^e` nhân mọi phần tử hiện có với `p^0..p^e`. Cuối cùng `sort`. Hữu ích khi cần **tất cả** ước, hoặc khi đã có sẵn bảng số nguyên tố.

````python
def kth_factor_prime(n, k):
    divs = [1]
    for p, e in factorize(n).items():
        divs = [d * p ** j for d in divs for j in range(e + 1)]
    divs.sort()
    return divs[k - 1] if k <= len(divs) else -1

class SolutionPrime:
    def kthFactor(self, n: int, k: int) -> int:
        return kth_factor_prime(n, k)

for (n, k), want in examples:
    print(kth_factor_prime(n, k), want)
divs360 = [1]
for p, e in factorize(360).items():
    divs360 = [d * p ** j for d in divs360 for j in range(e + 1)]
print("ước của 360 (sinh từ thừa số, chưa sort):", divs360)
````

**Độ phức tạp:** `O(√n)` để phân tích + `O(d log d)` sắp xếp; bộ nhớ `O(d)`.

## 8. TRACE từng bước

Trace Cách 4 với `n = 36` và vài giá trị `k`. Bảng dự kiến cho `k = 7`:

| lượt | i | 36 % i | là ước được đếm? | k sau |
|---|---|---|---|---|
| tăng | 1 | 0 | có (1) | 6 |
| tăng | 2 | 0 | có (2) | 5 |
| tăng | 3 | 0 | có (3) | 4 |
| tăng | 4 | 0 | có (4) | 3 |
| tăng | 5 | 1 | không | 3 |
| tăng | 6 | 0 | có (6) | 2 |
| giảm | 6 | 0 | **bỏ** (6·6 = 36) | 2 |
| giảm | 5 | 1 | không | 2 |
| giảm | 4 | 0 | có (36//4 = 9) | 1 |
| giảm | 3 | 0 | có (36//3 = 12) | **0 ⇒ trả 12** |

````python
def trace_sqrt(n, k):
    r = math.isqrt(n)
    print(f"n={n}, k={k}, r=isqrt(n)={r}")
    for i in range(1, r + 1):
        if n % i == 0:
            k -= 1
            print(f"  tăng i={i}: ước {i:4} -> k={k}")
            if k == 0:
                print("  => đáp án", i); return i
        else:
            print(f"  tăng i={i}: không chia hết")
    for i in range(r, 0, -1):
        if n % i == 0 and i * i != n:
            k -= 1
            print(f"  giảm i={i}: ước {n // i:4} -> k={k}")
            if k == 0:
                print("  => đáp án", n // i); return n // i
        elif n % i == 0:
            print(f"  giảm i={i}: bỏ qua vì {i}*{i}={n}")
        else:
            print(f"  giảm i={i}: không chia hết")
    print("  => không đủ ước, trả -1"); return -1

trace_sqrt(36, 7)
````

## 9. Code nộp LeetCode

Bản chính: Cách 4 (`O(√n)` thời gian, `O(1)` bộ nhớ) — trả lời đúng follow-up.

````python
import math

class Solution:
    def kthFactor(self, n: int, k: int) -> int:
        r = math.isqrt(n)
        for i in range(1, r + 1):              # nửa nhỏ: i <= sqrt(n)
            if n % i == 0:
                k -= 1
                if k == 0:
                    return i
        for i in range(r, 0, -1):              # nửa lớn: n // i, bỏ ước căn trùng
            if n % i == 0 and i * i != n:
                k -= 1
                if k == 0:
                    return n // i
        return -1

for (n, k), want in examples:
    assert Solution().kthFactor(n, k) == want
print("OK ví dụ đề")
````

## 10. Vì sao thuật toán đúng

Gọi `r = ⌊√n⌋`. Tập ước `D` của `n` chia thành `S = {d ∈ D : d ≤ r}` và `L = {d ∈ D : d > r}`.

1. **Song ánh `S \ {√n} → L`:** `d ↦ n // d`. Nếu `d ≤ r` và `d·d ≠ n` thì `d < √n` ⇒ `n/d > √n ≥ r` ⇒ `n // d ∈ L`. Ngược lại mọi `e ∈ L` có `n/e < √n` ⇒ `n/e ∈ S`. Nên `L = {n // d : d ∈ S, d·d ≠ n}`.
2. **Thứ tự:** lượt tăng liệt kê `S` tăng dần. Lượt giảm duyệt `d` giảm dần ⇒ `n // d` **tăng dần** ⇒ liệt kê `L` tăng dần. Mọi phần tử của `S` `< ` mọi phần tử của `L`.
3. Vậy thứ tự "lượt tăng rồi lượt giảm" chính là danh sách ước tăng dần, và bộ đếm `k` giảm đúng một mỗi ước ⇒ khi `k` về 0 ta đang ở ước thứ `k` ban đầu. Nếu hết mà `k > 0` ⇒ `|D| < k` ⇒ `-1`.

## 11. Phân tích Time / Space

`n ≤ 1000`, `d(n)` = số ước (`≤ 32` với `n ≤ 1000`; nói chung tăng rất chậm).

| Cách | Time | Space |
|---|---|---|
| 1. Liệt kê `1..n` | `O(n)` | `O(d(n))` |

## 13. Các lỗi người mới hay gặp

### Lỗi 1 — Đếm ước căn hai lần với số chính phương

Khi ghép nửa lớn `n // i` mà quên điều kiện `i * i != n`.

````python
def wrong1(n, k):
    r = math.isqrt(n)
    small = [i for i in range(1, r + 1) if n % i == 0]
    large = [n // i for i in reversed(small)]          # SAI: thiếu i*i != n
    divs = small + large
    return divs[k - 1] if k <= len(divs) else -1

for n, k in [(4, 3), (4, 4), (36, 6)]:
    print(f"n={n}, k={k}: sai={wrong1(n, k)} | đúng={Solution().kthFactor(n, k)}")
````

`n = 4`: danh sách sai `[1, 2, 2, 4]` ⇒ ước thứ 3 thành 2 (đúng là 4), và ước thứ 4 "tồn tại" (đúng là `-1`). **Sửa:** bỏ ước căn ở nửa lớn.

### Lỗi 2 — `range(1, n)` bỏ sót chính `n`

`range` không lấy cận phải. Quên `+ 1` ⇒ mất ước `n`.

````python
def wrong2(n, k):
    for i in range(1, n):                              # SAI: thiếu n
        if n % i == 0:
            k -= 1
            if k == 0:
                return i
    return -1

for n, k in [(7, 2), (1, 1), (12, 6)]:
    print(f"n={n}, k={k}: sai={wrong2(n, k)} | đúng={Solution().kthFactor(n, k)}")
````

### Lỗi 3 — Nhầm chỉ số 0/1

Đề đếm từ 1, list Python đánh số từ 0 ⇒ phải trả `divs[k - 1]`. Viết `divs[k]` vừa lệch một vị trí, vừa có thể `IndexError` (ta bắt lỗi để minh hoạ).

````python
def wrong3(n, k):
    divs = [i for i in range(1, n + 1) if n % i == 0]
    return divs[k] if k <= len(divs) else -1           # SAI: phải là k - 1

for n, k in [(12, 3), (4, 3)]:
    try:
        print(f"n={n}, k={k}: sai={wrong3(n, k)}", end="")
    except IndexError as e:
        print(f"n={n}, k={k}: sai=IndexError({e})", end="")
    print(f" | đúng={Solution().kthFactor(n, k)}")
````

### Lỗi 4 — Lượt giảm bắt đầu sai chỗ / sai thứ tự

Lượt hai duyệt `i` **tăng dần** và lấy `n // i` ⇒ nửa lớn bị liệt kê **giảm dần** ⇒ sai thứ tự.

````python
def wrong4(n, k):
    r = math.isqrt(n)
    for i in range(1, r + 1):
        if n % i == 0:
            k -= 1
            if k == 0:
                return i
    for i in range(1, r + 1):                          # SAI: phải duyệt giảm
        if n % i == 0 and i * i != n:
            k -= 1
            if k == 0:
                return n // i
    return -1

for n, k in [(12, 4), (12, 6), (36, 7)]:
    print(f"n={n}, k={k}: sai={wrong4(n, k)} | đúng={Solution().kthFactor(n, k)}")
````

### Lỗi 5 — Dùng `int(math.sqrt(n))` với số rất lớn

Với `n ≤ 1000` vô hại, nhưng ở biến thể `n` lớn có thể làm tròn **lên** (mục 3.3), khiến duyệt thêm một `i > √n` và đếm trùng ước. Luôn dùng `math.isqrt`.

## 14. Mẹo làm bài nhanh & nhận dạng pattern

- Gặp "ước số", "số ước", "tổng ước" ⇒ nghĩ ngay **cặp `(i, n // i)` với `i ≤ √n`**.
- Cần ước **theo thứ tự**: nửa nhỏ tăng + nửa lớn đọc ngược — không cần `sort`.
- Số chính phương: luôn tự hỏi "ước `√n` có bị đếm hai lần không?".
- Test nhanh: `n = 1`, số nguyên tố, số chính phương (`4`, `16`, `36`), `k` vừa đủ và `k` thừa một.
- `math.isqrt` cho căn nguyên chính xác.

## 15. Biến thể của bài

1. **Ước thứ `k` lớn nhất:** đối xứng — duyệt `n // i` với `i` tăng trước (Bài 4).
2. **Đếm số ước / tổng các ước:** `O(√n)` bằng cặp ước, hoặc công thức từ phân tích nguyên tố (`∏(e+1)`, `∏(p^(e+1)-1)/(p-1)`).
3. **LC 1390 Four Divisors, LC 1979, LC 2427 Number of Common Factors:** các bài đếm ước.
4. **Nhiều truy vấn với `n` khác nhau:** sàng ước (`for d in 1..N: for m in d, 2d, ...: divs[m].append(d)`) trong `O(N log N)`.
5. **`n` tới `10^12`:** vẫn `O(√n) = 10^6` phép chia — chấp nhận được.

````python
# Minh hoạ biến thể 4: sàng danh sách ước cho mọi n <= N, rồi trả lời nhiều truy vấn O(1)
N = 1000
divs_of = [[] for _ in range(N + 1)]
for d in range(1, N + 1):
    for m in range(d, N + 1, d):
        divs_of[m].append(d)                  # d tăng dần nên list đã sắp xếp
queries = [(12, 3), (7, 2), (4, 4), (840, 16)]
print([divs_of[n][k - 1] if k <= len(divs_of[n]) else -1 for n, k in queries])
````

## 16. So sánh các cách

| Cách | Ý tưởng | Time | Space | Khi nào dùng |
|---|---|---|---|---|
| 1. Liệt kê `1..n` | Gom mọi ước | `O(n)` | `O(d)` | `n` nhỏ, viết nhanh nhất |
| 2. Đếm `1..n` | Giảm `k`, dừng sớm | `O(n)` | `O(1)` | Như trên, đỡ bộ nhớ |
| 3. `√n` + list | Nửa nhỏ + nửa lớn đảo ngược | `O(√n)` | `O(d)` | Cần cả danh sách ước |
| 4. `√n` hai lượt | Đếm ngược `k` hai lượt | `O(√n)` | `O(1)` | Bản nộp tối ưu, follow-up |
| 5. Phân tích nguyên tố | Sinh ước từ `p^e` rồi sort | `O(√n + d log d)` | `O(d)` | Cần mọi ước / đã có sàng nguyên tố |
