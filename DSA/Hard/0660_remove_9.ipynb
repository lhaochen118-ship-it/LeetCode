## 3. Kiến thức nền

### 3.1. `str(x)` và toán tử `in` để kiểm tra chữ số

`'9' in str(x)` kiểm tra số `x` có chữ số 9 không. Đơn giản nhưng tốn `O(số chữ số)` mỗi lần và tạo chuỗi mới.

````python
for x in [8, 19, 90, 123, 1009]:
    print(x, "chứa 9?", '9' in str(x))
````

### 3.2. Lấy chữ số bằng `% 10` và `// 10`

Không cần chuỗi: `x % 10` là chữ số cuối, `x // 10` bỏ chữ số cuối. `divmod(x, 10)` trả cả hai cùng lúc.

````python
def has_digit(x, d):
    while x > 0:
        x, r = divmod(x, 10)
        if r == d:
            return True
    return False

print([has_digit(x, 9) for x in [8, 19, 90, 123, 1009]])
print(divmod(1234, 10))
````

### 3.3. Hệ cơ số `b`: đổi số sang cơ số `b`

Viết `n` trong cơ số `b`: lặp `n, r = divmod(n, b)`, các phần dư `r` là các chữ số từ **phải sang trái**. Ví dụ `10` trong cơ số 9: `10 = 1·9 + 1` → `"11"`.

````python
def to_base(n, b):
    if n == 0:
        return "0"
    digits = []
    while n > 0:
        n, r = divmod(n, b)
        digits.append(str(r))
    return ''.join(reversed(digits))

for n in [8, 9, 10, 80, 81]:
    print(n, "-> cơ số 9:", to_base(n, 9))
````

### 3.4. Đọc chuỗi trong cơ số `b`: `int(s, b)`

Ngược lại, `int("11", 9) = 10`. Và `int("11")` (cơ số 10 mặc định) = `11`. Chính sự "đọc nhầm cơ số" này là lời giải!

````python
print(int("11", 9), int("11"), int("100", 9), int(to_base(81, 9)))
````

### 3.5. Lũy thừa và số lượng chuỗi chữ số

Có `9^k` chuỗi gồm `k` chữ số lấy từ `{0, 1, ..., 8}` (cho phép số 0 đứng đầu). Dùng để đếm số không chứa 9 có tối đa `k` chữ số: `9^k` chuỗi, trừ chuỗi toàn 0 (số 0) → `9^k - 1` số dương.

````python
for k in range(1, 5):
    cnt = sum(1 for x in range(1, 10**k) if '9' not in str(x))
    print(f"tối đa {k} chữ số: {cnt} số, 9^{k} - 1 = {9**k - 1}")
````

### 3.6. Tìm kiếm nhị phân "giá trị nhỏ nhất thỏa điều kiện"

Nếu `P(x)` đơn điệu (sai…sai, đúng…đúng), tìm `x` nhỏ nhất có `P(x)` đúng bằng `while lo < hi: mid; if P(mid): hi = mid else: lo = mid + 1`.

````python
def first_true(lo, hi, P):
    while lo < hi:
        mid = (lo + hi) // 2
        if P(mid):
            hi = mid
        else:
            lo = mid + 1
    return lo

print(first_true(0, 100, lambda x: x * x >= 30))
````

## 4. Ý tưởng nền tảng / trực giác

### 4.1. Nhóm theo số chữ số

- 1 chữ số (không 9): `1..8` → **8** số (thứ tự 1–8).
- 2 chữ số: chữ số đầu `1..8` (8 cách), chữ số sau `0..8` (9 cách) → **72** số (thứ tự 9–80).
- 3 chữ số: `8 · 9 · 9 = 648` số (thứ tự 81–728).

Vậy số thứ `n = 100` có 3 chữ số; nó là số 3 chữ số thứ `100 - 80 = 20`.

### 4.2. Số thứ 20 trong nhóm 3 chữ số

Các số 3 chữ số không có 9, theo thứ tự: `100, 101, ..., 108, 110, ...`. Mỗi "chữ số hàng chục" gánh 9 số. Thứ 20 (đếm từ 1) ⇒ chỉ số 19 (đếm từ 0) = `0·81 + 2·9 + 1` → chữ số đầu là `1 + 0 = 1`, hàng chục `2`, hàng đơn vị `1` → `121`.

Kiểm tra bằng đổi cơ số: `100` trong cơ số 9 là `1·81 + 2·9 + 1` → `"121"`. Trùng khớp!

### 4.3. Trực giác cơ số 9

Đếm số trong cơ số 9: `1, 2, ..., 8, 10, 11, ..., 18, 20, ...` — **đúng y hệt** dãy của bài! Hệ cơ số 9 chỉ dùng chữ số `0..8`, không bao giờ có `9`. Đếm số trong cơ số 9 theo thứ tự = liệt kê mọi chuỗi chữ số không chứa 9 theo thứ tự tăng dần.

Do đó: **số thứ `n` của dãy = cách viết `n` trong cơ số 9**, đọc lại như số thập phân.

````python
seq_small = [x for x in range(1, 2000) if '9' not in str(x)]
for n in [1, 8, 9, 10, 80, 81, 100, 728, 729]:
    b9 = to_base(n, 9)
    print(f"n={n:>3}: cơ số 9 = {b9:>4} | phần tử thứ n của dãy thật = {seq_small[n - 1]}")
    assert int(b9) == seq_small[n - 1]
````

## 5. Cách 1 — Brute Force: duyệt và bỏ qua số chứa 9

**Ý tưởng.** Duyệt `x = 1, 2, 3, ...`, đếm những số không chứa 9; khi đếm đủ `n` thì trả `x`.

````python
def new_integer_brute(n):
    x = 0
    count = 0
    while count < n:
        x += 1
        if '9' not in str(x):
            count += 1
    return x

print(new_integer_brute(9), new_integer_brute(10), new_integer_brute(100))
assert new_integer_brute(9) == 10 and new_integer_brute(10) == 11
````

**Độ phức tạp.** Phải duyệt tới đáp án `x`. Với `n = 8·10^8`, đáp án cỡ `2·10^9` (xem mục 11) → hơn 2 tỉ vòng lặp, mỗi vòng `O(log x)` để tạo chuỗi → **nhiều phút** trong Python, vượt xa giới hạn thời gian. Space `O(1)`.

**Nút thắt.** Ta liệt kê **từng** số một. Nhưng số lượng số không chứa 9 trong một khoảng **đếm được bằng công thức** (mục 3.5) — không cần liệt kê.

````python
import time
for n in (10**4, 10**5, 10**6):
    t = time.perf_counter()
    new_integer_brute(n)
    dt = time.perf_counter() - t
    x_end = int(to_base(n, 9))                  # số vòng lặp = giá trị đáp án
    x_big = int(to_base(8 * 10**8, 9))
    print(f"n={n:>8}: {dt:.3f}s ({x_end} vòng) -> ước tính n = 8e8 ({x_big} vòng): ~{dt * x_big / x_end / 60:.0f} phút")
````

## 6. Bước đột phá

**Insight 1 (đếm thay vì liệt kê).** Đặt `cnt(x)` = số lượng số trong `[1, x]` không chứa 9. Tính được trong `O(số chữ số)` bằng cách duyệt chữ số của `x` từ trái sang phải. `cnt` không giảm → số thứ `n` là `x` **nhỏ nhất** có `cnt(x) >= n` → tìm kiếm nhị phân.

**Insight 2 (song ánh với cơ số 9).** Ánh xạ "chuỗi chữ số không chứa 9" ↔ "biểu diễn cơ số 9" là **song ánh bảo toàn thứ tự**:
- Cùng độ dài: thứ tự từ điển của chữ số `0..8` giống nhau trong cả hai hệ.
- Khác độ dài: chuỗi dài hơn lớn hơn trong cả hai hệ.

Vậy thứ hạng của số `x` (không chứa 9) trong dãy = giá trị của chuỗi chữ số của `x` đọc trong cơ số 9.

**Invariant khi đổi cơ số:** sau mỗi bước `n, r = divmod(n, 9)`, các chữ số đã lấy ra là các chữ số thấp nhất của `n` ban đầu trong cơ số 9.

**Quy tắc nếu…thì…:**
- Nếu cần số thứ `n` trong dãy bỏ một chữ số → nghĩ tới đổi cơ số `9`.
- Nếu chữ số bị bỏ **không phải 9** (ví dụ 4) → đổi cơ số 9 rồi **ánh xạ** chữ số `>= 4` thành chữ số `+1`.
- Nếu chữ số bị bỏ là `0` → không còn là "cơ số" thông thường (xem biến thể).

### 7.1. Cách 2 — Đếm theo chữ số + tìm kiếm nhị phân

**Giải thích hàm đếm.** Đếm số trong `[0, x]` không chứa 9 (rồi trừ 1 cho số 0). Duyệt các chữ số `d` của `x` từ trái sang phải, giữ nguyên tiền tố đã qua (tiền tố của `x`, chưa có 9):

- Ở vị trí có `rem` chữ số phía sau, đặt chữ số **nhỏ hơn** `d`: có `d` lựa chọn (`0..d-1`, đều khác 9 vì `d <= 9`), phần sau tự do: `9^rem` cách → cộng `d · 9^rem`.
- Nếu `d == 9`: không thể giữ tiền tố (chứa 9) → dừng.
- Nếu đi hết mà không gặp 9: cộng 1 cho chính `x`.

````python
def count_no9(x):
    """Số lượng số trong [1, x] không chứa chữ số 9."""
    s = str(x)
    L = len(s)
    total = 0
    for i, ch in enumerate(s):
        d = int(ch)
        rem = L - i - 1
        total += d * 9 ** rem          # đặt chữ số nhỏ hơn d ở vị trí này
        if d == 9:
            return total - 1           # tiền tố chứa 9: dừng; trừ số 0
    return total + 1 - 1               # +1 cho chính x, -1 cho số 0

for x in [8, 9, 10, 88, 99, 100, 1000]:
    print(f"cnt({x}) = {count_no9(x)} | brute = {sum(1 for y in range(1, x + 1) if '9' not in str(y))}")
````

**Đọc code hàm đếm:**

- `total += d * 9 ** rem`: các số cùng tiền tố, chữ số hiện tại `< d`, phần sau bất kỳ trong `0..8`.
- `d == 9`: các số có tiền tố này **và** chữ số hiện tại `= 9` đều chứa 9 → không đếm; các số có chữ số hiện tại `< 9` đã được cộng ở dòng trên → dừng.
- Đếm này bao gồm số 0 (chuỗi toàn 0) → trừ 1.

Giờ tìm kiếm nhị phân: `x` nhỏ nhất với `count_no9(x) >= n`. Cận trên an toàn: đáp án không vượt `n` viết trong cơ số 9 đọc thập phân ≤ `10^10` với `n ≤ 8·10^8`; dùng `hi = 10 ** (len(str(n)) + 2)` cho chắc.

````python
def new_integer_binary(n):
    lo, hi = 1, 10 ** (len(str(n)) + 2)
    while lo < hi:
        mid = (lo + hi) // 2
        if count_no9(mid) >= n:
            hi = mid
        else:
            lo = mid + 1
    return lo

print(new_integer_binary(9), new_integer_binary(10), new_integer_binary(100), new_integer_binary(8 * 10**8))
assert new_integer_binary(9) == 10 and new_integer_binary(10) == 11
````

**Vì sao kết quả không chứa 9?** `x*` nhỏ nhất có `cnt(x*) >= n` thỏa `cnt(x* - 1) < n <= cnt(x*)` → `cnt` tăng tại `x*` → `x*` được đếm → `x*` không chứa 9. (Cùng lập luận với bài Kth Smallest in Multiplication Table.)

**Code nộp LeetCode (Cách 2):**

````python
class SolutionBinary:
    def newInteger(self, n: int) -> int:
        def cnt(x: int) -> int:
            s = str(x)
            L = len(s)
            total = 0
            for i, ch in enumerate(s):
                d = ord(ch) - 48
                total += d * 9 ** (L - i - 1)
                if d == 9:
                    return total - 1
            return total

        lo, hi = 1, 10 ** 11
        while lo < hi:
            mid = (lo + hi) // 2
            if cnt(mid) >= n:
                hi = mid
            else:
                lo = mid + 1
        return lo

assert SolutionBinary().newInteger(9) == 10
assert SolutionBinary().newInteger(10) == 11
print("SolutionBinary OK")
````

**Độ phức tạp Cách 2:** `log2(10^11) ≈ 37` vòng × `O(11)` chữ số → Time `O(log² ans)` (thực tế vài trăm phép), Space `O(log ans)`.

### 7.2. Cách 3 — Dựng từng chữ số (không cần biết "cơ số 9")

**Giải thích.** Tư duy như mục 4.1–4.2 nhưng tổng quát:

1. Tìm số chữ số `L` của đáp án: trừ dần số lượng số có 1, 2, ... chữ số (`8 · 9^(L-1)`) cho tới khi `n` lọt vào nhóm `L` chữ số. Gọi `idx` = thứ hạng (đếm từ 0) trong nhóm.
2. Chữ số đầu: mỗi lựa chọn `1..8` gánh `9^(L-1)` số → chữ số đầu = `1 + idx // 9^(L-1)`.
3. Các chữ số sau: mỗi lựa chọn `0..8` gánh `9^(rem)` số → chữ số = `idx // 9^rem` (sau khi lấy phần dư ở bước trước).

````python
def new_integer_digits(n):
    L = 1
    while n > 8 * 9 ** (L - 1):            # bỏ qua cả nhóm L chữ số
        n -= 8 * 9 ** (L - 1)
        L += 1
    idx = n - 1                             # thứ hạng 0-based trong nhóm L chữ số
    block = 9 ** (L - 1)
    first, idx = divmod(idx, block)
    digits = [first + 1]
    for rem in range(L - 2, -1, -1):
        d, idx = divmod(idx, 9 ** rem)
        digits.append(d)
    return int(''.join(map(str, digits)))

print([new_integer_digits(n) for n in (1, 8, 9, 10, 80, 81, 100)])
assert new_integer_digits(100) == 121
````

**Đọc code:**

- Vòng `while`: nhóm `L` chữ số có `8 · 9^(L-1)` số. Nếu `n` lớn hơn, bỏ qua cả nhóm.
- `idx = n - 1`: chuyển sang đếm từ 0 cho tiện chia.
- Chữ số đầu dùng `first + 1` (không có số 0 đứng đầu); các chữ số sau dùng trực tiếp `d` trong `0..8`.

**Code nộp LeetCode (Cách 3):**

````python
class SolutionDigits:
    def newInteger(self, n: int) -> int:
        L, size = 1, 8
        while n > size:
            n -= size
            L += 1
            size *= 9
        idx = n - 1
        p = 9 ** (L - 1)
        first, idx = divmod(idx, p)
        res = first + 1
        while p > 1:
            p //= 9
            d, idx = divmod(idx, p)
            res = res * 10 + d
        return res

assert SolutionDigits().newInteger(9) == 10
assert SolutionDigits().newInteger(100) == 121
print("SolutionDigits OK")
````

**Độ phức tạp Cách 3:** Time `O(L) = O(log n)`, Space `O(1)`.

Nhận xét: nếu bạn cộng lại các bước, `first + 1` ở chữ số đầu và việc "trừ dần các nhóm" chính là **bù** cho việc chuỗi cơ số 9 không có số 0 đứng đầu... và kết quả trùng khớp cơ số 9. Cách 4 làm điều đó trực tiếp.

### 7.3. Cách 4 — Đổi `n` sang cơ số 9

**Giải thích.** Theo Insight 2: viết `n` trong cơ số 9, rồi đọc chuỗi đó như số thập phân.

```text
res = 0; base = 1
while n > 0:
    n, r = divmod(n, 9)
    res += r * base          # đặt chữ số r vào vị trí thập phân tương ứng
    base *= 10
return res
```

````python
def new_integer_base9(n):
    res, base = 0, 1
    while n > 0:
        n, r = divmod(n, 9)
        res += r * base
        base *= 10
    return res

print([new_integer_base9(n) for n in (1, 8, 9, 10, 80, 81, 100, 8 * 10**8)])
assert new_integer_base9(9) == 10 and new_integer_base9(10) == 11
````

**Đọc code:**

- `divmod(n, 9)`: chữ số thấp nhất trong cơ số 9 là `r`.
- `res += r * base` với `base = 1, 10, 100, ...`: đặt `r` vào hàng đơn vị, hàng chục, hàng trăm... **của hệ thập phân**. Đây chính là bước "đọc chuỗi cơ số 9 như số thập phân" mà không cần tạo chuỗi.
- Một dòng tương đương: `int(to_base(n, 9))`.

**Code nộp LeetCode (Cách 4):**

````python
class SolutionBase9:
    def newInteger(self, n: int) -> int:
        res, base = 0, 1
        while n:
            n, r = divmod(n, 9)
            res += r * base
            base *= 10
        return res

assert SolutionBase9().newInteger(9) == 10
print("SolutionBase9 OK")
````

**Độ phức tạp Cách 4:** `log_9(n) <= 10` vòng → Time `O(log n)`, Space `O(1)`.

## 8. TRACE từng bước

### 8.1. Đổi cơ số (Cách 4) với `n = 100`

| vòng | n trước | `divmod(n, 9)` | r | base | res |
|---|---|---|---|---|---|
| 1 | 100 | (11, 1) | 1 | 1 | 1 |
| 2 | 11 | (1, 2) | 2 | 10 | 21 |
| 3 | 1 | (0, 1) | 1 | 100 | 121 |

````python
def trace_base9(n):
    res, base, it = 0, 1, 0
    print(f"{'vòng':>4} | {'n trước':>10} | {'divmod':>14} | {'r':>2} | {'base':>12} | res")
    while n > 0:
        it += 1
        q, r = divmod(n, 9)
        res += r * base
        print(f"{it:>4} | {n:>10} | {str((q, r)):>14} | {r:>2} | {base:>12} | {res}")
        n = q
        base *= 10
    return res

assert trace_base9(100) == 121
print()
assert trace_base9(8 * 10**8) == new_integer_binary(8 * 10**8)
````

### 8.2. Hàm đếm (Cách 2) với `x = 121`

````python
def trace_count(x):
    s = str(x)
    L = len(s)
    total = 0
    print(f"đếm số trong [0, {x}] không chứa 9:")
    for i, ch in enumerate(s):
        d = int(ch)
        rem = L - i - 1
        add = d * 9 ** rem
        total += add
        print(f"  vị trí {i} (chữ số {d}, còn {rem} chữ số sau): chữ số nhỏ hơn {d} -> {d}·9^{rem} = {add}, tổng = {total}")
        if d == 9:
            print("  gặp chữ số 9 -> dừng")
            return total - 1
    print(f"  không gặp 9: +1 cho chính {x} -> {total + 1}; trừ số 0 -> {total}")
    return total

assert trace_count(121) == 100
assert trace_count(195) == count_no9(195)
````

`cnt(121) = 100` và `121` không chứa 9 → `121` là số thứ 100 ✓.

### 8.3. Tìm kiếm nhị phân (Cách 2) với `n = 100`

````python
def trace_binary(n, show=12):
    lo, hi, it = 1, 10 ** (len(str(n)) + 2), 0
    while lo < hi:
        it += 1
        mid = (lo + hi) // 2
        c = count_no9(mid)
        if it <= show:
            print(f"vòng {it:>2}: lo={lo:>6}, hi={hi:>6}, mid={mid:>6}, cnt(mid)={c:>5} -> {'hi = mid' if c >= n else 'lo = mid + 1'}")
        if c >= n:
            hi = mid
        else:
            lo = mid + 1
    print(f"kết quả: {lo} sau {it} vòng")
    return lo

assert trace_binary(100) == 121
````

## 9. Code nộp LeetCode

Lời giải chính: **đổi `n` sang cơ số 9** (Cách 4).

````python
class Solution:
    def newInteger(self, n: int) -> int:
        # Số thứ n của dãy không chứa chữ số 9 = n viết trong cơ số 9 (đọc như số thập phân)
        res, base = 0, 1
        while n > 0:
            n, r = divmod(n, 9)
            res += r * base
            base *= 10
        return res

sol = Solution()
assert sol.newInteger(9) == 10
assert sol.newInteger(10) == 11
print("Solution OK")
````

## 10. Vì sao thuật toán đúng

Gọi `A` là tập các số nguyên dương không chứa chữ số 9, sắp tăng dần: `a_1 < a_2 < ...`. Với số `x ∈ A` có chữ số thập phân `d_{L-1} ... d_1 d_0` (mọi `d_i ∈ {0..8}`), định nghĩa

```text
φ(x) = d_{L-1}·9^{L-1} + ... + d_1·9 + d_0     (đọc chuỗi chữ số của x trong cơ số 9)
```

**Bổ đề 1 (song ánh).** `φ: A → {1, 2, 3, ...}` là song ánh. Mỗi số nguyên dương `m` có **đúng một** biểu diễn cơ số 9 không có số 0 đứng đầu; chuỗi đó chỉ dùng chữ số `0..8`, đọc như số thập phân cho đúng một phần tử của `A`. Ngược lại mỗi `x ∈ A` cho một `φ(x)` dương.

**Bổ đề 2 (bảo toàn thứ tự).** `x < y` (trong `A`) ⇔ `φ(x) < φ(y)`.
- Nếu `x` có ít chữ số hơn `y`: `x < y`, và `φ(x) < 9^{L_x} <= 9^{L_y - 1} <= φ(y)` (chữ số đầu của `y` `>= 1`).
- Nếu cùng số chữ số: so sánh thập phân = so sánh từ điển chữ số = so sánh trong cơ số 9 (cùng tập chữ số `0..8`, cùng độ dài).

**Kết luận.** Song ánh tăng giữa hai tập sắp thứ tự tốt bắt đầu từ phần tử nhỏ nhất `φ(1) = 1` → `φ(a_n) = n` với mọi `n`, tức `a_n = φ^{-1}(n)` = biểu diễn cơ số 9 của `n` đọc như số thập phân — đúng thứ Cách 4 tính.

**Cách 2** đúng vì `cnt(x) = |A ∩ [1, x]|` (lập luận đếm ở mục 7.1) và `a_n` là `x` nhỏ nhất có `cnt(x) >= n`.

### Lỗi 1: Đọc ngược thứ tự chữ số khi đổi cơ số

**Code sai:** nối các phần dư theo thứ tự lấy ra (từ chữ số thấp) mà không đảo.

````python
def base9_reversed_wrong(n):
    s = ""
    while n > 0:
        n, r = divmod(n, 9)
        s += str(r)                    # SAI: chữ số thấp đứng đầu
    return int(s)

for n in (12, 15, 2024):
    print(n, "sai:", base9_reversed_wrong(n), "| đúng:", Solution().newInteger(n))
````

### Lỗi 2: Dùng `int(str(n), 9)` (ngược chiều)

**Code sai:** đọc `n` **như** số cơ số 9 rồi đổi sang thập phân — đó là phép **ngược** (tính thứ hạng của một số không chứa 9), và còn lỗi nếu `n` chứa chữ số 9.

````python
def wrong_direction(n):
    try:
        return int(str(n), 9)          # SAI chiều
    except ValueError as e:
        return f"ValueError ({e})"

for n in (10, 100, 19):
    print(n, "sai:", wrong_direction(n), "| đúng:", Solution().newInteger(n))
````

**Code đúng:** chuyển `n` **sang** cơ số 9 (divmod), rồi đọc như thập phân. (Phép `int(str(x), 9)` lại hữu ích cho câu hỏi ngược: "`x` là phần tử thứ mấy?")

### Lỗi 3: Lệch 1 khi dựng chữ số (quên `n - 1`)

**Code sai:** trong Cách 3 dùng `idx = n` thay vì `n - 1`.

````python
def digits_off_by_one_wrong(n):
    L = 1
    while n > 8 * 9 ** (L - 1):
        n -= 8 * 9 ** (L - 1)
        L += 1
    idx = n                             # SAI: phải là n - 1
    block = 9 ** (L - 1)
    first, idx = divmod(idx, block)
    digits = [first + 1]
    for rem in range(L - 2, -1, -1):
        d, idx = divmod(idx, 9 ** rem)
        digits.append(d)
    return int(''.join(map(str, digits)))

for n in (1, 8, 9, 80):
    print(n, "sai:", digits_off_by_one_wrong(n), "| đúng:", Solution().newInteger(n))
````

**Code đúng:** `idx = n - 1` (thứ hạng đếm từ 0).

### Lỗi 4: Hàm đếm quên trừ số 0 hoặc quên dừng khi gặp 9

Nếu không dừng khi gặp chữ số 9, các số có tiền tố chứa 9 bị đếm nhầm. Nếu không trừ số 0, `cnt` lệch 1 → tìm kiếm nhị phân trả về số **liền trước** đáp án đúng.

````python
def count_no_stop_wrong(x):
    s = str(x)
    L = len(s)
    total = 0
    for i, ch in enumerate(s):
        total += int(ch) * 9 ** (L - i - 1)       # SAI: không dừng khi gặp 9
    return total

for x in (95, 199):
    print(f"cnt({x}) sai: {count_no_stop_wrong(x)} | đúng: {count_no9(x)}")
````

## 14. Mẹo & nhận dạng pattern

- **"Số thứ `n` trong dãy bỏ đi các số chứa chữ số `d`"** → đổi cơ số 9 (+ ánh xạ chữ số nếu `d != 9`).
- **"Đếm số trong `[1, x]` thỏa điều kiện về chữ số"** → duyệt chữ số từ trái sang phải (digit DP); kết hợp tìm kiếm nhị phân để tìm phần tử thứ `n`.
- Thấy lũy thừa `9^k`, `8·9^(k-1)` khi đếm → nghi ngờ có cơ số ẩn.
- Song ánh **bảo toàn thứ tự** giữa hai tập → thứ hạng chuyển qua lại được.
- Bài cùng họ: 357 (Count Numbers with Unique Digits), 902 (Numbers At Most N Given Digit Set), 1012 (Numbers With Repeated Digits), 233 (Number of Digit One), 600 (Non-negative Integers without Consecutive Ones).

## 15. Biến thể

1. **Bỏ chữ số `d` (1 ≤ d ≤ 8)** — đổi `n` sang cơ số 9, rồi mỗi chữ số `r >= d` thay bằng `r + 1`.
2. **Câu hỏi ngược** — cho `x` không chứa 9, hỏi nó là phần tử thứ mấy: `int(str(x), 9)`.
3. **Bỏ chữ số 0** — dãy `1..9, 11..19, 21..` tương ứng **hệ cơ số 9 song ánh** (bijective base-9, chữ số 1..9).
4. **Bỏ nhiều chữ số** (ví dụ 4 và 7) — cơ số `10 - (số chữ số bị bỏ)` với ánh xạ chữ số tương ứng (nếu 0 không bị bỏ).

````python
def kth_without_digit(n, d):
    """Biến thể 1 & 4 (d là một chữ số 1..9): số thứ n không chứa chữ số d."""
    allowed = [c for c in range(10) if c != d]    # 0 luôn được phép
    res, base = 0, 1
    while n > 0:
        n, r = divmod(n, 9)
        res += allowed[r] * base
        base *= 10
    return res

for d in range(1, 10):
    seq_d = [x for x in range(1, 5000) if str(d) not in str(x)]
    for n in range(1, len(seq_d) + 1, 7):
        assert kth_without_digit(n, d) == seq_d[n - 1], (n, d)
print("Biến thể 1 OK; số thứ 20 không chứa chữ số 4:", kth_without_digit(20, 4))
````

````python
def rank_of(x):
    """Biến thể 2: x (không chứa 9) là phần tử thứ mấy."""
    return int(str(x), 9)

def kth_without_zero(n):
    """Biến thể 3: số thứ n không chứa chữ số 0 (cơ số 9 song ánh, chữ số 1..9)."""
    digits = []
    while n > 0:
        n -= 1
        n, r = divmod(n, 9)
        digits.append(r + 1)
    return int(''.join(map(str, reversed(digits))))

assert all(rank_of(Solution().newInteger(n)) == n for n in range(1, 3000))
seq0 = [x for x in range(1, 3000) if '0' not in str(x)]
assert all(kth_without_zero(n) == seq0[n - 1] for n in range(1, len(seq0) + 1))
print("Biến thể 2, 3 OK:", rank_of(121), kth_without_zero(10), kth_without_zero(81), kth_without_zero(82))
````
