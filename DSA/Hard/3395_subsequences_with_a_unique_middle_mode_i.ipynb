# LeetCode 3395 — Subsequences with a Unique Middle Mode I bằng Python

- Link: https://leetcode.com/problems/subsequences-with-a-unique-middle-mode-i/

- Chủ đề: **Tổ hợp đếm** với dãy con độ dài 5, cố định **phần tử giữa**, chia trường hợp theo số lần giá trị giữa xuất hiện, đếm "ba giá trị đôi một khác nhau" bằng **bao hàm – loại trừ**

Skeleton trên LeetCode (Python3):

```python
class Solution:
```

## 1. Mục tiêu học

1. Biết khi nào một dãy 5 phần tử có **mode duy nhất ở giữa**: giá trị giữa xuất hiện ≥ 3 lần, hoặc đúng 2 lần và ba phần tử còn lại đôi một khác nhau.
2. Cố định chỉ số giữa `i`; chọn 2 chỉ số trái, 2 chỉ số phải **độc lập**.
3. Đếm trường hợp "đúng 2 lần" bằng **bao hàm – loại trừ** trên các cặp có giá trị trùng.
4. Với `n ≤ 1000`, lời giải `O(n · D)` (`D` số giá trị khác nhau) là đủ; biết cách nâng lên `O(n)` (bản II — LC 3416).
5. Kiểm chứng bằng vét cạn `C(n, 5)` với `n` nhỏ.

## 2. Đề bài & ví dụ

Đếm số dãy con (theo chỉ số) độ dài 5 của `nums` mà phần tử ở giữa (`seq[2]`) là **mode duy nhất** — xuất hiện nhiều nhất và không hòa. Kết quả mod `10^9 + 7`.

**Ví dụ 1:** `[1,1,1,1,1,1]` → `6`. **Ví dụ 2:** `[1,2,2,3,3,4]` → `4`. **Ví dụ 3:** `[0,1,2,3,4,5,6,7,8]` → `0`.

**Ràng buộc:** `5 ≤ n ≤ 1000`, `−10^9 ≤ nums[i] ≤ 10^9`.

````python
from typing import List
from collections import Counter
from math import comb
import itertools

MOD = 10 ** 9 + 7
examples = [([1, 1, 1, 1, 1, 1], 6), ([1, 2, 2, 3, 3, 4], 4), (list(range(9)), 0)]
for a, exp in examples:
    print(a, '-> kỳ vọng', exp)
````

## 3. Kiến thức nền

### 3.1. Điều kiện mode duy nhất ở giữa

Gọi `t` = số lần giá trị giữa `x` xuất hiện trong dãy 5 phần tử. `t ≥ 3` ⇒ luôn được (các giá trị khác có ≤ 2 chỗ). `t = 2` ⇒ ba chỗ còn lại phải là ba giá trị **khác nhau**. `t = 1` ⇒ không bao giờ (một giá trị khác cũng có ≥ 1 lần).

````python
def is_good(seq):
    c = Counter(seq)
    m = seq[2]
    return all(c[m] > v for k, v in c.items() if k != m)

tests = [[1, 2, 2, 3, 4], [1, 2, 2, 3, 3], [2, 1, 2, 1, 2], [1, 2, 3, 4, 5], [7, 7, 7, 7, 7]]
print([(t, is_good(t)) for t in tests])
````

### 3.2. Đếm cặp khác giá trị bằng bao hàm – loại trừ

Trong một tập có `R` phần tử với số lần mỗi giá trị `r_v`: số cặp (không thứ tự) khác giá trị `= C(R, 2) − Σ C(r_v, 2)`. Số cặp khác giá trị và không chứa giá trị `v`: trừ thêm `r_v (R − r_v)`.

````python
vals = [5, 5, 6, 7, 7, 7]
cnt = Counter(vals); R = len(vals)
P = comb(R, 2) - sum(comb(c, 2) for c in cnt.values())
direct = sum(1 for i, j in itertools.combinations(range(R), 2) if vals[i] != vals[j])
print('cặp khác giá trị:', P, '=', direct)
````

## 4. Ý tưởng nền tảng / cách nghĩ trực giác

Ví dụ 2 `[1,2,2,3,3,4]`:

- Tâm `i = 2` (giá trị 2): trái `[1, 2]` có một số 2; phải `[3, 3, 4]`. Chọn số 2 trái + số 1 trái (bắt buộc vì chỉ có 2 phần tử) + hai phần tử phải khác nhau và khác 1, 2: cặp `(3, 4)` có `2` cách (hai số 3) ⇒ 2 dãy.
- Tâm `i = 3` (giá trị 3): đối xứng ⇒ 2 dãy.
- Các tâm khác: `t ≤ 1` hoặc không đủ phần tử ⇒ 0.

Tổng 4 ✓.

````python
a = [1, 2, 2, 3, 3, 4]
for idx in itertools.combinations(range(6), 5):
    seq = [a[i] for i in idx]
    print(idx, seq, '✓' if is_good(seq) else '')
````

## 5. Cách 1 — Brute Force: mọi bộ 5 chỉ số

````python
def count_brute(nums):
    return sum(1 for idx in itertools.combinations(range(len(nums)), 5)
               if is_good([nums[i] for i in idx])) % MOD

for a, exp in examples:
    print(count_brute(a), '| kỳ vọng', exp)
````

**Độ phức tạp:** `O(C(n, 5))` ≈ `8·10^{12}` với `n = 1000` — chỉ để đối chiếu với `n ≤ 15`.

## 6. Ý tưởng thuật toán / bước đột phá

Tại tâm `i`, giá trị `x`: bên trái có `a` số `x`, `L'` số khác `x`; bên phải `b` và `R'`.

- **`t ≥ 3`:** `Σ_{j+k ≥ 2} C(a, j) C(L', 2−j) C(b, k) C(R', 2−k)` (j, k ∈ {0, 1, 2}).
- **`t = 2`, thêm một `x` bên trái:** `a` cách chọn `x`, rồi một `y ≠ x` bên trái và cặp `{z1, z2}` bên phải, ba giá trị khác nhau. Với `y` có giá trị `v`: số cặp phải hợp lệ `= P_R − rc[v](R' − rc[v])`, với `P_R` = số cặp khác giá trị bên phải (đã loại `x`).
- **`t = 2`, thêm một `x` bên phải:** đối xứng.

Với `n ≤ 1000`, tính lại `Counter` hai bên cho mỗi tâm (`O(n)` mỗi tâm) ⇒ `O(n²)` tổng — đủ nhanh.

## 7. Các cách tiếp theo

### 7.1. Cách 2 — Theo tâm, duyệt **từng phần tử** `y` bên trái (`O(n²)`)

````python
def C2(x):
    return x * (x - 1) // 2

def count_per_element(nums):
    n, total = len(nums), 0
    for i in range(n):
        x = nums[i]
        left, right = nums[:i], nums[i + 1:]
        lc = Counter(v for v in left if v != x); rc = Counter(v for v in right if v != x)
        a, b = i - sum(lc.values()), len(right) - sum(rc.values())
        Lp, Rp = i - a, len(right) - b
        total += sum(comb(a, j) * comb(Lp, 2 - j) * comb(b, k) * comb(Rp, 2 - k)
                     for j in range(3) for k in range(3) if j + k >= 2)
        PR = C2(Rp) - sum(C2(c) for c in rc.values())
        PL = C2(Lp) - sum(C2(c) for c in lc.values())
        for y in left:                                      # y thuộc bên trái, khác x
            if y != x:
                total += a * (PR - rc[y] * (Rp - rc[y]))
        for z in right:
            if z != x:
                total += b * (PL - lc[z] * (Lp - lc[z]))
    return total % MOD

for a, exp in examples:
    print(count_per_element(a), '| kỳ vọng', exp)
````

### 7.2. Cách 3 — Theo tâm, gộp theo **giá trị** (lời giải chính, `O(n · D)`)

````python
def count_per_value(nums):
    n, total = len(nums), 0
    for i in range(n):
        x = nums[i]
        L, R = Counter(nums[:i]), Counter(nums[i + 1:])
        a, b = L.pop(x, 0), R.pop(x, 0)
        Lp, Rp = i - a, n - 1 - i - b
        total += sum(comb(a, j) * comb(Lp, 2 - j) * comb(b, k) * comb(Rp, 2 - k)
                     for j in range(3) for k in range(3) if j + k >= 2)
        PR = C2(Rp) - sum(C2(c) for c in R.values())
        PL = C2(Lp) - sum(C2(c) for c in L.values())
        total += a * (Lp * PR - sum(c * R.get(v, 0) * (Rp - R.get(v, 0)) for v, c in L.items()))
        total += b * (Rp * PL - sum(c * L.get(v, 0) * (Lp - L.get(v, 0)) for v, c in R.items()))
    return total % MOD

for a, exp in examples:
    print(count_per_value(a), '| kỳ vọng', exp)
````

### 7.3. Cách 4 — Duy trì các tổng khi dịch tâm (`O(n)`, cách của bản II)

Giữ `Σ cl·cr`, `Σ cl·cr²`, `Σ cl²·cr`, `Σ cl²`, `Σ cr²` trên mọi giá trị; khi tâm dịch qua phần tử `x` chỉ số hạng của `x` thay đổi. Các tổng "trừ `x`" lấy từ tổng toàn cục trừ số hạng của `x`.

````python
def count_fast(nums):
    n = len(nums)
    cl, cr = Counter(), Counter(nums)
    S1 = S2 = S4 = SL = 0
    SR = sum(c * c for c in cr.values())
    total = 0
    for i, x in enumerate(nums):
        l, r = cl[x], cr[x]
        S1 -= l * r; S2 -= l * r * r; S4 -= l * l * r; SR -= r * r
        r -= 1; cr[x] = r
        S1 += l * r; S2 += l * r * r; S4 += l * l * r; SR += r * r
        a, b = l, r
        Lp, Rp = i - a, n - 1 - i - b
        ca, cb, cL, cR = (1, a, C2(a)), (1, b, C2(b)), (1, Lp, C2(Lp)), (1, Rp, C2(Rp))
        total += sum(ca[j] * cL[2 - j] * cb[k] * cR[2 - k] for j in range(3) for k in range(3) if j + k >= 2)
        s1, s2, s4 = S1 - a * b, S2 - a * b * b, S4 - a * a * b
        PR = C2(Rp) - (SR - b * b - Rp) // 2
        PL = C2(Lp) - (SL - a * a - Lp) // 2
        total += a * (Lp * PR - (Rp * s1 - s2)) + b * (Rp * PL - (Lp * s1 - s4))
        S1 -= l * r; S2 -= l * r * r; S4 -= l * l * r; SL -= l * l
        l += 1; cl[x] = l
        S1 += l * r; S2 += l * r * r; S4 += l * l * r; SL += l * l
    return total % MOD

for a, exp in examples:
    print(count_fast(a), '| kỳ vọng', exp)
````

## 8. TRACE từng bước

Trace cách 3 trên `[2, 1, 2, 1, 2, 3]`: với mỗi tâm in số dãy con theo từng trường hợp.

````python
def trace(nums):
    n, total = len(nums), 0
    for i in range(n):
        x = nums[i]
        L, R = Counter(nums[:i]), Counter(nums[i + 1:])
        a, b = L.pop(x, 0), R.pop(x, 0)
        Lp, Rp = i - a, n - 1 - i - b
        A = sum(comb(a, j) * comb(Lp, 2 - j) * comb(b, k) * comb(Rp, 2 - k)
                for j in range(3) for k in range(3) if j + k >= 2)
        PR = C2(Rp) - sum(C2(c) for c in R.values()); PL = C2(Lp) - sum(C2(c) for c in L.values())
        B = a * (Lp * PR - sum(c * R.get(v, 0) * (Rp - R.get(v, 0)) for v, c in L.items()))
        Cc = b * (Rp * PL - sum(c * L.get(v, 0) * (Lp - L.get(v, 0)) for v, c in R.items()))
        total += A + B + Cc
        print(f"tâm {i} (x={x}): a={a}, L'={Lp}, b={b}, R'={Rp} -> t>=3: {A}, t=2 trái: {B}, t=2 phải: {Cc}")
    print('tổng:', total, '| vét cạn:', count_brute(nums))

trace([2, 1, 2, 1, 2, 3])
````

## 9. Code nộp LeetCode

````python
class Solution:
    def subsequencesWithMiddleMode(self, nums: List[int]) -> int:
        MOD = 10 ** 9 + 7
        n = len(nums)

        def c2(v):
            return v * (v - 1) // 2

        total = 0
        right = Counter(nums)
        left = Counter()
        for i, x in enumerate(nums):
            right[x] -= 1
            a, b = left[x], right[x]
            Lp, Rp = i - a, n - 1 - i - b
            ca, cb, cL, cR = (1, a, c2(a)), (1, b, c2(b)), (1, Lp, c2(Lp)), (1, Rp, c2(Rp))
            for j in range(3):
                for k in range(3):
                    if j + k >= 2:
                        total += ca[j] * cL[2 - j] * cb[k] * cR[2 - k]
            PR = c2(Rp) - sum(c2(c) for v, c in right.items() if v != x)
            PL = c2(Lp) - sum(c2(c) for v, c in left.items() if v != x)
            sL = 0
            for v, c in left.items():
                if v != x and c:
                    r = right.get(v, 0)
                    sL += c * r * (Rp - r)
            sR = 0
            for v, c in right.items():
                if v != x and c:
                    l = left.get(v, 0)
                    sR += c * l * (Lp - l)
            total += a * (Lp * PR - sL) + b * (Rp * PL - sR)
            left[x] += 1
        return total % MOD

for a, exp in examples:
    got = Solution().subsequencesWithMiddleMode(a)
    print(got, 'OK' if got == exp else 'SAI')
````

## 10. Vì sao thuật toán đúng

1. **Phân hoạch theo tâm:** mỗi dãy con có đúng một chỉ số giữa; với tâm cố định, dãy con ↔ (cặp chỉ số trái, cặp chỉ số phải).
2. **Điều kiện (3.1)** chia đúng ba trường hợp rời nhau: `t ≥ 3`; `t = 2` với `x` thêm ở trái; `t = 2` với `x` thêm ở phải.
3. **`t = 2` trái:** chọn `x` trái (`a` cách) độc lập với việc chọn `y` (trái, ≠ `x`) và `{z1, z2}` (phải, ≠ `x`). Số bộ `(y, {z1, z2})` có ba giá trị khác nhau `= Σ_y (P_R − rc[y](R' − rc[y]))` theo bao hàm – loại trừ (mục 3.2). Gộp theo giá trị `v` của `y` cho cách 3.
4. **Modulo** chỉ lấy cuối vì Python dùng số nguyên lớn (bản C++/Java phải lấy mod từng bước).

## 13. Các lỗi người mới hay gặp

### Lỗi 1 — Coi "giá trị giữa xuất hiện nhiều nhất" là đủ (cho phép hòa)

````python
def wrong_allow_tie(nums):
    def ok(seq):
        c = Counter(seq)
        return all(c[seq[2]] >= v for v in c.values())       # SAI: >= cho phép hòa
    return sum(1 for idx in itertools.combinations(range(len(nums)), 5) if ok([nums[i] for i in idx]))

print('sai :', wrong_allow_tie([1, 2, 2, 3, 3, 4]), '| đúng:', Solution().subsequencesWithMiddleMode([1, 2, 2, 3, 3, 4]))
````

**Vì sao sai:** đề yêu cầu mode **duy nhất**: `[1, 2, 2, 3, 3]` có hai mode (2 và 3).

### Lỗi 2 — Quên trường hợp `t = 2`

````python
def wrong_only_ge3(nums):
    n, total = len(nums), 0
    for i in range(n):
        x = nums[i]
        L, R = Counter(nums[:i]), Counter(nums[i + 1:])
        a, b = L[x], R[x]; Lp, Rp = i - a, n - 1 - i - b
        total += sum(comb(a, j) * comb(Lp, 2 - j) * comb(b, k) * comb(Rp, 2 - k)
                     for j in range(3) for k in range(3) if j + k >= 2)
    return total % MOD

print('sai :', wrong_only_ge3([1, 2, 2, 3, 3, 4]), '| đúng:', Solution().subsequencesWithMiddleMode([1, 2, 2, 3, 3, 4]))
````

**Vì sao sai:** giá trị giữa chỉ cần xuất hiện 2 lần nếu ba phần tử còn lại khác nhau.

### Lỗi 3 — Trường hợp `t = 2` chỉ yêu cầu `y`, `z1`, `z2` khác `x` (không yêu cầu đôi một khác nhau)

````python
def wrong_t2_not_distinct(nums):
    n, total = len(nums), 0
    for i in range(n):
        x = nums[i]
        L, R = Counter(nums[:i]), Counter(nums[i + 1:])
        a, b = L[x], R[x]; Lp, Rp = i - a, n - 1 - i - b
        total += sum(comb(a, j) * comb(Lp, 2 - j) * comb(b, k) * comb(Rp, 2 - k)
                     for j in range(3) for k in range(3) if j + k >= 2)
        total += a * Lp * C2(Rp) + b * Rp * C2(Lp)           # SAI
    return total % MOD

print('sai :', wrong_t2_not_distinct([1, 2, 2, 3, 3, 4]), '| đúng:', Solution().subsequencesWithMiddleMode([1, 2, 2, 3, 3, 4]))
````

**Vì sao sai:** nếu hai trong ba phần tử còn lại cùng giá trị thì giá trị đó cũng xuất hiện 2 lần ⇒ hòa.

### Lỗi 4 — Bỏ trường hợp `j = 2, k = 0` (hoặc `j = 0, k = 2`) khi đếm `t ≥ 3`

````python
def wrong_ge3_symmetric_only(nums):
    n, total = len(nums), 0
    for i in range(n):
        x = nums[i]
        L, R = Counter(nums[:i]), Counter(nums[i + 1:])
        a, b = L[x], R[x]; Lp, Rp = i - a, n - 1 - i - b
        total += sum(comb(a, j) * comb(Lp, 2 - j) * comb(b, k) * comb(Rp, 2 - k)
                     for j in range(1, 3) for k in range(1, 3))   # SAI: thiếu (2, 0), (0, 2)
    return total % MOD                     # (bỏ cả phần t = 2 để tập trung vào lỗi đếm t >= 3)

print('sai :', wrong_ge3_symmetric_only([1, 1, 1, 2, 3]), '| đúng:', Solution().subsequencesWithMiddleMode([1, 1, 1, 2, 3]))
````

**Vì sao sai:** ba số `x` có thể nằm lệch một bên (hai `x` bên trái và tâm), ví dụ `[1, 1, 1, 2, 3]`.

## 14. Mẹo làm bài nhanh & nhận dạng pattern

- Dãy con độ dài nhỏ cố định ⇒ cố định vị trí đặc biệt (giữa), đếm hai bên độc lập.
- "Mode duy nhất" ⇒ chia theo số lần xuất hiện của mode.
- "Đôi một khác nhau" ⇒ tổng trừ các cặp trùng (bao hàm – loại trừ).
- `n ≤ 1000` ⇒ `O(n · D)` hay `O(n²)` đều được; bản `10^5` cần duy trì tổng.

## 15. Biến thể của bài

1. **LC 3416 — bản II** (`n ≤ 10^5`, Premium): cách 4.
2. Độ dài 3 ⇒ chỉ cần `t ≥ 2`.
3. Đếm dãy con 5 phần tử mà giá trị giữa là **mode** (cho phép hòa) ⇒ thêm trường hợp `t = 2` với đúng một giá trị khác xuất hiện 2 lần, hoặc `t = 1` khi mọi giá trị khác xuất hiện 1 lần.
4. Mode duy nhất ở **vị trí bất kỳ**.

````python
# Biến thể 3: giá trị giữa là A mode (cho phép hòa) — vét cạn có cấu trúc để so
def count_mode_allow_tie(nums):
    n = len(nums)
    def ok(seq):
        c = Counter(seq)
        return c[seq[2]] == max(c.values())
    return sum(1 for idx in itertools.combinations(range(n), 5) if ok([nums[i] for i in idx]))

a = [1, 2, 2, 3, 3, 4]
print('cho phép hòa:', count_mode_allow_tie(a), '| mode duy nhất:', Solution().subsequencesWithMiddleMode(a))
````

````python
# Bài 1, 2, 3
assert Solution().subsequencesWithMiddleMode([4] * 7) == 21
assert Solution().subsequencesWithMiddleMode([1, 2, 1, 3, 1]) == 1 == count_brute([1, 2, 1, 3, 1])

def count_len3(nums):
    n, total = len(nums), 0
    L, R = Counter(), Counter(nums)
    for i, x in enumerate(nums):
        R[x] -= 1
        total += i * (n - 1 - i) - (i - L[x]) * (n - 1 - i - R[x])
        L[x] += 1
    return total

def brute_len3(nums):
    def ok(seq):
        c = Counter(seq); m = seq[1]
        return all(c[m] > v for k, v in c.items() if k != m)
    return sum(1 for idx in itertools.combinations(range(len(nums)), 3) if ok([nums[i] for i in idx]))

rng = random.Random(3)
for _ in range(400):
    a = [rng.randint(0, 3) for _ in range(rng.randint(3, 10))]
    assert count_len3(a) == brute_len3(a)
print('Bài 1: 21 | Bài 2: 1 | Bài 3 khớp vét cạn.')
````

````python
# Bài 4, 5
INV2 = pow(2, MOD - 2, MOD)

def count_per_value_mod(nums):
    n, total = len(nums), 0
    c2m = lambda v: v * (v - 1) % MOD * INV2 % MOD
    for i in range(n):
        x = nums[i]
        L, R = Counter(nums[:i]), Counter(nums[i + 1:])
        a, b = L.pop(x, 0), R.pop(x, 0)
        Lp, Rp = i - a, n - 1 - i - b
        ca, cb, cL, cR = (1, a, c2m(a)), (1, b, c2m(b)), (1, Lp, c2m(Lp)), (1, Rp, c2m(Rp))
        for j in range(3):
            for k in range(3):
                if j + k >= 2:
                    total = (total + ca[j] * cL[2 - j] % MOD * cb[k] % MOD * cR[2 - k]) % MOD
        PR = (c2m(Rp) - sum(c2m(c) for c in R.values())) % MOD
        PL = (c2m(Lp) - sum(c2m(c) for c in L.values())) % MOD
        sL = sum(c * R.get(v, 0) % MOD * (Rp - R.get(v, 0)) for v, c in L.items()) % MOD
        sR = sum(c * L.get(v, 0) % MOD * (Lp - L.get(v, 0)) for v, c in R.items()) % MOD
        total = (total + a * ((Lp * PR - sL) % MOD) + b * ((Rp * PL - sR) % MOD)) % MOD
    return total

def count_unique_mode_any(nums):
    def ok(seq):
        c = sorted(Counter(seq).values(), reverse=True)
        return len(c) == 1 or c[0] > c[1]
    return sum(1 for idx in itertools.combinations(range(len(nums)), 5) if ok([nums[i] for i in idx]))

rng = random.Random(5)
for _ in range(300):
    a = [rng.randint(0, 4) for _ in range(rng.randint(5, 12))]
    assert count_per_value_mod(a) == count_per_value(a)
big = [rng.randint(1, 20) for _ in range(400)]
assert count_per_value_mod(big) == Solution().subsequencesWithMiddleMode(big)
print('Bài 4 khớp. Bài 5 (vét cạn) với [1,1,2,2,3,1,4]:', count_unique_mode_any([1, 1, 2, 2, 3, 1, 4]))
````
