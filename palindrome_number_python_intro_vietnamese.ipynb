Mục tiêu: hiểu bài toán **số đối xứng / palindrome number**, biết nhiều cách làm từ dễ đến tối ưu, nắm mẹo xử lý edge case và tự luyện thêm bài tập.

## 1. Bài toán yêu cầu gì?

Ta cần kiểm tra một số nguyên `x` có phải là **palindrome** hay không.

Một số là palindrome nếu đọc từ trái sang phải hay từ phải sang trái đều giống nhau.

Ví dụ:

| x | Kết quả | Vì sao? |
|---:|:---:|---|
| `121` | `True` | đọc xuôi/ngược đều là 121 |
| `1221` | `True` | đọc xuôi/ngược đều là 1221 |
| `123` | `False` | đọc ngược thành 321 |
| `-121` | `False` | dấu `-` làm số đọc ngược không giống ban đầu |
| `10` | `False` | đọc ngược thành 01, tức là 1 |
| `0` | `True` | một chữ số luôn đối xứng |

## 2. Ý tưởng nền tảng

Có 3 hướng phổ biến:

1. **Đổi số thành chuỗi** rồi so sánh với chuỗi đảo ngược.
2. **Đảo toàn bộ số bằng toán học** rồi so sánh với số ban đầu.
3. **Chỉ đảo một nửa số** để tiết kiệm hơn, đây là cách rất hay cho bài này.

Với người mới, nên học theo thứ tự:

`Chuỗi` → `Đảo số bằng toán` → `Đảo nửa số tối ưu`.

## 3. Cách 1 — Đổi số thành chuỗi

Đây là cách dễ hiểu nhất.

Ý tưởng:

```python
s = str(x)
return s == s[::-1]
```

Trong Python:

- `str(x)` đổi số thành chuỗi.
- `s[::-1]` tạo chuỗi đảo ngược.
- Nếu chuỗi ban đầu bằng chuỗi đảo ngược thì số là palindrome.

````python
def is_palindrome_string(x: int) -> bool:
    s = str(x)
    return s == s[::-1]

print(is_palindrome_string(121))
print(is_palindrome_string(123))
print(is_palindrome_string(-121))
print(is_palindrome_string(0))
````

### Giải thích `s[::-1]`

Cú pháp cắt chuỗi tổng quát:

```python
s[start:end:step]
```

Khi viết:

```python
s[::-1]
```

nghĩa là:

- Không ghi `start`: lấy từ đầu/đuôi tùy chiều duyệt.
- Không ghi `end`: lấy đến hết.
- `step = -1`: đi ngược lại.

Nên `"121"[::-1]` cho ra `"121"`, còn `"123"[::-1]` cho ra `"321"`.

````python
s1 = "121"
s2 = "123"

print(s1, "->", s1[::-1])
print(s2, "->", s2[::-1])
````

### Code nộp LeetCode — Cách chuỗi

Cách này thường được chấp nhận, rất ngắn và dễ nhớ.

````python
class Solution:
    def isPalindrome(self, x: int) -> bool:
        s = str(x)
        return s == s[::-1]
````

- Bộ nhớ: `O(n)`, vì tạo chuỗi `s` và chuỗi đảo ngược.

Ưu điểm: cực dễ viết.

Nhược điểm: dùng thêm bộ nhớ chuỗi, không thuần toán học.

## 4. Cách 2 — Đảo toàn bộ số bằng toán học

Bây giờ ta không đổi sang chuỗi nữa.

Ví dụ cần đảo số `123` thành `321`.

Ta lấy từng chữ số cuối bằng `% 10`:

```python
123 % 10 = 3
12 % 10 = 2
1 % 10 = 1
```

Muốn bỏ chữ số cuối thì dùng `// 10`:

```python
123 // 10 = 12
12 // 10 = 1
1 // 10 = 0
```

Công thức đảo số:

```python
reversed_num = reversed_num * 10 + last_digit
```

Ví dụ:

```text
ban đầu reversed_num = 0
lấy 3: reversed_num = 0 * 10 + 3 = 3
lấy 2: reversed_num = 3 * 10 + 2 = 32
lấy 1: reversed_num = 32 * 10 + 1 = 321
```

````python
def reverse_number(x: int) -> int:
    reversed_num = 0
    while x > 0:
        last_digit = x % 10
        reversed_num = reversed_num * 10 + last_digit
        x //= 10
    return reversed_num

print(reverse_number(123))
print(reverse_number(1221))
````

### Áp dụng vào kiểm tra palindrome

Ta lưu lại số ban đầu vào biến `original`, sau đó đảo số `x`.

Nếu `original == reversed_num` thì số đối xứng.

Chú ý: số âm luôn không phải palindrome.

````python
def is_palindrome_math_full_reverse(x: int) -> bool:
    if x < 0:
        return False
    
    original = x
    reversed_num = 0
    
    while x > 0:
        last_digit = x % 10
        reversed_num = reversed_num * 10 + last_digit
        x //= 10
    
    return original == reversed_num

for value in [121, 1221, 123, -121, 10, 0]:
    print(value, "->", is_palindrome_math_full_reverse(value))
````

### Code nộp LeetCode — Đảo toàn bộ số

````python
class Solution:
    def isPalindrome(self, x: int) -> bool:
        if x < 0:
            return False
        
        original = x
        reversed_num = 0
        
        while x > 0:
            last_digit = x % 10
            reversed_num = reversed_num * 10 + last_digit
            x //= 10
        
        return original == reversed_num
````

### Độ phức tạp của cách đảo toàn bộ số

Nếu `x` có `n` chữ số:

- Thời gian: `O(n)`.
- Bộ nhớ: `O(1)`.

Ưu điểm: không dùng chuỗi.

Nhược điểm: với một số ngôn ngữ như C++/Java, đảo toàn bộ số có thể gặp vấn đề tràn số. Python thì số nguyên lớn linh hoạt hơn, nhưng trong phỏng vấn vẫn nên biết cách tối ưu hơn.

## 5. Cách 3 — Đảo một nửa số, kỹ thuật tối ưu nhất

Đây là kỹ thuật rất hay của bài này.

Thay vì đảo toàn bộ số, ta chỉ đảo **nửa sau** của số.

Ví dụ `1221`:

```text
x = 1221, reverted_half = 0

Lấy 1 từ cuối:
x = 122, reverted_half = 1

Lấy 2 từ cuối:
x = 12, reverted_half = 12

Dừng vì x <= reverted_half.
So sánh: x == reverted_half, tức 12 == 12 → True
```

Ví dụ `12321`:

```text
x = 12321, reverted_half = 0

Lấy 1:
x = 1232, reverted_half = 1

Lấy 2:
x = 123, reverted_half = 12

Lấy 3:
x = 12, reverted_half = 123

Dừng vì x <= reverted_half.
Vì số chữ số lẻ, chữ số giữa là 3 không cần so sánh.
So sánh: x == reverted_half // 10
12 == 123 // 10 == 12 → True
```

### Tại sao điều kiện dừng là `x > reverted_half`?

Ta đang bóc chữ số từ cuối của `x` và đưa sang `reverted_half`.

Ban đầu:

```text
x chứa toàn bộ số
reverted_half rỗng
```

Mỗi vòng lặp:

```python
reverted_half = reverted_half * 10 + x % 10
x //= 10
```

Tức là:

- `x` mất dần chữ số ở cuối.
- `reverted_half` nhận dần các chữ số đó theo thứ tự đảo ngược.

Khi `reverted_half` đã dài bằng hoặc dài hơn `x`, nghĩa là ta đã đi đến giữa số.

Vì vậy vòng lặp chạy khi:

```python
while x > reverted_half:
```

````python
def trace_half_reverse(x: int) -> None:
    print(f"Kiểm tra x = {x}")
    reverted_half = 0
    step = 1
    
    while x > reverted_half:
        last_digit = x % 10
        reverted_half = reverted_half * 10 + last_digit
        x //= 10
        print(f"Bước {step}: x = {x}, reverted_half = {reverted_half}")
        step += 1
    
    print("So sánh chẵn:", x, "==", reverted_half, "->", x == reverted_half)
    print("So sánh lẻ:", x, "==", reverted_half // 10, "->", x == reverted_half // 10)

trace_half_reverse(1221)
print()
trace_half_reverse(12321)
````

### Xử lý edge case trước khi đảo nửa số

Có 2 trường hợp cần loại sớm:

```python
if x < 0:
    return False
```

Vì số âm có dấu `-`, ví dụ `-121` đọc ngược không còn giống ban đầu.

```python
if x % 10 == 0 and x != 0:
    return False
```

Vì số tận cùng bằng 0 nhưng khác 0 thì không thể palindrome.

Ví dụ:

- `10` đọc ngược thành `01`, tức là `1`, không giống `10`.
- `100` đọc ngược thành `001`, tức là `1`, không giống `100`.

Riêng `0` là palindrome, nên phải có điều kiện `x != 0`.

### Code nộp LeetCode — Đảo một nửa số

````python
class Solution:
    def isPalindrome(self, x: int) -> bool:
        if x < 0:
            return False
        
        if x % 10 == 0 and x != 0:
            return False
        
        reverted_half = 0
        
        while x > reverted_half:
            reverted_half = reverted_half * 10 + x % 10
            x //= 10
        
        return x == reverted_half or x == reverted_half // 10
````

### Vì sao return có 2 trường hợp?

```python
return x == reverted_half or x == reverted_half // 10
```

Có 2 loại độ dài số:

### Trường hợp 1: Số có số chữ số chẵn

Ví dụ `1221`:

```text
x = 12
reverted_half = 12
```

Ta cần:

```python
x == reverted_half
```

### Trường hợp 2: Số có số chữ số lẻ

Ví dụ `12321`:

```text
x = 12
reverted_half = 123
```

Chữ số giữa `3` không cần so sánh, nên bỏ nó bằng `// 10`:

```python
reverted_half // 10 = 12
```

Ta cần:

```python
x == reverted_half // 10
```

### Độ phức tạp của cách đảo nửa số

Nếu `x` có `n` chữ số:

- Thời gian: `O(n)`, chính xác hơn là chạy khoảng `n / 2` vòng.
- Bộ nhớ: `O(1)`.

Đây là cách rất tốt để trình bày trong phỏng vấn vì:

- Không dùng chuỗi.
- Không đảo toàn bộ số.
- Tránh rủi ro tràn số ở các ngôn ngữ giới hạn kiểu số nguyên.

## 6. Bộ test kiểm tra

Ta nên test các nhóm sau:

1. Số dương palindrome: `121`, `1221`, `12321`.
2. Số dương không palindrome: `123`, `1231`.
3. Số âm: `-121`, `-1`.
4. Số tận cùng bằng 0: `10`, `100`.
5. Số một chữ số: `0`, `5`, `9`.

````python
def is_palindrome_half_reverse(x: int) -> bool:
    if x < 0:
        return False
    
    if x % 10 == 0 and x != 0:
        return False
    
    reverted_half = 0
    
    while x > reverted_half:
        reverted_half = reverted_half * 10 + x % 10
        x //= 10
    
    return x == reverted_half or x == reverted_half // 10


test_cases = [
    (121, True),
    (1221, True),
    (12321, True),
    (123, False),
    (1231, False),
    (-121, False),
    (-1, False),
    (10, False),
    (100, False),
    (0, True),
    (5, True),
    (9, True),
]

for x, expected in test_cases:
    result = is_palindrome_half_reverse(x)
    print(f"x = {x:>6} | result = {result:<5} | expected = {expected}")
    assert result == expected

print("Tất cả test đều PASS!")
````

## 8. Các lỗi người mới hay gặp

### Lỗi 1: Quên xử lý số âm

Sai:

```python
return str(x) == str(x)[::-1]
```

Thật ra với Python, `str(-121)` là `"-121"`, đảo lại là `"121-"`, nên vẫn ra False. Nhưng nếu dùng cách toán học mà quên số âm thì dễ lỗi logic.

### Lỗi 2: Quên lưu `original`

Sai:

```python
while x > 0:
    ...
return x == reversed_num
```

Sau vòng lặp, `x` đã bị chia về 0, nên không còn là số ban đầu nữa.

Phải làm:

```python
original = x
...
return original == reversed_num
```

### Lỗi 3: Không hiểu `//` khác `/`

Trong Python:

```python
123 / 10   # 12.3
123 // 10  # 12
```

Khi xử lý chữ số, ta cần `// 10`, không phải `/ 10`.

### Lỗi 4: Quên trường hợp `x = 0`

`0` là palindrome.

Nếu viết điều kiện:

```python
if x % 10 == 0:
    return False
```

thì `0` bị trả về False sai.

Phải viết:

```python
if x % 10 == 0 and x != 0:
    return False
```


## 13. Checklist tự kiểm tra trước khi submit

Trước khi nộp, hãy tự hỏi:

- Mình có xử lý `x < 0` chưa?
- Mình có xử lý `x = 0` đúng chưa?
- Mình có xử lý số tận cùng bằng 0 như `10`, `100` chưa?
- Nếu dùng đảo toàn bộ số, mình có lưu `original` chưa?
- Nếu dùng toán học, mình có dùng `// 10` thay vì `/ 10` chưa?
- Nếu dùng đảo nửa số, mình có xét cả số chữ số chẵn và lẻ chưa?
